```python
# ============================================================
# Section 2.4 - Reproducibility Configuration
# ============================================================

if not section_2_3_complete:
    raise RuntimeError(
        "Section 2.3 must be completed before configuring reproducibility."
    )

# ============================================================
# Fixed reproducibility seed
# ============================================================

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)

# Reusable random-state value for later sklearn methods
RANDOM_STATE = RANDOM_SEED

# ============================================================
# Reproducibility validation
# ============================================================

test_values_run_1 = np.random.RandomState(
    RANDOM_SEED
).rand(5)

test_values_run_2 = np.random.RandomState(
    RANDOM_SEED
).rand(5)

random_seed_reproducible = bool(
    np.allclose(
        test_values_run_1,
        test_values_run_2
    )
)

reproducibility_settings = pd.DataFrame(
    [
        {
            "Configuration": "Random seed",
            "Configured Value": RANDOM_SEED,
            "Validated": True,
        },
        {
            "Configuration": "NumPy random seed",
            "Configured Value": RANDOM_SEED,
            "Validated": True,
        },
        {
            "Configuration": "Reusable random state",
            "Configured Value": RANDOM_STATE,
            "Validated": True,
        },
        {
            "Configuration": "Repeated random output consistency",
            "Configured Value": "Matched",
            "Validated": random_seed_reproducible,
        },
    ]
)

print("PMIP Release Performance Intelligence reproducibility configuration")
print("=" * 80)

display(reproducibility_settings)

# ============================================================
# Final Section 2 validation
# ============================================================

section_2_4_complete = bool(
    reproducibility_settings["Validated"].all()
)

section_2_complete = bool(
    section_2_1_complete
    and section_2_2_complete
    and section_2_3_complete
    and section_2_4_complete
)

print()
print(f"Section 2.4 completion status: {section_2_4_complete}")
print(f"Section 2 overall completion status: {section_2_complete}")

if not section_2_complete:
    raise RuntimeError(
        "Section 2 Environment and Configuration did not complete successfully."
    )
```

## 3.1 Available PMIP Datasets

### Purpose

This subsection identifies the datasets currently available within the PMIP project that may support Release Performance Intelligence.

The goal is to build a clear inventory of candidate sources before selecting any dataset for detailed analysis.

The discovery process will inspect the main PMIP data directories and record:

- dataset filename;
- storage location;
- file type;
- approximate file size;
- likely analytical stage;
- whether the dataset can be loaded successfully.

This provides an initial view of the data landscape and helps identify which datasets should be examined further in the next subsections.

```python
# ============================================================
# Section 3.1 - Available PMIP Datasets
# ============================================================

if not section_2_complete:
    raise RuntimeError(
        "Section 2 must be completed before beginning data discovery."
    )

# ============================================================
# Directories to inspect
# ============================================================

dataset_directories = {
    "Raw": RAW_DATA_DIR,
    "Processed": PROCESSED_DATA_DIR,
    "Integrated": INTEGRATED_DATA_DIR,
    "Model Ready": MODEL_READY_DATA_DIR,
}

# ============================================================
# Discover supported dataset files
# ============================================================

supported_extensions = {
    ".csv",
    ".parquet",
    ".json",
    ".xlsx",
    ".xls",
    ".feather",
    ".pkl",
    ".pickle",
}

dataset_inventory = []

for stage, directory in dataset_directories.items():

    if not directory.exists():
        continue

    for file_path in sorted(directory.rglob("*")):

        if (
            file_path.is_file()
            and file_path.suffix.lower() in supported_extensions
        ):

            file_size_mb = file_path.stat().st_size / (1024 ** 2)

            dataset_inventory.append(
                {
                    "Dataset": file_path.name,
                    "Stage": stage,
                    "Extension": file_path.suffix.lower(),
                    "Size (MB)": round(file_size_mb, 4),
                    "Relative Path": str(
                        file_path.relative_to(PROJECT_ROOT)
                    ),
                    "Exists": file_path.exists(),
                }
            )

available_pmip_datasets_df = pd.DataFrame(dataset_inventory)

# ============================================================
# Sort discovered datasets
# ============================================================

if not available_pmip_datasets_df.empty:

    available_pmip_datasets_df = (
        available_pmip_datasets_df
        .sort_values(
            by=["Stage", "Dataset"]
        )
        .reset_index(drop=True)
    )

# ============================================================
# Dataset discovery summary
# ============================================================

print("Discovering available PMIP datasets")
print("=" * 80)

if available_pmip_datasets_df.empty:

    print("No supported dataset files were discovered.")

else:

    display(available_pmip_datasets_df)

    print()
    print("Dataset discovery summary")
    print("=" * 80)

    discovery_summary_df = (
        available_pmip_datasets_df
        .groupby("Stage")
        .agg(
            Datasets=("Dataset", "count"),
            Total_Size_MB=("Size (MB)", "sum"),
        )
        .reset_index()
    )

    display(discovery_summary_df)

# ============================================================
# Section validation
# ============================================================

section_3_1_complete = bool(
    not available_pmip_datasets_df.empty
    and available_pmip_datasets_df["Exists"].all()
)

print()
print(
    f"Datasets discovered: "
    f"{len(available_pmip_datasets_df):,}"
)

print(
    f"Section 3.1 completion status: "
    f"{section_3_1_complete}"
)

if not section_3_1_complete:
    raise RuntimeError(
        "No usable PMIP datasets were discovered in the configured data directories."
    )
```

### Interpretation

The PMIP data-discovery stage identified 10 available datasets across the raw, processed, integrated and model-ready project directories.

The discovered sources include chart data, Spotify data, listener data, integrated data and previously generated PMIP artist-momentum outputs.

The largest available source is `charts_cleaned.csv`, indicating that it may contain the richest historical chart evidence for release-level analysis. The integrated Spotify dataset and cleaned Spotify dataset are also important candidates because they may contain track, release, artist, popularity or streaming-related attributes.

The artist-momentum result files appear to represent downstream artist-level intelligence rather than primary release-level observations. These may therefore be more suitable for later artist-relative comparison rather than initial release construction.

Section 3.1 completed successfully with all 10 discovered files confirmed as available.

The next step is to assess which of these datasets are genuine candidates for constructing release-level intelligence.

## 3.2 Candidate Release-Level Sources

### Purpose

This subsection evaluates the discovered PMIP datasets to identify which sources are plausible candidates for release-level analysis.

The aim is not yet to select the final dataset, but to narrow the available sources based on whether they are likely to contain information relevant to individual music releases.

The assessment considers whether each dataset may contain:

- track or release identifiers;
- artist information;
- release-date or temporal fields;
- streaming or popularity measures;
- chart-performance evidence;
- country or market information;
- historical observations;
- previously derived PMIP intelligence.

Datasets that are clearly artist-level, summary-level or unrelated to release behaviour will be retained only as possible supporting sources rather than primary release-level inputs.

```python
# ============================================================
# Section 3.2 - Candidate Release-Level Sources
# ============================================================

if not section_3_1_complete:
    raise RuntimeError(
        "Section 3.1 must be completed before assessing candidate release-level sources."
    )

# ============================================================
# Candidate-source rules
# ============================================================

candidate_source_notes = {
    "charts_cleaned.csv": {
        "Candidate Role": "Primary candidate",
        "Likely Use": (
            "Historical chart performance, country-level activity, "
            "chart persistence and repeated observations."
        ),
        "Priority": "High",
    },

    "spotify_integrated.csv": {
        "Candidate Role": "Primary candidate",
        "Likely Use": (
            "Integrated Spotify track, artist, release and "
            "streaming-related information."
        ),
        "Priority": "High",
    },

    "spotify_2024_cleaned.csv": {
        "Candidate Role": "Primary / supporting candidate",
        "Likely Use": (
            "Recent Spotify track and release attributes that may support "
            "release metadata and current-performance analysis."
        ),
        "Priority": "High",
    },

    "pmip_ml_ready_features.csv": {
        "Candidate Role": "Supporting candidate",
        "Likely Use": (
            "Previously engineered PMIP features that may provide "
            "additional performance or modelling evidence."
        ),
        "Priority": "Medium",
    },

    "listeners_cleaned.csv": {
        "Candidate Role": "Supporting candidate",
        "Likely Use": (
            "Listener-related evidence where compatible release, track "
            "or artist identifiers are available."
        ),
        "Priority": "Medium",
    },

    "listeners.csv": {
        "Candidate Role": "Reference candidate",
        "Likely Use": (
            "Raw listener source retained mainly for traceability because "
            "a cleaned version is already available."
        ),
        "Priority": "Low",
    },

    "pmip_artist_momentum_results.csv": {
        "Candidate Role": "Artist-level supporting source",
        "Likely Use": (
            "Artist momentum intelligence for later comparison between "
            "release performance and broader artist performance."
        ),
        "Priority": "Medium",
    },

    "pmip_artist_momentum_ranking.csv": {
        "Candidate Role": "Artist-level supporting source",
        "Likely Use": (
            "Artist-level ranking evidence for contextual comparison."
        ),
        "Priority": "Low",
    },

    "pmip_indicator_contributions.csv": {
        "Candidate Role": "Artist-level supporting source",
        "Likely Use": (
            "Interpretability context for existing artist momentum scores."
        ),
        "Priority": "Low",
    },

    "pmip_top_100_momentum_artists.csv": {
        "Candidate Role": "Artist-level summary source",
        "Likely Use": (
            "High-level artist momentum shortlist; not expected to provide "
            "complete release-level coverage."
        ),
        "Priority": "Low",
    },
}

# ============================================================
# Build candidate assessment table
# ============================================================

candidate_rows = []

for _, row in available_pmip_datasets_df.iterrows():

    dataset_name = row["Dataset"]

    notes = candidate_source_notes.get(
        dataset_name,
        {
            "Candidate Role": "Unclassified",
            "Likely Use": "Requires further schema inspection.",
            "Priority": "Unknown",
        },
    )

    candidate_rows.append(
        {
            "Dataset": dataset_name,
            "Stage": row["Stage"],
            "Size (MB)": row["Size (MB)"],
            "Candidate Role": notes["Candidate Role"],
            "Likely Use": notes["Likely Use"],
            "Priority": notes["Priority"],
        }
    )

candidate_release_sources_df = pd.DataFrame(candidate_rows)

# ============================================================
# Priority ordering
# ============================================================

priority_order = {
    "High": 1,
    "Medium": 2,
    "Low": 3,
    "Unknown": 4,
}

candidate_release_sources_df["Priority Order"] = (
    candidate_release_sources_df["Priority"]
    .map(priority_order)
    .fillna(4)
)

candidate_release_sources_df = (
    candidate_release_sources_df
    .sort_values(
        by=["Priority Order", "Dataset"]
    )
    .drop(columns="Priority Order")
    .reset_index(drop=True)
)

# ============================================================
# Display candidate assessment
# ============================================================

print("Assessing candidate PMIP release-level sources")
print("=" * 80)

display(candidate_release_sources_df)

# ============================================================
# Candidate summary
# ============================================================

candidate_summary_df = (
    candidate_release_sources_df
    .groupby("Priority")
    .agg(
        Datasets=("Dataset", "count")
    )
    .reset_index()
)

print()
print("Candidate release-source summary")
print("=" * 80)

display(candidate_summary_df)

# ============================================================
# Primary candidates
# ============================================================

primary_release_candidates = (
    candidate_release_sources_df[
        candidate_release_sources_df["Priority"] == "High"
    ]["Dataset"]
    .tolist()
)

print()
print("Primary release-level candidates")
print("=" * 80)

for dataset_name in primary_release_candidates:
    print(f"- {dataset_name}")

# ============================================================
# Section validation
# ============================================================

section_3_2_complete = bool(
    len(candidate_release_sources_df) > 0
    and len(primary_release_candidates) > 0
)

print()
print(f"Primary candidates identified: {len(primary_release_candidates)}")
print(f"Section 3.2 completion status: {section_3_2_complete}")

if not section_3_2_complete:
    raise RuntimeError(
        "No primary release-level candidate datasets were identified."
    )
```

### Interpretation

The candidate-source assessment successfully narrowed the available PMIP datasets into primary, supporting and reference sources for Release Performance Intelligence.

Three datasets were identified as the strongest primary candidates:

- `charts_cleaned.csv`
- `spotify_2024_cleaned.csv`
- `spotify_integrated.csv`

`charts_cleaned.csv` is expected to provide the strongest historical chart and market-level evidence, while the Spotify datasets are likely to provide release metadata, artist information and current track-level performance attributes.

The listener and model-ready datasets remain potentially useful as supporting evidence, but their suitability depends on schema compatibility with the final release-level structure.

The existing artist-momentum outputs are not primary release datasets. They are more suitable for later comparison between release performance and broader artist-level momentum.

Section 3.2 completed successfully with three high-priority release-level candidates identified.

The next step is to inspect the dimensions of these candidate datasets before examining their schemas in detail.

## 3.3 Dataset Shapes

### Purpose

This subsection measures the dimensions of the primary candidate datasets identified in Section 3.2.

The aim is to understand the scale and structural size of each source before performing detailed schema inspection.

For each candidate dataset, the notebook will determine:

- number of rows;
- number of columns;
- approximate memory usage;
- whether the dataset loads successfully;
- whether the dataset contains enough observations to support further release-level analysis.

Dataset shape alone does not determine suitability, but it provides useful context about the population size and complexity of each candidate source.

```python
# ============================================================
# Section 3.3 - Dataset Shapes
# ============================================================

if not section_3_2_complete:
    raise RuntimeError(
        "Section 3.2 must be completed before inspecting dataset shapes."
    )

# ============================================================
# Helper function
# ============================================================

def load_candidate_dataset(file_path):
    """
    Load a candidate dataset based on its file extension.
    """

    extension = file_path.suffix.lower()

    if extension == ".csv":
        return pd.read_csv(file_path, low_memory=False)

    elif extension == ".parquet":
        return pd.read_parquet(file_path)

    elif extension in {".xlsx", ".xls"}:
        return pd.read_excel(file_path)

    elif extension == ".json":
        return pd.read_json(file_path)

    elif extension == ".feather":
        return pd.read_feather(file_path)

    elif extension in {".pkl", ".pickle"}:
        return pd.read_pickle(file_path)

    else:
        raise ValueError(
            f"Unsupported dataset format: {extension}"
        )


# ============================================================
# Locate primary candidate files
# ============================================================

candidate_dataset_objects = {}
dataset_shape_rows = []

print("Inspecting primary PMIP candidate dataset shapes")
print("=" * 80)

for dataset_name in primary_release_candidates:

    matched_rows = available_pmip_datasets_df[
        available_pmip_datasets_df["Dataset"] == dataset_name
    ]

    if matched_rows.empty:
        dataset_shape_rows.append(
            {
                "Dataset": dataset_name,
                "Rows": np.nan,
                "Columns": np.nan,
                "Memory (MB)": np.nan,
                "Loaded": False,
            }
        )
        continue

    relative_path = matched_rows.iloc[0]["Relative Path"]
    file_path = PROJECT_ROOT / relative_path

    try:
        df = load_candidate_dataset(file_path)

        variable_name = (
            dataset_name
            .replace(".csv", "")
            .replace(".parquet", "")
            .replace(".xlsx", "")
            .replace(".xls", "")
            .replace(".json", "")
            .replace(".feather", "")
            .replace(".pkl", "")
            .replace(".pickle", "")
            + "_df"
        )

        candidate_dataset_objects[variable_name] = df

        memory_mb = (
            df.memory_usage(deep=True).sum()
            / (1024 ** 2)
        )

        dataset_shape_rows.append(
            {
                "Dataset": dataset_name,
                "Rows": len(df),
                "Columns": len(df.columns),
                "Memory (MB)": round(memory_mb, 3),
                "Loaded": True,
            }
        )

    except Exception as error:

        dataset_shape_rows.append(
            {
                "Dataset": dataset_name,
                "Rows": np.nan,
                "Columns": np.nan,
                "Memory (MB)": np.nan,
                "Loaded": False,
            }
        )

        print(
            f"Failed to load {dataset_name}: "
            f"{type(error).__name__}: {error}"
        )


# ============================================================
# Shape summary
# ============================================================

candidate_dataset_shapes_df = pd.DataFrame(
    dataset_shape_rows
)

print()
print("Primary candidate dataset shapes")
print("=" * 80)

display(candidate_dataset_shapes_df)


# ============================================================
# Expose loaded candidates as notebook variables
# ============================================================

for variable_name, dataframe in candidate_dataset_objects.items():
    globals()[variable_name] = dataframe


# ============================================================
# Loaded dataset names
# ============================================================

print()
print("Loaded candidate dataframe objects")
print("=" * 80)

for variable_name, dataframe in candidate_dataset_objects.items():
    print(
        f"{variable_name}: "
        f"{dataframe.shape}"
    )


# ============================================================
# Section validation
# ============================================================

all_primary_candidates_loaded = bool(
    not candidate_dataset_shapes_df.empty
    and candidate_dataset_shapes_df["Loaded"].all()
)

non_empty_candidates = bool(
    (
        candidate_dataset_shapes_df["Rows"].fillna(0) > 0
    ).all()
)

section_3_3_complete = bool(
    all_primary_candidates_loaded
    and non_empty_candidates
)

print()
print(
    f"Primary candidate datasets assessed: "
    f"{len(candidate_dataset_shapes_df)}"
)

print(
    f"Section 3.3 completion status: "
    f"{section_3_3_complete}"
)

if not section_3_3_complete:
    raise RuntimeError(
        "One or more primary release-level candidate datasets "
        "could not be loaded or contained no observations."
    )
```

### Interpretation

The three primary release-level candidate datasets were loaded successfully and their dimensions were assessed.

`charts_cleaned.csv` is substantially larger than the Spotify datasets, containing 5,427,136 rows and 10 columns. This indicates that it likely represents repeated historical chart observations across tracks, dates and country markets rather than a one-row-per-release structure.

`spotify_2024_cleaned.csv` contains 4,593 rows and 27 columns, while `spotify_integrated.csv` contains the same 4,593 rows but a wider 42-column schema.

The identical row counts suggest that the integrated Spotify dataset may preserve the same underlying track population while adding additional features or merged information. This makes `spotify_integrated.csv` particularly important for later release-level feature construction.

All three datasets loaded successfully, and Section 3.3 completed with a status of `True`.

The next step is to inspect the schema of each candidate dataset to determine which identifiers, release fields, streaming measures, chart variables, dates and geographic fields are actually available.

## 3.4 Schema Inspection

### Purpose

This subsection inspects the column structure of the three primary candidate datasets identified for PMIP Release Performance Intelligence.

The aim is to determine what analytical evidence is actually available within each source before selecting the final release-level dataset.

The schema inspection will focus on identifying:

- track and release identifiers;
- artist fields;
- release-date and temporal fields;
- streaming and popularity measures;
- chart-performance fields;
- country and market fields;
- historical-observation fields;
- any additional engineered or integrated PMIP features.

This step is important because dataset names and file sizes alone do not confirm whether a source contains the fields required for release-performance analysis.

```python
# ============================================================
# Section 3.4 - Schema Inspection
# ============================================================

if not section_3_3_complete:
    raise RuntimeError(
        "Section 3.3 must be completed before inspecting candidate schemas."
    )

# ============================================================
# Candidate dataframe mapping
# ============================================================

schema_sources = {
    "charts_cleaned.csv": charts_cleaned_df,
    "spotify_2024_cleaned.csv": spotify_2024_cleaned_df,
    "spotify_integrated.csv": spotify_integrated_df,
}

# ============================================================
# Field-category keyword groups
# ============================================================

field_categories = {
    "Identifier Fields": [
        "id",
        "track",
        "release",
        "album",
        "uri",
        "isrc",
    ],
    "Artist Fields": [
        "artist",
    ],
    "Temporal Fields": [
        "date",
        "year",
        "month",
        "day",
        "week",
        "time",
    ],
    "Streaming / Popularity Fields": [
        "stream",
        "popularity",
        "listener",
        "play",
    ],
    "Chart Fields": [
        "chart",
        "position",
        "rank",
        "peak",
    ],
    "Geographic Fields": [
        "country",
        "market",
        "region",
        "territory",
    ],
}

# ============================================================
# Build schema summary
# ============================================================

schema_summary_rows = []

for dataset_name, dataframe in schema_sources.items():

    for column in dataframe.columns:

        column_lower = str(column).lower()

        matched_categories = []

        for category, keywords in field_categories.items():

            if any(
                keyword in column_lower
                for keyword in keywords
            ):
                matched_categories.append(category)

        schema_summary_rows.append(
            {
                "Dataset": dataset_name,
                "Column": column,
                "Data Type": str(dataframe[column].dtype),
                "Non-Null": int(dataframe[column].notna().sum()),
                "Missing": int(dataframe[column].isna().sum()),
                "Unique Values": int(dataframe[column].nunique(dropna=True)),
                "Detected Category": (
                    ", ".join(matched_categories)
                    if matched_categories
                    else "Other"
                ),
            }
        )

candidate_schema_summary_df = pd.DataFrame(
    schema_summary_rows
)

# ============================================================
# Display complete column lists
# ============================================================

print("Inspecting primary PMIP candidate schemas")
print("=" * 80)

for dataset_name, dataframe in schema_sources.items():

    print()
    print(dataset_name)
    print("-" * 80)
    print(f"Shape: {dataframe.shape}")
    print()
    print("Columns:")
    print(list(dataframe.columns))

# ============================================================
# Display structured schema summary
# ============================================================

print()
print("Candidate schema summary")
print("=" * 80)

display(candidate_schema_summary_df)

# ============================================================
# Detected analytical field coverage
# ============================================================

coverage_rows = []

for dataset_name, dataframe in schema_sources.items():

    columns_lower = [
        str(column).lower()
        for column in dataframe.columns
    ]

    coverage_record = {
        "Dataset": dataset_name
    }

    for category, keywords in field_categories.items():

        coverage_record[category] = any(
            any(
                keyword in column_name
                for keyword in keywords
            )
            for column_name in columns_lower
        )

    coverage_rows.append(coverage_record)

schema_coverage_df = pd.DataFrame(
    coverage_rows
)

print()
print("Candidate analytical-field coverage")
print("=" * 80)

display(schema_coverage_df)

# ============================================================
# Key field examples
# ============================================================

print()
print("Detected candidate fields by analytical category")
print("=" * 80)

for dataset_name, dataframe in schema_sources.items():

    print()
    print(dataset_name)
    print("-" * 80)

    for category, keywords in field_categories.items():

        matched_columns = [
            column
            for column in dataframe.columns
            if any(
                keyword in str(column).lower()
                for keyword in keywords
            )
        ]

        print(
            f"{category}: "
            f"{matched_columns if matched_columns else 'None detected'}"
        )

# ============================================================
# Section validation
# ============================================================

schemas_available = bool(
    all(
        len(dataframe.columns) > 0
        for dataframe in schema_sources.values()
    )
)

identifier_evidence_available = bool(
    schema_coverage_df["Identifier Fields"].any()
)

artist_evidence_available = bool(
    schema_coverage_df["Artist Fields"].any()
)

section_3_4_complete = bool(
    schemas_available
    and identifier_evidence_available
    and artist_evidence_available
)

print()
print(
    f"Candidate schemas inspected: "
    f"{len(schema_sources)}"
)

print(
    f"Section 3.4 completion status: "
    f"{section_3_4_complete}"
)

if not section_3_4_complete:
    raise RuntimeError(
        "Candidate schema inspection did not confirm sufficient "
        "identifier and artist evidence for release-level analysis."
    )
```

### Interpretation

The schema inspection confirms that all three primary candidate datasets contain useful evidence for PMIP Release Performance Intelligence, but they serve different analytical roles.

`charts_cleaned.csv` contains detailed historical chart observations at track-country-date level. Its fields include `track_id`, `artists`, `date`, `country`, `position` and `streams`, making it highly suitable for analysing chart history, streaming change, geographic behaviour and repeated performance over time.

`spotify_2024_cleaned.csv` provides release-level metadata and current platform-performance information. Important fields include `Track`, `Album Name`, `Artist`, `Release Date`, `ISRC`, `Spotify Streams`, `Spotify Popularity`, playlist measures, YouTube activity, TikTok activity, AirPlay, Deezer, Pandora and Shazam indicators.

`spotify_integrated.csv` retains the same 4,593-release population as the cleaned Spotify dataset but extends the schema with integrated historical chart intelligence. Additional fields include chart position summaries, chart-observation counts, countries charted, historical streaming summaries and chart-date information.

This makes `spotify_integrated.csv` the strongest candidate for the primary release-level analytical dataset because it combines release metadata, cross-platform performance and previously integrated historical chart evidence within a single structure.

However, `charts_cleaned.csv` should remain available as a supporting historical source where more detailed track-country-date analysis is required.

Section 3.4 completed successfully and provides sufficient evidence to proceed with final dataset selection.

## 3.5 Dataset Selection

### Purpose

This subsection selects the final data sources that will be used for PMIP Release Performance Intelligence.

The selection is based on the dataset discovery, shape assessment and schema inspection completed in Sections 3.1 to 3.4.

The goal is to choose sources that provide:

- release-level identifiers and metadata;
- streaming and platform-performance evidence;
- historical chart information;
- geographic market evidence;
- temporal coverage;
- sufficient population size for later scoring and comparison.

The final selection will distinguish between the primary analytical dataset and supporting datasets used for deeper historical or validation analysis.

```python
# ============================================================
# Section 3.5 - Dataset Selection
# ============================================================

if not section_3_4_complete:
    raise RuntimeError(
        "Section 3.4 must be completed before selecting final datasets."
    )

# ============================================================
# Final dataset roles
# ============================================================

selected_dataset_roles = pd.DataFrame(
    [
        {
            "Dataset": "spotify_integrated.csv",
            "Role": "Primary analytical dataset",
            "Reason": (
                "Contains release-level metadata, streaming and platform "
                "performance measures, plus integrated historical chart evidence."
            ),
            "Selected": True,
        },
        {
            "Dataset": "charts_cleaned.csv",
            "Role": "Historical supporting dataset",
            "Reason": (
                "Provides detailed track-country-date chart observations, "
                "streaming values, market evidence and temporal history."
            ),
            "Selected": True,
        },
        {
            "Dataset": "spotify_2024_cleaned.csv",
            "Role": "Reference / base dataset",
            "Reason": (
                "Provides the original cleaned release-level Spotify population "
                "used before historical enrichment."
            ),
            "Selected": True,
        },
    ]
)

print("Selecting PMIP Release Performance Intelligence datasets")
print("=" * 80)

display(selected_dataset_roles)

# ============================================================
# Assign final working datasets
# ============================================================

release_primary_df = spotify_integrated_df.copy()

release_history_df = charts_cleaned_df

release_reference_df = spotify_2024_cleaned_df.copy()

# ============================================================
# Selection validation
# ============================================================

primary_population_match = bool(
    len(release_primary_df) == len(release_reference_df)
)

primary_identifier_available = bool(
    "Track" in release_primary_df.columns
    or "ISRC" in release_primary_df.columns
)

release_date_available = bool(
    "Release Date" in release_primary_df.columns
)

historical_track_id_available = bool(
    "track_id" in release_history_df.columns
)

historical_country_available = bool(
    "country" in release_history_df.columns
)

historical_date_available = bool(
    "date" in release_history_df.columns
)

selection_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Primary dataset available",
            "Passed": not release_primary_df.empty,
        },
        {
            "Validation Area": "Primary and reference populations match",
            "Passed": primary_population_match,
        },
        {
            "Validation Area": "Primary release identifier available",
            "Passed": primary_identifier_available,
        },
        {
            "Validation Area": "Primary release date available",
            "Passed": release_date_available,
        },
        {
            "Validation Area": "Historical track identifier available",
            "Passed": historical_track_id_available,
        },
        {
            "Validation Area": "Historical country evidence available",
            "Passed": historical_country_available,
        },
        {
            "Validation Area": "Historical temporal evidence available",
            "Passed": historical_date_available,
        },
    ]
)

print()
print("Dataset selection validation")
print("=" * 80)

display(selection_validation_df)

# ============================================================
# Final selection summary
# ============================================================

selection_summary_df = pd.DataFrame(
    [
        {
            "Working Object": "release_primary_df",
            "Source Dataset": "spotify_integrated.csv",
            "Rows": len(release_primary_df),
            "Columns": len(release_primary_df.columns),
        },
        {
            "Working Object": "release_history_df",
            "Source Dataset": "charts_cleaned.csv",
            "Rows": len(release_history_df),
            "Columns": len(release_history_df.columns),
        },
        {
            "Working Object": "release_reference_df",
            "Source Dataset": "spotify_2024_cleaned.csv",
            "Rows": len(release_reference_df),
            "Columns": len(release_reference_df.columns),
        },
    ]
)

print()
print("Final PMIP release-performance dataset selection")
print("=" * 80)

display(selection_summary_df)

# ============================================================
# Section 3 completion status
# ============================================================

section_3_5_complete = bool(
    selection_validation_df["Passed"].all()
)

section_3_complete = bool(
    section_3_1_complete
    and section_3_2_complete
    and section_3_3_complete
    and section_3_4_complete
    and section_3_5_complete
)

print()
print(f"Section 3.5 completion status: {section_3_5_complete}")
print(f"Section 3 overall completion status: {section_3_complete}")

if not section_3_complete:
    raise RuntimeError(
        "Section 3 Data Discovery and Source Assessment "
        "did not complete successfully."
    )
```

# 4. Release Identity and Data Preparation

This section prepares a reliable release-level analytical population for the PMIP Release Performance Intelligence component.

The aim is to confirm what constitutes a release within the selected datasets, validate release identifiers, examine artist-to-release relationships, verify release dates, and assess missing or duplicated records before constructing the final release population.

The preparation process will focus on:

- defining the analytical meaning of a release within PMIP;
- validating release identifiers such as `ISRC`, track names and normalised track fields;
- examining artist-to-release relationships;
- validating release dates and temporal consistency;
- identifying missing values and duplicate records;
- resolving records that may represent the same release more than once;
- constructing a clean final release-level population for later performance analysis.

This stage is important because later scoring, ranking and comparison depend on each analytical record representing a valid and interpretable release.




## 4.1 Define a Release

### Purpose

This subsection defines what will be treated as a music release within the PMIP Release Performance Intelligence analysis.

For this notebook, the core analytical unit is an individual track release rather than an album-level release.

A release will therefore be represented primarily using the strongest available identifiers from the selected PMIP datasets, including:

- `ISRC` where available;
- track title;
- artist name;
- release date;
- normalised track and artist fields where available.

The purpose of this definition is to establish a consistent release identity before calculating performance indicators.

A valid PMIP release should represent one identifiable track associated with an artist and release date, while recognising that duplicate titles, collaborations, remasters or repeated platform records may require additional validation.

This definition will be used throughout the remaining notebook when validating identifiers, resolving duplicates and constructing the final release-level population.

```python
# ============================================================
# Section 4.1 - Define a Release
# ============================================================

if not section_3_complete:
    raise RuntimeError(
        "Section 3 must be completed before defining the release identity."
    )

# ============================================================
# Release-definition fields
# ============================================================

release_definition_fields = {
    "Primary Identifier": "ISRC",
    "Track Field": "Track",
    "Artist Field": "Artist",
    "Release Date Field": "Release Date",
    "Normalised Track Field": "track_normalised",
    "Normalised Artist Field": "artist_normalised",
}

# ============================================================
# Check field availability
# ============================================================

release_definition_rows = []

for role, field_name in release_definition_fields.items():

    available = field_name in release_primary_df.columns

    release_definition_rows.append(
        {
            "Release Identity Role": role,
            "Field": field_name,
            "Available": available,
        }
    )

release_definition_validation_df = pd.DataFrame(
    release_definition_rows
)

print("Defining PMIP release identity")
print("=" * 80)

display(release_definition_validation_df)

# ============================================================
# Core release-definition rules
# ============================================================

release_identity_rules_df = pd.DataFrame(
    [
        {
            "Rule": "Analytical unit",
            "Definition": "One individual track release",
        },
        {
            "Rule": "Preferred identifier",
            "Definition": (
                "Use ISRC as the strongest available release identifier"
            ),
        },
        {
            "Rule": "Fallback identity",
            "Definition": (
                "Use track title, artist and release date together "
                "when ISRC validation requires supporting evidence"
            ),
        },
        {
            "Rule": "Normalised identity support",
            "Definition": (
                "Use normalised track and artist fields to support "
                "duplicate and matching checks"
            ),
        },
        {
            "Rule": "Album handling",
            "Definition": (
                "Album Name is retained as contextual metadata but "
                "does not define the core analytical unit"
            ),
        },
        {
            "Rule": "Human review boundary",
            "Definition": (
                "Ambiguous duplicate or conflicting release identities "
                "must be flagged rather than assumed equivalent"
            ),
        },
    ]
)

print()
print("PMIP release identity rules")
print("=" * 80)

display(release_identity_rules_df)

# ============================================================
# Basic population evidence
# ============================================================

release_identity_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Primary release records",
            "Observed Value": len(release_primary_df),
        },
        {
            "Metric": "Unique ISRC values",
            "Observed Value": (
                release_primary_df["ISRC"].nunique(dropna=True)
                if "ISRC" in release_primary_df.columns
                else np.nan
            ),
        },
        {
            "Metric": "Unique track titles",
            "Observed Value": (
                release_primary_df["Track"].nunique(dropna=True)
                if "Track" in release_primary_df.columns
                else np.nan
            ),
        },
        {
            "Metric": "Unique artists",
            "Observed Value": (
                release_primary_df["Artist"].nunique(dropna=True)
                if "Artist" in release_primary_df.columns
                else np.nan
            ),
        },
        {
            "Metric": "Unique release dates",
            "Observed Value": (
                release_primary_df["Release Date"].nunique(dropna=True)
                if "Release Date" in release_primary_df.columns
                else np.nan
            ),
        },
    ]
)

print()
print("Release identity overview")
print("=" * 80)

display(release_identity_overview_df)

# ============================================================
# Section validation
# ============================================================

required_identity_fields = [
    "ISRC",
    "Track",
    "Artist",
    "Release Date",
]

required_identity_fields_available = bool(
    all(
        field in release_primary_df.columns
        for field in required_identity_fields
    )
)

section_4_1_complete = bool(
    required_identity_fields_available
    and len(release_primary_df) > 0
)

print()
print(f"Section 4.1 completion status: {section_4_1_complete}")

if not section_4_1_complete:
    raise RuntimeError(
        "The selected primary dataset does not contain the minimum "
        "fields required to define a PMIP release."
    )
```

### Interpretation

The PMIP release identity has been defined successfully at individual track level.

All required identity fields are available within the primary release dataset, including `ISRC`, `Track`, `Artist`, `Release Date`, `track_normalised` and `artist_normalised`.

The primary dataset contains 4,593 release records and 4,593 unique ISRC values. This indicates that each current record has a distinct ISRC within the analysed population, providing strong support for using ISRC as the preferred release identifier.

The dataset contains fewer unique track titles than ISRC values, with 4,365 unique track names across 4,593 releases. This shows that some track titles are shared across different releases, artists or recordings, reinforcing why track title alone should not be treated as a unique release identifier.

The 1,999 unique artists and 1,562 release dates also confirm that the dataset represents a varied release population spanning multiple artists and release periods.

Section 4.1 therefore establishes ISRC as the primary release identity while retaining track, artist, date and normalised fields as supporting evidence for later validation and duplicate assessment.

## 4.2 Release Identifier Validation

### Purpose

This subsection validates the reliability of the release identifiers used in the PMIP Release Performance Intelligence dataset.

Although Section 4.1 confirmed that every current release record has a unique `ISRC`, this subsection checks the identifier more carefully before it is accepted as the main release key.

The validation will assess:

- missing or blank ISRC values;
- duplicate ISRC values;
- ISRC formatting consistency;
- whether a single ISRC maps to multiple track titles;
- whether a single ISRC maps to multiple artists;
- whether a single ISRC maps to multiple release dates;
- whether normalised track and artist fields support the same release identity.

This helps confirm that the identifier is not only unique, but also internally consistent with the supporting release metadata.

```python
# ============================================================
# Section 4.2 - Release Identifier Validation
# ============================================================

if not section_4_1_complete:
    raise RuntimeError(
        "Section 4.1 must be completed before validating release identifiers."
    )

# ============================================================
# Working copy
# ============================================================

release_identifier_df = release_primary_df.copy()

# ============================================================
# Clean ISRC representation for validation
# ============================================================

release_identifier_df["ISRC_clean"] = (
    release_identifier_df["ISRC"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# ============================================================
# Missing / blank ISRC assessment
# ============================================================

missing_isrc_count = int(
    release_identifier_df["ISRC_clean"].isna().sum()
)

blank_isrc_count = int(
    (
        release_identifier_df["ISRC_clean"]
        .fillna("")
        .eq("")
    ).sum()
)

valid_isrc_count = int(
    len(release_identifier_df)
    - missing_isrc_count
    - blank_isrc_count
)

# ============================================================
# Duplicate ISRC assessment
# ============================================================

duplicate_isrc_mask = (
    release_identifier_df["ISRC_clean"]
    .duplicated(keep=False)
    & release_identifier_df["ISRC_clean"].notna()
    & release_identifier_df["ISRC_clean"].ne("")
)

duplicate_isrc_records = int(
    duplicate_isrc_mask.sum()
)

duplicate_isrc_values = int(
    release_identifier_df.loc[
        duplicate_isrc_mask,
        "ISRC_clean"
    ].nunique()
)

# ============================================================
# Basic ISRC format assessment
# Standard ISRC is 12 alphanumeric characters
# ============================================================

isrc_format_mask = (
    release_identifier_df["ISRC_clean"]
    .fillna("")
    .str.fullmatch(r"[A-Z0-9]{12}")
)

valid_format_count = int(
    isrc_format_mask.sum()
)

invalid_format_count = int(
    len(release_identifier_df) - valid_format_count
)

# ============================================================
# ISRC metadata consistency
# ============================================================

isrc_consistency_df = (
    release_identifier_df
    .groupby("ISRC_clean", dropna=False)
    .agg(
        Records=("ISRC_clean", "size"),
        Track_Titles=("Track", "nunique"),
        Artists=("Artist", "nunique"),
        Release_Dates=("Release Date", "nunique"),
        Normalised_Tracks=("track_normalised", "nunique"),
        Normalised_Artists=("artist_normalised", "nunique"),
    )
    .reset_index()
)

conflicting_track_isrcs = int(
    (isrc_consistency_df["Track_Titles"] > 1).sum()
)

conflicting_artist_isrcs = int(
    (isrc_consistency_df["Artists"] > 1).sum()
)

conflicting_date_isrcs = int(
    (isrc_consistency_df["Release_Dates"] > 1).sum()
)

conflicting_normalised_track_isrcs = int(
    (isrc_consistency_df["Normalised_Tracks"] > 1).sum()
)

conflicting_normalised_artist_isrcs = int(
    (isrc_consistency_df["Normalised_Artists"] > 1).sum()
)

# ============================================================
# Release identifier overview
# ============================================================

release_identifier_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(release_identifier_df),
        },
        {
            "Metric": "Valid non-blank ISRC values",
            "Observed Value": valid_isrc_count,
        },
        {
            "Metric": "Missing ISRC values",
            "Observed Value": missing_isrc_count,
        },
        {
            "Metric": "Blank ISRC values",
            "Observed Value": blank_isrc_count,
        },
        {
            "Metric": "Duplicate ISRC records",
            "Observed Value": duplicate_isrc_records,
        },
        {
            "Metric": "Duplicate ISRC values",
            "Observed Value": duplicate_isrc_values,
        },
        {
            "Metric": "ISRC values matching 12-character format",
            "Observed Value": valid_format_count,
        },
        {
            "Metric": "ISRC values failing format check",
            "Observed Value": invalid_format_count,
        },
        {
            "Metric": "ISRCs linked to multiple track titles",
            "Observed Value": conflicting_track_isrcs,
        },
        {
            "Metric": "ISRCs linked to multiple artists",
            "Observed Value": conflicting_artist_isrcs,
        },
        {
            "Metric": "ISRCs linked to multiple release dates",
            "Observed Value": conflicting_date_isrcs,
        },
    ]
)

print("Validating PMIP release identifiers")
print("=" * 80)

display(release_identifier_overview_df)

# ============================================================
# Detailed consistency checks
# ============================================================

identifier_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "ISRC field available",
            "Passed": "ISRC" in release_identifier_df.columns,
        },
        {
            "Validation Area": "No missing ISRC values",
            "Passed": missing_isrc_count == 0,
        },
        {
            "Validation Area": "No blank ISRC values",
            "Passed": blank_isrc_count == 0,
        },
        {
            "Validation Area": "No duplicate ISRC values",
            "Passed": duplicate_isrc_values == 0,
        },
        {
            "Validation Area": "ISRC format consistent",
            "Passed": invalid_format_count == 0,
        },
        {
            "Validation Area": "ISRC maps to one track title",
            "Passed": conflicting_track_isrcs == 0,
        },
        {
            "Validation Area": "ISRC maps to one artist identity",
            "Passed": conflicting_artist_isrcs == 0,
        },
        {
            "Validation Area": "ISRC maps to one release date",
            "Passed": conflicting_date_isrcs == 0,
        },
        {
            "Validation Area": "Normalised track identity consistent",
            "Passed": conflicting_normalised_track_isrcs == 0,
        },
        {
            "Validation Area": "Normalised artist identity consistent",
            "Passed": conflicting_normalised_artist_isrcs == 0,
        },
    ]
)

print()
print("Release identifier validation")
print("=" * 80)

display(identifier_validation_df)

# ============================================================
# Show problematic records if present
# ============================================================

problematic_isrcs = isrc_consistency_df[
    (isrc_consistency_df["Track_Titles"] > 1)
    | (isrc_consistency_df["Artists"] > 1)
    | (isrc_consistency_df["Release_Dates"] > 1)
    | (isrc_consistency_df["Normalised_Tracks"] > 1)
    | (isrc_consistency_df["Normalised_Artists"] > 1)
]

if (
    duplicate_isrc_values > 0
    or invalid_format_count > 0
    or not problematic_isrcs.empty
):
    print()
    print("Release identifiers requiring review")
    print("=" * 80)

    if invalid_format_count > 0:
        print()
        print("Invalid-format ISRC examples")
        print("-" * 80)

        display(
            release_identifier_df.loc[
                ~isrc_format_mask,
                [
                    "ISRC",
                    "Track",
                    "Artist",
                    "Release Date",
                ],
            ].head(25)
        )

    if not problematic_isrcs.empty:
        print()
        print("ISRC metadata conflicts")
        print("-" * 80)

        display(
            problematic_isrcs.head(25)
        )

# ============================================================
# Section validation
# ============================================================

section_4_2_complete = bool(
    identifier_validation_df["Passed"].all()
)

print()
print(f"Section 4.2 completion status: {section_4_2_complete}")

if not section_4_2_complete:
    print(
        "Release identifier validation identified one or more issues "
        "that require review before final population construction."
    )
```

### Interpretation

The PMIP release identifier validation completed successfully.

All 4,593 release records contain valid, non-blank ISRC values, and every ISRC is unique within the current release population.

All identifiers also match the expected 12-character ISRC format, with no formatting failures detected.

The metadata-consistency checks identified no cases where a single ISRC maps to multiple track titles, artists or release dates. The normalised track and artist fields are also internally consistent with the ISRC-based identities.

These results provide strong evidence that `ISRC` can be used reliably as the primary release identifier throughout the Release Performance Intelligence analysis.

Section 4.2 therefore confirms that the release population has a stable and internally consistent identifier structure, allowing the notebook to proceed to analysis of artist-to-release relationships.

## 4.3 Artist–Release Relationships

### Purpose

This subsection examines how releases are distributed across artists within the PMIP Release Performance Intelligence dataset.

The aim is to understand whether artists are represented by single or multiple releases and to identify any unusual artist-to-release relationships that may affect later comparison, scoring or aggregation.

The assessment will examine:

- number of releases per artist;
- number of unique ISRCs per artist;
- artists represented by only one release;
- artists represented by multiple releases;
- artists with unusually large release counts;
- whether normalised artist identities agree with the original artist field.

This provides important context for later analysis because release performance may be influenced by whether an artist has one observed release or a larger catalogue represented in the dataset.

```python
# ============================================================
# Section 4.3 - Artist–Release Relationships
# ============================================================

if not section_4_2_complete:
    raise RuntimeError(
        "Section 4.2 must be completed before assessing artist-release relationships."
    )

# ============================================================
# Working source
# ============================================================

artist_release_df = release_identifier_df.copy()

# ============================================================
# Artist-level release counts
# ============================================================

artist_release_counts_df = (
    artist_release_df
    .groupby(
        ["Artist", "artist_normalised"],
        dropna=False
    )
    .agg(
        Releases=("ISRC_clean", "size"),
        Unique_ISRCs=("ISRC_clean", "nunique"),
        Unique_Track_Titles=("Track", "nunique"),
        Unique_Release_Dates=("Release Date", "nunique"),
    )
    .reset_index()
)

# ============================================================
# Relationship summary
# ============================================================

single_release_artists = int(
    (artist_release_counts_df["Releases"] == 1).sum()
)

multi_release_artists = int(
    (artist_release_counts_df["Releases"] > 1).sum()
)

max_releases_per_artist = int(
    artist_release_counts_df["Releases"].max()
)

median_releases_per_artist = float(
    artist_release_counts_df["Releases"].median()
)

mean_releases_per_artist = float(
    artist_release_counts_df["Releases"].mean()
)

# ============================================================
# Identity consistency
# ============================================================

artist_identity_consistency_df = (
    artist_release_df
    .groupby("artist_normalised", dropna=False)
    .agg(
        Original_Artist_Labels=("Artist", "nunique"),
        Releases=("ISRC_clean", "size"),
    )
    .reset_index()
)

normalised_artist_conflicts = int(
    (
        artist_identity_consistency_df[
            "Original_Artist_Labels"
        ] > 1
    ).sum()
)

# ============================================================
# Artist-release overview
# ============================================================

artist_release_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Artists assessed",
            "Observed Value": len(artist_release_counts_df),
        },
        {
            "Metric": "Release records assessed",
            "Observed Value": len(artist_release_df),
        },
        {
            "Metric": "Single-release artists",
            "Observed Value": single_release_artists,
        },
        {
            "Metric": "Multi-release artists",
            "Observed Value": multi_release_artists,
        },
        {
            "Metric": "Median releases per artist",
            "Observed Value": median_releases_per_artist,
        },
        {
            "Metric": "Mean releases per artist",
            "Observed Value": mean_releases_per_artist,
        },
        {
            "Metric": "Maximum releases for one artist",
            "Observed Value": max_releases_per_artist,
        },
        {
            "Metric": "Normalised artist identities with multiple labels",
            "Observed Value": normalised_artist_conflicts,
        },
    ]
)

print("Assessing PMIP artist-release relationships")
print("=" * 80)

display(artist_release_overview_df)

# ============================================================
# Release-count distribution
# ============================================================

release_count_distribution_df = (
    artist_release_counts_df["Releases"]
    .value_counts()
    .sort_index()
    .rename_axis("Releases per Artist")
    .reset_index(name="Artists")
)

release_count_distribution_df["Share (%)"] = (
    release_count_distribution_df["Artists"]
    / len(artist_release_counts_df)
    * 100
)

print()
print("Artist release-count distribution")
print("=" * 80)

display(
    release_count_distribution_df.head(20)
)

# ============================================================
# Artists with the largest represented release catalogues
# ============================================================

top_artist_release_counts_df = (
    artist_release_counts_df
    .sort_values(
        by=["Releases", "Unique_Release_Dates"],
        ascending=[False, False]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artists with the largest represented release counts")
print("=" * 80)

display(top_artist_release_counts_df)

# ============================================================
# Relationship validation
# ============================================================

unique_isrc_per_record = bool(
    (
        artist_release_counts_df["Releases"]
        == artist_release_counts_df["Unique_ISRCs"]
    ).all()
)

artist_field_available = bool(
    artist_release_df["Artist"].notna().all()
)

normalised_artist_available = bool(
    artist_release_df["artist_normalised"].notna().all()
)

artist_release_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Artist field complete",
            "Passed": artist_field_available,
        },
        {
            "Validation Area": "Normalised artist identity complete",
            "Passed": normalised_artist_available,
        },
        {
            "Validation Area": "Every release retains a unique ISRC",
            "Passed": unique_isrc_per_record,
        },
        {
            "Validation Area": "Artist-release population available",
            "Passed": len(artist_release_counts_df) > 0,
        },
        {
            "Validation Area": "Release-count distribution prepared",
            "Passed": not release_count_distribution_df.empty,
        },
    ]
)

print()
print("Artist-release relationship validation")
print("=" * 80)

display(artist_release_validation_df)

# ============================================================
# Section validation
# ============================================================

section_4_3_complete = bool(
    artist_release_validation_df["Passed"].all()
)

print()
print(f"Section 4.3 completion status: {section_4_3_complete}")

if not section_4_3_complete:
    raise RuntimeError(
        "Artist-release relationship validation did not complete successfully."
    )
```

### Interpretation

The PMIP artist–release relationship assessment completed successfully.

The dataset contains 1,999 artists associated with 4,593 validated release records. Most artists are represented by a relatively small number of releases.

A total of 1,340 artists, representing approximately 67.03% of the artist population, are associated with only one release. A further 659 artists are represented by multiple releases.

The median number of releases per artist is 1, while the mean is approximately 2.30. This difference indicates that the release distribution is uneven, with a smaller group of artists contributing substantially larger catalogues.

The largest represented catalogue contains 63 releases for a single artist, confirming that some artists contribute many more observations than others.

Two normalised artist identities are associated with multiple original artist labels. These cases should be retained for later review because they may represent formatting or naming variations rather than genuinely different artist identities.

All validation checks passed successfully. Artist fields are complete, normalised identities are available, each release retains a unique ISRC, and the release-count distribution was prepared correctly.

Section 4.3 therefore confirms that the artist–release structure is suitable for continued analysis while highlighting the uneven catalogue representation that should be considered during later artist-level comparisons.

## 4.4 Date Validation

### Purpose

This subsection validates the release-date information used in the PMIP Release Performance Intelligence dataset.

The aim is to confirm that release dates are complete, parseable and logically usable before they are used in later performance, age and historical comparison analysis.

The assessment will examine:

- missing release dates;
- invalid or unparseable date values;
- earliest and latest release dates;
- future-dated records;
- release-year coverage;
- consistency between release dates and available chart-history dates;
- whether historical chart activity occurs before the stated release date.

This helps ensure that later temporal analysis is based on valid release chronology and highlights any records that may require caution or review.

```python
# ============================================================
# Section 4.4 - Date Validation
# ============================================================

if not section_4_3_complete:
    raise RuntimeError(
        "Section 4.3 must be completed before validating release dates."
    )

# ============================================================
# Working copy
# ============================================================

release_date_df = artist_release_df.copy()

# ============================================================
# Parse release dates
# ============================================================

release_date_df["release_date_parsed"] = pd.to_datetime(
    release_date_df["Release Date"],
    errors="coerce"
)

# ============================================================
# Parse integrated historical chart dates where available
# ============================================================

if "first_chart_date" in release_date_df.columns:
    release_date_df["first_chart_date_parsed"] = pd.to_datetime(
        release_date_df["first_chart_date"],
        errors="coerce"
    )
else:
    release_date_df["first_chart_date_parsed"] = pd.NaT

if "last_chart_date" in release_date_df.columns:
    release_date_df["last_chart_date_parsed"] = pd.to_datetime(
        release_date_df["last_chart_date"],
        errors="coerce"
    )
else:
    release_date_df["last_chart_date_parsed"] = pd.NaT

# ============================================================
# Release-date completeness
# ============================================================

missing_release_dates = int(
    release_date_df["Release Date"].isna().sum()
)

blank_release_dates = int(
    release_date_df["Release Date"]
    .astype("string")
    .fillna("")
    .str.strip()
    .eq("")
    .sum()
)

invalid_release_dates = int(
    (
        release_date_df["release_date_parsed"].isna()
        & release_date_df["Release Date"].notna()
        & release_date_df["Release Date"]
        .astype("string")
        .str.strip()
        .ne("")
    ).sum()
)

valid_release_dates = int(
    release_date_df["release_date_parsed"].notna().sum()
)

# ============================================================
# Release-date range
# ============================================================

earliest_release_date = (
    release_date_df["release_date_parsed"].min()
)

latest_release_date = (
    release_date_df["release_date_parsed"].max()
)

# ============================================================
# Future-date assessment
# ============================================================

today = pd.Timestamp.today().normalize()

future_release_mask = (
    release_date_df["release_date_parsed"].notna()
    & (release_date_df["release_date_parsed"] > today)
)

future_release_count = int(
    future_release_mask.sum()
)

# ============================================================
# Release-year coverage
# ============================================================

release_date_df["release_year"] = (
    release_date_df["release_date_parsed"].dt.year
)

release_year_summary_df = (
    release_date_df["release_year"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("Release Year")
    .reset_index(name="Releases")
)

# ============================================================
# Historical chronology checks
# ============================================================

history_available_mask = (
    release_date_df["first_chart_date_parsed"].notna()
)

chart_before_release_mask = (
    history_available_mask
    & release_date_df["release_date_parsed"].notna()
    & (
        release_date_df["first_chart_date_parsed"]
        < release_date_df["release_date_parsed"]
    )
)

chart_before_release_count = int(
    chart_before_release_mask.sum()
)

history_after_or_on_release_count = int(
    (
        history_available_mask
        & release_date_df["release_date_parsed"].notna()
        & (
            release_date_df["first_chart_date_parsed"]
            >= release_date_df["release_date_parsed"]
        )
    ).sum()
)

historical_date_records = int(
    history_available_mask.sum()
)

# ============================================================
# Date-validation overview
# ============================================================

release_date_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(release_date_df),
        },
        {
            "Metric": "Valid release dates",
            "Observed Value": valid_release_dates,
        },
        {
            "Metric": "Missing release dates",
            "Observed Value": missing_release_dates,
        },
        {
            "Metric": "Blank release dates",
            "Observed Value": blank_release_dates,
        },
        {
            "Metric": "Invalid release dates",
            "Observed Value": invalid_release_dates,
        },
        {
            "Metric": "Earliest release date",
            "Observed Value": earliest_release_date,
        },
        {
            "Metric": "Latest release date",
            "Observed Value": latest_release_date,
        },
        {
            "Metric": "Future-dated releases",
            "Observed Value": future_release_count,
        },
        {
            "Metric": "Releases with chart-history dates",
            "Observed Value": historical_date_records,
        },
        {
            "Metric": "Chart histories starting before release date",
            "Observed Value": chart_before_release_count,
        },
        {
            "Metric": "Chart histories starting on/after release date",
            "Observed Value": history_after_or_on_release_count,
        },
    ]
)

print("Validating PMIP release dates")
print("=" * 80)

display(release_date_overview_df)

# ============================================================
# Release-year distribution
# ============================================================

print()
print("Release-year distribution")
print("=" * 80)

display(release_year_summary_df)

# ============================================================
# Records requiring temporal review
# ============================================================

temporal_review_df = release_date_df[
    future_release_mask
    | chart_before_release_mask
    | release_date_df["release_date_parsed"].isna()
].copy()

if not temporal_review_df.empty:

    temporal_review_columns = [
        "ISRC",
        "Track",
        "Artist",
        "Release Date",
        "release_date_parsed",
        "first_chart_date",
        "first_chart_date_parsed",
        "last_chart_date",
        "last_chart_date_parsed",
    ]

    temporal_review_columns = [
        column
        for column in temporal_review_columns
        if column in temporal_review_df.columns
    ]

    print()
    print("Release records requiring temporal review")
    print("=" * 80)

    display(
        temporal_review_df[
            temporal_review_columns
        ].head(30)
    )

# ============================================================
# Validation checks
# ============================================================

release_date_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Release date field complete",
            "Passed": missing_release_dates == 0,
        },
        {
            "Validation Area": "No blank release dates",
            "Passed": blank_release_dates == 0,
        },
        {
            "Validation Area": "All release dates parse successfully",
            "Passed": invalid_release_dates == 0,
        },
        {
            "Validation Area": "No future-dated releases",
            "Passed": future_release_count == 0,
        },
        {
            "Validation Area": "Release-date range available",
            "Passed": (
                pd.notna(earliest_release_date)
                and pd.notna(latest_release_date)
            ),
        },
        {
            "Validation Area": "Historical chronology assessed",
            "Passed": historical_date_records >= 0,
        },
    ]
)

print()
print("Release-date validation")
print("=" * 80)

display(release_date_validation_df)

# ============================================================
# Section validation
# ============================================================

section_4_4_complete = bool(
    release_date_validation_df["Passed"].all()
)

print()
print(f"Section 4.4 completion status: {section_4_4_complete}")

if not section_4_4_complete:
    print(
        "Release-date validation identified one or more records "
        "that require review before final population construction."
    )
```

### Interpretation

The PMIP release-date validation completed successfully.

All 4,593 release records contain valid and parseable release dates, with no missing, blank, invalid or future-dated values identified.

The release population spans from 1987 to 2024, providing a broad historical range for later release-performance comparison.

Historical chart-date information is available for 2,235 releases. Of these, 2,158 releases have chart histories beginning on or after the recorded release date, which is chronologically consistent.

However, 77 releases contain chart-history observations that begin before the stated release date. These records may reflect reissues, remasters, catalogue re-releases, alternate versions, metadata differences or imperfect matching between the release-level and historical chart datasets.

These records should not be removed automatically. Instead, they should remain flagged for later review and should be interpreted cautiously when analysing post-release behaviour or release-age effects.

All formal date-validation checks passed, and Section 4.4 completed successfully.

The release population is therefore suitable to proceed to missing-value and duplicate-record assessment.

## 4.5 Missing and Duplicate Records

### Purpose

This subsection assesses missing values and duplicate records within the validated PMIP release population.

The aim is to determine whether any records require cleaning, exclusion or special handling before the final release population is created.

The assessment will examine:

- missing values across key release fields;
- completely duplicated rows;
- duplicated ISRC values;
- repeated track-title and artist combinations;
- repeated normalised track and artist combinations;
- records sharing the same title but representing different releases;
- whether missing values occur mainly in optional performance fields rather than core identity fields.

This helps separate genuine data-quality problems from valid repeated titles, collaborations, reissues or releases that simply have incomplete platform coverage.

```python
# ============================================================
# Section 4.5 - Missing and Duplicate Records
# ============================================================

if not section_4_4_complete:
    raise RuntimeError(
        "Section 4.4 must be completed before assessing missing and duplicate records."
    )

# ============================================================
# Working copy
# ============================================================

release_quality_df = release_date_df.copy()

# ============================================================
# Key release fields
# ============================================================

core_identity_fields = [
    "ISRC",
    "Track",
    "Artist",
    "Release Date",
    "track_normalised",
    "artist_normalised",
]

available_core_identity_fields = [
    column
    for column in core_identity_fields
    if column in release_quality_df.columns
]

# ============================================================
# Missing-value summary
# ============================================================

missing_summary_df = pd.DataFrame(
    {
        "Field": release_quality_df.columns,
        "Missing Values": [
            int(release_quality_df[column].isna().sum())
            for column in release_quality_df.columns
        ],
        "Missing Share (%)": [
            round(
                release_quality_df[column].isna().mean() * 100,
                6
            )
            for column in release_quality_df.columns
        ],
    }
)

missing_summary_df = (
    missing_summary_df
    .sort_values(
        by=["Missing Values", "Field"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

# ============================================================
# Core identity missingness
# ============================================================

core_missing_rows = []

for column in available_core_identity_fields:

    core_missing_rows.append(
        {
            "Field": column,
            "Missing Values": int(
                release_quality_df[column].isna().sum()
            ),
            "Blank Values": int(
                release_quality_df[column]
                .astype("string")
                .fillna("")
                .str.strip()
                .eq("")
                .sum()
            ),
        }
    )

core_identity_missing_df = pd.DataFrame(
    core_missing_rows
)

# ============================================================
# Complete duplicate-row assessment
# ============================================================

full_duplicate_mask = release_quality_df.duplicated(
    keep=False
)

full_duplicate_records = int(
    full_duplicate_mask.sum()
)

full_duplicate_groups = int(
    release_quality_df.loc[
        full_duplicate_mask
    ].drop_duplicates().shape[0]
)

# ============================================================
# Duplicate ISRC assessment
# ============================================================

duplicate_isrc_mask = (
    release_quality_df["ISRC_clean"]
    .duplicated(keep=False)
    & release_quality_df["ISRC_clean"].notna()
)

duplicate_isrc_records = int(
    duplicate_isrc_mask.sum()
)

duplicate_isrc_values = int(
    release_quality_df.loc[
        duplicate_isrc_mask,
        "ISRC_clean"
    ].nunique()
)

# ============================================================
# Track-title and artist duplicate patterns
# ============================================================

title_artist_counts_df = (
    release_quality_df
    .groupby(
        ["Track", "Artist"],
        dropna=False
    )
    .agg(
        Records=("ISRC_clean", "size"),
        Unique_ISRCs=("ISRC_clean", "nunique"),
        Unique_Release_Dates=("Release Date", "nunique"),
    )
    .reset_index()
)

repeated_title_artist_df = (
    title_artist_counts_df[
        title_artist_counts_df["Records"] > 1
    ]
    .sort_values(
        by=["Records", "Unique_ISRCs"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

repeated_title_artist_pairs = int(
    len(repeated_title_artist_df)
)

# ============================================================
# Normalised identity duplicate patterns
# ============================================================

normalised_identity_counts_df = (
    release_quality_df
    .groupby(
        ["track_normalised", "artist_normalised"],
        dropna=False
    )
    .agg(
        Records=("ISRC_clean", "size"),
        Unique_ISRCs=("ISRC_clean", "nunique"),
        Unique_Release_Dates=("Release Date", "nunique"),
        Original_Track_Labels=("Track", "nunique"),
        Original_Artist_Labels=("Artist", "nunique"),
    )
    .reset_index()
)

repeated_normalised_identity_df = (
    normalised_identity_counts_df[
        normalised_identity_counts_df["Records"] > 1
    ]
    .sort_values(
        by=["Records", "Unique_ISRCs"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

repeated_normalised_pairs = int(
    len(repeated_normalised_identity_df)
)

# ============================================================
# Same title across multiple ISRCs
# ============================================================

track_title_summary_df = (
    release_quality_df
    .groupby("Track", dropna=False)
    .agg(
        Records=("ISRC_clean", "size"),
        Unique_ISRCs=("ISRC_clean", "nunique"),
        Unique_Artists=("Artist", "nunique"),
    )
    .reset_index()
)

shared_title_multiple_isrc_df = (
    track_title_summary_df[
        track_title_summary_df["Unique_ISRCs"] > 1
    ]
    .sort_values(
        by=["Unique_ISRCs", "Records"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

shared_title_count = int(
    len(shared_title_multiple_isrc_df)
)

# ============================================================
# Overall data-quality overview
# ============================================================

missing_duplicate_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(release_quality_df),
        },
        {
            "Metric": "Core identity fields assessed",
            "Observed Value": len(available_core_identity_fields),
        },
        {
            "Metric": "Complete duplicate records",
            "Observed Value": full_duplicate_records,
        },
        {
            "Metric": "Complete duplicate groups",
            "Observed Value": full_duplicate_groups,
        },
        {
            "Metric": "Duplicate ISRC records",
            "Observed Value": duplicate_isrc_records,
        },
        {
            "Metric": "Duplicate ISRC values",
            "Observed Value": duplicate_isrc_values,
        },
        {
            "Metric": "Repeated track-title + artist pairs",
            "Observed Value": repeated_title_artist_pairs,
        },
        {
            "Metric": "Repeated normalised track + artist pairs",
            "Observed Value": repeated_normalised_pairs,
        },
        {
            "Metric": "Track titles linked to multiple ISRCs",
            "Observed Value": shared_title_count,
        },
    ]
)

print("Assessing PMIP missing and duplicate release records")
print("=" * 80)

display(missing_duplicate_overview_df)

# ============================================================
# Core identity missingness
# ============================================================

print()
print("Core release-identity missingness")
print("=" * 80)

display(core_identity_missing_df)

# ============================================================
# Fields containing missing values
# ============================================================

fields_with_missing_df = missing_summary_df[
    missing_summary_df["Missing Values"] > 0
].copy()

print()
print("Fields containing missing values")
print("=" * 80)

if fields_with_missing_df.empty:
    print("No missing values detected.")
else:
    display(fields_with_missing_df)

# ============================================================
# Repeated title + artist combinations
# ============================================================

if not repeated_title_artist_df.empty:

    print()
    print("Repeated track-title and artist combinations")
    print("=" * 80)

    display(
        repeated_title_artist_df.head(30)
    )

# ============================================================
# Repeated normalised identities
# ============================================================

if not repeated_normalised_identity_df.empty:

    print()
    print("Repeated normalised release identities")
    print("=" * 80)

    display(
        repeated_normalised_identity_df.head(30)
    )

# ============================================================
# Shared track titles across multiple ISRCs
# ============================================================

if not shared_title_multiple_isrc_df.empty:

    print()
    print("Track titles associated with multiple ISRC values")
    print("=" * 80)

    display(
        shared_title_multiple_isrc_df.head(30)
    )

# ============================================================
# Validation checks
# ============================================================

core_identity_complete = bool(
    (
        core_identity_missing_df["Missing Values"] == 0
    ).all()
    and (
        core_identity_missing_df["Blank Values"] == 0
    ).all()
)

no_full_duplicates = bool(
    full_duplicate_records == 0
)

no_duplicate_isrcs = bool(
    duplicate_isrc_values == 0
)

missing_duplicate_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Core release identity complete",
            "Passed": core_identity_complete,
        },
        {
            "Validation Area": "No complete duplicate rows",
            "Passed": no_full_duplicates,
        },
        {
            "Validation Area": "No duplicate ISRC values",
            "Passed": no_duplicate_isrcs,
        },
        {
            "Validation Area": "Repeated-title patterns assessed",
            "Passed": True,
        },
        {
            "Validation Area": "Normalised duplicate patterns assessed",
            "Passed": True,
        },
        {
            "Validation Area": "Missing-value profile prepared",
            "Passed": True,
        },
    ]
)

print()
print("Missing and duplicate record validation")
print("=" * 80)

display(missing_duplicate_validation_df)

# ============================================================
# Section validation
# ============================================================

section_4_5_complete = bool(
    missing_duplicate_validation_df["Passed"].all()
)

print()
print(f"Section 4.5 completion status: {section_4_5_complete}")

if not section_4_5_complete:
    print(
        "One or more core data-quality issues require attention "
        "before the final release population is created."
    )
```

### Interpretation

The PMIP missing-value and duplicate-record assessment completed successfully.

All core release identity fields remain complete, and no complete duplicate rows were identified within the validated release population.

No duplicate ISRC values were detected, confirming that each release continues to retain a unique primary identifier.

Repeated track-title and normalised identity patterns were assessed separately rather than being treated automatically as duplicates. This is important because repeated titles may represent valid alternative versions, reissues, collaborations or distinct recordings.

The missing-value profile was also prepared successfully. Any missing values that remain in optional platform-performance fields should therefore be interpreted as incomplete source coverage rather than invalid release identity.

Section 4.5 completed successfully, confirming that the release population is structurally suitable for final population construction.

## 4.6 Final Release Population

### Purpose

This subsection constructs and validates the final release-level analytical population for PMIP Release Performance Intelligence.

The final population is built from the validated primary dataset after completing release-definition, identifier, artist-relationship, date, missing-value and duplicate-record checks.

The aim is to ensure that every retained release:

- has a valid unique ISRC;
- has complete track, artist and release-date information;
- retains normalised track and artist identity fields;
- contains no complete duplicate record;
- remains traceable to the original primary dataset;
- preserves temporal-review flags where chronology requires caution.

The resulting dataframe will become the main release-level population used throughout the remaining performance analysis.

```python
# ============================================================
# Section 4.6 - Final Release Population
# ============================================================

if not section_4_5_complete:
    raise RuntimeError(
        "Section 4.5 must be completed before constructing the final release population."
    )

# ============================================================
# Start from validated release-quality population
# ============================================================

final_release_df = release_quality_df.copy()

# ============================================================
# Retain / standardise final identity fields
# ============================================================

final_release_df["release_id"] = (
    final_release_df["ISRC_clean"]
)

final_release_df["release_date"] = (
    final_release_df["release_date_parsed"]
)

# ============================================================
# Temporal review flag
# ============================================================

final_release_df["temporal_review_required"] = (
    final_release_df["first_chart_date_parsed"].notna()
    & final_release_df["release_date"].notna()
    & (
        final_release_df["first_chart_date_parsed"]
        < final_release_df["release_date"]
    )
)

# ============================================================
# Core identity completeness flag
# ============================================================

final_release_df["core_identity_complete"] = (
    final_release_df[
        [
            "release_id",
            "Track",
            "Artist",
            "release_date",
            "track_normalised",
            "artist_normalised",
        ]
    ]
    .notna()
    .all(axis=1)
)

# ============================================================
# Duplicate safety checks
# ============================================================

final_release_df["duplicate_release_id"] = (
    final_release_df["release_id"]
    .duplicated(keep=False)
)

final_release_df["complete_duplicate_record"] = (
    final_release_df.duplicated(
        subset=[
            "release_id",
            "Track",
            "Artist",
            "release_date",
        ],
        keep=False,
    )
)

# ============================================================
# Final retention rule
# ============================================================

final_release_df["retained_for_analysis"] = (
    final_release_df["core_identity_complete"]
    & ~final_release_df["duplicate_release_id"]
    & ~final_release_df["complete_duplicate_record"]
)

# ============================================================
# Final retained analytical population
# ============================================================

release_analysis_df = (
    final_release_df[
        final_release_df["retained_for_analysis"]
    ]
    .copy()
    .reset_index(drop=True)
)

# ============================================================
# Final population summary
# ============================================================

final_release_population_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Input release records",
            "Observed Value": len(final_release_df),
        },
        {
            "Metric": "Final retained releases",
            "Observed Value": len(release_analysis_df),
        },
        {
            "Metric": "Unique final release IDs",
            "Observed Value": release_analysis_df["release_id"].nunique(),
        },
        {
            "Metric": "Unique artists",
            "Observed Value": release_analysis_df["Artist"].nunique(),
        },
        {
            "Metric": "Unique track titles",
            "Observed Value": release_analysis_df["Track"].nunique(),
        },
        {
            "Metric": "Core-incomplete records excluded",
            "Observed Value": int(
                (~final_release_df["core_identity_complete"]).sum()
            ),
        },
        {
            "Metric": "Duplicate-ID records excluded",
            "Observed Value": int(
                final_release_df["duplicate_release_id"].sum()
            ),
        },
        {
            "Metric": "Complete duplicate records excluded",
            "Observed Value": int(
                final_release_df["complete_duplicate_record"].sum()
            ),
        },
        {
            "Metric": "Temporal-review releases retained",
            "Observed Value": int(
                release_analysis_df[
                    "temporal_review_required"
                ].sum()
            ),
        },
    ]
)

print("Constructing final PMIP release population")
print("=" * 80)

display(final_release_population_overview_df)

# ============================================================
# Final release population preview
# ============================================================

preview_columns = [
    "release_id",
    "Track",
    "Artist",
    "release_date",
    "track_normalised",
    "artist_normalised",
    "temporal_review_required",
    "retained_for_analysis",
]

preview_columns = [
    column
    for column in preview_columns
    if column in release_analysis_df.columns
]

print()
print("Final release population preview")
print("=" * 80)

display(
    release_analysis_df[
        preview_columns
    ].head(20)
)

# ============================================================
# Final population validation
# ============================================================

final_population_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Final release population available",
            "Passed": len(release_analysis_df) > 0,
        },
        {
            "Validation Area": "Release IDs remain unique",
            "Passed": (
                release_analysis_df["release_id"].nunique()
                == len(release_analysis_df)
            ),
        },
        {
            "Validation Area": "Core identity complete",
            "Passed": (
                release_analysis_df["core_identity_complete"].all()
            ),
        },
        {
            "Validation Area": "No duplicate release IDs",
            "Passed": (
                ~release_analysis_df["duplicate_release_id"]
            ).all(),
        },
        {
            "Validation Area": "No complete duplicate records",
            "Passed": (
                ~release_analysis_df["complete_duplicate_record"]
            ).all(),
        },
        {
            "Validation Area": "Temporal review flags retained",
            "Passed": (
                "temporal_review_required"
                in release_analysis_df.columns
            ),
        },
        {
            "Validation Area": "Original population preserved where valid",
            "Passed": (
                len(release_analysis_df)
                <= len(release_primary_df)
            ),
        },
    ]
)

print()
print("Final release population validation")
print("=" * 80)

display(final_population_validation_df)

# ============================================================
# Section 4 completion status
# ============================================================

section_4_6_complete = bool(
    final_population_validation_df["Passed"].all()
)

section_4_complete = bool(
    section_4_1_complete
    and section_4_2_complete
    and section_4_3_complete
    and section_4_4_complete
    and section_4_5_complete
    and section_4_6_complete
)

print()
print(f"Section 4.6 completion status: {section_4_6_complete}")
print(f"Section 4 overall completion status: {section_4_complete}")

if not section_4_complete:
    raise RuntimeError(
        "Section 4 Release Identity and Data Preparation "
        "did not complete successfully."
    )
```

### Interpretation

The final PMIP release-level analytical population has been constructed successfully.

All 4,593 validated release records were retained, meaning no releases were removed because of incomplete identity information, duplicate ISRC values or complete duplicate records.

Each retained release has a unique ISRC-based `release_id`, complete track and artist information, a valid release date, and normalised identity fields.

The final population contains 1,999 unique artists and 4,365 unique track titles.

A total of 77 releases remain flagged with `temporal_review_required = True` because their available chart history begins before the recorded release date. These records were intentionally retained rather than removed because the discrepancy may reflect reissues, remasters, alternate versions or metadata-matching differences.

The temporal-review flag therefore preserves important data-quality context while allowing the releases to remain available for broader performance analysis.

All final population validation checks passed successfully.

Section 4 is now complete, and `release_analysis_df` provides the validated release-level population that will be used throughout the remaining PMIP Release Performance Intelligence workflow.

# 5. Streaming Performance Intelligence

This section evaluates how strongly each validated release performs from a streaming perspective.

The aim is to move beyond raw stream counts and build a more balanced view of release performance by considering both the size and behaviour of streaming activity.

The analysis will focus on:

- overall streaming volume;
- change in streaming performance over time;
- streaming growth rate;
- persistence of streaming performance;
- stability and volatility of streaming behaviour;
- construction of a combined streaming-performance indicator.

This stage is important because a release with high total streams is not necessarily still growing, while a smaller release may show stronger recent momentum, persistence or stability.

The resulting streaming intelligence will later contribute to the broader PMIP Release Performance Intelligence framework.


## 5.1 Streaming Volume

### Purpose

This subsection measures the overall streaming volume associated with each validated release.

The aim is to establish the absolute scale of streaming performance before analysing change, growth, persistence or stability.

The assessment will examine:

- current Spotify stream volume;
- historical streaming totals where available;
- average and maximum historical streaming levels;
- distribution of streaming volume across releases;
- releases with unusually high or low streaming volume;
- missing streaming evidence;
- whether historical streaming coverage is available for each release.

Streaming volume provides an important baseline measure, but it will not be treated as a complete measure of release success on its own.



```python
# ============================================================
# Section 5.1 - Streaming Volume
# ============================================================

if not section_4_complete:
    raise RuntimeError(
        "Section 4 must be completed before beginning streaming performance analysis."
    )

# ============================================================
# Working copy
# ============================================================

streaming_volume_df = release_analysis_df.copy()

# ============================================================
# Streaming fields
# ============================================================

streaming_fields = [
    "Spotify Streams",
    "total_historical_streams",
    "average_historical_streams",
    "max_historical_streams",
]

available_streaming_fields = [
    column
    for column in streaming_fields
    if column in streaming_volume_df.columns
]

# ============================================================
# Numeric coercion
# ============================================================

for column in available_streaming_fields:
    streaming_volume_df[column] = pd.to_numeric(
        streaming_volume_df[column],
        errors="coerce"
    )

# ============================================================
# Missingness summary
# ============================================================

streaming_missing_rows = []

for column in available_streaming_fields:

    streaming_missing_rows.append(
        {
            "Streaming Field": column,
            "Available Values": int(
                streaming_volume_df[column].notna().sum()
            ),
            "Missing Values": int(
                streaming_volume_df[column].isna().sum()
            ),
            "Coverage (%)": round(
                streaming_volume_df[column].notna().mean() * 100,
                6
            ),
        }
    )

streaming_volume_coverage_df = pd.DataFrame(
    streaming_missing_rows
)

# ============================================================
# Primary streaming-volume field
# ============================================================

primary_streaming_field = (
    "Spotify Streams"
    if "Spotify Streams" in streaming_volume_df.columns
    else None
)

if primary_streaming_field is None:
    raise RuntimeError(
        "Spotify Streams is not available in the final release population."
    )

# ============================================================
# Streaming-volume descriptive statistics
# ============================================================

spotify_streams_valid = (
    streaming_volume_df[primary_streaming_field]
    .dropna()
)

streaming_volume_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(streaming_volume_df),
        },
        {
            "Metric": "Releases with Spotify stream evidence",
            "Observed Value": int(
                streaming_volume_df[
                    primary_streaming_field
                ].notna().sum()
            ),
        },
        {
            "Metric": "Releases missing Spotify stream evidence",
            "Observed Value": int(
                streaming_volume_df[
                    primary_streaming_field
                ].isna().sum()
            ),
        },
        {
            "Metric": "Minimum Spotify streams",
            "Observed Value": (
                float(spotify_streams_valid.min())
                if not spotify_streams_valid.empty
                else np.nan
            ),
        },
        {
            "Metric": "Median Spotify streams",
            "Observed Value": (
                float(spotify_streams_valid.median())
                if not spotify_streams_valid.empty
                else np.nan
            ),
        },
        {
            "Metric": "Mean Spotify streams",
            "Observed Value": (
                float(spotify_streams_valid.mean())
                if not spotify_streams_valid.empty
                else np.nan
            ),
        },
        {
            "Metric": "Maximum Spotify streams",
            "Observed Value": (
                float(spotify_streams_valid.max())
                if not spotify_streams_valid.empty
                else np.nan
            ),
        },
    ]
)

print("Assessing PMIP streaming volume")
print("=" * 80)

display(streaming_volume_summary_df)

# ============================================================
# Streaming-field coverage
# ============================================================

print()
print("Streaming evidence coverage")
print("=" * 80)

display(streaming_volume_coverage_df)

# ============================================================
# Streaming-volume percentiles
# ============================================================

streaming_percentiles = [
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
    0.99,
]

streaming_percentile_values = (
    spotify_streams_valid
    .quantile(streaming_percentiles)
)

streaming_volume_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
            "99%",
        ],
        "Spotify Streams": streaming_percentile_values.values,
    }
)

print()
print("Spotify streaming-volume percentiles")
print("=" * 80)

display(streaming_volume_percentiles_df)

# ============================================================
# Streaming-volume ranking
# ============================================================

streaming_volume_df["streaming_volume_rank"] = (
    streaming_volume_df[
        primary_streaming_field
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# Higher stream volume = higher percentile
streaming_volume_df["streaming_volume_percentile"] = (
    streaming_volume_df[
        primary_streaming_field
    ]
    .rank(
        method="average",
        pct=True,
        ascending=True
    )
    * 100
)

# ============================================================
# Streaming-volume classes
# ============================================================

valid_stream_mask = (
    streaming_volume_df[
        primary_streaming_field
    ].notna()
)

streaming_volume_df["streaming_volume_class"] = pd.NA

if valid_stream_mask.any():

    q50 = streaming_volume_df.loc[
        valid_stream_mask,
        primary_streaming_field
    ].quantile(0.50)

    q75 = streaming_volume_df.loc[
        valid_stream_mask,
        primary_streaming_field
    ].quantile(0.75)

    q90 = streaming_volume_df.loc[
        valid_stream_mask,
        primary_streaming_field
    ].quantile(0.90)

    q95 = streaming_volume_df.loc[
        valid_stream_mask,
        primary_streaming_field
    ].quantile(0.95)

    streaming_volume_df.loc[
        valid_stream_mask
        & (
            streaming_volume_df[
                primary_streaming_field
            ] < q50
        ),
        "streaming_volume_class"
    ] = "Lower Volume"

    streaming_volume_df.loc[
        valid_stream_mask
        & (
            streaming_volume_df[
                primary_streaming_field
            ] >= q50
        )
        & (
            streaming_volume_df[
                primary_streaming_field
            ] < q75
        ),
        "streaming_volume_class"
    ] = "Moderate Volume"

    streaming_volume_df.loc[
        valid_stream_mask
        & (
            streaming_volume_df[
                primary_streaming_field
            ] >= q75
        )
        & (
            streaming_volume_df[
                primary_streaming_field
            ] < q90
        ),
        "streaming_volume_class"
    ] = "High Volume"

    streaming_volume_df.loc[
        valid_stream_mask
        & (
            streaming_volume_df[
                primary_streaming_field
            ] >= q90
        )
        & (
            streaming_volume_df[
                primary_streaming_field
            ] < q95
        ),
        "streaming_volume_class"
    ] = "Very High Volume"

    streaming_volume_df.loc[
        valid_stream_mask
        & (
            streaming_volume_df[
                primary_streaming_field
            ] >= q95
        ),
        "streaming_volume_class"
    ] = "Exceptional Volume"

streaming_volume_df["streaming_volume_class"] = (
    streaming_volume_df[
        "streaming_volume_class"
    ].fillna("No Streaming Evidence")
)

# ============================================================
# Streaming-volume class distribution
# ============================================================

streaming_volume_class_distribution_df = (
    streaming_volume_df[
        "streaming_volume_class"
    ]
    .value_counts()
    .rename_axis("Streaming Volume Class")
    .reset_index(name="Releases")
)

streaming_volume_class_distribution_df["Share (%)"] = (
    streaming_volume_class_distribution_df["Releases"]
    / len(streaming_volume_df)
    * 100
)

print()
print("Streaming-volume class distribution")
print("=" * 80)

display(streaming_volume_class_distribution_df)

# ============================================================
# Highest streaming-volume releases
# ============================================================

top_streaming_volume_df = (
    streaming_volume_df[
        [
            "release_id",
            "Track",
            "Artist",
            primary_streaming_field,
            "streaming_volume_rank",
            "streaming_volume_percentile",
            "streaming_volume_class",
        ]
    ]
    .sort_values(
        by=primary_streaming_field,
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest streaming-volume releases")
print("=" * 80)

display(top_streaming_volume_df)

# ============================================================
# Section validation
# ============================================================

streaming_volume_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Final release population available",
            "Passed": len(streaming_volume_df) > 0,
        },
        {
            "Validation Area": "Spotify Streams field available",
            "Passed": primary_streaming_field is not None,
        },
        {
            "Validation Area": "Streaming evidence available",
            "Passed": streaming_volume_df[
                primary_streaming_field
            ].notna().any(),
        },
        {
            "Validation Area": "Streaming-volume ranking prepared",
            "Passed": "streaming_volume_rank" in streaming_volume_df.columns,
        },
        {
            "Validation Area": "Streaming-volume percentile prepared",
            "Passed": (
                "streaming_volume_percentile"
                in streaming_volume_df.columns
            ),
        },
        {
            "Validation Area": "Streaming-volume classes assigned",
            "Passed": (
                streaming_volume_df[
                    "streaming_volume_class"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Streaming-volume validation")
print("=" * 80)

display(streaming_volume_validation_df)

# ============================================================
# Section completion
# ============================================================

section_5_1_complete = bool(
    streaming_volume_validation_df["Passed"].all()
)

print()
print(f"Section 5.1 completion status: {section_5_1_complete}")

if not section_5_1_complete:
    raise RuntimeError(
        "Streaming Volume analysis did not complete successfully."
    )
```

### Interpretation

The PMIP streaming-volume assessment completed successfully.

Spotify streaming evidence is available for 4,485 of the 4,593 validated releases, representing approximately 97.65% coverage. Only 108 releases do not contain Spotify stream evidence.

The streaming distribution is strongly right-skewed. The median release has approximately 239.9 million Spotify streams, while the mean is approximately 447.4 million. This indicates that a relatively small number of extremely high-streaming releases increase the overall average.

Historical streaming evidence is available for 2,235 releases, representing approximately 48.66% of the final population. Historical measures therefore provide useful supporting evidence but cannot be treated as complete coverage across all releases.

## 5.2 Streaming Change

### Purpose

This subsection measures how streaming performance changes over time for releases with historical streaming evidence.

The aim is to move beyond accumulated stream volume and identify whether a release has increased, decreased or remained relatively stable across its observed history.

The assessment will examine:

- first observed historical streaming level;
- most recent observed historical streaming level;
- absolute streaming change;
- percentage streaming change;
- direction of change;
- releases with increasing, decreasing or stable streaming behaviour;
- coverage limitations where historical streaming evidence is unavailable.

Streaming change provides a more dynamic view of release performance and will later contribute to the broader streaming-growth and streaming-performance indicators.

```python
# ============================================================
# Section 5.2 - Streaming Change
# ============================================================

if not section_5_1_complete:
    raise RuntimeError(
        "Section 5.1 must be completed before analysing streaming change."
    )

# ============================================================
# Working source
# ============================================================

streaming_change_df = streaming_volume_df.copy()

# ============================================================
# Build historical release-level stream summaries from
# charts_cleaned.csv using track_id
# ============================================================

history_source = release_history_df.copy()

history_source["date"] = pd.to_datetime(
    history_source["date"],
    errors="coerce"
)

history_source["streams"] = pd.to_numeric(
    history_source["streams"],
    errors="coerce"
)

# Remove unusable historical observations
history_source = history_source[
    history_source["track_id"].notna()
    & history_source["date"].notna()
    & history_source["streams"].notna()
].copy()

# ============================================================
# First and last historical streaming observations per track
# ============================================================

history_source = history_source.sort_values(
    by=["track_id", "date"]
)

first_stream_df = (
    history_source
    .groupby("track_id", as_index=False)
    .first()[
        ["track_id", "date", "streams"]
    ]
    .rename(
        columns={
            "date": "first_stream_date",
            "streams": "first_historical_streams",
        }
    )
)

last_stream_df = (
    history_source
    .groupby("track_id", as_index=False)
    .last()[
        ["track_id", "date", "streams"]
    ]
    .rename(
        columns={
            "date": "last_stream_date",
            "streams": "last_historical_streams",
        }
    )
)

historical_stream_change_df = first_stream_df.merge(
    last_stream_df,
    on="track_id",
    how="inner",
)

# ============================================================
# Number of historical observations
# ============================================================

historical_observation_counts_df = (
    history_source
    .groupby("track_id")
    .size()
    .rename("historical_stream_observations")
    .reset_index()
)

historical_stream_change_df = historical_stream_change_df.merge(
    historical_observation_counts_df,
    on="track_id",
    how="left",
)

# ============================================================
# Calculate historical stream change
# ============================================================

historical_stream_change_df["absolute_stream_change"] = (
    historical_stream_change_df["last_historical_streams"]
    - historical_stream_change_df["first_historical_streams"]
)

historical_stream_change_df["percentage_stream_change"] = np.where(
    historical_stream_change_df["first_historical_streams"] > 0,
    (
        historical_stream_change_df["absolute_stream_change"]
        / historical_stream_change_df["first_historical_streams"]
    )
    * 100,
    np.nan,
)

historical_stream_change_df["stream_change_direction"] = np.select(
    [
        historical_stream_change_df["absolute_stream_change"] > 0,
        historical_stream_change_df["absolute_stream_change"] < 0,
        historical_stream_change_df["absolute_stream_change"] == 0,
    ],
    [
        "Increasing",
        "Decreasing",
        "Stable",
    ],
    default="Unavailable",
)

# ============================================================
# Attach historical change evidence to release population
#
# spotify_integrated.csv does not expose track_id directly,
# so use the integrated historical summaries to identify
# compatible release rows where possible.
# ============================================================

integrated_history_columns = [
    "total_historical_streams",
    "average_historical_streams",
    "max_historical_streams",
    "first_chart_date",
    "last_chart_date",
]

history_supported_mask = (
    streaming_change_df[
        integrated_history_columns
    ]
    .notna()
    .any(axis=1)
)

# Prepare fields for later population-level use
streaming_change_df["historical_stream_evidence_available"] = (
    history_supported_mask
)

# ============================================================
# Use integrated historical evidence to define a comparable
# change-strength measure where exact first/last streams
# cannot be joined safely by track_id.
# ============================================================

streaming_change_df["historical_stream_range"] = (
    streaming_change_df["max_historical_streams"]
    - streaming_change_df["average_historical_streams"]
)

streaming_change_df["relative_historical_stream_range_pct"] = np.where(
    streaming_change_df["average_historical_streams"] > 0,
    (
        streaming_change_df["historical_stream_range"]
        / streaming_change_df["average_historical_streams"]
    )
    * 100,
    np.nan,
)

# ============================================================
# Streaming change class
# ============================================================

streaming_change_df["streaming_change_class"] = "No Historical Evidence"

valid_change_mask = (
    streaming_change_df[
        "relative_historical_stream_range_pct"
    ].notna()
)

if valid_change_mask.any():

    q25 = streaming_change_df.loc[
        valid_change_mask,
        "relative_historical_stream_range_pct"
    ].quantile(0.25)

    q50 = streaming_change_df.loc[
        valid_change_mask,
        "relative_historical_stream_range_pct"
    ].quantile(0.50)

    q75 = streaming_change_df.loc[
        valid_change_mask,
        "relative_historical_stream_range_pct"
    ].quantile(0.75)

    streaming_change_df.loc[
        valid_change_mask
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] <= q25
        ),
        "streaming_change_class"
    ] = "Low Historical Variation"

    streaming_change_df.loc[
        valid_change_mask
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] > q25
        )
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] <= q50
        ),
        "streaming_change_class"
    ] = "Moderate Historical Variation"

    streaming_change_df.loc[
        valid_change_mask
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] > q50
        )
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] <= q75
        ),
        "streaming_change_class"
    ] = "High Historical Variation"

    streaming_change_df.loc[
        valid_change_mask
        & (
            streaming_change_df[
                "relative_historical_stream_range_pct"
            ] > q75
        ),
        "streaming_change_class"
    ] = "Very High Historical Variation"

# ============================================================
# Streaming change summary
# ============================================================

streaming_change_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(streaming_change_df),
        },
        {
            "Metric": "Releases with historical stream evidence",
            "Observed Value": int(
                streaming_change_df[
                    "historical_stream_evidence_available"
                ].sum()
            ),
        },
        {
            "Metric": "Releases without historical stream evidence",
            "Observed Value": int(
                (
                    ~streaming_change_df[
                        "historical_stream_evidence_available"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Median relative historical stream range (%)",
            "Observed Value": (
                streaming_change_df[
                    "relative_historical_stream_range_pct"
                ].median()
            ),
        },
        {
            "Metric": "Mean relative historical stream range (%)",
            "Observed Value": (
                streaming_change_df[
                    "relative_historical_stream_range_pct"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum relative historical stream range (%)",
            "Observed Value": (
                streaming_change_df[
                    "relative_historical_stream_range_pct"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP streaming change")
print("=" * 80)

display(streaming_change_summary_df)

# ============================================================
# Streaming change class distribution
# ============================================================

streaming_change_class_distribution_df = (
    streaming_change_df[
        "streaming_change_class"
    ]
    .value_counts()
    .rename_axis("Streaming Change Class")
    .reset_index(name="Releases")
)

streaming_change_class_distribution_df["Share (%)"] = (
    streaming_change_class_distribution_df["Releases"]
    / len(streaming_change_df)
    * 100
)

print()
print("Streaming-change class distribution")
print("=" * 80)

display(streaming_change_class_distribution_df)

# ============================================================
# Strongest historical variation
# ============================================================

top_streaming_change_df = (
    streaming_change_df[
        [
            "release_id",
            "Track",
            "Artist",
            "average_historical_streams",
            "max_historical_streams",
            "historical_stream_range",
            "relative_historical_stream_range_pct",
            "streaming_change_class",
        ]
    ]
    .sort_values(
        by="relative_historical_stream_range_pct",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest historical streaming variation")
print("=" * 80)

display(top_streaming_change_df)

# ============================================================
# Validation
# ============================================================

streaming_change_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.1 completed",
            "Passed": section_5_1_complete,
        },
        {
            "Validation Area": "Historical streaming evidence available",
            "Passed": streaming_change_df[
                "historical_stream_evidence_available"
            ].any(),
        },
        {
            "Validation Area": "Historical stream range calculated",
            "Passed": (
                "historical_stream_range"
                in streaming_change_df.columns
            ),
        },
        {
            "Validation Area": "Relative historical stream variation calculated",
            "Passed": (
                "relative_historical_stream_range_pct"
                in streaming_change_df.columns
            ),
        },
        {
            "Validation Area": "Streaming change classes assigned",
            "Passed": streaming_change_df[
                "streaming_change_class"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming-change validation")
print("=" * 80)

display(streaming_change_validation_df)

# ============================================================
# Completion
# ============================================================

section_5_2_complete = bool(
    streaming_change_validation_df["Passed"].all()
)

print()
print(f"Section 5.2 completion status: {section_5_2_complete}")

if not section_5_2_complete:
    raise RuntimeError(
        "Streaming Change analysis did not complete successfully."
    )
```

### Interpretation

The PMIP streaming-change assessment completed successfully.

Historical streaming evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not currently have sufficient historical stream evidence for this analysis.

Among releases with historical evidence, the median relative historical stream range is approximately 1,931.48%, while the mean is approximately 2,466.97%. This shows that historical streaming behaviour varies substantially across the observed release population.

The maximum relative historical stream range exceeds 23,589%, indicating that a small number of releases experienced extremely large differences between their average and maximum observed historical streaming levels.

The streaming-change classes separate releases into Low, Moderate, High and Very High Historical Variation groups. Releases without historical evidence remain explicitly identified rather than being assigned an artificial change level.

The class distribution is broadly balanced across the four historical-variation groups because percentile-based thresholds were used to define the categories.

It is important to note that this subsection measures the magnitude of historical streaming variation rather than the direction of growth. A release with very high variation may have experienced strong growth, sharp decline or substantial fluctuation over time.

Section 5.2 therefore provides a useful measure of historical streaming movement, while the next subsection will assess growth direction and rate more directly.

Section 5.2 completed successfully.

## 5.3 Streaming Growth Rate

### Purpose

This subsection measures the rate at which release-level streaming performance has grown or declined across the available historical chart period.

While Section 5.2 focused on the size of historical streaming variation, this subsection focuses more directly on growth direction and growth strength.

The assessment will estimate:

- the first observed historical streaming level;
- the last observed historical streaming level;
- absolute streaming growth;
- percentage streaming growth;
- chart-history duration in days;
- average daily streaming growth rate;
- growth direction;
- growth-rate classes for analytical comparison.

Because the final PMIP release population does not store raw chart-history observations directly, this subsection reconnects release records to their historical chart signatures using validated integrated historical evidence.

This allows PMIP to move from simple streaming size and variation toward a more dynamic measure of release momentum within the streaming-performance component.

```python
# ============================================================
# Section 5.3 - Streaming Growth Rate
# ============================================================

if not section_5_2_complete:
    raise RuntimeError(
        "Section 5.2 must be completed before analysing streaming growth rate."
    )

# ============================================================
# Working source
# ============================================================

streaming_growth_rate_df = streaming_change_df.copy()

# ============================================================
# Prepare historical chart source
# ============================================================

history_source = release_history_df.copy()

history_source["date"] = pd.to_datetime(
    history_source["date"],
    errors="coerce"
)

history_source["streams"] = pd.to_numeric(
    history_source["streams"],
    errors="coerce"
)

history_source = history_source[
    history_source["track_id"].notna()
    & history_source["date"].notna()
    & history_source["streams"].notna()
].copy()

history_source = history_source.sort_values(
    by=["track_id", "date"]
)

# ============================================================
# Build track-level historical growth signatures
# ============================================================

historical_growth_signature_df = (
    history_source
    .groupby("track_id")
    .agg(
        first_historical_stream_date=("date", "min"),
        last_historical_stream_date=("date", "max"),
        first_historical_streams=("streams", "first"),
        last_historical_streams=("streams", "last"),
        historical_stream_observations=("streams", "size"),
        total_historical_streams=("streams", "sum"),
        average_historical_streams=("streams", "mean"),
        max_historical_streams=("streams", "max"),
    )
    .reset_index()
)

# ============================================================
# Build signature keys so integrated release rows can reconnect
# safely to their historical chart summaries
# ============================================================

signature_key_cols = [
    "first_chart_date_key",
    "last_chart_date_key",
    "total_historical_streams_key",
    "average_historical_streams_key",
    "max_historical_streams_key",
]

historical_growth_signature_df["first_chart_date_key"] = (
    historical_growth_signature_df["first_historical_stream_date"]
    .dt.strftime("%Y-%m-%d")
)

historical_growth_signature_df["last_chart_date_key"] = (
    historical_growth_signature_df["last_historical_stream_date"]
    .dt.strftime("%Y-%m-%d")
)

historical_growth_signature_df["total_historical_streams_key"] = (
    historical_growth_signature_df["total_historical_streams"]
    .round(6)
)

historical_growth_signature_df["average_historical_streams_key"] = (
    historical_growth_signature_df["average_historical_streams"]
    .round(6)
)

historical_growth_signature_df["max_historical_streams_key"] = (
    historical_growth_signature_df["max_historical_streams"]
    .round(6)
)

historical_signature_counts_df = (
    historical_growth_signature_df
    .groupby(signature_key_cols)
    .size()
    .rename("signature_match_count")
    .reset_index()
)

historical_growth_signature_unique_df = (
    historical_growth_signature_df
    .sort_values(by=["track_id"])
    .drop_duplicates(subset=signature_key_cols, keep="first")
    .merge(
        historical_signature_counts_df,
        on=signature_key_cols,
        how="left"
    )
)

# ============================================================
# Prepare keys on the release-level PMIP population
# ============================================================

streaming_growth_rate_df["first_chart_date_parsed"] = pd.to_datetime(
    streaming_growth_rate_df["first_chart_date"],
    errors="coerce"
)

streaming_growth_rate_df["last_chart_date_parsed"] = pd.to_datetime(
    streaming_growth_rate_df["last_chart_date"],
    errors="coerce"
)

streaming_growth_rate_df["first_chart_date_key"] = (
    streaming_growth_rate_df["first_chart_date_parsed"]
    .dt.strftime("%Y-%m-%d")
)

streaming_growth_rate_df["last_chart_date_key"] = (
    streaming_growth_rate_df["last_chart_date_parsed"]
    .dt.strftime("%Y-%m-%d")
)

streaming_growth_rate_df["total_historical_streams_key"] = (
    pd.to_numeric(
        streaming_growth_rate_df["total_historical_streams"],
        errors="coerce"
    )
    .round(6)
)

streaming_growth_rate_df["average_historical_streams_key"] = (
    pd.to_numeric(
        streaming_growth_rate_df["average_historical_streams"],
        errors="coerce"
    )
    .round(6)
)

streaming_growth_rate_df["max_historical_streams_key"] = (
    pd.to_numeric(
        streaming_growth_rate_df["max_historical_streams"],
        errors="coerce"
    )
    .round(6)
)

# ============================================================
# Reconnect release rows to historical first/last stream values
# ============================================================

merge_columns = signature_key_cols + [
    "track_id",
    "first_historical_stream_date",
    "last_historical_stream_date",
    "first_historical_streams",
    "last_historical_streams",
    "historical_stream_observations",
    "signature_match_count",
]

streaming_growth_rate_df = streaming_growth_rate_df.merge(
    historical_growth_signature_unique_df[merge_columns],
    on=signature_key_cols,
    how="left"
)

streaming_growth_rate_df["historical_signature_matched"] = (
    streaming_growth_rate_df["track_id"].notna()
)

streaming_growth_rate_df["historical_signature_ambiguous"] = (
    streaming_growth_rate_df["signature_match_count"].fillna(0) > 1
)

# ============================================================
# Calculate growth-rate features
# ============================================================

streaming_growth_rate_df["chart_history_days"] = (
    streaming_growth_rate_df["last_historical_stream_date"]
    - streaming_growth_rate_df["first_historical_stream_date"]
).dt.days

streaming_growth_rate_df["absolute_stream_growth"] = (
    streaming_growth_rate_df["last_historical_streams"]
    - streaming_growth_rate_df["first_historical_streams"]
)

streaming_growth_rate_df["percentage_stream_growth"] = np.where(
    streaming_growth_rate_df["first_historical_streams"] > 0,
    (
        streaming_growth_rate_df["absolute_stream_growth"]
        / streaming_growth_rate_df["first_historical_streams"]
    )
    * 100,
    np.nan,
)

streaming_growth_rate_df["daily_stream_growth_rate"] = np.where(
    streaming_growth_rate_df["chart_history_days"] > 0,
    (
        streaming_growth_rate_df["absolute_stream_growth"]
        / streaming_growth_rate_df["chart_history_days"]
    ),
    np.nan,
)

streaming_growth_rate_df["stream_growth_direction"] = np.select(
    [
        streaming_growth_rate_df["absolute_stream_growth"] > 0,
        streaming_growth_rate_df["absolute_stream_growth"] < 0,
        streaming_growth_rate_df["absolute_stream_growth"] == 0,
    ],
    [
        "Increasing",
        "Decreasing",
        "Stable",
    ],
    default="Unavailable",
)

# ============================================================
# Growth-rate classes
# ============================================================

streaming_growth_rate_df["streaming_growth_rate_class"] = (
    "No Historical Match"
)

valid_growth_mask = (
    streaming_growth_rate_df["historical_signature_matched"]
    & streaming_growth_rate_df["percentage_stream_growth"].notna()
)

positive_growth_mask = (
    valid_growth_mask
    & (
        streaming_growth_rate_df["percentage_stream_growth"] > 0
    )
)

positive_growth_q50 = np.nan
positive_growth_q75 = np.nan

if positive_growth_mask.any():
    positive_growth_q50 = streaming_growth_rate_df.loc[
        positive_growth_mask,
        "percentage_stream_growth"
    ].quantile(0.50)

    positive_growth_q75 = streaming_growth_rate_df.loc[
        positive_growth_mask,
        "percentage_stream_growth"
    ].quantile(0.75)

streaming_growth_rate_df.loc[
    valid_growth_mask
    & (
        streaming_growth_rate_df["percentage_stream_growth"] < 0
    ),
    "streaming_growth_rate_class"
] = "Negative Growth"

streaming_growth_rate_df.loc[
    valid_growth_mask
    & (
        streaming_growth_rate_df["percentage_stream_growth"] == 0
    ),
    "streaming_growth_rate_class"
] = "Stable Growth"

if positive_growth_mask.any():
    streaming_growth_rate_df.loc[
        positive_growth_mask
        & (
            streaming_growth_rate_df["percentage_stream_growth"]
            <= positive_growth_q50
        ),
        "streaming_growth_rate_class"
    ] = "Moderate Positive Growth"

    streaming_growth_rate_df.loc[
        positive_growth_mask
        & (
            streaming_growth_rate_df["percentage_stream_growth"]
            > positive_growth_q50
        )
        & (
            streaming_growth_rate_df["percentage_stream_growth"]
            <= positive_growth_q75
        ),
        "streaming_growth_rate_class"
    ] = "High Positive Growth"

    streaming_growth_rate_df.loc[
        positive_growth_mask
        & (
            streaming_growth_rate_df["percentage_stream_growth"]
            > positive_growth_q75
        ),
        "streaming_growth_rate_class"
    ] = "Very High Positive Growth"

# ============================================================
# Summary
# ============================================================

streaming_growth_rate_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(streaming_growth_rate_df),
        },
        {
            "Metric": "Releases with historical signature match",
            "Observed Value": int(
                streaming_growth_rate_df[
                    "historical_signature_matched"
                ].sum()
            ),
        },
        {
            "Metric": "Releases without historical signature match",
            "Observed Value": int(
                (
                    ~streaming_growth_rate_df[
                        "historical_signature_matched"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Ambiguous historical signature matches",
            "Observed Value": int(
                streaming_growth_rate_df[
                    "historical_signature_ambiguous"
                ].sum()
            ),
        },
        {
            "Metric": "Median percentage stream growth",
            "Observed Value": (
                streaming_growth_rate_df[
                    "percentage_stream_growth"
                ].median()
            ),
        },
        {
            "Metric": "Mean percentage stream growth",
            "Observed Value": (
                streaming_growth_rate_df[
                    "percentage_stream_growth"
                ].mean()
            ),
        },
        {
            "Metric": "Median daily stream growth rate",
            "Observed Value": (
                streaming_growth_rate_df[
                    "daily_stream_growth_rate"
                ].median()
            ),
        },
        {
            "Metric": "Mean daily stream growth rate",
            "Observed Value": (
                streaming_growth_rate_df[
                    "daily_stream_growth_rate"
                ].mean()
            ),
        },
    ]
)

print("Assessing PMIP streaming growth rate")
print("=" * 80)

display(streaming_growth_rate_summary_df)

# ============================================================
# Growth direction distribution
# ============================================================

stream_growth_direction_distribution_df = (
    streaming_growth_rate_df["stream_growth_direction"]
    .value_counts(dropna=False)
    .rename_axis("Growth Direction")
    .reset_index(name="Releases")
)

stream_growth_direction_distribution_df["Share (%)"] = (
    stream_growth_direction_distribution_df["Releases"]
    / len(streaming_growth_rate_df)
    * 100
)

print()
print("Streaming-growth direction distribution")
print("=" * 80)

display(stream_growth_direction_distribution_df)

# ============================================================
# Growth-rate class distribution
# ============================================================

streaming_growth_rate_class_distribution_df = (
    streaming_growth_rate_df["streaming_growth_rate_class"]
    .value_counts(dropna=False)
    .rename_axis("Streaming Growth Rate Class")
    .reset_index(name="Releases")
)

streaming_growth_rate_class_distribution_df["Share (%)"] = (
    streaming_growth_rate_class_distribution_df["Releases"]
    / len(streaming_growth_rate_df)
    * 100
)

print()
print("Streaming-growth-rate class distribution")
print("=" * 80)

display(streaming_growth_rate_class_distribution_df)

# ============================================================
# Highest positive streaming growth
# ============================================================

top_streaming_growth_releases_df = (
    streaming_growth_rate_df[
        streaming_growth_rate_df["historical_signature_matched"]
        & streaming_growth_rate_df["percentage_stream_growth"].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "first_historical_stream_date",
            "last_historical_stream_date",
            "first_historical_streams",
            "last_historical_streams",
            "absolute_stream_growth",
            "percentage_stream_growth",
            "daily_stream_growth_rate",
            "streaming_growth_rate_class",
        ]
    ]
    .sort_values(
        by="percentage_stream_growth",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with the strongest positive streaming growth")
print("=" * 80)

display(top_streaming_growth_releases_df)

# ============================================================
# Validation
# ============================================================

streaming_growth_rate_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.2 completed",
            "Passed": section_5_2_complete,
        },
        {
            "Validation Area": "Historical signature matching prepared",
            "Passed": (
                "historical_signature_matched"
                in streaming_growth_rate_df.columns
            ),
        },
        {
            "Validation Area": "At least one historical signature matched",
            "Passed": streaming_growth_rate_df[
                "historical_signature_matched"
            ].any(),
        },
        {
            "Validation Area": "Absolute stream growth calculated",
            "Passed": (
                "absolute_stream_growth"
                in streaming_growth_rate_df.columns
            ),
        },
        {
            "Validation Area": "Percentage stream growth calculated",
            "Passed": (
                "percentage_stream_growth"
                in streaming_growth_rate_df.columns
            ),
        },
        {
            "Validation Area": "Daily stream growth rate calculated",
            "Passed": (
                "daily_stream_growth_rate"
                in streaming_growth_rate_df.columns
            ),
        },
        {
            "Validation Area": "Growth direction assigned",
            "Passed": streaming_growth_rate_df[
                "stream_growth_direction"
            ].notna().all(),
        },
        {
            "Validation Area": "Growth-rate classes assigned",
            "Passed": streaming_growth_rate_df[
                "streaming_growth_rate_class"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming-growth-rate validation")
print("=" * 80)

display(streaming_growth_rate_validation_df)

# ============================================================
# Completion
# ============================================================

section_5_3_complete = bool(
    streaming_growth_rate_validation_df["Passed"].all()
)

print()
print(f"Section 5.3 completion status: {section_5_3_complete}")

if not section_5_3_complete:
    raise RuntimeError(
        "Streaming Growth Rate analysis did not complete successfully."
    )
```

### Interpretation

The PMIP streaming-growth-rate assessment completed successfully.

Historical first-to-last streaming evidence was matched for 214 of the 4,593 validated releases. The remaining 4,379 releases did not receive a sufficiently reliable historical signature match and are therefore classified as having no historical growth-rate evidence.

This means that streaming growth rate is available for only a small subset of the overall release population. The results should therefore be interpreted as evidence for matched releases rather than as a complete description of all PMIP releases.

Among the matched releases, 131 show decreasing historical streaming levels, 49 show increasing streaming levels, and 34 remain stable across the observed period.

The median percentage streaming growth is approximately -26.23%, indicating that the typical matched release experienced a decline between its first and last observed historical streaming values.

However, the mean percentage growth is positive at approximately 218.55%. This large difference between the median and mean suggests that a small number of releases experienced extremely strong positive growth and pulled the average upward.

The strongest positive-growth releases show very large percentage increases because their first observed streaming values were relatively low compared with their later historical observations. These extreme values should therefore be interpreted together with the absolute growth amount and the length of the observed chart history.

The median daily streaming growth rate is negative, which is consistent with the larger number of declining releases among the matched population.

The growth-rate classes distinguish Negative Growth, Stable Growth, Moderate Positive Growth, High Positive Growth and Very High Positive Growth, while unmatched releases remain explicitly labelled as `No Historical Match`.

Section 5.3 therefore provides directional streaming-growth evidence for the subset of releases that could be matched reliably to historical chart records. Because coverage is limited, these growth-rate measures should later be combined with broader streaming-volume, persistence and stability evidence rather than used as a standalone performance measure.


## 5.4 Streaming Persistence

### Purpose

This subsection measures how consistently a release maintains streaming activity across its observed historical period.

While Section 5.3 focused on whether streaming levels increased or decreased, streaming persistence focuses on whether streaming activity is sustained across multiple historical observations rather than appearing only as a short-lived spike.

The assessment will examine:

- number of historical streaming observations;
- historical observation span;
- repeated streaming activity across time;
- consistency of non-zero streaming observations;
- persistence strength relative to the available history;
- releases with weak, moderate or strong sustained streaming presence.

Persistence is important because a release that maintains streaming activity over a longer period may represent more durable audience interest than a release that experiences only a brief peak.

```python
# ============================================================
# Section 5.4 - Streaming Persistence
# ============================================================

if not section_5_3_complete:
    raise RuntimeError(
        "Section 5.3 must be completed before analysing streaming persistence."
    )

# ============================================================
# Working source
# ============================================================

streaming_persistence_df = streaming_growth_rate_df.copy()

# ============================================================
# Historical observation evidence
# ============================================================

streaming_persistence_df["historical_observation_span_days"] = (
    streaming_persistence_df["last_historical_stream_date"]
    - streaming_persistence_df["first_historical_stream_date"]
).dt.days

streaming_persistence_df["repeated_historical_observations"] = (
    streaming_persistence_df["historical_stream_observations"]
    .fillna(0)
    .gt(1)
)

streaming_persistence_df["extended_historical_observations"] = (
    streaming_persistence_df["historical_stream_observations"]
    .fillna(0)
    .ge(5)
)

# ============================================================
# Persistence evidence score
#
# The score combines:
# - number of historical observations
# - observed history duration
#
# It intentionally does not depend on growth direction.
# ============================================================

valid_persistence_mask = (
    streaming_persistence_df["historical_signature_matched"]
    & streaming_persistence_df["historical_stream_observations"].notna()
)

streaming_persistence_df["observation_count_score"] = np.nan
streaming_persistence_df["history_span_score"] = np.nan

if valid_persistence_mask.any():

    max_observation_count = streaming_persistence_df.loc[
        valid_persistence_mask,
        "historical_stream_observations"
    ].max()

    max_history_span = streaming_persistence_df.loc[
        valid_persistence_mask,
        "historical_observation_span_days"
    ].max()

    if max_observation_count > 0:
        streaming_persistence_df.loc[
            valid_persistence_mask,
            "observation_count_score"
        ] = (
            streaming_persistence_df.loc[
                valid_persistence_mask,
                "historical_stream_observations"
            ]
            / max_observation_count
            * 100
        )

    if max_history_span > 0:
        streaming_persistence_df.loc[
            valid_persistence_mask,
            "history_span_score"
        ] = (
            streaming_persistence_df.loc[
                valid_persistence_mask,
                "historical_observation_span_days"
            ]
            / max_history_span
            * 100
        )

# ============================================================
# Combined persistence indicator
# ============================================================

streaming_persistence_df["streaming_persistence_score"] = np.where(
    valid_persistence_mask,
    (
        streaming_persistence_df["observation_count_score"] * 0.60
        + streaming_persistence_df["history_span_score"] * 0.40
    ),
    np.nan,
)

# ============================================================
# Persistence classes
# ============================================================

streaming_persistence_df["streaming_persistence_class"] = (
    "No Historical Match"
)

valid_score_mask = (
    streaming_persistence_df["streaming_persistence_score"].notna()
)

if valid_score_mask.any():

    q25 = streaming_persistence_df.loc[
        valid_score_mask,
        "streaming_persistence_score"
    ].quantile(0.25)

    q50 = streaming_persistence_df.loc[
        valid_score_mask,
        "streaming_persistence_score"
    ].quantile(0.50)

    q75 = streaming_persistence_df.loc[
        valid_score_mask,
        "streaming_persistence_score"
    ].quantile(0.75)

    streaming_persistence_df.loc[
        valid_score_mask
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] <= q25
        ),
        "streaming_persistence_class"
    ] = "Low Persistence"

    streaming_persistence_df.loc[
        valid_score_mask
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] > q25
        )
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] <= q50
        ),
        "streaming_persistence_class"
    ] = "Moderate Persistence"

    streaming_persistence_df.loc[
        valid_score_mask
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] > q50
        )
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] <= q75
        ),
        "streaming_persistence_class"
    ] = "High Persistence"

    streaming_persistence_df.loc[
        valid_score_mask
        & (
            streaming_persistence_df[
                "streaming_persistence_score"
            ] > q75
        ),
        "streaming_persistence_class"
    ] = "Very High Persistence"

# ============================================================
# Persistence summary
# ============================================================

streaming_persistence_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(streaming_persistence_df),
        },
        {
            "Metric": "Releases with persistence evidence",
            "Observed Value": int(
                valid_persistence_mask.sum()
            ),
        },
        {
            "Metric": "Releases without persistence evidence",
            "Observed Value": int(
                (~valid_persistence_mask).sum()
            ),
        },
        {
            "Metric": "Releases with repeated observations",
            "Observed Value": int(
                streaming_persistence_df[
                    "repeated_historical_observations"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with 5+ historical observations",
            "Observed Value": int(
                streaming_persistence_df[
                    "extended_historical_observations"
                ].sum()
            ),
        },
        {
            "Metric": "Median historical observation count",
            "Observed Value": (
                streaming_persistence_df.loc[
                    valid_persistence_mask,
                    "historical_stream_observations"
                ].median()
            ),
        },
        {
            "Metric": "Median historical observation span (days)",
            "Observed Value": (
                streaming_persistence_df.loc[
                    valid_persistence_mask,
                    "historical_observation_span_days"
                ].median()
            ),
        },
        {
            "Metric": "Median streaming persistence score",
            "Observed Value": (
                streaming_persistence_df[
                    "streaming_persistence_score"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP streaming persistence")
print("=" * 80)

display(streaming_persistence_summary_df)

# ============================================================
# Persistence class distribution
# ============================================================

streaming_persistence_class_distribution_df = (
    streaming_persistence_df[
        "streaming_persistence_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Streaming Persistence Class")
    .reset_index(name="Releases")
)

streaming_persistence_class_distribution_df["Share (%)"] = (
    streaming_persistence_class_distribution_df["Releases"]
    / len(streaming_persistence_df)
    * 100
)

print()
print("Streaming-persistence class distribution")
print("=" * 80)

display(streaming_persistence_class_distribution_df)

# ============================================================
# Highest-persistence releases
# ============================================================

top_streaming_persistence_df = (
    streaming_persistence_df[
        streaming_persistence_df[
            "streaming_persistence_score"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "historical_stream_observations",
            "historical_observation_span_days",
            "observation_count_score",
            "history_span_score",
            "streaming_persistence_score",
            "streaming_persistence_class",
        ]
    ]
    .sort_values(
        by="streaming_persistence_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest streaming persistence")
print("=" * 80)

display(top_streaming_persistence_df)

# ============================================================
# Validation
# ============================================================

streaming_persistence_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.3 completed",
            "Passed": section_5_3_complete,
        },
        {
            "Validation Area": "Historical observation counts available",
            "Passed": (
                "historical_stream_observations"
                in streaming_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Historical span calculated",
            "Passed": (
                "historical_observation_span_days"
                in streaming_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Persistence score calculated",
            "Passed": (
                "streaming_persistence_score"
                in streaming_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Persistence classes assigned",
            "Passed": streaming_persistence_df[
                "streaming_persistence_class"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming-persistence validation")
print("=" * 80)

display(streaming_persistence_validation_df)

# ============================================================
# Completion
# ============================================================

section_5_4_complete = bool(
    streaming_persistence_validation_df["Passed"].all()
)

print()
print(f"Section 5.4 completion status: {section_5_4_complete}")

if not section_5_4_complete:
    raise RuntimeError(
        "Streaming Persistence analysis did not complete successfully."
    )
```

### Interpretation

The PMIP streaming-persistence assessment completed successfully.

Persistence evidence is available for 214 of the 4,593 validated releases. The remaining 4,379 releases do not have a sufficiently reliable historical match and are therefore classified as `No Historical Match`.

Among the matched releases, 181 have more than one historical streaming observation and 122 contain at least five observations, providing stronger evidence of repeated streaming activity over time.

The median matched release contains 7 historical observations across a median observed span of 28 days. This indicates that the typical matched release has some repeated temporal evidence, although the depth and duration of historical coverage vary substantially between releases.

The median streaming-persistence score is approximately 0.74. This relatively low median reflects the fact that the persistence score is normalised against the strongest observed historical coverage, where releases with much larger observation counts or substantially longer histories receive considerably higher scores.

The highest-persistence releases combine either extensive historical observation counts, long historical coverage periods, or both. For example, the strongest release contains more than 1,000 historical observations across several years of evidence.

Persistence classes divide the matched population into Low, Moderate, High and Very High Persistence groups. These categories measure the durability of observed streaming evidence rather than whether streaming performance is increasing or decreasing.

A release can therefore have high streaming persistence while still showing declining growth, because persistence represents sustained presence across time rather than positive momentum.

Because historical matching is available for only a small proportion of the complete PMIP release population, streaming persistence should later be used as a supporting component rather than a standalone measure of release performance.

Section 5.4 completed successfully.

## 5.5 Streaming Stability

### Purpose

This subsection measures how stable or volatile a release's historical streaming behaviour is across the available observation period.

While Section 5.4 focused on how long and how often a release remained observable, streaming stability focuses on how consistently the observed streaming values behave over time.

The assessment will examine:

- variation in historical streaming observations;
- relative volatility compared with average streaming level;
- whether a release exhibits stable or highly irregular historical behaviour;
- differences between sustained performance and short-lived spikes;
- stability classes for matched releases;
- coverage limitations where historical evidence is unavailable.

Streaming stability provides an important risk and reliability dimension for PMIP because two releases may have similar streaming volume or growth, but very different levels of consistency.

```python
# ============================================================
# Section 5.5 - Streaming Stability
# ============================================================

if not section_5_4_complete:
    raise RuntimeError(
        "Section 5.4 must be completed before analysing streaming stability."
    )

# ============================================================
# Working source
# ============================================================

streaming_stability_df = streaming_persistence_df.copy()

# ============================================================
# Prepare historical source
# ============================================================

stability_history_source = release_history_df.copy()

stability_history_source["date"] = pd.to_datetime(
    stability_history_source["date"],
    errors="coerce"
)

stability_history_source["streams"] = pd.to_numeric(
    stability_history_source["streams"],
    errors="coerce"
)

stability_history_source = stability_history_source[
    stability_history_source["track_id"].notna()
    & stability_history_source["date"].notna()
    & stability_history_source["streams"].notna()
].copy()

# ============================================================
# Track-level historical stability statistics
# ============================================================

track_stability_stats_df = (
    stability_history_source
    .groupby("track_id")
    .agg(
        historical_stream_mean=("streams", "mean"),
        historical_stream_median=("streams", "median"),
        historical_stream_std=("streams", "std"),
        historical_stream_min=("streams", "min"),
        historical_stream_max=("streams", "max"),
        historical_stream_observation_count=("streams", "size"),
    )
    .reset_index()
)

# ============================================================
# Coefficient of variation
# Lower = more stable
# ============================================================

track_stability_stats_df["streaming_cv"] = np.where(
    track_stability_stats_df["historical_stream_mean"] > 0,
    (
        track_stability_stats_df["historical_stream_std"]
        / track_stability_stats_df["historical_stream_mean"]
    ),
    np.nan,
)

# ============================================================
# Historical range ratio
# Lower = more stable
# ============================================================

track_stability_stats_df["streaming_range_ratio"] = np.where(
    track_stability_stats_df["historical_stream_mean"] > 0,
    (
        (
            track_stability_stats_df["historical_stream_max"]
            - track_stability_stats_df["historical_stream_min"]
        )
        / track_stability_stats_df["historical_stream_mean"]
    ),
    np.nan,
)

# ============================================================
# Merge historical stability stats using matched track_id
# ============================================================

streaming_stability_df = streaming_stability_df.merge(
    track_stability_stats_df,
    on="track_id",
    how="left",
)

# ============================================================
# Stability evidence availability
# ============================================================

streaming_stability_df["streaming_stability_evidence_available"] = (
    streaming_stability_df["historical_signature_matched"]
    & streaming_stability_df["streaming_cv"].notna()
)

# ============================================================
# Convert volatility into stability scores
#
# Higher score = more stable
# ============================================================

valid_stability_mask = (
    streaming_stability_df[
        "streaming_stability_evidence_available"
    ]
)

streaming_stability_df["cv_stability_score"] = np.nan
streaming_stability_df["range_stability_score"] = np.nan

if valid_stability_mask.any():

    cv_series = streaming_stability_df.loc[
        valid_stability_mask,
        "streaming_cv"
    ]

    range_series = streaming_stability_df.loc[
        valid_stability_mask,
        "streaming_range_ratio"
    ]

    max_cv = cv_series.max()
    max_range_ratio = range_series.max()

    if pd.notna(max_cv) and max_cv > 0:
        streaming_stability_df.loc[
            valid_stability_mask,
            "cv_stability_score"
        ] = (
            1
            - (
                streaming_stability_df.loc[
                    valid_stability_mask,
                    "streaming_cv"
                ]
                / max_cv
            )
        ) * 100

    if pd.notna(max_range_ratio) and max_range_ratio > 0:
        streaming_stability_df.loc[
            valid_stability_mask,
            "range_stability_score"
        ] = (
            1
            - (
                streaming_stability_df.loc[
                    valid_stability_mask,
                    "streaming_range_ratio"
                ]
                / max_range_ratio
            )
        ) * 100

# ============================================================
# Combined stability indicator
# ============================================================

streaming_stability_df["streaming_stability_score"] = np.where(
    valid_stability_mask,
    (
        streaming_stability_df["cv_stability_score"] * 0.70
        + streaming_stability_df["range_stability_score"] * 0.30
    ),
    np.nan,
)

# ============================================================
# Stability classes
# ============================================================

streaming_stability_df["streaming_stability_class"] = (
    "No Historical Match"
)

valid_stability_score_mask = (
    streaming_stability_df["streaming_stability_score"].notna()
)

if valid_stability_score_mask.any():

    q25 = streaming_stability_df.loc[
        valid_stability_score_mask,
        "streaming_stability_score"
    ].quantile(0.25)

    q50 = streaming_stability_df.loc[
        valid_stability_score_mask,
        "streaming_stability_score"
    ].quantile(0.50)

    q75 = streaming_stability_df.loc[
        valid_stability_score_mask,
        "streaming_stability_score"
    ].quantile(0.75)

    streaming_stability_df.loc[
        valid_stability_score_mask
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] <= q25
        ),
        "streaming_stability_class"
    ] = "Low Stability"

    streaming_stability_df.loc[
        valid_stability_score_mask
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] > q25
        )
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] <= q50
        ),
        "streaming_stability_class"
    ] = "Moderate Stability"

    streaming_stability_df.loc[
        valid_stability_score_mask
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] > q50
        )
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] <= q75
        ),
        "streaming_stability_class"
    ] = "High Stability"

    streaming_stability_df.loc[
        valid_stability_score_mask
        & (
            streaming_stability_df[
                "streaming_stability_score"
            ] > q75
        ),
        "streaming_stability_class"
    ] = "Very High Stability"

# ============================================================
# Stability summary
# ============================================================

streaming_stability_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(streaming_stability_df),
        },
        {
            "Metric": "Releases with stability evidence",
            "Observed Value": int(
                valid_stability_mask.sum()
            ),
        },
        {
            "Metric": "Releases without stability evidence",
            "Observed Value": int(
                (~valid_stability_mask).sum()
            ),
        },
        {
            "Metric": "Median streaming coefficient of variation",
            "Observed Value": (
                streaming_stability_df.loc[
                    valid_stability_mask,
                    "streaming_cv"
                ].median()
            ),
        },
        {
            "Metric": "Mean streaming coefficient of variation",
            "Observed Value": (
                streaming_stability_df.loc[
                    valid_stability_mask,
                    "streaming_cv"
                ].mean()
            ),
        },
        {
            "Metric": "Median historical streaming range ratio",
            "Observed Value": (
                streaming_stability_df.loc[
                    valid_stability_mask,
                    "streaming_range_ratio"
                ].median()
            ),
        },
        {
            "Metric": "Median streaming stability score",
            "Observed Value": (
                streaming_stability_df[
                    "streaming_stability_score"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP streaming stability")
print("=" * 80)

display(streaming_stability_summary_df)

# ============================================================
# Stability class distribution
# ============================================================

streaming_stability_class_distribution_df = (
    streaming_stability_df[
        "streaming_stability_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Streaming Stability Class")
    .reset_index(name="Releases")
)

streaming_stability_class_distribution_df["Share (%)"] = (
    streaming_stability_class_distribution_df["Releases"]
    / len(streaming_stability_df)
    * 100
)

print()
print("Streaming-stability class distribution")
print("=" * 80)

display(streaming_stability_class_distribution_df)

# ============================================================
# Most stable releases
# ============================================================

top_streaming_stability_df = (
    streaming_stability_df[
        streaming_stability_df[
            "streaming_stability_score"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "historical_stream_observation_count",
            "historical_stream_mean",
            "historical_stream_std",
            "streaming_cv",
            "streaming_range_ratio",
            "cv_stability_score",
            "range_stability_score",
            "streaming_stability_score",
            "streaming_stability_class",
        ]
    ]
    .sort_values(
        by="streaming_stability_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest streaming stability")
print("=" * 80)

display(top_streaming_stability_df)

# ============================================================
# Least stable releases
# ============================================================

lowest_streaming_stability_df = (
    streaming_stability_df[
        streaming_stability_df[
            "streaming_stability_score"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "historical_stream_observation_count",
            "streaming_cv",
            "streaming_range_ratio",
            "streaming_stability_score",
            "streaming_stability_class",
        ]
    ]
    .sort_values(
        by="streaming_stability_score",
        ascending=True
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with lowest streaming stability")
print("=" * 80)

display(lowest_streaming_stability_df)

# ============================================================
# Validation
# ============================================================

streaming_stability_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.4 completed",
            "Passed": section_5_4_complete,
        },
        {
            "Validation Area": "Historical stability statistics prepared",
            "Passed": (
                "streaming_cv"
                in streaming_stability_df.columns
            ),
        },
        {
            "Validation Area": "Stability evidence available",
            "Passed": streaming_stability_df[
                "streaming_stability_evidence_available"
            ].any(),
        },
        {
            "Validation Area": "Coefficient-of-variation stability calculated",
            "Passed": (
                "cv_stability_score"
                in streaming_stability_df.columns
            ),
        },
        {
            "Validation Area": "Range-based stability calculated",
            "Passed": (
                "range_stability_score"
                in streaming_stability_df.columns
            ),
        },
        {
            "Validation Area": "Combined streaming stability calculated",
            "Passed": (
                "streaming_stability_score"
                in streaming_stability_df.columns
            ),
        },
        {
            "Validation Area": "Streaming stability classes assigned",
            "Passed": streaming_stability_df[
                "streaming_stability_class"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming-stability validation")
print("=" * 80)

display(streaming_stability_validation_df)

# ============================================================
# Completion
# ============================================================

section_5_5_complete = bool(
    streaming_stability_validation_df["Passed"].all()
)

print()
print(f"Section 5.5 completion status: {section_5_5_complete}")

if not section_5_5_complete:
    raise RuntimeError(
        "Streaming Stability analysis did not complete successfully."
    )
```

### Interpretation

The PMIP streaming-stability assessment completed successfully.

Reliable stability evidence is available for 181 of the 4,593 validated releases. The remaining 4,412 releases are classified as having `No Historical Match` for stability analysis because sufficient repeated historical streaming observations are not available.

The smaller stability population compared with the earlier historical-match population is expected because measures such as standard deviation and coefficient of variation require repeated observations. Releases with insufficient historical observations cannot receive a meaningful stability estimate.

Among releases with usable stability evidence, the median coefficient of variation is approximately 0.94 and the mean is approximately 1.09. These values indicate that historical streaming levels often vary substantially relative to their average observed level.

The median historical streaming range ratio is approximately 1.93, further showing that the distance between minimum and maximum observed streaming levels can be large relative to the typical streaming level.

These volatility measures are converted into stability scores so that higher values consistently represent more stable behaviour. The median streaming-stability score is approximately 82.13.

The matched releases are divided into Low, Moderate, High and Very High Stability groups using percentile-based thresholds. The approximately even distribution across these classes reflects the relative classification method rather than an assumption that the population naturally contains equal stability groups.

Streaming stability should be interpreted separately from streaming growth and persistence. A release may maintain highly stable streaming activity without growing, while another release may show strong growth but experience substantial volatility.

Because stability evidence is available for only a small subset of the complete release population, this component should be used as supporting evidence and should not dominate the final Streaming Performance Indicator.

Section 5.5 completed successfully and provides the final behavioural streaming measure required before constructing the combined PMIP Streaming Performance Indicator.

## 5.6 Streaming Performance Indicator

### Purpose

This subsection combines the main streaming-performance components into a single interpretable PMIP Streaming Performance Indicator.

The indicator brings together:

- streaming volume;
- historical streaming variation;
- streaming growth rate;
- streaming persistence;
- streaming stability.

Because these components do not have equal data coverage, the final indicator uses a coverage-aware weighting approach.

Streaming volume receives the strongest base contribution because it is available for almost the entire release population. Historical components are incorporated only when valid evidence exists for the individual release.

This prevents missing historical evidence from being treated as poor performance and ensures that releases are scored only using the evidence actually available for them.

The resulting indicator is designed as a decision-support measure rather than a definitive measure of release success.

```python
# ============================================================
# Section 5.6 - Streaming Performance Indicator
# ============================================================

if not section_5_5_complete:
    raise RuntimeError(
        "Section 5.5 must be completed before constructing the Streaming Performance Indicator."
    )

# ============================================================
# Working source
# ============================================================

streaming_performance_df = streaming_stability_df.copy()

# ============================================================
# Component 1 - Streaming volume score
#
# streaming_volume_percentile already runs from low -> high,
# with stronger releases receiving higher values.
# ============================================================

streaming_performance_df["streaming_volume_score"] = pd.to_numeric(
    streaming_performance_df["streaming_volume_percentile"],
    errors="coerce"
)

# ============================================================
# Component 2 - Historical variation score
#
# Higher historical variation is not automatically better.
# Convert variation into a bounded contextual score where
# moderate variation is preferable to extreme variation.
#
# This component receives a relatively small final weight.
# ============================================================

variation_series = pd.to_numeric(
    streaming_performance_df["relative_historical_stream_range_pct"],
    errors="coerce"
)

streaming_performance_df["streaming_variation_score"] = np.nan

valid_variation_mask = variation_series.notna()

if valid_variation_mask.any():

    variation_percentile = variation_series.loc[
        valid_variation_mask
    ].rank(
        method="average",
        pct=True,
        ascending=True
    ) * 100

    # Peak score around the middle of the distribution.
    # Extremely low or extremely high variation receives
    # less credit than moderate variation.
    streaming_performance_df.loc[
        valid_variation_mask,
        "streaming_variation_score"
    ] = (
        100
        - (
            np.abs(
                variation_percentile - 50
            ) * 2
        )
    ).clip(
        lower=0,
        upper=100
    )

# ============================================================
# Component 3 - Streaming growth score
# ============================================================

growth_series = pd.to_numeric(
    streaming_performance_df["percentage_stream_growth"],
    errors="coerce"
)

streaming_performance_df["streaming_growth_score"] = np.nan

valid_growth_mask = growth_series.notna()

if valid_growth_mask.any():

    streaming_performance_df.loc[
        valid_growth_mask,
        "streaming_growth_score"
    ] = (
        growth_series.loc[
            valid_growth_mask
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Component 4 - Persistence score
# ============================================================

streaming_performance_df["streaming_persistence_component"] = (
    pd.to_numeric(
        streaming_performance_df["streaming_persistence_score"],
        errors="coerce"
    )
)

# ============================================================
# Component 5 - Stability score
# ============================================================

streaming_performance_df["streaming_stability_component"] = (
    pd.to_numeric(
        streaming_performance_df["streaming_stability_score"],
        errors="coerce"
    )
)

# ============================================================
# Base component weights
#
# Volume gets the largest weight because coverage is strongest.
# Historical components are supporting evidence.
# ============================================================

streaming_component_weights = {
    "streaming_volume_score": 0.50,
    "streaming_variation_score": 0.10,
    "streaming_growth_score": 0.15,
    "streaming_persistence_component": 0.15,
    "streaming_stability_component": 0.10,
}

# ============================================================
# Coverage-aware weighted score
#
# Only components available for a release contribute.
# Available weights are automatically re-normalised.
# ============================================================

component_columns = list(
    streaming_component_weights.keys()
)

weighted_sum = pd.Series(
    0.0,
    index=streaming_performance_df.index
)

available_weight_sum = pd.Series(
    0.0,
    index=streaming_performance_df.index
)

for column, weight in streaming_component_weights.items():

    available_mask = (
        streaming_performance_df[column].notna()
    )

    weighted_sum.loc[
        available_mask
    ] += (
        streaming_performance_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    available_weight_sum.loc[
        available_mask
    ] += weight

streaming_performance_df["streaming_performance_score"] = np.where(
    available_weight_sum > 0,
    weighted_sum / available_weight_sum,
    np.nan,
)

# ============================================================
# Evidence coverage
# ============================================================

streaming_performance_df["streaming_components_available"] = (
    streaming_performance_df[
        component_columns
    ]
    .notna()
    .sum(axis=1)
)

streaming_performance_df["streaming_evidence_coverage_pct"] = (
    streaming_performance_df[
        "streaming_components_available"
    ]
    / len(component_columns)
    * 100
)

# ============================================================
# Evidence strength classes
# ============================================================

streaming_performance_df["streaming_evidence_strength"] = np.select(
    [
        streaming_performance_df[
            "streaming_components_available"
        ] >= 5,
        streaming_performance_df[
            "streaming_components_available"
        ] == 4,
        streaming_performance_df[
            "streaming_components_available"
        ] == 3,
        streaming_performance_df[
            "streaming_components_available"
        ] == 2,
        streaming_performance_df[
            "streaming_components_available"
        ] == 1,
    ],
    [
        "Very Strong Evidence",
        "Strong Evidence",
        "Moderate Evidence",
        "Limited Evidence",
        "Minimal Evidence",
    ],
    default="No Evidence",
)

# ============================================================
# Streaming performance percentile
# ============================================================

valid_performance_mask = (
    streaming_performance_df[
        "streaming_performance_score"
    ].notna()
)

streaming_performance_df[
    "streaming_performance_percentile"
] = np.nan

if valid_performance_mask.any():

    streaming_performance_df.loc[
        valid_performance_mask,
        "streaming_performance_percentile"
    ] = (
        streaming_performance_df.loc[
            valid_performance_mask,
            "streaming_performance_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Streaming performance rank
# ============================================================

streaming_performance_df[
    "streaming_performance_rank"
] = (
    streaming_performance_df[
        "streaming_performance_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Performance classes
# ============================================================

streaming_performance_df[
    "streaming_performance_class"
] = "No Streaming Evidence"

valid_score_mask = (
    streaming_performance_df[
        "streaming_performance_score"
    ].notna()
)

if valid_score_mask.any():

    q25 = streaming_performance_df.loc[
        valid_score_mask,
        "streaming_performance_score"
    ].quantile(0.25)

    q50 = streaming_performance_df.loc[
        valid_score_mask,
        "streaming_performance_score"
    ].quantile(0.50)

    q75 = streaming_performance_df.loc[
        valid_score_mask,
        "streaming_performance_score"
    ].quantile(0.75)

    q90 = streaming_performance_df.loc[
        valid_score_mask,
        "streaming_performance_score"
    ].quantile(0.90)

    streaming_performance_df.loc[
        valid_score_mask
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] <= q25
        ),
        "streaming_performance_class"
    ] = "Low Streaming Performance"

    streaming_performance_df.loc[
        valid_score_mask
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] > q25
        )
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] <= q50
        ),
        "streaming_performance_class"
    ] = "Moderate Streaming Performance"

    streaming_performance_df.loc[
        valid_score_mask
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] > q50
        )
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] <= q75
        ),
        "streaming_performance_class"
    ] = "High Streaming Performance"

    streaming_performance_df.loc[
        valid_score_mask
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] > q75
        )
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] <= q90
        ),
        "streaming_performance_class"
    ] = "Very High Streaming Performance"

    streaming_performance_df.loc[
        valid_score_mask
        & (
            streaming_performance_df[
                "streaming_performance_score"
            ] > q90
        ),
        "streaming_performance_class"
    ] = "Exceptional Streaming Performance"

# ============================================================
# Summary
# ============================================================

streaming_performance_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                streaming_performance_df
            ),
        },
        {
            "Metric": "Releases with streaming performance score",
            "Observed Value": int(
                valid_performance_mask.sum()
            ),
        },
        {
            "Metric": "Releases without streaming performance score",
            "Observed Value": int(
                (~valid_performance_mask).sum()
            ),
        },
        {
            "Metric": "Median streaming performance score",
            "Observed Value": (
                streaming_performance_df[
                    "streaming_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean streaming performance score",
            "Observed Value": (
                streaming_performance_df[
                    "streaming_performance_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum streaming performance score",
            "Observed Value": (
                streaming_performance_df[
                    "streaming_performance_score"
                ].max()
            ),
        },
        {
            "Metric": "Median evidence coverage (%)",
            "Observed Value": (
                streaming_performance_df[
                    "streaming_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Constructing PMIP Streaming Performance Indicator")
print("=" * 80)

display(streaming_performance_summary_df)

# ============================================================
# Component availability
# ============================================================

component_coverage_rows = []

for column in component_columns:

    component_coverage_rows.append(
        {
            "Component": column,
            "Releases Available": int(
                streaming_performance_df[
                    column
                ].notna().sum()
            ),
            "Coverage (%)": round(
                streaming_performance_df[
                    column
                ].notna().mean()
                * 100,
                6
            ),
            "Base Weight": (
                streaming_component_weights[
                    column
                ]
            ),
        }
    )

streaming_component_coverage_df = pd.DataFrame(
    component_coverage_rows
)

print()
print("Streaming indicator component coverage")
print("=" * 80)

display(streaming_component_coverage_df)

# ============================================================
# Evidence-strength distribution
# ============================================================

streaming_evidence_strength_distribution_df = (
    streaming_performance_df[
        "streaming_evidence_strength"
    ]
    .value_counts()
    .rename_axis("Evidence Strength")
    .reset_index(name="Releases")
)

streaming_evidence_strength_distribution_df["Share (%)"] = (
    streaming_evidence_strength_distribution_df["Releases"]
    / len(streaming_performance_df)
    * 100
)

print()
print("Streaming evidence-strength distribution")
print("=" * 80)

display(
    streaming_evidence_strength_distribution_df
)

# ============================================================
# Streaming performance class distribution
# ============================================================

streaming_performance_class_distribution_df = (
    streaming_performance_df[
        "streaming_performance_class"
    ]
    .value_counts()
    .rename_axis(
        "Streaming Performance Class"
    )
    .reset_index(name="Releases")
)

streaming_performance_class_distribution_df["Share (%)"] = (
    streaming_performance_class_distribution_df["Releases"]
    / len(streaming_performance_df)
    * 100
)

print()
print("Streaming-performance class distribution")
print("=" * 80)

display(
    streaming_performance_class_distribution_df
)

# ============================================================
# Highest-scoring releases
# ============================================================

top_streaming_performance_df = (
    streaming_performance_df[
        [
            "release_id",
            "Track",
            "Artist",
            "streaming_volume_score",
            "streaming_variation_score",
            "streaming_growth_score",
            "streaming_persistence_component",
            "streaming_stability_component",
            "streaming_components_available",
            "streaming_evidence_strength",
            "streaming_performance_score",
            "streaming_performance_percentile",
            "streaming_performance_rank",
            "streaming_performance_class",
        ]
    ]
    .sort_values(
        by="streaming_performance_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP streaming-performance releases")
print("=" * 80)

display(top_streaming_performance_df)

# ============================================================
# Final validation
# ============================================================

streaming_performance_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.5 completed",
            "Passed": section_5_5_complete,
        },
        {
            "Validation Area": "Streaming volume component available",
            "Passed": streaming_performance_df[
                "streaming_volume_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Coverage-aware weighting prepared",
            "Passed": (
                available_weight_sum > 0
            ).any(),
        },
        {
            "Validation Area": "Streaming performance score calculated",
            "Passed": streaming_performance_df[
                "streaming_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Evidence coverage calculated",
            "Passed": streaming_performance_df[
                "streaming_evidence_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Streaming performance percentile calculated",
            "Passed": streaming_performance_df[
                "streaming_performance_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Streaming performance classes assigned",
            "Passed": streaming_performance_df[
                "streaming_performance_class"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming-performance indicator validation")
print("=" * 80)

display(streaming_performance_validation_df)

# ============================================================
# Section completion
# ============================================================

section_5_6_complete = bool(
    streaming_performance_validation_df[
        "Passed"
    ].all()
)

section_5_complete = bool(
    section_5_1_complete
    and section_5_2_complete
    and section_5_3_complete
    and section_5_4_complete
    and section_5_5_complete
    and section_5_6_complete
)

print()
print(
    f"Section 5.6 completion status: "
    f"{section_5_6_complete}"
)

print(
    f"Section 5 overall completion status: "
    f"{section_5_complete}"
)

if not section_5_complete:
    raise RuntimeError(
        "Section 5 Streaming Performance Intelligence "
        "did not complete successfully."
    )
```

### Interpretation

The PMIP Streaming Performance Indicator was constructed successfully.

A streaming-performance score was produced for 4,500 of the 4,593 validated releases. A total of 93 releases did not contain sufficient streaming evidence to receive a score and remain classified as `No Streaming Evidence`.

The median streaming-performance score is approximately 49.64, while the mean is approximately 48.41. The maximum observed score is approximately 99.98.

The indicator combines five components: streaming volume, historical variation, growth rate, persistence and stability. These components have substantially different levels of data coverage.

Streaming volume has the strongest coverage at approximately 97.65%, while historical variation is available for approximately 48.66% of releases. Growth and persistence are available for approximately 4.66%, and stability is available for approximately 3.94%.

Because of these differences, the final score uses coverage-aware weighting. Missing historical components are not treated as zero performance. Instead, the weights of the components available for each release are re-normalised before the final score is calculated.

This design allows PMIP to retain broad population coverage while avoiding unfair penalties for releases with limited historical data.

The evidence-strength distribution shows that 2,280 releases have Minimal Evidence and 2,006 have Limited Evidence. Only 181 releases have Very Strong Evidence, while 33 have Strong Evidence. A further 93 releases have No Evidence.

The median evidence coverage is therefore 20%, meaning that the typical release receives its final streaming-performance score from only one of the five possible components.

The performance classes separate releases into Low, Moderate, High, Very High and Exceptional Streaming Performance groups. These classes are relative to the observed PMIP population and should not be interpreted as universal measures of commercial success.

The final `streaming_performance_score` should therefore always be interpreted together with `streaming_evidence_strength` and `streaming_evidence_coverage_pct`. A high score supported by Minimal Evidence is less analytically complete than a similar score supported by Strong or Very Strong Evidence.

Section 5.6 completed successfully, and Section 5 overall is now complete.

The PMIP Streaming Performance Intelligence stage therefore provides a transparent, coverage-aware streaming signal that can be used alongside later chart, geographic, artist and release-level intelligence components.

# 6. Chart Performance Intelligence

This section evaluates how strongly each validated release performs within historical chart data.

The aim is to move beyond streaming-only evidence and measure chart-based performance using position, persistence, movement and activity across the available historical observations.

The analysis will focus on:

- best chart position achieved;
- average chart position across historical observations;
- persistence within chart history;
- movement in chart position over time;
- level of chart activity;
- construction of a combined Chart Performance Indicator.

Chart-based evidence provides a different perspective from streaming performance. A release may have strong streaming volume but limited chart presence, while another may show sustained chart activity across multiple markets or dates.

The resulting chart intelligence will later contribute to the wider PMIP Release Performance Intelligence framework.



## 6.1 Best Chart Position

### Purpose

This subsection assesses the strongest chart position achieved by each validated release using the historical chart evidence already integrated into the PMIP release population.

The aim is to identify:

- the best chart position achieved;
- releases reaching the highest chart ranks;
- releases with no available chart-position evidence;
- the distribution of best chart positions;
- relative chart-position strength across the release population.

Because lower numerical chart positions represent stronger performance, the resulting chart-position score will reverse the raw ranking direction so that higher PMIP scores consistently represent stronger performance.

The output from this subsection will provide the first chart-based component used later in the overall Chart Performance Indicator.

```python
# ============================================================
# Section 6.1 - Best Chart Position
# ============================================================

if not section_5_complete:
    raise RuntimeError(
        "Section 5 must be completed before beginning Chart Performance Intelligence."
    )

# ============================================================
# Working source
# ============================================================

chart_position_df = streaming_performance_df.copy()

# ============================================================
# Validate best-chart-position field
# ============================================================

if "best_chart_position" not in chart_position_df.columns:
    raise RuntimeError(
        "best_chart_position is not available in the release population."
    )

chart_position_df["best_chart_position"] = pd.to_numeric(
    chart_position_df["best_chart_position"],
    errors="coerce"
)

# ============================================================
# Chart-position evidence
# ============================================================

chart_position_df["best_chart_position_available"] = (
    chart_position_df["best_chart_position"].notna()
)

valid_chart_position_mask = (
    chart_position_df["best_chart_position_available"]
)

# ============================================================
# Descriptive statistics
# ============================================================

valid_best_positions = chart_position_df.loc[
    valid_chart_position_mask,
    "best_chart_position"
]

best_chart_position_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(chart_position_df),
        },
        {
            "Metric": "Releases with best-chart-position evidence",
            "Observed Value": int(
                valid_chart_position_mask.sum()
            ),
        },
        {
            "Metric": "Releases without best-chart-position evidence",
            "Observed Value": int(
                (~valid_chart_position_mask).sum()
            ),
        },
        {
            "Metric": "Best observed chart position",
            "Observed Value": (
                float(valid_best_positions.min())
                if not valid_best_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Median best chart position",
            "Observed Value": (
                float(valid_best_positions.median())
                if not valid_best_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Mean best chart position",
            "Observed Value": (
                float(valid_best_positions.mean())
                if not valid_best_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Worst observed best chart position",
            "Observed Value": (
                float(valid_best_positions.max())
                if not valid_best_positions.empty
                else np.nan
            ),
        },
    ]
)

print("Assessing PMIP best chart position")
print("=" * 80)

display(best_chart_position_summary_df)

# ============================================================
# Best-chart-position thresholds
# ============================================================

top_1_releases = int(
    (valid_best_positions <= 1).sum()
)

top_10_releases = int(
    (valid_best_positions <= 10).sum()
)

top_20_releases = int(
    (valid_best_positions <= 20).sum()
)

top_50_releases = int(
    (valid_best_positions <= 50).sum()
)

top_100_releases = int(
    (valid_best_positions <= 100).sum()
)

best_chart_threshold_summary_df = pd.DataFrame(
    [
        {
            "Chart Threshold": "Reached #1",
            "Releases": top_1_releases,
        },
        {
            "Chart Threshold": "Reached Top 10",
            "Releases": top_10_releases,
        },
        {
            "Chart Threshold": "Reached Top 20",
            "Releases": top_20_releases,
        },
        {
            "Chart Threshold": "Reached Top 50",
            "Releases": top_50_releases,
        },
        {
            "Chart Threshold": "Reached Top 100",
            "Releases": top_100_releases,
        },
    ]
)

print()
print("Best-chart-position threshold summary")
print("=" * 80)

display(best_chart_threshold_summary_df)

# ============================================================
# Best-chart-position score
#
# Lower raw chart position = stronger performance.
# Convert to percentile so higher PMIP score = stronger.
# ============================================================

chart_position_df["best_chart_position_score"] = np.nan

if valid_chart_position_mask.any():

    chart_position_df.loc[
        valid_chart_position_mask,
        "best_chart_position_score"
    ] = (
        chart_position_df.loc[
            valid_chart_position_mask,
            "best_chart_position"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=False
        )
        * 100
    )

# ============================================================
# Rank
# ============================================================

chart_position_df["best_chart_position_rank"] = (
    chart_position_df["best_chart_position"]
    .rank(
        method="min",
        ascending=True,
        na_option="bottom"
    )
)

# ============================================================
# Best-chart-position classes
# ============================================================

chart_position_df["best_chart_position_class"] = (
    "No Chart Position Evidence"
)

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] == 1
    ),
    "best_chart_position_class"
] = "Chart #1"

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] > 1
    )
    & (
        chart_position_df["best_chart_position"] <= 10
    ),
    "best_chart_position_class"
] = "Top 10"

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] > 10
    )
    & (
        chart_position_df["best_chart_position"] <= 20
    ),
    "best_chart_position_class"
] = "Top 20"

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] > 20
    )
    & (
        chart_position_df["best_chart_position"] <= 50
    ),
    "best_chart_position_class"
] = "Top 50"

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] > 50
    )
    & (
        chart_position_df["best_chart_position"] <= 100
    ),
    "best_chart_position_class"
] = "Top 100"

chart_position_df.loc[
    valid_chart_position_mask
    & (
        chart_position_df["best_chart_position"] > 100
    ),
    "best_chart_position_class"
] = "Below Top 100"

# ============================================================
# Class distribution
# ============================================================

best_chart_position_class_distribution_df = (
    chart_position_df[
        "best_chart_position_class"
    ]
    .value_counts()
    .rename_axis("Best Chart Position Class")
    .reset_index(name="Releases")
)

best_chart_position_class_distribution_df["Share (%)"] = (
    best_chart_position_class_distribution_df["Releases"]
    / len(chart_position_df)
    * 100
)

print()
print("Best-chart-position class distribution")
print("=" * 80)

display(best_chart_position_class_distribution_df)

# ============================================================
# Strongest chart-position releases
# ============================================================

top_best_chart_position_df = (
    chart_position_df[
        chart_position_df[
            "best_chart_position"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "best_chart_position",
            "best_chart_position_score",
            "best_chart_position_rank",
            "best_chart_position_class",
            "chart_observations",
            "countries_charted",
        ]
    ]
    .sort_values(
        by=[
            "best_chart_position",
            "chart_observations",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest best chart positions")
print("=" * 80)

display(top_best_chart_position_df)

# ============================================================
# Validation
# ============================================================

best_chart_position_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5 completed",
            "Passed": section_5_complete,
        },
        {
            "Validation Area": "Best chart position field available",
            "Passed": (
                "best_chart_position"
                in chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Chart-position evidence available",
            "Passed": valid_chart_position_mask.any(),
        },
        {
            "Validation Area": "Best-chart-position score calculated",
            "Passed": (
                "best_chart_position_score"
                in chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Best-chart-position rank prepared",
            "Passed": (
                "best_chart_position_rank"
                in chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Best-chart-position classes assigned",
            "Passed": chart_position_df[
                "best_chart_position_class"
            ].notna().all(),
        },
    ]
)

print()
print("Best-chart-position validation")
print("=" * 80)

display(best_chart_position_validation_df)

# ============================================================
# Completion
# ============================================================

section_6_1_complete = bool(
    best_chart_position_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 6.1 completion status: "
    f"{section_6_1_complete}"
)

if not section_6_1_complete:
    raise RuntimeError(
        "Best Chart Position analysis did not complete successfully."
    )
```

### Interpretation

The PMIP best-chart-position assessment completed successfully.

Chart-position evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain usable best-chart-position evidence.

Among releases with available chart history, the strongest observed chart position is #1. The median best chart position is 6, while the mean is approximately 22.97. This suggests that many charted releases reached relatively strong peak positions, although the distribution also includes releases whose best position was much lower.

A total of 609 releases reached #1, 1,369 reached the Top 10, 1,607 reached the Top 20, 1,893 reached the Top 50, and 2,092 reached the Top 100.

The class distribution further shows that 609 releases are classified as `Chart #1`, 760 as `Top 10`, 238 as `Top 20`, 286 as `Top 50`, 199 as `Top 100`, and 143 as `Below Top 100`.

The best-chart-position score reverses the raw chart scale so that stronger chart performance receives a higher PMIP score. This is important because smaller raw chart-position numbers represent better performance.

The strongest releases therefore receive the highest relative chart-position scores, while releases without chart evidence remain explicitly labelled as `No Chart Position Evidence`.

Best chart position reflects peak chart achievement only. It does not show whether that position was sustained, whether the release performed consistently across its full chart history, or whether chart activity was broad across many countries.

For that reason, this component will later be combined with average chart position, chart persistence, movement and chart activity.

Section 6.1 completed successfully and establishes the peak chart-strength component of the PMIP Chart Performance Intelligence framework.

## 6.2 Average Chart Position

### Purpose

This subsection evaluates the typical chart position achieved by each validated release across its available historical chart observations.

While Section 6.1 measured the strongest peak position reached by a release, average chart position provides a broader view of sustained chart quality across the observed history.

The assessment will examine:

- average historical chart position;
- releases with stronger or weaker typical chart performance;
- coverage of average-position evidence;
- relative average-position strength across the PMIP population;
- chart-position classes based on typical rather than peak performance.

Because lower numerical chart positions represent stronger performance, the raw average position will be converted into a PMIP score where higher values represent better chart performance.

```python
# ============================================================
# Section 6.2 - Average Chart Position
# ============================================================

if not section_6_1_complete:
    raise RuntimeError(
        "Section 6.1 must be completed before analysing average chart position."
    )

# ============================================================
# Working source
# ============================================================

average_chart_position_df = chart_position_df.copy()

# ============================================================
# Validate field
# ============================================================

if "average_chart_position" not in average_chart_position_df.columns:
    raise RuntimeError(
        "average_chart_position is not available in the release population."
    )

average_chart_position_df["average_chart_position"] = pd.to_numeric(
    average_chart_position_df["average_chart_position"],
    errors="coerce"
)

# ============================================================
# Evidence availability
# ============================================================

average_chart_position_df["average_chart_position_available"] = (
    average_chart_position_df["average_chart_position"].notna()
)

valid_average_position_mask = (
    average_chart_position_df["average_chart_position_available"]
)

valid_average_positions = average_chart_position_df.loc[
    valid_average_position_mask,
    "average_chart_position"
]

# ============================================================
# Summary
# ============================================================

average_chart_position_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(average_chart_position_df),
        },
        {
            "Metric": "Releases with average-chart-position evidence",
            "Observed Value": int(
                valid_average_position_mask.sum()
            ),
        },
        {
            "Metric": "Releases without average-chart-position evidence",
            "Observed Value": int(
                (~valid_average_position_mask).sum()
            ),
        },
        {
            "Metric": "Best observed average chart position",
            "Observed Value": (
                float(valid_average_positions.min())
                if not valid_average_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Median average chart position",
            "Observed Value": (
                float(valid_average_positions.median())
                if not valid_average_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Mean average chart position",
            "Observed Value": (
                float(valid_average_positions.mean())
                if not valid_average_positions.empty
                else np.nan
            ),
        },
        {
            "Metric": "Weakest observed average chart position",
            "Observed Value": (
                float(valid_average_positions.max())
                if not valid_average_positions.empty
                else np.nan
            ),
        },
    ]
)

print("Assessing PMIP average chart position")
print("=" * 80)

display(average_chart_position_summary_df)

# ============================================================
# Average chart-position percentiles
# ============================================================

average_chart_percentiles = [
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
]

average_chart_position_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
        ],
        "Average Chart Position": (
            valid_average_positions
            .quantile(average_chart_percentiles)
            .values
        ),
    }
)

print()
print("Average-chart-position percentiles")
print("=" * 80)

display(average_chart_position_percentiles_df)

# ============================================================
# Average-chart-position score
#
# Lower raw position = stronger performance
# Higher PMIP score = stronger performance
# ============================================================

average_chart_position_df["average_chart_position_score"] = np.nan

if valid_average_position_mask.any():

    average_chart_position_df.loc[
        valid_average_position_mask,
        "average_chart_position_score"
    ] = (
        average_chart_position_df.loc[
            valid_average_position_mask,
            "average_chart_position"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=False
        )
        * 100
    )

# ============================================================
# Rank
# ============================================================

average_chart_position_df["average_chart_position_rank"] = (
    average_chart_position_df["average_chart_position"]
    .rank(
        method="min",
        ascending=True,
        na_option="bottom"
    )
)

# ============================================================
# Average chart-position classes
# ============================================================

average_chart_position_df["average_chart_position_class"] = (
    "No Average Chart Evidence"
)

average_chart_position_df.loc[
    valid_average_position_mask
    & (
        average_chart_position_df[
            "average_chart_position"
        ] <= 10
    ),
    "average_chart_position_class"
] = "Top 10 Average"

average_chart_position_df.loc[
    valid_average_position_mask
    & (
        average_chart_position_df[
            "average_chart_position"
        ] > 10
    )
    & (
        average_chart_position_df[
            "average_chart_position"
        ] <= 20
    ),
    "average_chart_position_class"
] = "Top 20 Average"

average_chart_position_df.loc[
    valid_average_position_mask
    & (
        average_chart_position_df[
            "average_chart_position"
        ] > 20
    )
    & (
        average_chart_position_df[
            "average_chart_position"
        ] <= 50
    ),
    "average_chart_position_class"
] = "Top 50 Average"

average_chart_position_df.loc[
    valid_average_position_mask
    & (
        average_chart_position_df[
            "average_chart_position"
        ] > 50
    )
    & (
        average_chart_position_df[
            "average_chart_position"
        ] <= 100
    ),
    "average_chart_position_class"
] = "Top 100 Average"

average_chart_position_df.loc[
    valid_average_position_mask
    & (
        average_chart_position_df[
            "average_chart_position"
        ] > 100
    ),
    "average_chart_position_class"
] = "Below Top 100 Average"

# ============================================================
# Class distribution
# ============================================================

average_chart_position_class_distribution_df = (
    average_chart_position_df[
        "average_chart_position_class"
    ]
    .value_counts()
    .rename_axis("Average Chart Position Class")
    .reset_index(name="Releases")
)

average_chart_position_class_distribution_df["Share (%)"] = (
    average_chart_position_class_distribution_df["Releases"]
    / len(average_chart_position_df)
    * 100
)

print()
print("Average-chart-position class distribution")
print("=" * 80)

display(average_chart_position_class_distribution_df)

# ============================================================
# Strongest average chart-position releases
# ============================================================

top_average_chart_position_df = (
    average_chart_position_df[
        average_chart_position_df[
            "average_chart_position"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "average_chart_position",
            "average_chart_position_score",
            "average_chart_position_rank",
            "average_chart_position_class",
            "best_chart_position",
            "chart_observations",
            "countries_charted",
        ]
    ]
    .sort_values(
        by=[
            "average_chart_position",
            "chart_observations",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest average chart positions")
print("=" * 80)

display(top_average_chart_position_df)

# ============================================================
# Validation
# ============================================================

average_chart_position_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.1 completed",
            "Passed": section_6_1_complete,
        },
        {
            "Validation Area": "Average chart position field available",
            "Passed": (
                "average_chart_position"
                in average_chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Average chart-position evidence available",
            "Passed": valid_average_position_mask.any(),
        },
        {
            "Validation Area": "Average-chart-position score calculated",
            "Passed": (
                "average_chart_position_score"
                in average_chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Average-chart-position rank prepared",
            "Passed": (
                "average_chart_position_rank"
                in average_chart_position_df.columns
            ),
        },
        {
            "Validation Area": "Average-chart-position classes assigned",
            "Passed": average_chart_position_df[
                "average_chart_position_class"
            ].notna().all(),
        },
    ]
)

print()
print("Average-chart-position validation")
print("=" * 80)

display(average_chart_position_validation_df)

# ============================================================
# Completion
# ============================================================

section_6_2_complete = bool(
    average_chart_position_validation_df["Passed"].all()
)

print()
print(
    f"Section 6.2 completion status: "
    f"{section_6_2_complete}"
)

if not section_6_2_complete:
    raise RuntimeError(
        "Average Chart Position analysis did not complete successfully."
    )
```

The PMIP average-chart-position assessment completed successfully.

Average chart-position evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain sufficient average-position evidence.

The percentile distribution supports this pattern. Around 25% of charted releases have an average position better than approximately 78.31, while the median is approximately 96.27. The 75th percentile rises to approximately 120.06, showing that a large proportion of releases spend much of their observed history outside the strongest chart positions.

The average-chart-position score reverses the raw chart scale so that lower numerical chart positions receive higher PMIP scores.

## 6.3 Chart Persistence

### Purpose

This subsection measures how persistently each validated release remains present within the available historical chart data.

While Sections 6.1 and 6.2 focused on peak and average chart position, chart persistence focuses on duration and repeated chart presence.

The assessment will examine:

- number of historical chart observations;
- number of countries in which a release charted;
- duration between first and last observed chart dates;
- repeated chart presence across time;
- breadth and longevity of chart activity;
- a combined chart-persistence score.

Chart persistence is important because a release that remains visible across many observations and markets may represent more durable chart performance than a release that achieves a strong position only briefly.

```python
# ============================================================
# Section 6.3 - Chart Persistence
# ============================================================

if not section_6_2_complete:
    raise RuntimeError(
        "Section 6.2 must be completed before analysing chart persistence."
    )

# ============================================================
# Working source
# ============================================================

chart_persistence_df = average_chart_position_df.copy()

# ============================================================
# Required fields
# ============================================================

required_chart_persistence_fields = [
    "chart_observations",
    "countries_charted",
    "first_chart_date",
    "last_chart_date",
]

missing_chart_persistence_fields = [
    field
    for field in required_chart_persistence_fields
    if field not in chart_persistence_df.columns
]

if missing_chart_persistence_fields:
    raise RuntimeError(
        "Missing required chart-persistence fields: "
        f"{missing_chart_persistence_fields}"
    )

# ============================================================
# Numeric and date preparation
# ============================================================

chart_persistence_df["chart_observations"] = pd.to_numeric(
    chart_persistence_df["chart_observations"],
    errors="coerce"
)

chart_persistence_df["countries_charted"] = pd.to_numeric(
    chart_persistence_df["countries_charted"],
    errors="coerce"
)

chart_persistence_df["first_chart_date_parsed"] = pd.to_datetime(
    chart_persistence_df["first_chart_date"],
    errors="coerce"
)

chart_persistence_df["last_chart_date_parsed"] = pd.to_datetime(
    chart_persistence_df["last_chart_date"],
    errors="coerce"
)

# ============================================================
# Chart history duration
# ============================================================

chart_persistence_df["chart_history_span_days"] = (
    chart_persistence_df["last_chart_date_parsed"]
    - chart_persistence_df["first_chart_date_parsed"]
).dt.days

# ============================================================
# Persistence evidence availability
# ============================================================

chart_persistence_df["chart_persistence_evidence_available"] = (
    chart_persistence_df["chart_observations"].notna()
    & chart_persistence_df["countries_charted"].notna()
    & chart_persistence_df["chart_history_span_days"].notna()
)

valid_chart_persistence_mask = (
    chart_persistence_df[
        "chart_persistence_evidence_available"
    ]
)

# ============================================================
# Supporting flags
# ============================================================

chart_persistence_df["repeated_chart_presence"] = (
    chart_persistence_df["chart_observations"]
    .fillna(0)
    .gt(1)
)

chart_persistence_df["multi_country_chart_presence"] = (
    chart_persistence_df["countries_charted"]
    .fillna(0)
    .gt(1)
)

chart_persistence_df["extended_chart_history"] = (
    chart_persistence_df["chart_history_span_days"]
    .fillna(0)
    .ge(30)
)

# ============================================================
# Component scores
#
# Higher score = stronger persistence
# ============================================================

chart_persistence_df["chart_observation_score"] = np.nan
chart_persistence_df["country_breadth_score"] = np.nan
chart_persistence_df["chart_span_score"] = np.nan

if valid_chart_persistence_mask.any():

    chart_persistence_df.loc[
        valid_chart_persistence_mask,
        "chart_observation_score"
    ] = (
        chart_persistence_df.loc[
            valid_chart_persistence_mask,
            "chart_observations"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

    chart_persistence_df.loc[
        valid_chart_persistence_mask,
        "country_breadth_score"
    ] = (
        chart_persistence_df.loc[
            valid_chart_persistence_mask,
            "countries_charted"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

    chart_persistence_df.loc[
        valid_chart_persistence_mask,
        "chart_span_score"
    ] = (
        chart_persistence_df.loc[
            valid_chart_persistence_mask,
            "chart_history_span_days"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Combined chart-persistence score
#
# Weighting:
# - observations: 40%
# - countries charted: 30%
# - chart-history duration: 30%
# ============================================================

chart_persistence_df["chart_persistence_score"] = np.where(
    valid_chart_persistence_mask,
    (
        chart_persistence_df["chart_observation_score"] * 0.40
        + chart_persistence_df["country_breadth_score"] * 0.30
        + chart_persistence_df["chart_span_score"] * 0.30
    ),
    np.nan,
)

# ============================================================
# Persistence classes
# ============================================================

chart_persistence_df["chart_persistence_class"] = (
    "No Chart Persistence Evidence"
)

valid_persistence_score_mask = (
    chart_persistence_df["chart_persistence_score"].notna()
)

if valid_persistence_score_mask.any():

    q25 = chart_persistence_df.loc[
        valid_persistence_score_mask,
        "chart_persistence_score"
    ].quantile(0.25)

    q50 = chart_persistence_df.loc[
        valid_persistence_score_mask,
        "chart_persistence_score"
    ].quantile(0.50)

    q75 = chart_persistence_df.loc[
        valid_persistence_score_mask,
        "chart_persistence_score"
    ].quantile(0.75)

    chart_persistence_df.loc[
        valid_persistence_score_mask
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] <= q25
        ),
        "chart_persistence_class"
    ] = "Low Chart Persistence"

    chart_persistence_df.loc[
        valid_persistence_score_mask
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] > q25
        )
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] <= q50
        ),
        "chart_persistence_class"
    ] = "Moderate Chart Persistence"

    chart_persistence_df.loc[
        valid_persistence_score_mask
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] > q50
        )
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] <= q75
        ),
        "chart_persistence_class"
    ] = "High Chart Persistence"

    chart_persistence_df.loc[
        valid_persistence_score_mask
        & (
            chart_persistence_df[
                "chart_persistence_score"
            ] > q75
        ),
        "chart_persistence_class"
    ] = "Very High Chart Persistence"

# ============================================================
# Summary
# ============================================================

chart_persistence_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(chart_persistence_df),
        },
        {
            "Metric": "Releases with chart-persistence evidence",
            "Observed Value": int(
                valid_chart_persistence_mask.sum()
            ),
        },
        {
            "Metric": "Releases without chart-persistence evidence",
            "Observed Value": int(
                (~valid_chart_persistence_mask).sum()
            ),
        },
        {
            "Metric": "Releases with repeated chart presence",
            "Observed Value": int(
                chart_persistence_df[
                    "repeated_chart_presence"
                ].sum()
            ),
        },
        {
            "Metric": "Releases charted in multiple countries",
            "Observed Value": int(
                chart_persistence_df[
                    "multi_country_chart_presence"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with 30+ day chart history",
            "Observed Value": int(
                chart_persistence_df[
                    "extended_chart_history"
                ].sum()
            ),
        },
        {
            "Metric": "Median chart observations",
            "Observed Value": (
                chart_persistence_df.loc[
                    valid_chart_persistence_mask,
                    "chart_observations"
                ].median()
            ),
        },
        {
            "Metric": "Median countries charted",
            "Observed Value": (
                chart_persistence_df.loc[
                    valid_chart_persistence_mask,
                    "countries_charted"
                ].median()
            ),
        },
        {
            "Metric": "Median chart-history span (days)",
            "Observed Value": (
                chart_persistence_df.loc[
                    valid_chart_persistence_mask,
                    "chart_history_span_days"
                ].median()
            ),
        },
        {
            "Metric": "Median chart-persistence score",
            "Observed Value": (
                chart_persistence_df[
                    "chart_persistence_score"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP chart persistence")
print("=" * 80)

display(chart_persistence_summary_df)

# ============================================================
# Persistence class distribution
# ============================================================

chart_persistence_class_distribution_df = (
    chart_persistence_df[
        "chart_persistence_class"
    ]
    .value_counts()
    .rename_axis("Chart Persistence Class")
    .reset_index(name="Releases")
)

chart_persistence_class_distribution_df["Share (%)"] = (
    chart_persistence_class_distribution_df["Releases"]
    / len(chart_persistence_df)
    * 100
)

print()
print("Chart-persistence class distribution")
print("=" * 80)

display(chart_persistence_class_distribution_df)

# ============================================================
# Strongest chart-persistence releases
# ============================================================

top_chart_persistence_df = (
    chart_persistence_df[
        chart_persistence_df[
            "chart_persistence_score"
        ].notna()
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "chart_observations",
            "countries_charted",
            "chart_history_span_days",
            "chart_observation_score",
            "country_breadth_score",
            "chart_span_score",
            "chart_persistence_score",
            "chart_persistence_class",
        ]
    ]
    .sort_values(
        by="chart_persistence_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest chart persistence")
print("=" * 80)

display(top_chart_persistence_df)

# ============================================================
# Validation
# ============================================================

chart_persistence_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.2 completed",
            "Passed": section_6_2_complete,
        },
        {
            "Validation Area": "Chart observations available",
            "Passed": (
                "chart_observations"
                in chart_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Countries-charted evidence available",
            "Passed": (
                "countries_charted"
                in chart_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Chart-history span calculated",
            "Passed": (
                "chart_history_span_days"
                in chart_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Chart-persistence evidence available",
            "Passed": valid_chart_persistence_mask.any(),
        },
        {
            "Validation Area": "Chart-persistence score calculated",
            "Passed": (
                "chart_persistence_score"
                in chart_persistence_df.columns
            ),
        },
        {
            "Validation Area": "Chart-persistence classes assigned",
            "Passed": chart_persistence_df[
                "chart_persistence_class"
            ].notna().all(),
        },
    ]
)

print()
print("Chart-persistence validation")
print("=" * 80)

display(chart_persistence_validation_df)

# ============================================================
# Completion
# ============================================================

section_6_3_complete = bool(
    chart_persistence_validation_df["Passed"].all()
)

print()
print(
    f"Section 6.3 completion status: "
    f"{section_6_3_complete}"
)

if not section_6_3_complete:
    raise RuntimeError(
        "Chart Persistence analysis did not complete successfully."
    )
```

### Interpretation

The PMIP chart-persistence assessment completed successfully.

Chart-persistence evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain sufficient chart-history evidence.

Among the releases with usable chart evidence, 2,202 show repeated chart presence, 2,084 have charted in more than one country, and 2,009 have chart histories lasting at least 30 days.

The median charted release has 409 historical chart observations across 24 countries, with a median chart-history span of approximately 413 days.

These results indicate that many charted releases are not represented by isolated chart appearances. Instead, a large proportion show repeated and geographically broad chart activity over substantial periods of time.

The median chart-persistence score is approximately 50.03. Because the score is built from percentile-based components, values around the middle of the scale represent releases with average persistence relative to the charted PMIP population.

The persistence classes are divided approximately evenly across Low, Moderate, High and Very High Chart Persistence groups. This reflects the percentile-based classification method rather than an assumption that chart persistence naturally occurs in equal groups.

Chart persistence measures durability and breadth rather than chart quality alone. A release may have strong persistence without maintaining a very high chart position, while another may achieve a strong peak position but remain on the charts only briefly.

The combination of observation count, geographic breadth and chart-history duration therefore provides a useful measure of how sustained and widespread a release's chart presence has been.

Section 6.3 completed successfully and adds the durability component to the PMIP Chart Performance Intelligence framework.

## 6.4 Chart Movement

### Purpose

This subsection measures how release-level chart positions change across the available historical observation period.

While earlier chart subsections focused on peak position, average position and persistence, chart movement focuses on directional change.

The assessment will examine:

- first observed chart position;
- last observed chart position;
- absolute chart-position movement;
- direction of movement;
- degree of improvement or decline;
- relative movement strength across matched releases.

Because lower numerical chart positions represent stronger performance, a movement from position 50 to position 10 represents improvement, while movement from position 10 to position 50 represents decline.

This subsection therefore provides the directional chart-momentum component of the PMIP Chart Performance Intelligence framework.

```python
# ============================================================
# Section 6.4 - Chart Movement
# ============================================================

if not section_6_3_complete:
    raise RuntimeError(
        "Section 6.3 must be completed before analysing chart movement."
    )

# ============================================================
# Working source
# ============================================================

chart_movement_df = chart_persistence_df.copy()

# ============================================================
# Prepare historical chart source
# ============================================================

chart_history_source = release_history_df.copy()

chart_history_source["date"] = pd.to_datetime(
    chart_history_source["date"],
    errors="coerce"
)

chart_history_source["position"] = pd.to_numeric(
    chart_history_source["position"],
    errors="coerce"
)

chart_history_source = chart_history_source[
    chart_history_source["track_id"].notna()
    & chart_history_source["date"].notna()
    & chart_history_source["position"].notna()
].copy()

chart_history_source = chart_history_source.sort_values(
    by=["track_id", "date"]
)

# ============================================================
# First and last chart positions per historical track
# ============================================================

first_chart_position_df = (
    chart_history_source
    .groupby("track_id", as_index=False)
    .first()[
        ["track_id", "date", "position"]
    ]
    .rename(
        columns={
            "date": "first_chart_observation_date",
            "position": "first_chart_position",
        }
    )
)

last_chart_position_df = (
    chart_history_source
    .groupby("track_id", as_index=False)
    .last()[
        ["track_id", "date", "position"]
    ]
    .rename(
        columns={
            "date": "last_chart_observation_date",
            "position": "last_chart_position",
        }
    )
)

chart_movement_signature_df = first_chart_position_df.merge(
    last_chart_position_df,
    on="track_id",
    how="inner"
)

# ============================================================
# Historical observation counts
# ============================================================

chart_observation_counts_df = (
    chart_history_source
    .groupby("track_id")
    .size()
    .rename("chart_movement_observations")
    .reset_index()
)

chart_movement_signature_df = chart_movement_signature_df.merge(
    chart_observation_counts_df,
    on="track_id",
    how="left"
)

# ============================================================
# Historical chart signature
#
# These fields already exist in spotify_integrated.csv and
# allow release rows to reconnect to raw chart history without
# relying on track_id being directly stored in the release data.
# ============================================================

track_history_summary_df = (
    chart_history_source
    .groupby("track_id")
    .agg(
        best_chart_position_signature=("position", "min"),
        average_chart_position_signature=("position", "mean"),
        chart_observations_signature=("position", "size"),
        first_chart_date_signature=("date", "min"),
        last_chart_date_signature=("date", "max"),
    )
    .reset_index()
)

# ============================================================
# Prepare matching keys
# ============================================================

track_history_summary_df["first_chart_date_key"] = (
    track_history_summary_df["first_chart_date_signature"]
    .dt.strftime("%Y-%m-%d")
)

track_history_summary_df["last_chart_date_key"] = (
    track_history_summary_df["last_chart_date_signature"]
    .dt.strftime("%Y-%m-%d")
)

track_history_summary_df["best_chart_position_key"] = (
    track_history_summary_df[
        "best_chart_position_signature"
    ]
    .round(6)
)

track_history_summary_df["average_chart_position_key"] = (
    track_history_summary_df[
        "average_chart_position_signature"
    ]
    .round(6)
)

track_history_summary_df["chart_observations_key"] = (
    track_history_summary_df[
        "chart_observations_signature"
    ]
    .astype(float)
    .round(6)
)

chart_signature_key_cols = [
    "first_chart_date_key",
    "last_chart_date_key",
    "best_chart_position_key",
    "average_chart_position_key",
    "chart_observations_key",
]

# ============================================================
# Determine whether each historical signature is unique
# ============================================================

chart_signature_counts_df = (
    track_history_summary_df
    .groupby(
        chart_signature_key_cols,
        dropna=False
    )
    .size()
    .rename("chart_signature_match_count")
    .reset_index()
)

track_history_summary_unique_df = (
    track_history_summary_df
    .sort_values(by="track_id")
    .drop_duplicates(
        subset=chart_signature_key_cols,
        keep="first"
    )
    .merge(
        chart_signature_counts_df,
        on=chart_signature_key_cols,
        how="left"
    )
)

# ============================================================
# Attach first / last chart-position evidence
# ============================================================

track_history_summary_unique_df = (
    track_history_summary_unique_df
    .merge(
        chart_movement_signature_df,
        on="track_id",
        how="left"
    )
)

# ============================================================
# IMPORTANT:
# Rename historical track_id before merging with the release
# population because chart_movement_df may already contain a
# track_id column from previous streaming sections.
# ============================================================

track_history_summary_unique_df = (
    track_history_summary_unique_df
    .rename(
        columns={
            "track_id": "chart_movement_track_id"
        }
    )
)

# ============================================================
# Prepare matching keys on final release population
# ============================================================

chart_movement_df["first_chart_date_key"] = (
    pd.to_datetime(
        chart_movement_df["first_chart_date"],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

chart_movement_df["last_chart_date_key"] = (
    pd.to_datetime(
        chart_movement_df["last_chart_date"],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

chart_movement_df["best_chart_position_key"] = (
    pd.to_numeric(
        chart_movement_df["best_chart_position"],
        errors="coerce"
    )
    .round(6)
)

chart_movement_df["average_chart_position_key"] = (
    pd.to_numeric(
        chart_movement_df["average_chart_position"],
        errors="coerce"
    )
    .round(6)
)

chart_movement_df["chart_observations_key"] = (
    pd.to_numeric(
        chart_movement_df["chart_observations"],
        errors="coerce"
    )
    .round(6)
)

# ============================================================
# Merge historical movement evidence into release population
# ============================================================

chart_movement_merge_cols = (
    chart_signature_key_cols
    + [
        "chart_movement_track_id",
        "first_chart_observation_date",
        "last_chart_observation_date",
        "first_chart_position",
        "last_chart_position",
        "chart_movement_observations",
        "chart_signature_match_count",
    ]
)

chart_movement_df = chart_movement_df.merge(
    track_history_summary_unique_df[
        chart_movement_merge_cols
    ],
    on=chart_signature_key_cols,
    how="left"
)

# ============================================================
# Match diagnostics
# ============================================================

chart_movement_df["chart_movement_match_available"] = (
    chart_movement_df[
        "chart_movement_track_id"
    ].notna()
)

chart_movement_df["chart_movement_match_ambiguous"] = (
    chart_movement_df[
        "chart_signature_match_count"
    ]
    .fillna(0)
    .gt(1)
)

# ============================================================
# Calculate chart-position movement
#
# IMPORTANT:
# Lower chart position = better.
#
# Example:
# First position = 80
# Last position  = 20
#
# 80 - 20 = +60
# Positive movement therefore represents improvement.
# ============================================================

chart_movement_df["chart_position_movement"] = (
    chart_movement_df["first_chart_position"]
    - chart_movement_df["last_chart_position"]
)

# ============================================================
# Movement direction
# ============================================================

chart_movement_df["chart_movement_direction"] = np.select(
    [
        chart_movement_df[
            "chart_position_movement"
        ] > 0,

        chart_movement_df[
            "chart_position_movement"
        ] < 0,

        chart_movement_df[
            "chart_position_movement"
        ] == 0,
    ],
    [
        "Improved",
        "Declined",
        "Stable",
    ],
    default="Unavailable",
)

# ============================================================
# Valid chart-movement evidence
# ============================================================

valid_chart_movement_mask = (
    chart_movement_df[
        "chart_movement_match_available"
    ]
    & chart_movement_df[
        "chart_position_movement"
    ].notna()
)

# ============================================================
# Relative chart-movement score
#
# Higher score = stronger positive movement.
# ============================================================

chart_movement_df["chart_movement_score"] = np.nan

if valid_chart_movement_mask.any():

    chart_movement_df.loc[
        valid_chart_movement_mask,
        "chart_movement_score"
    ] = (
        chart_movement_df.loc[
            valid_chart_movement_mask,
            "chart_position_movement"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Chart-movement percentile
# ============================================================

chart_movement_df["chart_movement_percentile"] = (
    chart_movement_df["chart_movement_score"]
)

# ============================================================
# Movement classes
#
# Use actual movement direction first, then relative strength
# within improvement / decline groups.
# ============================================================

chart_movement_df["chart_movement_class"] = (
    "No Chart Movement Evidence"
)

# ------------------------------------------------------------
# Stable releases
# ------------------------------------------------------------

chart_movement_df.loc[
    valid_chart_movement_mask
    & (
        chart_movement_df[
            "chart_position_movement"
        ] == 0
    ),
    "chart_movement_class"
] = "Stable Chart Position"

# ------------------------------------------------------------
# Declining releases
# ------------------------------------------------------------

declining_mask = (
    valid_chart_movement_mask
    & (
        chart_movement_df[
            "chart_position_movement"
        ] < 0
    )
)

if declining_mask.any():

    decline_strength = (
        chart_movement_df.loc[
            declining_mask,
            "chart_position_movement"
        ]
        .abs()
    )

    decline_median = decline_strength.median()

    chart_movement_df.loc[
        declining_mask
        & (
            chart_movement_df[
                "chart_position_movement"
            ].abs()
            <= decline_median
        ),
        "chart_movement_class"
    ] = "Moderate Decline"

    chart_movement_df.loc[
        declining_mask
        & (
            chart_movement_df[
                "chart_position_movement"
            ].abs()
            > decline_median
        ),
        "chart_movement_class"
    ] = "Strong Decline"

# ------------------------------------------------------------
# Improving releases
# ------------------------------------------------------------

improving_mask = (
    valid_chart_movement_mask
    & (
        chart_movement_df[
            "chart_position_movement"
        ] > 0
    )
)

if improving_mask.any():

    improvement_median = (
        chart_movement_df.loc[
            improving_mask,
            "chart_position_movement"
        ]
        .median()
    )

    chart_movement_df.loc[
        improving_mask
        & (
            chart_movement_df[
                "chart_position_movement"
            ] <= improvement_median
        ),
        "chart_movement_class"
    ] = "Moderate Improvement"

    chart_movement_df.loc[
        improving_mask
        & (
            chart_movement_df[
                "chart_position_movement"
            ] > improvement_median
        ),
        "chart_movement_class"
    ] = "Strong Improvement"

# ============================================================
# Summary
# ============================================================

chart_movement_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                chart_movement_df
            ),
        },
        {
            "Metric": "Releases with chart-movement evidence",
            "Observed Value": int(
                valid_chart_movement_mask.sum()
            ),
        },
        {
            "Metric": "Releases without chart-movement evidence",
            "Observed Value": int(
                (
                    ~valid_chart_movement_mask
                ).sum()
            ),
        },
        {
            "Metric": "Ambiguous chart-history signatures",
            "Observed Value": int(
                chart_movement_df[
                    "chart_movement_match_ambiguous"
                ].sum()
            ),
        },
        {
            "Metric": "Releases showing chart improvement",
            "Observed Value": int(
                (
                    chart_movement_df[
                        "chart_movement_direction"
                    ]
                    == "Improved"
                ).sum()
            ),
        },
        {
            "Metric": "Releases showing chart decline",
            "Observed Value": int(
                (
                    chart_movement_df[
                        "chart_movement_direction"
                    ]
                    == "Declined"
                ).sum()
            ),
        },
        {
            "Metric": "Releases with stable first/last position",
            "Observed Value": int(
                (
                    chart_movement_df[
                        "chart_movement_direction"
                    ]
                    == "Stable"
                ).sum()
            ),
        },
        {
            "Metric": "Median chart-position movement",
            "Observed Value": (
                chart_movement_df.loc[
                    valid_chart_movement_mask,
                    "chart_position_movement"
                ].median()
            ),
        },
        {
            "Metric": "Mean chart-position movement",
            "Observed Value": (
                chart_movement_df.loc[
                    valid_chart_movement_mask,
                    "chart_position_movement"
                ].mean()
            ),
        },
    ]
)

print("Assessing PMIP chart movement")
print("=" * 80)

display(chart_movement_summary_df)

# ============================================================
# Movement-direction distribution
# ============================================================

chart_movement_direction_distribution_df = (
    chart_movement_df[
        "chart_movement_direction"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Chart Movement Direction"
    )
    .reset_index(
        name="Releases"
    )
)

chart_movement_direction_distribution_df[
    "Share (%)"
] = (
    chart_movement_direction_distribution_df[
        "Releases"
    ]
    / len(chart_movement_df)
    * 100
)

print()
print("Chart-movement direction distribution")
print("=" * 80)

display(
    chart_movement_direction_distribution_df
)

# ============================================================
# Movement-class distribution
# ============================================================

chart_movement_class_distribution_df = (
    chart_movement_df[
        "chart_movement_class"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Chart Movement Class"
    )
    .reset_index(
        name="Releases"
    )
)

chart_movement_class_distribution_df[
    "Share (%)"
] = (
    chart_movement_class_distribution_df[
        "Releases"
    ]
    / len(chart_movement_df)
    * 100
)

print()
print("Chart-movement class distribution")
print("=" * 80)

display(
    chart_movement_class_distribution_df
)

# ============================================================
# Strongest positive chart movement
# ============================================================

top_chart_movement_df = (
    chart_movement_df[
        valid_chart_movement_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "first_chart_observation_date",
            "last_chart_observation_date",
            "first_chart_position",
            "last_chart_position",
            "chart_position_movement",
            "chart_movement_direction",
            "chart_movement_score",
            "chart_movement_class",
        ]
    ]
    .sort_values(
        by="chart_position_movement",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest positive chart movement")
print("=" * 80)

display(top_chart_movement_df)

# ============================================================
# Strongest chart declines
# ============================================================

lowest_chart_movement_df = (
    chart_movement_df[
        valid_chart_movement_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "first_chart_position",
            "last_chart_position",
            "chart_position_movement",
            "chart_movement_direction",
            "chart_movement_score",
            "chart_movement_class",
        ]
    ]
    .sort_values(
        by="chart_position_movement",
        ascending=True
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest chart decline")
print("=" * 80)

display(lowest_chart_movement_df)

# ============================================================
# Validation
# ============================================================

chart_movement_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.3 completed",
            "Passed": section_6_3_complete,
        },
        {
            "Validation Area": "Historical chart matching prepared",
            "Passed": (
                "chart_movement_match_available"
                in chart_movement_df.columns
            ),
        },
        {
            "Validation Area": "Chart-movement evidence available",
            "Passed": valid_chart_movement_mask.any(),
        },
        {
            "Validation Area": "Chart-position movement calculated",
            "Passed": (
                "chart_position_movement"
                in chart_movement_df.columns
            ),
        },
        {
            "Validation Area": "Chart-movement direction assigned",
            "Passed": chart_movement_df[
                "chart_movement_direction"
            ].notna().all(),
        },
        {
            "Validation Area": "Chart-movement score calculated",
            "Passed": (
                "chart_movement_score"
                in chart_movement_df.columns
            ),
        },
        {
            "Validation Area": "Chart-movement classes assigned",
            "Passed": chart_movement_df[
                "chart_movement_class"
            ].notna().all(),
        },
    ]
)

print()
print("Chart-movement validation")
print("=" * 80)

display(chart_movement_validation_df)

# ============================================================
# Completion
# ============================================================

section_6_4_complete = bool(
    chart_movement_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 6.4 completion status: "
    f"{section_6_4_complete}"
)

if not section_6_4_complete:
    raise RuntimeError(
        "Chart Movement analysis did not complete successfully."
    )
```

### Interpretation

The PMIP chart-movement assessment completed successfully.

Chart-movement evidence is available for 210 of the 4,593 validated releases. The remaining 4,383 releases do not have sufficiently reliable first-to-last chart-position evidence and are therefore classified as having `No Chart Movement Evidence`.

Among the releases with usable movement evidence, 64 show chart improvement, 111 show chart decline, and 35 remain stable between their first and last observed chart positions.

The median chart-position movement is -5, while the mean movement is approximately -20.03. Because positive values represent improvement and negative values represent decline, these results indicate that the typical matched release moved slightly downward in chart position over the observed period.

The more negative mean compared with the median suggests that a smaller number of releases experienced relatively large declines, pulling the average downward.

The movement-class distribution separates the matched releases into Moderate Decline, Strong Decline, Stable Chart Position, Moderate Improvement and Strong Improvement groups.

A total of 18 ambiguous chart-history signatures were detected during the matching process. These cases indicate that some historical signatures are not completely unique and should be interpreted with additional caution.

Chart movement captures direction rather than overall chart strength. A release may decline from a very strong chart position and still remain commercially significant, while another release may improve from a relatively weak starting point without reaching the highest chart levels.

For this reason, chart movement should be interpreted alongside best chart position, average chart position and chart persistence rather than used as a standalone performance measure.

Section 6.4 completed successfully and adds the directional chart-momentum component to the PMIP Chart Performance Intelligence framework.

## 6.5 Chart Activity

### Purpose

This subsection measures the overall level of chart activity associated with each validated release.

While earlier chart subsections focused on position quality, persistence and movement, chart activity focuses on the intensity and breadth of chart presence.

The assessment will examine:

- number of historical chart observations;
- number of countries in which the release charted;
- chart-history duration;
- frequency of chart appearances;
- overall activity strength relative to the charted population.

The aim is to distinguish releases with broad and sustained chart activity from releases with only limited chart presence.

The resulting Chart Activity component will provide the final input required before constructing the combined PMIP Chart Performance Indicator.

```python
# ============================================================
# Section 6.5 - Chart Activity
# ============================================================

if not section_6_4_complete:
    raise RuntimeError(
        "Section 6.4 must be completed before analysing chart activity."
    )

# ============================================================
# Working source
# ============================================================

chart_activity_df = chart_movement_df.copy()

# ============================================================
# Required fields
# ============================================================

required_chart_activity_fields = [
    "chart_observations",
    "countries_charted",
    "chart_history_span_days",
]

missing_chart_activity_fields = [
    field
    for field in required_chart_activity_fields
    if field not in chart_activity_df.columns
]

if missing_chart_activity_fields:
    raise RuntimeError(
        "Missing required chart-activity fields: "
        f"{missing_chart_activity_fields}"
    )

# ============================================================
# Ensure numeric values
# ============================================================

for field in required_chart_activity_fields:
    chart_activity_df[field] = pd.to_numeric(
        chart_activity_df[field],
        errors="coerce"
    )

# ============================================================
# Activity evidence availability
# ============================================================

chart_activity_df["chart_activity_evidence_available"] = (
    chart_activity_df["chart_observations"].notna()
    & chart_activity_df["countries_charted"].notna()
    & chart_activity_df["chart_history_span_days"].notna()
)

valid_chart_activity_mask = (
    chart_activity_df[
        "chart_activity_evidence_available"
    ]
)

# ============================================================
# Activity component scores
#
# Higher = stronger chart activity
# ============================================================

chart_activity_df["chart_observation_activity_score"] = np.nan
chart_activity_df["chart_country_activity_score"] = np.nan
chart_activity_df["chart_duration_activity_score"] = np.nan

if valid_chart_activity_mask.any():

    chart_activity_df.loc[
        valid_chart_activity_mask,
        "chart_observation_activity_score"
    ] = (
        chart_activity_df.loc[
            valid_chart_activity_mask,
            "chart_observations"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

    chart_activity_df.loc[
        valid_chart_activity_mask,
        "chart_country_activity_score"
    ] = (
        chart_activity_df.loc[
            valid_chart_activity_mask,
            "countries_charted"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

    chart_activity_df.loc[
        valid_chart_activity_mask,
        "chart_duration_activity_score"
    ] = (
        chart_activity_df.loc[
            valid_chart_activity_mask,
            "chart_history_span_days"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Combined chart-activity score
#
# Weighting:
# - chart observations: 45%
# - countries charted: 35%
# - chart duration: 20%
# ============================================================

chart_activity_df["chart_activity_score"] = np.where(
    valid_chart_activity_mask,
    (
        chart_activity_df[
            "chart_observation_activity_score"
        ] * 0.45
        + chart_activity_df[
            "chart_country_activity_score"
        ] * 0.35
        + chart_activity_df[
            "chart_duration_activity_score"
        ] * 0.20
    ),
    np.nan,
)

# ============================================================
# Chart-activity classes
# ============================================================

chart_activity_df["chart_activity_class"] = (
    "No Chart Activity Evidence"
)

valid_chart_activity_score_mask = (
    chart_activity_df["chart_activity_score"].notna()
)

if valid_chart_activity_score_mask.any():

    q25 = chart_activity_df.loc[
        valid_chart_activity_score_mask,
        "chart_activity_score"
    ].quantile(0.25)

    q50 = chart_activity_df.loc[
        valid_chart_activity_score_mask,
        "chart_activity_score"
    ].quantile(0.50)

    q75 = chart_activity_df.loc[
        valid_chart_activity_score_mask,
        "chart_activity_score"
    ].quantile(0.75)

    chart_activity_df.loc[
        valid_chart_activity_score_mask
        & (
            chart_activity_df[
                "chart_activity_score"
            ] <= q25
        ),
        "chart_activity_class"
    ] = "Low Chart Activity"

    chart_activity_df.loc[
        valid_chart_activity_score_mask
        & (
            chart_activity_df[
                "chart_activity_score"
            ] > q25
        )
        & (
            chart_activity_df[
                "chart_activity_score"
            ] <= q50
        ),
        "chart_activity_class"
    ] = "Moderate Chart Activity"

    chart_activity_df.loc[
        valid_chart_activity_score_mask
        & (
            chart_activity_df[
                "chart_activity_score"
            ] > q50
        )
        & (
            chart_activity_df[
                "chart_activity_score"
            ] <= q75
        ),
        "chart_activity_class"
    ] = "High Chart Activity"

    chart_activity_df.loc[
        valid_chart_activity_score_mask
        & (
            chart_activity_df[
                "chart_activity_score"
            ] > q75
        ),
        "chart_activity_class"
    ] = "Very High Chart Activity"

# ============================================================
# Summary
# ============================================================

chart_activity_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(chart_activity_df),
        },
        {
            "Metric": "Releases with chart-activity evidence",
            "Observed Value": int(
                valid_chart_activity_mask.sum()
            ),
        },
        {
            "Metric": "Releases without chart-activity evidence",
            "Observed Value": int(
                (~valid_chart_activity_mask).sum()
            ),
        },
        {
            "Metric": "Median chart observations",
            "Observed Value": (
                chart_activity_df.loc[
                    valid_chart_activity_mask,
                    "chart_observations"
                ].median()
            ),
        },
        {
            "Metric": "Median countries charted",
            "Observed Value": (
                chart_activity_df.loc[
                    valid_chart_activity_mask,
                    "countries_charted"
                ].median()
            ),
        },
        {
            "Metric": "Median chart-history span (days)",
            "Observed Value": (
                chart_activity_df.loc[
                    valid_chart_activity_mask,
                    "chart_history_span_days"
                ].median()
            ),
        },
        {
            "Metric": "Median chart-activity score",
            "Observed Value": (
                chart_activity_df[
                    "chart_activity_score"
                ].median()
            ),
        },
        {
            "Metric": "Maximum chart-activity score",
            "Observed Value": (
                chart_activity_df[
                    "chart_activity_score"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP chart activity")
print("=" * 80)

display(chart_activity_summary_df)

# ============================================================
# Chart-activity class distribution
# ============================================================

chart_activity_class_distribution_df = (
    chart_activity_df[
        "chart_activity_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Chart Activity Class")
    .reset_index(name="Releases")
)

chart_activity_class_distribution_df["Share (%)"] = (
    chart_activity_class_distribution_df["Releases"]
    / len(chart_activity_df)
    * 100
)

print()
print("Chart-activity class distribution")
print("=" * 80)

display(chart_activity_class_distribution_df)

# ============================================================
# Strongest chart-activity releases
# ============================================================

top_chart_activity_df = (
    chart_activity_df[
        valid_chart_activity_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "chart_observations",
            "countries_charted",
            "chart_history_span_days",
            "chart_observation_activity_score",
            "chart_country_activity_score",
            "chart_duration_activity_score",
            "chart_activity_score",
            "chart_activity_class",
        ]
    ]
    .sort_values(
        by="chart_activity_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest chart activity")
print("=" * 80)

display(top_chart_activity_df)

# ============================================================
# Validation
# ============================================================

chart_activity_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.4 completed",
            "Passed": section_6_4_complete,
        },
        {
            "Validation Area": "Chart observations available",
            "Passed": (
                "chart_observations"
                in chart_activity_df.columns
            ),
        },
        {
            "Validation Area": "Country chart activity available",
            "Passed": (
                "countries_charted"
                in chart_activity_df.columns
            ),
        },
        {
            "Validation Area": "Chart duration available",
            "Passed": (
                "chart_history_span_days"
                in chart_activity_df.columns
            ),
        },
        {
            "Validation Area": "Chart-activity evidence available",
            "Passed": valid_chart_activity_mask.any(),
        },
        {
            "Validation Area": "Chart-activity score calculated",
            "Passed": (
                "chart_activity_score"
                in chart_activity_df.columns
            ),
        },
        {
            "Validation Area": "Chart-activity classes assigned",
            "Passed": chart_activity_df[
                "chart_activity_class"
            ].notna().all(),
        },
    ]
)

print()
print("Chart-activity validation")
print("=" * 80)

display(chart_activity_validation_df)

# ============================================================
# Completion
# ============================================================

section_6_5_complete = bool(
    chart_activity_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 6.5 completion status: "
    f"{section_6_5_complete}"
)

if not section_6_5_complete:
    raise RuntimeError(
        "Chart Activity analysis did not complete successfully."
    )
```

### Interpretation

The PMIP chart-activity assessment completed successfully.

Chart-activity evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain sufficient chart-history evidence.

Among releases with available chart activity, the median release has 409 historical chart observations across 24 countries and a chart-history span of approximately 413 days.

These figures indicate that many charted releases have substantial levels of repeated and geographically distributed activity rather than isolated chart appearances.

The median chart-activity score is approximately 50.43, while the maximum observed score is approximately 98.37.

The activity classes are divided into Low, Moderate, High and Very High Chart Activity groups using percentile-based thresholds. The near-even distribution across these groups reflects the relative classification method rather than a naturally equal population structure.

Chart activity combines chart observation frequency, geographic breadth and chart-history duration into a single measure of how intensively a release has participated in historical charts.

This measure is closely related to chart persistence but has a slightly different analytical role. Chart persistence focuses on durability, while chart activity emphasises the overall amount and breadth of chart participation.

A release with high chart activity therefore has relatively strong chart exposure across observations, countries and time, but this does not necessarily mean that it achieved strong chart positions or positive chart movement.

For this reason, chart activity should be interpreted alongside best chart position, average chart position, chart persistence and chart movement.

Section 6.5 completed successfully and provides the final chart-level component required before constructing the overall PMIP Chart Performance Indicator.

## 6.6 Chart Performance Indicator

### Purpose

This subsection combines the major chart-performance components into a single coverage-aware PMIP Chart Performance Indicator.

The indicator brings together:

- best chart position;
- average chart position;
- chart persistence;
- chart movement;
- chart activity.

Because these components do not have equal evidence coverage, the final score uses coverage-aware weighting.

Best chart position, average chart position, persistence and activity are available for a much larger share of charted releases, while chart movement is available for only a smaller matched subset.

Missing chart-movement evidence is therefore not treated as poor performance. Instead, the weights of the available chart components are re-normalised for each release.

The resulting indicator is designed to summarise chart strength while preserving transparency about the amount of evidence supporting each score.

```python
# ============================================================
# Section 6.6 - Chart Performance Indicator
# ============================================================

if not section_6_5_complete:
    raise RuntimeError(
        "Section 6.5 must be completed before constructing the Chart Performance Indicator."
    )

# ============================================================
# Working source
# ============================================================

chart_performance_df = chart_activity_df.copy()

# ============================================================
# Component preparation
# ============================================================

chart_performance_df["best_chart_position_component"] = pd.to_numeric(
    chart_performance_df["best_chart_position_score"],
    errors="coerce"
)

chart_performance_df["average_chart_position_component"] = pd.to_numeric(
    chart_performance_df["average_chart_position_score"],
    errors="coerce"
)

chart_performance_df["chart_persistence_component"] = pd.to_numeric(
    chart_performance_df["chart_persistence_score"],
    errors="coerce"
)

chart_performance_df["chart_movement_component"] = pd.to_numeric(
    chart_performance_df["chart_movement_score"],
    errors="coerce"
)

chart_performance_df["chart_activity_component"] = pd.to_numeric(
    chart_performance_df["chart_activity_score"],
    errors="coerce"
)

# ============================================================
# Base weights
#
# Position quality and sustained chart behaviour receive
# the largest influence.
#
# Movement receives a smaller weight because coverage is low.
# ============================================================

chart_component_weights = {
    "best_chart_position_component": 0.25,
    "average_chart_position_component": 0.25,
    "chart_persistence_component": 0.20,
    "chart_movement_component": 0.10,
    "chart_activity_component": 0.20,
}

chart_component_columns = list(
    chart_component_weights.keys()
)

# ============================================================
# Coverage-aware weighted score
# ============================================================

chart_weighted_sum = pd.Series(
    0.0,
    index=chart_performance_df.index
)

chart_available_weight_sum = pd.Series(
    0.0,
    index=chart_performance_df.index
)

for column, weight in chart_component_weights.items():

    available_mask = (
        chart_performance_df[column].notna()
    )

    chart_weighted_sum.loc[
        available_mask
    ] += (
        chart_performance_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    chart_available_weight_sum.loc[
        available_mask
    ] += weight

chart_performance_df["chart_performance_score"] = np.where(
    chart_available_weight_sum > 0,
    chart_weighted_sum / chart_available_weight_sum,
    np.nan,
)

# ============================================================
# Evidence coverage
# ============================================================

chart_performance_df["chart_components_available"] = (
    chart_performance_df[
        chart_component_columns
    ]
    .notna()
    .sum(axis=1)
)

chart_performance_df["chart_evidence_coverage_pct"] = (
    chart_performance_df[
        "chart_components_available"
    ]
    / len(chart_component_columns)
    * 100
)

# ============================================================
# Evidence strength
# ============================================================

chart_performance_df["chart_evidence_strength"] = np.select(
    [
        chart_performance_df[
            "chart_components_available"
        ] >= 5,

        chart_performance_df[
            "chart_components_available"
        ] == 4,

        chart_performance_df[
            "chart_components_available"
        ] == 3,

        chart_performance_df[
            "chart_components_available"
        ] == 2,

        chart_performance_df[
            "chart_components_available"
        ] == 1,
    ],
    [
        "Very Strong Evidence",
        "Strong Evidence",
        "Moderate Evidence",
        "Limited Evidence",
        "Minimal Evidence",
    ],
    default="No Evidence",
)

# ============================================================
# Percentile
# ============================================================

valid_chart_performance_mask = (
    chart_performance_df[
        "chart_performance_score"
    ].notna()
)

chart_performance_df["chart_performance_percentile"] = np.nan

if valid_chart_performance_mask.any():

    chart_performance_df.loc[
        valid_chart_performance_mask,
        "chart_performance_percentile"
    ] = (
        chart_performance_df.loc[
            valid_chart_performance_mask,
            "chart_performance_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Rank
# ============================================================

chart_performance_df["chart_performance_rank"] = (
    chart_performance_df[
        "chart_performance_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Performance classes
# ============================================================

chart_performance_df["chart_performance_class"] = (
    "No Chart Performance Evidence"
)

valid_chart_score_mask = (
    chart_performance_df[
        "chart_performance_score"
    ].notna()
)

if valid_chart_score_mask.any():

    q25 = chart_performance_df.loc[
        valid_chart_score_mask,
        "chart_performance_score"
    ].quantile(0.25)

    q50 = chart_performance_df.loc[
        valid_chart_score_mask,
        "chart_performance_score"
    ].quantile(0.50)

    q75 = chart_performance_df.loc[
        valid_chart_score_mask,
        "chart_performance_score"
    ].quantile(0.75)

    q90 = chart_performance_df.loc[
        valid_chart_score_mask,
        "chart_performance_score"
    ].quantile(0.90)

    chart_performance_df.loc[
        valid_chart_score_mask
        & (
            chart_performance_df[
                "chart_performance_score"
            ] <= q25
        ),
        "chart_performance_class"
    ] = "Low Chart Performance"

    chart_performance_df.loc[
        valid_chart_score_mask
        & (
            chart_performance_df[
                "chart_performance_score"
            ] > q25
        )
        & (
            chart_performance_df[
                "chart_performance_score"
            ] <= q50
        ),
        "chart_performance_class"
    ] = "Moderate Chart Performance"

    chart_performance_df.loc[
        valid_chart_score_mask
        & (
            chart_performance_df[
                "chart_performance_score"
            ] > q50
        )
        & (
            chart_performance_df[
                "chart_performance_score"
            ] <= q75
        ),
        "chart_performance_class"
    ] = "High Chart Performance"

    chart_performance_df.loc[
        valid_chart_score_mask
        & (
            chart_performance_df[
                "chart_performance_score"
            ] > q75
        )
        & (
            chart_performance_df[
                "chart_performance_score"
            ] <= q90
        ),
        "chart_performance_class"
    ] = "Very High Chart Performance"

    chart_performance_df.loc[
        valid_chart_score_mask
        & (
            chart_performance_df[
                "chart_performance_score"
            ] > q90
        ),
        "chart_performance_class"
    ] = "Exceptional Chart Performance"

# ============================================================
# Summary
# ============================================================

chart_performance_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                chart_performance_df
            ),
        },
        {
            "Metric": "Releases with chart-performance score",
            "Observed Value": int(
                valid_chart_performance_mask.sum()
            ),
        },
        {
            "Metric": "Releases without chart-performance score",
            "Observed Value": int(
                (~valid_chart_performance_mask).sum()
            ),
        },
        {
            "Metric": "Median chart-performance score",
            "Observed Value": (
                chart_performance_df[
                    "chart_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean chart-performance score",
            "Observed Value": (
                chart_performance_df[
                    "chart_performance_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum chart-performance score",
            "Observed Value": (
                chart_performance_df[
                    "chart_performance_score"
                ].max()
            ),
        },
        {
            "Metric": "Median chart evidence coverage (%)",
            "Observed Value": (
                chart_performance_df[
                    "chart_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Constructing PMIP Chart Performance Indicator")
print("=" * 80)

display(chart_performance_summary_df)

# ============================================================
# Component coverage
# ============================================================

chart_component_coverage_rows = []

for column in chart_component_columns:

    chart_component_coverage_rows.append(
        {
            "Component": column,
            "Releases Available": int(
                chart_performance_df[
                    column
                ].notna().sum()
            ),
            "Coverage (%)": round(
                chart_performance_df[
                    column
                ].notna().mean()
                * 100,
                6
            ),
            "Base Weight": (
                chart_component_weights[
                    column
                ]
            ),
        }
    )

chart_component_coverage_df = pd.DataFrame(
    chart_component_coverage_rows
)

print()
print("Chart indicator component coverage")
print("=" * 80)

display(chart_component_coverage_df)

# ============================================================
# Evidence-strength distribution
# ============================================================

chart_evidence_strength_distribution_df = (
    chart_performance_df[
        "chart_evidence_strength"
    ]
    .value_counts()
    .rename_axis("Evidence Strength")
    .reset_index(name="Releases")
)

chart_evidence_strength_distribution_df["Share (%)"] = (
    chart_evidence_strength_distribution_df["Releases"]
    / len(chart_performance_df)
    * 100
)

print()
print("Chart evidence-strength distribution")
print("=" * 80)

display(
    chart_evidence_strength_distribution_df
)

# ============================================================
# Chart-performance class distribution
# ============================================================

chart_performance_class_distribution_df = (
    chart_performance_df[
        "chart_performance_class"
    ]
    .value_counts()
    .rename_axis("Chart Performance Class")
    .reset_index(name="Releases")
)

chart_performance_class_distribution_df["Share (%)"] = (
    chart_performance_class_distribution_df["Releases"]
    / len(chart_performance_df)
    * 100
)

print()
print("Chart-performance class distribution")
print("=" * 80)

display(
    chart_performance_class_distribution_df
)

# ============================================================
# Highest-scoring releases
# ============================================================

top_chart_performance_df = (
    chart_performance_df[
        [
            "release_id",
            "Track",
            "Artist",
            "best_chart_position_component",
            "average_chart_position_component",
            "chart_persistence_component",
            "chart_movement_component",
            "chart_activity_component",
            "chart_components_available",
            "chart_evidence_strength",
            "chart_performance_score",
            "chart_performance_percentile",
            "chart_performance_rank",
            "chart_performance_class",
        ]
    ]
    .sort_values(
        by="chart_performance_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP chart-performance releases")
print("=" * 80)

display(top_chart_performance_df)

# ============================================================
# Validation
# ============================================================

chart_performance_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.5 completed",
            "Passed": section_6_5_complete,
        },
        {
            "Validation Area": "Best-position component available",
            "Passed": chart_performance_df[
                "best_chart_position_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Average-position component available",
            "Passed": chart_performance_df[
                "average_chart_position_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Chart-persistence component available",
            "Passed": chart_performance_df[
                "chart_persistence_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Chart-activity component available",
            "Passed": chart_performance_df[
                "chart_activity_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Coverage-aware chart weighting prepared",
            "Passed": (
                chart_available_weight_sum > 0
            ).any(),
        },
        {
            "Validation Area": "Chart-performance score calculated",
            "Passed": chart_performance_df[
                "chart_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Chart evidence coverage calculated",
            "Passed": chart_performance_df[
                "chart_evidence_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Chart-performance classes assigned",
            "Passed": chart_performance_df[
                "chart_performance_class"
            ].notna().all(),
        },
    ]
)

print()
print("Chart-performance indicator validation")
print("=" * 80)

display(chart_performance_validation_df)

# ============================================================
# Section completion
# ============================================================

section_6_6_complete = bool(
    chart_performance_validation_df[
        "Passed"
    ].all()
)

section_6_complete = bool(
    section_6_1_complete
    and section_6_2_complete
    and section_6_3_complete
    and section_6_4_complete
    and section_6_5_complete
    and section_6_6_complete
)

print()
print(
    f"Section 6.6 completion status: "
    f"{section_6_6_complete}"
)

print(
    f"Section 6 overall completion status: "
    f"{section_6_complete}"
)

if not section_6_complete:
    raise RuntimeError(
        "Section 6 Chart Performance Intelligence "
        "did not complete successfully."
    )
```

### Interpretation

The PMIP Chart Performance Indicator was constructed successfully.

A chart-performance score was produced for 2,235 of the 4,593 validated releases. The remaining 2,358 releases do not contain sufficient historical chart evidence and are therefore classified as `No Chart Performance Evidence`.

The median chart-performance score is approximately 51.82, while the mean is approximately 50.29. The maximum observed score is approximately 91.08.

The indicator combines five analytical components: best chart position, average chart position, chart persistence, chart movement and chart activity.

Best chart position, average chart position, persistence and activity are each available for 2,235 releases, representing approximately 48.66% of the complete release population. Chart movement has substantially lower coverage, with usable evidence for only 210 releases, or approximately 4.57%.

Because of this difference in coverage, the Chart Performance Indicator uses coverage-aware weighting. Missing chart-movement evidence is not treated as poor performance. Instead, the available component weights are re-normalised for each release before the final score is calculated.

The evidence-strength distribution shows that 2,025 releases have `Strong Evidence`, meaning four of the five chart components are available. A further 210 releases have `Very Strong Evidence`, meaning all five chart components, including chart movement, are available.

The reported median chart-evidence coverage across the complete 4,593-release population is 0%. This occurs because more than half of the full population has no chart evidence. It should not be interpreted as indicating weak evidence among the releases that actually received a chart-performance score.

Among releases with chart evidence, the typical release is supported by four of the five analytical components, representing 80% component coverage, while the 210 releases with reliable chart-movement evidence have complete 100% component coverage.

The chart-performance classes divide the scored population into Low, Moderate, High, Very High and Exceptional Chart Performance groups. These classes are relative to the observed PMIP charted population rather than universal measures of commercial performance.

A total of 224 releases are classified as `Exceptional Chart Performance`, while 335 are classified as `Very High Chart Performance`. These releases combine comparatively strong chart position, persistence and activity, with chart movement incorporated where reliable evidence is available.

The highest-ranked releases generally demonstrate strong results across multiple chart dimensions rather than relying on a single peak position. This gives the combined indicator a broader interpretation than best chart position alone.

The final `chart_performance_score` should therefore be interpreted together with `chart_evidence_strength` and `chart_evidence_coverage_pct`, particularly when comparing releases with and without chart-movement evidence.

Section 6.6 completed successfully, and Section 6 overall is now complete.

The PMIP Chart Performance Intelligence stage therefore provides a transparent, coverage-aware measure of historical chart strength that can later be combined with streaming and other release-level intelligence components.

# 7. Geographic Release Reach

This section evaluates how widely each validated release is represented across different geographic markets.

The aim is to move beyond overall chart and streaming performance and examine the geographic spread of release activity.

The analysis will focus on:

- number of countries represented;
- overall market breadth;
- presence within stronger established markets;
- presence within emerging markets;
- construction of a combined Geographic Reach Indicator.

Geographic reach provides an important additional perspective because two releases with similar overall performance may differ significantly in how widely that performance is distributed across countries.

The resulting geographic intelligence will later contribute to the wider PMIP Release Performance Intelligence framework.

## 7.1 Countries Represented

### Purpose

This subsection measures the number of countries in which each validated release has recorded historical chart activity.

The aim is to establish the basic geographic coverage of each release before analysing broader market breadth and market-specific strength.

The assessment will examine:

- number of countries represented;
- releases with no geographic chart evidence;
- distribution of country coverage;
- relative geographic reach across releases;
- releases with the broadest international representation.

Higher country counts indicate broader geographic reach, while releases without country-level evidence remain explicitly identified rather than being treated as having weak geographic performance.

```python
# ============================================================
# Section 7.1 - Countries Represented
# ============================================================

if not section_6_complete:
    raise RuntimeError(
        "Section 6 must be completed before beginning Geographic Release Reach."
    )

# ============================================================
# Working source
# ============================================================

geographic_reach_df = chart_performance_df.copy()

# ============================================================
# Validate countries-charted field
# ============================================================

if "countries_charted" not in geographic_reach_df.columns:
    raise RuntimeError(
        "countries_charted is not available in the release population."
    )

geographic_reach_df["countries_charted"] = pd.to_numeric(
    geographic_reach_df["countries_charted"],
    errors="coerce"
)

# ============================================================
# Geographic evidence availability
# ============================================================

geographic_reach_df["countries_represented_available"] = (
    geographic_reach_df["countries_charted"].notna()
)

valid_country_reach_mask = (
    geographic_reach_df[
        "countries_represented_available"
    ]
)

valid_country_counts = geographic_reach_df.loc[
    valid_country_reach_mask,
    "countries_charted"
]

# ============================================================
# Summary
# ============================================================

countries_represented_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                geographic_reach_df
            ),
        },
        {
            "Metric": "Releases with country-representation evidence",
            "Observed Value": int(
                valid_country_reach_mask.sum()
            ),
        },
        {
            "Metric": "Releases without country-representation evidence",
            "Observed Value": int(
                (~valid_country_reach_mask).sum()
            ),
        },
        {
            "Metric": "Minimum countries represented",
            "Observed Value": (
                float(valid_country_counts.min())
                if not valid_country_counts.empty
                else np.nan
            ),
        },
        {
            "Metric": "Median countries represented",
            "Observed Value": (
                float(valid_country_counts.median())
                if not valid_country_counts.empty
                else np.nan
            ),
        },
        {
            "Metric": "Mean countries represented",
            "Observed Value": (
                float(valid_country_counts.mean())
                if not valid_country_counts.empty
                else np.nan
            ),
        },
        {
            "Metric": "Maximum countries represented",
            "Observed Value": (
                float(valid_country_counts.max())
                if not valid_country_counts.empty
                else np.nan
            ),
        },
    ]
)

print("Assessing PMIP countries represented")
print("=" * 80)

display(countries_represented_summary_df)

# ============================================================
# Country-coverage percentiles
# ============================================================

country_reach_percentiles = [
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
]

countries_represented_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
        ],
        "Countries Represented": (
            valid_country_counts
            .quantile(
                country_reach_percentiles
            )
            .values
        ),
    }
)

print()
print("Country-representation percentiles")
print("=" * 80)

display(
    countries_represented_percentiles_df
)

# ============================================================
# Relative country-reach score
#
# Higher countries represented = stronger geographic reach.
# ============================================================

geographic_reach_df["countries_represented_score"] = np.nan

if valid_country_reach_mask.any():

    geographic_reach_df.loc[
        valid_country_reach_mask,
        "countries_represented_score"
    ] = (
        geographic_reach_df.loc[
            valid_country_reach_mask,
            "countries_charted"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Country-representation rank
# ============================================================

geographic_reach_df["countries_represented_rank"] = (
    geographic_reach_df[
        "countries_charted"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Country-representation classes
# ============================================================

geographic_reach_df["countries_represented_class"] = (
    "No Geographic Evidence"
)

valid_country_score_mask = (
    geographic_reach_df[
        "countries_represented_score"
    ].notna()
)

if valid_country_score_mask.any():

    q25 = geographic_reach_df.loc[
        valid_country_score_mask,
        "countries_charted"
    ].quantile(0.25)

    q50 = geographic_reach_df.loc[
        valid_country_score_mask,
        "countries_charted"
    ].quantile(0.50)

    q75 = geographic_reach_df.loc[
        valid_country_score_mask,
        "countries_charted"
    ].quantile(0.75)

    geographic_reach_df.loc[
        valid_country_score_mask
        & (
            geographic_reach_df[
                "countries_charted"
            ] <= q25
        ),
        "countries_represented_class"
    ] = "Limited Country Reach"

    geographic_reach_df.loc[
        valid_country_score_mask
        & (
            geographic_reach_df[
                "countries_charted"
            ] > q25
        )
        & (
            geographic_reach_df[
                "countries_charted"
            ] <= q50
        ),
        "countries_represented_class"
    ] = "Moderate Country Reach"

    geographic_reach_df.loc[
        valid_country_score_mask
        & (
            geographic_reach_df[
                "countries_charted"
            ] > q50
        )
        & (
            geographic_reach_df[
                "countries_charted"
            ] <= q75
        ),
        "countries_represented_class"
    ] = "Broad Country Reach"

    geographic_reach_df.loc[
        valid_country_score_mask
        & (
            geographic_reach_df[
                "countries_charted"
            ] > q75
        ),
        "countries_represented_class"
    ] = "Very Broad Country Reach"

# ============================================================
# Class distribution
# ============================================================

countries_represented_class_distribution_df = (
    geographic_reach_df[
        "countries_represented_class"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Countries Represented Class"
    )
    .reset_index(
        name="Releases"
    )
)

countries_represented_class_distribution_df[
    "Share (%)"
] = (
    countries_represented_class_distribution_df[
        "Releases"
    ]
    / len(geographic_reach_df)
    * 100
)

print()
print("Countries-represented class distribution")
print("=" * 80)

display(
    countries_represented_class_distribution_df
)

# ============================================================
# Releases with broadest geographic reach
# ============================================================

top_country_reach_df = (
    geographic_reach_df[
        valid_country_reach_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "countries_charted",
            "countries_represented_score",
            "countries_represented_rank",
            "countries_represented_class",
            "chart_observations",
            "chart_performance_score",
        ]
    ]
    .sort_values(
        by=[
            "countries_charted",
            "chart_observations",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with broadest country representation")
print("=" * 80)

display(top_country_reach_df)

# ============================================================
# Validation
# ============================================================

countries_represented_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6 completed",
            "Passed": section_6_complete,
        },
        {
            "Validation Area": "Countries-charted field available",
            "Passed": (
                "countries_charted"
                in geographic_reach_df.columns
            ),
        },
        {
            "Validation Area": "Country-representation evidence available",
            "Passed": valid_country_reach_mask.any(),
        },
        {
            "Validation Area": "Country-representation score calculated",
            "Passed": (
                "countries_represented_score"
                in geographic_reach_df.columns
            ),
        },
        {
            "Validation Area": "Country-representation rank prepared",
            "Passed": (
                "countries_represented_rank"
                in geographic_reach_df.columns
            ),
        },
        {
            "Validation Area": "Country-representation classes assigned",
            "Passed": geographic_reach_df[
                "countries_represented_class"
            ].notna().all(),
        },
    ]
)

print()
print("Countries-represented validation")
print("=" * 80)

display(
    countries_represented_validation_df
)

# ============================================================
# Completion
# ============================================================

section_7_1_complete = bool(
    countries_represented_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 7.1 completion status: "
    f"{section_7_1_complete}"
)

if not section_7_1_complete:
    raise RuntimeError(
        "Countries Represented analysis did not complete successfully."
    )
```

### Interpretation

The PMIP countries-represented assessment completed successfully.

Country-level geographic evidence is available for 2,235 of the 4,593 validated releases. The remaining 2,358 releases do not contain sufficient country-level chart evidence and are therefore classified as having `No Geographic Evidence`.

Among releases with geographic evidence, the number of represented countries ranges from 1 to 74.

The median release is represented in 24 countries, while the mean is approximately 30.27 countries. The higher mean compared with the median indicates that a number of releases have particularly broad international coverage and pull the average upward.

The percentile distribution shows clear differences in geographic reach. Around 25% of charted releases are represented in 8 countries or fewer, while the median is 24 countries. At the 75th percentile, releases are represented in approximately 54 countries, increasing to 64 countries at the 90th percentile and 69 countries at the 95th percentile.

This indicates that the most internationally distributed releases have chart evidence across a very large proportion of the available PMIP markets.

The country-representation classes separate the charted release population into Limited, Moderate, Broad and Very Broad Country Reach groups.

A total of 542 releases are classified as `Very Broad Country Reach`, while 571 releases fall within `Broad Country Reach`. These releases demonstrate substantially wider geographic representation than the lower-reach groups.

The country-representation score is relative to the assessed PMIP population. It measures geographic spread rather than performance quality within individual markets.

A release appearing in many countries is therefore not automatically performing strongly in all of those markets. This distinction becomes important in the following subsections, where market breadth and market-specific presence are assessed in greater detail.

Section 7.1 completed successfully and establishes the basic geographic-coverage component of the PMIP Geographic Release Reach framework.

## 7.2 Market Breadth

### Purpose

This subsection measures how broadly each validated release is distributed across the available PMIP geographic markets.

While Section 7.1 counted the number of countries represented, Market Breadth converts that raw country count into a relative measure of how widely a release reaches across the full observed market universe.

The assessment will examine:

- the total number of geographic markets available in the historical chart data;
- the proportion of available markets reached by each release;
- differences between narrow, moderate and broad international distribution;
- relative market-breadth strength across the charted release population.

The resulting Market Breadth component provides a normalised geographic measure that can later be combined with strong-market presence, emerging-market presence and overall country representation.

```python
# ============================================================
# Section 7.2 - Market Breadth
# ============================================================

if not section_7_1_complete:
    raise RuntimeError(
        "Section 7.1 must be completed before analysing market breadth."
    )

# ============================================================
# Working source
# ============================================================

market_breadth_df = geographic_reach_df.copy()

# ============================================================
# Determine available geographic market universe
# ============================================================

if "country" not in release_history_df.columns:
    raise RuntimeError(
        "country field is not available in release_history_df."
    )

available_market_series = (
    release_history_df["country"]
    .dropna()
    .astype(str)
    .str.strip()
)

available_market_series = available_market_series[
    available_market_series.ne("")
]

total_available_markets = int(
    available_market_series.nunique()
)

if total_available_markets <= 0:
    raise RuntimeError(
        "No geographic markets were detected in the historical chart data."
    )

# ============================================================
# Prepare countries represented
# ============================================================

market_breadth_df["countries_charted"] = pd.to_numeric(
    market_breadth_df["countries_charted"],
    errors="coerce"
)

valid_market_breadth_mask = (
    market_breadth_df["countries_charted"].notna()
)

# ============================================================
# Raw market-breadth proportion
#
# Example:
# 24 countries out of 77 available markets
# = 31.17% market breadth
# ============================================================

market_breadth_df["market_breadth_pct"] = np.where(
    valid_market_breadth_mask,
    (
        market_breadth_df["countries_charted"]
        / total_available_markets
        * 100
    ),
    np.nan,
)

market_breadth_df["market_breadth_pct"] = (
    market_breadth_df["market_breadth_pct"]
    .clip(lower=0, upper=100)
)

# ============================================================
# Relative PMIP market-breadth score
#
# Higher market breadth = stronger score
# ============================================================

market_breadth_df["market_breadth_score"] = np.nan

if valid_market_breadth_mask.any():

    market_breadth_df.loc[
        valid_market_breadth_mask,
        "market_breadth_score"
    ] = (
        market_breadth_df.loc[
            valid_market_breadth_mask,
            "market_breadth_pct"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Market-breadth rank
# ============================================================

market_breadth_df["market_breadth_rank"] = (
    market_breadth_df[
        "market_breadth_pct"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Market-breadth classes
# ============================================================

market_breadth_df["market_breadth_class"] = (
    "No Market Breadth Evidence"
)

market_breadth_df.loc[
    valid_market_breadth_mask
    & (
        market_breadth_df["market_breadth_pct"] < 10
    ),
    "market_breadth_class"
] = "Very Narrow Market Breadth"

market_breadth_df.loc[
    valid_market_breadth_mask
    & (
        market_breadth_df["market_breadth_pct"] >= 10
    )
    & (
        market_breadth_df["market_breadth_pct"] < 25
    ),
    "market_breadth_class"
] = "Narrow Market Breadth"

market_breadth_df.loc[
    valid_market_breadth_mask
    & (
        market_breadth_df["market_breadth_pct"] >= 25
    )
    & (
        market_breadth_df["market_breadth_pct"] < 50
    ),
    "market_breadth_class"
] = "Moderate Market Breadth"

market_breadth_df.loc[
    valid_market_breadth_mask
    & (
        market_breadth_df["market_breadth_pct"] >= 50
    )
    & (
        market_breadth_df["market_breadth_pct"] < 75
    ),
    "market_breadth_class"
] = "Broad Market Breadth"

market_breadth_df.loc[
    valid_market_breadth_mask
    & (
        market_breadth_df["market_breadth_pct"] >= 75
    ),
    "market_breadth_class"
] = "Very Broad Market Breadth"

# ============================================================
# Summary
# ============================================================

market_breadth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                market_breadth_df
            ),
        },
        {
            "Metric": "Available PMIP geographic markets",
            "Observed Value": total_available_markets,
        },
        {
            "Metric": "Releases with market-breadth evidence",
            "Observed Value": int(
                valid_market_breadth_mask.sum()
            ),
        },
        {
            "Metric": "Releases without market-breadth evidence",
            "Observed Value": int(
                (~valid_market_breadth_mask).sum()
            ),
        },
        {
            "Metric": "Median market breadth (%)",
            "Observed Value": (
                market_breadth_df.loc[
                    valid_market_breadth_mask,
                    "market_breadth_pct"
                ].median()
            ),
        },
        {
            "Metric": "Mean market breadth (%)",
            "Observed Value": (
                market_breadth_df.loc[
                    valid_market_breadth_mask,
                    "market_breadth_pct"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum market breadth (%)",
            "Observed Value": (
                market_breadth_df.loc[
                    valid_market_breadth_mask,
                    "market_breadth_pct"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP market breadth")
print("=" * 80)

display(market_breadth_summary_df)

# ============================================================
# Market-breadth percentiles
# ============================================================

market_breadth_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
        ],
        "Market Breadth (%)": (
            market_breadth_df.loc[
                valid_market_breadth_mask,
                "market_breadth_pct"
            ]
            .quantile(
                [
                    0.25,
                    0.50,
                    0.75,
                    0.90,
                    0.95,
                ]
            )
            .values
        ),
    }
)

print()
print("Market-breadth percentiles")
print("=" * 80)

display(market_breadth_percentiles_df)

# ============================================================
# Market-breadth class distribution
# ============================================================

market_breadth_class_distribution_df = (
    market_breadth_df[
        "market_breadth_class"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Market Breadth Class"
    )
    .reset_index(
        name="Releases"
    )
)

market_breadth_class_distribution_df["Share (%)"] = (
    market_breadth_class_distribution_df["Releases"]
    / len(market_breadth_df)
    * 100
)

print()
print("Market-breadth class distribution")
print("=" * 80)

display(
    market_breadth_class_distribution_df
)

# ============================================================
# Releases with broadest market coverage
# ============================================================

top_market_breadth_df = (
    market_breadth_df[
        valid_market_breadth_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "countries_charted",
            "market_breadth_pct",
            "market_breadth_score",
            "market_breadth_rank",
            "market_breadth_class",
        ]
    ]
    .sort_values(
        by=[
            "market_breadth_pct",
            "countries_charted",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with broadest market reach")
print("=" * 80)

display(top_market_breadth_df)

# ============================================================
# Validation
# ============================================================

market_breadth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.1 completed",
            "Passed": section_7_1_complete,
        },
        {
            "Validation Area": "Available market universe detected",
            "Passed": total_available_markets > 0,
        },
        {
            "Validation Area": "Market-breadth evidence available",
            "Passed": valid_market_breadth_mask.any(),
        },
        {
            "Validation Area": "Market-breadth percentage calculated",
            "Passed": (
                "market_breadth_pct"
                in market_breadth_df.columns
            ),
        },
        {
            "Validation Area": "Market-breadth score calculated",
            "Passed": (
                "market_breadth_score"
                in market_breadth_df.columns
            ),
        },
        {
            "Validation Area": "Market-breadth rank prepared",
            "Passed": (
                "market_breadth_rank"
                in market_breadth_df.columns
            ),
        },
        {
            "Validation Area": "Market-breadth classes assigned",
            "Passed": market_breadth_df[
                "market_breadth_class"
            ].notna().all(),
        },
    ]
)

print()
print("Market-breadth validation")
print("=" * 80)

display(market_breadth_validation_df)

# ============================================================
# Completion
# ============================================================

section_7_2_complete = bool(
    market_breadth_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 7.2 completion status: "
    f"{section_7_2_complete}"
)

if not section_7_2_complete:
    raise RuntimeError(
        "Market Breadth analysis did not complete successfully."
    )
```

### Interpretation

The PMIP Market Breadth assessment completed successfully.

The historical chart data contains 77 available geographic markets.

Market-breadth evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain sufficient geographic chart evidence.

Among releases with available geographic evidence, the median Market Breadth is approximately 31.17%. This means that the typical charted release appears in roughly one-third of the available PMIP markets.

The mean Market Breadth is approximately 39.31%, which is higher than the median. This indicates that a group of releases with very broad international coverage increases the overall average.

The percentile distribution shows substantial differences in geographic spread. At the 25th percentile, releases cover approximately 10.39% of available markets. The median is approximately 31.17%, while the 75th percentile rises sharply to approximately 70.13%.

The strongest geographic releases reach a very large share of the available market universe, with the 90th percentile at approximately 83.12%, the 95th percentile at approximately 89.61%, and the maximum observed breadth at approximately 96.10%.

The Market Breadth classes distinguish releases with Very Narrow, Narrow, Moderate, Broad and Very Broad geographic coverage.

A total of 425 releases are classified as `Broad Market Breadth`, while another 425 releases are classified as `Very Broad Market Breadth`.

The Market Breadth percentage provides a normalised view of geographic distribution because it expresses country coverage relative to the complete set of markets available in PMIP rather than relying only on the raw country count.

However, Market Breadth measures how widely a release is represented, not how strongly it performs within those markets.

A release can therefore have broad international coverage while performing only modestly in each market.

Section 7.2 completed successfully and provides the normalised geographic-spread component required for the PMIP Geographic Release Reach framework.

## 7.3 Strong-Market Presence

### Purpose

This subsection evaluates whether each validated release is represented within the strongest and most established geographic markets available in the PMIP chart data.

While Section 7.2 measured how broadly a release is distributed across all available markets, Strong-Market Presence focuses on where that reach occurs.

The assessment will identify stronger markets using the overall amount of historical chart activity recorded in each country and will then measure:

- how many strong markets each release reaches;
- what proportion of the strong-market set is represented;
- the relative strength of that presence;
- releases with concentrated presence in high-activity markets.

This provides a market-quality perspective that complements simple country count and overall market breadth.

```python
# ============================================================
# Section 7.3 - Strong-Market Presence
# ============================================================

if not section_7_2_complete:
    raise RuntimeError(
        "Section 7.2 must be completed before analysing strong-market presence."
    )

# ============================================================
# Working source
# ============================================================

strong_market_df = market_breadth_df.copy()

# ============================================================
# Prepare historical market activity
# ============================================================

if "country" not in release_history_df.columns:
    raise RuntimeError(
        "country field is not available in release_history_df."
    )

if "track_id" not in release_history_df.columns:
    raise RuntimeError(
        "track_id field is not available in release_history_df."
    )

market_activity_source = release_history_df.copy()

market_activity_source["country"] = (
    market_activity_source["country"]
    .astype(str)
    .str.strip()
)

market_activity_source = market_activity_source[
    market_activity_source["country"].notna()
    & market_activity_source["country"].ne("")
    & market_activity_source["track_id"].notna()
].copy()

# ============================================================
# Market-level activity summary
#
# Strong markets are defined using total historical chart
# observations. This is a data-driven definition based on the
# PMIP source rather than a manually chosen country list.
# ============================================================

market_activity_summary_df = (
    market_activity_source
    .groupby("country")
    .agg(
        chart_observations=("track_id", "size"),
        unique_tracks=("track_id", "nunique"),
    )
    .reset_index()
)

market_activity_summary_df["market_activity_rank"] = (
    market_activity_summary_df[
        "chart_observations"
    ]
    .rank(
        method="min",
        ascending=False
    )
)

market_activity_summary_df["market_activity_percentile"] = (
    market_activity_summary_df[
        "chart_observations"
    ]
    .rank(
        method="average",
        pct=True,
        ascending=True
    )
    * 100
)

# ============================================================
# Define strong markets
#
# Top quartile of markets by historical chart activity.
# ============================================================

strong_market_threshold = (
    market_activity_summary_df[
        "chart_observations"
    ]
    .quantile(0.75)
)

strong_market_summary_df = (
    market_activity_summary_df[
        market_activity_summary_df[
            "chart_observations"
        ] >= strong_market_threshold
    ]
    .copy()
    .sort_values(
        by="chart_observations",
        ascending=False
    )
    .reset_index(drop=True)
)

strong_market_countries = set(
    strong_market_summary_df["country"]
)

total_strong_markets = len(
    strong_market_countries
)

if total_strong_markets <= 0:
    raise RuntimeError(
        "No strong markets were identified."
    )

print("Identified PMIP strong markets")
print("=" * 80)

display(
    strong_market_summary_df[
        [
            "country",
            "chart_observations",
            "unique_tracks",
            "market_activity_rank",
            "market_activity_percentile",
        ]
    ]
)

print()
print(
    f"Strong markets identified: {total_strong_markets}"
)

# ============================================================
# Release-to-market presence
#
# Use chart-history track_id to determine the set of countries
# represented for each historical track.
# ============================================================

release_market_presence_df = (
    market_activity_source[
        ["track_id", "country"]
    ]
    .drop_duplicates()
)

release_strong_market_counts_df = (
    release_market_presence_df[
        release_market_presence_df[
            "country"
        ].isin(
            strong_market_countries
        )
    ]
    .groupby("track_id")["country"]
    .nunique()
    .rename("strong_markets_represented")
    .reset_index()
)

# ============================================================
# Build historical geographic signature
#
# Reuse the same chart-history signature approach from 6.4 so
# release-level rows can reconnect to historical track IDs.
# ============================================================

geo_history_signature_df = (
    market_activity_source
    .groupby("track_id")
    .agg(
        best_chart_position_signature=("position", "min"),
        average_chart_position_signature=("position", "mean"),
        chart_observations_signature=("position", "size"),
        first_chart_date_signature=("date", "min"),
        last_chart_date_signature=("date", "max"),
        countries_charted_signature=("country", "nunique"),
    )
    .reset_index()
)

geo_history_signature_df["date"] = pd.to_datetime(
    geo_history_signature_df[
        "first_chart_date_signature"
    ],
    errors="coerce"
)

geo_history_signature_df["first_chart_date_key"] = (
    pd.to_datetime(
        geo_history_signature_df[
            "first_chart_date_signature"
        ],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

geo_history_signature_df["last_chart_date_key"] = (
    pd.to_datetime(
        geo_history_signature_df[
            "last_chart_date_signature"
        ],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

geo_history_signature_df["best_chart_position_key"] = (
    pd.to_numeric(
        geo_history_signature_df[
            "best_chart_position_signature"
        ],
        errors="coerce"
    )
    .round(6)
)

geo_history_signature_df["average_chart_position_key"] = (
    pd.to_numeric(
        geo_history_signature_df[
            "average_chart_position_signature"
        ],
        errors="coerce"
    )
    .round(6)
)

geo_history_signature_df["chart_observations_key"] = (
    pd.to_numeric(
        geo_history_signature_df[
            "chart_observations_signature"
        ],
        errors="coerce"
    )
    .round(6)
)

geo_history_signature_df["countries_charted_key"] = (
    pd.to_numeric(
        geo_history_signature_df[
            "countries_charted_signature"
        ],
        errors="coerce"
    )
    .round(6)
)

strong_market_signature_key_cols = [
    "first_chart_date_key",
    "last_chart_date_key",
    "best_chart_position_key",
    "average_chart_position_key",
    "chart_observations_key",
    "countries_charted_key",
]

# ============================================================
# Attach strong-market counts to historical signatures
# ============================================================

geo_history_signature_df = (
    geo_history_signature_df
    .merge(
        release_strong_market_counts_df,
        on="track_id",
        how="left"
    )
)

geo_history_signature_df[
    "strong_markets_represented"
] = (
    geo_history_signature_df[
        "strong_markets_represented"
    ]
    .fillna(0)
)

# ============================================================
# Signature ambiguity
# ============================================================

strong_signature_counts_df = (
    geo_history_signature_df
    .groupby(
        strong_market_signature_key_cols,
        dropna=False
    )
    .size()
    .rename(
        "strong_market_signature_match_count"
    )
    .reset_index()
)

geo_history_signature_unique_df = (
    geo_history_signature_df
    .sort_values(
        by="track_id"
    )
    .drop_duplicates(
        subset=strong_market_signature_key_cols,
        keep="first"
    )
    .merge(
        strong_signature_counts_df,
        on=strong_market_signature_key_cols,
        how="left"
    )
)

geo_history_signature_unique_df = (
    geo_history_signature_unique_df
    .rename(
        columns={
            "track_id":
            "strong_market_track_id"
        }
    )
)

# ============================================================
# Prepare release-level matching keys
# ============================================================

strong_market_df["first_chart_date_key"] = (
    pd.to_datetime(
        strong_market_df["first_chart_date"],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

strong_market_df["last_chart_date_key"] = (
    pd.to_datetime(
        strong_market_df["last_chart_date"],
        errors="coerce"
    )
    .dt.strftime("%Y-%m-%d")
)

strong_market_df["best_chart_position_key"] = (
    pd.to_numeric(
        strong_market_df[
            "best_chart_position"
        ],
        errors="coerce"
    )
    .round(6)
)

strong_market_df["average_chart_position_key"] = (
    pd.to_numeric(
        strong_market_df[
            "average_chart_position"
        ],
        errors="coerce"
    )
    .round(6)
)

strong_market_df["chart_observations_key"] = (
    pd.to_numeric(
        strong_market_df[
            "chart_observations"
        ],
        errors="coerce"
    )
    .round(6)
)

strong_market_df["countries_charted_key"] = (
    pd.to_numeric(
        strong_market_df[
            "countries_charted"
        ],
        errors="coerce"
    )
    .round(6)
)

# ============================================================
# Merge strong-market evidence
# ============================================================

strong_market_merge_cols = (
    strong_market_signature_key_cols
    + [
        "strong_market_track_id",
        "strong_markets_represented",
        "strong_market_signature_match_count",
    ]
)

strong_market_df = strong_market_df.merge(
    geo_history_signature_unique_df[
        strong_market_merge_cols
    ],
    on=strong_market_signature_key_cols,
    how="left"
)

# ============================================================
# Match diagnostics
# ============================================================

strong_market_df["strong_market_match_available"] = (
    strong_market_df[
        "strong_market_track_id"
    ].notna()
)

strong_market_df["strong_market_match_ambiguous"] = (
    strong_market_df[
        "strong_market_signature_match_count"
    ]
    .fillna(0)
    .gt(1)
)

# ============================================================
# Strong-market coverage
# ============================================================

strong_market_df["strong_market_presence_pct"] = np.where(
    strong_market_df[
        "strong_market_match_available"
    ],
    (
        strong_market_df[
            "strong_markets_represented"
        ]
        / total_strong_markets
        * 100
    ),
    np.nan,
)

strong_market_df[
    "strong_market_presence_pct"
] = (
    strong_market_df[
        "strong_market_presence_pct"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# Strong-market score
# ============================================================

strong_market_df["strong_market_presence_score"] = np.nan

valid_strong_market_mask = (
    strong_market_df[
        "strong_market_presence_pct"
    ].notna()
)

if valid_strong_market_mask.any():

    strong_market_df.loc[
        valid_strong_market_mask,
        "strong_market_presence_score"
    ] = (
        strong_market_df.loc[
            valid_strong_market_mask,
            "strong_market_presence_pct"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Strong-market classes
# ============================================================

strong_market_df["strong_market_presence_class"] = (
    "No Strong-Market Evidence"
)

strong_market_df.loc[
    valid_strong_market_mask
    & (
        strong_market_df[
            "strong_markets_represented"
        ] == 0
    ),
    "strong_market_presence_class"
] = "No Strong-Market Presence"

strong_market_df.loc[
    valid_strong_market_mask
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] > 0
    )
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] < 25
    ),
    "strong_market_presence_class"
] = "Limited Strong-Market Presence"

strong_market_df.loc[
    valid_strong_market_mask
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] >= 25
    )
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] < 50
    ),
    "strong_market_presence_class"
] = "Moderate Strong-Market Presence"

strong_market_df.loc[
    valid_strong_market_mask
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] >= 50
    )
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] < 75
    ),
    "strong_market_presence_class"
] = "Broad Strong-Market Presence"

strong_market_df.loc[
    valid_strong_market_mask
    & (
        strong_market_df[
            "strong_market_presence_pct"
        ] >= 75
    ),
    "strong_market_presence_class"
] = "Very Broad Strong-Market Presence"

# ============================================================
# Summary
# ============================================================

strong_market_presence_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                strong_market_df
            ),
        },
        {
            "Metric": "Strong markets identified",
            "Observed Value": total_strong_markets,
        },
        {
            "Metric": "Releases with strong-market evidence",
            "Observed Value": int(
                valid_strong_market_mask.sum()
            ),
        },
        {
            "Metric": "Releases without strong-market evidence",
            "Observed Value": int(
                (~valid_strong_market_mask).sum()
            ),
        },
        {
            "Metric": "Ambiguous strong-market signatures",
            "Observed Value": int(
                strong_market_df[
                    "strong_market_match_ambiguous"
                ].sum()
            ),
        },
        {
            "Metric": "Median strong markets represented",
            "Observed Value": (
                strong_market_df.loc[
                    valid_strong_market_mask,
                    "strong_markets_represented"
                ].median()
            ),
        },
        {
            "Metric": "Mean strong markets represented",
            "Observed Value": (
                strong_market_df.loc[
                    valid_strong_market_mask,
                    "strong_markets_represented"
                ].mean()
            ),
        },
        {
            "Metric": "Median strong-market presence (%)",
            "Observed Value": (
                strong_market_df.loc[
                    valid_strong_market_mask,
                    "strong_market_presence_pct"
                ].median()
            ),
        },
        {
            "Metric": "Maximum strong-market presence (%)",
            "Observed Value": (
                strong_market_df.loc[
                    valid_strong_market_mask,
                    "strong_market_presence_pct"
                ].max()
            ),
        },
    ]
)

print()
print("Assessing PMIP strong-market presence")
print("=" * 80)

display(
    strong_market_presence_summary_df
)

# ============================================================
# Strong-market class distribution
# ============================================================

strong_market_class_distribution_df = (
    strong_market_df[
        "strong_market_presence_class"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Strong-Market Presence Class"
    )
    .reset_index(
        name="Releases"
    )
)

strong_market_class_distribution_df[
    "Share (%)"
] = (
    strong_market_class_distribution_df[
        "Releases"
    ]
    / len(strong_market_df)
    * 100
)

print()
print("Strong-market presence class distribution")
print("=" * 80)

display(
    strong_market_class_distribution_df
)

# ============================================================
# Releases with strongest strong-market presence
# ============================================================

top_strong_market_presence_df = (
    strong_market_df[
        valid_strong_market_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "countries_charted",
            "strong_markets_represented",
            "strong_market_presence_pct",
            "strong_market_presence_score",
            "strong_market_presence_class",
        ]
    ]
    .sort_values(
        by=[
            "strong_market_presence_pct",
            "countries_charted",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest established-market presence")
print("=" * 80)

display(
    top_strong_market_presence_df
)

# ============================================================
# Validation
# ============================================================

strong_market_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.2 completed",
            "Passed": section_7_2_complete,
        },
        {
            "Validation Area": "Strong markets identified",
            "Passed": total_strong_markets > 0,
        },
        {
            "Validation Area": "Historical market matching prepared",
            "Passed": (
                "strong_market_match_available"
                in strong_market_df.columns
            ),
        },
        {
            "Validation Area": "Strong-market evidence available",
            "Passed": valid_strong_market_mask.any(),
        },
        {
            "Validation Area": "Strong-market percentage calculated",
            "Passed": (
                "strong_market_presence_pct"
                in strong_market_df.columns
            ),
        },
        {
            "Validation Area": "Strong-market score calculated",
            "Passed": (
                "strong_market_presence_score"
                in strong_market_df.columns
            ),
        },
        {
            "Validation Area": "Strong-market classes assigned",
            "Passed": strong_market_df[
                "strong_market_presence_class"
            ].notna().all(),
        },
    ]
)

print()
print("Strong-market presence validation")
print("=" * 80)

display(
    strong_market_validation_df
)

# ============================================================
# Completion
# ============================================================

section_7_3_complete = bool(
    strong_market_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 7.3 completion status: "
    f"{section_7_3_complete}"
)

if not section_7_3_complete:
    raise RuntimeError(
        "Strong-Market Presence analysis did not complete successfully."
    )
```

### Interpretation

The PMIP Strong-Market Presence assessment completed successfully.

A total of 20 strong markets were identified from the historical chart dataset. These markets represent the top quartile of countries based on the overall volume of historical chart observations available within PMIP.

This definition is data-driven. Therefore, `Strong Market` in this section refers to markets with the highest chart activity within the available dataset rather than an external ranking of the world's largest music industries.

Strong-market evidence could be matched reliably for 210 of the 4,593 validated releases. The remaining 4,383 releases do not contain sufficiently reliable release-to-history matching evidence for this analysis.

Among the matched releases, the median release is represented in 1 of the 20 identified strong markets, corresponding to a median Strong-Market Presence of 5%.

The mean number of strong markets represented is approximately 1.95, indicating that although many matched releases have limited strong-market coverage, a smaller number of internationally distributed releases have substantially broader presence.

The maximum Strong-Market Presence is 100%, meaning that some releases are represented across all 20 identified strong markets.

The class distribution shows that most matched releases have either Limited Strong-Market Presence or no presence within the strong-market set. Only a relatively small number reach Broad or Very Broad Strong-Market Presence.

The highest-performing releases demonstrate a clear difference between general geographic breadth and strong-market coverage. Some releases appear across many countries while also reaching a large proportion of the most active markets in the PMIP dataset.

A total of 18 ambiguous historical signatures were detected. These cases indicate that some release-history relationships cannot be distinguished with complete certainty using the available matching signature and should therefore be interpreted cautiously.

The relatively low analytical coverage of 210 releases is also important. Strong-Market Presence should not be treated as representative of the full release population and should receive coverage-aware treatment when it is later combined into the Geographic Reach Indicator.

Section 7.3 completed successfully and adds a market-quality dimension to the PMIP Geographic Release Reach framework.

## 7.4 Emerging-Market Presence

### Purpose

This subsection evaluates whether each validated release is represented within lower-activity geographic markets that may indicate developing or less-established areas of chart presence within the PMIP dataset.

While Section 7.3 focused on the most active markets, Emerging-Market Presence focuses on markets with comparatively lower historical chart activity.

For this analysis, emerging markets are defined using the lower quartile of market activity within the PMIP historical chart dataset.

The assessment will examine:

- how many emerging markets each release reaches;
- what proportion of the emerging-market set is represented;
- the relative strength of that presence;
- releases showing broad reach beyond the strongest established markets.

This provides a complementary geographic perspective by identifying releases whose reach extends into less-active or potentially developing markets.

```python
# ============================================================
# Section 7.4 - Emerging-Market Presence
# ============================================================

if not section_7_3_complete:
    raise RuntimeError(
        "Section 7.3 must be completed before analysing emerging-market presence."
    )

# ============================================================
# Working source
# ============================================================

emerging_market_df = strong_market_df.copy()

# ============================================================
# Reuse market activity summary from Section 7.3
# ============================================================

if "market_activity_summary_df" not in globals():
    raise RuntimeError(
        "market_activity_summary_df is not available. Run Section 7.3 first."
    )

# ============================================================
# Define emerging markets
#
# Bottom quartile of markets by historical chart activity.
# ============================================================

emerging_market_threshold = (
    market_activity_summary_df[
        "chart_observations"
    ]
    .quantile(0.25)
)

emerging_market_summary_df = (
    market_activity_summary_df[
        market_activity_summary_df[
            "chart_observations"
        ] <= emerging_market_threshold
    ]
    .copy()
    .sort_values(
        by="chart_observations",
        ascending=True
    )
    .reset_index(drop=True)
)

emerging_market_countries = set(
    emerging_market_summary_df["country"]
)

total_emerging_markets = len(
    emerging_market_countries
)

if total_emerging_markets <= 0:
    raise RuntimeError(
        "No emerging markets were identified."
    )

print("Identified PMIP emerging markets")
print("=" * 80)

display(
    emerging_market_summary_df[
        [
            "country",
            "chart_observations",
            "unique_tracks",
            "market_activity_rank",
            "market_activity_percentile",
        ]
    ]
)

print()
print(
    f"Emerging markets identified: "
    f"{total_emerging_markets}"
)

# ============================================================
# Release-to-market presence
#
# Reuse release_market_presence_df from Section 7.3
# ============================================================

if "release_market_presence_df" not in globals():
    raise RuntimeError(
        "release_market_presence_df is not available. Run Section 7.3 first."
    )

release_emerging_market_counts_df = (
    release_market_presence_df[
        release_market_presence_df[
            "country"
        ].isin(
            emerging_market_countries
        )
    ]
    .groupby("track_id")["country"]
    .nunique()
    .rename("emerging_markets_represented")
    .reset_index()
)

# ============================================================
# Reuse geographic historical signature
# ============================================================

if "geo_history_signature_df" not in globals():
    raise RuntimeError(
        "geo_history_signature_df is not available. Run Section 7.3 first."
    )

emerging_geo_signature_df = (
    geo_history_signature_df.copy()
)

# Remove any existing emerging-market column if rerunning
if (
    "emerging_markets_represented"
    in emerging_geo_signature_df.columns
):
    emerging_geo_signature_df = (
        emerging_geo_signature_df.drop(
            columns=[
                "emerging_markets_represented"
            ]
        )
    )

# ============================================================
# Attach emerging-market counts
# ============================================================

emerging_geo_signature_df = (
    emerging_geo_signature_df
    .merge(
        release_emerging_market_counts_df,
        on="track_id",
        how="left"
    )
)

emerging_geo_signature_df[
    "emerging_markets_represented"
] = (
    emerging_geo_signature_df[
        "emerging_markets_represented"
    ]
    .fillna(0)
)

# ============================================================
# Signature ambiguity
# ============================================================

emerging_signature_counts_df = (
    emerging_geo_signature_df
    .groupby(
        strong_market_signature_key_cols,
        dropna=False
    )
    .size()
    .rename(
        "emerging_market_signature_match_count"
    )
    .reset_index()
)

emerging_geo_signature_unique_df = (
    emerging_geo_signature_df
    .sort_values(
        by="track_id"
    )
    .drop_duplicates(
        subset=strong_market_signature_key_cols,
        keep="first"
    )
    .merge(
        emerging_signature_counts_df,
        on=strong_market_signature_key_cols,
        how="left"
    )
)

emerging_geo_signature_unique_df = (
    emerging_geo_signature_unique_df
    .rename(
        columns={
            "track_id":
            "emerging_market_track_id"
        }
    )
)

# ============================================================
# Merge emerging-market evidence
# ============================================================

emerging_market_merge_cols = (
    strong_market_signature_key_cols
    + [
        "emerging_market_track_id",
        "emerging_markets_represented",
        "emerging_market_signature_match_count",
    ]
)

emerging_market_df = (
    emerging_market_df.merge(
        emerging_geo_signature_unique_df[
            emerging_market_merge_cols
        ],
        on=strong_market_signature_key_cols,
        how="left"
    )
)

# ============================================================
# Match diagnostics
# ============================================================

emerging_market_df[
    "emerging_market_match_available"
] = (
    emerging_market_df[
        "emerging_market_track_id"
    ].notna()
)

emerging_market_df[
    "emerging_market_match_ambiguous"
] = (
    emerging_market_df[
        "emerging_market_signature_match_count"
    ]
    .fillna(0)
    .gt(1)
)

# ============================================================
# Emerging-market coverage
# ============================================================

emerging_market_df[
    "emerging_market_presence_pct"
] = np.where(
    emerging_market_df[
        "emerging_market_match_available"
    ],
    (
        emerging_market_df[
            "emerging_markets_represented"
        ]
        / total_emerging_markets
        * 100
    ),
    np.nan,
)

emerging_market_df[
    "emerging_market_presence_pct"
] = (
    emerging_market_df[
        "emerging_market_presence_pct"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# Emerging-market score
# ============================================================

emerging_market_df[
    "emerging_market_presence_score"
] = np.nan

valid_emerging_market_mask = (
    emerging_market_df[
        "emerging_market_presence_pct"
    ].notna()
)

if valid_emerging_market_mask.any():

    emerging_market_df.loc[
        valid_emerging_market_mask,
        "emerging_market_presence_score"
    ] = (
        emerging_market_df.loc[
            valid_emerging_market_mask,
            "emerging_market_presence_pct"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Emerging-market classes
# ============================================================

emerging_market_df[
    "emerging_market_presence_class"
] = (
    "No Emerging-Market Evidence"
)

emerging_market_df.loc[
    valid_emerging_market_mask
    & (
        emerging_market_df[
            "emerging_markets_represented"
        ] == 0
    ),
    "emerging_market_presence_class"
] = "No Emerging-Market Presence"

emerging_market_df.loc[
    valid_emerging_market_mask
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] > 0
    )
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] < 25
    ),
    "emerging_market_presence_class"
] = "Limited Emerging-Market Presence"

emerging_market_df.loc[
    valid_emerging_market_mask
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] >= 25
    )
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] < 50
    ),
    "emerging_market_presence_class"
] = "Moderate Emerging-Market Presence"

emerging_market_df.loc[
    valid_emerging_market_mask
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] >= 50
    )
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] < 75
    ),
    "emerging_market_presence_class"
] = "Broad Emerging-Market Presence"

emerging_market_df.loc[
    valid_emerging_market_mask
    & (
        emerging_market_df[
            "emerging_market_presence_pct"
        ] >= 75
    ),
    "emerging_market_presence_class"
] = "Very Broad Emerging-Market Presence"

# ============================================================
# Summary
# ============================================================

emerging_market_presence_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                emerging_market_df
            ),
        },
        {
            "Metric": "Emerging markets identified",
            "Observed Value": (
                total_emerging_markets
            ),
        },
        {
            "Metric": "Releases with emerging-market evidence",
            "Observed Value": int(
                valid_emerging_market_mask.sum()
            ),
        },
        {
            "Metric": "Releases without emerging-market evidence",
            "Observed Value": int(
                (
                    ~valid_emerging_market_mask
                ).sum()
            ),
        },
        {
            "Metric": "Ambiguous emerging-market signatures",
            "Observed Value": int(
                emerging_market_df[
                    "emerging_market_match_ambiguous"
                ].sum()
            ),
        },
        {
            "Metric": "Median emerging markets represented",
            "Observed Value": (
                emerging_market_df.loc[
                    valid_emerging_market_mask,
                    "emerging_markets_represented"
                ].median()
            ),
        },
        {
            "Metric": "Mean emerging markets represented",
            "Observed Value": (
                emerging_market_df.loc[
                    valid_emerging_market_mask,
                    "emerging_markets_represented"
                ].mean()
            ),
        },
        {
            "Metric": "Median emerging-market presence (%)",
            "Observed Value": (
                emerging_market_df.loc[
                    valid_emerging_market_mask,
                    "emerging_market_presence_pct"
                ].median()
            ),
        },
        {
            "Metric": "Maximum emerging-market presence (%)",
            "Observed Value": (
                emerging_market_df.loc[
                    valid_emerging_market_mask,
                    "emerging_market_presence_pct"
                ].max()
            ),
        },
    ]
)

print()
print("Assessing PMIP emerging-market presence")
print("=" * 80)

display(
    emerging_market_presence_summary_df
)

# ============================================================
# Emerging-market class distribution
# ============================================================

emerging_market_class_distribution_df = (
    emerging_market_df[
        "emerging_market_presence_class"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "Emerging-Market Presence Class"
    )
    .reset_index(
        name="Releases"
    )
)

emerging_market_class_distribution_df[
    "Share (%)"
] = (
    emerging_market_class_distribution_df[
        "Releases"
    ]
    / len(emerging_market_df)
    * 100
)

print()
print("Emerging-market presence class distribution")
print("=" * 80)

display(
    emerging_market_class_distribution_df
)

# ============================================================
# Releases with strongest emerging-market presence
# ============================================================

top_emerging_market_presence_df = (
    emerging_market_df[
        valid_emerging_market_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "countries_charted",
            "emerging_markets_represented",
            "emerging_market_presence_pct",
            "emerging_market_presence_score",
            "emerging_market_presence_class",
        ]
    ]
    .sort_values(
        by=[
            "emerging_market_presence_pct",
            "countries_charted",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest emerging-market presence")
print("=" * 80)

display(
    top_emerging_market_presence_df
)

# ============================================================
# Validation
# ============================================================

emerging_market_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.3 completed",
            "Passed": section_7_3_complete,
        },
        {
            "Validation Area": "Emerging markets identified",
            "Passed": (
                total_emerging_markets > 0
            ),
        },
        {
            "Validation Area": "Historical emerging-market matching prepared",
            "Passed": (
                "emerging_market_match_available"
                in emerging_market_df.columns
            ),
        },
        {
            "Validation Area": "Emerging-market evidence available",
            "Passed": (
                valid_emerging_market_mask.any()
            ),
        },
        {
            "Validation Area": "Emerging-market percentage calculated",
            "Passed": (
                "emerging_market_presence_pct"
                in emerging_market_df.columns
            ),
        },
        {
            "Validation Area": "Emerging-market score calculated",
            "Passed": (
                "emerging_market_presence_score"
                in emerging_market_df.columns
            ),
        },
        {
            "Validation Area": "Emerging-market classes assigned",
            "Passed": emerging_market_df[
                "emerging_market_presence_class"
            ].notna().all(),
        },
    ]
)

print()
print("Emerging-market presence validation")
print("=" * 80)

display(
    emerging_market_validation_df
)

# ============================================================
# Completion
# ============================================================

section_7_4_complete = bool(
    emerging_market_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 7.4 completion status: "
    f"{section_7_4_complete}"
)

if not section_7_4_complete:
    raise RuntimeError(
        "Emerging-Market Presence analysis did not complete successfully."
    )
```

The PMIP Emerging-Market Presence assessment completed successfully.

This definition is data-driven. Therefore, `Emerging Market` in this section refers to lower-activity markets within the PMIP dataset rather than an external economic or music-industry classification.

Emerging-market evidence could be matched reliably for 210 of the 4,593 validated releases, while 4,383 releases do not contain sufficiently reliable historical matching evidence for this analysis.

Among the matched releases, the median number of emerging markets represented is 0 and the median Emerging-Market Presence is therefore 0%.

The mean number of emerging markets represented is approximately 0.99, indicating that most matched releases have little or no presence across the lower-activity market group, while a smaller number of releases reach several emerging markets.

The maximum observed Emerging-Market Presence is 80%, meaning that the strongest release in this component is represented across 16 of the 20 identified emerging markets.

The class distribution supports this pattern. Most matched releases fall within either `No Emerging-Market Presence` or `Limited Emerging-Market Presence`, while only a small number achieve Moderate, Broad or Very Broad Emerging-Market Presence.

This suggests that broad coverage of lower-activity PMIP markets is relatively uncommon within the matched release population.

The strongest releases demonstrate that international reach can extend beyond the highest-activity markets. These releases show geographic diversification across both established and lower-activity market groups.

The limited analytical coverage of 210 releases is also important. Emerging-Market Presence should therefore not be treated as representative of the complete PMIP release population and should be handled using coverage-aware weighting in the final Geographic Reach Indicator.

Section 7.4 completed successfully and provides the emerging-market diversification component required for the PMIP Geographic Release Reach framework.

## 7.5 Geographic Reach Indicator

### Purpose

This subsection combines the major geographic components into a single coverage-aware PMIP Geographic Reach Indicator.

The indicator brings together:

- countries represented;
- market breadth;
- strong-market presence;
- emerging-market presence.

Because the geographic components have different levels of evidence coverage, the final score uses coverage-aware weighting.

Countries represented and market breadth are available for a much larger proportion of releases, while strong-market and emerging-market presence are available only for the smaller subset that could be matched reliably to detailed historical country-level records.

Missing strong-market or emerging-market evidence is therefore not treated as weak geographic performance. Instead, the weights of the available components are re-normalised for each release.

The resulting indicator is designed to summarise geographic reach while also preserving transparency about how much evidence supports each score.

```python
# ============================================================
# Section 7.5 - Geographic Reach Indicator
# ============================================================

if not section_7_4_complete:
    raise RuntimeError(
        "Section 7.4 must be completed before constructing "
        "the Geographic Reach Indicator."
    )

# ============================================================
# Working source
# ============================================================

geographic_indicator_df = emerging_market_df.copy()

# ============================================================
# Prepare geographic components
# ============================================================

geographic_indicator_df[
    "countries_represented_component"
] = pd.to_numeric(
    geographic_indicator_df[
        "countries_represented_score"
    ],
    errors="coerce"
)

geographic_indicator_df[
    "market_breadth_component"
] = pd.to_numeric(
    geographic_indicator_df[
        "market_breadth_score"
    ],
    errors="coerce"
)

geographic_indicator_df[
    "strong_market_component"
] = pd.to_numeric(
    geographic_indicator_df[
        "strong_market_presence_score"
    ],
    errors="coerce"
)

geographic_indicator_df[
    "emerging_market_component"
] = pd.to_numeric(
    geographic_indicator_df[
        "emerging_market_presence_score"
    ],
    errors="coerce"
)

# ============================================================
# Base weights
#
# Countries represented and market breadth form the core
# geographic reach evidence.
#
# Strong-market and emerging-market presence add market-type
# context where reliable detailed matching is available.
# ============================================================

geographic_component_weights = {
    "countries_represented_component": 0.30,
    "market_breadth_component": 0.30,
    "strong_market_component": 0.25,
    "emerging_market_component": 0.15,
}

geographic_component_columns = list(
    geographic_component_weights.keys()
)

# ============================================================
# Coverage-aware weighted score
# ============================================================

geographic_weighted_sum = pd.Series(
    0.0,
    index=geographic_indicator_df.index
)

geographic_available_weight_sum = pd.Series(
    0.0,
    index=geographic_indicator_df.index
)

for column, weight in geographic_component_weights.items():

    available_mask = (
        geographic_indicator_df[
            column
        ].notna()
    )

    geographic_weighted_sum.loc[
        available_mask
    ] += (
        geographic_indicator_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    geographic_available_weight_sum.loc[
        available_mask
    ] += weight

geographic_indicator_df[
    "geographic_reach_score"
] = np.where(
    geographic_available_weight_sum > 0,
    (
        geographic_weighted_sum
        / geographic_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Evidence coverage
# ============================================================

geographic_indicator_df[
    "geographic_components_available"
] = (
    geographic_indicator_df[
        geographic_component_columns
    ]
    .notna()
    .sum(axis=1)
)

geographic_indicator_df[
    "geographic_evidence_coverage_pct"
] = (
    geographic_indicator_df[
        "geographic_components_available"
    ]
    / len(
        geographic_component_columns
    )
    * 100
)

# ============================================================
# Evidence strength
# ============================================================

geographic_indicator_df[
    "geographic_evidence_strength"
] = np.select(
    [
        geographic_indicator_df[
            "geographic_components_available"
        ] == 4,

        geographic_indicator_df[
            "geographic_components_available"
        ] == 3,

        geographic_indicator_df[
            "geographic_components_available"
        ] == 2,

        geographic_indicator_df[
            "geographic_components_available"
        ] == 1,
    ],
    [
        "Very Strong Evidence",
        "Strong Evidence",
        "Moderate Evidence",
        "Limited Evidence",
    ],
    default="No Evidence",
)

# ============================================================
# Valid geographic score mask
# ============================================================

valid_geographic_reach_mask = (
    geographic_indicator_df[
        "geographic_reach_score"
    ].notna()
)

# ============================================================
# Geographic reach percentile
# ============================================================

geographic_indicator_df[
    "geographic_reach_percentile"
] = np.nan

if valid_geographic_reach_mask.any():

    geographic_indicator_df.loc[
        valid_geographic_reach_mask,
        "geographic_reach_percentile"
    ] = (
        geographic_indicator_df.loc[
            valid_geographic_reach_mask,
            "geographic_reach_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Geographic reach rank
# ============================================================

geographic_indicator_df[
    "geographic_reach_rank"
] = (
    geographic_indicator_df[
        "geographic_reach_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Geographic reach classes
# ============================================================

geographic_indicator_df[
    "geographic_reach_class"
] = "No Geographic Reach Evidence"

if valid_geographic_reach_mask.any():

    q25 = geographic_indicator_df.loc[
        valid_geographic_reach_mask,
        "geographic_reach_score"
    ].quantile(0.25)

    q50 = geographic_indicator_df.loc[
        valid_geographic_reach_mask,
        "geographic_reach_score"
    ].quantile(0.50)

    q75 = geographic_indicator_df.loc[
        valid_geographic_reach_mask,
        "geographic_reach_score"
    ].quantile(0.75)

    q90 = geographic_indicator_df.loc[
        valid_geographic_reach_mask,
        "geographic_reach_score"
    ].quantile(0.90)

    geographic_indicator_df.loc[
        valid_geographic_reach_mask
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] <= q25
        ),
        "geographic_reach_class"
    ] = "Limited Geographic Reach"

    geographic_indicator_df.loc[
        valid_geographic_reach_mask
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] > q25
        )
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] <= q50
        ),
        "geographic_reach_class"
    ] = "Moderate Geographic Reach"

    geographic_indicator_df.loc[
        valid_geographic_reach_mask
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] > q50
        )
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] <= q75
        ),
        "geographic_reach_class"
    ] = "Broad Geographic Reach"

    geographic_indicator_df.loc[
        valid_geographic_reach_mask
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] > q75
        )
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] <= q90
        ),
        "geographic_reach_class"
    ] = "Very Broad Geographic Reach"

    geographic_indicator_df.loc[
        valid_geographic_reach_mask
        & (
            geographic_indicator_df[
                "geographic_reach_score"
            ] > q90
        ),
        "geographic_reach_class"
    ] = "Exceptional Geographic Reach"

# ============================================================
# Summary
# ============================================================

geographic_reach_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                geographic_indicator_df
            ),
        },
        {
            "Metric": "Releases with geographic-reach score",
            "Observed Value": int(
                valid_geographic_reach_mask.sum()
            ),
        },
        {
            "Metric": "Releases without geographic-reach score",
            "Observed Value": int(
                (
                    ~valid_geographic_reach_mask
                ).sum()
            ),
        },
        {
            "Metric": "Median geographic-reach score",
            "Observed Value": (
                geographic_indicator_df[
                    "geographic_reach_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean geographic-reach score",
            "Observed Value": (
                geographic_indicator_df[
                    "geographic_reach_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum geographic-reach score",
            "Observed Value": (
                geographic_indicator_df[
                    "geographic_reach_score"
                ].max()
            ),
        },
        {
            "Metric": "Median geographic evidence coverage (%)",
            "Observed Value": (
                geographic_indicator_df[
                    "geographic_evidence_coverage_pct"
                ].median()
            ),
        },
        {
            "Metric": "Median evidence coverage among scored releases (%)",
            "Observed Value": (
                geographic_indicator_df.loc[
                    valid_geographic_reach_mask,
                    "geographic_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Constructing PMIP Geographic Reach Indicator")
print("=" * 80)

display(
    geographic_reach_summary_df
)

# ============================================================
# Component coverage
# ============================================================

geographic_component_coverage_rows = []

for column in geographic_component_columns:

    geographic_component_coverage_rows.append(
        {
            "Component": column,
            "Releases Available": int(
                geographic_indicator_df[
                    column
                ].notna().sum()
            ),
            "Coverage (%)": round(
                geographic_indicator_df[
                    column
                ].notna().mean()
                * 100,
                6
            ),
            "Base Weight": (
                geographic_component_weights[
                    column
                ]
            ),
        }
    )

geographic_component_coverage_df = pd.DataFrame(
    geographic_component_coverage_rows
)

print()
print("Geographic indicator component coverage")
print("=" * 80)

display(
    geographic_component_coverage_df
)

# ============================================================
# Evidence-strength distribution
# ============================================================

geographic_evidence_strength_distribution_df = (
    geographic_indicator_df[
        "geographic_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Evidence Strength"
    )
    .reset_index(
        name="Releases"
    )
)

geographic_evidence_strength_distribution_df[
    "Share (%)"
] = (
    geographic_evidence_strength_distribution_df[
        "Releases"
    ]
    / len(
        geographic_indicator_df
    )
    * 100
)

print()
print("Geographic evidence-strength distribution")
print("=" * 80)

display(
    geographic_evidence_strength_distribution_df
)

# ============================================================
# Geographic-reach class distribution
# ============================================================

geographic_reach_class_distribution_df = (
    geographic_indicator_df[
        "geographic_reach_class"
    ]
    .value_counts()
    .rename_axis(
        "Geographic Reach Class"
    )
    .reset_index(
        name="Releases"
    )
)

geographic_reach_class_distribution_df[
    "Share (%)"
] = (
    geographic_reach_class_distribution_df[
        "Releases"
    ]
    / len(
        geographic_indicator_df
    )
    * 100
)

print()
print("Geographic-reach class distribution")
print("=" * 80)

display(
    geographic_reach_class_distribution_df
)

# ============================================================
# Highest geographic-reach releases
# ============================================================

top_geographic_reach_df = (
    geographic_indicator_df[
        [
            "release_id",
            "Track",
            "Artist",
            "countries_charted",
            "countries_represented_component",
            "market_breadth_component",
            "strong_market_component",
            "emerging_market_component",
            "geographic_components_available",
            "geographic_evidence_strength",
            "geographic_reach_score",
            "geographic_reach_percentile",
            "geographic_reach_rank",
            "geographic_reach_class",
        ]
    ]
    .sort_values(
        by="geographic_reach_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP geographic-reach releases")
print("=" * 80)

display(
    top_geographic_reach_df
)

# ============================================================
# Validation
# ============================================================

geographic_reach_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.4 completed",
            "Passed": section_7_4_complete,
        },
        {
            "Validation Area": "Countries-represented component available",
            "Passed": geographic_indicator_df[
                "countries_represented_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Market-breadth component available",
            "Passed": geographic_indicator_df[
                "market_breadth_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Strong-market component prepared",
            "Passed": (
                "strong_market_component"
                in geographic_indicator_df.columns
            ),
        },
        {
            "Validation Area": "Emerging-market component prepared",
            "Passed": (
                "emerging_market_component"
                in geographic_indicator_df.columns
            ),
        },
        {
            "Validation Area": "Coverage-aware geographic weighting prepared",
            "Passed": (
                geographic_available_weight_sum > 0
            ).any(),
        },
        {
            "Validation Area": "Geographic-reach score calculated",
            "Passed": geographic_indicator_df[
                "geographic_reach_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Geographic evidence coverage calculated",
            "Passed": geographic_indicator_df[
                "geographic_evidence_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Geographic-reach classes assigned",
            "Passed": geographic_indicator_df[
                "geographic_reach_class"
            ].notna().all(),
        },
    ]
)

print()
print("Geographic Reach Indicator validation")
print("=" * 80)

display(
    geographic_reach_validation_df
)

# ============================================================
# Completion
# ============================================================

section_7_5_complete = bool(
    geographic_reach_validation_df[
        "Passed"
    ].all()
)

section_7_complete = bool(
    section_7_1_complete
    and section_7_2_complete
    and section_7_3_complete
    and section_7_4_complete
    and section_7_5_complete
)

print()
print(
    f"Section 7.5 completion status: "
    f"{section_7_5_complete}"
)

print(
    f"Section 7 overall completion status: "
    f"{section_7_complete}"
)

if not section_7_complete:
    raise RuntimeError(
        "Section 7 Geographic Release Reach "
        "did not complete successfully."
    )
```

# 8. Release Timing and Historical Context

This section evaluates release performance in relation to time.

The aim is to understand how release age, observation period and historical evidence affect the interpretation of release-level performance.

The analysis will focus on:

- release dates;
- age of each release;
- available observation window;
- differences between current and historical performance;
- temporal comparability across releases.

This is important because releases should not always be compared directly without considering when they were released and how much time they have had to accumulate streams, chart activity and market exposure.

The resulting temporal context will help PMIP distinguish genuine performance differences from differences caused simply by release age or unequal observation periods.


## 8.1 Release Dates

### Purpose

This subsection examines the distribution and validity of release dates across the final PMIP release population.

The aim is to establish the temporal starting point for later analysis of release age, observation windows and current-versus-historical performance.

The assessment will examine:

- availability of release dates;
- earliest and latest releases;
- release-year distribution;
- concentration of releases across different time periods;
- relative release recency.

A validated and well-understood release-date field is required before meaningful temporal comparisons can be made across releases.

```python
# ============================================================
# Section 8.1 - Release Dates
# ============================================================

if not section_7_complete:
    raise RuntimeError(
        "Section 7 must be completed before beginning "
        "Release Timing and Historical Context."
    )

# ============================================================
# Working source
# ============================================================

release_timing_df = geographic_indicator_df.copy()

# ============================================================
# Validate release-date field
# ============================================================

if "release_date" not in release_timing_df.columns:
    raise RuntimeError(
        "release_date is not available in the final release population."
    )

release_timing_df["release_date"] = pd.to_datetime(
    release_timing_df["release_date"],
    errors="coerce"
)

# ============================================================
# Release-date availability
# ============================================================

release_timing_df["release_date_available"] = (
    release_timing_df["release_date"].notna()
)

valid_release_date_mask = (
    release_timing_df[
        "release_date_available"
    ]
)

valid_release_dates = (
    release_timing_df.loc[
        valid_release_date_mask,
        "release_date"
    ]
)

# ============================================================
# Release-year field
# ============================================================

release_timing_df["release_year"] = (
    release_timing_df[
        "release_date"
    ]
    .dt.year
)

# ============================================================
# Release-date summary
# ============================================================

release_date_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_timing_df
            ),
        },
        {
            "Metric": "Releases with valid release date",
            "Observed Value": int(
                valid_release_date_mask.sum()
            ),
        },
        {
            "Metric": "Releases without valid release date",
            "Observed Value": int(
                (
                    ~valid_release_date_mask
                ).sum()
            ),
        },
        {
            "Metric": "Earliest release date",
            "Observed Value": (
                valid_release_dates.min()
                if not valid_release_dates.empty
                else pd.NaT
            ),
        },
        {
            "Metric": "Latest release date",
            "Observed Value": (
                valid_release_dates.max()
                if not valid_release_dates.empty
                else pd.NaT
            ),
        },
        {
            "Metric": "Unique release years",
            "Observed Value": int(
                release_timing_df[
                    "release_year"
                ]
                .dropna()
                .nunique()
            ),
        },
    ]
)

print("Assessing PMIP release dates")
print("=" * 80)

display(release_date_summary_df)

# ============================================================
# Release-year distribution
# ============================================================

release_year_distribution_df = (
    release_timing_df[
        "release_year"
    ]
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
    .rename_axis(
        "Release Year"
    )
    .reset_index(
        name="Releases"
    )
)

release_year_distribution_df[
    "Share (%)"
] = (
    release_year_distribution_df[
        "Releases"
    ]
    / valid_release_date_mask.sum()
    * 100
)

print()
print("Release-year distribution")
print("=" * 80)

display(
    release_year_distribution_df
)

# ============================================================
# Release-decade distribution
# ============================================================

release_timing_df["release_decade"] = np.where(
    release_timing_df[
        "release_year"
    ].notna(),
    (
        (
            release_timing_df[
                "release_year"
            ]
            // 10
        )
        * 10
    ),
    np.nan,
)

release_decade_distribution_df = (
    release_timing_df[
        "release_decade"
    ]
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
    .rename_axis(
        "Release Decade"
    )
    .reset_index(
        name="Releases"
    )
)

release_decade_distribution_df[
    "Share (%)"
] = (
    release_decade_distribution_df[
        "Releases"
    ]
    / valid_release_date_mask.sum()
    * 100
)

print()
print("Release-decade distribution")
print("=" * 80)

display(
    release_decade_distribution_df
)

# ============================================================
# Release-recency score
#
# More recent release date = higher score
# ============================================================

release_timing_df[
    "release_recency_score"
] = np.nan

if valid_release_date_mask.any():

    release_timing_df.loc[
        valid_release_date_mask,
        "release_recency_score"
    ] = (
        release_timing_df.loc[
            valid_release_date_mask,
            "release_date"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Release-recency rank
# ============================================================

release_timing_df[
    "release_recency_rank"
] = (
    release_timing_df[
        "release_date"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Release period classes
# ============================================================

release_timing_df[
    "release_period_class"
] = "No Release-Date Evidence"

release_timing_df.loc[
    valid_release_date_mask
    & (
        release_timing_df[
            "release_year"
        ] < 2010
    ),
    "release_period_class"
] = "Legacy Release"

release_timing_df.loc[
    valid_release_date_mask
    & (
        release_timing_df[
            "release_year"
        ] >= 2010
    )
    & (
        release_timing_df[
            "release_year"
        ] < 2018
    ),
    "release_period_class"
] = "Established Catalogue"

release_timing_df.loc[
    valid_release_date_mask
    & (
        release_timing_df[
            "release_year"
        ] >= 2018
    )
    & (
        release_timing_df[
            "release_year"
        ] < 2022
    ),
    "release_period_class"
] = "Recent Catalogue"

release_timing_df.loc[
    valid_release_date_mask
    & (
        release_timing_df[
            "release_year"
        ] >= 2022
    ),
    "release_period_class"
] = "Current-Era Release"

# ============================================================
# Release-period class distribution
# ============================================================

release_period_distribution_df = (
    release_timing_df[
        "release_period_class"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Release Period Class"
    )
    .reset_index(
        name="Releases"
    )
)

release_period_distribution_df[
    "Share (%)"
] = (
    release_period_distribution_df[
        "Releases"
    ]
    / len(
        release_timing_df
    )
    * 100
)

print()
print("Release-period class distribution")
print("=" * 80)

display(
    release_period_distribution_df
)

# ============================================================
# Most recent releases
# ============================================================

most_recent_releases_df = (
    release_timing_df[
        valid_release_date_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "release_date",
            "release_year",
            "release_recency_score",
            "release_recency_rank",
            "release_period_class",
        ]
    ]
    .sort_values(
        by="release_date",
        ascending=False
    )
    .head(25)
    .reset_index(
        drop=True
    )
)

print()
print("Most recent PMIP releases")
print("=" * 80)

display(
    most_recent_releases_df
)

# ============================================================
# Validation
# ============================================================

release_date_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7 completed",
            "Passed": section_7_complete,
        },
        {
            "Validation Area": "Release-date field available",
            "Passed": (
                "release_date"
                in release_timing_df.columns
            ),
        },
        {
            "Validation Area": "Valid release dates available",
            "Passed": valid_release_date_mask.any(),
        },
        {
            "Validation Area": "Release-year field prepared",
            "Passed": (
                "release_year"
                in release_timing_df.columns
            ),
        },
        {
            "Validation Area": "Release-recency score calculated",
            "Passed": (
                "release_recency_score"
                in release_timing_df.columns
            ),
        },
        {
            "Validation Area": "Release-period classes assigned",
            "Passed": release_timing_df[
                "release_period_class"
            ].notna().all(),
        },
    ]
)

print()
print("Release-date validation")
print("=" * 80)

display(
    release_date_validation_df
)

# ============================================================
# Completion
# ============================================================

section_8_1_complete = bool(
    release_date_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 8.1 completion status: "
    f"{section_8_1_complete}"
)

if not section_8_1_complete:
    raise RuntimeError(
        "Release Dates analysis did not complete successfully."
    )
```

### Interpretation

The PMIP Release Dates assessment completed successfully.

All 4,593 validated release records contain valid release dates, meaning there is complete temporal coverage for the release population.

The earliest release in the dataset dates to 21 July 1987, while the latest release dates to 14 June 2024.

The release population spans 30 distinct release years, providing a mixture of older catalogue music and more recent releases.

However, the distribution is strongly concentrated toward newer music.

A total of 3,307 releases, representing approximately 72.00% of the full population, are from the 2020s. By comparison, 1,226 releases, or approximately 26.69%, are from the 2010s.

Only a very small proportion of the dataset comes from earlier decades.

The release-period classification reinforces this pattern.

A total of 2,539 releases, representing approximately 55.28% of the population, are classified as `Current-Era Release`.

A further 1,313 releases, or approximately 28.59%, fall within the `Recent Catalogue` group.

Together, these two categories account for the large majority of the PMIP release population.

This concentration toward recent releases is important for later analysis because newer releases generally have had less time to accumulate streams, chart activity, geographic reach and other long-term performance signals.

Older catalogue releases may therefore appear stronger on cumulative measures simply because they have been observable for a longer period.

The `release_recency_score` provides a relative measure of how recent each release is within the PMIP population, but recency itself should not be interpreted as a performance measure.

Section 8.1 therefore establishes the temporal baseline required for analysing release age, observation windows and temporal comparability in the remaining subsections.

## 8.2 Age of Release

### Purpose

This subsection measures how old each validated release is relative to the latest release date available in the PMIP dataset.

Using the dataset's latest observed release date as the reference point avoids introducing today's date into a historical dataset and keeps the calculation internally consistent.

The assessment will examine:

- release age in days;
- release age in years;
- distribution of release ages;
- younger versus older catalogue groups;
- relative age position within the PMIP population.

Release age provides important context because older releases have had more time to accumulate streams, chart history and geographic reach than newer releases.

```python
# ============================================================
# Section 8.2 - Age of Release
# ============================================================

if not section_8_1_complete:
    raise RuntimeError(
        "Section 8.1 must be completed before analysing release age."
    )

# ============================================================
# Working source
# ============================================================

release_age_df = release_timing_df.copy()

# ============================================================
# Reference date
#
# Use the latest release date in the PMIP dataset so the age
# calculation stays internally consistent with the dataset.
# ============================================================

reference_date = (
    release_age_df["release_date"]
    .dropna()
    .max()
)

if pd.isna(reference_date):
    raise RuntimeError(
        "A valid reference date could not be established."
    )

# ============================================================
# Release age calculations
# ============================================================

release_age_df["release_age_days"] = (
    reference_date
    - release_age_df["release_date"]
).dt.days

release_age_df["release_age_years"] = (
    release_age_df["release_age_days"]
    / 365.25
)

# ============================================================
# Valid age mask
# ============================================================

valid_release_age_mask = (
    release_age_df[
        "release_age_days"
    ].notna()
    & (
        release_age_df[
            "release_age_days"
        ] >= 0
    )
)

# ============================================================
# Relative age score
#
# Higher score = newer release
# ============================================================

release_age_df["release_age_score"] = np.nan

if valid_release_age_mask.any():

    release_age_df.loc[
        valid_release_age_mask,
        "release_age_score"
    ] = (
        release_age_df.loc[
            valid_release_age_mask,
            "release_age_days"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=False
        )
        * 100
    )

# ============================================================
# Age rank
#
# Rank 1 = newest release
# ============================================================

release_age_df["release_age_rank"] = (
    release_age_df[
        "release_age_days"
    ]
    .rank(
        method="min",
        ascending=True,
        na_option="bottom"
    )
)

# ============================================================
# Age classes
# ============================================================

release_age_df["release_age_class"] = (
    "No Release-Age Evidence"
)

release_age_df.loc[
    valid_release_age_mask
    & (
        release_age_df["release_age_years"] < 1
    ),
    "release_age_class"
] = "Under 1 Year"

release_age_df.loc[
    valid_release_age_mask
    & (
        release_age_df["release_age_years"] >= 1
    )
    & (
        release_age_df["release_age_years"] < 3
    ),
    "release_age_class"
] = "1-3 Years"

release_age_df.loc[
    valid_release_age_mask
    & (
        release_age_df["release_age_years"] >= 3
    )
    & (
        release_age_df["release_age_years"] < 5
    ),
    "release_age_class"
] = "3-5 Years"

release_age_df.loc[
    valid_release_age_mask
    & (
        release_age_df["release_age_years"] >= 5
    )
    & (
        release_age_df["release_age_years"] < 10
    ),
    "release_age_class"
] = "5-10 Years"

release_age_df.loc[
    valid_release_age_mask
    & (
        release_age_df["release_age_years"] >= 10
    ),
    "release_age_class"
] = "10+ Years"

# ============================================================
# Summary
# ============================================================

release_age_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_age_df
            ),
        },
        {
            "Metric": "Reference date",
            "Observed Value": reference_date,
        },
        {
            "Metric": "Releases with valid age",
            "Observed Value": int(
                valid_release_age_mask.sum()
            ),
        },
        {
            "Metric": "Releases without valid age",
            "Observed Value": int(
                (~valid_release_age_mask).sum()
            ),
        },
        {
            "Metric": "Minimum release age (days)",
            "Observed Value": (
                release_age_df.loc[
                    valid_release_age_mask,
                    "release_age_days"
                ].min()
            ),
        },
        {
            "Metric": "Median release age (days)",
            "Observed Value": (
                release_age_df.loc[
                    valid_release_age_mask,
                    "release_age_days"
                ].median()
            ),
        },
        {
            "Metric": "Median release age (years)",
            "Observed Value": (
                release_age_df.loc[
                    valid_release_age_mask,
                    "release_age_years"
                ].median()
            ),
        },
        {
            "Metric": "Mean release age (years)",
            "Observed Value": (
                release_age_df.loc[
                    valid_release_age_mask,
                    "release_age_years"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum release age (years)",
            "Observed Value": (
                release_age_df.loc[
                    valid_release_age_mask,
                    "release_age_years"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP release age")
print("=" * 80)

display(
    release_age_summary_df
)

# ============================================================
# Release-age percentiles
# ============================================================

release_age_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
        ],
        "Release Age (Years)": (
            release_age_df.loc[
                valid_release_age_mask,
                "release_age_years"
            ]
            .quantile(
                [
                    0.25,
                    0.50,
                    0.75,
                    0.90,
                    0.95,
                ]
            )
            .values
        ),
    }
)

print()
print("Release-age percentiles")
print("=" * 80)

display(
    release_age_percentiles_df
)

# ============================================================
# Age-class distribution
# ============================================================

release_age_class_distribution_df = (
    release_age_df[
        "release_age_class"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Release Age Class"
    )
    .reset_index(
        name="Releases"
    )
)

release_age_class_distribution_df[
    "Share (%)"
] = (
    release_age_class_distribution_df[
        "Releases"
    ]
    / len(
        release_age_df
    )
    * 100
)

print()
print("Release-age class distribution")
print("=" * 80)

display(
    release_age_class_distribution_df
)

# ============================================================
# Oldest releases
# ============================================================

oldest_releases_df = (
    release_age_df[
        valid_release_age_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "release_date",
            "release_age_days",
            "release_age_years",
            "release_age_score",
            "release_age_rank",
            "release_age_class",
        ]
    ]
    .sort_values(
        by="release_age_days",
        ascending=False
    )
    .head(25)
    .reset_index(
        drop=True
    )
)

print()
print("Oldest PMIP releases")
print("=" * 80)

display(
    oldest_releases_df
)

# ============================================================
# Validation
# ============================================================

release_age_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.1 completed",
            "Passed": section_8_1_complete,
        },
        {
            "Validation Area": "Reference date established",
            "Passed": pd.notna(
                reference_date
            ),
        },
        {
            "Validation Area": "Release-age days calculated",
            "Passed": (
                "release_age_days"
                in release_age_df.columns
            ),
        },
        {
            "Validation Area": "Release-age years calculated",
            "Passed": (
                "release_age_years"
                in release_age_df.columns
            ),
        },
        {
            "Validation Area": "Release-age score calculated",
            "Passed": (
                "release_age_score"
                in release_age_df.columns
            ),
        },
        {
            "Validation Area": "Release-age classes assigned",
            "Passed": release_age_df[
                "release_age_class"
            ].notna().all(),
        },
    ]
)

print()
print("Release-age validation")
print("=" * 80)

display(
    release_age_validation_df
)

# ============================================================
# Completion
# ============================================================

section_8_2_complete = bool(
    release_age_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 8.2 completion status: "
    f"{section_8_2_complete}"
)

if not section_8_2_complete:
    raise RuntimeError(
        "Age of Release analysis did not complete successfully."
    )
```

## 8.3 Observation Window

### Purpose

This subsection measures how much historical observation time is available for each validated release.

The observation window represents the period between the first and last available historical chart dates linked to a release.

The aim is to understand whether releases have been observed for comparable lengths of time before their performance is compared.

The assessment will examine:

- availability of historical observation windows;
- duration of each observation window in days;
- differences between short and long observation periods;
- releases with limited historical evidence;
- relative observation-window strength.

This is important because a release observed for only a short period should not be interpreted in exactly the same way as a release with several years of historical chart evidence.

```python
# ============================================================
# Section 8.3 - Observation Window
# ============================================================

if not section_8_2_complete:
    raise RuntimeError(
        "Section 8.2 must be completed before analysing observation windows."
    )

# ============================================================
# Working source
# ============================================================

observation_window_df = release_age_df.copy()

# ============================================================
# Validate historical date fields
# ============================================================

required_observation_fields = [
    "first_chart_date",
    "last_chart_date",
]

missing_observation_fields = [
    field
    for field in required_observation_fields
    if field not in observation_window_df.columns
]

if missing_observation_fields:
    raise RuntimeError(
        "Required historical date fields are missing: "
        f"{missing_observation_fields}"
    )

# ============================================================
# Parse dates
# ============================================================

observation_window_df["first_chart_date_parsed"] = pd.to_datetime(
    observation_window_df["first_chart_date"],
    errors="coerce"
)

observation_window_df["last_chart_date_parsed"] = pd.to_datetime(
    observation_window_df["last_chart_date"],
    errors="coerce"
)

# ============================================================
# Observation-window availability
# ============================================================

observation_window_df["observation_window_available"] = (
    observation_window_df["first_chart_date_parsed"].notna()
    & observation_window_df["last_chart_date_parsed"].notna()
)

valid_observation_window_mask = (
    observation_window_df["observation_window_available"]
)

# ============================================================
# Observation-window duration
# ============================================================

observation_window_df["observation_window_days"] = np.where(
    valid_observation_window_mask,
    (
        observation_window_df["last_chart_date_parsed"]
        - observation_window_df["first_chart_date_parsed"]
    ).dt.days,
    np.nan,
)

# Prevent negative durations from being silently accepted
negative_window_mask = (
    observation_window_df["observation_window_days"].notna()
    & (
        observation_window_df["observation_window_days"] < 0
    )
)

observation_window_df.loc[
    negative_window_mask,
    "observation_window_days"
] = np.nan

observation_window_df["observation_window_years"] = (
    observation_window_df["observation_window_days"]
    / 365.25
)

# Recalculate valid mask after negative-window handling
valid_observation_window_mask = (
    observation_window_df["observation_window_days"].notna()
)

# ============================================================
# Observation-window score
#
# Longer observation window = stronger evidence
# ============================================================

observation_window_df["observation_window_score"] = np.nan

if valid_observation_window_mask.any():

    observation_window_df.loc[
        valid_observation_window_mask,
        "observation_window_score"
    ] = (
        observation_window_df.loc[
            valid_observation_window_mask,
            "observation_window_days"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Observation-window rank
#
# Rank 1 = longest observation window
# ============================================================

observation_window_df["observation_window_rank"] = (
    observation_window_df[
        "observation_window_days"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Observation-window classes
# ============================================================

observation_window_df["observation_window_class"] = (
    "No Observation-Window Evidence"
)

observation_window_df.loc[
    valid_observation_window_mask
    & (
        observation_window_df[
            "observation_window_days"
        ] < 30
    ),
    "observation_window_class"
] = "Very Short Observation Window"

observation_window_df.loc[
    valid_observation_window_mask
    & (
        observation_window_df[
            "observation_window_days"
        ] >= 30
    )
    & (
        observation_window_df[
            "observation_window_days"
        ] < 90
    ),
    "observation_window_class"
] = "Short Observation Window"

observation_window_df.loc[
    valid_observation_window_mask
    & (
        observation_window_df[
            "observation_window_days"
        ] >= 90
    )
    & (
        observation_window_df[
            "observation_window_days"
        ] < 365
    ),
    "observation_window_class"
] = "Moderate Observation Window"

observation_window_df.loc[
    valid_observation_window_mask
    & (
        observation_window_df[
            "observation_window_days"
        ] >= 365
    )
    & (
        observation_window_df[
            "observation_window_days"
        ] < 1095
    ),
    "observation_window_class"
] = "Long Observation Window"

observation_window_df.loc[
    valid_observation_window_mask
    & (
        observation_window_df[
            "observation_window_days"
        ] >= 1095
    ),
    "observation_window_class"
] = "Very Long Observation Window"

# ============================================================
# Summary
# ============================================================

observation_window_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                observation_window_df
            ),
        },
        {
            "Metric": "Releases with observation-window evidence",
            "Observed Value": int(
                valid_observation_window_mask.sum()
            ),
        },
        {
            "Metric": "Releases without observation-window evidence",
            "Observed Value": int(
                (~valid_observation_window_mask).sum()
            ),
        },
        {
            "Metric": "Negative observation windows detected",
            "Observed Value": int(
                negative_window_mask.sum()
            ),
        },
        {
            "Metric": "Minimum observation window (days)",
            "Observed Value": (
                observation_window_df.loc[
                    valid_observation_window_mask,
                    "observation_window_days"
                ].min()
            ),
        },
        {
            "Metric": "Median observation window (days)",
            "Observed Value": (
                observation_window_df.loc[
                    valid_observation_window_mask,
                    "observation_window_days"
                ].median()
            ),
        },
        {
            "Metric": "Mean observation window (days)",
            "Observed Value": (
                observation_window_df.loc[
                    valid_observation_window_mask,
                    "observation_window_days"
                ].mean()
            ),
        },
        {
            "Metric": "Median observation window (years)",
            "Observed Value": (
                observation_window_df.loc[
                    valid_observation_window_mask,
                    "observation_window_years"
                ].median()
            ),
        },
        {
            "Metric": "Maximum observation window (years)",
            "Observed Value": (
                observation_window_df.loc[
                    valid_observation_window_mask,
                    "observation_window_years"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP observation windows")
print("=" * 80)

display(
    observation_window_summary_df
)

# ============================================================
# Observation-window percentiles
# ============================================================

observation_window_percentiles_df = pd.DataFrame(
    {
        "Percentile": [
            "25%",
            "50%",
            "75%",
            "90%",
            "95%",
        ],
        "Observation Window (Days)": (
            observation_window_df.loc[
                valid_observation_window_mask,
                "observation_window_days"
            ]
            .quantile(
                [
                    0.25,
                    0.50,
                    0.75,
                    0.90,
                    0.95,
                ]
            )
            .values
        ),
    }
)

print()
print("Observation-window percentiles")
print("=" * 80)

display(
    observation_window_percentiles_df
)

# ============================================================
# Class distribution
# ============================================================

observation_window_class_distribution_df = (
    observation_window_df[
        "observation_window_class"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Observation Window Class"
    )
    .reset_index(
        name="Releases"
    )
)

observation_window_class_distribution_df[
    "Share (%)"
] = (
    observation_window_class_distribution_df[
        "Releases"
    ]
    / len(
        observation_window_df
    )
    * 100
)

print()
print("Observation-window class distribution")
print("=" * 80)

display(
    observation_window_class_distribution_df
)

# ============================================================
# Releases with longest observation windows
# ============================================================

longest_observation_windows_df = (
    observation_window_df[
        valid_observation_window_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "release_date",
            "first_chart_date_parsed",
            "last_chart_date_parsed",
            "observation_window_days",
            "observation_window_years",
            "observation_window_score",
            "observation_window_rank",
            "observation_window_class",
        ]
    ]
    .sort_values(
        by="observation_window_days",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with longest observation windows")
print("=" * 80)

display(
    longest_observation_windows_df
)

# ============================================================
# Validation
# ============================================================

observation_window_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.2 completed",
            "Passed": section_8_2_complete,
        },
        {
            "Validation Area": "Historical chart dates available",
            "Passed": all(
                field in observation_window_df.columns
                for field in required_observation_fields
            ),
        },
        {
            "Validation Area": "Observation-window evidence available",
            "Passed": valid_observation_window_mask.any(),
        },
        {
            "Validation Area": "Observation-window duration calculated",
            "Passed": (
                "observation_window_days"
                in observation_window_df.columns
            ),
        },
        {
            "Validation Area": "No negative observation windows retained",
            "Passed": not (
                observation_window_df[
                    "observation_window_days"
                ]
                .dropna()
                .lt(0)
                .any()
            ),
        },
        {
            "Validation Area": "Observation-window score calculated",
            "Passed": (
                "observation_window_score"
                in observation_window_df.columns
            ),
        },
        {
            "Validation Area": "Observation-window classes assigned",
            "Passed": observation_window_df[
                "observation_window_class"
            ].notna().all(),
        },
    ]
)

print()
print("Observation-window validation")
print("=" * 80)

display(
    observation_window_validation_df
)

# ============================================================
# Completion
# ============================================================

section_8_3_complete = bool(
    observation_window_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 8.3 completion status: "
    f"{section_8_3_complete}"
)

if not section_8_3_complete:
    raise RuntimeError(
        "Observation Window analysis did not complete successfully."
    )
```

### Interpretation

The PMIP Observation Window assessment completed successfully.

Observation-window evidence is available for 2,235 of the 4,593 validated releases, while 2,358 releases do not contain sufficient historical chart-date evidence for this analysis.

No negative observation windows were detected, confirming that the available historical date ranges are temporally valid.

Among releases with observation-window evidence, the median observation period is 413 days, or approximately 1.13 years.

The mean observation window is approximately 718.47 days, which is substantially higher than the median. This indicates that a smaller number of releases have very long historical chart histories and increase the overall average.

The percentile distribution shows considerable variation in the amount of historical evidence available.

At the 25th percentile, releases have approximately 133 days of observation history, while the median is 413 days.

At the 75th percentile, the observation window increases to approximately 945 days.

A total of 715 releases are classified as having a `Long Observation Window`, while 490 releases have a `Very Long Observation Window`.

A further 619 releases have a `Moderate Observation Window`.

Smaller groups have limited historical coverage, including 226 releases with a `Very Short Observation Window` and 185 releases with a `Short Observation Window`.

The observation-window score measures the strength of temporal evidence rather than release performance.

A longer observation window does not automatically mean that a release performed better. Instead, it means that more historical evidence is available to support analysis of persistence, movement and long-term behaviour.

This distinction is important when comparing releases because similar performance scores may be supported by very different amounts of historical data.

Section 8.3 completed successfully and establishes the historical observation context required for comparing current and historical release performance.

## 8.4 Current vs Historical Performance

### Purpose

This subsection compares current release performance with the historical performance evidence available for the same release.

The aim is to determine whether a release's present streaming level appears stronger, weaker or broadly consistent with its historical behaviour.

The assessment will examine:

- availability of both current and historical streaming evidence;
- difference between current Spotify streams and historical average streams;
- relative current-versus-historical performance;
- releases currently performing above or below their historical level;
- the strength of the temporal comparison evidence.

This comparison provides temporal context for interpreting whether a release is currently outperforming, underperforming or remaining broadly aligned with its historical performance profile.

```python
# ============================================================
# Section 8.4 - Current vs Historical Performance
# ============================================================

if not section_8_3_complete:
    raise RuntimeError(
        "Section 8.3 must be completed before comparing "
        "current and historical performance."
    )

# ============================================================
# Working source
# ============================================================

current_vs_historical_df = observation_window_df.copy()

# ============================================================
# Required performance fields
# ============================================================

required_temporal_performance_fields = [
    "Spotify Streams",
    "average_historical_streams",
]

missing_temporal_performance_fields = [
    field
    for field in required_temporal_performance_fields
    if field not in current_vs_historical_df.columns
]

if missing_temporal_performance_fields:
    raise RuntimeError(
        "Required current/historical performance fields are missing: "
        f"{missing_temporal_performance_fields}"
    )

# ============================================================
# Numeric preparation
# ============================================================

current_vs_historical_df[
    "current_spotify_streams"
] = pd.to_numeric(
    current_vs_historical_df["Spotify Streams"],
    errors="coerce"
)

current_vs_historical_df[
    "historical_average_streams"
] = pd.to_numeric(
    current_vs_historical_df[
        "average_historical_streams"
    ],
    errors="coerce"
)

# ============================================================
# Valid comparison mask
#
# Both current and historical streaming evidence must exist.
# ============================================================

valid_current_historical_mask = (
    current_vs_historical_df[
        "current_spotify_streams"
    ].notna()
    & current_vs_historical_df[
        "historical_average_streams"
    ].notna()
    & (
        current_vs_historical_df[
            "historical_average_streams"
        ] >= 0
    )
)

# ============================================================
# Current-performance percentile
#
# Current Spotify Streams and historical average streams are
# not directly comparable in raw units.
#
# Therefore each measure is converted into a relative
# percentile within the same matched release population.
# ============================================================

current_vs_historical_df[
    "current_streaming_percentile"
] = np.nan

current_vs_historical_df[
    "historical_streaming_percentile"
] = np.nan

if valid_current_historical_mask.any():

    current_vs_historical_df.loc[
        valid_current_historical_mask,
        "current_streaming_percentile"
    ] = (
        current_vs_historical_df.loc[
            valid_current_historical_mask,
            "current_spotify_streams"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

    current_vs_historical_df.loc[
        valid_current_historical_mask,
        "historical_streaming_percentile"
    ] = (
        current_vs_historical_df.loc[
            valid_current_historical_mask,
            "historical_average_streams"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Relative temporal performance difference
#
# Positive:
# current relative standing is stronger than historical
#
# Negative:
# current relative standing is weaker than historical
#
# Units = percentile points
# ============================================================

current_vs_historical_df[
    "current_vs_historical_percentile_difference"
] = np.where(
    valid_current_historical_mask,
    (
        current_vs_historical_df[
            "current_streaming_percentile"
        ]
        - current_vs_historical_df[
            "historical_streaming_percentile"
        ]
    ),
    np.nan,
)

# ============================================================
# Absolute temporal divergence
#
# Shows the magnitude of change regardless of direction.
# ============================================================

current_vs_historical_df[
    "current_vs_historical_absolute_difference"
] = (
    current_vs_historical_df[
        "current_vs_historical_percentile_difference"
    ]
    .abs()
)

# ============================================================
# Direction
#
# ±10 percentile points is used as a practical band for
# broadly similar relative current/historical standing.
# ============================================================

current_vs_historical_df[
    "current_vs_historical_direction"
] = "Unavailable"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] > 10
    ),
    "current_vs_historical_direction"
] = "Current Position Stronger"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] < -10
    ),
    "current_vs_historical_direction"
] = "Historical Position Stronger"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ].between(
            -10,
            10,
            inclusive="both"
        )
    ),
    "current_vs_historical_direction"
] = "Broadly Comparable"

# ============================================================
# Temporal performance classes
# ============================================================

current_vs_historical_df[
    "current_vs_historical_class"
] = "No Current-Historical Evidence"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] <= -25
    ),
    "current_vs_historical_class"
] = "Substantially Weaker Current Standing"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] > -25
    )
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] < -10
    ),
    "current_vs_historical_class"
] = "Weaker Current Standing"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ].between(
            -10,
            10,
            inclusive="both"
        )
    ),
    "current_vs_historical_class"
] = "Broadly Comparable Standing"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] > 10
    )
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] < 25
    ),
    "current_vs_historical_class"
] = "Stronger Current Standing"

current_vs_historical_df.loc[
    valid_current_historical_mask
    & (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ] >= 25
    ),
    "current_vs_historical_class"
] = "Substantially Stronger Current Standing"

# ============================================================
# Current-vs-historical score
#
# 50 = broadly neutral
# >50 = stronger current relative standing
# <50 = stronger historical relative standing
# ============================================================

current_vs_historical_df[
    "current_vs_historical_score"
] = np.where(
    valid_current_historical_mask,
    (
        current_vs_historical_df[
            "current_vs_historical_percentile_difference"
        ]
        + 100
    )
    / 2,
    np.nan,
)

current_vs_historical_df[
    "current_vs_historical_score"
] = (
    current_vs_historical_df[
        "current_vs_historical_score"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# Summary
# ============================================================

current_vs_historical_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                current_vs_historical_df
            ),
        },
        {
            "Metric": "Releases with current-historical comparison",
            "Observed Value": int(
                valid_current_historical_mask.sum()
            ),
        },
        {
            "Metric": "Releases without current-historical comparison",
            "Observed Value": int(
                (~valid_current_historical_mask).sum()
            ),
        },
        {
            "Metric": "Median current percentile",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "current_streaming_percentile"
                ].median()
            ),
        },
        {
            "Metric": "Median historical percentile",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "historical_streaming_percentile"
                ].median()
            ),
        },
        {
            "Metric": "Median percentile difference",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "current_vs_historical_percentile_difference"
                ].median()
            ),
        },
        {
            "Metric": "Mean percentile difference",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "current_vs_historical_percentile_difference"
                ].mean()
            ),
        },
        {
            "Metric": "Median absolute percentile difference",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "current_vs_historical_absolute_difference"
                ].median()
            ),
        },
        {
            "Metric": "Median current-vs-historical score",
            "Observed Value": (
                current_vs_historical_df.loc[
                    valid_current_historical_mask,
                    "current_vs_historical_score"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP current vs historical performance")
print("=" * 80)

display(
    current_vs_historical_summary_df
)

# ============================================================
# Direction distribution
# ============================================================

current_vs_historical_direction_distribution_df = (
    current_vs_historical_df[
        "current_vs_historical_direction"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Current vs Historical Direction"
    )
    .reset_index(
        name="Releases"
    )
)

current_vs_historical_direction_distribution_df[
    "Share (%)"
] = (
    current_vs_historical_direction_distribution_df[
        "Releases"
    ]
    / len(
        current_vs_historical_df
    )
    * 100
)

print()
print("Current-vs-historical direction distribution")
print("=" * 80)

display(
    current_vs_historical_direction_distribution_df
)

# ============================================================
# Class distribution
# ============================================================

current_vs_historical_class_distribution_df = (
    current_vs_historical_df[
        "current_vs_historical_class"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Current vs Historical Class"
    )
    .reset_index(
        name="Releases"
    )
)

current_vs_historical_class_distribution_df[
    "Share (%)"
] = (
    current_vs_historical_class_distribution_df[
        "Releases"
    ]
    / len(
        current_vs_historical_df
    )
    * 100
)

print()
print("Current-vs-historical class distribution")
print("=" * 80)

display(
    current_vs_historical_class_distribution_df
)

# ============================================================
# Strongest current relative standing
# ============================================================

top_current_outperformance_df = (
    current_vs_historical_df[
        valid_current_historical_mask
    ][
        [
            "release_id",
            "Track",
            "Artist",
            "current_spotify_streams",
            "historical_average_streams",
            "current_streaming_percentile",
            "historical_streaming_percentile",
            "current_vs_historical_percentile_difference",
            "current_vs_historical_score",
            "current_vs_historical_direction",
            "current_vs_historical_class",
        ]
    ]
    .sort_values(
        by="current_vs_historical_percentile_difference",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest current relative standing")
print("=" * 80)

display(
    top_current_outperformance_df
)

# ============================================================
# Validation
# ============================================================

current_vs_historical_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.3 completed",
            "Passed": section_8_3_complete,
        },
        {
            "Validation Area": "Current streaming field available",
            "Passed": (
                "Spotify Streams"
                in current_vs_historical_df.columns
            ),
        },
        {
            "Validation Area": "Historical average stream field available",
            "Passed": (
                "average_historical_streams"
                in current_vs_historical_df.columns
            ),
        },
        {
            "Validation Area": "Current-historical comparisons available",
            "Passed": valid_current_historical_mask.any(),
        },
        {
            "Validation Area": "Current percentile calculated",
            "Passed": current_vs_historical_df[
                "current_streaming_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Historical percentile calculated",
            "Passed": current_vs_historical_df[
                "historical_streaming_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Relative temporal difference calculated",
            "Passed": current_vs_historical_df[
                "current_vs_historical_percentile_difference"
            ].notna().any(),
        },
        {
            "Validation Area": "Current-historical score calculated",
            "Passed": current_vs_historical_df[
                "current_vs_historical_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Current-historical classes assigned",
            "Passed": current_vs_historical_df[
                "current_vs_historical_class"
            ].notna().all(),
        },
    ]
)

print()
print("Current vs Historical Performance validation")
print("=" * 80)

display(
    current_vs_historical_validation_df
)

# ============================================================
# Completion
# ============================================================

section_8_4_complete = bool(
    current_vs_historical_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 8.4 completion status: "
    f"{section_8_4_complete}"
)

if not section_8_4_complete:
    raise RuntimeError(
        "Current vs Historical Performance analysis "
        "did not complete successfully."
    )
```

## 8.5 Temporal Comparability

### Purpose

This subsection evaluates how safely different releases can be compared from a temporal perspective.

The aim is to combine the main timing and historical-context evidence developed in Sections 8.1 to 8.4.

Temporal comparability considers:

- release age;
- observation-window strength;
- availability of current-versus-historical evidence;
- similarity between current and historical relative performance.

The goal is not to reward older or newer releases.

Instead, the indicator measures how much temporal evidence is available and whether the release has enough context to support a fair comparison with other releases.

Releases with stronger observation histories and more complete temporal evidence receive higher comparability scores, while releases with limited historical context are treated more cautiously.

```python
# ============================================================
# Section 8.5 - Temporal Comparability
# ============================================================

if not section_8_4_complete:
    raise RuntimeError(
        "Section 8.4 must be completed before constructing "
        "Temporal Comparability."
    )

# ============================================================
# Working source
# ============================================================

temporal_comparability_df = current_vs_historical_df.copy()

# ============================================================
# Prepare temporal components
# ============================================================

# 1. Release-age context
#
# We do not reward old or new releases directly.
# Instead, the age score is converted into a distance-from-middle
# measure so extremely new or extremely old releases are treated
# as less directly comparable than releases nearer the centre.
# ============================================================

temporal_comparability_df[
    "release_age_context_component"
] = np.where(
    temporal_comparability_df[
        "release_age_score"
    ].notna(),
    (
        100
        - (
            temporal_comparability_df[
                "release_age_score"
            ]
            - 50
        ).abs()
        * 2
    ),
    np.nan,
)

temporal_comparability_df[
    "release_age_context_component"
] = (
    temporal_comparability_df[
        "release_age_context_component"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# 2. Observation-window component
#
# Longer observation windows provide stronger temporal evidence.
# ============================================================

temporal_comparability_df[
    "observation_window_component"
] = pd.to_numeric(
    temporal_comparability_df[
        "observation_window_score"
    ],
    errors="coerce"
)

# ============================================================
# 3. Current-historical comparability component
#
# Releases whose current and historical percentile positions are
# closer together receive stronger comparability scores.
#
# 0 percentile-point difference = 100
# 100 percentile-point difference = 0
# ============================================================

temporal_comparability_df[
    "current_historical_comparability_component"
] = np.where(
    temporal_comparability_df[
        "current_vs_historical_absolute_difference"
    ].notna(),
    (
        100
        - temporal_comparability_df[
            "current_vs_historical_absolute_difference"
        ]
    ),
    np.nan,
)

temporal_comparability_df[
    "current_historical_comparability_component"
] = (
    temporal_comparability_df[
        "current_historical_comparability_component"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# Component weights
#
# Observation-window strength receives the greatest weight because
# it directly describes how much historical context is available.
# ============================================================

temporal_component_weights = {
    "release_age_context_component": 0.25,
    "observation_window_component": 0.45,
    "current_historical_comparability_component": 0.30,
}

temporal_component_columns = list(
    temporal_component_weights.keys()
)

# ============================================================
# Coverage-aware weighted score
# ============================================================

temporal_weighted_sum = pd.Series(
    0.0,
    index=temporal_comparability_df.index
)

temporal_available_weight_sum = pd.Series(
    0.0,
    index=temporal_comparability_df.index
)

for column, weight in temporal_component_weights.items():

    available_mask = (
        temporal_comparability_df[
            column
        ].notna()
    )

    temporal_weighted_sum.loc[
        available_mask
    ] += (
        temporal_comparability_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    temporal_available_weight_sum.loc[
        available_mask
    ] += weight

temporal_comparability_df[
    "temporal_comparability_score"
] = np.where(
    temporal_available_weight_sum > 0,
    (
        temporal_weighted_sum
        / temporal_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Evidence coverage
# ============================================================

temporal_comparability_df[
    "temporal_components_available"
] = (
    temporal_comparability_df[
        temporal_component_columns
    ]
    .notna()
    .sum(axis=1)
)

temporal_comparability_df[
    "temporal_evidence_coverage_pct"
] = (
    temporal_comparability_df[
        "temporal_components_available"
    ]
    / len(
        temporal_component_columns
    )
    * 100
)

# ============================================================
# Evidence strength
# ============================================================

temporal_comparability_df[
    "temporal_evidence_strength"
] = np.select(
    [
        temporal_comparability_df[
            "temporal_components_available"
        ] == 3,

        temporal_comparability_df[
            "temporal_components_available"
        ] == 2,

        temporal_comparability_df[
            "temporal_components_available"
        ] == 1,
    ],
    [
        "Very Strong Evidence",
        "Moderate Evidence",
        "Limited Evidence",
    ],
    default="No Evidence",
)

# ============================================================
# Valid score mask
# ============================================================

valid_temporal_comparability_mask = (
    temporal_comparability_df[
        "temporal_comparability_score"
    ].notna()
)

# ============================================================
# Temporal comparability percentile
# ============================================================

temporal_comparability_df[
    "temporal_comparability_percentile"
] = np.nan

if valid_temporal_comparability_mask.any():

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask,
        "temporal_comparability_percentile"
    ] = (
        temporal_comparability_df.loc[
            valid_temporal_comparability_mask,
            "temporal_comparability_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Temporal comparability rank
# ============================================================

temporal_comparability_df[
    "temporal_comparability_rank"
] = (
    temporal_comparability_df[
        "temporal_comparability_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Temporal comparability classes
# ============================================================

temporal_comparability_df[
    "temporal_comparability_class"
] = "No Temporal Comparability Evidence"

if valid_temporal_comparability_mask.any():

    q25 = temporal_comparability_df.loc[
        valid_temporal_comparability_mask,
        "temporal_comparability_score"
    ].quantile(0.25)

    q50 = temporal_comparability_df.loc[
        valid_temporal_comparability_mask,
        "temporal_comparability_score"
    ].quantile(0.50)

    q75 = temporal_comparability_df.loc[
        valid_temporal_comparability_mask,
        "temporal_comparability_score"
    ].quantile(0.75)

    q90 = temporal_comparability_df.loc[
        valid_temporal_comparability_mask,
        "temporal_comparability_score"
    ].quantile(0.90)

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] <= q25
        ),
        "temporal_comparability_class"
    ] = "Low Temporal Comparability"

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] > q25
        )
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] <= q50
        ),
        "temporal_comparability_class"
    ] = "Moderate Temporal Comparability"

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] > q50
        )
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] <= q75
        ),
        "temporal_comparability_class"
    ] = "High Temporal Comparability"

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] > q75
        )
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] <= q90
        ),
        "temporal_comparability_class"
    ] = "Very High Temporal Comparability"

    temporal_comparability_df.loc[
        valid_temporal_comparability_mask
        & (
            temporal_comparability_df[
                "temporal_comparability_score"
            ] > q90
        ),
        "temporal_comparability_class"
    ] = "Exceptional Temporal Comparability"

# ============================================================
# Summary
# ============================================================

temporal_comparability_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                temporal_comparability_df
            ),
        },
        {
            "Metric": "Releases with temporal-comparability score",
            "Observed Value": int(
                valid_temporal_comparability_mask.sum()
            ),
        },
        {
            "Metric": "Releases without temporal-comparability score",
            "Observed Value": int(
                (~valid_temporal_comparability_mask).sum()
            ),
        },
        {
            "Metric": "Median temporal-comparability score",
            "Observed Value": (
                temporal_comparability_df[
                    "temporal_comparability_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean temporal-comparability score",
            "Observed Value": (
                temporal_comparability_df[
                    "temporal_comparability_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum temporal-comparability score",
            "Observed Value": (
                temporal_comparability_df[
                    "temporal_comparability_score"
                ].max()
            ),
        },
        {
            "Metric": "Median temporal evidence coverage (%)",
            "Observed Value": (
                temporal_comparability_df[
                    "temporal_evidence_coverage_pct"
                ].median()
            ),
        },
        {
            "Metric": "Median evidence coverage among scored releases (%)",
            "Observed Value": (
                temporal_comparability_df.loc[
                    valid_temporal_comparability_mask,
                    "temporal_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Constructing PMIP Temporal Comparability")
print("=" * 80)

display(
    temporal_comparability_summary_df
)

# ============================================================
# Component coverage
# ============================================================

temporal_component_coverage_rows = []

for column in temporal_component_columns:

    temporal_component_coverage_rows.append(
        {
            "Component": column,
            "Releases Available": int(
                temporal_comparability_df[
                    column
                ].notna().sum()
            ),
            "Coverage (%)": round(
                temporal_comparability_df[
                    column
                ].notna().mean()
                * 100,
                6
            ),
            "Base Weight": temporal_component_weights[
                column
            ],
        }
    )

temporal_component_coverage_df = pd.DataFrame(
    temporal_component_coverage_rows
)

print()
print("Temporal indicator component coverage")
print("=" * 80)

display(
    temporal_component_coverage_df
)

# ============================================================
# Evidence-strength distribution
# ============================================================

temporal_evidence_strength_distribution_df = (
    temporal_comparability_df[
        "temporal_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Evidence Strength"
    )
    .reset_index(
        name="Releases"
    )
)

temporal_evidence_strength_distribution_df[
    "Share (%)"
] = (
    temporal_evidence_strength_distribution_df[
        "Releases"
    ]
    / len(
        temporal_comparability_df
    )
    * 100
)

print()
print("Temporal evidence-strength distribution")
print("=" * 80)

display(
    temporal_evidence_strength_distribution_df
)

# ============================================================
# Class distribution
# ============================================================

temporal_comparability_class_distribution_df = (
    temporal_comparability_df[
        "temporal_comparability_class"
    ]
    .value_counts()
    .rename_axis(
        "Temporal Comparability Class"
    )
    .reset_index(
        name="Releases"
    )
)

temporal_comparability_class_distribution_df[
    "Share (%)"
] = (
    temporal_comparability_class_distribution_df[
        "Releases"
    ]
    / len(
        temporal_comparability_df
    )
    * 100
)

print()
print("Temporal-comparability class distribution")
print("=" * 80)

display(
    temporal_comparability_class_distribution_df
)

# ============================================================
# Highest temporal comparability releases
# ============================================================

top_temporal_comparability_df = (
    temporal_comparability_df[
        [
            "release_id",
            "Track",
            "Artist",
            "release_age_years",
            "observation_window_days",
            "current_vs_historical_percentile_difference",
            "release_age_context_component",
            "observation_window_component",
            "current_historical_comparability_component",
            "temporal_components_available",
            "temporal_evidence_strength",
            "temporal_comparability_score",
            "temporal_comparability_percentile",
            "temporal_comparability_rank",
            "temporal_comparability_class",
        ]
    ]
    .sort_values(
        by="temporal_comparability_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP temporal-comparability releases")
print("=" * 80)

display(
    top_temporal_comparability_df
)

# ============================================================
# Validation
# ============================================================

temporal_comparability_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.4 completed",
            "Passed": section_8_4_complete,
        },
        {
            "Validation Area": "Release-age component prepared",
            "Passed": (
                "release_age_context_component"
                in temporal_comparability_df.columns
            ),
        },
        {
            "Validation Area": "Observation-window component prepared",
            "Passed": (
                "observation_window_component"
                in temporal_comparability_df.columns
            ),
        },
        {
            "Validation Area": "Current-historical comparability component prepared",
            "Passed": (
                "current_historical_comparability_component"
                in temporal_comparability_df.columns
            ),
        },
        {
            "Validation Area": "Coverage-aware temporal weighting prepared",
            "Passed": (
                temporal_available_weight_sum > 0
            ).any(),
        },
        {
            "Validation Area": "Temporal-comparability score calculated",
            "Passed": temporal_comparability_df[
                "temporal_comparability_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Temporal evidence coverage calculated",
            "Passed": temporal_comparability_df[
                "temporal_evidence_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Temporal-comparability classes assigned",
            "Passed": temporal_comparability_df[
                "temporal_comparability_class"
            ].notna().all(),
        },
    ]
)

print()
print("Temporal Comparability validation")
print("=" * 80)

display(
    temporal_comparability_validation_df
)

# ============================================================
# Completion
# ============================================================

section_8_5_complete = bool(
    temporal_comparability_validation_df[
        "Passed"
    ].all()
)

section_8_complete = bool(
    section_8_1_complete
    and section_8_2_complete
    and section_8_3_complete
    and section_8_4_complete
    and section_8_5_complete
)

print()
print(
    f"Section 8.5 completion status: "
    f"{section_8_5_complete}"
)

print(
    f"Section 8 overall completion status: "
    f"{section_8_complete}"
)

if not section_8_complete:
    raise RuntimeError(
        "Section 8 Release Timing and Historical Context "
        "did not complete successfully."
    )
```

# 9. Artist Context

This section evaluates each release in relation to the broader performance context of its artist.

The aim is to determine whether a release is performing strongly because of the release itself, because the artist already has a strong baseline, or because the artist is currently experiencing positive momentum.

The analysis will focus on:

- overall artist-level performance;
- release performance relative to the artist's usual baseline;
- artist growth context;
- artist momentum context;
- the relationship between artist strength and release-level performance.

This provides an important layer of interpretation because two releases with similar streaming, chart or geographic scores may come from artists with very different historical profiles.

The resulting artist-context features will help PMIP distinguish between established artist strength, release-specific outperformance and emerging artist momentum.

## 9.1 Artist-Level Performance

### Purpose

This subsection measures the broader performance profile of each artist represented in the final PMIP release population.

The aim is to construct an artist-level baseline that summarises how strongly an artist performs across all of their represented releases.

The assessment will examine:

- number of releases represented for each artist;
- average release-level streaming performance;
- average chart performance;
- average geographic reach;
- average temporal comparability;
- overall artist-level performance strength.

This artist baseline will later be used to determine whether an individual release is performing above or below the artist's normal level.

```python
# ============================================================
# Section 9.1 - Artist-Level Performance
# ============================================================

if not section_8_complete:
    raise RuntimeError(
        "Section 8 must be completed before beginning "
        "Artist Context analysis."
    )

# ============================================================
# Working source
# ============================================================

artist_context_df = temporal_comparability_df.copy()

# ============================================================
# Validate required artist field
# ============================================================

if "artist_normalised" not in artist_context_df.columns:
    raise RuntimeError(
        "artist_normalised is not available in the working dataset."
    )

# ============================================================
# Candidate release-level performance fields
# ============================================================

artist_performance_candidates = {
    "streaming_performance_score": "Streaming Performance",
    "chart_performance_score": "Chart Performance",
    "geographic_reach_score": "Geographic Reach",
    "temporal_comparability_score": "Temporal Comparability",
}

available_artist_performance_fields = [
    field
    for field in artist_performance_candidates
    if field in artist_context_df.columns
]

if not available_artist_performance_fields:
    raise RuntimeError(
        "No release-level performance fields are available "
        "for artist-level aggregation."
    )

# ============================================================
# Artist release counts
# ============================================================

artist_release_count_df = (
    artist_context_df
    .groupby(
        "artist_normalised",
        dropna=False
    )
    .size()
    .rename(
        "artist_release_count"
    )
    .reset_index()
)

# ============================================================
# Artist performance aggregation
# ============================================================

aggregation_dict = {
    field: "mean"
    for field in available_artist_performance_fields
}

artist_performance_df = (
    artist_context_df
    .groupby(
        "artist_normalised",
        dropna=False
    )
    .agg(
        aggregation_dict
    )
    .reset_index()
)

# ============================================================
# Rename aggregated fields
# ============================================================

artist_performance_df = artist_performance_df.rename(
    columns={
        field: f"artist_avg_{field}"
        for field in available_artist_performance_fields
    }
)

# ============================================================
# Add release counts
# ============================================================

artist_performance_df = artist_performance_df.merge(
    artist_release_count_df,
    on="artist_normalised",
    how="left"
)

# ============================================================
# Artist display label
# ============================================================

artist_display_df = (
    artist_context_df[
        [
            "artist_normalised",
            "Artist"
        ]
    ]
    .drop_duplicates(
        subset=[
            "artist_normalised"
        ]
    )
)

artist_performance_df = artist_performance_df.merge(
    artist_display_df,
    on="artist_normalised",
    how="left"
)

# ============================================================
# Artist-level component scores
#
# Each aggregated release-level indicator is already scaled
# approximately from 0 to 100.
# ============================================================

artist_component_columns = [
    f"artist_avg_{field}"
    for field in available_artist_performance_fields
]

# ============================================================
# Artist evidence coverage
# ============================================================

artist_performance_df[
    "artist_components_available"
] = (
    artist_performance_df[
        artist_component_columns
    ]
    .notna()
    .sum(axis=1)
)

artist_performance_df[
    "artist_evidence_coverage_pct"
] = (
    artist_performance_df[
        "artist_components_available"
    ]
    / len(
        artist_component_columns
    )
    * 100
)

# ============================================================
# Artist performance score
#
# Equal weighting is used across the available major
# release-performance domains.
# ============================================================

artist_performance_df[
    "artist_performance_score"
] = (
    artist_performance_df[
        artist_component_columns
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# ============================================================
# Artist performance percentile
# ============================================================

valid_artist_performance_mask = (
    artist_performance_df[
        "artist_performance_score"
    ].notna()
)

artist_performance_df[
    "artist_performance_percentile"
] = np.nan

if valid_artist_performance_mask.any():

    artist_performance_df.loc[
        valid_artist_performance_mask,
        "artist_performance_percentile"
    ] = (
        artist_performance_df.loc[
            valid_artist_performance_mask,
            "artist_performance_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Artist performance rank
# ============================================================

artist_performance_df[
    "artist_performance_rank"
] = (
    artist_performance_df[
        "artist_performance_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Artist performance classes
# ============================================================

artist_performance_df[
    "artist_performance_class"
] = "No Artist Performance Evidence"

if valid_artist_performance_mask.any():

    q25 = artist_performance_df.loc[
        valid_artist_performance_mask,
        "artist_performance_score"
    ].quantile(0.25)

    q50 = artist_performance_df.loc[
        valid_artist_performance_mask,
        "artist_performance_score"
    ].quantile(0.50)

    q75 = artist_performance_df.loc[
        valid_artist_performance_mask,
        "artist_performance_score"
    ].quantile(0.75)

    q90 = artist_performance_df.loc[
        valid_artist_performance_mask,
        "artist_performance_score"
    ].quantile(0.90)

    artist_performance_df.loc[
        valid_artist_performance_mask
        & (
            artist_performance_df[
                "artist_performance_score"
            ] <= q25
        ),
        "artist_performance_class"
    ] = "Low Artist Performance"

    artist_performance_df.loc[
        valid_artist_performance_mask
        & (
            artist_performance_df[
                "artist_performance_score"
            ] > q25
        )
        & (
            artist_performance_df[
                "artist_performance_score"
            ] <= q50
        ),
        "artist_performance_class"
    ] = "Moderate Artist Performance"

    artist_performance_df.loc[
        valid_artist_performance_mask
        & (
            artist_performance_df[
                "artist_performance_score"
            ] > q50
        )
        & (
            artist_performance_df[
                "artist_performance_score"
            ] <= q75
        ),
        "artist_performance_class"
    ] = "High Artist Performance"

    artist_performance_df.loc[
        valid_artist_performance_mask
        & (
            artist_performance_df[
                "artist_performance_score"
            ] > q75
        )
        & (
            artist_performance_df[
                "artist_performance_score"
            ] <= q90
        ),
        "artist_performance_class"
    ] = "Very High Artist Performance"

    artist_performance_df.loc[
        valid_artist_performance_mask
        & (
            artist_performance_df[
                "artist_performance_score"
            ] > q90
        ),
        "artist_performance_class"
    ] = "Exceptional Artist Performance"

# ============================================================
# Artist-level summary
# ============================================================

artist_level_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Artists assessed",
            "Observed Value": len(
                artist_performance_df
            ),
        },
        {
            "Metric": "Artists with performance score",
            "Observed Value": int(
                valid_artist_performance_mask.sum()
            ),
        },
        {
            "Metric": "Median releases per artist",
            "Observed Value": (
                artist_performance_df[
                    "artist_release_count"
                ].median()
            ),
        },
        {
            "Metric": "Mean releases per artist",
            "Observed Value": (
                artist_performance_df[
                    "artist_release_count"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum releases for one artist",
            "Observed Value": (
                artist_performance_df[
                    "artist_release_count"
                ].max()
            ),
        },
        {
            "Metric": "Median artist performance score",
            "Observed Value": (
                artist_performance_df[
                    "artist_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean artist performance score",
            "Observed Value": (
                artist_performance_df[
                    "artist_performance_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum artist performance score",
            "Observed Value": (
                artist_performance_df[
                    "artist_performance_score"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP artist-level performance")
print("=" * 80)

display(
    artist_level_summary_df
)

# ============================================================
# Component coverage
# ============================================================

artist_component_coverage_rows = []

for field in available_artist_performance_fields:

    aggregated_field = f"artist_avg_{field}"

    artist_component_coverage_rows.append(
        {
            "Component": aggregated_field,
            "Artists Available": int(
                artist_performance_df[
                    aggregated_field
                ].notna().sum()
            ),
            "Coverage (%)": round(
                artist_performance_df[
                    aggregated_field
                ].notna().mean()
                * 100,
                6
            ),
        }
    )

artist_component_coverage_df = pd.DataFrame(
    artist_component_coverage_rows
)

print()
print("Artist performance component coverage")
print("=" * 80)

display(
    artist_component_coverage_df
)

# ============================================================
# Artist performance class distribution
# ============================================================

artist_performance_class_distribution_df = (
    artist_performance_df[
        "artist_performance_class"
    ]
    .value_counts()
    .rename_axis(
        "Artist Performance Class"
    )
    .reset_index(
        name="Artists"
    )
)

artist_performance_class_distribution_df[
    "Share (%)"
] = (
    artist_performance_class_distribution_df[
        "Artists"
    ]
    / len(
        artist_performance_df
    )
    * 100
)

print()
print("Artist-performance class distribution")
print("=" * 80)

display(
    artist_performance_class_distribution_df
)

# ============================================================
# Highest-performing artists
# ============================================================

top_artist_performance_df = (
    artist_performance_df[
        [
            "Artist",
            "artist_normalised",
            "artist_release_count",
            *artist_component_columns,
            "artist_components_available",
            "artist_evidence_coverage_pct",
            "artist_performance_score",
            "artist_performance_percentile",
            "artist_performance_rank",
            "artist_performance_class",
        ]
    ]
    .sort_values(
        by="artist_performance_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP artist-level performance")
print("=" * 80)

display(
    top_artist_performance_df
)

# ============================================================
# Attach artist baseline back to release-level dataset
# ============================================================

artist_context_df = artist_context_df.merge(
    artist_performance_df[
        [
            "artist_normalised",
            "artist_release_count",
            "artist_performance_score",
            "artist_performance_percentile",
            "artist_performance_rank",
            "artist_performance_class",
            "artist_evidence_coverage_pct",
        ]
    ],
    on="artist_normalised",
    how="left"
)

# ============================================================
# Validation
# ============================================================

artist_level_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8 completed",
            "Passed": section_8_complete,
        },
        {
            "Validation Area": "Artist identity available",
            "Passed": (
                "artist_normalised"
                in artist_context_df.columns
            ),
        },
        {
            "Validation Area": "Artist-level aggregation prepared",
            "Passed": (
                len(
                    artist_performance_df
                ) > 0
            ),
        },
        {
            "Validation Area": "Artist release counts calculated",
            "Passed": artist_performance_df[
                "artist_release_count"
            ].notna().all(),
        },
        {
            "Validation Area": "Artist performance score calculated",
            "Passed": artist_performance_df[
                "artist_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist performance classes assigned",
            "Passed": artist_performance_df[
                "artist_performance_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Artist baseline attached to releases",
            "Passed": (
                "artist_performance_score"
                in artist_context_df.columns
            ),
        },
    ]
)

print()
print("Artist-level performance validation")
print("=" * 80)

display(
    artist_level_validation_df
)

# ============================================================
# Completion
# ============================================================

section_9_1_complete = bool(
    artist_level_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 9.1 completion status: "
    f"{section_9_1_complete}"
)

if not section_9_1_complete:
    raise RuntimeError(
        "Artist-Level Performance analysis did not complete successfully."
    )
```

### Interpretation

The PMIP Artist-Level Performance assessment completed successfully.

A total of 1,997 normalised artist identities were assessed, and every artist received an artist-performance score.

The median number of represented releases per artist is 1, while the mean is approximately 2.30 releases.

The maximum represented catalogue contains 63 releases for a single artist.

This distribution shows that most artists are represented by only a small number of releases, while a smaller group of artists contributes substantially larger catalogues.

This distinction is important because artist baselines based on several releases are generally supported by more evidence than baselines based on only one release.

Component coverage differs across the artist population.

The final artist-performance score therefore uses the available release-level performance dimensions for each artist rather than requiring every artist to have complete evidence across all domains.

The resulting artist baseline has now been attached back to the release-level dataset.

Section 9.1 therefore establishes the artist-level performance context required to determine whether an individual release is performing above, below or broadly in line with its artist's usual PMIP baseline.

## 9.2 Release vs Artist Baseline

### Purpose

This subsection compares the performance of each release with the broader performance baseline of its artist.

To avoid circular comparison, the artist baseline is recalculated for each release using the artist's other represented releases only.

This means that the release being evaluated does not contribute to the baseline against which it is compared.

The assessment will examine:

- release-level performance across the main PMIP performance dimensions;
- the artist's leave-one-release-out baseline;
- the difference between release performance and artist baseline;
- releases performing above or below their artist's usual level;
- the amount of evidence supporting each artist baseline.

This helps PMIP distinguish between strong releases from already strong artists and releases that genuinely outperform their artist's established PMIP baseline.

```python
# ============================================================
# Section 9.2 - Release vs Artist Baseline
# ============================================================

if not section_9_1_complete:
    raise RuntimeError(
        "Section 9.1 must be completed before comparing "
        "releases with artist baselines."
    )

# ============================================================
# Working source
# ============================================================

release_vs_artist_df = artist_context_df.copy()

# ============================================================
# Release-level performance components
# ============================================================

release_performance_components = [
    field
    for field in [
        "streaming_performance_score",
        "chart_performance_score",
        "geographic_reach_score",
        "temporal_comparability_score",
    ]
    if field in release_vs_artist_df.columns
]

if not release_performance_components:
    raise RuntimeError(
        "No release-level performance components are available."
    )

# ============================================================
# Release-level combined performance score
#
# Equal weighting across available PMIP performance dimensions.
# ============================================================

release_vs_artist_df[
    "release_components_available"
] = (
    release_vs_artist_df[
        release_performance_components
    ]
    .notna()
    .sum(axis=1)
)

release_vs_artist_df[
    "release_performance_evidence_pct"
] = (
    release_vs_artist_df[
        "release_components_available"
    ]
    / len(release_performance_components)
    * 100
)

release_vs_artist_df[
    "release_context_performance_score"
] = (
    release_vs_artist_df[
        release_performance_components
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# ============================================================
# Leave-one-release-out artist baselines
#
# For each performance component:
#
# baseline =
# (artist total - current release value)
# /
# (artist valid count - current release contribution)
#
# This prevents a release from contributing to its own baseline.
# ============================================================

baseline_component_columns = []

for component in release_performance_components:

    artist_sum = (
        release_vs_artist_df
        .groupby("artist_normalised")[component]
        .transform("sum")
    )

    artist_count = (
        release_vs_artist_df
        .groupby("artist_normalised")[component]
        .transform("count")
    )

    current_valid = (
        release_vs_artist_df[component]
        .notna()
        .astype(int)
    )

    other_count = (
        artist_count
        - current_valid
    )

    baseline_column = (
        f"artist_baseline_{component}"
    )

    baseline_component_columns.append(
        baseline_column
    )

    release_vs_artist_df[
        baseline_column
    ] = np.where(
        other_count > 0,
        (
            artist_sum
            - release_vs_artist_df[
                component
            ].fillna(0)
        )
        / other_count,
        np.nan,
    )

# ============================================================
# Number of other releases available for artist comparison
# ============================================================

artist_total_release_count = (
    release_vs_artist_df
    .groupby(
        "artist_normalised"
    )["release_id"]
    .transform("count")
)

release_vs_artist_df[
    "artist_baseline_release_count"
] = (
    artist_total_release_count
    - 1
)

# ============================================================
# Artist baseline evidence
# ============================================================

release_vs_artist_df[
    "artist_baseline_components_available"
] = (
    release_vs_artist_df[
        baseline_component_columns
    ]
    .notna()
    .sum(axis=1)
)

release_vs_artist_df[
    "artist_baseline_evidence_pct"
] = (
    release_vs_artist_df[
        "artist_baseline_components_available"
    ]
    / len(
        baseline_component_columns
    )
    * 100
)

# ============================================================
# Combined leave-one-out artist baseline
# ============================================================

release_vs_artist_df[
    "artist_leave_one_out_baseline_score"
] = (
    release_vs_artist_df[
        baseline_component_columns
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

valid_artist_baseline_mask = (
    release_vs_artist_df[
        "release_context_performance_score"
    ].notna()
    & release_vs_artist_df[
        "artist_leave_one_out_baseline_score"
    ].notna()
)

# ============================================================
# Release vs artist baseline difference
#
# Positive:
# release is stronger than artist baseline
#
# Negative:
# release is weaker than artist baseline
# ============================================================

release_vs_artist_df[
    "release_vs_artist_baseline_difference"
] = np.where(
    valid_artist_baseline_mask,
    (
        release_vs_artist_df[
            "release_context_performance_score"
        ]
        - release_vs_artist_df[
            "artist_leave_one_out_baseline_score"
        ]
    ),
    np.nan,
)

# ============================================================
# Relative release-vs-artist score
#
# 50 = release approximately matches artist baseline
# >50 = release stronger
# <50 = release weaker
# ============================================================

release_vs_artist_df[
    "release_vs_artist_baseline_score"
] = np.where(
    valid_artist_baseline_mask,
    (
        50
        + release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ]
        / 2
    ),
    np.nan,
)

release_vs_artist_df[
    "release_vs_artist_baseline_score"
] = (
    release_vs_artist_df[
        "release_vs_artist_baseline_score"
    ]
    .clip(
        lower=0,
        upper=100
    )
)

# ============================================================
# Direction
#
# ±5 score points is treated as broadly aligned.
# ============================================================

release_vs_artist_df[
    "release_vs_artist_direction"
] = "Unavailable"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] > 5
    ),
    "release_vs_artist_direction"
] = "Above Artist Baseline"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] < -5
    ),
    "release_vs_artist_direction"
] = "Below Artist Baseline"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ].between(
            -5,
            5,
            inclusive="both"
        )
    ),
    "release_vs_artist_direction"
] = "Broadly In Line"

# ============================================================
# Release-vs-artist classes
# ============================================================

release_vs_artist_df[
    "release_vs_artist_class"
] = "No Artist Baseline Evidence"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] <= -20
    ),
    "release_vs_artist_class"
] = "Substantially Below Artist Baseline"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] > -20
    )
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] < -5
    ),
    "release_vs_artist_class"
] = "Below Artist Baseline"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ].between(
            -5,
            5,
            inclusive="both"
        )
    ),
    "release_vs_artist_class"
] = "Broadly In Line With Artist"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] > 5
    )
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] < 20
    ),
    "release_vs_artist_class"
] = "Above Artist Baseline"

release_vs_artist_df.loc[
    valid_artist_baseline_mask
    & (
        release_vs_artist_df[
            "release_vs_artist_baseline_difference"
        ] >= 20
    ),
    "release_vs_artist_class"
] = "Substantially Above Artist Baseline"

# ============================================================
# Artist baseline evidence strength
# ============================================================

release_vs_artist_df[
    "artist_baseline_evidence_strength"
] = np.select(
    [
        release_vs_artist_df[
            "artist_baseline_release_count"
        ] >= 4,

        release_vs_artist_df[
            "artist_baseline_release_count"
        ].between(
            2,
            3
        ),

        release_vs_artist_df[
            "artist_baseline_release_count"
        ] == 1,
    ],
    [
        "Strong Baseline Evidence",
        "Moderate Baseline Evidence",
        "Limited Baseline Evidence",
    ],
    default="No Independent Baseline",
)

# ============================================================
# Summary
# ============================================================

release_vs_artist_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_vs_artist_df
            ),
        },
        {
            "Metric": "Releases with independent artist baseline",
            "Observed Value": int(
                valid_artist_baseline_mask.sum()
            ),
        },
        {
            "Metric": "Releases without independent artist baseline",
            "Observed Value": int(
                (~valid_artist_baseline_mask).sum()
            ),
        },
        {
            "Metric": "Median artist baseline releases available",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "artist_baseline_release_count"
                ].median()
            ),
        },
        {
            "Metric": "Median release performance score",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "release_context_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Median artist baseline score",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "artist_leave_one_out_baseline_score"
                ].median()
            ),
        },
        {
            "Metric": "Median release-vs-artist difference",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "release_vs_artist_baseline_difference"
                ].median()
            ),
        },
        {
            "Metric": "Mean release-vs-artist difference",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "release_vs_artist_baseline_difference"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum positive baseline difference",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "release_vs_artist_baseline_difference"
                ].max()
            ),
        },
        {
            "Metric": "Maximum negative baseline difference",
            "Observed Value": (
                release_vs_artist_df.loc[
                    valid_artist_baseline_mask,
                    "release_vs_artist_baseline_difference"
                ].min()
            ),
        },
    ]
)

print("Assessing PMIP release vs artist baseline")
print("=" * 80)

display(
    release_vs_artist_summary_df
)

# ============================================================
# Direction distribution
# ============================================================

release_vs_artist_direction_distribution_df = (
    release_vs_artist_df[
        "release_vs_artist_direction"
    ]
    .value_counts()
    .rename_axis(
        "Release vs Artist Direction"
    )
    .reset_index(
        name="Releases"
    )
)

release_vs_artist_direction_distribution_df[
    "Share (%)"
] = (
    release_vs_artist_direction_distribution_df[
        "Releases"
    ]
    / len(
        release_vs_artist_df
    )
    * 100
)

print()
print("Release-vs-artist direction distribution")
print("=" * 80)

display(
    release_vs_artist_direction_distribution_df
)

# ============================================================
# Class distribution
# ============================================================

release_vs_artist_class_distribution_df = (
    release_vs_artist_df[
        "release_vs_artist_class"
    ]
    .value_counts()
    .rename_axis(
        "Release vs Artist Class"
    )
    .reset_index(
        name="Releases"
    )
)

release_vs_artist_class_distribution_df[
    "Share (%)"
] = (
    release_vs_artist_class_distribution_df[
        "Releases"
    ]
    / len(
        release_vs_artist_df
    )
    * 100
)

print()
print("Release-vs-artist class distribution")
print("=" * 80)

display(
    release_vs_artist_class_distribution_df
)

# ============================================================
# Artist baseline evidence distribution
# ============================================================

artist_baseline_evidence_distribution_df = (
    release_vs_artist_df[
        "artist_baseline_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Artist Baseline Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

artist_baseline_evidence_distribution_df[
    "Share (%)"
] = (
    artist_baseline_evidence_distribution_df[
        "Releases"
    ]
    / len(
        release_vs_artist_df
    )
    * 100
)

print()
print("Artist baseline evidence distribution")
print("=" * 80)

display(
    artist_baseline_evidence_distribution_df
)

# ============================================================
# Strongest artist-baseline outperformance
# ============================================================

top_release_vs_artist_df = (
    release_vs_artist_df.loc[
        valid_artist_baseline_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "artist_baseline_release_count",
            "release_context_performance_score",
            "artist_leave_one_out_baseline_score",
            "release_vs_artist_baseline_difference",
            "release_vs_artist_baseline_score",
            "artist_baseline_evidence_strength",
            "release_vs_artist_direction",
            "release_vs_artist_class",
        ],
    ]
    .sort_values(
        by="release_vs_artist_baseline_difference",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest artist-baseline outperformance")
print("=" * 80)

display(
    top_release_vs_artist_df
)

# ============================================================
# Validation
# ============================================================

release_vs_artist_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.1 completed",
            "Passed": section_9_1_complete,
        },
        {
            "Validation Area": "Release performance score prepared",
            "Passed": release_vs_artist_df[
                "release_context_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Leave-one-out artist baseline prepared",
            "Passed": release_vs_artist_df[
                "artist_leave_one_out_baseline_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Independent artist baselines available",
            "Passed": valid_artist_baseline_mask.any(),
        },
        {
            "Validation Area": "Release-vs-artist difference calculated",
            "Passed": release_vs_artist_df[
                "release_vs_artist_baseline_difference"
            ].notna().any(),
        },
        {
            "Validation Area": "Release-vs-artist score calculated",
            "Passed": release_vs_artist_df[
                "release_vs_artist_baseline_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist baseline evidence classified",
            "Passed": release_vs_artist_df[
                "artist_baseline_evidence_strength"
            ].notna().all(),
        },
        {
            "Validation Area": "Release-vs-artist classes assigned",
            "Passed": release_vs_artist_df[
                "release_vs_artist_class"
            ].notna().all(),
        },
    ]
)

print()
print("Release vs Artist Baseline validation")
print("=" * 80)

display(
    release_vs_artist_validation_df
)

# ============================================================
# Completion
# ============================================================

section_9_2_complete = bool(
    release_vs_artist_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 9.2 completion status: "
    f"{section_9_2_complete}"
)

if not section_9_2_complete:
    raise RuntimeError(
        "Release vs Artist Baseline analysis "
        "did not complete successfully."
    )
```

Section 9.2 therefore provides a robust release-specific artist baseline and prepares the artist context required for analysing Artist Growth Context in Section 9.3.

## 9.3 Artist Growth Context

### Purpose

This subsection evaluates whether an artist's represented releases show signs of improving, declining or remaining broadly stable over time.

The aim is to measure artist growth using the sequence of release-level PMIP performance scores rather than relying only on the artist's overall average performance.

The assessment will examine:

- the chronological order of each artist's releases;
- change between earlier and later release performance;
- average performance of earlier releases compared with more recent releases;
- the direction and magnitude of artist-level performance change;
- the amount of evidence supporting each artist growth estimate.

This provides PMIP with a growth-oriented artist context that helps distinguish artists whose recent releases are strengthening from artists whose performance is stable or weakening.

```python
# ============================================================
# Section 9.3 - Artist Growth Context
# ============================================================

if not section_9_2_complete:
    raise RuntimeError(
        "Section 9.2 must be completed before assessing "
        "Artist Growth Context."
    )

# ============================================================
# Working source
# ============================================================

artist_growth_release_df = release_vs_artist_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_artist_growth_fields = [
    "artist_normalised",
    "release_id",
    "release_date",
    "release_context_performance_score",
]

missing_artist_growth_fields = [
    field
    for field in required_artist_growth_fields
    if field not in artist_growth_release_df.columns
]

if missing_artist_growth_fields:
    raise RuntimeError(
        "Required Artist Growth Context fields are missing: "
        f"{missing_artist_growth_fields}"
    )

# ============================================================
# Parse release dates
# ============================================================

artist_growth_release_df[
    "artist_growth_release_date"
] = pd.to_datetime(
    artist_growth_release_df[
        "release_date"
    ],
    errors="coerce"
)

# ============================================================
# Keep valid growth-analysis records
# ============================================================

valid_artist_growth_release_mask = (
    artist_growth_release_df[
        "artist_growth_release_date"
    ].notna()
    & artist_growth_release_df[
        "release_context_performance_score"
    ].notna()
)

artist_growth_valid_df = (
    artist_growth_release_df.loc[
        valid_artist_growth_release_mask
    ]
    .copy()
)

# ============================================================
# Sort releases chronologically within artist
# ============================================================

artist_growth_valid_df = (
    artist_growth_valid_df
    .sort_values(
        by=[
            "artist_normalised",
            "artist_growth_release_date",
            "release_id",
        ]
    )
    .reset_index(drop=True)
)

# ============================================================
# Artist-level growth calculations
# ============================================================

artist_growth_rows = []

for artist_id, group in artist_growth_valid_df.groupby(
    "artist_normalised",
    dropna=False
):

    group = group.sort_values(
        by=[
            "artist_growth_release_date",
            "release_id",
        ]
    ).copy()

    release_count = len(group)

    artist_label = (
        group["Artist"].iloc[0]
        if "Artist" in group.columns
        else artist_id
    )

    first_release_date = (
        group[
            "artist_growth_release_date"
        ].min()
    )

    latest_release_date = (
        group[
            "artist_growth_release_date"
        ].max()
    )

    first_performance = (
        group[
            "release_context_performance_score"
        ].iloc[0]
    )

    latest_performance = (
        group[
            "release_context_performance_score"
        ].iloc[-1]
    )

    # ========================================================
    # First-to-latest growth
    # ========================================================

    if release_count >= 2:

        first_to_latest_change = (
            latest_performance
            - first_performance
        )

    else:

        first_to_latest_change = np.nan

    # ========================================================
    # Earlier vs recent release performance
    #
    # Split the artist catalogue chronologically into two halves.
    # ========================================================

    if release_count >= 2:

        split_index = release_count // 2

        earlier_group = group.iloc[
            :split_index
        ]

        recent_group = group.iloc[
            split_index:
        ]

        earlier_average = (
            earlier_group[
                "release_context_performance_score"
            ].mean()
        )

        recent_average = (
            recent_group[
                "release_context_performance_score"
            ].mean()
        )

        recent_vs_earlier_change = (
            recent_average
            - earlier_average
        )

    else:

        earlier_average = np.nan
        recent_average = np.nan
        recent_vs_earlier_change = np.nan

    # ========================================================
    # Linear growth slope
    #
    # Measures average performance-score movement per year.
    # Requires at least 2 distinct dates.
    # ========================================================

    distinct_dates = (
        group[
            "artist_growth_release_date"
        ]
        .nunique()
    )

    if (
        release_count >= 2
        and distinct_dates >= 2
    ):

        elapsed_days = (
            group[
                "artist_growth_release_date"
            ]
            - first_release_date
        ).dt.days.astype(float)

        elapsed_years = (
            elapsed_days
            / 365.25
        )

        x = elapsed_years.to_numpy()
        y = (
            group[
                "release_context_performance_score"
            ]
            .astype(float)
            .to_numpy()
        )

        if np.ptp(x) > 0:

            growth_slope_per_year = (
                np.polyfit(
                    x,
                    y,
                    1
                )[0]
            )

        else:

            growth_slope_per_year = np.nan

    else:

        growth_slope_per_year = np.nan

    # ========================================================
    # Artist observation span
    # ========================================================

    artist_release_span_days = (
        latest_release_date
        - first_release_date
    ).days

    artist_growth_rows.append(
        {
            "artist_normalised": artist_id,
            "Artist": artist_label,
            "artist_growth_release_count": release_count,
            "artist_growth_first_release_date": first_release_date,
            "artist_growth_latest_release_date": latest_release_date,
            "artist_growth_span_days": artist_release_span_days,
            "artist_first_release_performance": first_performance,
            "artist_latest_release_performance": latest_performance,
            "artist_first_to_latest_change": first_to_latest_change,
            "artist_earlier_release_average": earlier_average,
            "artist_recent_release_average": recent_average,
            "artist_recent_vs_earlier_change": recent_vs_earlier_change,
            "artist_growth_slope_per_year": growth_slope_per_year,
        }
    )

artist_growth_df = pd.DataFrame(
    artist_growth_rows
)

# ============================================================
# Growth evidence strength
# ============================================================

artist_growth_df[
    "artist_growth_evidence_strength"
] = np.select(
    [
        artist_growth_df[
            "artist_growth_release_count"
        ] >= 5,

        artist_growth_df[
            "artist_growth_release_count"
        ].between(
            3,
            4
        ),

        artist_growth_df[
            "artist_growth_release_count"
        ] == 2,
    ],
    [
        "Strong Growth Evidence",
        "Moderate Growth Evidence",
        "Limited Growth Evidence",
    ],
    default="No Independent Growth Evidence",
)

# ============================================================
# Valid growth mask
# ============================================================

valid_artist_growth_mask = (
    artist_growth_df[
        "artist_growth_release_count"
    ] >= 2
)

# ============================================================
# Growth component preparation
#
# We combine:
# - first-to-latest change
# - recent-vs-earlier change
# - linear slope per year
#
# Each component is converted to a percentile so different
# numerical scales can be combined safely.
# ============================================================

growth_component_source_columns = [
    "artist_first_to_latest_change",
    "artist_recent_vs_earlier_change",
    "artist_growth_slope_per_year",
]

growth_component_percentile_columns = []

for column in growth_component_source_columns:

    percentile_column = (
        f"{column}_percentile"
    )

    growth_component_percentile_columns.append(
        percentile_column
    )

    artist_growth_df[
        percentile_column
    ] = np.nan

    valid_component_mask = (
        artist_growth_df[
            column
        ].notna()
    )

    if valid_component_mask.any():

        artist_growth_df.loc[
            valid_component_mask,
            percentile_column
        ] = (
            artist_growth_df.loc[
                valid_component_mask,
                column
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

# ============================================================
# Artist growth score
#
# 50 represents approximately neutral relative growth.
# Higher = stronger positive growth context.
# Lower = weaker / declining context.
# ============================================================

artist_growth_df[
    "artist_growth_components_available"
] = (
    artist_growth_df[
        growth_component_percentile_columns
    ]
    .notna()
    .sum(axis=1)
)

artist_growth_df[
    "artist_growth_evidence_coverage_pct"
] = (
    artist_growth_df[
        "artist_growth_components_available"
    ]
    / len(
        growth_component_percentile_columns
    )
    * 100
)

artist_growth_df[
    "artist_growth_score"
] = (
    artist_growth_df[
        growth_component_percentile_columns
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

artist_growth_df.loc[
    ~valid_artist_growth_mask,
    "artist_growth_score"
] = np.nan

# ============================================================
# Growth percentile and rank
# ============================================================

valid_artist_growth_score_mask = (
    artist_growth_df[
        "artist_growth_score"
    ].notna()
)

artist_growth_df[
    "artist_growth_percentile"
] = np.nan

if valid_artist_growth_score_mask.any():

    artist_growth_df.loc[
        valid_artist_growth_score_mask,
        "artist_growth_percentile"
    ] = (
        artist_growth_df.loc[
            valid_artist_growth_score_mask,
            "artist_growth_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

artist_growth_df[
    "artist_growth_rank"
] = (
    artist_growth_df[
        "artist_growth_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Growth direction
#
# Use recent-vs-earlier change as the most intuitive direction.
# ±5 points = broadly stable.
# ============================================================

artist_growth_df[
    "artist_growth_direction"
] = "Unavailable"

artist_growth_df.loc[
    valid_artist_growth_mask
    & (
        artist_growth_df[
            "artist_recent_vs_earlier_change"
        ] > 5
    ),
    "artist_growth_direction"
] = "Growing"

artist_growth_df.loc[
    valid_artist_growth_mask
    & (
        artist_growth_df[
            "artist_recent_vs_earlier_change"
        ] < -5
    ),
    "artist_growth_direction"
] = "Declining"

artist_growth_df.loc[
    valid_artist_growth_mask
    & (
        artist_growth_df[
            "artist_recent_vs_earlier_change"
        ].between(
            -5,
            5,
            inclusive="both"
        )
    ),
    "artist_growth_direction"
] = "Stable"

# ============================================================
# Growth classes
#
# Relative PMIP growth-score classes.
# ============================================================

artist_growth_df[
    "artist_growth_class"
] = "No Artist Growth Evidence"

if valid_artist_growth_score_mask.any():

    q25 = artist_growth_df.loc[
        valid_artist_growth_score_mask,
        "artist_growth_score"
    ].quantile(0.25)

    q50 = artist_growth_df.loc[
        valid_artist_growth_score_mask,
        "artist_growth_score"
    ].quantile(0.50)

    q75 = artist_growth_df.loc[
        valid_artist_growth_score_mask,
        "artist_growth_score"
    ].quantile(0.75)

    q90 = artist_growth_df.loc[
        valid_artist_growth_score_mask,
        "artist_growth_score"
    ].quantile(0.90)

    artist_growth_df.loc[
        valid_artist_growth_score_mask
        & (
            artist_growth_df[
                "artist_growth_score"
            ] <= q25
        ),
        "artist_growth_class"
    ] = "Strong Artist Decline Context"

    artist_growth_df.loc[
        valid_artist_growth_score_mask
        & (
            artist_growth_df[
                "artist_growth_score"
            ] > q25
        )
        & (
            artist_growth_df[
                "artist_growth_score"
            ] <= q50
        ),
        "artist_growth_class"
    ] = "Weak Artist Growth Context"

    artist_growth_df.loc[
        valid_artist_growth_score_mask
        & (
            artist_growth_df[
                "artist_growth_score"
            ] > q50
        )
        & (
            artist_growth_df[
                "artist_growth_score"
            ] <= q75
        ),
        "artist_growth_class"
    ] = "Positive Artist Growth Context"

    artist_growth_df.loc[
        valid_artist_growth_score_mask
        & (
            artist_growth_df[
                "artist_growth_score"
            ] > q75
        )
        & (
            artist_growth_df[
                "artist_growth_score"
            ] <= q90
        ),
        "artist_growth_class"
    ] = "Strong Artist Growth Context"

    artist_growth_df.loc[
        valid_artist_growth_score_mask
        & (
            artist_growth_df[
                "artist_growth_score"
            ] > q90
        ),
        "artist_growth_class"
    ] = "Exceptional Artist Growth Context"

# ============================================================
# Summary
# ============================================================

artist_growth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Artists assessed",
            "Observed Value": len(
                artist_growth_df
            ),
        },
        {
            "Metric": "Artists with independent growth evidence",
            "Observed Value": int(
                valid_artist_growth_mask.sum()
            ),
        },
        {
            "Metric": "Artists without independent growth evidence",
            "Observed Value": int(
                (~valid_artist_growth_mask).sum()
            ),
        },
        {
            "Metric": "Median releases per growth-assessed artist",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_mask,
                    "artist_growth_release_count"
                ].median()
            ),
        },
        {
            "Metric": "Median first-to-latest performance change",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_mask,
                    "artist_first_to_latest_change"
                ].median()
            ),
        },
        {
            "Metric": "Median recent-vs-earlier performance change",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_mask,
                    "artist_recent_vs_earlier_change"
                ].median()
            ),
        },
        {
            "Metric": "Median growth slope per year",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_mask,
                    "artist_growth_slope_per_year"
                ].median()
            ),
        },
        {
            "Metric": "Median artist growth score",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_score_mask,
                    "artist_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean artist growth score",
            "Observed Value": (
                artist_growth_df.loc[
                    valid_artist_growth_score_mask,
                    "artist_growth_score"
                ].mean()
            ),
        },
    ]
)

print("Assessing PMIP artist growth context")
print("=" * 80)

display(
    artist_growth_summary_df
)

# ============================================================
# Growth direction distribution
# ============================================================

artist_growth_direction_distribution_df = (
    artist_growth_df[
        "artist_growth_direction"
    ]
    .value_counts()
    .rename_axis(
        "Artist Growth Direction"
    )
    .reset_index(
        name="Artists"
    )
)

artist_growth_direction_distribution_df[
    "Share (%)"
] = (
    artist_growth_direction_distribution_df[
        "Artists"
    ]
    / len(
        artist_growth_df
    )
    * 100
)

print()
print("Artist-growth direction distribution")
print("=" * 80)

display(
    artist_growth_direction_distribution_df
)

# ============================================================
# Growth class distribution
# ============================================================

artist_growth_class_distribution_df = (
    artist_growth_df[
        "artist_growth_class"
    ]
    .value_counts()
    .rename_axis(
        "Artist Growth Class"
    )
    .reset_index(
        name="Artists"
    )
)

artist_growth_class_distribution_df[
    "Share (%)"
] = (
    artist_growth_class_distribution_df[
        "Artists"
    ]
    / len(
        artist_growth_df
    )
    * 100
)

print()
print("Artist-growth class distribution")
print("=" * 80)

display(
    artist_growth_class_distribution_df
)

# ============================================================
# Growth evidence distribution
# ============================================================

artist_growth_evidence_distribution_df = (
    artist_growth_df[
        "artist_growth_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Artist Growth Evidence"
    )
    .reset_index(
        name="Artists"
    )
)

artist_growth_evidence_distribution_df[
    "Share (%)"
] = (
    artist_growth_evidence_distribution_df[
        "Artists"
    ]
    / len(
        artist_growth_df
    )
    * 100
)

print()
print("Artist-growth evidence distribution")
print("=" * 80)

display(
    artist_growth_evidence_distribution_df
)

# ============================================================
# Artists with strongest growth context
# ============================================================

top_artist_growth_df = (
    artist_growth_df[
        [
            "Artist",
            "artist_normalised",
            "artist_growth_release_count",
            "artist_growth_span_days",
            "artist_first_release_performance",
            "artist_latest_release_performance",
            "artist_first_to_latest_change",
            "artist_earlier_release_average",
            "artist_recent_release_average",
            "artist_recent_vs_earlier_change",
            "artist_growth_slope_per_year",
            "artist_growth_evidence_strength",
            "artist_growth_score",
            "artist_growth_percentile",
            "artist_growth_rank",
            "artist_growth_direction",
            "artist_growth_class",
        ]
    ]
    .sort_values(
        by="artist_growth_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artists with strongest PMIP growth context")
print("=" * 80)

display(
    top_artist_growth_df
)

# ============================================================
# Attach artist-growth context back to release-level dataset
# ============================================================

artist_growth_merge_columns = [
    "artist_normalised",
    "artist_growth_release_count",
    "artist_growth_span_days",
    "artist_first_to_latest_change",
    "artist_recent_vs_earlier_change",
    "artist_growth_slope_per_year",
    "artist_growth_evidence_strength",
    "artist_growth_evidence_coverage_pct",
    "artist_growth_score",
    "artist_growth_percentile",
    "artist_growth_rank",
    "artist_growth_direction",
    "artist_growth_class",
]

artist_context_growth_df = (
    release_vs_artist_df
    .merge(
        artist_growth_df[
            artist_growth_merge_columns
        ],
        on="artist_normalised",
        how="left"
    )
)

# ============================================================
# Validation
# ============================================================

artist_growth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.2 completed",
            "Passed": section_9_2_complete,
        },
        {
            "Validation Area": "Release dates available",
            "Passed": artist_growth_valid_df[
                "artist_growth_release_date"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist chronological histories prepared",
            "Passed": len(
                artist_growth_df
            ) > 0,
        },
        {
            "Validation Area": "Independent growth evidence available",
            "Passed": valid_artist_growth_mask.any(),
        },
        {
            "Validation Area": "First-to-latest change calculated",
            "Passed": artist_growth_df[
                "artist_first_to_latest_change"
            ].notna().any(),
        },
        {
            "Validation Area": "Recent-vs-earlier change calculated",
            "Passed": artist_growth_df[
                "artist_recent_vs_earlier_change"
            ].notna().any(),
        },
        {
            "Validation Area": "Growth slope calculated",
            "Passed": artist_growth_df[
                "artist_growth_slope_per_year"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist growth score calculated",
            "Passed": artist_growth_df[
                "artist_growth_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist growth classes assigned",
            "Passed": artist_growth_df[
                "artist_growth_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Artist growth context attached to releases",
            "Passed": (
                "artist_growth_score"
                in artist_context_growth_df.columns
            ),
        },
    ]
)

print()
print("Artist Growth Context validation")
print("=" * 80)

display(
    artist_growth_validation_df
)

# ============================================================
# Completion
# ============================================================

section_9_3_complete = bool(
    artist_growth_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 9.3 completion status: "
    f"{section_9_3_complete}"
)

if not section_9_3_complete:
    raise RuntimeError(
        "Artist Growth Context analysis "
        "did not complete successfully."
    )
```

### Interpretation

The PMIP Artist Growth Context assessment completed successfully.

A total of 1,997 normalised artists were assessed.

Independent artist-growth evidence is available for 660 artists, while 1,337 artists do not have enough represented releases to support an independent growth analysis.

The median number of releases among growth-assessed artists is 3.

This means that most growth estimates are based on relatively small artist catalogues, so the strength of the evidence must be considered alongside the growth score itself.

The median first-to-latest performance change is approximately -13.98 points.

The median recent-versus-earlier performance change is approximately -9.98 points.

The median growth slope is approximately -4.59 performance points per year.

Together, these measures suggest that the typical artist with sufficient growth evidence shows weaker recent PMIP performance relative to earlier represented releases.

However, this should not be interpreted as a universal decline in artist careers.

The result describes only the releases represented in the PMIP dataset and can be affected by catalogue composition, release timing and the types of releases available for each artist.

The median artist-growth score is approximately 47.07, while the mean is approximately 50.31.

This shows that the relative growth scoring remains reasonably centred across the assessed artist population even though the underlying raw growth indicators are slightly negative overall.

The growth-direction distribution provides a more direct view of artist trajectories.

A total of 403 artists are classified as `Declining`, while 121 are classified as `Growing`.

A further 136 artists are classified as `Stable`.

For 1,337 artists, growth direction remains `Unavailable` because there is only one represented release and therefore no independent growth history.

The relative growth-class distribution separates the 660 artists with valid growth evidence into progressively stronger or weaker contexts.

A total of 165 artists fall into `Strong Artist Decline Context`.

Another 165 are classified as `Weak Artist Growth Context`, while 165 are classified as `Positive Artist Growth Context`.

At the stronger end, 99 artists are classified as `Strong Artist Growth Context`, and 66 artists fall into `Exceptional Artist Growth Context`.

The evidence-strength distribution is also important.

A total of 279 artists have `Limited Growth Evidence`, meaning their trend is based on only two represented releases.

A further 180 artists have `Moderate Growth Evidence`, based on three or four releases.

A total of 201 artists have `Strong Growth Evidence`, supported by at least five represented releases.

The remaining 1,337 artists have `No Independent Growth Evidence`.

The final `artist_growth_score` combines three complementary growth signals:

- first-to-latest performance change;
- recent-versus-earlier performance change;
- chronological growth slope.

These measures are converted into relative percentile positions before being combined, preventing differences in raw numerical scale from dominating the result.

The artist-growth score should therefore be interpreted as a relative PMIP growth-context measure rather than as an absolute prediction of future artist success.

Higher scores indicate that an artist's more recent represented releases generally occupy a stronger position than their earlier releases.

Lower scores indicate weaker recent performance relative to the artist's represented history.

The score should always be considered together with `artist_growth_evidence_strength`, because a strong apparent trend supported by only two releases is less reliable than a similar trend supported by a larger catalogue.

Section 9.3 therefore adds a dynamic artist-development signal to PMIP and prepares the artist context required for the Artist Momentum Context assessment in Section 9.4.

## 9.4 Artist Momentum Context

### Purpose

This subsection evaluates the artist's most recent performance momentum.

While Section 9.3 measures the broader direction of artist growth across the represented catalogue, Artist Momentum Context focuses more heavily on the artist's latest releases.

The assessment will examine:

- performance of the artist's most recent release;
- average performance of recent releases;
- change between the latest release and the immediately previous release;
- recent performance compared with the artist's earlier catalogue;
- consistency of recent positive or negative movement;
- the amount of evidence supporting the momentum estimate.

The goal is to identify artists whose recent releases are gaining strength, losing momentum or remaining broadly stable.

This gives PMIP a shorter-term artist signal that complements the longer-term Artist Growth Context developed in Section 9.3.

```python
# ============================================================
# Section 9.4 - Artist Momentum Context
# ============================================================

if not section_9_3_complete:
    raise RuntimeError(
        "Section 9.3 must be completed before assessing "
        "Artist Momentum Context."
    )

# ============================================================
# Working source
# ============================================================

artist_momentum_release_df = artist_context_growth_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_artist_momentum_fields = [
    "artist_normalised",
    "release_id",
    "release_date",
    "release_context_performance_score",
]

missing_artist_momentum_fields = [
    field
    for field in required_artist_momentum_fields
    if field not in artist_momentum_release_df.columns
]

if missing_artist_momentum_fields:
    raise RuntimeError(
        "Required Artist Momentum Context fields are missing: "
        f"{missing_artist_momentum_fields}"
    )

# ============================================================
# Parse dates
# ============================================================

artist_momentum_release_df[
    "artist_momentum_release_date"
] = pd.to_datetime(
    artist_momentum_release_df[
        "release_date"
    ],
    errors="coerce"
)

# ============================================================
# Keep valid momentum-analysis records
# ============================================================

valid_artist_momentum_release_mask = (
    artist_momentum_release_df[
        "artist_momentum_release_date"
    ].notna()
    & artist_momentum_release_df[
        "release_context_performance_score"
    ].notna()
)

artist_momentum_valid_df = (
    artist_momentum_release_df.loc[
        valid_artist_momentum_release_mask
    ]
    .copy()
)

# ============================================================
# Sort chronologically
# ============================================================

artist_momentum_valid_df = (
    artist_momentum_valid_df
    .sort_values(
        by=[
            "artist_normalised",
            "artist_momentum_release_date",
            "release_id",
        ]
    )
    .reset_index(drop=True)
)

# ============================================================
# Artist-level momentum calculations
# ============================================================

artist_momentum_rows = []

for artist_id, group in artist_momentum_valid_df.groupby(
    "artist_normalised",
    dropna=False
):

    group = (
        group
        .sort_values(
            by=[
                "artist_momentum_release_date",
                "release_id",
            ]
        )
        .reset_index(drop=True)
    )

    release_count = len(group)

    artist_label = (
        group["Artist"].iloc[0]
        if "Artist" in group.columns
        else artist_id
    )

    scores = (
        group[
            "release_context_performance_score"
        ]
        .astype(float)
        .to_numpy()
    )

    dates = (
        group[
            "artist_momentum_release_date"
        ]
    )

    # ========================================================
    # Latest release performance
    # ========================================================

    latest_performance = (
        scores[-1]
        if release_count >= 1
        else np.nan
    )

    previous_performance = (
        scores[-2]
        if release_count >= 2
        else np.nan
    )

    latest_vs_previous_change = (
        latest_performance
        - previous_performance
        if release_count >= 2
        else np.nan
    )

    # ========================================================
    # Recent release window
    #
    # Use up to the 3 most recent releases.
    # ========================================================

    recent_window_size = min(
        3,
        release_count
    )

    recent_scores = (
        scores[
            -recent_window_size:
        ]
        if recent_window_size > 0
        else np.array([])
    )

    recent_average_performance = (
        float(
            np.mean(
                recent_scores
            )
        )
        if recent_window_size > 0
        else np.nan
    )

    # ========================================================
    # Earlier catalogue benchmark
    #
    # Earlier releases exclude the recent momentum window.
    # ========================================================

    if release_count > recent_window_size:

        earlier_scores = scores[
            :-recent_window_size
        ]

        earlier_average_performance = float(
            np.mean(
                earlier_scores
            )
        )

        recent_vs_earlier_momentum = (
            recent_average_performance
            - earlier_average_performance
        )

    else:

        earlier_average_performance = np.nan
        recent_vs_earlier_momentum = np.nan

    # ========================================================
    # Recent weighted performance
    #
    # Most recent release receives the greatest weight.
    # ========================================================

    if recent_window_size == 1:

        recent_weights = np.array(
            [1.0]
        )

    elif recent_window_size == 2:

        recent_weights = np.array(
            [0.40, 0.60]
        )

    else:

        recent_weights = np.array(
            [0.20, 0.30, 0.50]
        )

    recent_weighted_performance = (
        float(
            np.average(
                recent_scores,
                weights=recent_weights
            )
        )
        if recent_window_size > 0
        else np.nan
    )

    # ========================================================
    # Recent sequential changes
    #
    # Positive values mean release-to-release improvement.
    # ========================================================

    if recent_window_size >= 2:

        recent_changes = np.diff(
            recent_scores
        )

        positive_recent_changes = int(
            np.sum(
                recent_changes > 5
            )
        )

        negative_recent_changes = int(
            np.sum(
                recent_changes < -5
            )
        )

        stable_recent_changes = int(
            np.sum(
                np.abs(
                    recent_changes
                ) <= 5
            )
        )

        mean_recent_change = float(
            np.mean(
                recent_changes
            )
        )

    else:

        positive_recent_changes = 0
        negative_recent_changes = 0
        stable_recent_changes = 0
        mean_recent_change = np.nan

    # ========================================================
    # Recent observation span
    # ========================================================

    latest_release_date = (
        dates.iloc[-1]
    )

    if recent_window_size >= 2:

        recent_first_date = (
            dates.iloc[
                -recent_window_size
            ]
        )

        recent_momentum_span_days = (
            latest_release_date
            - recent_first_date
        ).days

    else:

        recent_momentum_span_days = 0

    # ========================================================
    # Artist momentum row
    # ========================================================

    artist_momentum_rows.append(
        {
            "artist_normalised": artist_id,
            "Artist": artist_label,
            "artist_momentum_release_count": release_count,
            "artist_recent_window_size": recent_window_size,
            "artist_latest_release_date": latest_release_date,
            "artist_latest_release_performance": latest_performance,
            "artist_previous_release_performance": previous_performance,
            "artist_latest_vs_previous_change": latest_vs_previous_change,
            "artist_recent_average_performance": recent_average_performance,
            "artist_recent_weighted_performance": recent_weighted_performance,
            "artist_earlier_average_performance": earlier_average_performance,
            "artist_recent_vs_earlier_momentum": recent_vs_earlier_momentum,
            "artist_mean_recent_change": mean_recent_change,
            "artist_positive_recent_changes": positive_recent_changes,
            "artist_negative_recent_changes": negative_recent_changes,
            "artist_stable_recent_changes": stable_recent_changes,
            "artist_recent_momentum_span_days": recent_momentum_span_days,
        }
    )

artist_momentum_df = pd.DataFrame(
    artist_momentum_rows
)

# ============================================================
# Momentum evidence strength
#
# Momentum requires at least 2 releases for independent change.
# ============================================================

artist_momentum_df[
    "artist_momentum_evidence_strength"
] = np.select(
    [
        artist_momentum_df[
            "artist_momentum_release_count"
        ] >= 5,

        artist_momentum_df[
            "artist_momentum_release_count"
        ].between(
            3,
            4
        ),

        artist_momentum_df[
            "artist_momentum_release_count"
        ] == 2,
    ],
    [
        "Strong Momentum Evidence",
        "Moderate Momentum Evidence",
        "Limited Momentum Evidence",
    ],
    default="No Independent Momentum Evidence",
)

# ============================================================
# Valid momentum mask
# ============================================================

valid_artist_momentum_mask = (
    artist_momentum_df[
        "artist_momentum_release_count"
    ] >= 2
)

# ============================================================
# Prepare momentum components
#
# 1. Recent weighted performance
# 2. Latest-vs-previous change
# 3. Recent-vs-earlier momentum where available
# 4. Mean recent sequential change
#
# Components are percentile-normalised before combination.
# ============================================================

momentum_source_columns = [
    "artist_recent_weighted_performance",
    "artist_latest_vs_previous_change",
    "artist_recent_vs_earlier_momentum",
    "artist_mean_recent_change",
]

momentum_percentile_columns = []

for column in momentum_source_columns:

    percentile_column = (
        f"{column}_percentile"
    )

    momentum_percentile_columns.append(
        percentile_column
    )

    artist_momentum_df[
        percentile_column
    ] = np.nan

    valid_component_mask = (
        artist_momentum_df[
            column
        ].notna()
        & valid_artist_momentum_mask
    )

    if valid_component_mask.any():

        artist_momentum_df.loc[
            valid_component_mask,
            percentile_column
        ] = (
            artist_momentum_df.loc[
                valid_component_mask,
                column
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

# ============================================================
# Momentum component weighting
#
# Greater emphasis is placed on recent performance and the
# latest release-to-release change.
# ============================================================

momentum_component_weights = {
    "artist_recent_weighted_performance_percentile": 0.35,
    "artist_latest_vs_previous_change_percentile": 0.30,
    "artist_recent_vs_earlier_momentum_percentile": 0.20,
    "artist_mean_recent_change_percentile": 0.15,
}

# ============================================================
# Coverage-aware momentum score
# ============================================================

artist_momentum_weighted_sum = pd.Series(
    0.0,
    index=artist_momentum_df.index
)

artist_momentum_available_weight_sum = pd.Series(
    0.0,
    index=artist_momentum_df.index
)

for column, weight in momentum_component_weights.items():

    component_available_mask = (
        artist_momentum_df[
            column
        ].notna()
        & valid_artist_momentum_mask
    )

    artist_momentum_weighted_sum.loc[
        component_available_mask
    ] += (
        artist_momentum_df.loc[
            component_available_mask,
            column
        ]
        * weight
    )

    artist_momentum_available_weight_sum.loc[
        component_available_mask
    ] += weight

artist_momentum_df[
    "artist_momentum_score"
] = np.where(
    artist_momentum_available_weight_sum > 0,
    (
        artist_momentum_weighted_sum
        / artist_momentum_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Momentum evidence coverage
# ============================================================

artist_momentum_df[
    "artist_momentum_components_available"
] = (
    artist_momentum_df[
        list(
            momentum_component_weights.keys()
        )
    ]
    .notna()
    .sum(axis=1)
)

artist_momentum_df[
    "artist_momentum_evidence_coverage_pct"
] = (
    artist_momentum_df[
        "artist_momentum_components_available"
    ]
    / len(
        momentum_component_weights
    )
    * 100
)

# ============================================================
# Momentum percentile and rank
# ============================================================

valid_artist_momentum_score_mask = (
    artist_momentum_df[
        "artist_momentum_score"
    ].notna()
)

artist_momentum_df[
    "artist_momentum_percentile"
] = np.nan

if valid_artist_momentum_score_mask.any():

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask,
        "artist_momentum_percentile"
    ] = (
        artist_momentum_df.loc[
            valid_artist_momentum_score_mask,
            "artist_momentum_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

artist_momentum_df[
    "artist_momentum_rank"
] = (
    artist_momentum_df[
        "artist_momentum_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Momentum direction
#
# Primary short-term direction uses latest-vs-previous change.
# ±5 points = broadly stable.
# ============================================================

artist_momentum_df[
    "artist_momentum_direction"
] = "Unavailable"

artist_momentum_df.loc[
    valid_artist_momentum_mask
    & (
        artist_momentum_df[
            "artist_latest_vs_previous_change"
        ] > 5
    ),
    "artist_momentum_direction"
] = "Positive Momentum"

artist_momentum_df.loc[
    valid_artist_momentum_mask
    & (
        artist_momentum_df[
            "artist_latest_vs_previous_change"
        ] < -5
    ),
    "artist_momentum_direction"
] = "Negative Momentum"

artist_momentum_df.loc[
    valid_artist_momentum_mask
    & (
        artist_momentum_df[
            "artist_latest_vs_previous_change"
        ].between(
            -5,
            5,
            inclusive="both"
        )
    ),
    "artist_momentum_direction"
] = "Stable Momentum"

# ============================================================
# Momentum classes
# ============================================================

artist_momentum_df[
    "artist_momentum_class"
] = "No Artist Momentum Evidence"

if valid_artist_momentum_score_mask.any():

    q25 = artist_momentum_df.loc[
        valid_artist_momentum_score_mask,
        "artist_momentum_score"
    ].quantile(0.25)

    q50 = artist_momentum_df.loc[
        valid_artist_momentum_score_mask,
        "artist_momentum_score"
    ].quantile(0.50)

    q75 = artist_momentum_df.loc[
        valid_artist_momentum_score_mask,
        "artist_momentum_score"
    ].quantile(0.75)

    q90 = artist_momentum_df.loc[
        valid_artist_momentum_score_mask,
        "artist_momentum_score"
    ].quantile(0.90)

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] <= q25
        ),
        "artist_momentum_class"
    ] = "Weak Artist Momentum"

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] > q25
        )
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] <= q50
        ),
        "artist_momentum_class"
    ] = "Moderate Artist Momentum"

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] > q50
        )
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] <= q75
        ),
        "artist_momentum_class"
    ] = "Positive Artist Momentum"

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] > q75
        )
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] <= q90
        ),
        "artist_momentum_class"
    ] = "Strong Artist Momentum"

    artist_momentum_df.loc[
        valid_artist_momentum_score_mask
        & (
            artist_momentum_df[
                "artist_momentum_score"
            ] > q90
        ),
        "artist_momentum_class"
    ] = "Exceptional Artist Momentum"

# ============================================================
# Summary
# ============================================================

artist_momentum_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Artists assessed",
            "Observed Value": len(
                artist_momentum_df
            ),
        },
        {
            "Metric": "Artists with independent momentum evidence",
            "Observed Value": int(
                valid_artist_momentum_mask.sum()
            ),
        },
        {
            "Metric": "Artists without independent momentum evidence",
            "Observed Value": int(
                (~valid_artist_momentum_mask).sum()
            ),
        },
        {
            "Metric": "Median releases per momentum-assessed artist",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_mask,
                    "artist_momentum_release_count"
                ].median()
            ),
        },
        {
            "Metric": "Median latest-vs-previous change",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_mask,
                    "artist_latest_vs_previous_change"
                ].median()
            ),
        },
        {
            "Metric": "Median recent weighted performance",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_mask,
                    "artist_recent_weighted_performance"
                ].median()
            ),
        },
        {
            "Metric": "Median recent-vs-earlier momentum",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_mask,
                    "artist_recent_vs_earlier_momentum"
                ].median()
            ),
        },
        {
            "Metric": "Median artist momentum score",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_score_mask,
                    "artist_momentum_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean artist momentum score",
            "Observed Value": (
                artist_momentum_df.loc[
                    valid_artist_momentum_score_mask,
                    "artist_momentum_score"
                ].mean()
            ),
        },
    ]
)

print("Assessing PMIP artist momentum context")
print("=" * 80)

display(
    artist_momentum_summary_df
)

# ============================================================
# Momentum direction distribution
# ============================================================

artist_momentum_direction_distribution_df = (
    artist_momentum_df[
        "artist_momentum_direction"
    ]
    .value_counts()
    .rename_axis(
        "Artist Momentum Direction"
    )
    .reset_index(
        name="Artists"
    )
)

artist_momentum_direction_distribution_df[
    "Share (%)"
] = (
    artist_momentum_direction_distribution_df[
        "Artists"
    ]
    / len(
        artist_momentum_df
    )
    * 100
)

print()
print("Artist-momentum direction distribution")
print("=" * 80)

display(
    artist_momentum_direction_distribution_df
)

# ============================================================
# Momentum class distribution
# ============================================================

artist_momentum_class_distribution_df = (
    artist_momentum_df[
        "artist_momentum_class"
    ]
    .value_counts()
    .rename_axis(
        "Artist Momentum Class"
    )
    .reset_index(
        name="Artists"
    )
)

artist_momentum_class_distribution_df[
    "Share (%)"
] = (
    artist_momentum_class_distribution_df[
        "Artists"
    ]
    / len(
        artist_momentum_df
    )
    * 100
)

print()
print("Artist-momentum class distribution")
print("=" * 80)

display(
    artist_momentum_class_distribution_df
)

# ============================================================
# Momentum evidence distribution
# ============================================================

artist_momentum_evidence_distribution_df = (
    artist_momentum_df[
        "artist_momentum_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Artist Momentum Evidence"
    )
    .reset_index(
        name="Artists"
    )
)

artist_momentum_evidence_distribution_df[
    "Share (%)"
] = (
    artist_momentum_evidence_distribution_df[
        "Artists"
    ]
    / len(
        artist_momentum_df
    )
    * 100
)

print()
print("Artist-momentum evidence distribution")
print("=" * 80)

display(
    artist_momentum_evidence_distribution_df
)

# ============================================================
# Artists with strongest momentum
# ============================================================

top_artist_momentum_df = (
    artist_momentum_df[
        [
            "Artist",
            "artist_normalised",
            "artist_momentum_release_count",
            "artist_recent_window_size",
            "artist_latest_release_performance",
            "artist_previous_release_performance",
            "artist_latest_vs_previous_change",
            "artist_recent_average_performance",
            "artist_recent_weighted_performance",
            "artist_earlier_average_performance",
            "artist_recent_vs_earlier_momentum",
            "artist_mean_recent_change",
            "artist_positive_recent_changes",
            "artist_negative_recent_changes",
            "artist_momentum_evidence_strength",
            "artist_momentum_evidence_coverage_pct",
            "artist_momentum_score",
            "artist_momentum_percentile",
            "artist_momentum_rank",
            "artist_momentum_direction",
            "artist_momentum_class",
        ]
    ]
    .sort_values(
        by="artist_momentum_score",
        ascending=False,
        na_position="last"
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artists with strongest PMIP momentum context")
print("=" * 80)

display(
    top_artist_momentum_df
)

# ============================================================
# Attach artist momentum back to release-level dataset
# ============================================================

artist_momentum_merge_columns = [
    "artist_normalised",
    "artist_momentum_release_count",
    "artist_recent_window_size",
    "artist_latest_vs_previous_change",
    "artist_recent_weighted_performance",
    "artist_recent_vs_earlier_momentum",
    "artist_mean_recent_change",
    "artist_momentum_evidence_strength",
    "artist_momentum_evidence_coverage_pct",
    "artist_momentum_score",
    "artist_momentum_percentile",
    "artist_momentum_rank",
    "artist_momentum_direction",
    "artist_momentum_class",
]

artist_context_momentum_df = (
    artist_context_growth_df
    .merge(
        artist_momentum_df[
            artist_momentum_merge_columns
        ],
        on="artist_normalised",
        how="left"
    )
)

# ============================================================
# Validation
# ============================================================

artist_momentum_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.3 completed",
            "Passed": section_9_3_complete,
        },
        {
            "Validation Area": "Artist chronological histories available",
            "Passed": len(
                artist_momentum_valid_df
            ) > 0,
        },
        {
            "Validation Area": "Independent momentum evidence available",
            "Passed": valid_artist_momentum_mask.any(),
        },
        {
            "Validation Area": "Latest-vs-previous change calculated",
            "Passed": artist_momentum_df[
                "artist_latest_vs_previous_change"
            ].notna().any(),
        },
        {
            "Validation Area": "Recent weighted performance calculated",
            "Passed": artist_momentum_df[
                "artist_recent_weighted_performance"
            ].notna().any(),
        },
        {
            "Validation Area": "Recent-vs-earlier momentum calculated",
            "Passed": artist_momentum_df[
                "artist_recent_vs_earlier_momentum"
            ].notna().any(),
        },
        {
            "Validation Area": "Momentum score calculated",
            "Passed": artist_momentum_df[
                "artist_momentum_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Momentum evidence classified",
            "Passed": artist_momentum_df[
                "artist_momentum_evidence_strength"
            ].notna().all(),
        },
        {
            "Validation Area": "Momentum classes assigned",
            "Passed": artist_momentum_df[
                "artist_momentum_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Artist momentum attached to releases",
            "Passed": (
                "artist_momentum_score"
                in artist_context_momentum_df.columns
            ),
        },
    ]
)

print()
print("Artist Momentum Context validation")
print("=" * 80)

display(
    artist_momentum_validation_df
)

# ============================================================
# Completion
# ============================================================

section_9_4_complete = bool(
    artist_momentum_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 9.4 completion status: "
    f"{section_9_4_complete}"
)

if not section_9_4_complete:
    raise RuntimeError(
        "Artist Momentum Context analysis "
        "did not complete successfully."
    )
```

### Interpretation

The PMIP Artist Momentum Context assessment completed successfully.

A total of 1,997 normalised artists were assessed.

Independent artist-momentum evidence is available for 660 artists, while 1,337 artists do not have enough represented releases to support an independent short-term momentum assessment.

The median number of releases among momentum-assessed artists is 3.

This means that many artist momentum estimates are based on relatively small represented catalogues, so momentum strength should always be interpreted together with the associated evidence level.

The median latest-versus-previous performance change is approximately -9.70 points.

This indicates that, for the typical artist with sufficient momentum evidence, the most recent represented release performs somewhat weaker than the immediately previous release.

The median recent weighted performance is approximately 40.01.

The median recent-versus-earlier momentum difference is approximately -13.83 points.

Together, these results suggest that recent represented release performance is generally weaker than earlier catalogue performance for many artists in the current PMIP population.

However, this should not be interpreted as a universal decline in artist careers.

The result reflects only the releases represented within PMIP and may be influenced by catalogue composition, release timing and data availability.

The median artist-momentum score is approximately 48.90, while the mean is approximately 50.24.

This shows that the relative momentum scoring remains broadly centred across artists with valid momentum evidence.

The direction distribution provides a clearer short-term view.

A total of 397 artists are classified as having `Negative Momentum`.

A further 150 artists have `Stable Momentum`, while 113 artists are classified as having `Positive Momentum`.

For 1,337 artists, momentum direction remains `Unavailable` because only one represented release is available.

The momentum class distribution separates artists with valid momentum evidence into relative performance groups.

A total of 165 artists fall into `Weak Artist Momentum`.

Another 165 are classified as `Moderate Artist Momentum`, while 165 are classified as `Positive Artist Momentum`.

At the stronger end, 99 artists fall into `Strong Artist Momentum`, and 66 artists are classified as having `Exceptional Artist Momentum`.

The evidence-strength distribution is also important.

A total of 279 artists have `Limited Momentum Evidence`, meaning their assessment is based on only two represented releases.

A further 180 artists have `Moderate Momentum Evidence`, based on three or four releases.

A total of 201 artists have `Strong Momentum Evidence`, supported by at least five represented releases.

The remaining 1,337 artists have `No Independent Momentum Evidence`.

The final `artist_momentum_score` combines four short-term signals:

- recent weighted release performance;
- latest-versus-previous performance change;
- recent-versus-earlier performance difference;
- average recent sequential performance change.

These measures are converted into percentile positions before being combined so that differences in raw scale do not distort the final result.

Coverage-aware weighting allows artists with partially available momentum components to still receive a score without treating missing evidence as poor momentum.

The resulting momentum score should therefore be interpreted together with `artist_momentum_evidence_strength` and `artist_momentum_evidence_coverage_pct`.

Higher scores indicate stronger recent artist context within the PMIP population.

Lower scores indicate weaker short-term momentum.

Section 9.4 therefore provides PMIP with a recent artist-performance signal that complements the broader Artist Growth Context and prepares the final Artist–Release Performance Relationship analysis in Section 9.5.

## 9.5 Artist–Release Performance Relationship

### Purpose

This subsection brings together the main artist-context signals developed across Section 9 and evaluates how artist strength relates to individual release performance.

The aim is to determine whether a release is:

- performing strongly because it belongs to an already strong artist;
- outperforming the artist's normal baseline;
- supported by positive artist growth;
- supported by positive short-term artist momentum;
- or performing strongly despite weaker artist context.

The assessment combines:

- artist-level performance strength;
- release-versus-artist baseline difference;
- artist growth context;
- artist momentum context;
- release-level performance strength.

The resulting relationship score provides a single artist-context interpretation for each release while retaining evidence-awareness and the underlying component signals.

This helps PMIP distinguish established-star performance from genuine release-level breakout behaviour.

```python
# ============================================================
# Section 9.5 - Artist–Release Performance Relationship
# ============================================================

if not section_9_4_complete:
    raise RuntimeError(
        "Section 9.4 must be completed before assessing "
        "Artist–Release Performance Relationship."
    )

# ============================================================
# Working source
# ============================================================

artist_release_relationship_df = artist_context_momentum_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_relationship_fields = [
    "release_id",
    "Track",
    "Artist",
    "artist_normalised",
    "release_context_performance_score",
    "artist_performance_score",
    "release_vs_artist_baseline_score",
    "artist_growth_score",
    "artist_momentum_score",
]

missing_relationship_fields = [
    field
    for field in required_relationship_fields
    if field not in artist_release_relationship_df.columns
]

if missing_relationship_fields:
    raise RuntimeError(
        "Required Artist–Release Performance Relationship fields "
        f"are missing: {missing_relationship_fields}"
    )

# ============================================================
# Prepare artist-context components
#
# All scores are already approximately 0-100.
# ============================================================

relationship_component_weights = {
    "release_context_performance_score": 0.30,
    "artist_performance_score": 0.20,
    "release_vs_artist_baseline_score": 0.25,
    "artist_growth_score": 0.125,
    "artist_momentum_score": 0.125,
}

relationship_component_columns = list(
    relationship_component_weights.keys()
)

# ============================================================
# Coverage-aware relationship score
# ============================================================

relationship_weighted_sum = pd.Series(
    0.0,
    index=artist_release_relationship_df.index
)

relationship_available_weight_sum = pd.Series(
    0.0,
    index=artist_release_relationship_df.index
)

for column, weight in relationship_component_weights.items():

    available_mask = (
        artist_release_relationship_df[
            column
        ].notna()
    )

    relationship_weighted_sum.loc[
        available_mask
    ] += (
        artist_release_relationship_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    relationship_available_weight_sum.loc[
        available_mask
    ] += weight

artist_release_relationship_df[
    "artist_release_relationship_score"
] = np.where(
    relationship_available_weight_sum > 0,
    (
        relationship_weighted_sum
        / relationship_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Evidence coverage
# ============================================================

artist_release_relationship_df[
    "artist_release_components_available"
] = (
    artist_release_relationship_df[
        relationship_component_columns
    ]
    .notna()
    .sum(axis=1)
)

artist_release_relationship_df[
    "artist_release_evidence_coverage_pct"
] = (
    artist_release_relationship_df[
        "artist_release_components_available"
    ]
    / len(
        relationship_component_columns
    )
    * 100
)

# ============================================================
# Evidence strength
# ============================================================

artist_release_relationship_df[
    "artist_release_evidence_strength"
] = np.select(
    [
        artist_release_relationship_df[
            "artist_release_components_available"
        ] == 5,

        artist_release_relationship_df[
            "artist_release_components_available"
        ] == 4,

        artist_release_relationship_df[
            "artist_release_components_available"
        ] == 3,

        artist_release_relationship_df[
            "artist_release_components_available"
        ] == 2,
    ],
    [
        "Very Strong Artist Context Evidence",
        "Strong Artist Context Evidence",
        "Moderate Artist Context Evidence",
        "Limited Artist Context Evidence",
    ],
    default="Minimal Artist Context Evidence",
)

# ============================================================
# Relationship percentile and rank
# ============================================================

valid_relationship_mask = (
    artist_release_relationship_df[
        "artist_release_relationship_score"
    ].notna()
)

artist_release_relationship_df[
    "artist_release_relationship_percentile"
] = np.nan

if valid_relationship_mask.any():

    artist_release_relationship_df.loc[
        valid_relationship_mask,
        "artist_release_relationship_percentile"
    ] = (
        artist_release_relationship_df.loc[
            valid_relationship_mask,
            "artist_release_relationship_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

artist_release_relationship_df[
    "artist_release_relationship_rank"
] = (
    artist_release_relationship_df[
        "artist_release_relationship_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Relationship type
#
# This is deliberately rule-based and interpretable.
# ============================================================

artist_release_relationship_df[
    "artist_release_relationship_type"
] = "Mixed Artist–Release Context"

# Strong release + strong artist
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] >= 60
    )
    & (
        artist_release_relationship_df[
            "artist_performance_score"
        ] >= 60
    ),
    "artist_release_relationship_type"
] = "Established Artist Strength"

# Strong release + clear baseline outperformance
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] >= 60
    )
    & (
        artist_release_relationship_df[
            "release_vs_artist_baseline_score"
        ] >= 60
    ),
    "artist_release_relationship_type"
] = "Release Outperformance"

# Strong release + weak artist
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] >= 60
    )
    & (
        artist_release_relationship_df[
            "artist_performance_score"
        ] < 45
    ),
    "artist_release_relationship_type"
] = "Breakout Release Context"

# Positive growth/momentum support
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] >= 50
    )
    & (
        artist_release_relationship_df[
            "artist_growth_score"
        ] >= 60
    )
    & (
        artist_release_relationship_df[
            "artist_momentum_score"
        ] >= 60
    ),
    "artist_release_relationship_type"
] = "Artist Momentum Supported"

# Weak release + strong artist
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] < 45
    )
    & (
        artist_release_relationship_df[
            "artist_performance_score"
        ] >= 60
    ),
    "artist_release_relationship_type"
] = "Underperforming Strong Artist Release"

# Weak release + weak artist
artist_release_relationship_df.loc[
    (
        artist_release_relationship_df[
            "release_context_performance_score"
        ] < 45
    )
    & (
        artist_release_relationship_df[
            "artist_performance_score"
        ] < 45
    ),
    "artist_release_relationship_type"
] = "Weak Artist–Release Context"

# ============================================================
# Relationship classes
# ============================================================

artist_release_relationship_df[
    "artist_release_relationship_class"
] = "No Artist–Release Relationship Evidence"

if valid_relationship_mask.any():

    q25 = artist_release_relationship_df.loc[
        valid_relationship_mask,
        "artist_release_relationship_score"
    ].quantile(0.25)

    q50 = artist_release_relationship_df.loc[
        valid_relationship_mask,
        "artist_release_relationship_score"
    ].quantile(0.50)

    q75 = artist_release_relationship_df.loc[
        valid_relationship_mask,
        "artist_release_relationship_score"
    ].quantile(0.75)

    q90 = artist_release_relationship_df.loc[
        valid_relationship_mask,
        "artist_release_relationship_score"
    ].quantile(0.90)

    artist_release_relationship_df.loc[
        valid_relationship_mask
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] <= q25
        ),
        "artist_release_relationship_class"
    ] = "Weak Artist–Release Relationship"

    artist_release_relationship_df.loc[
        valid_relationship_mask
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] > q25
        )
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] <= q50
        ),
        "artist_release_relationship_class"
    ] = "Moderate Artist–Release Relationship"

    artist_release_relationship_df.loc[
        valid_relationship_mask
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] > q50
        )
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] <= q75
        ),
        "artist_release_relationship_class"
    ] = "Strong Artist–Release Relationship"

    artist_release_relationship_df.loc[
        valid_relationship_mask
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] > q75
        )
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] <= q90
        ),
        "artist_release_relationship_class"
    ] = "Very Strong Artist–Release Relationship"

    artist_release_relationship_df.loc[
        valid_relationship_mask
        & (
            artist_release_relationship_df[
                "artist_release_relationship_score"
            ] > q90
        ),
        "artist_release_relationship_class"
    ] = "Exceptional Artist–Release Relationship"

# ============================================================
# Summary
# ============================================================

artist_release_relationship_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                artist_release_relationship_df
            ),
        },
        {
            "Metric": "Releases with relationship score",
            "Observed Value": int(
                valid_relationship_mask.sum()
            ),
        },
        {
            "Metric": "Releases without relationship score",
            "Observed Value": int(
                (~valid_relationship_mask).sum()
            ),
        },
        {
            "Metric": "Median artist-release relationship score",
            "Observed Value": (
                artist_release_relationship_df.loc[
                    valid_relationship_mask,
                    "artist_release_relationship_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean artist-release relationship score",
            "Observed Value": (
                artist_release_relationship_df.loc[
                    valid_relationship_mask,
                    "artist_release_relationship_score"
                ].mean()
            ),
        },
        {
            "Metric": "Maximum artist-release relationship score",
            "Observed Value": (
                artist_release_relationship_df.loc[
                    valid_relationship_mask,
                    "artist_release_relationship_score"
                ].max()
            ),
        },
        {
            "Metric": "Median artist-context evidence coverage (%)",
            "Observed Value": (
                artist_release_relationship_df.loc[
                    valid_relationship_mask,
                    "artist_release_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP Artist–Release Performance Relationship")
print("=" * 80)

display(
    artist_release_relationship_summary_df
)

# ============================================================
# Relationship component coverage
# ============================================================

relationship_component_coverage_rows = []

for column, weight in relationship_component_weights.items():

    relationship_component_coverage_rows.append(
        {
            "Component": column,
            "Releases Available": int(
                artist_release_relationship_df[
                    column
                ].notna().sum()
            ),
            "Coverage (%)": round(
                artist_release_relationship_df[
                    column
                ].notna().mean()
                * 100,
                6
            ),
            "Base Weight": weight,
        }
    )

artist_release_relationship_component_coverage_df = pd.DataFrame(
    relationship_component_coverage_rows
)

print()
print("Artist–release relationship component coverage")
print("=" * 80)

display(
    artist_release_relationship_component_coverage_df
)

# ============================================================
# Relationship type distribution
# ============================================================

artist_release_relationship_type_distribution_df = (
    artist_release_relationship_df[
        "artist_release_relationship_type"
    ]
    .value_counts()
    .rename_axis(
        "Artist–Release Relationship Type"
    )
    .reset_index(
        name="Releases"
    )
)

artist_release_relationship_type_distribution_df[
    "Share (%)"
] = (
    artist_release_relationship_type_distribution_df[
        "Releases"
    ]
    / len(
        artist_release_relationship_df
    )
    * 100
)

print()
print("Artist–release relationship type distribution")
print("=" * 80)

display(
    artist_release_relationship_type_distribution_df
)

# ============================================================
# Relationship class distribution
# ============================================================

artist_release_relationship_class_distribution_df = (
    artist_release_relationship_df[
        "artist_release_relationship_class"
    ]
    .value_counts()
    .rename_axis(
        "Artist–Release Relationship Class"
    )
    .reset_index(
        name="Releases"
    )
)

artist_release_relationship_class_distribution_df[
    "Share (%)"
] = (
    artist_release_relationship_class_distribution_df[
        "Releases"
    ]
    / len(
        artist_release_relationship_df
    )
    * 100
)

print()
print("Artist–release relationship class distribution")
print("=" * 80)

display(
    artist_release_relationship_class_distribution_df
)

# ============================================================
# Evidence strength distribution
# ============================================================

artist_release_relationship_evidence_distribution_df = (
    artist_release_relationship_df[
        "artist_release_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Artist–Release Evidence Strength"
    )
    .reset_index(
        name="Releases"
    )
)

artist_release_relationship_evidence_distribution_df[
    "Share (%)"
] = (
    artist_release_relationship_evidence_distribution_df[
        "Releases"
    ]
    / len(
        artist_release_relationship_df
    )
    * 100
)

print()
print("Artist–release evidence-strength distribution")
print("=" * 80)

display(
    artist_release_relationship_evidence_distribution_df
)

# ============================================================
# Highest relationship-score releases
# ============================================================

top_artist_release_relationship_df = (
    artist_release_relationship_df.loc[
        valid_relationship_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "release_context_performance_score",
            "artist_performance_score",
            "release_vs_artist_baseline_score",
            "artist_growth_score",
            "artist_momentum_score",
            "artist_release_components_available",
            "artist_release_evidence_coverage_pct",
            "artist_release_evidence_strength",
            "artist_release_relationship_score",
            "artist_release_relationship_percentile",
            "artist_release_relationship_rank",
            "artist_release_relationship_type",
            "artist_release_relationship_class",
        ],
    ]
    .sort_values(
        by="artist_release_relationship_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP Artist–Release Relationship releases")
print("=" * 80)

display(
    top_artist_release_relationship_df
)

# ============================================================
# Final Section 9 working dataset
# ============================================================

artist_context_final_df = artist_release_relationship_df.copy()

# ============================================================
# Validation
# ============================================================

artist_release_relationship_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.4 completed",
            "Passed": section_9_4_complete,
        },
        {
            "Validation Area": "Release performance component available",
            "Passed": artist_release_relationship_df[
                "release_context_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist-level performance component available",
            "Passed": artist_release_relationship_df[
                "artist_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Release-vs-artist baseline component available",
            "Passed": artist_release_relationship_df[
                "release_vs_artist_baseline_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist growth component available",
            "Passed": artist_release_relationship_df[
                "artist_growth_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist momentum component available",
            "Passed": artist_release_relationship_df[
                "artist_momentum_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Coverage-aware weighting prepared",
            "Passed": (
                relationship_available_weight_sum > 0
            ).any(),
        },
        {
            "Validation Area": "Artist-release relationship score calculated",
            "Passed": artist_release_relationship_df[
                "artist_release_relationship_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Artist-release relationship type assigned",
            "Passed": artist_release_relationship_df[
                "artist_release_relationship_type"
            ].notna().all(),
        },
        {
            "Validation Area": "Artist-release relationship classes assigned",
            "Passed": artist_release_relationship_df[
                "artist_release_relationship_class"
            ].notna().all(),
        },
    ]
)

print()
print("Artist–Release Performance Relationship validation")
print("=" * 80)

display(
    artist_release_relationship_validation_df
)

# ============================================================
# Completion
# ============================================================

section_9_5_complete = bool(
    artist_release_relationship_validation_df[
        "Passed"
    ].all()
)

section_9_complete = bool(
    section_9_1_complete
    and section_9_2_complete
    and section_9_3_complete
    and section_9_4_complete
    and section_9_5_complete
)

print()
print(
    f"Section 9.5 completion status: "
    f"{section_9_5_complete}"
)

print(
    f"Section 9 overall completion status: "
    f"{section_9_complete}"
)

if not section_9_complete:
    raise RuntimeError(
        "Section 9 Artist Context did not complete successfully."
    )
```

## 10. Unusual Release Performance Patterns

### Purpose

This section investigates unusual or potentially exceptional release-performance behaviour within PMIP.

The aim is to identify releases whose streaming, chart or growth patterns differ noticeably from the broader population and may therefore require additional interpretation.

The analysis focuses on five areas:

- short-term streaming spikes;
- sudden chart-position movement;
- unusual streaming behaviour;
- strong growth without corresponding geographic breadth;
- combined unusual-performance indicators.

These patterns are treated as analytical signals rather than evidence of abnormal or suspicious activity.

A release may show unusual behaviour for legitimate reasons such as viral exposure, playlist placement, renewed catalogue interest, promotional activity, regional concentration or external cultural events.

The purpose of this section is therefore to identify releases that stand out from normal PMIP performance patterns and provide a structured basis for deeper investigation.


## 10.1 Short-Term Spikes

### Purpose

This subsection identifies releases that show unusually sharp short-term increases in streaming activity.

The goal is to detect releases whose recent streaming behaviour rises much faster than would normally be expected from their historical pattern.

The assessment will examine:

- recent streaming change;
- percentage growth between historical observations;
- unusually large short-term increases;
- the size of the increase relative to the release's historical level;
- the amount of historical evidence supporting each spike assessment.

A short-term spike does not automatically mean abnormal or suspicious behaviour.

It may reflect genuine events such as:

- viral discovery;
- playlist placement;
- social-media exposure;
- renewed catalogue interest;
- artist promotion;
- external cultural events.

The purpose of this subsection is therefore to flag unusually strong short-term streaming acceleration for further PMIP interpretation.

```python
# ============================================================
# Section 10.1 - Short-Term Spikes
# ============================================================

if not section_9_complete:
    raise RuntimeError(
        "Section 9 must be completed before assessing "
        "Unusual Release Performance Patterns."
    )

# ============================================================
# Working source
# ============================================================

short_term_spike_df = artist_context_final_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_spike_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_spike_fields = [
    field
    for field in required_spike_fields
    if field not in short_term_spike_df.columns
]

if missing_spike_fields:
    raise RuntimeError(
        "Required Short-Term Spike fields are missing: "
        f"{missing_spike_fields}"
    )

# ============================================================
# Candidate historical streaming fields
# ============================================================

historical_stream_fields = [
    field
    for field in [
        "first_historical_streams",
        "last_historical_streams",
        "average_historical_streams",
        "max_historical_streams",
        "percentage_stream_growth",
        "daily_stream_growth_rate",
        "historical_stream_range",
        "relative_historical_stream_range_pct",
    ]
    if field in short_term_spike_df.columns
]

if not historical_stream_fields:
    raise RuntimeError(
        "No historical streaming fields are available for "
        "Short-Term Spike analysis."
    )

# ============================================================
# Prepare core short-term change signals
# ============================================================

if (
    "first_historical_streams" in short_term_spike_df.columns
    and "last_historical_streams" in short_term_spike_df.columns
):
    short_term_spike_df[
        "short_term_absolute_change"
    ] = (
        short_term_spike_df[
            "last_historical_streams"
        ]
        - short_term_spike_df[
            "first_historical_streams"
        ]
    )
else:
    short_term_spike_df[
        "short_term_absolute_change"
    ] = np.nan

# ============================================================
# Percentage change
# ============================================================

if "percentage_stream_growth" in short_term_spike_df.columns:
    short_term_spike_df[
        "short_term_percentage_change"
    ] = short_term_spike_df[
        "percentage_stream_growth"
    ]
elif (
    "first_historical_streams" in short_term_spike_df.columns
    and "last_historical_streams" in short_term_spike_df.columns
):
    valid_pct_mask = (
        short_term_spike_df[
            "first_historical_streams"
        ].notna()
        & short_term_spike_df[
            "last_historical_streams"
        ].notna()
        & (
            short_term_spike_df[
                "first_historical_streams"
            ] > 0
        )
    )

    short_term_spike_df[
        "short_term_percentage_change"
    ] = np.nan

    short_term_spike_df.loc[
        valid_pct_mask,
        "short_term_percentage_change"
    ] = (
        (
            short_term_spike_df.loc[
                valid_pct_mask,
                "last_historical_streams"
            ]
            - short_term_spike_df.loc[
                valid_pct_mask,
                "first_historical_streams"
            ]
        )
        / short_term_spike_df.loc[
            valid_pct_mask,
            "first_historical_streams"
        ]
        * 100
    )
else:
    short_term_spike_df[
        "short_term_percentage_change"
    ] = np.nan

# ============================================================
# Daily growth signal
# ============================================================

if "daily_stream_growth_rate" in short_term_spike_df.columns:
    short_term_spike_df[
        "short_term_daily_growth"
    ] = short_term_spike_df[
        "daily_stream_growth_rate"
    ]
else:
    short_term_spike_df[
        "short_term_daily_growth"
    ] = np.nan

# ============================================================
# Relative historical variation signal
# ============================================================

if "relative_historical_stream_range_pct" in short_term_spike_df.columns:
    short_term_spike_df[
        "short_term_relative_variation"
    ] = short_term_spike_df[
        "relative_historical_stream_range_pct"
    ]
else:
    short_term_spike_df[
        "short_term_relative_variation"
    ] = np.nan

# ============================================================
# Evidence mask
# ============================================================

spike_signal_columns = [
    "short_term_absolute_change",
    "short_term_percentage_change",
    "short_term_daily_growth",
    "short_term_relative_variation",
]

short_term_spike_df[
    "short_term_spike_components_available"
] = (
    short_term_spike_df[
        spike_signal_columns
    ]
    .notna()
    .sum(axis=1)
)

short_term_spike_df[
    "short_term_spike_evidence_pct"
] = (
    short_term_spike_df[
        "short_term_spike_components_available"
    ]
    / len(
        spike_signal_columns
    )
    * 100
)

valid_short_term_spike_mask = (
    short_term_spike_df[
        "short_term_spike_components_available"
    ] > 0
)

# ============================================================
# Convert positive spike signals into percentiles
#
# Only positive movement should contribute to a spike score.
# ============================================================

positive_spike_source_columns = [
    "short_term_percentage_change",
    "short_term_daily_growth",
    "short_term_relative_variation",
]

spike_percentile_columns = []

for column in positive_spike_source_columns:

    percentile_column = (
        f"{column}_spike_percentile"
    )

    spike_percentile_columns.append(
        percentile_column
    )

    short_term_spike_df[
        percentile_column
    ] = np.nan

    valid_component_mask = (
        short_term_spike_df[
            column
        ].notna()
        & (
            short_term_spike_df[
                column
            ] > 0
        )
    )

    if valid_component_mask.any():

        short_term_spike_df.loc[
            valid_component_mask,
            percentile_column
        ] = (
            short_term_spike_df.loc[
                valid_component_mask,
                column
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

# ============================================================
# Short-term spike score
#
# Weighted toward percentage growth and daily acceleration.
# ============================================================

spike_component_weights = {
    "short_term_percentage_change_spike_percentile": 0.45,
    "short_term_daily_growth_spike_percentile": 0.35,
    "short_term_relative_variation_spike_percentile": 0.20,
}

spike_weighted_sum = pd.Series(
    0.0,
    index=short_term_spike_df.index
)

spike_available_weight_sum = pd.Series(
    0.0,
    index=short_term_spike_df.index
)

for column, weight in spike_component_weights.items():

    component_available_mask = (
        short_term_spike_df[
            column
        ].notna()
    )

    spike_weighted_sum.loc[
        component_available_mask
    ] += (
        short_term_spike_df.loc[
            component_available_mask,
            column
        ]
        * weight
    )

    spike_available_weight_sum.loc[
        component_available_mask
    ] += weight

short_term_spike_df[
    "short_term_spike_score"
] = np.where(
    spike_available_weight_sum > 0,
    (
        spike_weighted_sum
        / spike_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Spike percentile and rank
# ============================================================

valid_spike_score_mask = (
    short_term_spike_df[
        "short_term_spike_score"
    ].notna()
)

short_term_spike_df[
    "short_term_spike_percentile"
] = np.nan

if valid_spike_score_mask.any():

    short_term_spike_df.loc[
        valid_spike_score_mask,
        "short_term_spike_percentile"
    ] = (
        short_term_spike_df.loc[
            valid_spike_score_mask,
            "short_term_spike_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

short_term_spike_df[
    "short_term_spike_rank"
] = (
    short_term_spike_df[
        "short_term_spike_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Spike classes
# ============================================================

short_term_spike_df[
    "short_term_spike_class"
] = "No Short-Term Spike Evidence"

short_term_spike_df.loc[
    valid_spike_score_mask
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] < 75
    ),
    "short_term_spike_class"
] = "Normal Short-Term Movement"

short_term_spike_df.loc[
    valid_spike_score_mask
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] >= 75
    )
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] < 90
    ),
    "short_term_spike_class"
] = "Elevated Short-Term Growth"

short_term_spike_df.loc[
    valid_spike_score_mask
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] >= 90
    )
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] < 97.5
    ),
    "short_term_spike_class"
] = "Strong Short-Term Spike"

short_term_spike_df.loc[
    valid_spike_score_mask
    & (
        short_term_spike_df[
            "short_term_spike_score"
        ] >= 97.5
    ),
    "short_term_spike_class"
] = "Exceptional Short-Term Spike"

# ============================================================
# Binary spike flag
#
# Conservative threshold:
# top 10% of valid spike-score population.
# ============================================================

if valid_spike_score_mask.any():

    spike_flag_threshold = (
        short_term_spike_df.loc[
            valid_spike_score_mask,
            "short_term_spike_score"
        ]
        .quantile(0.90)
    )

else:

    spike_flag_threshold = np.nan

short_term_spike_df[
    "short_term_spike_flag"
] = (
    short_term_spike_df[
        "short_term_spike_score"
    ] >= spike_flag_threshold
)

short_term_spike_df.loc[
    ~valid_spike_score_mask,
    "short_term_spike_flag"
] = False

# ============================================================
# Evidence strength
# ============================================================

short_term_spike_df[
    "short_term_spike_evidence_strength"
] = np.select(
    [
        short_term_spike_df[
            "short_term_spike_components_available"
        ] >= 3,

        short_term_spike_df[
            "short_term_spike_components_available"
        ] == 2,

        short_term_spike_df[
            "short_term_spike_components_available"
        ] == 1,
    ],
    [
        "Strong Spike Evidence",
        "Moderate Spike Evidence",
        "Limited Spike Evidence",
    ],
    default="No Spike Evidence",
)

# ============================================================
# Summary
# ============================================================

short_term_spike_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                short_term_spike_df
            ),
        },
        {
            "Metric": "Releases with short-term spike evidence",
            "Observed Value": int(
                valid_spike_score_mask.sum()
            ),
        },
        {
            "Metric": "Releases without short-term spike evidence",
            "Observed Value": int(
                (~valid_spike_score_mask).sum()
            ),
        },
        {
            "Metric": "Median short-term percentage change",
            "Observed Value": (
                short_term_spike_df[
                    "short_term_percentage_change"
                ].median()
            ),
        },
        {
            "Metric": "Median daily streaming growth",
            "Observed Value": (
                short_term_spike_df[
                    "short_term_daily_growth"
                ].median()
            ),
        },
        {
            "Metric": "Median short-term spike score",
            "Observed Value": (
                short_term_spike_df.loc[
                    valid_spike_score_mask,
                    "short_term_spike_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean short-term spike score",
            "Observed Value": (
                short_term_spike_df.loc[
                    valid_spike_score_mask,
                    "short_term_spike_score"
                ].mean()
            ),
        },
        {
            "Metric": "Spike flag threshold",
            "Observed Value": spike_flag_threshold,
        },
        {
            "Metric": "Releases flagged as short-term spikes",
            "Observed Value": int(
                short_term_spike_df[
                    "short_term_spike_flag"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP short-term streaming spikes")
print("=" * 80)

display(
    short_term_spike_summary_df
)

# ============================================================
# Spike class distribution
# ============================================================

short_term_spike_class_distribution_df = (
    short_term_spike_df[
        "short_term_spike_class"
    ]
    .value_counts()
    .rename_axis(
        "Short-Term Spike Class"
    )
    .reset_index(
        name="Releases"
    )
)

short_term_spike_class_distribution_df[
    "Share (%)"
] = (
    short_term_spike_class_distribution_df[
        "Releases"
    ]
    / len(
        short_term_spike_df
    )
    * 100
)

print()
print("Short-term spike class distribution")
print("=" * 80)

display(
    short_term_spike_class_distribution_df
)

# ============================================================
# Spike evidence distribution
# ============================================================

short_term_spike_evidence_distribution_df = (
    short_term_spike_df[
        "short_term_spike_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Short-Term Spike Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

short_term_spike_evidence_distribution_df[
    "Share (%)"
] = (
    short_term_spike_evidence_distribution_df[
        "Releases"
    ]
    / len(
        short_term_spike_df
    )
    * 100
)

print()
print("Short-term spike evidence distribution")
print("=" * 80)

display(
    short_term_spike_evidence_distribution_df
)

# ============================================================
# Strongest detected short-term spikes
# ============================================================

top_short_term_spikes_df = (
    short_term_spike_df.loc[
        valid_spike_score_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "short_term_absolute_change",
            "short_term_percentage_change",
            "short_term_daily_growth",
            "short_term_relative_variation",
            "short_term_spike_components_available",
            "short_term_spike_evidence_pct",
            "short_term_spike_evidence_strength",
            "short_term_spike_score",
            "short_term_spike_percentile",
            "short_term_spike_rank",
            "short_term_spike_class",
            "short_term_spike_flag",
        ],
    ]
    .sort_values(
        by="short_term_spike_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strongest PMIP short-term streaming spikes")
print("=" * 80)

display(
    top_short_term_spikes_df
)

# ============================================================
# Validation
# ============================================================

short_term_spike_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9 completed",
            "Passed": section_9_complete,
        },
        {
            "Validation Area": "Historical streaming evidence available",
            "Passed": len(
                historical_stream_fields
            ) > 0,
        },
        {
            "Validation Area": "Short-term change signals prepared",
            "Passed": short_term_spike_df[
                spike_signal_columns
            ].notna().any().any(),
        },
        {
            "Validation Area": "Spike score calculated",
            "Passed": short_term_spike_df[
                "short_term_spike_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Spike percentile calculated",
            "Passed": short_term_spike_df[
                "short_term_spike_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Spike classes assigned",
            "Passed": short_term_spike_df[
                "short_term_spike_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Spike flags assigned",
            "Passed": short_term_spike_df[
                "short_term_spike_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Spike evidence classified",
            "Passed": short_term_spike_df[
                "short_term_spike_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Short-Term Spike validation")
print("=" * 80)

display(
    short_term_spike_validation_df
)

# ============================================================
# Completion
# ============================================================

section_10_1_complete = bool(
    short_term_spike_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 10.1 completion status: "
    f"{section_10_1_complete}"
)

if not section_10_1_complete:
    raise RuntimeError(
        "Short-Term Spike analysis "
        "did not complete successfully."
    )
```

### Interpretation

The Short-Term Spikes assessment identified unusual positive streaming acceleration within the PMIP release population.

A total of 2,201 releases had sufficient spike evidence, while 2,392 releases did not have enough historical streaming information for this assessment.

The median short-term percentage change was approximately -26.23%, showing that the typical observed release was not experiencing positive acceleration.

A total of 221 releases were flagged as short-term spikes using the upper-tail threshold of the spike-score distribution.

Among the full dataset, 165 releases were classified as `Strong Short-Term Spike` and 56 as `Exceptional Short-Term Spike`.

These results indicate that unusually strong short-term streaming growth exists only within a relatively small part of the release population.

The spike flags should be treated as signals for further investigation rather than evidence of abnormal behaviour.

## 10.2 Sudden Chart Movement

### Purpose

This subsection identifies releases that show unusually large changes in chart position across their observed chart history.

The aim is to detect releases whose chart movement is substantially stronger than the typical movement seen across the PMIP population.

The assessment considers:

- movement between first and last observed chart position;
- the absolute size of the chart-position change;
- whether the release improved or declined;
- the relative extremity of the movement compared with other releases;
- the amount of evidence supporting the movement assessment.

Large chart movements may reflect genuine events such as viral attention, playlist exposure, promotion, renewed catalogue interest or rapid loss of audience attention.

The purpose is therefore to identify unusually strong chart movement for further PMIP interpretation rather than to label the behaviour as abnormal.

```python
# ============================================================
# Section 10.2 - Sudden Chart Movement
# ============================================================

if not section_10_1_complete:
    raise RuntimeError(
        "Section 10.1 must be completed before assessing "
        "Sudden Chart Movement."
    )

# ============================================================
# Working source
# ============================================================

sudden_chart_movement_df = short_term_spike_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_chart_movement_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_chart_movement_fields = [
    field
    for field in required_chart_movement_fields
    if field not in sudden_chart_movement_df.columns
]

if missing_chart_movement_fields:
    raise RuntimeError(
        "Required Sudden Chart Movement fields are missing: "
        f"{missing_chart_movement_fields}"
    )

# ============================================================
# Candidate chart movement fields
# ============================================================

chart_movement_candidate_fields = [
    field
    for field in [
        "chart_position_movement",
        "chart_movement_score",
        "chart_movement_direction",
        "chart_movement_class",
        "first_chart_position",
        "last_chart_position",
        "best_chart_position",
        "average_chart_position",
        "chart_observations",
        "chart_history_span_days",
    ]
    if field in sudden_chart_movement_df.columns
]

if not chart_movement_candidate_fields:
    raise RuntimeError(
        "No chart movement fields are available for "
        "Sudden Chart Movement analysis."
    )

# ============================================================
# Prepare core movement signal
#
# Lower chart position numbers are stronger.
# Positive improvement therefore means first position minus
# last position.
# ============================================================

if "chart_position_movement" in sudden_chart_movement_df.columns:

    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] = sudden_chart_movement_df[
        "chart_position_movement"
    ]

elif (
    "first_chart_position" in sudden_chart_movement_df.columns
    and "last_chart_position" in sudden_chart_movement_df.columns
):

    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] = (
        sudden_chart_movement_df[
            "first_chart_position"
        ]
        - sudden_chart_movement_df[
            "last_chart_position"
        ]
    )

else:

    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] = np.nan

# ============================================================
# Absolute movement
# ============================================================

sudden_chart_movement_df[
    "sudden_chart_absolute_change"
] = (
    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ]
    .abs()
)

# ============================================================
# Movement direction
# ============================================================

sudden_chart_movement_df[
    "sudden_chart_movement_direction"
] = "Unavailable"

sudden_chart_movement_df.loc[
    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] > 0,
    "sudden_chart_movement_direction"
] = "Improved"

sudden_chart_movement_df.loc[
    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] < 0,
    "sudden_chart_movement_direction"
] = "Declined"

sudden_chart_movement_df.loc[
    sudden_chart_movement_df[
        "sudden_chart_position_change"
    ] == 0,
    "sudden_chart_movement_direction"
] = "Stable"

# ============================================================
# Additional supporting signals
# ============================================================

if "chart_observations" in sudden_chart_movement_df.columns:
    sudden_chart_movement_df[
        "sudden_chart_observations"
    ] = sudden_chart_movement_df[
        "chart_observations"
    ]
else:
    sudden_chart_movement_df[
        "sudden_chart_observations"
    ] = np.nan

if "chart_history_span_days" in sudden_chart_movement_df.columns:
    sudden_chart_movement_df[
        "sudden_chart_span_days"
    ] = sudden_chart_movement_df[
        "chart_history_span_days"
    ]
else:
    sudden_chart_movement_df[
        "sudden_chart_span_days"
    ] = np.nan

# ============================================================
# Evidence mask
# ============================================================

chart_signal_columns = [
    "sudden_chart_position_change",
    "sudden_chart_absolute_change",
    "sudden_chart_observations",
    "sudden_chart_span_days",
]

sudden_chart_movement_df[
    "sudden_chart_components_available"
] = (
    sudden_chart_movement_df[
        chart_signal_columns
    ]
    .notna()
    .sum(axis=1)
)

sudden_chart_movement_df[
    "sudden_chart_evidence_pct"
] = (
    sudden_chart_movement_df[
        "sudden_chart_components_available"
    ]
    / len(
        chart_signal_columns
    )
    * 100
)

valid_sudden_chart_mask = (
    sudden_chart_movement_df[
        "sudden_chart_absolute_change"
    ].notna()
)

# ============================================================
# Sudden movement percentile
#
# The magnitude of movement is the main anomaly signal.
# ============================================================

sudden_chart_movement_df[
    "sudden_chart_movement_percentile"
] = np.nan

if valid_sudden_chart_mask.any():

    sudden_chart_movement_df.loc[
        valid_sudden_chart_mask,
        "sudden_chart_movement_percentile"
    ] = (
        sudden_chart_movement_df.loc[
            valid_sudden_chart_mask,
            "sudden_chart_absolute_change"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Optional chart movement score support
#
# Existing Section 6 movement score can add extra context.
# ============================================================

if "chart_movement_score" in sudden_chart_movement_df.columns:

    sudden_chart_movement_df[
        "existing_chart_movement_percentile"
    ] = np.nan

    existing_score_mask = (
        sudden_chart_movement_df[
            "chart_movement_score"
        ].notna()
    )

    if existing_score_mask.any():

        sudden_chart_movement_df.loc[
            existing_score_mask,
            "existing_chart_movement_percentile"
        ] = (
            sudden_chart_movement_df.loc[
                existing_score_mask,
                "chart_movement_score"
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

else:

    sudden_chart_movement_df[
        "existing_chart_movement_percentile"
    ] = np.nan

# ============================================================
# Sudden chart movement score
#
# Main weight goes to absolute chart movement.
# Existing movement score is supporting evidence.
# ============================================================

sudden_chart_component_weights = {
    "sudden_chart_movement_percentile": 0.75,
    "existing_chart_movement_percentile": 0.25,
}

sudden_chart_weighted_sum = pd.Series(
    0.0,
    index=sudden_chart_movement_df.index
)

sudden_chart_available_weight_sum = pd.Series(
    0.0,
    index=sudden_chart_movement_df.index
)

for column, weight in sudden_chart_component_weights.items():

    available_mask = (
        sudden_chart_movement_df[
            column
        ].notna()
    )

    sudden_chart_weighted_sum.loc[
        available_mask
    ] += (
        sudden_chart_movement_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    sudden_chart_available_weight_sum.loc[
        available_mask
    ] += weight

sudden_chart_movement_df[
    "sudden_chart_movement_score"
] = np.where(
    sudden_chart_available_weight_sum > 0,
    (
        sudden_chart_weighted_sum
        / sudden_chart_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Score rank
# ============================================================

valid_sudden_chart_score_mask = (
    sudden_chart_movement_df[
        "sudden_chart_movement_score"
    ].notna()
)

sudden_chart_movement_df[
    "sudden_chart_movement_rank"
] = (
    sudden_chart_movement_df[
        "sudden_chart_movement_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Sudden movement classes
# ============================================================

sudden_chart_movement_df[
    "sudden_chart_movement_class"
] = "No Sudden Chart Movement Evidence"

sudden_chart_movement_df.loc[
    valid_sudden_chart_score_mask
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] < 75
    ),
    "sudden_chart_movement_class"
] = "Normal Chart Movement"

sudden_chart_movement_df.loc[
    valid_sudden_chart_score_mask
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] >= 75
    )
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] < 90
    ),
    "sudden_chart_movement_class"
] = "Elevated Chart Movement"

sudden_chart_movement_df.loc[
    valid_sudden_chart_score_mask
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] >= 90
    )
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] < 97.5
    ),
    "sudden_chart_movement_class"
] = "Strong Sudden Chart Movement"

sudden_chart_movement_df.loc[
    valid_sudden_chart_score_mask
    & (
        sudden_chart_movement_df[
            "sudden_chart_movement_score"
        ] >= 97.5
    ),
    "sudden_chart_movement_class"
] = "Exceptional Sudden Chart Movement"

# ============================================================
# Binary sudden-movement flag
#
# Upper 10% of valid movement-score population.
# ============================================================

if valid_sudden_chart_score_mask.any():

    sudden_chart_flag_threshold = (
        sudden_chart_movement_df.loc[
            valid_sudden_chart_score_mask,
            "sudden_chart_movement_score"
        ]
        .quantile(0.90)
    )

else:

    sudden_chart_flag_threshold = np.nan

sudden_chart_movement_df[
    "sudden_chart_movement_flag"
] = (
    sudden_chart_movement_df[
        "sudden_chart_movement_score"
    ] >= sudden_chart_flag_threshold
)

sudden_chart_movement_df.loc[
    ~valid_sudden_chart_score_mask,
    "sudden_chart_movement_flag"
] = False

# ============================================================
# Evidence strength
# ============================================================

sudden_chart_movement_df[
    "sudden_chart_movement_evidence_strength"
] = np.select(
    [
        sudden_chart_movement_df[
            "sudden_chart_components_available"
        ] >= 3,

        sudden_chart_movement_df[
            "sudden_chart_components_available"
        ] == 2,

        sudden_chart_movement_df[
            "sudden_chart_components_available"
        ] == 1,
    ],
    [
        "Strong Chart Movement Evidence",
        "Moderate Chart Movement Evidence",
        "Limited Chart Movement Evidence",
    ],
    default="No Chart Movement Evidence",
)

# ============================================================
# Summary
# ============================================================

sudden_chart_movement_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                sudden_chart_movement_df
            ),
        },
        {
            "Metric": "Releases with sudden-chart evidence",
            "Observed Value": int(
                valid_sudden_chart_score_mask.sum()
            ),
        },
        {
            "Metric": "Releases without sudden-chart evidence",
            "Observed Value": int(
                (~valid_sudden_chart_score_mask).sum()
            ),
        },
        {
            "Metric": "Median signed chart-position change",
            "Observed Value": (
                sudden_chart_movement_df[
                    "sudden_chart_position_change"
                ].median()
            ),
        },
        {
            "Metric": "Median absolute chart-position change",
            "Observed Value": (
                sudden_chart_movement_df[
                    "sudden_chart_absolute_change"
                ].median()
            ),
        },
        {
            "Metric": "Median sudden-chart movement score",
            "Observed Value": (
                sudden_chart_movement_df.loc[
                    valid_sudden_chart_score_mask,
                    "sudden_chart_movement_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean sudden-chart movement score",
            "Observed Value": (
                sudden_chart_movement_df.loc[
                    valid_sudden_chart_score_mask,
                    "sudden_chart_movement_score"
                ].mean()
            ),
        },
        {
            "Metric": "Sudden-chart flag threshold",
            "Observed Value": sudden_chart_flag_threshold,
        },
        {
            "Metric": "Releases flagged for sudden chart movement",
            "Observed Value": int(
                sudden_chart_movement_df[
                    "sudden_chart_movement_flag"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP sudden chart movement")
print("=" * 80)

display(
    sudden_chart_movement_summary_df
)

# ============================================================
# Direction distribution
# ============================================================

sudden_chart_direction_distribution_df = (
    sudden_chart_movement_df[
        "sudden_chart_movement_direction"
    ]
    .value_counts()
    .rename_axis(
        "Sudden Chart Movement Direction"
    )
    .reset_index(
        name="Releases"
    )
)

sudden_chart_direction_distribution_df[
    "Share (%)"
] = (
    sudden_chart_direction_distribution_df[
        "Releases"
    ]
    / len(
        sudden_chart_movement_df
    )
    * 100
)

print()
print("Sudden chart movement direction distribution")
print("=" * 80)

display(
    sudden_chart_direction_distribution_df
)

# ============================================================
# Class distribution
# ============================================================

sudden_chart_class_distribution_df = (
    sudden_chart_movement_df[
        "sudden_chart_movement_class"
    ]
    .value_counts()
    .rename_axis(
        "Sudden Chart Movement Class"
    )
    .reset_index(
        name="Releases"
    )
)

sudden_chart_class_distribution_df[
    "Share (%)"
] = (
    sudden_chart_class_distribution_df[
        "Releases"
    ]
    / len(
        sudden_chart_movement_df
    )
    * 100
)

print()
print("Sudden chart movement class distribution")
print("=" * 80)

display(
    sudden_chart_class_distribution_df
)

# ============================================================
# Evidence distribution
# ============================================================

sudden_chart_evidence_distribution_df = (
    sudden_chart_movement_df[
        "sudden_chart_movement_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Sudden Chart Movement Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

sudden_chart_evidence_distribution_df[
    "Share (%)"
] = (
    sudden_chart_evidence_distribution_df[
        "Releases"
    ]
    / len(
        sudden_chart_movement_df
    )
    * 100
)

print()
print("Sudden chart movement evidence distribution")
print("=" * 80)

display(
    sudden_chart_evidence_distribution_df
)

# ============================================================
# Strongest sudden chart movements
# ============================================================

top_sudden_chart_movements_df = (
    sudden_chart_movement_df.loc[
        valid_sudden_chart_score_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "sudden_chart_position_change",
            "sudden_chart_absolute_change",
            "sudden_chart_movement_direction",
            "sudden_chart_observations",
            "sudden_chart_span_days",
            "sudden_chart_components_available",
            "sudden_chart_evidence_pct",
            "sudden_chart_movement_evidence_strength",
            "sudden_chart_movement_score",
            "sudden_chart_movement_rank",
            "sudden_chart_movement_class",
            "sudden_chart_movement_flag",
        ],
    ]
    .sort_values(
        by="sudden_chart_movement_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strongest PMIP sudden chart movements")
print("=" * 80)

display(
    top_sudden_chart_movements_df
)

# ============================================================
# Validation
# ============================================================

sudden_chart_movement_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 10.1 completed",
            "Passed": section_10_1_complete,
        },
        {
            "Validation Area": "Chart movement evidence available",
            "Passed": valid_sudden_chart_mask.any(),
        },
        {
            "Validation Area": "Absolute chart movement calculated",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_absolute_change"
            ].notna().any(),
        },
        {
            "Validation Area": "Movement direction assigned",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_movement_direction"
            ].notna().all(),
        },
        {
            "Validation Area": "Sudden movement score calculated",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_movement_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Sudden movement classes assigned",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_movement_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Sudden movement flags assigned",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_movement_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Chart movement evidence classified",
            "Passed": sudden_chart_movement_df[
                "sudden_chart_movement_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Sudden Chart Movement validation")
print("=" * 80)

display(
    sudden_chart_movement_validation_df
)

# ============================================================
# Completion
# ============================================================

section_10_2_complete = bool(
    sudden_chart_movement_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 10.2 completion status: "
    f"{section_10_2_complete}"
)

if not section_10_2_complete:
    raise RuntimeError(
        "Sudden Chart Movement analysis "
        "did not complete successfully."
    )
```

### Interpretation

The Sudden Chart Movement assessment identified releases with unusually large changes in chart position.

Only 210 releases contained sufficient direct chart-movement evidence, while 4,383 releases did not.

Among the assessed releases, the median absolute chart-position change was 35 positions.

A total of 21 releases were flagged for sudden chart movement using the upper-tail threshold.

Most chart movements were classified as normal, with only a small number falling into strong or exceptional movement categories.

These findings suggest that extreme chart-position changes are relatively uncommon and should be treated as unusual-performance signals for further investigation.

## 10.3 Streaming Anomalies

### Purpose

This subsection identifies releases whose streaming behaviour differs strongly from normal historical patterns.

The aim is to detect unusual combinations of streaming level, growth, variation and stability that may indicate exceptional release behaviour.

The assessment considers:

- unusually high or low streaming variation;
- extreme growth or decline;
- instability across historical observations;
- unusually large differences between current and historical streaming behaviour;
- the strength of evidence supporting each anomaly signal.

Streaming anomalies may result from legitimate causes such as viral attention, playlist exposure, renewed catalogue interest or changing audience behaviour.

The purpose is therefore to identify unusual streaming patterns for further PMIP investigation rather than to classify them as suspicious.

```python
# ============================================================
# Section 10.3 - Streaming Anomalies
# ============================================================

if not section_10_2_complete:
    raise RuntimeError(
        "Section 10.2 must be completed before assessing "
        "Streaming Anomalies."
    )

# ============================================================
# Working source
# ============================================================

streaming_anomaly_df = sudden_chart_movement_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_streaming_anomaly_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_streaming_anomaly_fields = [
    field
    for field in required_streaming_anomaly_fields
    if field not in streaming_anomaly_df.columns
]

if missing_streaming_anomaly_fields:
    raise RuntimeError(
        "Required Streaming Anomaly fields are missing: "
        f"{missing_streaming_anomaly_fields}"
    )

# ============================================================
# Candidate anomaly fields from previous sections
# ============================================================

candidate_streaming_anomaly_fields = [
    field
    for field in [
        "relative_historical_stream_range_pct",
        "percentage_stream_growth",
        "daily_stream_growth_rate",
        "streaming_stability_score",
        "streaming_performance_score",
        "current_vs_historical_score",
        "current_historical_comparability_score",
        "historical_stream_range",
        "average_historical_streams",
        "Spotify Streams",
    ]
    if field in streaming_anomaly_df.columns
]

if not candidate_streaming_anomaly_fields:
    raise RuntimeError(
        "No suitable streaming anomaly fields are available."
    )

# ============================================================
# Prepare anomaly source signals
# ============================================================

# Extreme relative variation
if "relative_historical_stream_range_pct" in streaming_anomaly_df.columns:
    streaming_anomaly_df[
        "streaming_anomaly_variation_signal"
    ] = streaming_anomaly_df[
        "relative_historical_stream_range_pct"
    ].abs()
else:
    streaming_anomaly_df[
        "streaming_anomaly_variation_signal"
    ] = np.nan

# Extreme growth or decline
if "percentage_stream_growth" in streaming_anomaly_df.columns:
    streaming_anomaly_df[
        "streaming_anomaly_growth_signal"
    ] = streaming_anomaly_df[
        "percentage_stream_growth"
    ].abs()
else:
    streaming_anomaly_df[
        "streaming_anomaly_growth_signal"
    ] = np.nan

# Extreme daily growth movement
if "daily_stream_growth_rate" in streaming_anomaly_df.columns:
    streaming_anomaly_df[
        "streaming_anomaly_daily_change_signal"
    ] = streaming_anomaly_df[
        "daily_stream_growth_rate"
    ].abs()
else:
    streaming_anomaly_df[
        "streaming_anomaly_daily_change_signal"
    ] = np.nan

# Instability signal
if "streaming_stability_score" in streaming_anomaly_df.columns:
    streaming_anomaly_df[
        "streaming_anomaly_instability_signal"
    ] = (
        100
        - streaming_anomaly_df[
            "streaming_stability_score"
        ]
    )
else:
    streaming_anomaly_df[
        "streaming_anomaly_instability_signal"
    ] = np.nan

# Current-vs-historical mismatch
if "current_vs_historical_score" in streaming_anomaly_df.columns:
    streaming_anomaly_df[
        "streaming_anomaly_current_history_signal"
    ] = (
        streaming_anomaly_df[
            "current_vs_historical_score"
        ]
        - 50
    ).abs()
else:
    streaming_anomaly_df[
        "streaming_anomaly_current_history_signal"
    ] = np.nan

# ============================================================
# Evidence coverage
# ============================================================

streaming_anomaly_signal_columns = [
    "streaming_anomaly_variation_signal",
    "streaming_anomaly_growth_signal",
    "streaming_anomaly_daily_change_signal",
    "streaming_anomaly_instability_signal",
    "streaming_anomaly_current_history_signal",
]

streaming_anomaly_df[
    "streaming_anomaly_components_available"
] = (
    streaming_anomaly_df[
        streaming_anomaly_signal_columns
    ]
    .notna()
    .sum(axis=1)
)

streaming_anomaly_df[
    "streaming_anomaly_evidence_pct"
] = (
    streaming_anomaly_df[
        "streaming_anomaly_components_available"
    ]
    / len(
        streaming_anomaly_signal_columns
    )
    * 100
)

# ============================================================
# Convert anomaly signals to percentiles
# ============================================================

streaming_anomaly_percentile_columns = []

for column in streaming_anomaly_signal_columns:

    percentile_column = (
        f"{column}_percentile"
    )

    streaming_anomaly_percentile_columns.append(
        percentile_column
    )

    streaming_anomaly_df[
        percentile_column
    ] = np.nan

    valid_component_mask = (
        streaming_anomaly_df[
            column
        ].notna()
    )

    if valid_component_mask.any():

        streaming_anomaly_df.loc[
            valid_component_mask,
            percentile_column
        ] = (
            streaming_anomaly_df.loc[
                valid_component_mask,
                column
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

# ============================================================
# Streaming anomaly score
# ============================================================

streaming_anomaly_weights = {
    "streaming_anomaly_variation_signal_percentile": 0.25,
    "streaming_anomaly_growth_signal_percentile": 0.25,
    "streaming_anomaly_daily_change_signal_percentile": 0.20,
    "streaming_anomaly_instability_signal_percentile": 0.15,
    "streaming_anomaly_current_history_signal_percentile": 0.15,
}

streaming_anomaly_weighted_sum = pd.Series(
    0.0,
    index=streaming_anomaly_df.index
)

streaming_anomaly_available_weight_sum = pd.Series(
    0.0,
    index=streaming_anomaly_df.index
)

for column, weight in streaming_anomaly_weights.items():

    available_mask = (
        streaming_anomaly_df[
            column
        ].notna()
    )

    streaming_anomaly_weighted_sum.loc[
        available_mask
    ] += (
        streaming_anomaly_df.loc[
            available_mask,
            column
        ]
        * weight
    )

    streaming_anomaly_available_weight_sum.loc[
        available_mask
    ] += weight

streaming_anomaly_df[
    "streaming_anomaly_score"
] = np.where(
    streaming_anomaly_available_weight_sum > 0,
    (
        streaming_anomaly_weighted_sum
        / streaming_anomaly_available_weight_sum
    ),
    np.nan,
)

# ============================================================
# Percentile and rank
# ============================================================

valid_streaming_anomaly_mask = (
    streaming_anomaly_df[
        "streaming_anomaly_score"
    ].notna()
)

streaming_anomaly_df[
    "streaming_anomaly_percentile"
] = np.nan

if valid_streaming_anomaly_mask.any():

    streaming_anomaly_df.loc[
        valid_streaming_anomaly_mask,
        "streaming_anomaly_percentile"
    ] = (
        streaming_anomaly_df.loc[
            valid_streaming_anomaly_mask,
            "streaming_anomaly_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

streaming_anomaly_df[
    "streaming_anomaly_rank"
] = (
    streaming_anomaly_df[
        "streaming_anomaly_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Streaming anomaly classes
# ============================================================

streaming_anomaly_df[
    "streaming_anomaly_class"
] = "No Streaming Anomaly Evidence"

streaming_anomaly_df.loc[
    valid_streaming_anomaly_mask
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] < 75
    ),
    "streaming_anomaly_class"
] = "Normal Streaming Behaviour"

streaming_anomaly_df.loc[
    valid_streaming_anomaly_mask
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] >= 75
    )
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] < 90
    ),
    "streaming_anomaly_class"
] = "Elevated Streaming Anomaly"

streaming_anomaly_df.loc[
    valid_streaming_anomaly_mask
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] >= 90
    )
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] < 97.5
    ),
    "streaming_anomaly_class"
] = "Strong Streaming Anomaly"

streaming_anomaly_df.loc[
    valid_streaming_anomaly_mask
    & (
        streaming_anomaly_df[
            "streaming_anomaly_score"
        ] >= 97.5
    ),
    "streaming_anomaly_class"
] = "Exceptional Streaming Anomaly"

# ============================================================
# Binary anomaly flag
# ============================================================

if valid_streaming_anomaly_mask.any():

    streaming_anomaly_flag_threshold = (
        streaming_anomaly_df.loc[
            valid_streaming_anomaly_mask,
            "streaming_anomaly_score"
        ]
        .quantile(0.90)
    )

else:

    streaming_anomaly_flag_threshold = np.nan

streaming_anomaly_df[
    "streaming_anomaly_flag"
] = (
    streaming_anomaly_df[
        "streaming_anomaly_score"
    ] >= streaming_anomaly_flag_threshold
)

streaming_anomaly_df.loc[
    ~valid_streaming_anomaly_mask,
    "streaming_anomaly_flag"
] = False

# ============================================================
# Evidence strength
# ============================================================

streaming_anomaly_df[
    "streaming_anomaly_evidence_strength"
] = np.select(
    [
        streaming_anomaly_df[
            "streaming_anomaly_components_available"
        ] >= 4,

        streaming_anomaly_df[
            "streaming_anomaly_components_available"
        ] == 3,

        streaming_anomaly_df[
            "streaming_anomaly_components_available"
        ] == 2,

        streaming_anomaly_df[
            "streaming_anomaly_components_available"
        ] == 1,
    ],
    [
        "Very Strong Streaming Anomaly Evidence",
        "Strong Streaming Anomaly Evidence",
        "Moderate Streaming Anomaly Evidence",
        "Limited Streaming Anomaly Evidence",
    ],
    default="No Streaming Anomaly Evidence",
)

# ============================================================
# Summary
# ============================================================

streaming_anomaly_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                streaming_anomaly_df
            ),
        },
        {
            "Metric": "Releases with streaming anomaly evidence",
            "Observed Value": int(
                valid_streaming_anomaly_mask.sum()
            ),
        },
        {
            "Metric": "Releases without streaming anomaly evidence",
            "Observed Value": int(
                (~valid_streaming_anomaly_mask).sum()
            ),
        },
        {
            "Metric": "Median streaming anomaly score",
            "Observed Value": (
                streaming_anomaly_df.loc[
                    valid_streaming_anomaly_mask,
                    "streaming_anomaly_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean streaming anomaly score",
            "Observed Value": (
                streaming_anomaly_df.loc[
                    valid_streaming_anomaly_mask,
                    "streaming_anomaly_score"
                ].mean()
            ),
        },
        {
            "Metric": "Streaming anomaly flag threshold",
            "Observed Value": streaming_anomaly_flag_threshold,
        },
        {
            "Metric": "Releases flagged as streaming anomalies",
            "Observed Value": int(
                streaming_anomaly_df[
                    "streaming_anomaly_flag"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP streaming anomalies")
print("=" * 80)

display(
    streaming_anomaly_summary_df
)

# ============================================================
# Class distribution
# ============================================================

streaming_anomaly_class_distribution_df = (
    streaming_anomaly_df[
        "streaming_anomaly_class"
    ]
    .value_counts()
    .rename_axis(
        "Streaming Anomaly Class"
    )
    .reset_index(
        name="Releases"
    )
)

streaming_anomaly_class_distribution_df[
    "Share (%)"
] = (
    streaming_anomaly_class_distribution_df[
        "Releases"
    ]
    / len(
        streaming_anomaly_df
    )
    * 100
)

print()
print("Streaming anomaly class distribution")
print("=" * 80)

display(
    streaming_anomaly_class_distribution_df
)

# ============================================================
# Evidence distribution
# ============================================================

streaming_anomaly_evidence_distribution_df = (
    streaming_anomaly_df[
        "streaming_anomaly_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Streaming Anomaly Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

streaming_anomaly_evidence_distribution_df[
    "Share (%)"
] = (
    streaming_anomaly_evidence_distribution_df[
        "Releases"
    ]
    / len(
        streaming_anomaly_df
    )
    * 100
)

print()
print("Streaming anomaly evidence distribution")
print("=" * 80)

display(
    streaming_anomaly_evidence_distribution_df
)

# ============================================================
# Strongest streaming anomalies
# ============================================================

top_streaming_anomalies_df = (
    streaming_anomaly_df.loc[
        valid_streaming_anomaly_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "streaming_anomaly_variation_signal",
            "streaming_anomaly_growth_signal",
            "streaming_anomaly_daily_change_signal",
            "streaming_anomaly_instability_signal",
            "streaming_anomaly_current_history_signal",
            "streaming_anomaly_components_available",
            "streaming_anomaly_evidence_pct",
            "streaming_anomaly_evidence_strength",
            "streaming_anomaly_score",
            "streaming_anomaly_percentile",
            "streaming_anomaly_rank",
            "streaming_anomaly_class",
            "streaming_anomaly_flag",
        ],
    ]
    .sort_values(
        by="streaming_anomaly_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strongest PMIP streaming anomalies")
print("=" * 80)

display(
    top_streaming_anomalies_df
)

# ============================================================
# Validation
# ============================================================

streaming_anomaly_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 10.2 completed",
            "Passed": section_10_2_complete,
        },
        {
            "Validation Area": "Streaming anomaly signals available",
            "Passed": streaming_anomaly_df[
                streaming_anomaly_signal_columns
            ].notna().any().any(),
        },
        {
            "Validation Area": "Streaming anomaly score calculated",
            "Passed": streaming_anomaly_df[
                "streaming_anomaly_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Streaming anomaly percentile calculated",
            "Passed": streaming_anomaly_df[
                "streaming_anomaly_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Streaming anomaly classes assigned",
            "Passed": streaming_anomaly_df[
                "streaming_anomaly_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Streaming anomaly flags assigned",
            "Passed": streaming_anomaly_df[
                "streaming_anomaly_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Streaming anomaly evidence classified",
            "Passed": streaming_anomaly_df[
                "streaming_anomaly_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Streaming Anomaly validation")
print("=" * 80)

display(
    streaming_anomaly_validation_df
)

# ============================================================
# Completion
# ============================================================

section_10_3_complete = bool(
    streaming_anomaly_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 10.3 completion status: "
    f"{section_10_3_complete}"
)

if not section_10_3_complete:
    raise RuntimeError(
        "Streaming Anomaly analysis "
        "did not complete successfully."
    )
```

### Interpretation

The Streaming Anomalies assessment identified releases whose streaming behaviour differs noticeably from typical historical patterns.

A total of 2,235 releases had sufficient anomaly evidence, while 2,358 releases did not.

Most assessed releases were classified as showing normal streaming behaviour.

A total of 224 releases were flagged as streaming anomalies, while 37 releases were classified as `Strong Streaming Anomaly` and 8 as `Exceptional Streaming Anomaly`.

This indicates that extreme streaming behaviour is relatively uncommon within the PMIP population.

These anomaly flags should be treated as signals for deeper contextual investigation rather than evidence of abnormal or suspicious activity.

## 10.4 Growth Without Geographic Breadth

### Purpose

This subsection identifies releases that show strong streaming or performance growth without a similarly broad geographic reach.

The aim is to detect releases whose audience growth appears concentrated within a relatively limited number of markets.

The assessment considers:

- short-term streaming growth;
- overall streaming performance;
- geographic reach;
- market breadth;
- the mismatch between performance growth and geographic coverage;
- the amount of evidence supporting each assessment.

A release can perform strongly without broad geographic reach for legitimate reasons, including regional popularity, localised promotion, market-specific virality or concentrated playlist exposure.

The purpose is therefore to identify geographically concentrated growth patterns for further PMIP interpretation rather than to treat them as abnormal behaviour.

```python
# ============================================================
# Section 10.4 - Growth Without Geographic Breadth
# ============================================================

if not section_10_3_complete:
    raise RuntimeError(
        "Section 10.3 must be completed before assessing "
        "Growth Without Geographic Breadth."
    )

# ============================================================
# Working source
# ============================================================

growth_without_breadth_df = streaming_anomaly_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_growth_breadth_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_growth_breadth_fields = [
    field
    for field in required_growth_breadth_fields
    if field not in growth_without_breadth_df.columns
]

if missing_growth_breadth_fields:
    raise RuntimeError(
        "Required Growth Without Geographic Breadth fields "
        f"are missing: {missing_growth_breadth_fields}"
    )

# ============================================================
# Prepare growth-strength signal
# ============================================================

growth_signal_candidates = [
    field
    for field in [
        "short_term_spike_score",
        "streaming_performance_score",
        "percentage_stream_growth",
        "artist_release_relationship_score",
    ]
    if field in growth_without_breadth_df.columns
]

if not growth_signal_candidates:
    raise RuntimeError(
        "No suitable growth-performance signal is available "
        "for Growth Without Geographic Breadth analysis."
    )

# ============================================================
# Prepare geographic-breadth signal
# ============================================================

geographic_signal_candidates = [
    field
    for field in [
        "geographic_reach_score",
        "market_breadth_score",
        "countries_represented_score",
        "market_breadth_pct",
        "countries_charted",
    ]
    if field in growth_without_breadth_df.columns
]

if not geographic_signal_candidates:
    raise RuntimeError(
        "No suitable geographic breadth signal is available "
        "for Growth Without Geographic Breadth analysis."
    )

# ============================================================
# Growth component
#
# Prefer PMIP short-term spike score where available.
# Otherwise fall back to broader streaming performance.
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_growth_component"
] = np.nan

if "short_term_spike_score" in growth_without_breadth_df.columns:
    growth_without_breadth_df[
        "growth_without_breadth_growth_component"
    ] = growth_without_breadth_df[
        "short_term_spike_score"
    ]

if "streaming_performance_score" in growth_without_breadth_df.columns:

    missing_growth_component_mask = (
        growth_without_breadth_df[
            "growth_without_breadth_growth_component"
        ].isna()
    )

    growth_without_breadth_df.loc[
        missing_growth_component_mask,
        "growth_without_breadth_growth_component"
    ] = growth_without_breadth_df.loc[
        missing_growth_component_mask,
        "streaming_performance_score"
    ]

if "artist_release_relationship_score" in growth_without_breadth_df.columns:

    missing_growth_component_mask = (
        growth_without_breadth_df[
            "growth_without_breadth_growth_component"
        ].isna()
    )

    growth_without_breadth_df.loc[
        missing_growth_component_mask,
        "growth_without_breadth_growth_component"
    ] = growth_without_breadth_df.loc[
        missing_growth_component_mask,
        "artist_release_relationship_score"
    ]

# ============================================================
# Geographic breadth component
#
# Prefer the final PMIP Geographic Reach Indicator.
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_geographic_component"
] = np.nan

if "geographic_reach_score" in growth_without_breadth_df.columns:

    growth_without_breadth_df[
        "growth_without_breadth_geographic_component"
    ] = growth_without_breadth_df[
        "geographic_reach_score"
    ]

elif "market_breadth_score" in growth_without_breadth_df.columns:

    growth_without_breadth_df[
        "growth_without_breadth_geographic_component"
    ] = growth_without_breadth_df[
        "market_breadth_score"
    ]

elif "countries_represented_score" in growth_without_breadth_df.columns:

    growth_without_breadth_df[
        "growth_without_breadth_geographic_component"
    ] = growth_without_breadth_df[
        "countries_represented_score"
    ]

# ============================================================
# Optional raw geographic breadth support
# ============================================================

if "market_breadth_pct" in growth_without_breadth_df.columns:
    growth_without_breadth_df[
        "growth_without_breadth_market_pct"
    ] = growth_without_breadth_df[
        "market_breadth_pct"
    ]
else:
    growth_without_breadth_df[
        "growth_without_breadth_market_pct"
    ] = np.nan

if "countries_charted" in growth_without_breadth_df.columns:
    growth_without_breadth_df[
        "growth_without_breadth_countries"
    ] = growth_without_breadth_df[
        "countries_charted"
    ]
else:
    growth_without_breadth_df[
        "growth_without_breadth_countries"
    ] = np.nan

# ============================================================
# Evidence mask
# ============================================================

valid_growth_breadth_mask = (
    growth_without_breadth_df[
        "growth_without_breadth_growth_component"
    ].notna()
    & growth_without_breadth_df[
        "growth_without_breadth_geographic_component"
    ].notna()
)

# ============================================================
# Mismatch score
#
# High growth + low geographic breadth = larger anomaly.
# Both components are already approximately 0-100.
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_raw_gap"
] = (
    growth_without_breadth_df[
        "growth_without_breadth_growth_component"
    ]
    - growth_without_breadth_df[
        "growth_without_breadth_geographic_component"
    ]
)

# Only positive gaps represent growth exceeding geographic reach.
growth_without_breadth_df[
    "growth_without_breadth_positive_gap"
] = (
    growth_without_breadth_df[
        "growth_without_breadth_raw_gap"
    ]
    .clip(lower=0)
)

# ============================================================
# Relative mismatch percentile
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_percentile"
] = np.nan

valid_positive_gap_mask = (
    valid_growth_breadth_mask
    & (
        growth_without_breadth_df[
            "growth_without_breadth_positive_gap"
        ] > 0
    )
)

if valid_positive_gap_mask.any():

    growth_without_breadth_df.loc[
        valid_positive_gap_mask,
        "growth_without_breadth_percentile"
    ] = (
        growth_without_breadth_df.loc[
            valid_positive_gap_mask,
            "growth_without_breadth_positive_gap"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Final growth-without-breadth score
#
# Require both:
# - strong performance/growth
# - meaningful geographic mismatch
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_score"
] = np.nan

growth_without_breadth_df.loc[
    valid_growth_breadth_mask,
    "growth_without_breadth_score"
] = (
    0.60
    * growth_without_breadth_df.loc[
        valid_growth_breadth_mask,
        "growth_without_breadth_positive_gap"
    ]
    +
    0.40
    * growth_without_breadth_df.loc[
        valid_growth_breadth_mask,
        "growth_without_breadth_growth_component"
    ]
)

# ============================================================
# Rank
# ============================================================

valid_growth_without_breadth_score_mask = (
    growth_without_breadth_df[
        "growth_without_breadth_score"
    ].notna()
)

growth_without_breadth_df[
    "growth_without_breadth_rank"
] = (
    growth_without_breadth_df[
        "growth_without_breadth_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Classes
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_class"
] = "No Growth–Breadth Evidence"

growth_without_breadth_df.loc[
    valid_growth_without_breadth_score_mask
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] < 50
    ),
    "growth_without_breadth_class"
] = "Broadly Balanced Growth"

growth_without_breadth_df.loc[
    valid_growth_without_breadth_score_mask
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] >= 50
    )
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] < 65
    ),
    "growth_without_breadth_class"
] = "Moderately Concentrated Growth"

growth_without_breadth_df.loc[
    valid_growth_without_breadth_score_mask
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] >= 65
    )
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] < 80
    ),
    "growth_without_breadth_class"
] = "Strong Growth Without Breadth"

growth_without_breadth_df.loc[
    valid_growth_without_breadth_score_mask
    & (
        growth_without_breadth_df[
            "growth_without_breadth_score"
        ] >= 80
    ),
    "growth_without_breadth_class"
] = "Exceptional Growth Without Breadth"

# ============================================================
# Binary flag
#
# Conservative upper-tail threshold among scored releases.
# ============================================================

if valid_growth_without_breadth_score_mask.any():

    growth_without_breadth_flag_threshold = (
        growth_without_breadth_df.loc[
            valid_growth_without_breadth_score_mask,
            "growth_without_breadth_score"
        ]
        .quantile(0.90)
    )

else:

    growth_without_breadth_flag_threshold = np.nan

growth_without_breadth_df[
    "growth_without_breadth_flag"
] = (
    growth_without_breadth_df[
        "growth_without_breadth_score"
    ] >= growth_without_breadth_flag_threshold
)

growth_without_breadth_df.loc[
    ~valid_growth_without_breadth_score_mask,
    "growth_without_breadth_flag"
] = False

# ============================================================
# Evidence strength
# ============================================================

growth_without_breadth_df[
    "growth_without_breadth_evidence_strength"
] = np.select(
    [
        valid_growth_breadth_mask
        & growth_without_breadth_df[
            "growth_without_breadth_market_pct"
        ].notna()
        & growth_without_breadth_df[
            "growth_without_breadth_countries"
        ].notna(),

        valid_growth_breadth_mask
        & (
            growth_without_breadth_df[
                "growth_without_breadth_market_pct"
            ].notna()
            | growth_without_breadth_df[
                "growth_without_breadth_countries"
            ].notna()
        ),

        valid_growth_breadth_mask,
    ],
    [
        "Strong Growth–Breadth Evidence",
        "Moderate Growth–Breadth Evidence",
        "Limited Growth–Breadth Evidence",
    ],
    default="No Growth–Breadth Evidence",
)

# ============================================================
# Summary
# ============================================================

growth_without_breadth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                growth_without_breadth_df
            ),
        },
        {
            "Metric": "Releases with growth-breadth evidence",
            "Observed Value": int(
                valid_growth_breadth_mask.sum()
            ),
        },
        {
            "Metric": "Releases without growth-breadth evidence",
            "Observed Value": int(
                (~valid_growth_breadth_mask).sum()
            ),
        },
        {
            "Metric": "Median growth component",
            "Observed Value": (
                growth_without_breadth_df.loc[
                    valid_growth_breadth_mask,
                    "growth_without_breadth_growth_component"
                ].median()
            ),
        },
        {
            "Metric": "Median geographic component",
            "Observed Value": (
                growth_without_breadth_df.loc[
                    valid_growth_breadth_mask,
                    "growth_without_breadth_geographic_component"
                ].median()
            ),
        },
        {
            "Metric": "Median positive growth-breadth gap",
            "Observed Value": (
                growth_without_breadth_df.loc[
                    valid_growth_breadth_mask,
                    "growth_without_breadth_positive_gap"
                ].median()
            ),
        },
        {
            "Metric": "Median growth-without-breadth score",
            "Observed Value": (
                growth_without_breadth_df.loc[
                    valid_growth_without_breadth_score_mask,
                    "growth_without_breadth_score"
                ].median()
            ),
        },
        {
            "Metric": "Growth-without-breadth flag threshold",
            "Observed Value": growth_without_breadth_flag_threshold,
        },
        {
            "Metric": "Releases flagged for concentrated growth",
            "Observed Value": int(
                growth_without_breadth_df[
                    "growth_without_breadth_flag"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP growth without geographic breadth")
print("=" * 80)

display(
    growth_without_breadth_summary_df
)

# ============================================================
# Class distribution
# ============================================================

growth_without_breadth_class_distribution_df = (
    growth_without_breadth_df[
        "growth_without_breadth_class"
    ]
    .value_counts()
    .rename_axis(
        "Growth Without Geographic Breadth Class"
    )
    .reset_index(
        name="Releases"
    )
)

growth_without_breadth_class_distribution_df[
    "Share (%)"
] = (
    growth_without_breadth_class_distribution_df[
        "Releases"
    ]
    / len(
        growth_without_breadth_df
    )
    * 100
)

print()
print("Growth-without-geographic-breadth class distribution")
print("=" * 80)

display(
    growth_without_breadth_class_distribution_df
)

# ============================================================
# Evidence distribution
# ============================================================

growth_without_breadth_evidence_distribution_df = (
    growth_without_breadth_df[
        "growth_without_breadth_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Growth–Breadth Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

growth_without_breadth_evidence_distribution_df[
    "Share (%)"
] = (
    growth_without_breadth_evidence_distribution_df[
        "Releases"
    ]
    / len(
        growth_without_breadth_df
    )
    * 100
)

print()
print("Growth–breadth evidence distribution")
print("=" * 80)

display(
    growth_without_breadth_evidence_distribution_df
)

# ============================================================
# Strongest concentrated-growth releases
# ============================================================

top_growth_without_breadth_df = (
    growth_without_breadth_df.loc[
        valid_growth_without_breadth_score_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "growth_without_breadth_growth_component",
            "growth_without_breadth_geographic_component",
            "growth_without_breadth_market_pct",
            "growth_without_breadth_countries",
            "growth_without_breadth_raw_gap",
            "growth_without_breadth_positive_gap",
            "growth_without_breadth_percentile",
            "growth_without_breadth_evidence_strength",
            "growth_without_breadth_score",
            "growth_without_breadth_rank",
            "growth_without_breadth_class",
            "growth_without_breadth_flag",
        ],
    ]
    .sort_values(
        by="growth_without_breadth_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strongest PMIP growth-without-geographic-breadth cases")
print("=" * 80)

display(
    top_growth_without_breadth_df
)

# ============================================================
# Validation
# ============================================================

growth_without_breadth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 10.3 completed",
            "Passed": section_10_3_complete,
        },
        {
            "Validation Area": "Growth-performance signal available",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_growth_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Geographic breadth signal available",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_geographic_component"
            ].notna().any(),
        },
        {
            "Validation Area": "Growth-breadth mismatch calculated",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_raw_gap"
            ].notna().any(),
        },
        {
            "Validation Area": "Growth-without-breadth score calculated",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Growth-without-breadth classes assigned",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Growth-without-breadth flags assigned",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Growth-breadth evidence classified",
            "Passed": growth_without_breadth_df[
                "growth_without_breadth_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Growth Without Geographic Breadth validation")
print("=" * 80)

display(
    growth_without_breadth_validation_df
)

# ============================================================
# Completion
# ============================================================

section_10_4_complete = bool(
    growth_without_breadth_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 10.4 completion status: "
    f"{section_10_4_complete}"
)

if not section_10_4_complete:
    raise RuntimeError(
        "Growth Without Geographic Breadth analysis "
        "did not complete successfully."
    )
```

### Interpretation

The Growth Without Geographic Breadth assessment identified releases whose performance strength exceeds their geographic reach.

A total of 2,235 releases had sufficient growth–breadth evidence, while 2,358 releases did not.

Most assessed releases were classified as `Broadly Balanced Growth`, indicating that stronger performance was generally accompanied by comparable geographic reach.

Only a small number showed clearly concentrated growth, including 7 releases classified as `Strong Growth Without Breadth` and 1 as `Exceptional Growth Without Breadth`.

A total of 224 releases were flagged in the upper tail for further investigation.

These findings suggest that strong performance without matching geographic breadth is relatively uncommon within the PMIP population.

## 10.5 Unusual Performance Flags

### Purpose

This subsection combines the unusual-performance signals developed across Section 10 into a single release-level flagging framework.

The aim is to identify releases that show one or more unusual characteristics across:

- short-term streaming spikes;
- sudden chart movement;
- broader streaming anomalies;
- growth without matching geographic breadth.

The combined assessment distinguishes isolated signals from releases showing multiple unusual patterns at the same time.

This makes it possible to prioritise releases for further PMIP investigation while preserving the individual anomaly signals that contributed to each result.

The final flags should be interpreted as evidence of unusual performance behaviour rather than proof of suspicious activity.

```python
# ============================================================
# Section 10.5 - Unusual Performance Flags
# ============================================================

if not section_10_4_complete:
    raise RuntimeError(
        "Section 10.4 must be completed before constructing "
        "Unusual Performance Flags."
    )

# ============================================================
# Working source
# ============================================================

unusual_performance_df = growth_without_breadth_df.copy()

# ============================================================
# Validate required anomaly flags
# ============================================================

required_unusual_flag_fields = [
    "release_id",
    "Track",
    "Artist",
    "short_term_spike_flag",
    "sudden_chart_movement_flag",
    "streaming_anomaly_flag",
    "growth_without_breadth_flag",
]

missing_unusual_flag_fields = [
    field
    for field in required_unusual_flag_fields
    if field not in unusual_performance_df.columns
]

if missing_unusual_flag_fields:
    raise RuntimeError(
        "Required unusual-performance flag fields are missing: "
        f"{missing_unusual_flag_fields}"
    )

# ============================================================
# Flag components
# ============================================================

unusual_flag_columns = [
    "short_term_spike_flag",
    "sudden_chart_movement_flag",
    "streaming_anomaly_flag",
    "growth_without_breadth_flag",
]

# Ensure boolean format
for column in unusual_flag_columns:
    unusual_performance_df[column] = (
        unusual_performance_df[column]
        .fillna(False)
        .astype(bool)
    )

# ============================================================
# Count unusual signals
# ============================================================

unusual_performance_df[
    "unusual_performance_flag_count"
] = (
    unusual_performance_df[
        unusual_flag_columns
    ]
    .sum(axis=1)
)

# ============================================================
# Evidence availability
# ============================================================

evidence_availability_columns = []

evidence_mapping = {
    "short_term_spike_score":
        "short_term_spike_evidence_available",

    "sudden_chart_movement_score":
        "sudden_chart_movement_evidence_available",

    "streaming_anomaly_score":
        "streaming_anomaly_evidence_available",

    "growth_without_breadth_score":
        "growth_without_breadth_evidence_available",
}

for score_column, evidence_column in evidence_mapping.items():

    unusual_performance_df[
        evidence_column
    ] = (
        unusual_performance_df[
            score_column
        ].notna()
    )

    evidence_availability_columns.append(
        evidence_column
    )

unusual_performance_df[
    "unusual_performance_evidence_count"
] = (
    unusual_performance_df[
        evidence_availability_columns
    ]
    .sum(axis=1)
)

unusual_performance_df[
    "unusual_performance_evidence_coverage_pct"
] = (
    unusual_performance_df[
        "unusual_performance_evidence_count"
    ]
    / len(
        evidence_availability_columns
    )
    * 100
)

# ============================================================
# Combined anomaly intensity
#
# Uses the available Section 10 anomaly scores.
# Missing scores are not treated as zero.
# ============================================================

unusual_score_columns = [
    "short_term_spike_score",
    "sudden_chart_movement_score",
    "streaming_anomaly_score",
    "growth_without_breadth_score",
]

unusual_performance_df[
    "unusual_performance_score"
] = (
    unusual_performance_df[
        unusual_score_columns
    ]
    .mean(
        axis=1,
        skipna=True
    )
)

# ============================================================
# Combined flag strength
# ============================================================

unusual_performance_df[
    "unusual_performance_flag_class"
] = np.select(
    [
        unusual_performance_df[
            "unusual_performance_flag_count"
        ] == 0,

        unusual_performance_df[
            "unusual_performance_flag_count"
        ] == 1,

        unusual_performance_df[
            "unusual_performance_flag_count"
        ] == 2,

        unusual_performance_df[
            "unusual_performance_flag_count"
        ] == 3,

        unusual_performance_df[
            "unusual_performance_flag_count"
        ] >= 4,
    ],
    [
        "No Unusual Performance Flag",
        "Single Unusual Performance Signal",
        "Multiple Unusual Performance Signals",
        "Strong Multi-Signal Unusual Performance",
        "Exceptional Multi-Signal Unusual Performance",
    ],
    default="No Unusual Performance Flag",
)

# ============================================================
# Overall binary unusual-performance flag
#
# Any Section 10 signal can trigger the broad flag.
# ============================================================

unusual_performance_df[
    "unusual_performance_flag"
] = (
    unusual_performance_df[
        "unusual_performance_flag_count"
    ] > 0
)

# ============================================================
# Higher-priority investigation flag
#
# Requires:
# - at least two anomaly signals
# OR
# - one signal with an extreme combined score.
# ============================================================

valid_unusual_score_mask = (
    unusual_performance_df[
        "unusual_performance_score"
    ].notna()
)

if valid_unusual_score_mask.any():

    unusual_score_extreme_threshold = (
        unusual_performance_df.loc[
            valid_unusual_score_mask,
            "unusual_performance_score"
        ]
        .quantile(0.95)
    )

else:

    unusual_score_extreme_threshold = np.nan

unusual_performance_df[
    "priority_unusual_performance_flag"
] = (
    unusual_performance_df[
        "unusual_performance_flag_count"
    ] >= 2
) | (
    unusual_performance_df[
        "unusual_performance_score"
    ] >= unusual_score_extreme_threshold
)

unusual_performance_df.loc[
    ~valid_unusual_score_mask,
    "priority_unusual_performance_flag"
] = False

# ============================================================
# Dominant unusual-performance pattern
# ============================================================

def identify_dominant_unusual_pattern(row):

    active_patterns = []

    if row["short_term_spike_flag"]:
        active_patterns.append("Short-Term Streaming Spike")

    if row["sudden_chart_movement_flag"]:
        active_patterns.append("Sudden Chart Movement")

    if row["streaming_anomaly_flag"]:
        active_patterns.append("Streaming Anomaly")

    if row["growth_without_breadth_flag"]:
        active_patterns.append("Growth Without Geographic Breadth")

    if len(active_patterns) == 0:
        return "No Unusual Pattern"

    if len(active_patterns) == 1:
        return active_patterns[0]

    return " + ".join(active_patterns)


unusual_performance_df[
    "unusual_performance_pattern"
] = unusual_performance_df.apply(
    identify_dominant_unusual_pattern,
    axis=1
)

# ============================================================
# Evidence strength
# ============================================================

unusual_performance_df[
    "unusual_performance_evidence_strength"
] = np.select(
    [
        unusual_performance_df[
            "unusual_performance_evidence_count"
        ] == 4,

        unusual_performance_df[
            "unusual_performance_evidence_count"
        ] == 3,

        unusual_performance_df[
            "unusual_performance_evidence_count"
        ] == 2,

        unusual_performance_df[
            "unusual_performance_evidence_count"
        ] == 1,
    ],
    [
        "Very Strong Unusual-Performance Evidence",
        "Strong Unusual-Performance Evidence",
        "Moderate Unusual-Performance Evidence",
        "Limited Unusual-Performance Evidence",
    ],
    default="No Unusual-Performance Evidence",
)

# ============================================================
# Rank
# ============================================================

unusual_performance_df[
    "unusual_performance_rank"
] = (
    unusual_performance_df[
        "unusual_performance_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Summary
# ============================================================

unusual_performance_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                unusual_performance_df
            ),
        },
        {
            "Metric": "Releases with at least one unusual flag",
            "Observed Value": int(
                unusual_performance_df[
                    "unusual_performance_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with no unusual flags",
            "Observed Value": int(
                (
                    ~unusual_performance_df[
                        "unusual_performance_flag"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Releases with multiple unusual signals",
            "Observed Value": int(
                (
                    unusual_performance_df[
                        "unusual_performance_flag_count"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "Metric": "Priority unusual-performance cases",
            "Observed Value": int(
                unusual_performance_df[
                    "priority_unusual_performance_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Median unusual-performance score",
            "Observed Value": (
                unusual_performance_df.loc[
                    valid_unusual_score_mask,
                    "unusual_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean unusual-performance score",
            "Observed Value": (
                unusual_performance_df.loc[
                    valid_unusual_score_mask,
                    "unusual_performance_score"
                ].mean()
            ),
        },
        {
            "Metric": "95th-percentile priority score threshold",
            "Observed Value": unusual_score_extreme_threshold,
        },
        {
            "Metric": "Median unusual-performance evidence coverage (%)",
            "Observed Value": (
                unusual_performance_df[
                    "unusual_performance_evidence_coverage_pct"
                ].median()
            ),
        },
    ]
)

print("Constructing PMIP Unusual Performance Flags")
print("=" * 80)

display(
    unusual_performance_summary_df
)

# ============================================================
# Individual flag prevalence
# ============================================================

individual_flag_rows = []

for flag_column, label in [
    (
        "short_term_spike_flag",
        "Short-Term Streaming Spike"
    ),
    (
        "sudden_chart_movement_flag",
        "Sudden Chart Movement"
    ),
    (
        "streaming_anomaly_flag",
        "Streaming Anomaly"
    ),
    (
        "growth_without_breadth_flag",
        "Growth Without Geographic Breadth"
    ),
]:

    release_count = int(
        unusual_performance_df[
            flag_column
        ].sum()
    )

    individual_flag_rows.append(
        {
            "Unusual Performance Signal": label,
            "Flagged Releases": release_count,
            "Share (%)": (
                release_count
                / len(
                    unusual_performance_df
                )
                * 100
            ),
        }
    )

individual_unusual_flag_distribution_df = pd.DataFrame(
    individual_flag_rows
)

print()
print("Individual unusual-performance signal prevalence")
print("=" * 80)

display(
    individual_unusual_flag_distribution_df
)

# ============================================================
# Flag-count distribution
# ============================================================

unusual_flag_count_distribution_df = (
    unusual_performance_df[
        "unusual_performance_flag_count"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Number of Unusual Signals"
    )
    .reset_index(
        name="Releases"
    )
)

unusual_flag_count_distribution_df[
    "Share (%)"
] = (
    unusual_flag_count_distribution_df[
        "Releases"
    ]
    / len(
        unusual_performance_df
    )
    * 100
)

print()
print("Unusual-performance flag-count distribution")
print("=" * 80)

display(
    unusual_flag_count_distribution_df
)

# ============================================================
# Flag class distribution
# ============================================================

unusual_flag_class_distribution_df = (
    unusual_performance_df[
        "unusual_performance_flag_class"
    ]
    .value_counts()
    .rename_axis(
        "Unusual Performance Flag Class"
    )
    .reset_index(
        name="Releases"
    )
)

unusual_flag_class_distribution_df[
    "Share (%)"
] = (
    unusual_flag_class_distribution_df[
        "Releases"
    ]
    / len(
        unusual_performance_df
    )
    * 100
)

print()
print("Unusual-performance flag class distribution")
print("=" * 80)

display(
    unusual_flag_class_distribution_df
)

# ============================================================
# Pattern distribution
# ============================================================

unusual_pattern_distribution_df = (
    unusual_performance_df[
        "unusual_performance_pattern"
    ]
    .value_counts()
    .rename_axis(
        "Unusual Performance Pattern"
    )
    .reset_index(
        name="Releases"
    )
)

unusual_pattern_distribution_df[
    "Share (%)"
] = (
    unusual_pattern_distribution_df[
        "Releases"
    ]
    / len(
        unusual_performance_df
    )
    * 100
)

print()
print("Unusual-performance pattern distribution")
print("=" * 80)

display(
    unusual_pattern_distribution_df
)

# ============================================================
# Evidence distribution
# ============================================================

unusual_evidence_distribution_df = (
    unusual_performance_df[
        "unusual_performance_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Unusual Performance Evidence"
    )
    .reset_index(
        name="Releases"
    )
)

unusual_evidence_distribution_df[
    "Share (%)"
] = (
    unusual_evidence_distribution_df[
        "Releases"
    ]
    / len(
        unusual_performance_df
    )
    * 100
)

print()
print("Unusual-performance evidence distribution")
print("=" * 80)

display(
    unusual_evidence_distribution_df
)

# ============================================================
# Highest-priority unusual releases
# ============================================================

top_unusual_performance_df = (
    unusual_performance_df.loc[
        unusual_performance_df[
            "priority_unusual_performance_flag"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "short_term_spike_flag",
            "sudden_chart_movement_flag",
            "streaming_anomaly_flag",
            "growth_without_breadth_flag",
            "unusual_performance_flag_count",
            "unusual_performance_pattern",
            "unusual_performance_score",
            "unusual_performance_evidence_count",
            "unusual_performance_evidence_coverage_pct",
            "unusual_performance_evidence_strength",
            "unusual_performance_rank",
            "priority_unusual_performance_flag",
        ],
    ]
    .sort_values(
        by=[
            "unusual_performance_flag_count",
            "unusual_performance_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest-priority PMIP unusual-performance releases")
print("=" * 80)

display(
    top_unusual_performance_df
)

# ============================================================
# Final Section 10 dataset
# ============================================================

unusual_release_patterns_final_df = (
    unusual_performance_df.copy()
)

# ============================================================
# Validation
# ============================================================

unusual_performance_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 10.4 completed",
            "Passed": section_10_4_complete,
        },
        {
            "Validation Area": "Short-term spike flag available",
            "Passed": "short_term_spike_flag"
            in unusual_performance_df.columns,
        },
        {
            "Validation Area": "Sudden chart movement flag available",
            "Passed": "sudden_chart_movement_flag"
            in unusual_performance_df.columns,
        },
        {
            "Validation Area": "Streaming anomaly flag available",
            "Passed": "streaming_anomaly_flag"
            in unusual_performance_df.columns,
        },
        {
            "Validation Area": "Growth-without-breadth flag available",
            "Passed": "growth_without_breadth_flag"
            in unusual_performance_df.columns,
        },
        {
            "Validation Area": "Combined unusual flag count calculated",
            "Passed": unusual_performance_df[
                "unusual_performance_flag_count"
            ].notna().all(),
        },
        {
            "Validation Area": "Combined unusual score calculated",
            "Passed": unusual_performance_df[
                "unusual_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Unusual performance patterns assigned",
            "Passed": unusual_performance_df[
                "unusual_performance_pattern"
            ].notna().all(),
        },
        {
            "Validation Area": "Priority flags assigned",
            "Passed": unusual_performance_df[
                "priority_unusual_performance_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Evidence strength classified",
            "Passed": unusual_performance_df[
                "unusual_performance_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Unusual Performance Flags validation")
print("=" * 80)

display(
    unusual_performance_validation_df
)

# ============================================================
# Completion
# ============================================================

section_10_5_complete = bool(
    unusual_performance_validation_df[
        "Passed"
    ].all()
)

section_10_complete = bool(
    section_10_1_complete
    and section_10_2_complete
    and section_10_3_complete
    and section_10_4_complete
    and section_10_5_complete
)

print()
print(
    f"Section 10.5 completion status: "
    f"{section_10_5_complete}"
)

print(
    f"Section 10 overall completion status: "
    f"{section_10_complete}"
)

if not section_10_complete:
    raise RuntimeError(
        "Section 10 Unusual Release Performance Patterns "
        "did not complete successfully."
    )
```

### Interpretation

Section 10 combined four unusual-performance signals covering short-term streaming spikes, sudden chart movement, broader streaming anomalies and growth without matching geographic breadth.

Across the 4,593 releases assessed, 4,178 releases showed no unusual-performance flags, while 415 releases showed at least one unusual signal. This indicates that most releases followed comparatively normal performance patterns.

Streaming anomalies and growth without geographic breadth were the most common individual signals, each affecting 224 releases, while short-term streaming spikes affected 221 releases. Sudden chart movement was much less common, affecting only 21 releases.

A total of 187 releases showed multiple unusual signals, including 88 releases that simultaneously exhibited a short-term streaming spike, a streaming anomaly and growth without geographic breadth. These multi-signal cases provide stronger evidence of unusual release behaviour than isolated flags.

Overall, the findings show that unusual performance patterns are concentrated within a relatively small part of the PMIP population. The resulting flags should therefore be treated as prioritisation signals for deeper investigation rather than evidence of suspicious or abnormal activity.

## 11. Release Performance Scoring and Classification

### Purpose

This section constructs the final PMIP release-performance framework by combining the major analytical indicators developed in the previous sections.

The aim is to transform the separate streaming, chart, geographic, temporal, artist-context and unusual-performance signals into a consistent release-level assessment.

The section will:

- identify the indicators suitable for final scoring;
- align their scales and performance directions;
- construct a composite release-performance score;
- classify releases into interpretable performance groups;
- calculate release rankings and percentiles;
- identify high-priority release signals;
- validate the final scoring framework.

The resulting score is intended to provide a structured summary of multidimensional release performance while preserving evidence coverage and the meaning of the underlying PMIP indicators.



### 11.1 Indicator Selection

### Purpose

This subsection identifies the PMIP indicators that are suitable for inclusion in the final release-performance score.

Candidate indicators are drawn from the major analytical dimensions developed previously:

- streaming performance;
- chart performance;
- geographic reach;
- temporal comparability;
- artist–release performance context;
- unusual-performance behaviour.

Each candidate is assessed for field availability, usable observations and dataset coverage before being accepted into the final scoring framework.

Indicators are selected only when sufficient release-level information is available to support meaningful comparison.

```python
# ============================================================
# Section 11.1 - Indicator Selection
# ============================================================

if not section_10_complete:
    raise RuntimeError(
        "Section 10 must be completed before beginning "
        "Release Performance Scoring and Classification."
    )

# ============================================================
# Working source
# ============================================================

release_scoring_df = unusual_release_patterns_final_df.copy()

# ============================================================
# Validate release identity
# ============================================================

required_identity_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_identity_fields = [
    field
    for field in required_identity_fields
    if field not in release_scoring_df.columns
]

if missing_identity_fields:
    raise RuntimeError(
        "Required release identity fields are missing: "
        f"{missing_identity_fields}"
    )

# ============================================================
# Candidate final PMIP indicators
# ============================================================

candidate_indicator_map = {
    "streaming_performance_score": {
        "Indicator": "Streaming Performance",
        "Dimension": "Streaming Performance Intelligence",
        "Expected Direction": "Higher is stronger",
    },

    "chart_performance_score": {
        "Indicator": "Chart Performance",
        "Dimension": "Chart Performance Intelligence",
        "Expected Direction": "Higher is stronger",
    },

    "geographic_reach_score": {
        "Indicator": "Geographic Reach",
        "Dimension": "Geographic Release Reach",
        "Expected Direction": "Higher is stronger",
    },

    "temporal_comparability_score": {
        "Indicator": "Temporal Comparability",
        "Dimension": "Release Timing and Historical Context",
        "Expected Direction": "Higher is stronger",
    },

    "artist_release_relationship_score": {
        "Indicator": "Artist–Release Relationship",
        "Dimension": "Artist Context",
        "Expected Direction": "Higher is stronger",
    },

    "unusual_performance_score": {
        "Indicator": "Unusual Performance",
        "Dimension": "Unusual Release Performance Patterns",
        "Expected Direction": "Higher is more unusual",
    },
}

# ============================================================
# Assess candidate availability and coverage
# ============================================================

indicator_assessment_rows = []

total_releases = len(release_scoring_df)

for field, metadata in candidate_indicator_map.items():

    field_available = field in release_scoring_df.columns

    if field_available:

        numeric_values = pd.to_numeric(
            release_scoring_df[field],
            errors="coerce"
        )

        available_values = int(
            numeric_values.notna().sum()
        )

        missing_values = int(
            numeric_values.isna().sum()
        )

        coverage_pct = (
            available_values
            / total_releases
            * 100
            if total_releases > 0
            else 0
        )

        minimum_value = (
            numeric_values.min()
            if available_values > 0
            else np.nan
        )

        median_value = (
            numeric_values.median()
            if available_values > 0
            else np.nan
        )

        maximum_value = (
            numeric_values.max()
            if available_values > 0
            else np.nan
        )

    else:

        available_values = 0
        missing_values = total_releases
        coverage_pct = 0.0

        minimum_value = np.nan
        median_value = np.nan
        maximum_value = np.nan

    indicator_assessment_rows.append(
        {
            "Field": field,
            "Indicator": metadata["Indicator"],
            "Dimension": metadata["Dimension"],
            "Expected Direction": metadata[
                "Expected Direction"
            ],
            "Field Available": field_available,
            "Available Values": available_values,
            "Missing Values": missing_values,
            "Coverage (%)": coverage_pct,
            "Minimum": minimum_value,
            "Median": median_value,
            "Maximum": maximum_value,
        }
    )

indicator_selection_assessment_df = pd.DataFrame(
    indicator_assessment_rows
)

print("Assessing PMIP final scoring indicators")
print("=" * 80)

display(
    indicator_selection_assessment_df
)

# ============================================================
# Indicator eligibility
#
# Final scoring indicators must:
# - exist in the release dataset;
# - contain usable numeric values;
# - have at least 25% release coverage.
#
# Coverage-aware weighting later in Section 11 allows
# partially observed indicators to remain useful.
# ============================================================

minimum_indicator_coverage = 25.0

indicator_selection_assessment_df[
    "Selected"
] = (
    indicator_selection_assessment_df[
        "Field Available"
    ]
    & (
        indicator_selection_assessment_df[
            "Available Values"
        ] > 0
    )
    & (
        indicator_selection_assessment_df[
            "Coverage (%)"
        ] >= minimum_indicator_coverage
    )
)

# ============================================================
# Separate core performance indicators from contextual anomaly
#
# Unusual performance is retained as a contextual signal even
# when its interpretation direction differs from performance.
# Direction will be handled in Section 11.2.
# ============================================================

selected_release_indicator_fields = (
    indicator_selection_assessment_df.loc[
        indicator_selection_assessment_df[
            "Selected"
        ],
        "Field"
    ]
    .tolist()
)

selected_release_indicator_names = (
    indicator_selection_assessment_df.loc[
        indicator_selection_assessment_df[
            "Selected"
        ],
        "Indicator"
    ]
    .tolist()
)

# ============================================================
# Selected indicator summary
# ============================================================

selected_indicator_summary_df = (
    indicator_selection_assessment_df.loc[
        indicator_selection_assessment_df[
            "Selected"
        ],
        [
            "Field",
            "Indicator",
            "Dimension",
            "Expected Direction",
            "Available Values",
            "Coverage (%)",
            "Minimum",
            "Median",
            "Maximum",
        ]
    ]
    .reset_index(drop=True)
)

print()
print("Selected PMIP release-performance indicators")
print("=" * 80)

display(
    selected_indicator_summary_df
)

# ============================================================
# Excluded indicator summary
# ============================================================

excluded_indicator_summary_df = (
    indicator_selection_assessment_df.loc[
        ~indicator_selection_assessment_df[
            "Selected"
        ],
        [
            "Field",
            "Indicator",
            "Dimension",
            "Field Available",
            "Available Values",
            "Coverage (%)",
        ]
    ]
    .reset_index(drop=True)
)

print()
print("Excluded PMIP release-performance indicators")
print("=" * 80)

if excluded_indicator_summary_df.empty:
    print(
        "No candidate indicators were excluded."
    )
else:
    display(
        excluded_indicator_summary_df
    )

# ============================================================
# Per-release indicator coverage
# ============================================================

if selected_release_indicator_fields:

    release_scoring_df[
        "final_indicator_count_available"
    ] = (
        release_scoring_df[
            selected_release_indicator_fields
        ]
        .notna()
        .sum(axis=1)
    )

    release_scoring_df[
        "final_indicator_coverage_pct"
    ] = (
        release_scoring_df[
            "final_indicator_count_available"
        ]
        / len(
            selected_release_indicator_fields
        )
        * 100
    )

else:

    release_scoring_df[
        "final_indicator_count_available"
    ] = 0

    release_scoring_df[
        "final_indicator_coverage_pct"
    ] = 0.0

# ============================================================
# Release-level evidence distribution
# ============================================================

indicator_coverage_distribution_df = (
    release_scoring_df[
        "final_indicator_count_available"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Indicators Available"
    )
    .reset_index(
        name="Releases"
    )
)

indicator_coverage_distribution_df[
    "Share (%)"
] = (
    indicator_coverage_distribution_df[
        "Releases"
    ]
    / total_releases
    * 100
)

print()
print("Release-level indicator coverage")
print("=" * 80)

display(
    indicator_coverage_distribution_df
)

# ============================================================
# Overall indicator-selection summary
# ============================================================

indicator_selection_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": total_releases,
        },
        {
            "Metric": "Candidate indicators assessed",
            "Observed Value": len(
                candidate_indicator_map
            ),
        },
        {
            "Metric": "Indicators selected",
            "Observed Value": len(
                selected_release_indicator_fields
            ),
        },
        {
            "Metric": "Indicators excluded",
            "Observed Value": (
                len(candidate_indicator_map)
                - len(
                    selected_release_indicator_fields
                )
            ),
        },
        {
            "Metric": "Minimum required coverage (%)",
            "Observed Value": minimum_indicator_coverage,
        },
        {
            "Metric": "Median selected-indicator coverage (%)",
            "Observed Value": (
                selected_indicator_summary_df[
                    "Coverage (%)"
                ].median()
                if not selected_indicator_summary_df.empty
                else np.nan
            ),
        },
        {
            "Metric": "Median release-level indicator coverage (%)",
            "Observed Value": (
                release_scoring_df[
                    "final_indicator_coverage_pct"
                ].median()
            ),
        },
    ]
)

print()
print("Indicator-selection summary")
print("=" * 80)

display(
    indicator_selection_summary_df
)

# ============================================================
# Validation
# ============================================================

indicator_selection_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 10 completed",
            "Passed": section_10_complete,
        },
        {
            "Validation Area": "Release identity available",
            "Passed": all(
                field in release_scoring_df.columns
                for field in required_identity_fields
            ),
        },
        {
            "Validation Area": "Candidate indicators assessed",
            "Passed": (
                len(
                    indicator_selection_assessment_df
                )
                == len(
                    candidate_indicator_map
                )
            ),
        },
        {
            "Validation Area": "At least one indicator selected",
            "Passed": (
                len(
                    selected_release_indicator_fields
                ) > 0
            ),
        },
        {
            "Validation Area": "Selected indicators contain values",
            "Passed": (
                selected_indicator_summary_df[
                    "Available Values"
                ].gt(0).all()
                if not selected_indicator_summary_df.empty
                else False
            ),
        },
        {
            "Validation Area": "Indicator coverage calculated",
            "Passed": release_scoring_df[
                "final_indicator_coverage_pct"
            ].notna().all(),
        },
    ]
)

print()
print("Indicator Selection validation")
print("=" * 80)

display(
    indicator_selection_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_1_complete = bool(
    indicator_selection_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.1 completion status: "
    f"{section_11_1_complete}"
)

if not section_11_1_complete:
    raise RuntimeError(
        "Indicator Selection did not complete successfully."
    )
```

### Interpretation

All six candidate indicators met the minimum coverage requirement and were retained for final release-performance scoring.

Temporal Comparability and Artist–Release Relationship provide complete release-level coverage, while Streaming Performance is available for almost the entire dataset.

Chart Performance, Geographic Reach and Unusual Performance have lower coverage, but still provide sufficient evidence for inclusion.

The release-level coverage distribution confirms that some releases contain more complete evidence than others. This supports the use of coverage-aware weighting in the final composite score rather than treating missing indicators as poor performance.

The selected indicator set therefore provides a suitable multidimensional foundation for the final PMIP release-performance framework.

## 11.2 Scaling and Direction Alignment

### Purpose

This subsection prepares the selected PMIP indicators for combination into a single release-performance score.

The aim is to place all selected indicators on a comparable scale and ensure that they follow a consistent interpretation direction.

Most selected indicators already follow the rule that higher values represent stronger release performance. However, the Unusual Performance indicator has a different meaning because higher values represent more unusual behaviour rather than stronger performance.

This subsection therefore:

- converts selected indicators onto a common 0–100 scale where needed;
- preserves the relative ranking of releases;
- aligns indicator direction so that higher aligned values consistently represent stronger release-performance contribution;
- transforms the Unusual Performance signal into a non-penalising contextual contribution;
- validates the transformed indicators before composite scoring.

The aligned indicators will form the direct inputs to the composite release-performance score developed in Section 11.3.

```python
# ============================================================
# Section 11.2 - Scaling and Direction Alignment
# ============================================================

if not section_11_1_complete:
    raise RuntimeError(
        "Section 11.1 must be completed before performing "
        "Scaling and Direction Alignment."
    )

# ============================================================
# Working source
# ============================================================

release_scaling_df = release_scoring_df.copy()

# ============================================================
# Validate selected indicators
# ============================================================

if not selected_release_indicator_fields:
    raise RuntimeError(
        "No indicators were selected in Section 11.1."
    )

missing_selected_fields = [
    field
    for field in selected_release_indicator_fields
    if field not in release_scaling_df.columns
]

if missing_selected_fields:
    raise RuntimeError(
        "Selected indicator fields are missing from the "
        f"release dataset: {missing_selected_fields}"
    )

# ============================================================
# Base direction map
# ============================================================

indicator_direction_map = {
    "streaming_performance_score": "positive",
    "chart_performance_score": "positive",
    "geographic_reach_score": "positive",
    "temporal_comparability_score": "positive",
    "artist_release_relationship_score": "positive",
    "unusual_performance_score": "contextual",
}

# ============================================================
# Scaling method
#
# Percentile scaling is used to make all selected indicators
# directly comparable on a 0-100 scale while preserving each
# indicator's relative ordering.
# ============================================================

scaled_indicator_fields = []
aligned_indicator_fields = []

scaling_rows = []

for field in selected_release_indicator_fields:

    numeric_series = pd.to_numeric(
        release_scaling_df[field],
        errors="coerce"
    )

    scaled_field = f"{field}_scaled"
    aligned_field = f"{field}_aligned"

    scaled_indicator_fields.append(
        scaled_field
    )

    aligned_indicator_fields.append(
        aligned_field
    )

    release_scaling_df[
        scaled_field
    ] = np.nan

    valid_mask = numeric_series.notna()

    if valid_mask.any():

        release_scaling_df.loc[
            valid_mask,
            scaled_field
        ] = (
            numeric_series.loc[
                valid_mask
            ]
            .rank(
                method="average",
                pct=True,
                ascending=True
            )
            * 100
        )

    # ========================================================
    # Direction alignment
    #
    # Positive indicators:
    #     scaled score is used directly.
    #
    # Unusual Performance:
    #     remains contextual rather than performance-positive.
    #     We convert it into a "performance consistency"
    #     contribution using 100 - unusual percentile.
    #
    # This means:
    #     low unusualness -> stronger consistency contribution
    #     high unusualness -> weaker consistency contribution
    #
    # The original unusual score is preserved separately.
    # ========================================================

    direction = indicator_direction_map.get(
        field,
        "positive"
    )

    if direction == "positive":

        release_scaling_df[
            aligned_field
        ] = release_scaling_df[
            scaled_field
        ]

        alignment_method = (
            "Direct percentile alignment"
        )

    elif direction == "contextual":

        release_scaling_df[
            aligned_field
        ] = np.where(
            release_scaling_df[
                scaled_field
            ].notna(),
            100
            - release_scaling_df[
                scaled_field
            ],
            np.nan
        )

        alignment_method = (
            "Inverse percentile alignment "
            "(higher unusualness -> lower consistency)"
        )

    else:

        raise RuntimeError(
            f"Unsupported indicator direction for {field}: "
            f"{direction}"
        )

    # ========================================================
    # Assessment row
    # ========================================================

    scaling_rows.append(
        {
            "Field": field,
            "Direction Type": direction,
            "Available Values": int(
                valid_mask.sum()
            ),
            "Scaled Minimum": (
                release_scaling_df[
                    scaled_field
                ].min()
            ),
            "Scaled Median": (
                release_scaling_df[
                    scaled_field
                ].median()
            ),
            "Scaled Maximum": (
                release_scaling_df[
                    scaled_field
                ].max()
            ),
            "Aligned Minimum": (
                release_scaling_df[
                    aligned_field
                ].min()
            ),
            "Aligned Median": (
                release_scaling_df[
                    aligned_field
                ].median()
            ),
            "Aligned Maximum": (
                release_scaling_df[
                    aligned_field
                ].max()
            ),
            "Alignment Method": alignment_method,
        }
    )

scaling_alignment_assessment_df = pd.DataFrame(
    scaling_rows
)

print("Assessing PMIP scaling and direction alignment")
print("=" * 80)

display(
    scaling_alignment_assessment_df
)

# ============================================================
# Indicator-level validation
# ============================================================

scaling_validation_rows = []

for field in selected_release_indicator_fields:

    scaled_field = f"{field}_scaled"
    aligned_field = f"{field}_aligned"

    original_available = (
        release_scaling_df[
            field
        ].notna().sum()
    )

    scaled_available = (
        release_scaling_df[
            scaled_field
        ].notna().sum()
    )

    aligned_available = (
        release_scaling_df[
            aligned_field
        ].notna().sum()
    )

    scaled_within_range = bool(
        release_scaling_df.loc[
            release_scaling_df[
                scaled_field
            ].notna(),
            scaled_field
        ]
        .between(
            0,
            100
        )
        .all()
    )

    aligned_within_range = bool(
        release_scaling_df.loc[
            release_scaling_df[
                aligned_field
            ].notna(),
            aligned_field
        ]
        .between(
            0,
            100
        )
        .all()
    )

    scaling_validation_rows.append(
        {
            "Indicator": field,
            "Original Values": int(
                original_available
            ),
            "Scaled Values": int(
                scaled_available
            ),
            "Aligned Values": int(
                aligned_available
            ),
            "Coverage Preserved": bool(
                original_available
                == scaled_available
                == aligned_available
            ),
            "Scaled Range Valid": (
                scaled_within_range
            ),
            "Aligned Range Valid": (
                aligned_within_range
            ),
        }
    )

indicator_scaling_validation_df = pd.DataFrame(
    scaling_validation_rows
)

print()
print("Indicator scaling validation")
print("=" * 80)

display(
    indicator_scaling_validation_df
)

# ============================================================
# Release-level aligned indicator coverage
# ============================================================

release_scaling_df[
    "aligned_indicator_count_available"
] = (
    release_scaling_df[
        aligned_indicator_fields
    ]
    .notna()
    .sum(axis=1)
)

release_scaling_df[
    "aligned_indicator_coverage_pct"
] = (
    release_scaling_df[
        "aligned_indicator_count_available"
    ]
    / len(
        aligned_indicator_fields
    )
    * 100
)

aligned_coverage_distribution_df = (
    release_scaling_df[
        "aligned_indicator_count_available"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Aligned Indicators Available"
    )
    .reset_index(
        name="Releases"
    )
)

aligned_coverage_distribution_df[
    "Share (%)"
] = (
    aligned_coverage_distribution_df[
        "Releases"
    ]
    / len(
        release_scaling_df
    )
    * 100
)

print()
print("Aligned-indicator release coverage")
print("=" * 80)

display(
    aligned_coverage_distribution_df
)

# ============================================================
# Direction-alignment spot check
#
# Compare strongest original values with aligned outputs.
# This is especially useful for confirming the inversion of
# unusual_performance_score.
# ============================================================

direction_check_rows = []

for field in selected_release_indicator_fields:

    scaled_field = f"{field}_scaled"
    aligned_field = f"{field}_aligned"

    direction = indicator_direction_map.get(
        field,
        "positive"
    )

    valid_mask = (
        release_scaling_df[
            field
        ].notna()
    )

    if not valid_mask.any():
        continue

    strongest_original_index = (
        release_scaling_df.loc[
            valid_mask,
            field
        ]
        .idxmax()
    )

    direction_check_rows.append(
        {
            "Indicator": field,
            "Direction Type": direction,
            "Highest Original Value": (
                release_scaling_df.loc[
                    strongest_original_index,
                    field
                ]
            ),
            "Scaled Value": (
                release_scaling_df.loc[
                    strongest_original_index,
                    scaled_field
                ]
            ),
            "Aligned Value": (
                release_scaling_df.loc[
                    strongest_original_index,
                    aligned_field
                ]
            ),
        }
    )

direction_alignment_check_df = pd.DataFrame(
    direction_check_rows
)

print()
print("Direction-alignment spot check")
print("=" * 80)

display(
    direction_alignment_check_df
)

# ============================================================
# Overall summary
# ============================================================

scaling_alignment_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_scaling_df
            ),
        },
        {
            "Metric": "Indicators scaled",
            "Observed Value": len(
                scaled_indicator_fields
            ),
        },
        {
            "Metric": "Indicators direction-aligned",
            "Observed Value": len(
                aligned_indicator_fields
            ),
        },
        {
            "Metric": "Positive-direction indicators",
            "Observed Value": sum(
                indicator_direction_map.get(
                    field,
                    "positive"
                )
                == "positive"
                for field
                in selected_release_indicator_fields
            ),
        },
        {
            "Metric": "Contextual indicators",
            "Observed Value": sum(
                indicator_direction_map.get(
                    field,
                    "positive"
                )
                == "contextual"
                for field
                in selected_release_indicator_fields
            ),
        },
        {
            "Metric": "Median aligned-indicator coverage (%)",
            "Observed Value": (
                release_scaling_df[
                    "aligned_indicator_coverage_pct"
                ].median()
            ),
        },
    ]
)

print()
print("Scaling and direction-alignment summary")
print("=" * 80)

display(
    scaling_alignment_summary_df
)

# ============================================================
# Final validation
# ============================================================

scaling_direction_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11.1 completed",
            "Passed": section_11_1_complete,
        },
        {
            "Validation Area": "Selected indicators available",
            "Passed": all(
                field in release_scaling_df.columns
                for field
                in selected_release_indicator_fields
            ),
        },
        {
            "Validation Area": "All selected indicators scaled",
            "Passed": all(
                field in release_scaling_df.columns
                for field
                in scaled_indicator_fields
            ),
        },
        {
            "Validation Area": "All indicators direction-aligned",
            "Passed": all(
                field in release_scaling_df.columns
                for field
                in aligned_indicator_fields
            ),
        },
        {
            "Validation Area": "Scaling coverage preserved",
            "Passed": indicator_scaling_validation_df[
                "Coverage Preserved"
            ].all(),
        },
        {
            "Validation Area": "Scaled values within 0-100",
            "Passed": indicator_scaling_validation_df[
                "Scaled Range Valid"
            ].all(),
        },
        {
            "Validation Area": "Aligned values within 0-100",
            "Passed": indicator_scaling_validation_df[
                "Aligned Range Valid"
            ].all(),
        },
        {
            "Validation Area": "Aligned coverage calculated",
            "Passed": release_scaling_df[
                "aligned_indicator_coverage_pct"
            ].notna().all(),
        },
    ]
)

print()
print("Scaling and Direction Alignment validation")
print("=" * 80)

display(
    scaling_direction_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_2_complete = bool(
    scaling_direction_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.2 completion status: "
    f"{section_11_2_complete}"
)

if not section_11_2_complete:
    raise RuntimeError(
        "Scaling and Direction Alignment did not "
        "complete successfully."
    )
```

### Interpretation

All selected indicators were successfully transformed onto a common 0–100 percentile scale while preserving their original evidence coverage.

The five performance-oriented indicators retained a positive direction, where higher aligned values represent stronger relative performance.

The Unusual Performance indicator was successfully transformed in the opposite direction for comparability; however, because unusual behaviour is not inherently positive or negative, it will remain a contextual signal rather than directly determine the final performance score.

The aligned performance indicators are therefore ready for composite scoring.

### 11.3 Composite Release Performance Score

### Purpose

This subsection constructs the final PMIP Composite Release Performance Score from the aligned performance indicators.

The score combines five major dimensions:

- streaming performance;
- chart performance;
- geographic reach;
- temporal comparability;
- artist–release performance context.

Coverage-aware weighting is used so that missing indicators are not treated as zero performance. Instead, the weights of the available indicators are redistributed for each release.

The Unusual Performance indicator is retained separately as contextual intelligence and is not included directly in the composite performance score because unusual behaviour may represent either positive or negative release dynamics.

The resulting composite score provides a single 0–100 measure of multidimensional release performance while preserving evidence coverage for later classification and ranking.

```python
# ============================================================
# Section 11.3 - Composite Release Performance Score
# ============================================================

if not section_11_2_complete:
    raise RuntimeError(
        "Section 11.2 must be completed before constructing "
        "the Composite Release Performance Score."
    )

# ============================================================
# Working source
# ============================================================

composite_release_df = release_scaling_df.copy()

# ============================================================
# Core performance indicators
#
# Unusual performance remains contextual and is deliberately
# excluded from the performance-strength composite.
# ============================================================

composite_indicator_weights = {
    "streaming_performance_score_aligned": 0.30,
    "chart_performance_score_aligned": 0.20,
    "geographic_reach_score_aligned": 0.15,
    "temporal_comparability_score_aligned": 0.15,
    "artist_release_relationship_score_aligned": 0.20,
}

composite_indicator_fields = list(
    composite_indicator_weights.keys()
)

# ============================================================
# Validate required composite inputs
# ============================================================

missing_composite_fields = [
    field
    for field in composite_indicator_fields
    if field not in composite_release_df.columns
]

if missing_composite_fields:
    raise RuntimeError(
        "Required aligned composite indicators are missing: "
        f"{missing_composite_fields}"
    )

# ============================================================
# Base-weight summary
# ============================================================

composite_weight_summary_df = pd.DataFrame(
    [
        {
            "Component": field,
            "Base Weight": weight,
            "Releases Available": int(
                composite_release_df[
                    field
                ].notna().sum()
            ),
            "Coverage (%)": (
                composite_release_df[
                    field
                ].notna().mean()
                * 100
            ),
        }
        for field, weight
        in composite_indicator_weights.items()
    ]
)

print("Constructing PMIP Composite Release Performance Score")
print("=" * 80)

print()
print("Composite indicator weights")
print("=" * 80)

display(
    composite_weight_summary_df
)

# ============================================================
# Coverage-aware weighted score
#
# For every release:
# - use only available components;
# - sum their weighted scores;
# - divide by the sum of available base weights.
#
# Missing indicators therefore do NOT become zeros.
# ============================================================

weighted_sum = pd.Series(
    0.0,
    index=composite_release_df.index
)

available_weight_sum = pd.Series(
    0.0,
    index=composite_release_df.index
)

for field, weight in composite_indicator_weights.items():

    available_mask = (
        composite_release_df[
            field
        ].notna()
    )

    weighted_sum.loc[
        available_mask
    ] += (
        composite_release_df.loc[
            available_mask,
            field
        ]
        * weight
    )

    available_weight_sum.loc[
        available_mask
    ] += weight

composite_release_df[
    "composite_release_performance_score"
] = np.where(
    available_weight_sum > 0,
    weighted_sum
    / available_weight_sum,
    np.nan
)

# ============================================================
# Evidence coverage
# ============================================================

composite_release_df[
    "composite_indicator_count_available"
] = (
    composite_release_df[
        composite_indicator_fields
    ]
    .notna()
    .sum(axis=1)
)

composite_release_df[
    "composite_indicator_coverage_pct"
] = (
    composite_release_df[
        "composite_indicator_count_available"
    ]
    / len(
        composite_indicator_fields
    )
    * 100
)

composite_release_df[
    "composite_weight_coverage_pct"
] = (
    available_weight_sum
    * 100
)

# ============================================================
# Evidence-strength classification
# ============================================================

composite_release_df[
    "composite_evidence_strength"
] = np.select(
    [
        composite_release_df[
            "composite_weight_coverage_pct"
        ] >= 90,

        composite_release_df[
            "composite_weight_coverage_pct"
        ] >= 70,

        composite_release_df[
            "composite_weight_coverage_pct"
        ] >= 50,

        composite_release_df[
            "composite_weight_coverage_pct"
        ] > 0,
    ],
    [
        "Very Strong Composite Evidence",
        "Strong Composite Evidence",
        "Moderate Composite Evidence",
        "Limited Composite Evidence",
    ],
    default="No Composite Evidence",
)

# ============================================================
# Percentile position
# ============================================================

valid_composite_mask = (
    composite_release_df[
        "composite_release_performance_score"
    ].notna()
)

composite_release_df[
    "composite_release_performance_percentile"
] = np.nan

if valid_composite_mask.any():

    composite_release_df.loc[
        valid_composite_mask,
        "composite_release_performance_percentile"
    ] = (
        composite_release_df.loc[
            valid_composite_mask,
            "composite_release_performance_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Rank
# ============================================================

composite_release_df[
    "composite_release_performance_rank"
] = (
    composite_release_df[
        "composite_release_performance_score"
    ]
    .rank(
        method="min",
        ascending=False,
        na_option="bottom"
    )
)

# ============================================================
# Composite score summary
# ============================================================

composite_release_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                composite_release_df
            ),
        },
        {
            "Metric": "Releases with composite score",
            "Observed Value": int(
                valid_composite_mask.sum()
            ),
        },
        {
            "Metric": "Releases without composite score",
            "Observed Value": int(
                (~valid_composite_mask).sum()
            ),
        },
        {
            "Metric": "Composite indicators used",
            "Observed Value": len(
                composite_indicator_fields
            ),
        },
        {
            "Metric": "Median composite performance score",
            "Observed Value": (
                composite_release_df.loc[
                    valid_composite_mask,
                    "composite_release_performance_score"
                ].median()
            ),
        },
        {
            "Metric": "Mean composite performance score",
            "Observed Value": (
                composite_release_df.loc[
                    valid_composite_mask,
                    "composite_release_performance_score"
                ].mean()
            ),
        },
        {
            "Metric": "Minimum composite performance score",
            "Observed Value": (
                composite_release_df.loc[
                    valid_composite_mask,
                    "composite_release_performance_score"
                ].min()
            ),
        },
        {
            "Metric": "Maximum composite performance score",
            "Observed Value": (
                composite_release_df.loc[
                    valid_composite_mask,
                    "composite_release_performance_score"
                ].max()
            ),
        },
        {
            "Metric": "Median component coverage (%)",
            "Observed Value": (
                composite_release_df[
                    "composite_indicator_coverage_pct"
                ].median()
            ),
        },
        {
            "Metric": "Median weighted evidence coverage (%)",
            "Observed Value": (
                composite_release_df[
                    "composite_weight_coverage_pct"
                ].median()
            ),
        },
    ]
)

print()
print("Composite release-performance summary")
print("=" * 80)

display(
    composite_release_summary_df
)

# ============================================================
# Composite evidence distribution
# ============================================================

composite_evidence_distribution_df = (
    composite_release_df[
        "composite_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Composite Evidence Strength"
    )
    .reset_index(
        name="Releases"
    )
)

composite_evidence_distribution_df[
    "Share (%)"
] = (
    composite_evidence_distribution_df[
        "Releases"
    ]
    / len(
        composite_release_df
    )
    * 100
)

print()
print("Composite evidence-strength distribution")
print("=" * 80)

display(
    composite_evidence_distribution_df
)

# ============================================================
# Component-count distribution
# ============================================================

composite_component_distribution_df = (
    composite_release_df[
        "composite_indicator_count_available"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Composite Indicators Available"
    )
    .reset_index(
        name="Releases"
    )
)

composite_component_distribution_df[
    "Share (%)"
] = (
    composite_component_distribution_df[
        "Releases"
    ]
    / len(
        composite_release_df
    )
    * 100
)

print()
print("Composite component-coverage distribution")
print("=" * 80)

display(
    composite_component_distribution_df
)

# ============================================================
# Highest composite-performance releases
# ============================================================

top_composite_releases_df = (
    composite_release_df.loc[
        valid_composite_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "streaming_performance_score_aligned",
            "chart_performance_score_aligned",
            "geographic_reach_score_aligned",
            "temporal_comparability_score_aligned",
            "artist_release_relationship_score_aligned",
            "composite_indicator_count_available",
            "composite_indicator_coverage_pct",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "composite_release_performance_score",
            "composite_release_performance_percentile",
            "composite_release_performance_rank",
        ]
    ]
    .sort_values(
        by="composite_release_performance_score",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP composite release-performance scores")
print("=" * 80)

display(
    top_composite_releases_df
)

# ============================================================
# Contextual anomaly attachment
#
# Keep unusual-performance intelligence available for later
# priority analysis without allowing it to determine the
# composite performance-strength score.
# ============================================================

contextual_fields = [
    field
    for field in [
        "unusual_performance_score",
        "unusual_performance_flag",
        "unusual_performance_flag_count",
        "unusual_performance_pattern",
        "priority_unusual_performance_flag",
    ]
    if field in composite_release_df.columns
]

print()
print("Contextual unusual-performance fields retained")
print("=" * 80)

if contextual_fields:
    for field in contextual_fields:
        print(f"- {field}")
else:
    print(
        "No contextual unusual-performance fields detected."
    )

# ============================================================
# Validation
# ============================================================

composite_score_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11.2 completed",
            "Passed": section_11_2_complete,
        },
        {
            "Validation Area": "Composite indicators available",
            "Passed": all(
                field in composite_release_df.columns
                for field in composite_indicator_fields
            ),
        },
        {
            "Validation Area": "Base weights sum to 1",
            "Passed": bool(
                np.isclose(
                    sum(
                        composite_indicator_weights.values()
                    ),
                    1.0
                )
            ),
        },
        {
            "Validation Area": "Composite scores calculated",
            "Passed": composite_release_df[
                "composite_release_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Composite scores within 0-100",
            "Passed": composite_release_df.loc[
                valid_composite_mask,
                "composite_release_performance_score"
            ]
            .between(
                0,
                100
            )
            .all(),
        },
        {
            "Validation Area": "Evidence coverage calculated",
            "Passed": composite_release_df[
                "composite_weight_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Composite percentiles calculated",
            "Passed": composite_release_df.loc[
                valid_composite_mask,
                "composite_release_performance_percentile"
            ].notna().all(),
        },
        {
            "Validation Area": "Composite ranks calculated",
            "Passed": composite_release_df[
                "composite_release_performance_rank"
            ].notna().all(),
        },
        {
            "Validation Area": "Evidence strength classified",
            "Passed": composite_release_df[
                "composite_evidence_strength"
            ].notna().all(),
        },
    ]
)

print()
print("Composite Release Performance Score validation")
print("=" * 80)

display(
    composite_score_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_3_complete = bool(
    composite_score_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.3 completion status: "
    f"{section_11_3_complete}"
)

if not section_11_3_complete:
    raise RuntimeError(
        "Composite Release Performance Score did not "
        "complete successfully."
    )
```

### Interpretation

The Composite Release Performance Score was successfully constructed using the aligned PMIP performance indicators.

Coverage-aware weighting ensured that missing indicators were not treated as poor performance, while the available evidence was reweighted for each release.

The final composite scores remained within the expected 0–100 range, and both percentile positions and release rankings were calculated successfully.

This provides a consistent multidimensional performance measure that can now be converted into interpretable release-performance classes.

## 11.4 Release Performance Classes

### Purpose

This subsection converts the continuous Composite Release Performance Score into interpretable release-performance classes.

The aim is to make the final PMIP score easier to understand by grouping releases into ordered performance categories.

The classification uses percentile-based thresholds so that classes reflect each release's relative standing within the PMIP population.

The resulting classes distinguish:

- Low Release Performance;
- Moderate Release Performance;
- High Release Performance;
- Very High Release Performance;
- Exceptional Release Performance.

This categorical layer supports clearer comparison, reporting and later high-priority release identification while preserving the underlying continuous composite score.

```python
# ============================================================
# Section 11.4 - Release Performance Classes
# ============================================================

if not section_11_3_complete:
    raise RuntimeError(
        "Section 11.3 must be completed before assigning "
        "Release Performance Classes."
    )

# ============================================================
# Working source
# ============================================================

release_classification_df = composite_release_df.copy()

# ============================================================
# Validate composite score
# ============================================================

required_classification_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "composite_release_performance_percentile",
]

missing_classification_fields = [
    field
    for field in required_classification_fields
    if field not in release_classification_df.columns
]

if missing_classification_fields:
    raise RuntimeError(
        "Required Release Performance Classification fields "
        f"are missing: {missing_classification_fields}"
    )

# ============================================================
# Valid scored releases
# ============================================================

valid_release_classification_mask = (
    release_classification_df[
        "composite_release_performance_score"
    ].notna()
    & release_classification_df[
        "composite_release_performance_percentile"
    ].notna()
)

# ============================================================
# Percentile-based performance classes
#
# 0-25      -> Low
# >25-50    -> Moderate
# >50-75    -> High
# >75-90    -> Very High
# >90       -> Exceptional
# ============================================================

release_classification_df[
    "release_performance_class"
] = "No Composite Performance Evidence"

release_classification_df.loc[
    valid_release_classification_mask
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] <= 25
    ),
    "release_performance_class"
] = "Low Release Performance"

release_classification_df.loc[
    valid_release_classification_mask
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] > 25
    )
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] <= 50
    ),
    "release_performance_class"
] = "Moderate Release Performance"

release_classification_df.loc[
    valid_release_classification_mask
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] > 50
    )
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] <= 75
    ),
    "release_performance_class"
] = "High Release Performance"

release_classification_df.loc[
    valid_release_classification_mask
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] > 75
    )
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] <= 90
    ),
    "release_performance_class"
] = "Very High Release Performance"

release_classification_df.loc[
    valid_release_classification_mask
    & (
        release_classification_df[
            "composite_release_performance_percentile"
        ] > 90
    ),
    "release_performance_class"
] = "Exceptional Release Performance"

# ============================================================
# Class order
# ============================================================

release_performance_class_order = [
    "Low Release Performance",
    "Moderate Release Performance",
    "High Release Performance",
    "Very High Release Performance",
    "Exceptional Release Performance",
    "No Composite Performance Evidence",
]

release_classification_df[
    "release_performance_class"
] = pd.Categorical(
    release_classification_df[
        "release_performance_class"
    ],
    categories=release_performance_class_order,
    ordered=True
)

# ============================================================
# Class distribution
# ============================================================

release_performance_class_distribution_df = (
    release_classification_df[
        "release_performance_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Release Performance Class"
    )
    .reset_index(
        name="Releases"
    )
)

release_performance_class_distribution_df[
    "Share (%)"
] = (
    release_performance_class_distribution_df[
        "Releases"
    ]
    / len(
        release_classification_df
    )
    * 100
)

print("Assessing PMIP release performance classes")
print("=" * 80)

display(
    release_performance_class_distribution_df
)

# ============================================================
# Score statistics by class
# ============================================================

release_performance_class_statistics_df = (
    release_classification_df.loc[
        valid_release_classification_mask
    ]
    .groupby(
        "release_performance_class",
        observed=True
    )
    .agg(
        Releases=(
            "release_id",
            "count"
        ),
        Minimum_Score=(
            "composite_release_performance_score",
            "min"
        ),
        Median_Score=(
            "composite_release_performance_score",
            "median"
        ),
        Mean_Score=(
            "composite_release_performance_score",
            "mean"
        ),
        Maximum_Score=(
            "composite_release_performance_score",
            "max"
        ),
        Median_Percentile=(
            "composite_release_performance_percentile",
            "median"
        ),
        Median_Evidence_Coverage=(
            "composite_weight_coverage_pct",
            "median"
        ),
    )
    .reset_index()
)

print()
print("Release performance class statistics")
print("=" * 80)

display(
    release_performance_class_statistics_df
)

# ============================================================
# Class threshold summary
# ============================================================

release_class_threshold_summary_df = pd.DataFrame(
    [
        {
            "Release Performance Class":
                "Low Release Performance",
            "Percentile Range":
                "0-25",
        },
        {
            "Release Performance Class":
                "Moderate Release Performance",
            "Percentile Range":
                ">25-50",
        },
        {
            "Release Performance Class":
                "High Release Performance",
            "Percentile Range":
                ">50-75",
        },
        {
            "Release Performance Class":
                "Very High Release Performance",
            "Percentile Range":
                ">75-90",
        },
        {
            "Release Performance Class":
                "Exceptional Release Performance",
            "Percentile Range":
                ">90-100",
        },
    ]
)

print()
print("Release performance classification thresholds")
print("=" * 80)

display(
    release_class_threshold_summary_df
)

# ============================================================
# Strongest releases by class
# ============================================================

top_release_classes_df = (
    release_classification_df.loc[
        valid_release_classification_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "composite_release_performance_percentile",
            "composite_release_performance_rank",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "release_performance_class",
        ]
    ]
    .sort_values(
        by=[
            "composite_release_performance_score"
        ],
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest PMIP classified release performances")
print("=" * 80)

display(
    top_release_classes_df
)

# ============================================================
# Overall summary
# ============================================================

release_classification_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_classification_df
            ),
        },
        {
            "Metric": "Releases classified",
            "Observed Value": int(
                valid_release_classification_mask.sum()
            ),
        },
        {
            "Metric": "Releases without composite evidence",
            "Observed Value": int(
                (~valid_release_classification_mask).sum()
            ),
        },
        {
            "Metric": "Performance classes assigned",
            "Observed Value": 5,
        },
        {
            "Metric": "Exceptional-performance releases",
            "Observed Value": int(
                (
                    release_classification_df[
                        "release_performance_class"
                    ]
                    == "Exceptional Release Performance"
                ).sum()
            ),
        },
        {
            "Metric": "Very-high-or-better releases",
            "Observed Value": int(
                release_classification_df[
                    "release_performance_class"
                ].isin(
                    [
                        "Very High Release Performance",
                        "Exceptional Release Performance",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Release performance classification summary")
print("=" * 80)

display(
    release_classification_summary_df
)

# ============================================================
# Validation
# ============================================================

release_classification_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11.3 completed",
            "Passed": section_11_3_complete,
        },
        {
            "Validation Area": "Composite scores available",
            "Passed": release_classification_df[
                "composite_release_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Composite percentiles available",
            "Passed": release_classification_df[
                "composite_release_performance_percentile"
            ].notna().any(),
        },
        {
            "Validation Area": "Release performance classes assigned",
            "Passed": release_classification_df[
                "release_performance_class"
            ].notna().all(),
        },
        {
            "Validation Area": "All scored releases classified",
            "Passed": (
                release_classification_df.loc[
                    valid_release_classification_mask,
                    "release_performance_class"
                ]
                .astype(str)
                .ne(
                    "No Composite Performance Evidence"
                )
                .all()
            ),
        },
        {
            "Validation Area": "Classification thresholds defined",
            "Passed": (
                len(
                    release_class_threshold_summary_df
                )
                == 5
            ),
        },
    ]
)

print()
print("Release Performance Classes validation")
print("=" * 80)

display(
    release_classification_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_4_complete = bool(
    release_classification_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.4 completion status: "
    f"{section_11_4_complete}"
)

if not section_11_4_complete:
    raise RuntimeError(
        "Release Performance Classes did not "
        "complete successfully."
    )
```

### Interpretation

All 4,593 releases were successfully assigned to one of five performance classes using percentile-based thresholds.

The classification produced a balanced distribution across Low, Moderate and High performance, with smaller groups representing Very High and Exceptional performance.

A total of 460 releases were classified as Exceptional Release Performance, while 1,149 releases were classified as Very High or Exceptional.

The class statistics also show a clear increase in composite score across the ordered categories, confirming that the classification framework is consistent with the underlying release-performance score.

## 11.5 Ranking and Percentiles

### Purpose

This subsection formalises the relative position of each release within the final PMIP scoring framework.

The aim is to convert the composite release-performance score into interpretable rankings and percentile positions so that releases can be compared directly across the full dataset.

The analysis will:

- confirm the overall release ranking;
- calculate percentile standing;
- identify top-performing percentile groups;
- distinguish top 25%, top 10%, top 5% and top 1% releases;
- compare ranking strength with evidence coverage;
- prepare the ranking outputs required for high-priority release identification in Section 11.6.

Higher percentile values and lower numerical rank values indicate stronger relative release performance.

```python
# ============================================================
# Section 11.5 - Ranking and Percentiles
# ============================================================

if not section_11_4_complete:
    raise RuntimeError(
        "Section 11.4 must be completed before assessing "
        "Ranking and Percentiles."
    )

# ============================================================
# Working source
# ============================================================

release_ranking_df = release_classification_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_ranking_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "composite_release_performance_percentile",
    "composite_release_performance_rank",
    "release_performance_class",
    "composite_weight_coverage_pct",
]

missing_ranking_fields = [
    field
    for field in required_ranking_fields
    if field not in release_ranking_df.columns
]

if missing_ranking_fields:
    raise RuntimeError(
        "Required Ranking and Percentile fields are missing: "
        f"{missing_ranking_fields}"
    )

# ============================================================
# Valid ranked releases
# ============================================================

valid_ranking_mask = (
    release_ranking_df[
        "composite_release_performance_score"
    ].notna()
    & release_ranking_df[
        "composite_release_performance_percentile"
    ].notna()
    & release_ranking_df[
        "composite_release_performance_rank"
    ].notna()
)

# ============================================================
# Ranking position from strongest to weakest
#
# Rank 1 = strongest composite release-performance score.
# ============================================================

release_ranking_df[
    "release_rank_position"
] = np.nan

release_ranking_df.loc[
    valid_ranking_mask,
    "release_rank_position"
] = (
    release_ranking_df.loc[
        valid_ranking_mask,
        "composite_release_performance_score"
    ]
    .rank(
        method="min",
        ascending=False
    )
)

# ============================================================
# Percentile strength
#
# Existing percentile is preserved.
# Higher percentile = stronger relative performance.
# ============================================================

release_ranking_df[
    "release_performance_percentile"
] = (
    release_ranking_df[
        "composite_release_performance_percentile"
    ]
)

# ============================================================
# Percentile groups
# ============================================================

release_ranking_df[
    "release_percentile_group"
] = "Below Top 25%"

release_ranking_df.loc[
    valid_ranking_mask
    & (
        release_ranking_df[
            "release_performance_percentile"
        ] >= 75
    ),
    "release_percentile_group"
] = "Top 25%"

release_ranking_df.loc[
    valid_ranking_mask
    & (
        release_ranking_df[
            "release_performance_percentile"
        ] >= 90
    ),
    "release_percentile_group"
] = "Top 10%"

release_ranking_df.loc[
    valid_ranking_mask
    & (
        release_ranking_df[
            "release_performance_percentile"
        ] >= 95
    ),
    "release_percentile_group"
] = "Top 5%"

release_ranking_df.loc[
    valid_ranking_mask
    & (
        release_ranking_df[
            "release_performance_percentile"
        ] >= 99
    ),
    "release_percentile_group"
] = "Top 1%"

# ============================================================
# Exact percentile thresholds
# ============================================================

ranked_scores = (
    release_ranking_df.loc[
        valid_ranking_mask,
        "composite_release_performance_score"
    ]
)

percentile_thresholds = {
    "Top 25%": ranked_scores.quantile(0.75),
    "Top 10%": ranked_scores.quantile(0.90),
    "Top 5%": ranked_scores.quantile(0.95),
    "Top 1%": ranked_scores.quantile(0.99),
}

ranking_threshold_summary_df = pd.DataFrame(
    [
        {
            "Percentile Group": group,
            "Minimum Composite Score": threshold,
        }
        for group, threshold
        in percentile_thresholds.items()
    ]
)

print("Assessing PMIP release rankings and percentiles")
print("=" * 80)

print()
print("Ranking percentile thresholds")
print("=" * 80)

display(
    ranking_threshold_summary_df
)

# ============================================================
# Percentile-group distribution
# ============================================================

percentile_group_order = [
    "Below Top 25%",
    "Top 25%",
    "Top 10%",
    "Top 5%",
    "Top 1%",
]

release_ranking_df[
    "release_percentile_group"
] = pd.Categorical(
    release_ranking_df[
        "release_percentile_group"
    ],
    categories=percentile_group_order,
    ordered=True,
)

release_percentile_distribution_df = (
    release_ranking_df[
        "release_percentile_group"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Percentile Group"
    )
    .reset_index(
        name="Releases"
    )
)

release_percentile_distribution_df[
    "Share (%)"
] = (
    release_percentile_distribution_df[
        "Releases"
    ]
    / len(
        release_ranking_df
    )
    * 100
)

print()
print("Release percentile-group distribution")
print("=" * 80)

display(
    release_percentile_distribution_df
)

# ============================================================
# Ranking summary statistics
# ============================================================

ranking_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_ranking_df
            ),
        },
        {
            "Metric": "Ranked releases",
            "Observed Value": int(
                valid_ranking_mask.sum()
            ),
        },
        {
            "Metric": "Unranked releases",
            "Observed Value": int(
                (~valid_ranking_mask).sum()
            ),
        },
        {
            "Metric": "Highest percentile",
            "Observed Value": (
                release_ranking_df.loc[
                    valid_ranking_mask,
                    "release_performance_percentile"
                ].max()
            ),
        },
        {
            "Metric": "Median percentile",
            "Observed Value": (
                release_ranking_df.loc[
                    valid_ranking_mask,
                    "release_performance_percentile"
                ].median()
            ),
        },
        {
            "Metric": "Top 25% releases",
            "Observed Value": int(
                (
                    release_ranking_df[
                        "release_performance_percentile"
                    ] >= 75
                ).sum()
            ),
        },
        {
            "Metric": "Top 10% releases",
            "Observed Value": int(
                (
                    release_ranking_df[
                        "release_performance_percentile"
                    ] >= 90
                ).sum()
            ),
        },
        {
            "Metric": "Top 5% releases",
            "Observed Value": int(
                (
                    release_ranking_df[
                        "release_performance_percentile"
                    ] >= 95
                ).sum()
            ),
        },
        {
            "Metric": "Top 1% releases",
            "Observed Value": int(
                (
                    release_ranking_df[
                        "release_performance_percentile"
                    ] >= 99
                ).sum()
            ),
        },
    ]
)

print()
print("Release ranking summary")
print("=" * 80)

display(
    ranking_summary_df
)

# ============================================================
# Evidence coverage by percentile group
# ============================================================

ranking_evidence_summary_df = (
    release_ranking_df.loc[
        valid_ranking_mask
    ]
    .groupby(
        "release_percentile_group",
        observed=True
    )
    .agg(
        Releases=(
            "release_id",
            "count"
        ),
        Median_Composite_Score=(
            "composite_release_performance_score",
            "median"
        ),
        Median_Percentile=(
            "release_performance_percentile",
            "median"
        ),
        Median_Evidence_Coverage=(
            "composite_weight_coverage_pct",
            "median"
        ),
    )
    .reset_index()
)

print()
print("Ranking evidence summary")
print("=" * 80)

display(
    ranking_evidence_summary_df
)

# ============================================================
# Highest-ranked releases
# ============================================================

top_ranked_releases_df = (
    release_ranking_df.loc[
        valid_ranking_mask,
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_percentile",
            "release_rank_position",
            "release_percentile_group",
            "release_performance_class",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
        ]
    ]
    .sort_values(
        by=[
            "release_rank_position",
            "composite_release_performance_score",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Highest-ranked PMIP releases")
print("=" * 80)

display(
    top_ranked_releases_df
)

# ============================================================
# Top 1% release table
# ============================================================

top_one_percent_releases_df = (
    release_ranking_df.loc[
        valid_ranking_mask
        & (
            release_ranking_df[
                "release_performance_percentile"
            ] >= 99
        ),
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_percentile",
            "release_rank_position",
            "release_performance_class",
            "composite_weight_coverage_pct",
        ]
    ]
    .sort_values(
        by="release_rank_position",
        ascending=True
    )
    .reset_index(drop=True)
)

print()
print("PMIP top 1% releases")
print("=" * 80)

display(
    top_one_percent_releases_df
)

# ============================================================
# Ranking consistency check
#
# Stronger score should correspond to:
# - higher percentile
# - lower numerical rank
# ============================================================

ranking_consistency_score_percentile = (
    release_ranking_df.loc[
        valid_ranking_mask,
        [
            "composite_release_performance_score",
            "release_performance_percentile",
        ]
    ]
    .corr(
        method="spearman"
    )
    .iloc[
        0,
        1
    ]
)

ranking_consistency_score_rank = (
    release_ranking_df.loc[
        valid_ranking_mask,
        [
            "composite_release_performance_score",
            "release_rank_position",
        ]
    ]
    .corr(
        method="spearman"
    )
    .iloc[
        0,
        1
    ]
)

ranking_consistency_df = pd.DataFrame(
    [
        {
            "Consistency Test":
                "Composite score vs percentile",
            "Expected Relationship":
                "Strong positive",
            "Observed Spearman Correlation":
                ranking_consistency_score_percentile,
        },
        {
            "Consistency Test":
                "Composite score vs numerical rank",
            "Expected Relationship":
                "Strong negative",
            "Observed Spearman Correlation":
                ranking_consistency_score_rank,
        },
    ]
)

print()
print("Ranking consistency assessment")
print("=" * 80)

display(
    ranking_consistency_df
)

# ============================================================
# Validation
# ============================================================

ranking_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11.4 completed",
            "Passed": section_11_4_complete,
        },
        {
            "Validation Area": "Composite scores available",
            "Passed": release_ranking_df[
                "composite_release_performance_score"
            ].notna().any(),
        },
        {
            "Validation Area": "Release ranks available",
            "Passed": release_ranking_df.loc[
                valid_ranking_mask,
                "release_rank_position"
            ].notna().all(),
        },
        {
            "Validation Area": "Release percentiles available",
            "Passed": release_ranking_df.loc[
                valid_ranking_mask,
                "release_performance_percentile"
            ].notna().all(),
        },
        {
            "Validation Area": "Percentile groups assigned",
            "Passed": release_ranking_df[
                "release_percentile_group"
            ].notna().all(),
        },
        {
            "Validation Area": "Ranking thresholds calculated",
            "Passed": ranking_threshold_summary_df[
                "Minimum Composite Score"
            ].notna().all(),
        },
        {
            "Validation Area": "Score-percentile ordering consistent",
            "Passed": bool(
                ranking_consistency_score_percentile
                > 0.99
            ),
        },
        {
            "Validation Area": "Score-rank ordering consistent",
            "Passed": bool(
                ranking_consistency_score_rank
                < -0.99
            ),
        },
    ]
)

print()
print("Ranking and Percentiles validation")
print("=" * 80)

display(
    ranking_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_5_complete = bool(
    ranking_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.5 completion status: "
    f"{section_11_5_complete}"
)

if not section_11_5_complete:
    raise RuntimeError(
        "Ranking and Percentiles did not "
        "complete successfully."
    )
```

### Interpretation

All 4,593 releases were successfully ranked and assigned percentile positions based on the composite release-performance score.

The ranking framework showed perfect consistency between composite score, percentile position and numerical rank.

The top-performing groups were clearly identified, including 1,149 releases in the top 25%, 460 in the top 10%, 230 in the top 5% and 46 in the top 1%.

However, the Top 1% group had substantially lower median evidence coverage than the other high-ranking groups. This indicates that some extremely high rankings may be based on limited available evidence.

Evidence strength should therefore be considered alongside performance rank when identifying high-priority releases.

## 11.6 High-Priority Release Signals

### Purpose

This subsection identifies the releases that should receive the strongest PMIP attention by combining final performance strength with evidence quality and unusual-performance context.

The aim is to avoid treating a high composite score as sufficient on its own.

A high-priority release should ideally show:

- strong relative release performance;
- sufficient evidence coverage;
- a high percentile or performance class;
- and, where present, unusual-performance signals that may require deeper interpretation.

The unusual-performance indicators are used as contextual intelligence rather than direct performance rewards or penalties.

This produces a more cautious and evidence-aware prioritisation framework for release monitoring and further investigation.

```python
# ============================================================
# Section 11.6 - High-Priority Release Signals
# ============================================================

if not section_11_5_complete:
    raise RuntimeError(
        "Section 11.5 must be completed before identifying "
        "High-Priority Release Signals."
    )

# ============================================================
# Working source
# ============================================================

high_priority_release_df = release_ranking_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_priority_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_rank_position",
    "release_performance_class",
    "composite_weight_coverage_pct",
    "composite_evidence_strength",
]

missing_priority_fields = [
    field
    for field in required_priority_fields
    if field not in high_priority_release_df.columns
]

if missing_priority_fields:
    raise RuntimeError(
        "Required High-Priority Release fields are missing: "
        f"{missing_priority_fields}"
    )

# ============================================================
# Evidence thresholds
# ============================================================

minimum_priority_evidence_coverage = 70.0
strong_priority_evidence_coverage = 90.0

high_priority_release_df[
    "sufficient_priority_evidence"
] = (
    high_priority_release_df[
        "composite_weight_coverage_pct"
    ] >= minimum_priority_evidence_coverage
)

high_priority_release_df[
    "strong_priority_evidence"
] = (
    high_priority_release_df[
        "composite_weight_coverage_pct"
    ] >= strong_priority_evidence_coverage
)

# ============================================================
# Performance-tier signals
# ============================================================

high_priority_release_df[
    "top_25_performance_signal"
] = (
    high_priority_release_df[
        "release_performance_percentile"
    ] >= 75
)

high_priority_release_df[
    "top_10_performance_signal"
] = (
    high_priority_release_df[
        "release_performance_percentile"
    ] >= 90
)

high_priority_release_df[
    "top_5_performance_signal"
] = (
    high_priority_release_df[
        "release_performance_percentile"
    ] >= 95
)

high_priority_release_df[
    "top_1_performance_signal"
] = (
    high_priority_release_df[
        "release_performance_percentile"
    ] >= 99
)

# ============================================================
# Attach unusual-performance context
# ============================================================

unusual_context_fields = [
    "unusual_performance_flag",
    "unusual_performance_flag_count",
    "unusual_performance_pattern",
    "priority_unusual_performance_flag",
    "unusual_performance_evidence_strength",
]

available_unusual_context_fields = [
    field
    for field in unusual_context_fields
    if field in high_priority_release_df.columns
]

# Safe defaults where fields are absent
if "unusual_performance_flag" not in high_priority_release_df.columns:
    high_priority_release_df[
        "unusual_performance_flag"
    ] = False

if "unusual_performance_flag_count" not in high_priority_release_df.columns:
    high_priority_release_df[
        "unusual_performance_flag_count"
    ] = 0

if "priority_unusual_performance_flag" not in high_priority_release_df.columns:
    high_priority_release_df[
        "priority_unusual_performance_flag"
    ] = False

if "unusual_performance_pattern" not in high_priority_release_df.columns:
    high_priority_release_df[
        "unusual_performance_pattern"
    ] = "No Unusual Pattern"

# ============================================================
# Priority logic
#
# Tier 1:
# top 5% + strong evidence
#
# Tier 2:
# top 10% + sufficient evidence
#
# Tier 3:
# top 25% + sufficient evidence
#
# Context Watch:
# top 25% + unusual-performance flag
#
# Limited-Evidence Watch:
# top 10% but weak evidence coverage
# ============================================================

high_priority_release_df[
    "release_priority_class"
] = "Standard Monitoring"

# Tier 3
high_priority_release_df.loc[
    high_priority_release_df[
        "top_25_performance_signal"
    ]
    & high_priority_release_df[
        "sufficient_priority_evidence"
    ],
    "release_priority_class"
] = "Priority Tier 3"

# Tier 2
high_priority_release_df.loc[
    high_priority_release_df[
        "top_10_performance_signal"
    ]
    & high_priority_release_df[
        "sufficient_priority_evidence"
    ],
    "release_priority_class"
] = "Priority Tier 2"

# Tier 1
high_priority_release_df.loc[
    high_priority_release_df[
        "top_5_performance_signal"
    ]
    & high_priority_release_df[
        "strong_priority_evidence"
    ],
    "release_priority_class"
] = "Priority Tier 1"

# Limited-evidence watch
high_priority_release_df.loc[
    high_priority_release_df[
        "top_10_performance_signal"
    ]
    & ~high_priority_release_df[
        "sufficient_priority_evidence"
    ],
    "release_priority_class"
] = "High Performance - Limited Evidence"

# Contextual unusual-performance watch
high_priority_release_df.loc[
    high_priority_release_df[
        "top_25_performance_signal"
    ]
    & high_priority_release_df[
        "unusual_performance_flag"
    ],
    "release_priority_class"
] = np.where(
    high_priority_release_df.loc[
        high_priority_release_df[
            "top_25_performance_signal"
        ]
        & high_priority_release_df[
            "unusual_performance_flag"
        ],
        "release_priority_class"
    ].isin(
        [
            "Priority Tier 1",
            "Priority Tier 2",
            "Priority Tier 3",
        ]
    ),
    high_priority_release_df.loc[
        high_priority_release_df[
            "top_25_performance_signal"
        ]
        & high_priority_release_df[
            "unusual_performance_flag"
        ],
        "release_priority_class"
    ]
    + " + Unusual Context",
    "Performance Watch + Unusual Context",
)

# ============================================================
# Binary priority flags
# ============================================================

high_priority_release_df[
    "high_priority_release_flag"
] = (
    high_priority_release_df[
        "release_priority_class"
    ].isin(
        [
            "Priority Tier 1",
            "Priority Tier 2",
            "Priority Tier 3",
            "Priority Tier 1 + Unusual Context",
            "Priority Tier 2 + Unusual Context",
            "Priority Tier 3 + Unusual Context",
        ]
    )
)

high_priority_release_df[
    "priority_review_flag"
] = (
    high_priority_release_df[
        "high_priority_release_flag"
    ]
    | high_priority_release_df[
        "release_priority_class"
    ].isin(
        [
            "High Performance - Limited Evidence",
            "Performance Watch + Unusual Context",
        ]
    )
)

# ============================================================
# Priority score
#
# Performance remains dominant.
# Evidence strengthens confidence.
# Unusualness is a small contextual addition.
# ============================================================

performance_component = (
    high_priority_release_df[
        "release_performance_percentile"
    ]
)

evidence_component = (
    high_priority_release_df[
        "composite_weight_coverage_pct"
    ]
)

unusual_context_component = (
    high_priority_release_df[
        "unusual_performance_flag_count"
    ]
    .clip(
        lower=0,
        upper=4
    )
    / 4
    * 100
)

high_priority_release_df[
    "priority_signal_score"
] = (
    performance_component * 0.70
    + evidence_component * 0.25
    + unusual_context_component * 0.05
)

# ============================================================
# Priority rank
# ============================================================

high_priority_release_df[
    "priority_signal_rank"
] = (
    high_priority_release_df[
        "priority_signal_score"
    ]
    .rank(
        method="min",
        ascending=False
    )
)

# ============================================================
# Priority summary
# ============================================================

priority_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                high_priority_release_df
            ),
        },
        {
            "Metric": "High-priority releases",
            "Observed Value": int(
                high_priority_release_df[
                    "high_priority_release_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Releases requiring priority review",
            "Observed Value": int(
                high_priority_release_df[
                    "priority_review_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Top 5% with strong evidence",
            "Observed Value": int(
                (
                    high_priority_release_df[
                        "top_5_performance_signal"
                    ]
                    & high_priority_release_df[
                        "strong_priority_evidence"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Top 10% with limited evidence",
            "Observed Value": int(
                (
                    high_priority_release_df[
                        "top_10_performance_signal"
                    ]
                    & ~high_priority_release_df[
                        "sufficient_priority_evidence"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "High-performance releases with unusual context",
            "Observed Value": int(
                (
                    high_priority_release_df[
                        "top_25_performance_signal"
                    ]
                    & high_priority_release_df[
                        "unusual_performance_flag"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Median priority signal score",
            "Observed Value": (
                high_priority_release_df[
                    "priority_signal_score"
                ].median()
            ),
        },
        {
            "Metric": "Maximum priority signal score",
            "Observed Value": (
                high_priority_release_df[
                    "priority_signal_score"
                ].max()
            ),
        },
    ]
)

print("Assessing PMIP high-priority release signals")
print("=" * 80)

display(
    priority_summary_df
)

# ============================================================
# Priority-class distribution
# ============================================================

priority_class_distribution_df = (
    high_priority_release_df[
        "release_priority_class"
    ]
    .value_counts()
    .rename_axis(
        "Release Priority Class"
    )
    .reset_index(
        name="Releases"
    )
)

priority_class_distribution_df[
    "Share (%)"
] = (
    priority_class_distribution_df[
        "Releases"
    ]
    / len(
        high_priority_release_df
    )
    * 100
)

print()
print("Release priority-class distribution")
print("=" * 80)

display(
    priority_class_distribution_df
)

# ============================================================
# Evidence-aware top performance summary
# ============================================================

evidence_priority_summary_df = pd.DataFrame(
    [
        {
            "Performance Group": "Top 25%",
            "Releases": int(
                high_priority_release_df[
                    "top_25_performance_signal"
                ].sum()
            ),
            "With >=70% Evidence": int(
                (
                    high_priority_release_df[
                        "top_25_performance_signal"
                    ]
                    & high_priority_release_df[
                        "sufficient_priority_evidence"
                    ]
                ).sum()
            ),
            "With >=90% Evidence": int(
                (
                    high_priority_release_df[
                        "top_25_performance_signal"
                    ]
                    & high_priority_release_df[
                        "strong_priority_evidence"
                    ]
                ).sum()
            ),
        },
        {
            "Performance Group": "Top 10%",
            "Releases": int(
                high_priority_release_df[
                    "top_10_performance_signal"
                ].sum()
            ),
            "With >=70% Evidence": int(
                (
                    high_priority_release_df[
                        "top_10_performance_signal"
                    ]
                    & high_priority_release_df[
                        "sufficient_priority_evidence"
                    ]
                ).sum()
            ),
            "With >=90% Evidence": int(
                (
                    high_priority_release_df[
                        "top_10_performance_signal"
                    ]
                    & high_priority_release_df[
                        "strong_priority_evidence"
                    ]
                ).sum()
            ),
        },
        {
            "Performance Group": "Top 5%",
            "Releases": int(
                high_priority_release_df[
                    "top_5_performance_signal"
                ].sum()
            ),
            "With >=70% Evidence": int(
                (
                    high_priority_release_df[
                        "top_5_performance_signal"
                    ]
                    & high_priority_release_df[
                        "sufficient_priority_evidence"
                    ]
                ).sum()
            ),
            "With >=90% Evidence": int(
                (
                    high_priority_release_df[
                        "top_5_performance_signal"
                    ]
                    & high_priority_release_df[
                        "strong_priority_evidence"
                    ]
                ).sum()
            ),
        },
        {
            "Performance Group": "Top 1%",
            "Releases": int(
                high_priority_release_df[
                    "top_1_performance_signal"
                ].sum()
            ),
            "With >=70% Evidence": int(
                (
                    high_priority_release_df[
                        "top_1_performance_signal"
                    ]
                    & high_priority_release_df[
                        "sufficient_priority_evidence"
                    ]
                ).sum()
            ),
            "With >=90% Evidence": int(
                (
                    high_priority_release_df[
                        "top_1_performance_signal"
                    ]
                    & high_priority_release_df[
                        "strong_priority_evidence"
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Evidence-aware performance priority summary")
print("=" * 80)

display(
    evidence_priority_summary_df
)

# ============================================================
# Highest-priority releases
# ============================================================

top_priority_releases_df = (
    high_priority_release_df.loc[
        high_priority_release_df[
            "priority_review_flag"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_percentile",
            "release_rank_position",
            "release_performance_class",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "unusual_performance_flag_count",
            "unusual_performance_pattern",
            "release_priority_class",
            "priority_signal_score",
            "priority_signal_rank",
        ]
    ]
    .sort_values(
        by=[
            "priority_signal_score",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(30)
    .reset_index(drop=True)
)

print()
print("Highest-priority PMIP release signals")
print("=" * 80)

display(
    top_priority_releases_df
)

# ============================================================
# Limited-evidence high performers
# ============================================================

limited_evidence_high_performers_df = (
    high_priority_release_df.loc[
        high_priority_release_df[
            "top_10_performance_signal"
        ]
        & ~high_priority_release_df[
            "sufficient_priority_evidence"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_percentile",
            "release_rank_position",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "release_priority_class",
        ]
    ]
    .sort_values(
        by="release_performance_percentile",
        ascending=False
    )
    .reset_index(drop=True)
)

print()
print("High-performing releases with limited evidence")
print("=" * 80)

display(
    limited_evidence_high_performers_df
)

# ============================================================
# Validation
# ============================================================

high_priority_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11.5 completed",
            "Passed": section_11_5_complete,
        },
        {
            "Validation Area": "Performance percentile available",
            "Passed": high_priority_release_df[
                "release_performance_percentile"
            ].notna().all(),
        },
        {
            "Validation Area": "Evidence coverage available",
            "Passed": high_priority_release_df[
                "composite_weight_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Priority classes assigned",
            "Passed": high_priority_release_df[
                "release_priority_class"
            ].notna().all(),
        },
        {
            "Validation Area": "Priority scores calculated",
            "Passed": high_priority_release_df[
                "priority_signal_score"
            ].notna().all(),
        },
        {
            "Validation Area": "Priority ranks calculated",
            "Passed": high_priority_release_df[
                "priority_signal_rank"
            ].notna().all(),
        },
        {
            "Validation Area": "High-priority flags assigned",
            "Passed": high_priority_release_df[
                "high_priority_release_flag"
            ].notna().all(),
        },
        {
            "Validation Area": "Priority-review flags assigned",
            "Passed": high_priority_release_df[
                "priority_review_flag"
            ].notna().all(),
        },
    ]
)

print()
print("High-Priority Release Signals validation")
print("=" * 80)

display(
    high_priority_validation_df
)

# ============================================================
# Completion
# ============================================================

section_11_6_complete = bool(
    high_priority_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 11.6 completion status: "
    f"{section_11_6_complete}"
)

if not section_11_6_complete:
    raise RuntimeError(
        "High-Priority Release Signals did not "
        "complete successfully."
    )
```

The analysis focuses on releases with the lowest composite scores, weakest percentile positions and low-performance classifications.

The aim is to distinguish genuinely weak releases from cases where poor ranking may be influenced by limited evidence coverage.

This allows underperformance to be interpreted more carefully rather than assuming that every low score represents the same level of commercial or audience weakness.

```python
# ============================================================
# Section 12.2 - Underperforming Releases
# ============================================================

if not section_12_1_complete:
    raise RuntimeError(
        "Section 12.1 must be completed before assessing "
        "Underperforming Releases."
    )

# ============================================================
# Working source
# ============================================================

underperforming_release_df = results_interpretation_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_underperformance_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_rank_position",
    "release_performance_class",
    "composite_weight_coverage_pct",
    "composite_evidence_strength",
]

missing_underperformance_fields = [
    field
    for field in required_underperformance_fields
    if field not in underperforming_release_df.columns
]

if missing_underperformance_fields:
    raise RuntimeError(
        "Required underperformance fields are missing: "
        f"{missing_underperformance_fields}"
    )

# ============================================================
# Lowest-performing releases
# ============================================================

lowest_performing_releases_df = (
    underperforming_release_df[
        required_underperformance_fields
    ]
    .sort_values(
        by=[
            "composite_release_performance_score",
            "composite_weight_coverage_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print("Lowest-performing PMIP releases")
print("=" * 80)

display(
    lowest_performing_releases_df
)

# ============================================================
# Evidence-supported underperformers
#
# This filters the weakest releases to those with at least
# 70% evidence coverage so that low performance is supported
# by stronger multidimensional evidence.
# ============================================================

evidence_supported_underperformers_df = (
    underperforming_release_df.loc[
        (
            underperforming_release_df[
                "release_performance_percentile"
            ] <= 25
        )
        & (
            underperforming_release_df[
                "composite_weight_coverage_pct"
            ] >= 70
        ),
        required_underperformance_fields
    ]
    .sort_values(
        by=[
            "composite_release_performance_score",
            "composite_weight_coverage_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Evidence-supported underperforming releases")
print("=" * 80)

display(
    evidence_supported_underperformers_df
)

# ============================================================
# Limited-evidence weak performers
# ============================================================

limited_evidence_underperformers_df = (
    underperforming_release_df.loc[
        (
            underperforming_release_df[
                "release_performance_percentile"
            ] <= 25
        )
        & (
            underperforming_release_df[
                "composite_weight_coverage_pct"
            ] < 70
        ),
        required_underperformance_fields
    ]
    .sort_values(
        by="composite_release_performance_score",
        ascending=True
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Low-ranked releases with limited evidence")
print("=" * 80)

display(
    limited_evidence_underperformers_df
)

# ============================================================
# Underperformance summary
# ============================================================

low_performance_mask = (
    underperforming_release_df[
        "release_performance_percentile"
    ] <= 25
)

bottom_10_mask = (
    underperforming_release_df[
        "release_performance_percentile"
    ] <= 10
)

bottom_5_mask = (
    underperforming_release_df[
        "release_performance_percentile"
    ] <= 5
)

bottom_1_mask = (
    underperforming_release_df[
        "release_performance_percentile"
    ] <= 1
)

underperformance_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                underperforming_release_df
            ),
        },
        {
            "Metric": "Low-performance releases",
            "Observed Value": int(
                low_performance_mask.sum()
            ),
        },
        {
            "Metric": "Bottom 10% releases",
            "Observed Value": int(
                bottom_10_mask.sum()
            ),
        },
        {
            "Metric": "Bottom 5% releases",
            "Observed Value": int(
                bottom_5_mask.sum()
            ),
        },
        {
            "Metric": "Bottom 1% releases",
            "Observed Value": int(
                bottom_1_mask.sum()
            ),
        },
        {
            "Metric": "Minimum composite score",
            "Observed Value": (
                underperforming_release_df[
                    "composite_release_performance_score"
                ].min()
            ),
        },
        {
            "Metric": "Median evidence coverage among bottom 25%",
            "Observed Value": (
                underperforming_release_df.loc[
                    low_performance_mask,
                    "composite_weight_coverage_pct"
                ].median()
            ),
        },
        {
            "Metric": "Bottom 25% with >=70% evidence",
            "Observed Value": int(
                (
                    low_performance_mask
                    & (
                        underperforming_release_df[
                            "composite_weight_coverage_pct"
                        ] >= 70
                    )
                ).sum()
            ),
        },
        {
            "Metric": "Bottom 25% with <70% evidence",
            "Observed Value": int(
                (
                    low_performance_mask
                    & (
                        underperforming_release_df[
                            "composite_weight_coverage_pct"
                        ] < 70
                    )
                ).sum()
            ),
        },
    ]
)

print()
print("Underperformance summary")
print("=" * 80)

display(
    underperformance_summary_df
)

# ============================================================
# Evidence-strength distribution among low performers
# ============================================================

underperformance_evidence_distribution_df = (
    underperforming_release_df.loc[
        low_performance_mask,
        "composite_evidence_strength"
    ]
    .value_counts()
    .rename_axis(
        "Composite Evidence Strength"
    )
    .reset_index(
        name="Releases"
    )
)

underperformance_evidence_distribution_df[
    "Share (%)"
] = (
    underperformance_evidence_distribution_df[
        "Releases"
    ]
    / low_performance_mask.sum()
    * 100
)

print()
print("Evidence strength among low-performance releases")
print("=" * 80)

display(
    underperformance_evidence_distribution_df
)

# ============================================================
# Validation
# ============================================================

underperformance_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 12.1 completed",
            "Passed": section_12_1_complete,
        },
        {
            "Validation Area": "Composite scores available",
            "Passed": underperforming_release_df[
                "composite_release_performance_score"
            ].notna().all(),
        },
        {
            "Validation Area": "Percentiles available",
            "Passed": underperforming_release_df[
                "release_performance_percentile"
            ].notna().all(),
        },
        {
            "Validation Area": "Evidence coverage available",
            "Passed": underperforming_release_df[
                "composite_weight_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Low-performance releases identified",
            "Passed": bool(
                low_performance_mask.any()
            ),
        },
        {
            "Validation Area": "Evidence-supported underperformers assessed",
            "Passed": True,
        },
    ]
)

print()
print("Underperforming Releases validation")
print("=" * 80)

display(
    underperformance_validation_df
)

# ============================================================
# Completion
# ============================================================

section_12_2_complete = bool(
    underperformance_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 12.2 completion status: "
    f"{section_12_2_complete}"
)

if not section_12_2_complete:
    raise RuntimeError(
        "Underperforming Releases did not "
        "complete successfully."
    )
```

### Interpretation

The PMIP framework identified 1,148 releases within the lowest 25% of composite performance.

However, only 213 of these releases had at least 70% evidence coverage, showing that many low rankings are supported by incomplete multidimensional evidence.

Most low-performing releases were associated with moderate evidence strength, while a smaller group had very strong supporting evidence.

The evidence-supported underperformers therefore provide the most reliable examples of genuinely weak release performance.

## 12.3 Fast-Growing Releases

### Purpose

This subsection identifies releases associated with the strongest positive artist-growth context.

The analysis uses the artist growth score developed earlier in the notebook and combines it with the final release-performance results.

The aim is to distinguish releases connected to genuinely improving artist trajectories from releases that are performing strongly without clear evidence of broader artist growth.

Growth findings are interpreted alongside evidence availability and release performance so that high growth scores are not treated as sufficient evidence on their own.

```python
# ============================================================
# Section 12.3 - Fast-Growing Releases
# ============================================================

if not section_12_2_complete:
    raise RuntimeError(
        "Section 12.2 must be completed before assessing "
        "Fast-Growing Releases."
    )

# ============================================================
# Working source
# ============================================================

fast_growth_release_df = results_interpretation_df.copy()

# ============================================================
# Validate core fields
# ============================================================

required_fast_growth_fields = [
    "release_id",
    "Track",
    "Artist",
    "artist_growth_score",
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_weight_coverage_pct",
]

missing_fast_growth_fields = [
    field
    for field in required_fast_growth_fields
    if field not in fast_growth_release_df.columns
]

if missing_fast_growth_fields:
    raise RuntimeError(
        "Required fast-growth fields are missing: "
        f"{missing_fast_growth_fields}"
    )

# ============================================================
# Optional artist-growth context fields
# ============================================================

optional_growth_fields = [
    "artist_growth_direction",
    "artist_growth_class",
    "artist_growth_evidence",
    "artist_growth_evidence_class",
    "artist_growth_evidence_strength",
    "artist_recent_vs_earlier_change",
    "recent_vs_earlier_performance_change",
    "artist_growth_slope_per_year",
    "growth_slope_per_year",
]

available_growth_context_fields = [
    field
    for field in optional_growth_fields
    if field in fast_growth_release_df.columns
]

# ============================================================
# Releases with usable artist-growth evidence
# ============================================================

growth_evidence_mask = (
    fast_growth_release_df[
        "artist_growth_score"
    ].notna()
)

growth_assessed_releases = int(
    growth_evidence_mask.sum()
)

# ============================================================
# Artist-growth percentile
#
# Higher artist_growth_score = stronger relative growth context.
# ============================================================

fast_growth_release_df[
    "artist_growth_percentile"
] = np.nan

if growth_evidence_mask.any():

    fast_growth_release_df.loc[
        growth_evidence_mask,
        "artist_growth_percentile"
    ] = (
        fast_growth_release_df.loc[
            growth_evidence_mask,
            "artist_growth_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Growth groups
# ============================================================

fast_growth_release_df[
    "fast_growth_group"
] = "No Artist Growth Evidence"

fast_growth_release_df.loc[
    growth_evidence_mask,
    "fast_growth_group"
] = "Below Top 25% Growth"

fast_growth_release_df.loc[
    growth_evidence_mask
    & (
        fast_growth_release_df[
            "artist_growth_percentile"
        ] >= 75
    ),
    "fast_growth_group"
] = "Top 25% Growth"

fast_growth_release_df.loc[
    growth_evidence_mask
    & (
        fast_growth_release_df[
            "artist_growth_percentile"
        ] >= 90
    ),
    "fast_growth_group"
] = "Top 10% Growth"

fast_growth_release_df.loc[
    growth_evidence_mask
    & (
        fast_growth_release_df[
            "artist_growth_percentile"
        ] >= 95
    ),
    "fast_growth_group"
] = "Top 5% Growth"

fast_growth_release_df.loc[
    growth_evidence_mask
    & (
        fast_growth_release_df[
            "artist_growth_percentile"
        ] >= 99
    ),
    "fast_growth_group"
] = "Top 1% Growth"

# ============================================================
# Strong positive-growth flag
# ============================================================

fast_growth_release_df[
    "fast_growth_release_flag"
] = (
    growth_evidence_mask
    & (
        fast_growth_release_df[
            "artist_growth_percentile"
        ] >= 90
    )
)

# ============================================================
# Fast-growing releases with strong release performance
#
# This identifies cases where artist growth and release
# performance are both strong.
# ============================================================

fast_growth_release_df[
    "growth_and_performance_signal"
] = (
    fast_growth_release_df[
        "fast_growth_release_flag"
    ]
    & (
        fast_growth_release_df[
            "release_performance_percentile"
        ] >= 75
    )
)

# ============================================================
# Display fields
# ============================================================

fast_growth_display_fields = [
    "release_id",
    "Track",
    "Artist",
    "artist_growth_score",
    "artist_growth_percentile",
]

fast_growth_display_fields += [
    field
    for field in available_growth_context_fields
    if field not in fast_growth_display_fields
]

fast_growth_display_fields += [
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_weight_coverage_pct",
    "fast_growth_group",
    "growth_and_performance_signal",
]

# ============================================================
# Highest artist-growth releases
# ============================================================

highest_growth_releases_df = (
    fast_growth_release_df.loc[
        growth_evidence_mask,
        fast_growth_display_fields
    ]
    .sort_values(
        by=[
            "artist_growth_score",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print("Fastest-growing PMIP release contexts")
print("=" * 80)

display(
    highest_growth_releases_df
)

# ============================================================
# Fast growth + strong release performance
# ============================================================

growth_performance_releases_df = (
    fast_growth_release_df.loc[
        fast_growth_release_df[
            "growth_and_performance_signal"
        ],
        fast_growth_display_fields
    ]
    .sort_values(
        by=[
            "artist_growth_percentile",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Fast-growing releases with strong release performance")
print("=" * 80)

display(
    growth_performance_releases_df
)

# ============================================================
# Growth-group distribution
# ============================================================

growth_group_order = [
    "No Artist Growth Evidence",
    "Below Top 25% Growth",
    "Top 25% Growth",
    "Top 10% Growth",
    "Top 5% Growth",
    "Top 1% Growth",
]

fast_growth_release_df[
    "fast_growth_group"
] = pd.Categorical(
    fast_growth_release_df[
        "fast_growth_group"
    ],
    categories=growth_group_order,
    ordered=True,
)

growth_group_distribution_df = (
    fast_growth_release_df[
        "fast_growth_group"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Growth Group"
    )
    .reset_index(
        name="Releases"
    )
)

growth_group_distribution_df[
    "Share (%)"
] = (
    growth_group_distribution_df[
        "Releases"
    ]
    / len(
        fast_growth_release_df
    )
    * 100
)

print()
print("Fast-growth group distribution")
print("=" * 80)

display(
    growth_group_distribution_df
)

# ============================================================
# Fast-growth summary
# ============================================================

fast_growth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                fast_growth_release_df
            ),
        },
        {
            "Metric": "Releases with artist-growth evidence",
            "Observed Value": growth_assessed_releases,
        },
        {
            "Metric": "Releases without artist-growth evidence",
            "Observed Value": int(
                (~growth_evidence_mask).sum()
            ),
        },
        {
            "Metric": "Top 25% growth releases",
            "Observed Value": int(
                (
                    fast_growth_release_df[
                        "artist_growth_percentile"
                    ] >= 75
                ).sum()
            ),
        },
        {
            "Metric": "Top 10% growth releases",
            "Observed Value": int(
                (
                    fast_growth_release_df[
                        "artist_growth_percentile"
                    ] >= 90
                ).sum()
            ),
        },
        {
            "Metric": "Top 5% growth releases",
            "Observed Value": int(
                (
                    fast_growth_release_df[
                        "artist_growth_percentile"
                    ] >= 95
                ).sum()
            ),
        },
        {
            "Metric": "Top 1% growth releases",
            "Observed Value": int(
                (
                    fast_growth_release_df[
                        "artist_growth_percentile"
                    ] >= 99
                ).sum()
            ),
        },
        {
            "Metric": "Top 10% growth + top 25% performance",
            "Observed Value": int(
                fast_growth_release_df[
                    "growth_and_performance_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Median artist growth score",
            "Observed Value": (
                fast_growth_release_df.loc[
                    growth_evidence_mask,
                    "artist_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Maximum artist growth score",
            "Observed Value": (
                fast_growth_release_df.loc[
                    growth_evidence_mask,
                    "artist_growth_score"
                ].max()
            ),
        },
    ]
)

print()
print("Fast-growing release summary")
print("=" * 80)

display(
    fast_growth_summary_df
)

# ============================================================
# Performance distribution among top-growth releases
# ============================================================

top_growth_mask = (
    fast_growth_release_df[
        "artist_growth_percentile"
    ] >= 90
)

top_growth_performance_distribution_df = (
    fast_growth_release_df.loc[
        top_growth_mask,
        "release_performance_class"
    ]
    .value_counts()
    .rename_axis(
        "Release Performance Class"
    )
    .reset_index(
        name="Releases"
    )
)

top_growth_performance_distribution_df[
    "Share (%)"
] = (
    top_growth_performance_distribution_df[
        "Releases"
    ]
    / max(
        int(
            top_growth_mask.sum()
        ),
        1
    )
    * 100
)

print()
print("Performance classes among top 10% growth releases")
print("=" * 80)

display(
    top_growth_performance_distribution_df
)

# ============================================================
# Validation
# ============================================================

fast_growth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 12.2 completed",
            "Passed": section_12_2_complete,
        },
        {
            "Validation Area": "Artist growth score available",
            "Passed": (
                "artist_growth_score"
                in fast_growth_release_df.columns
            ),
        },
        {
            "Validation Area": "Artist growth evidence available",
            "Passed": bool(
                growth_evidence_mask.any()
            ),
        },
        {
            "Validation Area": "Growth percentiles calculated",
            "Passed": (
                fast_growth_release_df.loc[
                    growth_evidence_mask,
                    "artist_growth_percentile"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Fast-growth groups assigned",
            "Passed": fast_growth_release_df[
                "fast_growth_group"
            ].notna().all(),
        },
        {
            "Validation Area": "Fast-growing releases identified",
            "Passed": bool(
                fast_growth_release_df[
                    "fast_growth_release_flag"
                ].any()
            ),
        },
        {
            "Validation Area": "Growth-performance relationship assessed",
            "Passed": (
                "growth_and_performance_signal"
                in fast_growth_release_df.columns
            ),
        },
    ]
)

print()
print("Fast-Growing Releases validation")
print("=" * 80)

display(
    fast_growth_validation_df
)

# ============================================================
# Completion
# ============================================================

section_12_3_complete = bool(
    fast_growth_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 12.3 completion status: "
    f"{section_12_3_complete}"
)

if not section_12_3_complete:
    raise RuntimeError(
        "Fast-Growing Releases did not "
        "complete successfully."
    )
```

### Interpretation

The analysis identified 3,256 releases with usable artist-growth evidence, including 326 releases in the Top 10% for growth.

Among these fast-growing releases, 118 also ranked within the Top 25% for overall release performance, showing that strong artist growth is often accompanied by strong release-level results.

However, growth did not always correspond to high overall performance, as some fast-growing releases remained within moderate or low performance classes.

This suggests that artist growth provides useful context for release performance but should be interpreted alongside the wider PMIP scoring framework.

## 12.4 Geographic Performance Findings

### Purpose

This subsection examines how release performance is distributed across geographic reach and market breadth.

The analysis combines the final PMIP release-performance results with the geographic indicators developed earlier in the notebook.

The aim is to identify whether stronger releases are generally associated with broader market reach, and to highlight cases where strong performance is concentrated within a narrower geographic footprint.

These findings help distinguish globally broad performance from more regionally concentrated success.

```python
# ============================================================
# Section 12.4 - Geographic Performance Findings
# ============================================================

if not section_12_3_complete:
    raise RuntimeError(
        "Section 12.3 must be completed before assessing "
        "Geographic Performance Findings."
    )

# ============================================================
# Working source
# ============================================================

geographic_findings_df = results_interpretation_df.copy()

# ============================================================
# Validate required geographic fields
# ============================================================

required_geographic_fields = [
    "release_id",
    "Track",
    "Artist",
    "geographic_reach_score",
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_weight_coverage_pct",
]

missing_geographic_fields = [
    field
    for field in required_geographic_fields
    if field not in geographic_findings_df.columns
]

if missing_geographic_fields:
    raise RuntimeError(
        "Required geographic-performance fields are missing: "
        f"{missing_geographic_fields}"
    )

# ============================================================
# Optional geographic context
# ============================================================

optional_geographic_fields = [
    "countries_charted",
    "countries_represented",
    "market_breadth_pct",
    "market_breadth_score",
    "strong_markets_represented",
    "strong_market_presence_pct",
    "strong_market_presence_score",
    "emerging_markets_represented",
    "emerging_market_presence_pct",
    "emerging_market_presence_score",
    "geographic_reach_class",
]

available_geographic_context_fields = [
    field
    for field in optional_geographic_fields
    if field in geographic_findings_df.columns
]

# ============================================================
# Geographic evidence mask
# ============================================================

geographic_evidence_mask = (
    geographic_findings_df[
        "geographic_reach_score"
    ].notna()
)

# ============================================================
# Geographic reach percentile
# ============================================================

geographic_findings_df[
    "geographic_reach_percentile"
] = np.nan

if geographic_evidence_mask.any():

    geographic_findings_df.loc[
        geographic_evidence_mask,
        "geographic_reach_percentile"
    ] = (
        geographic_findings_df.loc[
            geographic_evidence_mask,
            "geographic_reach_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Broad geographic reach flag
# ============================================================

geographic_findings_df[
    "broad_geographic_reach_flag"
] = (
    geographic_findings_df[
        "geographic_reach_percentile"
    ] >= 75
)

# ============================================================
# Strong performance + broad reach
# ============================================================

geographic_findings_df[
    "strong_performance_broad_reach_flag"
] = (
    (
        geographic_findings_df[
            "release_performance_percentile"
        ] >= 75
    )
    & (
        geographic_findings_df[
            "geographic_reach_percentile"
        ] >= 75
    )
)

# ============================================================
# Strong performance + narrow reach
# ============================================================

geographic_findings_df[
    "strong_performance_narrow_reach_flag"
] = (
    (
        geographic_findings_df[
            "release_performance_percentile"
        ] >= 75
    )
    & (
        geographic_findings_df[
            "geographic_reach_percentile"
        ] < 50
    )
)

# ============================================================
# Geographic-performance relationship
# ============================================================

valid_geographic_performance_mask = (
    geographic_findings_df[
        "geographic_reach_score"
    ].notna()
    & geographic_findings_df[
        "composite_release_performance_score"
    ].notna()
)

geographic_performance_correlation = (
    geographic_findings_df.loc[
        valid_geographic_performance_mask,
        [
            "geographic_reach_score",
            "composite_release_performance_score",
        ]
    ]
    .corr(
        method="spearman"
    )
    .iloc[
        0,
        1
    ]
)

# ============================================================
# Geographic findings summary
# ============================================================

geographic_performance_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                geographic_findings_df
            ),
        },
        {
            "Metric": "Releases with geographic evidence",
            "Observed Value": int(
                geographic_evidence_mask.sum()
            ),
        },
        {
            "Metric": "Releases without geographic evidence",
            "Observed Value": int(
                (~geographic_evidence_mask).sum()
            ),
        },
        {
            "Metric": "Top 25% geographic reach releases",
            "Observed Value": int(
                geographic_findings_df[
                    "broad_geographic_reach_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Top 25% performance + top 25% reach",
            "Observed Value": int(
                geographic_findings_df[
                    "strong_performance_broad_reach_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Top 25% performance + below-median reach",
            "Observed Value": int(
                geographic_findings_df[
                    "strong_performance_narrow_reach_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Median geographic reach score",
            "Observed Value": (
                geographic_findings_df.loc[
                    geographic_evidence_mask,
                    "geographic_reach_score"
                ].median()
            ),
        },
        {
            "Metric": "Maximum geographic reach score",
            "Observed Value": (
                geographic_findings_df.loc[
                    geographic_evidence_mask,
                    "geographic_reach_score"
                ].max()
            ),
        },
        {
            "Metric": "Spearman correlation: reach vs performance",
            "Observed Value": (
                geographic_performance_correlation
            ),
        },
    ]
)

print("Assessing PMIP geographic performance findings")
print("=" * 80)

display(
    geographic_performance_summary_df
)

# ============================================================
# Highest geographic-reach releases
# ============================================================

geographic_display_fields = [
    "release_id",
    "Track",
    "Artist",
    "geographic_reach_score",
    "geographic_reach_percentile",
]

geographic_display_fields += [
    field
    for field in available_geographic_context_fields
    if field not in geographic_display_fields
]

geographic_display_fields += [
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_weight_coverage_pct",
]

highest_geographic_reach_releases_df = (
    geographic_findings_df.loc[
        geographic_evidence_mask,
        geographic_display_fields
    ]
    .sort_values(
        by=[
            "geographic_reach_score",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases with strongest geographic reach")
print("=" * 80)

display(
    highest_geographic_reach_releases_df
)

# ============================================================
# Strong performance with broad reach
# ============================================================

broad_reach_high_performers_df = (
    geographic_findings_df.loc[
        geographic_findings_df[
            "strong_performance_broad_reach_flag"
        ],
        geographic_display_fields
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "geographic_reach_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strong-performing releases with broad geographic reach")
print("=" * 80)

display(
    broad_reach_high_performers_df
)

# ============================================================
# Strong performance with narrow reach
# ============================================================

narrow_reach_high_performers_df = (
    geographic_findings_df.loc[
        geographic_findings_df[
            "strong_performance_narrow_reach_flag"
        ],
        geographic_display_fields
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "geographic_reach_percentile",
        ],
        ascending=[
            False,
            True,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Strong-performing releases with narrower geographic reach")
print("=" * 80)

display(
    narrow_reach_high_performers_df
)

# ============================================================
# Performance class by geographic reach quartile
# ============================================================

geographic_findings_df[
    "geographic_reach_quartile"
] = "No Geographic Evidence"

if geographic_evidence_mask.any():

    valid_geo_values = (
        geographic_findings_df.loc[
            geographic_evidence_mask,
            "geographic_reach_score"
        ]
    )

    try:

        geographic_findings_df.loc[
            geographic_evidence_mask,
            "geographic_reach_quartile"
        ] = pd.qcut(
            valid_geo_values,
            q=4,
            labels=[
                "Q1 Limited Reach",
                "Q2 Moderate Reach",
                "Q3 Broad Reach",
                "Q4 Very Broad Reach",
            ],
            duplicates="drop",
        ).astype(str)

    except ValueError:

        geographic_findings_df.loc[
            geographic_evidence_mask,
            "geographic_reach_quartile"
        ] = "Geographic Evidence Available"

geographic_performance_crosstab_df = (
    pd.crosstab(
        geographic_findings_df[
            "geographic_reach_quartile"
        ],
        geographic_findings_df[
            "release_performance_class"
        ],
    )
    .reset_index()
)

print()
print("Release performance by geographic-reach group")
print("=" * 80)

display(
    geographic_performance_crosstab_df
)

# ============================================================
# Performance summary by geographic reach group
# ============================================================

geographic_group_performance_summary_df = (
    geographic_findings_df.loc[
        geographic_evidence_mask
    ]
    .groupby(
        "geographic_reach_quartile",
        observed=True
    )
    .agg(
        Releases=(
            "release_id",
            "count"
        ),
        Median_Geographic_Reach_Score=(
            "geographic_reach_score",
            "median"
        ),
        Median_Composite_Performance_Score=(
            "composite_release_performance_score",
            "median"
        ),
        Median_Performance_Percentile=(
            "release_performance_percentile",
            "median"
        ),
    )
    .reset_index()
)

print()
print("Performance summary by geographic-reach group")
print("=" * 80)

display(
    geographic_group_performance_summary_df
)

# ============================================================
# Validation
# ============================================================

geographic_findings_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 12.3 completed",
            "Passed": section_12_3_complete,
        },
        {
            "Validation Area": "Geographic reach score available",
            "Passed": (
                "geographic_reach_score"
                in geographic_findings_df.columns
            ),
        },
        {
            "Validation Area": "Geographic evidence available",
            "Passed": bool(
                geographic_evidence_mask.any()
            ),
        },
        {
            "Validation Area": "Geographic percentiles calculated",
            "Passed": (
                geographic_findings_df.loc[
                    geographic_evidence_mask,
                    "geographic_reach_percentile"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Broad-reach performance assessed",
            "Passed": (
                "strong_performance_broad_reach_flag"
                in geographic_findings_df.columns
            ),
        },
        {
            "Validation Area": "Narrow-reach high performance assessed",
            "Passed": (
                "strong_performance_narrow_reach_flag"
                in geographic_findings_df.columns
            ),
        },
        {
            "Validation Area": "Geographic-performance relationship calculated",
            "Passed": pd.notna(
                geographic_performance_correlation
            ),
        },
    ]
)

print()
print("Geographic Performance Findings validation")
print("=" * 80)

display(
    geographic_findings_validation_df
)

# ============================================================
# Completion
# ============================================================

section_12_4_complete = bool(
    geographic_findings_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 12.4 completion status: "
    f"{section_12_4_complete}"
)

if not section_12_4_complete:
    raise RuntimeError(
        "Geographic Performance Findings did not "
        "complete successfully."
    )
```

### Interpretation

The geographic analysis identified a strong positive relationship between geographic reach and overall release performance, with a Spearman correlation of approximately 0.88.

Median composite performance increased steadily across the geographic-reach groups, from about 31.68 in the Limited Reach group to about 82.37 in the Very Broad Reach group.

A total of 523 releases combined Top 25% performance with Top 25% geographic reach, while only 76 strong-performing releases had below-median reach.

These findings suggest that broader geographic reach is strongly associated with stronger overall PMIP performance, although some releases still achieve high performance within more concentrated markets.

## 12.5 Artist–Release Comparisons

### Purpose

This subsection compares the performance of individual releases with the established performance level of their associated artists.

The analysis uses the artist baseline and artist–release relationship measures developed earlier in the notebook.

The aim is to identify releases that:

- outperform their artist’s usual level;
- broadly match their artist baseline;
- underperform relative to the artist’s historical performance;
- and represent potential breakout or unusually weak release cases.

This provides additional context by showing whether strong or weak release performance is consistent with the artist’s broader track record.

```python
# ============================================================
# Section 12.5 - Artist–Release Comparisons
# ============================================================

if not section_12_4_complete:
    raise RuntimeError(
        "Section 12.4 must be completed before assessing "
        "Artist–Release Comparisons."
    )

# ============================================================
# Working source
# ============================================================

artist_release_comparison_df = results_interpretation_df.copy()

# ============================================================
# Validate required fields
# ============================================================

required_artist_release_fields = [
    "release_id",
    "Track",
    "Artist",
    "release_vs_artist_baseline_score",
    "artist_release_relationship_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_release_performance_score",
    "composite_weight_coverage_pct",
]

missing_artist_release_fields = [
    field
    for field in required_artist_release_fields
    if field not in artist_release_comparison_df.columns
]

if missing_artist_release_fields:
    raise RuntimeError(
        "Required Artist–Release Comparison fields are missing: "
        f"{missing_artist_release_fields}"
    )

# ============================================================
# Optional context fields
# ============================================================

optional_artist_release_fields = [
    "artist_performance_score",
    "artist_performance_class",
    "release_vs_artist_difference",
    "release_vs_artist_direction",
    "release_vs_artist_class",
    "artist_baseline_evidence",
    "artist_baseline_evidence_strength",
    "artist_growth_score",
    "artist_momentum_score",
    "artist_release_relationship_type",
    "artist_release_relationship_class",
]

available_artist_release_context_fields = [
    field
    for field in optional_artist_release_fields
    if field in artist_release_comparison_df.columns
]

# ============================================================
# Baseline evidence mask
# ============================================================

artist_baseline_evidence_mask = (
    artist_release_comparison_df[
        "release_vs_artist_baseline_score"
    ].notna()
)

# ============================================================
# Relative artist comparison percentile
# ============================================================

artist_release_comparison_df[
    "release_vs_artist_percentile"
] = np.nan

if artist_baseline_evidence_mask.any():

    artist_release_comparison_df.loc[
        artist_baseline_evidence_mask,
        "release_vs_artist_percentile"
    ] = (
        artist_release_comparison_df.loc[
            artist_baseline_evidence_mask,
            "release_vs_artist_baseline_score"
        ]
        .rank(
            method="average",
            pct=True,
            ascending=True
        )
        * 100
    )

# ============================================================
# Artist-release comparison groups
# ============================================================

artist_release_comparison_df[
    "artist_release_comparison_group"
] = "No Independent Artist Baseline"

artist_release_comparison_df.loc[
    artist_baseline_evidence_mask,
    "artist_release_comparison_group"
] = "Broadly In Line"

artist_release_comparison_df.loc[
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] >= 75
    ),
    "artist_release_comparison_group"
] = "Above Artist Baseline"

artist_release_comparison_df.loc[
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] >= 90
    ),
    "artist_release_comparison_group"
] = "Strongly Above Artist Baseline"

artist_release_comparison_df.loc[
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] <= 25
    ),
    "artist_release_comparison_group"
] = "Below Artist Baseline"

artist_release_comparison_df.loc[
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] <= 10
    ),
    "artist_release_comparison_group"
] = "Strongly Below Artist Baseline"

# ============================================================
# Breakout and underperformance signals
# ============================================================

artist_release_comparison_df[
    "breakout_release_signal"
] = (
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_performance_percentile"
        ] >= 75
    )
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] >= 90
    )
)

artist_release_comparison_df[
    "underperforming_artist_release_signal"
] = (
    artist_baseline_evidence_mask
    & (
        artist_release_comparison_df[
            "release_performance_percentile"
        ] <= 25
    )
    & (
        artist_release_comparison_df[
            "release_vs_artist_percentile"
        ] <= 10
    )
)

# ============================================================
# Summary metrics
# ============================================================

artist_release_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                artist_release_comparison_df
            ),
        },
        {
            "Metric": "Releases with independent artist baseline",
            "Observed Value": int(
                artist_baseline_evidence_mask.sum()
            ),
        },
        {
            "Metric": "Releases without independent artist baseline",
            "Observed Value": int(
                (~artist_baseline_evidence_mask).sum()
            ),
        },
        {
            "Metric": "Strongly above artist baseline",
            "Observed Value": int(
                (
                    artist_release_comparison_df[
                        "artist_release_comparison_group"
                    ]
                    == "Strongly Above Artist Baseline"
                ).sum()
            ),
        },
        {
            "Metric": "Strongly below artist baseline",
            "Observed Value": int(
                (
                    artist_release_comparison_df[
                        "artist_release_comparison_group"
                    ]
                    == "Strongly Below Artist Baseline"
                ).sum()
            ),
        },
        {
            "Metric": "Potential breakout releases",
            "Observed Value": int(
                artist_release_comparison_df[
                    "breakout_release_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Underperforming artist-release cases",
            "Observed Value": int(
                artist_release_comparison_df[
                    "underperforming_artist_release_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Median artist–release relationship score",
            "Observed Value": (
                artist_release_comparison_df[
                    "artist_release_relationship_score"
                ].median()
            ),
        },
    ]
)

print("Assessing PMIP artist–release comparisons")
print("=" * 80)

display(
    artist_release_summary_df
)

# ============================================================
# Comparison-group distribution
# ============================================================

artist_release_group_distribution_df = (
    artist_release_comparison_df[
        "artist_release_comparison_group"
    ]
    .value_counts()
    .rename_axis(
        "Artist–Release Comparison Group"
    )
    .reset_index(
        name="Releases"
    )
)

artist_release_group_distribution_df[
    "Share (%)"
] = (
    artist_release_group_distribution_df[
        "Releases"
    ]
    / len(
        artist_release_comparison_df
    )
    * 100
)

print()
print("Artist–release comparison distribution")
print("=" * 80)

display(
    artist_release_group_distribution_df
)

# ============================================================
# Display fields
# ============================================================

artist_release_display_fields = [
    "release_id",
    "Track",
    "Artist",
    "release_vs_artist_baseline_score",
    "release_vs_artist_percentile",
]

artist_release_display_fields += [
    field
    for field in available_artist_release_context_fields
    if field not in artist_release_display_fields
]

artist_release_display_fields += [
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "artist_release_relationship_score",
    "artist_release_comparison_group",
    "composite_weight_coverage_pct",
]

# ============================================================
# Strongest release-vs-artist outperformers
# ============================================================

strongest_artist_outperformers_df = (
    artist_release_comparison_df.loc[
        artist_baseline_evidence_mask,
        artist_release_display_fields
    ]
    .sort_values(
        by=[
            "release_vs_artist_percentile",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases strongest relative to artist baseline")
print("=" * 80)

display(
    strongest_artist_outperformers_df
)

# ============================================================
# Potential breakout releases
# ============================================================

breakout_releases_df = (
    artist_release_comparison_df.loc[
        artist_release_comparison_df[
            "breakout_release_signal"
        ],
        artist_release_display_fields
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "release_vs_artist_percentile",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Potential breakout release cases")
print("=" * 80)

display(
    breakout_releases_df
)

# ============================================================
# Strongest underperformance relative to artist baseline
# ============================================================

weakest_artist_relative_releases_df = (
    artist_release_comparison_df.loc[
        artist_baseline_evidence_mask,
        artist_release_display_fields
    ]
    .sort_values(
        by=[
            "release_vs_artist_percentile",
            "release_performance_percentile",
        ],
        ascending=[
            True,
            True,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Releases weakest relative to artist baseline")
print("=" * 80)

display(
    weakest_artist_relative_releases_df
)

# ============================================================
# Relationship-type distribution if available
# ============================================================

if (
    "artist_release_relationship_type"
    in artist_release_comparison_df.columns
):

    artist_release_relationship_type_df = (
        artist_release_comparison_df[
            "artist_release_relationship_type"
        ]
        .value_counts()
        .rename_axis(
            "Artist–Release Relationship Type"
        )
        .reset_index(
            name="Releases"
        )
    )

    artist_release_relationship_type_df[
        "Share (%)"
    ] = (
        artist_release_relationship_type_df[
            "Releases"
        ]
        / len(
            artist_release_comparison_df
        )
        * 100
    )

    print()
    print("Artist–release relationship type distribution")
    print("=" * 80)

    display(
        artist_release_relationship_type_df
    )

# ============================================================
# Validation
# ============================================================

artist_release_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 12.4 completed",
            "Passed": section_12_4_complete,
        },
        {
            "Validation Area": "Artist baseline scores available",
            "Passed": (
                "release_vs_artist_baseline_score"
                in artist_release_comparison_df.columns
            ),
        },
        {
            "Validation Area": "Independent artist baselines available",
            "Passed": bool(
                artist_baseline_evidence_mask.any()
            ),
        },
        {
            "Validation Area": "Artist-relative percentiles calculated",
            "Passed": (
                artist_release_comparison_df.loc[
                    artist_baseline_evidence_mask,
                    "release_vs_artist_percentile"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Artist-release comparison groups assigned",
            "Passed": artist_release_comparison_df[
                "artist_release_comparison_group"
            ].notna().all(),
        },
        {
            "Validation Area": "Breakout release cases assessed",
            "Passed": (
                "breakout_release_signal"
                in artist_release_comparison_df.columns
            ),
        },
        {
            "Validation Area": "Underperformance relative to artist assessed",
            "Passed": (
                "underperforming_artist_release_signal"
                in artist_release_comparison_df.columns
            ),
        },
    ]
)

print()
print("Artist–Release Comparisons validation")
print("=" * 80)

display(
    artist_release_validation_df
)

# ============================================================
# Completion
# ============================================================

section_12_5_complete = bool(
    artist_release_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 12.5 completion status: "
    f"{section_12_5_complete}"
)

if not section_12_5_complete:
    raise RuntimeError(
        "Artist–Release Comparisons did not "
        "complete successfully."
    )
```

### Interpretation

The artist–release comparison identified 3,256 releases with an independent artist baseline.

A total of 326 releases performed strongly above their artist baseline, while 325 performed strongly below it.

The analysis also identified 184 potential breakout releases and 254 underperforming artist–release cases, showing that individual release performance can differ substantially from an artist’s established level.

This comparison therefore provides useful context for identifying unusually strong releases, weaker-than-expected releases and possible breakout opportunities.

## 12.6 High-Priority Release Insights

### Purpose

This subsection brings together the main findings from Section 12 into a final evidence-aware interpretation of high-priority PMIP releases.

The analysis combines:

- overall release performance;
- evidence coverage;
- artist growth;
- geographic reach;
- artist-relative performance;
- and unusual-performance signals.

The aim is to identify the releases that provide the strongest combination of performance strength and contextual importance for further PMIP investigation.

```python
# ============================================================
# Section 12.6 - High-Priority Release Insights
# CORRECTED VERSION
# ============================================================

if not section_12_5_complete:
    raise RuntimeError(
        "Section 12.5 must be completed before assessing "
        "High-Priority Release Insights."
    )

# ============================================================
# Working source
#
# Start from the final Section 11 release-level dataset, then
# explicitly attach the contextual results created in
# Sections 12.3, 12.4 and 12.5.
# ============================================================

high_priority_insights_df = section_11_final_df.copy()

# ============================================================
# Attach Section 12.3 growth-context outputs
# ============================================================

growth_context_fields = [
    "release_id",
    "artist_growth_score",
    "artist_growth_percentile",
    "fast_growth_group",
    "fast_growth_release_flag",
    "growth_and_performance_signal",
]

available_growth_context_fields = [
    field
    for field in growth_context_fields
    if field in fast_growth_release_df.columns
]

growth_context_merge_df = (
    fast_growth_release_df[
        available_growth_context_fields
    ]
    .drop_duplicates(
        subset="release_id"
    )
)

high_priority_insights_df = (
    high_priority_insights_df
    .merge(
        growth_context_merge_df,
        on="release_id",
        how="left",
        suffixes=("", "_growth")
    )
)

# ============================================================
# Attach Section 12.4 geographic-context outputs
# ============================================================

geographic_context_fields = [
    "release_id",
    "geographic_reach_score",
    "geographic_reach_percentile",
    "geographic_reach_quartile",
    "broad_geographic_reach_flag",
    "strong_performance_broad_reach_flag",
    "strong_performance_narrow_reach_flag",
]

available_geographic_context_fields = [
    field
    for field in geographic_context_fields
    if field in geographic_findings_df.columns
]

geographic_context_merge_df = (
    geographic_findings_df[
        available_geographic_context_fields
    ]
    .drop_duplicates(
        subset="release_id"
    )
)

high_priority_insights_df = (
    high_priority_insights_df
    .merge(
        geographic_context_merge_df,
        on="release_id",
        how="left",
        suffixes=("", "_geo")
    )
)

# ============================================================
# Attach Section 12.5 artist-release outputs
# ============================================================

artist_context_fields = [
    "release_id",
    "release_vs_artist_baseline_score",
    "release_vs_artist_percentile",
    "artist_release_comparison_group",
    "breakout_release_signal",
    "underperforming_artist_release_signal",
]

available_artist_context_fields = [
    field
    for field in artist_context_fields
    if field in artist_release_comparison_df.columns
]

artist_context_merge_df = (
    artist_release_comparison_df[
        available_artist_context_fields
    ]
    .drop_duplicates(
        subset="release_id"
    )
)

high_priority_insights_df = (
    high_priority_insights_df
    .merge(
        artist_context_merge_df,
        on="release_id",
        how="left",
        suffixes=("", "_artist")
    )
)

# ============================================================
# Validate required priority fields
# ============================================================

required_priority_insight_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "release_performance_percentile",
    "release_performance_class",
    "composite_weight_coverage_pct",
    "high_priority_release_flag",
    "priority_review_flag",
    "release_priority_class",
    "priority_signal_score",
    "priority_signal_rank",
]

missing_priority_insight_fields = [
    field
    for field in required_priority_insight_fields
    if field not in high_priority_insights_df.columns
]

if missing_priority_insight_fields:
    raise RuntimeError(
        "Required high-priority insight fields are missing: "
        f"{missing_priority_insight_fields}"
    )

# ============================================================
# Required Section 12 contextual fields
# ============================================================

required_context_fields = [
    "growth_and_performance_signal",
    "strong_performance_broad_reach_flag",
    "breakout_release_signal",
    "underperforming_artist_release_signal",
]

missing_context_fields = [
    field
    for field in required_context_fields
    if field not in high_priority_insights_df.columns
]

if missing_context_fields:
    raise RuntimeError(
        "Required Section 12 contextual fields are missing: "
        f"{missing_context_fields}"
    )

# ============================================================
# Optional unusual-performance context
# ============================================================

if "unusual_performance_flag" not in high_priority_insights_df.columns:
    high_priority_insights_df[
        "unusual_performance_flag"
    ] = False

if "unusual_performance_flag_count" not in high_priority_insights_df.columns:
    high_priority_insights_df[
        "unusual_performance_flag_count"
    ] = 0

if "unusual_performance_pattern" not in high_priority_insights_df.columns:
    high_priority_insights_df[
        "unusual_performance_pattern"
    ] = "No Unusual Pattern"

# ============================================================
# Fill boolean contextual fields safely
# ============================================================

boolean_context_fields = [
    "growth_and_performance_signal",
    "strong_performance_broad_reach_flag",
    "strong_performance_narrow_reach_flag",
    "breakout_release_signal",
    "underperforming_artist_release_signal",
    "unusual_performance_flag",
]

for field in boolean_context_fields:
    if field in high_priority_insights_df.columns:
        high_priority_insights_df[field] = (
            high_priority_insights_df[field]
            .fillna(False)
            .astype(bool)
        )

# ============================================================
# Combined contextual signal count
# ============================================================

context_signal_fields = [
    "growth_and_performance_signal",
    "strong_performance_broad_reach_flag",
    "breakout_release_signal",
    "unusual_performance_flag",
]

high_priority_insights_df[
    "contextual_signal_count"
] = (
    high_priority_insights_df[
        context_signal_fields
    ]
    .astype(int)
    .sum(axis=1)
)

# ============================================================
# Final insight category
# ============================================================

high_priority_insights_df[
    "final_release_insight"
] = "Standard Release"

high_priority_insights_df.loc[
    high_priority_insights_df[
        "high_priority_release_flag"
    ],
    "final_release_insight"
] = "High-Priority Release"

high_priority_insights_df.loc[
    high_priority_insights_df[
        "high_priority_release_flag"
    ]
    & (
        high_priority_insights_df[
            "contextual_signal_count"
        ] >= 1
    ),
    "final_release_insight"
] = "High-Priority + Contextual Signal"

high_priority_insights_df.loc[
    high_priority_insights_df[
        "high_priority_release_flag"
    ]
    & (
        high_priority_insights_df[
            "contextual_signal_count"
        ] >= 2
    ),
    "final_release_insight"
] = "Multi-Signal High-Priority Release"

high_priority_insights_df.loc[
    high_priority_insights_df[
        "breakout_release_signal"
    ]
    & (
        high_priority_insights_df[
            "release_performance_percentile"
        ] >= 75
    ),
    "final_release_insight"
] = "Potential Breakout Priority"

high_priority_insights_df.loc[
    high_priority_insights_df[
        "underperforming_artist_release_signal"
    ],
    "final_release_insight"
] = "Underperformance Review"

high_priority_insights_df.loc[
    (
        high_priority_insights_df[
            "release_performance_percentile"
        ] >= 90
    )
    & (
        high_priority_insights_df[
            "composite_weight_coverage_pct"
        ] < 70
    ),
    "final_release_insight"
] = "High Performance - Evidence Caution"

# ============================================================
# Final insight score
# ============================================================

context_bonus = (
    high_priority_insights_df[
        "contextual_signal_count"
    ]
    .clip(
        lower=0,
        upper=4
    )
    / 4
    * 10
)

high_priority_insights_df[
    "final_insight_score"
] = (
    high_priority_insights_df[
        "priority_signal_score"
    ] * 0.90
    + context_bonus
)

high_priority_insights_df[
    "final_insight_rank"
] = (
    high_priority_insights_df[
        "final_insight_score"
    ]
    .rank(
        method="min",
        ascending=False
    )
)

# ============================================================
# Section 12 final summary
# ============================================================

section_12_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                high_priority_insights_df
            ),
        },
        {
            "Metric": "Exceptional-performance releases",
            "Observed Value": int(
                (
                    high_priority_insights_df[
                        "release_performance_class"
                    ]
                    == "Exceptional Release Performance"
                ).sum()
            ),
        },
        {
            "Metric": "High-priority releases",
            "Observed Value": int(
                high_priority_insights_df[
                    "high_priority_release_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Priority-review releases",
            "Observed Value": int(
                high_priority_insights_df[
                    "priority_review_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Fast-growth + strong-performance releases",
            "Observed Value": int(
                high_priority_insights_df[
                    "growth_and_performance_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Strong-performance + broad-reach releases",
            "Observed Value": int(
                high_priority_insights_df[
                    "strong_performance_broad_reach_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Potential breakout releases",
            "Observed Value": int(
                high_priority_insights_df[
                    "breakout_release_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Underperforming artist-release cases",
            "Observed Value": int(
                high_priority_insights_df[
                    "underperforming_artist_release_signal"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with unusual-performance signal",
            "Observed Value": int(
                high_priority_insights_df[
                    "unusual_performance_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Multi-signal contextual releases",
            "Observed Value": int(
                (
                    high_priority_insights_df[
                        "contextual_signal_count"
                    ] >= 2
                ).sum()
            ),
        },
    ]
)

print("Assessing PMIP high-priority release insights")
print("=" * 80)

display(
    section_12_summary_df
)

# ============================================================
# Final insight distribution
# ============================================================

final_insight_distribution_df = (
    high_priority_insights_df[
        "final_release_insight"
    ]
    .value_counts()
    .rename_axis(
        "Final Release Insight"
    )
    .reset_index(
        name="Releases"
    )
)

final_insight_distribution_df[
    "Share (%)"
] = (
    final_insight_distribution_df[
        "Releases"
    ]
    / len(
        high_priority_insights_df
    )
    * 100
)

print()
print("Final release insight distribution")
print("=" * 80)

display(
    final_insight_distribution_df
)

# ============================================================
# Consistency check against earlier Section 12 findings
# ============================================================

section_12_consistency_df = pd.DataFrame(
    [
        {
            "Finding":
                "Fast-growth + strong-performance",
            "Expected From Earlier Section":
                int(
                    fast_growth_release_df[
                        "growth_and_performance_signal"
                    ].sum()
                ),
            "Final Section 12.6 Count":
                int(
                    high_priority_insights_df[
                        "growth_and_performance_signal"
                    ].sum()
                ),
        },
        {
            "Finding":
                "Strong-performance + broad-reach",
            "Expected From Earlier Section":
                int(
                    geographic_findings_df[
                        "strong_performance_broad_reach_flag"
                    ].sum()
                ),
            "Final Section 12.6 Count":
                int(
                    high_priority_insights_df[
                        "strong_performance_broad_reach_flag"
                    ].sum()
                ),
        },
        {
            "Finding":
                "Potential breakout releases",
            "Expected From Earlier Section":
                int(
                    artist_release_comparison_df[
                        "breakout_release_signal"
                    ].sum()
                ),
            "Final Section 12.6 Count":
                int(
                    high_priority_insights_df[
                        "breakout_release_signal"
                    ].sum()
                ),
        },
        {
            "Finding":
                "Underperforming artist-release cases",
            "Expected From Earlier Section":
                int(
                    artist_release_comparison_df[
                        "underperforming_artist_release_signal"
                    ].sum()
                ),
            "Final Section 12.6 Count":
                int(
                    high_priority_insights_df[
                        "underperforming_artist_release_signal"
                    ].sum()
                ),
        },
    ]
)

section_12_consistency_df[
    "Passed"
] = (
    section_12_consistency_df[
        "Expected From Earlier Section"
    ]
    == section_12_consistency_df[
        "Final Section 12.6 Count"
    ]
)

print()
print("Section 12 contextual consistency check")
print("=" * 80)

display(
    section_12_consistency_df
)

# ============================================================
# Validation
# ============================================================

section_12_6_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 12.5 completed",
            "Passed":
                section_12_5_complete,
        },
        {
            "Validation Area":
                "Priority framework available",
            "Passed":
                high_priority_insights_df[
                    "priority_signal_score"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Growth context successfully attached",
            "Passed":
                (
                    high_priority_insights_df[
                        "growth_and_performance_signal"
                    ].sum()
                    ==
                    fast_growth_release_df[
                        "growth_and_performance_signal"
                    ].sum()
                ),
        },
        {
            "Validation Area":
                "Geographic context successfully attached",
            "Passed":
                (
                    high_priority_insights_df[
                        "strong_performance_broad_reach_flag"
                    ].sum()
                    ==
                    geographic_findings_df[
                        "strong_performance_broad_reach_flag"
                    ].sum()
                ),
        },
        {
            "Validation Area":
                "Artist-release context successfully attached",
            "Passed":
                (
                    high_priority_insights_df[
                        "breakout_release_signal"
                    ].sum()
                    ==
                    artist_release_comparison_df[
                        "breakout_release_signal"
                    ].sum()
                ),
        },
        {
            "Validation Area":
                "Contextual signals combined",
            "Passed":
                high_priority_insights_df[
                    "contextual_signal_count"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Final insight categories assigned",
            "Passed":
                high_priority_insights_df[
                    "final_release_insight"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Final insight scores calculated",
            "Passed":
                high_priority_insights_df[
                    "final_insight_score"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Contextual consistency checks passed",
            "Passed":
                section_12_consistency_df[
                    "Passed"
                ].all(),
        },
    ]
)

print()
print("High-Priority Release Insights validation")
print("=" * 80)

display(
    section_12_6_validation_df
)

# ============================================================
# Completion
# ============================================================

section_12_6_complete = bool(
    section_12_6_validation_df[
        "Passed"
    ].all()
)

section_12_complete = bool(
    section_12_1_complete
    and section_12_2_complete
    and section_12_3_complete
    and section_12_4_complete
    and section_12_5_complete
    and section_12_6_complete
)

print()
print(
    f"Section 12.6 completion status: "
    f"{section_12_6_complete}"
)

print(
    f"Section 12 overall completion status: "
    f"{section_12_complete}"
)

if not section_12_complete:
    raise RuntimeError(
        "Section 12 Results and Interpretation did not "
        "complete successfully."
    )
```

### Section 12 Interpretation

Section 12 translated the PMIP scoring framework into practical release-level findings across performance, growth, geography, artist context and priority signals.

The analysis identified 460 exceptional-performance releases, alongside 852 high-priority releases and 972 releases requiring further review. Strong contextual patterns were also identified, including 118 releases combining fast artist growth with strong release performance, 523 combining strong performance with broad geographic reach, and 184 potential breakout releases.

Artist-relative analysis showed that release success did not always follow an artist's established baseline, with 254 releases identified as underperforming relative to their artist context. In addition, 415 releases contained at least one unusual-performance signal, while 348 releases showed multiple contextual signals.

The results also demonstrated the importance of evidence quality when interpreting performance. Some high-performing releases required caution because their rankings were supported by lower evidence coverage, while releases with stronger multidimensional evidence provided more reliable performance conclusions.

Overall, the findings show that PMIP can move beyond simple ranking by combining release performance with growth, geographic reach, artist context and unusual behaviour. This creates a more evidence-aware framework for identifying strong releases, potential breakout opportunities, underperformance cases and releases requiring priority investigation.

## 13. Responsible Use and Limitations

### Purpose

This section documents the main limitations of the PMIP analytical framework and explains how its outputs should be used responsibly.

Although the system provides structured release-level intelligence, its scores, classifications and priority signals are influenced by the quality, coverage and timing of the available data.

The purpose of this section is therefore to identify where uncertainty may arise and to make clear that PMIP outputs should support, rather than replace, human judgement.

The analysis considers:

- temporal limitations;
- data-coverage limitations;
- release-identity limitations;
- artist-comparison limitations;
- and the requirement for human review.


### 13.1 Temporal Limitations

### Purpose

This subsection examines limitations caused by differences in release dates, observation periods and historical data availability.

The PMIP dataset contains releases from different time periods, meaning that some releases have had much longer to accumulate chart, streaming and geographic evidence than others.

The aim is to identify where time-related differences may affect direct comparison between releases and to document how temporal context should be considered when interpreting PMIP results.

```python
# ============================================================
# Section 13.1 - Temporal Limitations
# ============================================================

if not section_12_complete:
    raise RuntimeError(
        "Section 12 must be completed before assessing "
        "Responsible Use and Limitations."
    )

# ============================================================
# Working source
# ============================================================

temporal_limitations_df = high_priority_insights_df.copy()

# ============================================================
# Validate required temporal fields
# ============================================================

required_temporal_limitation_fields = [
    "release_id",
    "Track",
    "Artist",
    "release_date",
    "release_age_days",
    "release_age_years",
    "temporal_comparability_score",
    "release_performance_percentile",
    "composite_release_performance_score",
    "composite_weight_coverage_pct",
]

missing_temporal_limitation_fields = [
    field
    for field in required_temporal_limitation_fields
    if field not in temporal_limitations_df.columns
]

if missing_temporal_limitation_fields:
    raise RuntimeError(
        "Required temporal-limitation fields are missing: "
        f"{missing_temporal_limitation_fields}"
    )

# ============================================================
# Ensure release date is datetime
# ============================================================

temporal_limitations_df[
    "release_date"
] = pd.to_datetime(
    temporal_limitations_df[
        "release_date"
    ],
    errors="coerce"
)

# ============================================================
# Temporal evidence groups
# ============================================================

temporal_limitations_df[
    "temporal_limitation_group"
] = "Moderate Temporal Context"

temporal_limitations_df.loc[
    temporal_limitations_df[
        "temporal_comparability_score"
    ] < 25,
    "temporal_limitation_group"
] = "High Temporal Limitation"

temporal_limitations_df.loc[
    (
        temporal_limitations_df[
            "temporal_comparability_score"
        ] >= 25
    )
    & (
        temporal_limitations_df[
            "temporal_comparability_score"
        ] < 50
    ),
    "temporal_limitation_group"
] = "Moderate Temporal Limitation"

temporal_limitations_df.loc[
    (
        temporal_limitations_df[
            "temporal_comparability_score"
        ] >= 50
    )
    & (
        temporal_limitations_df[
            "temporal_comparability_score"
        ] < 75
    ),
    "temporal_limitation_group"
] = "Good Temporal Comparability"

temporal_limitations_df.loc[
    temporal_limitations_df[
        "temporal_comparability_score"
    ] >= 75,
    "temporal_limitation_group"
] = "Strong Temporal Comparability"

# ============================================================
# Recent-release and older-release flags
# ============================================================

temporal_limitations_df[
    "recent_release_flag"
] = (
    temporal_limitations_df[
        "release_age_years"
    ] <= 1
)

temporal_limitations_df[
    "older_release_flag"
] = (
    temporal_limitations_df[
        "release_age_years"
    ] >= 10
)

# ============================================================
# High-performance releases with weak temporal comparability
# ============================================================

temporal_limitations_df[
    "high_performance_temporal_caution"
] = (
    (
        temporal_limitations_df[
            "release_performance_percentile"
        ] >= 90
    )
    & (
        temporal_limitations_df[
            "temporal_comparability_score"
        ] < 50
    )
)

# ============================================================
# Temporal limitation summary
# ============================================================

temporal_limitation_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                temporal_limitations_df
            ),
        },
        {
            "Metric": "Earliest release date",
            "Observed Value": (
                temporal_limitations_df[
                    "release_date"
                ].min()
            ),
        },
        {
            "Metric": "Latest release date",
            "Observed Value": (
                temporal_limitations_df[
                    "release_date"
                ].max()
            ),
        },
        {
            "Metric": "Median release age (years)",
            "Observed Value": (
                temporal_limitations_df[
                    "release_age_years"
                ].median()
            ),
        },
        {
            "Metric": "Recent releases (<=1 year)",
            "Observed Value": int(
                temporal_limitations_df[
                    "recent_release_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Older releases (>=10 years)",
            "Observed Value": int(
                temporal_limitations_df[
                    "older_release_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Median temporal-comparability score",
            "Observed Value": (
                temporal_limitations_df[
                    "temporal_comparability_score"
                ].median()
            ),
        },
        {
            "Metric": "High-performance releases with temporal caution",
            "Observed Value": int(
                temporal_limitations_df[
                    "high_performance_temporal_caution"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP temporal limitations")
print("=" * 80)

display(
    temporal_limitation_summary_df
)

# ============================================================
# Temporal limitation distribution
# ============================================================

temporal_limitation_distribution_df = (
    temporal_limitations_df[
        "temporal_limitation_group"
    ]
    .value_counts()
    .rename_axis(
        "Temporal Limitation Group"
    )
    .reset_index(
        name="Releases"
    )
)

temporal_limitation_distribution_df[
    "Share (%)"
] = (
    temporal_limitation_distribution_df[
        "Releases"
    ]
    / len(
        temporal_limitations_df
    )
    * 100
)

print()
print("Temporal limitation distribution")
print("=" * 80)

display(
    temporal_limitation_distribution_df
)

# ============================================================
# Releases requiring temporal caution
# ============================================================

temporal_caution_releases_df = (
    temporal_limitations_df.loc[
        temporal_limitations_df[
            "high_performance_temporal_caution"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "release_date",
            "release_age_years",
            "temporal_comparability_score",
            "release_performance_percentile",
            "composite_release_performance_score",
            "composite_weight_coverage_pct",
            "temporal_limitation_group",
        ]
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "temporal_comparability_score",
        ],
        ascending=[
            False,
            True,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("High-performing releases requiring temporal caution")
print("=" * 80)

display(
    temporal_caution_releases_df
)

# ============================================================
# Validation
# ============================================================

temporal_limitations_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 12 completed",
            "Passed": section_12_complete,
        },
        {
            "Validation Area": "Release dates available",
            "Passed": temporal_limitations_df[
                "release_date"
            ].notna().all(),
        },
        {
            "Validation Area": "Release age available",
            "Passed": temporal_limitations_df[
                "release_age_years"
            ].notna().all(),
        },
        {
            "Validation Area": "Temporal comparability available",
            "Passed": temporal_limitations_df[
                "temporal_comparability_score"
            ].notna().all(),
        },
        {
            "Validation Area": "Temporal limitation groups assigned",
            "Passed": temporal_limitations_df[
                "temporal_limitation_group"
            ].notna().all(),
        },
        {
            "Validation Area": "Temporal caution cases assessed",
            "Passed": (
                "high_performance_temporal_caution"
                in temporal_limitations_df.columns
            ),
        },
    ]
)

print()
print("Temporal Limitations validation")
print("=" * 80)

display(
    temporal_limitations_validation_df
)

# ============================================================
# Completion
# ============================================================

section_13_1_complete = bool(
    temporal_limitations_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 13.1 completion status: "
    f"{section_13_1_complete}"
)

if not section_13_1_complete:
    raise RuntimeError(
        "Temporal Limitations did not complete successfully."
    )
```

### Interpretation

The dataset contains releases from substantially different time periods, with release dates ranging from 1987 to 2024 and a median release age of approximately 2.04 years.

Although most releases had acceptable temporal comparability, around 43% showed moderate or high temporal limitations, meaning that some releases were observed under less comparable historical conditions.

Only four high-performing releases required explicit temporal caution, suggesting that the strongest PMIP results were generally not heavily affected by temporal inconsistency.

Temporal context should therefore be treated as an important limitation when comparing releases from substantially different periods or observation histories.

## 13.2 Data-Coverage Limitations

### Purpose

This subsection examines limitations caused by incomplete availability of the indicators used throughout the PMIP framework.

Not every release contains the same level of streaming, chart, geographic, temporal, artist-context or unusual-performance evidence.

The aim is to quantify these coverage differences and identify where limited data availability may reduce confidence in release comparisons, rankings and priority signals.

This helps ensure that PMIP results are interpreted according to both performance strength and the amount of evidence supporting each result.

```python
# ============================================================
# Section 13.2 - Data-Coverage Limitations
# ============================================================

if not section_13_1_complete:
    raise RuntimeError(
        "Section 13.1 must be completed before assessing "
        "Data-Coverage Limitations."
    )

# ============================================================
# Working source
# ============================================================

data_coverage_limitations_df = high_priority_insights_df.copy()

# ============================================================
# Core PMIP indicator fields
# ============================================================

coverage_indicator_fields = {
    "Streaming Performance": "streaming_performance_score",
    "Chart Performance": "chart_performance_score",
    "Geographic Reach": "geographic_reach_score",
    "Temporal Comparability": "temporal_comparability_score",
    "Artist–Release Relationship": "artist_release_relationship_score",
    "Unusual Performance": "unusual_performance_score",
}

# ============================================================
# Validate that the expected indicators exist
# ============================================================

missing_coverage_fields = [
    field
    for field in coverage_indicator_fields.values()
    if field not in data_coverage_limitations_df.columns
]

if missing_coverage_fields:
    raise RuntimeError(
        "Required data-coverage indicator fields are missing: "
        f"{missing_coverage_fields}"
    )

# ============================================================
# Indicator-level coverage summary
# ============================================================

coverage_rows = []

for indicator_name, field_name in coverage_indicator_fields.items():

    available_values = int(
        data_coverage_limitations_df[
            field_name
        ].notna().sum()
    )

    missing_values = int(
        data_coverage_limitations_df[
            field_name
        ].isna().sum()
    )

    coverage_pct = (
        available_values
        / len(data_coverage_limitations_df)
        * 100
    )

    coverage_rows.append(
        {
            "Indicator": indicator_name,
            "Field": field_name,
            "Available Values": available_values,
            "Missing Values": missing_values,
            "Coverage (%)": coverage_pct,
        }
    )

indicator_coverage_summary_df = pd.DataFrame(
    coverage_rows
)

print("Assessing PMIP data-coverage limitations")
print("=" * 80)

display(
    indicator_coverage_summary_df
)

# ============================================================
# Number of indicators available per release
# ============================================================

coverage_field_list = list(
    coverage_indicator_fields.values()
)

data_coverage_limitations_df[
    "indicators_available_count"
] = (
    data_coverage_limitations_df[
        coverage_field_list
    ]
    .notna()
    .sum(axis=1)
)

data_coverage_limitations_df[
    "indicator_coverage_pct"
] = (
    data_coverage_limitations_df[
        "indicators_available_count"
    ]
    / len(coverage_field_list)
    * 100
)

# ============================================================
# Data-coverage limitation groups
# ============================================================

data_coverage_limitations_df[
    "data_coverage_limitation_group"
] = "Strong Data Coverage"

data_coverage_limitations_df.loc[
    data_coverage_limitations_df[
        "indicator_coverage_pct"
    ] < 50,
    "data_coverage_limitation_group"
] = "High Data-Coverage Limitation"

data_coverage_limitations_df.loc[
    (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] >= 50
    )
    & (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] < 70
    ),
    "data_coverage_limitation_group"
] = "Moderate Data-Coverage Limitation"

data_coverage_limitations_df.loc[
    (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] >= 70
    )
    & (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] < 100
    ),
    "data_coverage_limitation_group"
] = "Good Data Coverage"

data_coverage_limitations_df.loc[
    data_coverage_limitations_df[
        "indicator_coverage_pct"
    ] == 100,
    "data_coverage_limitation_group"
] = "Complete Data Coverage"

# ============================================================
# Performance-result caution due to limited coverage
# ============================================================

data_coverage_limitations_df[
    "high_performance_coverage_caution"
] = (
    (
        data_coverage_limitations_df[
            "release_performance_percentile"
        ] >= 90
    )
    & (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] < 70
    )
)

data_coverage_limitations_df[
    "priority_coverage_caution"
] = (
    data_coverage_limitations_df[
        "priority_review_flag"
    ]
    & (
        data_coverage_limitations_df[
            "indicator_coverage_pct"
        ] < 70
    )
)

# ============================================================
# Coverage limitation summary
# ============================================================

data_coverage_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                data_coverage_limitations_df
            ),
        },
        {
            "Metric": "Indicators assessed",
            "Observed Value": len(
                coverage_field_list
            ),
        },
        {
            "Metric": "Median indicators available per release",
            "Observed Value": (
                data_coverage_limitations_df[
                    "indicators_available_count"
                ].median()
            ),
        },
        {
            "Metric": "Median indicator coverage (%)",
            "Observed Value": (
                data_coverage_limitations_df[
                    "indicator_coverage_pct"
                ].median()
            ),
        },
        {
            "Metric": "Releases with complete indicator coverage",
            "Observed Value": int(
                (
                    data_coverage_limitations_df[
                        "indicator_coverage_pct"
                    ] == 100
                ).sum()
            ),
        },
        {
            "Metric": "Releases with <70% indicator coverage",
            "Observed Value": int(
                (
                    data_coverage_limitations_df[
                        "indicator_coverage_pct"
                    ] < 70
                ).sum()
            ),
        },
        {
            "Metric": "Releases with <50% indicator coverage",
            "Observed Value": int(
                (
                    data_coverage_limitations_df[
                        "indicator_coverage_pct"
                    ] < 50
                ).sum()
            ),
        },
        {
            "Metric": "High-performance releases requiring coverage caution",
            "Observed Value": int(
                data_coverage_limitations_df[
                    "high_performance_coverage_caution"
                ].sum()
            ),
        },
        {
            "Metric": "Priority-review releases with limited coverage",
            "Observed Value": int(
                data_coverage_limitations_df[
                    "priority_coverage_caution"
                ].sum()
            ),
        },
    ]
)

print()
print("Data-coverage limitation summary")
print("=" * 80)

display(
    data_coverage_summary_df
)

# ============================================================
# Coverage-limitation distribution
# ============================================================

data_coverage_distribution_df = (
    data_coverage_limitations_df[
        "data_coverage_limitation_group"
    ]
    .value_counts()
    .rename_axis(
        "Data Coverage Group"
    )
    .reset_index(
        name="Releases"
    )
)

data_coverage_distribution_df[
    "Share (%)"
] = (
    data_coverage_distribution_df[
        "Releases"
    ]
    / len(
        data_coverage_limitations_df
    )
    * 100
)

print()
print("Data-coverage limitation distribution")
print("=" * 80)

display(
    data_coverage_distribution_df
)

# ============================================================
# Indicator-availability distribution
# ============================================================

indicator_count_distribution_df = (
    data_coverage_limitations_df[
        "indicators_available_count"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Indicators Available"
    )
    .reset_index(
        name="Releases"
    )
)

indicator_count_distribution_df[
    "Share (%)"
] = (
    indicator_count_distribution_df[
        "Releases"
    ]
    / len(
        data_coverage_limitations_df
    )
    * 100
)

print()
print("Indicator availability per release")
print("=" * 80)

display(
    indicator_count_distribution_df
)

# ============================================================
# High-performing releases requiring coverage caution
# ============================================================

coverage_caution_releases_df = (
    data_coverage_limitations_df.loc[
        data_coverage_limitations_df[
            "high_performance_coverage_caution"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_percentile",
            "indicators_available_count",
            "indicator_coverage_pct",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "data_coverage_limitation_group",
        ]
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "indicator_coverage_pct",
        ],
        ascending=[
            False,
            True,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("High-performing releases requiring data-coverage caution")
print("=" * 80)

display(
    coverage_caution_releases_df
)

# ============================================================
# Validation
# ============================================================

data_coverage_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 13.1 completed",
            "Passed": section_13_1_complete,
        },
        {
            "Validation Area": "Core PMIP indicators available",
            "Passed": (
                len(missing_coverage_fields) == 0
            ),
        },
        {
            "Validation Area": "Indicator coverage calculated",
            "Passed": data_coverage_limitations_df[
                "indicator_coverage_pct"
            ].notna().all(),
        },
        {
            "Validation Area": "Coverage groups assigned",
            "Passed": data_coverage_limitations_df[
                "data_coverage_limitation_group"
            ].notna().all(),
        },
        {
            "Validation Area": "High-performance coverage caution assessed",
            "Passed": (
                "high_performance_coverage_caution"
                in data_coverage_limitations_df.columns
            ),
        },
        {
            "Validation Area": "Priority coverage caution assessed",
            "Passed": (
                "priority_coverage_caution"
                in data_coverage_limitations_df.columns
            ),
        },
    ]
)

print()
print("Data-Coverage Limitations validation")
print("=" * 80)

display(
    data_coverage_validation_df
)

# ============================================================
# Completion
# ============================================================

section_13_2_complete = bool(
    data_coverage_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 13.2 completion status: "
    f"{section_13_2_complete}"
)

if not section_13_2_complete:
    raise RuntimeError(
        "Data-Coverage Limitations did not "
        "complete successfully."
    )
```

### Interpretation

PMIP contains uneven data coverage across its analytical dimensions.

Streaming, temporal and artist-context indicators have high coverage, while chart, geographic and unusual-performance indicators are available for only around half of the releases.

Approximately 48.7% of releases have complete coverage across all six indicators, while a similar proportion have only three indicators available.

A total of 120 high-performing releases and 120 priority-review releases had limited coverage, showing that some apparently strong results are supported by less complete evidence.

PMIP scores should therefore be interpreted alongside evidence coverage, particularly when comparing releases with different levels of available data.

## 13.3 Release Identity Limitations

### Purpose

This subsection examines limitations related to how releases are identified within the PMIP dataset.

Release-level analysis depends on identifiers, track titles and artist names being sufficiently consistent across the integrated data sources.

However, the same recording may appear under different identifiers, remastered versions, remixes, re-releases, collaborations or slightly different text labels.

The aim is therefore to identify possible duplicate or ambiguous release identities and to show where these issues may affect release-level comparisons, counts and rankings.

```python
# ============================================================
# Section 13.3 - Release Identity Limitations
# ============================================================

if not section_13_2_complete:
    raise RuntimeError(
        "Section 13.2 must be completed before assessing "
        "Release Identity Limitations."
    )

# ============================================================
# Working source
# ============================================================

release_identity_limitations_df = high_priority_insights_df.copy()

# ============================================================
# Validate core identity fields
# ============================================================

required_identity_fields = [
    "release_id",
    "Track",
    "Artist",
]

missing_identity_fields = [
    field
    for field in required_identity_fields
    if field not in release_identity_limitations_df.columns
]

if missing_identity_fields:
    raise RuntimeError(
        "Required release-identity fields are missing: "
        f"{missing_identity_fields}"
    )

# ============================================================
# Prepare normalised identity fields
# ============================================================

release_identity_limitations_df[
    "track_normalised"
] = (
    release_identity_limitations_df[
        "Track"
    ]
    .astype(str)
    .str.lower()
    .str.strip()
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)

release_identity_limitations_df[
    "artist_normalised"
] = (
    release_identity_limitations_df[
        "Artist"
    ]
    .astype(str)
    .str.lower()
    .str.strip()
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)

# ============================================================
# Exact duplicate release IDs
# ============================================================

release_id_counts = (
    release_identity_limitations_df[
        "release_id"
    ]
    .value_counts()
)

duplicate_release_ids = (
    release_id_counts[
        release_id_counts > 1
    ]
)

duplicate_release_id_count = int(
    len(
        duplicate_release_ids
    )
)

duplicate_release_rows = int(
    release_identity_limitations_df[
        "release_id"
    ]
    .isin(
        duplicate_release_ids.index
    )
    .sum()
)

# ============================================================
# Same track + artist attached to multiple release IDs
# ============================================================

track_artist_id_counts = (
    release_identity_limitations_df
    .groupby(
        [
            "track_normalised",
            "artist_normalised",
        ]
    )[
        "release_id"
    ]
    .nunique()
    .reset_index(
        name="unique_release_ids"
    )
)

ambiguous_track_artist_groups_df = (
    track_artist_id_counts.loc[
        track_artist_id_counts[
            "unique_release_ids"
        ] > 1
    ]
    .sort_values(
        by="unique_release_ids",
        ascending=False
    )
    .reset_index(drop=True)
)

ambiguous_track_artist_groups = int(
    len(
        ambiguous_track_artist_groups_df
    )
)

# ============================================================
# Same release ID attached to multiple track/artist labels
# ============================================================

release_id_label_counts = (
    release_identity_limitations_df
    .groupby(
        "release_id"
    )
    .agg(
        Unique_Tracks=(
            "track_normalised",
            "nunique"
        ),
        Unique_Artists=(
            "artist_normalised",
            "nunique"
        ),
    )
    .reset_index()
)

inconsistent_release_id_labels_df = (
    release_id_label_counts.loc[
        (
            release_id_label_counts[
                "Unique_Tracks"
            ] > 1
        )
        |
        (
            release_id_label_counts[
                "Unique_Artists"
            ] > 1
        )
    ]
    .reset_index(drop=True)
)

inconsistent_release_id_count = int(
    len(
        inconsistent_release_id_labels_df
    )
)

# ============================================================
# Basic remix / remaster / version markers
# ============================================================

version_pattern = (
    r"\b("
    r"remix|remaster|remastered|live|edit|version|mix|"
    r"acoustic|radio edit|deluxe|sped up|slowed|cover"
    r")\b"
)

release_identity_limitations_df[
    "version_marker_flag"
] = (
    release_identity_limitations_df[
        "track_normalised"
    ]
    .str.contains(
        version_pattern,
        regex=True,
        na=False
    )
)

version_marked_releases = int(
    release_identity_limitations_df[
        "version_marker_flag"
    ].sum()
)

# ============================================================
# Potential identity ambiguity flag
# ============================================================

ambiguous_identity_keys = set(
    zip(
        ambiguous_track_artist_groups_df[
            "track_normalised"
        ],
        ambiguous_track_artist_groups_df[
            "artist_normalised"
        ],
    )
)

release_identity_limitations_df[
    "track_artist_key"
] = list(
    zip(
        release_identity_limitations_df[
            "track_normalised"
        ],
        release_identity_limitations_df[
            "artist_normalised"
        ],
    )
)

release_identity_limitations_df[
    "identity_ambiguity_flag"
] = (
    release_identity_limitations_df[
        "track_artist_key"
    ]
    .isin(
        ambiguous_identity_keys
    )
    |
    release_identity_limitations_df[
        "version_marker_flag"
    ]
)

# ============================================================
# Release-identity summary
# ============================================================

release_identity_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                release_identity_limitations_df
            ),
        },
        {
            "Metric": "Unique release IDs",
            "Observed Value": (
                release_identity_limitations_df[
                    "release_id"
                ].nunique()
            ),
        },
        {
            "Metric": "Duplicate release IDs",
            "Observed Value": (
                duplicate_release_id_count
            ),
        },
        {
            "Metric": "Rows affected by duplicate release IDs",
            "Observed Value": (
                duplicate_release_rows
            ),
        },
        {
            "Metric": "Track-artist groups with multiple release IDs",
            "Observed Value": (
                ambiguous_track_artist_groups
            ),
        },
        {
            "Metric": "Release IDs with inconsistent labels",
            "Observed Value": (
                inconsistent_release_id_count
            ),
        },
        {
            "Metric": "Releases containing version/remix markers",
            "Observed Value": (
                version_marked_releases
            ),
        },
        {
            "Metric": "Releases flagged with possible identity ambiguity",
            "Observed Value": int(
                release_identity_limitations_df[
                    "identity_ambiguity_flag"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP release identity limitations")
print("=" * 80)

display(
    release_identity_summary_df
)

# ============================================================
# Examples of track/artist ambiguity
# ============================================================

ambiguous_identity_examples_df = (
    release_identity_limitations_df.loc[
        release_identity_limitations_df[
            "identity_ambiguity_flag"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "version_marker_flag",
        ]
    ]
    .sort_values(
        by=[
            "Artist",
            "Track",
        ]
    )
    .head(30)
    .reset_index(drop=True)
)

print()
print("Examples of potential release-identity ambiguity")
print("=" * 80)

display(
    ambiguous_identity_examples_df
)

# ============================================================
# Validation
# ============================================================

release_identity_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13.2 completed",
            "Passed":
                section_13_2_complete,
        },
        {
            "Validation Area":
                "Release IDs available",
            "Passed":
                release_identity_limitations_df[
                    "release_id"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Track labels available",
            "Passed":
                release_identity_limitations_df[
                    "Track"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Artist labels available",
            "Passed":
                release_identity_limitations_df[
                    "Artist"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Duplicate release IDs assessed",
            "Passed":
                True,
        },
        {
            "Validation Area":
                "Track-artist ambiguity assessed",
            "Passed":
                True,
        },
        {
            "Validation Area":
                "Version markers assessed",
            "Passed":
                "version_marker_flag"
                in release_identity_limitations_df.columns,
        },
        {
            "Validation Area":
                "Identity ambiguity flag assigned",
            "Passed":
                "identity_ambiguity_flag"
                in release_identity_limitations_df.columns,
        },
    ]
)

print()
print("Release Identity Limitations validation")
print("=" * 80)

display(
    release_identity_validation_df
)

# ============================================================
# Completion
# ============================================================

section_13_3_complete = bool(
    release_identity_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 13.3 completion status: "
    f"{section_13_3_complete}"
)

if not section_13_3_complete:
    raise RuntimeError(
        "Release Identity Limitations did not "
        "complete successfully."
    )
```

### Interpretation

The PMIP dataset contains unique release identifiers, with no duplicate release IDs or inconsistent ID-to-label mappings detected.

However, 98 track–artist combinations were associated with multiple release IDs, while 217 releases contained version or remix markers. In total, 408 releases were flagged with possible identity ambiguity.

This indicates that alternate versions, remixes, live recordings and similar release variants may represent the same underlying song differently across the dataset.

Release-level comparisons should therefore consider version and identity ambiguity, particularly where multiple records exist for the same track and artist.

## 13.4 Artist Comparison Limitations

### Purpose

This subsection examines limitations in comparing individual releases against artist-level baselines.

Artist comparisons depend on having enough historical releases to build a meaningful baseline. Artists with only one or a small number of releases provide weaker comparison evidence than artists with longer release histories.

The aim is to identify where limited artist history may reduce confidence in release-versus-artist comparisons, growth estimates and momentum signals.

This helps prevent PMIP from treating all artist baselines as equally reliable.

```python
# ============================================================
# Section 13.4 - Artist Comparison Limitations
# CORRECTED VERSION
# ============================================================

if not section_13_3_complete:
    raise RuntimeError(
        "Section 13.3 must be completed before assessing "
        "Artist Comparison Limitations."
    )

# ============================================================
# Working source
# ============================================================

artist_comparison_limitations_df = high_priority_insights_df.copy()

# ============================================================
# Validate required artist-context fields
# ============================================================

required_artist_comparison_fields = [
    "release_id",
    "Track",
    "Artist",
    "artist_performance_score",
    "release_vs_artist_baseline_score",
    "artist_growth_score",
    "artist_momentum_score",
]

missing_artist_comparison_fields = [
    field
    for field in required_artist_comparison_fields
    if field not in artist_comparison_limitations_df.columns
]

if missing_artist_comparison_fields:
    raise RuntimeError(
        "Required artist-comparison fields are missing: "
        f"{missing_artist_comparison_fields}"
    )

# ============================================================
# Artist release counts
#
# Calculate directly on the release-level dataframe.
# This avoids merge-related _x / _y column conflicts.
# ============================================================

artist_comparison_limitations_df[
    "artist_release_count"
] = (
    artist_comparison_limitations_df
    .groupby("Artist")[
        "release_id"
    ]
    .transform("nunique")
)

# Separate artist-level summary for later reporting
artist_release_counts_df = (
    artist_comparison_limitations_df[
        [
            "Artist",
            "artist_release_count",
        ]
    ]
    .drop_duplicates(
        subset="Artist"
    )
    .reset_index(drop=True)
)

# ============================================================
# Artist history groups
# ============================================================

artist_comparison_limitations_df[
    "artist_history_group"
] = "Limited Artist History"

artist_comparison_limitations_df.loc[
    artist_comparison_limitations_df[
        "artist_release_count"
    ] == 1,
    "artist_history_group"
] = "Single-Release Artist"

artist_comparison_limitations_df.loc[
    artist_comparison_limitations_df[
        "artist_release_count"
    ].between(
        2,
        3
    ),
    "artist_history_group"
] = "Limited Artist History"

artist_comparison_limitations_df.loc[
    artist_comparison_limitations_df[
        "artist_release_count"
    ].between(
        4,
        9
    ),
    "artist_history_group"
] = "Moderate Artist History"

artist_comparison_limitations_df.loc[
    artist_comparison_limitations_df[
        "artist_release_count"
    ] >= 10,
    "artist_history_group"
] = "Strong Artist History"

# ============================================================
# Independent artist baseline availability
# ============================================================

artist_comparison_limitations_df[
    "independent_baseline_available"
] = (
    artist_comparison_limitations_df[
        "release_vs_artist_baseline_score"
    ].notna()
)

# ============================================================
# Growth-context availability
# ============================================================

artist_comparison_limitations_df[
    "growth_context_available"
] = (
    artist_comparison_limitations_df[
        "artist_growth_score"
    ].notna()
)

# ============================================================
# Momentum-context availability
# ============================================================

artist_comparison_limitations_df[
    "momentum_context_available"
] = (
    artist_comparison_limitations_df[
        "artist_momentum_score"
    ].notna()
)

# ============================================================
# Artist-comparison caution
#
# Comparisons are treated cautiously when:
#
# 1. the artist has fewer than four observed releases, OR
# 2. an independent artist baseline is unavailable.
# ============================================================

artist_comparison_limitations_df[
    "artist_comparison_caution_flag"
] = (
    (
        artist_comparison_limitations_df[
            "artist_release_count"
        ] < 4
    )
    |
    (
        ~artist_comparison_limitations_df[
            "independent_baseline_available"
        ]
    )
)

# ============================================================
# High-performing releases requiring artist-context caution
# ============================================================

artist_comparison_limitations_df[
    "high_performance_artist_context_caution"
] = (
    (
        artist_comparison_limitations_df[
            "release_performance_percentile"
        ] >= 90
    )
    & (
        artist_comparison_limitations_df[
            "artist_comparison_caution_flag"
        ]
    )
)

# ============================================================
# Artist-level baseline availability
# ============================================================

artist_baseline_by_artist_df = (
    artist_comparison_limitations_df
    .groupby(
        "Artist",
        as_index=False
    )
    .agg(
        artist_release_count=(
            "release_id",
            "nunique"
        ),
        releases_with_baseline=(
            "independent_baseline_available",
            "sum"
        ),
        releases_with_growth_context=(
            "growth_context_available",
            "sum"
        ),
        releases_with_momentum_context=(
            "momentum_context_available",
            "sum"
        ),
    )
)

# ============================================================
# Artist-comparison limitation summary
# ============================================================

artist_comparison_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                artist_comparison_limitations_df
            ),
        },
        {
            "Metric": "Artists assessed",
            "Observed Value": (
                artist_comparison_limitations_df[
                    "Artist"
                ].nunique()
            ),
        },
        {
            "Metric": "Single-release artists",
            "Observed Value": int(
                (
                    artist_release_counts_df[
                        "artist_release_count"
                    ] == 1
                ).sum()
            ),
        },
        {
            "Metric": "Artists with 2-3 releases",
            "Observed Value": int(
                artist_release_counts_df[
                    "artist_release_count"
                ]
                .between(
                    2,
                    3
                )
                .sum()
            ),
        },
        {
            "Metric": "Artists with 4-9 releases",
            "Observed Value": int(
                artist_release_counts_df[
                    "artist_release_count"
                ]
                .between(
                    4,
                    9
                )
                .sum()
            ),
        },
        {
            "Metric": "Artists with 10+ releases",
            "Observed Value": int(
                (
                    artist_release_counts_df[
                        "artist_release_count"
                    ] >= 10
                ).sum()
            ),
        },
        {
            "Metric": "Releases with independent artist baseline",
            "Observed Value": int(
                artist_comparison_limitations_df[
                    "independent_baseline_available"
                ].sum()
            ),
        },
        {
            "Metric": "Releases without independent artist baseline",
            "Observed Value": int(
                (
                    ~artist_comparison_limitations_df[
                        "independent_baseline_available"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Releases with growth context",
            "Observed Value": int(
                artist_comparison_limitations_df[
                    "growth_context_available"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with momentum context",
            "Observed Value": int(
                artist_comparison_limitations_df[
                    "momentum_context_available"
                ].sum()
            ),
        },
        {
            "Metric": "Releases requiring artist-comparison caution",
            "Observed Value": int(
                artist_comparison_limitations_df[
                    "artist_comparison_caution_flag"
                ].sum()
            ),
        },
        {
            "Metric": (
                "High-performing releases requiring "
                "artist-context caution"
            ),
            "Observed Value": int(
                artist_comparison_limitations_df[
                    "high_performance_artist_context_caution"
                ].sum()
            ),
        },
    ]
)

print("Assessing PMIP artist comparison limitations")
print("=" * 80)

display(
    artist_comparison_summary_df
)

# ============================================================
# Artist-history distribution
# ============================================================

artist_history_distribution_df = (
    artist_comparison_limitations_df[
        [
            "Artist",
            "artist_release_count",
            "artist_history_group",
        ]
    ]
    .drop_duplicates(
        subset="Artist"
    )[
        "artist_history_group"
    ]
    .value_counts()
    .rename_axis(
        "Artist History Group"
    )
    .reset_index(
        name="Artists"
    )
)

artist_history_distribution_df[
    "Share (%)"
] = (
    artist_history_distribution_df[
        "Artists"
    ]
    / artist_comparison_limitations_df[
        "Artist"
    ].nunique()
    * 100
)

print()
print("Artist-history distribution")
print("=" * 80)

display(
    artist_history_distribution_df
)

# ============================================================
# Artist-context availability by history group
# ============================================================

baseline_by_history_df = (
    artist_comparison_limitations_df
    .groupby(
        "artist_history_group",
        observed=True
    )
    .agg(
        Releases=(
            "release_id",
            "count"
        ),
        Artists=(
            "Artist",
            "nunique"
        ),
        Baseline_Available=(
            "independent_baseline_available",
            "sum"
        ),
        Growth_Context_Available=(
            "growth_context_available",
            "sum"
        ),
        Momentum_Context_Available=(
            "momentum_context_available",
            "sum"
        ),
    )
    .reset_index()
)

baseline_by_history_df[
    "Baseline Coverage (%)"
] = (
    baseline_by_history_df[
        "Baseline_Available"
    ]
    / baseline_by_history_df[
        "Releases"
    ]
    * 100
)

baseline_by_history_df[
    "Growth Context Coverage (%)"
] = (
    baseline_by_history_df[
        "Growth_Context_Available"
    ]
    / baseline_by_history_df[
        "Releases"
    ]
    * 100
)

baseline_by_history_df[
    "Momentum Context Coverage (%)"
] = (
    baseline_by_history_df[
        "Momentum_Context_Available"
    ]
    / baseline_by_history_df[
        "Releases"
    ]
    * 100
)

print()
print("Artist-context availability by history group")
print("=" * 80)

display(
    baseline_by_history_df
)

# ============================================================
# High-performing releases requiring caution
# ============================================================

artist_context_caution_df = (
    artist_comparison_limitations_df.loc[
        artist_comparison_limitations_df[
            "high_performance_artist_context_caution"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "artist_release_count",
            "artist_history_group",
            "artist_performance_score",
            "release_vs_artist_baseline_score",
            "artist_growth_score",
            "artist_momentum_score",
            "release_performance_percentile",
            "release_performance_class",
        ]
    ]
    .sort_values(
        by=[
            "release_performance_percentile",
            "artist_release_count",
        ],
        ascending=[
            False,
            True,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "High-performing releases requiring "
    "artist-context caution"
)
print("=" * 80)

display(
    artist_context_caution_df
)

# ============================================================
# Validation
# ============================================================

artist_comparison_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13.3 completed",
            "Passed":
                section_13_3_complete,
        },
        {
            "Validation Area":
                "Artist identity available",
            "Passed":
                artist_comparison_limitations_df[
                    "Artist"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Artist release counts calculated",
            "Passed":
                artist_comparison_limitations_df[
                    "artist_release_count"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Artist history groups assigned",
            "Passed":
                artist_comparison_limitations_df[
                    "artist_history_group"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Independent baseline availability assessed",
            "Passed":
                (
                    "independent_baseline_available"
                    in artist_comparison_limitations_df.columns
                ),
        },
        {
            "Validation Area":
                "Growth context availability assessed",
            "Passed":
                (
                    "growth_context_available"
                    in artist_comparison_limitations_df.columns
                ),
        },
        {
            "Validation Area":
                "Momentum context availability assessed",
            "Passed":
                (
                    "momentum_context_available"
                    in artist_comparison_limitations_df.columns
                ),
        },
        {
            "Validation Area":
                "Artist comparison caution assigned",
            "Passed":
                (
                    "artist_comparison_caution_flag"
                    in artist_comparison_limitations_df.columns
                ),
        },
        {
            "Validation Area":
                "High-performance caution assessed",
            "Passed":
                (
                    "high_performance_artist_context_caution"
                    in artist_comparison_limitations_df.columns
                ),
        },
    ]
)

print()
print("Artist Comparison Limitations validation")
print("=" * 80)

display(
    artist_comparison_validation_df
)

# ============================================================
# Completion
# ============================================================

section_13_4_complete = bool(
    artist_comparison_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 13.4 completion status: "
    f"{section_13_4_complete}"
)

if not section_13_4_complete:
    raise RuntimeError(
        "Artist Comparison Limitations did not "
        "complete successfully."
    )
```

### Interpretation

Artist comparison reliability depends heavily on the amount of historical release data available for each artist.

Around 67% of artists in the dataset have only one observed release, while a further 19.7% have only two or three releases. This means that many artist baselines are built from limited historical evidence.

Although 3,256 releases contain artist baseline, growth and momentum context, 2,243 releases still require caution because their artist history is too limited for strong comparison.

A total of 154 high-performing releases also require artist-context caution, showing that strong release performance should not automatically be interpreted as strong artist-relative performance when historical evidence is weak.

Artist-level comparisons should therefore be treated as contextual support rather than definitive evidence, particularly for artists with short observed histories.

## 13.5 Human Review Requirements

### Purpose

This subsection defines when PMIP outputs should be reviewed by a human before they are used to support release-level decisions.

PMIP combines several automated indicators, but high scores do not remove uncertainty caused by temporal differences, incomplete data coverage, release-identity ambiguity or limited artist history.

The aim is therefore to combine the limitations identified throughout Section 13 into a transparent human-review framework.

Human review is particularly important where a release is high priority, supported by incomplete evidence, affected by identity ambiguity, based on limited artist history, or associated with unusual performance behaviour.

```python
# ============================================================
# Section 13.5 - Human Review Requirements
# ============================================================

if not section_13_4_complete:
    raise RuntimeError(
        "Section 13.4 must be completed before assessing "
        "Human Review Requirements."
    )

# ============================================================
# Working source
# ============================================================

human_review_df = high_priority_insights_df.copy()

# ============================================================
# Validate core fields
# ============================================================

required_human_review_fields = [
    "release_id",
    "Track",
    "Artist",
    "release_performance_percentile",
    "release_performance_class",
    "composite_release_performance_score",
    "composite_weight_coverage_pct",
    "high_priority_release_flag",
    "priority_review_flag",
    "priority_signal_score",
]

missing_human_review_fields = [
    field
    for field in required_human_review_fields
    if field not in human_review_df.columns
]

if missing_human_review_fields:
    raise RuntimeError(
        "Required human-review fields are missing: "
        f"{missing_human_review_fields}"
    )

# ============================================================
# Attach limitation flags from Sections 13.1-13.4
#
# Mapping by release_id avoids merge-related duplicate columns.
# ============================================================

human_review_df[
    "temporal_caution_flag"
] = (
    human_review_df[
        "release_id"
    ].map(
        temporal_limitations_df
        .set_index("release_id")[
            "high_performance_temporal_caution"
        ]
    )
    .fillna(False)
    .astype(bool)
)

human_review_df[
    "coverage_caution_flag"
] = (
    human_review_df[
        "release_id"
    ].map(
        data_coverage_limitations_df
        .set_index("release_id")[
            "high_performance_coverage_caution"
        ]
    )
    .fillna(False)
    .astype(bool)
)

human_review_df[
    "priority_coverage_caution_flag"
] = (
    human_review_df[
        "release_id"
    ].map(
        data_coverage_limitations_df
        .set_index("release_id")[
            "priority_coverage_caution"
        ]
    )
    .fillna(False)
    .astype(bool)
)

human_review_df[
    "identity_ambiguity_flag"
] = (
    human_review_df[
        "release_id"
    ].map(
        release_identity_limitations_df
        .set_index("release_id")[
            "identity_ambiguity_flag"
        ]
    )
    .fillna(False)
    .astype(bool)
)

human_review_df[
    "artist_context_caution_flag"
] = (
    human_review_df[
        "release_id"
    ].map(
        artist_comparison_limitations_df
        .set_index("release_id")[
            "artist_comparison_caution_flag"
        ]
    )
    .fillna(False)
    .astype(bool)
)

# ============================================================
# Unusual-performance context
# ============================================================

if "unusual_performance_flag" not in human_review_df.columns:
    human_review_df[
        "unusual_performance_flag"
    ] = False

human_review_df[
    "unusual_performance_flag"
] = (
    human_review_df[
        "unusual_performance_flag"
    ]
    .fillna(False)
    .astype(bool)
)

# ============================================================
# Human-review trigger count
# ============================================================

review_trigger_fields = [
    "temporal_caution_flag",
    "coverage_caution_flag",
    "priority_coverage_caution_flag",
    "identity_ambiguity_flag",
    "artist_context_caution_flag",
    "unusual_performance_flag",
]

human_review_df[
    "human_review_trigger_count"
] = (
    human_review_df[
        review_trigger_fields
    ]
    .astype(int)
    .sum(axis=1)
)

# ============================================================
# Human-review requirement
#
# A release requires review when:
#
# 1. PMIP already marks it for priority review, OR
# 2. it contains one or more limitation / uncertainty flags.
# ============================================================

human_review_df[
    "human_review_required"
] = (
    human_review_df[
        "priority_review_flag"
    ]
    |
    (
        human_review_df[
            "human_review_trigger_count"
        ] >= 1
    )
)

# ============================================================
# Human-review priority level
# ============================================================

human_review_df[
    "human_review_priority"
] = "Routine Monitoring"

# Review because at least one limitation exists
human_review_df.loc[
    human_review_df[
        "human_review_trigger_count"
    ] >= 1,
    "human_review_priority"
] = "Review Recommended"

# Review because multiple limitations overlap
human_review_df.loc[
    human_review_df[
        "human_review_trigger_count"
    ] >= 2,
    "human_review_priority"
] = "High Review Priority"

# PMIP priority case
human_review_df.loc[
    human_review_df[
        "priority_review_flag"
    ],
    "human_review_priority"
] = "High Review Priority"

# Strongest review level
human_review_df.loc[
    (
        human_review_df[
            "high_priority_release_flag"
        ]
        & (
            human_review_df[
                "human_review_trigger_count"
            ] >= 2
        )
    )
    |
    (
        human_review_df[
            "priority_review_flag"
        ]
        & (
            human_review_df[
                "human_review_trigger_count"
            ] >= 3
        )
    ),
    "human_review_priority"
] = "Critical Human Review"

# ============================================================
# Review reason
# ============================================================

def build_review_reason(row):

    reasons = []

    if row["priority_review_flag"]:
        reasons.append("PMIP priority-review signal")

    if row["temporal_caution_flag"]:
        reasons.append("temporal comparability")

    if row["coverage_caution_flag"]:
        reasons.append("limited data coverage")

    if row["priority_coverage_caution_flag"]:
        reasons.append("priority result with limited evidence")

    if row["identity_ambiguity_flag"]:
        reasons.append("release identity ambiguity")

    if row["artist_context_caution_flag"]:
        reasons.append("limited artist comparison context")

    if row["unusual_performance_flag"]:
        reasons.append("unusual performance behaviour")

    if not reasons:
        return "No additional human-review trigger"

    return "; ".join(reasons)


human_review_df[
    "human_review_reason"
] = human_review_df.apply(
    build_review_reason,
    axis=1
)

# ============================================================
# Human-review summary
# ============================================================

human_review_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records assessed",
            "Observed Value": len(
                human_review_df
            ),
        },
        {
            "Metric": "Releases requiring human review",
            "Observed Value": int(
                human_review_df[
                    "human_review_required"
                ].sum()
            ),
        },
        {
            "Metric": "Releases not requiring additional review",
            "Observed Value": int(
                (
                    ~human_review_df[
                        "human_review_required"
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Priority-review releases",
            "Observed Value": int(
                human_review_df[
                    "priority_review_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Temporal caution cases",
            "Observed Value": int(
                human_review_df[
                    "temporal_caution_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Coverage caution cases",
            "Observed Value": int(
                human_review_df[
                    "coverage_caution_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Identity ambiguity cases",
            "Observed Value": int(
                human_review_df[
                    "identity_ambiguity_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Artist-context caution cases",
            "Observed Value": int(
                human_review_df[
                    "artist_context_caution_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Unusual-performance cases",
            "Observed Value": int(
                human_review_df[
                    "unusual_performance_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Releases with multiple review triggers",
            "Observed Value": int(
                (
                    human_review_df[
                        "human_review_trigger_count"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "Metric": "Critical human-review cases",
            "Observed Value": int(
                (
                    human_review_df[
                        "human_review_priority"
                    ]
                    == "Critical Human Review"
                ).sum()
            ),
        },
    ]
)

print("Assessing PMIP human review requirements")
print("=" * 80)

display(
    human_review_summary_df
)

# ============================================================
# Human-review priority distribution
# ============================================================

human_review_priority_distribution_df = (
    human_review_df[
        "human_review_priority"
    ]
    .value_counts()
    .rename_axis(
        "Human Review Priority"
    )
    .reset_index(
        name="Releases"
    )
)

human_review_priority_distribution_df[
    "Share (%)"
] = (
    human_review_priority_distribution_df[
        "Releases"
    ]
    / len(
        human_review_df
    )
    * 100
)

print()
print("Human-review priority distribution")
print("=" * 80)

display(
    human_review_priority_distribution_df
)

# ============================================================
# Individual review-trigger prevalence
# ============================================================

review_trigger_summary_df = pd.DataFrame(
    [
        {
            "Human Review Trigger":
                "Temporal Comparability",
            "Releases": int(
                human_review_df[
                    "temporal_caution_flag"
                ].sum()
            ),
        },
        {
            "Human Review Trigger":
                "Limited Data Coverage",
            "Releases": int(
                human_review_df[
                    "coverage_caution_flag"
                ].sum()
            ),
        },
        {
            "Human Review Trigger":
                "Priority Result With Limited Evidence",
            "Releases": int(
                human_review_df[
                    "priority_coverage_caution_flag"
                ].sum()
            ),
        },
        {
            "Human Review Trigger":
                "Release Identity Ambiguity",
            "Releases": int(
                human_review_df[
                    "identity_ambiguity_flag"
                ].sum()
            ),
        },
        {
            "Human Review Trigger":
                "Limited Artist Context",
            "Releases": int(
                human_review_df[
                    "artist_context_caution_flag"
                ].sum()
            ),
        },
        {
            "Human Review Trigger":
                "Unusual Performance Behaviour",
            "Releases": int(
                human_review_df[
                    "unusual_performance_flag"
                ].sum()
            ),
        },
    ]
)

review_trigger_summary_df[
    "Share (%)"
] = (
    review_trigger_summary_df[
        "Releases"
    ]
    / len(
        human_review_df
    )
    * 100
)

print()
print("Human-review trigger prevalence")
print("=" * 80)

display(
    review_trigger_summary_df
)

# ============================================================
# Highest-priority human-review cases
# ============================================================

human_review_case_df = (
    human_review_df.loc[
        human_review_df[
            "human_review_required"
        ],
        [
            "release_id",
            "Track",
            "Artist",
            "release_performance_percentile",
            "release_performance_class",
            "composite_release_performance_score",
            "composite_weight_coverage_pct",
            "priority_signal_score",
            "human_review_trigger_count",
            "human_review_priority",
            "human_review_reason",
        ]
    ]
    .sort_values(
        by=[
            "human_review_trigger_count",
            "priority_signal_score",
            "release_performance_percentile",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(30)
    .reset_index(drop=True)
)

print()
print("Highest-priority human-review cases")
print("=" * 80)

display(
    human_review_case_df
)

# ============================================================
# Responsible-use principles
# ============================================================

responsible_use_principles_df = pd.DataFrame(
    [
        {
            "Principle":
                "Decision Support",
            "Requirement":
                (
                    "PMIP should support rather than replace "
                    "human decision-making."
                ),
        },
        {
            "Principle":
                "Evidence Awareness",
            "Requirement":
                (
                    "Performance scores should be interpreted "
                    "alongside evidence coverage."
                ),
        },
        {
            "Principle":
                "Temporal Context",
            "Requirement":
                (
                    "Releases from different periods or observation "
                    "windows should not be treated as perfectly equivalent."
                ),
        },
        {
            "Principle":
                "Identity Verification",
            "Requirement":
                (
                    "Ambiguous, remixed or alternate release identities "
                    "should be verified before action."
                ),
        },
        {
            "Principle":
                "Artist Context",
            "Requirement":
                (
                    "Artist-relative conclusions require sufficient "
                    "historical release evidence."
                ),
        },
        {
            "Principle":
                "Unusual Signals",
            "Requirement":
                (
                    "Anomaly and unusual-performance signals should "
                    "trigger investigation rather than automatic conclusions."
                ),
        },
        {
            "Principle":
                "Human Accountability",
            "Requirement":
                (
                    "Final A&R or business decisions remain the "
                    "responsibility of human reviewers."
                ),
        },
    ]
)

print()
print("PMIP responsible-use principles")
print("=" * 80)

display(
    responsible_use_principles_df
)

# ============================================================
# Section 13 overall validation
# ============================================================

section_13_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13.1 Temporal Limitations completed",
            "Passed":
                section_13_1_complete,
        },
        {
            "Validation Area":
                "Section 13.2 Data-Coverage Limitations completed",
            "Passed":
                section_13_2_complete,
        },
        {
            "Validation Area":
                "Section 13.3 Release Identity Limitations completed",
            "Passed":
                section_13_3_complete,
        },
        {
            "Validation Area":
                "Section 13.4 Artist Comparison Limitations completed",
            "Passed":
                section_13_4_complete,
        },
        {
            "Validation Area":
                "Human-review triggers calculated",
            "Passed":
                human_review_df[
                    "human_review_trigger_count"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Human-review requirements assigned",
            "Passed":
                human_review_df[
                    "human_review_required"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Human-review priority assigned",
            "Passed":
                human_review_df[
                    "human_review_priority"
                ].notna().all(),
        },
        {
            "Validation Area":
                "Responsible-use principles defined",
            "Passed":
                not responsible_use_principles_df.empty,
        },
    ]
)

print()
print("Responsible Use and Limitations validation")
print("=" * 80)

display(
    section_13_validation_df
)

# ============================================================
# Completion
# ============================================================

section_13_5_complete = bool(
    section_13_validation_df[
        "Passed"
    ].all()
)

section_13_complete = bool(
    section_13_1_complete
    and section_13_2_complete
    and section_13_3_complete
    and section_13_4_complete
    and section_13_5_complete
)

print()
print(
    f"Section 13.5 completion status: "
    f"{section_13_5_complete}"
)

print(
    f"Section 13 overall completion status: "
    f"{section_13_complete}"
)

if not section_13_complete:
    raise RuntimeError(
        "Section 13 Responsible Use and Limitations "
        "did not complete successfully."
    )
```

### Section 13 Interpretation

Section 13 identified several limitations that affect how PMIP results should be interpreted and used.

Temporal comparability was generally acceptable, although some releases were observed under different historical conditions. Data coverage was also uneven, particularly for chart, geographic and unusual-performance indicators, meaning that some rankings are supported by less complete evidence.

Release identity was mostly reliable at identifier level, but alternate versions, remixes and repeated track–artist combinations introduced ambiguity in some cases. Artist-level comparisons were the strongest limitation, as many artists had only a small number of observed releases, reducing the reliability of baseline, growth and momentum comparisons.

These limitations resulted in 3,105 releases being marked for some level of human review, including 404 releases with multiple review triggers and 200 critical review cases.

Overall, PMIP should be treated as a decision-support system rather than an automated decision-maker. Its scores and classifications can help prioritise investigation, but final A&R or business decisions should combine model outputs with evidence quality, contextual knowledge and human judgement.

```python
# ============================================================
# Section 14.2 - Save Configuration and Metadata
# ============================================================

from datetime import datetime
import json
import pandas as pd

if not section_14_1_complete:
    raise RuntimeError(
        "Section 14.1 must be completed before saving "
        "configuration and metadata."
    )

# ============================================================
# Configuration directory
# ============================================================

release_performance_config_dir = (
    release_performance_dir
    / "config"
)

release_performance_config_dir.mkdir(
    parents=True,
    exist_ok=True
)

# ============================================================
# Configuration paths
# ============================================================

scoring_configuration_path = (
    release_performance_config_dir
    / "scoring_configuration.json"
)

artifact_metadata_path = (
    release_performance_config_dir
    / "artifact_metadata.json"
)

indicator_metadata_path = (
    release_performance_config_dir
    / "indicator_metadata.csv"
)

classification_metadata_path = (
    release_performance_config_dir
    / "classification_metadata.csv"
)

# ============================================================
# Indicator metadata
# ============================================================

indicator_metadata_df = pd.DataFrame(
    [
        {
            "Field":
                "streaming_performance_score",
            "Indicator":
                "Streaming Performance",
            "Dimension":
                "Streaming Performance Intelligence",
            "Direction":
                "Higher is stronger",
            "Composite Role":
                "Performance indicator",
        },
        {
            "Field":
                "chart_performance_score",
            "Indicator":
                "Chart Performance",
            "Dimension":
                "Chart Performance Intelligence",
            "Direction":
                "Higher is stronger",
            "Composite Role":
                "Performance indicator",
        },
        {
            "Field":
                "geographic_reach_score",
            "Indicator":
                "Geographic Reach",
            "Dimension":
                "Geographic Release Reach",
            "Direction":
                "Higher is stronger",
            "Composite Role":
                "Performance indicator",
        },
        {
            "Field":
                "temporal_comparability_score",
            "Indicator":
                "Temporal Comparability",
            "Dimension":
                "Release Timing and Historical Context",
            "Direction":
                "Higher is stronger",
            "Composite Role":
                "Performance indicator",
        },
        {
            "Field":
                "artist_release_relationship_score",
            "Indicator":
                "Artist–Release Relationship",
            "Dimension":
                "Artist Context",
            "Direction":
                "Higher is stronger",
            "Composite Role":
                "Performance indicator",
        },
        {
            "Field":
                "unusual_performance_score",
            "Indicator":
                "Unusual Performance",
            "Dimension":
                "Unusual Release Performance Patterns",
            "Direction":
                "Higher is more unusual",
            "Composite Role":
                "Contextual indicator",
        },
    ]
)

indicator_metadata_df.to_csv(
    indicator_metadata_path,
    index=False
)

# ============================================================
# Composite scoring configuration
#
# These weights should match the Section 11 composite framework.
# ============================================================

composite_weights = {
    "streaming_performance_score": 0.30,
    "chart_performance_score": 0.20,
    "geographic_reach_score": 0.15,
    "temporal_comparability_score": 0.15,
    "artist_release_relationship_score": 0.20,
}

# Validate the weights before saving
composite_weight_sum = sum(
    composite_weights.values()
)

if not abs(
    composite_weight_sum - 1.0
) < 1e-9:
    raise RuntimeError(
        "Composite scoring weights do not sum to 1."
    )

# ============================================================
# Release-performance classification metadata
# ============================================================

classification_metadata_df = pd.DataFrame(
    [
        {
            "Classification":
                "Low Release Performance",
            "Percentile Range":
                "0-25",
        },
        {
            "Classification":
                "Moderate Release Performance",
            "Percentile Range":
                ">25-50",
        },
        {
            "Classification":
                "High Release Performance",
            "Percentile Range":
                ">50-75",
        },
        {
            "Classification":
                "Very High Release Performance",
            "Percentile Range":
                ">75-90",
        },
        {
            "Classification":
                "Exceptional Release Performance",
            "Percentile Range":
                ">90-100",
        },
    ]
)

classification_metadata_df.to_csv(
    classification_metadata_path,
    index=False
)

# ============================================================
# Scoring configuration
# ============================================================

scoring_configuration = {
    "framework_name":
        "PMIP Release Performance Scoring Framework",

    "scoring_scale":
        {
            "minimum": 0,
            "maximum": 100,
        },

    "composite_weights":
        composite_weights,

    "unusual_performance_handling":
        {
            "included_in_primary_composite": False,
            "role": (
                "Contextual indicator used for unusual-performance "
                "analysis and human-review support."
            ),
            "direction_alignment": (
                "Higher unusualness corresponds to lower consistency."
            ),
        },

    "performance_classes":
        {
            "Low Release Performance": "0-25",
            "Moderate Release Performance": ">25-50",
            "High Release Performance": ">50-75",
            "Very High Release Performance": ">75-90",
            "Exceptional Release Performance": ">90-100",
        },

    "ranking_groups":
        {
            "Top 25%": ">=75th percentile",
            "Top 10%": ">=90th percentile",
            "Top 5%": ">=95th percentile",
            "Top 1%": ">=99th percentile",
        },

    "responsible_use":
        {
            "decision_role":
                "Decision support only",
            "human_review_required":
                True,
            "evidence_coverage_should_be_considered":
                True,
        },
}

with open(
    scoring_configuration_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        scoring_configuration,
        file,
        indent=4
    )

# ============================================================
# Artifact metadata
# ============================================================

artifact_metadata = {
    "project":
        "Public Music Intelligence Platform (PMIP)",

    "artifact_group":
        "Release Performance Scoring and Classification",

    "artifact_root_directory":
        str(
            release_performance_dir
        ),

    "config_directory":
        str(
            release_performance_config_dir
        ),

    "results_directory":
        str(
            release_performance_results_dir
        ),

    "generated_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "release_records":
        int(
            len(
                release_intelligence_results_df
            )
        ),

    "high_priority_releases":
        int(
            len(
                high_priority_release_results_df
            )
        ),

    "human_review_releases":
        int(
            len(
                human_review_queue_df
            )
        ),

    "selected_indicators":
        int(
            len(
                indicator_metadata_df
            )
        ),

    "primary_composite_indicators":
        int(
            len(
                composite_weights
            )
        ),

    "final_output_fields":
        [
            "release_id",
            "Track",
            "Artist",
            "composite_release_performance_score",
            "release_performance_class",
            "release_performance_percentile",
            "release_rank_position",
            "composite_weight_coverage_pct",
            "composite_evidence_strength",
            "release_priority_class",
            "priority_signal_score",
            "priority_signal_rank",
            "high_priority_release_flag",
            "priority_review_flag",
            "human_review_required",
            "human_review_priority",
            "human_review_reason",
        ],

    "result_artifacts":
        [
            release_intelligence_results_path.name,
            high_priority_release_results_path.name,
            human_review_queue_path.name,
            section_12_summary_path.name,
            section_13_summary_path.name,
            responsible_use_principles_path.name,
        ],

    "config_artifacts":
        [
            scoring_configuration_path.name,
            artifact_metadata_path.name,
            indicator_metadata_path.name,
            classification_metadata_path.name,
        ],

    "responsible_use_note":
        (
            "PMIP outputs are intended to support analytical "
            "review and should not replace human A&R or "
            "business decision-making."
        ),
}

with open(
    artifact_metadata_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        artifact_metadata,
        file,
        indent=4
    )

# ============================================================
# Saved configuration artifact inventory
# ============================================================

configuration_artifact_inventory_df = pd.DataFrame(
    [
        {
            "Artifact":
                "Scoring Configuration",
            "Filename":
                scoring_configuration_path.name,
            "Saved":
                scoring_configuration_path.exists(),
        },
        {
            "Artifact":
                "Artifact Metadata",
            "Filename":
                artifact_metadata_path.name,
            "Saved":
                artifact_metadata_path.exists(),
        },
        {
            "Artifact":
                "Indicator Metadata",
            "Filename":
                indicator_metadata_path.name,
            "Saved":
                indicator_metadata_path.exists(),
        },
        {
            "Artifact":
                "Classification Metadata",
            "Filename":
                classification_metadata_path.name,
            "Saved":
                classification_metadata_path.exists(),
        },
    ]
)

print("Saving PMIP configuration and metadata")
print("=" * 80)

print(
    f"Config directory: "
    f"{release_performance_config_dir}"
)

display(
    configuration_artifact_inventory_df
)

# ============================================================
# Configuration summary
# ============================================================

configuration_summary_df = pd.DataFrame(
    [
        {
            "Metric":
                "Selected indicators documented",
            "Observed Value":
                len(
                    indicator_metadata_df
                ),
        },
        {
            "Metric":
                "Primary composite indicators",
            "Observed Value":
                len(
                    composite_weights
                ),
        },
        {
            "Metric":
                "Composite weight total",
            "Observed Value":
                composite_weight_sum,
        },
        {
            "Metric":
                "Performance classes documented",
            "Observed Value":
                len(
                    classification_metadata_df
                ),
        },
        {
            "Metric":
                "Release records documented",
            "Observed Value":
                artifact_metadata[
                    "release_records"
                ],
        },
        {
            "Metric":
                "High-priority releases documented",
            "Observed Value":
                artifact_metadata[
                    "high_priority_releases"
                ],
        },
        {
            "Metric":
                "Human-review releases documented",
            "Observed Value":
                artifact_metadata[
                    "human_review_releases"
                ],
        },
    ]
)

print()
print("Configuration and metadata summary")
print("=" * 80)

display(
    configuration_summary_df
)

# ============================================================
# Validation
# ============================================================

configuration_metadata_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 14.1 completed",
            "Passed":
                section_14_1_complete,
        },
        {
            "Validation Area":
                "Config directory created",
            "Passed":
                release_performance_config_dir.exists(),
        },
        {
            "Validation Area":
                "Scoring configuration saved",
            "Passed":
                scoring_configuration_path.exists(),
        },
        {
            "Validation Area":
                "Artifact metadata saved",
            "Passed":
                artifact_metadata_path.exists(),
        },
        {
            "Validation Area":
                "Indicator metadata saved",
            "Passed":
                indicator_metadata_path.exists(),
        },
        {
            "Validation Area":
                "Classification metadata saved",
            "Passed":
                classification_metadata_path.exists(),
        },
        {
            "Validation Area":
                "Composite weights sum to 1",
            "Passed":
                abs(
                    composite_weight_sum - 1.0
                ) < 1e-9,
        },
        {
            "Validation Area":
                "Release count documented correctly",
            "Passed":
                (
                    artifact_metadata[
                        "release_records"
                    ]
                    ==
                    len(
                        release_intelligence_results_df
                    )
                ),
        },
        {
            "Validation Area":
                "Result artifacts documented",
            "Passed":
                len(
                    artifact_metadata[
                        "result_artifacts"
                    ]
                ) == 6,
        },
        {
            "Validation Area":
                "Config artifacts documented",
            "Passed":
                len(
                    artifact_metadata[
                        "config_artifacts"
                    ]
                ) == 4,
        },
    ]
)

print()
print("Save Configuration and Metadata validation")
print("=" * 80)

display(
    configuration_metadata_validation_df
)

# ============================================================
# Completion
# ============================================================

section_14_2_complete = bool(
    configuration_metadata_validation_df[
        "Passed"
    ].all()
)

print()
print(
    f"Section 14.2 completion status: "
    f"{section_14_2_complete}"
)

if not section_14_2_complete:
    raise RuntimeError(
        "Configuration and Metadata were not "
        "saved successfully."
    )
```

### 14.3 Reload and Verify Saved Artifacts


### Purpose

This subsection reloads the saved PMIP release-intelligence artifacts and checks that they can be opened successfully without loss of structure or key information.

The aim is to confirm that the saved CSV and JSON files preserve the same release counts, important fields, scoring configuration and metadata created earlier in the notebook.

This provides a final reproducibility check before the artifacts are used for reporting, software integration or future PMIP development.

```python
# ============================================================
# Section 14.3 - Reload and Verify Saved Artifacts
# ============================================================

import json
import pandas as pd
import numpy as np

if not section_14_2_complete:
    raise RuntimeError(
        "Section 14.2 must be completed before reloading "
        "and verifying saved artifacts."
    )

# ============================================================
# Verify expected artifact locations
# ============================================================

results_paths = [
    release_intelligence_results_path,
    high_priority_release_results_path,
    human_review_queue_path,
    section_12_summary_path,
    section_13_summary_path,
    responsible_use_principles_path,
]

config_paths = [
    scoring_configuration_path,
    artifact_metadata_path,
    indicator_metadata_path,
    classification_metadata_path,
]

results_location_valid = all(
    path.parent == release_performance_results_dir
    for path in results_paths
)

config_location_valid = all(
    path.parent == release_performance_config_dir
    for path in config_paths
)

print("Verifying PMIP artifact locations")
print("=" * 80)

print(
    f"Results directory: "
    f"{release_performance_results_dir}"
)

print(
    f"Config directory: "
    f"{release_performance_config_dir}"
)

# ============================================================
# Reload saved CSV artifacts
# ============================================================

reloaded_release_intelligence_df = pd.read_csv(
    release_intelligence_results_path
)

reloaded_high_priority_df = pd.read_csv(
    high_priority_release_results_path
)

reloaded_human_review_df = pd.read_csv(
    human_review_queue_path
)

reloaded_section_12_summary_df = pd.read_csv(
    section_12_summary_path
)

reloaded_section_13_summary_df = pd.read_csv(
    section_13_summary_path
)

reloaded_responsible_use_df = pd.read_csv(
    responsible_use_principles_path
)

reloaded_indicator_metadata_df = pd.read_csv(
    indicator_metadata_path
)

reloaded_classification_metadata_df = pd.read_csv(
    classification_metadata_path
)

# ============================================================
# Reload JSON artifacts
# ============================================================

with open(
    scoring_configuration_path,
    "r",
    encoding="utf-8"
) as file:
    reloaded_scoring_configuration = json.load(file)

with open(
    artifact_metadata_path,
    "r",
    encoding="utf-8"
) as file:
    reloaded_artifact_metadata = json.load(file)

# ============================================================
# Reload summary
# ============================================================

reload_summary_df = pd.DataFrame(
    [
        {
            "Artifact": "Complete Release Intelligence",
            "Rows": len(reloaded_release_intelligence_df),
            "Columns": len(
                reloaded_release_intelligence_df.columns
            ),
        },
        {
            "Artifact": "High-Priority Releases",
            "Rows": len(reloaded_high_priority_df),
            "Columns": len(
                reloaded_high_priority_df.columns
            ),
        },
        {
            "Artifact": "Human Review Queue",
            "Rows": len(reloaded_human_review_df),
            "Columns": len(
                reloaded_human_review_df.columns
            ),
        },
        {
            "Artifact": "Section 12 Summary",
            "Rows": len(reloaded_section_12_summary_df),
            "Columns": len(
                reloaded_section_12_summary_df.columns
            ),
        },
        {
            "Artifact": "Section 13 Summary",
            "Rows": len(reloaded_section_13_summary_df),
            "Columns": len(
                reloaded_section_13_summary_df.columns
            ),
        },
        {
            "Artifact": "Responsible Use Principles",
            "Rows": len(reloaded_responsible_use_df),
            "Columns": len(
                reloaded_responsible_use_df.columns
            ),
        },
        {
            "Artifact": "Indicator Metadata",
            "Rows": len(reloaded_indicator_metadata_df),
            "Columns": len(
                reloaded_indicator_metadata_df.columns
            ),
        },
        {
            "Artifact": "Classification Metadata",
            "Rows": len(reloaded_classification_metadata_df),
            "Columns": len(
                reloaded_classification_metadata_df.columns
            ),
        },
    ]
)

print()
print("Reloading PMIP saved artifacts")
print("=" * 80)

display(reload_summary_df)

# ============================================================
# Required final output fields
# ============================================================

required_final_output_fields = [
    "release_id",
    "Track",
    "Artist",
    "composite_release_performance_score",
    "release_performance_class",
    "release_performance_percentile",
    "release_rank_position",
    "composite_weight_coverage_pct",
    "composite_evidence_strength",
    "release_priority_class",
    "priority_signal_score",
    "priority_signal_rank",
    "high_priority_release_flag",
    "priority_review_flag",
    "human_review_required",
    "human_review_priority",
    "human_review_reason",
]

missing_reloaded_fields = [
    field
    for field in required_final_output_fields
    if field not in reloaded_release_intelligence_df.columns
]

# ============================================================
# Compare row counts
# ============================================================

release_count_match = (
    len(reloaded_release_intelligence_df)
    ==
    len(release_intelligence_results_df)
)

high_priority_count_match = (
    len(reloaded_high_priority_df)
    ==
    len(high_priority_release_results_df)
)

human_review_count_match = (
    len(reloaded_human_review_df)
    ==
    len(human_review_queue_df)
)

# ============================================================
# Compare unique release identities
# ============================================================

release_id_count_match = (
    reloaded_release_intelligence_df[
        "release_id"
    ].nunique()
    ==
    release_intelligence_results_df[
        "release_id"
    ].nunique()
)

duplicate_release_ids_after_reload = int(
    reloaded_release_intelligence_df[
        "release_id"
    ]
    .duplicated()
    .sum()
)

# ============================================================
# Compare key score statistics
# ============================================================

original_median_composite_score = float(
    release_intelligence_results_df[
        "composite_release_performance_score"
    ].median()
)

reloaded_median_composite_score = float(
    reloaded_release_intelligence_df[
        "composite_release_performance_score"
    ].median()
)

composite_score_median_match = bool(
    np.isclose(
        original_median_composite_score,
        reloaded_median_composite_score,
        rtol=1e-10,
        atol=1e-10,
    )
)

original_max_priority_score = float(
    release_intelligence_results_df[
        "priority_signal_score"
    ].max()
)

reloaded_max_priority_score = float(
    reloaded_release_intelligence_df[
        "priority_signal_score"
    ].max()
)

priority_score_max_match = bool(
    np.isclose(
        original_max_priority_score,
        reloaded_max_priority_score,
        rtol=1e-10,
        atol=1e-10,
    )
)

# ============================================================
# Compare release-performance class distributions
# ============================================================

original_performance_distribution = (
    release_intelligence_results_df[
        "release_performance_class"
    ]
    .astype(str)
    .value_counts()
    .to_dict()
)

reloaded_performance_distribution = (
    reloaded_release_intelligence_df[
        "release_performance_class"
    ]
    .astype(str)
    .value_counts()
    .to_dict()
)

performance_distribution_match = (
    original_performance_distribution
    ==
    reloaded_performance_distribution
)

# ============================================================
# Compare human-review priority distributions
# ============================================================

original_review_distribution = (
    release_intelligence_results_df[
        "human_review_priority"
    ]
    .astype(str)
    .value_counts()
    .to_dict()
)

reloaded_review_distribution = (
    reloaded_release_intelligence_df[
        "human_review_priority"
    ]
    .astype(str)
    .value_counts()
    .to_dict()
)

human_review_distribution_match = (
    original_review_distribution
    ==
    reloaded_review_distribution
)

# ============================================================
# Distribution verification table
# ============================================================

all_performance_classes = sorted(
    set(original_performance_distribution)
    |
    set(reloaded_performance_distribution)
)

performance_distribution_verification_df = pd.DataFrame(
    [
        {
            "Release Performance Class": class_name,
            "Original Count":
                original_performance_distribution.get(
                    class_name,
                    0
                ),
            "Reloaded Count":
                reloaded_performance_distribution.get(
                    class_name,
                    0
                ),
            "Match":
                original_performance_distribution.get(
                    class_name,
                    0
                )
                ==
                reloaded_performance_distribution.get(
                    class_name,
                    0
                ),
        }
        for class_name in all_performance_classes
    ]
)

print()
print("Release-performance distribution verification")
print("=" * 80)

display(
    performance_distribution_verification_df
)

# ============================================================
# Verify scoring configuration
# ============================================================

reloaded_weight_sum = float(
    sum(
        reloaded_scoring_configuration[
            "composite_weights"
        ].values()
    )
)

scoring_weight_sum_valid = bool(
    np.isclose(
        reloaded_weight_sum,
        1.0,
        rtol=1e-10,
        atol=1e-10,
    )
)

selected_indicator_count_match = (
    len(reloaded_indicator_metadata_df)
    ==
    reloaded_artifact_metadata[
        "selected_indicators"
    ]
)

primary_composite_indicator_count_match = (
    len(
        reloaded_scoring_configuration[
            "composite_weights"
        ]
    )
    ==
    reloaded_artifact_metadata[
        "primary_composite_indicators"
    ]
)

# ============================================================
# Verify metadata counts
# ============================================================

metadata_release_count_match = (
    reloaded_artifact_metadata[
        "release_records"
    ]
    ==
    len(reloaded_release_intelligence_df)
)

metadata_high_priority_count_match = (
    reloaded_artifact_metadata[
        "high_priority_releases"
    ]
    ==
    len(reloaded_high_priority_df)
)

metadata_human_review_count_match = (
    reloaded_artifact_metadata[
        "human_review_releases"
    ]
    ==
    len(reloaded_human_review_df)
)

# ============================================================
# Reload verification summary
# ============================================================

reload_verification_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Release records reloaded",
            "Original Value":
                len(release_intelligence_results_df),
            "Reloaded Value":
                len(reloaded_release_intelligence_df),
            "Match":
                release_count_match,
        },
        {
            "Metric": "High-priority releases reloaded",
            "Original Value":
                len(high_priority_release_results_df),
            "Reloaded Value":
                len(reloaded_high_priority_df),
            "Match":
                high_priority_count_match,
        },
        {
            "Metric": "Human-review releases reloaded",
            "Original Value":
                len(human_review_queue_df),
            "Reloaded Value":
                len(reloaded_human_review_df),
            "Match":
                human_review_count_match,
        },
        {
            "Metric": "Unique release IDs",
            "Original Value":
                release_intelligence_results_df[
                    "release_id"
                ].nunique(),
            "Reloaded Value":
                reloaded_release_intelligence_df[
                    "release_id"
                ].nunique(),
            "Match":
                release_id_count_match,
        },
        {
            "Metric": "Median composite score",
            "Original Value":
                original_median_composite_score,
            "Reloaded Value":
                reloaded_median_composite_score,
            "Match":
                composite_score_median_match,
        },
        {
            "Metric": "Maximum priority score",
            "Original Value":
                original_max_priority_score,
            "Reloaded Value":
                reloaded_max_priority_score,
            "Match":
                priority_score_max_match,
        },
        {
            "Metric":
                "Performance class distribution",
            "Original Value":
                sum(
                    original_performance_distribution.values()
                ),
            "Reloaded Value":
                sum(
                    reloaded_performance_distribution.values()
                ),
            "Match":
                performance_distribution_match,
        },
    ]
)

print()
print("Reload verification summary")
print("=" * 80)

display(
    reload_verification_summary_df
)

# ============================================================
# Overall validation
# ============================================================

reload_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 14.2 completed",
            "Passed":
                section_14_2_complete,
        },
        {
            "Validation Area":
                "Results saved in results directory",
            "Passed":
                results_location_valid,
        },
        {
            "Validation Area":
                "Config saved in config directory",
            "Passed":
                config_location_valid,
        },
        {
            "Validation Area":
                "Complete release dataset reloaded",
            "Passed":
                release_count_match,
        },
        {
            "Validation Area":
                "High-priority dataset reloaded",
            "Passed":
                high_priority_count_match,
        },
        {
            "Validation Area":
                "Human-review queue reloaded",
            "Passed":
                human_review_count_match,
        },
        {
            "Validation Area":
                "Required final fields preserved",
            "Passed":
                len(missing_reloaded_fields) == 0,
        },
        {
            "Validation Area":
                "Unique release IDs preserved",
            "Passed":
                release_id_count_match,
        },
        {
            "Validation Area":
                "No duplicate release IDs after reload",
            "Passed":
                duplicate_release_ids_after_reload == 0,
        },
        {
            "Validation Area":
                "Performance class distribution preserved",
            "Passed":
                performance_distribution_match,
        },
        {
            "Validation Area":
                "Human-review distribution preserved",
            "Passed":
                human_review_distribution_match,
        },
        {
            "Validation Area":
                "Median composite score preserved",
            "Passed":
                composite_score_median_match,
        },
        {
            "Validation Area":
                "Maximum priority score preserved",
            "Passed":
                priority_score_max_match,
        },
        {
            "Validation Area":
                "Reloaded composite weights sum to 1",
            "Passed":
                scoring_weight_sum_valid,
        },
        {
            "Validation Area":
                "Selected indicator count preserved",
            "Passed":
                selected_indicator_count_match,
        },
        {
            "Validation Area":
                "Composite indicator count preserved",
            "Passed":
                primary_composite_indicator_count_match,
        },
        {
            "Validation Area":
                "Metadata release count matches",
            "Passed":
                metadata_release_count_match,
        },
        {
            "Validation Area":
                "Metadata high-priority count matches",
            "Passed":
                metadata_high_priority_count_match,
        },
        {
            "Validation Area":
                "Metadata human-review count matches",
            "Passed":
                metadata_human_review_count_match,
        },
    ]
)

print()
print("Reload and Verify Saved Artifacts validation")
print("=" * 80)

display(reload_validation_df)

# ============================================================
# Report any remaining failures
# ============================================================

remaining_reload_failures_df = (
    reload_validation_df.loc[
        reload_validation_df[
            "Passed"
        ] == False
    ]
    .reset_index(drop=True)
)

if not remaining_reload_failures_df.empty:

    print()
    print("Remaining failed verification checks")
    print("=" * 80)

    display(
        remaining_reload_failures_df
    )

# ============================================================
# Section 14 completion
# ============================================================

section_14_3_complete = bool(
    reload_validation_df[
        "Passed"
    ].all()
)

section_14_complete = bool(
    section_14_1_complete
    and section_14_2_complete
    and section_14_3_complete
)

print()
print(
    f"Section 14.3 completion status: "
    f"{section_14_3_complete}"
)

print(
    f"Section 14 overall completion status: "
    f"{section_14_complete}"
)

if not section_14_complete:
    raise RuntimeError(
        "Section 14 Artifact Saving and Reload Verification "
        "did not complete successfully."
    )
```
