# PMIP Market Growth Detection

## Public Music Intelligence Platform

### Purpose

Develop the market growth detection component of PMIP to identify countries and music markets where artists and tracks are showing meaningful signs of increasing performance over time.

The workflow will analyse historical streaming and chart behaviour across markets, distinguish sustained growth from short-term spikes, construct interpretable growth indicators, compare growth patterns between countries, and prepare validated market-growth intelligence for later integration into the PMIP backend and dashboard.


## Objectives

The objectives of the PMIP Market Growth Detection workflow are to:

1. Define what constitutes meaningful market growth within PMIP.
2. Analyse historical artist and track performance across countries and markets.
3. Measure changes in streaming performance over time.
4. Analyse changes in chart position, chart activity and persistence.
5. Distinguish short-term spikes from sustained growth patterns.
6. Develop interpretable streaming, chart, persistence and stability indicators.
7. Compare market growth rates across countries.
8. Investigate relationships between geographic reach and performance growth.
9. Identify emerging markets for individual tracks and artists.
10. Construct market growth scores and classifications.
11. Rank markets using validated growth evidence.
12. Prepare country-, artist- and track-level growth findings.
13. Document temporal, coverage and cross-country comparison limitations.
14. Define responsible-use and human-review requirements.
15. Save, validate and reload market-growth artifacts.
16. Prepare the completed market-growth component for PMIP software integration.

## 1. Introduction and Objectives

This section establishes the purpose, objectives, scope, and analytical boundaries of the PMIP Market Growth Detection workflow.

The workflow focuses on identifying meaningful changes in artist and track performance within observed country markets over time while keeping clear boundaries between historical evidence, growth signals, and future predictions.


### 1.3 Scope and Analytical Boundaries

#### Purpose

Define exactly what the PMIP Market Growth Detection component will analyse and clarify what its outputs can and cannot represent.

The analysis will focus on historical changes in streaming performance, chart behaviour, persistence, volatility, and geographic market activity for artists and tracks.

Market growth will be treated as an **observed historical performance pattern** supported by the available PMIP data.

The workflow will not assume that increasing performance guarantees future growth, commercial success, artist popularity, or investment value.

Growth signals will therefore be designed as interpretable decision-support evidence that can later be reviewed alongside other PMIP intelligence components.

#### Interpretation

Section 1.3 successfully defines the scope and analytical boundaries of the PMIP Market Growth Detection workflow.

A total of **10 analytical scope areas** were established.

The workflow will analyse market growth using historical evidence at the:

- track level,
- artist level,
- and observed country-market level.

Streaming growth is included as a core signal.

This means the workflow will examine how observed streaming performance changes over time within markets.

However, increasing streams will not be treated as proof of guaranteed future growth or as evidence that a particular event or market caused the increase.

Chart performance is also included.

The workflow will analyse changes in:

- chart position,
- chart activity,
- ranking behaviour,
- and chart persistence.

Chart improvement alone will not be interpreted as complete commercial success.

Market growth will be evaluated using **temporal ordering**.

This means observations must be placed in the correct historical sequence before growth is calculated.

The workflow will therefore compare earlier and later performance rather than treating unordered observations as evidence of change.

A major analytical objective is to distinguish **sustained growth from short-term spikes**.

Repeated or persistent improvement will carry more meaning than a single sudden increase.

This prevents isolated performance events from automatically being classified as genuine market growth.

Geographic growth comparisons are also included.

Artist and track performance can be compared across observed country markets, but the workflow recognises that countries may differ in:

- audience size,
- observation volume,
- platform activity,
- chart structure,
- and market behaviour.

Cross-country growth values must therefore remain context-sensitive.

The workflow will also identify **emerging markets** where artists or tracks demonstrate meaningful historical improvement.

These signals are intended to support PMIP discovery and prioritisation.

They must not be treated as automatic recommendations to invest, promote, sign, or allocate commercial resources.

Growth scores and classifications will be developed later in the workflow.

These scores will summarise multiple validated growth signals, but they will not represent a universal measure of artist or track quality.

The individual components behind each score must remain visible and interpretable.

The Market Growth Detection component is also explicitly separated from forecasting.

Its purpose is to describe and classify **historical growth behaviour**.

It does not directly predict future success.

Forecasting may later use these growth indicators as supporting features, but future outcomes remain outside the scope of this notebook.

A total of **8 interpretation principles** were also defined.

These require PMIP market-growth analysis to preserve:

- observed evidence,
- temporal ordering,
- persistence,
- contextual information,
- non-causal interpretation,
- uncertainty around future outcomes,
- component-level transparency,
- and human review.

## 2. Data Source Assessment

This section identifies and evaluates the historical datasets required for PMIP Market Growth Detection.

The assessment focuses on whether the available data contains sufficient temporal, geographic, streaming, chart, artist, and track information to support reliable growth analysis.

### 2.1 Candidate Historical Datasets

#### Purpose

Identify the PMIP datasets that may support market-growth analysis and check whether they contain the core information required for historical performance tracking.

The assessment will focus on:

- time or date fields,
- country or market fields,
- track identifiers,
- artist identifiers,
- streaming metrics,
- chart-position metrics,
- and sufficient historical observations.


```python
import pandas as pd

print("Assessing PMIP candidate historical datasets")
print("=" * 100)

# ------------------------------------------------------------
# 1. Discover dataframe variables currently available
# ------------------------------------------------------------

candidate_dataframes = {}

for variable_name, value in list(globals().items()):
    if isinstance(value, pd.DataFrame):
        candidate_dataframes[variable_name] = value


print()
print("Available dataframe candidates")
print("=" * 100)

candidate_summary = []

for variable_name, dataframe in candidate_dataframes.items():

    candidate_summary.append(
        {
            "Variable": variable_name,
            "Rows": len(dataframe),
            "Columns": len(dataframe.columns),
        }
    )

candidate_summary_df = (
    pd.DataFrame(candidate_summary)
    .sort_values(
        by=["Rows", "Columns"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

display(candidate_summary_df)


# ------------------------------------------------------------
# 2. Search for growth-relevant fields
# ------------------------------------------------------------

growth_keywords = [
    "date",
    "time",
    "timestamp",
    "country",
    "market",
    "track",
    "artist",
    "stream",
    "position",
    "rank",
    "chart",
]

candidate_field_records = []

for variable_name, dataframe in candidate_dataframes.items():

    matched_columns = []

    for column in dataframe.columns:

        column_lower = str(column).lower()

        if any(
            keyword in column_lower
            for keyword in growth_keywords
        ):
            matched_columns.append(str(column))

    candidate_field_records.append(
        {
            "Variable": variable_name,
            "Rows": len(dataframe),
            "Columns": len(dataframe.columns),
            "Matched Growth-Relevant Fields": matched_columns,
            "Matched Field Count": len(matched_columns),
        }
    )

candidate_growth_fields_df = (
    pd.DataFrame(candidate_field_records)
    .sort_values(
        by=["Matched Field Count", "Rows"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

print()
print("Growth-relevant field assessment")
print("=" * 100)

display(candidate_growth_fields_df)


# ------------------------------------------------------------
# 3. Define candidate suitability
# ------------------------------------------------------------

def has_keyword(columns, keywords):

    columns_lower = [
        str(column).lower()
        for column in columns
    ]

    return any(
        any(
            keyword in column
            for keyword in keywords
        )
        for column in columns_lower
    )


suitability_records = []

for variable_name, dataframe in candidate_dataframes.items():

    columns = dataframe.columns

    has_temporal = has_keyword(
        columns,
        ["date", "time", "timestamp"]
    )

    has_geographic = has_keyword(
        columns,
        ["country", "market"]
    )

    has_track = has_keyword(
        columns,
        ["track"]
    )

    has_artist = has_keyword(
        columns,
        ["artist"]
    )

    has_streaming = has_keyword(
        columns,
        ["stream"]
    )

    has_chart = has_keyword(
        columns,
        ["position", "rank", "chart"]
    )

    suitability_score = sum(
        [
            has_temporal,
            has_geographic,
            has_track,
            has_artist,
            has_streaming,
            has_chart,
        ]
    )

    suitability_records.append(
        {
            "Variable": variable_name,
            "Temporal Field": has_temporal,
            "Geographic Field": has_geographic,
            "Track Field": has_track,
            "Artist Field": has_artist,
            "Streaming Field": has_streaming,
            "Chart Field": has_chart,
            "Suitability Score": suitability_score,
        }
    )

candidate_suitability_df = (
    pd.DataFrame(suitability_records)
    .sort_values(
        by="Suitability Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print()
print("Candidate historical dataset suitability")
print("=" * 100)

display(candidate_suitability_df)


# ------------------------------------------------------------
# 4. Diagnostics
# ------------------------------------------------------------

candidate_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Dataframes discovered",
            "Observed Value": len(candidate_dataframes),
        },
        {
            "Diagnostic": "Datasets with temporal fields",
            "Observed Value": int(
                candidate_suitability_df[
                    "Temporal Field"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with geographic fields",
            "Observed Value": int(
                candidate_suitability_df[
                    "Geographic Field"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with streaming fields",
            "Observed Value": int(
                candidate_suitability_df[
                    "Streaming Field"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with chart fields",
            "Observed Value": int(
                candidate_suitability_df[
                    "Chart Field"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with full 6-field suitability",
            "Observed Value": int(
                (
                    candidate_suitability_df[
                        "Suitability Score"
                    ] == 6
                ).sum()
            ),
        },
    ]
)

print()
print("Candidate dataset diagnostics")
print("=" * 100)

display(candidate_diagnostics_df)


# ------------------------------------------------------------
# 5. Completion status
# ------------------------------------------------------------

section_2_1_complete = bool(
    len(candidate_dataframes) > 0
)

if not section_2_1_complete:
    raise RuntimeError(
        "Section 2.1 could not identify any dataframe "
        "candidates in the current notebook environment."
    )

print()
print(
    f"Section 2.1 completion status: "
    f"{section_2_1_complete}"
)

print(
    "Candidate historical datasets have been identified "
    "and assessed for market-growth relevance."
)
```

```python
from pathlib import Path

print("Inspecting available PMIP data files")
print("=" * 100)

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

data_directories = [
    project_root / "data" / "raw",
    project_root / "data" / "processed",
    project_root / "data" / "integrated",
    project_root / "data" / "model_ready",
]

file_records = []

for directory in data_directories:

    if not directory.exists():
        continue

    for path in sorted(directory.rglob("*")):

        if path.is_file():

            file_records.append(
                {
                    "Directory": str(directory.relative_to(project_root)),
                    "File Name": path.name,
                    "Extension": path.suffix.lower(),
                    "Size MB": round(
                        path.stat().st_size / (1024 ** 2),
                        3
                    ),
                    "Relative Path": str(
                        path.relative_to(project_root)
                    ),
                }
            )

available_data_files_df = pd.DataFrame(file_records)

print()
print("Available PMIP data files")
print("=" * 100)

if available_data_files_df.empty:

    print("No files found in the expected PMIP data directories.")

else:

    display(
        available_data_files_df[
            available_data_files_df["Extension"].isin(
                [".csv", ".parquet", ".json"]
            )
        ].reset_index(drop=True)
    )

print()
print(
    f"Data files discovered: "
    f"{len(available_data_files_df)}"
)
```

#### Interpretation

### 2.2 Required Columns

#### Purpose

Identify the fields required for reliable market-growth analysis and determine which available PMIP datasets contain them.

The assessment will focus on temporal, geographic, track, artist, streaming, and chart-performance fields.

```python
import pandas as pd
from pathlib import Path

print("Assessing PMIP market growth required columns")
print("=" * 100)

# ------------------------------------------------------------
# 1. Define required analytical field groups
# ------------------------------------------------------------

required_field_groups = {
    "Temporal": [
        "date",
        "time",
        "timestamp",
        "week",
        "month",
        "year",
    ],
    "Geographic": [
        "country",
        "market",
        "region",
    ],
    "Track": [
        "track",
        "track_id",
        "track_name",
    ],
    "Artist": [
        "artist",
        "artist_id",
        "artist_name",
        "artists",
    ],
    "Streaming": [
        "stream",
        "streams",
    ],
    "Chart": [
        "position",
        "rank",
        "chart",
    ],
}


# ------------------------------------------------------------
# 2. Inspect available CSV schemas safely
# ------------------------------------------------------------

schema_records = []

csv_files_df = available_data_files_df[
    available_data_files_df["Extension"] == ".csv"
].copy()

for _, row in csv_files_df.iterrows():

    relative_path = row["Relative Path"]
    file_path = project_root / relative_path

    try:

        sample_df = pd.read_csv(
            file_path,
            nrows=5,
            low_memory=False
        )

        columns = list(sample_df.columns)

        schema_records.append(
            {
                "File Name": row["File Name"],
                "Relative Path": relative_path,
                "Column Count": len(columns),
                "Columns": columns,
                "Read Successful": True,
            }
        )

    except Exception as error:

        schema_records.append(
            {
                "File Name": row["File Name"],
                "Relative Path": relative_path,
                "Column Count": 0,
                "Columns": [],
                "Read Successful": False,
            }
        )


dataset_schema_df = pd.DataFrame(schema_records)

print()
print("Available dataset schemas")
print("=" * 100)

display(
    dataset_schema_df[
        [
            "File Name",
            "Column Count",
            "Columns",
            "Read Successful",
        ]
    ]
)


# ------------------------------------------------------------
# 3. Check each dataset for required field groups
# ------------------------------------------------------------

def find_matching_columns(columns, keywords):

    matches = []

    for column in columns:

        column_lower = str(column).lower()

        if any(
            keyword in column_lower
            for keyword in keywords
        ):
            matches.append(column)

    return matches


requirement_records = []

for _, row in dataset_schema_df.iterrows():

    columns = row["Columns"]

    record = {
        "File Name": row["File Name"],
    }

    available_group_count = 0

    for group_name, keywords in required_field_groups.items():

        matches = find_matching_columns(
            columns,
            keywords
        )

        record[f"{group_name} Fields"] = matches
        record[f"{group_name} Available"] = len(matches) > 0

        if matches:
            available_group_count += 1

    record["Required Groups Available"] = available_group_count
    record["Full Growth Suitability"] = (
        available_group_count
        == len(required_field_groups)
    )

    requirement_records.append(record)


required_columns_assessment_df = (
    pd.DataFrame(requirement_records)
    .sort_values(
        by="Required Groups Available",
        ascending=False
    )
    .reset_index(drop=True)
)

print()
print("Required market-growth field assessment")
print("=" * 100)

display(required_columns_assessment_df)


# ------------------------------------------------------------
# 4. Diagnostics
# ------------------------------------------------------------

required_columns_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "CSV datasets assessed",
            "Observed Value": len(dataset_schema_df),
        },
        {
            "Diagnostic": "Readable dataset schemas",
            "Observed Value": int(
                dataset_schema_df[
                    "Read Successful"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with temporal fields",
            "Observed Value": int(
                required_columns_assessment_df[
                    "Temporal Available"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with geographic fields",
            "Observed Value": int(
                required_columns_assessment_df[
                    "Geographic Available"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with streaming fields",
            "Observed Value": int(
                required_columns_assessment_df[
                    "Streaming Available"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets with chart fields",
            "Observed Value": int(
                required_columns_assessment_df[
                    "Chart Available"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Datasets satisfying all required groups",
            "Observed Value": int(
                required_columns_assessment_df[
                    "Full Growth Suitability"
                ].sum()
            ),
        },
    ]
)

print()
print("Required column diagnostics")
print("=" * 100)

display(required_columns_diagnostics_df)
```

#### Interpretation

The required-column assessment confirms that the PMIP project contains suitable fields for market-growth analysis.

`charts_cleaned.csv` is the main historical dataset because it contains temporal, geographic, track, artist, streaming, and chart-position fields.

Two datasets satisfy all six required field groups, but `charts_cleaned.csv` provides the strongest direct evidence for historical growth analysis.

Section 2.2 is therefore complete.

### 2.3 Temporal Coverage Assessment

#### Purpose

Assess whether the historical chart data provides sufficient date coverage and repeated observations to support reliable market-growth detection over time.

```python
import pandas as pd
from pathlib import Path

print("Assessing PMIP temporal coverage")
print("=" * 100)

# ------------------------------------------------------------
# 1. Locate primary historical dataset
# ------------------------------------------------------------

charts_path = project_root / "data" / "processed" / "charts_cleaned.csv"

if not charts_path.exists():
    raise FileNotFoundError(
        f"Expected historical dataset not found: {charts_path}"
    )


# ------------------------------------------------------------
# 2. Load only fields needed for temporal assessment
# ------------------------------------------------------------

usecols = [
    "date",
    "country",
    "track_id",
]

temporal_df = pd.read_csv(
    charts_path,
    usecols=usecols,
    low_memory=False
)

print()
print("Historical temporal source")
print("=" * 100)

print(f"Rows loaded: {len(temporal_df):,}")
print(f"Columns loaded: {list(temporal_df.columns)}")


# ------------------------------------------------------------
# 3. Parse and validate dates
# ------------------------------------------------------------

temporal_df["date"] = pd.to_datetime(
    temporal_df["date"],
    errors="coerce"
)

valid_date_count = int(
    temporal_df["date"].notna().sum()
)

missing_date_count = int(
    temporal_df["date"].isna().sum()
)

min_date = temporal_df["date"].min()
max_date = temporal_df["date"].max()

date_span_days = (
    max_date - min_date
).days if pd.notna(min_date) and pd.notna(max_date) else 0


# ------------------------------------------------------------
# 4. Assess observation frequency
# ------------------------------------------------------------

unique_dates = temporal_df["date"].nunique()

country_count = temporal_df["country"].nunique()

track_count = temporal_df["track_id"].nunique()

track_country_history_df = (
    temporal_df
    .dropna(
        subset=[
            "date",
            "country",
            "track_id",
        ]
    )
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .agg(
        observations=("date", "size"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        unique_dates=("date", "nunique"),
    )
    .reset_index()
)

track_country_history_df[
    "history_span_days"
] = (
    track_country_history_df["last_date"]
    - track_country_history_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 5. Temporal coverage summary
# ------------------------------------------------------------

temporal_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Valid dated observations",
            "Observed Value": valid_date_count,
        },
        {
            "Metric": "Missing / invalid dates",
            "Observed Value": missing_date_count,
        },
        {
            "Metric": "Earliest observation",
            "Observed Value": min_date,
        },
        {
            "Metric": "Latest observation",
            "Observed Value": max_date,
        },
        {
            "Metric": "Dataset span in days",
            "Observed Value": date_span_days,
        },
        {
            "Metric": "Unique observation dates",
            "Observed Value": unique_dates,
        },
        {
            "Metric": "Country markets",
            "Observed Value": country_count,
        },
        {
            "Metric": "Unique tracks",
            "Observed Value": track_count,
        },
        {
            "Metric": "Track-country histories",
            "Observed Value": len(
                track_country_history_df
            ),
        },
    ]
)

print()
print("Temporal coverage overview")
print("=" * 100)

display(temporal_summary_df)


# ------------------------------------------------------------
# 6. Track-country history diagnostics
# ------------------------------------------------------------

history_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Median observations per track-country history",
            "Observed Value": float(
                track_country_history_df[
                    "observations"
                ].median()
            ),
        },
        {
            "Diagnostic": "Median unique dates per track-country history",
            "Observed Value": float(
                track_country_history_df[
                    "unique_dates"
                ].median()
            ),
        },
        {
            "Diagnostic": "Median history span in days",
            "Observed Value": float(
                track_country_history_df[
                    "history_span_days"
                ].median()
            ),
        },
        {
            "Diagnostic": "Histories with at least 2 dates",
            "Observed Value": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 5 dates",
            "Observed Value": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 5
                ).sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 10 dates",
            "Observed Value": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 10
                ).sum()
            ),
        },
    ]
)

print()
print("Track-country temporal diagnostics")
print("=" * 100)

display(history_diagnostics_df)


# ------------------------------------------------------------
# 7. Validation
# ------------------------------------------------------------

temporal_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Date availability",
            "Requirement": (
                "The historical dataset must contain "
                "valid dated observations."
            ),
            "Observed Evidence": (
                f"{valid_date_count:,} valid dated observations"
            ),
            "Passed": valid_date_count > 0,
        },
        {
            "Validation Area": "Temporal span",
            "Requirement": (
                "The dataset must span more than one observation date."
            ),
            "Observed Evidence": (
                f"{unique_dates:,} unique dates across "
                f"{date_span_days:,} days"
            ),
            "Passed": unique_dates > 1,
        },
        {
            "Validation Area": "Repeated track-country histories",
            "Requirement": (
                "At least some track-country histories must contain "
                "multiple observation dates."
            ),
            "Observed Evidence": (
                f"{int((track_country_history_df['unique_dates'] >= 2).sum()):,} "
                "histories contain at least 2 dates"
            ),
            "Passed": bool(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 2
                ).any()
            ),
        },
        {
            "Validation Area": "Market coverage",
            "Requirement": (
                "Historical observations must span multiple country markets."
            ),
            "Observed Evidence": (
                f"{country_count:,} country markets represented"
            ),
            "Passed": country_count > 1,
        },
    ]
)

print()
print("Temporal coverage validation")
print("=" * 100)

display(temporal_validation_df)


section_2_3_complete = bool(
    temporal_validation_df["Passed"].all()
)

if not section_2_3_complete:
    raise AssertionError(
        "Section 2.3 Temporal Coverage Assessment "
        "validation failed."
    )

print()
print(
    f"Section 2.3 completion status: "
    f"{section_2_3_complete}"
)

print(
    "Historical temporal coverage is suitable for "
    "continued PMIP market-growth assessment."
)
```

#### Interpretation

The historical chart dataset contains **5,427,136 valid dated observations** covering **515 unique dates** from **2013-04-28 to 2023-04-06**.

Repeated track-country histories are common, with **262,916 histories containing at least two observation dates**, so temporal growth can be measured reliably for a substantial part of the dataset.

The source currently contains **77 market values**; this will be examined further in Section 2.4 to confirm the valid country-market population.

Section 2.3 is complete.

### 2.4 Country and Market Coverage

#### Purpose

Assess the geographic coverage of the historical chart dataset and identify the valid country markets available for PMIP market-growth analysis.

This step also checks for non-country, missing, or unusually sparse market values before cross-country growth comparisons are performed.

```python
print("Assessing PMIP country and market coverage")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prepare country-level coverage
# ------------------------------------------------------------

country_coverage_df = (
    temporal_df
    .dropna(subset=["country"])
    .groupby("country")
    .agg(
        observations=("track_id", "size"),
        unique_tracks=("track_id", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        unique_dates=("date", "nunique"),
    )
    .reset_index()
)

country_coverage_df["history_span_days"] = (
    country_coverage_df["last_date"]
    - country_coverage_df["first_date"]
).dt.days

country_coverage_df = (
    country_coverage_df
    .sort_values(
        by="observations",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. Country-market overview
# ------------------------------------------------------------

print()
print("Country-market coverage overview")
print("=" * 100)

print(
    f"Distinct market values: "
    f"{country_coverage_df['country'].nunique():,}"
)

print(
    f"Total observations represented: "
    f"{country_coverage_df['observations'].sum():,}"
)

print(
    f"Median observations per market: "
    f"{country_coverage_df['observations'].median():,.1f}"
)

print(
    f"Median unique tracks per market: "
    f"{country_coverage_df['unique_tracks'].median():,.1f}"
)

print(
    f"Median unique dates per market: "
    f"{country_coverage_df['unique_dates'].median():,.1f}"
)


# ------------------------------------------------------------
# 3. Inspect market values
# ------------------------------------------------------------

print()
print("Observed country-market values")
print("=" * 100)

display(
    country_coverage_df[
        [
            "country",
            "observations",
            "unique_tracks",
            "unique_dates",
            "first_date",
            "last_date",
        ]
    ]
)


# ------------------------------------------------------------
# 4. Identify potentially non-country market values
# ------------------------------------------------------------

country_labels = (
    country_coverage_df["country"]
    .astype(str)
    .str.strip()
)

potential_non_country_mask = (
    country_labels.str.len().ne(2)
)

potential_non_country_markets_df = (
    country_coverage_df[
        potential_non_country_mask
    ]
    .copy()
    .reset_index(drop=True)
)

print()
print("Potential non-country or unusual market values")
print("=" * 100)

if potential_non_country_markets_df.empty:
    print(
        "No unusual market labels detected using "
        "the two-character country-code rule."
    )
else:
    display(potential_non_country_markets_df)


# ------------------------------------------------------------
# 5. Coverage distribution
# ------------------------------------------------------------

coverage_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Distinct market values",
            "Observed Value": len(country_coverage_df),
        },
        {
            "Metric": "Minimum observations",
            "Observed Value": int(
                country_coverage_df[
                    "observations"
                ].min()
            ),
        },
        {
            "Metric": "Median observations",
            "Observed Value": float(
                country_coverage_df[
                    "observations"
                ].median()
            ),
        },
        {
            "Metric": "Maximum observations",
            "Observed Value": int(
                country_coverage_df[
                    "observations"
                ].max()
            ),
        },
        {
            "Metric": "Minimum unique tracks",
            "Observed Value": int(
                country_coverage_df[
                    "unique_tracks"
                ].min()
            ),
        },
        {
            "Metric": "Median unique tracks",
            "Observed Value": float(
                country_coverage_df[
                    "unique_tracks"
                ].median()
            ),
        },
        {
            "Metric": "Maximum unique tracks",
            "Observed Value": int(
                country_coverage_df[
                    "unique_tracks"
                ].max()
            ),
        },
        {
            "Metric": "Potential unusual market values",
            "Observed Value": len(
                potential_non_country_markets_df
            ),
        },
    ]
)

print()
print("Country-market coverage diagnostics")
print("=" * 100)

display(coverage_summary_df)


# ------------------------------------------------------------
# 6. Validation
# ------------------------------------------------------------

country_coverage_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Market population availability",
            "Requirement": (
                "The dataset must contain multiple "
                "observed country markets."
            ),
            "Observed Evidence": (
                f"{len(country_coverage_df):,} "
                "distinct market values"
            ),
            "Passed": len(country_coverage_df) > 1,
        },
        {
            "Validation Area": "Observation coverage",
            "Requirement": (
                "Every retained market must contain "
                "at least one observation."
            ),
            "Observed Evidence": (
                f"Minimum market observations: "
                f"{int(country_coverage_df['observations'].min()):,}"
            ),
            "Passed": bool(
                (
                    country_coverage_df[
                        "observations"
                    ] > 0
                ).all()
            ),
        },
        {
            "Validation Area": "Track coverage",
            "Requirement": (
                "Every retained market must contain "
                "at least one unique track."
            ),
            "Observed Evidence": (
                f"Minimum unique tracks: "
                f"{int(country_coverage_df['unique_tracks'].min()):,}"
            ),
            "Passed": bool(
                (
                    country_coverage_df[
                        "unique_tracks"
                    ] > 0
                ).all()
            ),
        },
        {
            "Validation Area": "Temporal market coverage",
            "Requirement": (
                "Markets must contain valid dated "
                "historical observations."
            ),
            "Observed Evidence": (
                f"Minimum unique dates in a market: "
                f"{int(country_coverage_df['unique_dates'].min()):,}"
            ),
            "Passed": bool(
                (
                    country_coverage_df[
                        "unique_dates"
                    ] > 0
                ).all()
            ),
        },
    ]
)

print()
print("Country and market coverage validation")
print("=" * 100)

display(country_coverage_validation_df)


section_2_4_complete = bool(
    country_coverage_validation_df["Passed"].all()
)

if not section_2_4_complete:
    raise AssertionError(
        "Section 2.4 Country and Market Coverage "
        "validation failed."
    )

print()
print(
    f"Section 2.4 completion status: "
    f"{section_2_4_complete}"
)

print(
    "Country and market coverage has been assessed "
    "for PMIP market-growth analysis."
)
```

```python
print("Finalising validated PMIP country-market population")
print("=" * 100)

# ------------------------------------------------------------
# Exclude aggregate non-country market
# ------------------------------------------------------------

validated_country_coverage_df = (
    country_coverage_df[
        country_coverage_df["country"] != "global"
    ]
    .copy()
    .reset_index(drop=True)
)

valid_country_codes = set(
    validated_country_coverage_df["country"]
)

print()
print(
    f"Original market values: "
    f"{len(country_coverage_df):,}"
)

print(
    f"Aggregate/non-country markets excluded: "
    f"{len(potential_non_country_markets_df):,}"
)

print(
    f"Validated country markets: "
    f"{len(validated_country_coverage_df):,}"
)

section_2_4_complete = bool(
    len(validated_country_coverage_df) == 76
)

if not section_2_4_complete:
    raise AssertionError(
        "Expected 76 validated country markets after "
        "excluding aggregate market values."
    )

print()
print(
    f"Section 2.4 completion status: "
    f"{section_2_4_complete}"
)
```

#### Interpretation

The historical dataset contains **77 market values**, including one aggregate `global` market.

After excluding the non-country aggregate, the validated PMIP market-growth population contains **76 country markets**.

Section 2.4 is complete.

### 2.5 Entity-Level Coverage

#### Purpose

Assess whether the historical dataset contains sufficient track- and artist-level coverage across validated country markets to support market-growth analysis at multiple entity levels.

```python
print("Assessing PMIP entity-level coverage")
print("=" * 100)

# ------------------------------------------------------------
# 1. Load entity fields from the historical dataset
# ------------------------------------------------------------

entity_usecols = [
    "track_id",
    "artists",
    "country",
    "date",
]

entity_df = pd.read_csv(
    charts_path,
    usecols=entity_usecols,
    low_memory=False
)

entity_df["date"] = pd.to_datetime(
    entity_df["date"],
    errors="coerce"
)

# Retain validated country markets only
entity_df = entity_df[
    entity_df["country"].isin(valid_country_codes)
].copy()


# ------------------------------------------------------------
# 2. Track-level coverage
# ------------------------------------------------------------

track_coverage_df = (
    entity_df
    .dropna(subset=["track_id"])
    .groupby("track_id")
    .agg(
        observations=("track_id", "size"),
        country_markets=("country", "nunique"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
    )
    .reset_index()
)

track_coverage_df["history_span_days"] = (
    track_coverage_df["last_date"]
    - track_coverage_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 3. Artist-level coverage
# ------------------------------------------------------------

artist_coverage_df = (
    entity_df
    .dropna(subset=["artists"])
    .groupby("artists")
    .agg(
        observations=("artists", "size"),
        country_markets=("country", "nunique"),
        unique_tracks=("track_id", "nunique"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
    )
    .reset_index()
)

artist_coverage_df["history_span_days"] = (
    artist_coverage_df["last_date"]
    - artist_coverage_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 4. Entity-level summary
# ------------------------------------------------------------

entity_coverage_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Validated country markets",
            "Observed Value": len(valid_country_codes),
        },
        {
            "Metric": "Unique tracks",
            "Observed Value": track_coverage_df["track_id"].nunique(),
        },
        {
            "Metric": "Unique artist representations",
            "Observed Value": artist_coverage_df["artists"].nunique(),
        },
        {
            "Metric": "Median markets per track",
            "Observed Value": float(
                track_coverage_df["country_markets"].median()
            ),
        },
        {
            "Metric": "Median dates per track",
            "Observed Value": float(
                track_coverage_df["unique_dates"].median()
            ),
        },
        {
            "Metric": "Median markets per artist",
            "Observed Value": float(
                artist_coverage_df["country_markets"].median()
            ),
        },
        {
            "Metric": "Median tracks per artist",
            "Observed Value": float(
                artist_coverage_df["unique_tracks"].median()
            ),
        },
        {
            "Metric": "Median dates per artist",
            "Observed Value": float(
                artist_coverage_df["unique_dates"].median()
            ),
        },
    ]
)

print()
print("Entity-level coverage overview")
print("=" * 100)

display(entity_coverage_summary_df)


# ------------------------------------------------------------
# 5. Diagnostics
# ------------------------------------------------------------

entity_coverage_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Tracks with at least 2 observation dates",
            "Observed Value": int(
                (
                    track_coverage_df["unique_dates"] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Tracks observed in at least 2 markets",
            "Observed Value": int(
                (
                    track_coverage_df["country_markets"] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artists with at least 2 observation dates",
            "Observed Value": int(
                (
                    artist_coverage_df["unique_dates"] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artists observed in at least 2 markets",
            "Observed Value": int(
                (
                    artist_coverage_df["country_markets"] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artists represented by at least 2 tracks",
            "Observed Value": int(
                (
                    artist_coverage_df["unique_tracks"] >= 2
                ).sum()
            ),
        },
    ]
)

print()
print("Entity-level coverage diagnostics")
print("=" * 100)

display(entity_coverage_diagnostics_df)


# ------------------------------------------------------------
# 6. Validation
# ------------------------------------------------------------

entity_coverage_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Track population availability",
            "Requirement": (
                "The dataset must contain track-level observations."
            ),
            "Observed Evidence": (
                f"{track_coverage_df['track_id'].nunique():,} "
                "unique tracks"
            ),
            "Passed": track_coverage_df["track_id"].nunique() > 0,
        },
        {
            "Validation Area": "Artist population availability",
            "Requirement": (
                "The dataset must contain artist-level observations."
            ),
            "Observed Evidence": (
                f"{artist_coverage_df['artists'].nunique():,} "
                "artist representations"
            ),
            "Passed": artist_coverage_df["artists"].nunique() > 0,
        },
        {
            "Validation Area": "Track temporal coverage",
            "Requirement": (
                "At least some tracks must contain repeated dates."
            ),
            "Observed Evidence": (
                f"{int((track_coverage_df['unique_dates'] >= 2).sum()):,} "
                "tracks contain at least 2 dates"
            ),
            "Passed": bool(
                (
                    track_coverage_df["unique_dates"] >= 2
                ).any()
            ),
        },
        {
            "Validation Area": "Artist temporal coverage",
            "Requirement": (
                "At least some artists must contain repeated dates."
            ),
            "Observed Evidence": (
                f"{int((artist_coverage_df['unique_dates'] >= 2).sum()):,} "
                "artists contain at least 2 dates"
            ),
            "Passed": bool(
                (
                    artist_coverage_df["unique_dates"] >= 2
                ).any()
            ),
        },
        {
            "Validation Area": "Multi-market entity coverage",
            "Requirement": (
                "At least some tracks and artists must appear "
                "across multiple country markets."
            ),
            "Observed Evidence": (
                f"{int((track_coverage_df['country_markets'] >= 2).sum()):,} "
                "multi-market tracks / "
                f"{int((artist_coverage_df['country_markets'] >= 2).sum()):,} "
                "multi-market artists"
            ),
            "Passed": bool(
                (
                    track_coverage_df["country_markets"] >= 2
                ).any()
                and
                (
                    artist_coverage_df["country_markets"] >= 2
                ).any()
            ),
        },
    ]
)

print()
print("Entity-level coverage validation")
print("=" * 100)

display(entity_coverage_validation_df)


section_2_5_complete = bool(
    entity_coverage_validation_df["Passed"].all()
)

section_2_complete = bool(
    section_2_3_complete
    and section_2_4_complete
    and section_2_5_complete
)

if not section_2_5_complete:
    raise AssertionError(
        "Section 2.5 Entity-Level Coverage validation failed."
    )

print()
print(
    f"Section 2.5 completion status: "
    f"{section_2_5_complete}"
)

print(
    f"Section 2 overall completion status: "
    f"{section_2_complete}"
)

print(
    "Track- and artist-level historical coverage "
    "is available for PMIP market-growth analysis."
)
```

### 3.3 Track-Level Growth

#### Purpose

Define and calculate track-level market-growth behaviour using historical streaming and chart observations within validated country markets.

The analysis will focus on repeated temporal change for each track-country history.

```python
print("Preparing PMIP track-level market growth histories")
print("=" * 100)

# ------------------------------------------------------------
# 1. Load track-level historical growth fields
# ------------------------------------------------------------

track_growth_usecols = [
    "date",
    "country",
    "track_id",
    "streams",
    "position",
]

track_growth_df = pd.read_csv(
    charts_path,
    usecols=track_growth_usecols,
    low_memory=False
)

track_growth_df["date"] = pd.to_datetime(
    track_growth_df["date"],
    errors="coerce"
)

# Retain validated country markets only
track_growth_df = track_growth_df[
    track_growth_df["country"].isin(valid_country_codes)
].copy()


# ------------------------------------------------------------
# 2. Clean required numeric fields
# ------------------------------------------------------------

track_growth_df["streams"] = pd.to_numeric(
    track_growth_df["streams"],
    errors="coerce"
)

track_growth_df["position"] = pd.to_numeric(
    track_growth_df["position"],
    errors="coerce"
)

track_growth_df = track_growth_df.dropna(
    subset=[
        "date",
        "country",
        "track_id",
    ]
)


# ------------------------------------------------------------
# 3. Order histories correctly
# ------------------------------------------------------------

track_growth_df = track_growth_df.sort_values(
    by=[
        "track_id",
        "country",
        "date",
    ]
).reset_index(drop=True)


# ------------------------------------------------------------
# 4. Calculate observation-to-observation changes
# ------------------------------------------------------------

group_keys = [
    "track_id",
    "country",
]

track_growth_df["previous_streams"] = (
    track_growth_df
    .groupby(group_keys)["streams"]
    .shift(1)
)

track_growth_df["previous_position"] = (
    track_growth_df
    .groupby(group_keys)["position"]
    .shift(1)
)

track_growth_df["stream_change"] = (
    track_growth_df["streams"]
    - track_growth_df["previous_streams"]
)

track_growth_df["stream_pct_change"] = (
    track_growth_df
    .groupby(group_keys)["streams"]
    .pct_change(fill_method=None)
)

# Lower numerical chart position is better.
track_growth_df["chart_improvement"] = (
    track_growth_df["previous_position"]
    - track_growth_df["position"]
)

track_growth_df["positive_stream_change"] = (
    track_growth_df["stream_change"] > 0
)

track_growth_df["positive_chart_change"] = (
    track_growth_df["chart_improvement"] > 0
)


# ------------------------------------------------------------
# 5. Aggregate track-country growth histories
# ------------------------------------------------------------

track_country_growth_df = (
    track_growth_df
    .groupby(group_keys)
    .agg(
        observations=("date", "size"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        first_streams=("streams", "first"),
        last_streams=("streams", "last"),
        mean_streams=("streams", "mean"),
        median_streams=("streams", "median"),
        stream_change_mean=("stream_change", "mean"),
        stream_pct_change_median=("stream_pct_change", "median"),
        positive_stream_changes=("positive_stream_change", "sum"),
        first_position=("position", "first"),
        last_position=("position", "last"),
        mean_position=("position", "mean"),
        chart_improvement_mean=("chart_improvement", "mean"),
        positive_chart_changes=("positive_chart_change", "sum"),
    )
    .reset_index()
)

track_country_growth_df["history_span_days"] = (
    track_country_growth_df["last_date"]
    - track_country_growth_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 6. Calculate directional growth evidence
# ------------------------------------------------------------

track_country_growth_df["stream_growth_direction"] = (
    track_country_growth_df["last_streams"]
    - track_country_growth_df["first_streams"]
)

track_country_growth_df["chart_growth_direction"] = (
    track_country_growth_df["first_position"]
    - track_country_growth_df["last_position"]
)

track_country_growth_df["stream_growth_supported"] = (
    track_country_growth_df["stream_growth_direction"] > 0
)

track_country_growth_df["chart_growth_supported"] = (
    track_country_growth_df["chart_growth_direction"] > 0
)

track_country_growth_df["repeated_history"] = (
    track_country_growth_df["unique_dates"] >= 2
)


# ------------------------------------------------------------
# 7. Overview
# ------------------------------------------------------------

track_growth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Track-country histories",
            "Observed Value": len(track_country_growth_df),
        },
        {
            "Metric": "Unique tracks",
            "Observed Value": track_country_growth_df[
                "track_id"
            ].nunique(),
        },
        {
            "Metric": "Validated country markets",
            "Observed Value": track_country_growth_df[
                "country"
            ].nunique(),
        },
        {
            "Metric": "Histories with repeated dates",
            "Observed Value": int(
                track_country_growth_df[
                    "repeated_history"
                ].sum()
            ),
        },
        {
            "Metric": "Histories with positive stream direction",
            "Observed Value": int(
                track_country_growth_df[
                    "stream_growth_supported"
                ].sum()
            ),
        },
        {
            "Metric": "Histories with positive chart direction",
            "Observed Value": int(
                track_country_growth_df[
                    "chart_growth_supported"
                ].sum()
            ),
        },
        {
            "Metric": "Median history span days",
            "Observed Value": float(
                track_country_growth_df[
                    "history_span_days"
                ].median()
            ),
        },
    ]
)

print()
print("Track-level growth overview")
print("=" * 100)

display(track_growth_summary_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

track_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track-country histories assessed",
            "Observed Value": len(track_country_growth_df),
        },
        {
            "Diagnostic": "Histories with at least 2 dates",
            "Observed Value": int(
                (
                    track_country_growth_df[
                        "unique_dates"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 5 dates",
            "Observed Value": int(
                (
                    track_country_growth_df[
                        "unique_dates"
                    ] >= 5
                ).sum()
            ),
        },
        {
            "Diagnostic": "Positive stream-growth histories",
            "Observed Value": int(
                track_country_growth_df[
                    "stream_growth_supported"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Positive chart-growth histories",
            "Observed Value": int(
                track_country_growth_df[
                    "chart_growth_supported"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories positive in both dimensions",
            "Observed Value": int(
                (
                    track_country_growth_df[
                        "stream_growth_supported"
                    ]
                    &
                    track_country_growth_df[
                        "chart_growth_supported"
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Track-level growth diagnostics")
print("=" * 100)

display(track_growth_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

track_growth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Track population availability",
            "Requirement": (
                "Track-level growth histories must be available."
            ),
            "Observed Evidence": (
                f"{track_country_growth_df['track_id'].nunique():,} "
                "unique tracks assessed"
            ),
            "Passed": (
                track_country_growth_df["track_id"].nunique() > 0
            ),
        },
        {
            "Validation Area": "Country-market preservation",
            "Requirement": (
                "Track growth analysis must retain validated "
                "country markets only."
            ),
            "Observed Evidence": (
                f"{track_country_growth_df['country'].nunique():,} "
                "validated markets represented"
            ),
            "Passed": (
                track_country_growth_df["country"].nunique()
                == len(valid_country_codes)
            ),
        },
        {
            "Validation Area": "Temporal ordering",
            "Requirement": (
                "Growth must be calculated from ordered "
                "historical observations."
            ),
            "Observed Evidence": (
                "Track-country observations sorted by date"
            ),
            "Passed": True,
        },
        {
            "Validation Area": "Repeated history availability",
            "Requirement": (
                "At least some track-country histories must "
                "contain repeated dates."
            ),
            "Observed Evidence": (
                f"{int(track_country_growth_df['repeated_history'].sum()):,} "
                "histories contain repeated dates"
            ),
            "Passed": bool(
                track_country_growth_df[
                    "repeated_history"
                ].any()
            ),
        },
        {
            "Validation Area": "Streaming growth evidence",
            "Requirement": (
                "Track histories must provide measurable "
                "stream-growth direction."
            ),
            "Observed Evidence": (
                f"{int(track_country_growth_df['stream_growth_supported'].sum()):,} "
                "positive stream-growth histories"
            ),
            "Passed": bool(
                track_country_growth_df[
                    "stream_growth_direction"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Chart growth evidence",
            "Requirement": (
                "Track histories must provide measurable "
                "chart-growth direction."
            ),
            "Observed Evidence": (
                f"{int(track_country_growth_df['chart_growth_supported'].sum()):,} "
                "positive chart-growth histories"
            ),
            "Passed": bool(
                track_country_growth_df[
                    "chart_growth_direction"
                ].notna().any()
            ),
        },
    ]
)

print()
print("Track-level growth validation")
print("=" * 100)

display(track_growth_validation_df)


section_3_3_complete = bool(
    track_growth_validation_df["Passed"].all()
)

if not section_3_3_complete:
    raise AssertionError(
        "Section 3.3 Track-Level Growth validation failed."
    )

print()
print(
    f"Section 3.3 completion status: "
    f"{section_3_3_complete}"
)

print(
    "Track-level historical growth evidence has been prepared "
    "for PMIP market-growth detection."
)
```

#### Interpretation

Track-level growth evidence was successfully prepared for **110,198 tracks** across **76 validated country markets**.

A substantial number of track-country histories contain repeated dates, with **82,762 showing positive stream direction** and **46,341 showing positive chart direction**.

Section 3.3 is complete.

### 3.4 Artist-Level Growth

#### Purpose

Prepare artist-level market-growth evidence by aggregating historical track performance across validated country markets.

The analysis will measure whether an artist shows repeated streaming and chart improvement over time within each market.

```python
print("Preparing PMIP artist-level market growth histories")
print("=" * 100)

# ------------------------------------------------------------
# 1. Load artist-level historical growth fields
# ------------------------------------------------------------

artist_growth_usecols = [
    "date",
    "country",
    "artists",
    "track_id",
    "streams",
    "position",
]

artist_growth_df = pd.read_csv(
    charts_path,
    usecols=artist_growth_usecols,
    low_memory=False
)

artist_growth_df["date"] = pd.to_datetime(
    artist_growth_df["date"],
    errors="coerce"
)

artist_growth_df["streams"] = pd.to_numeric(
    artist_growth_df["streams"],
    errors="coerce"
)

artist_growth_df["position"] = pd.to_numeric(
    artist_growth_df["position"],
    errors="coerce"
)

artist_growth_df = artist_growth_df[
    artist_growth_df["country"].isin(valid_country_codes)
].copy()

artist_growth_df = artist_growth_df.dropna(
    subset=[
        "date",
        "country",
        "artists",
    ]
)


# ------------------------------------------------------------
# 2. Aggregate artist-country-date performance
# ------------------------------------------------------------

artist_country_date_df = (
    artist_growth_df
    .groupby(
        [
            "artists",
            "country",
            "date",
        ]
    )
    .agg(
        daily_streams=("streams", "sum"),
        mean_chart_position=("position", "mean"),
        tracks_observed=("track_id", "nunique"),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Order artist-country histories
# ------------------------------------------------------------

artist_country_date_df = (
    artist_country_date_df
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

artist_group_keys = [
    "artists",
    "country",
]


# ------------------------------------------------------------
# 4. Calculate historical changes
# ------------------------------------------------------------

artist_country_date_df["previous_streams"] = (
    artist_country_date_df
    .groupby(artist_group_keys)["daily_streams"]
    .shift(1)
)

artist_country_date_df["previous_position"] = (
    artist_country_date_df
    .groupby(artist_group_keys)["mean_chart_position"]
    .shift(1)
)

artist_country_date_df["stream_change"] = (
    artist_country_date_df["daily_streams"]
    - artist_country_date_df["previous_streams"]
)

artist_country_date_df["stream_pct_change"] = (
    artist_country_date_df
    .groupby(artist_group_keys)["daily_streams"]
    .pct_change(fill_method=None)
)

artist_country_date_df["chart_improvement"] = (
    artist_country_date_df["previous_position"]
    - artist_country_date_df["mean_chart_position"]
)

artist_country_date_df["positive_stream_change"] = (
    artist_country_date_df["stream_change"] > 0
)

artist_country_date_df["positive_chart_change"] = (
    artist_country_date_df["chart_improvement"] > 0
)


# ------------------------------------------------------------
# 5. Aggregate artist-country growth histories
# ------------------------------------------------------------

artist_country_growth_df = (
    artist_country_date_df
    .groupby(artist_group_keys)
    .agg(
        observations=("date", "size"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        first_streams=("daily_streams", "first"),
        last_streams=("daily_streams", "last"),
        mean_streams=("daily_streams", "mean"),
        median_streams=("daily_streams", "median"),
        stream_change_mean=("stream_change", "mean"),
        stream_pct_change_median=("stream_pct_change", "median"),
        positive_stream_changes=("positive_stream_change", "sum"),
        first_position=("mean_chart_position", "first"),
        last_position=("mean_chart_position", "last"),
        mean_position=("mean_chart_position", "mean"),
        chart_improvement_mean=("chart_improvement", "mean"),
        positive_chart_changes=("positive_chart_change", "sum"),
        median_tracks_observed=("tracks_observed", "median"),
    )
    .reset_index()
)

artist_country_growth_df["history_span_days"] = (
    artist_country_growth_df["last_date"]
    - artist_country_growth_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 6. Calculate directional artist-growth evidence
# ------------------------------------------------------------

artist_country_growth_df["stream_growth_direction"] = (
    artist_country_growth_df["last_streams"]
    - artist_country_growth_df["first_streams"]
)

artist_country_growth_df["chart_growth_direction"] = (
    artist_country_growth_df["first_position"]
    - artist_country_growth_df["last_position"]
)

artist_country_growth_df["stream_growth_supported"] = (
    artist_country_growth_df["stream_growth_direction"] > 0
)

artist_country_growth_df["chart_growth_supported"] = (
    artist_country_growth_df["chart_growth_direction"] > 0
)

artist_country_growth_df["repeated_history"] = (
    artist_country_growth_df["unique_dates"] >= 2
)


# ------------------------------------------------------------
# 7. Artist-level overview
# ------------------------------------------------------------

artist_growth_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Artist-country histories",
            "Observed Value": len(artist_country_growth_df),
        },
        {
            "Metric": "Unique artist representations",
            "Observed Value": artist_country_growth_df[
                "artists"
            ].nunique(),
        },
        {
            "Metric": "Validated country markets",
            "Observed Value": artist_country_growth_df[
                "country"
            ].nunique(),
        },
        {
            "Metric": "Histories with repeated dates",
            "Observed Value": int(
                artist_country_growth_df[
                    "repeated_history"
                ].sum()
            ),
        },
        {
            "Metric": "Histories with positive stream direction",
            "Observed Value": int(
                artist_country_growth_df[
                    "stream_growth_supported"
                ].sum()
            ),
        },
        {
            "Metric": "Histories with positive chart direction",
            "Observed Value": int(
                artist_country_growth_df[
                    "chart_growth_supported"
                ].sum()
            ),
        },
        {
            "Metric": "Median history span days",
            "Observed Value": float(
                artist_country_growth_df[
                    "history_span_days"
                ].median()
            ),
        },
    ]
)

print()
print("Artist-level growth overview")
print("=" * 100)

display(artist_growth_summary_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

artist_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Artist-country histories assessed",
            "Observed Value": len(artist_country_growth_df),
        },
        {
            "Diagnostic": "Histories with at least 2 dates",
            "Observed Value": int(
                (
                    artist_country_growth_df[
                        "unique_dates"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 5 dates",
            "Observed Value": int(
                (
                    artist_country_growth_df[
                        "unique_dates"
                    ] >= 5
                ).sum()
            ),
        },
        {
            "Diagnostic": "Positive stream-growth histories",
            "Observed Value": int(
                artist_country_growth_df[
                    "stream_growth_supported"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Positive chart-growth histories",
            "Observed Value": int(
                artist_country_growth_df[
                    "chart_growth_supported"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories positive in both dimensions",
            "Observed Value": int(
                (
                    artist_country_growth_df[
                        "stream_growth_supported"
                    ]
                    &
                    artist_country_growth_df[
                        "chart_growth_supported"
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Artist-level growth diagnostics")
print("=" * 100)

display(artist_growth_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

artist_growth_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Artist population availability",
            "Requirement": (
                "Artist-level growth histories must be available."
            ),
            "Observed Evidence": (
                f"{artist_country_growth_df['artists'].nunique():,} "
                "artist representations assessed"
            ),
            "Passed": (
                artist_country_growth_df["artists"].nunique() > 0
            ),
        },
        {
            "Validation Area": "Country-market preservation",
            "Requirement": (
                "Artist growth analysis must retain validated "
                "country markets only."
            ),
            "Observed Evidence": (
                f"{artist_country_growth_df['country'].nunique():,} "
                "validated markets represented"
            ),
            "Passed": (
                artist_country_growth_df["country"].nunique()
                == len(valid_country_codes)
            ),
        },
        {
            "Validation Area": "Temporal ordering",
            "Requirement": (
                "Artist growth must be calculated from ordered "
                "historical observations."
            ),
            "Observed Evidence": (
                "Artist-country observations sorted by date"
            ),
            "Passed": True,
        },
        {
            "Validation Area": "Repeated history availability",
            "Requirement": (
                "At least some artist-country histories must "
                "contain repeated dates."
            ),
            "Observed Evidence": (
                f"{int(artist_country_growth_df['repeated_history'].sum()):,} "
                "histories contain repeated dates"
            ),
            "Passed": bool(
                artist_country_growth_df[
                    "repeated_history"
                ].any()
            ),
        },
        {
            "Validation Area": "Streaming growth evidence",
            "Requirement": (
                "Artist histories must provide measurable "
                "stream-growth direction."
            ),
            "Observed Evidence": (
                f"{int(artist_country_growth_df['stream_growth_supported'].sum()):,} "
                "positive stream-growth histories"
            ),
            "Passed": bool(
                artist_country_growth_df[
                    "stream_growth_direction"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Chart growth evidence",
            "Requirement": (
                "Artist histories must provide measurable "
                "chart-growth direction."
            ),
            "Observed Evidence": (
                f"{int(artist_country_growth_df['chart_growth_supported'].sum()):,} "
                "positive chart-growth histories"
            ),
            "Passed": bool(
                artist_country_growth_df[
                    "chart_growth_direction"
                ].notna().any()
            ),
        },
    ]
)

print()
print("Artist-level growth validation")
print("=" * 100)

display(artist_growth_validation_df)


section_3_4_complete = bool(
    artist_growth_validation_df["Passed"].all()
)

section_3_complete = bool(
    section_3_1_complete
    and section_3_2_complete
    and section_3_3_complete
    and section_3_4_complete
)

if not section_3_4_complete:
    raise AssertionError(
        "Section 3.4 Artist-Level Growth validation failed."
    )

print()
print(
    f"Section 3.4 completion status: "
    f"{section_3_4_complete}"
)

print(
    f"Section 3 overall completion status: "
    f"{section_3_complete}"
)

print(
    "Artist-level historical growth evidence has been prepared "
    "for PMIP market-growth detection."
)
```

#### Interpretation

Artist-level growth evidence was prepared for **49,843 artist representations** across **76 validated country markets**.

There are **61,730 artist-country histories with positive streaming growth**, while **32,289 show positive chart improvement**.

Section 3.4 and Section 3 are complete.

```python
print("Defining PMIP market-growth interpretation boundaries")
print("=" * 100)

# ------------------------------------------------------------
# 1. Define interpretation boundaries
# ------------------------------------------------------------

growth_interpretation_boundaries_df = pd.DataFrame(
    [
        {
            "Boundary Area": "Historical Evidence",
            "PMIP Can Support": (
                "Describe observed historical growth, decline, "
                "persistence and market movement."
            ),
            "PMIP Must Not Be Interpreted As": (
                "A guarantee that the same growth pattern will continue."
            ),
        },
        {
            "Boundary Area": "Causality",
            "PMIP Can Support": (
                "Identify associations between time, streams, charts "
                "and market activity."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Proof that a specific country, campaign, event or "
                "platform caused the observed growth."
            ),
        },
        {
            "Boundary Area": "Short-Term Spikes",
            "PMIP Can Support": (
                "Detect sharp temporary changes and compare them "
                "with sustained growth patterns."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Evidence that every temporary increase represents "
                "meaningful long-term growth."
            ),
        },
        {
            "Boundary Area": "Market Comparisons",
            "PMIP Can Support": (
                "Compare observed growth behaviour across validated "
                "country markets."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Proof that country markets are identical in size, "
                "coverage, audience or commercial value."
            ),
        },
        {
            "Boundary Area": "Artist and Track Identity",
            "PMIP Can Support": (
                "Analyse the markets where artists and tracks show "
                "observed performance growth."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Verified nationality, origin or home-market evidence."
            ),
        },
        {
            "Boundary Area": "Commercial Decisions",
            "PMIP Can Support": (
                "Provide structured growth signals for A&R and "
                "market-prioritisation review."
            ),
            "PMIP Must Not Be Interpreted As": (
                "An automatic instruction to sign, promote, invest "
                "or allocate budget."
            ),
        },
        {
            "Boundary Area": "Composite Growth Scores",
            "PMIP Can Support": (
                "Combine multiple validated growth dimensions into "
                "interpretable summary indicators."
            ),
            "PMIP Must Not Be Interpreted As": (
                "A universal measure of artist quality, popularity "
                "or commercial success."
            ),
        },
        {
            "Boundary Area": "Sparse Histories",
            "PMIP Can Support": (
                "Flag entities with limited historical evidence "
                "for cautious interpretation."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Strong evidence of growth or decline when temporal "
                "support is insufficient."
            ),
        },
        {
            "Boundary Area": "Volatility",
            "PMIP Can Support": (
                "Measure instability and distinguish volatile movement "
                "from more consistent performance."
            ),
            "PMIP Must Not Be Interpreted As": (
                "Meaningful sustained growth when changes are highly unstable."
            ),
        },
        {
            "Boundary Area": "Human Judgement",
            "PMIP Can Support": (
                "Provide transparent evidence to support expert review."
            ),
            "PMIP Must Not Be Interpreted As": (
                "A replacement for A&R, market, commercial or "
                "domain-expert judgement."
            ),
        },
    ]
)

print()
print("PMIP market-growth interpretation boundary framework")
print("=" * 100)

display(growth_interpretation_boundaries_df)


# ------------------------------------------------------------
# 2. Define responsible interpretation principles
# ------------------------------------------------------------

growth_responsible_use_df = pd.DataFrame(
    [
        {
            "Principle": "Observed evidence only",
            "Application": (
                "Interpret growth using validated historical PMIP data."
            ),
        },
        {
            "Principle": "No causal inference",
            "Application": (
                "Do not convert association into proof of cause."
            ),
        },
        {
            "Principle": "No guaranteed forecast",
            "Application": (
                "Historical growth must not be presented as certain "
                "future growth."
            ),
        },
        {
            "Principle": "Persistence before strength",
            "Application": (
                "Repeated growth evidence should carry more weight "
                "than isolated increases."
            ),
        },
        {
            "Principle": "Context preservation",
            "Application": (
                "Retain country coverage, history length and volatility "
                "during interpretation."
            ),
        },
        {
            "Principle": "Human review",
            "Application": (
                "High-priority growth signals must support rather than "
                "replace professional judgement."
            ),
        },
    ]
)

print()
print("Responsible market-growth interpretation principles")
print("=" * 100)

display(growth_responsible_use_df)


# ------------------------------------------------------------
# 3. Diagnostics
# ------------------------------------------------------------

growth_boundary_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Interpretation boundaries defined",
            "Observed Value": len(
                growth_interpretation_boundaries_df
            ),
        },
        {
            "Diagnostic": "Responsible-use principles defined",
            "Observed Value": len(
                growth_responsible_use_df
            ),
        },
        {
            "Diagnostic": "Causal boundary defined",
            "Observed Value": bool(
                "Causality"
                in growth_interpretation_boundaries_df[
                    "Boundary Area"
                ].values
            ),
        },
        {
            "Diagnostic": "Forecast boundary defined",
            "Observed Value": bool(
                "Historical Evidence"
                in growth_interpretation_boundaries_df[
                    "Boundary Area"
                ].values
            ),
        },
        {
            "Diagnostic": "Commercial decision boundary defined",
            "Observed Value": bool(
                "Commercial Decisions"
                in growth_interpretation_boundaries_df[
                    "Boundary Area"
                ].values
            ),
        },
        {
            "Diagnostic": "Human judgement retained",
            "Observed Value": bool(
                "Human Judgement"
                in growth_interpretation_boundaries_df[
                    "Boundary Area"
                ].values
            ),
        },
    ]
)

print()
print("Growth interpretation boundary diagnostics")
print("=" * 100)

display(growth_boundary_diagnostics_df)


# ------------------------------------------------------------
# 4. Validation
# ------------------------------------------------------------

growth_boundary_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Boundary completeness",
            "Requirement": (
                "Each interpretation boundary must define "
                "supported and restricted interpretation."
            ),
            "Observed Evidence": (
                f"{len(growth_interpretation_boundaries_df)} "
                "boundaries defined"
            ),
            "Passed": bool(
                growth_interpretation_boundaries_df[
                    [
                        "Boundary Area",
                        "PMIP Can Support",
                        "PMIP Must Not Be Interpreted As",
                    ]
                ]
                .notna()
                .all()
                .all()
            ),
        },
        {
            "Validation Area": "Responsible-use coverage",
            "Requirement": (
                "At least five responsible interpretation principles "
                "must be defined."
            ),
            "Observed Evidence": (
                f"{len(growth_responsible_use_df)} "
                "principles defined"
            ),
            "Passed": len(growth_responsible_use_df) >= 5,
        },
        {
            "Validation Area": "Critical boundary coverage",
            "Requirement": (
                "Causal, forecast, commercial and human-review "
                "boundaries must be represented."
            ),
            "Observed Evidence": (
                "Critical interpretation boundaries represented"
            ),
            "Passed": bool(
                {
                    "Causality",
                    "Historical Evidence",
                    "Commercial Decisions",
                    "Human Judgement",
                }.issubset(
                    set(
                        growth_interpretation_boundaries_df[
                            "Boundary Area"
                        ]
                    )
                )
            ),
        },
    ]
)

print()
print("Growth interpretation boundary validation")
print("=" * 100)

display(growth_boundary_validation_df)


section_3_6_complete = bool(
    growth_boundary_validation_df["Passed"].all()
)

section_3_complete = bool(
    section_3_1_complete
    and section_3_2_complete
    and section_3_3_complete
    and section_3_4_complete
    and section_3_6_complete
)

if not section_3_6_complete:
    raise AssertionError(
        "Section 3.6 Growth Interpretation Boundaries validation failed."
    )

print()
print(
    f"Section 3.6 completion status: "
    f"{section_3_6_complete}"
)

print(
    f"Section 3 overall completion status: "
    f"{section_3_complete}"
)

print(
    "PMIP market-growth interpretation boundaries "
    "have been defined successfully."
)
```

## 4. Historical Performance Preparation

This section prepares the historical PMIP chart data for reliable market-growth analysis.

The preparation focuses on temporal ordering, track-country and artist-country histories, missing observation periods, and minimum evidence requirements before growth metrics are calculated.

### 4.1 Temporal Ordering

#### Purpose

Ensure all historical observations are arranged in the correct chronological order within each track-country and artist-country history.

Correct temporal ordering is required before calculating changes, trends, persistence, or market-growth indicators.

```python
print("Preparing PMIP historical temporal ordering")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 3 completion
# ------------------------------------------------------------

if not section_3_complete:
    raise RuntimeError(
        "Section 3 must be complete before historical "
        "performance preparation begins."
    )

print(
    f"Section 3 overall completion status: "
    f"{section_3_complete}"
)


# ------------------------------------------------------------
# 2. Prepare validated historical source
# ------------------------------------------------------------

historical_preparation_df = (
    artist_growth_df[
        [
            "date",
            "country",
            "artists",
            "track_id",
            "streams",
            "position",
        ]
    ]
    .copy()
)

historical_preparation_df = historical_preparation_df.dropna(
    subset=[
        "date",
        "country",
        "track_id",
        "artists",
    ]
)

historical_preparation_df = historical_preparation_df[
    historical_preparation_df["country"].isin(
        valid_country_codes
    )
].copy()


# ------------------------------------------------------------
# 3. Check original ordering
# ------------------------------------------------------------

track_order_check_before = (
    historical_preparation_df[
        [
            "track_id",
            "country",
            "date",
        ]
    ]
    .reset_index(drop=True)
)

track_sorted_before = (
    track_order_check_before
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

track_order_already_valid = bool(
    track_order_check_before.equals(
        track_sorted_before
    )
)


# ------------------------------------------------------------
# 4. Apply canonical temporal ordering
# ------------------------------------------------------------

historical_preparation_df = (
    historical_preparation_df
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. Validate track-country temporal ordering
# ------------------------------------------------------------

track_temporal_validation = (
    historical_preparation_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )["date"]
    .apply(
        lambda x: x.is_monotonic_increasing
    )
)

track_histories_ordered = int(
    track_temporal_validation.sum()
)

track_histories_total = len(
    track_temporal_validation
)

track_temporal_order_valid = bool(
    track_temporal_validation.all()
)


# ------------------------------------------------------------
# 6. Prepare artist-country-date temporal source
# ------------------------------------------------------------

artist_temporal_df = (
    historical_preparation_df
    .groupby(
        [
            "artists",
            "country",
            "date",
        ],
        as_index=False
    )
    .agg(
        daily_streams=("streams", "sum"),
        mean_position=("position", "mean"),
        tracks_observed=("track_id", "nunique"),
    )
)

artist_temporal_df = (
    artist_temporal_df
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. Validate artist-country temporal ordering
# ------------------------------------------------------------

artist_temporal_validation = (
    artist_temporal_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )["date"]
    .apply(
        lambda x: x.is_monotonic_increasing
    )
)

artist_histories_ordered = int(
    artist_temporal_validation.sum()
)

artist_histories_total = len(
    artist_temporal_validation
)

artist_temporal_order_valid = bool(
    artist_temporal_validation.all()
)


# ------------------------------------------------------------
# 8. Temporal ordering overview
# ------------------------------------------------------------

temporal_ordering_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Historical observations prepared",
            "Observed Value": len(
                historical_preparation_df
            ),
        },
        {
            "Metric": "Validated country markets",
            "Observed Value": historical_preparation_df[
                "country"
            ].nunique(),
        },
        {
            "Metric": "Unique tracks",
            "Observed Value": historical_preparation_df[
                "track_id"
            ].nunique(),
        },
        {
            "Metric": "Unique artist representations",
            "Observed Value": historical_preparation_df[
                "artists"
            ].nunique(),
        },
        {
            "Metric": "Track-country histories",
            "Observed Value": track_histories_total,
        },
        {
            "Metric": "Artist-country histories",
            "Observed Value": artist_histories_total,
        },
    ]
)

print()
print("Temporal ordering overview")
print("=" * 100)

display(temporal_ordering_summary_df)


# ------------------------------------------------------------
# 9. Diagnostics
# ------------------------------------------------------------

temporal_ordering_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Original source already ordered",
            "Observed Value": track_order_already_valid,
        },
        {
            "Diagnostic": "Track-country histories ordered",
            "Observed Value": track_histories_ordered,
        },
        {
            "Diagnostic": "Track-country histories assessed",
            "Observed Value": track_histories_total,
        },
        {
            "Diagnostic": "Track temporal ordering valid",
            "Observed Value": track_temporal_order_valid,
        },
        {
            "Diagnostic": "Artist-country histories ordered",
            "Observed Value": artist_histories_ordered,
        },
        {
            "Diagnostic": "Artist-country histories assessed",
            "Observed Value": artist_histories_total,
        },
        {
            "Diagnostic": "Artist temporal ordering valid",
            "Observed Value": artist_temporal_order_valid,
        },
    ]
)

print()
print("Temporal ordering diagnostics")
print("=" * 100)

display(temporal_ordering_diagnostics_df)


# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

section_4_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 3 completion",
            "Requirement": (
                "Market Growth Definition must already "
                "be complete."
            ),
            "Observed Evidence": (
                f"Section 3 overall completion status: "
                f"{section_3_complete}"
            ),
            "Passed": section_3_complete,
        },
        {
            "Validation Area": "Country-market preservation",
            "Requirement": (
                "Historical preparation must retain only "
                "validated country markets."
            ),
            "Observed Evidence": (
                f"{historical_preparation_df['country'].nunique():,} "
                "validated markets retained"
            ),
            "Passed": (
                historical_preparation_df[
                    "country"
                ].nunique()
                == len(valid_country_codes)
            ),
        },
        {
            "Validation Area": "Track temporal ordering",
            "Requirement": (
                "Every track-country history must be "
                "chronologically ordered."
            ),
            "Observed Evidence": (
                f"{track_histories_ordered:,} of "
                f"{track_histories_total:,} histories ordered"
            ),
            "Passed": track_temporal_order_valid,
        },
        {
            "Validation Area": "Artist temporal ordering",
            "Requirement": (
                "Every artist-country history must be "
                "chronologically ordered."
            ),
            "Observed Evidence": (
                f"{artist_histories_ordered:,} of "
                f"{artist_histories_total:,} histories ordered"
            ),
            "Passed": artist_temporal_order_valid,
        },
        {
            "Validation Area": "Date completeness",
            "Requirement": (
                "Prepared historical observations must "
                "contain valid dates."
            ),
            "Observed Evidence": (
                f"{historical_preparation_df['date'].notna().sum():,} "
                "valid dated observations"
            ),
            "Passed": bool(
                historical_preparation_df[
                    "date"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Temporal ordering validation")
print("=" * 100)

display(section_4_1_validation_df)


section_4_1_complete = bool(
    section_4_1_validation_df[
        "Passed"
    ].all()
)

if not section_4_1_complete:
    raise AssertionError(
        "Section 4.1 Temporal Ordering validation failed."
    )

print()
print(
    f"Section 4.1 completion status: "
    f"{section_4_1_complete}"
)

print(
    "Historical observations are correctly ordered "
    "for track-country and artist-country analysis."
)
```

#### Interpretation

The historical PMIP data was successfully reordered into valid chronological sequence.

All **353,282 track-country histories** and **184,681 artist-country histories** are now correctly ordered by date across **76 validated country markets**.

Section 4.1 is complete.

### 4.2 Track-Country Histories

#### Purpose

Construct validated track-country histories that summarise each track's observed performance within each country market.

These histories will provide the structured temporal units required for later growth, persistence, volatility, and trend analysis.

```python
print("Preparing PMIP track-country historical profiles")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 4.1 completion
# ------------------------------------------------------------

if not section_4_1_complete:
    raise RuntimeError(
        "Section 4.1 Temporal Ordering must be complete "
        "before track-country histories are prepared."
    )

print(
    f"Section 4.1 completion status: "
    f"{section_4_1_complete}"
)


# ------------------------------------------------------------
# 2. Aggregate track-country histories
# ------------------------------------------------------------

track_country_history_df = (
    historical_preparation_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .agg(
        observations=("date", "size"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),

        first_streams=("streams", "first"),
        last_streams=("streams", "last"),
        mean_streams=("streams", "mean"),
        median_streams=("streams", "median"),
        min_streams=("streams", "min"),
        max_streams=("streams", "max"),

        first_position=("position", "first"),
        last_position=("position", "last"),
        mean_position=("position", "mean"),
        median_position=("position", "median"),
        best_position=("position", "min"),
        worst_position=("position", "max"),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. History duration
# ------------------------------------------------------------

track_country_history_df["history_span_days"] = (
    track_country_history_df["last_date"]
    - track_country_history_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 4. Observation-density indicators
# ------------------------------------------------------------

track_country_history_df["repeated_history"] = (
    track_country_history_df["unique_dates"] >= 2
)

track_country_history_df["five_date_history"] = (
    track_country_history_df["unique_dates"] >= 5
)

track_country_history_df["ten_date_history"] = (
    track_country_history_df["unique_dates"] >= 10
)


# ------------------------------------------------------------
# 5. Track-country history overview
# ------------------------------------------------------------

track_country_history_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Track-country histories",
            "Observed Value": len(
                track_country_history_df
            ),
        },
        {
            "Metric": "Unique tracks",
            "Observed Value": track_country_history_df[
                "track_id"
            ].nunique(),
        },
        {
            "Metric": "Validated country markets",
            "Observed Value": track_country_history_df[
                "country"
            ].nunique(),
        },
        {
            "Metric": "Median observations per history",
            "Observed Value": float(
                track_country_history_df[
                    "observations"
                ].median()
            ),
        },
        {
            "Metric": "Median unique dates per history",
            "Observed Value": float(
                track_country_history_df[
                    "unique_dates"
                ].median()
            ),
        },
        {
            "Metric": "Median history span days",
            "Observed Value": float(
                track_country_history_df[
                    "history_span_days"
                ].median()
            ),
        },
    ]
)

print()
print("Track-country history overview")
print("=" * 100)

display(track_country_history_summary_df)


# ------------------------------------------------------------
# 6. History-length distribution
# ------------------------------------------------------------

track_country_history_distribution_df = pd.DataFrame(
    [
        {
            "History Group": "Single-date history",
            "Track-Country Histories": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] == 1
                ).sum()
            ),
        },
        {
            "History Group": "At least 2 dates",
            "Track-Country Histories": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "History Group": "At least 5 dates",
            "Track-Country Histories": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 5
                ).sum()
            ),
        },
        {
            "History Group": "At least 10 dates",
            "Track-Country Histories": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 10
                ).sum()
            ),
        },
        {
            "History Group": "At least 25 dates",
            "Track-Country Histories": int(
                (
                    track_country_history_df[
                        "unique_dates"
                    ] >= 25
                ).sum()
            ),
        },
    ]
)

track_country_history_distribution_df["Share (%)"] = (
    track_country_history_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_country_history_df)
    * 100
)

print()
print("Track-country history-length distribution")
print("=" * 100)

display(track_country_history_distribution_df)


# ------------------------------------------------------------
# 7. Diagnostics
# ------------------------------------------------------------

track_country_history_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track-country histories prepared",
            "Observed Value": len(
                track_country_history_df
            ),
        },
        {
            "Diagnostic": "Unique tracks retained",
            "Observed Value": track_country_history_df[
                "track_id"
            ].nunique(),
        },
        {
            "Diagnostic": "Country markets retained",
            "Observed Value": track_country_history_df[
                "country"
            ].nunique(),
        },
        {
            "Diagnostic": "Histories with at least 2 dates",
            "Observed Value": int(
                track_country_history_df[
                    "repeated_history"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 5 dates",
            "Observed Value": int(
                track_country_history_df[
                    "five_date_history"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 10 dates",
            "Observed Value": int(
                track_country_history_df[
                    "ten_date_history"
                ].sum()
            ),
        },
    ]
)

print()
print("Track-country history diagnostics")
print("=" * 100)

display(track_country_history_diagnostics_df)


# ------------------------------------------------------------
# 8. Validation
# ------------------------------------------------------------

section_4_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 4.1 completion",
            "Requirement": (
                "Temporal Ordering must already be complete."
            ),
            "Observed Evidence": (
                f"Section 4.1 completion status: "
                f"{section_4_1_complete}"
            ),
            "Passed": section_4_1_complete,
        },
        {
            "Validation Area": "Track population preservation",
            "Requirement": (
                "All validated tracks must remain represented."
            ),
            "Observed Evidence": (
                f"{track_country_history_df['track_id'].nunique():,} "
                "unique tracks represented"
            ),
            "Passed": (
                track_country_history_df[
                    "track_id"
                ].nunique()
                == historical_preparation_df[
                    "track_id"
                ].nunique()
            ),
        },
        {
            "Validation Area": "Country-market preservation",
            "Requirement": (
                "All validated country markets must remain represented."
            ),
            "Observed Evidence": (
                f"{track_country_history_df['country'].nunique():,} "
                "country markets represented"
            ),
            "Passed": (
                track_country_history_df[
                    "country"
                ].nunique()
                == len(valid_country_codes)
            ),
        },
        {
            "Validation Area": "Track-country uniqueness",
            "Requirement": (
                "Each track-country pair must appear exactly once."
            ),
            "Observed Evidence": (
                f"{len(track_country_history_df):,} unique "
                "track-country histories"
            ),
            "Passed": bool(
                ~track_country_history_df.duplicated(
                    subset=[
                        "track_id",
                        "country",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "History date validity",
            "Requirement": (
                "Each history must contain valid first and last dates."
            ),
            "Observed Evidence": (
                f"{track_country_history_df['first_date'].notna().sum():,} "
                "histories with valid dates"
            ),
            "Passed": bool(
                track_country_history_df[
                    [
                        "first_date",
                        "last_date",
                    ]
                ]
                .notna()
                .all()
                .all()
            ),
        },
        {
            "Validation Area": "Repeated-history availability",
            "Requirement": (
                "At least some track-country histories must contain "
                "multiple observation dates."
            ),
            "Observed Evidence": (
                f"{int(track_country_history_df['repeated_history'].sum()):,} "
                "histories contain at least 2 dates"
            ),
            "Passed": bool(
                track_country_history_df[
                    "repeated_history"
                ].any()
            ),
        },
    ]
)

print()
print("Track-country history validation")
print("=" * 100)

display(section_4_2_validation_df)


section_4_2_complete = bool(
    section_4_2_validation_df[
        "Passed"
    ].all()
)

if not section_4_2_complete:
    raise AssertionError(
        "Section 4.2 Track-Country Histories "
        "validation failed."
    )

print()
print(
    f"Section 4.2 completion status: "
    f"{section_4_2_complete}"
)

print(
    "Track-country historical profiles are prepared "
    "for subsequent PMIP growth analysis."
)
```

#### Interpretation

The dataset contains **353,282 unique track-country histories** across **110,198 tracks** and **76 validated country markets**.

Most histories contain repeated observations, with **258,493 histories having at least two dates**, which provides a strong base for later growth analysis.

Section 4.2 is complete.

### 4.3 Artist-Country Histories

#### Purpose

Construct artist-country historical profiles by aggregating artist performance across tracks within each validated country market.

These histories will support later analysis of artist growth, persistence, volatility, and emerging-market behaviour.

```python
print("Preparing PMIP artist-country historical profiles")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 4.2 completion
# ------------------------------------------------------------

if not section_4_2_complete:
    raise RuntimeError(
        "Section 4.2 Track-Country Histories must be complete "
        "before artist-country histories are prepared."
    )

print(
    f"Section 4.2 completion status: "
    f"{section_4_2_complete}"
)


# ------------------------------------------------------------
# 2. Build artist-country-date histories
# ------------------------------------------------------------

artist_country_date_history_df = (
    historical_preparation_df
    .groupby(
        [
            "artists",
            "country",
            "date",
        ]
    )
    .agg(
        daily_streams=("streams", "sum"),
        mean_position=("position", "mean"),
        median_position=("position", "median"),
        best_position=("position", "min"),
        worst_position=("position", "max"),
        tracks_observed=("track_id", "nunique"),
        observations=("track_id", "size"),
    )
    .reset_index()
)

artist_country_date_history_df = (
    artist_country_date_history_df
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Aggregate artist-country histories
# ------------------------------------------------------------

artist_country_history_df = (
    artist_country_date_history_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )
    .agg(
        observations=("date", "size"),
        unique_dates=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),

        first_streams=("daily_streams", "first"),
        last_streams=("daily_streams", "last"),
        mean_streams=("daily_streams", "mean"),
        median_streams=("daily_streams", "median"),
        min_streams=("daily_streams", "min"),
        max_streams=("daily_streams", "max"),

        first_position=("mean_position", "first"),
        last_position=("mean_position", "last"),
        mean_position=("mean_position", "mean"),
        median_position=("median_position", "median"),
        best_position=("best_position", "min"),
        worst_position=("worst_position", "max"),

        median_tracks_observed=("tracks_observed", "median"),
        max_tracks_observed=("tracks_observed", "max"),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 4. History duration
# ------------------------------------------------------------

artist_country_history_df["history_span_days"] = (
    artist_country_history_df["last_date"]
    - artist_country_history_df["first_date"]
).dt.days


# ------------------------------------------------------------
# 5. Observation-density indicators
# ------------------------------------------------------------

artist_country_history_df["repeated_history"] = (
    artist_country_history_df["unique_dates"] >= 2
)

artist_country_history_df["five_date_history"] = (
    artist_country_history_df["unique_dates"] >= 5
)

artist_country_history_df["ten_date_history"] = (
    artist_country_history_df["unique_dates"] >= 10
)


# ------------------------------------------------------------
# 6. Artist-country history overview
# ------------------------------------------------------------

artist_country_history_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Artist-country histories",
            "Observed Value": len(
                artist_country_history_df
            ),
        },
        {
            "Metric": "Unique artist representations",
            "Observed Value": artist_country_history_df[
                "artists"
            ].nunique(),
        },
        {
            "Metric": "Validated country markets",
            "Observed Value": artist_country_history_df[
                "country"
            ].nunique(),
        },
        {
            "Metric": "Median observations per history",
            "Observed Value": float(
                artist_country_history_df[
                    "observations"
                ].median()
            ),
        },
        {
            "Metric": "Median unique dates per history",
            "Observed Value": float(
                artist_country_history_df[
                    "unique_dates"
                ].median()
            ),
        },
        {
            "Metric": "Median history span days",
            "Observed Value": float(
                artist_country_history_df[
                    "history_span_days"
                ].median()
            ),
        },
    ]
)

print()
print("Artist-country history overview")
print("=" * 100)

display(artist_country_history_summary_df)


# ------------------------------------------------------------
# 7. History-length distribution
# ------------------------------------------------------------

artist_country_history_distribution_df = pd.DataFrame(
    [
        {
            "History Group": "Single-date history",
            "Artist-Country Histories": int(
                (
                    artist_country_history_df[
                        "unique_dates"
                    ] == 1
                ).sum()
            ),
        },
        {
            "History Group": "At least 2 dates",
            "Artist-Country Histories": int(
                (
                    artist_country_history_df[
                        "unique_dates"
                    ] >= 2
                ).sum()
            ),
        },
        {
            "History Group": "At least 5 dates",
            "Artist-Country Histories": int(
                (
                    artist_country_history_df[
                        "unique_dates"
                    ] >= 5
                ).sum()
            ),
        },
        {
            "History Group": "At least 10 dates",
            "Artist-Country Histories": int(
                (
                    artist_country_history_df[
                        "unique_dates"
                    ] >= 10
                ).sum()
            ),
        },
        {
            "History Group": "At least 25 dates",
            "Artist-Country Histories": int(
                (
                    artist_country_history_df[
                        "unique_dates"
                    ] >= 25
                ).sum()
            ),
        },
    ]
)

artist_country_history_distribution_df["Share (%)"] = (
    artist_country_history_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_country_history_df)
    * 100
)

print()
print("Artist-country history-length distribution")
print("=" * 100)

display(artist_country_history_distribution_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

artist_country_history_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Artist-country histories prepared",
            "Observed Value": len(
                artist_country_history_df
            ),
        },
        {
            "Diagnostic": "Unique artist representations retained",
            "Observed Value": artist_country_history_df[
                "artists"
            ].nunique(),
        },
        {
            "Diagnostic": "Country markets retained",
            "Observed Value": artist_country_history_df[
                "country"
            ].nunique(),
        },
        {
            "Diagnostic": "Histories with at least 2 dates",
            "Observed Value": int(
                artist_country_history_df[
                    "repeated_history"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 5 dates",
            "Observed Value": int(
                artist_country_history_df[
                    "five_date_history"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Histories with at least 10 dates",
            "Observed Value": int(
                artist_country_history_df[
                    "ten_date_history"
                ].sum()
            ),
        },
    ]
)

print()
print("Artist-country history diagnostics")
print("=" * 100)

display(artist_country_history_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_4_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 4.2 completion",
            "Requirement": (
                "Track-Country Histories must already be complete."
            ),
            "Observed Evidence": (
                f"Section 4.2 completion status: "
                f"{section_4_2_complete}"
            ),
            "Passed": section_4_2_complete,
        },
        {
            "Validation Area": "Artist population preservation",
            "Requirement": (
                "All validated artist representations must remain represented."
            ),
            "Observed Evidence": (
                f"{artist_country_history_df['artists'].nunique():,} "
                "artist representations retained"
            ),
            "Passed": (
                artist_country_history_df[
                    "artists"
                ].nunique()
                == historical_preparation_df[
                    "artists"
                ].nunique()
            ),
        },
        {
            "Validation Area": "Country-market preservation",
            "Requirement": (
                "All validated country markets must remain represented."
            ),
            "Observed Evidence": (
                f"{artist_country_history_df['country'].nunique():,} "
                "country markets represented"
            ),
            "Passed": (
                artist_country_history_df[
                    "country"
                ].nunique()
                == len(valid_country_codes)
            ),
        },
        {
            "Validation Area": "Artist-country uniqueness",
            "Requirement": (
                "Each artist-country pair must appear exactly once."
            ),
            "Observed Evidence": (
                f"{len(artist_country_history_df):,} unique "
                "artist-country histories"
            ),
            "Passed": bool(
                ~artist_country_history_df.duplicated(
                    subset=[
                        "artists",
                        "country",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "History date validity",
            "Requirement": (
                "Each artist-country history must contain "
                "valid first and last dates."
            ),
            "Observed Evidence": (
                f"{artist_country_history_df['first_date'].notna().sum():,} "
                "histories with valid dates"
            ),
            "Passed": bool(
                artist_country_history_df[
                    [
                        "first_date",
                        "last_date",
                    ]
                ]
                .notna()
                .all()
                .all()
            ),
        },
        {
            "Validation Area": "Repeated-history availability",
            "Requirement": (
                "At least some artist-country histories must contain "
                "multiple observation dates."
            ),
            "Observed Evidence": (
                f"{int(artist_country_history_df['repeated_history'].sum()):,} "
                "histories contain at least 2 dates"
            ),
            "Passed": bool(
                artist_country_history_df[
                    "repeated_history"
                ].any()
            ),
        },
    ]
)

print()
print("Artist-country history validation")
print("=" * 100)

display(section_4_3_validation_df)


section_4_3_complete = bool(
    section_4_3_validation_df[
        "Passed"
    ].all()
)

if not section_4_3_complete:
    raise AssertionError(
        "Section 4.3 Artist-Country Histories "
        "validation failed."
    )

print()
print(
    f"Section 4.3 completion status: "
    f"{section_4_3_complete}"
)

print(
    "Artist-country historical profiles are prepared "
    "for subsequent PMIP growth analysis."
)
```

#### Interpretation

The dataset contains **184,681 artist-country histories** across **49,843 artist representations** and **76 validated country markets**.

Most artist-country histories contain repeated observations, with **148,700 histories having at least two dates**, providing a strong base for later growth analysis.

Section 4.3 is complete.

### 4.4 Missing Period Assessment

#### Purpose

Assess gaps between consecutive historical observations within track-country and artist-country histories.

This helps identify irregular observation periods that could affect the reliability of growth, persistence, and trend calculations.

```python
print("Assessing PMIP missing historical observation periods")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 4.3 completion
# ------------------------------------------------------------

if not section_4_3_complete:
    raise RuntimeError(
        "Section 4.3 Artist-Country Histories must be complete "
        "before missing-period assessment begins."
    )

print(
    f"Section 4.3 completion status: "
    f"{section_4_3_complete}"
)


# ------------------------------------------------------------
# 2. Track-country observation gaps
# ------------------------------------------------------------

track_gap_df = (
    historical_preparation_df[
        [
            "track_id",
            "country",
            "date",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

track_gap_df["previous_date"] = (
    track_gap_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )["date"]
    .shift(1)
)

track_gap_df["gap_days"] = (
    track_gap_df["date"]
    - track_gap_df["previous_date"]
).dt.days

track_valid_gaps = (
    track_gap_df["gap_days"]
    .dropna()
)


# ------------------------------------------------------------
# 3. Artist-country observation gaps
# ------------------------------------------------------------

artist_gap_df = (
    artist_country_date_history_df[
        [
            "artists",
            "country",
            "date",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

artist_gap_df["previous_date"] = (
    artist_gap_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )["date"]
    .shift(1)
)

artist_gap_df["gap_days"] = (
    artist_gap_df["date"]
    - artist_gap_df["previous_date"]
).dt.days

artist_valid_gaps = (
    artist_gap_df["gap_days"]
    .dropna()
)


# ------------------------------------------------------------
# 4. Define gap categories
# ------------------------------------------------------------

def classify_gap(gap):
    if pd.isna(gap):
        return "No Previous Observation"
    elif gap <= 7:
        return "0-7 Days"
    elif gap <= 14:
        return "8-14 Days"
    elif gap <= 30:
        return "15-30 Days"
    elif gap <= 90:
        return "31-90 Days"
    else:
        return "More Than 90 Days"


track_gap_df["gap_class"] = (
    track_gap_df["gap_days"]
    .apply(classify_gap)
)

artist_gap_df["gap_class"] = (
    artist_gap_df["gap_days"]
    .apply(classify_gap)
)


# ------------------------------------------------------------
# 5. Track-country gap distribution
# ------------------------------------------------------------

track_gap_distribution_df = (
    track_gap_df[
        track_gap_df["gap_days"].notna()
    ]["gap_class"]
    .value_counts()
    .rename_axis("Gap Class")
    .reset_index(name="Track Observation Gaps")
)

track_gap_distribution_df["Share (%)"] = (
    track_gap_distribution_df[
        "Track Observation Gaps"
    ]
    / track_gap_distribution_df[
        "Track Observation Gaps"
    ].sum()
    * 100
)

print()
print("Track-country observation-gap distribution")
print("=" * 100)

display(track_gap_distribution_df)


# ------------------------------------------------------------
# 6. Artist-country gap distribution
# ------------------------------------------------------------

artist_gap_distribution_df = (
    artist_gap_df[
        artist_gap_df["gap_days"].notna()
    ]["gap_class"]
    .value_counts()
    .rename_axis("Gap Class")
    .reset_index(name="Artist Observation Gaps")
)

artist_gap_distribution_df["Share (%)"] = (
    artist_gap_distribution_df[
        "Artist Observation Gaps"
    ]
    / artist_gap_distribution_df[
        "Artist Observation Gaps"
    ].sum()
    * 100
)

print()
print("Artist-country observation-gap distribution")
print("=" * 100)

display(artist_gap_distribution_df)


# ------------------------------------------------------------
# 7. Missing-period overview
# ------------------------------------------------------------

missing_period_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Track-country gaps assessed",
            "Observed Value": len(track_valid_gaps),
        },
        {
            "Metric": "Median track gap days",
            "Observed Value": float(
                track_valid_gaps.median()
            ),
        },
        {
            "Metric": "Maximum track gap days",
            "Observed Value": int(
                track_valid_gaps.max()
            ),
        },
        {
            "Metric": "Track gaps > 30 days",
            "Observed Value": int(
                (track_valid_gaps > 30).sum()
            ),
        },
        {
            "Metric": "Artist-country gaps assessed",
            "Observed Value": len(artist_valid_gaps),
        },
        {
            "Metric": "Median artist gap days",
            "Observed Value": float(
                artist_valid_gaps.median()
            ),
        },
        {
            "Metric": "Maximum artist gap days",
            "Observed Value": int(
                artist_valid_gaps.max()
            ),
        },
        {
            "Metric": "Artist gaps > 30 days",
            "Observed Value": int(
                (artist_valid_gaps > 30).sum()
            ),
        },
    ]
)

print()
print("Missing-period assessment overview")
print("=" * 100)

display(missing_period_summary_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

missing_period_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track-country histories assessed",
            "Observed Value": track_gap_df[
                ["track_id", "country"]
            ].drop_duplicates().shape[0],
        },
        {
            "Diagnostic": "Artist-country histories assessed",
            "Observed Value": artist_gap_df[
                ["artists", "country"]
            ].drop_duplicates().shape[0],
        },
        {
            "Diagnostic": "Track observation gaps calculated",
            "Observed Value": len(track_valid_gaps),
        },
        {
            "Diagnostic": "Artist observation gaps calculated",
            "Observed Value": len(artist_valid_gaps),
        },
        {
            "Diagnostic": "Track gaps above 90 days",
            "Observed Value": int(
                (track_valid_gaps > 90).sum()
            ),
        },
        {
            "Diagnostic": "Artist gaps above 90 days",
            "Observed Value": int(
                (artist_valid_gaps > 90).sum()
            ),
        },
    ]
)

print()
print("Missing-period diagnostics")
print("=" * 100)

display(missing_period_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_4_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 4.3 completion",
            "Requirement": (
                "Artist-Country Histories must already be complete."
            ),
            "Observed Evidence": (
                f"Section 4.3 completion status: "
                f"{section_4_3_complete}"
            ),
            "Passed": section_4_3_complete,
        },
        {
            "Validation Area": "Track gap availability",
            "Requirement": (
                "Repeated track-country histories must provide "
                "measurable observation gaps."
            ),
            "Observed Evidence": (
                f"{len(track_valid_gaps):,} track gaps calculated"
            ),
            "Passed": len(track_valid_gaps) > 0,
        },
        {
            "Validation Area": "Artist gap availability",
            "Requirement": (
                "Repeated artist-country histories must provide "
                "measurable observation gaps."
            ),
            "Observed Evidence": (
                f"{len(artist_valid_gaps):,} artist gaps calculated"
            ),
            "Passed": len(artist_valid_gaps) > 0,
        },
        {
            "Validation Area": "Track gap numeric validity",
            "Requirement": (
                "Track observation gaps must be non-negative."
            ),
            "Observed Evidence": (
                f"Minimum track gap: "
                f"{int(track_valid_gaps.min())} days"
            ),
            "Passed": bool(
                (track_valid_gaps >= 0).all()
            ),
        },
        {
            "Validation Area": "Artist gap numeric validity",
            "Requirement": (
                "Artist observation gaps must be non-negative."
            ),
            "Observed Evidence": (
                f"Minimum artist gap: "
                f"{int(artist_valid_gaps.min())} days"
            ),
            "Passed": bool(
                (artist_valid_gaps >= 0).all()
            ),
        },
    ]
)

print()
print("Missing-period assessment validation")
print("=" * 100)

display(section_4_4_validation_df)


section_4_4_complete = bool(
    section_4_4_validation_df[
        "Passed"
    ].all()
)

if not section_4_4_complete:
    raise AssertionError(
        "Section 4.4 Missing Period Assessment "
        "validation failed."
    )

print()
print(
    f"Section 4.4 completion status: "
    f"{section_4_4_complete}"
)

print(
    "Historical observation gaps have been assessed "
    "for PMIP track-country and artist-country histories."
)
```

#### Interpretation

Most observation gaps are short, with a **median gap of 7 days** for both track-country and artist-country histories.

However, some histories contain gaps longer than 90 days, so later growth calculations should include minimum observation requirements before strong conclusions are made.

Section 4.4 is complete.

### 4.5 Observation Sufficiency Rules

#### Purpose

Define the minimum historical evidence required before PMIP classifies track-country or artist-country growth.

These rules prevent sparse histories from being treated as reliable evidence of sustained market growth.

```python
print("Defining PMIP observation sufficiency rules")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 4.4 completion
# ------------------------------------------------------------

if not section_4_4_complete:
    raise RuntimeError(
        "Section 4.4 Missing Period Assessment must be complete "
        "before observation sufficiency rules are defined."
    )

print(
    f"Section 4.4 completion status: "
    f"{section_4_4_complete}"
)


# ------------------------------------------------------------
# 2. Define sufficiency thresholds
# ------------------------------------------------------------

MIN_DATES_BASIC = 2
MIN_DATES_GROWTH = 5
MIN_DATES_STRONG = 10

MAX_MEDIAN_GAP_BASIC = 90
MAX_MEDIAN_GAP_STRONG = 30


observation_sufficiency_rules_df = pd.DataFrame(
    [
        {
            "Evidence Level": "Insufficient",
            "Minimum Dates": 1,
            "Maximum Median Gap Days": None,
            "Interpretation": (
                "Not enough repeated historical evidence "
                "for reliable market-growth assessment."
            ),
        },
        {
            "Evidence Level": "Basic",
            "Minimum Dates": MIN_DATES_BASIC,
            "Maximum Median Gap Days": MAX_MEDIAN_GAP_BASIC,
            "Interpretation": (
                "Enough evidence for limited directional analysis."
            ),
        },
        {
            "Evidence Level": "Growth Eligible",
            "Minimum Dates": MIN_DATES_GROWTH,
            "Maximum Median Gap Days": MAX_MEDIAN_GAP_BASIC,
            "Interpretation": (
                "Suitable for standard PMIP growth analysis."
            ),
        },
        {
            "Evidence Level": "Strong Historical Support",
            "Minimum Dates": MIN_DATES_STRONG,
            "Maximum Median Gap Days": MAX_MEDIAN_GAP_STRONG,
            "Interpretation": (
                "Provides stronger evidence for sustained-growth "
                "interpretation."
            ),
        },
    ]
)

print()
print("PMIP observation sufficiency framework")
print("=" * 100)

display(observation_sufficiency_rules_df)


# ------------------------------------------------------------
# 3. Track-country gap summaries
# ------------------------------------------------------------

track_gap_summary_df = (
    track_gap_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .agg(
        median_gap_days=("gap_days", "median"),
        max_gap_days=("gap_days", "max"),
    )
    .reset_index()
)

track_sufficiency_df = (
    track_country_history_df
    .merge(
        track_gap_summary_df,
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 4. Artist-country gap summaries
# ------------------------------------------------------------

artist_gap_summary_df = (
    artist_gap_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )
    .agg(
        median_gap_days=("gap_days", "median"),
        max_gap_days=("gap_days", "max"),
    )
    .reset_index()
)

artist_sufficiency_df = (
    artist_country_history_df
    .merge(
        artist_gap_summary_df,
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 5. Classify track-country evidence sufficiency
# ------------------------------------------------------------

def classify_observation_sufficiency(
    unique_dates,
    median_gap_days
):
    if unique_dates < MIN_DATES_BASIC:
        return "Insufficient"

    if (
        unique_dates >= MIN_DATES_STRONG
        and pd.notna(median_gap_days)
        and median_gap_days <= MAX_MEDIAN_GAP_STRONG
    ):
        return "Strong Historical Support"

    if (
        unique_dates >= MIN_DATES_GROWTH
        and pd.notna(median_gap_days)
        and median_gap_days <= MAX_MEDIAN_GAP_BASIC
    ):
        return "Growth Eligible"

    return "Basic"


track_sufficiency_df["evidence_level"] = (
    track_sufficiency_df.apply(
        lambda row: classify_observation_sufficiency(
            row["unique_dates"],
            row["median_gap_days"],
        ),
        axis=1,
    )
)

artist_sufficiency_df["evidence_level"] = (
    artist_sufficiency_df.apply(
        lambda row: classify_observation_sufficiency(
            row["unique_dates"],
            row["median_gap_days"],
        ),
        axis=1,
    )
)


# ------------------------------------------------------------
# 6. Track sufficiency distribution
# ------------------------------------------------------------

track_sufficiency_distribution_df = (
    track_sufficiency_df[
        "evidence_level"
    ]
    .value_counts()
    .rename_axis("Evidence Level")
    .reset_index(name="Track-Country Histories")
)

track_sufficiency_distribution_df["Share (%)"] = (
    track_sufficiency_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_sufficiency_df)
    * 100
)

print()
print("Track-country observation sufficiency")
print("=" * 100)

display(track_sufficiency_distribution_df)


# ------------------------------------------------------------
# 7. Artist sufficiency distribution
# ------------------------------------------------------------

artist_sufficiency_distribution_df = (
    artist_sufficiency_df[
        "evidence_level"
    ]
    .value_counts()
    .rename_axis("Evidence Level")
    .reset_index(name="Artist-Country Histories")
)

artist_sufficiency_distribution_df["Share (%)"] = (
    artist_sufficiency_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_sufficiency_df)
    * 100
)

print()
print("Artist-country observation sufficiency")
print("=" * 100)

display(artist_sufficiency_distribution_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

observation_sufficiency_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track-country histories assessed",
            "Observed Value": len(track_sufficiency_df),
        },
        {
            "Diagnostic": "Track histories growth eligible",
            "Observed Value": int(
                track_sufficiency_df[
                    "evidence_level"
                ].isin(
                    [
                        "Growth Eligible",
                        "Strong Historical Support",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Track histories insufficient",
            "Observed Value": int(
                (
                    track_sufficiency_df[
                        "evidence_level"
                    ] == "Insufficient"
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist-country histories assessed",
            "Observed Value": len(artist_sufficiency_df),
        },
        {
            "Diagnostic": "Artist histories growth eligible",
            "Observed Value": int(
                artist_sufficiency_df[
                    "evidence_level"
                ].isin(
                    [
                        "Growth Eligible",
                        "Strong Historical Support",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories insufficient",
            "Observed Value": int(
                (
                    artist_sufficiency_df[
                        "evidence_level"
                    ] == "Insufficient"
                ).sum()
            ),
        },
    ]
)

print()
print("Observation sufficiency diagnostics")
print("=" * 100)

display(observation_sufficiency_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_4_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 4.4 completion",
            "Requirement": (
                "Missing Period Assessment must already be complete."
            ),
            "Observed Evidence": (
                f"Section 4.4 completion status: "
                f"{section_4_4_complete}"
            ),
            "Passed": section_4_4_complete,
        },
        {
            "Validation Area": "Sufficiency rule completeness",
            "Requirement": (
                "Observation sufficiency levels must be defined."
            ),
            "Observed Evidence": (
                f"{len(observation_sufficiency_rules_df)} "
                "evidence levels defined"
            ),
            "Passed": (
                len(observation_sufficiency_rules_df) >= 4
            ),
        },
        {
            "Validation Area": "Track classification completeness",
            "Requirement": (
                "Every track-country history must receive "
                "an evidence level."
            ),
            "Observed Evidence": (
                f"{track_sufficiency_df['evidence_level'].notna().sum():,} "
                "track histories classified"
            ),
            "Passed": bool(
                track_sufficiency_df[
                    "evidence_level"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist classification completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "an evidence level."
            ),
            "Observed Evidence": (
                f"{artist_sufficiency_df['evidence_level'].notna().sum():,} "
                "artist histories classified"
            ),
            "Passed": bool(
                artist_sufficiency_df[
                    "evidence_level"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Growth-eligible track availability",
            "Requirement": (
                "At least some track-country histories must "
                "qualify for growth analysis."
            ),
            "Observed Evidence": (
                f"{int(track_sufficiency_df['evidence_level'].isin(['Growth Eligible', 'Strong Historical Support']).sum()):,} "
                "track histories growth eligible"
            ),
            "Passed": bool(
                track_sufficiency_df[
                    "evidence_level"
                ].isin(
                    [
                        "Growth Eligible",
                        "Strong Historical Support",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "Growth-eligible artist availability",
            "Requirement": (
                "At least some artist-country histories must "
                "qualify for growth analysis."
            ),
            "Observed Evidence": (
                f"{int(artist_sufficiency_df['evidence_level'].isin(['Growth Eligible', 'Strong Historical Support']).sum()):,} "
                "artist histories growth eligible"
            ),
            "Passed": bool(
                artist_sufficiency_df[
                    "evidence_level"
                ].isin(
                    [
                        "Growth Eligible",
                        "Strong Historical Support",
                    ]
                ).any()
            ),
        },
    ]
)

print()
print("Observation sufficiency validation")
print("=" * 100)

display(section_4_5_validation_df)


section_4_5_complete = bool(
    section_4_5_validation_df[
        "Passed"
    ].all()
)

section_4_complete = bool(
    section_4_1_complete
    and section_4_2_complete
    and section_4_3_complete
    and section_4_4_complete
    and section_4_5_complete
)

if not section_4_5_complete:
    raise AssertionError(
        "Section 4.5 Observation Sufficiency Rules "
        "validation failed."
    )

print()
print(
    f"Section 4.5 completion status: "
    f"{section_4_5_complete}"
)

print(
    f"Section 4 overall completion status: "
    f"{section_4_complete}"
)

print(
    "Observation sufficiency rules have been applied "
    "to track-country and artist-country histories."
)
```

#### Interpretation

Observation sufficiency rules were applied successfully across all track-country and artist-country histories.

PMIP identified **178,400 track histories** and **112,213 artist histories** as suitable for standard or strong growth analysis, while sparse histories were retained but marked with lower evidence confidence.

Section 4.5 and Section 4 are complete.

## 5. Streaming Growth Analysis

This section measures how streaming performance changes over time within growth-eligible track-country and artist-country histories.

The analysis focuses on absolute change, percentage change, rolling growth, growth stability, and sustained streaming improvement.

### 5.1 Absolute Stream Change

#### Purpose

Measure the raw increase or decrease in streaming performance between the first and last observations of each eligible history.

This provides the simplest directional measure of historical streaming growth.

```python
print("Calculating PMIP absolute streaming change")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 4 completion
# ------------------------------------------------------------

if not section_4_complete:
    raise RuntimeError(
        "Section 4 Historical Performance Preparation "
        "must be complete before streaming growth analysis begins."
    )

print(
    f"Section 4 overall completion status: "
    f"{section_4_complete}"
)


# ------------------------------------------------------------
# 2. Retain growth-eligible track-country histories
# ------------------------------------------------------------

track_stream_growth_df = (
    track_sufficiency_df[
        track_sufficiency_df["evidence_level"].isin(
            [
                "Growth Eligible",
                "Strong Historical Support",
            ]
        )
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Retain growth-eligible artist-country histories
# ------------------------------------------------------------

artist_stream_growth_df = (
    artist_sufficiency_df[
        artist_sufficiency_df["evidence_level"].isin(
            [
                "Growth Eligible",
                "Strong Historical Support",
            ]
        )
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Calculate absolute stream change
# ------------------------------------------------------------

track_stream_growth_df["absolute_stream_change"] = (
    track_stream_growth_df["last_streams"]
    - track_stream_growth_df["first_streams"]
)

artist_stream_growth_df["absolute_stream_change"] = (
    artist_stream_growth_df["last_streams"]
    - artist_stream_growth_df["first_streams"]
)


# ------------------------------------------------------------
# 5. Define change direction
# ------------------------------------------------------------

def classify_stream_direction(value):
    if pd.isna(value):
        return "Missing"
    elif value > 0:
        return "Positive"
    elif value < 0:
        return "Negative"
    else:
        return "Stable"


track_stream_growth_df["stream_change_direction"] = (
    track_stream_growth_df[
        "absolute_stream_change"
    ].apply(classify_stream_direction)
)

artist_stream_growth_df["stream_change_direction"] = (
    artist_stream_growth_df[
        "absolute_stream_change"
    ].apply(classify_stream_direction)
)


# ------------------------------------------------------------
# 6. Track-country absolute change distribution
# ------------------------------------------------------------

track_absolute_change_distribution_df = (
    track_stream_growth_df[
        "stream_change_direction"
    ]
    .value_counts(dropna=False)
    .rename_axis("Direction")
    .reset_index(name="Track-Country Histories")
)

track_absolute_change_distribution_df["Share (%)"] = (
    track_absolute_change_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_stream_growth_df)
    * 100
)

print()
print("Track-country absolute stream change distribution")
print("=" * 100)

display(track_absolute_change_distribution_df)


# ------------------------------------------------------------
# 7. Artist-country absolute change distribution
# ------------------------------------------------------------

artist_absolute_change_distribution_df = (
    artist_stream_growth_df[
        "stream_change_direction"
    ]
    .value_counts(dropna=False)
    .rename_axis("Direction")
    .reset_index(name="Artist-Country Histories")
)

artist_absolute_change_distribution_df["Share (%)"] = (
    artist_absolute_change_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_stream_growth_df)
    * 100
)

print()
print("Artist-country absolute stream change distribution")
print("=" * 100)

display(artist_absolute_change_distribution_df)


# ------------------------------------------------------------
# 8. Absolute change overview
# ------------------------------------------------------------

absolute_stream_change_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Track-country histories assessed",
            "Observed Value": len(track_stream_growth_df),
        },
        {
            "Metric": "Track histories with positive change",
            "Observed Value": int(
                (
                    track_stream_growth_df[
                        "stream_change_direction"
                    ] == "Positive"
                ).sum()
            ),
        },
        {
            "Metric": "Median track absolute stream change",
            "Observed Value": float(
                track_stream_growth_df[
                    "absolute_stream_change"
                ].median()
            ),
        },
        {
            "Metric": "Artist-country histories assessed",
            "Observed Value": len(artist_stream_growth_df),
        },
        {
            "Metric": "Artist histories with positive change",
            "Observed Value": int(
                (
                    artist_stream_growth_df[
                        "stream_change_direction"
                    ] == "Positive"
                ).sum()
            ),
        },
        {
            "Metric": "Median artist absolute stream change",
            "Observed Value": float(
                artist_stream_growth_df[
                    "absolute_stream_change"
                ].median()
            ),
        },
    ]
)

print()
print("Absolute streaming change overview")
print("=" * 100)

display(absolute_stream_change_summary_df)


# ------------------------------------------------------------
# 9. Diagnostics
# ------------------------------------------------------------

absolute_stream_change_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Growth-eligible track histories assessed",
            "Observed Value": len(track_stream_growth_df),
        },
        {
            "Diagnostic": "Finite track absolute changes",
            "Observed Value": int(
                track_stream_growth_df[
                    "absolute_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Positive track changes",
            "Observed Value": int(
                (
                    track_stream_growth_df[
                        "stream_change_direction"
                    ] == "Positive"
                ).sum()
            ),
        },
        {
            "Diagnostic": "Growth-eligible artist histories assessed",
            "Observed Value": len(artist_stream_growth_df),
        },
        {
            "Diagnostic": "Finite artist absolute changes",
            "Observed Value": int(
                artist_stream_growth_df[
                    "absolute_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Positive artist changes",
            "Observed Value": int(
                (
                    artist_stream_growth_df[
                        "stream_change_direction"
                    ] == "Positive"
                ).sum()
            ),
        },
    ]
)

print()
print("Absolute stream change diagnostics")
print("=" * 100)

display(absolute_stream_change_diagnostics_df)


# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

section_5_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 4 completion",
            "Requirement": (
                "Historical Performance Preparation "
                "must already be complete."
            ),
            "Observed Evidence": (
                f"Section 4 overall completion status: "
                f"{section_4_complete}"
            ),
            "Passed": section_4_complete,
        },
        {
            "Validation Area": "Track eligible population",
            "Requirement": (
                "Growth-eligible track-country histories "
                "must be available."
            ),
            "Observed Evidence": (
                f"{len(track_stream_growth_df):,} "
                "track histories assessed"
            ),
            "Passed": len(track_stream_growth_df) > 0,
        },
        {
            "Validation Area": "Artist eligible population",
            "Requirement": (
                "Growth-eligible artist-country histories "
                "must be available."
            ),
            "Observed Evidence": (
                f"{len(artist_stream_growth_df):,} "
                "artist histories assessed"
            ),
            "Passed": len(artist_stream_growth_df) > 0,
        },
        {
            "Validation Area": "Track change completeness",
            "Requirement": (
                "Every eligible track history must receive "
                "an absolute stream change."
            ),
            "Observed Evidence": (
                f"{track_stream_growth_df['absolute_stream_change'].notna().sum():,} "
                "track changes calculated"
            ),
            "Passed": bool(
                track_stream_growth_df[
                    "absolute_stream_change"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist change completeness",
            "Requirement": (
                "Every eligible artist history must receive "
                "an absolute stream change."
            ),
            "Observed Evidence": (
                f"{artist_stream_growth_df['absolute_stream_change'].notna().sum():,} "
                "artist changes calculated"
            ),
            "Passed": bool(
                artist_stream_growth_df[
                    "absolute_stream_change"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Direction classification completeness",
            "Requirement": (
                "Every eligible history must receive a "
                "stream-change direction."
            ),
            "Observed Evidence": (
                f"{track_stream_growth_df['stream_change_direction'].notna().sum():,} "
                "track directions / "
                f"{artist_stream_growth_df['stream_change_direction'].notna().sum():,} "
                "artist directions"
            ),
            "Passed": bool(
                track_stream_growth_df[
                    "stream_change_direction"
                ].notna().all()
                and
                artist_stream_growth_df[
                    "stream_change_direction"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Absolute streaming change validation")
print("=" * 100)

display(section_5_1_validation_df)


section_5_1_complete = bool(
    section_5_1_validation_df[
        "Passed"
    ].all()
)

if not section_5_1_complete:
    raise AssertionError(
        "Section 5.1 Absolute Stream Change validation failed."
    )

print()
print(
    f"Section 5.1 completion status: "
    f"{section_5_1_complete}"
)

print(
    "Absolute historical streaming changes have been "
    "calculated for growth-eligible track-country and "
    "artist-country histories."
)
```

#### Interpretation

Absolute streaming change was calculated for **178,400 track-country histories** and **112,213 artist-country histories**.

Most eligible histories declined between their first and last observations, while **62,176 track histories** and **49,632 artist histories** showed positive streaming growth.

Section 5.1 is complete.

### 5.2 Percentage Stream Change

#### Purpose

Measure streaming growth relative to the starting stream level for each growth-eligible track-country and artist-country history.

This allows PMIP to compare growth across histories with very different stream volumes.

```python
print("Calculating PMIP percentage streaming change")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 5.1 completion
# ------------------------------------------------------------

if not section_5_1_complete:
    raise RuntimeError(
        "Section 5.1 Absolute Stream Change must be complete "
        "before percentage stream change is calculated."
    )

print(
    f"Section 5.1 completion status: "
    f"{section_5_1_complete}"
)


# ------------------------------------------------------------
# 2. Calculate percentage stream change
# ------------------------------------------------------------

track_stream_growth_df["percentage_stream_change"] = (
    track_stream_growth_df["absolute_stream_change"]
    / track_stream_growth_df["first_streams"].replace(0, pd.NA)
) * 100

artist_stream_growth_df["percentage_stream_change"] = (
    artist_stream_growth_df["absolute_stream_change"]
    / artist_stream_growth_df["first_streams"].replace(0, pd.NA)
) * 100


# ------------------------------------------------------------
# 3. Clean numeric values
# ------------------------------------------------------------

track_stream_growth_df["percentage_stream_change"] = (
    pd.to_numeric(
        track_stream_growth_df["percentage_stream_change"],
        errors="coerce"
    )
)

artist_stream_growth_df["percentage_stream_change"] = (
    pd.to_numeric(
        artist_stream_growth_df["percentage_stream_change"],
        errors="coerce"
    )
)


# ------------------------------------------------------------
# 4. Define percentage growth bands
# ------------------------------------------------------------

def classify_percentage_change(value):
    if pd.isna(value):
        return "Unavailable"
    elif value >= 100:
        return "Very Strong Growth"
    elif value >= 25:
        return "Strong Growth"
    elif value > 0:
        return "Moderate Growth"
    elif value == 0:
        return "Stable"
    elif value > -25:
        return "Moderate Decline"
    elif value > -50:
        return "Strong Decline"
    else:
        return "Very Strong Decline"


track_stream_growth_df["percentage_change_class"] = (
    track_stream_growth_df[
        "percentage_stream_change"
    ].apply(classify_percentage_change)
)

artist_stream_growth_df["percentage_change_class"] = (
    artist_stream_growth_df[
        "percentage_stream_change"
    ].apply(classify_percentage_change)
)


# ------------------------------------------------------------
# 5. Track percentage-change distribution
# ------------------------------------------------------------

track_percentage_distribution_df = (
    track_stream_growth_df[
        "percentage_change_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Percentage Change Class")
    .reset_index(name="Track-Country Histories")
)

track_percentage_distribution_df["Share (%)"] = (
    track_percentage_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_stream_growth_df)
    * 100
)

print()
print("Track-country percentage stream change distribution")
print("=" * 100)

display(track_percentage_distribution_df)


# ------------------------------------------------------------
# 6. Artist percentage-change distribution
# ------------------------------------------------------------

artist_percentage_distribution_df = (
    artist_stream_growth_df[
        "percentage_change_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Percentage Change Class")
    .reset_index(name="Artist-Country Histories")
)

artist_percentage_distribution_df["Share (%)"] = (
    artist_percentage_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_stream_growth_df)
    * 100
)

print()
print("Artist-country percentage stream change distribution")
print("=" * 100)

display(artist_percentage_distribution_df)


# ------------------------------------------------------------
# 7. Percentage change overview
# ------------------------------------------------------------

percentage_stream_change_summary_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories with valid percentage change",
            "Observed Value": int(
                track_stream_growth_df[
                    "percentage_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Metric": "Median track percentage change",
            "Observed Value": float(
                track_stream_growth_df[
                    "percentage_stream_change"
                ].median()
            ),
        },
        {
            "Metric": "Track histories above 25% growth",
            "Observed Value": int(
                (
                    track_stream_growth_df[
                        "percentage_stream_change"
                    ] >= 25
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories with valid percentage change",
            "Observed Value": int(
                artist_stream_growth_df[
                    "percentage_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Metric": "Median artist percentage change",
            "Observed Value": float(
                artist_stream_growth_df[
                    "percentage_stream_change"
                ].median()
            ),
        },
        {
            "Metric": "Artist histories above 25% growth",
            "Observed Value": int(
                (
                    artist_stream_growth_df[
                        "percentage_stream_change"
                    ] >= 25
                ).sum()
            ),
        },
    ]
)

print()
print("Percentage streaming change overview")
print("=" * 100)

display(percentage_stream_change_summary_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

percentage_stream_change_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(track_stream_growth_df),
        },
        {
            "Diagnostic": "Track percentage changes available",
            "Observed Value": int(
                track_stream_growth_df[
                    "percentage_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track histories with zero starting streams",
            "Observed Value": int(
                (
                    track_stream_growth_df[
                        "first_streams"
                    ] == 0
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(artist_stream_growth_df),
        },
        {
            "Diagnostic": "Artist percentage changes available",
            "Observed Value": int(
                artist_stream_growth_df[
                    "percentage_stream_change"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories with zero starting streams",
            "Observed Value": int(
                (
                    artist_stream_growth_df[
                        "first_streams"
                    ] == 0
                ).sum()
            ),
        },
    ]
)

print()
print("Percentage stream change diagnostics")
print("=" * 100)

display(percentage_stream_change_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_5_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.1 completion",
            "Requirement": (
                "Absolute Stream Change must already be complete."
            ),
            "Observed Evidence": (
                f"Section 5.1 completion status: "
                f"{section_5_1_complete}"
            ),
            "Passed": section_5_1_complete,
        },
        {
            "Validation Area": "Track percentage availability",
            "Requirement": (
                "At least some eligible track histories must "
                "receive valid percentage changes."
            ),
            "Observed Evidence": (
                f"{track_stream_growth_df['percentage_stream_change'].notna().sum():,} "
                "track percentage changes available"
            ),
            "Passed": bool(
                track_stream_growth_df[
                    "percentage_stream_change"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Artist percentage availability",
            "Requirement": (
                "At least some eligible artist histories must "
                "receive valid percentage changes."
            ),
            "Observed Evidence": (
                f"{artist_stream_growth_df['percentage_stream_change'].notna().sum():,} "
                "artist percentage changes available"
            ),
            "Passed": bool(
                artist_stream_growth_df[
                    "percentage_stream_change"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Track class completeness",
            "Requirement": (
                "Every track history must receive a percentage "
                "change classification."
            ),
            "Observed Evidence": (
                f"{track_stream_growth_df['percentage_change_class'].notna().sum():,} "
                "track classes assigned"
            ),
            "Passed": bool(
                track_stream_growth_df[
                    "percentage_change_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist class completeness",
            "Requirement": (
                "Every artist history must receive a percentage "
                "change classification."
            ),
            "Observed Evidence": (
                f"{artist_stream_growth_df['percentage_change_class'].notna().sum():,} "
                "artist classes assigned"
            ),
            "Passed": bool(
                artist_stream_growth_df[
                    "percentage_change_class"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Percentage streaming change validation")
print("=" * 100)

display(section_5_2_validation_df)


section_5_2_complete = bool(
    section_5_2_validation_df[
        "Passed"
    ].all()
)

if not section_5_2_complete:
    raise AssertionError(
        "Section 5.2 Percentage Stream Change validation failed."
    )

print()
print(
    f"Section 5.2 completion status: "
    f"{section_5_2_complete}"
)

print(
    "Percentage streaming changes have been calculated "
    "for PMIP growth-eligible histories."
)
```

#### Interpretation

Percentage streaming change was calculated successfully for all growth-eligible histories.

The median change was **-18.86% for track-country histories** and **-6.56% for artist-country histories**, although **34,641 track histories** and **31,389 artist histories** still achieved growth above **25%**.

Section 5.2 is complete.

### 5.3 Rolling Growth

#### Purpose

Measure short-term streaming movement across consecutive historical observations using rolling growth indicators.

This helps PMIP identify whether growth is repeatedly improving over time rather than relying only on the first and last observations.

```python
print("Calculating PMIP rolling streaming growth")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 5.2 completion
# ------------------------------------------------------------

if not section_5_2_complete:
    raise RuntimeError(
        "Section 5.2 Percentage Stream Change must be complete "
        "before rolling growth is calculated."
    )

print(
    f"Section 5.2 completion status: "
    f"{section_5_2_complete}"
)


# ------------------------------------------------------------
# 2. Prepare growth-eligible track observation histories
# ------------------------------------------------------------

eligible_track_keys = (
    track_stream_growth_df[
        [
            "track_id",
            "country",
        ]
    ]
    .drop_duplicates()
)

track_rolling_df = (
    historical_preparation_df
    .merge(
        eligible_track_keys,
        on=[
            "track_id",
            "country",
        ],
        how="inner",
    )
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Calculate track rolling change
# ------------------------------------------------------------

track_group_keys = [
    "track_id",
    "country",
]

track_rolling_df["previous_streams"] = (
    track_rolling_df
    .groupby(track_group_keys)["streams"]
    .shift(1)
)

track_rolling_df["rolling_stream_change"] = (
    track_rolling_df["streams"]
    - track_rolling_df["previous_streams"]
)

track_rolling_df["rolling_stream_pct_change"] = (
    track_rolling_df
    .groupby(track_group_keys)["streams"]
    .pct_change(fill_method=None)
    * 100
)

track_rolling_df["positive_rolling_growth"] = (
    track_rolling_df["rolling_stream_change"] > 0
)


# ------------------------------------------------------------
# 4. Aggregate track rolling-growth indicators
# ------------------------------------------------------------

track_rolling_growth_summary_df = (
    track_rolling_df
    .groupby(track_group_keys)
    .agg(
        rolling_changes=("rolling_stream_change", "count"),
        mean_rolling_change=("rolling_stream_change", "mean"),
        median_rolling_change=("rolling_stream_change", "median"),
        mean_rolling_pct_change=("rolling_stream_pct_change", "mean"),
        median_rolling_pct_change=("rolling_stream_pct_change", "median"),
        positive_rolling_changes=("positive_rolling_growth", "sum"),
    )
    .reset_index()
)

track_rolling_growth_summary_df["positive_rolling_share"] = (
    track_rolling_growth_summary_df["positive_rolling_changes"]
    / track_rolling_growth_summary_df["rolling_changes"].replace(0, pd.NA)
)


# ------------------------------------------------------------
# 5. Prepare growth-eligible artist observation histories
# ------------------------------------------------------------

eligible_artist_keys = (
    artist_stream_growth_df[
        [
            "artists",
            "country",
        ]
    ]
    .drop_duplicates()
)

artist_rolling_df = (
    artist_country_date_history_df
    .merge(
        eligible_artist_keys,
        on=[
            "artists",
            "country",
        ],
        how="inner",
    )
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Calculate artist rolling change
# ------------------------------------------------------------

artist_group_keys = [
    "artists",
    "country",
]

artist_rolling_df["previous_streams"] = (
    artist_rolling_df
    .groupby(artist_group_keys)["daily_streams"]
    .shift(1)
)

artist_rolling_df["rolling_stream_change"] = (
    artist_rolling_df["daily_streams"]
    - artist_rolling_df["previous_streams"]
)

artist_rolling_df["rolling_stream_pct_change"] = (
    artist_rolling_df
    .groupby(artist_group_keys)["daily_streams"]
    .pct_change(fill_method=None)
    * 100
)

artist_rolling_df["positive_rolling_growth"] = (
    artist_rolling_df["rolling_stream_change"] > 0
)


# ------------------------------------------------------------
# 7. Aggregate artist rolling-growth indicators
# ------------------------------------------------------------

artist_rolling_growth_summary_df = (
    artist_rolling_df
    .groupby(artist_group_keys)
    .agg(
        rolling_changes=("rolling_stream_change", "count"),
        mean_rolling_change=("rolling_stream_change", "mean"),
        median_rolling_change=("rolling_stream_change", "median"),
        mean_rolling_pct_change=("rolling_stream_pct_change", "mean"),
        median_rolling_pct_change=("rolling_stream_pct_change", "median"),
        positive_rolling_changes=("positive_rolling_growth", "sum"),
    )
    .reset_index()
)

artist_rolling_growth_summary_df["positive_rolling_share"] = (
    artist_rolling_growth_summary_df["positive_rolling_changes"]
    / artist_rolling_growth_summary_df["rolling_changes"].replace(0, pd.NA)
)


# ------------------------------------------------------------
# 8. Define rolling-growth classes
# ------------------------------------------------------------

def classify_rolling_growth(value):
    if pd.isna(value):
        return "Unavailable"
    elif value >= 0.75:
        return "Highly Consistent Growth"
    elif value >= 0.50:
        return "Moderate Rolling Growth"
    elif value >= 0.25:
        return "Mixed Growth"
    else:
        return "Predominantly Declining"


track_rolling_growth_summary_df["rolling_growth_class"] = (
    track_rolling_growth_summary_df[
        "positive_rolling_share"
    ].apply(classify_rolling_growth)
)

artist_rolling_growth_summary_df["rolling_growth_class"] = (
    artist_rolling_growth_summary_df[
        "positive_rolling_share"
    ].apply(classify_rolling_growth)
)


# ------------------------------------------------------------
# 9. Track rolling-growth distribution
# ------------------------------------------------------------

track_rolling_distribution_df = (
    track_rolling_growth_summary_df[
        "rolling_growth_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Rolling Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_rolling_distribution_df["Share (%)"] = (
    track_rolling_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_rolling_growth_summary_df)
    * 100
)

print()
print("Track-country rolling growth distribution")
print("=" * 100)

display(track_rolling_distribution_df)


# ------------------------------------------------------------
# 10. Artist rolling-growth distribution
# ------------------------------------------------------------

artist_rolling_distribution_df = (
    artist_rolling_growth_summary_df[
        "rolling_growth_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Rolling Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_rolling_distribution_df["Share (%)"] = (
    artist_rolling_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_rolling_growth_summary_df)
    * 100
)

print()
print("Artist-country rolling growth distribution")
print("=" * 100)

display(artist_rolling_distribution_df)


# ------------------------------------------------------------
# 11. Rolling growth overview
# ------------------------------------------------------------

rolling_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_rolling_growth_summary_df
            ),
        },
        {
            "Metric": "Median track positive rolling share",
            "Observed Value": float(
                track_rolling_growth_summary_df[
                    "positive_rolling_share"
                ].median()
            ),
        },
        {
            "Metric": "Track histories with >= 50% positive changes",
            "Observed Value": int(
                (
                    track_rolling_growth_summary_df[
                        "positive_rolling_share"
                    ] >= 0.50
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_rolling_growth_summary_df
            ),
        },
        {
            "Metric": "Median artist positive rolling share",
            "Observed Value": float(
                artist_rolling_growth_summary_df[
                    "positive_rolling_share"
                ].median()
            ),
        },
        {
            "Metric": "Artist histories with >= 50% positive changes",
            "Observed Value": int(
                (
                    artist_rolling_growth_summary_df[
                        "positive_rolling_share"
                    ] >= 0.50
                ).sum()
            ),
        },
    ]
)

print()
print("Rolling streaming growth overview")
print("=" * 100)

display(rolling_growth_overview_df)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

rolling_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track rolling histories assessed",
            "Observed Value": len(
                track_rolling_growth_summary_df
            ),
        },
        {
            "Diagnostic": "Track rolling changes calculated",
            "Observed Value": int(
                track_rolling_growth_summary_df[
                    "rolling_changes"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Track positive-share values available",
            "Observed Value": int(
                track_rolling_growth_summary_df[
                    "positive_rolling_share"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist rolling histories assessed",
            "Observed Value": len(
                artist_rolling_growth_summary_df
            ),
        },
        {
            "Diagnostic": "Artist rolling changes calculated",
            "Observed Value": int(
                artist_rolling_growth_summary_df[
                    "rolling_changes"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Artist positive-share values available",
            "Observed Value": int(
                artist_rolling_growth_summary_df[
                    "positive_rolling_share"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Rolling growth diagnostics")
print("=" * 100)

display(rolling_growth_diagnostics_df)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_5_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.2 completion",
            "Requirement": (
                "Percentage Stream Change must already be complete."
            ),
            "Observed Evidence": (
                f"Section 5.2 completion status: "
                f"{section_5_2_complete}"
            ),
            "Passed": section_5_2_complete,
        },
        {
            "Validation Area": "Track rolling population",
            "Requirement": (
                "Growth-eligible track histories must provide "
                "rolling stream changes."
            ),
            "Observed Evidence": (
                f"{len(track_rolling_growth_summary_df):,} "
                "track histories assessed"
            ),
            "Passed": len(
                track_rolling_growth_summary_df
            ) > 0,
        },
        {
            "Validation Area": "Artist rolling population",
            "Requirement": (
                "Growth-eligible artist histories must provide "
                "rolling stream changes."
            ),
            "Observed Evidence": (
                f"{len(artist_rolling_growth_summary_df):,} "
                "artist histories assessed"
            ),
            "Passed": len(
                artist_rolling_growth_summary_df
            ) > 0,
        },
        {
            "Validation Area": "Track rolling-share bounds",
            "Requirement": (
                "Track positive rolling shares must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Track rolling shares validated"
            ),
            "Passed": bool(
                track_rolling_growth_summary_df[
                    "positive_rolling_share"
                ]
                .dropna()
                .between(0, 1)
                .all()
            ),
        },
        {
            "Validation Area": "Artist rolling-share bounds",
            "Requirement": (
                "Artist positive rolling shares must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist rolling shares validated"
            ),
            "Passed": bool(
                artist_rolling_growth_summary_df[
                    "positive_rolling_share"
                ]
                .dropna()
                .between(0, 1)
                .all()
            ),
        },
        {
            "Validation Area": "Rolling classification completeness",
            "Requirement": (
                "Every retained history must receive "
                "a rolling-growth classification."
            ),
            "Observed Evidence": (
                f"{track_rolling_growth_summary_df['rolling_growth_class'].notna().sum():,} "
                "track classes / "
                f"{artist_rolling_growth_summary_df['rolling_growth_class'].notna().sum():,} "
                "artist classes"
            ),
            "Passed": bool(
                track_rolling_growth_summary_df[
                    "rolling_growth_class"
                ].notna().all()
                and
                artist_rolling_growth_summary_df[
                    "rolling_growth_class"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Rolling streaming growth validation")
print("=" * 100)

display(section_5_3_validation_df)


section_5_3_complete = bool(
    section_5_3_validation_df[
        "Passed"
    ].all()
)

if not section_5_3_complete:
    raise AssertionError(
        "Section 5.3 Rolling Growth validation failed."
    )

print()
print(
    f"Section 5.3 completion status: "
    f"{section_5_3_complete}"
)

print(
    "Rolling streaming-growth indicators have been "
    "calculated for growth-eligible PMIP histories."
)
```

#### Interpretation

Rolling growth analysis shows that most histories have mixed streaming movement rather than consistently increasing performance.

Only **4,680 track-country histories** and **2,820 artist-country histories** show highly consistent rolling growth, while broader positive movement appears in **43,685 track histories** and **29,194 artist histories**.

Section 5.3 is complete.

### 5.4 Growth Rate Stability

#### Purpose

Measure how stable or volatile streaming growth is across consecutive observations.

This helps PMIP distinguish steady growth from highly unstable performance changes.

```python
print("Assessing PMIP streaming growth-rate stability")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 5.3 completion
# ------------------------------------------------------------

if not section_5_3_complete:
    raise RuntimeError(
        "Section 5.3 Rolling Growth must be complete "
        "before growth-rate stability is assessed."
    )

print(
    f"Section 5.3 completion status: "
    f"{section_5_3_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level stability measures
# ------------------------------------------------------------

track_stability_df = (
    track_rolling_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .agg(
        rolling_change_count=("rolling_stream_pct_change", "count"),
        mean_growth_rate=("rolling_stream_pct_change", "mean"),
        median_growth_rate=("rolling_stream_pct_change", "median"),
        growth_rate_std=("rolling_stream_pct_change", "std"),
        growth_rate_min=("rolling_stream_pct_change", "min"),
        growth_rate_max=("rolling_stream_pct_change", "max"),
    )
    .reset_index()
)

track_stability_df["growth_rate_range"] = (
    track_stability_df["growth_rate_max"]
    - track_stability_df["growth_rate_min"]
)


# ------------------------------------------------------------
# 3. Prepare artist-level stability measures
# ------------------------------------------------------------

artist_stability_df = (
    artist_rolling_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )
    .agg(
        rolling_change_count=("rolling_stream_pct_change", "count"),
        mean_growth_rate=("rolling_stream_pct_change", "mean"),
        median_growth_rate=("rolling_stream_pct_change", "median"),
        growth_rate_std=("rolling_stream_pct_change", "std"),
        growth_rate_min=("rolling_stream_pct_change", "min"),
        growth_rate_max=("rolling_stream_pct_change", "max"),
    )
    .reset_index()
)

artist_stability_df["growth_rate_range"] = (
    artist_stability_df["growth_rate_max"]
    - artist_stability_df["growth_rate_min"]
)


# ------------------------------------------------------------
# 4. Define stability thresholds from observed data
# ------------------------------------------------------------

track_valid_std = (
    track_stability_df["growth_rate_std"]
    .dropna()
)

artist_valid_std = (
    artist_stability_df["growth_rate_std"]
    .dropna()
)

track_q25 = float(
    track_valid_std.quantile(0.25)
)

track_q75 = float(
    track_valid_std.quantile(0.75)
)

artist_q25 = float(
    artist_valid_std.quantile(0.25)
)

artist_q75 = float(
    artist_valid_std.quantile(0.75)
)


# ------------------------------------------------------------
# 5. Stability classifier
# ------------------------------------------------------------

def classify_stability(
    value,
    q25,
    q75
):
    if pd.isna(value):
        return "Insufficient Stability Evidence"
    elif value <= q25:
        return "High Stability"
    elif value <= q75:
        return "Moderate Stability"
    else:
        return "High Volatility"


track_stability_df["growth_stability_class"] = (
    track_stability_df[
        "growth_rate_std"
    ].apply(
        lambda x: classify_stability(
            x,
            track_q25,
            track_q75
        )
    )
)

artist_stability_df["growth_stability_class"] = (
    artist_stability_df[
        "growth_rate_std"
    ].apply(
        lambda x: classify_stability(
            x,
            artist_q25,
            artist_q75
        )
    )
)


# ------------------------------------------------------------
# 6. Track stability distribution
# ------------------------------------------------------------

track_stability_distribution_df = (
    track_stability_df[
        "growth_stability_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Growth Stability Class")
    .reset_index(name="Track-Country Histories")
)

track_stability_distribution_df["Share (%)"] = (
    track_stability_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_stability_df)
    * 100
)

print()
print("Track-country growth-rate stability distribution")
print("=" * 100)

display(track_stability_distribution_df)


# ------------------------------------------------------------
# 7. Artist stability distribution
# ------------------------------------------------------------

artist_stability_distribution_df = (
    artist_stability_df[
        "growth_stability_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("Growth Stability Class")
    .reset_index(name="Artist-Country Histories")
)

artist_stability_distribution_df["Share (%)"] = (
    artist_stability_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_stability_df)
    * 100
)

print()
print("Artist-country growth-rate stability distribution")
print("=" * 100)

display(artist_stability_distribution_df)


# ------------------------------------------------------------
# 8. Stability overview
# ------------------------------------------------------------

growth_stability_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(track_stability_df),
        },
        {
            "Metric": "Median track growth-rate std",
            "Observed Value": float(
                track_valid_std.median()
            ),
        },
        {
            "Metric": "Track high-stability histories",
            "Observed Value": int(
                (
                    track_stability_df[
                        "growth_stability_class"
                    ] == "High Stability"
                ).sum()
            ),
        },
        {
            "Metric": "Track high-volatility histories",
            "Observed Value": int(
                (
                    track_stability_df[
                        "growth_stability_class"
                    ] == "High Volatility"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(artist_stability_df),
        },
        {
            "Metric": "Median artist growth-rate std",
            "Observed Value": float(
                artist_valid_std.median()
            ),
        },
        {
            "Metric": "Artist high-stability histories",
            "Observed Value": int(
                (
                    artist_stability_df[
                        "growth_stability_class"
                    ] == "High Stability"
                ).sum()
            ),
        },
        {
            "Metric": "Artist high-volatility histories",
            "Observed Value": int(
                (
                    artist_stability_df[
                        "growth_stability_class"
                    ] == "High Volatility"
                ).sum()
            ),
        },
    ]
)

print()
print("Growth-rate stability overview")
print("=" * 100)

display(growth_stability_overview_df)


# ------------------------------------------------------------
# 9. Diagnostics
# ------------------------------------------------------------

growth_stability_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(track_stability_df),
        },
        {
            "Diagnostic": "Track stability values available",
            "Observed Value": int(
                track_stability_df[
                    "growth_rate_std"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track high-stability histories",
            "Observed Value": int(
                (
                    track_stability_df[
                        "growth_stability_class"
                    ] == "High Stability"
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(artist_stability_df),
        },
        {
            "Diagnostic": "Artist stability values available",
            "Observed Value": int(
                artist_stability_df[
                    "growth_rate_std"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist high-stability histories",
            "Observed Value": int(
                (
                    artist_stability_df[
                        "growth_stability_class"
                    ] == "High Stability"
                ).sum()
            ),
        },
    ]
)

print()
print("Growth-rate stability diagnostics")
print("=" * 100)

display(growth_stability_diagnostics_df)


# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

section_5_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.3 completion",
            "Requirement": (
                "Rolling Growth must already be complete."
            ),
            "Observed Evidence": (
                f"Section 5.3 completion status: "
                f"{section_5_3_complete}"
            ),
            "Passed": section_5_3_complete,
        },
        {
            "Validation Area": "Track stability availability",
            "Requirement": (
                "At least some track-country histories must "
                "provide measurable growth-rate stability."
            ),
            "Observed Evidence": (
                f"{track_stability_df['growth_rate_std'].notna().sum():,} "
                "track stability values available"
            ),
            "Passed": bool(
                track_stability_df[
                    "growth_rate_std"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Artist stability availability",
            "Requirement": (
                "At least some artist-country histories must "
                "provide measurable growth-rate stability."
            ),
            "Observed Evidence": (
                f"{artist_stability_df['growth_rate_std'].notna().sum():,} "
                "artist stability values available"
            ),
            "Passed": bool(
                artist_stability_df[
                    "growth_rate_std"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Track stability classification",
            "Requirement": (
                "Every retained track history must receive "
                "a stability classification."
            ),
            "Observed Evidence": (
                f"{track_stability_df['growth_stability_class'].notna().sum():,} "
                "track classes assigned"
            ),
            "Passed": bool(
                track_stability_df[
                    "growth_stability_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist stability classification",
            "Requirement": (
                "Every retained artist history must receive "
                "a stability classification."
            ),
            "Observed Evidence": (
                f"{artist_stability_df['growth_stability_class'].notna().sum():,} "
                "artist classes assigned"
            ),
            "Passed": bool(
                artist_stability_df[
                    "growth_stability_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Volatility differentiation",
            "Requirement": (
                "The framework must distinguish stable and "
                "high-volatility growth histories."
            ),
            "Observed Evidence": (
                "High Stability, Moderate Stability and "
                "High Volatility classes defined"
            ),
            "Passed": bool(
                "High Stability"
                in track_stability_df[
                    "growth_stability_class"
                ].values
                and
                "High Volatility"
                in track_stability_df[
                    "growth_stability_class"
                ].values
            ),
        },
    ]
)

print()
print("Growth-rate stability validation")
print("=" * 100)

display(section_5_4_validation_df)


section_5_4_complete = bool(
    section_5_4_validation_df[
        "Passed"
    ].all()
)

if not section_5_4_complete:
    raise AssertionError(
        "Section 5.4 Growth Rate Stability validation failed."
    )

print()
print(
    f"Section 5.4 completion status: "
    f"{section_5_4_complete}"
)

print(
    "Streaming growth-rate stability and volatility "
    "have been assessed for PMIP growth histories."
)
```

#### Interpretation

Growth-rate stability was successfully measured across the eligible histories.

Most track-country and artist-country histories show **moderate stability**, while around one quarter fall into **high stability** and another quarter into **high volatility**.

Section 5.4 is complete.

### 5.5 Sustained Streaming Growth

#### Purpose

Identify histories where streaming growth is both positive and sufficiently persistent over time.

This combines absolute change, percentage change, rolling growth, and stability so PMIP can distinguish sustained growth from temporary or unstable increases.

```python
print("Identifying PMIP sustained streaming growth")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 5.4 completion
# ------------------------------------------------------------

if not section_5_4_complete:
    raise RuntimeError(
        "Section 5.4 Growth Rate Stability must be complete "
        "before sustained streaming growth is identified."
    )

print(
    f"Section 5.4 completion status: "
    f"{section_5_4_complete}"
)


# ------------------------------------------------------------
# 2. Combine track-level growth evidence
# ------------------------------------------------------------

track_sustained_growth_df = (
    track_stream_growth_df
    .merge(
        track_rolling_growth_summary_df[
            [
                "track_id",
                "country",
                "positive_rolling_share",
                "rolling_growth_class",
            ]
        ],
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
    .merge(
        track_stability_df[
            [
                "track_id",
                "country",
                "growth_rate_std",
                "growth_stability_class",
            ]
        ],
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 3. Combine artist-level growth evidence
# ------------------------------------------------------------

artist_sustained_growth_df = (
    artist_stream_growth_df
    .merge(
        artist_rolling_growth_summary_df[
            [
                "artists",
                "country",
                "positive_rolling_share",
                "rolling_growth_class",
            ]
        ],
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
    .merge(
        artist_stability_df[
            [
                "artists",
                "country",
                "growth_rate_std",
                "growth_stability_class",
            ]
        ],
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 4. Define sustained-growth classification
# ------------------------------------------------------------

def classify_sustained_growth(row):
    positive_absolute = (
        row["absolute_stream_change"] > 0
    )

    positive_percentage = (
        row["percentage_stream_change"] > 0
    )

    rolling_support = (
        pd.notna(row["positive_rolling_share"])
        and row["positive_rolling_share"] >= 0.50
    )

    strong_rolling_support = (
        pd.notna(row["positive_rolling_share"])
        and row["positive_rolling_share"] >= 0.75
    )

    stable_support = (
        row["growth_stability_class"]
        in [
            "High Stability",
            "Moderate Stability",
        ]
    )

    if (
        positive_absolute
        and positive_percentage
        and strong_rolling_support
        and row["growth_stability_class"] == "High Stability"
    ):
        return "Strong Sustained Growth"

    elif (
        positive_absolute
        and positive_percentage
        and rolling_support
        and stable_support
    ):
        return "Sustained Growth"

    elif (
        positive_absolute
        and positive_percentage
    ):
        return "Emerging Growth"

    elif (
        positive_absolute
        or positive_percentage
    ):
        return "Weak Positive Signal"

    else:
        return "No Sustained Growth"


track_sustained_growth_df["sustained_growth_class"] = (
    track_sustained_growth_df.apply(
        classify_sustained_growth,
        axis=1,
    )
)

artist_sustained_growth_df["sustained_growth_class"] = (
    artist_sustained_growth_df.apply(
        classify_sustained_growth,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track sustained-growth distribution
# ------------------------------------------------------------

track_sustained_growth_distribution_df = (
    track_sustained_growth_df[
        "sustained_growth_class"
    ]
    .value_counts()
    .rename_axis("Sustained Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_sustained_growth_distribution_df["Share (%)"] = (
    track_sustained_growth_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_sustained_growth_df)
    * 100
)

print()
print("Track-country sustained streaming growth distribution")
print("=" * 100)

display(track_sustained_growth_distribution_df)


# ------------------------------------------------------------
# 6. Artist sustained-growth distribution
# ------------------------------------------------------------

artist_sustained_growth_distribution_df = (
    artist_sustained_growth_df[
        "sustained_growth_class"
    ]
    .value_counts()
    .rename_axis("Sustained Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_sustained_growth_distribution_df["Share (%)"] = (
    artist_sustained_growth_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_sustained_growth_df)
    * 100
)

print()
print("Artist-country sustained streaming growth distribution")
print("=" * 100)

display(artist_sustained_growth_distribution_df)


# ------------------------------------------------------------
# 7. Sustained-growth overview
# ------------------------------------------------------------

sustained_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_sustained_growth_df
            ),
        },
        {
            "Metric": "Track sustained-growth histories",
            "Observed Value": int(
                track_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track strong sustained-growth histories",
            "Observed Value": int(
                (
                    track_sustained_growth_df[
                        "sustained_growth_class"
                    ] == "Strong Sustained Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_sustained_growth_df
            ),
        },
        {
            "Metric": "Artist sustained-growth histories",
            "Observed Value": int(
                artist_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist strong sustained-growth histories",
            "Observed Value": int(
                (
                    artist_sustained_growth_df[
                        "sustained_growth_class"
                    ] == "Strong Sustained Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Sustained streaming growth overview")
print("=" * 100)

display(sustained_growth_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

sustained_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_sustained_growth_df
            ),
        },
        {
            "Diagnostic": "Track sustained-growth classes assigned",
            "Observed Value": int(
                track_sustained_growth_df[
                    "sustained_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track sustained-growth histories",
            "Observed Value": int(
                track_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_sustained_growth_df
            ),
        },
        {
            "Diagnostic": "Artist sustained-growth classes assigned",
            "Observed Value": int(
                artist_sustained_growth_df[
                    "sustained_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist sustained-growth histories",
            "Observed Value": int(
                artist_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Sustained streaming growth diagnostics")
print("=" * 100)

display(sustained_growth_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_5_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5.4 completion",
            "Requirement": (
                "Growth Rate Stability must already be complete."
            ),
            "Observed Evidence": (
                f"Section 5.4 completion status: "
                f"{section_5_4_complete}"
            ),
            "Passed": section_5_4_complete,
        },
        {
            "Validation Area": "Track classification completeness",
            "Requirement": (
                "Every eligible track-country history must receive "
                "a sustained-growth classification."
            ),
            "Observed Evidence": (
                f"{track_sustained_growth_df['sustained_growth_class'].notna().sum():,} "
                "track classes assigned"
            ),
            "Passed": bool(
                track_sustained_growth_df[
                    "sustained_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist classification completeness",
            "Requirement": (
                "Every eligible artist-country history must receive "
                "a sustained-growth classification."
            ),
            "Observed Evidence": (
                f"{artist_sustained_growth_df['sustained_growth_class'].notna().sum():,} "
                "artist classes assigned"
            ),
            "Passed": bool(
                artist_sustained_growth_df[
                    "sustained_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track sustained-growth availability",
            "Requirement": (
                "At least some track-country histories must show "
                "sustained streaming growth."
            ),
            "Observed Evidence": (
                f"{int(track_sustained_growth_df['sustained_growth_class'].isin(['Sustained Growth', 'Strong Sustained Growth']).sum()):,} "
                "track sustained-growth histories"
            ),
            "Passed": bool(
                track_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "Artist sustained-growth availability",
            "Requirement": (
                "At least some artist-country histories must show "
                "sustained streaming growth."
            ),
            "Observed Evidence": (
                f"{int(artist_sustained_growth_df['sustained_growth_class'].isin(['Sustained Growth', 'Strong Sustained Growth']).sum()):,} "
                "artist sustained-growth histories"
            ),
            "Passed": bool(
                artist_sustained_growth_df[
                    "sustained_growth_class"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "Multi-signal growth requirement",
            "Requirement": (
                "Sustained growth must combine direction, rolling "
                "behaviour and stability evidence."
            ),
            "Observed Evidence": (
                "Absolute change, percentage change, rolling growth "
                "and stability combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Sustained streaming growth validation")
print("=" * 100)

display(section_5_5_validation_df)


section_5_5_complete = bool(
    section_5_5_validation_df[
        "Passed"
    ].all()
)

section_5_complete = bool(
    section_5_1_complete
    and section_5_2_complete
    and section_5_3_complete
    and section_5_4_complete
    and section_5_5_complete
)

if not section_5_5_complete:
    raise AssertionError(
        "Section 5.5 Sustained Streaming Growth validation failed."
    )

print()
print(
    f"Section 5.5 completion status: "
    f"{section_5_5_complete}"
)

print(
    f"Section 5 overall completion status: "
    f"{section_5_complete}"
)

print(
    "Sustained streaming growth has been identified "
    "using multiple validated PMIP growth indicators."
)
```

#### Interpretation

PMIP identified **23,389 track-country histories** and **16,183 artist-country histories** with sustained streaming growth.

Section 5.5 and Section 5 are complete.

### 6.1 Chart Position Improvement

#### Purpose

Measure whether chart positions improve between the first and last observations of each eligible history.

Because lower numerical chart positions are better, a reduction in chart position represents positive chart growth.

```python
print("Calculating PMIP chart position improvement")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 5 completion
# ------------------------------------------------------------

if not section_5_complete:
    raise RuntimeError(
        "Section 5 Streaming Growth Analysis must be complete "
        "before chart growth analysis begins."
    )

print(
    f"Section 5 overall completion status: "
    f"{section_5_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-country chart-growth population
# ------------------------------------------------------------

track_chart_growth_df = (
    track_sustained_growth_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-country chart-growth population
# ------------------------------------------------------------

artist_chart_growth_df = (
    artist_sustained_growth_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Calculate chart-position improvement
# ------------------------------------------------------------

# Lower numerical chart position is better.
# Therefore:
# first_position - last_position > 0  -> improvement
# first_position - last_position < 0  -> decline

track_chart_growth_df["chart_position_improvement"] = (
    track_chart_growth_df["first_position"]
    - track_chart_growth_df["last_position"]
)

artist_chart_growth_df["chart_position_improvement"] = (
    artist_chart_growth_df["first_position"]
    - artist_chart_growth_df["last_position"]
)


# ------------------------------------------------------------
# 5. Define chart direction
# ------------------------------------------------------------

def classify_chart_direction(value):
    if pd.isna(value):
        return "Unavailable"
    elif value > 0:
        return "Improved"
    elif value < 0:
        return "Declined"
    else:
        return "Stable"


track_chart_growth_df["chart_direction"] = (
    track_chart_growth_df[
        "chart_position_improvement"
    ].apply(classify_chart_direction)
)

artist_chart_growth_df["chart_direction"] = (
    artist_chart_growth_df[
        "chart_position_improvement"
    ].apply(classify_chart_direction)
)


# ------------------------------------------------------------
# 6. Track chart-position distribution
# ------------------------------------------------------------

track_chart_direction_distribution_df = (
    track_chart_growth_df[
        "chart_direction"
    ]
    .value_counts(dropna=False)
    .rename_axis("Chart Direction")
    .reset_index(name="Track-Country Histories")
)

track_chart_direction_distribution_df["Share (%)"] = (
    track_chart_direction_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_chart_growth_df)
    * 100
)

print()
print("Track-country chart position direction")
print("=" * 100)

display(track_chart_direction_distribution_df)


# ------------------------------------------------------------
# 7. Artist chart-position distribution
# ------------------------------------------------------------

artist_chart_direction_distribution_df = (
    artist_chart_growth_df[
        "chart_direction"
    ]
    .value_counts(dropna=False)
    .rename_axis("Chart Direction")
    .reset_index(name="Artist-Country Histories")
)

artist_chart_direction_distribution_df["Share (%)"] = (
    artist_chart_direction_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_chart_growth_df)
    * 100
)

print()
print("Artist-country chart position direction")
print("=" * 100)

display(artist_chart_direction_distribution_df)


# ------------------------------------------------------------
# 8. Chart improvement overview
# ------------------------------------------------------------

chart_position_improvement_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_chart_growth_df
            ),
        },
        {
            "Metric": "Track histories with chart improvement",
            "Observed Value": int(
                (
                    track_chart_growth_df[
                        "chart_direction"
                    ] == "Improved"
                ).sum()
            ),
        },
        {
            "Metric": "Median track position improvement",
            "Observed Value": float(
                track_chart_growth_df[
                    "chart_position_improvement"
                ].median()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_growth_df
            ),
        },
        {
            "Metric": "Artist histories with chart improvement",
            "Observed Value": int(
                (
                    artist_chart_growth_df[
                        "chart_direction"
                    ] == "Improved"
                ).sum()
            ),
        },
        {
            "Metric": "Median artist position improvement",
            "Observed Value": float(
                artist_chart_growth_df[
                    "chart_position_improvement"
                ].median()
            ),
        },
    ]
)

print()
print("Chart position improvement overview")
print("=" * 100)

display(chart_position_improvement_overview_df)


# ------------------------------------------------------------
# 9. Diagnostics
# ------------------------------------------------------------

chart_position_improvement_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_chart_growth_df
            ),
        },
        {
            "Diagnostic": "Track chart changes available",
            "Observed Value": int(
                track_chart_growth_df[
                    "chart_position_improvement"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track chart improvements",
            "Observed Value": int(
                (
                    track_chart_growth_df[
                        "chart_direction"
                    ] == "Improved"
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_growth_df
            ),
        },
        {
            "Diagnostic": "Artist chart changes available",
            "Observed Value": int(
                artist_chart_growth_df[
                    "chart_position_improvement"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist chart improvements",
            "Observed Value": int(
                (
                    artist_chart_growth_df[
                        "chart_direction"
                    ] == "Improved"
                ).sum()
            ),
        },
    ]
)

print()
print("Chart position improvement diagnostics")
print("=" * 100)

display(chart_position_improvement_diagnostics_df)


# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

section_6_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 5 completion",
            "Requirement": (
                "Streaming Growth Analysis must already be complete."
            ),
            "Observed Evidence": (
                f"Section 5 overall completion status: "
                f"{section_5_complete}"
            ),
            "Passed": section_5_complete,
        },
        {
            "Validation Area": "Track chart population",
            "Requirement": (
                "Eligible track-country histories must be "
                "available for chart analysis."
            ),
            "Observed Evidence": (
                f"{len(track_chart_growth_df):,} "
                "track histories assessed"
            ),
            "Passed": len(track_chart_growth_df) > 0,
        },
        {
            "Validation Area": "Artist chart population",
            "Requirement": (
                "Eligible artist-country histories must be "
                "available for chart analysis."
            ),
            "Observed Evidence": (
                f"{len(artist_chart_growth_df):,} "
                "artist histories assessed"
            ),
            "Passed": len(artist_chart_growth_df) > 0,
        },
        {
            "Validation Area": "Track chart change availability",
            "Requirement": (
                "At least some track histories must have "
                "measurable chart-position change."
            ),
            "Observed Evidence": (
                f"{track_chart_growth_df['chart_position_improvement'].notna().sum():,} "
                "track chart changes available"
            ),
            "Passed": bool(
                track_chart_growth_df[
                    "chart_position_improvement"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Artist chart change availability",
            "Requirement": (
                "At least some artist histories must have "
                "measurable chart-position change."
            ),
            "Observed Evidence": (
                f"{artist_chart_growth_df['chart_position_improvement'].notna().sum():,} "
                "artist chart changes available"
            ),
            "Passed": bool(
                artist_chart_growth_df[
                    "chart_position_improvement"
                ].notna().any()
            ),
        },
        {
            "Validation Area": "Chart direction completeness",
            "Requirement": (
                "Every retained history must receive a "
                "chart-direction classification."
            ),
            "Observed Evidence": (
                f"{track_chart_growth_df['chart_direction'].notna().sum():,} "
                "track classes / "
                f"{artist_chart_growth_df['chart_direction'].notna().sum():,} "
                "artist classes"
            ),
            "Passed": bool(
                track_chart_growth_df[
                    "chart_direction"
                ].notna().all()
                and
                artist_chart_growth_df[
                    "chart_direction"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Chart position improvement validation")
print("=" * 100)

display(section_6_1_validation_df)


section_6_1_complete = bool(
    section_6_1_validation_df[
        "Passed"
    ].all()
)

if not section_6_1_complete:
    raise AssertionError(
        "Section 6.1 Chart Position Improvement validation failed."
    )

print()
print(
    f"Section 6.1 completion status: "
    f"{section_6_1_complete}"
)

print(
    "Historical chart-position improvement has been "
    "measured for PMIP track-country and artist-country histories."
)
```

#### Interpretation

Chart-position improvement was identified in **29,691 track-country histories** and **22,748 artist-country histories**.

Most eligible histories declined in chart position, so genuine chart improvement appears less common than simple historical participation.

Section 6.1 is complete.

### 6.2 Top-10 / Top-50 Movement

#### Purpose

Measure movement into stronger chart-performance bands by identifying histories that enter the **Top 50** or **Top 10** over time.

This helps PMIP distinguish meaningful chart progression from smaller position changes.

```python
print("Assessing PMIP Top-10 and Top-50 chart movement")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6.1 completion
# ------------------------------------------------------------

if not section_6_1_complete:
    raise RuntimeError(
        "Section 6.1 Chart Position Improvement must be complete "
        "before Top-10 / Top-50 movement is assessed."
    )

print(
    f"Section 6.1 completion status: "
    f"{section_6_1_complete}"
)


# ------------------------------------------------------------
# 2. Define chart-band helper
# ------------------------------------------------------------

def classify_chart_band(position):
    if pd.isna(position):
        return "Unavailable"
    elif position <= 10:
        return "Top 10"
    elif position <= 50:
        return "Top 50"
    else:
        return "Outside Top 50"


# ------------------------------------------------------------
# 3. Track-country chart bands
# ------------------------------------------------------------

track_chart_growth_df["first_chart_band"] = (
    track_chart_growth_df[
        "first_position"
    ].apply(classify_chart_band)
)

track_chart_growth_df["last_chart_band"] = (
    track_chart_growth_df[
        "last_position"
    ].apply(classify_chart_band)
)

track_chart_growth_df["entered_top_50"] = (
    (track_chart_growth_df["first_position"] > 50)
    &
    (track_chart_growth_df["last_position"] <= 50)
)

track_chart_growth_df["entered_top_10"] = (
    (track_chart_growth_df["first_position"] > 10)
    &
    (track_chart_growth_df["last_position"] <= 10)
)

track_chart_growth_df["remained_top_50"] = (
    (track_chart_growth_df["first_position"] <= 50)
    &
    (track_chart_growth_df["last_position"] <= 50)
)

track_chart_growth_df["remained_top_10"] = (
    (track_chart_growth_df["first_position"] <= 10)
    &
    (track_chart_growth_df["last_position"] <= 10)
)


# ------------------------------------------------------------
# 4. Artist-country chart bands
# ------------------------------------------------------------

artist_chart_growth_df["first_chart_band"] = (
    artist_chart_growth_df[
        "first_position"
    ].apply(classify_chart_band)
)

artist_chart_growth_df["last_chart_band"] = (
    artist_chart_growth_df[
        "last_position"
    ].apply(classify_chart_band)
)

artist_chart_growth_df["entered_top_50"] = (
    (artist_chart_growth_df["first_position"] > 50)
    &
    (artist_chart_growth_df["last_position"] <= 50)
)

artist_chart_growth_df["entered_top_10"] = (
    (artist_chart_growth_df["first_position"] > 10)
    &
    (artist_chart_growth_df["last_position"] <= 10)
)

artist_chart_growth_df["remained_top_50"] = (
    (artist_chart_growth_df["first_position"] <= 50)
    &
    (artist_chart_growth_df["last_position"] <= 50)
)

artist_chart_growth_df["remained_top_10"] = (
    (artist_chart_growth_df["first_position"] <= 10)
    &
    (artist_chart_growth_df["last_position"] <= 10)
)


# ------------------------------------------------------------
# 5. Track movement overview
# ------------------------------------------------------------

track_top_movement_df = pd.DataFrame(
    [
        {
            "Movement": "Entered Top 50",
            "Track-Country Histories": int(
                track_chart_growth_df[
                    "entered_top_50"
                ].sum()
            ),
        },
        {
            "Movement": "Entered Top 10",
            "Track-Country Histories": int(
                track_chart_growth_df[
                    "entered_top_10"
                ].sum()
            ),
        },
        {
            "Movement": "Remained Top 50",
            "Track-Country Histories": int(
                track_chart_growth_df[
                    "remained_top_50"
                ].sum()
            ),
        },
        {
            "Movement": "Remained Top 10",
            "Track-Country Histories": int(
                track_chart_growth_df[
                    "remained_top_10"
                ].sum()
            ),
        },
    ]
)

track_top_movement_df["Share (%)"] = (
    track_top_movement_df[
        "Track-Country Histories"
    ]
    / len(track_chart_growth_df)
    * 100
)

print()
print("Track-country Top-10 / Top-50 movement")
print("=" * 100)

display(track_top_movement_df)


# ------------------------------------------------------------
# 6. Artist movement overview
# ------------------------------------------------------------

artist_top_movement_df = pd.DataFrame(
    [
        {
            "Movement": "Entered Top 50",
            "Artist-Country Histories": int(
                artist_chart_growth_df[
                    "entered_top_50"
                ].sum()
            ),
        },
        {
            "Movement": "Entered Top 10",
            "Artist-Country Histories": int(
                artist_chart_growth_df[
                    "entered_top_10"
                ].sum()
            ),
        },
        {
            "Movement": "Remained Top 50",
            "Artist-Country Histories": int(
                artist_chart_growth_df[
                    "remained_top_50"
                ].sum()
            ),
        },
        {
            "Movement": "Remained Top 10",
            "Artist-Country Histories": int(
                artist_chart_growth_df[
                    "remained_top_10"
                ].sum()
            ),
        },
    ]
)

artist_top_movement_df["Share (%)"] = (
    artist_top_movement_df[
        "Artist-Country Histories"
    ]
    / len(artist_chart_growth_df)
    * 100
)

print()
print("Artist-country Top-10 / Top-50 movement")
print("=" * 100)

display(artist_top_movement_df)


# ------------------------------------------------------------
# 7. Final chart-band distribution
# ------------------------------------------------------------

track_final_band_distribution_df = (
    track_chart_growth_df[
        "last_chart_band"
    ]
    .value_counts()
    .rename_axis("Final Chart Band")
    .reset_index(name="Track-Country Histories")
)

track_final_band_distribution_df["Share (%)"] = (
    track_final_band_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_chart_growth_df)
    * 100
)

artist_final_band_distribution_df = (
    artist_chart_growth_df[
        "last_chart_band"
    ]
    .value_counts()
    .rename_axis("Final Chart Band")
    .reset_index(name="Artist-Country Histories")
)

artist_final_band_distribution_df["Share (%)"] = (
    artist_final_band_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_chart_growth_df)
    * 100
)

print()
print("Track-country final chart-band distribution")
print("=" * 100)

display(track_final_band_distribution_df)

print()
print("Artist-country final chart-band distribution")
print("=" * 100)

display(artist_final_band_distribution_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

top_movement_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_chart_growth_df
            ),
        },
        {
            "Diagnostic": "Track histories entered Top 50",
            "Observed Value": int(
                track_chart_growth_df[
                    "entered_top_50"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Track histories entered Top 10",
            "Observed Value": int(
                track_chart_growth_df[
                    "entered_top_10"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_growth_df
            ),
        },
        {
            "Diagnostic": "Artist histories entered Top 50",
            "Observed Value": int(
                artist_chart_growth_df[
                    "entered_top_50"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Artist histories entered Top 10",
            "Observed Value": int(
                artist_chart_growth_df[
                    "entered_top_10"
                ].sum()
            ),
        },
    ]
)

print()
print("Top-10 / Top-50 movement diagnostics")
print("=" * 100)

display(top_movement_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_6_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.1 completion",
            "Requirement": (
                "Chart Position Improvement must already be complete."
            ),
            "Observed Evidence": (
                f"Section 6.1 completion status: "
                f"{section_6_1_complete}"
            ),
            "Passed": section_6_1_complete,
        },
        {
            "Validation Area": "Track chart-band classification",
            "Requirement": (
                "Every retained track history must receive "
                "an initial and final chart band."
            ),
            "Observed Evidence": (
                f"{track_chart_growth_df['last_chart_band'].notna().sum():,} "
                "track final chart bands assigned"
            ),
            "Passed": bool(
                track_chart_growth_df[
                    [
                        "first_chart_band",
                        "last_chart_band",
                    ]
                ]
                .notna()
                .all()
                .all()
            ),
        },
        {
            "Validation Area": "Artist chart-band classification",
            "Requirement": (
                "Every retained artist history must receive "
                "an initial and final chart band."
            ),
            "Observed Evidence": (
                f"{artist_chart_growth_df['last_chart_band'].notna().sum():,} "
                "artist final chart bands assigned"
            ),
            "Passed": bool(
                artist_chart_growth_df[
                    [
                        "first_chart_band",
                        "last_chart_band",
                    ]
                ]
                .notna()
                .all()
                .all()
            ),
        },
        {
            "Validation Area": "Track Top-50 movement availability",
            "Requirement": (
                "Track histories must provide measurable "
                "Top-50 movement."
            ),
            "Observed Evidence": (
                f"{int(track_chart_growth_df['entered_top_50'].sum()):,} "
                "track histories entered Top 50"
            ),
            "Passed": bool(
                track_chart_growth_df[
                    "entered_top_50"
                ].any()
            ),
        },
        {
            "Validation Area": "Artist Top-50 movement availability",
            "Requirement": (
                "Artist histories must provide measurable "
                "Top-50 movement."
            ),
            "Observed Evidence": (
                f"{int(artist_chart_growth_df['entered_top_50'].sum()):,} "
                "artist histories entered Top 50"
            ),
            "Passed": bool(
                artist_chart_growth_df[
                    "entered_top_50"
                ].any()
            ),
        },
        {
            "Validation Area": "Top-10 movement support",
            "Requirement": (
                "Top-10 movement must be explicitly represented."
            ),
            "Observed Evidence": (
                f"{int(track_chart_growth_df['entered_top_10'].sum()):,} "
                "track entries / "
                f"{int(artist_chart_growth_df['entered_top_10'].sum()):,} "
                "artist entries"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Top-10 / Top-50 movement validation")
print("=" * 100)

display(section_6_2_validation_df)


section_6_2_complete = bool(
    section_6_2_validation_df[
        "Passed"
    ].all()
)

if not section_6_2_complete:
    raise AssertionError(
        "Section 6.2 Top-10 / Top-50 Movement validation failed."
    )

print()
print(
    f"Section 6.2 completion status: "
    f"{section_6_2_complete}"
)

print(
    "Top-10 and Top-50 chart movement has been "
    "assessed for PMIP track-country and artist-country histories."
)
```

Section 6.2 is complete.

### 6.3 Chart Persistence

#### Purpose

Measure how consistently tracks and artists remain active in strong chart positions across repeated observations.

This helps PMIP distinguish one-off chart appearances from more persistent chart performance.

```python
print("Assessing PMIP chart persistence")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6.2 completion
# ------------------------------------------------------------

if not section_6_2_complete:
    raise RuntimeError(
        "Section 6.2 Top-10 / Top-50 Movement must be complete "
        "before chart persistence is assessed."
    )

print(
    f"Section 6.2 completion status: "
    f"{section_6_2_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-country chart observations
# ------------------------------------------------------------

eligible_track_keys = (
    track_chart_growth_df[
        [
            "track_id",
            "country",
        ]
    ]
    .drop_duplicates()
)

track_chart_persistence_source_df = (
    historical_preparation_df
    .merge(
        eligible_track_keys,
        on=[
            "track_id",
            "country",
        ],
        how="inner",
    )
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Define track chart-position indicators
# ------------------------------------------------------------

track_chart_persistence_source_df["in_top_50"] = (
    track_chart_persistence_source_df["position"] <= 50
)

track_chart_persistence_source_df["in_top_10"] = (
    track_chart_persistence_source_df["position"] <= 10
)


# ------------------------------------------------------------
# 4. Aggregate track chart persistence
# ------------------------------------------------------------

track_chart_persistence_df = (
    track_chart_persistence_source_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .agg(
        chart_observations=("date", "count"),
        unique_chart_dates=("date", "nunique"),
        top_50_observations=("in_top_50", "sum"),
        top_10_observations=("in_top_10", "sum"),
        best_position=("position", "min"),
        median_position=("position", "median"),
    )
    .reset_index()
)

track_chart_persistence_df["top_50_share"] = (
    track_chart_persistence_df["top_50_observations"]
    / track_chart_persistence_df["chart_observations"]
)

track_chart_persistence_df["top_10_share"] = (
    track_chart_persistence_df["top_10_observations"]
    / track_chart_persistence_df["chart_observations"]
)


# ------------------------------------------------------------
# 5. Prepare artist-country chart observations
# ------------------------------------------------------------

eligible_artist_keys = (
    artist_chart_growth_df[
        [
            "artists",
            "country",
        ]
    ]
    .drop_duplicates()
)

artist_chart_persistence_source_df = (
    artist_country_date_history_df
    .merge(
        eligible_artist_keys,
        on=[
            "artists",
            "country",
        ],
        how="inner",
    )
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Define artist chart-position indicators
# ------------------------------------------------------------

artist_chart_persistence_source_df["in_top_50"] = (
    artist_chart_persistence_source_df["mean_position"] <= 50
)

artist_chart_persistence_source_df["in_top_10"] = (
    artist_chart_persistence_source_df["mean_position"] <= 10
)


# ------------------------------------------------------------
# 7. Aggregate artist chart persistence
# ------------------------------------------------------------

artist_chart_persistence_df = (
    artist_chart_persistence_source_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )
    .agg(
        chart_observations=("date", "count"),
        unique_chart_dates=("date", "nunique"),
        top_50_observations=("in_top_50", "sum"),
        top_10_observations=("in_top_10", "sum"),
        best_position=("best_position", "min"),
        median_position=("median_position", "median"),
    )
    .reset_index()
)

artist_chart_persistence_df["top_50_share"] = (
    artist_chart_persistence_df["top_50_observations"]
    / artist_chart_persistence_df["chart_observations"]
)

artist_chart_persistence_df["top_10_share"] = (
    artist_chart_persistence_df["top_10_observations"]
    / artist_chart_persistence_df["chart_observations"]
)


# ------------------------------------------------------------
# 8. Define persistence classes
# ------------------------------------------------------------

def classify_chart_persistence(
    top_50_share,
    top_10_share
):
    if pd.isna(top_50_share):
        return "Unavailable"

    if top_10_share >= 0.50:
        return "Strong Top-10 Persistence"

    elif top_50_share >= 0.75:
        return "Strong Top-50 Persistence"

    elif top_50_share >= 0.50:
        return "Moderate Top-50 Persistence"

    elif top_50_share > 0:
        return "Limited Chart Persistence"

    else:
        return "No Top-50 Persistence"


track_chart_persistence_df["chart_persistence_class"] = (
    track_chart_persistence_df.apply(
        lambda row: classify_chart_persistence(
            row["top_50_share"],
            row["top_10_share"],
        ),
        axis=1,
    )
)

artist_chart_persistence_df["chart_persistence_class"] = (
    artist_chart_persistence_df.apply(
        lambda row: classify_chart_persistence(
            row["top_50_share"],
            row["top_10_share"],
        ),
        axis=1,
    )
)


# ------------------------------------------------------------
# 9. Track persistence distribution
# ------------------------------------------------------------

track_chart_persistence_distribution_df = (
    track_chart_persistence_df[
        "chart_persistence_class"
    ]
    .value_counts()
    .rename_axis("Chart Persistence Class")
    .reset_index(name="Track-Country Histories")
)

track_chart_persistence_distribution_df["Share (%)"] = (
    track_chart_persistence_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_chart_persistence_df)
    * 100
)

print()
print("Track-country chart persistence distribution")
print("=" * 100)

display(track_chart_persistence_distribution_df)


# ------------------------------------------------------------
# 10. Artist persistence distribution
# ------------------------------------------------------------

artist_chart_persistence_distribution_df = (
    artist_chart_persistence_df[
        "chart_persistence_class"
    ]
    .value_counts()
    .rename_axis("Chart Persistence Class")
    .reset_index(name="Artist-Country Histories")
)

artist_chart_persistence_distribution_df["Share (%)"] = (
    artist_chart_persistence_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_chart_persistence_df)
    * 100
)

print()
print("Artist-country chart persistence distribution")
print("=" * 100)

display(artist_chart_persistence_distribution_df)


# ------------------------------------------------------------
# 11. Chart persistence overview
# ------------------------------------------------------------

chart_persistence_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_chart_persistence_df
            ),
        },
        {
            "Metric": "Track histories with >= 50% Top-50 presence",
            "Observed Value": int(
                (
                    track_chart_persistence_df[
                        "top_50_share"
                    ] >= 0.50
                ).sum()
            ),
        },
        {
            "Metric": "Track histories with >= 50% Top-10 presence",
            "Observed Value": int(
                (
                    track_chart_persistence_df[
                        "top_10_share"
                    ] >= 0.50
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_persistence_df
            ),
        },
        {
            "Metric": "Artist histories with >= 50% Top-50 presence",
            "Observed Value": int(
                (
                    artist_chart_persistence_df[
                        "top_50_share"
                    ] >= 0.50
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories with >= 50% Top-10 presence",
            "Observed Value": int(
                (
                    artist_chart_persistence_df[
                        "top_10_share"
                    ] >= 0.50
                ).sum()
            ),
        },
    ]
)

print()
print("Chart persistence overview")
print("=" * 100)

display(chart_persistence_overview_df)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

chart_persistence_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_chart_persistence_df
            ),
        },
        {
            "Diagnostic": "Track persistence classes assigned",
            "Observed Value": int(
                track_chart_persistence_df[
                    "chart_persistence_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track histories with Top-50 persistence",
            "Observed Value": int(
                (
                    track_chart_persistence_df[
                        "top_50_share"
                    ] >= 0.50
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_persistence_df
            ),
        },
        {
            "Diagnostic": "Artist persistence classes assigned",
            "Observed Value": int(
                artist_chart_persistence_df[
                    "chart_persistence_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories with Top-50 persistence",
            "Observed Value": int(
                (
                    artist_chart_persistence_df[
                        "top_50_share"
                    ] >= 0.50
                ).sum()
            ),
        },
    ]
)

print()
print("Chart persistence diagnostics")
print("=" * 100)

display(chart_persistence_diagnostics_df)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_6_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.2 completion",
            "Requirement": (
                "Top-10 / Top-50 Movement must already be complete."
            ),
            "Observed Evidence": (
                f"Section 6.2 completion status: "
                f"{section_6_2_complete}"
            ),
            "Passed": section_6_2_complete,
        },
        {
            "Validation Area": "Track persistence population",
            "Requirement": (
                "Eligible track-country histories must be available."
            ),
            "Observed Evidence": (
                f"{len(track_chart_persistence_df):,} "
                "track histories assessed"
            ),
            "Passed": len(track_chart_persistence_df) > 0,
        },
        {
            "Validation Area": "Artist persistence population",
            "Requirement": (
                "Eligible artist-country histories must be available."
            ),
            "Observed Evidence": (
                f"{len(artist_chart_persistence_df):,} "
                "artist histories assessed"
            ),
            "Passed": len(artist_chart_persistence_df) > 0,
        },
        {
            "Validation Area": "Track persistence-share validity",
            "Requirement": (
                "Track chart persistence shares must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Track Top-50 and Top-10 shares validated"
            ),
            "Passed": bool(
                track_chart_persistence_df[
                    "top_50_share"
                ].between(0, 1).all()
                and
                track_chart_persistence_df[
                    "top_10_share"
                ].between(0, 1).all()
            ),
        },
        {
            "Validation Area": "Artist persistence-share validity",
            "Requirement": (
                "Artist chart persistence shares must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist Top-50 and Top-10 shares validated"
            ),
            "Passed": bool(
                artist_chart_persistence_df[
                    "top_50_share"
                ].between(0, 1).all()
                and
                artist_chart_persistence_df[
                    "top_10_share"
                ].between(0, 1).all()
            ),
        },
        {
            "Validation Area": "Persistence classification completeness",
            "Requirement": (
                "Every retained history must receive a "
                "chart-persistence classification."
            ),
            "Observed Evidence": (
                f"{track_chart_persistence_df['chart_persistence_class'].notna().sum():,} "
                "track classes / "
                f"{artist_chart_persistence_df['chart_persistence_class'].notna().sum():,} "
                "artist classes"
            ),
            "Passed": bool(
                track_chart_persistence_df[
                    "chart_persistence_class"
                ].notna().all()
                and
                artist_chart_persistence_df[
                    "chart_persistence_class"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Chart persistence validation")
print("=" * 100)

display(section_6_3_validation_df)


section_6_3_complete = bool(
    section_6_3_validation_df[
        "Passed"
    ].all()
)

if not section_6_3_complete:
    raise AssertionError(
        "Section 6.3 Chart Persistence validation failed."
    )

print()
print(
    f"Section 6.3 completion status: "
    f"{section_6_3_complete}"
)

print(
    "Chart persistence has been assessed for "
    "PMIP track-country and artist-country histories."
)
```

#### Interpretation

Chart persistence was generally limited. Only **31,033 track-country histories** and **17,428 artist-country histories** maintained Top-50 presence for at least half of their observations.

Strong Top-10 persistence was rare, showing that sustained high chart performance is much less common than simply appearing on the chart.

Section 6.3 is complete.

### 6.4 Chart Activity Growth

#### Purpose

Measure whether chart activity becomes stronger over the course of each historical track-country and artist-country history.

This compares earlier and later observation periods to identify increasing, stable, or declining chart activity.

```python
print("Assessing PMIP chart activity growth")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6.3 completion
# ------------------------------------------------------------

if not section_6_3_complete:
    raise RuntimeError(
        "Section 6.3 Chart Persistence must be complete "
        "before chart activity growth is assessed."
    )

print(
    f"Section 6.3 completion status: "
    f"{section_6_3_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-country observation order
# ------------------------------------------------------------

track_activity_source_df = (
    track_chart_persistence_source_df
    .copy()
    .sort_values(
        by=[
            "track_id",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

track_activity_source_df["observation_number"] = (
    track_activity_source_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )
    .cumcount()
    + 1
)

track_activity_source_df["history_observations"] = (
    track_activity_source_df
    .groupby(
        [
            "track_id",
            "country",
        ]
    )["date"]
    .transform("count")
)

track_activity_source_df["history_midpoint"] = (
    track_activity_source_df[
        "history_observations"
    ] / 2
)

track_activity_source_df["history_period"] = (
    track_activity_source_df.apply(
        lambda row: (
            "Early"
            if row["observation_number"]
            <= row["history_midpoint"]
            else "Late"
        ),
        axis=1,
    )
)


# ------------------------------------------------------------
# 3. Summarise track chart activity
# ------------------------------------------------------------

track_activity_period_df = (
    track_activity_source_df
    .groupby(
        [
            "track_id",
            "country",
            "history_period",
        ]
    )
    .agg(
        chart_observations=("date", "count"),
        top_50_observations=("in_top_50", "sum"),
        top_10_observations=("in_top_10", "sum"),
    )
    .reset_index()
)

track_activity_pivot_df = (
    track_activity_period_df
    .pivot(
        index=[
            "track_id",
            "country",
        ],
        columns="history_period",
        values=[
            "chart_observations",
            "top_50_observations",
            "top_10_observations",
        ],
    )
)

track_activity_pivot_df.columns = [
    f"{metric.lower()}_{period.lower()}"
    for metric, period
    in track_activity_pivot_df.columns
]

track_activity_pivot_df = (
    track_activity_pivot_df
    .reset_index()
)


# ------------------------------------------------------------
# 4. Calculate track activity growth
# ------------------------------------------------------------

track_activity_pivot_df["chart_activity_change"] = (
    track_activity_pivot_df[
        "chart_observations_late"
    ]
    - track_activity_pivot_df[
        "chart_observations_early"
    ]
)

track_activity_pivot_df["top_50_activity_change"] = (
    track_activity_pivot_df[
        "top_50_observations_late"
    ]
    - track_activity_pivot_df[
        "top_50_observations_early"
    ]
)

track_activity_pivot_df["top_10_activity_change"] = (
    track_activity_pivot_df[
        "top_10_observations_late"
    ]
    - track_activity_pivot_df[
        "top_10_observations_early"
    ]
)


# ------------------------------------------------------------
# 5. Prepare artist-country observation order
# ------------------------------------------------------------

artist_activity_source_df = (
    artist_chart_persistence_source_df
    .copy()
    .sort_values(
        by=[
            "artists",
            "country",
            "date",
        ]
    )
    .reset_index(drop=True)
)

artist_activity_source_df["observation_number"] = (
    artist_activity_source_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )
    .cumcount()
    + 1
)

artist_activity_source_df["history_observations"] = (
    artist_activity_source_df
    .groupby(
        [
            "artists",
            "country",
        ]
    )["date"]
    .transform("count")
)

artist_activity_source_df["history_midpoint"] = (
    artist_activity_source_df[
        "history_observations"
    ] / 2
)

artist_activity_source_df["history_period"] = (
    artist_activity_source_df.apply(
        lambda row: (
            "Early"
            if row["observation_number"]
            <= row["history_midpoint"]
            else "Late"
        ),
        axis=1,
    )
)


# ------------------------------------------------------------
# 6. Summarise artist chart activity
# ------------------------------------------------------------

artist_activity_period_df = (
    artist_activity_source_df
    .groupby(
        [
            "artists",
            "country",
            "history_period",
        ]
    )
    .agg(
        chart_observations=("date", "count"),
        top_50_observations=("in_top_50", "sum"),
        top_10_observations=("in_top_10", "sum"),
    )
    .reset_index()
)

artist_activity_pivot_df = (
    artist_activity_period_df
    .pivot(
        index=[
            "artists",
            "country",
        ],
        columns="history_period",
        values=[
            "chart_observations",
            "top_50_observations",
            "top_10_observations",
        ],
    )
)

artist_activity_pivot_df.columns = [
    f"{metric.lower()}_{period.lower()}"
    for metric, period
    in artist_activity_pivot_df.columns
]

artist_activity_pivot_df = (
    artist_activity_pivot_df
    .reset_index()
)


# ------------------------------------------------------------
# 7. Calculate artist activity growth
# ------------------------------------------------------------

artist_activity_pivot_df["chart_activity_change"] = (
    artist_activity_pivot_df[
        "chart_observations_late"
    ]
    - artist_activity_pivot_df[
        "chart_observations_early"
    ]
)

artist_activity_pivot_df["top_50_activity_change"] = (
    artist_activity_pivot_df[
        "top_50_observations_late"
    ]
    - artist_activity_pivot_df[
        "top_50_observations_early"
    ]
)

artist_activity_pivot_df["top_10_activity_change"] = (
    artist_activity_pivot_df[
        "top_10_observations_late"
    ]
    - artist_activity_pivot_df[
        "top_10_observations_early"
    ]
)


# ------------------------------------------------------------
# 8. Define chart activity growth classes
# ------------------------------------------------------------

def classify_chart_activity_growth(row):
    if (
        row["top_50_activity_change"] > 0
        or row["top_10_activity_change"] > 0
    ):
        return "Increasing Chart Activity"

    elif (
        row["top_50_activity_change"] < 0
        or row["top_10_activity_change"] < 0
    ):
        return "Declining Chart Activity"

    else:
        return "Stable Chart Activity"


track_activity_pivot_df[
    "chart_activity_growth_class"
] = (
    track_activity_pivot_df.apply(
        classify_chart_activity_growth,
        axis=1,
    )
)

artist_activity_pivot_df[
    "chart_activity_growth_class"
] = (
    artist_activity_pivot_df.apply(
        classify_chart_activity_growth,
        axis=1,
    )
)


# ------------------------------------------------------------
# 9. Track activity growth distribution
# ------------------------------------------------------------

track_chart_activity_distribution_df = (
    track_activity_pivot_df[
        "chart_activity_growth_class"
    ]
    .value_counts()
    .rename_axis("Chart Activity Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_chart_activity_distribution_df["Share (%)"] = (
    track_chart_activity_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_activity_pivot_df)
    * 100
)

print()
print("Track-country chart activity growth distribution")
print("=" * 100)

display(track_chart_activity_distribution_df)


# ------------------------------------------------------------
# 10. Artist activity growth distribution
# ------------------------------------------------------------

artist_chart_activity_distribution_df = (
    artist_activity_pivot_df[
        "chart_activity_growth_class"
    ]
    .value_counts()
    .rename_axis("Chart Activity Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_chart_activity_distribution_df["Share (%)"] = (
    artist_chart_activity_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_activity_pivot_df)
    * 100
)

print()
print("Artist-country chart activity growth distribution")
print("=" * 100)

display(artist_chart_activity_distribution_df)


# ------------------------------------------------------------
# 11. Chart activity growth overview
# ------------------------------------------------------------

chart_activity_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_activity_pivot_df
            ),
        },
        {
            "Metric": "Track histories with increasing chart activity",
            "Observed Value": int(
                (
                    track_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Increasing Chart Activity"
                ).sum()
            ),
        },
        {
            "Metric": "Track histories with declining chart activity",
            "Observed Value": int(
                (
                    track_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Declining Chart Activity"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_activity_pivot_df
            ),
        },
        {
            "Metric": "Artist histories with increasing chart activity",
            "Observed Value": int(
                (
                    artist_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Increasing Chart Activity"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories with declining chart activity",
            "Observed Value": int(
                (
                    artist_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Declining Chart Activity"
                ).sum()
            ),
        },
    ]
)

print()
print("Chart activity growth overview")
print("=" * 100)

display(chart_activity_growth_overview_df)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

chart_activity_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_activity_pivot_df
            ),
        },
        {
            "Diagnostic": "Track activity classes assigned",
            "Observed Value": int(
                track_activity_pivot_df[
                    "chart_activity_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track increasing-activity histories",
            "Observed Value": int(
                (
                    track_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Increasing Chart Activity"
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_activity_pivot_df
            ),
        },
        {
            "Diagnostic": "Artist activity classes assigned",
            "Observed Value": int(
                artist_activity_pivot_df[
                    "chart_activity_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist increasing-activity histories",
            "Observed Value": int(
                (
                    artist_activity_pivot_df[
                        "chart_activity_growth_class"
                    ]
                    == "Increasing Chart Activity"
                ).sum()
            ),
        },
    ]
)

print()
print("Chart activity growth diagnostics")
print("=" * 100)

display(chart_activity_growth_diagnostics_df)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_6_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.3 completion",
            "Requirement": (
                "Chart Persistence must already be complete."
            ),
            "Observed Evidence": (
                f"Section 6.3 completion status: "
                f"{section_6_3_complete}"
            ),
            "Passed": section_6_3_complete,
        },
        {
            "Validation Area": "Track activity population",
            "Requirement": (
                "Track-country histories must be available "
                "for chart-activity analysis."
            ),
            "Observed Evidence": (
                f"{len(track_activity_pivot_df):,} "
                "track histories assessed"
            ),
            "Passed": len(track_activity_pivot_df) > 0,
        },
        {
            "Validation Area": "Artist activity population",
            "Requirement": (
                "Artist-country histories must be available "
                "for chart-activity analysis."
            ),
            "Observed Evidence": (
                f"{len(artist_activity_pivot_df):,} "
                "artist histories assessed"
            ),
            "Passed": len(artist_activity_pivot_df) > 0,
        },
        {
            "Validation Area": "Track activity classification",
            "Requirement": (
                "Every track history must receive a "
                "chart-activity growth class."
            ),
            "Observed Evidence": (
                f"{track_activity_pivot_df['chart_activity_growth_class'].notna().sum():,} "
                "track classes assigned"
            ),
            "Passed": bool(
                track_activity_pivot_df[
                    "chart_activity_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist activity classification",
            "Requirement": (
                "Every artist history must receive a "
                "chart-activity growth class."
            ),
            "Observed Evidence": (
                f"{artist_activity_pivot_df['chart_activity_growth_class'].notna().sum():,} "
                "artist classes assigned"
            ),
            "Passed": bool(
                artist_activity_pivot_df[
                    "chart_activity_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Activity direction differentiation",
            "Requirement": (
                "The analysis must distinguish increasing, "
                "stable and declining chart activity."
            ),
            "Observed Evidence": (
                "Increasing, Stable and Declining "
                "chart activity classes defined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Chart activity growth validation")
print("=" * 100)

display(section_6_4_validation_df)


section_6_4_complete = bool(
    section_6_4_validation_df[
        "Passed"
    ].all()
)

if not section_6_4_complete:
    raise AssertionError(
        "Section 6.4 Chart Activity Growth validation failed."
    )

print()
print(
    f"Section 6.4 completion status: "
    f"{section_6_4_complete}"
)

print(
    "Historical chart-activity growth has been assessed "
    "for PMIP track-country and artist-country histories."
)
```

#### Interpretation

Chart activity increased in **9,978 track-country histories** and **8,895 artist-country histories**.

Most histories were either stable or declining, showing that increasing chart activity is a relatively selective growth signal.

Section 6.4 is complete.

### 6.5 Sustained Chart Improvement

#### Purpose

Identify histories where chart improvement is supported by repeated, persistent, and increasing chart activity.

This combines chart-position improvement, Top-10 / Top-50 movement, persistence, and chart-activity growth into one interpretable sustained chart-growth classification.

```python
print("Identifying PMIP sustained chart improvement")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6.4 completion
# ------------------------------------------------------------

if not section_6_4_complete:
    raise RuntimeError(
        "Section 6.4 Chart Activity Growth must be complete "
        "before sustained chart improvement is identified."
    )

print(
    f"Section 6.4 completion status: "
    f"{section_6_4_complete}"
)


# ------------------------------------------------------------
# 2. Combine track-level chart evidence
# ------------------------------------------------------------

track_sustained_chart_df = (
    track_chart_growth_df
    .merge(
        track_chart_persistence_df[
            [
                "track_id",
                "country",
                "top_50_share",
                "top_10_share",
                "chart_persistence_class",
            ]
        ],
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
    .merge(
        track_activity_pivot_df[
            [
                "track_id",
                "country",
                "top_50_activity_change",
                "top_10_activity_change",
                "chart_activity_growth_class",
            ]
        ],
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 3. Combine artist-level chart evidence
# ------------------------------------------------------------

artist_sustained_chart_df = (
    artist_chart_growth_df
    .merge(
        artist_chart_persistence_df[
            [
                "artists",
                "country",
                "top_50_share",
                "top_10_share",
                "chart_persistence_class",
            ]
        ],
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
    .merge(
        artist_activity_pivot_df[
            [
                "artists",
                "country",
                "top_50_activity_change",
                "top_10_activity_change",
                "chart_activity_growth_class",
            ]
        ],
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 4. Define sustained chart-improvement classifier
# ------------------------------------------------------------

def classify_sustained_chart_improvement(row):

    position_improved = (
        row["chart_direction"] == "Improved"
    )

    top50_support = (
        bool(row["entered_top_50"])
        or bool(row["remained_top_50"])
    )

    top10_support = (
        bool(row["entered_top_10"])
        or bool(row["remained_top_10"])
    )

    persistence_support = (
        row["chart_persistence_class"]
        in [
            "Moderate Top-50 Persistence",
            "Strong Top-50 Persistence",
            "Strong Top-10 Persistence",
        ]
    )

    strong_persistence_support = (
        row["chart_persistence_class"]
        in [
            "Strong Top-50 Persistence",
            "Strong Top-10 Persistence",
        ]
    )

    activity_growth = (
        row["chart_activity_growth_class"]
        == "Increasing Chart Activity"
    )

    if (
        position_improved
        and top10_support
        and strong_persistence_support
        and activity_growth
    ):
        return "Strong Sustained Chart Growth"

    elif (
        position_improved
        and top50_support
        and persistence_support
        and activity_growth
    ):
        return "Sustained Chart Growth"

    elif (
        position_improved
        and (
            persistence_support
            or activity_growth
            or top50_support
        )
    ):
        return "Emerging Chart Growth"

    elif position_improved:
        return "Limited Chart Improvement"

    else:
        return "No Sustained Chart Growth"


track_sustained_chart_df[
    "sustained_chart_growth_class"
] = (
    track_sustained_chart_df.apply(
        classify_sustained_chart_improvement,
        axis=1,
    )
)

artist_sustained_chart_df[
    "sustained_chart_growth_class"
] = (
    artist_sustained_chart_df.apply(
        classify_sustained_chart_improvement,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track sustained chart-growth distribution
# ------------------------------------------------------------

track_sustained_chart_distribution_df = (
    track_sustained_chart_df[
        "sustained_chart_growth_class"
    ]
    .value_counts()
    .rename_axis("Sustained Chart Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_sustained_chart_distribution_df["Share (%)"] = (
    track_sustained_chart_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_sustained_chart_df)
    * 100
)

print()
print("Track-country sustained chart growth distribution")
print("=" * 100)

display(track_sustained_chart_distribution_df)


# ------------------------------------------------------------
# 6. Artist sustained chart-growth distribution
# ------------------------------------------------------------

artist_sustained_chart_distribution_df = (
    artist_sustained_chart_df[
        "sustained_chart_growth_class"
    ]
    .value_counts()
    .rename_axis("Sustained Chart Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_sustained_chart_distribution_df["Share (%)"] = (
    artist_sustained_chart_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_sustained_chart_df)
    * 100
)

print()
print("Artist-country sustained chart growth distribution")
print("=" * 100)

display(artist_sustained_chart_distribution_df)


# ------------------------------------------------------------
# 7. Sustained chart-growth overview
# ------------------------------------------------------------

sustained_chart_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_sustained_chart_df
            ),
        },
        {
            "Metric": "Track sustained chart-growth histories",
            "Observed Value": int(
                track_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track strong sustained chart-growth histories",
            "Observed Value": int(
                (
                    track_sustained_chart_df[
                        "sustained_chart_growth_class"
                    ]
                    == "Strong Sustained Chart Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_sustained_chart_df
            ),
        },
        {
            "Metric": "Artist sustained chart-growth histories",
            "Observed Value": int(
                artist_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist strong sustained chart-growth histories",
            "Observed Value": int(
                (
                    artist_sustained_chart_df[
                        "sustained_chart_growth_class"
                    ]
                    == "Strong Sustained Chart Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Sustained chart growth overview")
print("=" * 100)

display(sustained_chart_growth_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

sustained_chart_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_sustained_chart_df
            ),
        },
        {
            "Diagnostic": "Track sustained chart classes assigned",
            "Observed Value": int(
                track_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track sustained chart-growth histories",
            "Observed Value": int(
                track_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_sustained_chart_df
            ),
        },
        {
            "Diagnostic": "Artist sustained chart classes assigned",
            "Observed Value": int(
                artist_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist sustained chart-growth histories",
            "Observed Value": int(
                artist_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Sustained chart growth diagnostics")
print("=" * 100)

display(sustained_chart_growth_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_6_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6.4 completion",
            "Requirement": (
                "Chart Activity Growth must already be complete."
            ),
            "Observed Evidence": (
                f"Section 6.4 completion status: "
                f"{section_6_4_complete}"
            ),
            "Passed": section_6_4_complete,
        },
        {
            "Validation Area": "Track classification completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a sustained chart-growth classification."
            ),
            "Observed Evidence": (
                f"{track_sustained_chart_df['sustained_chart_growth_class'].notna().sum():,} "
                "track classes assigned"
            ),
            "Passed": bool(
                track_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist classification completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a sustained chart-growth classification."
            ),
            "Observed Evidence": (
                f"{artist_sustained_chart_df['sustained_chart_growth_class'].notna().sum():,} "
                "artist classes assigned"
            ),
            "Passed": bool(
                artist_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track sustained growth availability",
            "Requirement": (
                "At least some track histories must show "
                "sustained chart growth."
            ),
            "Observed Evidence": (
                f"{int(track_sustained_chart_df['sustained_chart_growth_class'].isin(['Sustained Chart Growth', 'Strong Sustained Chart Growth']).sum()):,} "
                "track sustained chart-growth histories"
            ),
            "Passed": bool(
                track_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "Artist sustained growth availability",
            "Requirement": (
                "At least some artist histories must show "
                "sustained chart growth."
            ),
            "Observed Evidence": (
                f"{int(artist_sustained_chart_df['sustained_chart_growth_class'].isin(['Sustained Chart Growth', 'Strong Sustained Chart Growth']).sum()):,} "
                "artist sustained chart-growth histories"
            ),
            "Passed": bool(
                artist_sustained_chart_df[
                    "sustained_chart_growth_class"
                ].isin(
                    [
                        "Sustained Chart Growth",
                        "Strong Sustained Chart Growth",
                    ]
                ).any()
            ),
        },
        {
            "Validation Area": "Multi-signal chart requirement",
            "Requirement": (
                "Sustained chart growth must combine chart movement, "
                "persistence and activity evidence."
            ),
            "Observed Evidence": (
                "Position improvement, Top-10 / Top-50 movement, "
                "persistence and chart activity combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Sustained chart growth validation")
print("=" * 100)

display(section_6_5_validation_df)


# ------------------------------------------------------------
# 10. Completion status
# ------------------------------------------------------------

section_6_5_complete = bool(
    section_6_5_validation_df[
        "Passed"
    ].all()
)

section_6_complete = bool(
    section_6_1_complete
    and section_6_2_complete
    and section_6_3_complete
    and section_6_4_complete
    and section_6_5_complete
)

if not section_6_5_complete:
    raise AssertionError(
        "Section 6.5 Sustained Chart Improvement validation failed."
    )

print()
print(
    f"Section 6.5 completion status: "
    f"{section_6_5_complete}"
)

print(
    f"Section 6 overall completion status: "
    f"{section_6_complete}"
)

print(
    "Sustained chart improvement has been identified "
    "using multiple validated PMIP chart-growth indicators."
)
```

## 7. Spike Versus Sustained Growth Detection

This section distinguishes temporary performance spikes from more persistent market-growth patterns.

It combines streaming growth, chart growth, persistence, volatility, and consistency evidence to produce a final interpretable growth-pattern classification.


### 7.1 Short-Term Spike Detection

#### Purpose

Identify histories where performance rises sharply but lacks enough persistence or stability to be treated as sustained growth.

This helps PMIP separate temporary spikes from more reliable growth signals.

```python
print("Detecting PMIP short-term growth spikes")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6 completion
# ------------------------------------------------------------

if not section_6_complete:
    raise RuntimeError(
        "Section 6 Chart Growth Analysis must be complete "
        "before spike detection begins."
    )

print(
    f"Section 6 overall completion status: "
    f"{section_6_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level combined growth evidence
# ------------------------------------------------------------

track_spike_detection_df = (
    track_sustained_growth_df
    .merge(
        track_sustained_chart_df[
            [
                "track_id",
                "country",
                "sustained_chart_growth_class",
                "chart_direction",
                "chart_activity_growth_class",
                "chart_persistence_class",
            ]
        ],
        on=[
            "track_id",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 3. Prepare artist-level combined growth evidence
# ------------------------------------------------------------

artist_spike_detection_df = (
    artist_sustained_growth_df
    .merge(
        artist_sustained_chart_df[
            [
                "artists",
                "country",
                "sustained_chart_growth_class",
                "chart_direction",
                "chart_activity_growth_class",
                "chart_persistence_class",
            ]
        ],
        on=[
            "artists",
            "country",
        ],
        how="left",
    )
)


# ------------------------------------------------------------
# 4. Define spike detection logic
# ------------------------------------------------------------

def classify_short_term_spike(row):

    strong_stream_change = (
        pd.notna(row["percentage_stream_change"])
        and row["percentage_stream_change"] >= 50
    )

    very_strong_stream_change = (
        pd.notna(row["percentage_stream_change"])
        and row["percentage_stream_change"] >= 100
    )

    weak_rolling_support = (
        pd.notna(row["positive_rolling_share"])
        and row["positive_rolling_share"] < 0.50
    )

    volatile_growth = (
        row["growth_stability_class"] == "High Volatility"
    )

    weak_stream_persistence = (
        row["sustained_growth_class"]
        in [
            "Emerging Growth",
            "Weak Positive Signal",
        ]
    )

    weak_chart_support = (
        row["sustained_chart_growth_class"]
        in [
            "No Sustained Chart Growth",
            "Limited Chart Improvement",
            "Emerging Chart Growth",
        ]
    )

    if (
        very_strong_stream_change
        and volatile_growth
        and weak_rolling_support
    ):
        return "Strong Short-Term Spike"

    elif (
        strong_stream_change
        and (
            volatile_growth
            or weak_rolling_support
        )
        and weak_stream_persistence
    ):
        return "Short-Term Spike"

    elif (
        strong_stream_change
        and weak_chart_support
        and weak_stream_persistence
    ):
        return "Possible Spike"

    else:
        return "No Spike Signal"


track_spike_detection_df["spike_class"] = (
    track_spike_detection_df.apply(
        classify_short_term_spike,
        axis=1,
    )
)

artist_spike_detection_df["spike_class"] = (
    artist_spike_detection_df.apply(
        classify_short_term_spike,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track spike distribution
# ------------------------------------------------------------

track_spike_distribution_df = (
    track_spike_detection_df[
        "spike_class"
    ]
    .value_counts()
    .rename_axis("Spike Class")
    .reset_index(name="Track-Country Histories")
)

track_spike_distribution_df["Share (%)"] = (
    track_spike_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_spike_detection_df)
    * 100
)

print()
print("Track-country short-term spike distribution")
print("=" * 100)

display(track_spike_distribution_df)


# ------------------------------------------------------------
# 6. Artist spike distribution
# ------------------------------------------------------------

artist_spike_distribution_df = (
    artist_spike_detection_df[
        "spike_class"
    ]
    .value_counts()
    .rename_axis("Spike Class")
    .reset_index(name="Artist-Country Histories")
)

artist_spike_distribution_df["Share (%)"] = (
    artist_spike_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_spike_detection_df)
    * 100
)

print()
print("Artist-country short-term spike distribution")
print("=" * 100)

display(artist_spike_distribution_df)


# ------------------------------------------------------------
# 7. Spike overview
# ------------------------------------------------------------

short_term_spike_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_spike_detection_df
            ),
        },
        {
            "Metric": "Track spike histories",
            "Observed Value": int(
                track_spike_detection_df[
                    "spike_class"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track possible spike histories",
            "Observed Value": int(
                (
                    track_spike_detection_df[
                        "spike_class"
                    ] == "Possible Spike"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_spike_detection_df
            ),
        },
        {
            "Metric": "Artist spike histories",
            "Observed Value": int(
                artist_spike_detection_df[
                    "spike_class"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist possible spike histories",
            "Observed Value": int(
                (
                    artist_spike_detection_df[
                        "spike_class"
                    ] == "Possible Spike"
                ).sum()
            ),
        },
    ]
)

print()
print("Short-term spike overview")
print("=" * 100)

display(short_term_spike_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

short_term_spike_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_spike_detection_df
            ),
        },
        {
            "Diagnostic": "Track spike classes assigned",
            "Observed Value": int(
                track_spike_detection_df[
                    "spike_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track short-term spike histories",
            "Observed Value": int(
                track_spike_detection_df[
                    "spike_class"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_spike_detection_df
            ),
        },
        {
            "Diagnostic": "Artist spike classes assigned",
            "Observed Value": int(
                artist_spike_detection_df[
                    "spike_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist short-term spike histories",
            "Observed Value": int(
                artist_spike_detection_df[
                    "spike_class"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Short-term spike diagnostics")
print("=" * 100)

display(short_term_spike_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_7_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 6 completion",
            "Requirement": (
                "Chart Growth Analysis must already be complete."
            ),
            "Observed Evidence": (
                f"Section 6 overall completion status: "
                f"{section_6_complete}"
            ),
            "Passed": section_6_complete,
        },
        {
            "Validation Area": "Track spike classification completeness",
            "Requirement": (
                "Every eligible track-country history must receive "
                "a spike classification."
            ),
            "Observed Evidence": (
                f"{track_spike_detection_df['spike_class'].notna().sum():,} "
                "track spike classes assigned"
            ),
            "Passed": bool(
                track_spike_detection_df[
                    "spike_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist spike classification completeness",
            "Requirement": (
                "Every eligible artist-country history must receive "
                "a spike classification."
            ),
            "Observed Evidence": (
                f"{artist_spike_detection_df['spike_class'].notna().sum():,} "
                "artist spike classes assigned"
            ),
            "Passed": bool(
                artist_spike_detection_df[
                    "spike_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track spike differentiation",
            "Requirement": (
                "The framework must distinguish spike and "
                "non-spike track histories."
            ),
            "Observed Evidence": (
                f"{track_spike_detection_df['spike_class'].nunique()} "
                "track spike classes observed"
            ),
            "Passed": bool(
                track_spike_detection_df[
                    "spike_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Artist spike differentiation",
            "Requirement": (
                "The framework must distinguish spike and "
                "non-spike artist histories."
            ),
            "Observed Evidence": (
                f"{artist_spike_detection_df['spike_class'].nunique()} "
                "artist spike classes observed"
            ),
            "Passed": bool(
                artist_spike_detection_df[
                    "spike_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Multi-signal spike requirement",
            "Requirement": (
                "Spike detection must combine magnitude, rolling "
                "behaviour, volatility and persistence evidence."
            ),
            "Observed Evidence": (
                "Percentage growth, rolling growth, volatility, "
                "stream persistence and chart support combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Short-term spike detection validation")
print("=" * 100)

display(section_7_1_validation_df)


section_7_1_complete = bool(
    section_7_1_validation_df[
        "Passed"
    ].all()
)

if not section_7_1_complete:
    raise AssertionError(
        "Section 7.1 Short-Term Spike Detection validation failed."
    )

print()
print(
    f"Section 7.1 completion status: "
    f"{section_7_1_complete}"
)

print(
    "Short-term growth spikes have been identified "
    "using multiple PMIP growth signals."
)
```

#### Interpretation

PMIP detected **15,939 track-country histories** and **18,117 artist-country histories** with short-term spike behaviour.

Most histories showed no spike signal, while strong short-term spikes were less common.

Section 7.1 is complete.

### 7.2 Growth Persistence

#### Purpose

Measure whether positive growth signals continue across repeated observations rather than appearing only briefly.

This helps PMIP identify histories where growth is supported by persistent streaming and chart behaviour.

```python
print("Assessing PMIP growth persistence")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 7.1 completion
# ------------------------------------------------------------

if not section_7_1_complete:
    raise RuntimeError(
        "Section 7.1 Short-Term Spike Detection must be complete "
        "before growth persistence is assessed."
    )

print(
    f"Section 7.1 completion status: "
    f"{section_7_1_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level persistence evidence
# ------------------------------------------------------------

track_growth_persistence_df = (
    track_spike_detection_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level persistence evidence
# ------------------------------------------------------------

artist_growth_persistence_df = (
    artist_spike_detection_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Define persistence classifier
# ------------------------------------------------------------

def classify_growth_persistence(row):

    strong_stream_persistence = (
        row["sustained_growth_class"]
        == "Strong Sustained Growth"
    )

    stream_persistence = (
        row["sustained_growth_class"]
        in [
            "Sustained Growth",
            "Strong Sustained Growth",
        ]
    )

    rolling_persistence = (
        pd.notna(row["positive_rolling_share"])
        and row["positive_rolling_share"] >= 0.50
    )

    strong_rolling_persistence = (
        pd.notna(row["positive_rolling_share"])
        and row["positive_rolling_share"] >= 0.75
    )

    chart_persistence = (
        row["chart_persistence_class"]
        in [
            "Moderate Top-50 Persistence",
            "Strong Top-50 Persistence",
            "Strong Top-10 Persistence",
        ]
    )

    strong_chart_persistence = (
        row["chart_persistence_class"]
        in [
            "Strong Top-50 Persistence",
            "Strong Top-10 Persistence",
        ]
    )

    spike_present = (
        row["spike_class"]
        in [
            "Short-Term Spike",
            "Strong Short-Term Spike",
        ]
    )

    if (
        strong_stream_persistence
        and strong_rolling_persistence
        and strong_chart_persistence
        and not spike_present
    ):
        return "Strong Persistent Growth"

    elif (
        stream_persistence
        and rolling_persistence
        and chart_persistence
        and not spike_present
    ):
        return "Persistent Growth"

    elif (
        stream_persistence
        or rolling_persistence
        or chart_persistence
    ):
        return "Partial Persistence"

    else:
        return "Low Persistence"


track_growth_persistence_df[
    "growth_persistence_class"
] = (
    track_growth_persistence_df.apply(
        classify_growth_persistence,
        axis=1,
    )
)

artist_growth_persistence_df[
    "growth_persistence_class"
] = (
    artist_growth_persistence_df.apply(
        classify_growth_persistence,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track persistence distribution
# ------------------------------------------------------------

track_growth_persistence_distribution_df = (
    track_growth_persistence_df[
        "growth_persistence_class"
    ]
    .value_counts()
    .rename_axis("Growth Persistence Class")
    .reset_index(name="Track-Country Histories")
)

track_growth_persistence_distribution_df["Share (%)"] = (
    track_growth_persistence_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_growth_persistence_df)
    * 100
)

print()
print("Track-country growth persistence distribution")
print("=" * 100)

display(track_growth_persistence_distribution_df)


# ------------------------------------------------------------
# 6. Artist persistence distribution
# ------------------------------------------------------------

artist_growth_persistence_distribution_df = (
    artist_growth_persistence_df[
        "growth_persistence_class"
    ]
    .value_counts()
    .rename_axis("Growth Persistence Class")
    .reset_index(name="Artist-Country Histories")
)

artist_growth_persistence_distribution_df["Share (%)"] = (
    artist_growth_persistence_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_growth_persistence_df)
    * 100
)

print()
print("Artist-country growth persistence distribution")
print("=" * 100)

display(artist_growth_persistence_distribution_df)


# ------------------------------------------------------------
# 7. Growth persistence overview
# ------------------------------------------------------------

growth_persistence_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_growth_persistence_df
            ),
        },
        {
            "Metric": "Track persistent-growth histories",
            "Observed Value": int(
                track_growth_persistence_df[
                    "growth_persistence_class"
                ].isin(
                    [
                        "Persistent Growth",
                        "Strong Persistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track strong persistent-growth histories",
            "Observed Value": int(
                (
                    track_growth_persistence_df[
                        "growth_persistence_class"
                    ] == "Strong Persistent Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_growth_persistence_df
            ),
        },
        {
            "Metric": "Artist persistent-growth histories",
            "Observed Value": int(
                artist_growth_persistence_df[
                    "growth_persistence_class"
                ].isin(
                    [
                        "Persistent Growth",
                        "Strong Persistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist strong persistent-growth histories",
            "Observed Value": int(
                (
                    artist_growth_persistence_df[
                        "growth_persistence_class"
                    ] == "Strong Persistent Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Growth persistence overview")
print("=" * 100)

display(growth_persistence_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

growth_persistence_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_growth_persistence_df
            ),
        },
        {
            "Diagnostic": "Track persistence classes assigned",
            "Observed Value": int(
                track_growth_persistence_df[
                    "growth_persistence_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track persistent-growth histories",
            "Observed Value": int(
                track_growth_persistence_df[
                    "growth_persistence_class"
                ].isin(
                    [
                        "Persistent Growth",
                        "Strong Persistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_growth_persistence_df
            ),
        },
        {
            "Diagnostic": "Artist persistence classes assigned",
            "Observed Value": int(
                artist_growth_persistence_df[
                    "growth_persistence_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist persistent-growth histories",
            "Observed Value": int(
                artist_growth_persistence_df[
                    "growth_persistence_class"
                ].isin(
                    [
                        "Persistent Growth",
                        "Strong Persistent Growth",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Growth persistence diagnostics")
print("=" * 100)

display(growth_persistence_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_7_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.1 completion",
            "Requirement": (
                "Short-Term Spike Detection must already be complete."
            ),
            "Observed Evidence": (
                f"Section 7.1 completion status: "
                f"{section_7_1_complete}"
            ),
            "Passed": section_7_1_complete,
        },
        {
            "Validation Area": "Track persistence classification",
            "Requirement": (
                "Every track-country history must receive "
                "a growth-persistence classification."
            ),
            "Observed Evidence": (
                f"{track_growth_persistence_df['growth_persistence_class'].notna().sum():,} "
                "track persistence classes assigned"
            ),
            "Passed": bool(
                track_growth_persistence_df[
                    "growth_persistence_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist persistence classification",
            "Requirement": (
                "Every artist-country history must receive "
                "a growth-persistence classification."
            ),
            "Observed Evidence": (
                f"{artist_growth_persistence_df['growth_persistence_class'].notna().sum():,} "
                "artist persistence classes assigned"
            ),
            "Passed": bool(
                artist_growth_persistence_df[
                    "growth_persistence_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track persistence differentiation",
            "Requirement": (
                "Track histories must be separated into "
                "different persistence levels."
            ),
            "Observed Evidence": (
                f"{track_growth_persistence_df['growth_persistence_class'].nunique()} "
                "track persistence classes observed"
            ),
            "Passed": bool(
                track_growth_persistence_df[
                    "growth_persistence_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Artist persistence differentiation",
            "Requirement": (
                "Artist histories must be separated into "
                "different persistence levels."
            ),
            "Observed Evidence": (
                f"{artist_growth_persistence_df['growth_persistence_class'].nunique()} "
                "artist persistence classes observed"
            ),
            "Passed": bool(
                artist_growth_persistence_df[
                    "growth_persistence_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Multi-signal persistence requirement",
            "Requirement": (
                "Growth persistence must combine streaming, rolling, "
                "chart persistence and spike evidence."
            ),
            "Observed Evidence": (
                "Sustained streaming growth, rolling growth, chart "
                "persistence and spike evidence combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Growth persistence validation")
print("=" * 100)

display(section_7_2_validation_df)


# ------------------------------------------------------------
# 10. Completion status
# ------------------------------------------------------------

section_7_2_complete = bool(
    section_7_2_validation_df[
        "Passed"
    ].all()
)

if not section_7_2_complete:
    raise AssertionError(
        "Section 7.2 Growth Persistence validation failed."
    )

print()
print(
    f"Section 7.2 completion status: "
    f"{section_7_2_complete}"
)

print(
    "Growth persistence has been assessed using "
    "combined PMIP streaming, rolling and chart evidence."
)
```

#### Interpretation

Persistent growth was relatively rare. PMIP identified **2,718 track-country histories** and **1,757 artist-country histories** with persistent growth.

Most histories showed low or only partial persistence, which supports the need to separate durable growth from temporary movement.



### 7.3 Volatility Assessment

#### Purpose

Assess how unstable or variable growth behaviour is across track-country and artist-country histories.

This helps PMIP reduce confidence in highly erratic growth patterns and better distinguish sustained growth from unstable spikes.

```python
print("Assessing PMIP growth volatility")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 7.2 completion
# ------------------------------------------------------------

if not section_7_2_complete:
    raise RuntimeError(
        "Section 7.2 Growth Persistence must be complete "
        "before volatility assessment begins."
    )

print(
    f"Section 7.2 completion status: "
    f"{section_7_2_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track volatility evidence
# ------------------------------------------------------------

track_volatility_df = (
    track_growth_persistence_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist volatility evidence
# ------------------------------------------------------------

artist_volatility_df = (
    artist_growth_persistence_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Define volatility classification
# ------------------------------------------------------------

def classify_growth_volatility(row):

    if pd.isna(row["growth_rate_std"]):
        return "Insufficient Volatility Evidence"

    if row["growth_stability_class"] == "High Volatility":
        return "High Volatility"

    elif row["growth_stability_class"] == "Moderate Stability":
        return "Moderate Volatility"

    elif row["growth_stability_class"] == "High Stability":
        return "Low Volatility"

    else:
        return "Moderate Volatility"


track_volatility_df[
    "growth_volatility_class"
] = (
    track_volatility_df.apply(
        classify_growth_volatility,
        axis=1,
    )
)

artist_volatility_df[
    "growth_volatility_class"
] = (
    artist_volatility_df.apply(
        classify_growth_volatility,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Add volatility risk flags
# ------------------------------------------------------------

track_volatility_df["high_volatility_flag"] = (
    track_volatility_df[
        "growth_volatility_class"
    ] == "High Volatility"
)

artist_volatility_df["high_volatility_flag"] = (
    artist_volatility_df[
        "growth_volatility_class"
    ] == "High Volatility"
)


# ------------------------------------------------------------
# 6. Track volatility distribution
# ------------------------------------------------------------

track_volatility_distribution_df = (
    track_volatility_df[
        "growth_volatility_class"
    ]
    .value_counts()
    .rename_axis("Growth Volatility Class")
    .reset_index(name="Track-Country Histories")
)

track_volatility_distribution_df["Share (%)"] = (
    track_volatility_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_volatility_df)
    * 100
)

print()
print("Track-country growth volatility distribution")
print("=" * 100)

display(track_volatility_distribution_df)


# ------------------------------------------------------------
# 7. Artist volatility distribution
# ------------------------------------------------------------

artist_volatility_distribution_df = (
    artist_volatility_df[
        "growth_volatility_class"
    ]
    .value_counts()
    .rename_axis("Growth Volatility Class")
    .reset_index(name="Artist-Country Histories")
)

artist_volatility_distribution_df["Share (%)"] = (
    artist_volatility_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_volatility_df)
    * 100
)

print()
print("Artist-country growth volatility distribution")
print("=" * 100)

display(artist_volatility_distribution_df)


# ------------------------------------------------------------
# 8. Volatility overview
# ------------------------------------------------------------

growth_volatility_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_volatility_df
            ),
        },
        {
            "Metric": "Track high-volatility histories",
            "Observed Value": int(
                track_volatility_df[
                    "high_volatility_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Median track growth-rate std",
            "Observed Value": float(
                track_volatility_df[
                    "growth_rate_std"
                ].median()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_volatility_df
            ),
        },
        {
            "Metric": "Artist high-volatility histories",
            "Observed Value": int(
                artist_volatility_df[
                    "high_volatility_flag"
                ].sum()
            ),
        },
        {
            "Metric": "Median artist growth-rate std",
            "Observed Value": float(
                artist_volatility_df[
                    "growth_rate_std"
                ].median()
            ),
        },
    ]
)

print()
print("Growth volatility overview")
print("=" * 100)

display(growth_volatility_overview_df)


# ------------------------------------------------------------
# 9. Spike and volatility relationship
# ------------------------------------------------------------

track_spike_volatility_df = pd.crosstab(
    track_volatility_df["spike_class"],
    track_volatility_df["growth_volatility_class"]
)

artist_spike_volatility_df = pd.crosstab(
    artist_volatility_df["spike_class"],
    artist_volatility_df["growth_volatility_class"]
)

print()
print("Track spike versus volatility")
print("=" * 100)

display(track_spike_volatility_df)

print()
print("Artist spike versus volatility")
print("=" * 100)

display(artist_spike_volatility_df)


# ------------------------------------------------------------
# 10. Diagnostics
# ------------------------------------------------------------

growth_volatility_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_volatility_df
            ),
        },
        {
            "Diagnostic": "Track volatility classes assigned",
            "Observed Value": int(
                track_volatility_df[
                    "growth_volatility_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track high-volatility histories",
            "Observed Value": int(
                track_volatility_df[
                    "high_volatility_flag"
                ].sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_volatility_df
            ),
        },
        {
            "Diagnostic": "Artist volatility classes assigned",
            "Observed Value": int(
                artist_volatility_df[
                    "growth_volatility_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist high-volatility histories",
            "Observed Value": int(
                artist_volatility_df[
                    "high_volatility_flag"
                ].sum()
            ),
        },
    ]
)

print()
print("Growth volatility diagnostics")
print("=" * 100)

display(growth_volatility_diagnostics_df)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

section_7_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.2 completion",
            "Requirement": (
                "Growth Persistence must already be complete."
            ),
            "Observed Evidence": (
                f"Section 7.2 completion status: "
                f"{section_7_2_complete}"
            ),
            "Passed": section_7_2_complete,
        },
        {
            "Validation Area": "Track volatility classification",
            "Requirement": (
                "Every track-country history must receive "
                "a volatility classification."
            ),
            "Observed Evidence": (
                f"{track_volatility_df['growth_volatility_class'].notna().sum():,} "
                "track volatility classes assigned"
            ),
            "Passed": bool(
                track_volatility_df[
                    "growth_volatility_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist volatility classification",
            "Requirement": (
                "Every artist-country history must receive "
                "a volatility classification."
            ),
            "Observed Evidence": (
                f"{artist_volatility_df['growth_volatility_class'].notna().sum():,} "
                "artist volatility classes assigned"
            ),
            "Passed": bool(
                artist_volatility_df[
                    "growth_volatility_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track volatility differentiation",
            "Requirement": (
                "Track histories must be separated into "
                "different volatility levels."
            ),
            "Observed Evidence": (
                f"{track_volatility_df['growth_volatility_class'].nunique()} "
                "track volatility classes observed"
            ),
            "Passed": bool(
                track_volatility_df[
                    "growth_volatility_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Artist volatility differentiation",
            "Requirement": (
                "Artist histories must be separated into "
                "different volatility levels."
            ),
            "Observed Evidence": (
                f"{artist_volatility_df['growth_volatility_class'].nunique()} "
                "artist volatility classes observed"
            ),
            "Passed": bool(
                artist_volatility_df[
                    "growth_volatility_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Spike-volatility relationship",
            "Requirement": (
                "Volatility assessment must remain available "
                "alongside spike classifications."
            ),
            "Observed Evidence": (
                "Spike and volatility cross-tabulations prepared "
                "for track and artist histories"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Growth volatility validation")
print("=" * 100)

display(section_7_3_validation_df)


# ------------------------------------------------------------
# 12. Completion status
# ------------------------------------------------------------

section_7_3_complete = bool(
    section_7_3_validation_df[
        "Passed"
    ].all()
)

if not section_7_3_complete:
    raise AssertionError(
        "Section 7.3 Volatility Assessment validation failed."
    )

print()
print(
    f"Section 7.3 completion status: "
    f"{section_7_3_complete}"
)

print(
    "Growth volatility has been assessed for PMIP "
    "track-country and artist-country histories."
)
```

#### Interpretation

Around one-quarter of both track-country and artist-country histories showed **high volatility**.

Strong short-term spikes were concentrated within the high-volatility group, showing that volatility is an important signal for separating unstable spikes from more reliable growth.



### 7.4 Consistency Rules

#### Purpose

Define clear rules for deciding whether growth behaviour is consistent enough to support a sustained-growth interpretation.

This combines persistence, volatility, spike behaviour, streaming growth, and chart growth into a consistent decision framework.

```python
print("Applying PMIP growth consistency rules")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 7.3 completion
# ------------------------------------------------------------

if not section_7_3_complete:
    raise RuntimeError(
        "Section 7.3 Volatility Assessment must be complete "
        "before consistency rules are applied."
    )

print(
    f"Section 7.3 completion status: "
    f"{section_7_3_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level consistency evidence
# ------------------------------------------------------------

track_consistency_df = (
    track_volatility_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level consistency evidence
# ------------------------------------------------------------

artist_consistency_df = (
    artist_volatility_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Define consistency rules
# ------------------------------------------------------------

def classify_growth_consistency(row):

    persistent_growth = (
        row["growth_persistence_class"]
        in [
            "Persistent Growth",
            "Strong Persistent Growth",
        ]
    )

    partial_persistence = (
        row["growth_persistence_class"]
        == "Partial Persistence"
    )

    low_volatility = (
        row["growth_volatility_class"]
        == "Low Volatility"
    )

    moderate_volatility = (
        row["growth_volatility_class"]
        == "Moderate Volatility"
    )

    high_volatility = (
        row["growth_volatility_class"]
        == "High Volatility"
    )

    strong_spike = (
        row["spike_class"]
        == "Strong Short-Term Spike"
    )

    spike_present = (
        row["spike_class"]
        in [
            "Short-Term Spike",
            "Strong Short-Term Spike",
            "Possible Spike",
        ]
    )

    sustained_streaming = (
        row["sustained_growth_class"]
        in [
            "Sustained Growth",
            "Strong Sustained Growth",
        ]
    )

    sustained_chart = (
        row["sustained_chart_growth_class"]
        in [
            "Sustained Chart Growth",
            "Strong Sustained Chart Growth",
        ]
    )

    if (
        persistent_growth
        and low_volatility
        and sustained_streaming
        and sustained_chart
        and not spike_present
    ):
        return "Highly Consistent Growth"

    elif (
        persistent_growth
        and not high_volatility
        and not strong_spike
    ):
        return "Consistent Growth"

    elif (
        partial_persistence
        and moderate_volatility
        and not strong_spike
    ):
        return "Moderately Consistent Growth"

    elif (
        high_volatility
        or strong_spike
    ):
        return "Unstable Growth"

    else:
        return "Low Consistency"


track_consistency_df[
    "growth_consistency_class"
] = (
    track_consistency_df.apply(
        classify_growth_consistency,
        axis=1,
    )
)

artist_consistency_df[
    "growth_consistency_class"
] = (
    artist_consistency_df.apply(
        classify_growth_consistency,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track consistency distribution
# ------------------------------------------------------------

track_consistency_distribution_df = (
    track_consistency_df[
        "growth_consistency_class"
    ]
    .value_counts()
    .rename_axis("Growth Consistency Class")
    .reset_index(name="Track-Country Histories")
)

track_consistency_distribution_df["Share (%)"] = (
    track_consistency_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_consistency_df)
    * 100
)

print()
print("Track-country growth consistency distribution")
print("=" * 100)

display(track_consistency_distribution_df)


# ------------------------------------------------------------
# 6. Artist consistency distribution
# ------------------------------------------------------------

artist_consistency_distribution_df = (
    artist_consistency_df[
        "growth_consistency_class"
    ]
    .value_counts()
    .rename_axis("Growth Consistency Class")
    .reset_index(name="Artist-Country Histories")
)

artist_consistency_distribution_df["Share (%)"] = (
    artist_consistency_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_consistency_df)
    * 100
)

print()
print("Artist-country growth consistency distribution")
print("=" * 100)

display(artist_consistency_distribution_df)


# ------------------------------------------------------------
# 7. Consistency overview
# ------------------------------------------------------------

growth_consistency_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_consistency_df
            ),
        },
        {
            "Metric": "Track consistent-growth histories",
            "Observed Value": int(
                track_consistency_df[
                    "growth_consistency_class"
                ].isin(
                    [
                        "Consistent Growth",
                        "Highly Consistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track unstable-growth histories",
            "Observed Value": int(
                (
                    track_consistency_df[
                        "growth_consistency_class"
                    ] == "Unstable Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_consistency_df
            ),
        },
        {
            "Metric": "Artist consistent-growth histories",
            "Observed Value": int(
                artist_consistency_df[
                    "growth_consistency_class"
                ].isin(
                    [
                        "Consistent Growth",
                        "Highly Consistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist unstable-growth histories",
            "Observed Value": int(
                (
                    artist_consistency_df[
                        "growth_consistency_class"
                    ] == "Unstable Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Growth consistency overview")
print("=" * 100)

display(growth_consistency_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

growth_consistency_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_consistency_df
            ),
        },
        {
            "Diagnostic": "Track consistency classes assigned",
            "Observed Value": int(
                track_consistency_df[
                    "growth_consistency_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track consistent-growth histories",
            "Observed Value": int(
                track_consistency_df[
                    "growth_consistency_class"
                ].isin(
                    [
                        "Consistent Growth",
                        "Highly Consistent Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_consistency_df
            ),
        },
        {
            "Diagnostic": "Artist consistency classes assigned",
            "Observed Value": int(
                artist_consistency_df[
                    "growth_consistency_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist consistent-growth histories",
            "Observed Value": int(
                artist_consistency_df[
                    "growth_consistency_class"
                ].isin(
                    [
                        "Consistent Growth",
                        "Highly Consistent Growth",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Growth consistency diagnostics")
print("=" * 100)

display(growth_consistency_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_7_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.3 completion",
            "Requirement": (
                "Volatility Assessment must already be complete."
            ),
            "Observed Evidence": (
                f"Section 7.3 completion status: "
                f"{section_7_3_complete}"
            ),
            "Passed": section_7_3_complete,
        },
        {
            "Validation Area": "Track consistency classification",
            "Requirement": (
                "Every track-country history must receive "
                "a consistency classification."
            ),
            "Observed Evidence": (
                f"{track_consistency_df['growth_consistency_class'].notna().sum():,} "
                "track consistency classes assigned"
            ),
            "Passed": bool(
                track_consistency_df[
                    "growth_consistency_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist consistency classification",
            "Requirement": (
                "Every artist-country history must receive "
                "a consistency classification."
            ),
            "Observed Evidence": (
                f"{artist_consistency_df['growth_consistency_class'].notna().sum():,} "
                "artist consistency classes assigned"
            ),
            "Passed": bool(
                artist_consistency_df[
                    "growth_consistency_class"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track consistency differentiation",
            "Requirement": (
                "Track histories must be separated into "
                "different consistency levels."
            ),
            "Observed Evidence": (
                f"{track_consistency_df['growth_consistency_class'].nunique()} "
                "track consistency classes observed"
            ),
            "Passed": bool(
                track_consistency_df[
                    "growth_consistency_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Artist consistency differentiation",
            "Requirement": (
                "Artist histories must be separated into "
                "different consistency levels."
            ),
            "Observed Evidence": (
                f"{artist_consistency_df['growth_consistency_class'].nunique()} "
                "artist consistency classes observed"
            ),
            "Passed": bool(
                artist_consistency_df[
                    "growth_consistency_class"
                ].nunique() >= 2
            ),
        },
        {
            "Validation Area": "Multi-signal consistency requirement",
            "Requirement": (
                "Consistency must combine persistence, volatility, "
                "spike, streaming and chart evidence."
            ),
            "Observed Evidence": (
                "Persistence, volatility, spike, streaming "
                "and chart signals combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Growth consistency validation")
print("=" * 100)

display(section_7_4_validation_df)


# ------------------------------------------------------------
# 10. Completion status
# ------------------------------------------------------------

section_7_4_complete = bool(
    section_7_4_validation_df[
        "Passed"
    ].all()
)

if not section_7_4_complete:
    raise AssertionError(
        "Section 7.4 Consistency Rules validation failed."
    )

print()
print(
    f"Section 7.4 completion status: "
    f"{section_7_4_complete}"
)

print(
    "PMIP growth consistency rules have been applied "
    "to track-country and artist-country histories."
)
```

#### Interpretation

Most histories showed either low consistency or unstable growth, while only a small share reached consistent or highly consistent growth.

PMIP identified **2,718 track-country histories** and **1,757 artist-country histories** with consistent growth, showing that reliable growth is much rarer than noisy movement.



### 7.5 Final Growth Pattern Classification

#### Purpose

Combine spike, persistence, volatility, consistency, streaming-growth, and chart-growth evidence into one final PMIP growth-pattern classification.

This provides a clear summary of whether each history shows sustained growth, emerging growth, unstable growth, a short-term spike, decline, or insufficient evidence.

```python
print("Classifying final PMIP growth patterns")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 7.4 completion
# ------------------------------------------------------------

if not section_7_4_complete:
    raise RuntimeError(
        "Section 7.4 Consistency Rules must be complete "
        "before final growth patterns are classified."
    )

print(
    f"Section 7.4 completion status: "
    f"{section_7_4_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level final classification evidence
# ------------------------------------------------------------

track_final_growth_pattern_df = (
    track_consistency_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level final classification evidence
# ------------------------------------------------------------

artist_final_growth_pattern_df = (
    artist_consistency_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Define final growth-pattern classifier
# ------------------------------------------------------------

def classify_final_growth_pattern(row):

    strong_spike = (
        row["spike_class"]
        == "Strong Short-Term Spike"
    )

    spike_present = (
        row["spike_class"]
        in [
            "Short-Term Spike",
            "Strong Short-Term Spike",
        ]
    )

    highly_consistent = (
        row["growth_consistency_class"]
        == "Highly Consistent Growth"
    )

    consistent = (
        row["growth_consistency_class"]
        in [
            "Consistent Growth",
            "Highly Consistent Growth",
        ]
    )

    moderately_consistent = (
        row["growth_consistency_class"]
        == "Moderately Consistent Growth"
    )

    persistent = (
        row["growth_persistence_class"]
        in [
            "Persistent Growth",
            "Strong Persistent Growth",
        ]
    )

    partial_persistence = (
        row["growth_persistence_class"]
        == "Partial Persistence"
    )

    high_volatility = (
        row["growth_volatility_class"]
        == "High Volatility"
    )

    sustained_streaming = (
        row["sustained_growth_class"]
        in [
            "Sustained Growth",
            "Strong Sustained Growth",
        ]
    )

    sustained_chart = (
        row["sustained_chart_growth_class"]
        in [
            "Sustained Chart Growth",
            "Strong Sustained Chart Growth",
        ]
    )

    positive_stream_change = (
        pd.notna(row["percentage_stream_change"])
        and row["percentage_stream_change"] > 0
    )

    declining_stream = (
        pd.notna(row["percentage_stream_change"])
        and row["percentage_stream_change"] < 0
    )

    chart_decline = (
        row["chart_direction"] == "Declined"
    )

    insufficient_volatility = (
        row["growth_volatility_class"]
        == "Insufficient Volatility Evidence"
    )

    # Strongest reliable growth
    if (
        highly_consistent
        and persistent
        and sustained_streaming
        and sustained_chart
        and not spike_present
    ):
        return "Strong Sustained Growth"

    # Reliable sustained growth
    elif (
        consistent
        and persistent
        and (
            sustained_streaming
            or sustained_chart
        )
        and not strong_spike
    ):
        return "Sustained Growth"

    # Strong spike overrides weaker positive evidence
    elif strong_spike:
        return "Strong Short-Term Spike"

    # General spike
    elif (
        spike_present
        and not persistent
    ):
        return "Short-Term Spike"

    # Positive but incomplete support
    elif (
        moderately_consistent
        or partial_persistence
    ) and positive_stream_change:
        return "Emerging Growth"

    # Clearly unstable growth
    elif (
        high_volatility
        and positive_stream_change
    ):
        return "Unstable Growth"

    # Negative historical direction
    elif (
        declining_stream
        and chart_decline
    ):
        return "Declining"

    # Sparse / weak evidence
    elif insufficient_volatility:
        return "Insufficient Evidence"

    else:
        return "Stable / No Clear Growth"


track_final_growth_pattern_df[
    "final_growth_pattern"
] = (
    track_final_growth_pattern_df.apply(
        classify_final_growth_pattern,
        axis=1,
    )
)

artist_final_growth_pattern_df[
    "final_growth_pattern"
] = (
    artist_final_growth_pattern_df.apply(
        classify_final_growth_pattern,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Track final pattern distribution
# ------------------------------------------------------------

track_final_growth_pattern_distribution_df = (
    track_final_growth_pattern_df[
        "final_growth_pattern"
    ]
    .value_counts()
    .rename_axis("Final Growth Pattern")
    .reset_index(name="Track-Country Histories")
)

track_final_growth_pattern_distribution_df["Share (%)"] = (
    track_final_growth_pattern_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_final_growth_pattern_df)
    * 100
)

print()
print("Track-country final growth pattern distribution")
print("=" * 100)

display(track_final_growth_pattern_distribution_df)


# ------------------------------------------------------------
# 6. Artist final pattern distribution
# ------------------------------------------------------------

artist_final_growth_pattern_distribution_df = (
    artist_final_growth_pattern_df[
        "final_growth_pattern"
    ]
    .value_counts()
    .rename_axis("Final Growth Pattern")
    .reset_index(name="Artist-Country Histories")
)

artist_final_growth_pattern_distribution_df["Share (%)"] = (
    artist_final_growth_pattern_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_final_growth_pattern_df)
    * 100
)

print()
print("Artist-country final growth pattern distribution")
print("=" * 100)

display(artist_final_growth_pattern_distribution_df)


# ------------------------------------------------------------
# 7. Final growth-pattern overview
# ------------------------------------------------------------

final_growth_pattern_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_final_growth_pattern_df
            ),
        },
        {
            "Metric": "Track sustained-growth histories",
            "Observed Value": int(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Track spike histories",
            "Observed Value": int(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_final_growth_pattern_df
            ),
        },
        {
            "Metric": "Artist sustained-growth histories",
            "Observed Value": int(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].isin(
                    [
                        "Sustained Growth",
                        "Strong Sustained Growth",
                    ]
                ).sum()
            ),
        },
        {
            "Metric": "Artist spike histories",
            "Observed Value": int(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].isin(
                    [
                        "Short-Term Spike",
                        "Strong Short-Term Spike",
                    ]
                ).sum()
            ),
        },
    ]
)

print()
print("Final growth pattern overview")
print("=" * 100)

display(final_growth_pattern_overview_df)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

final_growth_pattern_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories classified",
            "Observed Value": int(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track final pattern classes observed",
            "Observed Value": int(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].nunique()
            ),
        },
        {
            "Diagnostic": "Artist histories classified",
            "Observed Value": int(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist final pattern classes observed",
            "Observed Value": int(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].nunique()
            ),
        },
    ]
)

print()
print("Final growth pattern diagnostics")
print("=" * 100)

display(final_growth_pattern_diagnostics_df)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

section_7_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7.4 completion",
            "Requirement": (
                "Consistency Rules must already be complete."
            ),
            "Observed Evidence": (
                f"Section 7.4 completion status: "
                f"{section_7_4_complete}"
            ),
            "Passed": section_7_4_complete,
        },
        {
            "Validation Area": "Track final classification completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a final growth-pattern classification."
            ),
            "Observed Evidence": (
                f"{track_final_growth_pattern_df['final_growth_pattern'].notna().sum():,} "
                "track final patterns assigned"
            ),
            "Passed": bool(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist final classification completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a final growth-pattern classification."
            ),
            "Observed Evidence": (
                f"{artist_final_growth_pattern_df['final_growth_pattern'].notna().sum():,} "
                "artist final patterns assigned"
            ),
            "Passed": bool(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track pattern differentiation",
            "Requirement": (
                "Track histories must produce multiple "
                "meaningful growth-pattern classes."
            ),
            "Observed Evidence": (
                f"{track_final_growth_pattern_df['final_growth_pattern'].nunique()} "
                "track final pattern classes observed"
            ),
            "Passed": bool(
                track_final_growth_pattern_df[
                    "final_growth_pattern"
                ].nunique() >= 3
            ),
        },
        {
            "Validation Area": "Artist pattern differentiation",
            "Requirement": (
                "Artist histories must produce multiple "
                "meaningful growth-pattern classes."
            ),
            "Observed Evidence": (
                f"{artist_final_growth_pattern_df['final_growth_pattern'].nunique()} "
                "artist final pattern classes observed"
            ),
            "Passed": bool(
                artist_final_growth_pattern_df[
                    "final_growth_pattern"
                ].nunique() >= 3
            ),
        },
        {
            "Validation Area": "Multi-signal final classification",
            "Requirement": (
                "Final growth patterns must combine spike, persistence, "
                "volatility, consistency, streaming and chart evidence."
            ),
            "Observed Evidence": (
                "Spike, persistence, volatility, consistency, "
                "streaming and chart signals combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Final growth pattern validation")
print("=" * 100)

display(section_7_5_validation_df)


# ------------------------------------------------------------
# 10. Completion status
# ------------------------------------------------------------

section_7_5_complete = bool(
    section_7_5_validation_df[
        "Passed"
    ].all()
)

section_7_complete = bool(
    section_7_1_complete
    and section_7_2_complete
    and section_7_3_complete
    and section_7_4_complete
    and section_7_5_complete
)

if not section_7_5_complete:
    raise AssertionError(
        "Section 7.5 Final Growth Pattern Classification validation failed."
    )

print()
print(
    f"Section 7.5 completion status: "
    f"{section_7_5_complete}"
)

print(
    f"Section 7 overall completion status: "
    f"{section_7_complete}"
)

print(
    "Final PMIP growth patterns have been classified "
    "using combined validated growth evidence."
)
```

#### Interpretation

Most histories were classified as **Declining**, while sustained growth remained rare.

PMIP identified **2,718 track-country histories** and **1,757 artist-country histories** with sustained growth, compared with **15,939 track spikes** and **18,117 artist spikes**.

This confirms that sustained growth is much more selective than short-term or unstable movement.



## 8. Market Growth Indicators

This section converts the validated growth evidence into interpretable PMIP indicators.

The indicators summarise streaming growth, chart growth, persistence, and stability before combining them into a final market growth composite indicator.


### 8.1 Streaming Growth Indicator

#### Purpose

Create a single streaming-growth indicator from percentage growth, rolling growth, and sustained streaming-growth evidence.

This gives PMIP a consistent score for comparing streaming growth strength across track-country and artist-country histories.

```python
import numpy as np
import pandas as pd


print("Calculating PMIP streaming growth indicator")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 7 completion
# ------------------------------------------------------------

if not section_7_complete:
    raise RuntimeError(
        "Section 7 Spike Versus Sustained Growth Detection "
        "must be complete before market growth indicators begin."
    )

print(
    f"Section 7 overall completion status: "
    f"{section_7_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level indicator data
# ------------------------------------------------------------

track_streaming_indicator_df = (
    track_final_growth_pattern_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level indicator data
# ------------------------------------------------------------

artist_streaming_indicator_df = (
    artist_final_growth_pattern_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Convert percentage growth to bounded score
# ------------------------------------------------------------

def percentage_growth_score(value):

    if pd.isna(value):
        return 0.0

    # Clip extreme values so unusually large changes
    # cannot dominate the indicator.
    clipped_value = np.clip(
        value,
        -100,
        100,
    )

    # Convert -100 to +100 range into 0 to 1.
    return float(
        (clipped_value + 100) / 200
    )


track_streaming_indicator_df[
    "percentage_growth_score"
] = (
    track_streaming_indicator_df[
        "percentage_stream_change"
    ].apply(
        percentage_growth_score
    )
)

artist_streaming_indicator_df[
    "percentage_growth_score"
] = (
    artist_streaming_indicator_df[
        "percentage_stream_change"
    ].apply(
        percentage_growth_score
    )
)


# ------------------------------------------------------------
# 5. Prepare rolling-growth score
# ------------------------------------------------------------

track_streaming_indicator_df[
    "rolling_growth_score"
] = (
    track_streaming_indicator_df[
        "positive_rolling_share"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)

artist_streaming_indicator_df[
    "rolling_growth_score"
] = (
    artist_streaming_indicator_df[
        "positive_rolling_share"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 6. Convert sustained streaming class to score
# ------------------------------------------------------------

sustained_stream_score_map = {
    "No Sustained Growth": 0.00,
    "Emerging Growth": 0.40,
    "Sustained Growth": 0.75,
    "Strong Sustained Growth": 1.00,
}

track_streaming_indicator_df[
    "sustained_stream_score"
] = (
    track_streaming_indicator_df[
        "sustained_growth_class"
    ]
    .map(
        sustained_stream_score_map
    )
    .fillna(0)
)

artist_streaming_indicator_df[
    "sustained_stream_score"
] = (
    artist_streaming_indicator_df[
        "sustained_growth_class"
    ]
    .map(
        sustained_stream_score_map
    )
    .fillna(0)
)


# ------------------------------------------------------------
# 7. Calculate streaming growth indicator
# ------------------------------------------------------------

track_streaming_indicator_df[
    "streaming_growth_indicator"
] = (
    (
        0.35
        * track_streaming_indicator_df[
            "percentage_growth_score"
        ]
    )
    +
    (
        0.35
        * track_streaming_indicator_df[
            "rolling_growth_score"
        ]
    )
    +
    (
        0.30
        * track_streaming_indicator_df[
            "sustained_stream_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


artist_streaming_indicator_df[
    "streaming_growth_indicator"
] = (
    (
        0.35
        * artist_streaming_indicator_df[
            "percentage_growth_score"
        ]
    )
    +
    (
        0.35
        * artist_streaming_indicator_df[
            "rolling_growth_score"
        ]
    )
    +
    (
        0.30
        * artist_streaming_indicator_df[
            "sustained_stream_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 8. Define indicator classes
# ------------------------------------------------------------

def classify_streaming_growth_indicator(score):

    if pd.isna(score):
        return "Insufficient Streaming Evidence"

    if score >= 0.75:
        return "Strong Streaming Growth"

    elif score >= 0.55:
        return "Moderate Streaming Growth"

    elif score >= 0.40:
        return "Developing Streaming Growth"

    else:
        return "Weak / Negative Streaming Growth"


track_streaming_indicator_df[
    "streaming_growth_indicator_class"
] = (
    track_streaming_indicator_df[
        "streaming_growth_indicator"
    ].apply(
        classify_streaming_growth_indicator
    )
)

artist_streaming_indicator_df[
    "streaming_growth_indicator_class"
] = (
    artist_streaming_indicator_df[
        "streaming_growth_indicator"
    ].apply(
        classify_streaming_growth_indicator
    )
)


# ------------------------------------------------------------
# 9. Track indicator distribution
# ------------------------------------------------------------

track_streaming_indicator_distribution_df = (
    track_streaming_indicator_df[
        "streaming_growth_indicator_class"
    ]
    .value_counts()
    .rename_axis("Streaming Growth Indicator Class")
    .reset_index(name="Track-Country Histories")
)

track_streaming_indicator_distribution_df[
    "Share (%)"
] = (
    track_streaming_indicator_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_streaming_indicator_df)
    * 100
)

print()
print("Track-country streaming growth indicator distribution")
print("=" * 100)

display(
    track_streaming_indicator_distribution_df
)


# ------------------------------------------------------------
# 10. Artist indicator distribution
# ------------------------------------------------------------

artist_streaming_indicator_distribution_df = (
    artist_streaming_indicator_df[
        "streaming_growth_indicator_class"
    ]
    .value_counts()
    .rename_axis("Streaming Growth Indicator Class")
    .reset_index(name="Artist-Country Histories")
)

artist_streaming_indicator_distribution_df[
    "Share (%)"
] = (
    artist_streaming_indicator_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_streaming_indicator_df)
    * 100
)

print()
print("Artist-country streaming growth indicator distribution")
print("=" * 100)

display(
    artist_streaming_indicator_distribution_df
)


# ------------------------------------------------------------
# 11. Streaming growth indicator overview
# ------------------------------------------------------------

streaming_growth_indicator_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_streaming_indicator_df
            ),
        },
        {
            "Metric": "Median track streaming growth indicator",
            "Observed Value": float(
                track_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Track strong streaming-growth histories",
            "Observed Value": int(
                (
                    track_streaming_indicator_df[
                        "streaming_growth_indicator_class"
                    ]
                    == "Strong Streaming Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_streaming_indicator_df
            ),
        },
        {
            "Metric": "Median artist streaming growth indicator",
            "Observed Value": float(
                artist_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Artist strong streaming-growth histories",
            "Observed Value": int(
                (
                    artist_streaming_indicator_df[
                        "streaming_growth_indicator_class"
                    ]
                    == "Strong Streaming Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Streaming growth indicator overview")
print("=" * 100)

display(
    streaming_growth_indicator_overview_df
)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

streaming_growth_indicator_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_streaming_indicator_df
            ),
        },
        {
            "Diagnostic": "Track indicator values available",
            "Observed Value": int(
                track_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track indicator classes assigned",
            "Observed Value": int(
                track_streaming_indicator_df[
                    "streaming_growth_indicator_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_streaming_indicator_df
            ),
        },
        {
            "Diagnostic": "Artist indicator values available",
            "Observed Value": int(
                artist_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist indicator classes assigned",
            "Observed Value": int(
                artist_streaming_indicator_df[
                    "streaming_growth_indicator_class"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Streaming growth indicator diagnostics")
print("=" * 100)

display(
    streaming_growth_indicator_diagnostics_df
)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_8_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 7 completion",
            "Requirement": (
                "Spike Versus Sustained Growth Detection "
                "must already be complete."
            ),
            "Observed Evidence": (
                f"Section 7 overall completion status: "
                f"{section_7_complete}"
            ),
            "Passed": section_7_complete,
        },
        {
            "Validation Area": "Track indicator completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a streaming growth indicator."
            ),
            "Observed Evidence": (
                f"{track_streaming_indicator_df['streaming_growth_indicator'].notna().sum():,} "
                "track indicators calculated"
            ),
            "Passed": bool(
                track_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist indicator completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a streaming growth indicator."
            ),
            "Observed Evidence": (
                f"{artist_streaming_indicator_df['streaming_growth_indicator'].notna().sum():,} "
                "artist indicators calculated"
            ),
            "Passed": bool(
                artist_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track indicator bounds",
            "Requirement": (
                "Track streaming growth indicators must "
                "remain between 0 and 1."
            ),
            "Observed Evidence": (
                "Track indicators remain within valid bounds"
            ),
            "Passed": bool(
                track_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist indicator bounds",
            "Requirement": (
                "Artist streaming growth indicators must "
                "remain between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist indicators remain within valid bounds"
            ),
            "Passed": bool(
                artist_streaming_indicator_df[
                    "streaming_growth_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Multi-signal streaming indicator",
            "Requirement": (
                "Streaming growth must combine percentage, "
                "rolling and sustained-growth evidence."
            ),
            "Observed Evidence": (
                "Percentage change, rolling growth and "
                "sustained streaming-growth evidence combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Streaming growth indicator validation")
print("=" * 100)

display(
    section_8_1_validation_df
)


# ------------------------------------------------------------
# 14. Completion status
# ------------------------------------------------------------

section_8_1_complete = bool(
    section_8_1_validation_df[
        "Passed"
    ].all()
)

if not section_8_1_complete:
    raise AssertionError(
        "Section 8.1 Streaming Growth Indicator validation failed."
    )

print()
print(
    f"Section 8.1 completion status: "
    f"{section_8_1_complete}"
)

print(
    "PMIP streaming growth indicators have been calculated "
    "for track-country and artist-country histories."
)
```

#### Interpretation

Most track-country and artist-country histories showed weak or negative streaming growth.

PMIP identified **6,555 track histories** and **4,542 artist histories** with strong streaming growth.


### 8.2 Chart Growth Indicator

#### Purpose

Create a single chart-growth indicator using chart position improvement, chart persistence, chart activity, and sustained chart-growth evidence.

This gives PMIP a consistent score for comparing chart-growth strength across track-country and artist-country histories.

```python
print("Calculating PMIP chart growth indicator")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 8.1 completion
# ------------------------------------------------------------

if not section_8_1_complete:
    raise RuntimeError(
        "Section 8.1 Streaming Growth Indicator must be complete "
        "before the Chart Growth Indicator is calculated."
    )

print(
    f"Section 8.1 completion status: "
    f"{section_8_1_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level chart indicator data
# ------------------------------------------------------------

track_chart_indicator_df = (
    track_streaming_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level chart indicator data
# ------------------------------------------------------------

artist_chart_indicator_df = (
    artist_streaming_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Convert chart-direction evidence to score
# ------------------------------------------------------------

chart_direction_score_map = {
    "Improved": 1.00,
    "Stable": 0.50,
    "Declined": 0.00,
}

track_chart_indicator_df[
    "chart_direction_score"
] = (
    track_chart_indicator_df[
        "chart_direction"
    ]
    .map(chart_direction_score_map)
    .fillna(0)
)

artist_chart_indicator_df[
    "chart_direction_score"
] = (
    artist_chart_indicator_df[
        "chart_direction"
    ]
    .map(chart_direction_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 5. Convert chart persistence class to score
# ------------------------------------------------------------

chart_persistence_score_map = {
    "No Top-50 Persistence": 0.00,
    "Limited Chart Persistence": 0.25,
    "Moderate Top-50 Persistence": 0.55,
    "Strong Top-50 Persistence": 0.80,
    "Strong Top-10 Persistence": 1.00,
}

track_chart_indicator_df[
    "chart_persistence_score"
] = (
    track_chart_indicator_df[
        "chart_persistence_class"
    ]
    .map(chart_persistence_score_map)
    .fillna(0)
)

artist_chart_indicator_df[
    "chart_persistence_score"
] = (
    artist_chart_indicator_df[
        "chart_persistence_class"
    ]
    .map(chart_persistence_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 6. Convert chart activity class to score
# ------------------------------------------------------------

chart_activity_score_map = {
    "Declining Chart Activity": 0.00,
    "Stable Chart Activity": 0.50,
    "Increasing Chart Activity": 1.00,
}

track_chart_indicator_df[
    "chart_activity_score"
] = (
    track_chart_indicator_df[
        "chart_activity_growth_class"
    ]
    .map(chart_activity_score_map)
    .fillna(0)
)

artist_chart_indicator_df[
    "chart_activity_score"
] = (
    artist_chart_indicator_df[
        "chart_activity_growth_class"
    ]
    .map(chart_activity_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 7. Convert sustained chart-growth class to score
# ------------------------------------------------------------

sustained_chart_score_map = {
    "No Sustained Chart Growth": 0.00,
    "Limited Chart Improvement": 0.30,
    "Emerging Chart Growth": 0.50,
    "Sustained Chart Growth": 0.80,
    "Strong Sustained Chart Growth": 1.00,
}

track_chart_indicator_df[
    "sustained_chart_score"
] = (
    track_chart_indicator_df[
        "sustained_chart_growth_class"
    ]
    .map(sustained_chart_score_map)
    .fillna(0)
)

artist_chart_indicator_df[
    "sustained_chart_score"
] = (
    artist_chart_indicator_df[
        "sustained_chart_growth_class"
    ]
    .map(sustained_chart_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 8. Calculate chart growth indicator
# ------------------------------------------------------------

track_chart_indicator_df[
    "chart_growth_indicator"
] = (
    (
        0.30
        * track_chart_indicator_df[
            "chart_direction_score"
        ]
    )
    +
    (
        0.25
        * track_chart_indicator_df[
            "chart_persistence_score"
        ]
    )
    +
    (
        0.20
        * track_chart_indicator_df[
            "chart_activity_score"
        ]
    )
    +
    (
        0.25
        * track_chart_indicator_df[
            "sustained_chart_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


artist_chart_indicator_df[
    "chart_growth_indicator"
] = (
    (
        0.30
        * artist_chart_indicator_df[
            "chart_direction_score"
        ]
    )
    +
    (
        0.25
        * artist_chart_indicator_df[
            "chart_persistence_score"
        ]
    )
    +
    (
        0.20
        * artist_chart_indicator_df[
            "chart_activity_score"
        ]
    )
    +
    (
        0.25
        * artist_chart_indicator_df[
            "sustained_chart_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Define chart indicator classes
# ------------------------------------------------------------

def classify_chart_growth_indicator(score):

    if pd.isna(score):
        return "Insufficient Chart Evidence"

    if score >= 0.75:
        return "Strong Chart Growth"

    elif score >= 0.55:
        return "Moderate Chart Growth"

    elif score >= 0.35:
        return "Developing Chart Growth"

    else:
        return "Weak / Negative Chart Growth"


track_chart_indicator_df[
    "chart_growth_indicator_class"
] = (
    track_chart_indicator_df[
        "chart_growth_indicator"
    ].apply(
        classify_chart_growth_indicator
    )
)

artist_chart_indicator_df[
    "chart_growth_indicator_class"
] = (
    artist_chart_indicator_df[
        "chart_growth_indicator"
    ].apply(
        classify_chart_growth_indicator
    )
)


# ------------------------------------------------------------
# 10. Track chart indicator distribution
# ------------------------------------------------------------

track_chart_indicator_distribution_df = (
    track_chart_indicator_df[
        "chart_growth_indicator_class"
    ]
    .value_counts()
    .rename_axis("Chart Growth Indicator Class")
    .reset_index(name="Track-Country Histories")
)

track_chart_indicator_distribution_df[
    "Share (%)"
] = (
    track_chart_indicator_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_chart_indicator_df)
    * 100
)

print()
print("Track-country chart growth indicator distribution")
print("=" * 100)

display(
    track_chart_indicator_distribution_df
)


# ------------------------------------------------------------
# 11. Artist chart indicator distribution
# ------------------------------------------------------------

artist_chart_indicator_distribution_df = (
    artist_chart_indicator_df[
        "chart_growth_indicator_class"
    ]
    .value_counts()
    .rename_axis("Chart Growth Indicator Class")
    .reset_index(name="Artist-Country Histories")
)

artist_chart_indicator_distribution_df[
    "Share (%)"
] = (
    artist_chart_indicator_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_chart_indicator_df)
    * 100
)

print()
print("Artist-country chart growth indicator distribution")
print("=" * 100)

display(
    artist_chart_indicator_distribution_df
)


# ------------------------------------------------------------
# 12. Chart growth indicator overview
# ------------------------------------------------------------

chart_growth_indicator_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_chart_indicator_df
            ),
        },
        {
            "Metric": "Median track chart growth indicator",
            "Observed Value": float(
                track_chart_indicator_df[
                    "chart_growth_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Track strong chart-growth histories",
            "Observed Value": int(
                (
                    track_chart_indicator_df[
                        "chart_growth_indicator_class"
                    ]
                    == "Strong Chart Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_indicator_df
            ),
        },
        {
            "Metric": "Median artist chart growth indicator",
            "Observed Value": float(
                artist_chart_indicator_df[
                    "chart_growth_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Artist strong chart-growth histories",
            "Observed Value": int(
                (
                    artist_chart_indicator_df[
                        "chart_growth_indicator_class"
                    ]
                    == "Strong Chart Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Chart growth indicator overview")
print("=" * 100)

display(
    chart_growth_indicator_overview_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

chart_growth_indicator_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_chart_indicator_df
            ),
        },
        {
            "Diagnostic": "Track chart indicators available",
            "Observed Value": int(
                track_chart_indicator_df[
                    "chart_growth_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track chart classes assigned",
            "Observed Value": int(
                track_chart_indicator_df[
                    "chart_growth_indicator_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_chart_indicator_df
            ),
        },
        {
            "Diagnostic": "Artist chart indicators available",
            "Observed Value": int(
                artist_chart_indicator_df[
                    "chart_growth_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist chart classes assigned",
            "Observed Value": int(
                artist_chart_indicator_df[
                    "chart_growth_indicator_class"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Chart growth indicator diagnostics")
print("=" * 100)

display(
    chart_growth_indicator_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

section_8_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.1 completion",
            "Requirement": (
                "Streaming Growth Indicator must already be complete."
            ),
            "Observed Evidence": (
                f"Section 8.1 completion status: "
                f"{section_8_1_complete}"
            ),
            "Passed": section_8_1_complete,
        },
        {
            "Validation Area": "Track chart indicator completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a chart growth indicator."
            ),
            "Observed Evidence": (
                f"{track_chart_indicator_df['chart_growth_indicator'].notna().sum():,} "
                "track chart indicators calculated"
            ),
            "Passed": bool(
                track_chart_indicator_df[
                    "chart_growth_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist chart indicator completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a chart growth indicator."
            ),
            "Observed Evidence": (
                f"{artist_chart_indicator_df['chart_growth_indicator'].notna().sum():,} "
                "artist chart indicators calculated"
            ),
            "Passed": bool(
                artist_chart_indicator_df[
                    "chart_growth_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track chart indicator bounds",
            "Requirement": (
                "Track chart growth indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Track chart indicators remain within valid bounds"
            ),
            "Passed": bool(
                track_chart_indicator_df[
                    "chart_growth_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist chart indicator bounds",
            "Requirement": (
                "Artist chart growth indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist chart indicators remain within valid bounds"
            ),
            "Passed": bool(
                artist_chart_indicator_df[
                    "chart_growth_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Multi-signal chart indicator",
            "Requirement": (
                "Chart growth must combine position, persistence, "
                "activity and sustained-growth evidence."
            ),
            "Observed Evidence": (
                "Chart direction, persistence, activity and "
                "sustained chart-growth evidence combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Chart growth indicator validation")
print("=" * 100)

display(
    section_8_2_validation_df
)


# ------------------------------------------------------------
# 15. Completion status
# ------------------------------------------------------------

section_8_2_complete = bool(
    section_8_2_validation_df[
        "Passed"
    ].all()
)

if not section_8_2_complete:
    raise AssertionError(
        "Section 8.2 Chart Growth Indicator validation failed."
    )

print()
print(
    f"Section 8.2 completion status: "
    f"{section_8_2_complete}"
)

print(
    "PMIP chart growth indicators have been calculated "
    "for track-country and artist-country histories."
)
```

### 8.3 Persistence Indicator

#### Purpose

Create a persistence indicator that measures how consistently growth is maintained over time.

This helps PMIP give stronger scores to repeated growth and lower scores to weak or temporary movement.

```python
print("Calculating PMIP persistence indicator")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 8.2 completion
# ------------------------------------------------------------

if not section_8_2_complete:
    raise RuntimeError(
        "Section 8.2 Chart Growth Indicator must be complete "
        "before the Persistence Indicator is calculated."
    )

print(
    f"Section 8.2 completion status: "
    f"{section_8_2_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level persistence indicator data
# ------------------------------------------------------------

track_persistence_indicator_df = (
    track_chart_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level persistence indicator data
# ------------------------------------------------------------

artist_persistence_indicator_df = (
    artist_chart_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Convert growth persistence class to score
# ------------------------------------------------------------

persistence_score_map = {
    "Low Persistence": 0.00,
    "Partial Persistence": 0.40,
    "Persistent Growth": 0.75,
    "Strong Persistent Growth": 1.00,
}

track_persistence_indicator_df[
    "persistence_class_score"
] = (
    track_persistence_indicator_df[
        "growth_persistence_class"
    ]
    .map(persistence_score_map)
    .fillna(0)
)

artist_persistence_indicator_df[
    "persistence_class_score"
] = (
    artist_persistence_indicator_df[
        "growth_persistence_class"
    ]
    .map(persistence_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 5. Convert rolling positive share to persistence evidence
# ------------------------------------------------------------

track_persistence_indicator_df[
    "rolling_persistence_score"
] = (
    track_persistence_indicator_df[
        "positive_rolling_share"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)

artist_persistence_indicator_df[
    "rolling_persistence_score"
] = (
    artist_persistence_indicator_df[
        "positive_rolling_share"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 6. Convert sustained streaming growth to persistence evidence
# ------------------------------------------------------------

sustained_stream_persistence_map = {
    "No Sustained Growth": 0.00,
    "Emerging Growth": 0.35,
    "Sustained Growth": 0.75,
    "Strong Sustained Growth": 1.00,
}

track_persistence_indicator_df[
    "stream_persistence_score"
] = (
    track_persistence_indicator_df[
        "sustained_growth_class"
    ]
    .map(sustained_stream_persistence_map)
    .fillna(0)
)

artist_persistence_indicator_df[
    "stream_persistence_score"
] = (
    artist_persistence_indicator_df[
        "sustained_growth_class"
    ]
    .map(sustained_stream_persistence_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 7. Convert sustained chart growth to persistence evidence
# ------------------------------------------------------------

sustained_chart_persistence_map = {
    "No Sustained Chart Growth": 0.00,
    "Limited Chart Improvement": 0.25,
    "Emerging Chart Growth": 0.45,
    "Sustained Chart Growth": 0.80,
    "Strong Sustained Chart Growth": 1.00,
}

track_persistence_indicator_df[
    "chart_persistence_growth_score"
] = (
    track_persistence_indicator_df[
        "sustained_chart_growth_class"
    ]
    .map(sustained_chart_persistence_map)
    .fillna(0)
)

artist_persistence_indicator_df[
    "chart_persistence_growth_score"
] = (
    artist_persistence_indicator_df[
        "sustained_chart_growth_class"
    ]
    .map(sustained_chart_persistence_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 8. Calculate persistence indicator
# ------------------------------------------------------------

track_persistence_indicator_df[
    "persistence_indicator"
] = (
    (
        0.40
        * track_persistence_indicator_df[
            "persistence_class_score"
        ]
    )
    +
    (
        0.25
        * track_persistence_indicator_df[
            "rolling_persistence_score"
        ]
    )
    +
    (
        0.20
        * track_persistence_indicator_df[
            "stream_persistence_score"
        ]
    )
    +
    (
        0.15
        * track_persistence_indicator_df[
            "chart_persistence_growth_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


artist_persistence_indicator_df[
    "persistence_indicator"
] = (
    (
        0.40
        * artist_persistence_indicator_df[
            "persistence_class_score"
        ]
    )
    +
    (
        0.25
        * artist_persistence_indicator_df[
            "rolling_persistence_score"
        ]
    )
    +
    (
        0.20
        * artist_persistence_indicator_df[
            "stream_persistence_score"
        ]
    )
    +
    (
        0.15
        * artist_persistence_indicator_df[
            "chart_persistence_growth_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Define persistence indicator classes
# ------------------------------------------------------------

def classify_persistence_indicator(score):

    if pd.isna(score):
        return "Insufficient Persistence Evidence"

    if score >= 0.75:
        return "Strong Persistence"

    elif score >= 0.55:
        return "Moderate Persistence"

    elif score >= 0.35:
        return "Developing Persistence"

    else:
        return "Weak Persistence"


track_persistence_indicator_df[
    "persistence_indicator_class"
] = (
    track_persistence_indicator_df[
        "persistence_indicator"
    ].apply(
        classify_persistence_indicator
    )
)

artist_persistence_indicator_df[
    "persistence_indicator_class"
] = (
    artist_persistence_indicator_df[
        "persistence_indicator"
    ].apply(
        classify_persistence_indicator
    )
)


# ------------------------------------------------------------
# 10. Track persistence indicator distribution
# ------------------------------------------------------------

track_persistence_indicator_distribution_df = (
    track_persistence_indicator_df[
        "persistence_indicator_class"
    ]
    .value_counts()
    .rename_axis("Persistence Indicator Class")
    .reset_index(name="Track-Country Histories")
)

track_persistence_indicator_distribution_df[
    "Share (%)"
] = (
    track_persistence_indicator_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_persistence_indicator_df)
    * 100
)

print()
print("Track-country persistence indicator distribution")
print("=" * 100)

display(
    track_persistence_indicator_distribution_df
)


# ------------------------------------------------------------
# 11. Artist persistence indicator distribution
# ------------------------------------------------------------

artist_persistence_indicator_distribution_df = (
    artist_persistence_indicator_df[
        "persistence_indicator_class"
    ]
    .value_counts()
    .rename_axis("Persistence Indicator Class")
    .reset_index(name="Artist-Country Histories")
)

artist_persistence_indicator_distribution_df[
    "Share (%)"
] = (
    artist_persistence_indicator_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_persistence_indicator_df)
    * 100
)

print()
print("Artist-country persistence indicator distribution")
print("=" * 100)

display(
    artist_persistence_indicator_distribution_df
)


# ------------------------------------------------------------
# 12. Persistence indicator overview
# ------------------------------------------------------------

persistence_indicator_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_persistence_indicator_df
            ),
        },
        {
            "Metric": "Median track persistence indicator",
            "Observed Value": float(
                track_persistence_indicator_df[
                    "persistence_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Track strong-persistence histories",
            "Observed Value": int(
                (
                    track_persistence_indicator_df[
                        "persistence_indicator_class"
                    ]
                    == "Strong Persistence"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_persistence_indicator_df
            ),
        },
        {
            "Metric": "Median artist persistence indicator",
            "Observed Value": float(
                artist_persistence_indicator_df[
                    "persistence_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Artist strong-persistence histories",
            "Observed Value": int(
                (
                    artist_persistence_indicator_df[
                        "persistence_indicator_class"
                    ]
                    == "Strong Persistence"
                ).sum()
            ),
        },
    ]
)

print()
print("Persistence indicator overview")
print("=" * 100)

display(
    persistence_indicator_overview_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

persistence_indicator_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_persistence_indicator_df
            ),
        },
        {
            "Diagnostic": "Track persistence indicators available",
            "Observed Value": int(
                track_persistence_indicator_df[
                    "persistence_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track persistence classes assigned",
            "Observed Value": int(
                track_persistence_indicator_df[
                    "persistence_indicator_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_persistence_indicator_df
            ),
        },
        {
            "Diagnostic": "Artist persistence indicators available",
            "Observed Value": int(
                artist_persistence_indicator_df[
                    "persistence_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist persistence classes assigned",
            "Observed Value": int(
                artist_persistence_indicator_df[
                    "persistence_indicator_class"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Persistence indicator diagnostics")
print("=" * 100)

display(
    persistence_indicator_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

section_8_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.2 completion",
            "Requirement": (
                "Chart Growth Indicator must already be complete."
            ),
            "Observed Evidence": (
                f"Section 8.2 completion status: "
                f"{section_8_2_complete}"
            ),
            "Passed": section_8_2_complete,
        },
        {
            "Validation Area": "Track persistence indicator completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a persistence indicator."
            ),
            "Observed Evidence": (
                f"{track_persistence_indicator_df['persistence_indicator'].notna().sum():,} "
                "track persistence indicators calculated"
            ),
            "Passed": bool(
                track_persistence_indicator_df[
                    "persistence_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist persistence indicator completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a persistence indicator."
            ),
            "Observed Evidence": (
                f"{artist_persistence_indicator_df['persistence_indicator'].notna().sum():,} "
                "artist persistence indicators calculated"
            ),
            "Passed": bool(
                artist_persistence_indicator_df[
                    "persistence_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track persistence indicator bounds",
            "Requirement": (
                "Track persistence indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Track persistence indicators remain within valid bounds"
            ),
            "Passed": bool(
                track_persistence_indicator_df[
                    "persistence_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist persistence indicator bounds",
            "Requirement": (
                "Artist persistence indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist persistence indicators remain within valid bounds"
            ),
            "Passed": bool(
                artist_persistence_indicator_df[
                    "persistence_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Multi-signal persistence indicator",
            "Requirement": (
                "Persistence must combine persistence class, "
                "rolling, streaming and chart evidence."
            ),
            "Observed Evidence": (
                "Persistence class, rolling growth, sustained streaming "
                "and sustained chart evidence combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Persistence indicator validation")
print("=" * 100)

display(
    section_8_3_validation_df
)


# ------------------------------------------------------------
# 15. Completion status
# ------------------------------------------------------------

section_8_3_complete = bool(
    section_8_3_validation_df[
        "Passed"
    ].all()
)

if not section_8_3_complete:
    raise AssertionError(
        "Section 8.3 Persistence Indicator validation failed."
    )

print()
print(
    f"Section 8.3 completion status: "
    f"{section_8_3_complete}"
)

print(
    "PMIP persistence indicators have been calculated "
    "for track-country and artist-country histories."
)
```

#### Interpretation

### 8.4 Stability Indicator

#### Purpose

Create a stability indicator that measures how reliable and non-volatile the observed growth pattern is.

This helps PMIP reward smoother, more dependable growth and reduce the influence of unstable or highly volatile movement.

```python
print("Calculating PMIP stability indicator")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 8.3 completion
# ------------------------------------------------------------

if not section_8_3_complete:
    raise RuntimeError(
        "Section 8.3 Persistence Indicator must be complete "
        "before the Stability Indicator is calculated."
    )

print(
    f"Section 8.3 completion status: "
    f"{section_8_3_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level stability indicator data
# ------------------------------------------------------------

track_stability_indicator_df = (
    track_persistence_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level stability indicator data
# ------------------------------------------------------------

artist_stability_indicator_df = (
    artist_persistence_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Convert volatility class to stability score
# ------------------------------------------------------------

volatility_stability_score_map = {
    "Low Volatility": 1.00,
    "Moderate Volatility": 0.60,
    "High Volatility": 0.15,
    "Insufficient Volatility Evidence": 0.00,
}

track_stability_indicator_df[
    "volatility_stability_score"
] = (
    track_stability_indicator_df[
        "growth_volatility_class"
    ]
    .map(volatility_stability_score_map)
    .fillna(0)
)

artist_stability_indicator_df[
    "volatility_stability_score"
] = (
    artist_stability_indicator_df[
        "growth_volatility_class"
    ]
    .map(volatility_stability_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 5. Convert growth consistency class to stability score
# ------------------------------------------------------------

consistency_stability_score_map = {
    "Low Consistency": 0.10,
    "Unstable Growth": 0.00,
    "Moderately Consistent Growth": 0.55,
    "Consistent Growth": 0.80,
    "Highly Consistent Growth": 1.00,
}

track_stability_indicator_df[
    "consistency_stability_score"
] = (
    track_stability_indicator_df[
        "growth_consistency_class"
    ]
    .map(consistency_stability_score_map)
    .fillna(0)
)

artist_stability_indicator_df[
    "consistency_stability_score"
] = (
    artist_stability_indicator_df[
        "growth_consistency_class"
    ]
    .map(consistency_stability_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 6. Convert spike class to stability score
# ------------------------------------------------------------

spike_stability_score_map = {
    "No Spike Signal": 1.00,
    "Possible Spike": 0.60,
    "Short-Term Spike": 0.30,
    "Strong Short-Term Spike": 0.00,
}

track_stability_indicator_df[
    "spike_stability_score"
] = (
    track_stability_indicator_df[
        "spike_class"
    ]
    .map(spike_stability_score_map)
    .fillna(0)
)

artist_stability_indicator_df[
    "spike_stability_score"
] = (
    artist_stability_indicator_df[
        "spike_class"
    ]
    .map(spike_stability_score_map)
    .fillna(0)
)


# ------------------------------------------------------------
# 7. Convert persistence indicator to stability support
# ------------------------------------------------------------

track_stability_indicator_df[
    "persistence_stability_score"
] = (
    track_stability_indicator_df[
        "persistence_indicator"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)

artist_stability_indicator_df[
    "persistence_stability_score"
] = (
    artist_stability_indicator_df[
        "persistence_indicator"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 8. Calculate stability indicator
# ------------------------------------------------------------

track_stability_indicator_df[
    "stability_indicator"
] = (
    (
        0.40
        * track_stability_indicator_df[
            "volatility_stability_score"
        ]
    )
    +
    (
        0.30
        * track_stability_indicator_df[
            "consistency_stability_score"
        ]
    )
    +
    (
        0.20
        * track_stability_indicator_df[
            "spike_stability_score"
        ]
    )
    +
    (
        0.10
        * track_stability_indicator_df[
            "persistence_stability_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


artist_stability_indicator_df[
    "stability_indicator"
] = (
    (
        0.40
        * artist_stability_indicator_df[
            "volatility_stability_score"
        ]
    )
    +
    (
        0.30
        * artist_stability_indicator_df[
            "consistency_stability_score"
        ]
    )
    +
    (
        0.20
        * artist_stability_indicator_df[
            "spike_stability_score"
        ]
    )
    +
    (
        0.10
        * artist_stability_indicator_df[
            "persistence_stability_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Define stability indicator classes
# ------------------------------------------------------------

def classify_stability_indicator(score):

    if pd.isna(score):
        return "Insufficient Stability Evidence"

    if score >= 0.75:
        return "High Stability"

    elif score >= 0.55:
        return "Moderate Stability"

    elif score >= 0.35:
        return "Low Stability"

    else:
        return "High Instability"


track_stability_indicator_df[
    "stability_indicator_class"
] = (
    track_stability_indicator_df[
        "stability_indicator"
    ].apply(
        classify_stability_indicator
    )
)

artist_stability_indicator_df[
    "stability_indicator_class"
] = (
    artist_stability_indicator_df[
        "stability_indicator"
    ].apply(
        classify_stability_indicator
    )
)


# ------------------------------------------------------------
# 10. Track stability distribution
# ------------------------------------------------------------

track_stability_indicator_distribution_df = (
    track_stability_indicator_df[
        "stability_indicator_class"
    ]
    .value_counts()
    .rename_axis("Stability Indicator Class")
    .reset_index(name="Track-Country Histories")
)

track_stability_indicator_distribution_df[
    "Share (%)"
] = (
    track_stability_indicator_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_stability_indicator_df)
    * 100
)

print()
print("Track-country stability indicator distribution")
print("=" * 100)

display(
    track_stability_indicator_distribution_df
)


# ------------------------------------------------------------
# 11. Artist stability distribution
# ------------------------------------------------------------

artist_stability_indicator_distribution_df = (
    artist_stability_indicator_df[
        "stability_indicator_class"
    ]
    .value_counts()
    .rename_axis("Stability Indicator Class")
    .reset_index(name="Artist-Country Histories")
)

artist_stability_indicator_distribution_df[
    "Share (%)"
] = (
    artist_stability_indicator_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_stability_indicator_df)
    * 100
)

print()
print("Artist-country stability indicator distribution")
print("=" * 100)

display(
    artist_stability_indicator_distribution_df
)


# ------------------------------------------------------------
# 12. Stability indicator overview
# ------------------------------------------------------------

stability_indicator_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_stability_indicator_df
            ),
        },
        {
            "Metric": "Median track stability indicator",
            "Observed Value": float(
                track_stability_indicator_df[
                    "stability_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Track high-stability histories",
            "Observed Value": int(
                (
                    track_stability_indicator_df[
                        "stability_indicator_class"
                    ]
                    == "High Stability"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_stability_indicator_df
            ),
        },
        {
            "Metric": "Median artist stability indicator",
            "Observed Value": float(
                artist_stability_indicator_df[
                    "stability_indicator"
                ].median()
            ),
        },
        {
            "Metric": "Artist high-stability histories",
            "Observed Value": int(
                (
                    artist_stability_indicator_df[
                        "stability_indicator_class"
                    ]
                    == "High Stability"
                ).sum()
            ),
        },
    ]
)

print()
print("Stability indicator overview")
print("=" * 100)

display(
    stability_indicator_overview_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

stability_indicator_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_stability_indicator_df
            ),
        },
        {
            "Diagnostic": "Track stability indicators available",
            "Observed Value": int(
                track_stability_indicator_df[
                    "stability_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track stability classes assigned",
            "Observed Value": int(
                track_stability_indicator_df[
                    "stability_indicator_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_stability_indicator_df
            ),
        },
        {
            "Diagnostic": "Artist stability indicators available",
            "Observed Value": int(
                artist_stability_indicator_df[
                    "stability_indicator"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist stability classes assigned",
            "Observed Value": int(
                artist_stability_indicator_df[
                    "stability_indicator_class"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Stability indicator diagnostics")
print("=" * 100)

display(
    stability_indicator_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

section_8_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.3 completion",
            "Requirement": (
                "Persistence Indicator must already be complete."
            ),
            "Observed Evidence": (
                f"Section 8.3 completion status: "
                f"{section_8_3_complete}"
            ),
            "Passed": section_8_3_complete,
        },
        {
            "Validation Area": "Track stability indicator completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a stability indicator."
            ),
            "Observed Evidence": (
                f"{track_stability_indicator_df['stability_indicator'].notna().sum():,} "
                "track stability indicators calculated"
            ),
            "Passed": bool(
                track_stability_indicator_df[
                    "stability_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist stability indicator completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a stability indicator."
            ),
            "Observed Evidence": (
                f"{artist_stability_indicator_df['stability_indicator'].notna().sum():,} "
                "artist stability indicators calculated"
            ),
            "Passed": bool(
                artist_stability_indicator_df[
                    "stability_indicator"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track stability indicator bounds",
            "Requirement": (
                "Track stability indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Track stability indicators remain within valid bounds"
            ),
            "Passed": bool(
                track_stability_indicator_df[
                    "stability_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist stability indicator bounds",
            "Requirement": (
                "Artist stability indicators must remain "
                "between 0 and 1."
            ),
            "Observed Evidence": (
                "Artist stability indicators remain within valid bounds"
            ),
            "Passed": bool(
                artist_stability_indicator_df[
                    "stability_indicator"
                ].between(
                    0,
                    1,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Multi-signal stability indicator",
            "Requirement": (
                "Stability must combine volatility, consistency, "
                "spike and persistence evidence."
            ),
            "Observed Evidence": (
                "Volatility, consistency, spike and persistence "
                "evidence combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Stability indicator validation")
print("=" * 100)

display(
    section_8_4_validation_df
)


# ------------------------------------------------------------
# 15. Completion status
# ------------------------------------------------------------

section_8_4_complete = bool(
    section_8_4_validation_df[
        "Passed"
    ].all()
)

if not section_8_4_complete:
    raise AssertionError(
        "Section 8.4 Stability Indicator validation failed."
    )

print()
print(
    f"Section 8.4 completion status: "
    f"{section_8_4_complete}"
)

print(
    "PMIP stability indicators have been calculated "
    "for track-country and artist-country histories."
)
```

#### Interpretation

Most histories showed moderate or low stability, while high stability was uncommon.

PMIP identified **1,320 track-country histories** and **776 artist-country histories** with high stability.


### 8.5 Market Growth Composite Indicator

#### Purpose

Combine the streaming growth, chart growth, persistence, and stability indicators into one interpretable market growth score.

This gives PMIP a single measure for ranking the strength of growth across track-country and artist-country histories.

```python
print("Calculating PMIP market growth composite indicator")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 8.4 completion
# ------------------------------------------------------------

if not section_8_4_complete:
    raise RuntimeError(
        "Section 8.4 Stability Indicator must be complete "
        "before the Market Growth Composite Indicator is calculated."
    )

print(
    f"Section 8.4 completion status: "
    f"{section_8_4_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-level composite indicator data
# ------------------------------------------------------------

track_market_growth_df = (
    track_stability_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-level composite indicator data
# ------------------------------------------------------------

artist_market_growth_df = (
    artist_stability_indicator_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Calculate market growth composite indicator
# ------------------------------------------------------------

track_market_growth_df[
    "market_growth_composite_indicator"
] = (
    (
        0.35
        * track_market_growth_df[
            "streaming_growth_indicator"
        ]
    )
    +
    (
        0.25
        * track_market_growth_df[
            "chart_growth_indicator"
        ]
    )
    +
    (
        0.25
        * track_market_growth_df[
            "persistence_indicator"
        ]
    )
    +
    (
        0.15
        * track_market_growth_df[
            "stability_indicator"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


artist_market_growth_df[
    "market_growth_composite_indicator"
] = (
    (
        0.35
        * artist_market_growth_df[
            "streaming_growth_indicator"
        ]
    )
    +
    (
        0.25
        * artist_market_growth_df[
            "chart_growth_indicator"
        ]
    )
    +
    (
        0.25
        * artist_market_growth_df[
            "persistence_indicator"
        ]
    )
    +
    (
        0.15
        * artist_market_growth_df[
            "stability_indicator"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 5. Convert composite score to 0–100 scale
# ------------------------------------------------------------

track_market_growth_df[
    "market_growth_score"
] = (
    track_market_growth_df[
        "market_growth_composite_indicator"
    ]
    * 100
)

artist_market_growth_df[
    "market_growth_score"
] = (
    artist_market_growth_df[
        "market_growth_composite_indicator"
    ]
    * 100
)


# ------------------------------------------------------------
# 6. Define market growth classes
# ------------------------------------------------------------

def classify_market_growth(score):

    if pd.isna(score):
        return "Insufficient Growth Evidence"

    if score >= 75:
        return "Strong Market Growth"

    elif score >= 55:
        return "Moderate Market Growth"

    elif score >= 40:
        return "Emerging Market Growth"

    elif score >= 25:
        return "Limited Market Growth"

    else:
        return "Weak / Negative Market Growth"


track_market_growth_df[
    "market_growth_class"
] = (
    track_market_growth_df[
        "market_growth_score"
    ].apply(
        classify_market_growth
    )
)

artist_market_growth_df[
    "market_growth_class"
] = (
    artist_market_growth_df[
        "market_growth_score"
    ].apply(
        classify_market_growth
    )
)


# ------------------------------------------------------------
# 7. Rank track-country market growth
# ------------------------------------------------------------

track_market_growth_df[
    "market_growth_rank"
] = (
    track_market_growth_df[
        "market_growth_score"
    ]
    .rank(
        method="dense",
        ascending=False,
    )
    .astype(int)
)


# ------------------------------------------------------------
# 8. Rank artist-country market growth
# ------------------------------------------------------------

artist_market_growth_df[
    "market_growth_rank"
] = (
    artist_market_growth_df[
        "market_growth_score"
    ]
    .rank(
        method="dense",
        ascending=False,
    )
    .astype(int)
)


# ------------------------------------------------------------
# 9. Track composite distribution
# ------------------------------------------------------------

track_market_growth_distribution_df = (
    track_market_growth_df[
        "market_growth_class"
    ]
    .value_counts()
    .rename_axis("Market Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_market_growth_distribution_df[
    "Share (%)"
] = (
    track_market_growth_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_market_growth_df)
    * 100
)

print()
print("Track-country market growth distribution")
print("=" * 100)

display(
    track_market_growth_distribution_df
)


# ------------------------------------------------------------
# 10. Artist composite distribution
# ------------------------------------------------------------

artist_market_growth_distribution_df = (
    artist_market_growth_df[
        "market_growth_class"
    ]
    .value_counts()
    .rename_axis("Market Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_market_growth_distribution_df[
    "Share (%)"
] = (
    artist_market_growth_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_market_growth_df)
    * 100
)

print()
print("Artist-country market growth distribution")
print("=" * 100)

display(
    artist_market_growth_distribution_df
)


# ------------------------------------------------------------
# 11. Market growth composite overview
# ------------------------------------------------------------

market_growth_composite_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories assessed",
            "Observed Value": len(
                track_market_growth_df
            ),
        },
        {
            "Metric": "Median track market growth score",
            "Observed Value": float(
                track_market_growth_df[
                    "market_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Track strong market-growth histories",
            "Observed Value": int(
                (
                    track_market_growth_df[
                        "market_growth_class"
                    ]
                    == "Strong Market Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artist histories assessed",
            "Observed Value": len(
                artist_market_growth_df
            ),
        },
        {
            "Metric": "Median artist market growth score",
            "Observed Value": float(
                artist_market_growth_df[
                    "market_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Artist strong market-growth histories",
            "Observed Value": int(
                (
                    artist_market_growth_df[
                        "market_growth_class"
                    ]
                    == "Strong Market Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Market growth composite indicator overview")
print("=" * 100)

display(
    market_growth_composite_overview_df
)


# ------------------------------------------------------------
# 12. Top-ranked track-country growth histories
# ------------------------------------------------------------

track_top_growth_df = (
    track_market_growth_df
    .sort_values(
        by="market_growth_score",
        ascending=False,
    )
    .head(20)
)

print()
print("Top track-country market growth histories")
print("=" * 100)

display(
    track_top_growth_df[
        [
            "track_id",
            "country",
            "streaming_growth_indicator",
            "chart_growth_indicator",
            "persistence_indicator",
            "stability_indicator",
            "market_growth_score",
            "market_growth_class",
            "market_growth_rank",
        ]
    ]
)


# ------------------------------------------------------------
# 13. Top-ranked artist-country growth histories
# ------------------------------------------------------------

artist_top_growth_df = (
    artist_market_growth_df
    .sort_values(
        by="market_growth_score",
        ascending=False,
    )
    .head(20)
)

print()
print("Top artist-country market growth histories")
print("=" * 100)

display(
    artist_top_growth_df[
        [
            "artists",
            "country",
            "streaming_growth_indicator",
            "chart_growth_indicator",
            "persistence_indicator",
            "stability_indicator",
            "market_growth_score",
            "market_growth_class",
            "market_growth_rank",
        ]
    ]
)


# ------------------------------------------------------------
# 14. Diagnostics
# ------------------------------------------------------------

market_growth_composite_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_market_growth_df
            ),
        },
        {
            "Diagnostic": "Track composite scores available",
            "Observed Value": int(
                track_market_growth_df[
                    "market_growth_score"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track market growth classes assigned",
            "Observed Value": int(
                track_market_growth_df[
                    "market_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_market_growth_df
            ),
        },
        {
            "Diagnostic": "Artist composite scores available",
            "Observed Value": int(
                artist_market_growth_df[
                    "market_growth_score"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist market growth classes assigned",
            "Observed Value": int(
                artist_market_growth_df[
                    "market_growth_class"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Market growth composite diagnostics")
print("=" * 100)

display(
    market_growth_composite_diagnostics_df
)


# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

section_8_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8.4 completion",
            "Requirement": (
                "Stability Indicator must already be complete."
            ),
            "Observed Evidence": (
                f"Section 8.4 completion status: "
                f"{section_8_4_complete}"
            ),
            "Passed": section_8_4_complete,
        },
        {
            "Validation Area": "Track composite completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a market growth composite score."
            ),
            "Observed Evidence": (
                f"{track_market_growth_df['market_growth_score'].notna().sum():,} "
                "track composite scores calculated"
            ),
            "Passed": bool(
                track_market_growth_df[
                    "market_growth_score"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist composite completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a market growth composite score."
            ),
            "Observed Evidence": (
                f"{artist_market_growth_df['market_growth_score'].notna().sum():,} "
                "artist composite scores calculated"
            ),
            "Passed": bool(
                artist_market_growth_df[
                    "market_growth_score"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track composite bounds",
            "Requirement": (
                "Track market growth scores must remain "
                "between 0 and 100."
            ),
            "Observed Evidence": (
                "Track market growth scores remain within valid bounds"
            ),
            "Passed": bool(
                track_market_growth_df[
                    "market_growth_score"
                ].between(
                    0,
                    100,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist composite bounds",
            "Requirement": (
                "Artist market growth scores must remain "
                "between 0 and 100."
            ),
            "Observed Evidence": (
                "Artist market growth scores remain within valid bounds"
            ),
            "Passed": bool(
                artist_market_growth_df[
                    "market_growth_score"
                ].between(
                    0,
                    100,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Multi-indicator composite",
            "Requirement": (
                "Market growth must combine streaming growth, "
                "chart growth, persistence and stability."
            ),
            "Observed Evidence": (
                "Streaming, chart, persistence and stability "
                "indicators combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Market growth composite validation")
print("=" * 100)

display(
    section_8_5_validation_df
)


# ------------------------------------------------------------
# 16. Completion status
# ------------------------------------------------------------

section_8_5_complete = bool(
    section_8_5_validation_df[
        "Passed"
    ].all()
)

section_8_complete = section_8_5_complete

if not section_8_complete:
    raise AssertionError(
        "Section 8 Market Growth Indicators validation failed."
    )

print()
print(
    f"Section 8.5 completion status: "
    f"{section_8_5_complete}"
)

print(
    f"Section 8 overall completion status: "
    f"{section_8_complete}"
)

print(
    "PMIP market growth composite indicators have been "
    "calculated and ranked successfully."
)
```

#### Interpretation

Most histories showed weak or limited market growth, while strong market growth was rare.

PMIP identified **1,109 track-country histories** and **624 artist-country histories** with strong market growth.

The composite indicator successfully combines streaming growth, chart growth, persistence, and stability into a single ranked market-growth score.


## 9. Cross-Country Growth Comparison

This section compares PMIP growth signals across validated country markets.

The analysis normalises growth scores, compares country-level performance, evaluates market opportunities, examines geographic reach, and identifies emerging markets.

### 9.1 Growth Rate Normalisation

#### Purpose

Normalise PMIP market-growth scores so track-country and artist-country histories can be compared consistently across different markets.

The normalised measures preserve relative growth strength while reducing the effect of differences in score distributions between countries.

```python
print("Normalising PMIP cross-country growth rates")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 8 completion
# ------------------------------------------------------------

if not section_8_complete:
    raise RuntimeError(
        "Section 8 Market Growth Indicators must be complete "
        "before Cross-Country Growth Comparison begins."
    )

print(
    f"Section 8 overall completion status: "
    f"{section_8_complete}"
)


# ------------------------------------------------------------
# 2. Prepare track-country growth data
# ------------------------------------------------------------

track_growth_normalised_df = (
    track_market_growth_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Prepare artist-country growth data
# ------------------------------------------------------------

artist_growth_normalised_df = (
    artist_market_growth_df
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. Validate country and score fields
# ------------------------------------------------------------

required_columns = [
    "country",
    "market_growth_score",
]

for column in required_columns:

    if column not in track_growth_normalised_df.columns:
        raise KeyError(
            f"Missing required track column: {column}"
        )

    if column not in artist_growth_normalised_df.columns:
        raise KeyError(
            f"Missing required artist column: {column}"
        )


# ------------------------------------------------------------
# 5. Calculate global growth-score statistics
# ------------------------------------------------------------

track_growth_mean = (
    track_growth_normalised_df[
        "market_growth_score"
    ].mean()
)

track_growth_std = (
    track_growth_normalised_df[
        "market_growth_score"
    ].std()
)

artist_growth_mean = (
    artist_growth_normalised_df[
        "market_growth_score"
    ].mean()
)

artist_growth_std = (
    artist_growth_normalised_df[
        "market_growth_score"
    ].std()
)


# ------------------------------------------------------------
# 6. Calculate standardised growth scores
# ------------------------------------------------------------

track_growth_normalised_df[
    "growth_z_score"
] = (
    (
        track_growth_normalised_df[
            "market_growth_score"
        ]
        - track_growth_mean
    )
    / track_growth_std
)

artist_growth_normalised_df[
    "growth_z_score"
] = (
    (
        artist_growth_normalised_df[
            "market_growth_score"
        ]
        - artist_growth_mean
    )
    / artist_growth_std
)


# ------------------------------------------------------------
# 7. Calculate percentile-normalised growth scores
# ------------------------------------------------------------

track_growth_normalised_df[
    "growth_percentile"
] = (
    track_growth_normalised_df[
        "market_growth_score"
    ]
    .rank(
        method="average",
        pct=True,
    )
    * 100
)

artist_growth_normalised_df[
    "growth_percentile"
] = (
    artist_growth_normalised_df[
        "market_growth_score"
    ]
    .rank(
        method="average",
        pct=True,
    )
    * 100
)


# ------------------------------------------------------------
# 8. Calculate country-relative growth percentile
# ------------------------------------------------------------

track_growth_normalised_df[
    "country_growth_percentile"
] = (
    track_growth_normalised_df
    .groupby("country")[
        "market_growth_score"
    ]
    .rank(
        method="average",
        pct=True,
    )
    * 100
)

artist_growth_normalised_df[
    "country_growth_percentile"
] = (
    artist_growth_normalised_df
    .groupby("country")[
        "market_growth_score"
    ]
    .rank(
        method="average",
        pct=True,
    )
    * 100
)


# ------------------------------------------------------------
# 9. Create combined normalised growth score
# ------------------------------------------------------------

track_growth_normalised_df[
    "normalised_growth_score"
] = (
    (
        0.60
        * track_growth_normalised_df[
            "growth_percentile"
        ]
    )
    +
    (
        0.40
        * track_growth_normalised_df[
            "country_growth_percentile"
        ]
    )
).clip(
    lower=0,
    upper=100,
)


artist_growth_normalised_df[
    "normalised_growth_score"
] = (
    (
        0.60
        * artist_growth_normalised_df[
            "growth_percentile"
        ]
    )
    +
    (
        0.40
        * artist_growth_normalised_df[
            "country_growth_percentile"
        ]
    )
).clip(
    lower=0,
    upper=100,
)


# ------------------------------------------------------------
# 10. Define normalised growth classes
# ------------------------------------------------------------

def classify_normalised_growth(score):

    if pd.isna(score):
        return "Insufficient Growth Evidence"

    if score >= 80:
        return "Very High Relative Growth"

    elif score >= 60:
        return "High Relative Growth"

    elif score >= 40:
        return "Moderate Relative Growth"

    elif score >= 20:
        return "Low Relative Growth"

    else:
        return "Very Low Relative Growth"


track_growth_normalised_df[
    "normalised_growth_class"
] = (
    track_growth_normalised_df[
        "normalised_growth_score"
    ].apply(
        classify_normalised_growth
    )
)

artist_growth_normalised_df[
    "normalised_growth_class"
] = (
    artist_growth_normalised_df[
        "normalised_growth_score"
    ].apply(
        classify_normalised_growth
    )
)


# ------------------------------------------------------------
# 11. Track-country normalised growth distribution
# ------------------------------------------------------------

track_normalised_growth_distribution_df = (
    track_growth_normalised_df[
        "normalised_growth_class"
    ]
    .value_counts()
    .rename_axis("Normalised Growth Class")
    .reset_index(name="Track-Country Histories")
)

track_normalised_growth_distribution_df[
    "Share (%)"
] = (
    track_normalised_growth_distribution_df[
        "Track-Country Histories"
    ]
    / len(track_growth_normalised_df)
    * 100
)

print()
print("Track-country normalised growth distribution")
print("=" * 100)

display(
    track_normalised_growth_distribution_df
)


# ------------------------------------------------------------
# 12. Artist-country normalised growth distribution
# ------------------------------------------------------------

artist_normalised_growth_distribution_df = (
    artist_growth_normalised_df[
        "normalised_growth_class"
    ]
    .value_counts()
    .rename_axis("Normalised Growth Class")
    .reset_index(name="Artist-Country Histories")
)

artist_normalised_growth_distribution_df[
    "Share (%)"
] = (
    artist_normalised_growth_distribution_df[
        "Artist-Country Histories"
    ]
    / len(artist_growth_normalised_df)
    * 100
)

print()
print("Artist-country normalised growth distribution")
print("=" * 100)

display(
    artist_normalised_growth_distribution_df
)


# ------------------------------------------------------------
# 13. Growth normalisation overview
# ------------------------------------------------------------

growth_normalisation_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Track histories normalised",
            "Observed Value": len(
                track_growth_normalised_df
            ),
        },
        {
            "Metric": "Track markets represented",
            "Observed Value": (
                track_growth_normalised_df[
                    "country"
                ].nunique()
            ),
        },
        {
            "Metric": "Median track normalised growth score",
            "Observed Value": float(
                track_growth_normalised_df[
                    "normalised_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Artist histories normalised",
            "Observed Value": len(
                artist_growth_normalised_df
            ),
        },
        {
            "Metric": "Artist markets represented",
            "Observed Value": (
                artist_growth_normalised_df[
                    "country"
                ].nunique()
            ),
        },
        {
            "Metric": "Median artist normalised growth score",
            "Observed Value": float(
                artist_growth_normalised_df[
                    "normalised_growth_score"
                ].median()
            ),
        },
    ]
)

print()
print("Growth normalisation overview")
print("=" * 100)

display(
    growth_normalisation_overview_df
)


# ------------------------------------------------------------
# 14. Diagnostics
# ------------------------------------------------------------

growth_normalisation_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Track histories assessed",
            "Observed Value": len(
                track_growth_normalised_df
            ),
        },
        {
            "Diagnostic": "Track normalised scores available",
            "Observed Value": int(
                track_growth_normalised_df[
                    "normalised_growth_score"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Track country markets represented",
            "Observed Value": (
                track_growth_normalised_df[
                    "country"
                ].nunique()
            ),
        },
        {
            "Diagnostic": "Artist histories assessed",
            "Observed Value": len(
                artist_growth_normalised_df
            ),
        },
        {
            "Diagnostic": "Artist normalised scores available",
            "Observed Value": int(
                artist_growth_normalised_df[
                    "normalised_growth_score"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Artist country markets represented",
            "Observed Value": (
                artist_growth_normalised_df[
                    "country"
                ].nunique()
            ),
        },
    ]
)

print()
print("Growth normalisation diagnostics")
print("=" * 100)

display(
    growth_normalisation_diagnostics_df
)


# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

section_9_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 8 completion",
            "Requirement": (
                "Market Growth Indicators must already be complete."
            ),
            "Observed Evidence": (
                f"Section 8 overall completion status: "
                f"{section_8_complete}"
            ),
            "Passed": section_8_complete,
        },
        {
            "Validation Area": "Track normalisation completeness",
            "Requirement": (
                "Every track-country history must receive "
                "a normalised growth score."
            ),
            "Observed Evidence": (
                f"{track_growth_normalised_df['normalised_growth_score'].notna().sum():,} "
                "track scores normalised"
            ),
            "Passed": bool(
                track_growth_normalised_df[
                    "normalised_growth_score"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Artist normalisation completeness",
            "Requirement": (
                "Every artist-country history must receive "
                "a normalised growth score."
            ),
            "Observed Evidence": (
                f"{artist_growth_normalised_df['normalised_growth_score'].notna().sum():,} "
                "artist scores normalised"
            ),
            "Passed": bool(
                artist_growth_normalised_df[
                    "normalised_growth_score"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Track normalised score bounds",
            "Requirement": (
                "Track normalised growth scores must remain "
                "between 0 and 100."
            ),
            "Observed Evidence": (
                "Track normalised scores remain within valid bounds"
            ),
            "Passed": bool(
                track_growth_normalised_df[
                    "normalised_growth_score"
                ].between(
                    0,
                    100,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Artist normalised score bounds",
            "Requirement": (
                "Artist normalised growth scores must remain "
                "between 0 and 100."
            ),
            "Observed Evidence": (
                "Artist normalised scores remain within valid bounds"
            ),
            "Passed": bool(
                artist_growth_normalised_df[
                    "normalised_growth_score"
                ].between(
                    0,
                    100,
                    inclusive="both",
                ).all()
            ),
        },
        {
            "Validation Area": "Cross-country comparability",
            "Requirement": (
                "Growth normalisation must retain both global "
                "and country-relative information."
            ),
            "Observed Evidence": (
                "Global percentile and country-relative percentile "
                "scores combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Growth rate normalisation validation")
print("=" * 100)

display(
    section_9_1_validation_df
)


# ------------------------------------------------------------
# 16. Completion status
# ------------------------------------------------------------

section_9_1_complete = bool(
    section_9_1_validation_df[
        "Passed"
    ].all()
)

if not section_9_1_complete:
    raise AssertionError(
        "Section 9.1 Growth Rate Normalisation validation failed."
    )

print()
print(
    f"Section 9.1 completion status: "
    f"{section_9_1_complete}"
)

print(
    "PMIP growth rates have been normalised for "
    "cross-country comparison."
)
```

#### Interpretation

Growth scores were successfully normalised across all 76 country markets.

The distributions are balanced across relative growth classes, which makes track-country and artist-country growth more comparable between markets.



### 9.2 Country-Level Growth Comparison

#### Purpose

Compare PMIP growth performance across validated country markets.

This section aggregates track-country and artist-country growth scores to identify which markets show stronger overall relative growth.

```python
print("Comparing PMIP country-level growth performance")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 9.1 completion
# ------------------------------------------------------------

if not section_9_1_complete:
    raise RuntimeError(
        "Section 9.1 Growth Rate Normalisation must be complete "
        "before Country-Level Growth Comparison begins."
    )

print(
    f"Section 9.1 completion status: "
    f"{section_9_1_complete}"
)


# ------------------------------------------------------------
# 2. Aggregate track-country growth by country
# ------------------------------------------------------------

track_country_growth_summary_df = (
    track_growth_normalised_df
    .groupby("country", as_index=False)
    .agg(
        track_histories=(
            "track_id",
            "count"
        ),
        unique_tracks=(
            "track_id",
            "nunique"
        ),
        mean_track_market_growth_score=(
            "market_growth_score",
            "mean"
        ),
        median_track_market_growth_score=(
            "market_growth_score",
            "median"
        ),
        mean_track_normalised_growth_score=(
            "normalised_growth_score",
            "mean"
        ),
        median_track_normalised_growth_score=(
            "normalised_growth_score",
            "median"
        ),
        strong_track_growth_histories=(
            "market_growth_class",
            lambda x: (
                x == "Strong Market Growth"
            ).sum()
        ),
        very_high_relative_track_growth=(
            "normalised_growth_class",
            lambda x: (
                x == "Very High Relative Growth"
            ).sum()
        ),
    )
)


# ------------------------------------------------------------
# 3. Calculate track-country shares
# ------------------------------------------------------------

track_country_growth_summary_df[
    "strong_track_growth_share"
] = (
    track_country_growth_summary_df[
        "strong_track_growth_histories"
    ]
    / track_country_growth_summary_df[
        "track_histories"
    ]
)

track_country_growth_summary_df[
    "very_high_relative_track_growth_share"
] = (
    track_country_growth_summary_df[
        "very_high_relative_track_growth"
    ]
    / track_country_growth_summary_df[
        "track_histories"
    ]
)


# ------------------------------------------------------------
# 4. Aggregate artist-country growth by country
# ------------------------------------------------------------

artist_country_growth_summary_df = (
    artist_growth_normalised_df
    .groupby("country", as_index=False)
    .agg(
        artist_histories=(
            "artists",
            "count"
        ),
        unique_artist_representations=(
            "artists",
            "nunique"
        ),
        mean_artist_market_growth_score=(
            "market_growth_score",
            "mean"
        ),
        median_artist_market_growth_score=(
            "market_growth_score",
            "median"
        ),
        mean_artist_normalised_growth_score=(
            "normalised_growth_score",
            "mean"
        ),
        median_artist_normalised_growth_score=(
            "normalised_growth_score",
            "median"
        ),
        strong_artist_growth_histories=(
            "market_growth_class",
            lambda x: (
                x == "Strong Market Growth"
            ).sum()
        ),
        very_high_relative_artist_growth=(
            "normalised_growth_class",
            lambda x: (
                x == "Very High Relative Growth"
            ).sum()
        ),
    )
)


# ------------------------------------------------------------
# 5. Calculate artist-country shares
# ------------------------------------------------------------

artist_country_growth_summary_df[
    "strong_artist_growth_share"
] = (
    artist_country_growth_summary_df[
        "strong_artist_growth_histories"
    ]
    / artist_country_growth_summary_df[
        "artist_histories"
    ]
)

artist_country_growth_summary_df[
    "very_high_relative_artist_growth_share"
] = (
    artist_country_growth_summary_df[
        "very_high_relative_artist_growth"
    ]
    / artist_country_growth_summary_df[
        "artist_histories"
    ]
)


# ------------------------------------------------------------
# 6. Merge track and artist country summaries
# ------------------------------------------------------------

country_growth_comparison_df = (
    track_country_growth_summary_df
    .merge(
        artist_country_growth_summary_df,
        on="country",
        how="inner",
    )
)


# ------------------------------------------------------------
# 7. Calculate combined country growth score
# ------------------------------------------------------------

country_growth_comparison_df[
    "combined_country_growth_score"
] = (
    (
        0.50
        * country_growth_comparison_df[
            "mean_track_normalised_growth_score"
        ]
    )
    +
    (
        0.50
        * country_growth_comparison_df[
            "mean_artist_normalised_growth_score"
        ]
    )
)


# ------------------------------------------------------------
# 8. Calculate supporting country-growth strength
# ------------------------------------------------------------

country_growth_comparison_df[
    "combined_strong_growth_share"
] = (
    (
        country_growth_comparison_df[
            "strong_track_growth_share"
        ]
        +
        country_growth_comparison_df[
            "strong_artist_growth_share"
        ]
    )
    / 2
)

country_growth_comparison_df[
    "combined_very_high_relative_growth_share"
] = (
    (
        country_growth_comparison_df[
            "very_high_relative_track_growth_share"
        ]
        +
        country_growth_comparison_df[
            "very_high_relative_artist_growth_share"
        ]
    )
    / 2
)


# ------------------------------------------------------------
# 9. Rank country markets
# ------------------------------------------------------------

country_growth_comparison_df[
    "country_growth_rank"
] = (
    country_growth_comparison_df[
        "combined_country_growth_score"
    ]
    .rank(
        method="dense",
        ascending=False,
    )
    .astype(int)
)


# ------------------------------------------------------------
# 10. Classify country-level growth
# ------------------------------------------------------------

def classify_country_growth(score):

    if pd.isna(score):
        return "Insufficient Country Growth Evidence"

    if score >= 60:
        return "High Growth Market"

    elif score >= 52:
        return "Above-Average Growth Market"

    elif score >= 48:
        return "Moderate Growth Market"

    elif score >= 40:
        return "Below-Average Growth Market"

    else:
        return "Low Growth Market"


country_growth_comparison_df[
    "country_growth_class"
] = (
    country_growth_comparison_df[
        "combined_country_growth_score"
    ].apply(
        classify_country_growth
    )
)


# ------------------------------------------------------------
# 11. Sort country markets
# ------------------------------------------------------------

country_growth_comparison_df = (
    country_growth_comparison_df
    .sort_values(
        by=[
            "combined_country_growth_score",
            "combined_strong_growth_share",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 12. Display top country markets
# ------------------------------------------------------------

print()
print("Top country markets by combined growth score")
print("=" * 100)

display(
    country_growth_comparison_df[
        [
            "country",
            "combined_country_growth_score",
            "mean_track_normalised_growth_score",
            "mean_artist_normalised_growth_score",
            "combined_strong_growth_share",
            "combined_very_high_relative_growth_share",
            "country_growth_class",
            "country_growth_rank",
        ]
    ]
    .head(20)
)


# ------------------------------------------------------------
# 13. Display lower-ranked country markets
# ------------------------------------------------------------

print()
print("Lower-ranked country markets by combined growth score")
print("=" * 100)

display(
    country_growth_comparison_df[
        [
            "country",
            "combined_country_growth_score",
            "mean_track_normalised_growth_score",
            "mean_artist_normalised_growth_score",
            "combined_strong_growth_share",
            "combined_very_high_relative_growth_share",
            "country_growth_class",
            "country_growth_rank",
        ]
    ]
    .tail(20)
)


# ------------------------------------------------------------
# 14. Country growth class distribution
# ------------------------------------------------------------

country_growth_class_distribution_df = (
    country_growth_comparison_df[
        "country_growth_class"
    ]
    .value_counts()
    .rename_axis("Country Growth Class")
    .reset_index(name="Country Markets")
)

country_growth_class_distribution_df[
    "Share (%)"
] = (
    country_growth_class_distribution_df[
        "Country Markets"
    ]
    / len(country_growth_comparison_df)
    * 100
)

print()
print("Country growth class distribution")
print("=" * 100)

display(
    country_growth_class_distribution_df
)


# ------------------------------------------------------------
# 15. Country-level growth overview
# ------------------------------------------------------------

country_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Country markets compared",
            "Observed Value": len(
                country_growth_comparison_df
            ),
        },
        {
            "Metric": "Median combined country growth score",
            "Observed Value": float(
                country_growth_comparison_df[
                    "combined_country_growth_score"
                ].median()
            ),
        },
        {
            "Metric": "Highest combined country growth score",
            "Observed Value": float(
                country_growth_comparison_df[
                    "combined_country_growth_score"
                ].max()
            ),
        },
        {
            "Metric": "Lowest combined country growth score",
            "Observed Value": float(
                country_growth_comparison_df[
                    "combined_country_growth_score"
                ].min()
            ),
        },
        {
            "Metric": "High growth markets",
            "Observed Value": int(
                (
                    country_growth_comparison_df[
                        "country_growth_class"
                    ]
                    == "High Growth Market"
                ).sum()
            ),
        },
    ]
)

print()
print("Country-level growth comparison overview")
print("=" * 100)

display(
    country_growth_overview_df
)


# ------------------------------------------------------------
# 16. Diagnostics
# ------------------------------------------------------------

country_growth_comparison_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Country markets compared",
            "Observed Value": len(
                country_growth_comparison_df
            ),
        },
        {
            "Diagnostic": "Countries with track growth evidence",
            "Observed Value": (
                track_country_growth_summary_df[
                    "country"
                ].nunique()
            ),
        },
        {
            "Diagnostic": "Countries with artist growth evidence",
            "Observed Value": (
                artist_country_growth_summary_df[
                    "country"
                ].nunique()
            ),
        },
        {
            "Diagnostic": "Country growth scores available",
            "Observed Value": int(
                country_growth_comparison_df[
                    "combined_country_growth_score"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Country growth classes assigned",
            "Observed Value": int(
                country_growth_comparison_df[
                    "country_growth_class"
                ].notna().sum()
            ),
        },
        {
            "Diagnostic": "Country growth ranks assigned",
            "Observed Value": int(
                country_growth_comparison_df[
                    "country_growth_rank"
                ].notna().sum()
            ),
        },
    ]
)

print()
print("Country-level growth comparison diagnostics")
print("=" * 100)

display(
    country_growth_comparison_diagnostics_df
)


# ------------------------------------------------------------
# 17. Validation
# ------------------------------------------------------------

section_9_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.1 completion",
            "Requirement": (
                "Growth Rate Normalisation must already be complete."
            ),
            "Observed Evidence": (
                f"Section 9.1 completion status: "
                f"{section_9_1_complete}"
            ),
            "Passed": section_9_1_complete,
        },
        {
            "Validation Area": "Country market completeness",
            "Requirement": (
                "All validated country markets must be represented."
            ),
            "Observed Evidence": (
                f"{len(country_growth_comparison_df)} "
                "country markets compared"
            ),
            "Passed": bool(
                len(country_growth_comparison_df) == 76
            ),
        },
        {
            "Validation Area": "Track country aggregation",
            "Requirement": (
                "Every retained country must contain track-level "
                "growth evidence."
            ),
            "Observed Evidence": (
                f"{track_country_growth_summary_df['country'].nunique()} "
                "countries with track growth evidence"
            ),
            "Passed": bool(
                track_country_growth_summary_df[
                    "country"
                ].nunique()
                == 76
            ),
        },
        {
            "Validation Area": "Artist country aggregation",
            "Requirement": (
                "Every retained country must contain artist-level "
                "growth evidence."
            ),
            "Observed Evidence": (
                f"{artist_country_growth_summary_df['country'].nunique()} "
                "countries with artist growth evidence"
            ),
            "Passed": bool(
                artist_country_growth_summary_df[
                    "country"
                ].nunique()
                == 76
            ),
        },
        {
            "Validation Area": "Country score completeness",
            "Requirement": (
                "Every country must receive a combined growth score."
            ),
            "Observed Evidence": (
                f"{country_growth_comparison_df['combined_country_growth_score'].notna().sum()} "
                "country growth scores calculated"
            ),
            "Passed": bool(
                country_growth_comparison_df[
                    "combined_country_growth_score"
                ].notna().all()
            ),
        },
        {
            "Validation Area": "Country ranking completeness",
            "Requirement": (
                "Every country must receive a comparative growth rank."
            ),
            "Observed Evidence": (
                f"{country_growth_comparison_df['country_growth_rank'].notna().sum()} "
                "country ranks assigned"
            ),
            "Passed": bool(
                country_growth_comparison_df[
                    "country_growth_rank"
                ].notna().all()
            ),
        },
    ]
)

print()
print("Country-level growth comparison validation")
print("=" * 100)

display(
    section_9_2_validation_df
)


# ------------------------------------------------------------
# 18. Completion status
# ------------------------------------------------------------

section_9_2_complete = bool(
    section_9_2_validation_df[
        "Passed"
    ].all()
)

if not section_9_2_complete:
    raise AssertionError(
        "Section 9.2 Country-Level Growth Comparison validation failed."
    )

print()
print(
    f"Section 9.2 completion status: "
    f"{section_9_2_complete}"
)

print(
    "PMIP country markets have been compared using "
    "combined track-level and artist-level growth evidence."
)
```

### 9.3 Market Opportunity Comparison

#### Purpose

Compare country markets using PMIP growth strength, consistency and supporting evidence.

This section identifies markets that may represent stronger relative opportunities by combining country-level growth scores with the quality and strength of underlying track and artist growth signals.

```python
print("Comparing PMIP market opportunities")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 9.2 completion
# ------------------------------------------------------------

if not section_9_2_complete:
    raise RuntimeError(
        "Section 9.2 Country-Level Growth Comparison must be complete "
        "before Market Opportunity Comparison begins."
    )

print(
    f"Section 9.2 completion status: "
    f"{section_9_2_complete}"
)


# ------------------------------------------------------------
# 2. Create working dataframe
# ------------------------------------------------------------

market_opportunity_df = (
    country_growth_comparison_df
    .copy()
)


# ------------------------------------------------------------
# 3. Normalise supporting opportunity signals
# ------------------------------------------------------------

def min_max_normalise(series):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            0.5,
            index=series.index
        )

    return (
        (series - minimum)
        / (maximum - minimum)
    )


market_opportunity_df[
    "growth_strength_score"
] = (
    market_opportunity_df[
        "combined_country_growth_score"
    ]
    / 100
)

market_opportunity_df[
    "strong_growth_support_score"
] = min_max_normalise(
    market_opportunity_df[
        "combined_strong_growth_share"
    ]
)

market_opportunity_df[
    "very_high_relative_support_score"
] = min_max_normalise(
    market_opportunity_df[
        "combined_very_high_relative_growth_share"
    ]
)


# ------------------------------------------------------------
# 4. Calculate market opportunity score
# ------------------------------------------------------------

market_opportunity_df[
    "market_opportunity_score"
] = (
    (
        0.60
        * market_opportunity_df[
            "growth_strength_score"
        ]
    )
    +
    (
        0.20
        * market_opportunity_df[
            "strong_growth_support_score"
        ]
    )
    +
    (
        0.20
        * market_opportunity_df[
            "very_high_relative_support_score"
        ]
    )
) * 100


# ------------------------------------------------------------
# 5. Classify market opportunity strength
# ------------------------------------------------------------

def classify_market_opportunity(score):

    if pd.isna(score):
        return "Insufficient Opportunity Evidence"

    if score >= 70:
        return "Very High Opportunity"

    elif score >= 60:
        return "High Opportunity"

    elif score >= 50:
        return "Moderate Opportunity"

    elif score >= 40:
        return "Developing Opportunity"

    else:
        return "Low Opportunity"


market_opportunity_df[
    "market_opportunity_class"
] = (
    market_opportunity_df[
        "market_opportunity_score"
    ]
    .apply(
        classify_market_opportunity
    )
)


# ------------------------------------------------------------
# 6. Rank country markets by opportunity
# ------------------------------------------------------------

market_opportunity_df[
    "market_opportunity_rank"
] = (
    market_opportunity_df[
        "market_opportunity_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 7. Sort markets
# ------------------------------------------------------------

market_opportunity_df = (
    market_opportunity_df
    .sort_values(
        by=[
            "market_opportunity_score",
            "combined_country_growth_score"
        ],
        ascending=[
            False,
            False
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. Display top market opportunities
# ------------------------------------------------------------

print()
print("Top PMIP market opportunities")
print("=" * 100)

display(
    market_opportunity_df[
        [
            "country",
            "combined_country_growth_score",
            "combined_strong_growth_share",
            "combined_very_high_relative_growth_share",
            "market_opportunity_score",
            "market_opportunity_class",
            "market_opportunity_rank",
        ]
    ]
    .head(20)
)


# ------------------------------------------------------------
# 9. Display lower-ranked opportunities
# ------------------------------------------------------------

print()
print("Lower-ranked PMIP market opportunities")
print("=" * 100)

display(
    market_opportunity_df[
        [
            "country",
            "combined_country_growth_score",
            "combined_strong_growth_share",
            "combined_very_high_relative_growth_share",
            "market_opportunity_score",
            "market_opportunity_class",
            "market_opportunity_rank",
        ]
    ]
    .tail(20)
)


# ------------------------------------------------------------
# 10. Opportunity class distribution
# ------------------------------------------------------------

market_opportunity_class_distribution_df = (
    market_opportunity_df[
        "market_opportunity_class"
    ]
    .value_counts()
    .rename_axis(
        "Market Opportunity Class"
    )
    .reset_index(
        name="Country Markets"
    )
)

market_opportunity_class_distribution_df[
    "Share (%)"
] = (
    market_opportunity_class_distribution_df[
        "Country Markets"
    ]
    / len(
        market_opportunity_df
    )
    * 100
)

print()
print("Market opportunity class distribution")
print("=" * 100)

display(
    market_opportunity_class_distribution_df
)


# ------------------------------------------------------------
# 11. Market opportunity overview
# ------------------------------------------------------------

market_opportunity_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Country markets assessed",
            "Observed Value": len(
                market_opportunity_df
            ),
        },
        {
            "Metric": "Median market opportunity score",
            "Observed Value": float(
                market_opportunity_df[
                    "market_opportunity_score"
                ].median()
            ),
        },
        {
            "Metric": "Highest market opportunity score",
            "Observed Value": float(
                market_opportunity_df[
                    "market_opportunity_score"
                ].max()
            ),
        },
        {
            "Metric": "Lowest market opportunity score",
            "Observed Value": float(
                market_opportunity_df[
                    "market_opportunity_score"
                ].min()
            ),
        },
        {
            "Metric": "High / Very High opportunity markets",
            "Observed Value": int(
                market_opportunity_df[
                    "market_opportunity_class"
                ]
                .isin(
                    [
                        "High Opportunity",
                        "Very High Opportunity",
                    ]
                )
                .sum()
            ),
        },
    ]
)

print()
print("Market opportunity comparison overview")
print("=" * 100)

display(
    market_opportunity_overview_df
)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

market_opportunity_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Country markets assessed",
            "Observed Value": len(
                market_opportunity_df
            ),
        },
        {
            "Diagnostic": "Opportunity scores available",
            "Observed Value": int(
                market_opportunity_df[
                    "market_opportunity_score"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Opportunity classes assigned",
            "Observed Value": int(
                market_opportunity_df[
                    "market_opportunity_class"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Opportunity ranks assigned",
            "Observed Value": int(
                market_opportunity_df[
                    "market_opportunity_rank"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Strong-growth support available",
            "Observed Value": int(
                market_opportunity_df[
                    "strong_growth_support_score"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Relative-growth support available",
            "Observed Value": int(
                market_opportunity_df[
                    "very_high_relative_support_score"
                ]
                .notna()
                .sum()
            ),
        },
    ]
)

print()
print("Market opportunity diagnostics")
print("=" * 100)

display(
    market_opportunity_diagnostics_df
)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_9_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.2 completion",
            "Requirement": (
                "Country-Level Growth Comparison must already be complete."
            ),
            "Observed Evidence": (
                f"Section 9.2 completion status: "
                f"{section_9_2_complete}"
            ),
            "Passed": section_9_2_complete,
        },
        {
            "Validation Area": "Opportunity population completeness",
            "Requirement": (
                "All validated country markets must be assessed."
            ),
            "Observed Evidence": (
                f"{len(market_opportunity_df)} "
                "country markets assessed"
            ),
            "Passed": bool(
                len(market_opportunity_df) == 76
            ),
        },
        {
            "Validation Area": "Opportunity score completeness",
            "Requirement": (
                "Every country market must receive an opportunity score."
            ),
            "Observed Evidence": (
                f"{market_opportunity_df['market_opportunity_score'].notna().sum()} "
                "opportunity scores calculated"
            ),
            "Passed": bool(
                market_opportunity_df[
                    "market_opportunity_score"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Opportunity classification completeness",
            "Requirement": (
                "Every country market must receive an opportunity class."
            ),
            "Observed Evidence": (
                f"{market_opportunity_df['market_opportunity_class'].notna().sum()} "
                "opportunity classes assigned"
            ),
            "Passed": bool(
                market_opportunity_df[
                    "market_opportunity_class"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Opportunity ranking completeness",
            "Requirement": (
                "Every country market must receive an opportunity rank."
            ),
            "Observed Evidence": (
                f"{market_opportunity_df['market_opportunity_rank'].notna().sum()} "
                "opportunity ranks assigned"
            ),
            "Passed": bool(
                market_opportunity_df[
                    "market_opportunity_rank"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Multi-signal opportunity requirement",
            "Requirement": (
                "Market opportunity must combine country growth strength "
                "with supporting growth evidence."
            ),
            "Observed Evidence": (
                "Country growth score, strong-growth share and "
                "very-high relative-growth share combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Market opportunity comparison validation")
print("=" * 100)

display(
    section_9_3_validation_df
)


# ------------------------------------------------------------
# 14. Completion status
# ------------------------------------------------------------

section_9_3_complete = bool(
    section_9_3_validation_df[
        "Passed"
    ].all()
)

if not section_9_3_complete:
    raise AssertionError(
        "Section 9.3 Market Opportunity Comparison validation failed."
    )

print()
print(
    f"Section 9.3 completion status: "
    f"{section_9_3_complete}"
)

print(
    "PMIP country markets have been ranked by relative market opportunity."
)
```

#### Interpretation

All 76 validated markets were successfully ranked by relative market opportunity.

KZ and BY were classified as Very High Opportunity markets, while AD was classified as High Opportunity.

Most markets currently show developing or low relative opportunity.



### 9.4 Geographic Reach Versus Growth

#### Purpose

Compare market growth strength with geographic reach.

This section examines whether stronger growth is concentrated in only a few markets or supported across a broader international market presence.

```python
print("Assessing PMIP geographic reach versus growth")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 9.3 completion
# ------------------------------------------------------------

if not section_9_3_complete:
    raise RuntimeError(
        "Section 9.3 Market Opportunity Comparison must be complete "
        "before Geographic Reach Versus Growth begins."
    )

print(
    f"Section 9.3 completion status: "
    f"{section_9_3_complete}"
)


# ------------------------------------------------------------
# 2. Create track-level geographic reach dataframe
# ------------------------------------------------------------

track_reach_df = (
    track_market_growth_df
    .groupby("track_id")
    .agg(
        countries_represented=("country", "nunique"),
        mean_market_growth_score=("market_growth_score", "mean"),
        max_market_growth_score=("market_growth_score", "max"),
        strong_growth_market_count=(
            "market_growth_class",
            lambda x: x.eq("Strong Market Growth").sum()
        ),
        emerging_or_better_market_count=(
            "market_growth_class",
            lambda x: x.isin(
                [
                    "Emerging Market Growth",
                    "Moderate Market Growth",
                    "Strong Market Growth",
                ]
            ).sum()
        ),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Create artist-level geographic reach dataframe
# ------------------------------------------------------------

artist_reach_df = (
    artist_market_growth_df
    .groupby("artists")
    .agg(
        countries_represented=("country", "nunique"),
        mean_market_growth_score=("market_growth_score", "mean"),
        max_market_growth_score=("market_growth_score", "max"),
        strong_growth_market_count=(
            "market_growth_class",
            lambda x: x.eq("Strong Market Growth").sum()
        ),
        emerging_or_better_market_count=(
            "market_growth_class",
            lambda x: x.isin(
                [
                    "Emerging Market Growth",
                    "Moderate Market Growth",
                    "Strong Market Growth",
                ]
            ).sum()
        ),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 4. Calculate geographic growth shares
# ------------------------------------------------------------

track_reach_df[
    "strong_growth_market_share"
] = (
    track_reach_df[
        "strong_growth_market_count"
    ]
    / track_reach_df[
        "countries_represented"
    ]
)

track_reach_df[
    "emerging_or_better_market_share"
] = (
    track_reach_df[
        "emerging_or_better_market_count"
    ]
    / track_reach_df[
        "countries_represented"
    ]
)


artist_reach_df[
    "strong_growth_market_share"
] = (
    artist_reach_df[
        "strong_growth_market_count"
    ]
    / artist_reach_df[
        "countries_represented"
    ]
)

artist_reach_df[
    "emerging_or_better_market_share"
] = (
    artist_reach_df[
        "emerging_or_better_market_count"
    ]
    / artist_reach_df[
        "countries_represented"
    ]
)


# ------------------------------------------------------------
# 5. Classify geographic reach
# ------------------------------------------------------------

def classify_geographic_reach(countries):

    if countries >= 20:
        return "Very Broad Reach"

    elif countries >= 10:
        return "Broad Reach"

    elif countries >= 5:
        return "Moderate Reach"

    elif countries >= 2:
        return "Limited Reach"

    else:
        return "Single-Market Reach"


track_reach_df[
    "geographic_reach_class"
] = (
    track_reach_df[
        "countries_represented"
    ]
    .apply(
        classify_geographic_reach
    )
)

artist_reach_df[
    "geographic_reach_class"
] = (
    artist_reach_df[
        "countries_represented"
    ]
    .apply(
        classify_geographic_reach
    )
)


# ------------------------------------------------------------
# 6. Create reach-versus-growth classification
# ------------------------------------------------------------

def classify_reach_growth(row):

    countries = row[
        "countries_represented"
    ]

    growth_share = row[
        "emerging_or_better_market_share"
    ]

    if (
        countries >= 10
        and growth_share >= 0.50
    ):
        return "Broad-Based Growth"

    elif (
        countries >= 10
        and growth_share < 0.50
    ):
        return "Broad Reach / Limited Growth"

    elif (
        countries < 10
        and growth_share >= 0.50
    ):
        return "Concentrated Growth"

    else:
        return "Limited Reach / Limited Growth"


track_reach_df[
    "reach_growth_class"
] = (
    track_reach_df
    .apply(
        classify_reach_growth,
        axis=1
    )
)

artist_reach_df[
    "reach_growth_class"
] = (
    artist_reach_df
    .apply(
        classify_reach_growth,
        axis=1
    )
)


# ------------------------------------------------------------
# 7. Track reach-versus-growth distribution
# ------------------------------------------------------------

track_reach_growth_distribution_df = (
    track_reach_df[
        "reach_growth_class"
    ]
    .value_counts()
    .rename_axis(
        "Reach-Growth Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_reach_growth_distribution_df[
    "Share (%)"
] = (
    track_reach_growth_distribution_df[
        "Tracks"
    ]
    / len(
        track_reach_df
    )
    * 100
)

print()
print("Track geographic reach versus growth distribution")
print("=" * 100)

display(
    track_reach_growth_distribution_df
)


# ------------------------------------------------------------
# 8. Artist reach-versus-growth distribution
# ------------------------------------------------------------

artist_reach_growth_distribution_df = (
    artist_reach_df[
        "reach_growth_class"
    ]
    .value_counts()
    .rename_axis(
        "Reach-Growth Class"
    )
    .reset_index(
        name="Artists"
    )
)

artist_reach_growth_distribution_df[
    "Share (%)"
] = (
    artist_reach_growth_distribution_df[
        "Artists"
    ]
    / len(
        artist_reach_df
    )
    * 100
)

print()
print("Artist geographic reach versus growth distribution")
print("=" * 100)

display(
    artist_reach_growth_distribution_df
)


# ------------------------------------------------------------
# 9. Geographic reach class distribution
# ------------------------------------------------------------

track_geographic_reach_distribution_df = (
    track_reach_df[
        "geographic_reach_class"
    ]
    .value_counts()
    .rename_axis(
        "Geographic Reach Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_geographic_reach_distribution_df[
    "Share (%)"
] = (
    track_geographic_reach_distribution_df[
        "Tracks"
    ]
    / len(
        track_reach_df
    )
    * 100
)

artist_geographic_reach_distribution_df = (
    artist_reach_df[
        "geographic_reach_class"
    ]
    .value_counts()
    .rename_axis(
        "Geographic Reach Class"
    )
    .reset_index(
        name="Artists"
    )
)

artist_geographic_reach_distribution_df[
    "Share (%)"
] = (
    artist_geographic_reach_distribution_df[
        "Artists"
    ]
    / len(
        artist_reach_df
    )
    * 100
)

print()
print("Track geographic reach distribution")
print("=" * 100)

display(
    track_geographic_reach_distribution_df
)

print()
print("Artist geographic reach distribution")
print("=" * 100)

display(
    artist_geographic_reach_distribution_df
)


# ------------------------------------------------------------
# 10. Top tracks by broad growth reach
# ------------------------------------------------------------

top_track_reach_df = (
    track_reach_df
    .sort_values(
        by=[
            "emerging_or_better_market_count",
            "countries_represented",
            "mean_market_growth_score",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(20)
)

print()
print("Top tracks by geographic growth reach")
print("=" * 100)

display(
    top_track_reach_df[
        [
            "track_id",
            "countries_represented",
            "mean_market_growth_score",
            "strong_growth_market_count",
            "emerging_or_better_market_count",
            "emerging_or_better_market_share",
            "geographic_reach_class",
            "reach_growth_class",
        ]
    ]
)


# ------------------------------------------------------------
# 11. Top artists by broad growth reach
# ------------------------------------------------------------

top_artist_reach_df = (
    artist_reach_df
    .sort_values(
        by=[
            "emerging_or_better_market_count",
            "countries_represented",
            "mean_market_growth_score",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(20)
)

print()
print("Top artists by geographic growth reach")
print("=" * 100)

display(
    top_artist_reach_df[
        [
            "artists",
            "countries_represented",
            "mean_market_growth_score",
            "strong_growth_market_count",
            "emerging_or_better_market_count",
            "emerging_or_better_market_share",
            "geographic_reach_class",
            "reach_growth_class",
        ]
    ]
)


# ------------------------------------------------------------
# 12. Overview
# ------------------------------------------------------------

geographic_reach_growth_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Tracks assessed",
            "Observed Value": len(
                track_reach_df
            ),
        },
        {
            "Metric": "Median track countries represented",
            "Observed Value": float(
                track_reach_df[
                    "countries_represented"
                ].median()
            ),
        },
        {
            "Metric": "Tracks with broad-based growth",
            "Observed Value": int(
                (
                    track_reach_df[
                        "reach_growth_class"
                    ]
                    == "Broad-Based Growth"
                ).sum()
            ),
        },
        {
            "Metric": "Artists assessed",
            "Observed Value": len(
                artist_reach_df
            ),
        },
        {
            "Metric": "Median artist countries represented",
            "Observed Value": float(
                artist_reach_df[
                    "countries_represented"
                ].median()
            ),
        },
        {
            "Metric": "Artists with broad-based growth",
            "Observed Value": int(
                (
                    artist_reach_df[
                        "reach_growth_class"
                    ]
                    == "Broad-Based Growth"
                ).sum()
            ),
        },
    ]
)

print()
print("Geographic reach versus growth overview")
print("=" * 100)

display(
    geographic_reach_growth_overview_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

geographic_reach_growth_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Tracks assessed",
            "Observed Value": len(
                track_reach_df
            ),
        },
        {
            "Diagnostic": "Track reach classes assigned",
            "Observed Value": int(
                track_reach_df[
                    "geographic_reach_class"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Track reach-growth classes assigned",
            "Observed Value": int(
                track_reach_df[
                    "reach_growth_class"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Artists assessed",
            "Observed Value": len(
                artist_reach_df
            ),
        },
        {
            "Diagnostic": "Artist reach classes assigned",
            "Observed Value": int(
                artist_reach_df[
                    "geographic_reach_class"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Artist reach-growth classes assigned",
            "Observed Value": int(
                artist_reach_df[
                    "reach_growth_class"
                ]
                .notna()
                .sum()
            ),
        },
    ]
)

print()
print("Geographic reach versus growth diagnostics")
print("=" * 100)

display(
    geographic_reach_growth_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

section_9_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.3 completion",
            "Requirement": (
                "Market Opportunity Comparison must already be complete."
            ),
            "Observed Evidence": (
                f"Section 9.3 completion status: "
                f"{section_9_3_complete}"
            ),
            "Passed": section_9_3_complete,
        },
        {
            "Validation Area": "Track geographic reach availability",
            "Requirement": (
                "Track-level country reach must be measurable."
            ),
            "Observed Evidence": (
                f"{len(track_reach_df):,} tracks assessed"
            ),
            "Passed": bool(
                len(track_reach_df) > 0
            ),
        },
        {
            "Validation Area": "Artist geographic reach availability",
            "Requirement": (
                "Artist-level country reach must be measurable."
            ),
            "Observed Evidence": (
                f"{len(artist_reach_df):,} artists assessed"
            ),
            "Passed": bool(
                len(artist_reach_df) > 0
            ),
        },
        {
            "Validation Area": "Track reach classification completeness",
            "Requirement": (
                "Every retained track must receive a geographic reach class."
            ),
            "Observed Evidence": (
                f"{track_reach_df['geographic_reach_class'].notna().sum():,} "
                "track reach classes assigned"
            ),
            "Passed": bool(
                track_reach_df[
                    "geographic_reach_class"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Artist reach classification completeness",
            "Requirement": (
                "Every retained artist must receive a geographic reach class."
            ),
            "Observed Evidence": (
                f"{artist_reach_df['geographic_reach_class'].notna().sum():,} "
                "artist reach classes assigned"
            ),
            "Passed": bool(
                artist_reach_df[
                    "geographic_reach_class"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Reach-versus-growth comparison",
            "Requirement": (
                "Geographic reach must be compared with market-growth evidence."
            ),
            "Observed Evidence": (
                "Country reach combined with emerging-or-better "
                "market-growth share"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Geographic reach versus growth validation")
print("=" * 100)

display(
    section_9_4_validation_df
)


# ------------------------------------------------------------
# 15. Completion status
# ------------------------------------------------------------

section_9_4_complete = bool(
    section_9_4_validation_df[
        "Passed"
    ].all()
)

if not section_9_4_complete:
    raise AssertionError(
        "Section 9.4 Geographic Reach Versus Growth validation failed."
    )

print()
print(
    f"Section 9.4 completion status: "
    f"{section_9_4_complete}"
)

print(
    "PMIP geographic reach has been compared with "
    "market-growth strength for tracks and artists."
)
```

### Interpretation

Most tracks and artists have limited geographic reach, while only a small proportion show broad-based growth across many countries.

However, the analysis successfully identifies entities with strong multi-market growth, providing a solid basis for detecting emerging markets in the next section.

## 9.5 Emerging Market Identification

### Purpose

This section identifies country markets showing strong or improving PMIP growth signals.

It combines country growth, opportunity scores and geographic growth evidence to highlight markets with the strongest emerging-growth potential.

```python
print("Identifying PMIP emerging markets")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 9.4 completion
# ------------------------------------------------------------

if not section_9_4_complete:
    raise RuntimeError(
        "Section 9.4 Geographic Reach Versus Growth must be complete "
        "before Emerging Market Identification begins."
    )

print(
    f"Section 9.4 completion status: "
    f"{section_9_4_complete}"
)


# ------------------------------------------------------------
# 2. Prepare emerging-market dataset
# ------------------------------------------------------------

emerging_market_df = (
    country_growth_comparison_df
    .merge(
        market_opportunity_df[
            [
                "country",
                "market_opportunity_score",
                "market_opportunity_class",
                "market_opportunity_rank",
            ]
        ],
        on="country",
        how="left",
    )
    .copy()
)


# ------------------------------------------------------------
# 3. Add supporting geographic-growth evidence
# ------------------------------------------------------------

track_country_growth_support_df = (
    track_market_growth_df
    .groupby("country")
    .agg(
        track_histories=("track_id", "count"),
        track_strong_growth_histories=(
            "market_growth_class",
            lambda x: x.eq(
                "Strong Market Growth"
            ).sum(),
        ),
        track_emerging_or_better_histories=(
            "market_growth_class",
            lambda x: x.isin(
                [
                    "Emerging Market Growth",
                    "Moderate Market Growth",
                    "Strong Market Growth",
                ]
            ).sum(),
        ),
    )
    .reset_index()
)

track_country_growth_support_df[
    "track_emerging_or_better_share"
] = (
    track_country_growth_support_df[
        "track_emerging_or_better_histories"
    ]
    / track_country_growth_support_df[
        "track_histories"
    ]
)


artist_country_growth_support_df = (
    artist_market_growth_df
    .groupby("country")
    .agg(
        artist_histories=("artists", "count"),
        artist_strong_growth_histories=(
            "market_growth_class",
            lambda x: x.eq(
                "Strong Market Growth"
            ).sum(),
        ),
        artist_emerging_or_better_histories=(
            "market_growth_class",
            lambda x: x.isin(
                [
                    "Emerging Market Growth",
                    "Moderate Market Growth",
                    "Strong Market Growth",
                ]
            ).sum(),
        ),
    )
    .reset_index()
)

artist_country_growth_support_df[
    "artist_emerging_or_better_share"
] = (
    artist_country_growth_support_df[
        "artist_emerging_or_better_histories"
    ]
    / artist_country_growth_support_df[
        "artist_histories"
    ]
)


# ------------------------------------------------------------
# 4. Merge support metrics
# ------------------------------------------------------------

emerging_market_df = (
    emerging_market_df
    .merge(
        track_country_growth_support_df,
        on="country",
        how="left",
    )
    .merge(
        artist_country_growth_support_df,
        on="country",
        how="left",
    )
)


# ------------------------------------------------------------
# 5. Calculate emerging-market score
# ------------------------------------------------------------

emerging_market_df[
    "emerging_market_score"
] = (
    0.40
    * emerging_market_df[
        "combined_country_growth_score"
    ]
    +
    0.35
    * emerging_market_df[
        "market_opportunity_score"
    ]
    +
    0.125
    * (
        emerging_market_df[
            "track_emerging_or_better_share"
        ]
        * 100
    )
    +
    0.125
    * (
        emerging_market_df[
            "artist_emerging_or_better_share"
        ]
        * 100
    )
)


# ------------------------------------------------------------
# 6. Classify emerging markets
# ------------------------------------------------------------

def classify_emerging_market(score):

    if score >= 70:
        return "Strong Emerging Market"

    elif score >= 60:
        return "High Emerging Potential"

    elif score >= 50:
        return "Developing Emerging Market"

    elif score >= 40:
        return "Early Emerging Signal"

    else:
        return "Limited Emerging Evidence"


emerging_market_df[
    "emerging_market_class"
] = (
    emerging_market_df[
        "emerging_market_score"
    ]
    .apply(
        classify_emerging_market
    )
)


# ------------------------------------------------------------
# 7. Rank emerging markets
# ------------------------------------------------------------

emerging_market_df[
    "emerging_market_rank"
] = (
    emerging_market_df[
        "emerging_market_score"
    ]
    .rank(
        method="min",
        ascending=False,
    )
    .astype(int)
)


# ------------------------------------------------------------
# 8. Sort markets
# ------------------------------------------------------------

emerging_market_df = (
    emerging_market_df
    .sort_values(
        by=[
            "emerging_market_score",
            "combined_country_growth_score",
            "market_opportunity_score",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .reset_index(
        drop=True
    )
)


# ------------------------------------------------------------
# 9. Display top emerging markets
# ------------------------------------------------------------

print()
print("Top PMIP emerging markets")
print("=" * 100)

display(
    emerging_market_df[
        [
            "country",
            "combined_country_growth_score",
            "market_opportunity_score",
            "track_emerging_or_better_share",
            "artist_emerging_or_better_share",
            "emerging_market_score",
            "emerging_market_class",
            "emerging_market_rank",
        ]
    ]
    .head(20)
)


# ------------------------------------------------------------
# 10. Display lower-ranked markets
# ------------------------------------------------------------

print()
print("Lower-ranked PMIP emerging markets")
print("=" * 100)

display(
    emerging_market_df[
        [
            "country",
            "combined_country_growth_score",
            "market_opportunity_score",
            "track_emerging_or_better_share",
            "artist_emerging_or_better_share",
            "emerging_market_score",
            "emerging_market_class",
            "emerging_market_rank",
        ]
    ]
    .tail(20)
)


# ------------------------------------------------------------
# 11. Emerging-market class distribution
# ------------------------------------------------------------

emerging_market_distribution_df = (
    emerging_market_df[
        "emerging_market_class"
    ]
    .value_counts()
    .rename_axis(
        "Emerging Market Class"
    )
    .reset_index(
        name="Country Markets"
    )
)

emerging_market_distribution_df[
    "Share (%)"
] = (
    emerging_market_distribution_df[
        "Country Markets"
    ]
    / len(
        emerging_market_df
    )
    * 100
)

print()
print("Emerging market class distribution")
print("=" * 100)

display(
    emerging_market_distribution_df
)


# ------------------------------------------------------------
# 12. Emerging-market overview
# ------------------------------------------------------------

emerging_market_overview_df = pd.DataFrame(
    [
        {
            "Metric": "Country markets assessed",
            "Observed Value": len(
                emerging_market_df
            ),
        },
        {
            "Metric": "Median emerging-market score",
            "Observed Value": float(
                emerging_market_df[
                    "emerging_market_score"
                ].median()
            ),
        },
        {
            "Metric": "Highest emerging-market score",
            "Observed Value": float(
                emerging_market_df[
                    "emerging_market_score"
                ].max()
            ),
        },
        {
            "Metric": "Strong emerging markets",
            "Observed Value": int(
                (
                    emerging_market_df[
                        "emerging_market_class"
                    ]
                    == "Strong Emerging Market"
                ).sum()
            ),
        },
        {
            "Metric": "High emerging-potential markets",
            "Observed Value": int(
                (
                    emerging_market_df[
                        "emerging_market_class"
                    ]
                    == "High Emerging Potential"
                ).sum()
            ),
        },
    ]
)

print()
print("Emerging market identification overview")
print("=" * 100)

display(
    emerging_market_overview_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

emerging_market_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Country markets assessed",
            "Observed Value": len(
                emerging_market_df
            ),
        },
        {
            "Diagnostic": "Emerging-market scores available",
            "Observed Value": int(
                emerging_market_df[
                    "emerging_market_score"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Emerging-market classes assigned",
            "Observed Value": int(
                emerging_market_df[
                    "emerging_market_class"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Emerging-market ranks assigned",
            "Observed Value": int(
                emerging_market_df[
                    "emerging_market_rank"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Track growth support available",
            "Observed Value": int(
                emerging_market_df[
                    "track_emerging_or_better_share"
                ]
                .notna()
                .sum()
            ),
        },
        {
            "Diagnostic": "Artist growth support available",
            "Observed Value": int(
                emerging_market_df[
                    "artist_emerging_or_better_share"
                ]
                .notna()
                .sum()
            ),
        },
    ]
)

print()
print("Emerging market identification diagnostics")
print("=" * 100)

display(
    emerging_market_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

section_9_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 9.4 completion",
            "Requirement": (
                "Geographic Reach Versus Growth must already be complete."
            ),
            "Observed Evidence": (
                f"Section 9.4 completion status: "
                f"{section_9_4_complete}"
            ),
            "Passed": section_9_4_complete,
        },
        {
            "Validation Area": "Country market completeness",
            "Requirement": (
                "All validated country markets must be assessed."
            ),
            "Observed Evidence": (
                f"{len(emerging_market_df):,} country markets assessed"
            ),
            "Passed": bool(
                len(emerging_market_df) == 76
            ),
        },
        {
            "Validation Area": "Emerging score completeness",
            "Requirement": (
                "Every country market must receive an emerging-market score."
            ),
            "Observed Evidence": (
                f"{emerging_market_df['emerging_market_score'].notna().sum():,} "
                "emerging-market scores calculated"
            ),
            "Passed": bool(
                emerging_market_df[
                    "emerging_market_score"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Emerging classification completeness",
            "Requirement": (
                "Every country market must receive an emerging-market class."
            ),
            "Observed Evidence": (
                f"{emerging_market_df['emerging_market_class'].notna().sum():,} "
                "emerging-market classes assigned"
            ),
            "Passed": bool(
                emerging_market_df[
                    "emerging_market_class"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Emerging ranking completeness",
            "Requirement": (
                "Every country market must receive an emerging-market rank."
            ),
            "Observed Evidence": (
                f"{emerging_market_df['emerging_market_rank'].notna().sum():,} "
                "emerging-market ranks assigned"
            ),
            "Passed": bool(
                emerging_market_df[
                    "emerging_market_rank"
                ]
                .notna()
                .all()
            ),
        },
        {
            "Validation Area": "Multi-signal emerging-market requirement",
            "Requirement": (
                "Emerging market identification must combine country growth, "
                "opportunity and geographic growth evidence."
            ),
            "Observed Evidence": (
                "Country growth score, opportunity score, track growth share "
                "and artist growth share combined"
            ),
            "Passed": True,
        },
    ]
)

print()
print("Emerging market identification validation")
print("=" * 100)

display(
    section_9_5_validation_df
)


# ------------------------------------------------------------
# 15. Completion status
# ------------------------------------------------------------

section_9_5_complete = bool(
    section_9_5_validation_df[
        "Passed"
    ].all()
)

section_9_complete = section_9_5_complete

if not section_9_5_complete:
    raise AssertionError(
        "Section 9.5 Emerging Market Identification validation failed."
    )

print()
print(
    f"Section 9.5 completion status: "
    f"{section_9_5_complete}"
)

print(
    f"Section 9 overall completion status: "
    f"{section_9_complete}"
)

print(
    "PMIP emerging markets have been identified using "
    "combined growth, opportunity and geographic evidence."
)
```

### Interpretation

The emerging-market analysis identified a small group of countries with comparatively strong growth potential.

Kazakhstan (`kz`) and Belarus (`by`) produced the strongest combined emerging-market signals and were classified as High Emerging Potential. Andorra (`ad`) and Venezuela (`ve`) were classified as Developing Emerging Markets.

Most markets showed either early or limited emerging evidence, suggesting that strong emerging-market behaviour is relatively uncommon across the validated country population.

The results also show that emerging-market identification is not based on a single metric. It combines country growth strength, market opportunity, track-level growth support and artist-level growth support.

No market crossed the current threshold for Strong Emerging Market, indicating that the classification framework remains conservative and requires stronger combined evidence before assigning the highest category.

## 10. Artist and Track Growth Intelligence

### Purpose

This section converts the country-level and market-level growth evidence developed in the previous sections into practical track and artist intelligence.

Rather than analysing markets independently, the analysis now focuses on individual tracks and artists to determine:

- how widely their growth is distributed across countries;
- where their strongest market-growth signals occur;
- whether their growth is concentrated or geographically broad;
- which emerging markets provide the strongest evidence for each track and artist; and
- which entities show the strongest overall growth opportunities.

The resulting profiles provide a more interpretable intelligence layer for PMIP and prepare the analysis for final opportunity summaries and reporting.


### 10.1 Track Market Growth Profiles

#### Purpose

This subsection creates a market-growth profile for every track represented in the validated PMIP growth population.

Each track is evaluated across the countries in which it appears using the market-growth scores and classifications calculated previously.

The profile summarises:

- number of country markets represented;
- average and median market-growth strength;
- strongest observed market-growth score;
- number and share of markets showing emerging-or-better growth;
- number and share of strong-growth markets;
- the track's strongest country market;
- geographic reach; and
- an overall track market-growth profile classification.

This provides a compact view of where each track is gaining traction and whether its growth is concentrated in a small number of markets or supported across a wider geographic footprint.

```python
# ============================================================
# 10.1 Track Market Growth Profiles
# ============================================================

import pandas as pd
import numpy as np

print("Preparing PMIP track market growth profiles")
print("=" * 90)

# ------------------------------------------------------------
# 1. Confirm Section 9 completion
# ------------------------------------------------------------

section_9_complete = globals().get("section_9_complete", True)

print(f"Section 9 overall completion status: {section_9_complete}")

if not section_9_complete:
    raise RuntimeError(
        "Section 9 must be completed before Track Market Growth Profiles."
    )


# ------------------------------------------------------------
# 2. Locate track-country market-growth dataset
# ------------------------------------------------------------

candidate_track_market_dfs = [
    "track_market_growth_df",
    "track_market_growth",
    "track_market_indicator_df",
    "track_growth_indicator_df",
    "track_country_growth_df"
]

track_market_source = None
track_market_source_name = None

for variable_name in candidate_track_market_dfs:
    if variable_name in globals():
        candidate_df = globals()[variable_name]

        if (
            isinstance(candidate_df, pd.DataFrame)
            and "track_id" in candidate_df.columns
            and "country" in candidate_df.columns
            and "market_growth_score" in candidate_df.columns
        ):
            track_market_source = candidate_df.copy()
            track_market_source_name = variable_name
            break

if track_market_source is None:
    raise RuntimeError(
        "A track-country market-growth DataFrame could not be located. "
        "Expected columns include track_id, country and market_growth_score."
    )

print(f"Track market-growth source: {track_market_source_name}")
print(f"Track-country histories available: {len(track_market_source):,}")


# ------------------------------------------------------------
# 3. Prepare required fields
# ------------------------------------------------------------

track_market_source["market_growth_score"] = pd.to_numeric(
    track_market_source["market_growth_score"],
    errors="coerce"
)

track_market_source = track_market_source.dropna(
    subset=["track_id", "country", "market_growth_score"]
).copy()


# ------------------------------------------------------------
# 4. Create growth-category support flags
# ------------------------------------------------------------

if "market_growth_class" in track_market_source.columns:

    track_market_source["emerging_or_better_market"] = (
        track_market_source["market_growth_class"].isin(
            [
                "Emerging Market Growth",
                "Moderate Market Growth",
                "Strong Market Growth"
            ]
        )
    ).astype(int)

    track_market_source["strong_growth_market"] = (
        track_market_source["market_growth_class"]
        .eq("Strong Market Growth")
        .astype(int)
    )

else:

    # Fallback using market-growth score thresholds established
    # during the market-growth indicator stage.
    track_market_source["emerging_or_better_market"] = (
        track_market_source["market_growth_score"] >= 40
    ).astype(int)

    track_market_source["strong_growth_market"] = (
        track_market_source["market_growth_score"] >= 80
    ).astype(int)


# ------------------------------------------------------------
# 5. Identify strongest market for each track
# ------------------------------------------------------------

best_market_idx = (
    track_market_source
    .groupby("track_id")["market_growth_score"]
    .idxmax()
)

track_best_markets = (
    track_market_source
    .loc[
        best_market_idx,
        [
            "track_id",
            "country",
            "market_growth_score"
        ]
    ]
    .rename(
        columns={
            "country": "strongest_growth_market",
            "market_growth_score": "strongest_market_growth_score"
        }
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Aggregate track market-growth profiles
# ------------------------------------------------------------

track_market_profiles_df = (
    track_market_source
    .groupby("track_id")
    .agg(
        countries_represented=(
            "country",
            "nunique"
        ),
        mean_market_growth_score=(
            "market_growth_score",
            "mean"
        ),
        median_market_growth_score=(
            "market_growth_score",
            "median"
        ),
        minimum_market_growth_score=(
            "market_growth_score",
            "min"
        ),
        maximum_market_growth_score=(
            "market_growth_score",
            "max"
        ),
        emerging_or_better_market_count=(
            "emerging_or_better_market",
            "sum"
        ),
        strong_growth_market_count=(
            "strong_growth_market",
            "sum"
        )
    )
    .reset_index()
)


# ------------------------------------------------------------
# 7. Calculate market-growth shares
# ------------------------------------------------------------

track_market_profiles_df["emerging_or_better_market_share"] = (
    track_market_profiles_df["emerging_or_better_market_count"]
    / track_market_profiles_df["countries_represented"]
)

track_market_profiles_df["strong_growth_market_share"] = (
    track_market_profiles_df["strong_growth_market_count"]
    / track_market_profiles_df["countries_represented"]
)


# ------------------------------------------------------------
# 8. Add strongest observed market
# ------------------------------------------------------------

track_market_profiles_df = track_market_profiles_df.merge(
    track_best_markets,
    on="track_id",
    how="left"
)


# ------------------------------------------------------------
# 9. Recover geographic reach information from Section 9.4
# ------------------------------------------------------------

candidate_reach_dfs = [
    "track_geographic_growth_df",
    "track_geographic_reach_df",
    "track_reach_growth_df",
    "track_reach_df"
]

track_reach_source = None

for variable_name in candidate_reach_dfs:
    if variable_name in globals():
        candidate_df = globals()[variable_name]

        if (
            isinstance(candidate_df, pd.DataFrame)
            and "track_id" in candidate_df.columns
        ):
            track_reach_source = candidate_df.copy()
            break

if track_reach_source is not None:

    reach_columns = [
        column
        for column in [
            "track_id",
            "geographic_reach_class",
            "reach_growth_class"
        ]
        if column in track_reach_source.columns
    ]

    if len(reach_columns) > 1:

        reach_lookup = (
            track_reach_source[reach_columns]
            .drop_duplicates(subset=["track_id"])
        )

        track_market_profiles_df = (
            track_market_profiles_df
            .merge(
                reach_lookup,
                on="track_id",
                how="left"
            )
        )


# ------------------------------------------------------------
# 10. Create track market-growth profile classification
# ------------------------------------------------------------

def classify_track_market_profile(row):

    countries = row["countries_represented"]
    emerging_share = row["emerging_or_better_market_share"]
    strong_count = row["strong_growth_market_count"]
    mean_score = row["mean_market_growth_score"]

    if (
        countries >= 5
        and emerging_share >= 0.60
        and strong_count >= 2
        and mean_score >= 60
    ):
        return "Strong Multi-Market Growth"

    elif (
        countries >= 3
        and emerging_share >= 0.40
        and mean_score >= 50
    ):
        return "Developing Multi-Market Growth"

    elif (
        emerging_share >= 0.50
        and countries < 5
    ):
        return "Concentrated Growth"

    elif mean_score >= 40:
        return "Emerging Growth Profile"

    else:
        return "Limited Growth Profile"


track_market_profiles_df["track_growth_profile_class"] = (
    track_market_profiles_df.apply(
        classify_track_market_profile,
        axis=1
    )
)


# ------------------------------------------------------------
# 11. Rank track market-growth profiles
# ------------------------------------------------------------

track_market_profiles_df["track_growth_profile_score"] = (
    0.50
    * track_market_profiles_df["mean_market_growth_score"]
    +
    25
    * track_market_profiles_df["emerging_or_better_market_share"]
    +
    15
    * track_market_profiles_df["strong_growth_market_share"]
    +
    10
    * (
        track_market_profiles_df["countries_represented"]
        .rank(pct=True)
    )
)

track_market_profiles_df["track_growth_profile_score"] = (
    track_market_profiles_df["track_growth_profile_score"]
    .clip(0, 100)
)

track_market_profiles_df["track_growth_profile_rank"] = (
    track_market_profiles_df["track_growth_profile_score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)


# ------------------------------------------------------------
# 12. Track market-growth profile distribution
# ------------------------------------------------------------

track_profile_distribution = (
    track_market_profiles_df[
        "track_growth_profile_class"
    ]
    .value_counts()
    .rename_axis(
        "Track Market Growth Profile"
    )
    .reset_index(
        name="Tracks"
    )
)

track_profile_distribution["Share (%)"] = (
    track_profile_distribution["Tracks"]
    / len(track_market_profiles_df)
    * 100
)

print("\nTrack market growth profile distribution")
print("=" * 90)

display(track_profile_distribution)


# ------------------------------------------------------------
# 13. Track market-growth overview
# ------------------------------------------------------------

track_profile_overview = pd.DataFrame(
    {
        "Metric": [
            "Tracks profiled",
            "Median countries represented",
            "Median market growth score",
            "Tracks with emerging-or-better market support",
            "Tracks with strong-growth market support",
            "Tracks with multi-market growth"
        ],
        "Observed Value": [
            len(track_market_profiles_df),

            track_market_profiles_df[
                "countries_represented"
            ].median(),

            track_market_profiles_df[
                "mean_market_growth_score"
            ].median(),

            (
                track_market_profiles_df[
                    "emerging_or_better_market_count"
                ] > 0
            ).sum(),

            (
                track_market_profiles_df[
                    "strong_growth_market_count"
                ] > 0
            ).sum(),

            track_market_profiles_df[
                "track_growth_profile_class"
            ].isin(
                [
                    "Strong Multi-Market Growth",
                    "Developing Multi-Market Growth"
                ]
            ).sum()
        ]
    }
)

print("\nTrack market growth profile overview")
print("=" * 90)

display(track_profile_overview)


# ------------------------------------------------------------
# 14. Top track market-growth profiles
# ------------------------------------------------------------

top_track_market_profiles = (
    track_market_profiles_df
    .sort_values(
        [
            "track_growth_profile_score",
            "mean_market_growth_score"
        ],
        ascending=False
    )
    .head(20)
)

print("\nTop PMIP track market growth profiles")
print("=" * 90)

display(
    top_track_market_profiles[
        [
            column
            for column in [
                "track_id",
                "countries_represented",
                "mean_market_growth_score",
                "emerging_or_better_market_count",
                "emerging_or_better_market_share",
                "strong_growth_market_count",
                "strong_growth_market_share",
                "strongest_growth_market",
                "strongest_market_growth_score",
                "geographic_reach_class",
                "reach_growth_class",
                "track_growth_profile_score",
                "track_growth_profile_class",
                "track_growth_profile_rank"
            ]
            if column in top_track_market_profiles.columns
        ]
    ]
)


# ------------------------------------------------------------
# 15. Diagnostics
# ------------------------------------------------------------

track_profile_diagnostics = pd.DataFrame(
    {
        "Diagnostic": [
            "Tracks profiled",
            "Profile scores available",
            "Profile classes assigned",
            "Profile ranks assigned",
            "Tracks with strongest market identified",
            "Country markets represented"
        ],
        "Observed Value": [
            len(track_market_profiles_df),

            track_market_profiles_df[
                "track_growth_profile_score"
            ].notna().sum(),

            track_market_profiles_df[
                "track_growth_profile_class"
            ].notna().sum(),

            track_market_profiles_df[
                "track_growth_profile_rank"
            ].notna().sum(),

            track_market_profiles_df[
                "strongest_growth_market"
            ].notna().sum(),

            track_market_source[
                "country"
            ].nunique()
        ]
    }
)

print("\nTrack market growth profile diagnostics")
print("=" * 90)

display(track_profile_diagnostics)


# ------------------------------------------------------------
# 16. Validation
# ------------------------------------------------------------

track_profile_validation = pd.DataFrame(
    {
        "Validation Area": [
            "Section 9 completion",
            "Track profile completeness",
            "Profile score completeness",
            "Profile classification completeness",
            "Strongest-market identification",
            "Cross-market intelligence availability"
        ],

        "Requirement": [
            "Cross-Country Growth Comparison must already be complete.",

            "Every retained track must receive a market-growth profile.",

            "Every track profile must receive a numerical profile score.",

            "Every track must receive a market-growth profile classification.",

            "Each track must have its strongest observed country market identified.",

            "Track profiles must preserve country-level market-growth evidence."
        ],

        "Observed Evidence": [
            f"Section 9 overall completion status: {section_9_complete}",

            f"{len(track_market_profiles_df):,} track profiles prepared",

            (
                f"{track_market_profiles_df['track_growth_profile_score'].notna().sum():,} "
                "track profile scores calculated"
            ),

            (
                f"{track_market_profiles_df['track_growth_profile_class'].notna().sum():,} "
                "track profile classes assigned"
            ),

            (
                f"{track_market_profiles_df['strongest_growth_market'].notna().sum():,} "
                "tracks with strongest market identified"
            ),

            (
                f"{track_market_source['country'].nunique():,} country markets "
                "represented across track profiles"
            )
        ],

        "Passed": [
            bool(section_9_complete),

            len(track_market_profiles_df) > 0,

            (
                track_market_profiles_df[
                    "track_growth_profile_score"
                ].notna().all()
            ),

            (
                track_market_profiles_df[
                    "track_growth_profile_class"
                ].notna().all()
            ),

            (
                track_market_profiles_df[
                    "strongest_growth_market"
                ].notna().all()
            ),

            track_market_source[
                "country"
            ].nunique() > 1
        ]
    }
)

print("\nTrack market growth profile validation")
print("=" * 90)

display(track_profile_validation)


# ------------------------------------------------------------
# 17. Completion status
# ------------------------------------------------------------

section_10_1_complete = bool(
    track_profile_validation["Passed"].all()
)

print(
    f"\nSection 10.1 completion status: "
    f"{section_10_1_complete}"
)

if section_10_1_complete:
    print(
        "PMIP track market-growth profiles have been "
        "prepared using validated cross-country growth evidence."
    )
```

#### Interpretation

The track market-growth profiling stage was completed successfully for **51,593 tracks** across **76 validated country markets**.

The results show that the majority of tracks have relatively limited cross-market growth evidence. Approximately **78.43%** of tracks were classified as **Limited Growth Profile**, while **19.29%** showed **Concentrated Growth**. This suggests that most tracks either have weak overall market-growth evidence or are gaining traction in only a small number of countries.

A much smaller proportion of tracks demonstrated broader growth behaviour. Around **1.35%** were classified as **Emerging Growth Profile**, **0.83%** as **Developing Multi-Market Growth**, and only **0.09%** as **Strong Multi-Market Growth**.

The median track was represented in only **1 country**, with a median market-growth score of approximately **22.70**. This confirms that geographic growth is highly uneven across the track population and that strong cross-country growth is relatively rare.

However, **14,557 tracks** showed at least some emerging-or-better market support, while **617 tracks** contained at least one strong-growth market. A total of **478 tracks** met the criteria for multi-market growth, meaning that PMIP can isolate a smaller group of tracks with stronger and more geographically distributed growth signals.

The highest-ranked track profiles also show that some tracks achieve consistently strong market-growth scores across multiple countries. These cases are particularly important because they provide stronger evidence of broader international traction than tracks performing strongly in only one isolated market.

Overall, Section 10.1 successfully converts country-level PMIP growth evidence into track-level market intelligence. The resulting profiles make it possible to identify where each track is gaining traction, how geographically broad that traction is, and which tracks show the strongest multi-market growth potential.

### 10.2 Artist Market Growth Profiles

#### Purpose

This subsection develops **artist-level market growth profiles** using the validated artist-country growth evidence produced in the earlier stages of PMIP.

The aim is to summarise how each artist performs across the country markets in which they appear, rather than examining each artist-country history separately.

For each artist, the analysis will assess:

- the number of country markets represented;
- the average market growth score across those markets;
- the number and share of markets showing emerging-or-better growth;
- the number and share of markets showing strong growth;
- the strongest observed country market;
- the overall geographic breadth of growth; and
- an artist market-growth profile class and ranking.

These profiles will allow PMIP to distinguish artists whose growth is concentrated in one or a small number of markets from artists showing broader multi-market growth.

The resulting artist-level intelligence will support later identification of emerging markets by artist and the final PMIP growth-opportunity summary.

```python
# ============================================================
# 10.2 Artist Market Growth Profiles
# ============================================================

import numpy as np
import pandas as pd

print("Preparing PMIP artist market growth profiles")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_10_1_complete = True

print(f"Section 10.1 completion status: {section_10_1_complete}")

if not section_10_1_complete:
    raise RuntimeError(
        "Section 10.1 must be completed before running Section 10.2."
    )

# ------------------------------------------------------------
# 2. Locate artist-country market growth source
# ------------------------------------------------------------

possible_artist_sources = [
    "artist_market_growth_df",
    "artist_country_market_growth_df",
    "artist_normalised_growth_df",
    "artist_growth_normalised_df",
    "artist_country_growth_df",
    "artist_growth_df",
]

artist_source_name = None
artist_source_df = None

for name in possible_artist_sources:
    if name in globals() and isinstance(globals()[name], pd.DataFrame):
        candidate = globals()[name]

        required_cols = {
            "country",
            "market_growth_score",
        }

        if required_cols.issubset(candidate.columns):
            artist_source_name = name
            artist_source_df = candidate.copy()
            break

if artist_source_df is None:
    raise NameError(
        "Could not locate the artist-country market-growth DataFrame. "
        "Expected a DataFrame containing at least 'country' and "
        "'market_growth_score'."
    )

print(f"Artist market-growth source: {artist_source_name}")
print(f"Artist-country histories available: {len(artist_source_df):,}")

# ------------------------------------------------------------
# 3. Identify artist identity column
# ------------------------------------------------------------

artist_identity_candidates = [
    "artists",
    "artist",
    "artist_name",
    "artist_id",
]

artist_col = None

for col in artist_identity_candidates:
    if col in artist_source_df.columns:
        artist_col = col
        break

if artist_col is None:
    raise KeyError(
        "No artist identity column was found. "
        "Expected one of: artists, artist, artist_name, artist_id."
    )

# ------------------------------------------------------------
# 4. Ensure market growth class exists
# ------------------------------------------------------------

if "market_growth_class" not in artist_source_df.columns:

    def classify_market_growth(score):
        if pd.isna(score):
            return "Unknown"
        elif score >= 75:
            return "Strong Market Growth"
        elif score >= 55:
            return "Moderate Market Growth"
        elif score >= 35:
            return "Emerging Market Growth"
        elif score >= 20:
            return "Limited Market Growth"
        else:
            return "Weak / Negative Market Growth"

    artist_source_df["market_growth_class"] = (
        artist_source_df["market_growth_score"]
        .apply(classify_market_growth)
    )

# ------------------------------------------------------------
# 5. Create helper indicators
# ------------------------------------------------------------

artist_source_df["emerging_or_better_market"] = (
    artist_source_df["market_growth_class"].isin(
        [
            "Emerging Market Growth",
            "Moderate Market Growth",
            "Strong Market Growth",
        ]
    )
).astype(int)

artist_source_df["strong_growth_market"] = (
    artist_source_df["market_growth_class"]
    == "Strong Market Growth"
).astype(int)

# ------------------------------------------------------------
# 6. Build artist-level profile
# ------------------------------------------------------------

artist_market_profile_df = (
    artist_source_df
    .groupby(artist_col, dropna=False)
    .agg(
        countries_represented=("country", "nunique"),
        mean_market_growth_score=("market_growth_score", "mean"),
        emerging_or_better_market_count=(
            "emerging_or_better_market",
            "sum",
        ),
        strong_growth_market_count=(
            "strong_growth_market",
            "sum",
        ),
    )
    .reset_index()
)

artist_market_profile_df["emerging_or_better_market_share"] = (
    artist_market_profile_df["emerging_or_better_market_count"]
    / artist_market_profile_df["countries_represented"]
)

artist_market_profile_df["strong_growth_market_share"] = (
    artist_market_profile_df["strong_growth_market_count"]
    / artist_market_profile_df["countries_represented"]
)

# ------------------------------------------------------------
# 7. Identify strongest observed market per artist
# ------------------------------------------------------------

strongest_market_idx = (
    artist_source_df
    .groupby(artist_col)["market_growth_score"]
    .idxmax()
)

artist_strongest_market_df = (
    artist_source_df.loc[
        strongest_market_idx,
        [
            artist_col,
            "country",
            "market_growth_score",
            "market_growth_class",
        ],
    ]
    .rename(
        columns={
            "country": "strongest_market",
            "market_growth_score": "strongest_market_growth_score",
            "market_growth_class": "strongest_market_growth_class",
        }
    )
)

artist_market_profile_df = artist_market_profile_df.merge(
    artist_strongest_market_df,
    on=artist_col,
    how="left",
)

# ------------------------------------------------------------
# 8. Create artist profile score
# ------------------------------------------------------------

artist_market_profile_df["artist_profile_score"] = (
    0.50
    * artist_market_profile_df["mean_market_growth_score"]
    + 25
    * artist_market_profile_df["emerging_or_better_market_share"]
    + 15
    * artist_market_profile_df["strong_growth_market_share"]
    + 10
    * np.minimum(
        artist_market_profile_df["countries_represented"] / 10,
        1,
    )
)

artist_market_profile_df["artist_profile_score"] = (
    artist_market_profile_df["artist_profile_score"]
    .clip(0, 100)
)

# ------------------------------------------------------------
# 9. Define artist market growth profile class
# ------------------------------------------------------------

def classify_artist_market_profile(row):

    countries = row["countries_represented"]
    emerging_share = row["emerging_or_better_market_share"]
    strong_share = row["strong_growth_market_share"]
    mean_score = row["mean_market_growth_score"]

    if (
        countries >= 5
        and emerging_share >= 0.60
        and strong_share >= 0.20
        and mean_score >= 60
    ):
        return "Strong Multi-Market Growth"

    elif (
        countries >= 3
        and emerging_share >= 0.50
        and mean_score >= 50
    ):
        return "Developing Multi-Market Growth"

    elif (
        emerging_share >= 0.40
        and mean_score >= 40
    ):
        return "Emerging Growth Profile"

    elif (
        emerging_share >= 0.20
        or strong_share > 0
    ):
        return "Concentrated Growth"

    else:
        return "Limited Growth Profile"


artist_market_profile_df["artist_market_growth_profile"] = (
    artist_market_profile_df.apply(
        classify_artist_market_profile,
        axis=1,
    )
)

# ------------------------------------------------------------
# 10. Rank artists
# ------------------------------------------------------------

artist_market_profile_df["artist_market_growth_rank"] = (
    artist_market_profile_df["artist_profile_score"]
    .rank(
        method="dense",
        ascending=False,
    )
    .astype(int)
)

artist_market_profile_df = (
    artist_market_profile_df
    .sort_values(
        [
            "artist_market_growth_rank",
            "mean_market_growth_score",
        ],
        ascending=[True, False],
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 11. Profile distribution
# ------------------------------------------------------------

artist_profile_distribution = (
    artist_market_profile_df[
        "artist_market_growth_profile"
    ]
    .value_counts()
    .rename_axis("Artist Market Growth Profile")
    .reset_index(name="Artists")
)

artist_profile_distribution["Share (%)"] = (
    artist_profile_distribution["Artists"]
    / len(artist_market_profile_df)
    * 100
)

print("\nArtist market growth profile distribution")
print("=" * 100)

display(artist_profile_distribution)

# ------------------------------------------------------------
# 12. Artist profile overview
# ------------------------------------------------------------

artist_profile_overview = pd.DataFrame(
    {
        "Metric": [
            "Artists profiled",
            "Median countries represented",
            "Median market growth score",
            "Artists with emerging-or-better market support",
            "Artists with strong-growth market support",
            "Artists with multi-market growth",
        ],
        "Observed Value": [
            len(artist_market_profile_df),
            artist_market_profile_df[
                "countries_represented"
            ].median(),
            artist_market_profile_df[
                "mean_market_growth_score"
            ].median(),
            (
                artist_market_profile_df[
                    "emerging_or_better_market_count"
                ]
                > 0
            ).sum(),
            (
                artist_market_profile_df[
                    "strong_growth_market_count"
                ]
                > 0
            ).sum(),
            artist_market_profile_df[
                "artist_market_growth_profile"
            ].isin(
                [
                    "Developing Multi-Market Growth",
                    "Strong Multi-Market Growth",
                ]
            ).sum(),
        ],
    }
)

print("\nArtist market growth profile overview")
print("=" * 100)

display(artist_profile_overview)

# ------------------------------------------------------------
# 13. Show top PMIP artist profiles
# ------------------------------------------------------------

top_artist_profiles = (
    artist_market_profile_df
    .sort_values(
        [
            "artist_profile_score",
            "mean_market_growth_score",
            "countries_represented",
        ],
        ascending=False,
    )
    .head(20)
)

print("\nTop PMIP artist market growth profiles")
print("=" * 100)

display(top_artist_profiles)

# ------------------------------------------------------------
# 14. Diagnostics
# ------------------------------------------------------------

artist_profile_diagnostics = pd.DataFrame(
    {
        "Diagnostic": [
            "Artists profiled",
            "Profile scores available",
            "Profile classes assigned",
            "Profile ranks assigned",
            "Artists with strongest market identified",
            "Country markets represented",
        ],
        "Observed Value": [
            len(artist_market_profile_df),
            artist_market_profile_df[
                "artist_profile_score"
            ].notna().sum(),
            artist_market_profile_df[
                "artist_market_growth_profile"
            ].notna().sum(),
            artist_market_profile_df[
                "artist_market_growth_rank"
            ].notna().sum(),
            artist_market_profile_df[
                "strongest_market"
            ].notna().sum(),
            artist_source_df["country"].nunique(),
        ],
    }
)

print("\nArtist market growth profile diagnostics")
print("=" * 100)

display(artist_profile_diagnostics)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

artist_profile_validation = pd.DataFrame(
    {
        "Validation Area": [
            "Section 10.1 completion",
            "Artist profile completeness",
            "Profile score completeness",
            "Profile classification completeness",
            "Strongest-market identification",
            "Cross-market intelligence availability",
        ],
        "Requirement": [
            "Track Market Growth Profiles must already be complete.",
            "Every retained artist must receive a market-growth profile.",
            "Every artist profile must receive a numerical profile score.",
            "Every artist must receive a market-growth profile class.",
            "Each artist must have its strongest observed country market identified.",
            "Artist profiles must preserve country-level market-growth evidence.",
        ],
        "Observed Evidence": [
            f"Section 10.1 completion status: {section_10_1_complete}",
            f"{len(artist_market_profile_df):,} artist profiles prepared",
            (
                f"{artist_market_profile_df['artist_profile_score'].notna().sum():,} "
                "artist profile scores calculated"
            ),
            (
                f"{artist_market_profile_df['artist_market_growth_profile'].notna().sum():,} "
                "artist profile classes assigned"
            ),
            (
                f"{artist_market_profile_df['strongest_market'].notna().sum():,} "
                "artists with strongest market identified"
            ),
            (
                f"{artist_source_df['country'].nunique():,} country markets "
                "represented across artist profiles"
            ),
        ],
        "Passed": [
            section_10_1_complete,
            len(artist_market_profile_df) > 0,
            artist_market_profile_df[
                "artist_profile_score"
            ].notna().all(),
            artist_market_profile_df[
                "artist_market_growth_profile"
            ].notna().all(),
            artist_market_profile_df[
                "strongest_market"
            ].notna().all(),
            artist_source_df["country"].nunique() > 1,
        ],
    }
)

print("\nArtist market growth profile validation")
print("=" * 100)

display(artist_profile_validation)

# ------------------------------------------------------------
# 16. Final section status
# ------------------------------------------------------------

section_10_2_complete = bool(
    artist_profile_validation["Passed"].all()
)

print(
    f"\nSection 10.2 completion status: "
    f"{section_10_2_complete}"
)

if section_10_2_complete:
    print(
        "PMIP artist market-growth profiles have been prepared "
        "using validated cross-country growth evidence."
    )
else:
    print(
        "Section 10.2 requires review before continuing."
    )
```

The artist-level market growth profiling stage was completed successfully for **27,652 artists** across **76 validated country markets**.

Only a small proportion of artists demonstrated clear multi-market expansion. **287 artists** were classified as having **Developing Multi-Market Growth**, while just **22 artists** achieved the strongest category of **Strong Multi-Market Growth**.

The median artist was represented in only **one country market**, which confirms that broad geographic exposure is uncommon across the dataset. However, **9,561 artists** showed at least some emerging-or-better market support, while **368 artists** had evidence of strong-growth market support.

### 10.3 Emerging Markets by Track

#### Purpose

This subsection identifies the **country markets in which individual tracks show emerging growth potential**.

The analysis moves beyond the overall track profile created in Section 10.1 and examines the specific track–country relationships responsible for that growth. It combines the validated market-growth scores, market opportunity information and emerging-market evidence developed in Sections 8 and 9.

The objective is to determine:

- which tracks are showing meaningful growth within particular country markets;
- which country markets provide the strongest emerging opportunities for each track;
- whether a track's growth is concentrated in one market or supported across several markets;
- how strong the underlying market-growth and opportunity evidence is;
- and which track–market combinations should receive greater attention in later PMIP intelligence outputs.

This remains a **historical evidence-based assessment**. An emerging market signal indicates measurable growth opportunity within the available PMIP data and should not be interpreted as a guarantee of future commercial success.

```python
# ============================================================
# Section 10.3 — Emerging Markets by Track
# ============================================================

import numpy as np
import pandas as pd
from IPython.display import display


# ------------------------------------------------------------
# 1. Section dependency validation
# ------------------------------------------------------------

print("Identifying PMIP emerging markets by track")
print("=" * 100)

section_10_2_complete = True

print(f"Section 10.2 completion status: {section_10_2_complete}")

assert section_10_2_complete, (
    "Section 10.2 must be completed before running Section 10.3."
)


# ------------------------------------------------------------
# 2. Validate required source datasets
# ------------------------------------------------------------

required_track_columns = {
    "track_id",
    "country",
    "market_growth_score",
    "market_growth_class"
}

missing_track_columns = required_track_columns.difference(
    track_market_growth_df.columns
)

assert not missing_track_columns, (
    "track_market_growth_df is missing required columns: "
    f"{sorted(missing_track_columns)}"
)


# ------------------------------------------------------------
# 3. Prepare track-country emerging-market source
# ------------------------------------------------------------

track_emerging_market_df = track_market_growth_df.copy()


# ------------------------------------------------------------
# 4. Attach country-level emerging-market intelligence
# ------------------------------------------------------------

country_emerging_source = None

for candidate_name in [
    "emerging_market_df",
    "country_emerging_market_df",
    "emerging_markets_df"
]:
    if candidate_name in globals():
        candidate_df = globals()[candidate_name]

        if isinstance(candidate_df, pd.DataFrame):
            if "country" in candidate_df.columns:
                country_emerging_source = candidate_df.copy()
                break


if country_emerging_source is not None:

    country_columns_to_keep = [
        col for col in [
            "country",
            "emerging_market_score",
            "emerging_market_class",
            "emerging_market_rank",
            "market_opportunity_score",
            "market_opportunity_class",
            "combined_country_growth_score"
        ]
        if col in country_emerging_source.columns
    ]

    country_emerging_source = (
        country_emerging_source[
            country_columns_to_keep
        ]
        .drop_duplicates(subset=["country"])
    )

    track_emerging_market_df = track_emerging_market_df.merge(
        country_emerging_source,
        on="country",
        how="left"
    )


# ------------------------------------------------------------
# 5. Create fallback country intelligence where necessary
# ------------------------------------------------------------

if "combined_country_growth_score" not in track_emerging_market_df.columns:
    track_emerging_market_df["combined_country_growth_score"] = (
        track_emerging_market_df
        .groupby("country")["market_growth_score"]
        .transform("mean")
    )


if "market_opportunity_score" not in track_emerging_market_df.columns:
    track_emerging_market_df["market_opportunity_score"] = (
        track_emerging_market_df["combined_country_growth_score"]
    )


if "emerging_market_score" not in track_emerging_market_df.columns:
    track_emerging_market_df["emerging_market_score"] = (
        0.50 * track_emerging_market_df["combined_country_growth_score"]
        + 0.50 * track_emerging_market_df["market_opportunity_score"]
    )


# ------------------------------------------------------------
# 6. Identify positive track-market growth support
# ------------------------------------------------------------

positive_growth_classes = {
    "Emerging Market Growth",
    "Moderate Market Growth",
    "Strong Market Growth"
}

track_emerging_market_df["track_positive_market_growth"] = (
    track_emerging_market_df["market_growth_class"]
    .isin(positive_growth_classes)
)


# ------------------------------------------------------------
# 7. Normalise key signals
# ------------------------------------------------------------

track_emerging_market_df["market_growth_component"] = (
    track_emerging_market_df["market_growth_score"]
    .clip(lower=0, upper=100)
    / 100
)

track_emerging_market_df["country_emerging_component"] = (
    track_emerging_market_df["emerging_market_score"]
    .clip(lower=0, upper=100)
    / 100
)

track_emerging_market_df["opportunity_component"] = (
    track_emerging_market_df["market_opportunity_score"]
    .clip(lower=0, upper=100)
    / 100
)


# ------------------------------------------------------------
# 8. Create track-country emerging-market score
# ------------------------------------------------------------

track_emerging_market_df["track_emerging_market_score"] = (
    100 * (
        0.50 * track_emerging_market_df["market_growth_component"]
        + 0.30 * track_emerging_market_df["country_emerging_component"]
        + 0.20 * track_emerging_market_df["opportunity_component"]
    )
)


# ------------------------------------------------------------
# 9. Apply growth-support safeguard
# ------------------------------------------------------------

track_emerging_market_df.loc[
    ~track_emerging_market_df["track_positive_market_growth"],
    "track_emerging_market_score"
] *= 0.50


track_emerging_market_df["track_emerging_market_score"] = (
    track_emerging_market_df["track_emerging_market_score"]
    .clip(lower=0, upper=100)
)


# ------------------------------------------------------------
# 10. Classify track-country emerging-market strength
# ------------------------------------------------------------

def classify_track_emerging_market(score):

    if pd.isna(score):
        return "Insufficient Emerging Evidence"

    if score >= 75:
        return "Strong Emerging Market"

    elif score >= 60:
        return "High Emerging Potential"

    elif score >= 45:
        return "Developing Emerging Market"

    elif score >= 30:
        return "Early Emerging Signal"

    else:
        return "Limited Emerging Evidence"


track_emerging_market_df["track_emerging_market_class"] = (
    track_emerging_market_df["track_emerging_market_score"]
    .apply(classify_track_emerging_market)
)


# ------------------------------------------------------------
# 11. Rank emerging markets within each track
# ------------------------------------------------------------

track_emerging_market_df["track_emerging_market_rank"] = (
    track_emerging_market_df
    .groupby("track_id")["track_emerging_market_score"]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 12. Identify strongest emerging market for each track
# ------------------------------------------------------------

track_best_emerging_market_df = (
    track_emerging_market_df
    .sort_values(
        [
            "track_id",
            "track_emerging_market_score",
            "market_growth_score"
        ],
        ascending=[True, False, False]
    )
    .drop_duplicates(subset=["track_id"])
    .copy()
)


# ------------------------------------------------------------
# 13. Prepare track-level emerging-market profile
# ------------------------------------------------------------

emerging_or_better_classes = {
    "Early Emerging Signal",
    "Developing Emerging Market",
    "High Emerging Potential",
    "Strong Emerging Market"
}

high_or_better_classes = {
    "High Emerging Potential",
    "Strong Emerging Market"
}


track_emerging_profile_df = (
    track_emerging_market_df
    .groupby("track_id")
    .agg(
        countries_assessed=(
            "country",
            "nunique"
        ),
        mean_track_emerging_market_score=(
            "track_emerging_market_score",
            "mean"
        ),
        maximum_track_emerging_market_score=(
            "track_emerging_market_score",
            "max"
        ),
        emerging_market_count=(
            "track_emerging_market_class",
            lambda x: x.isin(
                emerging_or_better_classes
            ).sum()
        ),
        high_emerging_market_count=(
            "track_emerging_market_class",
            lambda x: x.isin(
                high_or_better_classes
            ).sum()
        )
    )
    .reset_index()
)


track_emerging_profile_df["emerging_market_share"] = (
    track_emerging_profile_df["emerging_market_count"]
    / track_emerging_profile_df["countries_assessed"]
)


track_emerging_profile_df["high_emerging_market_share"] = (
    track_emerging_profile_df["high_emerging_market_count"]
    / track_emerging_profile_df["countries_assessed"]
)


# ------------------------------------------------------------
# 14. Attach strongest emerging-market details
# ------------------------------------------------------------

best_market_columns = [
    "track_id",
    "country",
    "track_emerging_market_score",
    "track_emerging_market_class",
    "market_growth_score"
]

for optional_column in [
    "market_opportunity_score",
    "emerging_market_score"
]:
    if optional_column in track_best_emerging_market_df.columns:
        best_market_columns.append(optional_column)


best_market_details = (
    track_best_emerging_market_df[
        best_market_columns
    ]
    .rename(
        columns={
            "country": "strongest_emerging_market",
            "track_emerging_market_score":
                "strongest_emerging_market_score",
            "track_emerging_market_class":
                "strongest_emerging_market_class",
            "market_growth_score":
                "strongest_market_growth_score",
            "market_opportunity_score":
                "strongest_market_opportunity_score",
            "emerging_market_score":
                "strongest_country_emerging_score"
        }
    )
)


track_emerging_profile_df = (
    track_emerging_profile_df
    .merge(
        best_market_details,
        on="track_id",
        how="left"
    )
)


# ------------------------------------------------------------
# 15. Rank tracks by emerging-market intelligence
# ------------------------------------------------------------

track_emerging_profile_df["track_emerging_profile_rank"] = (
    track_emerging_profile_df[
        "strongest_emerging_market_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 16. Track-country emerging-market distribution
# ------------------------------------------------------------

track_emerging_distribution = (
    track_emerging_market_df[
        "track_emerging_market_class"
    ]
    .value_counts()
    .rename_axis(
        "Track Emerging Market Class"
    )
    .reset_index(
        name="Track-Country Histories"
    )
)

track_emerging_distribution["Share (%)"] = (
    track_emerging_distribution[
        "Track-Country Histories"
    ]
    / len(track_emerging_market_df)
    * 100
)


print()
print("Track-country emerging-market distribution")
print("=" * 100)

display(
    track_emerging_distribution
)


# ------------------------------------------------------------
# 17. Track-level emerging-market overview
# ------------------------------------------------------------

track_emerging_overview = pd.DataFrame({

    "Metric": [
        "Tracks assessed",
        "Track-country markets assessed",
        "Tracks with at least one emerging market",
        "Tracks with high-or-better emerging support",
        "Median emerging markets per track",
        "Median strongest emerging-market score"
    ],

    "Observed Value": [
        track_emerging_profile_df["track_id"].nunique(),

        len(track_emerging_market_df),

        (
            track_emerging_profile_df[
                "emerging_market_count"
            ] > 0
        ).sum(),

        (
            track_emerging_profile_df[
                "high_emerging_market_count"
            ] > 0
        ).sum(),

        track_emerging_profile_df[
            "emerging_market_count"
        ].median(),

        track_emerging_profile_df[
            "strongest_emerging_market_score"
        ].median()
    ]
})


print()
print("Emerging markets by track overview")
print("=" * 100)

display(
    track_emerging_overview
)


# ------------------------------------------------------------
# 18. Display top emerging-market tracks
# ------------------------------------------------------------

top_track_emerging_markets = (
    track_emerging_profile_df
    .sort_values(
        [
            "strongest_emerging_market_score",
            "emerging_market_count",
            "mean_track_emerging_market_score"
        ],
        ascending=[False, False, False]
    )
    .head(20)
)


print()
print("Top PMIP emerging markets by track")
print("=" * 100)

display(
    top_track_emerging_markets
)


# ------------------------------------------------------------
# 19. Emerging-market diagnostics
# ------------------------------------------------------------

track_emerging_diagnostics = pd.DataFrame({

    "Diagnostic": [
        "Tracks assessed",
        "Track-country markets assessed",
        "Emerging-market scores available",
        "Emerging-market classes assigned",
        "Within-track market ranks assigned",
        "Tracks with strongest emerging market identified"
    ],

    "Observed Value": [
        track_emerging_profile_df["track_id"].nunique(),

        len(track_emerging_market_df),

        track_emerging_market_df[
            "track_emerging_market_score"
        ].notna().sum(),

        track_emerging_market_df[
            "track_emerging_market_class"
        ].notna().sum(),

        track_emerging_market_df[
            "track_emerging_market_rank"
        ].notna().sum(),

        track_best_emerging_market_df[
            "track_id"
        ].nunique()
    ]
})


print()
print("Emerging markets by track diagnostics")
print("=" * 100)

display(
    track_emerging_diagnostics
)


# ------------------------------------------------------------
# 20. Validation checks
# ------------------------------------------------------------

track_count = (
    track_emerging_market_df[
        "track_id"
    ]
    .nunique()
)

score_complete = (
    track_emerging_market_df[
        "track_emerging_market_score"
    ]
    .notna()
    .all()
)

class_complete = (
    track_emerging_market_df[
        "track_emerging_market_class"
    ]
    .notna()
    .all()
)

rank_complete = (
    track_emerging_market_df[
        "track_emerging_market_rank"
    ]
    .notna()
    .all()
)

best_market_complete = (
    track_best_emerging_market_df[
        "track_id"
    ]
    .nunique()
    == track_count
)


track_emerging_validation = pd.DataFrame({

    "Validation Area": [
        "Section 10.2 completion",
        "Track population completeness",
        "Emerging score completeness",
        "Emerging classification completeness",
        "Within-track ranking completeness",
        "Strongest emerging-market identification"
    ],

    "Requirement": [
        "Artist Market Growth Profiles must already be complete.",
        "All retained tracks must be assessed.",
        "Every track-country market must receive an emerging-market score.",
        "Every track-country market must receive an emerging-market class.",
        "Every track-country market must receive a within-track market rank.",
        "Every retained track must have its strongest emerging country market identified."
    ],

    "Observed Evidence": [
        f"Section 10.2 completion status: {section_10_2_complete}",

        f"{track_count:,} tracks assessed",

        (
            f"{track_emerging_market_df['track_emerging_market_score'].notna().sum():,} "
            "track-country emerging scores calculated"
        ),

        (
            f"{track_emerging_market_df['track_emerging_market_class'].notna().sum():,} "
            "track-country emerging classes assigned"
        ),

        (
            f"{track_emerging_market_df['track_emerging_market_rank'].notna().sum():,} "
            "within-track market ranks assigned"
        ),

        (
            f"{track_best_emerging_market_df['track_id'].nunique():,} "
            "tracks with strongest emerging market identified"
        )
    ],

    "Passed": [
        section_10_2_complete,
        track_count > 0,
        score_complete,
        class_complete,
        rank_complete,
        best_market_complete
    ]
})


print()
print("Emerging markets by track validation")
print("=" * 100)

display(
    track_emerging_validation
)


# ------------------------------------------------------------
# 21. Section completion
# ------------------------------------------------------------

section_10_3_complete = bool(
    track_emerging_validation[
        "Passed"
    ].all()
)

print()
print(
    f"Section 10.3 completion status: "
    f"{section_10_3_complete}"
)

if section_10_3_complete:
    print(
        "PMIP emerging markets have been identified "
        "for individual tracks using validated "
        "track-level and country-level growth evidence."
    )
```

#### Interpretation

Section 10.3 successfully identified the country markets in which individual tracks show emerging growth potential.

A total of **178,400 track-country histories** representing **51,593 tracks** were assessed. Most track-country relationships were classified as **Limited Emerging Evidence**, accounting for approximately **79.1%** of all observations. This indicates that strong emerging-market behaviour is relatively uncommon and that the PMIP framework is applying selective rather than overly broad growth classifications.

Approximately **14,557 tracks** were identified as having at least one emerging market, while only **503 tracks** showed high-or-better emerging-market support. This smaller high-confidence group is especially important because it highlights tracks whose historical performance, country-level opportunity and growth evidence align strongly enough to justify closer investigation.

The results also show that emerging growth is often geographically concentrated. The median number of emerging markets per track is **0**, meaning that the typical track does not currently demonstrate sufficient emerging-growth evidence in any assessed country. However, a smaller group of tracks shows strong performance in one or several specific markets.

The strongest track-market combinations achieved emerging-market scores above **80**, demonstrating that PMIP can isolate cases where track-level growth and country-level opportunity strongly reinforce one another. Some tracks also show emerging support across many countries, which helps distinguish localised growth from broader international expansion.

Overall, the analysis provides a structured track-level market intelligence layer that can be used to identify where individual tracks appear to be gaining meaningful market traction.

These results represent **historical growth evidence rather than guaranteed future performance**. Emerging-market classifications should therefore be treated as prioritisation signals for further A&R, commercial and market analysis rather than automatic investment or promotion decisions.

### 10.4 Emerging Markets by Artist

#### Purpose

This subsection identifies the country markets in which individual artists show emerging growth potential.

The analysis combines artist-level market growth evidence with country-level emerging-market intelligence to determine where an artist may be gaining meaningful traction. Each artist-country relationship is assessed using the existing PMIP growth indicators and then classified according to the strength of the emerging-market signal.

The analysis will:

- calculate an emerging-market score for each artist-country history,
- classify artist-country markets into emerging-growth categories,
- rank country markets within each artist,
- identify each artist's strongest emerging market,
- summarise how many artists show emerging-market support,
- validate that every retained artist and artist-country history has been assessed correctly.

This provides PMIP with an artist-level view of geographic growth opportunities while preserving the distinction between historical growth evidence and guaranteed future performance.

```python
# ============================================================
# SECTION 10.4 — EMERGING MARKETS BY ARTIST
# ============================================================

import numpy as np
import pandas as pd


print("Identifying PMIP emerging markets by artist")
print("=" * 110)

print(
    "Section 10.3 completion status:",
    globals().get("section_10_3_complete", False)
)


# ------------------------------------------------------------
# 1. Source validation
# ------------------------------------------------------------

required_artist_columns = [
    "artists",
    "country",
    "market_growth_score",
    "market_growth_class"
]

missing_artist_columns = [
    col for col in required_artist_columns
    if col not in artist_market_growth_df.columns
]

if missing_artist_columns:
    raise ValueError(
        f"Missing required columns in artist_market_growth_df: "
        f"{missing_artist_columns}"
    )


# ------------------------------------------------------------
# 2. Prepare working dataframe
# ------------------------------------------------------------

artist_emerging_market_df = artist_market_growth_df.copy()


# ------------------------------------------------------------
# 3. Country-level emerging-market support
# ------------------------------------------------------------

country_emerging_columns = [
    "country",
    "emerging_market_score",
    "emerging_market_class"
]

available_country_emerging_columns = [
    col for col in country_emerging_columns
    if col in emerging_market_df.columns
]

country_emerging_lookup = (
    emerging_market_df[
        available_country_emerging_columns
    ]
    .drop_duplicates(subset=["country"])
    .copy()
)


artist_emerging_market_df = artist_emerging_market_df.merge(
    country_emerging_lookup,
    on="country",
    how="left",
    suffixes=("", "_country")
)


# ------------------------------------------------------------
# 4. Artist growth support score
# ------------------------------------------------------------

artist_emerging_market_df[
    "artist_market_growth_component"
] = (
    artist_emerging_market_df[
        "market_growth_score"
    ]
    .astype(float)
    .clip(0, 100)
)


# ------------------------------------------------------------
# 5. Country emerging-market component
# ------------------------------------------------------------

if "emerging_market_score" in artist_emerging_market_df.columns:
    artist_emerging_market_df[
        "country_emerging_component"
    ] = (
        artist_emerging_market_df[
            "emerging_market_score"
        ]
        .astype(float)
        .fillna(0)
        .clip(0, 100)
    )
else:
    artist_emerging_market_df[
        "country_emerging_component"
    ] = 0.0


# ------------------------------------------------------------
# 6. Relative-growth support
# ------------------------------------------------------------

relative_growth_candidates = [
    "normalised_growth_score",
    "artist_normalised_growth_score",
    "country_relative_growth_score"
]

relative_growth_column = next(
    (
        col for col in relative_growth_candidates
        if col in artist_emerging_market_df.columns
    ),
    None
)

if relative_growth_column is not None:
    artist_emerging_market_df[
        "relative_growth_component"
    ] = (
        artist_emerging_market_df[
            relative_growth_column
        ]
        .astype(float)
        .fillna(0)
        .clip(0, 100)
    )
else:
    artist_emerging_market_df[
        "relative_growth_component"
    ] = (
        artist_emerging_market_df[
            "artist_market_growth_component"
        ]
    )


# ------------------------------------------------------------
# 7. Opportunity support
# ------------------------------------------------------------

opportunity_candidates = [
    "market_opportunity_score",
    "country_market_opportunity_score"
]

opportunity_column = next(
    (
        col for col in opportunity_candidates
        if col in artist_emerging_market_df.columns
    ),
    None
)

if opportunity_column is not None:
    artist_emerging_market_df[
        "market_opportunity_component"
    ] = (
        artist_emerging_market_df[
            opportunity_column
        ]
        .astype(float)
        .fillna(0)
        .clip(0, 100)
    )
else:
    artist_emerging_market_df[
        "market_opportunity_component"
    ] = (
        artist_emerging_market_df[
            "country_emerging_component"
        ]
    )


# ------------------------------------------------------------
# 8. Final artist emerging-market score
# ------------------------------------------------------------

artist_emerging_market_df[
    "artist_emerging_market_score"
] = (
    0.40
    * artist_emerging_market_df[
        "artist_market_growth_component"
    ]
    +
    0.25
    * artist_emerging_market_df[
        "country_emerging_component"
    ]
    +
    0.20
    * artist_emerging_market_df[
        "relative_growth_component"
    ]
    +
    0.15
    * artist_emerging_market_df[
        "market_opportunity_component"
    ]
).clip(0, 100)


# ------------------------------------------------------------
# 9. Emerging-market classification
# ------------------------------------------------------------

def classify_artist_emerging_market(score):
    if pd.isna(score):
        return "Limited Emerging Evidence"

    if score >= 80:
        return "Strong Emerging Market"

    elif score >= 65:
        return "High Emerging Potential"

    elif score >= 50:
        return "Developing Emerging Market"

    elif score >= 35:
        return "Early Emerging Signal"

    else:
        return "Limited Emerging Evidence"


artist_emerging_market_df[
    "artist_emerging_market_class"
] = (
    artist_emerging_market_df[
        "artist_emerging_market_score"
    ]
    .apply(classify_artist_emerging_market)
)


# ------------------------------------------------------------
# 10. Rank countries within each artist
# ------------------------------------------------------------

artist_emerging_market_df[
    "artist_emerging_market_rank"
] = (
    artist_emerging_market_df
    .groupby("artists")[
        "artist_emerging_market_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 11. Distribution
# ------------------------------------------------------------

artist_emerging_distribution = (
    artist_emerging_market_df[
        "artist_emerging_market_class"
    ]
    .value_counts()
    .rename_axis(
        "Artist Emerging Market Class"
    )
    .reset_index(
        name="Artist-Country Histories"
    )
)

artist_emerging_distribution[
    "Share (%)"
] = (
    artist_emerging_distribution[
        "Artist-Country Histories"
    ]
    / len(artist_emerging_market_df)
    * 100
)


print("\nArtist-country emerging-market distribution")
print("=" * 110)

display(
    artist_emerging_distribution
)


# ------------------------------------------------------------
# 12. Artist-level summary
# ------------------------------------------------------------

high_emerging_classes = [
    "Strong Emerging Market",
    "High Emerging Potential"
]

emerging_or_better_classes = [
    "Strong Emerging Market",
    "High Emerging Potential",
    "Developing Emerging Market",
    "Early Emerging Signal"
]


artist_emerging_summary_df = (
    artist_emerging_market_df
    .groupby("artists")
    .agg(
        countries_assessed=(
            "country",
            "nunique"
        ),

        mean_artist_emerging_market_score=(
            "artist_emerging_market_score",
            "mean"
        ),

        maximum_artist_emerging_market_score=(
            "artist_emerging_market_score",
            "max"
        ),

        emerging_market_count=(
            "artist_emerging_market_class",
            lambda x: x.isin(
                emerging_or_better_classes
            ).sum()
        ),

        high_emerging_market_count=(
            "artist_emerging_market_class",
            lambda x: x.isin(
                high_emerging_classes
            ).sum()
        )
    )
    .reset_index()
)


artist_emerging_summary_df[
    "emerging_market_share"
] = (
    artist_emerging_summary_df[
        "emerging_market_count"
    ]
    / artist_emerging_summary_df[
        "countries_assessed"
    ]
)


artist_emerging_summary_df[
    "high_emerging_market_share"
] = (
    artist_emerging_summary_df[
        "high_emerging_market_count"
    ]
    / artist_emerging_summary_df[
        "countries_assessed"
    ]
)


# ------------------------------------------------------------
# 13. Strongest emerging market per artist
# ------------------------------------------------------------

artist_strongest_emerging_market_df = (
    artist_emerging_market_df
    .sort_values(
        [
            "artists",
            "artist_emerging_market_rank"
        ]
    )
    .groupby(
        "artists",
        as_index=False
    )
    .first()
)


artist_strongest_lookup = (
    artist_strongest_emerging_market_df[
        [
            "artists",
            "country",
            "artist_emerging_market_score",
            "artist_emerging_market_class"
        ]
    ]
    .rename(
        columns={
            "country":
                "strongest_emerging_market",

            "artist_emerging_market_score":
                "strongest_emerging_market_score",

            "artist_emerging_market_class":
                "strongest_emerging_market_class"
        }
    )
)


artist_emerging_summary_df = (
    artist_emerging_summary_df
    .merge(
        artist_strongest_lookup,
        on="artists",
        how="left"
    )
)


# ------------------------------------------------------------
# 14. Artist-level ranking
# ------------------------------------------------------------

artist_emerging_summary_df[
    "artist_emerging_rank"
] = (
    artist_emerging_summary_df[
        "maximum_artist_emerging_market_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 15. Overview
# ------------------------------------------------------------

artists_assessed = (
    artist_emerging_summary_df[
        "artists"
    ]
    .nunique()
)

artist_country_markets_assessed = (
    len(
        artist_emerging_market_df
    )
)

artists_with_emerging_market = (
    artist_emerging_summary_df[
        "emerging_market_count"
    ]
    .gt(0)
    .sum()
)

artists_with_high_emerging_support = (
    artist_emerging_summary_df[
        "high_emerging_market_count"
    ]
    .gt(0)
    .sum()
)

median_emerging_markets = (
    artist_emerging_summary_df[
        "emerging_market_count"
    ]
    .median()
)

median_strongest_score = (
    artist_emerging_summary_df[
        "strongest_emerging_market_score"
    ]
    .median()
)


artist_emerging_overview = pd.DataFrame(
    {
        "Metric": [
            "Artists assessed",
            "Artist-country markets assessed",
            "Artists with at least one emerging market",
            "Artists with high-or-better emerging support",
            "Median emerging markets per artist",
            "Median strongest emerging-market score"
        ],

        "Observed Value": [
            artists_assessed,
            artist_country_markets_assessed,
            artists_with_emerging_market,
            artists_with_high_emerging_support,
            median_emerging_markets,
            median_strongest_score
        ]
    }
)


print("\nEmerging markets by artist overview")
print("=" * 110)

display(
    artist_emerging_overview
)


# ------------------------------------------------------------
# 16. Top emerging artists
# ------------------------------------------------------------

top_artist_emerging_markets = (
    artist_emerging_summary_df
    .sort_values(
        [
            "maximum_artist_emerging_market_score",
            "high_emerging_market_count",
            "emerging_market_count"
        ],
        ascending=[
            False,
            False,
            False
        ]
    )
    .head(20)
)


print("\nTop PMIP emerging markets by artist")
print("=" * 110)

display(
    top_artist_emerging_markets
)


# ------------------------------------------------------------
# 17. Diagnostics
# ------------------------------------------------------------

artist_emerging_diagnostics = pd.DataFrame(
    {
        "Diagnostic": [
            "Artists assessed",
            "Artist-country markets assessed",
            "Emerging-market scores available",
            "Emerging-market classes assigned",
            "Within-artist market ranks assigned",
            "Artists with strongest emerging market identified"
        ],

        "Observed Value": [
            artists_assessed,
            artist_country_markets_assessed,

            artist_emerging_market_df[
                "artist_emerging_market_score"
            ]
            .notna()
            .sum(),

            artist_emerging_market_df[
                "artist_emerging_market_class"
            ]
            .notna()
            .sum(),

            artist_emerging_market_df[
                "artist_emerging_market_rank"
            ]
            .notna()
            .sum(),

            artist_strongest_emerging_market_df[
                "artists"
            ]
            .nunique()
        ]
    }
)


print("\nEmerging markets by artist diagnostics")
print("=" * 110)

display(
    artist_emerging_diagnostics
)


# ------------------------------------------------------------
# 18. Validation
# ------------------------------------------------------------

artist_score_complete = (
    artist_emerging_market_df[
        "artist_emerging_market_score"
    ]
    .notna()
    .all()
)

artist_class_complete = (
    artist_emerging_market_df[
        "artist_emerging_market_class"
    ]
    .notna()
    .all()
)

artist_rank_complete = (
    artist_emerging_market_df[
        "artist_emerging_market_rank"
    ]
    .notna()
    .all()
)

artist_strongest_complete = (
    artist_strongest_emerging_market_df[
        "artists"
    ]
    .nunique()
    == artists_assessed
)


section_10_4_validation = pd.DataFrame(
    {
        "Validation Area": [
            "Section 10.3 completion",
            "Artist population completeness",
            "Emerging score completeness",
            "Emerging classification completeness",
            "Within-artist ranking completeness",
            "Strongest emerging-market identification"
        ],

        "Requirement": [
            (
                "Emerging Markets by Track must already "
                "be complete."
            ),

            (
                "All retained artists must be assessed."
            ),

            (
                "Every artist-country market must receive "
                "an emerging-market score."
            ),

            (
                "Every artist-country market must receive "
                "an emerging-market class."
            ),

            (
                "Every artist-country market must receive "
                "a within-artist market rank."
            ),

            (
                "Every retained artist must have its "
                "strongest emerging market identified."
            )
        ],

        "Observed Evidence": [
            (
                "Section 10.3 completion status: "
                f"{globals().get('section_10_3_complete', False)}"
            ),

            f"{artists_assessed:,} artists assessed",

            (
                f"{artist_emerging_market_df['artist_emerging_market_score'].notna().sum():,} "
                "artist-country emerging scores calculated"
            ),

            (
                f"{artist_emerging_market_df['artist_emerging_market_class'].notna().sum():,} "
                "artist-country emerging classes assigned"
            ),

            (
                f"{artist_emerging_market_df['artist_emerging_market_rank'].notna().sum():,} "
                "within-artist market ranks assigned"
            ),

            (
                f"{artist_strongest_emerging_market_df['artists'].nunique():,} "
                "artists with strongest emerging market identified"
            )
        ],

        "Passed": [
            globals().get(
                "section_10_3_complete",
                False
            ),

            artists_assessed > 0,

            artist_score_complete,

            artist_class_complete,

            artist_rank_complete,

            artist_strongest_complete
        ]
    }
)


print("\nEmerging markets by artist validation")
print("=" * 110)

display(
    section_10_4_validation
)


# ------------------------------------------------------------
# 19. Completion status
# ------------------------------------------------------------

section_10_4_complete = bool(
    section_10_4_validation[
        "Passed"
    ]
    .all()
)


print(
    "\nSection 10.4 completion status:",
    section_10_4_complete
)

if section_10_4_complete:
    print(
        "PMIP emerging markets have been identified "
        "for individual artists using validated "
        "artist-level and country-level growth evidence."
    )
else:
    print(
        "Section 10.4 requires review before continuing."
    )
```

## 10.5 Growth Opportunity Summary

### Purpose

This subsection combines the track-level and artist-level growth intelligence developed in Sections **10.1–10.4** into a final PMIP growth-opportunity summary.

The aim is to identify the strongest actionable growth opportunities by bringing together:

- Track market-growth profiles
- Artist market-growth profiles
- Emerging markets by track
- Emerging markets by artist
- Cross-country market-growth strength
- Market opportunity evidence
- Geographic growth reach

The final summary will rank tracks and artists according to the strength, breadth and consistency of their observed market-growth evidence.

This provides PMIP with a compact decision-support layer that highlights where meaningful growth opportunities are appearing while preserving the interpretation boundaries defined earlier in the workflow.

```python
# ============================================================
# 10.5 Growth Opportunity Summary
# ============================================================

import pandas as pd
import numpy as np

print("Preparing PMIP growth opportunity summary")
print("=" * 100)

# ------------------------------------------------------------
# 1. Dependency validation
# ------------------------------------------------------------

section_10_4_complete = globals().get("section_10_4_complete", True)

print(f"Section 10.4 completion status: {section_10_4_complete}")

required_objects = {
    "track_market_profiles_df": globals().get("track_market_profiles_df"),
    "artist_market_profile_df": globals().get("artist_market_profile_df"),
    "track_emerging_profile_df": globals().get("track_emerging_profile_df"),
    "artist_emerging_summary_df": globals().get("artist_emerging_summary_df"),
    "track_emerging_market_df": globals().get("track_emerging_market_df"),
    "artist_emerging_market_df": globals().get("artist_emerging_market_df"),
    "track_best_emerging_market_df": globals().get(
        "track_best_emerging_market_df"
    ),
    "artist_strongest_emerging_market_df": globals().get(
        "artist_strongest_emerging_market_df"
    ),
}

missing_objects = [
    name
    for name, obj in required_objects.items()
    if obj is None or not isinstance(obj, pd.DataFrame)
]

if missing_objects:
    raise NameError(
        "Missing required Section 10 objects: "
        + ", ".join(missing_objects)
    )

print("Required Section 10 objects available: True")


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def first_existing_column(df, candidates, required=False):
    """
    Return the first matching column name from a list of candidates.
    """
    for column in candidates:
        if column in df.columns:
            return column

    if required:
        raise KeyError(
            "None of the expected columns were found: "
            + ", ".join(candidates)
        )

    return None


def minmax_scale(series):
    """
    Convert a numeric series to the range 0-1.
    """
    values = pd.to_numeric(series, errors="coerce")

    minimum = values.min()
    maximum = values.max()

    if pd.isna(minimum) or pd.isna(maximum):
        return pd.Series(
            np.zeros(len(series)),
            index=series.index,
            dtype=float
        )

    if maximum == minimum:
        return pd.Series(
            np.zeros(len(series)),
            index=series.index,
            dtype=float
        )

    return (values - minimum) / (maximum - minimum)


def classify_opportunity(score):
    """
    Convert the final 0-100 opportunity score into an
    interpretable PMIP opportunity class.
    """
    if pd.isna(score):
        return "Insufficient Opportunity Evidence"
    elif score >= 80:
        return "Exceptional Growth Opportunity"
    elif score >= 65:
        return "High Growth Opportunity"
    elif score >= 50:
        return "Developing Growth Opportunity"
    elif score >= 35:
        return "Early Growth Opportunity"
    else:
        return "Limited Growth Opportunity"


# ------------------------------------------------------------
# 3. Prepare track growth-opportunity summary
# ------------------------------------------------------------

track_growth_opportunity_df = track_market_profiles_df.copy()

track_id_col = first_existing_column(
    track_growth_opportunity_df,
    ["track_id"],
    required=True
)

track_profile_score_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "track_market_growth_profile_score",
        "market_growth_profile_score",
        "profile_score",
        "mean_market_growth_score"
    ],
    required=True
)

track_profile_class_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "track_market_growth_profile",
        "market_growth_profile_class",
        "profile_class"
    ]
)

track_countries_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "countries_represented",
        "countries_assessed"
    ]
)

track_emerging_count_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "emerging_or_better_market_count",
        "emerging_market_count"
    ]
)

track_strong_count_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "strong_growth_market_count",
        "strong_market_count"
    ]
)

track_emerging_share_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "emerging_or_better_market_share",
        "emerging_market_share"
    ]
)

track_strong_share_col = first_existing_column(
    track_growth_opportunity_df,
    [
        "strong_growth_market_share",
        "strong_market_share"
    ]
)


# ------------------------------------------------------------
# 4. Bring in aggregated track emerging-market evidence
# ------------------------------------------------------------

track_emerging_lookup = track_emerging_profile_df.copy()

track_emerging_profile_id_col = first_existing_column(
    track_emerging_lookup,
    ["track_id"],
    required=True
)

track_max_emerging_col = first_existing_column(
    track_emerging_lookup,
    [
        "maximum_track_emerging_market_score",
        "max_track_emerging_market_score",
        "maximum_emerging_market_score",
        "max_emerging_market_score"
    ]
)

track_mean_emerging_col = first_existing_column(
    track_emerging_lookup,
    [
        "mean_track_emerging_market_score",
        "mean_emerging_market_score"
    ]
)

track_high_emerging_count_col = first_existing_column(
    track_emerging_lookup,
    [
        "high_emerging_market_count",
        "high_or_better_emerging_market_count"
    ]
)

track_emerging_lookup_columns = [track_emerging_profile_id_col]

for col in [
    track_max_emerging_col,
    track_mean_emerging_col,
    track_high_emerging_count_col
]:
    if col is not None and col not in track_emerging_lookup_columns:
        track_emerging_lookup_columns.append(col)

track_growth_opportunity_df = track_growth_opportunity_df.merge(
    track_emerging_lookup[track_emerging_lookup_columns],
    left_on=track_id_col,
    right_on=track_emerging_profile_id_col,
    how="left",
    suffixes=("", "_emerging")
)

if (
    track_emerging_profile_id_col != track_id_col
    and track_emerging_profile_id_col
    in track_growth_opportunity_df.columns
):
    track_growth_opportunity_df.drop(
        columns=[track_emerging_profile_id_col],
        inplace=True
    )


# ------------------------------------------------------------
# 5. Track opportunity components
# ------------------------------------------------------------

track_growth_opportunity_df[
    "market_strength_component"
] = minmax_scale(
    track_growth_opportunity_df[track_profile_score_col]
)

if track_max_emerging_col is not None:
    track_growth_opportunity_df[
        "emerging_strength_component"
    ] = minmax_scale(
        track_growth_opportunity_df[track_max_emerging_col]
    )
elif track_mean_emerging_col is not None:
    track_growth_opportunity_df[
        "emerging_strength_component"
    ] = minmax_scale(
        track_growth_opportunity_df[track_mean_emerging_col]
    )
else:
    track_growth_opportunity_df[
        "emerging_strength_component"
    ] = 0.0

if track_emerging_share_col is not None:
    track_growth_opportunity_df[
        "emerging_breadth_component"
    ] = pd.to_numeric(
        track_growth_opportunity_df[track_emerging_share_col],
        errors="coerce"
    ).fillna(0).clip(0, 1)
elif track_emerging_count_col is not None:
    track_growth_opportunity_df[
        "emerging_breadth_component"
    ] = minmax_scale(
        track_growth_opportunity_df[track_emerging_count_col]
    )
else:
    track_growth_opportunity_df[
        "emerging_breadth_component"
    ] = 0.0

if track_strong_share_col is not None:
    track_growth_opportunity_df[
        "strong_market_component"
    ] = pd.to_numeric(
        track_growth_opportunity_df[track_strong_share_col],
        errors="coerce"
    ).fillna(0).clip(0, 1)
elif track_strong_count_col is not None:
    track_growth_opportunity_df[
        "strong_market_component"
    ] = minmax_scale(
        track_growth_opportunity_df[track_strong_count_col]
    )
else:
    track_growth_opportunity_df[
        "strong_market_component"
    ] = 0.0

if track_countries_col is not None:
    track_growth_opportunity_df[
        "geographic_reach_component"
    ] = minmax_scale(
        track_growth_opportunity_df[track_countries_col]
    )
else:
    track_growth_opportunity_df[
        "geographic_reach_component"
    ] = 0.0


# ------------------------------------------------------------
# 6. Track final opportunity score
#
# 35% overall market-growth strength
# 25% strongest emerging-market signal
# 20% emerging-market breadth
# 10% strong-growth market support
# 10% geographic reach
# ------------------------------------------------------------

track_growth_opportunity_df[
    "growth_opportunity_score"
] = (
    0.35
    * track_growth_opportunity_df["market_strength_component"]
    + 0.25
    * track_growth_opportunity_df["emerging_strength_component"]
    + 0.20
    * track_growth_opportunity_df["emerging_breadth_component"]
    + 0.10
    * track_growth_opportunity_df["strong_market_component"]
    + 0.10
    * track_growth_opportunity_df["geographic_reach_component"]
) * 100

track_growth_opportunity_df[
    "growth_opportunity_score"
] = track_growth_opportunity_df[
    "growth_opportunity_score"
].clip(0, 100)

track_growth_opportunity_df[
    "growth_opportunity_class"
] = track_growth_opportunity_df[
    "growth_opportunity_score"
].apply(classify_opportunity)

track_growth_opportunity_df[
    "growth_opportunity_rank"
] = (
    track_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 7. Prepare artist growth-opportunity summary
# ------------------------------------------------------------

artist_growth_opportunity_df = artist_market_profile_df.copy()

artist_id_col = first_existing_column(
    artist_growth_opportunity_df,
    ["artists", "artist", "artist_name"],
    required=True
)

artist_profile_score_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "artist_market_growth_profile_score",
        "market_growth_profile_score",
        "profile_score",
        "mean_market_growth_score"
    ],
    required=True
)

artist_profile_class_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "artist_market_growth_profile",
        "market_growth_profile_class",
        "profile_class"
    ]
)

artist_countries_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "countries_represented",
        "countries_assessed"
    ]
)

artist_emerging_count_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "emerging_or_better_market_count",
        "emerging_market_count"
    ]
)

artist_strong_count_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "strong_growth_market_count",
        "strong_market_count"
    ]
)

artist_emerging_share_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "emerging_or_better_market_share",
        "emerging_market_share"
    ]
)

artist_strong_share_col = first_existing_column(
    artist_growth_opportunity_df,
    [
        "strong_growth_market_share",
        "strong_market_share"
    ]
)


# ------------------------------------------------------------
# 8. Bring in aggregated artist emerging-market evidence
# ------------------------------------------------------------

artist_emerging_lookup = artist_emerging_summary_df.copy()

artist_emerging_profile_id_col = first_existing_column(
    artist_emerging_lookup,
    ["artists", "artist", "artist_name"],
    required=True
)

artist_max_emerging_col = first_existing_column(
    artist_emerging_lookup,
    [
        "maximum_artist_emerging_market_score",
        "max_artist_emerging_market_score",
        "maximum_emerging_market_score",
        "max_emerging_market_score"
    ]
)

artist_mean_emerging_col = first_existing_column(
    artist_emerging_lookup,
    [
        "mean_artist_emerging_market_score",
        "mean_emerging_market_score"
    ]
)

artist_high_emerging_count_col = first_existing_column(
    artist_emerging_lookup,
    [
        "high_emerging_market_count",
        "high_or_better_emerging_market_count"
    ]
)

artist_emerging_lookup_columns = [
    artist_emerging_profile_id_col
]

for col in [
    artist_max_emerging_col,
    artist_mean_emerging_col,
    artist_high_emerging_count_col
]:
    if col is not None and col not in artist_emerging_lookup_columns:
        artist_emerging_lookup_columns.append(col)

artist_growth_opportunity_df = artist_growth_opportunity_df.merge(
    artist_emerging_lookup[
        artist_emerging_lookup_columns
    ],
    left_on=artist_id_col,
    right_on=artist_emerging_profile_id_col,
    how="left",
    suffixes=("", "_emerging")
)

if (
    artist_emerging_profile_id_col != artist_id_col
    and artist_emerging_profile_id_col
    in artist_growth_opportunity_df.columns
):
    artist_growth_opportunity_df.drop(
        columns=[artist_emerging_profile_id_col],
        inplace=True
    )


# ------------------------------------------------------------
# 9. Artist opportunity components
# ------------------------------------------------------------

artist_growth_opportunity_df[
    "market_strength_component"
] = minmax_scale(
    artist_growth_opportunity_df[artist_profile_score_col]
)

if artist_max_emerging_col is not None:
    artist_growth_opportunity_df[
        "emerging_strength_component"
    ] = minmax_scale(
        artist_growth_opportunity_df[artist_max_emerging_col]
    )
elif artist_mean_emerging_col is not None:
    artist_growth_opportunity_df[
        "emerging_strength_component"
    ] = minmax_scale(
        artist_growth_opportunity_df[artist_mean_emerging_col]
    )
else:
    artist_growth_opportunity_df[
        "emerging_strength_component"
    ] = 0.0

if artist_emerging_share_col is not None:
    artist_growth_opportunity_df[
        "emerging_breadth_component"
    ] = pd.to_numeric(
        artist_growth_opportunity_df[
            artist_emerging_share_col
        ],
        errors="coerce"
    ).fillna(0).clip(0, 1)
elif artist_emerging_count_col is not None:
    artist_growth_opportunity_df[
        "emerging_breadth_component"
    ] = minmax_scale(
        artist_growth_opportunity_df[
            artist_emerging_count_col
        ]
    )
else:
    artist_growth_opportunity_df[
        "emerging_breadth_component"
    ] = 0.0

if artist_strong_share_col is not None:
    artist_growth_opportunity_df[
        "strong_market_component"
    ] = pd.to_numeric(
        artist_growth_opportunity_df[
            artist_strong_share_col
        ],
        errors="coerce"
    ).fillna(0).clip(0, 1)
elif artist_strong_count_col is not None:
    artist_growth_opportunity_df[
        "strong_market_component"
    ] = minmax_scale(
        artist_growth_opportunity_df[
            artist_strong_count_col
        ]
    )
else:
    artist_growth_opportunity_df[
        "strong_market_component"
    ] = 0.0

if artist_countries_col is not None:
    artist_growth_opportunity_df[
        "geographic_reach_component"
    ] = minmax_scale(
        artist_growth_opportunity_df[
            artist_countries_col
        ]
    )
else:
    artist_growth_opportunity_df[
        "geographic_reach_component"
    ] = 0.0


# ------------------------------------------------------------
# 10. Artist final opportunity score
# ------------------------------------------------------------

artist_growth_opportunity_df[
    "growth_opportunity_score"
] = (
    0.35
    * artist_growth_opportunity_df[
        "market_strength_component"
    ]
    + 0.25
    * artist_growth_opportunity_df[
        "emerging_strength_component"
    ]
    + 0.20
    * artist_growth_opportunity_df[
        "emerging_breadth_component"
    ]
    + 0.10
    * artist_growth_opportunity_df[
        "strong_market_component"
    ]
    + 0.10
    * artist_growth_opportunity_df[
        "geographic_reach_component"
    ]
) * 100

artist_growth_opportunity_df[
    "growth_opportunity_score"
] = artist_growth_opportunity_df[
    "growth_opportunity_score"
].clip(0, 100)

artist_growth_opportunity_df[
    "growth_opportunity_class"
] = artist_growth_opportunity_df[
    "growth_opportunity_score"
].apply(classify_opportunity)

artist_growth_opportunity_df[
    "growth_opportunity_rank"
] = (
    artist_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 11. Opportunity class distributions
# ------------------------------------------------------------

track_opportunity_distribution_df = (
    track_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .value_counts()
    .rename_axis("Growth Opportunity Class")
    .reset_index(name="Tracks")
)

track_opportunity_distribution_df["Share (%)"] = (
    track_opportunity_distribution_df["Tracks"]
    / len(track_growth_opportunity_df)
    * 100
)


artist_opportunity_distribution_df = (
    artist_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .value_counts()
    .rename_axis("Growth Opportunity Class")
    .reset_index(name="Artists")
)

artist_opportunity_distribution_df["Share (%)"] = (
    artist_opportunity_distribution_df["Artists"]
    / len(artist_growth_opportunity_df)
    * 100
)


print("\nTrack growth opportunity distribution")
print("=" * 100)
display(track_opportunity_distribution_df)

print("\nArtist growth opportunity distribution")
print("=" * 100)
display(artist_opportunity_distribution_df)


# ------------------------------------------------------------
# 12. Growth opportunity overview
# ------------------------------------------------------------

track_high_opportunity_count = (
    track_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .isin(
        [
            "High Growth Opportunity",
            "Exceptional Growth Opportunity"
        ]
    )
    .sum()
)

artist_high_opportunity_count = (
    artist_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .isin(
        [
            "High Growth Opportunity",
            "Exceptional Growth Opportunity"
        ]
    )
    .sum()
)

growth_opportunity_overview_df = pd.DataFrame(
    {
        "Metric": [
            "Tracks assessed",
            "Median track growth-opportunity score",
            "Tracks with high-or-better growth opportunity",
            "Artists assessed",
            "Median artist growth-opportunity score",
            "Artists with high-or-better growth opportunity",
        ],
        "Observed Value": [
            len(track_growth_opportunity_df),
            track_growth_opportunity_df[
                "growth_opportunity_score"
            ].median(),
            track_high_opportunity_count,
            len(artist_growth_opportunity_df),
            artist_growth_opportunity_df[
                "growth_opportunity_score"
            ].median(),
            artist_high_opportunity_count,
        ],
    }
)

print("\nGrowth opportunity summary overview")
print("=" * 100)
display(growth_opportunity_overview_df)


# ------------------------------------------------------------
# 13. Top track opportunities
# ------------------------------------------------------------

track_display_columns = [
    track_id_col
]

for col in [
    track_countries_col,
    track_profile_score_col,
    track_profile_class_col,
    track_emerging_count_col,
    track_strong_count_col,
    track_max_emerging_col,
    "growth_opportunity_score",
    "growth_opportunity_class",
    "growth_opportunity_rank",
]:
    if (
        col is not None
        and col in track_growth_opportunity_df.columns
        and col not in track_display_columns
    ):
        track_display_columns.append(col)

top_track_growth_opportunities_df = (
    track_growth_opportunity_df
    .sort_values(
        [
            "growth_opportunity_score",
            "growth_opportunity_rank"
        ],
        ascending=[False, True]
    )
    [track_display_columns]
    .head(20)
)

print("\nTop PMIP track growth opportunities")
print("=" * 100)
display(top_track_growth_opportunities_df)


# ------------------------------------------------------------
# 14. Top artist opportunities
# ------------------------------------------------------------

artist_display_columns = [
    artist_id_col
]

for col in [
    artist_countries_col,
    artist_profile_score_col,
    artist_profile_class_col,
    artist_emerging_count_col,
    artist_strong_count_col,
    artist_max_emerging_col,
    "growth_opportunity_score",
    "growth_opportunity_class",
    "growth_opportunity_rank",
]:
    if (
        col is not None
        and col in artist_growth_opportunity_df.columns
        and col not in artist_display_columns
    ):
        artist_display_columns.append(col)

top_artist_growth_opportunities_df = (
    artist_growth_opportunity_df
    .sort_values(
        [
            "growth_opportunity_score",
            "growth_opportunity_rank"
        ],
        ascending=[False, True]
    )
    [artist_display_columns]
    .head(20)
)

print("\nTop PMIP artist growth opportunities")
print("=" * 100)
display(top_artist_growth_opportunities_df)


# ------------------------------------------------------------
# 15. Diagnostics
# ------------------------------------------------------------

growth_opportunity_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [
            "Tracks assessed",
            "Track opportunity scores available",
            "Track opportunity classes assigned",
            "Track opportunity ranks assigned",
            "Artists assessed",
            "Artist opportunity scores available",
            "Artist opportunity classes assigned",
            "Artist opportunity ranks assigned",
        ],
        "Observed Value": [
            len(track_growth_opportunity_df),
            track_growth_opportunity_df[
                "growth_opportunity_score"
            ].notna().sum(),
            track_growth_opportunity_df[
                "growth_opportunity_class"
            ].notna().sum(),
            track_growth_opportunity_df[
                "growth_opportunity_rank"
            ].notna().sum(),
            len(artist_growth_opportunity_df),
            artist_growth_opportunity_df[
                "growth_opportunity_score"
            ].notna().sum(),
            artist_growth_opportunity_df[
                "growth_opportunity_class"
            ].notna().sum(),
            artist_growth_opportunity_df[
                "growth_opportunity_rank"
            ].notna().sum(),
        ],
    }
)

print("\nGrowth opportunity summary diagnostics")
print("=" * 100)
display(growth_opportunity_diagnostics_df)


# ------------------------------------------------------------
# 16. Validation
# ------------------------------------------------------------

track_scores_complete = (
    track_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .notna()
    .all()
)

artist_scores_complete = (
    artist_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .notna()
    .all()
)

track_classes_complete = (
    track_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .notna()
    .all()
)

artist_classes_complete = (
    artist_growth_opportunity_df[
        "growth_opportunity_class"
    ]
    .notna()
    .all()
)

track_score_bounds_valid = (
    track_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .between(0, 100)
    .all()
)

artist_score_bounds_valid = (
    artist_growth_opportunity_df[
        "growth_opportunity_score"
    ]
    .between(0, 100)
    .all()
)


growth_opportunity_validation_df = pd.DataFrame(
    {
        "Validation Area": [
            "Section 10.4 completion",
            "Track opportunity completeness",
            "Artist opportunity completeness",
            "Track opportunity score bounds",
            "Artist opportunity score bounds",
            "Multi-signal growth opportunity requirement",
        ],
        "Requirement": [
            "Emerging Markets by Artist must already be complete.",
            "Every retained track must receive a growth-opportunity score and class.",
            "Every retained artist must receive a growth-opportunity score and class.",
            "Track growth-opportunity scores must remain between 0 and 100.",
            "Artist growth-opportunity scores must remain between 0 and 100.",
            "Growth opportunity must combine market strength, emerging-market evidence, breadth and geographic reach.",
        ],
        "Observed Evidence": [
            f"Section 10.4 completion status: {section_10_4_complete}",
            f"{len(track_growth_opportunity_df):,} track opportunities calculated",
            f"{len(artist_growth_opportunity_df):,} artist opportunities calculated",
            "Track growth-opportunity scores remain within valid bounds",
            "Artist growth-opportunity scores remain within valid bounds",
            "Market-growth strength, emerging-market strength, breadth, strong-market support and geographic reach combined",
        ],
        "Passed": [
            bool(section_10_4_complete),
            bool(
                track_scores_complete
                and track_classes_complete
            ),
            bool(
                artist_scores_complete
                and artist_classes_complete
            ),
            bool(track_score_bounds_valid),
            bool(artist_score_bounds_valid),
            True,
        ],
    }
)

print("\nGrowth opportunity summary validation")
print("=" * 100)
display(growth_opportunity_validation_df)


# ------------------------------------------------------------
# 17. Section completion
# ------------------------------------------------------------

section_10_5_complete = bool(
    growth_opportunity_validation_df["Passed"].all()
)

section_10_complete = section_10_5_complete

print()
print(
    f"Section 10.5 completion status: "
    f"{section_10_5_complete}"
)
print(
    f"Section 10 overall completion status: "
    f"{section_10_complete}"
)

if section_10_complete:
    print(
        "PMIP track and artist growth opportunities have been "
        "summarised using validated market-growth, emerging-market "
        "and geographic evidence."
    )
```

# 11. Market Growth Scoring and Classification

This section converts the validated growth intelligence developed in the previous stages into final PMIP growth scores, classifications and rankings.

The analysis combines market-growth strength, emerging-market evidence, geographic reach and the growth-opportunity intelligence produced in Section 10. These signals will be consolidated into interpretable scores that allow tracks and artists to be compared consistently.

The section will:

- construct final growth scores for tracks and artists;
- validate the numerical behaviour and completeness of those scores;
- assign interpretable growth classes;
- rank tracks and artists according to growth potential; and
- identify the highest-priority growth signals for PMIP decision support.


## 11.1 Growth Score Construction

### Purpose

The purpose of this subsection is to construct a final PMIP growth score for every retained track and artist.

The score combines several complementary pieces of evidence produced earlier in the notebook rather than relying on a single growth measure. The main inputs are:

- the growth-opportunity score produced in Section 10.5;
- average market-growth strength;
- the strength of the best emerging-market signal; and
- the breadth of geographic market coverage.

Each component is converted to a comparable 0–100 scale before being combined into the final PMIP growth score.

A larger score represents stronger overall evidence that a track or artist is demonstrating meaningful and commercially relevant market growth.

```python
# ============================================================
# 11.1 Growth Score Construction
# ============================================================

import numpy as np
import pandas as pd


print("Constructing PMIP final growth scores")
print("=" * 95)

section_10_complete = bool(
    globals().get(
        "section_10_complete",
        globals().get("section_10_overall_complete", True)
    )
)

print(
    f"Section 10 overall completion status: "
    f"{section_10_complete}"
)


# ============================================================
# 1. Helper functions
# ============================================================

def find_full_dataframe(required_columns):
    """
    Search currently available notebook DataFrames for those
    containing all required columns.

    The LARGEST matching DataFrame is selected so that PMIP
    uses the complete Section 10 population rather than a
    Top-20 display table.
    """

    required_columns = set(required_columns)

    # Snapshot protects against:
    # RuntimeError: dictionary changed size during iteration
    namespace_snapshot = list(globals().items())

    candidates = []

    for name, obj in namespace_snapshot:

        if not isinstance(obj, pd.DataFrame):
            continue

        if not required_columns.issubset(set(obj.columns)):
            continue

        # Exclude obvious presentation/diagnostic tables
        name_lower = name.lower()

        excluded_terms = [
            "top_",
            "distribution",
            "diagnostic",
            "validation",
            "overview"
        ]

        if any(
            term in name_lower
            for term in excluded_terms
        ):
            continue

        candidates.append(
            {
                "name": name,
                "dataframe": obj,
                "rows": len(obj),
                "columns": len(obj.columns)
            }
        )

    if not candidates:
        return None, None, pd.DataFrame()

    candidate_summary_df = pd.DataFrame(
        [
            {
                "DataFrame": item["name"],
                "Rows": item["rows"],
                "Columns": item["columns"]
            }
            for item in candidates
        ]
    ).sort_values(
        ["Rows", "Columns"],
        ascending=[False, False]
    )

    # Select the largest matching DataFrame
    selected = sorted(
        candidates,
        key=lambda item: (
            item["rows"],
            item["columns"]
        ),
        reverse=True
    )[0]

    return (
        selected["name"],
        selected["dataframe"].copy(),
        candidate_summary_df
    )


def normalise_0_100(series):
    """
    Convert a numerical series to the 0–100 range.
    """

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    minimum = values.min()
    maximum = values.max()

    if pd.isna(minimum) or pd.isna(maximum):
        return pd.Series(
            np.nan,
            index=series.index
        )

    if maximum == minimum:
        return pd.Series(
            50.0,
            index=series.index
        )

    return (
        (
            values - minimum
        )
        /
        (
            maximum - minimum
        )
        * 100
    )


def safe_existing_0_100(series):
    """
    Preserve scores already expressed on a 0–100 scale.
    Otherwise convert them to 0–100.
    """

    values = pd.to_numeric(
        series,
        errors="coerce"
    )

    valid_values = values.dropna()

    if valid_values.empty:
        return pd.Series(
            np.nan,
            index=series.index
        )

    if (
        valid_values.min() >= 0
        and valid_values.max() <= 100
    ):
        return values.clip(
            lower=0,
            upper=100
        )

    return normalise_0_100(
        values
    )


# ============================================================
# 2. Locate COMPLETE Section 10 track opportunity dataset
# ============================================================

track_required_columns = [
    "track_id",
    "countries_represented",
    "mean_market_growth_score",
    "maximum_track_emerging_market_score",
    "growth_opportunity_score"
]


(
    track_source_name,
    track_growth_score_df,
    track_candidate_summary_df
) = find_full_dataframe(
    track_required_columns
)


if track_growth_score_df is None:
    raise NameError(
        "Could not locate the complete Section 10 track "
        "growth-opportunity DataFrame.\n\n"
        "Required columns:\n- "
        + "\n- ".join(
            track_required_columns
        )
    )


# ============================================================
# 3. Locate COMPLETE Section 10 artist opportunity dataset
# ============================================================

artist_required_columns = [
    "artists",
    "countries_represented",
    "mean_market_growth_score",
    "maximum_artist_emerging_market_score",
    "growth_opportunity_score"
]


(
    artist_source_name,
    artist_growth_score_df,
    artist_candidate_summary_df
) = find_full_dataframe(
    artist_required_columns
)


if artist_growth_score_df is None:
    raise NameError(
        "Could not locate the complete Section 10 artist "
        "growth-opportunity DataFrame.\n\n"
        "Required columns:\n- "
        + "\n- ".join(
            artist_required_columns
        )
    )


# ============================================================
# 4. Report selected full sources
# ============================================================

print()

print(
    f"Track score source: "
    f"{track_source_name}"
)

print(
    f"Artist score source: "
    f"{artist_source_name}"
)

print(
    f"Tracks available: "
    f"{len(track_growth_score_df):,}"
)

print(
    f"Artists available: "
    f"{len(artist_growth_score_df):,}"
)


print()
print("Matching track source candidates")
print("=" * 95)

display(
    track_candidate_summary_df
)


print()
print("Matching artist source candidates")
print("=" * 95)

display(
    artist_candidate_summary_df
)


# ============================================================
# 5. Population protection
# ============================================================

if len(track_growth_score_df) <= 20:

    raise ValueError(
        "The selected track DataFrame contains only "
        f"{len(track_growth_score_df):,} rows. "
        "This appears to be a Top-N presentation table rather "
        "than the complete Section 10 track population."
    )


if len(artist_growth_score_df) <= 20:

    raise ValueError(
        "The selected artist DataFrame contains only "
        f"{len(artist_growth_score_df):,} rows. "
        "This appears to be a Top-N presentation table rather "
        "than the complete Section 10 artist population."
    )


# ============================================================
# 6. Prepare track scoring components
# ============================================================

track_growth_score_df[
    "opportunity_component"
] = safe_existing_0_100(
    track_growth_score_df[
        "growth_opportunity_score"
    ]
)


track_growth_score_df[
    "market_growth_component"
] = safe_existing_0_100(
    track_growth_score_df[
        "mean_market_growth_score"
    ]
)


track_growth_score_df[
    "emerging_market_component"
] = safe_existing_0_100(
    track_growth_score_df[
        "maximum_track_emerging_market_score"
    ]
)


track_growth_score_df[
    "geographic_reach_component"
] = normalise_0_100(
    track_growth_score_df[
        "countries_represented"
    ]
)


# ============================================================
# 7. Construct final track PMIP growth score
# ============================================================

track_growth_score_df[
    "pmip_growth_score"
] = (

    0.50
    * track_growth_score_df[
        "opportunity_component"
    ]

    +

    0.25
    * track_growth_score_df[
        "market_growth_component"
    ]

    +

    0.15
    * track_growth_score_df[
        "emerging_market_component"
    ]

    +

    0.10
    * track_growth_score_df[
        "geographic_reach_component"
    ]
)


track_growth_score_df[
    "pmip_growth_score"
] = (
    track_growth_score_df[
        "pmip_growth_score"
    ]
    .clip(
        lower=0,
        upper=100
    )
    .round(6)
)


# ============================================================
# 8. Prepare artist scoring components
# ============================================================

artist_growth_score_df[
    "opportunity_component"
] = safe_existing_0_100(
    artist_growth_score_df[
        "growth_opportunity_score"
    ]
)


artist_growth_score_df[
    "market_growth_component"
] = safe_existing_0_100(
    artist_growth_score_df[
        "mean_market_growth_score"
    ]
)


artist_growth_score_df[
    "emerging_market_component"
] = safe_existing_0_100(
    artist_growth_score_df[
        "maximum_artist_emerging_market_score"
    ]
)


artist_growth_score_df[
    "geographic_reach_component"
] = normalise_0_100(
    artist_growth_score_df[
        "countries_represented"
    ]
)


# ============================================================
# 9. Construct final artist PMIP growth score
# ============================================================

artist_growth_score_df[
    "pmip_growth_score"
] = (

    0.50
    * artist_growth_score_df[
        "opportunity_component"
    ]

    +

    0.25
    * artist_growth_score_df[
        "market_growth_component"
    ]

    +

    0.15
    * artist_growth_score_df[
        "emerging_market_component"
    ]

    +

    0.10
    * artist_growth_score_df[
        "geographic_reach_component"
    ]
)


artist_growth_score_df[
    "pmip_growth_score"
] = (
    artist_growth_score_df[
        "pmip_growth_score"
    ]
    .clip(
        lower=0,
        upper=100
    )
    .round(6)
)


# ============================================================
# 10. Track score summary
# ============================================================

print()
print("Track PMIP growth score summary")
print("=" * 95)

display(
    track_growth_score_df[
        "pmip_growth_score"
    ]
    .describe()
    .to_frame(
        name="Observed Value"
    )
)


# ============================================================
# 11. Artist score summary
# ============================================================

print()
print("Artist PMIP growth score summary")
print("=" * 95)

display(
    artist_growth_score_df[
        "pmip_growth_score"
    ]
    .describe()
    .to_frame(
        name="Observed Value"
    )
)


# ============================================================
# 12. Growth score construction overview
# ============================================================

growth_score_overview_df = pd.DataFrame(
    {
        "Metric": [

            "Tracks scored",

            "Median track PMIP growth score",

            "Mean track PMIP growth score",

            "Maximum track PMIP growth score",

            "Artists scored",

            "Median artist PMIP growth score",

            "Mean artist PMIP growth score",

            "Maximum artist PMIP growth score"
        ],

        "Observed Value": [

            len(
                track_growth_score_df
            ),

            track_growth_score_df[
                "pmip_growth_score"
            ].median(),

            track_growth_score_df[
                "pmip_growth_score"
            ].mean(),

            track_growth_score_df[
                "pmip_growth_score"
            ].max(),

            len(
                artist_growth_score_df
            ),

            artist_growth_score_df[
                "pmip_growth_score"
            ].median(),

            artist_growth_score_df[
                "pmip_growth_score"
            ].mean(),

            artist_growth_score_df[
                "pmip_growth_score"
            ].max()
        ]
    }
)


print()
print(
    "PMIP growth score construction overview"
)
print("=" * 95)

display(
    growth_score_overview_df
)


# ============================================================
# 13. Top PMIP track growth scores
# ============================================================

track_display_columns = [

    column

    for column in [

        "track_id",

        "countries_represented",

        "mean_market_growth_score",

        "maximum_track_emerging_market_score",

        "growth_opportunity_score",

        "pmip_growth_score"
    ]

    if column
    in track_growth_score_df.columns
]


top_track_growth_scores_df = (
    track_growth_score_df[
        track_display_columns
    ]
    .sort_values(
        "pmip_growth_score",
        ascending=False
    )
    .head(20)
)


print()
print(
    "Top PMIP track growth scores"
)
print("=" * 95)

display(
    top_track_growth_scores_df
)


# ============================================================
# 14. Top PMIP artist growth scores
# ============================================================

artist_display_columns = [

    column

    for column in [

        "artists",

        "countries_represented",

        "mean_market_growth_score",

        "maximum_artist_emerging_market_score",

        "growth_opportunity_score",

        "pmip_growth_score"
    ]

    if column
    in artist_growth_score_df.columns
]


top_artist_growth_scores_df = (
    artist_growth_score_df[
        artist_display_columns
    ]
    .sort_values(
        "pmip_growth_score",
        ascending=False
    )
    .head(20)
)


print()
print(
    "Top PMIP artist growth scores"
)
print("=" * 95)

display(
    top_artist_growth_scores_df
)


# ============================================================
# 15. Diagnostics
# ============================================================

growth_score_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [

            "Tracks assessed",

            "Track PMIP growth scores available",

            "Track scores within 0–100 bounds",

            "Artists assessed",

            "Artist PMIP growth scores available",

            "Artist scores within 0–100 bounds"
        ],

        "Observed Value": [

            len(
                track_growth_score_df
            ),

            track_growth_score_df[
                "pmip_growth_score"
            ].notna().sum(),

            track_growth_score_df[
                "pmip_growth_score"
            ]
            .between(
                0,
                100,
                inclusive="both"
            )
            .sum(),

            len(
                artist_growth_score_df
            ),

            artist_growth_score_df[
                "pmip_growth_score"
            ].notna().sum(),

            artist_growth_score_df[
                "pmip_growth_score"
            ]
            .between(
                0,
                100,
                inclusive="both"
            )
            .sum()
        ]
    }
)


print()
print(
    "Growth score construction diagnostics"
)
print("=" * 95)

display(
    growth_score_diagnostics_df
)


# ============================================================
# 16. Validation conditions
# ============================================================

track_scores_complete = (
    track_growth_score_df[
        "pmip_growth_score"
    ]
    .notna()
    .all()
)


artist_scores_complete = (
    artist_growth_score_df[
        "pmip_growth_score"
    ]
    .notna()
    .all()
)


track_scores_valid = (
    track_growth_score_df[
        "pmip_growth_score"
    ]
    .between(
        0,
        100,
        inclusive="both"
    )
    .all()
)


artist_scores_valid = (
    artist_growth_score_df[
        "pmip_growth_score"
    ]
    .between(
        0,
        100,
        inclusive="both"
    )
    .all()
)


track_population_valid = (
    len(
        track_growth_score_df
    )
    > 20
)


artist_population_valid = (
    len(
        artist_growth_score_df
    )
    > 20
)


# ============================================================
# 17. Validation table
# ============================================================

growth_score_validation_df = pd.DataFrame(
    {
        "Validation Area": [

            "Section 10 completion",

            "Track population completeness",

            "Artist population completeness",

            "Track growth score completeness",

            "Artist growth score completeness",

            "Track growth score bounds",

            "Artist growth score bounds",

            "Multi-signal score construction"
        ],

        "Requirement": [

            (
                "Artist and Track Growth Intelligence "
                "must already be complete."
            ),

            (
                "The complete retained track population "
                "must be used rather than a Top-N table."
            ),

            (
                "The complete retained artist population "
                "must be used rather than a Top-N table."
            ),

            (
                "Every retained track must receive "
                "a PMIP growth score."
            ),

            (
                "Every retained artist must receive "
                "a PMIP growth score."
            ),

            (
                "Track PMIP growth scores must remain "
                "between 0 and 100."
            ),

            (
                "Artist PMIP growth scores must remain "
                "between 0 and 100."
            ),

            (
                "The PMIP growth score must combine "
                "opportunity, market growth, emerging-market "
                "strength and geographic reach."
            )
        ],

        "Observed Evidence": [

            (
                "Section 10 overall completion status: "
                f"{section_10_complete}"
            ),

            (
                f"{len(track_growth_score_df):,} "
                "tracks included in score construction"
            ),

            (
                f"{len(artist_growth_score_df):,} "
                "artists included in score construction"
            ),

            (
                f"{track_growth_score_df['pmip_growth_score'].notna().sum():,} "
                "track growth scores calculated"
            ),

            (
                f"{artist_growth_score_df['pmip_growth_score'].notna().sum():,} "
                "artist growth scores calculated"
            ),

            (
                "Track PMIP growth scores remain "
                "within valid 0–100 bounds"
            ),

            (
                "Artist PMIP growth scores remain "
                "within valid 0–100 bounds"
            ),

            (
                "Growth opportunity, market-growth strength, "
                "emerging-market strength and geographic "
                "reach combined"
            )
        ],

        "Passed": [

            section_10_complete,

            track_population_valid,

            artist_population_valid,

            track_scores_complete,

            artist_scores_complete,

            track_scores_valid,

            artist_scores_valid,

            True
        ]
    }
)


print()
print(
    "Growth score construction validation"
)
print("=" * 95)

display(
    growth_score_validation_df
)


# ============================================================
# 18. Section completion
# ============================================================

section_11_1_complete = bool(
    growth_score_validation_df[
        "Passed"
    ].all()
)


print()

print(
    f"Section 11.1 completion status: "
    f"{section_11_1_complete}"
)


if section_11_1_complete:

    print(
        "PMIP final growth scores have been constructed "
        "for the complete retained track and artist "
        "populations using combined validated "
        "market-growth intelligence."
    )

else:

    print(
        "Section 11.1 requires review before proceeding "
        "to Growth Score Validation."
    )
```

## 11.2 Growth Score Validation

### Purpose

This subsection validates the PMIP final growth scores created in Section 11.1 before they are used for growth classification and ranking.

The validation assesses whether the track and artist growth scores are complete, remain within the required 0–100 range, preserve meaningful variation, and behave consistently with their underlying market-growth indicators.

The analysis also examines the relationship between the final PMIP growth score and its component signals — growth opportunity, market growth, emerging-market strength, and geographic reach — to ensure that the composite score represents balanced multi-signal growth intelligence rather than being driven disproportionately by a single feature.

Successful validation in this subsection will confirm that the PMIP growth scores are sufficiently reliable for subsequent growth-class assignment and market ranking.

```python
# ============================================================
# 11.2 Growth Score Validation
# ============================================================

import numpy as np
import pandas as pd


print("Validating PMIP final growth scores")
print("=" * 95)

section_11_1_complete = bool(
    globals().get(
        "section_11_1_complete",
        False
    )
)

print(
    f"Section 11.1 completion status: "
    f"{section_11_1_complete}"
)


if not section_11_1_complete:
    raise RuntimeError(
        "Section 11.1 must be complete before "
        "Growth Score Validation can proceed."
    )


# ============================================================
# 1. Confirm required DataFrames
# ============================================================

required_objects = [
    "track_growth_score_df",
    "artist_growth_score_df"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Section 11.1 objects: "
        + ", ".join(missing_objects)
    )


track_validation_df = (
    track_growth_score_df.copy()
)

artist_validation_df = (
    artist_growth_score_df.copy()
)


# ============================================================
# 2. Required component columns
# ============================================================

component_columns = [
    "opportunity_component",
    "market_growth_component",
    "emerging_market_component",
    "geographic_reach_component"
]

required_score_columns = (
    component_columns
    + ["pmip_growth_score"]
)


missing_track_columns = [
    column
    for column in required_score_columns
    if column not in track_validation_df.columns
]

missing_artist_columns = [
    column
    for column in required_score_columns
    if column not in artist_validation_df.columns
]


if missing_track_columns:
    raise KeyError(
        "Missing required track validation columns: "
        + ", ".join(missing_track_columns)
    )

if missing_artist_columns:
    raise KeyError(
        "Missing required artist validation columns: "
        + ", ".join(missing_artist_columns)
    )


# ============================================================
# 3. Completeness and bounds checks
# ============================================================

track_score_count = (
    track_validation_df[
        "pmip_growth_score"
    ]
    .notna()
    .sum()
)

artist_score_count = (
    artist_validation_df[
        "pmip_growth_score"
    ]
    .notna()
    .sum()
)


track_score_complete = (
    track_score_count
    == len(track_validation_df)
)

artist_score_complete = (
    artist_score_count
    == len(artist_validation_df)
)


track_score_bounds_valid = (
    track_validation_df[
        "pmip_growth_score"
    ]
    .between(
        0,
        100,
        inclusive="both"
    )
    .all()
)

artist_score_bounds_valid = (
    artist_validation_df[
        "pmip_growth_score"
    ]
    .between(
        0,
        100,
        inclusive="both"
    )
    .all()
)


# ============================================================
# 4. Distribution statistics
# ============================================================

track_distribution_summary_df = (
    track_validation_df[
        "pmip_growth_score"
    ]
    .describe()
    .to_frame(
        name="Observed Value"
    )
)

artist_distribution_summary_df = (
    artist_validation_df[
        "pmip_growth_score"
    ]
    .describe()
    .to_frame(
        name="Observed Value"
    )
)


print()
print("Track growth score distribution")
print("=" * 95)

display(
    track_distribution_summary_df
)


print()
print("Artist growth score distribution")
print("=" * 95)

display(
    artist_distribution_summary_df
)


# ============================================================
# 5. Distribution spread diagnostics
# ============================================================

track_std = (
    track_validation_df[
        "pmip_growth_score"
    ]
    .std()
)

artist_std = (
    artist_validation_df[
        "pmip_growth_score"
    ]
    .std()
)


track_score_range = (
    track_validation_df[
        "pmip_growth_score"
    ].max()
    -
    track_validation_df[
        "pmip_growth_score"
    ].min()
)

artist_score_range = (
    artist_validation_df[
        "pmip_growth_score"
    ].max()
    -
    artist_validation_df[
        "pmip_growth_score"
    ].min()
)


track_variation_valid = bool(
    track_std > 0
    and track_score_range > 0
)

artist_variation_valid = bool(
    artist_std > 0
    and artist_score_range > 0
)


# ============================================================
# 6. Track component correlations
# ============================================================

track_correlation_columns = (
    component_columns
    + ["pmip_growth_score"]
)

track_growth_correlation_df = (
    track_validation_df[
        track_correlation_columns
    ]
    .corr()
)


print()
print("Track PMIP growth-score correlations")
print("=" * 95)

display(
    track_growth_correlation_df
)


# ============================================================
# 7. Artist component correlations
# ============================================================

artist_growth_correlation_df = (
    artist_validation_df[
        track_correlation_columns
    ]
    .corr()
)


print()
print("Artist PMIP growth-score correlations")
print("=" * 95)

display(
    artist_growth_correlation_df
)


# ============================================================
# 8. Extract component-to-final-score correlations
# ============================================================

track_component_relationship_df = pd.DataFrame(
    {
        "Component": component_columns,

        "Correlation With PMIP Growth Score": [
            track_growth_correlation_df.loc[
                column,
                "pmip_growth_score"
            ]
            for column in component_columns
        ]
    }
).sort_values(
    "Correlation With PMIP Growth Score",
    ascending=False
)


artist_component_relationship_df = pd.DataFrame(
    {
        "Component": component_columns,

        "Correlation With PMIP Growth Score": [
            artist_growth_correlation_df.loc[
                column,
                "pmip_growth_score"
            ]
            for column in component_columns
        ]
    }
).sort_values(
    "Correlation With PMIP Growth Score",
    ascending=False
)


print()
print(
    "Track component relationship with final growth score"
)
print("=" * 95)

display(
    track_component_relationship_df
)


print()
print(
    "Artist component relationship with final growth score"
)
print("=" * 95)

display(
    artist_component_relationship_df
)


# ============================================================
# 9. High-score evidence validation
# ============================================================

track_high_threshold = (
    track_validation_df[
        "pmip_growth_score"
    ]
    .quantile(0.90)
)

artist_high_threshold = (
    artist_validation_df[
        "pmip_growth_score"
    ]
    .quantile(0.90)
)


track_high_scores_df = (
    track_validation_df[
        track_validation_df[
            "pmip_growth_score"
        ]
        >= track_high_threshold
    ]
)

artist_high_scores_df = (
    artist_validation_df[
        artist_validation_df[
            "pmip_growth_score"
        ]
        >= artist_high_threshold
    ]
)


track_high_component_means = (
    track_high_scores_df[
        component_columns
        + ["pmip_growth_score"]
    ]
    .mean()
    .to_frame(
        name="Top 10% Mean"
    )
)


track_overall_component_means = (
    track_validation_df[
        component_columns
        + ["pmip_growth_score"]
    ]
    .mean()
    .to_frame(
        name="Overall Mean"
    )
)


track_high_score_evidence_df = (
    track_overall_component_means
    .join(
        track_high_component_means
    )
)

track_high_score_evidence_df[
    "Difference"
] = (
    track_high_score_evidence_df[
        "Top 10% Mean"
    ]
    -
    track_high_score_evidence_df[
        "Overall Mean"
    ]
)


artist_high_component_means = (
    artist_high_scores_df[
        component_columns
        + ["pmip_growth_score"]
    ]
    .mean()
    .to_frame(
        name="Top 10% Mean"
    )
)


artist_overall_component_means = (
    artist_validation_df[
        component_columns
        + ["pmip_growth_score"]
    ]
    .mean()
    .to_frame(
        name="Overall Mean"
    )
)


artist_high_score_evidence_df = (
    artist_overall_component_means
    .join(
        artist_high_component_means
    )
)

artist_high_score_evidence_df[
    "Difference"
] = (
    artist_high_score_evidence_df[
        "Top 10% Mean"
    ]
    -
    artist_high_score_evidence_df[
        "Overall Mean"
    ]
)


print()
print(
    "Track high-score evidence comparison"
)
print("=" * 95)

display(
    track_high_score_evidence_df
)


print()
print(
    "Artist high-score evidence comparison"
)
print("=" * 95)

display(
    artist_high_score_evidence_df
)


# ============================================================
# 10. Validate high scores have stronger evidence
# ============================================================

track_high_signal_valid = bool(
    (
        track_high_score_evidence_df.loc[
            component_columns,
            "Difference"
        ]
        > 0
    )
    .sum()
    >= 3
)

artist_high_signal_valid = bool(
    (
        artist_high_score_evidence_df.loc[
            component_columns,
            "Difference"
        ]
        > 0
    )
    .sum()
    >= 3
)


# ============================================================
# 11. Component dominance assessment
# ============================================================

track_abs_correlations = (
    track_component_relationship_df[
        "Correlation With PMIP Growth Score"
    ]
    .abs()
)

artist_abs_correlations = (
    artist_component_relationship_df[
        "Correlation With PMIP Growth Score"
    ]
    .abs()
)


track_max_component_correlation = (
    track_abs_correlations.max()
)

artist_max_component_correlation = (
    artist_abs_correlations.max()
)


track_min_component_correlation = (
    track_abs_correlations.min()
)

artist_min_component_correlation = (
    artist_abs_correlations.min()
)


track_multi_signal_valid = bool(
    (
        track_abs_correlations
        > 0.05
    )
    .sum()
    >= 3
)

artist_multi_signal_valid = bool(
    (
        artist_abs_correlations
        > 0.05
    )
    .sum()
    >= 3
)


# ============================================================
# 12. Validation overview
# ============================================================

growth_score_validation_overview_df = pd.DataFrame(
    {
        "Metric": [

            "Track scores assessed",

            "Track score standard deviation",

            "Track score range",

            "Track 90th-percentile threshold",

            "Maximum track component correlation",

            "Minimum track component correlation",

            "Artist scores assessed",

            "Artist score standard deviation",

            "Artist score range",

            "Artist 90th-percentile threshold",

            "Maximum artist component correlation",

            "Minimum artist component correlation"
        ],

        "Observed Value": [

            len(
                track_validation_df
            ),

            track_std,

            track_score_range,

            track_high_threshold,

            track_max_component_correlation,

            track_min_component_correlation,

            len(
                artist_validation_df
            ),

            artist_std,

            artist_score_range,

            artist_high_threshold,

            artist_max_component_correlation,

            artist_min_component_correlation
        ]
    }
)


print()
print(
    "Growth score validation overview"
)
print("=" * 95)

display(
    growth_score_validation_overview_df
)


# ============================================================
# 13. Validation diagnostics
# ============================================================

growth_score_validation_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [

            "Track scores available",

            "Artist scores available",

            "Track scores within 0–100",

            "Artist scores within 0–100",

            "Track score variation detected",

            "Artist score variation detected",

            "Track high-score evidence strengthened",

            "Artist high-score evidence strengthened",

            "Track multi-signal relationship retained",

            "Artist multi-signal relationship retained"
        ],

        "Observed Value": [

            track_score_count,

            artist_score_count,

            int(
                track_validation_df[
                    "pmip_growth_score"
                ]
                .between(
                    0,
                    100,
                    inclusive="both"
                )
                .sum()
            ),

            int(
                artist_validation_df[
                    "pmip_growth_score"
                ]
                .between(
                    0,
                    100,
                    inclusive="both"
                )
                .sum()
            ),

            track_variation_valid,

            artist_variation_valid,

            track_high_signal_valid,

            artist_high_signal_valid,

            track_multi_signal_valid,

            artist_multi_signal_valid
        ]
    }
)


print()
print(
    "Growth score validation diagnostics"
)
print("=" * 95)

display(
    growth_score_validation_diagnostics_df
)


# ============================================================
# 14. Formal validation table
# ============================================================

growth_score_validation_checks_df = pd.DataFrame(
    {
        "Validation Area": [

            "Section 11.1 completion",

            "Track score completeness",

            "Artist score completeness",

            "Track score bounds",

            "Artist score bounds",

            "Track score variation",

            "Artist score variation",

            "Track high-score evidence",

            "Artist high-score evidence",

            "Track multi-signal behaviour",

            "Artist multi-signal behaviour"
        ],

        "Requirement": [

            (
                "Growth Score Construction must "
                "already be complete."
            ),

            (
                "Every retained track must have "
                "a PMIP growth score."
            ),

            (
                "Every retained artist must have "
                "a PMIP growth score."
            ),

            (
                "All track PMIP growth scores must "
                "remain between 0 and 100."
            ),

            (
                "All artist PMIP growth scores must "
                "remain between 0 and 100."
            ),

            (
                "Track growth scores must retain "
                "meaningful numerical variation."
            ),

            (
                "Artist growth scores must retain "
                "meaningful numerical variation."
            ),

            (
                "High-scoring tracks should show "
                "stronger underlying growth evidence."
            ),

            (
                "High-scoring artists should show "
                "stronger underlying growth evidence."
            ),

            (
                "Track growth scores must retain "
                "relationships with multiple input signals."
            ),

            (
                "Artist growth scores must retain "
                "relationships with multiple input signals."
            )
        ],

        "Observed Evidence": [

            (
                "Section 11.1 completion status: "
                f"{section_11_1_complete}"
            ),

            (
                f"{track_score_count:,} of "
                f"{len(track_validation_df):,} "
                "track scores available"
            ),

            (
                f"{artist_score_count:,} of "
                f"{len(artist_validation_df):,} "
                "artist scores available"
            ),

            (
                "Track PMIP growth scores remain "
                "within valid 0–100 bounds"
            ),

            (
                "Artist PMIP growth scores remain "
                "within valid 0–100 bounds"
            ),

            (
                f"Track score standard deviation = "
                f"{track_std:.6f}, "
                f"range = {track_score_range:.6f}"
            ),

            (
                f"Artist score standard deviation = "
                f"{artist_std:.6f}, "
                f"range = {artist_score_range:.6f}"
            ),

            (
                "Top-decile track scores show stronger "
                "component evidence across multiple signals"
            ),

            (
                "Top-decile artist scores show stronger "
                "component evidence across multiple signals"
            ),

            (
                "Track PMIP growth score remains related "
                "to multiple component signals"
            ),

            (
                "Artist PMIP growth score remains related "
                "to multiple component signals"
            )
        ],

        "Passed": [

            section_11_1_complete,

            track_score_complete,

            artist_score_complete,

            track_score_bounds_valid,

            artist_score_bounds_valid,

            track_variation_valid,

            artist_variation_valid,

            track_high_signal_valid,

            artist_high_signal_valid,

            track_multi_signal_valid,

            artist_multi_signal_valid
        ]
    }
)


print()
print(
    "Growth score validation"
)
print("=" * 95)

display(
    growth_score_validation_checks_df
)


# ============================================================
# 15. Section completion
# ============================================================

section_11_2_complete = bool(
    growth_score_validation_checks_df[
        "Passed"
    ]
    .all()
)


print()

print(
    f"Section 11.2 completion status: "
    f"{section_11_2_complete}"
)


if section_11_2_complete:

    print(
        "PMIP final growth scores have been validated "
        "for completeness, bounds, variation and "
        "multi-signal behaviour."
    )

else:

    print(
        "Growth score validation identified one or more "
        "conditions requiring review before growth-class "
        "construction."
    )
```

## 11.3 Growth Classes

### Purpose

This subsection converts the validated PMIP growth scores into interpretable growth classes for tracks and artists.

While the numerical PMIP growth score provides detailed ranking information, categorical growth classes make the results easier to interpret and communicate. The classification separates records into distinct levels of growth strength, ranging from weak or limited growth through to exceptional growth opportunities.

The class boundaries are designed around the observed PMIP score distributions while preserving meaningful differences between lower-growth, developing, moderate-growth, high-growth, and exceptional-growth cases.

This classification will provide the foundation for the market-ranking and high-priority growth-signal analysis performed in the following subsections.

```python
# ============================================================
# 11.3 Growth Classes
# ============================================================

import numpy as np
import pandas as pd


print("Classifying PMIP growth scores")
print("=" * 95)

section_11_2_complete = bool(
    globals().get(
        "section_11_2_complete",
        False
    )
)

print(
    f"Section 11.2 completion status: "
    f"{section_11_2_complete}"
)


if not section_11_2_complete:
    raise RuntimeError(
        "Section 11.2 must be complete before "
        "Growth Classes can be created."
    )


# ============================================================
# 1. Confirm required score DataFrames
# ============================================================

required_objects = [
    "track_growth_score_df",
    "artist_growth_score_df"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Section 11 objects: "
        + ", ".join(missing_objects)
    )


track_growth_class_df = (
    track_growth_score_df.copy()
)

artist_growth_class_df = (
    artist_growth_score_df.copy()
)


# ============================================================
# 2. Confirm score column
# ============================================================

score_column = "pmip_growth_score"

if score_column not in track_growth_class_df.columns:
    raise KeyError(
        f"Track score column '{score_column}' is missing."
    )

if score_column not in artist_growth_class_df.columns:
    raise KeyError(
        f"Artist score column '{score_column}' is missing."
    )


# ============================================================
# 3. Define PMIP growth classes
# ============================================================
#
# Thresholds:
#
# < 10      = Low Growth
# 10–<25    = Developing Growth
# 25–<45    = Moderate Growth
# 45–<65    = High Growth
# >= 65     = Exceptional Growth
#
# These thresholds preserve meaningful separation across
# the observed PMIP score distributions.
# ============================================================

def classify_pmip_growth(score):

    if pd.isna(score):
        return "Unclassified"

    if score < 10:
        return "Low Growth"

    elif score < 25:
        return "Developing Growth"

    elif score < 45:
        return "Moderate Growth"

    elif score < 65:
        return "High Growth"

    else:
        return "Exceptional Growth"


track_growth_class_df[
    "pmip_growth_class"
] = (
    track_growth_class_df[
        score_column
    ]
    .apply(
        classify_pmip_growth
    )
)


artist_growth_class_df[
    "pmip_growth_class"
] = (
    artist_growth_class_df[
        score_column
    ]
    .apply(
        classify_pmip_growth
    )
)


# ============================================================
# 4. Ordered growth classes
# ============================================================

growth_class_order = [
    "Low Growth",
    "Developing Growth",
    "Moderate Growth",
    "High Growth",
    "Exceptional Growth"
]


track_growth_class_df[
    "pmip_growth_class"
] = pd.Categorical(
    track_growth_class_df[
        "pmip_growth_class"
    ],
    categories=growth_class_order,
    ordered=True
)


artist_growth_class_df[
    "pmip_growth_class"
] = pd.Categorical(
    artist_growth_class_df[
        "pmip_growth_class"
    ],
    categories=growth_class_order,
    ordered=True
)


# ============================================================
# 5. Track growth-class distribution
# ============================================================

track_growth_class_distribution_df = (
    track_growth_class_df[
        "pmip_growth_class"
    ]
    .value_counts(
        sort=False
    )
    .rename(
        "Tracks"
    )
    .reset_index()
    .rename(
        columns={
            "index": "PMIP Growth Class"
        }
    )
)


track_growth_class_distribution_df[
    "Share (%)"
] = (
    track_growth_class_distribution_df[
        "Tracks"
    ]
    / len(track_growth_class_df)
    * 100
)


print()
print("Track PMIP growth-class distribution")
print("=" * 95)

display(
    track_growth_class_distribution_df
)


# ============================================================
# 6. Artist growth-class distribution
# ============================================================

artist_growth_class_distribution_df = (
    artist_growth_class_df[
        "pmip_growth_class"
    ]
    .value_counts(
        sort=False
    )
    .rename(
        "Artists"
    )
    .reset_index()
    .rename(
        columns={
            "index": "PMIP Growth Class"
        }
    )
)


artist_growth_class_distribution_df[
    "Share (%)"
] = (
    artist_growth_class_distribution_df[
        "Artists"
    ]
    / len(artist_growth_class_df)
    * 100
)


print()
print("Artist PMIP growth-class distribution")
print("=" * 95)

display(
    artist_growth_class_distribution_df
)


# ============================================================
# 7. Growth-class score summaries
# ============================================================

track_growth_class_summary_df = (
    track_growth_class_df
    .groupby(
        "pmip_growth_class",
        observed=False
    )[
        score_column
    ]
    .agg(
        Count="count",
        Mean="mean",
        Median="median",
        Minimum="min",
        Maximum="max"
    )
    .reset_index()
)


artist_growth_class_summary_df = (
    artist_growth_class_df
    .groupby(
        "pmip_growth_class",
        observed=False
    )[
        score_column
    ]
    .agg(
        Count="count",
        Mean="mean",
        Median="median",
        Minimum="min",
        Maximum="max"
    )
    .reset_index()
)


print()
print("Track PMIP growth-class score summary")
print("=" * 95)

display(
    track_growth_class_summary_df
)


print()
print("Artist PMIP growth-class score summary")
print("=" * 95)

display(
    artist_growth_class_summary_df
)


# ============================================================
# 8. High-or-better growth populations
# ============================================================

track_high_growth_mask = (
    track_growth_class_df[
        "pmip_growth_class"
    ]
    .isin(
        [
            "High Growth",
            "Exceptional Growth"
        ]
    )
)


artist_high_growth_mask = (
    artist_growth_class_df[
        "pmip_growth_class"
    ]
    .isin(
        [
            "High Growth",
            "Exceptional Growth"
        ]
    )
)


track_high_growth_count = int(
    track_high_growth_mask.sum()
)

artist_high_growth_count = int(
    artist_high_growth_mask.sum()
)


track_exceptional_growth_count = int(
    (
        track_growth_class_df[
            "pmip_growth_class"
        ]
        == "Exceptional Growth"
    )
    .sum()
)


artist_exceptional_growth_count = int(
    (
        artist_growth_class_df[
            "pmip_growth_class"
        ]
        == "Exceptional Growth"
    )
    .sum()
)


# ============================================================
# 9. Overview
# ============================================================

growth_class_overview_df = pd.DataFrame(
    {
        "Metric": [

            "Tracks classified",

            "Track growth classes observed",

            "Tracks with high-or-better growth",

            "Tracks with exceptional growth",

            "Artists classified",

            "Artist growth classes observed",

            "Artists with high-or-better growth",

            "Artists with exceptional growth"
        ],

        "Observed Value": [

            len(
                track_growth_class_df
            ),

            track_growth_class_df[
                "pmip_growth_class"
            ]
            .nunique(),

            track_high_growth_count,

            track_exceptional_growth_count,

            len(
                artist_growth_class_df
            ),

            artist_growth_class_df[
                "pmip_growth_class"
            ]
            .nunique(),

            artist_high_growth_count,

            artist_exceptional_growth_count
        ]
    }
)


print()
print("PMIP growth-class overview")
print("=" * 95)

display(
    growth_class_overview_df
)


# ============================================================
# 10. Diagnostics
# ============================================================

growth_class_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [

            "Tracks assessed",

            "Track growth classes assigned",

            "Track growth classes observed",

            "Artists assessed",

            "Artist growth classes assigned",

            "Artist growth classes observed"
        ],

        "Observed Value": [

            len(
                track_growth_class_df
            ),

            track_growth_class_df[
                "pmip_growth_class"
            ]
            .notna()
            .sum(),

            track_growth_class_df[
                "pmip_growth_class"
            ]
            .nunique(),

            len(
                artist_growth_class_df
            ),

            artist_growth_class_df[
                "pmip_growth_class"
            ]
            .notna()
            .sum(),

            artist_growth_class_df[
                "pmip_growth_class"
            ]
            .nunique()
        ]
    }
)


print()
print("Growth-class diagnostics")
print("=" * 95)

display(
    growth_class_diagnostics_df
)


# ============================================================
# 11. Validation checks
# ============================================================

track_class_complete = bool(
    track_growth_class_df[
        "pmip_growth_class"
    ]
    .notna()
    .all()
)

artist_class_complete = bool(
    artist_growth_class_df[
        "pmip_growth_class"
    ]
    .notna()
    .all()
)


track_classes_valid = bool(
    set(
        track_growth_class_df[
            "pmip_growth_class"
        ]
        .dropna()
        .astype(str)
        .unique()
    )
    .issubset(
        set(
            growth_class_order
        )
    )
)


artist_classes_valid = bool(
    set(
        artist_growth_class_df[
            "pmip_growth_class"
        ]
        .dropna()
        .astype(str)
        .unique()
    )
    .issubset(
        set(
            growth_class_order
        )
    )
)


track_multiple_classes = bool(
    track_growth_class_df[
        "pmip_growth_class"
    ]
    .nunique()
    >= 3
)


artist_multiple_classes = bool(
    artist_growth_class_df[
        "pmip_growth_class"
    ]
    .nunique()
    >= 3
)


growth_class_validation_df = pd.DataFrame(
    {
        "Validation Area": [

            "Section 11.2 completion",

            "Track classification completeness",

            "Artist classification completeness",

            "Track class validity",

            "Artist class validity",

            "Track class differentiation",

            "Artist class differentiation"
        ],

        "Requirement": [

            (
                "Growth Score Validation must "
                "already be complete."
            ),

            (
                "Every retained track must receive "
                "a PMIP growth class."
            ),

            (
                "Every retained artist must receive "
                "a PMIP growth class."
            ),

            (
                "All track growth classes must belong "
                "to the approved PMIP growth taxonomy."
            ),

            (
                "All artist growth classes must belong "
                "to the approved PMIP growth taxonomy."
            ),

            (
                "Track scores must produce multiple "
                "meaningful growth classes."
            ),

            (
                "Artist scores must produce multiple "
                "meaningful growth classes."
            )
        ],

        "Observed Evidence": [

            (
                "Section 11.2 completion status: "
                f"{section_11_2_complete}"
            ),

            (
                f"{len(track_growth_class_df):,} "
                "track growth classes assigned"
            ),

            (
                f"{len(artist_growth_class_df):,} "
                "artist growth classes assigned"
            ),

            (
                "Track classes conform to the "
                "five-level PMIP growth taxonomy"
            ),

            (
                "Artist classes conform to the "
                "five-level PMIP growth taxonomy"
            ),

            (
                f"{track_growth_class_df['pmip_growth_class'].nunique()} "
                "track growth classes observed"
            ),

            (
                f"{artist_growth_class_df['pmip_growth_class'].nunique()} "
                "artist growth classes observed"
            )
        ],

        "Passed": [

            section_11_2_complete,

            track_class_complete,

            artist_class_complete,

            track_classes_valid,

            artist_classes_valid,

            track_multiple_classes,

            artist_multiple_classes
        ]
    }
)


print()
print("Growth-class validation")
print("=" * 95)

display(
    growth_class_validation_df
)


# ============================================================
# 12. Section completion
# ============================================================

section_11_3_complete = bool(
    growth_class_validation_df[
        "Passed"
    ]
    .all()
)


print()

print(
    f"Section 11.3 completion status: "
    f"{section_11_3_complete}"
)


if section_11_3_complete:

    print(
        "PMIP growth classes have been assigned "
        "successfully to all retained tracks and artists."
    )

else:

    print(
        "Growth-class construction identified one or more "
        "conditions requiring review."
    )
```

## 11.4 Market Ranking

### Purpose

This subsection ranks all retained PMIP tracks and artists according to their validated final growth scores.

The ranking converts the PMIP growth score into an ordered intelligence output, allowing stronger growth opportunities to be prioritised over weaker ones. Each track and artist receives an overall PMIP growth rank, while its previously assigned growth class is retained for interpretation.

The analysis will also identify the highest-ranked track and artist opportunities, compare ranking distributions, and validate that every retained record receives a unique and consistent ranking position.

```python
# ============================================================
# 11.4 Market Ranking
# ============================================================

import pandas as pd
import numpy as np


print("Ranking PMIP growth opportunities")
print("=" * 95)

section_11_3_complete = bool(
    globals().get(
        "section_11_3_complete",
        False
    )
)

print(
    f"Section 11.3 completion status: "
    f"{section_11_3_complete}"
)


if not section_11_3_complete:
    raise RuntimeError(
        "Section 11.3 must be complete before "
        "Market Ranking can be performed."
    )


# ============================================================
# 1. Confirm required DataFrames
# ============================================================

required_objects = [
    "track_growth_class_df",
    "artist_growth_class_df"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Section 11 objects: "
        + ", ".join(missing_objects)
    )


track_market_ranking_df = (
    track_growth_class_df.copy()
)

artist_market_ranking_df = (
    artist_growth_class_df.copy()
)


# ============================================================
# 2. Confirm required score/class columns
# ============================================================

required_columns = [
    "pmip_growth_score",
    "pmip_growth_class"
]


for column in required_columns:

    if column not in track_market_ranking_df.columns:
        raise KeyError(
            f"Track ranking source is missing '{column}'."
        )

    if column not in artist_market_ranking_df.columns:
        raise KeyError(
            f"Artist ranking source is missing '{column}'."
        )


# ============================================================
# 3. Assign overall PMIP growth ranks
# ============================================================
#
# Higher PMIP growth score = stronger growth opportunity.
# Dense ranking keeps equal scores on the same rank.
# ============================================================

track_market_ranking_df[
    "pmip_growth_rank"
] = (
    track_market_ranking_df[
        "pmip_growth_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


artist_market_ranking_df[
    "pmip_growth_rank"
] = (
    artist_market_ranking_df[
        "pmip_growth_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ============================================================
# 4. Percentile ranking
# ============================================================

track_market_ranking_df[
    "pmip_growth_percentile"
] = (
    track_market_ranking_df[
        "pmip_growth_score"
    ]
    .rank(
        method="average",
        pct=True
    )
    * 100
)


artist_market_ranking_df[
    "pmip_growth_percentile"
] = (
    artist_market_ranking_df[
        "pmip_growth_score"
    ]
    .rank(
        method="average",
        pct=True
    )
    * 100
)


# ============================================================
# 5. Ranking tier
# ============================================================

def ranking_tier(percentile):

    if pd.isna(percentile):
        return "Unclassified"

    if percentile >= 99:
        return "Top 1%"

    elif percentile >= 95:
        return "Top 5%"

    elif percentile >= 90:
        return "Top 10%"

    elif percentile >= 75:
        return "Top 25%"

    elif percentile >= 50:
        return "Upper Half"

    else:
        return "Lower Half"


track_market_ranking_df[
    "pmip_ranking_tier"
] = (
    track_market_ranking_df[
        "pmip_growth_percentile"
    ]
    .apply(
        ranking_tier
    )
)


artist_market_ranking_df[
    "pmip_ranking_tier"
] = (
    artist_market_ranking_df[
        "pmip_growth_percentile"
    ]
    .apply(
        ranking_tier
    )
)


# ============================================================
# 6. Sort by final growth rank
# ============================================================

track_market_ranking_df = (
    track_market_ranking_df
    .sort_values(
        by=[
            "pmip_growth_rank",
            "pmip_growth_score"
        ],
        ascending=[
            True,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)


artist_market_ranking_df = (
    artist_market_ranking_df
    .sort_values(
        by=[
            "pmip_growth_rank",
            "pmip_growth_score"
        ],
        ascending=[
            True,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 7. Top ranked tracks
# ============================================================

track_display_columns = [
    column
    for column in [
        "track_id",
        "countries_represented",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        "growth_opportunity_score",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "pmip_ranking_tier"
    ]
    if column in track_market_ranking_df.columns
]


top_track_market_ranking_df = (
    track_market_ranking_df[
        track_display_columns
    ]
    .head(20)
)


print()
print("Top PMIP track market rankings")
print("=" * 95)

display(
    top_track_market_ranking_df
)


# ============================================================
# 8. Top ranked artists
# ============================================================

artist_display_columns = [
    column
    for column in [
        "artists",
        "countries_represented",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
        "growth_opportunity_score",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "pmip_ranking_tier"
    ]
    if column in artist_market_ranking_df.columns
]


top_artist_market_ranking_df = (
    artist_market_ranking_df[
        artist_display_columns
    ]
    .head(20)
)


print()
print("Top PMIP artist market rankings")
print("=" * 95)

display(
    top_artist_market_ranking_df
)


# ============================================================
# 9. Track ranking-tier distribution
# ============================================================

track_ranking_tier_distribution_df = (
    track_market_ranking_df[
        "pmip_ranking_tier"
    ]
    .value_counts()
    .rename(
        "Tracks"
    )
    .reset_index()
    .rename(
        columns={
            "index": "PMIP Ranking Tier"
        }
    )
)


track_ranking_tier_distribution_df[
    "Share (%)"
] = (
    track_ranking_tier_distribution_df[
        "Tracks"
    ]
    / len(
        track_market_ranking_df
    )
    * 100
)


print()
print("Track PMIP ranking-tier distribution")
print("=" * 95)

display(
    track_ranking_tier_distribution_df
)


# ============================================================
# 10. Artist ranking-tier distribution
# ============================================================

artist_ranking_tier_distribution_df = (
    artist_market_ranking_df[
        "pmip_ranking_tier"
    ]
    .value_counts()
    .rename(
        "Artists"
    )
    .reset_index()
    .rename(
        columns={
            "index": "PMIP Ranking Tier"
        }
    )
)


artist_ranking_tier_distribution_df[
    "Share (%)"
] = (
    artist_ranking_tier_distribution_df[
        "Artists"
    ]
    / len(
        artist_market_ranking_df
    )
    * 100
)


print()
print("Artist PMIP ranking-tier distribution")
print("=" * 95)

display(
    artist_ranking_tier_distribution_df
)


# ============================================================
# 11. Ranking overview
# ============================================================

ranking_overview_df = pd.DataFrame(
    {
        "Metric": [

            "Tracks ranked",

            "Highest track PMIP growth score",

            "Track Top 1% population",

            "Track Top 10% population",

            "Artists ranked",

            "Highest artist PMIP growth score",

            "Artist Top 1% population",

            "Artist Top 10% population"
        ],

        "Observed Value": [

            len(
                track_market_ranking_df
            ),

            track_market_ranking_df[
                "pmip_growth_score"
            ]
            .max(),

            (
                track_market_ranking_df[
                    "pmip_ranking_tier"
                ]
                == "Top 1%"
            )
            .sum(),

            (
                track_market_ranking_df[
                    "pmip_growth_percentile"
                ]
                >= 90
            )
            .sum(),

            len(
                artist_market_ranking_df
            ),

            artist_market_ranking_df[
                "pmip_growth_score"
            ]
            .max(),

            (
                artist_market_ranking_df[
                    "pmip_ranking_tier"
                ]
                == "Top 1%"
            )
            .sum(),

            (
                artist_market_ranking_df[
                    "pmip_growth_percentile"
                ]
                >= 90
            )
            .sum()
        ]
    }
)


print()
print("PMIP market ranking overview")
print("=" * 95)

display(
    ranking_overview_df
)


# ============================================================
# 12. Diagnostics
# ============================================================

ranking_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [

            "Tracks assessed",

            "Track ranks assigned",

            "Track percentile values assigned",

            "Artists assessed",

            "Artist ranks assigned",

            "Artist percentile values assigned"
        ],

        "Observed Value": [

            len(
                track_market_ranking_df
            ),

            track_market_ranking_df[
                "pmip_growth_rank"
            ]
            .notna()
            .sum(),

            track_market_ranking_df[
                "pmip_growth_percentile"
            ]
            .notna()
            .sum(),

            len(
                artist_market_ranking_df
            ),

            artist_market_ranking_df[
                "pmip_growth_rank"
            ]
            .notna()
            .sum(),

            artist_market_ranking_df[
                "pmip_growth_percentile"
            ]
            .notna()
            .sum()
        ]
    }
)


print()
print("Market ranking diagnostics")
print("=" * 95)

display(
    ranking_diagnostics_df
)


# ============================================================
# 13. Validation
# ============================================================

track_ranking_complete = bool(
    track_market_ranking_df[
        "pmip_growth_rank"
    ]
    .notna()
    .all()
)

artist_ranking_complete = bool(
    artist_market_ranking_df[
        "pmip_growth_rank"
    ]
    .notna()
    .all()
)


track_percentile_valid = bool(
    track_market_ranking_df[
        "pmip_growth_percentile"
    ]
    .between(
        0,
        100
    )
    .all()
)


artist_percentile_valid = bool(
    artist_market_ranking_df[
        "pmip_growth_percentile"
    ]
    .between(
        0,
        100
    )
    .all()
)


track_rank_direction_valid = bool(
    track_market_ranking_df[
        "pmip_growth_score"
    ]
    .is_monotonic_decreasing
)


artist_rank_direction_valid = bool(
    artist_market_ranking_df[
        "pmip_growth_score"
    ]
    .is_monotonic_decreasing
)


ranking_validation_df = pd.DataFrame(
    {
        "Validation Area": [

            "Section 11.3 completion",

            "Track ranking completeness",

            "Artist ranking completeness",

            "Track percentile bounds",

            "Artist percentile bounds",

            "Track ranking direction",

            "Artist ranking direction"
        ],

        "Requirement": [

            (
                "Growth Classes must already "
                "be complete."
            ),

            (
                "Every retained track must receive "
                "a PMIP growth rank."
            ),

            (
                "Every retained artist must receive "
                "a PMIP growth rank."
            ),

            (
                "All track growth percentiles must "
                "remain between 0 and 100."
            ),

            (
                "All artist growth percentiles must "
                "remain between 0 and 100."
            ),

            (
                "Higher track PMIP growth scores must "
                "receive stronger ranking positions."
            ),

            (
                "Higher artist PMIP growth scores must "
                "receive stronger ranking positions."
            )
        ],

        "Observed Evidence": [

            (
                "Section 11.3 completion status: "
                f"{section_11_3_complete}"
            ),

            (
                f"{len(track_market_ranking_df):,} "
                "track ranks assigned"
            ),

            (
                f"{len(artist_market_ranking_df):,} "
                "artist ranks assigned"
            ),

            (
                "Track growth percentiles remain "
                "within valid 0–100 bounds"
            ),

            (
                "Artist growth percentiles remain "
                "within valid 0–100 bounds"
            ),

            (
                "Track ranking decreases consistently "
                "with PMIP growth score"
            ),

            (
                "Artist ranking decreases consistently "
                "with PMIP growth score"
            )
        ],

        "Passed": [

            section_11_3_complete,

            track_ranking_complete,

            artist_ranking_complete,

            track_percentile_valid,

            artist_percentile_valid,

            track_rank_direction_valid,

            artist_rank_direction_valid
        ]
    }
)


print()
print("Market ranking validation")
print("=" * 95)

display(
    ranking_validation_df
)


# ============================================================
# 14. Section completion
# ============================================================

section_11_4_complete = bool(
    ranking_validation_df[
        "Passed"
    ]
    .all()
)


print()

print(
    f"Section 11.4 completion status: "
    f"{section_11_4_complete}"
)


if section_11_4_complete:

    print(
        "PMIP tracks and artists have been ranked "
        "successfully according to validated growth strength."
    )

else:

    print(
        "Market ranking identified one or more "
        "conditions requiring review."
    )
```

###  Interpretation

The validated PMIP growth scores were successfully converted into comparative rankings for the complete retained track and artist populations.

All **51,593 tracks** and **27,652 artists** received a PMIP growth rank, percentile position, and ranking tier. Higher PMIP growth scores consistently produced stronger ranking positions, confirming that the ranking framework preserves the ordering established during growth-score construction.

The highest-ranked track achieved a PMIP growth score of approximately **80.16**, while the highest-ranked artist achieved approximately **80.58**. These represent the strongest growth opportunities currently identified within their respective PMIP populations.

For tracks, approximately **516** records fall within the **Top 1%**, while approximately **5,160** fall within the overall **Top 10%** of the ranked population. For artists, approximately **277** fall within the **Top 1%**, while approximately **2,766** fall within the overall **Top 10%**.

The ranking tiers provide an additional interpretation layer beyond the five growth classes. A record may therefore be described both by its absolute growth class, such as *Exceptional Growth*, and by its relative position within the complete PMIP population, such as *Top 1%* or *Top 5%*. This makes it easier to distinguish particularly strong opportunities even when multiple records belong to the same growth class.

The highest-ranked records are dominated by the **Exceptional Growth** class, which is consistent with the intended PMIP scoring structure. However, ranking is determined directly from the continuous PMIP growth score rather than from the class label alone, allowing finer differentiation between records within the same class.

All validation checks passed. Ranking coverage is complete, percentile values remain within valid bounds, and higher PMIP growth scores correspond correctly with stronger ranking positions.

Section 11.4 is therefore complete. These rankings can now be combined with growth class, emerging-market evidence, market-growth strength, geographic reach and opportunity measures in Section 11.5 to identify the highest-priority PMIP growth signals.

## 11.5 High-Priority Growth Signals

### Purpose

This subsection identifies the highest-priority PMIP growth opportunities by combining the final growth score and ranking with the underlying market-intelligence signals developed throughout the analysis.

Rather than relying on ranking alone, the high-priority framework considers multiple forms of evidence including growth class, PMIP percentile position, market-growth strength, emerging-market potential, growth-opportunity strength and geographic reach.

The objective is to separate ordinary high-scoring records from opportunities supported by several independent growth signals. Tracks and artists will therefore be assigned priority scores, priority classes and final PMIP priority ranks.

This provides the final decision-oriented output of the Market Growth Scoring and Classification stage.

```python
# ============================================================
# 11.5 High-Priority Growth Signals
# ============================================================

import pandas as pd
import numpy as np


print("Identifying PMIP high-priority growth signals")
print("=" * 100)

section_11_4_complete = bool(
    globals().get(
        "section_11_4_complete",
        False
    )
)

print(
    f"Section 11.4 completion status: "
    f"{section_11_4_complete}"
)


if not section_11_4_complete:
    raise RuntimeError(
        "Section 11.4 must be complete before "
        "High-Priority Growth Signals can be identified."
    )


# ============================================================
# 1. Confirm required ranking DataFrames
# ============================================================

required_objects = [
    "track_market_ranking_df",
    "artist_market_ranking_df"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required Section 11 ranking objects: "
        + ", ".join(missing_objects)
    )


track_priority_df = (
    track_market_ranking_df.copy()
)

artist_priority_df = (
    artist_market_ranking_df.copy()
)


# ============================================================
# 2. Helper function for safe numeric columns
# ============================================================

def safe_numeric_series(df, column, default=0.0):

    if column in df.columns:

        return (
            pd.to_numeric(
                df[column],
                errors="coerce"
            )
            .fillna(default)
        )

    return pd.Series(
        default,
        index=df.index,
        dtype=float
    )


# ============================================================
# 3. Prepare track priority components
# ============================================================

track_priority_df[
    "priority_growth_score_component"
] = (
    safe_numeric_series(
        track_priority_df,
        "pmip_growth_score"
    )
    .clip(
        0,
        100
    )
)


track_priority_df[
    "priority_percentile_component"
] = (
    safe_numeric_series(
        track_priority_df,
        "pmip_growth_percentile"
    )
    .clip(
        0,
        100
    )
)


track_priority_df[
    "priority_opportunity_component"
] = (
    safe_numeric_series(
        track_priority_df,
        "growth_opportunity_score"
    )
    .clip(
        0,
        100
    )
)


track_priority_df[
    "priority_market_growth_component"
] = (
    safe_numeric_series(
        track_priority_df,
        "mean_market_growth_score"
    )
    .clip(
        0,
        100
    )
)


track_priority_df[
    "priority_emerging_market_component"
] = (
    safe_numeric_series(
        track_priority_df,
        "maximum_track_emerging_market_score"
    )
    .clip(
        0,
        100
    )
)


# Geographic breadth is useful but should not overpower
# genuine growth strength.

if "countries_represented" in track_priority_df.columns:

    max_track_countries = max(
        float(
            safe_numeric_series(
                track_priority_df,
                "countries_represented"
            ).max()
        ),
        1.0
    )

    track_priority_df[
        "priority_reach_component"
    ] = (
        safe_numeric_series(
            track_priority_df,
            "countries_represented"
        )
        / max_track_countries
        * 100
    ).clip(
        0,
        100
    )

else:

    track_priority_df[
        "priority_reach_component"
    ] = 0.0


# ============================================================
# 4. Prepare artist priority components
# ============================================================

artist_priority_df[
    "priority_growth_score_component"
] = (
    safe_numeric_series(
        artist_priority_df,
        "pmip_growth_score"
    )
    .clip(
        0,
        100
    )
)


artist_priority_df[
    "priority_percentile_component"
] = (
    safe_numeric_series(
        artist_priority_df,
        "pmip_growth_percentile"
    )
    .clip(
        0,
        100
    )
)


artist_priority_df[
    "priority_opportunity_component"
] = (
    safe_numeric_series(
        artist_priority_df,
        "growth_opportunity_score"
    )
    .clip(
        0,
        100
    )
)


artist_priority_df[
    "priority_market_growth_component"
] = (
    safe_numeric_series(
        artist_priority_df,
        "mean_market_growth_score"
    )
    .clip(
        0,
        100
    )
)


artist_priority_df[
    "priority_emerging_market_component"
] = (
    safe_numeric_series(
        artist_priority_df,
        "maximum_artist_emerging_market_score"
    )
    .clip(
        0,
        100
    )
)


if "countries_represented" in artist_priority_df.columns:

    max_artist_countries = max(
        float(
            safe_numeric_series(
                artist_priority_df,
                "countries_represented"
            ).max()
        ),
        1.0
    )

    artist_priority_df[
        "priority_reach_component"
    ] = (
        safe_numeric_series(
            artist_priority_df,
            "countries_represented"
        )
        / max_artist_countries
        * 100
    ).clip(
        0,
        100
    )

else:

    artist_priority_df[
        "priority_reach_component"
    ] = 0.0


# ============================================================
# 5. Construct PMIP priority score
# ============================================================
#
# Weighting:
#
# Final PMIP growth score     30%
# Relative percentile        20%
# Growth opportunity         20%
# Market growth strength     15%
# Emerging-market strength   10%
# Geographic reach            5%
#
# The final growth score remains the largest component,
# while the additional signals provide supporting evidence.
# ============================================================

priority_weights = {
    "growth_score": 0.30,
    "percentile": 0.20,
    "opportunity": 0.20,
    "market_growth": 0.15,
    "emerging_market": 0.10,
    "reach": 0.05
}


track_priority_df[
    "pmip_priority_score"
] = (

    track_priority_df[
        "priority_growth_score_component"
    ]
    * priority_weights[
        "growth_score"
    ]

    +

    track_priority_df[
        "priority_percentile_component"
    ]
    * priority_weights[
        "percentile"
    ]

    +

    track_priority_df[
        "priority_opportunity_component"
    ]
    * priority_weights[
        "opportunity"
    ]

    +

    track_priority_df[
        "priority_market_growth_component"
    ]
    * priority_weights[
        "market_growth"
    ]

    +

    track_priority_df[
        "priority_emerging_market_component"
    ]
    * priority_weights[
        "emerging_market"
    ]

    +

    track_priority_df[
        "priority_reach_component"
    ]
    * priority_weights[
        "reach"
    ]
).clip(
    0,
    100
)


artist_priority_df[
    "pmip_priority_score"
] = (

    artist_priority_df[
        "priority_growth_score_component"
    ]
    * priority_weights[
        "growth_score"
    ]

    +

    artist_priority_df[
        "priority_percentile_component"
    ]
    * priority_weights[
        "percentile"
    ]

    +

    artist_priority_df[
        "priority_opportunity_component"
    ]
    * priority_weights[
        "opportunity"
    ]

    +

    artist_priority_df[
        "priority_market_growth_component"
    ]
    * priority_weights[
        "market_growth"
    ]

    +

    artist_priority_df[
        "priority_emerging_market_component"
    ]
    * priority_weights[
        "emerging_market"
    ]

    +

    artist_priority_df[
        "priority_reach_component"
    ]
    * priority_weights[
        "reach"
    ]
).clip(
    0,
    100
)


# ============================================================
# 6. Supporting-signal count
# ============================================================
#
# Count how many strong signals are present.
# This prevents one strong metric alone from automatically
# becoming a high-priority PMIP opportunity.
# ============================================================

track_priority_df[
    "priority_signal_count"
] = (

    (
        track_priority_df[
            "pmip_growth_score"
        ]
        >= 45
    ).astype(int)

    +

    (
        track_priority_df[
            "pmip_growth_percentile"
        ]
        >= 90
    ).astype(int)

    +

    (
        track_priority_df[
            "priority_opportunity_component"
        ]
        >= 45
    ).astype(int)

    +

    (
        track_priority_df[
            "priority_market_growth_component"
        ]
        >= 50
    ).astype(int)

    +

    (
        track_priority_df[
            "priority_emerging_market_component"
        ]
        >= 50
    ).astype(int)

    +

    (
        track_priority_df[
            "priority_reach_component"
        ]
        >= 25
    ).astype(int)
)


artist_priority_df[
    "priority_signal_count"
] = (

    (
        artist_priority_df[
            "pmip_growth_score"
        ]
        >= 45
    ).astype(int)

    +

    (
        artist_priority_df[
            "pmip_growth_percentile"
        ]
        >= 90
    ).astype(int)

    +

    (
        artist_priority_df[
            "priority_opportunity_component"
        ]
        >= 45
    ).astype(int)

    +

    (
        artist_priority_df[
            "priority_market_growth_component"
        ]
        >= 50
    ).astype(int)

    +

    (
        artist_priority_df[
            "priority_emerging_market_component"
        ]
        >= 50
    ).astype(int)

    +

    (
        artist_priority_df[
            "priority_reach_component"
        ]
        >= 25
    ).astype(int)
)


# ============================================================
# 7. Priority classification
# ============================================================

def classify_priority(score, signal_count):

    if pd.isna(score):
        return "Unclassified"

    if (
        score >= 75
        and signal_count >= 5
    ):
        return "Critical Priority"

    elif (
        score >= 60
        and signal_count >= 4
    ):
        return "High Priority"

    elif (
        score >= 45
        and signal_count >= 3
    ):
        return "Elevated Priority"

    elif (
        score >= 30
        and signal_count >= 2
    ):
        return "Developing Priority"

    else:
        return "Low Priority"


track_priority_df[
    "pmip_priority_class"
] = [
    classify_priority(
        score,
        signals
    )
    for score, signals in zip(
        track_priority_df[
            "pmip_priority_score"
        ],
        track_priority_df[
            "priority_signal_count"
        ]
    )
]


artist_priority_df[
    "pmip_priority_class"
] = [
    classify_priority(
        score,
        signals
    )
    for score, signals in zip(
        artist_priority_df[
            "pmip_priority_score"
        ],
        artist_priority_df[
            "priority_signal_count"
        ]
    )
]


# ============================================================
# 8. Final priority ranking
# ============================================================

track_priority_df[
    "pmip_priority_rank"
] = (
    track_priority_df[
        "pmip_priority_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


artist_priority_df[
    "pmip_priority_rank"
] = (
    artist_priority_df[
        "pmip_priority_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


track_priority_df = (
    track_priority_df
    .sort_values(
        [
            "pmip_priority_rank",
            "pmip_growth_rank"
        ]
    )
    .reset_index(
        drop=True
    )
)


artist_priority_df = (
    artist_priority_df
    .sort_values(
        [
            "pmip_priority_rank",
            "pmip_growth_rank"
        ]
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 9. Priority distributions
# ============================================================

priority_class_order = [
    "Critical Priority",
    "High Priority",
    "Elevated Priority",
    "Developing Priority",
    "Low Priority"
]


track_priority_distribution_df = (
    track_priority_df[
        "pmip_priority_class"
    ]
    .value_counts()
    .reindex(
        priority_class_order,
        fill_value=0
    )
    .rename(
        "Tracks"
    )
    .reset_index()
    .rename(
        columns={
            "pmip_priority_class":
                "PMIP Priority Class"
        }
    )
)


track_priority_distribution_df[
    "Share (%)"
] = (
    track_priority_distribution_df[
        "Tracks"
    ]
    / len(
        track_priority_df
    )
    * 100
)


artist_priority_distribution_df = (
    artist_priority_df[
        "pmip_priority_class"
    ]
    .value_counts()
    .reindex(
        priority_class_order,
        fill_value=0
    )
    .rename(
        "Artists"
    )
    .reset_index()
    .rename(
        columns={
            "pmip_priority_class":
                "PMIP Priority Class"
        }
    )
)


artist_priority_distribution_df[
    "Share (%)"
] = (
    artist_priority_distribution_df[
        "Artists"
    ]
    / len(
        artist_priority_df
    )
    * 100
)


print()
print("Track high-priority growth-signal distribution")
print("=" * 100)

display(
    track_priority_distribution_df
)


print()
print("Artist high-priority growth-signal distribution")
print("=" * 100)

display(
    artist_priority_distribution_df
)


# ============================================================
# 10. Top PMIP track priority signals
# ============================================================

track_priority_display_columns = [
    column
    for column in [
        "track_id",
        "countries_represented",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        "priority_signal_count",
        "pmip_priority_score",
        "pmip_priority_class",
        "pmip_priority_rank"
    ]
    if column in track_priority_df.columns
]


top_track_priority_df = (
    track_priority_df[
        track_priority_display_columns
    ]
    .head(20)
)


print()
print("Top PMIP track priority signals")
print("=" * 100)

display(
    top_track_priority_df
)


# ============================================================
# 11. Top PMIP artist priority signals
# ============================================================

artist_priority_display_columns = [
    column
    for column in [
        "artists",
        "countries_represented",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
        "priority_signal_count",
        "pmip_priority_score",
        "pmip_priority_class",
        "pmip_priority_rank"
    ]
    if column in artist_priority_df.columns
]


top_artist_priority_df = (
    artist_priority_df[
        artist_priority_display_columns
    ]
    .head(20)
)


print()
print("Top PMIP artist priority signals")
print("=" * 100)

display(
    top_artist_priority_df
)


# ============================================================
# 12. Priority overview
# ============================================================

track_high_priority_count = (
    track_priority_df[
        "pmip_priority_class"
    ]
    .isin(
        [
            "Critical Priority",
            "High Priority"
        ]
    )
    .sum()
)


artist_high_priority_count = (
    artist_priority_df[
        "pmip_priority_class"
    ]
    .isin(
        [
            "Critical Priority",
            "High Priority"
        ]
    )
    .sum()
)


priority_overview_df = pd.DataFrame(
    {
        "Metric": [

            "Tracks assessed",

            "Median track priority score",

            "Tracks with High-or-Critical Priority",

            "Critical-priority tracks",

            "Artists assessed",

            "Median artist priority score",

            "Artists with High-or-Critical Priority",

            "Critical-priority artists"
        ],

        "Observed Value": [

            len(
                track_priority_df
            ),

            track_priority_df[
                "pmip_priority_score"
            ]
            .median(),

            track_high_priority_count,

            (
                track_priority_df[
                    "pmip_priority_class"
                ]
                == "Critical Priority"
            )
            .sum(),

            len(
                artist_priority_df
            ),

            artist_priority_df[
                "pmip_priority_score"
            ]
            .median(),

            artist_high_priority_count,

            (
                artist_priority_df[
                    "pmip_priority_class"
                ]
                == "Critical Priority"
            )
            .sum()
        ]
    }
)


print()
print("High-priority growth signal overview")
print("=" * 100)

display(
    priority_overview_df
)


# ============================================================
# 13. Diagnostics
# ============================================================

priority_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [

            "Tracks assessed",

            "Track priority scores available",

            "Track priority classes assigned",

            "Track priority ranks assigned",

            "Artists assessed",

            "Artist priority scores available",

            "Artist priority classes assigned",

            "Artist priority ranks assigned"
        ],

        "Observed Value": [

            len(
                track_priority_df
            ),

            track_priority_df[
                "pmip_priority_score"
            ]
            .notna()
            .sum(),

            track_priority_df[
                "pmip_priority_class"
            ]
            .notna()
            .sum(),

            track_priority_df[
                "pmip_priority_rank"
            ]
            .notna()
            .sum(),

            len(
                artist_priority_df
            ),

            artist_priority_df[
                "pmip_priority_score"
            ]
            .notna()
            .sum(),

            artist_priority_df[
                "pmip_priority_class"
            ]
            .notna()
            .sum(),

            artist_priority_df[
                "pmip_priority_rank"
            ]
            .notna()
            .sum()
        ]
    }
)


print()
print("High-priority growth signal diagnostics")
print("=" * 100)

display(
    priority_diagnostics_df
)


# ============================================================
# 14. Validation
# ============================================================

track_priority_complete = bool(
    track_priority_df[
        "pmip_priority_score"
    ]
    .notna()
    .all()
)

artist_priority_complete = bool(
    artist_priority_df[
        "pmip_priority_score"
    ]
    .notna()
    .all()
)


track_priority_bounds = bool(
    track_priority_df[
        "pmip_priority_score"
    ]
    .between(
        0,
        100
    )
    .all()
)


artist_priority_bounds = bool(
    artist_priority_df[
        "pmip_priority_score"
    ]
    .between(
        0,
        100
    )
    .all()
)


track_priority_classes_complete = bool(
    track_priority_df[
        "pmip_priority_class"
    ]
    .notna()
    .all()
)


artist_priority_classes_complete = bool(
    artist_priority_df[
        "pmip_priority_class"
    ]
    .notna()
    .all()
)


track_priority_ranks_complete = bool(
    track_priority_df[
        "pmip_priority_rank"
    ]
    .notna()
    .all()
)


artist_priority_ranks_complete = bool(
    artist_priority_df[
        "pmip_priority_rank"
    ]
    .notna()
    .all()
)


priority_validation_df = pd.DataFrame(
    {
        "Validation Area": [

            "Section 11.4 completion",

            "Track priority completeness",

            "Artist priority completeness",

            "Track priority score bounds",

            "Artist priority score bounds",

            "Track priority classification completeness",

            "Artist priority classification completeness",

            "Track priority ranking completeness",

            "Artist priority ranking completeness",

            "Multi-signal priority construction"
        ],

        "Requirement": [

            (
                "Market Ranking must already "
                "be complete."
            ),

            (
                "Every retained track must receive "
                "a PMIP priority score."
            ),

            (
                "Every retained artist must receive "
                "a PMIP priority score."
            ),

            (
                "Track priority scores must remain "
                "between 0 and 100."
            ),

            (
                "Artist priority scores must remain "
                "between 0 and 100."
            ),

            (
                "Every retained track must receive "
                "a PMIP priority class."
            ),

            (
                "Every retained artist must receive "
                "a PMIP priority class."
            ),

            (
                "Every retained track must receive "
                "a priority ranking."
            ),

            (
                "Every retained artist must receive "
                "a priority ranking."
            ),

            (
                "Priority identification must combine "
                "multiple validated PMIP growth signals."
            )
        ],

        "Observed Evidence": [

            (
                "Section 11.4 completion status: "
                f"{section_11_4_complete}"
            ),

            (
                f"{track_priority_df['pmip_priority_score'].notna().sum():,} "
                "track priority scores calculated"
            ),

            (
                f"{artist_priority_df['pmip_priority_score'].notna().sum():,} "
                "artist priority scores calculated"
            ),

            (
                "Track PMIP priority scores remain "
                "within valid 0–100 bounds"
            ),

            (
                "Artist PMIP priority scores remain "
                "within valid 0–100 bounds"
            ),

            (
                f"{track_priority_df['pmip_priority_class'].notna().sum():,} "
                "track priority classes assigned"
            ),

            (
                f"{artist_priority_df['pmip_priority_class'].notna().sum():,} "
                "artist priority classes assigned"
            ),

            (
                f"{track_priority_df['pmip_priority_rank'].notna().sum():,} "
                "track priority ranks assigned"
            ),

            (
                f"{artist_priority_df['pmip_priority_rank'].notna().sum():,} "
                "artist priority ranks assigned"
            ),

            (
                "Growth score, percentile, opportunity, "
                "market growth, emerging-market evidence "
                "and geographic reach combined"
            )
        ],

        "Passed": [

            section_11_4_complete,

            track_priority_complete,

            artist_priority_complete,

            track_priority_bounds,

            artist_priority_bounds,

            track_priority_classes_complete,

            artist_priority_classes_complete,

            track_priority_ranks_complete,

            artist_priority_ranks_complete,

            True
        ]
    }
)


print()
print("High-priority growth signal validation")
print("=" * 100)

display(
    priority_validation_df
)


# ============================================================
# 15. Section completion
# ============================================================

section_11_5_complete = bool(
    priority_validation_df[
        "Passed"
    ]
    .all()
)


section_11_complete = bool(
    section_11_1_complete
    and section_11_2_complete
    and section_11_3_complete
    and section_11_4_complete
    and section_11_5_complete
)


print()

print(
    f"Section 11.5 completion status: "
    f"{section_11_5_complete}"
)

print(
    f"Section 11 overall completion status: "
    f"{section_11_complete}"
)


if section_11_complete:

    print(
        "PMIP high-priority growth signals have been "
        "identified using combined validated growth, "
        "market, emerging-market and geographic evidence."
    )

else:

    print(
        "Section 11 contains one or more conditions "
        "requiring review."
    )
```

The highest-ranked priority records are also dominated by the **Exceptional Growth** class. This is consistent with the PMIP scoring structure because these records already show strong overall growth performance, while the priority framework provides an additional decision layer by checking whether supporting market and geographic signals reinforce that strength.

The results also show that some highly ranked records have relatively limited geographic reach. This is important because strong performance within a small number of markets can still represent a significant opportunity when other growth signals are sufficiently strong. Geographic reach therefore contributes supporting evidence without automatically excluding concentrated but high-performing opportunities.

### 12.1 Country-Level Findings

#### Purpose

This subsection summarises the main country-level findings produced by the PMIP growth-intelligence framework.

The aim is to identify which country markets demonstrate the strongest overall growth conditions, the highest market opportunities, and the strongest emerging-market potential.

The analysis brings together the validated outputs produced in earlier sections, including:

- combined country growth performance,
- market opportunity scores,
- emerging-market scores,
- country growth classifications,
- market opportunity classifications,
- and comparative country rankings.

The purpose is not to create new growth metrics, but to interpret the existing validated PMIP results and convert them into clear country-level intelligence that can support market prioritisation and strategic decision-making.


```python
# ============================================================
# 12.1 Country-Level Findings
# ============================================================

import pandas as pd
import numpy as np

print("Summarising PMIP country-level findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Confirm Section 11 completion
# ------------------------------------------------------------

section_11_complete = bool(
    globals().get("section_11_5_complete", False)
    or globals().get("section_11_complete", False)
)

print(f"Section 11 completion status: {section_11_complete}")

if not section_11_complete:
    raise RuntimeError(
        "Section 11 must be completed before interpreting final PMIP results."
    )


# ------------------------------------------------------------
# 2. Locate the country-level result source
# ------------------------------------------------------------

country_candidates = [
    "emerging_market_df",
    "market_opportunity_df",
    "country_growth_df",
    "country_growth_comparison_df",
]

country_source_name = None
country_results_df = None

for name in country_candidates:
    obj = globals().get(name)

    if isinstance(obj, pd.DataFrame) and "country" in obj.columns:
        country_source_name = name
        country_results_df = obj.copy()
        break

if country_results_df is None:
    raise NameError(
        "Could not locate a suitable PMIP country-level DataFrame."
    )

print(f"Country findings source: {country_source_name}")
print(f"Country markets available: {len(country_results_df):,}")


# ------------------------------------------------------------
# 3. Detect useful country-level result columns
# ------------------------------------------------------------

desired_columns = [
    "country",
    "combined_country_growth_score",
    "country_growth_class",
    "country_growth_rank",
    "market_opportunity_score",
    "market_opportunity_class",
    "market_opportunity_rank",
    "emerging_market_score",
    "emerging_market_class",
    "emerging_market_rank",
    "combined_strong_growth_share",
    "combined_very_high_relative_growth_share",
    "track_emerging_or_better_share",
    "artist_emerging_or_better_share",
]

available_columns = [
    col for col in desired_columns
    if col in country_results_df.columns
]

country_findings_df = (
    country_results_df[available_columns]
    .drop_duplicates(subset=["country"])
    .copy()
)


# ------------------------------------------------------------
# 4. Prepare country-level ranking views
# ------------------------------------------------------------

def top_country_view(df, score_column, n=15):

    if score_column not in df.columns:
        return pd.DataFrame()

    return (
        df.sort_values(
            score_column,
            ascending=False,
            na_position="last"
        )
        .head(n)
        .reset_index(drop=True)
    )


top_country_growth_df = top_country_view(
    country_findings_df,
    "combined_country_growth_score",
    15
)

top_market_opportunity_countries_df = top_country_view(
    country_findings_df,
    "market_opportunity_score",
    15
)

top_emerging_market_countries_df = top_country_view(
    country_findings_df,
    "emerging_market_score",
    15
)


# ------------------------------------------------------------
# 5. Display strongest country growth markets
# ------------------------------------------------------------

print("\nTop country markets by overall growth")
print("=" * 100)

growth_display_columns = [
    col for col in [
        "country",
        "combined_country_growth_score",
        "country_growth_class",
        "country_growth_rank",
        "combined_strong_growth_share",
        "combined_very_high_relative_growth_share",
    ]
    if col in top_country_growth_df.columns
]

display(
    top_country_growth_df[growth_display_columns]
)


# ------------------------------------------------------------
# 6. Display strongest market opportunities
# ------------------------------------------------------------

print("\nTop country markets by growth opportunity")
print("=" * 100)

opportunity_display_columns = [
    col for col in [
        "country",
        "market_opportunity_score",
        "market_opportunity_class",
        "market_opportunity_rank",
        "combined_country_growth_score",
        "combined_strong_growth_share",
    ]
    if col in top_market_opportunity_countries_df.columns
]

display(
    top_market_opportunity_countries_df[
        opportunity_display_columns
    ]
)


# ------------------------------------------------------------
# 7. Display strongest emerging markets
# ------------------------------------------------------------

print("\nTop PMIP emerging country markets")
print("=" * 100)

emerging_display_columns = [
    col for col in [
        "country",
        "emerging_market_score",
        "emerging_market_class",
        "emerging_market_rank",
        "market_opportunity_score",
        "track_emerging_or_better_share",
        "artist_emerging_or_better_share",
    ]
    if col in top_emerging_market_countries_df.columns
]

display(
    top_emerging_market_countries_df[
        emerging_display_columns
    ]
)


# ------------------------------------------------------------
# 8. Country-level class distributions
# ------------------------------------------------------------

country_class_distributions = {}

class_columns = [
    "country_growth_class",
    "market_opportunity_class",
    "emerging_market_class",
]

for column in class_columns:

    if column in country_findings_df.columns:

        distribution = (
            country_findings_df[column]
            .value_counts(dropna=False)
            .rename_axis(column)
            .reset_index(name="Country Markets")
        )

        distribution["Share (%)"] = (
            distribution["Country Markets"]
            / len(country_findings_df)
            * 100
        )

        country_class_distributions[column] = distribution


if "country_growth_class" in country_class_distributions:

    print("\nCountry growth class distribution")
    print("=" * 100)

    display(
        country_class_distributions[
            "country_growth_class"
        ]
    )


if "market_opportunity_class" in country_class_distributions:

    print("\nMarket opportunity class distribution")
    print("=" * 100)

    display(
        country_class_distributions[
            "market_opportunity_class"
        ]
    )


if "emerging_market_class" in country_class_distributions:

    print("\nEmerging market class distribution")
    print("=" * 100)

    display(
        country_class_distributions[
            "emerging_market_class"
        ]
    )


# ------------------------------------------------------------
# 9. Country-level findings overview
# ------------------------------------------------------------

overview_rows = [
    {
        "Metric": "Country markets assessed",
        "Observed Value": len(country_findings_df),
    }
]

if "combined_country_growth_score" in country_findings_df.columns:

    overview_rows.extend([
        {
            "Metric": "Median combined country growth score",
            "Observed Value":
                country_findings_df[
                    "combined_country_growth_score"
                ].median(),
        },
        {
            "Metric": "Highest combined country growth score",
            "Observed Value":
                country_findings_df[
                    "combined_country_growth_score"
                ].max(),
        },
    ])


if "market_opportunity_score" in country_findings_df.columns:

    overview_rows.extend([
        {
            "Metric": "Median market opportunity score",
            "Observed Value":
                country_findings_df[
                    "market_opportunity_score"
                ].median(),
        },
        {
            "Metric": "Highest market opportunity score",
            "Observed Value":
                country_findings_df[
                    "market_opportunity_score"
                ].max(),
        },
    ])


if "emerging_market_score" in country_findings_df.columns:

    overview_rows.extend([
        {
            "Metric": "Median emerging-market score",
            "Observed Value":
                country_findings_df[
                    "emerging_market_score"
                ].median(),
        },
        {
            "Metric": "Highest emerging-market score",
            "Observed Value":
                country_findings_df[
                    "emerging_market_score"
                ].max(),
        },
    ])


country_findings_overview_df = pd.DataFrame(
    overview_rows
)

print("\nCountry-level findings overview")
print("=" * 100)

display(country_findings_overview_df)


# ------------------------------------------------------------
# 10. Diagnostics
# ------------------------------------------------------------

country_findings_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Country markets assessed",
        "Country identifiers available",
        "Country growth scores available",
        "Market opportunity scores available",
        "Emerging-market scores available",
        "Country-level ranking views prepared",
    ],
    "Observed Value": [
        len(country_findings_df),

        country_findings_df["country"]
        .notna()
        .sum(),

        (
            country_findings_df[
                "combined_country_growth_score"
            ]
            .notna()
            .sum()
            if "combined_country_growth_score"
            in country_findings_df.columns
            else 0
        ),

        (
            country_findings_df[
                "market_opportunity_score"
            ]
            .notna()
            .sum()
            if "market_opportunity_score"
            in country_findings_df.columns
            else 0
        ),

        (
            country_findings_df[
                "emerging_market_score"
            ]
            .notna()
            .sum()
            if "emerging_market_score"
            in country_findings_df.columns
            else 0
        ),

        all([
            isinstance(
                top_country_growth_df,
                pd.DataFrame
            ),
            isinstance(
                top_market_opportunity_countries_df,
                pd.DataFrame
            ),
            isinstance(
                top_emerging_market_countries_df,
                pd.DataFrame
            ),
        ]),
    ]
})

print("\nCountry-level findings diagnostics")
print("=" * 100)

display(country_findings_diagnostics_df)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

country_count = len(country_findings_df)

country_findings_validation_df = pd.DataFrame({
    "Validation Area": [
        "Section 11 completion",
        "Country market availability",
        "Country identifier completeness",
        "Country growth intelligence availability",
        "Market opportunity intelligence availability",
        "Emerging-market intelligence availability",
    ],

    "Requirement": [
        "Market Growth Scoring and Classification must already be complete.",
        "Validated country markets must remain available for interpretation.",
        "Every country market must retain a valid country identifier.",
        "Country-level growth intelligence must remain available.",
        "Market-opportunity intelligence must remain available.",
        "Emerging-market intelligence must remain available.",
    ],

    "Observed Evidence": [
        f"Section 11 overall completion status: {section_11_complete}",

        f"{country_count:,} country markets available",

        (
            f"{country_findings_df['country'].notna().sum():,} "
            "country identifiers available"
        ),

        (
            f"{country_findings_df['combined_country_growth_score'].notna().sum():,} "
            "country growth scores available"
            if "combined_country_growth_score"
            in country_findings_df.columns
            else "Country growth score column unavailable"
        ),

        (
            f"{country_findings_df['market_opportunity_score'].notna().sum():,} "
            "market opportunity scores available"
            if "market_opportunity_score"
            in country_findings_df.columns
            else "Market opportunity score column unavailable"
        ),

        (
            f"{country_findings_df['emerging_market_score'].notna().sum():,} "
            "emerging-market scores available"
            if "emerging_market_score"
            in country_findings_df.columns
            else "Emerging-market score column unavailable"
        ),
    ],

    "Passed": [
        section_11_complete,

        country_count > 0,

        country_findings_df[
            "country"
        ].notna().all(),

        (
            country_findings_df[
                "combined_country_growth_score"
            ].notna().all()
            if "combined_country_growth_score"
            in country_findings_df.columns
            else False
        ),

        (
            country_findings_df[
                "market_opportunity_score"
            ].notna().all()
            if "market_opportunity_score"
            in country_findings_df.columns
            else False
        ),

        (
            country_findings_df[
                "emerging_market_score"
            ].notna().all()
            if "emerging_market_score"
            in country_findings_df.columns
            else False
        ),
    ],
})


print("\nCountry-level findings validation")
print("=" * 100)

display(country_findings_validation_df)


# ------------------------------------------------------------
# 12. Completion status
# ------------------------------------------------------------

section_12_1_complete = bool(
    country_findings_validation_df[
        "Passed"
    ].all()
)

print(
    f"\nSection 12.1 completion status: "
    f"{section_12_1_complete}"
)

if section_12_1_complete:

    print(
        "PMIP country-level findings have been "
        "summarised successfully from validated "
        "growth, opportunity and emerging-market intelligence."
    )

else:

    print(
        "Section 12.1 requires review before "
        "continuing to artist-level findings."
    )
```

### Interpretation

The country-level results show that PMIP assessed **76 country markets** and successfully retained complete growth, opportunity, and emerging-market intelligence for every market.

The overall country-growth comparison shows a fairly mixed market landscape. Most countries fall into the **Moderate Growth Market** category, representing **39 markets (51.32%)**, while **24 markets (31.58%)** are classified as **Above-Average Growth Markets**. Only **3 markets (3.95%)** reached the **High Growth Market** category, showing that very strong market-wide growth is relatively uncommon within the analysed data.

The strongest country-level growth was observed for **Kazakhstan (KZ)**, with a combined country growth score of approximately **65.90**, followed by **Andorra (AD)** at **64.78** and **Belarus (BY)** at **62.90**. These three markets form the highest-growth group identified by the PMIP framework. Their rankings are supported not only by their overall growth scores, but also by evidence of strong and very-high relative growth across track and artist activity.

Market opportunity results provide an additional layer of interpretation. Kazakhstan recorded the highest market opportunity score at approximately **79.54**, followed by Belarus at **70.96** and Andorra at **65.23**. Kazakhstan and Belarus were therefore classified as **Very High Opportunity** markets, while Andorra was classified as **High Opportunity**. Across all 76 markets, however, **48 markets (63.16%)** remained in the Low Opportunity category and another **23 markets (30.26%)** were classified as Developing Opportunity. This suggests that the strongest opportunities are concentrated within a relatively small group of countries.

The emerging-market results show a similar pattern. Kazakhstan achieved the highest emerging-market score at approximately **69.55**, with Belarus following at **64.33**. Both were classified as having **High Emerging Potential**. Andorra and Venezuela were classified as **Developing Emerging Markets**, while a further group of markets showed **Early Emerging Signals**.

Overall, these findings demonstrate that market size or existing visibility alone should not determine growth priority. PMIP highlights countries where multiple signals agree, including overall growth strength, relative growth, market opportunity, and emerging-market behaviour. Kazakhstan and Belarus stand out particularly strongly because they appear near the top across several independent country-level measures rather than performing well on only one indicator.

This provides useful market intelligence for identifying countries where artist development, release promotion, audience targeting, or further A&R investigation may deserve greater attention.

## 12.2 Artist-Level Findings

### Purpose

This subsection summarises the most important artist-level growth findings produced by the PMIP analysis.

The objective is to move from the detailed artist-level calculations developed in earlier sections to a concise intelligence view showing:

- the highest-ranked artists by final PMIP growth score;
- artists with the strongest overall growth opportunities;
- artists showing the strongest emerging-market signals;
- the distribution of artist PMIP growth classes;
- the distribution of artist priority classes;
- and the overall strength of artist-level growth evidence.

These findings combine the validated artist-level market-growth, emerging-market, opportunity, ranking, and priority signals produced throughout the PMIP workflow.

The results are intended to highlight artists whose current market behaviour may justify further A&R investigation, market expansion analysis, promotional attention, or strategic monitoring.

```python
# ============================================================
# 12.2 ARTIST-LEVEL FINDINGS
# ============================================================

import pandas as pd
import numpy as np

print("Summarising PMIP artist-level findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Confirm Section 11 completion
# ------------------------------------------------------------

section_11_complete = bool(
    globals().get("section_11_complete", False)
    or globals().get("section_11_overall_complete", False)
    or globals().get("section_11_5_complete", False)
)

print(f"Section 11 completion status: {section_11_complete}")


# ------------------------------------------------------------
# 2. Helper: safely locate the best available DataFrame
# ------------------------------------------------------------

def find_dataframe(
    preferred_names=None,
    required_columns=None,
    minimum_rows=1
):
    preferred_names = preferred_names or []
    required_columns = required_columns or []

    # First check preferred names
    for name in preferred_names:
        obj = globals().get(name)

        if isinstance(obj, pd.DataFrame):
            if (
                len(obj) >= minimum_rows
                and all(col in obj.columns for col in required_columns)
            ):
                return name, obj.copy()

    # Then inspect all available DataFrames safely
    for name, obj in list(globals().items()):
        if not isinstance(obj, pd.DataFrame):
            continue

        if len(obj) < minimum_rows:
            continue

        if all(col in obj.columns for col in required_columns):
            return name, obj.copy()

    return None, None


# ------------------------------------------------------------
# 3. Locate final artist growth intelligence
# ------------------------------------------------------------

artist_growth_source_name, artist_findings_df = find_dataframe(
    preferred_names=[
        "artist_priority_df",
        "artist_growth_score_df",
        "artist_growth_opportunity_df",
        "artist_market_ranking_df"
    ],
    required_columns=[
        "artists",
        "pmip_growth_score"
    ],
    minimum_rows=100
)

if artist_findings_df is None:
    raise NameError(
        "Could not locate the complete artist-level PMIP growth DataFrame."
    )

print(f"Artist findings source: {artist_growth_source_name}")
print(f"Artists available: {len(artist_findings_df):,}")


# ------------------------------------------------------------
# 4. Ensure numeric fields are usable
# ------------------------------------------------------------

numeric_candidates = [
    "pmip_growth_score",
    "pmip_growth_rank",
    "pmip_growth_percentile",
    "growth_opportunity_score",
    "mean_market_growth_score",
    "maximum_artist_emerging_market_score",
    "priority_score",
    "countries_represented"
]

for col in numeric_candidates:
    if col in artist_findings_df.columns:
        artist_findings_df[col] = pd.to_numeric(
            artist_findings_df[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# 5. Top artists by final PMIP growth score
# ------------------------------------------------------------

growth_columns = [
    col for col in [
        "artists",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
        "countries_represented"
    ]
    if col in artist_findings_df.columns
]

top_artist_growth_df = (
    artist_findings_df
    .sort_values(
        "pmip_growth_score",
        ascending=False
    )
    .head(15)[growth_columns]
    .reset_index(drop=True)
)

print("\nTop artists by PMIP growth score")
print("=" * 100)
display(top_artist_growth_df)


# ------------------------------------------------------------
# 6. Top artists by growth opportunity
# ------------------------------------------------------------

if "growth_opportunity_score" in artist_findings_df.columns:

    opportunity_columns = [
        col for col in [
            "artists",
            "growth_opportunity_score",
            "growth_opportunity_class",
            "pmip_growth_score",
            "pmip_growth_class",
            "pmip_growth_rank",
            "mean_market_growth_score",
            "maximum_artist_emerging_market_score",
            "countries_represented"
        ]
        if col in artist_findings_df.columns
    ]

    top_artist_opportunity_df = (
        artist_findings_df
        .sort_values(
            "growth_opportunity_score",
            ascending=False
        )
        .head(15)[opportunity_columns]
        .reset_index(drop=True)
    )

    print("\nTop artists by growth opportunity")
    print("=" * 100)
    display(top_artist_opportunity_df)

else:
    top_artist_opportunity_df = pd.DataFrame()

    print("\nGrowth opportunity score not available in selected source.")


# ------------------------------------------------------------
# 7. Top artists by emerging-market strength
# ------------------------------------------------------------

if "maximum_artist_emerging_market_score" in artist_findings_df.columns:

    emerging_columns = [
        col for col in [
            "artists",
            "maximum_artist_emerging_market_score",
            "growth_opportunity_score",
            "pmip_growth_score",
            "pmip_growth_class",
            "pmip_growth_rank",
            "mean_market_growth_score",
            "countries_represented"
        ]
        if col in artist_findings_df.columns
    ]

    top_artist_emerging_df = (
        artist_findings_df
        .sort_values(
            "maximum_artist_emerging_market_score",
            ascending=False
        )
        .head(15)[emerging_columns]
        .reset_index(drop=True)
    )

    print("\nTop artists by emerging-market strength")
    print("=" * 100)
    display(top_artist_emerging_df)

else:
    top_artist_emerging_df = pd.DataFrame()

    print("\nEmerging-market score not available in selected source.")


# ------------------------------------------------------------
# 8. Artist PMIP growth-class distribution
# ------------------------------------------------------------

if "pmip_growth_class" in artist_findings_df.columns:

    artist_growth_class_distribution_df = (
        artist_findings_df["pmip_growth_class"]
        .value_counts(dropna=False)
        .rename_axis("PMIP Growth Class")
        .reset_index(name="Artists")
    )

    artist_growth_class_distribution_df["Share (%)"] = (
        artist_growth_class_distribution_df["Artists"]
        / len(artist_findings_df)
        * 100
    )

    print("\nArtist PMIP growth-class distribution")
    print("=" * 100)
    display(artist_growth_class_distribution_df)

else:
    artist_growth_class_distribution_df = pd.DataFrame()


# ------------------------------------------------------------
# 9. Artist priority-class distribution
# ------------------------------------------------------------

priority_column = None

for candidate in [
    "pmip_priority_class",
    "priority_class",
    "PMIP Priority Class"
]:
    if candidate in artist_findings_df.columns:
        priority_column = candidate
        break

if priority_column is not None:

    artist_priority_distribution_df = (
        artist_findings_df[priority_column]
        .value_counts(dropna=False)
        .rename_axis("PMIP Priority Class")
        .reset_index(name="Artists")
    )

    artist_priority_distribution_df["Share (%)"] = (
        artist_priority_distribution_df["Artists"]
        / len(artist_findings_df)
        * 100
    )

    print("\nArtist PMIP priority-class distribution")
    print("=" * 100)
    display(artist_priority_distribution_df)

else:
    artist_priority_distribution_df = pd.DataFrame()

    print("\nArtist priority-class column not available in selected source.")


# ------------------------------------------------------------
# 10. Artist-level findings overview
# ------------------------------------------------------------

artist_overview_rows = [
    {
        "Metric": "Artists assessed",
        "Observed Value": len(artist_findings_df)
    },
    {
        "Metric": "Median artist PMIP growth score",
        "Observed Value": artist_findings_df[
            "pmip_growth_score"
        ].median()
    },
    {
        "Metric": "Mean artist PMIP growth score",
        "Observed Value": artist_findings_df[
            "pmip_growth_score"
        ].mean()
    },
    {
        "Metric": "Highest artist PMIP growth score",
        "Observed Value": artist_findings_df[
            "pmip_growth_score"
        ].max()
    }
]

if "growth_opportunity_score" in artist_findings_df.columns:
    artist_overview_rows.extend([
        {
            "Metric": "Median artist growth-opportunity score",
            "Observed Value": artist_findings_df[
                "growth_opportunity_score"
            ].median()
        },
        {
            "Metric": "Highest artist growth-opportunity score",
            "Observed Value": artist_findings_df[
                "growth_opportunity_score"
            ].max()
        }
    ])

if "maximum_artist_emerging_market_score" in artist_findings_df.columns:
    artist_overview_rows.extend([
        {
            "Metric": "Median strongest emerging-market score",
            "Observed Value": artist_findings_df[
                "maximum_artist_emerging_market_score"
            ].median()
        },
        {
            "Metric": "Highest emerging-market score",
            "Observed Value": artist_findings_df[
                "maximum_artist_emerging_market_score"
            ].max()
        }
    ])

artist_findings_overview_df = pd.DataFrame(
    artist_overview_rows
)

print("\nArtist-level findings overview")
print("=" * 100)
display(artist_findings_overview_df)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

artist_findings_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Artists assessed",
        "Artist identifiers available",
        "PMIP growth scores available",
        "PMIP growth classes available",
        "Growth opportunity intelligence available",
        "Emerging-market intelligence available",
        "Artist ranking view prepared"
    ],
    "Observed Value": [
        len(artist_findings_df),

        int(
            artist_findings_df["artists"]
            .notna()
            .sum()
        ),

        int(
            artist_findings_df["pmip_growth_score"]
            .notna()
            .sum()
        ),

        (
            int(
                artist_findings_df["pmip_growth_class"]
                .notna()
                .sum()
            )
            if "pmip_growth_class" in artist_findings_df.columns
            else 0
        ),

        (
            int(
                artist_findings_df["growth_opportunity_score"]
                .notna()
                .sum()
            )
            if "growth_opportunity_score" in artist_findings_df.columns
            else 0
        ),

        (
            int(
                artist_findings_df[
                    "maximum_artist_emerging_market_score"
                ]
                .notna()
                .sum()
            )
            if "maximum_artist_emerging_market_score"
            in artist_findings_df.columns
            else 0
        ),

        True
    ]
})

print("\nArtist-level findings diagnostics")
print("=" * 100)
display(artist_findings_diagnostics_df)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

artist_count = len(artist_findings_df)

artist_identifier_complete = (
    artist_findings_df["artists"]
    .notna()
    .sum()
    == artist_count
)

artist_score_complete = (
    artist_findings_df["pmip_growth_score"]
    .notna()
    .sum()
    == artist_count
)

artist_score_bounds_valid = (
    artist_findings_df["pmip_growth_score"]
    .dropna()
    .between(0, 100)
    .all()
)

artist_growth_classes_available = (
    "pmip_growth_class" in artist_findings_df.columns
)

opportunity_available = (
    "growth_opportunity_score"
    in artist_findings_df.columns
)

emerging_available = (
    "maximum_artist_emerging_market_score"
    in artist_findings_df.columns
)

artist_findings_validation_df = pd.DataFrame({
    "Validation Area": [
        "Section 11 completion",
        "Artist population availability",
        "Artist identifier completeness",
        "Artist growth intelligence availability",
        "Artist growth score bounds",
        "Growth opportunity intelligence availability",
        "Emerging-market intelligence availability"
    ],

    "Requirement": [
        "Market Growth Scoring and Classification must already be complete.",

        "The retained artist population must remain available for final interpretation.",

        "Every retained artist must retain a valid artist identifier.",

        "Every retained artist must retain a PMIP final growth score and growth classification.",

        "Artist PMIP growth scores must remain between 0 and 100.",

        "Artist growth-opportunity evidence must remain available for interpretation.",

        "Artist emerging-market evidence must remain available for interpretation."
    ],

    "Observed Evidence": [
        f"Section 11 overall completion status: {section_11_complete}",

        f"{artist_count:,} artists available",

        (
            f"{artist_findings_df['artists'].notna().sum():,} "
            "artist identifiers available"
        ),

        (
            f"{artist_findings_df['pmip_growth_score'].notna().sum():,} "
            "artist PMIP growth scores available"
        ),

        "Artist PMIP growth scores remain within valid 0–100 bounds",

        (
            f"{artist_findings_df['growth_opportunity_score'].notna().sum():,} "
            "artist growth-opportunity scores available"
            if opportunity_available
            else "Artist growth-opportunity score unavailable"
        ),

        (
            f"{artist_findings_df['maximum_artist_emerging_market_score'].notna().sum():,} "
            "artist emerging-market scores available"
            if emerging_available
            else "Artist emerging-market score unavailable"
        )
    ],

    "Passed": [
        section_11_complete,
        artist_count > 0,
        artist_identifier_complete,
        artist_score_complete and artist_growth_classes_available,
        artist_score_bounds_valid,
        opportunity_available,
        emerging_available
    ]
})

print("\nArtist-level findings validation")
print("=" * 100)
display(artist_findings_validation_df)


# ------------------------------------------------------------
# 13. Completion flag
# ------------------------------------------------------------

section_12_2_complete = bool(
    artist_findings_validation_df["Passed"].all()
)

print()
print(
    f"Section 12.2 completion status: "
    f"{section_12_2_complete}"
)

if section_12_2_complete:
    print(
        "PMIP artist-level findings have been summarised "
        "successfully from validated growth, opportunity, "
        "emerging-market and ranking intelligence."
    )
else:
    print(
        "Section 12.2 requires review because one or more "
        "artist-level validation checks did not pass."
    )
```

## 12.3 Track-Level Findings

### Purpose

This subsection summarises the final PMIP intelligence produced for individual tracks.

The analysis brings together each track's final PMIP growth score, growth classification, ranking position, growth-opportunity evidence, emerging-market strength, market coverage, and priority classification.

The purpose is to identify:

- the strongest tracks by final PMIP growth score;
- tracks with the highest growth opportunities;
- tracks showing the strongest emerging-market evidence;
- the distribution of final PMIP growth classes;
- the distribution of track priority classes; and
- whether the complete retained track population remains available for final interpretation.

This subsection uses the validated outputs produced in Sections 10 and 11 and does not construct any new growth signals.

```python
# ============================================================
# Section 12.3 — Track-Level Findings
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display


print("Summarising PMIP track-level findings")
print("=" * 95)


# ------------------------------------------------------------
# 1. Confirm previous section completion
# ------------------------------------------------------------

section_11_complete = globals().get("section_11_complete", True)

print(f"Section 11 completion status: {section_11_complete}")


if not section_11_complete:
    raise RuntimeError(
        "Section 11 must be completed before Section 12.3 can run."
    )


# ------------------------------------------------------------
# 2. Locate the complete track-level intelligence source
# ------------------------------------------------------------

track_source_candidates = [
    "track_priority_df",
    "track_ranking_df",
    "track_growth_score_df",
    "track_growth_opportunity_df",
]

track_findings_source_name = None
track_findings_source_df = None

for candidate_name in track_source_candidates:
    candidate_obj = globals().get(candidate_name)

    if (
        isinstance(candidate_obj, pd.DataFrame)
        and len(candidate_obj) > 0
    ):
        if "track_id" in candidate_obj.columns:
            track_findings_source_name = candidate_name
            track_findings_source_df = candidate_obj.copy()
            break


if track_findings_source_df is None:
    raise NameError(
        "No valid track-level PMIP intelligence DataFrame could be found."
    )


print(f"Track findings source: {track_findings_source_name}")
print(f"Tracks available: {len(track_findings_source_df):,}")


# ------------------------------------------------------------
# 3. Helper for selecting existing columns safely
# ------------------------------------------------------------

def existing_columns(df, columns):
    return [column for column in columns if column in df.columns]


# ------------------------------------------------------------
# 4. Top tracks by final PMIP growth score
# ------------------------------------------------------------

track_growth_columns = existing_columns(
    track_findings_source_df,
    [
        "track_id",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        "countries_represented",
        "pmip_priority_class",
        "priority_score",
    ],
)

if "pmip_growth_score" not in track_growth_columns:
    raise KeyError(
        "The track source does not contain 'pmip_growth_score'."
    )

top_tracks_by_growth = (
    track_findings_source_df[
        track_growth_columns
    ]
    .sort_values(
        "pmip_growth_score",
        ascending=False,
    )
    .head(15)
    .copy()
)


print("\nTop tracks by PMIP growth score")
print("=" * 95)

display(top_tracks_by_growth)


# ------------------------------------------------------------
# 5. Top tracks by growth opportunity
# ------------------------------------------------------------

track_opportunity_columns = existing_columns(
    track_findings_source_df,
    [
        "track_id",
        "growth_opportunity_score",
        "growth_opportunity_class",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        "countries_represented",
        "pmip_priority_class",
    ],
)


if "growth_opportunity_score" in track_opportunity_columns:

    top_tracks_by_opportunity = (
        track_findings_source_df[
            track_opportunity_columns
        ]
        .sort_values(
            "growth_opportunity_score",
            ascending=False,
        )
        .head(15)
        .copy()
    )

    print("\nTop tracks by growth opportunity")
    print("=" * 95)

    display(top_tracks_by_opportunity)

else:

    top_tracks_by_opportunity = pd.DataFrame()

    print(
        "\nGrowth-opportunity score not available in the selected track source."
    )


# ------------------------------------------------------------
# 6. Top tracks by emerging-market strength
# ------------------------------------------------------------

track_emerging_columns = existing_columns(
    track_findings_source_df,
    [
        "track_id",
        "maximum_track_emerging_market_score",
        "growth_opportunity_score",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "mean_market_growth_score",
        "countries_represented",
        "pmip_priority_class",
    ],
)


if "maximum_track_emerging_market_score" in track_emerging_columns:

    top_tracks_by_emerging_strength = (
        track_findings_source_df[
            track_emerging_columns
        ]
        .sort_values(
            "maximum_track_emerging_market_score",
            ascending=False,
        )
        .head(15)
        .copy()
    )

    print("\nTop tracks by emerging-market strength")
    print("=" * 95)

    display(top_tracks_by_emerging_strength)

else:

    top_tracks_by_emerging_strength = pd.DataFrame()

    print(
        "\nEmerging-market strength score not available in the selected track source."
    )


# ------------------------------------------------------------
# 7. PMIP growth-class distribution
# ------------------------------------------------------------

if "pmip_growth_class" in track_findings_source_df.columns:

    track_growth_class_distribution = (
        track_findings_source_df[
            "pmip_growth_class"
        ]
        .value_counts(dropna=False)
        .rename_axis("PMIP Growth Class")
        .reset_index(name="Tracks")
    )

    track_growth_class_distribution["Share (%)"] = (
        track_growth_class_distribution["Tracks"]
        / len(track_findings_source_df)
        * 100
    )

    print("\nTrack PMIP growth-class distribution")
    print("=" * 95)

    display(track_growth_class_distribution)

else:

    track_growth_class_distribution = pd.DataFrame()


# ------------------------------------------------------------
# 8. PMIP priority-class distribution
# ------------------------------------------------------------

priority_column = None

for candidate_priority_column in [
    "pmip_priority_class",
    "priority_class",
]:
    if candidate_priority_column in track_findings_source_df.columns:
        priority_column = candidate_priority_column
        break


if priority_column is not None:

    track_priority_class_distribution = (
        track_findings_source_df[
            priority_column
        ]
        .value_counts(dropna=False)
        .rename_axis("PMIP Priority Class")
        .reset_index(name="Tracks")
    )

    track_priority_class_distribution["Share (%)"] = (
        track_priority_class_distribution["Tracks"]
        / len(track_findings_source_df)
        * 100
    )

    print("\nTrack PMIP priority-class distribution")
    print("=" * 95)

    display(track_priority_class_distribution)

else:

    track_priority_class_distribution = pd.DataFrame()


# ------------------------------------------------------------
# 9. Track-level findings overview
# ------------------------------------------------------------

track_overview_rows = [
    {
        "Metric": "Tracks assessed",
        "Observed Value": len(track_findings_source_df),
    }
]


if "pmip_growth_score" in track_findings_source_df.columns:

    track_growth_scores = pd.to_numeric(
        track_findings_source_df["pmip_growth_score"],
        errors="coerce",
    )

    track_overview_rows.extend(
        [
            {
                "Metric": "Median track PMIP growth score",
                "Observed Value": track_growth_scores.median(),
            },
            {
                "Metric": "Mean track PMIP growth score",
                "Observed Value": track_growth_scores.mean(),
            },
            {
                "Metric": "Highest track PMIP growth score",
                "Observed Value": track_growth_scores.max(),
            },
        ]
    )


if "growth_opportunity_score" in track_findings_source_df.columns:

    opportunity_scores = pd.to_numeric(
        track_findings_source_df["growth_opportunity_score"],
        errors="coerce",
    )

    track_overview_rows.extend(
        [
            {
                "Metric": "Median track growth-opportunity score",
                "Observed Value": opportunity_scores.median(),
            },
            {
                "Metric": "Highest track growth-opportunity score",
                "Observed Value": opportunity_scores.max(),
            },
        ]
    )


if "maximum_track_emerging_market_score" in track_findings_source_df.columns:

    emerging_scores = pd.to_numeric(
        track_findings_source_df[
            "maximum_track_emerging_market_score"
        ],
        errors="coerce",
    )

    track_overview_rows.extend(
        [
            {
                "Metric": "Median strongest emerging-market score",
                "Observed Value": emerging_scores.median(),
            },
            {
                "Metric": "Highest emerging-market score",
                "Observed Value": emerging_scores.max(),
            },
        ]
    )


track_level_findings_overview_df = pd.DataFrame(
    track_overview_rows
)


print("\nTrack-level findings overview")
print("=" * 95)

display(track_level_findings_overview_df)


# ------------------------------------------------------------
# 10. Diagnostics
# ------------------------------------------------------------

track_level_findings_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [
            "Tracks assessed",
            "Track identifiers available",
            "PMIP growth scores available",
            "PMIP growth classes available",
            "Growth opportunity intelligence available",
            "Emerging-market intelligence available",
            "Track ranking view prepared",
        ],
        "Observed Value": [
            len(track_findings_source_df),

            (
                track_findings_source_df["track_id"].notna().sum()
                if "track_id" in track_findings_source_df.columns
                else 0
            ),

            (
                track_findings_source_df["pmip_growth_score"].notna().sum()
                if "pmip_growth_score" in track_findings_source_df.columns
                else 0
            ),

            (
                track_findings_source_df["pmip_growth_class"].notna().sum()
                if "pmip_growth_class" in track_findings_source_df.columns
                else 0
            ),

            (
                track_findings_source_df[
                    "growth_opportunity_score"
                ].notna().sum()
                if "growth_opportunity_score"
                in track_findings_source_df.columns
                else 0
            ),

            (
                track_findings_source_df[
                    "maximum_track_emerging_market_score"
                ].notna().sum()
                if "maximum_track_emerging_market_score"
                in track_findings_source_df.columns
                else 0
            ),

            len(top_tracks_by_growth) > 0,
        ],
    }
)


print("\nTrack-level findings diagnostics")
print("=" * 95)

display(track_level_findings_diagnostics_df)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

track_count = len(track_findings_source_df)


track_identifier_complete = (
    "track_id" in track_findings_source_df.columns
    and track_findings_source_df["track_id"].notna().sum()
    == track_count
)


track_growth_score_complete = (
    "pmip_growth_score" in track_findings_source_df.columns
    and track_findings_source_df[
        "pmip_growth_score"
    ].notna().sum()
    == track_count
)


track_growth_class_complete = (
    "pmip_growth_class" in track_findings_source_df.columns
    and track_findings_source_df[
        "pmip_growth_class"
    ].notna().sum()
    == track_count
)


track_growth_score_bounds_valid = (
    "pmip_growth_score" in track_findings_source_df.columns
    and pd.to_numeric(
        track_findings_source_df["pmip_growth_score"],
        errors="coerce",
    )
    .between(0, 100)
    .all()
)


track_opportunity_available = (
    "growth_opportunity_score"
    in track_findings_source_df.columns
    and track_findings_source_df[
        "growth_opportunity_score"
    ].notna().sum()
    == track_count
)


track_emerging_available = (
    "maximum_track_emerging_market_score"
    in track_findings_source_df.columns
    and track_findings_source_df[
        "maximum_track_emerging_market_score"
    ].notna().sum()
    == track_count
)


track_level_findings_validation_df = pd.DataFrame(
    [
        {
            "Validation Area": "Section 11 completion",
            "Requirement":
                "Market Growth Scoring and Classification must already be complete.",
            "Observed Evidence":
                f"Section 11 overall completion status: {section_11_complete}",
            "Passed": bool(section_11_complete),
        },
        {
            "Validation Area": "Track population availability",
            "Requirement":
                "The retained track population must remain available for interpretation.",
            "Observed Evidence":
                f"{track_count:,} tracks available",
            "Passed": track_count > 0,
        },
        {
            "Validation Area": "Track identifier completeness",
            "Requirement":
                "Every retained track must retain a valid track identifier.",
            "Observed Evidence":
                (
                    f"{track_findings_source_df['track_id'].notna().sum():,} "
                    "track identifiers available"
                    if "track_id" in track_findings_source_df.columns
                    else "Track identifier column unavailable"
                ),
            "Passed": bool(track_identifier_complete),
        },
        {
            "Validation Area": "Track growth intelligence availability",
            "Requirement":
                "Every retained track must retain a final PMIP growth score.",
            "Observed Evidence":
                (
                    f"{track_findings_source_df['pmip_growth_score'].notna().sum():,} "
                    "track PMIP growth scores available"
                    if "pmip_growth_score" in track_findings_source_df.columns
                    else "PMIP growth score unavailable"
                ),
            "Passed": bool(track_growth_score_complete),
        },
        {
            "Validation Area": "Track growth score bounds",
            "Requirement":
                "Track PMIP growth scores must remain between 0 and 100.",
            "Observed Evidence":
                "Track PMIP growth scores remain within valid 0–100 bounds",
            "Passed": bool(track_growth_score_bounds_valid),
        },
        {
            "Validation Area": "Growth opportunity intelligence availability",
            "Requirement":
                "Track growth-opportunity evidence must remain available.",
            "Observed Evidence":
                (
                    f"{track_findings_source_df['growth_opportunity_score'].notna().sum():,} "
                    "track growth-opportunity scores available"
                    if "growth_opportunity_score"
                    in track_findings_source_df.columns
                    else "Growth-opportunity intelligence unavailable"
                ),
            "Passed": bool(track_opportunity_available),
        },
        {
            "Validation Area": "Emerging-market intelligence availability",
            "Requirement":
                "Track emerging-market evidence must remain available.",
            "Observed Evidence":
                (
                    f"{track_findings_source_df['maximum_track_emerging_market_score'].notna().sum():,} "
                    "track emerging-market scores available"
                    if "maximum_track_emerging_market_score"
                    in track_findings_source_df.columns
                    else "Emerging-market intelligence unavailable"
                ),
            "Passed": bool(track_emerging_available),
        },
    ]
)


print("\nTrack-level findings validation")
print("=" * 95)

display(track_level_findings_validation_df)


# ------------------------------------------------------------
# 12. Completion status
# ------------------------------------------------------------

section_12_3_complete = bool(
    track_level_findings_validation_df["Passed"].all()
)


print(
    f"\nSection 12.3 completion status: "
    f"{section_12_3_complete}"
)


if section_12_3_complete:

    print(
        "PMIP track-level findings have been summarised successfully "
        "from validated growth, opportunity, emerging-market and "
        "ranking intelligence."
    )
```

## 12.4 Growth Comparisons

### Purpose

This subsection compares the final PMIP growth intelligence across the country, artist, and track levels.

The aim is to determine whether growth strength appears consistently across the different levels of analysis and to identify important differences between broad market conditions and individual artist or track performance.

The comparison focuses on:

- overall PMIP growth-score distributions for tracks and artists;
- differences between track-level and artist-level growth classes;
- differences in growth-opportunity strength;
- differences in emerging-market evidence;
- the relationship between country-level market strength and the final artist and track growth results;

### Interpretation

The cross-level comparison shows that PMIP is producing consistent but clearly differentiated growth intelligence across country, track, and artist levels.

At country level, the median growth score is approximately **50.99**, which is substantially higher than the median track score of **13.12** and the median artist score of **19.93**. This reflects the fact that country-level scores summarise wider market conditions, while individual tracks and artists are assessed more selectively based on their own growth evidence.

Artist-level growth appears stronger than track-level growth overall. Approximately **10.49% of artists** fall within the High Growth or Exceptional Growth categories, compared with approximately **9.94% of tracks**. Artists also show a slightly larger Exceptional Growth share, at approximately **0.59%**, compared with **0.48%** for tracks.

The class distributions reveal an important difference between the two populations. Tracks contain a much larger Low Growth share at approximately **31.42%**, while only **11.38% of artists** fall within this class. In contrast, artists are more concentrated within Developing Growth and Moderate Growth. This suggests that artist-level aggregation can capture broader momentum across multiple tracks even when individual track performance remains uneven.

Growth-opportunity results show a similar pattern. Artists have a higher median opportunity score of approximately **15.44**, compared with **9.62** for tracks. Country-level market opportunity is substantially higher, with a median score of approximately **38.70**, reflecting broader market-level conditions rather than individual performance.

Emerging-market evidence is also stronger at artist level than at track level. The median artist emerging-market score is approximately **33.53**, compared with **15.97** for tracks. This indicates that artists may accumulate emerging-market strength across several releases or territories even when a single track has more limited international evidence.

The maximum scores show that individual tracks and artists can outperform the overall strength of their country environments. Maximum PMIP growth scores exceed **80** for both tracks and artists, while the highest country growth score is approximately **65.90**. This demonstrates that strong individual opportunities can exist even when broader market conditions are less extreme.

Priority distributions remain selective. Approximately **88.08% of tracks** and **86.47% of artists** remain within the Low Priority category, while only small proportions reach High or Critical Priority. This is desirable because the PMIP priority framework is intended to isolate a relatively small number of high-confidence growth signals rather than classify a large proportion of the population as urgent opportunities.

Overall, the results show that country, artist, and track intelligence complement one another rather than producing identical conclusions. Country analysis provides the broader market context, artist analysis captures aggregated momentum, and track analysis identifies more specific release-level growth signals. This supports the use of PMIP as a multi-level music intelligence framework.

### 12.5 High-Priority Market Growth Insights

#### Purpose

The purpose of this subsection is to identify and summarise the strongest PMIP growth signals across countries, tracks, and artists.

This step brings together the validated growth scores, opportunity scores, emerging-market strength, ranking information, and PMIP priority classes developed in the previous sections.

The analysis will focus on:

- identifying the highest-priority track and artist opportunities;
- comparing Critical, High, Elevated, Developing, and Low Priority populations;
- examining the strongest country-market opportunities supporting these signals;
- identifying cases where strong growth, opportunity, and emerging-market evidence occur together;
- summarising the most important actionable growth intelligence produced by PMIP; and
- validating that the final high-priority insight views remain consistent with the completed PMIP scoring and ranking framework.

This subsection provides the final interpreted intelligence layer of Section 12 before the overall results are consolidated.

```python
# ============================================================
# SECTION 12.5 - HIGH-PRIORITY MARKET GROWTH INSIGHTS
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display


print("Summarising PMIP high-priority market growth insights")
print("=" * 105)

section_12_4_complete = bool(
    globals().get("section_12_4_complete", False)
    or globals().get("section_12_4_completed", False)
)

print(f"Section 12.4 completion status: {section_12_4_complete}")


# ------------------------------------------------------------
# 1. Validate required objects
# ------------------------------------------------------------

required_objects = {
    "track_priority_df": globals().get("track_priority_df"),
    "artist_priority_df": globals().get("artist_priority_df"),
    "emerging_market_df": globals().get("emerging_market_df"),
}

missing_objects = [
    name
    for name, obj in required_objects.items()
    if not isinstance(obj, pd.DataFrame)
]

if missing_objects:
    raise NameError(
        "Missing required Section 12.5 DataFrames: "
        + ", ".join(missing_objects)
    )


track_insight_source = track_priority_df.copy()
artist_insight_source = artist_priority_df.copy()
country_insight_source = emerging_market_df.copy()


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def first_existing_column(df, candidates):
    for col in candidates:
        if col in df.columns:
            return col
    return None


def safe_numeric(series):
    return pd.to_numeric(series, errors="coerce")


def print_heading(text):
    print(f"\n{text}")
    print("=" * 105)


# ------------------------------------------------------------
# 3. Identify important columns
# ------------------------------------------------------------

track_priority_col = first_existing_column(
    track_insight_source,
    ["pmip_priority_class", "priority_class", "PMIP Priority Class"]
)

artist_priority_col = first_existing_column(
    artist_insight_source,
    ["pmip_priority_class", "priority_class", "PMIP Priority Class"]
)

track_priority_score_col = first_existing_column(
    track_insight_source,
    ["pmip_priority_score", "priority_score"]
)

artist_priority_score_col = first_existing_column(
    artist_insight_source,
    ["pmip_priority_score", "priority_score"]
)

track_rank_col = first_existing_column(
    track_insight_source,
    ["pmip_growth_rank", "pmip_priority_rank"]
)

artist_rank_col = first_existing_column(
    artist_insight_source,
    ["pmip_growth_rank", "pmip_priority_rank"]
)

country_rank_col = first_existing_column(
    country_insight_source,
    ["emerging_market_rank", "market_opportunity_rank", "country_growth_rank"]
)


# ------------------------------------------------------------
# 4. High-priority track population
# ------------------------------------------------------------

priority_order = [
    "Critical Priority",
    "High Priority",
    "Elevated Priority",
    "Developing Priority",
    "Low Priority",
]

if track_priority_col is not None:

    track_high_priority_df = track_insight_source[
        track_insight_source[track_priority_col].isin(
            ["Critical Priority", "High Priority", "Elevated Priority"]
        )
    ].copy()

else:
    track_high_priority_df = track_insight_source.copy()


if artist_priority_col is not None:

    artist_high_priority_df = artist_insight_source[
        artist_insight_source[artist_priority_col].isin(
            ["Critical Priority", "High Priority", "Elevated Priority"]
        )
    ].copy()

else:
    artist_high_priority_df = artist_insight_source.copy()


# ------------------------------------------------------------
# 5. Rank strongest high-priority track signals
# ------------------------------------------------------------

track_sort_cols = [
    col for col in [
        track_priority_score_col,
        "pmip_growth_score",
        "growth_opportunity_score",
        "maximum_track_emerging_market_score",
        "mean_market_growth_score",
    ]
    if col is not None and col in track_high_priority_df.columns
]

if track_sort_cols:
    track_high_priority_df = track_high_priority_df.sort_values(
        track_sort_cols,
        ascending=[False] * len(track_sort_cols)
    )

top_track_priority_insights_df = track_high_priority_df.head(20).copy()


track_display_cols = [
    col for col in [
        "track_id",
        "countries_represented",
        "pmip_growth_score",
        "pmip_growth_class",
        track_rank_col,
        track_priority_score_col,
        track_priority_col,
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
    ]
    if col is not None and col in top_track_priority_insights_df.columns
]

print_heading("Top high-priority PMIP track growth signals")
display(top_track_priority_insights_df[track_display_cols])


# ------------------------------------------------------------
# 6. Rank strongest high-priority artist signals
# ------------------------------------------------------------

artist_sort_cols = [
    col for col in [
        artist_priority_score_col,
        "pmip_growth_score",
        "growth_opportunity_score",
        "maximum_artist_emerging_market_score",
        "mean_market_growth_score",
    ]
    if col is not None and col in artist_high_priority_df.columns
]

if artist_sort_cols:
    artist_high_priority_df = artist_high_priority_df.sort_values(
        artist_sort_cols,
        ascending=[False] * len(artist_sort_cols)
    )

top_artist_priority_insights_df = artist_high_priority_df.head(20).copy()


artist_display_cols = [
    col for col in [
        "artists",
        "countries_represented",
        "pmip_growth_score",
        "pmip_growth_class",
        artist_rank_col,
        artist_priority_score_col,
        artist_priority_col,
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
    ]
    if col is not None and col in top_artist_priority_insights_df.columns
]

print_heading("Top high-priority PMIP artist growth signals")
display(top_artist_priority_insights_df[artist_display_cols])


# ------------------------------------------------------------
# 7. Strongest supporting country-market opportunities
# ------------------------------------------------------------

country_sort_cols = [
    col for col in [
        "emerging_market_score",
        "market_opportunity_score",
        "combined_country_growth_score",
    ]
    if col in country_insight_source.columns
]

if country_sort_cols:
    top_country_market_insights_df = (
        country_insight_source
        .sort_values(
            country_sort_cols,
            ascending=[False] * len(country_sort_cols)
        )
        .head(20)
        .copy()
    )
else:
    top_country_market_insights_df = country_insight_source.head(20).copy()


country_display_cols = [
    col for col in [
        "country",
        "combined_country_growth_score",
        "country_growth_class",
        "country_growth_rank",
        "market_opportunity_score",
        "market_opportunity_class",
        "market_opportunity_rank",
        "emerging_market_score",
        "emerging_market_class",
        country_rank_col,
        "track_emerging_or_better_share",
        "artist_emerging_or_better_share",
    ]
    if col is not None and col in top_country_market_insights_df.columns
]

print_heading("Top supporting PMIP country-market growth signals")
display(top_country_market_insights_df[country_display_cols])


# ------------------------------------------------------------
# 8. Priority-class distributions
# ------------------------------------------------------------

if track_priority_col is not None:

    track_priority_distribution_df = (
        track_insight_source[track_priority_col]
        .value_counts()
        .reindex(priority_order)
        .dropna()
        .reset_index()
    )

    track_priority_distribution_df.columns = [
        "PMIP Priority Class",
        "Tracks"
    ]

    track_priority_distribution_df["Share (%)"] = (
        track_priority_distribution_df["Tracks"]
        / len(track_insight_source)
        * 100
    )

    print_heading("Track high-priority signal distribution")
    display(track_priority_distribution_df)


if artist_priority_col is not None:

    artist_priority_distribution_df = (
        artist_insight_source[artist_priority_col]
        .value_counts()
        .reindex(priority_order)
        .dropna()
        .reset_index()
    )

    artist_priority_distribution_df.columns = [
        "PMIP Priority Class",
        "Artists"
    ]

    artist_priority_distribution_df["Share (%)"] = (
        artist_priority_distribution_df["Artists"]
        / len(artist_insight_source)
        * 100
    )

    print_heading("Artist high-priority signal distribution")
    display(artist_priority_distribution_df)


# ------------------------------------------------------------
# 9. Multi-signal high-priority evidence
# ------------------------------------------------------------

track_multi_signal_columns = [
    col for col in [
        "pmip_growth_score",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        track_priority_score_col,
    ]
    if col is not None and col in track_insight_source.columns
]

artist_multi_signal_columns = [
    col for col in [
        "pmip_growth_score",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
        artist_priority_score_col,
    ]
    if col is not None and col in artist_insight_source.columns
]


track_thresholds = {}
for col in track_multi_signal_columns:
    track_thresholds[col] = safe_numeric(
        track_insight_source[col]
    ).quantile(0.90)


artist_thresholds = {}
for col in artist_multi_signal_columns:
    artist_thresholds[col] = safe_numeric(
        artist_insight_source[col]
    ).quantile(0.90)


if track_multi_signal_columns:

    track_signal_flags = pd.DataFrame(index=track_insight_source.index)

    for col in track_multi_signal_columns:
        track_signal_flags[col] = (
            safe_numeric(track_insight_source[col])
            >= track_thresholds[col]
        )

    track_insight_source[
        "high_signal_component_count"
    ] = track_signal_flags.sum(axis=1)

    track_multi_signal_insights_df = (
        track_insight_source
        .sort_values(
            [
                "high_signal_component_count",
                "pmip_growth_score"
            ],
            ascending=[False, False]
        )
        .head(20)
        .copy()
    )

    print_heading("Strongest multi-signal track opportunities")

    display_cols = [
        col for col in [
            "track_id",
            "high_signal_component_count",
            "pmip_growth_score",
            track_priority_col,
            "growth_opportunity_score",
            "mean_market_growth_score",
            "maximum_track_emerging_market_score",
        ]
        if col is not None and col in track_multi_signal_insights_df.columns
    ]

    display(track_multi_signal_insights_df[display_cols])


if artist_multi_signal_columns:

    artist_signal_flags = pd.DataFrame(index=artist_insight_source.index)

    for col in artist_multi_signal_columns:
        artist_signal_flags[col] = (
            safe_numeric(artist_insight_source[col])
            >= artist_thresholds[col]
        )

    artist_insight_source[
        "high_signal_component_count"
    ] = artist_signal_flags.sum(axis=1)

    artist_multi_signal_insights_df = (
        artist_insight_source
        .sort_values(
            [
                "high_signal_component_count",
                "pmip_growth_score"
            ],
            ascending=[False, False]
        )
        .head(20)
        .copy()
    )

    print_heading("Strongest multi-signal artist opportunities")

    display_cols = [
        col for col in [
            "artists",
            "high_signal_component_count",
            "pmip_growth_score",
            artist_priority_col,
            "growth_opportunity_score",
            "mean_market_growth_score",
            "maximum_artist_emerging_market_score",
        ]
        if col is not None and col in artist_multi_signal_insights_df.columns
    ]

    display(artist_multi_signal_insights_df[display_cols])


# ------------------------------------------------------------
# 10. High-priority insight overview
# ------------------------------------------------------------

track_high_or_critical = (
    track_insight_source[track_priority_col].isin(
        ["High Priority", "Critical Priority"]
    ).sum()
    if track_priority_col is not None
    else np.nan
)

artist_high_or_critical = (
    artist_insight_source[artist_priority_col].isin(
        ["High Priority", "Critical Priority"]
    ).sum()
    if artist_priority_col is not None
    else np.nan
)

track_critical = (
    (track_insight_source[track_priority_col] == "Critical Priority").sum()
    if track_priority_col is not None
    else np.nan
)

artist_critical = (
    (artist_insight_source[artist_priority_col] == "Critical Priority").sum()
    if artist_priority_col is not None
    else np.nan
)

country_high_emerging = (
    country_insight_source["emerging_market_class"]
    .isin(["High Emerging Potential", "Strong Emerging Market"])
    .sum()
    if "emerging_market_class" in country_insight_source.columns
    else np.nan
)


high_priority_overview_df = pd.DataFrame({
    "Metric": [
        "Tracks assessed",
        "Tracks with High-or-Critical Priority",
        "Critical-priority tracks",
        "Artists assessed",
        "Artists with High-or-Critical Priority",
        "Critical-priority artists",
        "Country markets assessed",
        "High emerging-potential country markets",
    ],
    "Observed Value": [
        len(track_insight_source),
        track_high_or_critical,
        track_critical,
        len(artist_insight_source),
        artist_high_or_critical,
        artist_critical,
        len(country_insight_source),
        country_high_emerging,
    ]
})

print_heading("High-priority market growth insight overview")
display(high_priority_overview_df)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

high_priority_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Track population available",
        "Track priority classes available",
        "Track PMIP growth scores available",
        "Artist population available",
        "Artist priority classes available",
        "Artist PMIP growth scores available",
        "Country market intelligence available",
        "Cross-level priority insight views prepared",
    ],
    "Observed Value": [
        len(track_insight_source),
        track_priority_col is not None,
        "pmip_growth_score" in track_insight_source.columns,
        len(artist_insight_source),
        artist_priority_col is not None,
        "pmip_growth_score" in artist_insight_source.columns,
        len(country_insight_source),
        True,
    ]
})

print_heading("High-priority market growth insight diagnostics")
display(high_priority_diagnostics_df)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

track_priority_complete = (
    track_priority_col is not None
    and track_insight_source[track_priority_col].notna().all()
)

artist_priority_complete = (
    artist_priority_col is not None
    and artist_insight_source[artist_priority_col].notna().all()
)

track_growth_complete = (
    "pmip_growth_score" in track_insight_source.columns
    and track_insight_source["pmip_growth_score"].notna().all()
)

artist_growth_complete = (
    "pmip_growth_score" in artist_insight_source.columns
    and artist_insight_source["pmip_growth_score"].notna().all()
)

country_intelligence_complete = (
    "country" in country_insight_source.columns
    and country_insight_source["country"].notna().all()
)

multi_signal_views_complete = (
    "high_signal_component_count" in track_insight_source.columns
    and "high_signal_component_count" in artist_insight_source.columns
)


high_priority_validation_df = pd.DataFrame({
    "Validation Area": [
        "Section 12.4 completion",
        "Track priority intelligence completeness",
        "Artist priority intelligence completeness",
        "Track growth intelligence completeness",
        "Artist growth intelligence completeness",
        "Country market intelligence availability",
        "Multi-signal high-priority analysis",
    ],
    "Requirement": [
        "Growth Comparisons must already be complete.",
        "Every retained track must retain a PMIP priority classification.",
        "Every retained artist must retain a PMIP priority classification.",
        "Every retained track must retain its validated PMIP growth score.",
        "Every retained artist must retain its validated PMIP growth score.",
        "Validated country-market intelligence must remain available.",
        "High-priority insights must combine multiple growth-intelligence signals.",
    ],
    "Observed Evidence": [
        f"Section 12.4 completion status: {section_12_4_complete}",
        f"{track_insight_source[track_priority_col].notna().sum():,} track priority classes available"
        if track_priority_col is not None else "Track priority class unavailable",
        f"{artist_insight_source[artist_priority_col].notna().sum():,} artist priority classes available"
        if artist_priority_col is not None else "Artist priority class unavailable",
        f"{track_insight_source['pmip_growth_score'].notna().sum():,} track PMIP growth scores available"
        if "pmip_growth_score" in track_insight_source.columns else "Track growth score unavailable",
        f"{artist_insight_source['pmip_growth_score'].notna().sum():,} artist PMIP growth scores available"
        if "pmip_growth_score" in artist_insight_source.columns else "Artist growth score unavailable",
        f"{len(country_insight_source):,} country markets available",
        "Growth score, opportunity, market growth, emerging-market and priority evidence combined",
    ],
    "Passed": [
        section_12_4_complete,
        track_priority_complete,
        artist_priority_complete,
        track_growth_complete,
        artist_growth_complete,
        country_intelligence_complete,
        multi_signal_views_complete,
    ]
})

print_heading("High-priority market growth insight validation")
display(high_priority_validation_df)


# ------------------------------------------------------------
# 13. Completion status
# ------------------------------------------------------------

section_12_5_complete = bool(
    high_priority_validation_df["Passed"].all()
)

section_12_complete = section_12_5_complete

print(f"\nSection 12.5 completion status: {section_12_5_complete}")
print(f"Section 12 overall completion status: {section_12_complete}")

if section_12_5_complete:
    print(
        "PMIP high-priority market growth insights have been summarised "
        "successfully across country, track and artist intelligence."
    )
else:
    print(
        "Section 12.5 requires review because one or more validation "
        "requirements were not satisfied."
    )
```

## 13. Responsible Use and Limitations

### Purpose

This section documents the main limitations, interpretation boundaries, and responsible-use requirements of the PMIP market-growth intelligence framework.

The aim is to make clear that PMIP outputs should be treated as decision-support evidence rather than absolute predictions or automated commercial decisions.

The section evaluates five important areas:

- **13.1 Temporal Limitations** — how the available observation period affects growth interpretation.
- **13.2 Data-Coverage Limitations** — where missing, sparse, or uneven data may reduce confidence.
- **13.3 Cross-Country Comparison Limitations** — why country-level scores may not always be perfectly comparable.
- **13.4 Growth Interpretation Boundaries** — what PMIP growth scores can and cannot be interpreted to mean.
- **13.5 Human Review Requirements** — why A&R, marketing, and commercial decisions should still include expert human judgement.

The objective is to ensure that PMIP is used responsibly, transparently, and within the boundaries supported by the available data and modelling approach.

### 13.1 Temporal Limitations

#### Purpose

This subsection examines the limitations created by the time period covered by the PMIP datasets.

Growth signals may change depending on how much historical data is available, how frequently observations are recorded, and whether recent behaviour differs from earlier periods.

The aim is to verify the temporal coverage available in the retained track-level and artist-level growth histories and to identify where short observation windows may make growth conclusions less reliable.

```python
# ============================================================
# Section 13.1 - Temporal Limitations
# ============================================================

import pandas as pd
import numpy as np

print("Assessing PMIP temporal limitations")
print("=" * 90)


# ------------------------------------------------------------
# 1. Section 12 dependency
# ------------------------------------------------------------

section_12_status = bool(
    globals().get("section_12_complete", True)
)

print(f"Section 12 overall completion status: {section_12_status}")


# ------------------------------------------------------------
# 2. Find available temporal DataFrames
# ------------------------------------------------------------

track_candidate_names = [
    "track_country_date_df",
    "track_country_date_history_df",
    "track_date_df",
    "track_date_history_df",
    "track_country_history_df"
]

artist_candidate_names = [
    "artist_country_date_df",
    "artist_country_date_history_df",
    "artist_date_df",
    "artist_date_history_df",
    "artist_country_history_df"
]


def find_existing_dataframes(candidate_names):
    found = []

    for name in candidate_names:
        if (
            name in globals()
            and isinstance(globals()[name], pd.DataFrame)
            and not globals()[name].empty
        ):
            found.append((name, globals()[name]))

    return found


track_candidates = find_existing_dataframes(track_candidate_names)
artist_candidates = find_existing_dataframes(artist_candidate_names)

print("\nAvailable track temporal candidates")
print("=" * 90)

if track_candidates:
    for name, df in track_candidates:
        print(f"{name}: {df.shape}")
else:
    print("No candidate track temporal DataFrames found.")

print("\nAvailable artist temporal candidates")
print("=" * 90)

if artist_candidates:
    for name, df in artist_candidates:
        print(f"{name}: {df.shape}")
else:
    print("No candidate artist temporal DataFrames found.")


# ------------------------------------------------------------
# 3. Detect genuine calendar-date columns
# ------------------------------------------------------------

def find_real_date_column(df):
    """
    Detect a genuine calendar-date column.

    Numeric count fields such as:
        unique_dates
        date_count
        observations
    are deliberately excluded.
    """

    excluded_count_words = [
        "unique_dates",
        "date_count",
        "dates_count",
        "number_of_dates",
        "num_dates",
        "observation_count",
        "observations"
    ]

    preferred_names = [
        "date",
        "chart_date",
        "snapshot_date",
        "observation_date",
        "week_date",
        "month_date",
        "timestamp",
        "datetime"
    ]

    # First try exact preferred names
    for col in preferred_names:
        if col in df.columns:

            series = df[col]

            # Already datetime
            if pd.api.types.is_datetime64_any_dtype(series):
                return col

            # Only attempt conversion for non-numeric data
            if not pd.api.types.is_numeric_dtype(series):

                converted = pd.to_datetime(
                    series,
                    errors="coerce"
                )

                success_rate = converted.notna().mean()

                if success_rate >= 0.50:
                    return col

    # Then inspect other date-like column names
    for col in df.columns:

        col_lower = str(col).lower()

        if col_lower in excluded_count_words:
            continue

        if not any(
            word in col_lower
            for word in ["date", "time", "timestamp"]
        ):
            continue

        series = df[col]

        # Do NOT convert plain numeric columns into timestamps
        if pd.api.types.is_numeric_dtype(series):
            continue

        converted = pd.to_datetime(
            series,
            errors="coerce"
        )

        success_rate = converted.notna().mean()

        if success_rate >= 0.50:
            return col

    return None


def select_temporal_source(candidates):
    """
    Prefer a DataFrame containing an actual calendar date.
    """

    for name, df in candidates:

        date_col = find_real_date_column(df)

        if date_col is not None:
            return name, df.copy(), date_col

    return None, None, None


track_date_source_name, track_date_df, track_date_col = (
    select_temporal_source(track_candidates)
)

artist_date_source_name, artist_date_df, artist_date_col = (
    select_temporal_source(artist_candidates)
)


print("\nDetected genuine temporal sources")
print("=" * 90)

print(f"Track date source:  {track_date_source_name}")
print(f"Track date column:  {track_date_col}")

print(f"Artist date source: {artist_date_source_name}")
print(f"Artist date column: {artist_date_col}")


# ------------------------------------------------------------
# 4. Find identifier columns
# ------------------------------------------------------------

def detect_identifier(df, candidates):

    if df is None:
        return None

    for col in candidates:
        if col in df.columns:
            return col

    return None


track_id_col = detect_identifier(
    track_date_df,
    ["track_id", "track", "id"]
)

artist_id_col = detect_identifier(
    artist_date_df,
    ["artists", "artist", "artist_id"]
)

print(f"\nDetected track identifier:  {track_id_col}")
print(f"Detected artist identifier: {artist_id_col}")


# ------------------------------------------------------------
# 5. Genuine track calendar coverage
# ------------------------------------------------------------

track_min_date = pd.NaT
track_max_date = pd.NaT
track_unique_calendar_dates = np.nan

track_calendar_available = bool(
    track_date_df is not None
    and track_date_col is not None
)

if track_calendar_available:

    track_date_df[track_date_col] = pd.to_datetime(
        track_date_df[track_date_col],
        errors="coerce"
    )

    valid_track_dates = (
        track_date_df[track_date_col]
        .dropna()
    )

    if not valid_track_dates.empty:

        track_min_date = valid_track_dates.min()
        track_max_date = valid_track_dates.max()
        track_unique_calendar_dates = valid_track_dates.nunique()

print("\nTrack calendar coverage")
print("=" * 90)

if track_calendar_available and pd.notna(track_min_date):

    print(f"Earliest track observation: {track_min_date}")
    print(f"Latest track observation:   {track_max_date}")
    print(
        f"Unique track calendar dates: "
        f"{int(track_unique_calendar_dates):,}"
    )

else:

    print(
        "No genuine track calendar-date field was available "
        "in the retained temporal sources."
    )

    print(
        "Track temporal limitations will therefore be assessed "
        "using available history-length evidence instead."
    )


# ------------------------------------------------------------
# 6. Genuine artist calendar coverage
# ------------------------------------------------------------

artist_min_date = pd.NaT
artist_max_date = pd.NaT
artist_unique_calendar_dates = np.nan

artist_calendar_available = bool(
    artist_date_df is not None
    and artist_date_col is not None
)

if artist_calendar_available:

    artist_date_df[artist_date_col] = pd.to_datetime(
        artist_date_df[artist_date_col],
        errors="coerce"
    )

    valid_artist_dates = (
        artist_date_df[artist_date_col]
        .dropna()
    )

    if not valid_artist_dates.empty:

        artist_min_date = valid_artist_dates.min()
        artist_max_date = valid_artist_dates.max()
        artist_unique_calendar_dates = (
            valid_artist_dates.nunique()
        )

print("\nArtist calendar coverage")
print("=" * 90)

if artist_calendar_available and pd.notna(artist_min_date):

    print(f"Earliest artist observation: {artist_min_date}")
    print(f"Latest artist observation:   {artist_max_date}")
    print(
        f"Unique artist calendar dates: "
        f"{int(artist_unique_calendar_dates):,}"
    )

else:

    print(
        "No genuine artist calendar-date field was available "
        "in the retained temporal sources."
    )


# ------------------------------------------------------------
# 7. Track history-length evidence
# ------------------------------------------------------------

track_history_source_name = None
track_history_lengths = None


# Prefer an existing summary field such as unique_dates
for name, df in track_candidates:

    if "unique_dates" in df.columns:

        possible_track_id = detect_identifier(
            df,
            ["track_id", "track", "id"]
        )

        if possible_track_id is not None:

            track_history_source_name = name

            track_history_lengths = (
                df[
                    [possible_track_id, "unique_dates"]
                ]
                .dropna()
                .groupby(possible_track_id)["unique_dates"]
                .max()
            )

            track_history_lengths = pd.to_numeric(
                track_history_lengths,
                errors="coerce"
            ).dropna()

            break


# Fallback: calculate from genuine dates
if (
    track_history_lengths is None
    and track_date_df is not None
    and track_id_col is not None
    and track_date_col is not None
):

    track_history_source_name = track_date_source_name

    track_history_lengths = (
        track_date_df
        .dropna(
            subset=[
                track_id_col,
                track_date_col
            ]
        )
        .groupby(track_id_col)[track_date_col]
        .nunique()
    )


if track_history_lengths is None:
    track_history_lengths = pd.Series(dtype=float)


print("\nTrack history-length source")
print("=" * 90)
print(track_history_source_name)

print("\nTrack history-length summary")
print("=" * 90)

if len(track_history_lengths) > 0:

    display(
        track_history_lengths
        .describe()
        .to_frame("Observed Dates")
    )

else:

    print("Track history-length evidence unavailable.")


# ------------------------------------------------------------
# 8. Artist history-length evidence
# ------------------------------------------------------------

artist_history_source_name = None
artist_history_lengths = None


# Prefer direct date history because genuine artist dates exist
if (
    artist_date_df is not None
    and artist_id_col is not None
    and artist_date_col is not None
):

    artist_history_source_name = artist_date_source_name

    artist_history_lengths = (
        artist_date_df
        .dropna(
            subset=[
                artist_id_col,
                artist_date_col
            ]
        )
        .groupby(artist_id_col)[artist_date_col]
        .nunique()
    )


# Fallback to unique_dates
if artist_history_lengths is None:

    for name, df in artist_candidates:

        if "unique_dates" in df.columns:

            possible_artist_id = detect_identifier(
                df,
                ["artists", "artist", "artist_id"]
            )

            if possible_artist_id is not None:

                artist_history_source_name = name

                artist_history_lengths = (
                    df[
                        [
                            possible_artist_id,
                            "unique_dates"
                        ]
                    ]
                    .dropna()
                    .groupby(
                        possible_artist_id
                    )["unique_dates"]
                    .max()
                )

                artist_history_lengths = pd.to_numeric(
                    artist_history_lengths,
                    errors="coerce"
                ).dropna()

                break


if artist_history_lengths is None:
    artist_history_lengths = pd.Series(dtype=float)


print("\nArtist history-length source")
print("=" * 90)
print(artist_history_source_name)

print("\nArtist history-length summary")
print("=" * 90)

if len(artist_history_lengths) > 0:

    display(
        artist_history_lengths
        .describe()
        .to_frame("Observed Dates")
    )

else:

    print("Artist history-length evidence unavailable.")


# ------------------------------------------------------------
# 9. Short-history assessment
# ------------------------------------------------------------

short_history_threshold = 3


def calculate_short_history(series, threshold):

    if len(series) == 0:
        return 0, np.nan

    count = int(
        (series <= threshold).sum()
    )

    share = (
        count / len(series)
    ) * 100

    return count, share


track_short_count, track_short_share = (
    calculate_short_history(
        track_history_lengths,
        short_history_threshold
    )
)

artist_short_count, artist_short_share = (
    calculate_short_history(
        artist_history_lengths,
        short_history_threshold
    )
)


# ------------------------------------------------------------
# 10. Temporal limitation overview
# ------------------------------------------------------------

def safe_date_string(value):

    if pd.isna(value):
        return "Not available"

    return str(value)


def safe_numeric(series, statistic):

    if len(series) == 0:
        return np.nan

    if statistic == "median":
        return float(series.median())

    if statistic == "mean":
        return float(series.mean())

    if statistic == "min":
        return float(series.min())

    if statistic == "max":
        return float(series.max())

    return np.nan


temporal_limitation_overview_df = pd.DataFrame({

    "Metric": [

        "Track calendar-date coverage available",
        "Earliest track observation",
        "Latest track observation",
        "Unique track calendar dates",

        "Track histories assessed",
        "Mean track history length",
        "Median track history length",
        "Maximum track history length",
        f"Tracks with <= {short_history_threshold} observations",
        f"Tracks with <= {short_history_threshold} observations (%)",

        "Artist calendar-date coverage available",
        "Earliest artist observation",
        "Latest artist observation",
        "Unique artist calendar dates",

        "Artist histories assessed",
        "Mean artist history length",
        "Median artist history length",
        "Maximum artist history length",
        f"Artists with <= {short_history_threshold} observations",
        f"Artists with <= {short_history_threshold} observations (%)"
    ],

    "Observed Value": [

        track_calendar_available,
        safe_date_string(track_min_date),
        safe_date_string(track_max_date),
        (
            int(track_unique_calendar_dates)
            if pd.notna(track_unique_calendar_dates)
            else "Not available"
        ),

        len(track_history_lengths),
        safe_numeric(
            track_history_lengths,
            "mean"
        ),
        safe_numeric(
            track_history_lengths,
            "median"
        ),
        safe_numeric(
            track_history_lengths,
            "max"
        ),
        track_short_count,
        track_short_share,

        artist_calendar_available,
        safe_date_string(artist_min_date),
        safe_date_string(artist_max_date),
        (
            int(artist_unique_calendar_dates)
            if pd.notna(artist_unique_calendar_dates)
            else "Not available"
        ),

        len(artist_history_lengths),
        safe_numeric(
            artist_history_lengths,
            "mean"
        ),
        safe_numeric(
            artist_history_lengths,
            "median"
        ),
        safe_numeric(
            artist_history_lengths,
            "max"
        ),
        artist_short_count,
        artist_short_share
    ]
})


print("\nTemporal limitation overview")
print("=" * 90)
display(temporal_limitation_overview_df)


# ------------------------------------------------------------
# 11. Explicit limitation flags
# ------------------------------------------------------------

track_short_history_risk = bool(
    pd.notna(track_short_share)
    and track_short_share > 25
)

artist_short_history_risk = bool(
    pd.notna(artist_short_share)
    and artist_short_share > 25
)

track_calendar_limitation = (
    not track_calendar_available
)

artist_calendar_limitation = (
    not artist_calendar_available
)


temporal_limitation_flags_df = pd.DataFrame({

    "Temporal Limitation": [

        "Track calendar coverage unavailable",
        "Artist calendar coverage unavailable",
        "Track short-history concentration",
        "Artist short-history concentration"
    ],

    "Observed Evidence": [

        (
            "No genuine track calendar-date field retained"
            if track_calendar_limitation
            else (
                f"Track calendar coverage runs from "
                f"{track_min_date.date()} to "
                f"{track_max_date.date()}"
            )
        ),

        (
            "No genuine artist calendar-date field retained"
            if artist_calendar_limitation
            else (
                f"Artist calendar coverage runs from "
                f"{artist_min_date.date()} to "
                f"{artist_max_date.date()}"
            )
        ),

        (
            f"{track_short_share:.2f}% of assessed track histories "
            f"contain <= {short_history_threshold} observations"
            if pd.notna(track_short_share)
            else "Track short-history share unavailable"
        ),

        (
            f"{artist_short_share:.2f}% of assessed artist histories "
            f"contain <= {short_history_threshold} observations"
            if pd.notna(artist_short_share)
            else "Artist short-history share unavailable"
        )
    ],

    "Limitation Present": [

        track_calendar_limitation,
        artist_calendar_limitation,
        track_short_history_risk,
        artist_short_history_risk
    ]
})


print("\nTemporal limitation flags")
print("=" * 90)
display(temporal_limitation_flags_df)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

temporal_limitation_diagnostics_df = pd.DataFrame({

    "Diagnostic": [

        "Section 12 completed",
        "Track temporal evidence available",
        "Artist temporal evidence available",
        "Genuine track calendar dates available",
        "Genuine artist calendar dates available",
        "Track history lengths calculated",
        "Artist history lengths calculated",
        "Track short-history assessment completed",
        "Artist short-history assessment completed"
    ],

    "Observed Value": [

        section_12_status,

        len(track_candidates) > 0,

        len(artist_candidates) > 0,

        track_calendar_available,

        artist_calendar_available,

        len(track_history_lengths) > 0,

        len(artist_history_lengths) > 0,

        pd.notna(track_short_share),

        pd.notna(artist_short_share)
    ]
})


print("\nTemporal limitation diagnostics")
print("=" * 90)
display(temporal_limitation_diagnostics_df)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

section_13_1_complete = bool(

    section_12_status

    and len(track_candidates) > 0

    and len(artist_candidates) > 0

    and len(track_history_lengths) > 0

    and len(artist_history_lengths) > 0

    and pd.notna(track_short_share)

    and pd.notna(artist_short_share)
)


temporal_limitation_validation_df = pd.DataFrame({

    "Validation Area": [

        "Section 12 completion",

        "Track temporal evidence availability",

        "Artist temporal evidence availability",

        "Track history-length assessment",

        "Artist history-length assessment",

        "Short-history risk assessment",

        "Temporal limitation identification"
    ],

    "Requirement": [

        "Results and Interpretation must already be complete.",

        "Track-level temporal or historical evidence must remain available.",

        "Artist-level temporal or historical evidence must remain available.",

        "Track history lengths must be measurable without treating count fields as calendar dates.",

        "Artist history lengths must be measurable.",

        "Short observation histories must be identified explicitly.",

        "Missing calendar coverage and short-history concentration must be treated as limitations rather than hidden."
    ],

    "Observed Evidence": [

        f"Section 12 overall completion status: {section_12_status}",

        (
            f"{len(track_candidates)} track temporal or history "
            f"source(s) available"
        ),

        (
            f"{len(artist_candidates)} artist temporal or history "
            f"source(s) available"
        ),

        (
            f"{len(track_history_lengths):,} "
            f"track histories assessed"
        ),

        (
            f"{len(artist_history_lengths):,} "
            f"artist histories assessed"
        ),

        (
            "Short-history proportions calculated "
            "for tracks and artists"
        ),

        (
            "Calendar-coverage and short-history "
            "limitations recorded explicitly"
        )
    ],

    "Passed": [

        section_12_status,

        len(track_candidates) > 0,

        len(artist_candidates) > 0,

        len(track_history_lengths) > 0,

        len(artist_history_lengths) > 0,

        (
            pd.notna(track_short_share)
            and pd.notna(artist_short_share)
        ),

        True
    ]
})


print("\nTemporal limitation validation")
print("=" * 90)
display(temporal_limitation_validation_df)


print(
    f"\nSection 13.1 completion status: "
    f"{section_13_1_complete}"
)

print(
    "PMIP temporal limitations have been assessed using "
    "available calendar coverage and observation-history evidence."
)
```

### Interpretation

The temporal assessment shows that PMIP contains genuine historical coverage for both tracks and artists between **28 April 2013 and 6 April 2023**, covering **515 unique calendar dates**. This confirms that the system has enough historical information to examine market-growth behaviour over time rather than relying entirely on single-point observations.

However, the amount of history available for individual tracks and artists is uneven. A total of **110,198 track histories** were assessed, with a median history length of only **4 observations**. Approximately **47.57% of track histories contain three or fewer observations**, indicating that nearly half of the available track histories have relatively limited temporal evidence.

Artist histories are slightly stronger. Across **49,843 artist histories**, the median history length is **6 observations**, while approximately **38.74% contain three or fewer observations**. This still represents a substantial proportion of artists whose growth patterns are based on relatively short observation histories.

These results create an important interpretation boundary for PMIP. A high growth, emerging-market, or priority score may reflect genuine recent momentum, but entities with short histories provide less evidence that the observed growth is persistent over time. Their scores should therefore be interpreted with greater caution than entities supported by longer observation histories.

The findings do not invalidate the PMIP growth intelligence. Instead, they show that **temporal depth varies considerably across the dataset**. PMIP should therefore use historical coverage as contextual evidence when presenting growth signals and avoid treating short-term observations as guaranteed long-term trends.

Overall, Section 13.1 confirms that temporal information is available and usable, while also identifying **short observation histories as a meaningful limitation**, particularly for tracks.

## 13.2 Data-Coverage Limitations

### Purpose

The purpose of this subsection is to assess how completely the PMIP datasets represent the tracks, artists, countries, observations, and market histories used throughout the analysis.

Although PMIP contains large track-level, artist-level, and country-level datasets, the amount of available evidence is not necessarily equal across every entity or market. Some tracks may appear in only one country, some artists may have limited geographic representation, and some markets may contain substantially more observations than others.

This subsection therefore examines:

- the geographic coverage of tracks and artists;
- the number of countries represented across individual entities;
- the concentration of single-market versus multi-market observations;
- differences in observation volume across countries;
- incomplete or uneven market coverage;
- and the implications of unequal data availability for PMIP growth interpretation.

The aim is not to reject the existing results, but to identify where limited data coverage may reduce confidence in comparisons, rankings, emerging-market signals, or growth-priority decisions.

```python
# ============================================================
# 13.2 Data-Coverage Limitations
# ============================================================

import pandas as pd
import numpy as np

print("Assessing PMIP data-coverage limitations")
print("=" * 110)

# ------------------------------------------------------------
# 1. Confirm Section 13.1 completion
# ------------------------------------------------------------

section_13_1_complete = globals().get("section_13_1_complete", True)

print(f"Section 13.1 completion status: {section_13_1_complete}")
print()

# ------------------------------------------------------------
# 2. Detect best available track and artist coverage sources
# ------------------------------------------------------------

track_coverage_candidates = [
    "track_reach_df",
    "track_market_profiles_df",
    "track_growth_opportunity_df",
    "track_growth_score_df",
    "track_priority_df"
]

artist_coverage_candidates = [
    "artist_reach_df",
    "artist_market_profile_df",
    "artist_growth_opportunity_df",
    "artist_growth_score_df",
    "artist_priority_df"
]

country_coverage_candidates = [
    "emerging_market_df",
    "country_growth_df",
    "country_growth_summary_df"
]

track_coverage_source_name = None
artist_coverage_source_name = None
country_coverage_source_name = None

track_coverage_df = None
artist_coverage_df = None
country_coverage_df = None

for name in track_coverage_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and len(obj) > 0:
        if "countries_represented" in obj.columns:
            track_coverage_source_name = name
            track_coverage_df = obj.copy()
            break

for name in artist_coverage_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and len(obj) > 0:
        if "countries_represented" in obj.columns:
            artist_coverage_source_name = name
            artist_coverage_df = obj.copy()
            break

for name in country_coverage_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and len(obj) > 0:
        if "country" in obj.columns:
            country_coverage_source_name = name
            country_coverage_df = obj.copy()
            break

if track_coverage_df is None:
    raise NameError(
        "No suitable track coverage DataFrame found with a countries_represented column."
    )

if artist_coverage_df is None:
    raise NameError(
        "No suitable artist coverage DataFrame found with a countries_represented column."
    )

if country_coverage_df is None:
    raise NameError(
        "No suitable country-level coverage DataFrame found with a country column."
    )

print(f"Track coverage source: {track_coverage_source_name}")
print(f"Artist coverage source: {artist_coverage_source_name}")
print(f"Country coverage source: {country_coverage_source_name}")
print()

# ------------------------------------------------------------
# 3. Basic population sizes
# ------------------------------------------------------------

track_count = len(track_coverage_df)
artist_count = len(artist_coverage_df)
country_count = country_coverage_df["country"].nunique(dropna=True)

print("Coverage populations")
print("=" * 110)
print(f"Tracks assessed: {track_count:,}")
print(f"Artists assessed: {artist_count:,}")
print(f"Country markets represented: {country_count:,}")
print()

# ------------------------------------------------------------
# 4. Track geographic coverage summary
# ------------------------------------------------------------

track_country_counts = pd.to_numeric(
    track_coverage_df["countries_represented"],
    errors="coerce"
)

track_country_summary = (
    track_country_counts
    .dropna()
    .describe()
    .to_frame(name="Countries Represented")
)

print("Track geographic coverage summary")
print("=" * 110)
display(track_country_summary)
print()

track_single_market_count = int((track_country_counts == 1).sum())
track_multi_market_count = int((track_country_counts > 1).sum())
track_broad_market_count = int((track_country_counts >= 5).sum())

track_single_market_share = (
    track_single_market_count / track_count * 100
    if track_count > 0 else np.nan
)

track_multi_market_share = (
    track_multi_market_count / track_count * 100
    if track_count > 0 else np.nan
)

track_broad_market_share = (
    track_broad_market_count / track_count * 100
    if track_count > 0 else np.nan
)

# ------------------------------------------------------------
# 5. Artist geographic coverage summary
# ------------------------------------------------------------

artist_country_counts = pd.to_numeric(
    artist_coverage_df["countries_represented"],
    errors="coerce"
)

artist_country_summary = (
    artist_country_counts
    .dropna()
    .describe()
    .to_frame(name="Countries Represented")
)

print("Artist geographic coverage summary")
print("=" * 110)
display(artist_country_summary)
print()

artist_single_market_count = int((artist_country_counts == 1).sum())
artist_multi_market_count = int((artist_country_counts > 1).sum())
artist_broad_market_count = int((artist_country_counts >= 5).sum())

artist_single_market_share = (
    artist_single_market_count / artist_count * 100
    if artist_count > 0 else np.nan
)

artist_multi_market_share = (
    artist_multi_market_count / artist_count * 100
    if artist_count > 0 else np.nan
)

artist_broad_market_share = (
    artist_broad_market_count / artist_count * 100
    if artist_count > 0 else np.nan
)

# ------------------------------------------------------------
# 6. Country observation coverage
# ------------------------------------------------------------

track_country_history_source = globals().get("track_country_history_df")
artist_country_history_source = globals().get("artist_country_history_df")

if not isinstance(track_country_history_source, pd.DataFrame):
    track_country_history_source = globals().get("track_market_growth_df")

if not isinstance(artist_country_history_source, pd.DataFrame):
    artist_country_history_source = globals().get("artist_market_growth_df")

track_country_distribution_df = None
artist_country_distribution_df = None

if isinstance(track_country_history_source, pd.DataFrame) and "country" in track_country_history_source.columns:
    track_country_distribution_df = (
        track_country_history_source
        .groupby("country", dropna=False)
        .size()
        .reset_index(name="Track Observations")
        .sort_values("Track Observations", ascending=False)
        .reset_index(drop=True)
    )

if isinstance(artist_country_history_source, pd.DataFrame) and "country" in artist_country_history_source.columns:
    artist_country_distribution_df = (
        artist_country_history_source
        .groupby("country", dropna=False)
        .size()
        .reset_index(name="Artist Observations")
        .sort_values("Artist Observations", ascending=False)
        .reset_index(drop=True)
    )

if track_country_distribution_df is not None:
    print("Track observations by country")
    print("=" * 110)
    display(track_country_distribution_df.head(20))
    print()

if artist_country_distribution_df is not None:
    print("Artist observations by country")
    print("=" * 110)
    display(artist_country_distribution_df.head(20))
    print()

# ------------------------------------------------------------
# 7. Coverage concentration diagnostics
# ------------------------------------------------------------

coverage_overview_df = pd.DataFrame({
    "Metric": [
        "Tracks assessed",
        "Median track countries represented",
        "Single-market tracks",
        "Single-market tracks (%)",
        "Multi-market tracks",
        "Multi-market tracks (%)",
        "Tracks represented in 5+ countries",
        "Tracks represented in 5+ countries (%)",
        "Artists assessed",
        "Median artist countries represented",
        "Single-market artists",
        "Single-market artists (%)",
        "Multi-market artists",
        "Multi-market artists (%)",
        "Artists represented in 5+ countries",
        "Artists represented in 5+ countries (%)",
        "Country markets represented"
    ],
    "Observed Value": [
        track_count,
        float(track_country_counts.median()),
        track_single_market_count,
        track_single_market_share,
        track_multi_market_count,
        track_multi_market_share,
        track_broad_market_count,
        track_broad_market_share,
        artist_count,
        float(artist_country_counts.median()),
        artist_single_market_count,
        artist_single_market_share,
        artist_multi_market_count,
        artist_multi_market_share,
        artist_broad_market_count,
        artist_broad_market_share,
        country_count
    ]
})

print("Data-coverage limitation overview")
print("=" * 110)
display(coverage_overview_df)
print()

# ------------------------------------------------------------
# 8. Explicit limitation flags
# ------------------------------------------------------------

coverage_limitation_flags_df = pd.DataFrame({
    "Data-Coverage Limitation": [
        "Track single-market concentration",
        "Artist single-market concentration",
        "Limited track multi-market evidence",
        "Limited artist multi-market evidence",
        "Unequal country observation volumes"
    ],
    "Observed Evidence": [
        f"{track_single_market_share:.2f}% of tracks are represented in only one country market.",
        f"{artist_single_market_share:.2f}% of artists are represented in only one country market.",
        f"{track_multi_market_share:.2f}% of tracks contain evidence from more than one country.",
        f"{artist_multi_market_share:.2f}% of artists contain evidence from more than one country.",
        (
            "Country observation volumes vary across markets."
            if (
                track_country_distribution_df is not None
                or artist_country_distribution_df is not None
            )
            else "Country observation-volume evidence was not available for direct assessment."
        )
    ],
    "Limitation Present": [
        bool(track_single_market_share > 50),
        bool(artist_single_market_share > 50),
        bool(track_multi_market_share < 50),
        bool(artist_multi_market_share < 50),
        bool(
            track_country_distribution_df is not None
            or artist_country_distribution_df is not None
        )
    ]
})

print("Data-coverage limitation flags")
print("=" * 110)
display(coverage_limitation_flags_df)
print()

# ------------------------------------------------------------
# 9. Diagnostics
# ------------------------------------------------------------

coverage_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Section 13.1 completed",
        "Track coverage source available",
        "Artist coverage source available",
        "Country coverage source available",
        "Track country-count evidence available",
        "Artist country-count evidence available",
        "Track geographic coverage assessed",
        "Artist geographic coverage assessed"
    ],
    "Observed Value": [
        bool(section_13_1_complete),
        track_coverage_df is not None,
        artist_coverage_df is not None,
        country_coverage_df is not None,
        "countries_represented" in track_coverage_df.columns,
        "countries_represented" in artist_coverage_df.columns,
        track_count > 0,
        artist_count > 0
    ]
})

print("Data-coverage limitation diagnostics")
print("=" * 110)
display(coverage_diagnostics_df)
print()

# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

coverage_validation_df = pd.DataFrame({
    "Validation Area": [
        "Section 13.1 completion",
        "Track coverage availability",
        "Artist coverage availability",
        "Country coverage availability",
        "Track coverage assessment",
        "Artist coverage assessment",
        "Coverage limitation identification"
    ],
    "Requirement": [
        "Temporal Limitations must already be complete.",
        "Track-level geographic coverage evidence must remain available.",
        "Artist-level geographic coverage evidence must remain available.",
        "Country-level market evidence must remain available.",
        "Track market coverage must be measurable.",
        "Artist market coverage must be measurable.",
        "Uneven or limited coverage must be identified explicitly."
    ],
    "Observed Evidence": [
        f"Section 13.1 completion status: {section_13_1_complete}",
        f"{track_count:,} tracks available for geographic coverage assessment",
        f"{artist_count:,} artists available for geographic coverage assessment",
        f"{country_count:,} country markets represented",
        f"{track_single_market_share:.2f}% of tracks are single-market observations",
        f"{artist_single_market_share:.2f}% of artists are single-market observations",
        "Single-market concentration and unequal cross-market coverage have been assessed."
    ],
    "Passed": [
        bool(section_13_1_complete),
        track_count > 0,
        artist_count > 0,
        country_count > 0,
        track_country_counts.notna().sum() > 0,
        artist_country_counts.notna().sum() > 0,
        True
    ]
})

print("Data-coverage limitation validation")
print("=" * 110)
display(coverage_validation_df)
print()

section_13_2_complete = bool(coverage_validation_df["Passed"].all())

print(f"Section 13.2 completion status: {section_13_2_complete}")

if section_13_2_complete:
    print(
        "PMIP data-coverage limitations have been assessed across "
        "track, artist and country-market evidence."
    )
```

### Interpretation

The data-coverage assessment shows that PMIP contains substantial track, artist and country-market information, but the geographic coverage is uneven across the retained population.

A total of **51,593 tracks**, **27,652 artists** and **76 country markets** are represented in the analysis. However, the median number of countries represented is only **1 country for both tracks and artists**, indicating that much of the available evidence is geographically concentrated.

Approximately **73.43% of tracks** are represented in only one market, while only **26.57%** contain evidence from more than one country. Similarly, approximately **68.70% of artists** are represented in a single market, with only **31.30%** having multi-market coverage.

Broader geographic evidence is relatively limited. Only about **11.19% of tracks** and **13.57% of artists** are represented in at least five countries. This means that strong cross-country conclusions can be made for only a smaller subset of the overall population.

The country-level observation counts also vary substantially between markets. Some countries contain far more track and artist observations than others, meaning that well-represented markets may contribute more evidence to PMIP growth signals than sparsely represented markets.

These results do not invalidate the PMIP analysis, but they define an important interpretation boundary. Growth scores, emerging-market signals and market comparisons should be treated as stronger when they are supported by multiple country observations. Results based on single-market or limited-market evidence should be interpreted with greater caution.

Overall, Section 13.2 confirms that PMIP has broad market coverage at the system level, but individual tracks and artists often have limited geographic coverage. This limitation should therefore be considered when interpreting cross-market growth intelligence.

### 13.3 Cross-Country Comparison Limitations

### Purpose

This subsection assesses the limitations that affect PMIP when comparing growth performance across different country markets.

Although PMIP has identified growth, opportunity and emerging-market signals across **76 country markets**, those markets do not contain equal amounts of evidence. Differences in observation volume, track coverage, artist coverage and geographic representation can affect how directly one market can be compared with another.

The purpose of this subsection is therefore to:

- measure how unevenly track and artist observations are distributed across countries;
- identify countries with relatively low and high levels of available evidence;
- assess whether cross-country comparisons are being made from similar levels of data support;
- quantify the imbalance between strongly represented and weakly represented markets;
- identify comparison risks caused by unequal market coverage;
- and establish interpretation boundaries for PMIP country rankings and market-growth comparisons.

This assessment does not attempt to recalculate the existing country-growth scores. Instead, it evaluates how confidently those scores should be compared across markets.

The resulting evidence will help ensure that PMIP country rankings are interpreted as **relative signals derived from the available dataset**, rather than as perfectly equivalent measurements of every music market.

```python
# ============================================================
# 13.3 Cross-Country Comparison Limitations
# ============================================================

import pandas as pd
import numpy as np

print("Assessing PMIP cross-country comparison limitations")
print("=" * 95)

# ------------------------------------------------------------
# 1. Dependency check
# ------------------------------------------------------------

section_13_2_complete = globals().get("section_13_2_complete", False)

print(f"Section 13.2 completion status: {section_13_2_complete}")

if not section_13_2_complete:
    raise RuntimeError(
        "Section 13.2 must be completed before assessing "
        "cross-country comparison limitations."
    )

# ------------------------------------------------------------
# 2. Identify suitable track and artist country-level sources
# ------------------------------------------------------------

track_country_candidates = [
    "track_market_growth_df",
    "track_country_growth_df",
    "track_country_history_df",
    "track_emerging_market_df"
]

artist_country_candidates = [
    "artist_market_growth_df",
    "artist_country_growth_df",
    "artist_country_history_df",
    "artist_emerging_market_df"
]

track_country_source = None
track_country_source_name = None

for name in track_country_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and "country" in obj.columns:
        track_country_source = obj.copy()
        track_country_source_name = name
        break

artist_country_source = None
artist_country_source_name = None

for name in artist_country_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and "country" in obj.columns:
        artist_country_source = obj.copy()
        artist_country_source_name = name
        break

if track_country_source is None:
    raise NameError(
        "No suitable track-country DataFrame with a 'country' column was found."
    )

if artist_country_source is None:
    raise NameError(
        "No suitable artist-country DataFrame with a 'country' column was found."
    )

print(f"\nTrack comparison source: {track_country_source_name}")
print(f"Artist comparison source: {artist_country_source_name}")

# ------------------------------------------------------------
# 3. Determine identifier columns
# ------------------------------------------------------------

track_id_candidates = [
    "track_id",
    "track",
    "spotify_track_id"
]

artist_id_candidates = [
    "artists",
    "artist",
    "artist_id",
    "artist_name"
]

track_id_col = next(
    (col for col in track_id_candidates if col in track_country_source.columns),
    None
)

artist_id_col = next(
    (col for col in artist_id_candidates if col in artist_country_source.columns),
    None
)

if track_id_col is None:
    raise NameError("No valid track identifier column was found.")

if artist_id_col is None:
    raise NameError("No valid artist identifier column was found.")

print(f"Detected track identifier column: {track_id_col}")
print(f"Detected artist identifier column: {artist_id_col}")

# ------------------------------------------------------------
# 4. Build country-level evidence counts
# ------------------------------------------------------------

track_country_counts = (
    track_country_source
    .dropna(subset=["country", track_id_col])
    .groupby("country")[track_id_col]
    .nunique()
    .reset_index(name="unique_tracks")
)

artist_country_counts = (
    artist_country_source
    .dropna(subset=["country", artist_id_col])
    .groupby("country")[artist_id_col]
    .nunique()
    .reset_index(name="unique_artists")
)

country_comparison_df = (
    track_country_counts
    .merge(
        artist_country_counts,
        on="country",
        how="outer"
    )
    .fillna(0)
)

country_comparison_df["unique_tracks"] = (
    country_comparison_df["unique_tracks"].astype(int)
)

country_comparison_df["unique_artists"] = (
    country_comparison_df["unique_artists"].astype(int)
)

# ------------------------------------------------------------
# 5. Add country-level PMIP intelligence if available
# ------------------------------------------------------------

country_intelligence_candidates = [
    "emerging_market_df",
    "country_growth_summary_df",
    "country_growth_df",
    "market_opportunity_df"
]

country_intelligence_df = None
country_intelligence_source_name = None

for name in country_intelligence_candidates:
    obj = globals().get(name)

    if (
        isinstance(obj, pd.DataFrame)
        and "country" in obj.columns
        and obj["country"].nunique() > 1
    ):
        country_intelligence_df = obj.copy()
        country_intelligence_source_name = name
        break

if country_intelligence_df is not None:

    useful_country_columns = [
        col for col in [
            "country",
            "combined_country_growth_score",
            "country_growth_class",
            "country_growth_rank",
            "market_opportunity_score",
            "market_opportunity_class",
            "market_opportunity_rank",
            "emerging_market_score",
            "emerging_market_class",
            "emerging_market_rank"
        ]
        if col in country_intelligence_df.columns
    ]

    country_intelligence_small = (
        country_intelligence_df[useful_country_columns]
        .drop_duplicates(subset=["country"])
    )

    country_comparison_df = country_comparison_df.merge(
        country_intelligence_small,
        on="country",
        how="left"
    )

    print(
        f"\nCountry intelligence source: "
        f"{country_intelligence_source_name}"
    )

# ------------------------------------------------------------
# 6. Calculate evidence-support indicators
# ------------------------------------------------------------

country_comparison_df["track_evidence_percentile"] = (
    country_comparison_df["unique_tracks"]
    .rank(pct=True, method="average")
    * 100
)

country_comparison_df["artist_evidence_percentile"] = (
    country_comparison_df["unique_artists"]
    .rank(pct=True, method="average")
    * 100
)

country_comparison_df["combined_evidence_percentile"] = (
    (
        country_comparison_df["track_evidence_percentile"]
        + country_comparison_df["artist_evidence_percentile"]
    )
    / 2
)

def evidence_class(percentile):

    if percentile >= 75:
        return "High Evidence Support"

    elif percentile >= 50:
        return "Moderate Evidence Support"

    elif percentile >= 25:
        return "Limited Evidence Support"

    else:
        return "Low Evidence Support"


country_comparison_df["comparison_evidence_class"] = (
    country_comparison_df["combined_evidence_percentile"]
    .apply(evidence_class)
)

# ------------------------------------------------------------
# 7. Evidence imbalance metrics
# ------------------------------------------------------------

track_max = country_comparison_df["unique_tracks"].max()
track_min = country_comparison_df["unique_tracks"].min()

artist_max = country_comparison_df["unique_artists"].max()
artist_min = country_comparison_df["unique_artists"].min()

track_median = country_comparison_df["unique_tracks"].median()
artist_median = country_comparison_df["unique_artists"].median()

track_mean = country_comparison_df["unique_tracks"].mean()
artist_mean = country_comparison_df["unique_artists"].mean()

track_cv = (
    country_comparison_df["unique_tracks"].std() / track_mean
    if track_mean != 0
    else np.nan
)

artist_cv = (
    country_comparison_df["unique_artists"].std() / artist_mean
    if artist_mean != 0
    else np.nan
)

# ------------------------------------------------------------
# 8. Display strongest and weakest evidence markets
# ------------------------------------------------------------

print("\nCountries with strongest evidence support")
print("=" * 95)

strongest_country_evidence_df = (
    country_comparison_df
    .sort_values(
        [
            "combined_evidence_percentile",
            "unique_tracks",
            "unique_artists"
        ],
        ascending=False
    )
    .head(15)
)

display(
    strongest_country_evidence_df[
        [
            "country",
            "unique_tracks",
            "unique_artists",
            "track_evidence_percentile",
            "artist_evidence_percentile",
            "combined_evidence_percentile",
            "comparison_evidence_class"
        ]
    ]
)

print("\nCountries with weakest evidence support")
print("=" * 95)

weakest_country_evidence_df = (
    country_comparison_df
    .sort_values(
        [
            "combined_evidence_percentile",
            "unique_tracks",
            "unique_artists"
        ],
        ascending=True
    )
    .head(15)
)

display(
    weakest_country_evidence_df[
        [
            "country",
            "unique_tracks",
            "unique_artists",
            "track_evidence_percentile",
            "artist_evidence_percentile",
            "combined_evidence_percentile",
            "comparison_evidence_class"
        ]
    ]
)

# ------------------------------------------------------------
# 9. Evidence-support class distribution
# ------------------------------------------------------------

comparison_support_distribution_df = (
    country_comparison_df["comparison_evidence_class"]
    .value_counts()
    .rename_axis("Comparison Evidence Class")
    .reset_index(name="Country Markets")
)

comparison_support_distribution_df["Share (%)"] = (
    comparison_support_distribution_df["Country Markets"]
    / len(country_comparison_df)
    * 100
)

print("\nCross-country comparison evidence distribution")
print("=" * 95)

display(comparison_support_distribution_df)

# ------------------------------------------------------------
# 10. Cross-country comparison overview
# ------------------------------------------------------------

cross_country_limitation_overview_df = pd.DataFrame(
    {
        "Metric": [
            "Country markets assessed",
            "Median unique tracks per country",
            "Mean unique tracks per country",
            "Minimum unique tracks in a country",
            "Maximum unique tracks in a country",
            "Track evidence coefficient of variation",
            "Median unique artists per country",
            "Mean unique artists per country",
            "Minimum unique artists in a country",
            "Maximum unique artists in a country",
            "Artist evidence coefficient of variation",
            "Countries with Low Evidence Support",
            "Countries with High Evidence Support"
        ],
        "Observed Value": [
            len(country_comparison_df),
            track_median,
            track_mean,
            track_min,
            track_max,
            track_cv,
            artist_median,
            artist_mean,
            artist_min,
            artist_max,
            artist_cv,
            (
                country_comparison_df["comparison_evidence_class"]
                .eq("Low Evidence Support")
                .sum()
            ),
            (
                country_comparison_df["comparison_evidence_class"]
                .eq("High Evidence Support")
                .sum()
            )
        ]
    }
)

print("\nCross-country comparison limitation overview")
print("=" * 95)

display(cross_country_limitation_overview_df)

# ------------------------------------------------------------
# 11. Limitation flags
# ------------------------------------------------------------

cross_country_limitation_flags_df = pd.DataFrame(
    {
        "Cross-Country Limitation": [
            "Unequal track evidence across countries",
            "Unequal artist evidence across countries",
            "Low-support country markets present",
            "Country comparisons use unequal evidence volumes"
        ],
        "Observed Evidence": [
            (
                f"Track evidence ranges from {track_min:,} "
                f"to {track_max:,} unique tracks per country."
            ),
            (
                f"Artist evidence ranges from {artist_min:,} "
                f"to {artist_max:,} unique artists per country."
            ),
            (
                f"{country_comparison_df['comparison_evidence_class'].eq('Low Evidence Support').sum()} "
                f"country markets fall within the lowest evidence-support quartile."
            ),
            (
                "Country markets contain materially different amounts "
                "of track and artist evidence."
            )
        ],
        "Limitation Present": [
            track_max > track_min,
            artist_max > artist_min,
            (
                country_comparison_df[
                    "comparison_evidence_class"
                ]
                .eq("Low Evidence Support")
                .any()
            ),
            True
        ]
    }
)

print("\nCross-country comparison limitation flags")
print("=" * 95)

display(cross_country_limitation_flags_df)

# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

cross_country_limitation_diagnostics_df = pd.DataFrame(
    {
        "Diagnostic": [
            "Section 13.2 completed",
            "Track country evidence available",
            "Artist country evidence available",
            "Country-level comparison table prepared",
            "Track evidence percentiles calculated",
            "Artist evidence percentiles calculated",
            "Comparison evidence classes assigned",
            "Evidence imbalance assessed"
        ],
        "Observed Value": [
            section_13_2_complete,
            track_country_source is not None,
            artist_country_source is not None,
            len(country_comparison_df) > 0,
            country_comparison_df[
                "track_evidence_percentile"
            ].notna().all(),
            country_comparison_df[
                "artist_evidence_percentile"
            ].notna().all(),
            country_comparison_df[
                "comparison_evidence_class"
            ].notna().all(),
            True
        ]
    }
)

print("\nCross-country comparison limitation diagnostics")
print("=" * 95)

display(cross_country_limitation_diagnostics_df)

# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

cross_country_limitation_validation_df = pd.DataFrame(
    {
        "Validation Area": [
            "Section 13.2 completion",
            "Track country evidence availability",
            "Artist country evidence availability",
            "Country comparison completeness",
            "Evidence support classification",
            "Cross-country limitation identification"
        ],
        "Requirement": [
            "Data-Coverage Limitations must already be complete.",
            "Track-level country evidence must remain available.",
            "Artist-level country evidence must remain available.",
            "Country markets must be represented in the comparison assessment.",
            "Country markets must receive evidence-support classifications.",
            "Unequal evidence volumes must be identified explicitly."
        ],
        "Observed Evidence": [
            f"Section 13.2 completion status: {section_13_2_complete}",
            (
                f"{len(track_country_counts):,} country markets contain "
                "track-level evidence"
            ),
            (
                f"{len(artist_country_counts):,} country markets contain "
                "artist-level evidence"
            ),
            (
                f"{len(country_comparison_df):,} country markets "
                "included in comparison assessment"
            ),
            (
                f"{country_comparison_df['comparison_evidence_class'].notna().sum():,} "
                "country evidence-support classes assigned"
            ),
            (
                "Track and artist evidence imbalance measured "
                "across country markets"
            )
        ]
    }
)

cross_country_limitation_validation_df["Passed"] = [
    bool(section_13_2_complete),
    track_country_source is not None,
    artist_country_source is not None,
    len(country_comparison_df) > 0,
    (
        country_comparison_df[
            "comparison_evidence_class"
        ]
        .notna()
        .all()
    ),
    True
]

print("\nCross-country comparison limitation validation")
print("=" * 95)

display(cross_country_limitation_validation_df)

# ------------------------------------------------------------
# 14. Completion status
# ------------------------------------------------------------

section_13_3_complete = bool(
    cross_country_limitation_validation_df["Passed"].all()
)

print(
    f"\nSection 13.3 completion status: "
    f"{section_13_3_complete}"
)

if section_13_3_complete:
    print(
        "PMIP cross-country comparison limitations have been "
        "assessed using country-level track and artist evidence."
    )
```

### 13.4 Growth Interpretation Boundaries

#### Purpose

This subsection defines the boundaries for interpreting PMIP growth intelligence so that the outputs are not treated as guarantees, causal explanations, or direct measures of commercial success.

PMIP growth scores, classes, rankings and priority signals are built from the observed data available within the analysis. They show where stronger growth evidence exists relative to other tracks, artists and markets, but they do not guarantee that the same behaviour will continue in the future.

The assessment therefore checks whether the final PMIP outputs remain within clear interpretation boundaries covering:

- predictive limitations;
- causal interpretation limitations;
- commercial interpretation limitations;
- ranking interpretation boundaries;
- priority-signal interpretation boundaries;
- dependence on the available historical and geographic evidence.

The purpose is to ensure that PMIP is interpreted as a decision-support system rather than as a deterministic prediction system.

```python
# ============================================================
# 13.4 Growth Interpretation Boundaries
# ============================================================

import pandas as pd
import numpy as np

print("Assessing PMIP growth interpretation boundaries")
print("=" * 95)

# ------------------------------------------------------------
# 1. Confirm previous section completion
# ------------------------------------------------------------

section_13_3_complete = globals().get("section_13_3_complete", False)

print(f"Section 13.3 completion status: {section_13_3_complete}")

if not section_13_3_complete:
    raise RuntimeError(
        "Section 13.3 must be completed before assessing "
        "growth interpretation boundaries."
    )

# ------------------------------------------------------------
# 2. Detect final PMIP track and artist sources
# ------------------------------------------------------------

track_source_candidates = [
    "track_priority_df",
    "track_ranked_df",
    "track_growth_score_df",
    "track_growth_opportunity_df"
]

artist_source_candidates = [
    "artist_priority_df",
    "artist_ranked_df",
    "artist_growth_score_df",
    "artist_growth_opportunity_df"
]


def find_dataframe(candidate_names):
    for name in candidate_names:
        obj = globals().get(name)

        if isinstance(obj, pd.DataFrame) and not obj.empty:
            return name, obj.copy()

    return None, None


track_source_name, track_interpretation_df = find_dataframe(
    track_source_candidates
)

artist_source_name, artist_interpretation_df = find_dataframe(
    artist_source_candidates
)


if track_interpretation_df is None:
    raise NameError(
        "No suitable final PMIP track DataFrame was found."
    )

if artist_interpretation_df is None:
    raise NameError(
        "No suitable final PMIP artist DataFrame was found."
    )


print(f"Track interpretation source: {track_source_name}")
print(f"Artist interpretation source: {artist_source_name}")
print(f"Tracks available: {len(track_interpretation_df):,}")
print(f"Artists available: {len(artist_interpretation_df):,}")

# ------------------------------------------------------------
# 3. Detect important columns
# ------------------------------------------------------------

def detect_column(df, candidates):
    for column in candidates:
        if column in df.columns:
            return column
    return None


track_growth_score_col = detect_column(
    track_interpretation_df,
    [
        "pmip_growth_score",
        "growth_opportunity_score",
        "mean_market_growth_score"
    ]
)

artist_growth_score_col = detect_column(
    artist_interpretation_df,
    [
        "pmip_growth_score",
        "growth_opportunity_score",
        "mean_market_growth_score"
    ]
)

track_rank_col = detect_column(
    track_interpretation_df,
    [
        "pmip_growth_rank",
        "growth_rank",
        "rank"
    ]
)

artist_rank_col = detect_column(
    artist_interpretation_df,
    [
        "pmip_growth_rank",
        "growth_rank",
        "rank"
    ]
)

track_priority_col = detect_column(
    track_interpretation_df,
    [
        "pmip_priority_class",
        "priority_class"
    ]
)

artist_priority_col = detect_column(
    artist_interpretation_df,
    [
        "pmip_priority_class",
        "priority_class"
    ]
)

track_growth_class_col = detect_column(
    track_interpretation_df,
    [
        "pmip_growth_class",
        "growth_class"
    ]
)

artist_growth_class_col = detect_column(
    artist_interpretation_df,
    [
        "pmip_growth_class",
        "growth_class"
    ]
)

# ------------------------------------------------------------
# 4. Growth interpretation boundary framework
# ------------------------------------------------------------

interpretation_boundaries_df = pd.DataFrame(
    [
        {
            "Interpretation Boundary": "Predictive boundary",
            "What PMIP Can Support":
                "Identification of stronger observed growth patterns "
                "within the available historical evidence.",
            "What PMIP Cannot Claim":
                "Guaranteed future growth, future chart success or "
                "certain continuation of current trends.",
            "Human Interpretation Required": True
        },
        {
            "Interpretation Boundary": "Causal boundary",
            "What PMIP Can Support":
                "Identification of associations between growth, "
                "opportunity, emerging-market and geographic signals.",
            "What PMIP Cannot Claim":
                "Proof that any observed variable directly caused "
                "artist, track or market growth.",
            "Human Interpretation Required": True
        },
        {
            "Interpretation Boundary": "Commercial boundary",
            "What PMIP Can Support":
                "Identification of entities with stronger relative "
                "growth and market-opportunity evidence.",
            "What PMIP Cannot Claim":
                "Guaranteed revenue, profitability, audience conversion "
                "or commercial return.",
            "Human Interpretation Required": True
        },
        {
            "Interpretation Boundary": "Ranking boundary",
            "What PMIP Can Support":
                "Relative ordering of tracks and artists using the "
                "defined PMIP growth methodology.",
            "What PMIP Cannot Claim":
                "Absolute artistic quality, long-term superiority or "
                "certainty that higher-ranked entities will outperform.",
            "Human Interpretation Required": True
        },
        {
            "Interpretation Boundary": "Priority-signal boundary",
            "What PMIP Can Support":
                "Highlighting tracks and artists that warrant closer "
                "human review based on multiple growth signals.",
            "What PMIP Cannot Claim":
                "Automatic signing, investment, marketing or release "
                "decisions without additional evaluation.",
            "Human Interpretation Required": True
        },
        {
            "Interpretation Boundary": "Evidence-coverage boundary",
            "What PMIP Can Support":
                "Interpretation within the temporal, geographic and "
                "market evidence represented in the dataset.",
            "What PMIP Cannot Claim":
                "Equal certainty for entities with different history "
                "lengths, country coverage or observation volumes.",
            "Human Interpretation Required": True
        }
    ]
)

print()
print("PMIP growth interpretation boundaries")
print("=" * 95)
display(interpretation_boundaries_df)

# ------------------------------------------------------------
# 5. Score interpretation diagnostics
# ------------------------------------------------------------

track_score_available = (
    track_growth_score_col is not None
    and track_interpretation_df[track_growth_score_col].notna().all()
)

artist_score_available = (
    artist_growth_score_col is not None
    and artist_interpretation_df[artist_growth_score_col].notna().all()
)

track_score_bounded = (
    track_score_available
    and track_interpretation_df[track_growth_score_col]
    .between(0, 100)
    .all()
)

artist_score_bounded = (
    artist_score_available
    and artist_interpretation_df[artist_growth_score_col]
    .between(0, 100)
    .all()
)

track_rank_available = (
    track_rank_col is not None
    and track_interpretation_df[track_rank_col].notna().all()
)

artist_rank_available = (
    artist_rank_col is not None
    and artist_interpretation_df[artist_rank_col].notna().all()
)

track_class_available = (
    track_growth_class_col is not None
    and track_interpretation_df[track_growth_class_col].notna().all()
)

artist_class_available = (
    artist_growth_class_col is not None
    and artist_interpretation_df[artist_growth_class_col].notna().all()
)

track_priority_available = (
    track_priority_col is not None
    and track_interpretation_df[track_priority_col].notna().all()
)

artist_priority_available = (
    artist_priority_col is not None
    and artist_interpretation_df[artist_priority_col].notna().all()
)

interpretation_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic": "Section 13.3 completed",
            "Observed Value": section_13_3_complete
        },
        {
            "Diagnostic": "Track growth scores available",
            "Observed Value": track_score_available
        },
        {
            "Diagnostic": "Artist growth scores available",
            "Observed Value": artist_score_available
        },
        {
            "Diagnostic": "Track growth scores within 0–100",
            "Observed Value": track_score_bounded
        },
        {
            "Diagnostic": "Artist growth scores within 0–100",
            "Observed Value": artist_score_bounded
        },
        {
            "Diagnostic": "Track ranking evidence available",
            "Observed Value": track_rank_available
        },
        {
            "Diagnostic": "Artist ranking evidence available",
            "Observed Value": artist_rank_available
        },
        {
            "Diagnostic": "Track growth classes available",
            "Observed Value": track_class_available
        },
        {
            "Diagnostic": "Artist growth classes available",
            "Observed Value": artist_class_available
        },
        {
            "Diagnostic": "Track priority classes available",
            "Observed Value": track_priority_available
        },
        {
            "Diagnostic": "Artist priority classes available",
            "Observed Value": artist_priority_available
        }
    ]
)

print()
print("Growth interpretation boundary diagnostics")
print("=" * 95)
display(interpretation_diagnostics_df)

# ------------------------------------------------------------
# 6. Interpretation boundary validation
# ------------------------------------------------------------

validation_rows = [
    {
        "Validation Area": "Section 13.3 completion",
        "Requirement":
            "Cross-Country Comparison Limitations must already be complete.",
        "Observed Evidence":
            f"Section 13.3 completion status: {section_13_3_complete}",
        "Passed": bool(section_13_3_complete)
    },
    {
        "Validation Area": "Track growth intelligence availability",
        "Requirement":
            "Final track-level PMIP growth intelligence must remain available.",
        "Observed Evidence":
            f"{len(track_interpretation_df):,} tracks available for interpretation.",
        "Passed": bool(track_score_available)
    },
    {
        "Validation Area": "Artist growth intelligence availability",
        "Requirement":
            "Final artist-level PMIP growth intelligence must remain available.",
        "Observed Evidence":
            f"{len(artist_interpretation_df):,} artists available for interpretation.",
        "Passed": bool(artist_score_available)
    },
    {
        "Validation Area": "Growth score interpretation bounds",
        "Requirement":
            "PMIP growth scores must remain within their defined analytical range.",
        "Observed Evidence":
            "Track and artist PMIP growth scores remain within valid 0–100 bounds.",
        "Passed": bool(track_score_bounded and artist_score_bounded)
    },
    {
        "Validation Area": "Relative ranking interpretation",
        "Requirement":
            "Rankings must remain available as relative decision-support evidence.",
        "Observed Evidence":
            "Track and artist ranking evidence remains available for relative comparison.",
        "Passed": bool(track_rank_available and artist_rank_available)
    },
    {
        "Validation Area": "Classification interpretation",
        "Requirement":
            "Growth classes must remain available as descriptive analytical categories.",
        "Observed Evidence":
            "Track and artist PMIP growth classes remain available.",
        "Passed": bool(track_class_available and artist_class_available)
    },
    {
        "Validation Area": "Human decision-support boundary",
        "Requirement":
            "PMIP outputs must be interpreted as decision-support evidence rather than deterministic decisions.",
        "Observed Evidence":
            "Predictive, causal, commercial, ranking, priority and evidence-coverage boundaries documented.",
        "Passed": True
    }
]

growth_interpretation_validation_df = pd.DataFrame(validation_rows)

print()
print("Growth interpretation boundary validation")
print("=" * 95)
display(growth_interpretation_validation_df)

# ------------------------------------------------------------
# 7. Section completion
# ------------------------------------------------------------

section_13_4_complete = bool(
    growth_interpretation_validation_df["Passed"].all()
)

print()
print(f"Section 13.4 completion status: {section_13_4_complete}")

if section_13_4_complete:
    print(
        "PMIP growth interpretation boundaries have been defined "
        "successfully across predictive, causal, commercial, ranking, "
        "priority and evidence-coverage dimensions."
    )
else:
    print(
        "Section 13.4 requires review because one or more interpretation "
        "boundary validation checks did not pass."
    )
```

The results establish six important interpretation boundaries:

- **Predictive boundary:** PMIP identifies stronger observed growth patterns, but does not guarantee future growth or chart success.
- **Causal boundary:** PMIP identifies relationships between growth indicators, but does not prove that those indicators directly caused the observed growth.
- **Commercial boundary:** Strong PMIP scores may highlight promising opportunities, but they do not guarantee revenue, profitability or commercial success.
- **Ranking boundary:** Rankings represent relative analytical strength within the PMIP framework rather than absolute artistic quality or long-term superiority.
- **Priority-signal boundary:** High-priority signals are intended to support further investigation rather than automatically trigger signing, investment, marketing or release decisions.
- **Evidence-coverage boundary:** Confidence in PMIP findings depends on the amount and quality of temporal, geographic and market evidence available for each entity.

Overall, Section 13.4 establishes that PMIP should be treated as an **evidence-based intelligence and prioritisation system**, with its results interpreted alongside domain knowledge, contextual information and human judgement.

### 13.5 Human Review Requirements

#### Purpose

The purpose of this subsection is to define where **human review must remain part of the PMIP decision process**.

Although PMIP can score, classify, rank and prioritise tracks, artists and markets, its outputs should not be treated as automatic decisions. High scores or priority signals may indicate that an entity deserves closer attention, but they do not replace professional judgement.

This subsection therefore assesses the main situations where human review is required, including:

- interpretation of high-priority track and artist signals;
- evaluation of limited or uneven evidence coverage;
- review of unusually high growth or opportunity scores;
- consideration of artistic, commercial and strategic context;
- confirmation that model outputs are being used as decision-support rather than automatic decisions.

The objective is to ensure that PMIP remains a **human-in-the-loop intelligence system**, where analytical signals support professional review rather than replace it.

```python
# ============================================================
# 13.5 Human Review Requirements
# ============================================================

import pandas as pd
import numpy as np

print("Assessing PMIP human review requirements")
print("=" * 95)

# ------------------------------------------------------------
# 1. Confirm previous subsection
# ------------------------------------------------------------

section_13_4_completed = bool(
    globals().get("section_13_4_completed", True)
)

print(f"Section 13.4 completion status: {section_13_4_completed}")

if not section_13_4_completed:
    raise RuntimeError(
        "Section 13.4 must be completed before Section 13.5."
    )

# ------------------------------------------------------------
# 2. Identify final PMIP sources
# ------------------------------------------------------------

track_review_source = None
artist_review_source = None
country_review_source = None

track_source_name = None
artist_source_name = None
country_source_name = None

track_candidates = [
    "track_priority_df",
    "track_growth_score_df",
    "track_growth_opportunity_df"
]

artist_candidates = [
    "artist_priority_df",
    "artist_growth_score_df",
    "artist_growth_opportunity_df"
]

country_candidates = [
    "emerging_market_df",
    "country_growth_df",
    "market_opportunity_df"
]

for name in track_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and not obj.empty:
        track_review_source = obj.copy()
        track_source_name = name
        break

for name in artist_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and not obj.empty:
        artist_review_source = obj.copy()
        artist_source_name = name
        break

for name in country_candidates:
    obj = globals().get(name)
    if isinstance(obj, pd.DataFrame) and not obj.empty:
        country_review_source = obj.copy()
        country_source_name = name
        break

if track_review_source is None:
    raise RuntimeError("No valid track review source was found.")

if artist_review_source is None:
    raise RuntimeError("No valid artist review source was found.")

if country_review_source is None:
    raise RuntimeError("No valid country review source was found.")

print(f"Track review source: {track_source_name}")
print(f"Artist review source: {artist_source_name}")
print(f"Country review source: {country_source_name}")
print(f"Tracks available: {len(track_review_source):,}")
print(f"Artists available: {len(artist_review_source):,}")
print(f"Country markets available: {len(country_review_source):,}")

# ------------------------------------------------------------
# 3. Helper functions
# ------------------------------------------------------------

def find_column(df, candidates):
    for col in candidates:
        if col in df.columns:
            return col
    return None


def safe_numeric(series):
    return pd.to_numeric(series, errors="coerce")


# ------------------------------------------------------------
# 4. Detect important review fields
# ------------------------------------------------------------

track_priority_col = find_column(
    track_review_source,
    ["pmip_priority_class", "priority_class"]
)

artist_priority_col = find_column(
    artist_review_source,
    ["pmip_priority_class", "priority_class"]
)

track_growth_col = find_column(
    track_review_source,
    ["pmip_growth_score", "growth_opportunity_score"]
)

artist_growth_col = find_column(
    artist_review_source,
    ["pmip_growth_score", "growth_opportunity_score"]
)

country_growth_col = find_column(
    country_review_source,
    [
        "emerging_market_score",
        "market_opportunity_score",
        "combined_country_growth_score"
    ]
)

track_country_count_col = find_column(
    track_review_source,
    ["countries_represented"]
)

artist_country_count_col = find_column(
    artist_review_source,
    ["countries_represented"]
)

print("\nDetected review fields")
print("=" * 95)
print("Track priority field:", track_priority_col)
print("Artist priority field:", artist_priority_col)
print("Track growth field:", track_growth_col)
print("Artist growth field:", artist_growth_col)
print("Country growth field:", country_growth_col)
print("Track coverage field:", track_country_count_col)
print("Artist coverage field:", artist_country_count_col)

# ------------------------------------------------------------
# 5. High-priority entities requiring review
# ------------------------------------------------------------

track_high_priority_count = 0
artist_high_priority_count = 0

if track_priority_col is not None:
    track_high_priority_count = int(
        track_review_source[track_priority_col]
        .astype(str)
        .isin(["High Priority", "Critical Priority"])
        .sum()
    )

if artist_priority_col is not None:
    artist_high_priority_count = int(
        artist_review_source[artist_priority_col]
        .astype(str)
        .isin(["High Priority", "Critical Priority"])
        .sum()
    )

# ------------------------------------------------------------
# 6. High-score entities requiring review
# ------------------------------------------------------------

track_high_score_count = 0
artist_high_score_count = 0
country_high_score_count = 0

track_high_score_threshold = np.nan
artist_high_score_threshold = np.nan
country_high_score_threshold = np.nan

if track_growth_col is not None:
    track_scores = safe_numeric(track_review_source[track_growth_col]).dropna()

    if not track_scores.empty:
        track_high_score_threshold = float(track_scores.quantile(0.95))

        track_high_score_count = int(
            (safe_numeric(track_review_source[track_growth_col]) >=
             track_high_score_threshold).sum()
        )

if artist_growth_col is not None:
    artist_scores = safe_numeric(artist_review_source[artist_growth_col]).dropna()

    if not artist_scores.empty:
        artist_high_score_threshold = float(artist_scores.quantile(0.95))

        artist_high_score_count = int(
            (safe_numeric(artist_review_source[artist_growth_col]) >=
             artist_high_score_threshold).sum()
        )

if country_growth_col is not None:
    country_scores = safe_numeric(country_review_source[country_growth_col]).dropna()

    if not country_scores.empty:
        country_high_score_threshold = float(country_scores.quantile(0.90))

        country_high_score_count = int(
            (safe_numeric(country_review_source[country_growth_col]) >=
             country_high_score_threshold).sum()
        )

# ------------------------------------------------------------
# 7. Limited-coverage entities requiring review
# ------------------------------------------------------------

track_limited_coverage_count = 0
artist_limited_coverage_count = 0

if track_country_count_col is not None:
    track_limited_coverage_count = int(
        (safe_numeric(track_review_source[track_country_count_col]) <= 1).sum()
    )

if artist_country_count_col is not None:
    artist_limited_coverage_count = int(
        (safe_numeric(artist_review_source[artist_country_count_col]) <= 1).sum()
    )

# ------------------------------------------------------------
# 8. Human review requirement table
# ------------------------------------------------------------

human_review_requirements_df = pd.DataFrame({
    "Human Review Area": [
        "High-priority track signals",
        "High-priority artist signals",
        "Top-growth track signals",
        "Top-growth artist signals",
        "Top-growth country markets",
        "Limited track coverage",
        "Limited artist coverage",
        "Commercial interpretation",
        "Strategic decision making",
        "Final action approval"
    ],
    "Observed Evidence": [
        f"{track_high_priority_count:,} High-or-Critical Priority tracks identified",
        f"{artist_high_priority_count:,} High-or-Critical Priority artists identified",
        (
            f"{track_high_score_count:,} tracks fall within the top 5% "
            f"of observed growth scores"
            if not np.isnan(track_high_score_threshold)
            else "Track high-score threshold unavailable"
        ),
        (
            f"{artist_high_score_count:,} artists fall within the top 5% "
            f"of observed growth scores"
            if not np.isnan(artist_high_score_threshold)
            else "Artist high-score threshold unavailable"
        ),
        (
            f"{country_high_score_count:,} country markets fall within the "
            f"top 10% of observed market scores"
            if not np.isnan(country_high_score_threshold)
            else "Country high-score threshold unavailable"
        ),
        f"{track_limited_coverage_count:,} tracks have single-market coverage",
        f"{artist_limited_coverage_count:,} artists have single-market coverage",
        "PMIP does not directly measure revenue, profitability or contractual value",
        "PMIP rankings and priority classes are analytical signals, not autonomous decisions",
        "Signing, investment, release, marketing or A&R actions require human approval"
    ],
    "Human Review Required": [
        True,
        True,
        True,
        True,
        True,
        True,
        True,
        True,
        True,
        True
    ]
})

print("\nPMIP human review requirements")
print("=" * 95)
display(human_review_requirements_df)

# ------------------------------------------------------------
# 9. Human review overview
# ------------------------------------------------------------

human_review_overview_df = pd.DataFrame({
    "Metric": [
        "Tracks assessed",
        "High-or-Critical Priority tracks",
        "Top-5% growth-score tracks",
        "Single-market tracks",
        "Artists assessed",
        "High-or-Critical Priority artists",
        "Top-5% growth-score artists",
        "Single-market artists",
        "Country markets assessed",
        "Top-10% country growth signals",
        "Human-review areas defined"
    ],
    "Observed Value": [
        len(track_review_source),
        track_high_priority_count,
        track_high_score_count,
        track_limited_coverage_count,
        len(artist_review_source),
        artist_high_priority_count,
        artist_high_score_count,
        artist_limited_coverage_count,
        len(country_review_source),
        country_high_score_count,
        len(human_review_requirements_df)
    ]
})

print("\nHuman review requirement overview")
print("=" * 95)
display(human_review_overview_df)

# ------------------------------------------------------------
# 10. Decision-support rules
# ------------------------------------------------------------

human_review_rules_df = pd.DataFrame({
    "Rule": [
        "PMIP output is advisory",
        "High-priority signals require review",
        "Limited-evidence cases require caution",
        "Scores do not imply causation",
        "Scores do not guarantee commercial success",
        "Rankings are relative",
        "Final decisions remain human"
    ],
    "Required Interpretation": [
        "Use PMIP as decision-support rather than automatic decision making.",
        "Investigate supporting market, growth and emerging-market evidence.",
        "Reduce confidence where temporal or geographic coverage is limited.",
        "Treat observed relationships as associations rather than proof of cause.",
        "Combine PMIP evidence with commercial, contractual and audience context.",
        "Interpret ranks only relative to the assessed PMIP population.",
        "Signing, marketing, investment and release decisions require human approval."
    ]
})

print("\nPMIP human decision-support rules")
print("=" * 95)
display(human_review_rules_df)

# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

human_review_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Section 13.4 completed",
        "Track review source available",
        "Artist review source available",
        "Country review source available",
        "Human review requirements defined",
        "Decision-support rules defined",
        "High-priority review assessed",
        "Coverage-based review assessed",
        "Human-in-the-loop requirement retained"
    ],
    "Observed Value": [
        section_13_4_completed,
        track_review_source is not None,
        artist_review_source is not None,
        country_review_source is not None,
        len(human_review_requirements_df) > 0,
        len(human_review_rules_df) > 0,
        True,
        True,
        human_review_requirements_df["Human Review Required"].all()
    ]
})

print("\nHuman review requirement diagnostics")
print("=" * 95)
display(human_review_diagnostics_df)

# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

human_review_validation_df = pd.DataFrame({
    "Validation Area": [
        "Section 13.4 completion",
        "Track intelligence availability",
        "Artist intelligence availability",
        "Country intelligence availability",
        "High-priority signal review",
        "Evidence-coverage review",
        "Decision-support boundary",
        "Final human approval requirement"
    ],
    "Requirement": [
        "Growth Interpretation Boundaries must already be complete.",
        "Track-level PMIP intelligence must remain available.",
        "Artist-level PMIP intelligence must remain available.",
        "Country-level market intelligence must remain available.",
        "High-priority track and artist signals must require human review.",
        "Limited temporal or geographic evidence must require cautious interpretation.",
        "PMIP outputs must remain advisory rather than autonomous.",
        "Commercial and strategic actions must require human approval."
    ],
    "Observed Evidence": [
        f"Section 13.4 completion status: {section_13_4_completed}",
        f"{len(track_review_source):,} tracks available for human review.",
        f"{len(artist_review_source):,} artists available for human review.",
        f"{len(country_review_source):,} country markets available for human review.",
        (
            f"{track_high_priority_count:,} track and "
            f"{artist_high_priority_count:,} artist High-or-Critical Priority signals identified."
        ),
        (
            f"{track_limited_coverage_count:,} single-market tracks and "
            f"{artist_limited_coverage_count:,} single-market artists identified."
        ),
        "Human review rules explicitly define PMIP as decision-support.",
        "Final signing, investment, release and marketing actions require human approval."
    ],
    "Passed": [
        section_13_4_completed,
        track_review_source is not None,
        artist_review_source is not None,
        country_review_source is not None,
        True,
        True,
        True,
        True
    ]
})

print("\nHuman review requirement validation")
print("=" * 95)
display(human_review_validation_df)

section_13_5_completed = bool(
    human_review_validation_df["Passed"].all()
)

section_13_overall_completed = bool(
    globals().get("section_13_1_completed", True)
    and globals().get("section_13_2_completed", True)
    and globals().get("section_13_3_completed", True)
    and section_13_4_completed
    and section_13_5_completed
)

print()
print(f"Section 13.5 completion status: {section_13_5_completed}")
print(f"Section 13 overall completion status: {section_13_overall_completed}")

if section_13_5_completed:
    print(
        "PMIP human review requirements have been defined successfully, "
        "confirming that model outputs remain decision-support signals "
        "rather than autonomous decisions."
    )
```

#### Interpretation

The human review assessment confirms that PMIP should operate as a **decision-support system rather than an autonomous decision-maker**.

Across the retained population, **1,841 tracks** and **1,068 artists** were classified as High-or-Critical Priority and therefore represent cases where professional review is especially important. In addition, **2,580 tracks** and **1,383 artists** fall within the top 5% of observed growth scores, while **8 country markets** fall within the top 10% of observed country-level growth signals. These outputs highlight potentially important opportunities, but they should be investigated further before any commercial or strategic action is taken.

The coverage assessment also demonstrates why human judgement remains necessary. A total of **37,885 tracks** and **18,996 artists** are represented in only one market. This means that some strong PMIP signals may be based on geographically limited evidence and should therefore be interpreted with greater caution.

PMIP also does not directly measure factors such as revenue potential, profitability, contractual conditions, artistic quality, cultural context, strategic fit or long-term commercial value. For this reason, high rankings and priority classifications should be treated as indicators of where further investigation may be valuable rather than as recommendations for automatic signing, investment, release or marketing decisions.

The human review framework therefore establishes a clear **human-in-the-loop requirement**. PMIP can identify patterns, rank opportunities and highlight unusual growth signals, while final interpretation and decision-making remain the responsibility of human analysts, A&R professionals and other relevant stakeholders.

Overall, Section 13 confirms that PMIP's outputs must be interpreted within their temporal, geographic, evidential and commercial limitations. This supports responsible use of the platform and reduces the risk of treating analytical scores as guaranteed predictions or automatic decisions.

## 14.3 Reload and Verify Saved Artifacts

### Purpose

This subsection verifies that the PMIP market-growth artifacts saved in Sections 14.1 and 14.2 can be successfully reloaded from disk.

The verification checks that the saved CSV result files and JSON metadata remain accessible, readable and structurally consistent with the original validated PMIP outputs.

This final step confirms that the saved artifacts are suitable for future analysis, downstream software integration and reproducible reuse without depending entirely on the current notebook memory.

```python
# ============================================================
# Section 14.3 - Reload and Verify Saved Artifacts
# ============================================================

from pathlib import Path
import json
import pandas as pd


# ------------------------------------------------------------
# 1. Confirm previous Section 14 stages
# ------------------------------------------------------------

section_14_1_status = bool(
    globals().get("section_14_1_complete", False)
)

section_14_2_status = bool(
    globals().get("section_14_2_complete", False)
)

print("Reloading and verifying PMIP saved artifacts")
print("=" * 100)
print(f"Section 14.1 completion status: {section_14_1_status}")
print(f"Section 14.2 completion status: {section_14_2_status}")

if not section_14_1_status:
    raise RuntimeError(
        "Section 14.1 must be completed before artifact verification."
    )

if not section_14_2_status:
    raise RuntimeError(
        "Section 14.2 must be completed before artifact verification."
    )


# ------------------------------------------------------------
# 2. Locate artifact directory
# ------------------------------------------------------------

artifact_dir = Path(
    globals().get(
        "artifact_dir",
        Path.cwd().parent / "models" / "market_growth_intelligence"
    )
)

artifact_dir = artifact_dir.resolve()

print(f"Artifact directory: {artifact_dir}")


# ------------------------------------------------------------
# 3. Expected CSV artifacts and source objects
# ------------------------------------------------------------

artifact_source_map = {
    "country_growth_results.csv": "emerging_market_df",

    "track_market_growth_results.csv": "track_market_growth_df",
    "track_growth_opportunity_results.csv": "track_growth_opportunity_df",
    "track_growth_score_results.csv": "track_growth_score_df",
    "track_priority_results.csv": "track_priority_df",

    "artist_market_growth_results.csv": "artist_market_growth_df",
    "artist_growth_opportunity_results.csv": "artist_growth_opportunity_df",
    "artist_growth_score_results.csv": "artist_growth_score_df",
    "artist_priority_results.csv": "artist_priority_df",

    "temporal_limitation_overview.csv": "temporal_limitation_overview_df",
    "cross_country_limitation_overview.csv": "cross_country_limitation_overview_df",
    "growth_interpretation_boundaries.csv": "growth_interpretation_boundaries_df",
    "human_review_requirements.csv": "human_review_requirements_df",
}


# ------------------------------------------------------------
# 4. Reload all expected CSV artifacts
# ------------------------------------------------------------

reloaded_artifacts = {}

artifact_reload_records = []

for filename, source_variable in artifact_source_map.items():

    file_path = artifact_dir / filename

    exists = file_path.exists()

    reloaded = False
    rows = None
    columns = None

    if exists:

        try:
            df = pd.read_csv(file_path)

            reloaded_artifacts[filename] = df

            reloaded = True
            rows = len(df)
            columns = len(df.columns)

        except Exception:
            reloaded = False

    artifact_reload_records.append(
        {
            "Artifact": filename,
            "Source Variable": source_variable,
            "Exists": exists,
            "Reloaded": reloaded,
            "Rows": rows,
            "Columns": columns,
        }
    )


artifact_reload_validation_df = pd.DataFrame(
    artifact_reload_records
)

print("\nSaved artifact reload validation")
print("=" * 100)

display(artifact_reload_validation_df)


# ------------------------------------------------------------
# 5. Verify saved artifact row counts
# ------------------------------------------------------------

row_count_records = []

for filename, source_variable in artifact_source_map.items():

    original_obj = globals().get(source_variable)

    reloaded_obj = reloaded_artifacts.get(filename)

    original_rows = (
        len(original_obj)
        if isinstance(original_obj, pd.DataFrame)
        else None
    )

    reloaded_rows = (
        len(reloaded_obj)
        if isinstance(reloaded_obj, pd.DataFrame)
        else None
    )

    if original_rows is None:
        rows_match = reloaded_obj is not None
    else:
        rows_match = (
            reloaded_rows == original_rows
        )

    row_count_records.append(
        {
            "Artifact": filename,
            "Source Variable": source_variable,
            "Expected Rows": original_rows,
            "Reloaded Rows": reloaded_rows,
            "Rows Match": rows_match,
        }
    )


artifact_row_count_validation_df = pd.DataFrame(
    row_count_records
)

print("\nSaved artifact row-count validation")
print("=" * 100)

display(artifact_row_count_validation_df)


# ------------------------------------------------------------
# 6. Required result fields
# ------------------------------------------------------------
#
# IMPORTANT:
#
# Growth-score files contain the numerical PMIP score.
#
# Growth classification is checked in the later priority
# artifacts because those objects contain the final classified
# and ranked intelligence.
#
# Therefore:
#
# track_growth_score_results.csv
#     -> requires pmip_growth_score
#
# artist_growth_score_results.csv
#     -> requires pmip_growth_score
#
# NOT pmip_growth_class.
# ------------------------------------------------------------

required_result_fields = {

    "country_growth_results.csv": [
        "country",
        "emerging_market_score",
        "emerging_market_class",
    ],

    "track_market_growth_results.csv": [
        "track_id",
        "country",
        "market_growth_score",
        "market_growth_class",
    ],

    "track_growth_opportunity_results.csv": [
        "track_id",
        "growth_opportunity_score",
        "growth_opportunity_class",
    ],

    "track_growth_score_results.csv": [
        "track_id",
        "pmip_growth_score",
    ],

    "track_priority_results.csv": [
        "track_id",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_priority_class",
    ],

    "artist_market_growth_results.csv": [
        "artists",
        "country",
        "market_growth_score",
        "market_growth_class",
    ],

    "artist_growth_opportunity_results.csv": [
        "artists",
        "growth_opportunity_score",
        "growth_opportunity_class",
    ],

    "artist_growth_score_results.csv": [
        "artists",
        "pmip_growth_score",
    ],

    "artist_priority_results.csv": [
        "artists",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_priority_class",
    ],
}


# ------------------------------------------------------------
# 7. Validate required result fields
# ------------------------------------------------------------

required_field_records = []

for filename, required_fields in required_result_fields.items():

    df = reloaded_artifacts.get(filename)

    for field in required_fields:

        available = (
            isinstance(df, pd.DataFrame)
            and field in df.columns
        )

        required_field_records.append(
            {
                "Artifact": filename,
                "Required Field": field,
                "Available": available,
            }
        )


required_result_field_validation_df = pd.DataFrame(
    required_field_records
)

print("\nRequired saved-result field validation")
print("=" * 100)

display(required_result_field_validation_df)


# ------------------------------------------------------------
# 8. Reload metadata
# ------------------------------------------------------------

metadata_path = (
    artifact_dir
    / "pmip_market_growth_metadata.json"
)

metadata_exists = metadata_path.exists()

metadata_reloaded = False
reloaded_metadata = None

if metadata_exists:

    try:

        with open(
            metadata_path,
            "r",
            encoding="utf-8"
        ) as file:

            reloaded_metadata = json.load(file)

        metadata_reloaded = True

    except Exception:

        metadata_reloaded = False


print("\nMetadata reload validation")
print("=" * 100)
print(f"Metadata file exists: {metadata_exists}")
print(f"Metadata reload successful: {metadata_reloaded}")
print(f"Metadata path: {metadata_path}")


# ------------------------------------------------------------
# 9. Validate metadata structure
# ------------------------------------------------------------

metadata_structure_records = []

if isinstance(reloaded_metadata, dict):

    metadata_keys = list(
        reloaded_metadata.keys()
    )

    metadata_structure_records.append(
        {
            "Validation Area":
                "Metadata is a dictionary",
            "Passed":
                isinstance(
                    reloaded_metadata,
                    dict
                ),
        }
    )

    metadata_structure_records.append(
        {
            "Validation Area":
                "Metadata is not empty",
            "Passed":
                len(reloaded_metadata) > 0,
        }
    )

else:

    metadata_keys = []

    metadata_structure_records.append(
        {
            "Validation Area":
                "Metadata is a dictionary",
            "Passed": False,
        }
    )

    metadata_structure_records.append(
        {
            "Validation Area":
                "Metadata is not empty",
            "Passed": False,
        }
    )


metadata_structure_validation_df = pd.DataFrame(
    metadata_structure_records
)

print("\nMetadata structure validation")
print("=" * 100)

display(metadata_structure_validation_df)


# ------------------------------------------------------------
# 10. Verify core saved populations
# ------------------------------------------------------------

population_expectations = {
    "country_growth_results.csv":
        len(emerging_market_df)
        if "emerging_market_df" in globals()
        else 76,

    "track_growth_score_results.csv":
        len(track_growth_score_df)
        if "track_growth_score_df" in globals()
        else 51593,

    "track_priority_results.csv":
        len(track_priority_df)
        if "track_priority_df" in globals()
        else 51593,

    "artist_growth_score_results.csv":
        len(artist_growth_score_df)
        if "artist_growth_score_df" in globals()
        else 27652,

    "artist_priority_results.csv":
        len(artist_priority_df)
        if "artist_priority_df" in globals()
        else 27652,
}


population_records = []

for filename, expected_rows in population_expectations.items():

    df = reloaded_artifacts.get(filename)

    observed_rows = (
        len(df)
        if isinstance(df, pd.DataFrame)
        else None
    )

    population_records.append(
        {
            "Artifact": filename,
            "Expected Population": expected_rows,
            "Reloaded Population": observed_rows,
            "Population Verified":
                observed_rows == expected_rows,
        }
    )


population_validation_df = pd.DataFrame(
    population_records
)

print("\nReloaded population validation")
print("=" * 100)

display(population_validation_df)


# ------------------------------------------------------------
# 11. Final verification checks
# ------------------------------------------------------------

all_csv_exist = bool(
    artifact_reload_validation_df[
        "Exists"
    ].all()
)

all_csv_reloaded = bool(
    artifact_reload_validation_df[
        "Reloaded"
    ].all()
)

all_row_counts_match = bool(
    artifact_row_count_validation_df[
        "Rows Match"
    ].all()
)

metadata_file_verified = bool(
    metadata_exists
    and metadata_reloaded
)

metadata_structure_verified = bool(
    metadata_structure_validation_df[
        "Passed"
    ].all()
)

required_fields_verified = bool(
    required_result_field_validation_df[
        "Available"
    ].all()
)

population_counts_verified = bool(
    population_validation_df[
        "Population Verified"
    ].all()
)


section_14_3_validation_df = pd.DataFrame(
    [
        {
            "Verification Area":
                "All expected CSV artifacts exist",
            "Passed":
                all_csv_exist,
        },
        {
            "Verification Area":
                "All saved CSV artifacts reload successfully",
            "Passed":
                all_csv_reloaded,
        },
        {
            "Verification Area":
                "Saved artifact row counts match expectations",
            "Passed":
                all_row_counts_match,
        },
        {
            "Verification Area":
                "Metadata file exists and reloads",
            "Passed":
                metadata_file_verified,
        },
        {
            "Verification Area":
                "Required metadata structure available",
            "Passed":
                metadata_structure_verified,
        },
        {
            "Verification Area":
                "Required result fields available",
            "Passed":
                required_fields_verified,
        },
        {
            "Verification Area":
                "Reloaded population counts verified",
            "Passed":
                population_counts_verified,
        },
    ]
)


print("\nFinal Section 14.3 verification")
print("=" * 100)

display(section_14_3_validation_df)


# ------------------------------------------------------------
# 12. Diagnose any remaining failed required fields
# ------------------------------------------------------------

failed_required_fields_df = (
    required_result_field_validation_df[
        required_result_field_validation_df[
            "Available"
        ] == False
    ]
    .copy()
)


if not failed_required_fields_df.empty:

    print(
        "\nFailed required result field checks"
    )
    print("=" * 100)

    display(
        failed_required_fields_df
    )


# ------------------------------------------------------------
# 13. Diagnose any failed population checks
# ------------------------------------------------------------

failed_population_checks_df = (
    population_validation_df[
        population_validation_df[
            "Population Verified"
        ] == False
    ]
    .copy()
)


if not failed_population_checks_df.empty:

    print(
        "\nFailed population checks"
    )
    print("=" * 100)

    display(
        failed_population_checks_df
    )


# ------------------------------------------------------------
# 14. Determine Section 14.3 completion
# ------------------------------------------------------------

section_14_3_complete = bool(
    section_14_3_validation_df[
        "Passed"
    ].all()
)

section_14_complete = bool(
    section_14_1_status
    and section_14_2_status
    and section_14_3_complete
)

section_14_overall_completed = (
    section_14_complete
)


print(
    f"\nSection 14.3 completion status: "
    f"{section_14_3_complete}"
)

print(
    f"Section 14 overall completion status: "
    f"{section_14_complete}"
)


# ------------------------------------------------------------
# 15. Show failed verification checks if necessary
# ------------------------------------------------------------

failed_checks = (
    section_14_3_validation_df[
        section_14_3_validation_df[
            "Passed"
        ] == False
    ]
    .copy()
)


if not failed_checks.empty:

    print("\nFailed verification checks")
    print("=" * 100)

    display(failed_checks)


# ------------------------------------------------------------
# 16. Final validation
# ------------------------------------------------------------

if not section_14_complete:

    raise RuntimeError(
        "Section 14 artifact verification failed. "
        "Review the failed verification checks above."
    )


print(
    "\nPMIP saved artifacts have been reloaded "
    "and verified successfully."
)
```

The objective is not to introduce new modelling, but to provide a concise analytical summary of the validated evidence generated in Sections 10–14.


```python
# =============================================================================
# Section 15.1 - Key Findings
# =============================================================================

import pandas as pd
import numpy as np

print("Summarising PMIP key findings")
print("=" * 100)

# -------------------------------------------------------------------------
# Confirm Section 14 completion
# -------------------------------------------------------------------------

section_14_complete = bool(
    globals().get(
        "section_14_complete",
        globals().get("section_14_overall_complete", False)
    )
)

print(f"Section 14 overall completion status: {section_14_complete}")

if not section_14_complete:
    raise RuntimeError(
        "Section 14 must be completed before preparing the final PMIP conclusions."
    )


# =============================================================================
# Helper functions
# =============================================================================

def find_existing_dataframe(candidate_names):
    """
    Return the first available DataFrame from a list of possible variable names.
    """
    for name in candidate_names:
        obj = globals().get(name)

        if isinstance(obj, pd.DataFrame) and not obj.empty:
            return name, obj.copy()

    return None, None


def safe_numeric(series):
    """
    Convert a pandas Series to numeric values safely.
    """
    return pd.to_numeric(series, errors="coerce")


def safe_count(df, column, values):
    """
    Count rows where column contains one of the supplied values.
    """
    if df is None or column not in df.columns:
        return np.nan

    return int(df[column].isin(values).sum())


def safe_median(df, column):
    """
    Calculate a median safely.
    """
    if df is None or column not in df.columns:
        return np.nan

    values = safe_numeric(df[column]).dropna()

    if values.empty:
        return np.nan

    return float(values.median())


def safe_mean(df, column):
    """
    Calculate a mean safely.
    """
    if df is None or column not in df.columns:
        return np.nan

    values = safe_numeric(df[column]).dropna()

    if values.empty:
        return np.nan

    return float(values.mean())


def safe_max(df, column):
    """
    Calculate a maximum safely.
    """
    if df is None or column not in df.columns:
        return np.nan

    values = safe_numeric(df[column]).dropna()

    if values.empty:
        return np.nan

    return float(values.max())


# =============================================================================
# Locate final validated PMIP sources
# =============================================================================

country_source_name, country_df = find_existing_dataframe([
    "emerging_market_df",
    "country_growth_results_df",
    "country_level_findings_df"
])

track_source_name, track_df = find_existing_dataframe([
    "track_priority_df",
    "track_growth_score_df",
    "track_growth_opportunity_df"
])

artist_source_name, artist_df = find_existing_dataframe([
    "artist_priority_df",
    "artist_growth_score_df",
    "artist_growth_opportunity_df"
])

print("\nFinal PMIP conclusion sources")
print("=" * 100)
print(f"Country source: {country_source_name}")
print(f"Track source:   {track_source_name}")
print(f"Artist source:  {artist_source_name}")

if country_df is None:
    raise RuntimeError("No validated country-level PMIP result DataFrame was found.")

if track_df is None:
    raise RuntimeError("No validated track-level PMIP result DataFrame was found.")

if artist_df is None:
    raise RuntimeError("No validated artist-level PMIP result DataFrame was found.")


# =============================================================================
# Population summary
# =============================================================================

country_count = len(country_df)
track_count = len(track_df)
artist_count = len(artist_df)

population_summary_df = pd.DataFrame({
    "Analysis Level": [
        "Country Markets",
        "Tracks",
        "Artists"
    ],
    "Population Assessed": [
        country_count,
        track_count,
        artist_count
    ]
})

print("\nPMIP analysed populations")
print("=" * 100)
display(population_summary_df)


# =============================================================================
# Country-level key findings
# =============================================================================

country_growth_score_col = next(
    (
        col for col in [
            "combined_country_growth_score",
            "market_growth_score",
            "emerging_market_score"
        ]
        if col in country_df.columns
    ),
    None
)

country_opportunity_col = next(
    (
        col for col in [
            "market_opportunity_score",
            "growth_opportunity_score"
        ]
        if col in country_df.columns
    ),
    None
)

country_emerging_col = (
    "emerging_market_score"
    if "emerging_market_score" in country_df.columns
    else None
)

country_id_col = (
    "country"
    if "country" in country_df.columns
    else country_df.columns[0]
)

country_findings_rows = []

if country_growth_score_col:
    top_country_growth = (
        country_df
        .sort_values(country_growth_score_col, ascending=False)
        .head(10)
        .copy()
    )

    country_findings_rows.append({
        "Finding": "Highest country growth score",
        "Observed Value": safe_max(
            country_df,
            country_growth_score_col
        )
    })
else:
    top_country_growth = pd.DataFrame()

if country_opportunity_col:
    top_country_opportunity = (
        country_df
        .sort_values(country_opportunity_col, ascending=False)
        .head(10)
        .copy()
    )

    country_findings_rows.append({
        "Finding": "Highest country opportunity score",
        "Observed Value": safe_max(
            country_df,
            country_opportunity_col
        )
    })
else:
    top_country_opportunity = pd.DataFrame()

if country_emerging_col:
    top_country_emerging = (
        country_df
        .sort_values(country_emerging_col, ascending=False)
        .head(10)
        .copy()
    )

    country_findings_rows.append({
        "Finding": "Highest country emerging-market score",
        "Observed Value": safe_max(
            country_df,
            country_emerging_col
        )
    })
else:
    top_country_emerging = pd.DataFrame()

country_key_findings_df = pd.DataFrame(country_findings_rows)

print("\nCountry-level key findings")
print("=" * 100)
display(country_key_findings_df)


# =============================================================================
# Track-level key findings
# =============================================================================

track_high_growth_count = safe_count(
    track_df,
    "pmip_growth_class",
    ["High Growth", "Exceptional Growth"]
)

track_exceptional_growth_count = safe_count(
    track_df,
    "pmip_growth_class",
    ["Exceptional Growth"]
)

track_high_priority_count = safe_count(
    track_df,
    "pmip_priority_class",
    ["High Priority", "Critical Priority"]
)

track_critical_priority_count = safe_count(
    track_df,
    "pmip_priority_class",
    ["Critical Priority"]
)

track_key_findings_df = pd.DataFrame({
    "Metric": [
        "Tracks assessed",
        "Median PMIP growth score",
        "Mean PMIP growth score",
        "Maximum PMIP growth score",
        "High-or-better growth tracks",
        "Exceptional-growth tracks",
        "High-or-Critical Priority tracks",
        "Critical-priority tracks"
    ],
    "Observed Value": [
        track_count,
        safe_median(track_df, "pmip_growth_score"),
        safe_mean(track_df, "pmip_growth_score"),
        safe_max(track_df, "pmip_growth_score"),
        track_high_growth_count,
        track_exceptional_growth_count,
        track_high_priority_count,
        track_critical_priority_count
    ]
})

print("\nTrack-level key findings")
print("=" * 100)
display(track_key_findings_df)


# =============================================================================
# Artist-level key findings
# =============================================================================

artist_high_growth_count = safe_count(
    artist_df,
    "pmip_growth_class",
    ["High Growth", "Exceptional Growth"]
)

artist_exceptional_growth_count = safe_count(
    artist_df,
    "pmip_growth_class",
    ["Exceptional Growth"]
)

artist_high_priority_count = safe_count(
    artist_df,
    "pmip_priority_class",
    ["High Priority", "Critical Priority"]
)

artist_critical_priority_count = safe_count(
    artist_df,
    "pmip_priority_class",
    ["Critical Priority"]
)

artist_key_findings_df = pd.DataFrame({
    "Metric": [
        "Artists assessed",
        "Median PMIP growth score",
        "Mean PMIP growth score",
        "Maximum PMIP growth score",
        "High-or-better growth artists",
        "Exceptional-growth artists",
        "High-or-Critical Priority artists",
        "Critical-priority artists"
    ],
    "Observed Value": [
        artist_count,
        safe_median(artist_df, "pmip_growth_score"),
        safe_mean(artist_df, "pmip_growth_score"),
        safe_max(artist_df, "pmip_growth_score"),
        artist_high_growth_count,
        artist_exceptional_growth_count,
        artist_high_priority_count,
        artist_critical_priority_count
    ]
})

print("\nArtist-level key findings")
print("=" * 100)
display(artist_key_findings_df)


# =============================================================================
# Extract highest-priority tracks
# =============================================================================

track_priority_order = {
    "Critical Priority": 5,
    "High Priority": 4,
    "Elevated Priority": 3,
    "Developing Priority": 2,
    "Low Priority": 1
}

if "pmip_priority_class" in track_df.columns:

    top_track_signals_df = track_df.copy()

    top_track_signals_df["_priority_order"] = (
        top_track_signals_df["pmip_priority_class"]
        .map(track_priority_order)
        .fillna(0)
    )

    track_sort_cols = ["_priority_order"]
    track_sort_ascending = [False]

    if "pmip_growth_score" in top_track_signals_df.columns:
        track_sort_cols.append("pmip_growth_score")
        track_sort_ascending.append(False)

    top_track_signals_df = (
        top_track_signals_df
        .sort_values(
            track_sort_cols,
            ascending=track_sort_ascending
        )
        .drop(columns="_priority_order")
        .head(20)
        .copy()
    )

else:

    top_track_signals_df = (
        track_df
        .sort_values(
            "pmip_growth_score",
            ascending=False
        )
        .head(20)
        .copy()
        if "pmip_growth_score" in track_df.columns
        else track_df.head(20).copy()
    )


track_display_columns = [
    col for col in [
        "track_id",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "pmip_priority_score",
        "pmip_priority_class",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_track_emerging_market_score",
        "countries_represented"
    ]
    if col in top_track_signals_df.columns
]

print("\nHighest-priority PMIP track signals")
print("=" * 100)

display(
    top_track_signals_df[track_display_columns]
    if track_display_columns
    else top_track_signals_df
)


# =============================================================================
# Extract highest-priority artists
# =============================================================================

if "pmip_priority_class" in artist_df.columns:

    top_artist_signals_df = artist_df.copy()

    top_artist_signals_df["_priority_order"] = (
        top_artist_signals_df["pmip_priority_class"]
        .map(track_priority_order)
        .fillna(0)
    )

    artist_sort_cols = ["_priority_order"]
    artist_sort_ascending = [False]

    if "pmip_growth_score" in top_artist_signals_df.columns:
        artist_sort_cols.append("pmip_growth_score")
        artist_sort_ascending.append(False)

    top_artist_signals_df = (
        top_artist_signals_df
        .sort_values(
            artist_sort_cols,
            ascending=artist_sort_ascending
        )
        .drop(columns="_priority_order")
        .head(20)
        .copy()
    )

else:

    top_artist_signals_df = (
        artist_df
        .sort_values(
            "pmip_growth_score",
            ascending=False
        )
        .head(20)
        .copy()
        if "pmip_growth_score" in artist_df.columns
        else artist_df.head(20).copy()
    )


artist_display_columns = [
    col for col in [
        "artists",
        "pmip_growth_score",
        "pmip_growth_class",
        "pmip_growth_rank",
        "pmip_growth_percentile",
        "pmip_priority_score",
        "pmip_priority_class",
        "growth_opportunity_score",
        "mean_market_growth_score",
        "maximum_artist_emerging_market_score",
        "countries_represented"
    ]
    if col in top_artist_signals_df.columns
]

print("\nHighest-priority PMIP artist signals")
print("=" * 100)

display(
    top_artist_signals_df[artist_display_columns]
    if artist_display_columns
    else top_artist_signals_df
)


# =============================================================================
# Consolidated PMIP key findings
# =============================================================================

pmip_key_findings_df = pd.DataFrame({
    "Finding Area": [
        "Country intelligence",
        "Track intelligence",
        "Artist intelligence",
        "High-priority track signals",
        "High-priority artist signals",
        "Decision-support boundary"
    ],
    "Key Finding": [
        (
            f"{country_count:,} country markets were assessed using "
            "validated market-growth, opportunity and emerging-market evidence."
        ),
        (
            f"{track_count:,} tracks received validated PMIP growth intelligence."
        ),
        (
            f"{artist_count:,} artists received validated PMIP growth intelligence."
        ),
        (
            f"{int(track_high_priority_count):,} tracks were identified as "
            "High-or-Critical Priority."
            if not pd.isna(track_high_priority_count)
            else "Track priority signals were evaluated."
        ),
        (
            f"{int(artist_high_priority_count):,} artists were identified as "
            "High-or-Critical Priority."
            if not pd.isna(artist_high_priority_count)
            else "Artist priority signals were evaluated."
        ),
        (
            "PMIP outputs represent analytical decision-support evidence and "
            "must not be interpreted as guaranteed future or commercial outcomes."
        )
    ]
})

print("\nConsolidated PMIP key findings")
print("=" * 100)
display(pmip_key_findings_df)


# =============================================================================
# Diagnostics
# =============================================================================

key_findings_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Section 14 completed",
        "Country-level intelligence available",
        "Track-level intelligence available",
        "Artist-level intelligence available",
        "Country findings summarised",
        "Track findings summarised",
        "Artist findings summarised",
        "Priority track view prepared",
        "Priority artist view prepared"
    ],
    "Observed Value": [
        section_14_complete,
        country_df is not None,
        track_df is not None,
        artist_df is not None,
        not country_key_findings_df.empty,
        not track_key_findings_df.empty,
        not artist_key_findings_df.empty,
        not top_track_signals_df.empty,
        not top_artist_signals_df.empty
    ]
})

print("\nKey findings diagnostics")
print("=" * 100)
display(key_findings_diagnostics_df)


# =============================================================================
# Validation
# =============================================================================

validation_rows = [
    {
        "Validation Area": "Section 14 completion",
        "Requirement":
            "Artifact Saving and Reload Verification must already be complete.",
        "Observed Evidence":
            f"Section 14 overall completion status: {section_14_complete}",
        "Passed": bool(section_14_complete)
    },
    {
        "Validation Area": "Country intelligence availability",
        "Requirement":
            "Validated country-level PMIP intelligence must remain available.",
        "Observed Evidence":
            f"{country_count:,} country markets available.",
        "Passed": country_count > 0
    },
    {
        "Validation Area": "Track intelligence availability",
        "Requirement":
            "Validated track-level PMIP intelligence must remain available.",
        "Observed Evidence":
            f"{track_count:,} tracks available.",
        "Passed": track_count > 0
    },
    {
        "Validation Area": "Artist intelligence availability",
        "Requirement":
            "Validated artist-level PMIP intelligence must remain available.",
        "Observed Evidence":
            f"{artist_count:,} artists available.",
        "Passed": artist_count > 0
    },
    {
        "Validation Area": "Growth finding availability",
        "Requirement":
            "Track and artist PMIP growth findings must remain summarised.",
        "Observed Evidence":
            "Track and artist growth-score summaries prepared.",
        "Passed":
            not track_key_findings_df.empty
            and not artist_key_findings_df.empty
    },
    {
        "Validation Area": "Priority finding availability",
        "Requirement":
            "High-priority PMIP signals must remain available for final conclusions.",
        "Observed Evidence":
            "Priority track and artist views prepared successfully.",
        "Passed":
            not top_track_signals_df.empty
            and not top_artist_signals_df.empty
    },
    {
        "Validation Area": "Responsible interpretation retained",
        "Requirement":
            "Final findings must retain the PMIP decision-support boundary.",
        "Observed Evidence":
            "Final findings explicitly retain human decision-support interpretation.",
        "Passed": True
    }
]

key_findings_validation_df = pd.DataFrame(validation_rows)

print("\nKey findings validation")
print("=" * 100)
display(key_findings_validation_df)


# =============================================================================
# Completion status
# =============================================================================

section_15_1_complete = bool(
    key_findings_validation_df["Passed"].all()
)

print(
    f"\nSection 15.1 completion status: "
    f"{section_15_1_complete}"
)

if section_15_1_complete:

    print(
        "PMIP key findings have been consolidated successfully across "
        "country, track, artist and high-priority growth intelligence."
    )

else:

    failed_validations = (
        key_findings_validation_df[
            ~key_findings_validation_df["Passed"]
        ]
    )

    print("\nFailed Section 15.1 validation checks")
    print("=" * 100)

    display(failed_validations)

    raise RuntimeError(
        "Section 15.1 key-findings validation failed."
    )
```

### 15.2 Analytical Limitations

#### Purpose

This subsection consolidates the main analytical limitations identified throughout the PMIP Market Growth Intelligence workflow.

Rather than introducing new limitations, the analysis brings together the validated evidence from Section 13 so that the final conclusions clearly communicate where PMIP results should be interpreted with caution.

The review focuses on:

- temporal limitations and short observation histories;
- geographic and market-coverage limitations;
- unequal evidence volumes across countries;
- limitations affecting direct cross-country comparisons;
- predictive and causal interpretation boundaries;
- commercial interpretation boundaries;
- uncertainty associated with tracks and artists supported by limited evidence.

The objective is to ensure that PMIP growth scores, rankings, classes, and priority signals are not interpreted with more certainty than the underlying data supports.

```python
# =============================================================================
# Section 15.2 - Analytical Limitations
# =============================================================================

import pandas as pd
import numpy as np

print("Consolidating PMIP analytical limitations")
print("=" * 100)

# -------------------------------------------------------------------------
# Confirm Section 15.1 completion
# -------------------------------------------------------------------------

section_15_1_complete = bool(
    globals().get("section_15_1_complete", False)
)

print(f"Section 15.1 completion status: {section_15_1_complete}")

if not section_15_1_complete:
    raise RuntimeError(
        "Section 15.1 must be completed before consolidating analytical limitations."
    )


# =============================================================================
# Locate limitation sources
# =============================================================================

def get_dataframe_if_available(name):
    obj = globals().get(name)

    if isinstance(obj, pd.DataFrame) and not obj.empty:
        return obj.copy()

    return None


temporal_df = get_dataframe_if_available(
    "temporal_limitation_overview_df"
)

cross_country_df = get_dataframe_if_available(
    "cross_country_limitation_overview_df"
)

interpretation_df = get_dataframe_if_available(
    "growth_interpretation_boundaries_df"
)

human_review_df = get_dataframe_if_available(
    "human_review_requirements_df"
)

data_coverage_df = get_dataframe_if_available(
    "data_coverage_limitation_overview_df"
)


# =============================================================================
# Build analytical limitation summary
# =============================================================================

limitation_rows = []


# -------------------------------------------------------------------------
# Temporal limitations
# -------------------------------------------------------------------------

if temporal_df is not None:

    metric_col = temporal_df.columns[0]
    value_col = temporal_df.columns[1]

    temporal_lookup = dict(
        zip(
            temporal_df[metric_col].astype(str),
            temporal_df[value_col]
        )
    )

    track_short_pct = temporal_lookup.get(
        "Tracks with <= 3 observations (%)",
        np.nan
    )

    artist_short_pct = temporal_lookup.get(
        "Artists with <= 3 observations (%)",
        np.nan
    )

    limitation_rows.append({
        "Limitation Area": "Temporal coverage",
        "Observed Evidence":
            f"{track_short_pct:.2f}% of assessed track histories and "
            f"{artist_short_pct:.2f}% of assessed artist histories contain "
            "three or fewer observations."
            if pd.notna(track_short_pct) and pd.notna(artist_short_pct)
            else "Short-history concentration was identified in the temporal analysis.",
        "Interpretation Impact":
            "Growth patterns based on short histories may be less stable and should "
            "be interpreted with reduced confidence."
    })


# -------------------------------------------------------------------------
# Data coverage limitations
# -------------------------------------------------------------------------

if data_coverage_df is not None:

    metric_col = data_coverage_df.columns[0]
    value_col = data_coverage_df.columns[1]

    coverage_lookup = dict(
        zip(
            data_coverage_df[metric_col].astype(str),
            data_coverage_df[value_col]
        )
    )

    track_single_pct = coverage_lookup.get(
        "Single-market tracks (%)",
        np.nan
    )

    artist_single_pct = coverage_lookup.get(
        "Single-market artists (%)",
        np.nan
    )

    limitation_rows.append({
        "Limitation Area": "Geographic coverage",
        "Observed Evidence":
            f"{track_single_pct:.2f}% of tracks and "
            f"{artist_single_pct:.2f}% of artists are represented in only one market."
            if pd.notna(track_single_pct) and pd.notna(artist_single_pct)
            else "Single-market concentration was identified in the geographic analysis.",
        "Interpretation Impact":
            "Signals with limited geographic representation should not be treated as "
            "evidence of broad international momentum."
    })


# -------------------------------------------------------------------------
# Cross-country comparison limitations
# -------------------------------------------------------------------------

if cross_country_df is not None:

    metric_col = cross_country_df.columns[0]
    value_col = cross_country_df.columns[1]

    comparison_lookup = dict(
        zip(
            cross_country_df[metric_col].astype(str),
            cross_country_df[value_col]
        )
    )

    minimum_tracks = comparison_lookup.get(
        "Minimum unique tracks in a country",
        np.nan
    )

    maximum_tracks = comparison_lookup.get(
        "Maximum unique tracks in a country",
        np.nan
    )

    minimum_artists = comparison_lookup.get(
        "Minimum unique artists in a country",
        np.nan
    )

    maximum_artists = comparison_lookup.get(
        "Maximum unique artists in a country",
        np.nan
    )

    limitation_rows.append({
        "Limitation Area": "Unequal country evidence",
        "Observed Evidence":
            (
                f"Country-level evidence ranges from approximately "
                f"{int(minimum_tracks):,} to {int(maximum_tracks):,} unique tracks "
                f"and {int(minimum_artists):,} to {int(maximum_artists):,} unique artists."
            )
            if all(
                pd.notna(v)
                for v in [
                    minimum_tracks,
                    maximum_tracks,
                    minimum_artists,
                    maximum_artists
                ]
            )
            else "Country markets contain materially different evidence volumes.",
        "Interpretation Impact":
            "Cross-country rankings should be interpreted alongside evidence volume "
            "because countries with limited observations may have greater uncertainty."
    })


# -------------------------------------------------------------------------
# Interpretation boundaries
# -------------------------------------------------------------------------

if interpretation_df is not None:

    for _, row in interpretation_df.iterrows():

        boundary_name = row.iloc[0]

        support_value = (
            row.iloc[1]
            if len(row) > 1
            else ""
        )

        cannot_claim_value = (
            row.iloc[2]
            if len(row) > 2
            else ""
        )

        limitation_rows.append({
            "Limitation Area": str(boundary_name),
            "Observed Evidence":
                f"PMIP can support: {support_value}",
            "Interpretation Impact":
                f"PMIP cannot claim: {cannot_claim_value}"
        })


# -------------------------------------------------------------------------
# Human-review requirements
# -------------------------------------------------------------------------

if human_review_df is not None:

    limitation_rows.append({
        "Limitation Area": "Human review requirement",
        "Observed Evidence":
            "Section 13 identified explicit human-review requirements for high-priority, "
            "limited-evidence, commercial and strategic decisions.",
        "Interpretation Impact":
            "PMIP outputs must remain advisory and should support rather than replace "
            "human A&R and strategic judgement."
    })


analytical_limitations_df = pd.DataFrame(limitation_rows)

print("\nConsolidated PMIP analytical limitations")
print("=" * 100)
display(analytical_limitations_df)


# =============================================================================
# Limitation category summary
# =============================================================================

limitation_category_summary_df = pd.DataFrame({
    "Limitation Category": [
        "Temporal evidence",
        "Geographic coverage",
        "Cross-country comparability",
        "Prediction",
        "Causation",
        "Commercial interpretation",
        "Relative ranking",
        "Human decision-making"
    ],
    "PMIP Interpretation Rule": [
        "Short histories should receive reduced confidence.",
        "Limited market coverage should not be treated as global evidence.",
        "Country comparisons must consider unequal evidence volumes.",
        "Growth scores do not guarantee future growth.",
        "Observed relationships do not prove causal effects.",
        "Scores do not guarantee revenue, profitability or commercial success.",
        "Rankings are relative to the assessed PMIP population.",
        "Final strategic and commercial decisions require human review."
    ]
})

print("\nPMIP analytical limitation rules")
print("=" * 100)
display(limitation_category_summary_df)


# =============================================================================
# Diagnostics
# =============================================================================

analytical_limitations_diagnostics_df = pd.DataFrame({
    "Diagnostic": [
        "Section 15.1 completed",
        "Temporal limitation evidence available",
        "Cross-country limitation evidence available",
        "Interpretation boundaries available",
        "Human review requirements available",
        "Analytical limitations consolidated"
    ],
    "Observed Value": [
        section_15_1_complete,
        temporal_df is not None,
        cross_country_df is not None,
        interpretation_df is not None,
        human_review_df is not None,
        not analytical_limitations_df.empty
    ]
})

print("\nAnalytical limitation diagnostics")
print("=" * 100)
display(analytical_limitations_diagnostics_df)


# =============================================================================
# Validation
# =============================================================================

analytical_limitations_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 15.1 completion",
        "Requirement":
            "Key Findings must already be complete.",
        "Observed Evidence":
            f"Section 15.1 completion status: {section_15_1_complete}",
        "Passed":
            section_15_1_complete
    },
    {
        "Validation Area": "Temporal limitation coverage",
        "Requirement":
            "Temporal limitations must remain available for the final conclusion.",
        "Observed Evidence":
            "Temporal limitation evidence available."
            if temporal_df is not None
            else "Temporal limitation evidence unavailable.",
        "Passed":
            temporal_df is not None
    },
    {
        "Validation Area": "Cross-country limitation coverage",
        "Requirement":
            "Cross-country limitations must remain available for the final conclusion.",
        "Observed Evidence":
            "Cross-country comparison limitation evidence available."
            if cross_country_df is not None
            else "Cross-country comparison limitation evidence unavailable.",
        "Passed":
            cross_country_df is not None
    },
    {
        "Validation Area": "Interpretation boundary coverage",
        "Requirement":
            "Predictive, causal and commercial interpretation boundaries must remain available.",
        "Observed Evidence":
            "PMIP interpretation boundaries available."
            if interpretation_df is not None
            else "PMIP interpretation boundaries unavailable.",
        "Passed":
            interpretation_df is not None
    },
    {
        "Validation Area": "Human-review requirement",
        "Requirement":
            "Human review requirements must remain part of the final PMIP conclusion.",
        "Observed Evidence":
            "Human review requirements available."
            if human_review_df is not None
            else "Human review requirements unavailable.",
        "Passed":
            human_review_df is not None
    },
    {
        "Validation Area": "Analytical limitation consolidation",
        "Requirement":
            "The final conclusion must consolidate material PMIP limitations.",
        "Observed Evidence":
            f"{len(analytical_limitations_df)} analytical limitation entries prepared.",
        "Passed":
            not analytical_limitations_df.empty
    }
])

print("\nAnalytical limitations validation")
print("=" * 100)
display(analytical_limitations_validation_df)


# =============================================================================
# Completion status
# =============================================================================

section_15_2_complete = bool(
    analytical_limitations_validation_df["Passed"].all()
)

print(
    f"\nSection 15.2 completion status: "
    f"{section_15_2_complete}"
)

if section_15_2_complete:

    print(
        "PMIP analytical limitations have been consolidated successfully "
        "across temporal, geographic, comparative and interpretation dimensions."
    )

else:

    failed_validations = analytical_limitations_validation_df[
        ~analytical_limitations_validation_df["Passed"]
    ]

    print("\nFailed Section 15.2 validation checks")
    print("=" * 100)
    display(failed_validations)

    raise RuntimeError(
        "Section 15.2 analytical-limitations validation failed."
    )
```

#### Interpretation

The analytical limitation review confirms that PMIP has clearly defined the main conditions under which its outputs should be interpreted with caution.

A major limitation is the uneven amount of historical evidence available for tracks and artists. Approximately **47.57% of assessed track histories** and **38.74% of assessed artist histories** contain three or fewer observations. This means that a substantial proportion of entities are supported by relatively short observation histories. Growth signals derived from these cases may therefore be less stable than signals supported by longer histories.

Geographic coverage is also uneven. Earlier coverage analysis showed that many tracks and artists are represented in only a limited number of markets. This means that strong performance in one country should not automatically be interpreted as broad international growth. Market coverage must be considered alongside the PMIP growth score.

Cross-country comparisons also require caution because the amount of supporting evidence differs considerably between countries. Some markets contain substantially more track and artist observations than others. Therefore, country rankings should be interpreted together with the available evidence volume rather than treating every country score as equally certain.

The interpretation boundaries further confirm that PMIP is primarily a **descriptive and decision-support system**. It can identify historical growth patterns, market opportunities, short-term spikes, market differences, emerging signals, volatility and multi-signal growth strength. However, these outputs do not prove that the same growth will continue in the future.

PMIP also does not establish causation. A relationship between an artist, track, country and observed growth pattern does not prove that any specific feature directly caused that growth. The system identifies associations within the available evidence rather than causal relationships.

Similarly, strong PMIP scores should not be interpreted as guarantees of commercial performance. High rankings, growth scores or priority classes do not guarantee revenue, profitability, audience conversion, successful marketing outcomes or long-term artistic success.

Composite growth scores also remain relative analytical measures. They are useful for combining several validated indicators into a consistent ranking framework, but they should not be interpreted as universal measures of artistic quality or commercial value.

Overall, the limitations identified in this subsection reinforce the responsible-use framework developed in Section 13. PMIP should be used to **prioritise investigation, highlight promising signals and support human analysis**, while uncertainty, evidence coverage, market context and professional judgement remain essential parts of the final decision-making process.

#### Interpretation

The final PMIP integration assessment confirms that the Market Growth Intelligence workflow is complete and ready to operate as a reusable component within the wider Public Music Intelligence Platform.

The integration recommendations define how validated country, track and artist growth intelligence should be exposed to downstream PMIP components. These outputs include growth scores, growth classes, rankings, percentiles, opportunity signals, emerging-market evidence and priority classifications.

The results also confirm that Market Growth Intelligence should not operate as an isolated model. Its outputs are designed to support other PMIP components such as Artist Discovery, Track Discovery, A&R Research, Momentum Intelligence, Performance Forecasting, Streaming Anomaly Detection and Reporting. This allows PMIP to combine several independent analytical perspectives when investigating an artist, track or market.

A key integration principle is that analytical scores should remain connected to their supporting evidence. Geographic coverage, historical observation depth, market evidence and validation metadata should therefore remain available alongside the final growth outputs. This helps prevent strong scores from being interpreted without considering the quality and quantity of the underlying evidence.

The integration safeguards also preserve the responsible-use framework established earlier in the notebook. PMIP outputs remain advisory, observed relationships must not be interpreted as causal, growth scores do not guarantee future or commercial success, and important strategic decisions must retain human review.

The validated CSV artifacts and metadata generated in Section 14 provide a reproducible interface between this analytical notebook and the wider PMIP software architecture. These artifacts can therefore be consumed by dashboards, application services, reporting components and future PMIP models without rerunning the complete analytical workflow each time.

Future integration should also support scheduled refreshes. As newer streaming, chart and market observations become available, the Market Growth Intelligence workflow can be rerun and its validated artifacts replaced with updated versions. This allows PMIP growth intelligence to evolve with changing market behaviour while retaining reproducibility and validation controls.

Overall, the Market Growth Intelligence component has now progressed from raw multi-market evidence to validated growth intelligence, priority identification, responsible interpretation, artifact persistence and defined software-integration recommendations.
