# PMIP Geographic Performance Intelligence

## Purpose

This notebook develops the geographic performance intelligence component of the Public Music Intelligence Platform (PMIP).

The objective is to analyse how artists and tracks perform across different countries and geographic markets using the validated historical music-performance data available within PMIP.

The analysis focuses on identifying where artists and tracks perform most strongly, how widely they reach across international markets, how chart and streaming performance differ between countries, and whether geographic expansion patterns can be observed over time.

The notebook also develops reusable geographic performance indicators, country-level summaries and structured outputs that can later support PMIP backend services, dashboards, geographic visualisations and map-based analytics.

Where the available data supports temporal analysis, the workflow will investigate changes in geographic reach and market penetration over time. All conclusions will remain bounded by the observed data coverage and will not infer unsupported causal explanations for geographic performance differences.


## Objectives

The objectives of this notebook are to:

1. Identify and validate the geographic information available within the historical PMIP chart and streaming datasets.

2. Assess the completeness, consistency and historical coverage of country-level observations.

3. Analyse artist and track performance across different countries and geographic markets.

4. Measure the number of countries in which individual tracks and artists have appeared.

5. Compare chart positions, streaming volumes and other performance indicators between markets.

6. Identify countries where particular artists and tracks perform most strongly.

7. Quantify geographic differences in artist popularity and track performance.

8. Develop geographic performance indicators for artists and tracks.

9. Measure international reach, geographic concentration and market penetration.

10. Distinguish artists with predominantly local performance from those with broader international reach.

11. Investigate geographic expansion and contraction patterns where repeated dated observations support temporal analysis.

12. Produce country-level artist and track performance summaries.

13. Prepare analytical structures suitable for geographic visualisations, rankings and map-based PMIP features.

14. Evaluate the reliability and limitations of the geographic intelligence outputs.

15. Document responsible interpretation boundaries for country-level and artist-level geographic comparisons.

16. Save validated geographic intelligence outputs and metadata for later PMIP software integration.

# 1. Introduction

## Purpose

This section introduces the PMIP Geographic Performance Intelligence workflow and defines the analytical context, scope and core questions that guide the geographic analysis.

The goal is to establish what geographic performance intelligence means within PMIP before analysing datasets, engineering features or developing geographic indicators.

This section does not transform data or calculate performance metrics. It defines the analytical foundation for the sections that follow.

## 1.1 Geographic Intelligence Context

Geographic performance intelligence examines how music performance differs across countries and markets.

Within PMIP, this means understanding where artists and tracks perform strongly, how widely they reach across international markets, how concentrated their performance is within particular countries, and how geographic performance changes over time.

Streaming and chart performance can vary substantially between countries because audiences, market sizes, platform usage, release strategies, cultural preferences and historical artist exposure are different across regions.

For this reason, a high stream count in one market should not automatically be interpreted in the same way as the same stream count in another market.

The PMIP geographic intelligence component therefore focuses on relative and contextual performance rather than raw stream totals alone.

The analysis will combine country coverage, streaming performance, chart position, artist reach, track reach, market penetration and geographic concentration to create a more complete view of international music performance.


## 1.2 Analytical Scope

The geographic intelligence workflow focuses on artist, track, country and time-based performance patterns available within the validated PMIP historical datasets.

The analysis will examine:

- country-level streaming and chart performance,
- the number of countries in which artists and tracks appear,
- artist and track market penetration,
- geographic concentration of performance,
- differences between local and international reach,
- strongest-performing markets for artists and tracks,
- cross-market performance differences,
- new-market entry and geographic expansion,
- geographic contraction where historical evidence supports it,
- country-level and artist-level geographic summaries.

The workflow will only make temporal claims where repeated dated observations are available.

The analysis will not assume that geographic performance differences have a specific real-world cause unless additional external evidence supports that interpretation.


## 1.3 Geographic Intelligence Questions

The geographic performance intelligence component is designed to answer the following questions:

1. Which countries are represented within the PMIP historical data?

2. Which artists and tracks have the widest geographic reach?

3. In which countries do particular artists perform most strongly?

4. In which countries do particular tracks perform most strongly?

5. How does streaming performance differ across markets?

6. How does chart performance differ across markets?

7. Which artists show broad international reach?

8. Which artists appear concentrated within a smaller number of markets?

9. Which tracks demonstrate strong cross-country penetration?

10. Are there identifiable patterns of geographic expansion or contraction over time?

11. Which countries contribute most strongly to an artist's overall historical performance?

12. How concentrated or diversified is an artist's geographic performance?

13. Can geographic performance indicators be produced reliably enough for PMIP backend and dashboard integration?

14. What data-coverage limitations must be considered before interpreting country-level differences?

# 2. Data Source Assessment

## Purpose

This section identifies and validates the PMIP datasets that can support geographic performance intelligence.

The goal is to confirm which available datasets contain the required country, artist, track, date, streaming and chart-position information before geographic analysis begins.

This section also checks whether the candidate datasets provide sufficient historical and geographic coverage for later country-level, artist-level and track-level analysis.

No geographic indicators are calculated in this section. The focus is dataset suitability, schema validation and analytical readiness.

## 2.1 Candidate Dataset Review

The first step is to inspect the available PMIP datasets and determine which of them contain information relevant to geographic performance analysis.

Candidate datasets should be assessed for the presence of fields such as:

- reporting date,
- country or market identifier,
- artist,
- track identifier,
- track name where available,
- streaming volume,
- chart position,
- repeated observations across countries and dates.

Priority should be given to validated integrated or processed datasets that preserve observation-level geographic information.

The review should also confirm whether the data supports:

- cross-country artist comparisons,
- cross-country track comparisons,
- country-level aggregation,
- historical geographic analysis,
- market-entry and expansion analysis.

Datasets that do not contain country-level information may still support later contextual analysis, but they should not be treated as primary geographic-intelligence sources.

```python
from pathlib import Path
import pandas as pd

print("Inspecting candidate PMIP datasets for geographic intelligence")
print("=" * 90)

# ------------------------------------------------------------
# Repository paths
# ------------------------------------------------------------

REPO_ROOT = Path("/Users/richardakole/Documents/Now/PMIP")
DATA_ROOT = REPO_ROOT / "data"

candidate_directories = [
    DATA_ROOT / "processed",
    DATA_ROOT / "integrated",
    DATA_ROOT / "model_ready",
]

# ------------------------------------------------------------
# Discover candidate tabular files
# ------------------------------------------------------------

candidate_files = []

for directory in candidate_directories:
    if not directory.exists():
        continue

    for pattern in ("*.csv", "*.parquet"):
        candidate_files.extend(sorted(directory.glob(pattern)))

print(f"Candidate files discovered: {len(candidate_files)}")

if not candidate_files:
    raise RuntimeError(
        "No candidate CSV or Parquet datasets were found in the processed, "
        "integrated or model_ready directories."
    )

# ------------------------------------------------------------
# Geographic field candidates
# ------------------------------------------------------------

field_groups = {
    "date": [
        "date",
        "reporting_date",
        "week",
        "week_start",
        "chart_date",
    ],
    "country": [
        "country",
        "market",
        "country_code",
        "region",
    ],
    "artist": [
        "artist",
        "artists",
        "artist_name",
    ],
    "track": [
        "track_id",
        "track",
        "track_name",
        "title",
    ],
    "streams": [
        "streams",
        "stream_count",
        "streaming_count",
    ],
    "position": [
        "position",
        "chart_position",
        "rank",
    ],
}

def first_matching_column(columns, candidates):
    lower_map = {str(col).lower(): col for col in columns}

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]

    return None

# ------------------------------------------------------------
# Inspect each candidate dataset
# ------------------------------------------------------------

records = []

for path in candidate_files:
    try:
        if path.suffix.lower() == ".csv":
            sample_df = pd.read_csv(path, nrows=5)
        elif path.suffix.lower() == ".parquet":
            sample_df = pd.read_parquet(path)
            sample_df = sample_df.head(5)
        else:
            continue

        detected = {
            group: first_matching_column(sample_df.columns, candidates)
            for group, candidates in field_groups.items()
        }

        required_hits = sum(
            detected[group] is not None
            for group in ["date", "country", "artist", "track"]
        )

        performance_hits = sum(
            detected[group] is not None
            for group in ["streams", "position"]
        )

        geographic_candidate = (
            detected["country"] is not None
            and detected["artist"] is not None
            and detected["track"] is not None
        )

        records.append({
            "Dataset": path.name,
            "Directory": path.parent.name,
            "Sample Columns": len(sample_df.columns),
            "Date Field": detected["date"],
            "Country Field": detected["country"],
            "Artist Field": detected["artist"],
            "Track Field": detected["track"],
            "Streams Field": detected["streams"],
            "Position Field": detected["position"],
            "Identity Fields Found": required_hits,
            "Performance Fields Found": performance_hits,
            "Geographic Candidate": geographic_candidate,
            "Path": str(path),
        })

    except Exception as exc:
        records.append({
            "Dataset": path.name,
            "Directory": path.parent.name,
            "Sample Columns": None,
            "Date Field": None,
            "Country Field": None,
            "Artist Field": None,
            "Track Field": None,
            "Streams Field": None,
            "Position Field": None,
            "Identity Fields Found": 0,
            "Performance Fields Found": 0,
            "Geographic Candidate": False,
            "Path": str(path),
            "Inspection Error": str(exc),
        })

candidate_review_df = pd.DataFrame(records)

candidate_review_df = candidate_review_df.sort_values(
    by=[
        "Geographic Candidate",
        "Identity Fields Found",
        "Performance Fields Found",
    ],
    ascending=[False, False, False],
).reset_index(drop=True)

print()
print("Candidate geographic dataset review")
print("=" * 90)

display(candidate_review_df)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

geographic_candidates = candidate_review_df[
    candidate_review_df["Geographic Candidate"] == True
]

print()
print("Geographic dataset suitability summary")
print("=" * 90)

print(f"Datasets inspected: {len(candidate_review_df)}")
print(f"Geographic candidates identified: {len(geographic_candidates)}")

if len(geographic_candidates) > 0:
    print()
    print("Highest-priority geographic candidates:")

    display(
        geographic_candidates[
            [
                "Dataset",
                "Directory",
                "Date Field",
                "Country Field",
                "Artist Field",
                "Track Field",
                "Streams Field",
                "Position Field",
            ]
        ].head(10)
    )
else:
    print(
        "No dataset currently contains the minimum country + artist + track "
        "structure required for geographic intelligence."
    )
```

### Interpretation

The candidate dataset review identified five available PMIP datasets across the processed, integrated and model-ready data directories.

Only `charts_cleaned.csv` contains the complete geographic observation structure required for this workflow. It includes:

- a reporting date,
- country information,
- artist information,
- track identifiers,
- stream counts,
- chart positions.

This makes `charts_cleaned.csv` the primary dataset for PMIP geographic performance intelligence.

The remaining datasets do not contain the minimum combination of country, artist and track-level geographic fields required for direct cross-market analysis.

Although these datasets may still provide additional artist, track or contextual information later in the workflow, they should not be treated as primary geographic-intelligence sources.

The geographic analysis will therefore proceed using `charts_cleaned.csv` as the validated candidate source, subject to more detailed field, schema and observation-level validation in the following subsections.

## 2.2 Geographic Field Identification

### Purpose

This subsection performs a detailed inspection of the geographic and performance fields available within the selected `charts_cleaned.csv` dataset.

The objective is to confirm the exact column structure, data types and representative values for the fields that will support geographic intelligence.

The review focuses on:

- reporting date,
- country,
- artist,
- track identifier,
- streams,
- chart position,
- any additional fields that may provide useful geographic or analytical context.

No geographic metrics are calculated at this stage. The goal is to understand the validated source structure before defining the analytical population.

```python
from pathlib import Path
import pandas as pd
import numpy as np

print("Inspecting geographic fields in the selected PMIP source")
print("=" * 90)

# ------------------------------------------------------------
# Source path
# ------------------------------------------------------------

REPO_ROOT = Path("/Users/richardakole/Documents/Now/PMIP")
CHARTS_PATH = REPO_ROOT / "data" / "processed" / "charts_cleaned.csv"

if not CHARTS_PATH.exists():
    raise FileNotFoundError(
        f"Expected geographic source not found: {CHARTS_PATH}"
    )

# ------------------------------------------------------------
# Load source
# ------------------------------------------------------------

charts_geo_df = pd.read_csv(CHARTS_PATH)

print(f"Source dataset: {CHARTS_PATH.name}")
print(f"Rows: {len(charts_geo_df):,}")
print(f"Columns: {len(charts_geo_df.columns):,}")

# ------------------------------------------------------------
# Complete schema
# ------------------------------------------------------------

schema_df = pd.DataFrame({
    "Column": charts_geo_df.columns,
    "Data Type": [str(charts_geo_df[col].dtype) for col in charts_geo_df.columns],
    "Non-Missing": [
        int(charts_geo_df[col].notna().sum())
        for col in charts_geo_df.columns
    ],
    "Missing": [
        int(charts_geo_df[col].isna().sum())
        for col in charts_geo_df.columns
    ],
    "Missing (%)": [
        float(charts_geo_df[col].isna().mean() * 100)
        for col in charts_geo_df.columns
    ],
    "Unique Values": [
        int(charts_geo_df[col].nunique(dropna=True))
        for col in charts_geo_df.columns
    ],
})

print()
print("Complete source schema")
print("=" * 90)
display(schema_df)

# ------------------------------------------------------------
# Required geographic fields
# ------------------------------------------------------------

required_fields = [
    "date",
    "country",
    "artists",
    "track_id",
    "streams",
    "position",
]

missing_required_fields = [
    col for col in required_fields
    if col not in charts_geo_df.columns
]

if missing_required_fields:
    raise RuntimeError(
        "Selected geographic source is missing required fields: "
        + ", ".join(missing_required_fields)
    )

print()
print("Required geographic fields confirmed")
print("=" * 90)

required_field_summary = schema_df[
    schema_df["Column"].isin(required_fields)
].copy()

display(required_field_summary)

# ------------------------------------------------------------
# Representative values
# ------------------------------------------------------------

print()
print("Representative geographic values")
print("=" * 90)

for col in required_fields:
    values = (
        charts_geo_df[col]
        .dropna()
        .astype(str)
        .drop_duplicates()
        .head(10)
        .tolist()
    )

    print(f"{col}:")
    print(values)
    print()

# ------------------------------------------------------------
# Core geographic coverage
# ------------------------------------------------------------

date_series = pd.to_datetime(
    charts_geo_df["date"],
    errors="coerce"
)

country_series = (
    charts_geo_df["country"]
    .astype("string")
    .str.strip()
)

artist_series = (
    charts_geo_df["artists"]
    .astype("string")
    .str.strip()
)

track_series = (
    charts_geo_df["track_id"]
    .astype("string")
    .str.strip()
)

streams_series = pd.to_numeric(
    charts_geo_df["streams"],
    errors="coerce"
)

position_series = pd.to_numeric(
    charts_geo_df["position"],
    errors="coerce"
)

print("Core geographic coverage summary")
print("=" * 90)

print(
    f"Valid reporting dates: "
    f"{date_series.notna().sum():,} / {len(charts_geo_df):,}"
)

print(
    f"Reporting period: "
    f"{date_series.min().date()} to {date_series.max().date()}"
)

print(
    f"Unique reporting dates: "
    f"{date_series.nunique():,}"
)

print(
    f"Unique countries: "
    f"{country_series.nunique(dropna=True):,}"
)

print(
    f"Unique artist representations: "
    f"{artist_series.nunique(dropna=True):,}"
)

print(
    f"Unique tracks: "
    f"{track_series.nunique(dropna=True):,}"
)

print(
    f"Finite stream values: "
    f"{np.isfinite(streams_series).sum():,}"
)

print(
    f"Finite chart-position values: "
    f"{np.isfinite(position_series).sum():,}"
)

# ------------------------------------------------------------
# Country examples
# ------------------------------------------------------------

country_examples = (
    country_series
    .dropna()
    .value_counts()
    .head(20)
    .rename_axis("Country")
    .reset_index(name="Observations")
)

print()
print("Most frequently represented countries")
print("=" * 90)

display(country_examples)

# ------------------------------------------------------------
# Validation flags
# ------------------------------------------------------------

field_validation_df = pd.DataFrame({
    "Field": required_fields,
    "Exists": [
        col in charts_geo_df.columns
        for col in required_fields
    ],
    "Has Non-Missing Values": [
        charts_geo_df[col].notna().any()
        for col in required_fields
    ],
    "Unique Values": [
        int(charts_geo_df[col].nunique(dropna=True))
        for col in required_fields
    ],
})

print()
print("Geographic field validation")
print("=" * 90)

display(field_validation_df)

section_2_2_complete = bool(
    field_validation_df["Exists"].all()
    and field_validation_df["Has Non-Missing Values"].all()
)

if not section_2_2_complete:
    raise AssertionError(
        "Section 2.2 geographic field validation failed."
    )

print()
print("All Section 2.2 geographic field checks passed.")
print(f"Section 2.2 completion status: {section_2_2_complete}")
```

### Interpretation

The detailed field inspection confirms that `charts_cleaned.csv` contains the complete observation structure required for PMIP geographic performance intelligence.

The dataset contains **5,427,136 observations across 10 fields**, with no missing values in any column. The six required geographic-analysis fields — `date`, `country`, `artists`, `track_id`, `streams` and `position` — are all present and fully populated.

The historical source covers **515 reporting dates** from **28 April 2013 to 6 April 2023**, providing almost ten years of observations for temporal geographic analysis.

Geographic coverage is substantial, with **77 country values** represented in the dataset. This supports cross-country comparisons, market-level aggregation and later analysis of international reach and geographic concentration.

The source also contains:

- **110,198 unique track identifiers**,
- **49,843 unique artist representations**,
- **947,912 distinct stream-count values**,
- **358 distinct chart-position values**.

The `artists` field contains string representations of artist lists, including collaborations involving multiple artists. Therefore, the current count of 49,843 should be interpreted as unique stored artist representations rather than necessarily 49,843 individual artists. Artist identity handling will require explicit validation before artist-level geographic aggregation.

Country values are represented using short country codes such as `de`, `jp`, `us`, `se` and `gb`. These codes appear suitable for geographic grouping, but their format and consistency should be validated before they are used for country naming, mapping or external geographic visualisation.

The most frequently represented countries have broadly similar observation volumes, which is encouraging for cross-market analysis. However, complete country coverage and temporal consistency must still be assessed before interpreting differences between markets.

All Section 2.2 geographic field checks passed, so the selected source can proceed to stricter required-field and value-level validation in Section 2.3.

## 2.3 Required Field Validation

### Purpose

This subsection validates the analytical quality of the required geographic fields rather than checking only whether the columns exist.

The objective is to confirm that dates, country codes, artist values, track identifiers, stream counts and chart positions contain structurally valid values suitable for later aggregation and geographic comparison.

The validation checks:

- reporting-date parsing and chronological boundaries,
- country-code structure and consistency,
- blank or malformed artist representations,
- blank or malformed track identifiers,
- stream-count validity,
- chart-position validity,
- duplicated observation structures,
- combinations of fields that may identify repeated geographic chart observations.

No observations are removed in this subsection. Any detected issues are documented before the analytical population is defined later in the workflow.

```python
import ast
import numpy as np
import pandas as pd

print("Validating required PMIP geographic fields")
print("=" * 100)

if "charts_geo_df" not in globals():
    raise RuntimeError(
        "charts_geo_df is not available. Run Section 2.2 before Section 2.3."
    )

geo_validation_df = charts_geo_df.copy()

# ------------------------------------------------------------
# 1. Date validation
# ------------------------------------------------------------

parsed_dates = pd.to_datetime(
    geo_validation_df["date"],
    errors="coerce"
)

invalid_date_mask = parsed_dates.isna()

# ------------------------------------------------------------
# 2. Country / market validation
# ------------------------------------------------------------

country_clean = (
    geo_validation_df["country"]
    .astype("string")
    .str.strip()
    .str.lower()
)

blank_country_mask = (
    country_clean.isna()
    | country_clean.eq("")
)

# Standard two-letter country-code diagnostic.
# IMPORTANT:
# This is descriptive only because the PMIP source may contain
# non-standard platform market identifiers or special territories.
standard_country_code_mask = country_clean.str.fullmatch(
    r"[a-z]{2}",
    na=False
)

nonstandard_country_code_mask = (
    ~blank_country_mask
    & ~standard_country_code_mask
)

nonstandard_country_values = (
    country_clean[
        nonstandard_country_code_mask
    ]
    .value_counts(dropna=False)
    .rename_axis("Country / Market Value")
    .reset_index(name="Observations")
)

# ------------------------------------------------------------
# 3. Artist validation
# ------------------------------------------------------------

artist_clean = (
    geo_validation_df["artists"]
    .astype("string")
    .str.strip()
)

blank_artist_mask = (
    artist_clean.isna()
    | artist_clean.eq("")
)

def valid_artist_representation(value):
    if pd.isna(value):
        return False

    text = str(value).strip()

    if text == "":
        return False

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            valid_members = [
                str(member).strip()
                for member in parsed
                if str(member).strip() != ""
            ]

            return len(valid_members) > 0

    except (ValueError, SyntaxError):
        pass

    # Plain-text artist representations are also retained.
    return len(text) > 0

unique_artist_values = artist_clean.dropna().unique()

artist_format_lookup = {
    value: valid_artist_representation(value)
    for value in unique_artist_values
}

invalid_artist_format_mask = (
    ~blank_artist_mask
    & ~artist_clean.map(artist_format_lookup).fillna(False)
)

# ------------------------------------------------------------
# 4. Track identifier validation
# ------------------------------------------------------------

track_clean = (
    geo_validation_df["track_id"]
    .astype("string")
    .str.strip()
)

blank_track_mask = (
    track_clean.isna()
    | track_clean.eq("")
)

# Spotify-style track IDs are commonly 22 alphanumeric characters.
# This is a diagnostic only and is NOT used as an exclusion rule.
spotify_track_format_mask = track_clean.str.fullmatch(
    r"[A-Za-z0-9]{22}",
    na=False
)

nonstandard_track_id_mask = (
    ~blank_track_mask
    & ~spotify_track_format_mask
)

# ------------------------------------------------------------
# 5. Stream validation
# ------------------------------------------------------------

streams_numeric = pd.to_numeric(
    geo_validation_df["streams"],
    errors="coerce"
)

nonfinite_stream_mask = ~np.isfinite(streams_numeric)

negative_stream_mask = (
    np.isfinite(streams_numeric)
    & (streams_numeric < 0)
)

zero_stream_mask = (
    np.isfinite(streams_numeric)
    & (streams_numeric == 0)
)

# ------------------------------------------------------------
# 6. Chart-position validation
# ------------------------------------------------------------

position_numeric = pd.to_numeric(
    geo_validation_df["position"],
    errors="coerce"
)

nonfinite_position_mask = ~np.isfinite(position_numeric)

nonpositive_position_mask = (
    np.isfinite(position_numeric)
    & (position_numeric <= 0)
)

noninteger_position_mask = (
    np.isfinite(position_numeric)
    & (position_numeric % 1 != 0)
)

position_min = (
    float(position_numeric.min())
    if np.isfinite(position_numeric).any()
    else np.nan
)

position_max = (
    float(position_numeric.max())
    if np.isfinite(position_numeric).any()
    else np.nan
)

# ------------------------------------------------------------
# 7. Duplicate / observation-key diagnostics
# ------------------------------------------------------------

exact_duplicate_mask = geo_validation_df.duplicated(
    keep=False
)

candidate_observation_key = [
    "date",
    "country",
    "track_id",
]

candidate_key_summary = (
    geo_validation_df
    .groupby(
        candidate_observation_key,
        dropna=False
    )
    .size()
    .rename("rows_per_candidate_key")
)

duplicate_candidate_keys = int(
    (candidate_key_summary > 1).sum()
)

max_rows_per_candidate_key = int(
    candidate_key_summary.max()
)

# ------------------------------------------------------------
# Required-field validation
# ------------------------------------------------------------

validation_records = [
    {
        "Validation Area": "Date parsing",
        "Requirement": "Every reporting date must parse successfully.",
        "Issue Count": int(invalid_date_mask.sum()),
        "Passed": int(invalid_date_mask.sum()) == 0,
    },
    {
        "Validation Area": "Country / market availability",
        "Requirement": "Country or market values must not be blank.",
        "Issue Count": int(blank_country_mask.sum()),
        "Passed": int(blank_country_mask.sum()) == 0,
    },
    {
        "Validation Area": "Artist availability",
        "Requirement": "Artist representations must not be blank.",
        "Issue Count": int(blank_artist_mask.sum()),
        "Passed": int(blank_artist_mask.sum()) == 0,
    },
    {
        "Validation Area": "Artist representation structure",
        "Requirement": "Artist representations must contain usable artist information.",
        "Issue Count": int(invalid_artist_format_mask.sum()),
        "Passed": int(invalid_artist_format_mask.sum()) == 0,
    },
    {
        "Validation Area": "Track identifier availability",
        "Requirement": "Track identifiers must not be blank.",
        "Issue Count": int(blank_track_mask.sum()),
        "Passed": int(blank_track_mask.sum()) == 0,
    },
    {
        "Validation Area": "Finite stream values",
        "Requirement": "Streams must be finite numeric values.",
        "Issue Count": int(nonfinite_stream_mask.sum()),
        "Passed": int(nonfinite_stream_mask.sum()) == 0,
    },
    {
        "Validation Area": "Non-negative stream values",
        "Requirement": "Streams must not be negative.",
        "Issue Count": int(negative_stream_mask.sum()),
        "Passed": int(negative_stream_mask.sum()) == 0,
    },
    {
        "Validation Area": "Finite chart positions",
        "Requirement": "Chart positions must be finite numeric values.",
        "Issue Count": int(nonfinite_position_mask.sum()),
        "Passed": int(nonfinite_position_mask.sum()) == 0,
    },
    {
        "Validation Area": "Positive chart positions",
        "Requirement": "Chart positions must be greater than zero.",
        "Issue Count": int(nonpositive_position_mask.sum()),
        "Passed": int(nonpositive_position_mask.sum()) == 0,
    },
    {
        "Validation Area": "Integer chart positions",
        "Requirement": "Chart positions should be whole-number ranks.",
        "Issue Count": int(noninteger_position_mask.sum()),
        "Passed": int(noninteger_position_mask.sum()) == 0,
    },
]

required_field_validation_df = pd.DataFrame(
    validation_records
)

print()
print("Required-field validation")
print("=" * 100)
display(required_field_validation_df)

# ------------------------------------------------------------
# Country-code diagnostic
# ------------------------------------------------------------

print()
print("Country / market code diagnostic")
print("=" * 100)

print(
    f"Unique country / market values: "
    f"{country_clean.nunique(dropna=True):,}"
)

print(
    f"Standard two-letter values: "
    f"{country_clean[standard_country_code_mask].nunique(dropna=True):,}"
)

print(
    f"Non-standard country / market values: "
    f"{len(nonstandard_country_values):,}"
)

if len(nonstandard_country_values) > 0:
    print()
    print(
        "The following values do not match the standard "
        "two-letter format. They are retained for investigation:"
    )
    display(nonstandard_country_values)
else:
    print("All observed market values use the standard two-letter format.")

# ------------------------------------------------------------
# Structural diagnostics
# ------------------------------------------------------------

structural_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Reporting period start",
        "Observed Value": str(parsed_dates.min().date()),
    },
    {
        "Diagnostic": "Reporting period end",
        "Observed Value": str(parsed_dates.max().date()),
    },
    {
        "Diagnostic": "Unique reporting dates",
        "Observed Value": f"{parsed_dates.nunique():,}",
    },
    {
        "Diagnostic": "Unique country / market values",
        "Observed Value": f"{country_clean.nunique(dropna=True):,}",
    },
    {
        "Diagnostic": "Non-standard country / market values",
        "Observed Value": f"{len(nonstandard_country_values):,}",
    },
    {
        "Diagnostic": "Minimum chart position",
        "Observed Value": f"{position_min:,.0f}",
    },
    {
        "Diagnostic": "Maximum chart position",
        "Observed Value": f"{position_max:,.0f}",
    },
    {
        "Diagnostic": "Zero-stream observations",
        "Observed Value": f"{int(zero_stream_mask.sum()):,}",
    },
    {
        "Diagnostic": "Non-standard Spotify-style track IDs",
        "Observed Value": f"{int(nonstandard_track_id_mask.sum()):,}",
    },
    {
        "Diagnostic": "Exact duplicate rows",
        "Observed Value": f"{int(exact_duplicate_mask.sum()):,}",
    },
    {
        "Diagnostic": "Repeated date-country-track keys",
        "Observed Value": f"{duplicate_candidate_keys:,}",
    },
    {
        "Diagnostic": "Maximum rows for one date-country-track key",
        "Observed Value": f"{max_rows_per_candidate_key:,}",
    },
])

print()
print("Structural diagnostics")
print("=" * 100)
display(structural_diagnostics_df)

# ------------------------------------------------------------
# Repeated candidate-key examples
# ------------------------------------------------------------

if duplicate_candidate_keys > 0:
    repeated_key_examples = (
        candidate_key_summary[
            candidate_key_summary > 1
        ]
        .sort_values(ascending=False)
        .head(20)
        .reset_index()
    )

    print()
    print("Example repeated date-country-track keys")
    print("=" * 100)
    display(repeated_key_examples)

# ------------------------------------------------------------
# Completion status
# ------------------------------------------------------------

section_2_3_complete = bool(
    required_field_validation_df["Passed"].all()
)

if not section_2_3_complete:
    failed_checks = required_field_validation_df.loc[
        ~required_field_validation_df["Passed"],
        "Validation Area"
    ].tolist()

    raise AssertionError(
        "Section 2.3 required-field validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print("All Section 2.3 required-field validation checks passed.")
print(f"Section 2.3 completion status: {section_2_3_complete}")

print(
    "Country-code format, track-ID format and repeated observation-key "
    "patterns remain diagnostics for Section 2.4 rather than exclusion rules."
)
```

### Interpretation

Section 2.3 confirms that the selected geographic source is structurally suitable for PMIP geographic performance intelligence.

All required-field validation checks passed. Reporting dates parse successfully, country and market values are populated, artist representations contain usable information, track identifiers are present and structurally valid, stream values are numeric and non-negative, and chart positions are finite positive integers.

The candidate observation key formed by `date + country + track_id` is unique across the dataset. No repeated keys or exact duplicate rows were identified, which strongly suggests that each row represents a distinct track observation for a particular reporting date and geographic market.

The country field contains **77 unique values**. Of these, **76 follow the expected two-letter country-code structure**, while one value — `global` — represents a non-country aggregate market containing **88,456 observations**.

The `global` value should not be treated as an individual country during country-level geographic comparisons. However, it may provide useful aggregate reference information and will therefore be retained separately until the analytical population is formally defined.

Only **7 observations contain zero streams**. These observations are not automatically removed because zero may represent a legitimate recorded value. Their treatment will be reviewed when the geographic analytical population and performance metrics are constructed.

Chart positions range from **1 to 358**. Because these ranks are directly observed in the validated source, no unsupported Top-200 restriction will be introduced.

Overall, the required geographic fields are analytically usable and the observation structure is sufficiently consistent to proceed to Section 2.4, where the dataset schema, observation key and treatment of the `global` market will be formally defined.

## 2.4 Observation-Key and Schema Validation

### Purpose

This subsection formally validates the observation structure that will underpin the PMIP geographic intelligence workflow.

The previous validation indicates that `date + country + track_id` uniquely identifies every row in the selected geographic source. This subsection confirms that structure explicitly and documents the meaning of country-level and global observations.

The analysis will:

- validate the proposed geographic observation key,
- confirm row-level uniqueness,
- distinguish country markets from the `global` aggregate market,
- assess whether track identity remains consistent across markets,
- validate artist and track-name consistency,
- establish the canonical schema used by later geographic analysis,
- create a validated source table without modifying the original dataset.

The `global` market will be retained as a separate analytical category but excluded from metrics that specifically measure country reach, country concentration or international market penetration.

```python
import ast
import numpy as np
import pandas as pd

print("Validating PMIP geographic observation key and canonical schema")
print("=" * 100)

if "geo_validation_df" not in globals():
    raise RuntimeError(
        "geo_validation_df is not available. Run Section 2.3 before Section 2.4."
    )

source_df = geo_validation_df.copy()

# ------------------------------------------------------------
# Canonical field preparation
# ------------------------------------------------------------

canonical_geo_df = pd.DataFrame({
    "date": pd.to_datetime(
        source_df["date"],
        errors="coerce"
    ),
    "country": (
        source_df["country"]
        .astype("string")
        .str.strip()
        .str.lower()
    ),
    "track_id": (
        source_df["track_id"]
        .astype("string")
        .str.strip()
    ),
    "track_name": (
        source_df["name"]
        .astype("string")
        .str.strip()
    ),
    "artists_raw": (
        source_df["artists"]
        .astype("string")
        .str.strip()
    ),
    "artist_genres": (
        source_df["artist_genres"]
        .astype("string")
        .str.strip()
    ),
    "streams": pd.to_numeric(
        source_df["streams"],
        errors="coerce"
    ),
    "position": pd.to_numeric(
        source_df["position"],
        errors="coerce"
    ),
    "duration": pd.to_numeric(
        source_df["duration"],
        errors="coerce"
    ),
    "explicit": source_df["explicit"].astype("bool"),
})

# ------------------------------------------------------------
# Market classification
# ------------------------------------------------------------

canonical_geo_df["market_type"] = np.where(
    canonical_geo_df["country"].eq("global"),
    "Global aggregate",
    "Country market"
)

canonical_geo_df["is_global_market"] = (
    canonical_geo_df["country"].eq("global")
)

canonical_geo_df["is_country_market"] = (
    ~canonical_geo_df["is_global_market"]
)

# ------------------------------------------------------------
# Artist parsing
# ------------------------------------------------------------

def parse_artist_list(value):
    if pd.isna(value):
        return []

    text = str(value).strip()

    if text == "":
        return []

    try:
        parsed = ast.literal_eval(text)

        if isinstance(parsed, list):
            return [
                str(item).strip()
                for item in parsed
                if str(item).strip() != ""
            ]

    except (ValueError, SyntaxError):
        pass

    return [text]

unique_artist_strings = canonical_geo_df[
    "artists_raw"
].dropna().unique()

artist_parse_lookup = {
    value: parse_artist_list(value)
    for value in unique_artist_strings
}

canonical_geo_df["artist_list"] = (
    canonical_geo_df["artists_raw"]
    .map(artist_parse_lookup)
)

canonical_geo_df["artist_count"] = (
    canonical_geo_df["artist_list"]
    .map(len)
)

canonical_geo_df["primary_artist"] = (
    canonical_geo_df["artist_list"]
    .map(lambda values: values[0] if values else pd.NA)
)

# ------------------------------------------------------------
# Observation-key validation
# ------------------------------------------------------------

geographic_observation_key = [
    "date",
    "country",
    "track_id",
]

duplicate_key_mask = canonical_geo_df.duplicated(
    subset=geographic_observation_key,
    keep=False
)

duplicate_key_rows = int(
    duplicate_key_mask.sum()
)

unique_key_count = int(
    canonical_geo_df[
        geographic_observation_key
    ]
    .drop_duplicates()
    .shape[0]
)

observation_key_unique = (
    unique_key_count == len(canonical_geo_df)
)

# ------------------------------------------------------------
# Track identity consistency
# ------------------------------------------------------------

track_identity_summary = (
    canonical_geo_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_names=("track_name", "nunique"),
        artist_representations=("artists_raw", "nunique"),
        explicit_values=("explicit", "nunique"),
        duration_values=("duration", "nunique"),
        observations=("track_id", "size"),
        country_markets=(
            "country",
            lambda s: s[s != "global"].nunique()
        ),
        has_global_market=(
            "is_global_market",
            "max"
        ),
    )
    .reset_index()
)

track_name_conflicts = int(
    (track_identity_summary["track_names"] > 1).sum()
)

artist_identity_conflicts = int(
    (
        track_identity_summary[
            "artist_representations"
        ] > 1
    ).sum()
)

explicit_identity_conflicts = int(
    (
        track_identity_summary[
            "explicit_values"
        ] > 1
    ).sum()
)

duration_identity_conflicts = int(
    (
        track_identity_summary[
            "duration_values"
        ] > 1
    ).sum()
)

# ------------------------------------------------------------
# Market population summary
# ------------------------------------------------------------

market_population_summary_df = pd.DataFrame([
    {
        "Population": "All observations",
        "Observations": len(canonical_geo_df),
        "Unique Markets": canonical_geo_df["country"].nunique(),
        "Unique Tracks": canonical_geo_df["track_id"].nunique(),
    },
    {
        "Population": "Country-market observations",
        "Observations": int(
            canonical_geo_df["is_country_market"].sum()
        ),
        "Unique Markets": int(
            canonical_geo_df.loc[
                canonical_geo_df["is_country_market"],
                "country"
            ].nunique()
        ),
        "Unique Tracks": int(
            canonical_geo_df.loc[
                canonical_geo_df["is_country_market"],
                "track_id"
            ].nunique()
        ),
    },
    {
        "Population": "Global aggregate observations",
        "Observations": int(
            canonical_geo_df["is_global_market"].sum()
        ),
        "Unique Markets": int(
            canonical_geo_df.loc[
                canonical_geo_df["is_global_market"],
                "country"
            ].nunique()
        ),
        "Unique Tracks": int(
            canonical_geo_df.loc[
                canonical_geo_df["is_global_market"],
                "track_id"
            ].nunique()
        ),
    },
])

print()
print("Market population structure")
print("=" * 100)
display(market_population_summary_df)

# ------------------------------------------------------------
# Schema summary
# ------------------------------------------------------------

canonical_schema_df = pd.DataFrame({
    "Field": canonical_geo_df.columns,
    "Data Type": [
        str(canonical_geo_df[col].dtype)
        for col in canonical_geo_df.columns
    ],
    "Non-Missing": [
        int(canonical_geo_df[col].notna().sum())
        for col in canonical_geo_df.columns
    ],
    "Unique Values": [
        int(
            canonical_geo_df[col]
            .astype(str)
            .nunique(dropna=True)
        )
        for col in canonical_geo_df.columns
    ],
})

print()
print("Canonical geographic schema")
print("=" * 100)
display(canonical_schema_df)

# ------------------------------------------------------------
# Track identity diagnostics
# ------------------------------------------------------------

track_identity_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Unique track IDs",
        "Observed Value": f"{canonical_geo_df['track_id'].nunique():,}",
    },
    {
        "Diagnostic": "Track IDs with multiple track names",
        "Observed Value": f"{track_name_conflicts:,}",
    },
    {
        "Diagnostic": "Track IDs with multiple artist representations",
        "Observed Value": f"{artist_identity_conflicts:,}",
    },
    {
        "Diagnostic": "Track IDs with inconsistent explicit flags",
        "Observed Value": f"{explicit_identity_conflicts:,}",
    },
    {
        "Diagnostic": "Track IDs with multiple duration values",
        "Observed Value": f"{duration_identity_conflicts:,}",
    },
])

print()
print("Track identity diagnostics")
print("=" * 100)
display(track_identity_diagnostics_df)

# ------------------------------------------------------------
# Validation register
# ------------------------------------------------------------

validation_records = [
    {
        "Validation Area": "Canonical date availability",
        "Requirement": "Every canonical row must retain a valid reporting date.",
        "Observed Evidence": (
            f"{canonical_geo_df['date'].notna().sum():,} "
            f"of {len(canonical_geo_df):,} valid dates"
        ),
        "Passed": canonical_geo_df["date"].notna().all(),
    },
    {
        "Validation Area": "Canonical market availability",
        "Requirement": "Every canonical row must retain a market identifier.",
        "Observed Evidence": (
            f"{canonical_geo_df['country'].notna().sum():,} "
            f"of {len(canonical_geo_df):,} market values"
        ),
        "Passed": canonical_geo_df["country"].notna().all(),
    },
    {
        "Validation Area": "Canonical track availability",
        "Requirement": "Every canonical row must retain a track identifier.",
        "Observed Evidence": (
            f"{canonical_geo_df['track_id'].notna().sum():,} "
            f"of {len(canonical_geo_df):,} track IDs"
        ),
        "Passed": canonical_geo_df["track_id"].notna().all(),
    },
    {
        "Validation Area": "Observation-key uniqueness",
        "Requirement": "date + country + track_id must uniquely identify each row.",
        "Observed Evidence": (
            f"{unique_key_count:,} unique keys across "
            f"{len(canonical_geo_df):,} rows"
        ),
        "Passed": observation_key_unique,
    },
    {
        "Validation Area": "Duplicate-key exclusion",
        "Requirement": "No rows should share the validated geographic observation key.",
        "Observed Evidence": (
            f"{duplicate_key_rows:,} rows involved in repeated keys"
        ),
        "Passed": duplicate_key_rows == 0,
    },
    {
        "Validation Area": "Country/global separation",
        "Requirement": "Global aggregate observations must be distinguishable from country markets.",
        "Observed Evidence": (
            f"{int(canonical_geo_df['is_global_market'].sum()):,} global rows; "
            f"{int(canonical_geo_df['is_country_market'].sum()):,} country-market rows"
        ),
        "Passed": (
            canonical_geo_df["market_type"]
            .isin(
                [
                    "Global aggregate",
                    "Country market",
                ]
            )
            .all()
        ),
    },
    {
        "Validation Area": "Artist parsing",
        "Requirement": "Every row must retain at least one usable artist identity.",
        "Observed Evidence": (
            f"{int((canonical_geo_df['artist_count'] > 0).sum()):,} "
            f"of {len(canonical_geo_df):,} rows contain parsed artists"
        ),
        "Passed": (
            canonical_geo_df["artist_count"] > 0
        ).all(),
    },
]

section_2_4_validation_df = pd.DataFrame(
    validation_records
)

print()
print("Observation-key and schema validation")
print("=" * 100)
display(section_2_4_validation_df)

# ------------------------------------------------------------
# Canonical country-only analytical source
# ------------------------------------------------------------

country_geo_df = (
    canonical_geo_df.loc[
        canonical_geo_df["is_country_market"]
    ]
    .copy()
    .reset_index(drop=True)
)

global_geo_df = (
    canonical_geo_df.loc[
        canonical_geo_df["is_global_market"]
    ]
    .copy()
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Completion status
# ------------------------------------------------------------

section_2_4_complete = bool(
    section_2_4_validation_df["Passed"].all()
)

if not section_2_4_complete:
    failed_checks = section_2_4_validation_df.loc[
        ~section_2_4_validation_df["Passed"],
        "Validation Area"
    ].tolist()

    raise AssertionError(
        "Section 2.4 observation-key and schema validation failed for: "
        + ", ".join(failed_checks)
    )

section_2_overall_complete = bool(
    globals().get("section_2_2_complete", False)
    and globals().get("section_2_3_complete", False)
    and section_2_4_complete
)

print()
print("All Section 2.4 observation-key and schema validation checks passed.")
print(f"Section 2.4 completion status: {section_2_4_complete}")
print(f"Section 2 overall completion status: {section_2_overall_complete}")
print()
print(
    f"Country-market analytical rows prepared: "
    f"{len(country_geo_df):,}"
)
print(
    f"Global aggregate reference rows retained separately: "
    f"{len(global_geo_df):,}"
)
print(
    "The original charts_cleaned.csv dataframe has not been modified."
)
```

### Interpretation

Section 2.4 confirms the observation structure and canonical schema that will be used throughout the PMIP geographic intelligence workflow.

The combination of `date + country + track_id` uniquely identifies every observation in the selected source. All **5,427,136 rows** have unique geographic observation keys, with no repeated keys detected.

The source population has now been separated into two analytically distinct groups:

- **5,338,680 country-market observations**
- **88,456 global aggregate observations**

The `global` market is retained as a separate reference population rather than being treated as an individual country. This prevents global aggregate observations from artificially increasing country-reach, market-penetration or geographic-concentration metrics.

Track identity is highly consistent across the dataset. All **110,198 unique track identifiers** retain a single track name, artist representation, explicit-status value and duration value. This provides strong support for reliable track-level geographic aggregation.

Artist-list parsing also succeeded for every observation, allowing collaboration information to be retained while also deriving a primary artist field for later artist-level analysis.

The canonical geographic schema now provides validated date, market, track, artist and performance fields together with explicit country/global market indicators.

`country_geo_df` will therefore serve as the primary analytical population for geographic comparisons, while `global_geo_df` will remain available for aggregate reference analysis.

All Section 2.4 validation checks passed, and **Section 2 is complete**. The workflow can now proceed to Section 3, where the depth and consistency of geographic coverage across countries, tracks, artists and reporting periods will be evaluated.

# 3. Geographic Data Coverage

## Purpose

This section evaluates how completely and consistently the validated PMIP country-market dataset represents geographic music performance.

The objective is to determine whether the available data provides sufficient coverage across countries, artists, tracks and reporting periods for reliable geographic comparison.

The analysis focuses on:

- the number of observations available for each country,
- the number of reporting dates represented in each market,
- the number of unique artists and tracks observed by country,
- differences in market coverage depth,
- countries with relatively sparse or dense historical representation,
- potential imbalances that could influence later geographic indicators.

The global aggregate population remains excluded from country-specific coverage metrics so that market reach and geographic representation are based only on actual country markets.

## 3.1 Country Coverage

### Purpose

This subsection measures the depth of historical coverage available for each validated country market.

The analysis compares countries using observation counts, reporting-date coverage, unique track coverage and unique artist-representation coverage.

The goal is to identify whether some markets are substantially more represented than others before later comparing streaming performance, chart position, market penetration or geographic reach.

```python
import pandas as pd
import numpy as np

print("Assessing PMIP country-level geographic coverage")
print("=" * 100)

if "country_geo_df" not in globals():
    raise RuntimeError(
        "country_geo_df is not available. Run Section 2.4 before Section 3.1."
    )

coverage_source_df = country_geo_df.copy()

# ------------------------------------------------------------
# Country-level coverage summary
# ------------------------------------------------------------

country_coverage_df = (
    coverage_source_df
    .groupby(
        "country",
        dropna=False
    )
    .agg(
        observations=("track_id", "size"),
        reporting_dates=("date", "nunique"),
        unique_tracks=("track_id", "nunique"),
        unique_artist_representations=("artists_raw", "nunique"),
        unique_primary_artists=("primary_artist", "nunique"),
        total_streams=("streams", "sum"),
        median_streams=("streams", "median"),
        mean_chart_position=("position", "mean"),
        median_chart_position=("position", "median"),
    )
    .reset_index()
)

# ------------------------------------------------------------
# Coverage percentages
# ------------------------------------------------------------

total_country_observations = len(coverage_source_df)

total_reporting_dates = int(
    coverage_source_df["date"].nunique()
)

total_unique_tracks = int(
    coverage_source_df["track_id"].nunique()
)

total_artist_representations = int(
    coverage_source_df["artists_raw"].nunique()
)

country_coverage_df["observation_share_pct"] = (
    country_coverage_df["observations"]
    / total_country_observations
    * 100
)

country_coverage_df["date_coverage_pct"] = (
    country_coverage_df["reporting_dates"]
    / total_reporting_dates
    * 100
)

country_coverage_df["track_coverage_pct"] = (
    country_coverage_df["unique_tracks"]
    / total_unique_tracks
    * 100
)

country_coverage_df["artist_representation_coverage_pct"] = (
    country_coverage_df["unique_artist_representations"]
    / total_artist_representations
    * 100
)

# ------------------------------------------------------------
# Coverage balance diagnostics
# ------------------------------------------------------------

observation_median = float(
    country_coverage_df["observations"].median()
)

observation_q1 = float(
    country_coverage_df["observations"].quantile(0.25)
)

observation_q3 = float(
    country_coverage_df["observations"].quantile(0.75)
)

observation_iqr = observation_q3 - observation_q1

lower_sparse_boundary = max(
    0,
    observation_q1 - 1.5 * observation_iqr
)

upper_dense_boundary = (
    observation_q3 + 1.5 * observation_iqr
)

country_coverage_df["coverage_classification"] = np.select(
    [
        country_coverage_df["observations"] < lower_sparse_boundary,
        country_coverage_df["observations"] > upper_dense_boundary,
    ],
    [
        "Relatively sparse",
        "Relatively dense",
    ],
    default="Typical coverage"
)

# ------------------------------------------------------------
# Overall summary
# ------------------------------------------------------------

print()
print("Country coverage overview")
print("=" * 100)

print(f"Country markets assessed: {len(country_coverage_df):,}")
print(f"Country-market observations: {total_country_observations:,}")
print(f"Reporting dates represented: {total_reporting_dates:,}")
print(f"Unique tracks represented: {total_unique_tracks:,}")
print(
    f"Unique artist representations: "
    f"{total_artist_representations:,}"
)

print()
print("Observation-count distribution across countries")
print("=" * 100)

print(
    f"Minimum observations in one country: "
    f"{country_coverage_df['observations'].min():,}"
)

print(
    f"Median observations per country: "
    f"{observation_median:,.1f}"
)

print(
    f"Maximum observations in one country: "
    f"{country_coverage_df['observations'].max():,}"
)

print(
    f"Q1 observations: "
    f"{observation_q1:,.1f}"
)

print(
    f"Q3 observations: "
    f"{observation_q3:,.1f}"
)

print(
    f"Relative sparse boundary: "
    f"{lower_sparse_boundary:,.1f}"
)

print(
    f"Relative dense boundary: "
    f"{upper_dense_boundary:,.1f}"
)

# ------------------------------------------------------------
# Most represented countries
# ------------------------------------------------------------

top_country_coverage_df = (
    country_coverage_df
    .sort_values(
        "observations",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Most represented country markets")
print("=" * 100)

display(
    top_country_coverage_df[
        [
            "country",
            "observations",
            "observation_share_pct",
            "reporting_dates",
            "date_coverage_pct",
            "unique_tracks",
            "unique_artist_representations",
            "coverage_classification",
        ]
    ]
)

# ------------------------------------------------------------
# Least represented countries
# ------------------------------------------------------------

bottom_country_coverage_df = (
    country_coverage_df
    .sort_values(
        "observations",
        ascending=True
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Least represented country markets")
print("=" * 100)

display(
    bottom_country_coverage_df[
        [
            "country",
            "observations",
            "observation_share_pct",
            "reporting_dates",
            "date_coverage_pct",
            "unique_tracks",
            "unique_artist_representations",
            "coverage_classification",
        ]
    ]
)

# ------------------------------------------------------------
# Date coverage diagnostics
# ------------------------------------------------------------

date_coverage_summary_df = pd.DataFrame([
    {
        "Diagnostic": "Countries with complete reporting-date coverage",
        "Observed Value": int(
            (
                country_coverage_df["reporting_dates"]
                == total_reporting_dates
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries with >= 95% date coverage",
        "Observed Value": int(
            (
                country_coverage_df["date_coverage_pct"]
                >= 95
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries with >= 75% date coverage",
        "Observed Value": int(
            (
                country_coverage_df["date_coverage_pct"]
                >= 75
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries with < 50% date coverage",
        "Observed Value": int(
            (
                country_coverage_df["date_coverage_pct"]
                < 50
            ).sum()
        ),
    },
])

print()
print("Country reporting-date coverage diagnostics")
print("=" * 100)

display(date_coverage_summary_df)

# ------------------------------------------------------------
# Country coverage validation
# ------------------------------------------------------------

section_3_1_validation_df = pd.DataFrame([
    {
        "Validation Area": "Country population availability",
        "Requirement": "At least one validated country market must be available.",
        "Observed Evidence": (
            f"{len(country_coverage_df):,} country markets assessed"
        ),
        "Passed": len(country_coverage_df) > 0,
    },
    {
        "Validation Area": "Country observation preservation",
        "Requirement": "Coverage aggregation must preserve all country-market observations.",
        "Observed Evidence": (
            f"{int(country_coverage_df['observations'].sum()):,} "
            f"aggregated rows from {total_country_observations:,} source rows"
        ),
        "Passed": (
            int(country_coverage_df["observations"].sum())
            == total_country_observations
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": "The global aggregate market must not appear as a country.",
        "Observed Evidence": (
            f"'global' present in country table: "
            f"{'global' in set(country_coverage_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_coverage_df["country"].astype(str)
            )
        ),
    },
    {
        "Validation Area": "Finite coverage metrics",
        "Requirement": "Country coverage metrics must contain finite values.",
        "Observed Evidence": (
            f"{len(country_coverage_df):,} country summaries created"
        ),
        "Passed": bool(
            np.isfinite(
                country_coverage_df[
                    [
                        "observations",
                        "reporting_dates",
                        "unique_tracks",
                        "observation_share_pct",
                        "date_coverage_pct",
                    ]
                ].to_numpy(dtype=float)
            ).all()
        ),
    },
])

print()
print("Country coverage validation")
print("=" * 100)

display(section_3_1_validation_df)

section_3_1_complete = bool(
    section_3_1_validation_df["Passed"].all()
)

if not section_3_1_complete:
    failed_checks = section_3_1_validation_df.loc[
        ~section_3_1_validation_df["Passed"],
        "Validation Area"
    ].tolist()

    raise AssertionError(
        "Section 3.1 country coverage validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print("All Section 3.1 country coverage validation checks passed.")
print(f"Section 3.1 completion status: {section_3_1_complete}")
print(
    "Country coverage classifications are descriptive diagnostics "
    "and do not yet determine inclusion or exclusion from later analysis."
)
```

### Interpretation

Section 3.1 confirms that the PMIP geographic source provides substantial country-level coverage across the validated historical dataset.

A total of **76 country markets** are represented across **5,338,680 country-market observations**, covering **515 reporting dates**, **110,198 unique tracks** and **49,843 unique artist representations**.

Coverage is relatively balanced across many of the most established markets. Countries such as `pl`, `ch`, `us`, `se`, `no`, `de`, `mx`, `be`, `fr`, `it`, `au`, `pt`, `es`, `gb`, `fi`, `ie`, `sg`, `dk`, `nz` and `nl` each contain roughly 89,000 to 90,000 observations and all have complete coverage across the 515 reporting dates.

However, geographic coverage is not uniform across the entire dataset.

Several markets have substantially fewer historical observations. In particular, `ve`, `by`, `ad` and `pk` fall below the descriptive sparse-coverage boundary. Other markets such as `kz`, `ng`, `ru`, `kr`, `ua`, `sa`, `ae`, `eg`, `za`, `ma` and `in` also have materially shorter or thinner historical coverage than the best-represented markets.

Temporal coverage reinforces this difference:

- **27 countries** are represented on all 515 reporting dates.
- **32 countries** have at least 95% reporting-date coverage.
- **54 countries** have at least 75% reporting-date coverage.
- **15 countries** have less than 50% reporting-date coverage.

This means later cross-country comparisons must distinguish between performance differences and data-coverage differences. A market with fewer observations or a shorter historical period should not automatically be interpreted as having weaker artist or track performance.

The `global` aggregate market remains correctly excluded from country-level coverage calculations, and all validated country-market observations were preserved during aggregation.

All Section 3.1 validation checks passed. Country coverage classifications remain descriptive diagnostics rather than exclusion rules, and the workflow can proceed to artist- and track-level geographic coverage assessment.

## 3.2 Artist and Track Geographic Coverage

### Purpose

This subsection evaluates how widely individual artists and tracks are represented across the validated PMIP country markets.

The objective is to measure the geographic breadth of observed music performance before later developing international-reach, market-penetration and geographic-concentration indicators.

The analysis examines:

- the number of country markets associated with each track,
- the number of country markets associated with each artist representation,
- the number of reporting dates available for each entity,
- the number of observations supporting each track and artist,
- the distribution of local, multi-market and international coverage,
- whether artist and track geographic coverage is sufficiently broad for later cross-market analysis.

These measures describe data coverage only. They do not yet represent geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Assessing artist and track geographic coverage")
print("=" * 100)

if "country_geo_df" not in globals():
    raise RuntimeError(
        "country_geo_df is not available. Run Section 2.4 before Section 3.2."
    )

geo_entity_source_df = country_geo_df.copy()

total_country_markets = int(
    geo_entity_source_df["country"].nunique()
)

total_reporting_dates = int(
    geo_entity_source_df["date"].nunique()
)

# ------------------------------------------------------------
# Track geographic coverage
# ------------------------------------------------------------

track_geographic_coverage_df = (
    geo_entity_source_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_name=("track_name", "first"),
        artists_raw=("artists_raw", "first"),
        observations=("track_id", "size"),
        country_markets=("country", "nunique"),
        reporting_dates=("date", "nunique"),
        first_observed_date=("date", "min"),
        last_observed_date=("date", "max"),
        total_streams=("streams", "sum"),
        median_streams=("streams", "median"),
        best_chart_position=("position", "min"),
        median_chart_position=("position", "median"),
    )
    .reset_index()
)

track_geographic_coverage_df["market_coverage_pct"] = (
    track_geographic_coverage_df["country_markets"]
    / total_country_markets
    * 100
)

track_geographic_coverage_df["date_coverage_pct"] = (
    track_geographic_coverage_df["reporting_dates"]
    / total_reporting_dates
    * 100
)

track_geographic_coverage_df["coverage_span_days"] = (
    track_geographic_coverage_df["last_observed_date"]
    - track_geographic_coverage_df["first_observed_date"]
).dt.days

# ------------------------------------------------------------
# Artist-representation geographic coverage
# ------------------------------------------------------------

artist_geographic_coverage_df = (
    geo_entity_source_df
    .groupby(
        "artists_raw",
        dropna=False
    )
    .agg(
        observations=("artists_raw", "size"),
        country_markets=("country", "nunique"),
        reporting_dates=("date", "nunique"),
        unique_tracks=("track_id", "nunique"),
        first_observed_date=("date", "min"),
        last_observed_date=("date", "max"),
        total_streams=("streams", "sum"),
        median_streams=("streams", "median"),
        best_chart_position=("position", "min"),
        median_chart_position=("position", "median"),
    )
    .reset_index()
)

artist_geographic_coverage_df["market_coverage_pct"] = (
    artist_geographic_coverage_df["country_markets"]
    / total_country_markets
    * 100
)

artist_geographic_coverage_df["date_coverage_pct"] = (
    artist_geographic_coverage_df["reporting_dates"]
    / total_reporting_dates
    * 100
)

artist_geographic_coverage_df["coverage_span_days"] = (
    artist_geographic_coverage_df["last_observed_date"]
    - artist_geographic_coverage_df["first_observed_date"]
).dt.days

# ------------------------------------------------------------
# Coverage classifications
# ------------------------------------------------------------

def classify_market_reach(series):
    return np.select(
        [
            series == 1,
            series.between(2, 5),
            series.between(6, 15),
            series.between(16, 30),
            series > 30,
        ],
        [
            "Single-market",
            "Limited multi-market",
            "Moderate international",
            "Broad international",
            "Very broad international",
        ],
        default="Unclassified"
    )

track_geographic_coverage_df["coverage_classification"] = (
    classify_market_reach(
        track_geographic_coverage_df["country_markets"]
    )
)

artist_geographic_coverage_df["coverage_classification"] = (
    classify_market_reach(
        artist_geographic_coverage_df["country_markets"]
    )
)

# ------------------------------------------------------------
# Track coverage overview
# ------------------------------------------------------------

print()
print("Track geographic coverage overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_coverage_df):,}"
)

print(
    f"Minimum countries per track: "
    f"{track_geographic_coverage_df['country_markets'].min():,}"
)

print(
    f"Median countries per track: "
    f"{track_geographic_coverage_df['country_markets'].median():,.1f}"
)

print(
    f"Mean countries per track: "
    f"{track_geographic_coverage_df['country_markets'].mean():,.2f}"
)

print(
    f"Maximum countries per track: "
    f"{track_geographic_coverage_df['country_markets'].max():,}"
)

print()
print("Track geographic coverage classification")
print("=" * 100)

track_coverage_class_summary_df = (
    track_geographic_coverage_df[
        "coverage_classification"
    ]
    .value_counts()
    .rename_axis("Coverage Classification")
    .reset_index(name="Tracks")
)

track_coverage_class_summary_df["Share (%)"] = (
    track_coverage_class_summary_df["Tracks"]
    / len(track_geographic_coverage_df)
    * 100
)

display(track_coverage_class_summary_df)

# ------------------------------------------------------------
# Artist coverage overview
# ------------------------------------------------------------

print()
print("Artist-representation geographic coverage overview")
print("=" * 100)

print(
    f"Artist representations assessed: "
    f"{len(artist_geographic_coverage_df):,}"
)

print(
    f"Minimum countries per artist representation: "
    f"{artist_geographic_coverage_df['country_markets'].min():,}"
)

print(
    f"Median countries per artist representation: "
    f"{artist_geographic_coverage_df['country_markets'].median():,.1f}"
)

print(
    f"Mean countries per artist representation: "
    f"{artist_geographic_coverage_df['country_markets'].mean():,.2f}"
)

print(
    f"Maximum countries per artist representation: "
    f"{artist_geographic_coverage_df['country_markets'].max():,}"
)

print()
print("Artist geographic coverage classification")
print("=" * 100)

artist_coverage_class_summary_df = (
    artist_geographic_coverage_df[
        "coverage_classification"
    ]
    .value_counts()
    .rename_axis("Coverage Classification")
    .reset_index(name="Artist Representations")
)

artist_coverage_class_summary_df["Share (%)"] = (
    artist_coverage_class_summary_df[
        "Artist Representations"
    ]
    / len(artist_geographic_coverage_df)
    * 100
)

display(artist_coverage_class_summary_df)

# ------------------------------------------------------------
# Widest track coverage
# ------------------------------------------------------------

top_track_geographic_coverage_df = (
    track_geographic_coverage_df
    .sort_values(
        [
            "country_markets",
            "observations",
        ],
        ascending=[False, False]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Tracks with the widest observed geographic coverage")
print("=" * 100)

display(
    top_track_geographic_coverage_df[
        [
            "track_id",
            "track_name",
            "artists_raw",
            "country_markets",
            "market_coverage_pct",
            "reporting_dates",
            "observations",
            "coverage_classification",
        ]
    ]
)

# ------------------------------------------------------------
# Widest artist coverage
# ------------------------------------------------------------

top_artist_geographic_coverage_df = (
    artist_geographic_coverage_df
    .sort_values(
        [
            "country_markets",
            "observations",
        ],
        ascending=[False, False]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Artist representations with the widest observed geographic coverage")
print("=" * 100)

display(
    top_artist_geographic_coverage_df[
        [
            "artists_raw",
            "country_markets",
            "market_coverage_pct",
            "reporting_dates",
            "unique_tracks",
            "observations",
            "coverage_classification",
        ]
    ]
)

# ------------------------------------------------------------
# Distribution diagnostics
# ------------------------------------------------------------

coverage_distribution_df = pd.DataFrame([
    {
        "Entity": "Tracks",
        "Total Entities": len(track_geographic_coverage_df),
        "Single-Market": int(
            (
                track_geographic_coverage_df["country_markets"]
                == 1
            ).sum()
        ),
        "At Least 5 Markets": int(
            (
                track_geographic_coverage_df["country_markets"]
                >= 5
            ).sum()
        ),
        "At Least 10 Markets": int(
            (
                track_geographic_coverage_df["country_markets"]
                >= 10
            ).sum()
        ),
        "At Least 25 Markets": int(
            (
                track_geographic_coverage_df["country_markets"]
                >= 25
            ).sum()
        ),
        "At Least 50 Markets": int(
            (
                track_geographic_coverage_df["country_markets"]
                >= 50
            ).sum()
        ),
    },
    {
        "Entity": "Artist representations",
        "Total Entities": len(artist_geographic_coverage_df),
        "Single-Market": int(
            (
                artist_geographic_coverage_df["country_markets"]
                == 1
            ).sum()
        ),
        "At Least 5 Markets": int(
            (
                artist_geographic_coverage_df["country_markets"]
                >= 5
            ).sum()
        ),
        "At Least 10 Markets": int(
            (
                artist_geographic_coverage_df["country_markets"]
                >= 10
            ).sum()
        ),
        "At Least 25 Markets": int(
            (
                artist_geographic_coverage_df["country_markets"]
                >= 25
            ).sum()
        ),
        "At Least 50 Markets": int(
            (
                artist_geographic_coverage_df["country_markets"]
                >= 50
            ).sum()
        ),
    },
])

print()
print("Geographic breadth diagnostics")
print("=" * 100)

display(coverage_distribution_df)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

section_3_2_validation_df = pd.DataFrame([
    {
        "Validation Area": "Track coverage completeness",
        "Requirement": (
            "Every validated track must receive a geographic coverage summary."
        ),
        "Observed Evidence": (
            f"{len(track_geographic_coverage_df):,} track summaries "
            f"for {geo_entity_source_df['track_id'].nunique():,} unique tracks"
        ),
        "Passed": (
            len(track_geographic_coverage_df)
            == geo_entity_source_df["track_id"].nunique()
        ),
    },
    {
        "Validation Area": "Artist coverage completeness",
        "Requirement": (
            "Every validated artist representation must receive "
            "a geographic coverage summary."
        ),
        "Observed Evidence": (
            f"{len(artist_geographic_coverage_df):,} artist summaries "
            f"for {geo_entity_source_df['artists_raw'].nunique():,} "
            f"unique artist representations"
        ),
        "Passed": (
            len(artist_geographic_coverage_df)
            == geo_entity_source_df["artists_raw"].nunique()
        ),
    },
    {
        "Validation Area": "Track market bounds",
        "Requirement": (
            "No track may appear in more countries than the "
            "validated 76-market population."
        ),
        "Observed Evidence": (
            f"Maximum observed track coverage: "
            f"{track_geographic_coverage_df['country_markets'].max():,}"
        ),
        "Passed": (
            track_geographic_coverage_df[
                "country_markets"
            ].max()
            <= total_country_markets
        ),
    },
    {
        "Validation Area": "Artist market bounds",
        "Requirement": (
            "No artist representation may appear in more countries "
            "than the validated 76-market population."
        ),
        "Observed Evidence": (
            f"Maximum observed artist coverage: "
            f"{artist_geographic_coverage_df['country_markets'].max():,}"
        ),
        "Passed": (
            artist_geographic_coverage_df[
                "country_markets"
            ].max()
            <= total_country_markets
        ),
    },
    {
        "Validation Area": "Global market exclusion",
        "Requirement": (
            "Geographic breadth calculations must exclude "
            "the global aggregate market."
        ),
        "Observed Evidence": (
            f"Country-market population used: "
            f"{total_country_markets} markets"
        ),
        "Passed": total_country_markets == 76,
    },
])

print()
print("Artist and track geographic coverage validation")
print("=" * 100)

display(section_3_2_validation_df)

section_3_2_complete = bool(
    section_3_2_validation_df["Passed"].all()
)

if not section_3_2_complete:
    failed_checks = section_3_2_validation_df.loc[
        ~section_3_2_validation_df["Passed"],
        "Validation Area"
    ].tolist()

    raise AssertionError(
        "Section 3.2 artist and track geographic coverage "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 3.2 artist and track geographic coverage "
    "validation checks passed."
)
print(
    f"Section 3.2 completion status: "
    f"{section_3_2_complete}"
)
print(
    "These measures describe observed geographic coverage only "
    "and are not yet geographic performance scores."
)
```

### Interpretation

Section 3.2 confirms that geographic coverage is highly uneven across individual tracks and artist representations.

A total of **110,198 tracks** and **49,843 artist representations** were assessed across the validated 76-country market population.

For tracks:

- **73.58%** appear in only one country market.
- **17.55%** appear in between two and five markets.
- **4.11%** show moderate international coverage across 6 to 15 markets.
- **2.55%** show broad international coverage across 16 to 30 markets.
- **2.22%** appear in more than 30 markets.

The median track is therefore represented in only **one country**, while the mean is **3.21 countries**, showing a strongly right-skewed distribution in which a relatively small group of internationally successful tracks appears across many markets.

The widest-observed tracks reach as many as **73 of the 76 available country markets**, equivalent to approximately **96% geographic coverage**. This demonstrates that the dataset is capable of representing genuinely global or near-global track performance when it occurs.

Artist-level coverage follows a similar pattern. Around **69.88% of artist representations** are observed in only one market, while relatively small proportions achieve broad or very broad international coverage.

The median artist representation appears in **one country**, with an average of **3.71 countries**. However, the maximum artist coverage reaches all **76 validated markets**, confirming that the dataset contains examples of artists with complete observed geographic reach.

The geographic breadth diagnostics further show that:

- **11,257 tracks** appear in at least five markets,
- **7,288 tracks** appear in at least ten markets,
- **3,148 tracks** appear in at least 25 markets,
- **1,114 tracks** appear in at least 50 markets.

For artist representations:

- **6,142** appear in at least five markets,
- **4,080** appear in at least ten markets,
- **1,733** appear in at least 25 markets,
- **664** appear in at least 50 markets.

These results are important for later PMIP geographic intelligence because broad market presence should not be treated as the normal condition. Most artists and tracks have limited observed geographic reach, while international expansion is concentrated among a much smaller group.

The measures in this subsection describe **coverage rather than performance**. A track appearing in many countries is not automatically performing strongly in all of them, and a single-market track is not automatically weak.

All Section 3.2 validation checks passed, so the workflow can now proceed to historical coverage analysis by country.

## 3.3 Historical Coverage by Country

### Purpose

This subsection evaluates how country-level coverage changes through time.

The objective is to determine when each market enters the historical dataset, how long it remains represented, and whether reporting continuity differs across countries.

The analysis examines:

- first and last reporting dates for each country,
- historical coverage span,
- number and percentage of reporting dates represented,
- gaps in country reporting histories,
- late-entering markets,
- countries with continuous or fragmented historical coverage.

These diagnostics are required so that later geographic performance comparisons do not confuse differences in market history with differences in artist or track performance.

```python
import pandas as pd
import numpy as np

print("Assessing historical geographic coverage by country")
print("=" * 100)

if "country_geo_df" not in globals():
    raise RuntimeError(
        "country_geo_df is not available. Run Section 2.4 before Section 3.3."
    )

historical_source_df = country_geo_df.copy()

all_reporting_dates = (
    historical_source_df["date"]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)

total_reporting_dates = int(
    all_reporting_dates.nunique()
)

dataset_start_date = all_reporting_dates.min()
dataset_end_date = all_reporting_dates.max()

# ------------------------------------------------------------
# Country historical coverage
# ------------------------------------------------------------

country_historical_coverage_df = (
    historical_source_df
    .groupby(
        "country",
        dropna=False
    )
    .agg(
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        reporting_dates=("date", "nunique"),
        observations=("track_id", "size"),
        unique_tracks=("track_id", "nunique"),
        unique_artist_representations=("artists_raw", "nunique"),
    )
    .reset_index()
)

country_historical_coverage_df["coverage_span_days"] = (
    country_historical_coverage_df["last_reporting_date"]
    - country_historical_coverage_df["first_reporting_date"]
).dt.days

country_historical_coverage_df["date_coverage_pct"] = (
    country_historical_coverage_df["reporting_dates"]
    / total_reporting_dates
    * 100
)

country_historical_coverage_df["entry_delay_days"] = (
    country_historical_coverage_df["first_reporting_date"]
    - dataset_start_date
).dt.days

country_historical_coverage_df["exit_gap_days"] = (
    dataset_end_date
    - country_historical_coverage_df["last_reporting_date"]
).dt.days

# ------------------------------------------------------------
# Missing reporting dates within each country's active span
# ------------------------------------------------------------

country_gap_records = []

for country, country_df in historical_source_df.groupby(
    "country",
    dropna=False
):
    observed_dates = (
        country_df["date"]
        .drop_duplicates()
        .sort_values()
    )

    first_date = observed_dates.min()
    last_date = observed_dates.max()

    active_dataset_dates = all_reporting_dates[
        (all_reporting_dates >= first_date)
        & (all_reporting_dates <= last_date)
    ]

    missing_active_dates = (
        len(active_dataset_dates)
        - observed_dates.nunique()
    )

    country_gap_records.append(
        {
            "country": country,
            "active_period_possible_dates": len(active_dataset_dates),
            "observed_reporting_dates": observed_dates.nunique(),
            "missing_dates_within_active_span": int(
                missing_active_dates
            ),
            "active_span_coverage_pct": (
                observed_dates.nunique()
                / len(active_dataset_dates)
                * 100
                if len(active_dataset_dates) > 0
                else np.nan
            ),
        }
    )

country_gap_df = pd.DataFrame(
    country_gap_records
)

country_historical_coverage_df = (
    country_historical_coverage_df
    .merge(
        country_gap_df,
        on="country",
        how="left",
        validate="one_to_one"
    )
)

# ------------------------------------------------------------
# Historical classification
# ------------------------------------------------------------

country_historical_coverage_df[
    "historical_coverage_class"
] = np.select(
    [
        country_historical_coverage_df[
            "date_coverage_pct"
        ] >= 95,

        country_historical_coverage_df[
            "date_coverage_pct"
        ].between(75, 95, inclusive="left"),

        country_historical_coverage_df[
            "date_coverage_pct"
        ].between(50, 75, inclusive="left"),

        country_historical_coverage_df[
            "date_coverage_pct"
        ] < 50,
    ],
    [
        "Near-complete history",
        "Strong history",
        "Partial history",
        "Limited history",
    ],
    default="Unclassified"
)

# ------------------------------------------------------------
# Overview
# ------------------------------------------------------------

print()
print("Historical country coverage overview")
print("=" * 100)

print(
    f"Dataset reporting period: "
    f"{dataset_start_date.date()} to {dataset_end_date.date()}"
)

print(
    f"Total reporting dates available: "
    f"{total_reporting_dates:,}"
)

print(
    f"Countries assessed: "
    f"{len(country_historical_coverage_df):,}"
)

print(
    f"Earliest country entry: "
    f"{country_historical_coverage_df['first_reporting_date'].min().date()}"
)

print(
    f"Latest country entry: "
    f"{country_historical_coverage_df['first_reporting_date'].max().date()}"
)

# ------------------------------------------------------------
# Classification summary
# ------------------------------------------------------------

historical_class_summary_df = (
    country_historical_coverage_df[
        "historical_coverage_class"
    ]
    .value_counts()
    .rename_axis("Historical Coverage Class")
    .reset_index(name="Countries")
)

historical_class_summary_df["Share (%)"] = (
    historical_class_summary_df["Countries"]
    / len(country_historical_coverage_df)
    * 100
)

print()
print("Historical coverage classification")
print("=" * 100)

display(historical_class_summary_df)

# ------------------------------------------------------------
# Earliest entering countries
# ------------------------------------------------------------

earliest_country_entry_df = (
    country_historical_coverage_df
    .sort_values(
        [
            "first_reporting_date",
            "country",
        ]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Earliest represented country markets")
print("=" * 100)

display(
    earliest_country_entry_df[
        [
            "country",
            "first_reporting_date",
            "last_reporting_date",
            "reporting_dates",
            "date_coverage_pct",
            "missing_dates_within_active_span",
            "active_span_coverage_pct",
            "historical_coverage_class",
        ]
    ]
)

# ------------------------------------------------------------
# Latest entering countries
# ------------------------------------------------------------

latest_country_entry_df = (
    country_historical_coverage_df
    .sort_values(
        [
            "first_reporting_date",
            "country",
        ],
        ascending=[False, True]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Latest entering country markets")
print("=" * 100)

display(
    latest_country_entry_df[
        [
            "country",
            "first_reporting_date",
            "last_reporting_date",
            "reporting_dates",
            "date_coverage_pct",
            "entry_delay_days",
            "missing_dates_within_active_span",
            "active_span_coverage_pct",
            "historical_coverage_class",
        ]
    ]
)

# ------------------------------------------------------------
# Reporting-gap diagnostics
# ------------------------------------------------------------

country_gap_summary_df = (
    country_historical_coverage_df
    .sort_values(
        "missing_dates_within_active_span",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Countries with the largest reporting gaps inside their active span")
print("=" * 100)

display(
    country_gap_summary_df[
        [
            "country",
            "reporting_dates",
            "active_period_possible_dates",
            "missing_dates_within_active_span",
            "active_span_coverage_pct",
            "first_reporting_date",
            "last_reporting_date",
        ]
    ]
)

# ------------------------------------------------------------
# Historical diagnostics
# ------------------------------------------------------------

historical_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Countries covering full dataset period",
        "Observed Value": int(
            (
                (
                    country_historical_coverage_df[
                        "first_reporting_date"
                    ] == dataset_start_date
                )
                &
                (
                    country_historical_coverage_df[
                        "last_reporting_date"
                    ] == dataset_end_date
                )
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries with no internal reporting gaps",
        "Observed Value": int(
            (
                country_historical_coverage_df[
                    "missing_dates_within_active_span"
                ] == 0
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries with at least one internal gap",
        "Observed Value": int(
            (
                country_historical_coverage_df[
                    "missing_dates_within_active_span"
                ] > 0
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries entering after dataset start",
        "Observed Value": int(
            (
                country_historical_coverage_df[
                    "entry_delay_days"
                ] > 0
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries ending before dataset end",
        "Observed Value": int(
            (
                country_historical_coverage_df[
                    "exit_gap_days"
                ] > 0
            ).sum()
        ),
    },
])

print()
print("Historical coverage diagnostics")
print("=" * 100)

display(historical_diagnostics_df)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

section_3_3_validation_df = pd.DataFrame([
    {
        "Validation Area": "Country history completeness",
        "Requirement": (
            "Every validated country market must receive "
            "a historical coverage summary."
        ),
        "Observed Evidence": (
            f"{len(country_historical_coverage_df):,} summaries "
            f"for {historical_source_df['country'].nunique():,} markets"
        ),
        "Passed": (
            len(country_historical_coverage_df)
            == historical_source_df["country"].nunique()
        ),
    },
    {
        "Validation Area": "Date-boundary validity",
        "Requirement": (
            "Country histories must remain inside the "
            "validated PMIP reporting period."
        ),
        "Observed Evidence": (
            f"{dataset_start_date.date()} to "
            f"{dataset_end_date.date()}"
        ),
        "Passed": bool(
            (
                country_historical_coverage_df[
                    "first_reporting_date"
                ] >= dataset_start_date
            ).all()
            and
            (
                country_historical_coverage_df[
                    "last_reporting_date"
                ] <= dataset_end_date
            ).all()
        ),
    },
    {
        "Validation Area": "Chronological ordering",
        "Requirement": (
            "Each country's first reporting date must not "
            "occur after its last reporting date."
        ),
        "Observed Evidence": (
            f"{len(country_historical_coverage_df):,} "
            f"country histories evaluated"
        ),
        "Passed": bool(
            (
                country_historical_coverage_df[
                    "first_reporting_date"
                ]
                <=
                country_historical_coverage_df[
                    "last_reporting_date"
                ]
            ).all()
        ),
    },
    {
        "Validation Area": "Coverage percentage bounds",
        "Requirement": (
            "Historical date-coverage percentages must remain "
            "between 0% and 100%."
        ),
        "Observed Evidence": (
            f"Observed range: "
            f"{country_historical_coverage_df['date_coverage_pct'].min():.2f}% "
            f"to "
            f"{country_historical_coverage_df['date_coverage_pct'].max():.2f}%"
        ),
        "Passed": bool(
            country_historical_coverage_df[
                "date_coverage_pct"
            ].between(0, 100).all()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Historical country analysis must exclude "
            "the global aggregate market."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(country_historical_coverage_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_historical_coverage_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Historical country coverage validation")
print("=" * 100)

display(section_3_3_validation_df)

section_3_3_complete = bool(
    section_3_3_validation_df["Passed"].all()
)

if not section_3_3_complete:
    failed_checks = section_3_3_validation_df.loc[
        ~section_3_3_validation_df["Passed"],
        "Validation Area"
    ].tolist()

    raise AssertionError(
        "Section 3.3 historical country coverage validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 3.3 historical country coverage "
    "validation checks passed."
)
print(
    f"Section 3.3 completion status: "
    f"{section_3_3_complete}"
)
print(
    "Historical coverage differences remain explicit diagnostics "
    "for later geographic performance interpretation."
)
```

### Interpretation

Section 3.3 confirms that historical geographic coverage differs substantially across PMIP country markets, particularly in terms of when countries first appear in the dataset.

The validated historical period runs from **28 April 2013 to 6 April 2023**, covering **515 reporting dates** across **76 country markets**.

Historical coverage can be divided into four broad groups:

- **32 countries (42.11%)** have near-complete historical coverage.
- **22 countries (28.95%)** have strong historical coverage.
- **7 countries (9.21%)** have partial historical coverage.
- **15 countries (19.74%)** have limited historical coverage.

A major finding is that **48 of the 76 countries entered the dataset after the overall dataset start date**. This means historical coverage differences are driven largely by staggered market entry rather than widespread missing observations after a country becomes active.

Only **12 countries** contain at least one reporting gap within their active historical span, while **64 countries** have continuous coverage once they begin appearing in the dataset.

Several established markets, including `at`, `au`, `be`, `ch`, `de`, `dk`, `ee`, `es`, `fi`, `fr`, `gb`, `hk`, `ie`, `is`, `it`, `lt`, `lv`, `mx` and others, are represented from the beginning of the dataset and maintain near-complete historical coverage.

By contrast, some markets entered much later. For example:

- `pk` begins in May 2022,
- `ve` begins in March 2022,
- `by`, `kz` and `ng` begin in February 2022,
- `kr` begins in February 2021,
- `ru` and `ua` begin in July 2020,
- several Middle Eastern and African markets begin in 2019,
- `jp` and `th` begin in 2017.

These later entry dates explain why some countries have substantially fewer total observations than long-running markets.

Most countries also show strong continuity once active. Even many late-entry countries achieve close to **100% active-span coverage**, meaning they are consistently represented from their entry date onward.

One important exception is `ad`, which contains only **120 observed reporting dates from 352 possible dates within its active span**, giving approximately **34.09% active-span coverage**. This represents a genuine internal coverage limitation and should be treated separately from markets that simply entered the dataset late.

Three countries also stop appearing before the overall dataset end date, showing that historical coverage is not perfectly symmetrical across all markets.

These findings mean that later PMIP geographic comparisons must consider both:

1. **historical depth** — how long a country has been present in the dataset, and
2. **active-period continuity** — how consistently the country is represented once it becomes available.

A country with a shorter history should not automatically be interpreted as having weaker artist or track performance than a country with ten years of observations.

All Section 3.3 validation checks passed. Historical coverage differences will therefore remain explicit context for later geographic performance indicators and comparisons.

## 3.4 Missingness and Data-Quality Assessment

### Purpose

This subsection evaluates whether the validated country-market dataset contains remaining data-quality issues that could distort later geographic performance analysis.

The analysis focuses on:

- missing values across the canonical geographic schema,
- zero-stream observations,
- unusual or extreme chart-position values,
- countries with weak historical continuity,
- countries with very low observation support,
- sparse country-date coverage,
- whether any country-market combinations require additional caution before performance comparison.

The objective is not to remove valid observations automatically, but to identify where geographic interpretation may need stronger safeguards.

```python
import pandas as pd
import numpy as np

print("Assessing geographic missingness and data quality")
print("=" * 100)

if "country_geo_df" not in globals():
    raise RuntimeError(
        "country_geo_df is not available. Run Section 2.4 before Section 3.4."
    )

if "country_historical_coverage_df" not in globals():
    raise RuntimeError(
        "country_historical_coverage_df is not available. "
        "Run Section 3.3 before Section 3.4."
    )

quality_source_df = country_geo_df.copy()

# ------------------------------------------------------------
# 1. Canonical missingness assessment
# ------------------------------------------------------------

quality_fields = [
    "date",
    "country",
    "track_id",
    "track_name",
    "artists_raw",
    "streams",
    "position",
    "duration",
    "explicit",
    "primary_artist",
]

missingness_records = []

for column in quality_fields:
    missing_count = int(
        quality_source_df[column].isna().sum()
    )

    missing_pct = (
        missing_count
        / len(quality_source_df)
        * 100
    )

    missingness_records.append(
        {
            "Field": column,
            "Missing Observations": missing_count,
            "Missing (%)": missing_pct,
            "Non-Missing Observations": int(
                quality_source_df[column].notna().sum()
            ),
        }
    )

geographic_missingness_df = pd.DataFrame(
    missingness_records
)

print()
print("Canonical geographic missingness")
print("=" * 100)
display(geographic_missingness_df)

# ------------------------------------------------------------
# 2. Stream-value diagnostics
# ------------------------------------------------------------

streams_numeric = pd.to_numeric(
    quality_source_df["streams"],
    errors="coerce"
)

stream_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Non-finite stream values",
        "Observed Value": int(
            (~np.isfinite(streams_numeric)).sum()
        ),
    },
    {
        "Diagnostic": "Negative stream values",
        "Observed Value": int(
            (
                np.isfinite(streams_numeric)
                & (streams_numeric < 0)
            ).sum()
        ),
    },
    {
        "Diagnostic": "Zero-stream observations",
        "Observed Value": int(
            (
                np.isfinite(streams_numeric)
                & (streams_numeric == 0)
            ).sum()
        ),
    },
    {
        "Diagnostic": "Positive-stream observations",
        "Observed Value": int(
            (
                np.isfinite(streams_numeric)
                & (streams_numeric > 0)
            ).sum()
        ),
    },
])

print()
print("Stream-value diagnostics")
print("=" * 100)
display(stream_diagnostics_df)

# ------------------------------------------------------------
# 3. Zero-stream observation review
# ------------------------------------------------------------

zero_stream_df = (
    quality_source_df.loc[
        streams_numeric.eq(0),
        [
            "date",
            "country",
            "track_id",
            "track_name",
            "artists_raw",
            "streams",
            "position",
        ]
    ]
    .copy()
    .sort_values(
        [
            "date",
            "country",
            "track_id",
        ]
    )
    .reset_index(drop=True)
)

print()
print("Zero-stream observation review")
print("=" * 100)

if len(zero_stream_df) > 0:
    display(zero_stream_df)
else:
    print("No zero-stream observations found.")

# ------------------------------------------------------------
# 4. Chart-position diagnostics
# ------------------------------------------------------------

position_numeric = pd.to_numeric(
    quality_source_df["position"],
    errors="coerce"
)

position_summary_df = pd.DataFrame([
    {
        "Diagnostic": "Minimum chart position",
        "Observed Value": float(
            position_numeric.min()
        ),
    },
    {
        "Diagnostic": "Median chart position",
        "Observed Value": float(
            position_numeric.median()
        ),
    },
    {
        "Diagnostic": "95th percentile chart position",
        "Observed Value": float(
            position_numeric.quantile(0.95)
        ),
    },
    {
        "Diagnostic": "99th percentile chart position",
        "Observed Value": float(
            position_numeric.quantile(0.99)
        ),
    },
    {
        "Diagnostic": "Maximum chart position",
        "Observed Value": float(
            position_numeric.max()
        ),
    },
])

print()
print("Chart-position diagnostics")
print("=" * 100)
display(position_summary_df)

# ------------------------------------------------------------
# 5. Country-level quality diagnostics
# ------------------------------------------------------------

country_quality_df = (
    quality_source_df
    .groupby(
        "country",
        dropna=False
    )
    .agg(
        observations=("track_id", "size"),
        reporting_dates=("date", "nunique"),
        unique_tracks=("track_id", "nunique"),
        unique_artist_representations=("artists_raw", "nunique"),
        zero_stream_rows=(
            "streams",
            lambda s: int(
                (
                    pd.to_numeric(
                        s,
                        errors="coerce"
                    ) == 0
                ).sum()
            )
        ),
        minimum_position=("position", "min"),
        maximum_position=("position", "max"),
        median_position=("position", "median"),
    )
    .reset_index()
)

country_quality_df = (
    country_quality_df
    .merge(
        country_historical_coverage_df[
            [
                "country",
                "date_coverage_pct",
                "active_span_coverage_pct",
                "missing_dates_within_active_span",
                "historical_coverage_class",
            ]
        ],
        on="country",
        how="left",
        validate="one_to_one"
    )
)

# ------------------------------------------------------------
# 6. Country quality-risk flags
# ------------------------------------------------------------

observation_q1 = float(
    country_quality_df["observations"].quantile(0.25)
)

observation_q3 = float(
    country_quality_df["observations"].quantile(0.75)
)

observation_iqr = (
    observation_q3
    - observation_q1
)

sparse_observation_boundary = max(
    0,
    observation_q1
    - 1.5 * observation_iqr
)

country_quality_df[
    "low_observation_support"
] = (
    country_quality_df["observations"]
    < sparse_observation_boundary
)

country_quality_df[
    "low_date_coverage"
] = (
    country_quality_df["date_coverage_pct"]
    < 50
)

country_quality_df[
    "internal_gap_present"
] = (
    country_quality_df[
        "missing_dates_within_active_span"
    ] > 0
)

country_quality_df[
    "material_internal_gap"
] = (
    country_quality_df[
        "active_span_coverage_pct"
    ] < 95
)

country_quality_df[
    "zero_stream_present"
] = (
    country_quality_df[
        "zero_stream_rows"
    ] > 0
)

country_quality_df[
    "quality_risk_count"
] = (
    country_quality_df[
        [
            "low_observation_support",
            "low_date_coverage",
            "internal_gap_present",
            "material_internal_gap",
            "zero_stream_present",
        ]
    ]
    .astype(int)
    .sum(axis=1)
)

country_quality_df[
    "quality_review_priority"
] = np.select(
    [
        country_quality_df[
            "quality_risk_count"
        ] >= 3,

        country_quality_df[
            "quality_risk_count"
        ] == 2,

        country_quality_df[
            "quality_risk_count"
        ] == 1,
    ],
    [
        "High review priority",
        "Moderate review priority",
        "Limited review priority",
    ],
    default="No material issue identified"
)

# ------------------------------------------------------------
# 7. Highest-risk country summaries
# ------------------------------------------------------------

highest_quality_risk_df = (
    country_quality_df
    .sort_values(
        [
            "quality_risk_count",
            "active_span_coverage_pct",
            "observations",
        ],
        ascending=[
            False,
            True,
            True,
        ]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Countries requiring the strongest data-quality caution")
print("=" * 100)

display(
    highest_quality_risk_df[
        [
            "country",
            "observations",
            "reporting_dates",
            "date_coverage_pct",
            "active_span_coverage_pct",
            "missing_dates_within_active_span",
            "zero_stream_rows",
            "low_observation_support",
            "low_date_coverage",
            "material_internal_gap",
            "quality_risk_count",
            "quality_review_priority",
        ]
    ]
)

# ------------------------------------------------------------
# 8. Country quality summary
# ------------------------------------------------------------

quality_priority_summary_df = (
    country_quality_df[
        "quality_review_priority"
    ]
    .value_counts()
    .rename_axis(
        "Quality Review Priority"
    )
    .reset_index(
        name="Countries"
    )
)

quality_priority_summary_df[
    "Share (%)"
] = (
    quality_priority_summary_df[
        "Countries"
    ]
    / len(country_quality_df)
    * 100
)

print()
print("Country data-quality review summary")
print("=" * 100)

display(
    quality_priority_summary_df
)

# ------------------------------------------------------------
# 9. Data-quality validation
# ------------------------------------------------------------

all_required_missing = int(
    geographic_missingness_df[
        "Missing Observations"
    ].sum()
)

section_3_4_validation_df = pd.DataFrame([
    {
        "Validation Area": "Required-field completeness",
        "Requirement": (
            "Validated canonical geographic fields must "
            "remain non-missing."
        ),
        "Observed Evidence": (
            f"{all_required_missing:,} missing values "
            f"across required canonical fields"
        ),
        "Passed": (
            all_required_missing == 0
        ),
    },
    {
        "Validation Area": "Stream validity",
        "Requirement": (
            "Stream values must remain finite and non-negative."
        ),
        "Observed Evidence": (
            f"{int((~np.isfinite(streams_numeric)).sum()):,} "
            f"non-finite; "
            f"{int(((streams_numeric < 0) & np.isfinite(streams_numeric)).sum()):,} "
            f"negative"
        ),
        "Passed": bool(
            np.isfinite(
                streams_numeric
            ).all()
            and
            (
                streams_numeric >= 0
            ).all()
        ),
    },
    {
        "Validation Area": "Chart-position validity",
        "Requirement": (
            "Chart positions must remain finite positive ranks."
        ),
        "Observed Evidence": (
            f"Observed position range: "
            f"{int(position_numeric.min())} to "
            f"{int(position_numeric.max())}"
        ),
        "Passed": bool(
            np.isfinite(
                position_numeric
            ).all()
            and
            (
                position_numeric > 0
            ).all()
        ),
    },
    {
        "Validation Area": "Country quality coverage",
        "Requirement": (
            "Every validated country market must receive "
            "a quality assessment."
        ),
        "Observed Evidence": (
            f"{len(country_quality_df):,} quality summaries "
            f"for {quality_source_df['country'].nunique():,} countries"
        ),
        "Passed": (
            len(country_quality_df)
            == quality_source_df[
                "country"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Country-level quality diagnostics must exclude "
            "the global aggregate market."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(country_quality_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_quality_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Geographic data-quality validation")
print("=" * 100)

display(
    section_3_4_validation_df
)

section_3_4_complete = bool(
    section_3_4_validation_df[
        "Passed"
    ].all()
)

if not section_3_4_complete:
    failed_checks = (
        section_3_4_validation_df.loc[
            ~section_3_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 3.4 geographic data-quality "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 3.4 geographic data-quality "
    "validation checks passed."
)
print(
    f"Section 3.4 completion status: "
    f"{section_3_4_complete}"
)
print(
    "Quality-risk classifications are interpretation safeguards "
    "and do not automatically remove country markets or observations."
)
```

### Interpretation

Section 3.4 confirms that the validated country-market dataset is structurally clean and suitable for continued geographic analysis.

All required canonical geographic fields contain complete data across the **5,338,680 country-market observations**. No missing values were found in the validated date, country, track, artist, stream, chart-position, duration, explicit-status or primary-artist fields.

Stream values are also structurally valid. No non-finite or negative stream values were identified. Only **7 observations contain zero streams**, representing an extremely small proportion of the full analytical population. These observations are retained because zero may represent a legitimate recorded value rather than an error.

Chart positions remain valid positive ranks ranging from **1 to 358**. The median chart position is **97**, while the 95th and 99th percentiles are approximately **191** and **199**, showing that most observations fall within the more commonly represented chart range even though higher ranks are present in the source.

Country-level quality assessment reveals that most geographic markets do not present major structural concerns:

- **53 countries (69.74%)** show no material data-quality issue.
- **11 countries (14.47%)** require limited review.
- **9 countries (11.84%)** require moderate review.
- **3 countries (3.95%)** receive high review priority.

The strongest quality concerns arise mainly from limited historical coverage, low observation support or internal reporting gaps rather than missing core fields.

For example, `ad` combines very low observation support, low overall historical coverage and substantial internal missingness, making it one of the least reliable markets for direct historical comparison. Other markets such as `by`, `ve`, `pk`, `kr`, `ru` and several later-entry countries also require caution because their historical depth is much shorter than that of long-running markets.

These quality-risk classifications are interpretive safeguards only. They do not automatically remove countries or observations from the PMIP geographic intelligence workflow.

Overall, the dataset is highly complete at the row level, while the main remaining quality concerns are geographic and temporal coverage differences between markets.

All Section 3.4 validation checks passed, and the workflow can proceed to Section 3.5, where these coverage risks will be consolidated into explicit geographic-analysis limitations.

## 3.5 Geographic Coverage Limitations

### Purpose

This subsection consolidates the main geographic coverage limitations identified across Sections 3.1 to 3.4.

The objective is to define clear interpretation safeguards before later PMIP geographic performance indicators are calculated.

The analysis documents limitations relating to:

- unequal country-level observation depth,
- staggered market-entry dates,
- incomplete historical coverage,
- internal reporting gaps,
- uneven artist and track geographic representation,
- sparse support in selected country markets,
- the distinction between country-level and global aggregate observations,
- the risk of confusing data availability with music performance.

These limitations do not automatically invalidate the geographic dataset. Instead, they define the conditions under which later geographic comparisons should be interpreted cautiously.

```python
import pandas as pd
import numpy as np

print("Consolidating PMIP geographic coverage limitations")
print("=" * 100)

required_objects = [
    "country_coverage_df",
    "track_geographic_coverage_df",
    "artist_geographic_coverage_df",
    "country_historical_coverage_df",
    "country_quality_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required Section 3 objects are missing: "
        + ", ".join(missing_objects)
    )

# ------------------------------------------------------------
# 1. Core limitation evidence
# ------------------------------------------------------------

total_countries = int(
    country_coverage_df["country"].nunique()
)

countries_below_50_date_coverage = int(
    (
        country_historical_coverage_df[
            "date_coverage_pct"
        ] < 50
    ).sum()
)

countries_entering_late = int(
    (
        country_historical_coverage_df[
            "entry_delay_days"
        ] > 0
    ).sum()
)

countries_with_internal_gaps = int(
    (
        country_historical_coverage_df[
            "missing_dates_within_active_span"
        ] > 0
    ).sum()
)

countries_with_material_internal_gap = int(
    (
        country_historical_coverage_df[
            "active_span_coverage_pct"
        ] < 95
    ).sum()
)

countries_high_review = int(
    (
        country_quality_df[
            "quality_review_priority"
        ] == "High review priority"
    ).sum()
)

countries_moderate_or_high_review = int(
    country_quality_df[
        "quality_review_priority"
    ].isin(
        [
            "Moderate review priority",
            "High review priority",
        ]
    ).sum()
)

single_market_tracks = int(
    (
        track_geographic_coverage_df[
            "country_markets"
        ] == 1
    ).sum()
)

single_market_track_share = (
    single_market_tracks
    / len(track_geographic_coverage_df)
    * 100
)

single_market_artists = int(
    (
        artist_geographic_coverage_df[
            "country_markets"
        ] == 1
    ).sum()
)

single_market_artist_share = (
    single_market_artists
    / len(artist_geographic_coverage_df)
    * 100
)

tracks_50_plus_markets = int(
    (
        track_geographic_coverage_df[
            "country_markets"
        ] >= 50
    ).sum()
)

artists_50_plus_markets = int(
    (
        artist_geographic_coverage_df[
            "country_markets"
        ] >= 50
    ).sum()
)

minimum_country_observations = int(
    country_coverage_df[
        "observations"
    ].min()
)

maximum_country_observations = int(
    country_coverage_df[
        "observations"
    ].max()
)

country_observation_ratio = (
    maximum_country_observations
    / minimum_country_observations
    if minimum_country_observations > 0
    else np.nan
)

# ------------------------------------------------------------
# 2. Coverage-limitation register
# ------------------------------------------------------------

geographic_coverage_limitations_df = pd.DataFrame([
    {
        "Limitation": "Unequal country observation depth",
        "Observed Evidence": (
            f"Country observations range from "
            f"{minimum_country_observations:,} to "
            f"{maximum_country_observations:,}; "
            f"largest/smallest ratio ≈ "
            f"{country_observation_ratio:.2f}x"
        ),
        "Analytical Risk": (
            "Countries with larger historical populations may appear "
            "more prominent simply because they have more observations."
        ),
        "Required Safeguard": (
            "Use rates, normalised indicators and minimum-support checks "
            "alongside raw counts."
        ),
    },
    {
        "Limitation": "Staggered market entry",
        "Observed Evidence": (
            f"{countries_entering_late} of {total_countries} countries "
            f"entered after the dataset start date."
        ),
        "Analytical Risk": (
            "Later-entry markets have less opportunity to accumulate "
            "tracks, artists and observations."
        ),
        "Required Safeguard": (
            "Compare markets using overlapping time periods or "
            "coverage-adjusted measures where appropriate."
        ),
    },
    {
        "Limitation": "Limited historical coverage",
        "Observed Evidence": (
            f"{countries_below_50_date_coverage} countries have "
            f"less than 50% reporting-date coverage."
        ),
        "Analytical Risk": (
            "Long-term comparisons may unfairly disadvantage markets "
            "with short historical records."
        ),
        "Required Safeguard": (
            "Retain historical coverage percentages and first/last "
            "reporting dates in later market summaries."
        ),
    },
    {
        "Limitation": "Internal reporting gaps",
        "Observed Evidence": (
            f"{countries_with_internal_gaps} countries contain at least "
            f"one reporting gap within their active span; "
            f"{countries_with_material_internal_gap} fall below 95% "
            f"active-span coverage."
        ),
        "Analytical Risk": (
            "Interrupted country histories may distort time-based "
            "performance trends."
        ),
        "Required Safeguard": (
            "Flag fragmented market histories before trend or growth "
            "interpretation."
        ),
    },
    {
        "Limitation": "Track geographic sparsity",
        "Observed Evidence": (
            f"{single_market_tracks:,} tracks "
            f"({single_market_track_share:.2f}%) appear in only one market, "
            f"while only {tracks_50_plus_markets:,} appear in at least 50."
        ),
        "Analytical Risk": (
            "Most tracks do not provide enough international coverage "
            "for broad cross-market conclusions."
        ),
        "Required Safeguard": (
            "Only compute broad international indicators when adequate "
            "market support exists."
        ),
    },
    {
        "Limitation": "Artist geographic sparsity",
        "Observed Evidence": (
            f"{single_market_artists:,} artist representations "
            f"({single_market_artist_share:.2f}%) appear in one market, "
            f"while only {artists_50_plus_markets:,} appear in at least 50."
        ),
        "Analytical Risk": (
            "Artist-level geographic breadth can be strongly influenced "
            "by limited observed market coverage."
        ),
        "Required Safeguard": (
            "Separate local, limited multi-market and broad international "
            "artist populations in later analysis."
        ),
    },
    {
        "Limitation": "Country quality-risk concentration",
        "Observed Evidence": (
            f"{countries_moderate_or_high_review} countries require "
            f"moderate or high review; {countries_high_review} are "
            f"high-review priority."
        ),
        "Analytical Risk": (
            "Some markets contain combinations of low support, short history "
            "or reporting gaps that weaken direct comparability."
        ),
        "Required Safeguard": (
            "Carry forward country-level quality-review labels into later "
            "performance outputs."
        ),
    },
    {
        "Limitation": "Global aggregate is not a country market",
        "Observed Evidence": (
            "88,456 global aggregate observations were separated from "
            "country-level analysis in Section 2.4."
        ),
        "Analytical Risk": (
            "Treating the global aggregate as a country would inflate "
            "market-reach and penetration measures."
        ),
        "Required Safeguard": (
            "Keep global aggregate outputs separate from country-count "
            "and international-reach metrics."
        ),
    },
    {
        "Limitation": "Coverage does not equal performance",
        "Observed Evidence": (
            "Sections 3.1 to 3.4 describe observation availability, "
            "not music-performance strength."
        ),
        "Analytical Risk": (
            "High observation counts or broad geographic presence may be "
            "misread as superior streaming or chart performance."
        ),
        "Required Safeguard": (
            "Keep geographic coverage indicators separate from later "
            "performance indicators."
        ),
    },
])

print()
print("Geographic coverage limitation register")
print("=" * 100)

display(
    geographic_coverage_limitations_df
)

# ------------------------------------------------------------
# 3. Country interpretation safeguards
# ------------------------------------------------------------

country_interpretation_safeguards_df = (
    country_quality_df[
        [
            "country",
            "observations",
            "date_coverage_pct",
            "active_span_coverage_pct",
            "missing_dates_within_active_span",
            "quality_review_priority",
        ]
    ]
    .copy()
)

country_interpretation_safeguards_df[
    "historical_comparison_allowed"
] = (
    country_interpretation_safeguards_df[
        "date_coverage_pct"
    ] >= 50
)

country_interpretation_safeguards_df[
    "trend_interpretation_allowed"
] = (
    country_interpretation_safeguards_df[
        "active_span_coverage_pct"
    ] >= 95
)

country_interpretation_safeguards_df[
    "direct_comparison_caution"
] = np.where(
    country_interpretation_safeguards_df[
        "quality_review_priority"
    ].isin(
        [
            "Moderate review priority",
            "High review priority",
        ]
    ),
    True,
    False
)

print()
print("Country interpretation safeguard summary")
print("=" * 100)

safeguard_summary_df = pd.DataFrame([
    {
        "Safeguard": "Historical comparison supported",
        "Countries": int(
            country_interpretation_safeguards_df[
                "historical_comparison_allowed"
            ].sum()
        ),
    },
    {
        "Safeguard": "Historical comparison limited",
        "Countries": int(
            (
                ~country_interpretation_safeguards_df[
                    "historical_comparison_allowed"
                ]
            ).sum()
        ),
    },
    {
        "Safeguard": "Trend interpretation supported",
        "Countries": int(
            country_interpretation_safeguards_df[
                "trend_interpretation_allowed"
            ].sum()
        ),
    },
    {
        "Safeguard": "Trend interpretation requires caution",
        "Countries": int(
            (
                ~country_interpretation_safeguards_df[
                    "trend_interpretation_allowed"
                ]
            ).sum()
        ),
    },
    {
        "Safeguard": "Direct comparison caution required",
        "Countries": int(
            country_interpretation_safeguards_df[
                "direct_comparison_caution"
            ].sum()
        ),
    },
])

display(
    safeguard_summary_df
)

# ------------------------------------------------------------
# 4. Validation
# ------------------------------------------------------------

section_3_5_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 3.1 completion",
        "Requirement": (
            "Country coverage assessment must already be complete."
        ),
        "Observed Evidence": (
            f"Section 3.1 completion status: "
            f"{globals().get('section_3_1_complete', False)}"
        ),
        "Passed": bool(
            globals().get(
                "section_3_1_complete",
                False
            )
        ),
    },
    {
        "Validation Area": "Section 3.2 completion",
        "Requirement": (
            "Artist and track geographic coverage must already be complete."
        ),
        "Observed Evidence": (
            f"Section 3.2 completion status: "
            f"{globals().get('section_3_2_complete', False)}"
        ),
        "Passed": bool(
            globals().get(
                "section_3_2_complete",
                False
            )
        ),
    },
    {
        "Validation Area": "Section 3.3 completion",
        "Requirement": (
            "Historical country coverage must already be complete."
        ),
        "Observed Evidence": (
            f"Section 3.3 completion status: "
            f"{globals().get('section_3_3_complete', False)}"
        ),
        "Passed": bool(
            globals().get(
                "section_3_3_complete",
                False
            )
        ),
    },
    {
        "Validation Area": "Section 3.4 completion",
        "Requirement": (
            "Data-quality assessment must already be complete."
        ),
        "Observed Evidence": (
            f"Section 3.4 completion status: "
            f"{globals().get('section_3_4_complete', False)}"
        ),
        "Passed": bool(
            globals().get(
                "section_3_4_complete",
                False
            )
        ),
    },
    {
        "Validation Area": "Limitation coverage",
        "Requirement": (
            "The limitation register must document the major "
            "coverage risks identified in Section 3."
        ),
        "Observed Evidence": (
            f"{len(geographic_coverage_limitations_df)} "
            f"limitations documented"
        ),
        "Passed": (
            len(
                geographic_coverage_limitations_df
            ) >= 8
        ),
    },
    {
        "Validation Area": "Country safeguard coverage",
        "Requirement": (
            "Every validated country must receive interpretation safeguards."
        ),
        "Observed Evidence": (
            f"{len(country_interpretation_safeguards_df):,} "
            f"safeguard records for {total_countries:,} countries"
        ),
        "Passed": (
            len(
                country_interpretation_safeguards_df
            ) == total_countries
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Country safeguard tables must not include the global aggregate."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(country_interpretation_safeguards_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_interpretation_safeguards_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Geographic coverage limitation validation")
print("=" * 100)

display(
    section_3_5_validation_df
)

section_3_5_complete = bool(
    section_3_5_validation_df[
        "Passed"
    ].all()
)

if not section_3_5_complete:
    failed_checks = (
        section_3_5_validation_df.loc[
            ~section_3_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 3.5 geographic coverage limitation "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

section_3_overall_complete = bool(
    globals().get("section_3_1_complete", False)
    and globals().get("section_3_2_complete", False)
    and globals().get("section_3_3_complete", False)
    and globals().get("section_3_4_complete", False)
    and section_3_5_complete
)

print()
print(
    "All Section 3.5 geographic coverage limitation "
    "validation checks passed."
)
print(
    f"Section 3.5 completion status: "
    f"{section_3_5_complete}"
)
print(
    f"Section 3 overall completion status: "
    f"{section_3_overall_complete}"
)
print()
print(
    "Geographic coverage limitations and safeguards are now "
    "explicitly documented for later PMIP performance analysis."
)
```

### Interpretation

Section 3.5 consolidates the geographic coverage risks identified throughout Section 3 and converts them into explicit safeguards for later PMIP analysis.

The geographic dataset provides strong overall coverage, but country histories and entity-level representation are not uniform. Observation counts range from relatively sparse markets with approximately **1,300 observations** to highly represented markets with more than **90,000 observations**.

A major limitation is staggered country entry. **48 of the 76 validated country markets** entered the dataset after the overall historical start date, meaning that some markets have had considerably less time to accumulate tracks, artists and observations.

Historical coverage is also uneven. **15 countries** contain less than 50% of the complete reporting-date history, while **12 countries** contain at least one internal reporting gap. These differences mean that raw historical counts should not be compared across markets without considering the available observation period.

Geographic representation at track and artist level is particularly sparse. Approximately **73.58% of tracks** and **69.88% of artist representations** appear in only one country market. Broad international coverage therefore represents a relatively uncommon pattern rather than the normal structure of the dataset.

Country-level quality assessment shows that most markets remain suitable for analysis, although **12 countries** require moderate or high direct-comparison caution. These markets should retain their coverage and quality metadata whenever later performance summaries are produced.

The interpretation safeguards establish that:

- **61 countries** currently have sufficient historical depth for broader historical comparison.
- **15 countries** should be treated as historically limited.
- **75 countries** have sufficient continuity for trend interpretation.
- **1 country** requires particular caution for longitudinal trend analysis.
- **12 countries** require explicit caution during direct market comparison.

These safeguards do not remove countries from PMIP. They instead ensure that later geographic performance indicators are interpreted relative to the amount and quality of available evidence.

The `global` aggregate population remains explicitly separate from the 76 country markets and must not contribute to country-count, market-penetration or international-reach calculations.

Most importantly, geographic coverage must remain conceptually separate from geographic performance. A country, artist or track having more observations does not by itself mean stronger streaming performance, greater popularity or better market success.

All Section 3.5 validation checks passed, and **Section 3 is complete**. The validated coverage summaries, historical diagnostics, quality assessments and interpretation safeguards are ready to support the next stage of PMIP geographic performance analysis.

## 4.1 Validated Analytical Population

### Purpose

This subsection defines the final observation-level population that will be used for PMIP geographic performance analysis.

The analysis verifies that:

- only validated country-market observations are included,
- global aggregate observations remain separate,
- the canonical geographic observation key remains unique,
- required performance and identity fields remain complete,
- Section 3 coverage and quality safeguards remain available,

The resulting analytical population will act as the source for all country-level, artist-country and track-country aggregations in the remainder of Section 4.

```python
import pandas as pd
import numpy as np

print("Preparing validated PMIP geographic analytical population")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

required_objects = [
    "country_geo_df",
    "country_interpretation_safeguards_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required validated objects are missing: "
        + ", ".join(missing_objects)
    )

if not globals().get(
    "section_3_overall_complete",
    False
):
    raise RuntimeError(
        "Section 3 is not fully complete. "
        "Complete Section 3 before running Section 4.1."
    )

print(
    f"Section 3 overall completion status: "
    f"{section_3_overall_complete}"
)

# ------------------------------------------------------------
# 2. Build analytical population
# ------------------------------------------------------------

geographic_analysis_df = (
    country_geo_df
    .copy()
)

print()
print("Validated country-market source")
print("=" * 100)

print(
    f"Source observations: "
    f"{len(geographic_analysis_df):,}"
)

print(
    f"Country markets: "
    f"{geographic_analysis_df['country'].nunique():,}"
)

print(
    f"Reporting dates: "
    f"{geographic_analysis_df['date'].nunique():,}"
)

print(
    f"Unique tracks: "
    f"{geographic_analysis_df['track_id'].nunique():,}"
)

print(
    f"Unique artist representations: "
    f"{geographic_analysis_df['artists_raw'].nunique():,}"
)

# ------------------------------------------------------------
# 3. Attach country safeguards
# ------------------------------------------------------------

safeguard_columns = [
    "country",
    "historical_comparison_allowed",
    "trend_interpretation_allowed",
    "direct_comparison_caution",
]

geographic_analysis_df = (
    geographic_analysis_df
    .merge(
        country_interpretation_safeguards_df[
            safeguard_columns
        ],
        on="country",
        how="left",
        validate="many_to_one"
    )
)

# ------------------------------------------------------------
# 4. Canonical analytical fields
# ------------------------------------------------------------

required_analysis_fields = [
    "date",
    "country",
    "track_id",
    "track_name",
    "artists_raw",
    "primary_artist",
    "streams",
    "position",
    "historical_comparison_allowed",
    "trend_interpretation_allowed",
    "direct_comparison_caution",
]

missing_analysis_fields = [
    column
    for column in required_analysis_fields
    if column not in geographic_analysis_df.columns
]

if missing_analysis_fields:
    raise RuntimeError(
        "Validated geographic analytical population is "
        "missing required fields: "
        + ", ".join(missing_analysis_fields)
    )

# ------------------------------------------------------------
# 5. Observation-key validation
# ------------------------------------------------------------

geographic_analysis_df[
    "observation_key"
] = (
    geographic_analysis_df[
        "date"
    ].astype(str)
    + "|"
    + geographic_analysis_df[
        "country"
    ].astype(str)
    + "|"
    + geographic_analysis_df[
        "track_id"
    ].astype(str)
)

duplicate_key_rows = int(
    geographic_analysis_df[
        "observation_key"
    ].duplicated(
        keep=False
    ).sum()
)

unique_observation_keys = int(
    geographic_analysis_df[
        "observation_key"
    ].nunique()
)

print()
print("Analytical observation-key validation")
print("=" * 100)

print(
    f"Analytical rows: "
    f"{len(geographic_analysis_df):,}"
)

print(
    f"Unique observation keys: "
    f"{unique_observation_keys:,}"
)

print(
    f"Rows involved in duplicate keys: "
    f"{duplicate_key_rows:,}"
)

# ------------------------------------------------------------
# 6. Required-field completeness
# ------------------------------------------------------------

analysis_missingness_df = pd.DataFrame([
    {
        "Field": column,
        "Missing Observations": int(
            geographic_analysis_df[
                column
            ].isna().sum()
        ),
        "Non-Missing Observations": int(
            geographic_analysis_df[
                column
            ].notna().sum()
        ),
    }
    for column in required_analysis_fields
])

print()
print("Analytical population field completeness")
print("=" * 100)

display(
    analysis_missingness_df
)

# ------------------------------------------------------------
# 7. Performance-value diagnostics
# ------------------------------------------------------------

stream_values = pd.to_numeric(
    geographic_analysis_df[
        "streams"
    ],
    errors="coerce"
)

position_values = pd.to_numeric(
    geographic_analysis_df[
        "position"
    ],
    errors="coerce"
)

performance_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Finite stream values",
        "Observed Value": int(
            np.isfinite(
                stream_values
            ).sum()
        ),
    },
    {
        "Diagnostic": "Zero-stream observations",
        "Observed Value": int(
            (
                stream_values == 0
            ).sum()
        ),
    },
    {
        "Diagnostic": "Negative stream observations",
        "Observed Value": int(
            (
                stream_values < 0
            ).sum()
        ),
    },
    {
        "Diagnostic": "Finite chart positions",
        "Observed Value": int(
            np.isfinite(
                position_values
            ).sum()
        ),
    },
    {
        "Diagnostic": "Minimum chart position",
        "Observed Value": float(
            position_values.min()
        ),
    },
    {
        "Diagnostic": "Maximum chart position",
        "Observed Value": float(
            position_values.max()
        ),
    },
])

print()
print("Performance-value diagnostics")
print("=" * 100)

display(
    performance_diagnostics_df
)

# ------------------------------------------------------------
# 8. Country safeguard population summary
# ------------------------------------------------------------

country_safeguard_population_df = (
    geographic_analysis_df[
        [
            "country",
            "historical_comparison_allowed",
            "trend_interpretation_allowed",
            "direct_comparison_caution",
        ]
    ]
    .drop_duplicates(
        subset="country"
    )
    .reset_index(drop=True)
)

safeguard_population_summary_df = pd.DataFrame([
    {
        "Safeguard": "Historical comparison supported",
        "Countries": int(
            country_safeguard_population_df[
                "historical_comparison_allowed"
            ].sum()
        ),
    },
    {
        "Safeguard": "Historical comparison limited",
        "Countries": int(
            (
                ~country_safeguard_population_df[
                    "historical_comparison_allowed"
                ]
            ).sum()
        ),
    },
    {
        "Safeguard": "Trend interpretation supported",
        "Countries": int(
            country_safeguard_population_df[
                "trend_interpretation_allowed"
            ].sum()
        ),
    },
    {
        "Safeguard": "Trend interpretation limited",
        "Countries": int(
            (
                ~country_safeguard_population_df[
                    "trend_interpretation_allowed"
                ]
            ).sum()
        ),
    },
    {
        "Safeguard": "Direct comparison caution required",
        "Countries": int(
            country_safeguard_population_df[
                "direct_comparison_caution"
            ].sum()
        ),
    },
])

print()
print("Preserved geographic interpretation safeguards")
print("=" * 100)

display(
    safeguard_population_summary_df
)

# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

total_missing_required = int(
    analysis_missingness_df[
        "Missing Observations"
    ].sum()
)

global_present = (
    "global"
    in set(
        geographic_analysis_df[
            "country"
        ].astype(str)
    )
)

section_4_1_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 3 completion",
        "Requirement": (
            "Geographic coverage validation must be complete "
            "before performance preparation."
        ),
        "Observed Evidence": (
            f"Section 3 overall completion status: "
            f"{section_3_overall_complete}"
        ),
        "Passed": bool(
            section_3_overall_complete
        ),
    },
    {
        "Validation Area": "Analytical population preservation",
        "Requirement": (
            "Section 4.1 must preserve every validated "
            "country-market observation."
        ),
        "Observed Evidence": (
            f"{len(geographic_analysis_df):,} analytical rows "
            f"from {len(country_geo_df):,} validated source rows"
        ),
        "Passed": (
            len(
                geographic_analysis_df
            )
            == len(
                country_geo_df
            )
        ),
    },
    {
        "Validation Area": "Observation-key uniqueness",
        "Requirement": (
            "date + country + track_id must remain unique."
        ),
        "Observed Evidence": (
            f"{duplicate_key_rows:,} rows involved in duplicate keys"
        ),
        "Passed": (
            duplicate_key_rows == 0
        ),
    },
    {
        "Validation Area": "Required-field completeness",
        "Requirement": (
            "All performance and identity fields must remain available."
        ),
        "Observed Evidence": (
            f"{total_missing_required:,} missing values "
            f"across required analytical fields"
        ),
        "Passed": (
            total_missing_required == 0
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "The global aggregate must remain outside "
            "the country performance population."
        ),
        "Observed Evidence": (
            f"'global' present in analytical population: "
            f"{global_present}"
        ),
        "Passed": (
            not global_present
        ),
    },
    {
        "Validation Area": "Stream validity",
        "Requirement": (
            "Stream values must remain finite and non-negative."
        ),
        "Observed Evidence": (
            f"{int((~np.isfinite(stream_values)).sum()):,} non-finite; "
            f"{int((stream_values < 0).sum()):,} negative"
        ),
        "Passed": bool(
            np.isfinite(
                stream_values
            ).all()
            and
            (
                stream_values >= 0
            ).all()
        ),
    },
    {
        "Validation Area": "Chart-position validity",
        "Requirement": (
            "Chart positions must remain finite and positive."
        ),
        "Observed Evidence": (
            f"Observed position range: "
            f"{int(position_values.min())} to "
            f"{int(position_values.max())}"
        ),
        "Passed": bool(
            np.isfinite(
                position_values
            ).all()
            and
            (
                position_values > 0
            ).all()
        ),
    },
    {
        "Validation Area": "Country safeguard preservation",
        "Requirement": (
            "Every country must retain Section 3 interpretation safeguards."
        ),
        "Observed Evidence": (
            f"{len(country_safeguard_population_df):,} "
            f"country safeguard records"
        ),
        "Passed": (
            len(
                country_safeguard_population_df
            )
            == geographic_analysis_df[
                "country"
            ].nunique()
        ),
    },
])

print()
print("Validated analytical population checks")
print("=" * 100)

display(
    section_4_1_validation_df
)

section_4_1_complete = bool(
    section_4_1_validation_df[
        "Passed"
    ].all()
)

if not section_4_1_complete:
    failed_checks = (
        section_4_1_validation_df.loc[
            ~section_4_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 4.1 validated analytical population "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 4.1 validated analytical population "
    "checks passed."
)
print(
    f"Section 4.1 completion status: "
    f"{section_4_1_complete}"
)
print(
    f"Validated geographic analytical population prepared: "
    f"{len(geographic_analysis_df):,} observations."
)
print(
    "No country observations were excluded solely because "
    "of limited historical coverage."
)
print(
    "The analytical population is ready for "
    "Section 4.2 Country-Level Aggregation."
)
```

## 4.2 Country-Level Aggregation

### Purpose

This subsection aggregates the validated observation-level geographic population into one analytical summary for each country market.

The objective is to create the country-level performance foundation required for later PMIP market comparisons and geographic intelligence indicators.

The aggregation measures:

- total streaming volume,
- mean and median streaming performance,
- chart-position performance,
- number of unique tracks,
- number of unique artists,
- reporting-date coverage,
- number of chart observations,
- top-chart frequency,
- country-level observation support,
- retained interpretation safeguards.

These summaries describe observed country-level performance and support. They do not yet represent a final PMIP geographic performance ranking or score.

```python
import pandas as pd
import numpy as np

print("Preparing PMIP country-level performance aggregation")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

if "geographic_analysis_df" not in globals():
    raise RuntimeError(
        "geographic_analysis_df is not available. "
        "Run Section 4.1 before Section 4.2."
    )

if not globals().get(
    "section_4_1_complete",
    False
):
    raise RuntimeError(
        "Section 4.1 is not complete. "
        "Run and validate Section 4.1 before Section 4.2."
    )

source_df = geographic_analysis_df.copy()

print(
    f"Section 4.1 completion status: "
    f"{section_4_1_complete}"
)

# ------------------------------------------------------------
# 2. Country-level aggregation
# ------------------------------------------------------------

country_performance_df = (
    source_df
    .groupby(
        "country",
        dropna=False
    )
    .agg(
        observations=("track_id", "size"),
        reporting_dates=("date", "nunique"),
        unique_tracks=("track_id", "nunique"),
        unique_artist_representations=("artists_raw", "nunique"),
        unique_primary_artists=("primary_artist", "nunique"),
        total_streams=("streams", "sum"),
        mean_streams=("streams", "mean"),
        median_streams=("streams", "median"),
        maximum_streams=("streams", "max"),
        minimum_chart_position=("position", "min"),
        mean_chart_position=("position", "mean"),
        median_chart_position=("position", "median"),
        maximum_chart_position=("position", "max"),
        historical_comparison_allowed=(
            "historical_comparison_allowed",
            "first"
        ),
        trend_interpretation_allowed=(
            "trend_interpretation_allowed",
            "first"
        ),
        direct_comparison_caution=(
            "direct_comparison_caution",
            "first"
        ),
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. Additional country performance indicators
# ------------------------------------------------------------

country_performance_df[
    "top_10_observations"
] = (
    source_df[
        source_df["position"] <= 10
    ]
    .groupby("country")
    .size()
    .reindex(
        country_performance_df["country"],
        fill_value=0
    )
    .to_numpy()
)

country_performance_df[
    "top_50_observations"
] = (
    source_df[
        source_df["position"] <= 50
    ]
    .groupby("country")
    .size()
    .reindex(
        country_performance_df["country"],
        fill_value=0
    )
    .to_numpy()
)

country_performance_df[
    "number_one_observations"
] = (
    source_df[
        source_df["position"] == 1
    ]
    .groupby("country")
    .size()
    .reindex(
        country_performance_df["country"],
        fill_value=0
    )
    .to_numpy()
)

country_performance_df[
    "top_10_rate_pct"
] = (
    country_performance_df[
        "top_10_observations"
    ]
    / country_performance_df[
        "observations"
    ]
    * 100
)

country_performance_df[
    "top_50_rate_pct"
] = (
    country_performance_df[
        "top_50_observations"
    ]
    / country_performance_df[
        "observations"
    ]
    * 100
)

country_performance_df[
    "number_one_rate_pct"
] = (
    country_performance_df[
        "number_one_observations"
    ]
    / country_performance_df[
        "observations"
    ]
    * 100
)

country_performance_df[
    "streams_per_track"
] = (
    country_performance_df[
        "total_streams"
    ]
    / country_performance_df[
        "unique_tracks"
    ]
)

country_performance_df[
    "streams_per_artist_representation"
] = (
    country_performance_df[
        "total_streams"
    ]
    / country_performance_df[
        "unique_artist_representations"
    ]
)

# ------------------------------------------------------------
# 4. Coverage-adjusted fields
# ------------------------------------------------------------

if "country_historical_coverage_df" in globals():
    country_performance_df = (
        country_performance_df
        .merge(
            country_historical_coverage_df[
                [
                    "country",
                    "first_reporting_date",
                    "last_reporting_date",
                    "date_coverage_pct",
                    "active_span_coverage_pct",
                    "historical_coverage_class",
                ]
            ],
            on="country",
            how="left",
            validate="one_to_one"
        )
    )

if "country_quality_df" in globals():
    country_performance_df = (
        country_performance_df
        .merge(
            country_quality_df[
                [
                    "country",
                    "quality_review_priority",
                    "quality_risk_count",
                ]
            ],
            on="country",
            how="left",
            validate="one_to_one"
        )
    )

# ------------------------------------------------------------
# 5. Country-level overview
# ------------------------------------------------------------

print()
print("Country-level performance overview")
print("=" * 100)

print(
    f"Country summaries prepared: "
    f"{len(country_performance_df):,}"
)

print(
    f"Total source observations represented: "
    f"{int(country_performance_df['observations'].sum()):,}"
)

print(
    f"Total streams represented: "
    f"{country_performance_df['total_streams'].sum():,.0f}"
)

print(
    f"Median country observations: "
    f"{country_performance_df['observations'].median():,.1f}"
)

print(
    f"Median country reporting dates: "
    f"{country_performance_df['reporting_dates'].median():,.1f}"
)

# ------------------------------------------------------------
# 6. Highest total-stream markets
# ------------------------------------------------------------

top_total_stream_markets_df = (
    country_performance_df
    .sort_values(
        "total_streams",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Country markets with highest total observed streams")
print("=" * 100)

display(
    top_total_stream_markets_df[
        [
            "country",
            "total_streams",
            "observations",
            "reporting_dates",
            "unique_tracks",
            "unique_artist_representations",
            "median_streams",
            "median_chart_position",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 7. Strongest median-stream markets
# ------------------------------------------------------------

top_median_stream_markets_df = (
    country_performance_df
    .sort_values(
        "median_streams",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Country markets with highest median observed streams")
print("=" * 100)

display(
    top_median_stream_markets_df[
        [
            "country",
            "median_streams",
            "mean_streams",
            "total_streams",
            "observations",
            "unique_tracks",
            "date_coverage_pct",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 8. Strongest chart-position markets
# ------------------------------------------------------------

top_chart_markets_df = (
    country_performance_df
    .sort_values(
        [
            "median_chart_position",
            "top_10_rate_pct",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Country markets with strongest observed chart-position profile")
print("=" * 100)

display(
    top_chart_markets_df[
        [
            "country",
            "median_chart_position",
            "mean_chart_position",
            "top_10_rate_pct",
            "top_50_rate_pct",
            "number_one_rate_pct",
            "observations",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 9. Performance support diagnostics
# ------------------------------------------------------------

country_performance_support_df = pd.DataFrame([
    {
        "Diagnostic": "Countries represented",
        "Observed Value": len(
            country_performance_df
        ),
    },
    {
        "Diagnostic": "Countries with >= 50% historical date coverage",
        "Observed Value": int(
            (
                country_performance_df[
                    "date_coverage_pct"
                ] >= 50
            ).sum()
        ),
    },
    {
        "Diagnostic": "Countries requiring direct comparison caution",
        "Observed Value": int(
            country_performance_df[
                "direct_comparison_caution"
            ].sum()
        ),
    },
    {
        "Diagnostic": "Countries with high or moderate quality review",
        "Observed Value": int(
            country_performance_df[
                "quality_review_priority"
            ].isin(
                [
                    "High review priority",
                    "Moderate review priority",
                ]
            ).sum()
        ),
    },
])

print()
print("Country performance support diagnostics")
print("=" * 100)

display(
    country_performance_support_df
)

# ------------------------------------------------------------
# 10. Validation
# ------------------------------------------------------------

numeric_validation_columns = [
    "observations",
    "reporting_dates",
    "unique_tracks",
    "unique_artist_representations",
    "total_streams",
    "mean_streams",
    "median_streams",
    "minimum_chart_position",
    "mean_chart_position",
    "median_chart_position",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "streams_per_track",
    "streams_per_artist_representation",
]

finite_country_metrics = bool(
    np.isfinite(
        country_performance_df[
            numeric_validation_columns
        ].to_numpy(
            dtype=float
        )
    ).all()
)

section_4_2_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 4.1 completion",
        "Requirement": (
            "Validated analytical population must be complete."
        ),
        "Observed Evidence": (
            f"Section 4.1 completion status: "
            f"{section_4_1_complete}"
        ),
        "Passed": bool(
            section_4_1_complete
        ),
    },
    {
        "Validation Area": "Country summary completeness",
        "Requirement": (
            "Every validated country market must receive "
            "one country performance summary."
        ),
        "Observed Evidence": (
            f"{len(country_performance_df):,} summaries "
            f"for {source_df['country'].nunique():,} countries"
        ),
        "Passed": (
            len(
                country_performance_df
            )
            == source_df[
                "country"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Observation preservation",
        "Requirement": (
            "Country aggregation must preserve all source "
            "observations through grouped counts."
        ),
        "Observed Evidence": (
            f"{int(country_performance_df['observations'].sum()):,} "
            f"grouped observations from {len(source_df):,} source rows"
        ),
        "Passed": (
            int(
                country_performance_df[
                    "observations"
                ].sum()
            )
            == len(
                source_df
            )
        ),
    },
    {
        "Validation Area": "Stream preservation",
        "Requirement": (
            "Aggregated total streams must reconcile with "
            "the validated source population."
        ),
        "Observed Evidence": (
            f"Grouped streams: "
            f"{country_performance_df['total_streams'].sum():,.0f}; "
            f"source streams: {source_df['streams'].sum():,.0f}"
        ),
        "Passed": bool(
            np.isclose(
                country_performance_df[
                    "total_streams"
                ].sum(),
                source_df[
                    "streams"
                ].sum(),
            )
        ),
    },
    {
        "Validation Area": "Finite country metrics",
        "Requirement": (
            "Country-level performance metrics must be finite."
        ),
        "Observed Evidence": (
            f"{len(numeric_validation_columns)} "
            f"numeric metric fields assessed"
        ),
        "Passed": finite_country_metrics,
    },
    {
        "Validation Area": "Safeguard preservation",
        "Requirement": (
            "Country-level summaries must retain Section 3 "
            "interpretation safeguards."
        ),
        "Observed Evidence": (
            f"{country_performance_df['direct_comparison_caution'].notna().sum():,} "
            f"country safeguard values retained"
        ),
        "Passed": bool(
            country_performance_df[
                [
                    "historical_comparison_allowed",
                    "trend_interpretation_allowed",
                    "direct_comparison_caution",
                ]
            ]
            .notna()
            .all()
            .all()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Country aggregation must exclude the global aggregate."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(country_performance_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_performance_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Country-level aggregation validation")
print("=" * 100)

display(
    section_4_2_validation_df
)

section_4_2_complete = bool(
    section_4_2_validation_df[
        "Passed"
    ].all()
)

if not section_4_2_complete:
    failed_checks = (
        section_4_2_validation_df.loc[
            ~section_4_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 4.2 country-level aggregation "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 4.2 country-level aggregation "
    "validation checks passed."
)
print(
    f"Section 4.2 completion status: "
    f"{section_4_2_complete}"
)
print(
    f"Country-level analytical table prepared: "
    f"{len(country_performance_df):,} rows."
)
print(
    "Country metrics remain descriptive preparation outputs "
    "and are not yet a final PMIP geographic performance ranking."
)
print(
    "The workflow is ready for "
    "Section 4.3 Artist-Country Aggregation."
)
```

### Interpretation

Section 4.2 successfully converts the validated observation-level geographic population into **76 country-level performance summaries**.

The country summaries now provide a consistent performance foundation containing:

The United States has the largest total observed streaming volume and the highest median observed streams among the strongest represented markets. Other markets such as Brazil, Mexico, Germany and the United Kingdom also show high aggregate streaming volumes.

However, raw country rankings must not yet be interpreted as final PMIP geographic performance rankings.

Some markets with apparently strong chart-position or streaming profiles have weaker historical or observation support. For example, markets such as `ad`, `pk`, `ve`, `in` and `ru` appear strongly on selected metrics but also carry comparison cautions or shorter historical coverage.

All Section 4.2 validation checks passed. The country-level analytical table is therefore ready to support later geographic intelligence calculations, but it should still be treated as a descriptive preparation layer rather than a final market ranking.

## 4.3 Artist-Country Aggregation

### Purpose

This subsection creates a validated analytical summary for each observed artist-country combination.

The objective is to measure how individual artists perform within specific country markets while preserving the geographic coverage safeguards established earlier in the workflow.

The aggregation will capture:

- artist-country observation counts,
- number of unique tracks,
- reporting-date support,
- total, mean and median streaming performance,
- chart-position performance,
- Top 10 and Top 50 frequency,
- number-one chart frequency,
- first and last observed reporting dates,
- country-level interpretation safeguards.

These outputs will form the foundation for later analysis of local strength, international reach and artist-specific geographic performance.

```python
import pandas as pd
import numpy as np

print("Preparing PMIP artist-country performance aggregation")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

if "geographic_analysis_df" not in globals():
    raise RuntimeError(
        "geographic_analysis_df is not available. "
        "Run Section 4.1 before Section 4.3."
    )

if not globals().get(
    "section_4_2_complete",
    False
):
    raise RuntimeError(
        "Section 4.2 is not complete. "
        "Run and validate Section 4.2 before Section 4.3."
    )

source_df = geographic_analysis_df.copy()

print(
    f"Section 4.2 completion status: "
    f"{section_4_2_complete}"
)

# ------------------------------------------------------------
# 2. Artist-country aggregation
# ------------------------------------------------------------

artist_country_performance_df = (
    source_df
    .groupby(
        [
            "artists_raw",
            "country",
        ],
        dropna=False
    )
    .agg(
        observations=("track_id", "size"),
        reporting_dates=("date", "nunique"),
        unique_tracks=("track_id", "nunique"),
        total_streams=("streams", "sum"),
        mean_streams=("streams", "mean"),
        median_streams=("streams", "median"),
        maximum_streams=("streams", "max"),
        minimum_chart_position=("position", "min"),
        mean_chart_position=("position", "mean"),
        median_chart_position=("position", "median"),
        maximum_chart_position=("position", "max"),
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        historical_comparison_allowed=(
            "historical_comparison_allowed",
            "first"
        ),
        trend_interpretation_allowed=(
            "trend_interpretation_allowed",
            "first"
        ),
        direct_comparison_caution=(
            "direct_comparison_caution",
            "first"
        ),
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. Additional artist-country chart indicators
# ------------------------------------------------------------

artist_country_key = [
    "artists_raw",
    "country",
]

top_10_artist_country = (
    source_df.loc[
        source_df["position"] <= 10
    ]
    .groupby(
        artist_country_key
    )
    .size()
)

top_50_artist_country = (
    source_df.loc[
        source_df["position"] <= 50
    ]
    .groupby(
        artist_country_key
    )
    .size()
)

number_one_artist_country = (
    source_df.loc[
        source_df["position"] == 1
    ]
    .groupby(
        artist_country_key
    )
    .size()
)

artist_country_index = pd.MultiIndex.from_frame(
    artist_country_performance_df[
        artist_country_key
    ]
)

artist_country_performance_df[
    "top_10_observations"
] = (
    top_10_artist_country
    .reindex(
        artist_country_index,
        fill_value=0
    )
    .to_numpy()
)

artist_country_performance_df[
    "top_50_observations"
] = (
    top_50_artist_country
    .reindex(
        artist_country_index,
        fill_value=0
    )
    .to_numpy()
)

artist_country_performance_df[
    "number_one_observations"
] = (
    number_one_artist_country
    .reindex(
        artist_country_index,
        fill_value=0
    )
    .to_numpy()
)

# ------------------------------------------------------------
# 4. Artist-country rates and support measures
# ------------------------------------------------------------

artist_country_performance_df[
    "top_10_rate_pct"
] = (
    artist_country_performance_df[
        "top_10_observations"
    ]
    / artist_country_performance_df[
        "observations"
    ]
    * 100
)

artist_country_performance_df[
    "top_50_rate_pct"
] = (
    artist_country_performance_df[
        "top_50_observations"
    ]
    / artist_country_performance_df[
        "observations"
    ]
    * 100
)

artist_country_performance_df[
    "number_one_rate_pct"
] = (
    artist_country_performance_df[
        "number_one_observations"
    ]
    / artist_country_performance_df[
        "observations"
    ]
    * 100
)

artist_country_performance_df[
    "streams_per_track"
] = (
    artist_country_performance_df[
        "total_streams"
    ]
    / artist_country_performance_df[
        "unique_tracks"
    ]
)

artist_country_performance_df[
    "observation_span_days"
] = (
    artist_country_performance_df[
        "last_reporting_date"
    ]
    - artist_country_performance_df[
        "first_reporting_date"
    ]
).dt.days

# ------------------------------------------------------------
# 5. Artist-country support classification
# ------------------------------------------------------------

artist_country_performance_df[
    "support_classification"
] = np.select(
    [
        artist_country_performance_df[
            "observations"
        ] >= 100,

        artist_country_performance_df[
            "observations"
        ] >= 25,

        artist_country_performance_df[
            "observations"
        ] >= 5,
    ],
    [
        "Strong support",
        "Moderate support",
        "Limited support",
    ],
    default="Very limited support"
)

# ------------------------------------------------------------
# 6. Artist geographic breadth within aggregated table
# ------------------------------------------------------------

artist_market_counts = (
    artist_country_performance_df
    .groupby(
        "artists_raw"
    )["country"]
    .nunique()
)

artist_country_performance_df[
    "artist_country_markets"
] = (
    artist_country_performance_df[
        "artists_raw"
    ]
    .map(
        artist_market_counts
    )
)

artist_country_performance_df[
    "artist_market_coverage_pct"
] = (
    artist_country_performance_df[
        "artist_country_markets"
    ]
    / source_df[
        "country"
    ].nunique()
    * 100
)

# ------------------------------------------------------------
# 7. Country quality metadata
# ------------------------------------------------------------

if "country_quality_df" in globals():

    artist_country_performance_df = (
        artist_country_performance_df
        .merge(
            country_quality_df[
                [
                    "country",
                    "quality_review_priority",
                    "quality_risk_count",
                ]
            ],
            on="country",
            how="left",
            validate="many_to_one"
        )
    )

if "country_historical_coverage_df" in globals():

    artist_country_performance_df = (
        artist_country_performance_df
        .merge(
            country_historical_coverage_df[
                [
                    "country",
                    "date_coverage_pct",
                    "active_span_coverage_pct",
                    "historical_coverage_class",
                ]
            ],
            on="country",
            how="left",
            validate="many_to_one"
        )
    )

# ------------------------------------------------------------
# 8. Artist-country overview
# ------------------------------------------------------------

print()
print("Artist-country performance overview")
print("=" * 100)

print(
    f"Artist-country summaries prepared: "
    f"{len(artist_country_performance_df):,}"
)

print(
    f"Unique artist representations: "
    f"{artist_country_performance_df['artists_raw'].nunique():,}"
)

print(
    f"Country markets represented: "
    f"{artist_country_performance_df['country'].nunique():,}"
)

print(
    f"Total grouped observations represented: "
    f"{int(artist_country_performance_df['observations'].sum()):,}"
)

print(
    f"Total grouped streams represented: "
    f"{artist_country_performance_df['total_streams'].sum():,.0f}"
)

print(
    f"Median observations per artist-country pair: "
    f"{artist_country_performance_df['observations'].median():,.1f}"
)

# ------------------------------------------------------------
# 9. Strongest artist-country combinations by total streams
# ------------------------------------------------------------

top_artist_country_total_streams_df = (
    artist_country_performance_df
    .sort_values(
        "total_streams",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Artist-country combinations with highest total observed streams"
)
print("=" * 100)

display(
    top_artist_country_total_streams_df[
        [
            "artists_raw",
            "country",
            "total_streams",
            "observations",
            "reporting_dates",
            "unique_tracks",
            "median_streams",
            "median_chart_position",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 10. Strongest median-stream artist-country combinations
# ------------------------------------------------------------

# Require some minimum observational support so that one-off
# artist-country observations do not dominate the display.
supported_artist_country_df = (
    artist_country_performance_df.loc[
        artist_country_performance_df[
            "observations"
        ] >= 25
    ]
    .copy()
)

top_artist_country_median_streams_df = (
    supported_artist_country_df
    .sort_values(
        [
            "median_streams",
            "observations",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Supported artist-country combinations with highest median streams"
)
print("=" * 100)

display(
    top_artist_country_median_streams_df[
        [
            "artists_raw",
            "country",
            "median_streams",
            "mean_streams",
            "total_streams",
            "observations",
            "unique_tracks",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 11. Strongest chart-profile artist-country combinations
# ------------------------------------------------------------

top_artist_country_chart_df = (
    supported_artist_country_df
    .sort_values(
        [
            "median_chart_position",
            "top_10_rate_pct",
            "observations",
        ],
        ascending=[
            True,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Supported artist-country combinations with strongest chart profiles"
)
print("=" * 100)

display(
    top_artist_country_chart_df[
        [
            "artists_raw",
            "country",
            "median_chart_position",
            "mean_chart_position",
            "top_10_rate_pct",
            "top_50_rate_pct",
            "number_one_rate_pct",
            "observations",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Artist-country support distribution
# ------------------------------------------------------------

artist_country_support_summary_df = (
    artist_country_performance_df[
        "support_classification"
    ]
    .value_counts()
    .rename_axis(
        "Support Classification"
    )
    .reset_index(
        name="Artist-Country Pairs"
    )
)

artist_country_support_summary_df[
    "Share (%)"
] = (
    artist_country_support_summary_df[
        "Artist-Country Pairs"
    ]
    / len(
        artist_country_performance_df
    )
    * 100
)

print()
print("Artist-country analytical support distribution")
print("=" * 100)

display(
    artist_country_support_summary_df
)

# ------------------------------------------------------------
# 13. Artist geographic breadth diagnostics
# ------------------------------------------------------------

artist_breadth_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Unique artist representations",
        "Observed Value": int(
            artist_country_performance_df[
                "artists_raw"
            ].nunique()
        ),
    },
    {
        "Diagnostic": "Artist-country combinations",
        "Observed Value": int(
            len(
                artist_country_performance_df
            )
        ),
    },
    {
        "Diagnostic": "Artists appearing in one country",
        "Observed Value": int(
            (
                artist_market_counts == 1
            ).sum()
        ),
    },
    {
        "Diagnostic": "Artists appearing in at least 5 countries",
        "Observed Value": int(
            (
                artist_market_counts >= 5
            ).sum()
        ),
    },
    {
        "Diagnostic": "Artists appearing in at least 10 countries",
        "Observed Value": int(
            (
                artist_market_counts >= 10
            ).sum()
        ),
    },
    {
        "Diagnostic": "Artists appearing in at least 25 countries",
        "Observed Value": int(
            (
                artist_market_counts >= 25
            ).sum()
        ),
    },
    {
        "Diagnostic": "Artists appearing in at least 50 countries",
        "Observed Value": int(
            (
                artist_market_counts >= 50
            ).sum()
        ),
    },
    {
        "Diagnostic": "Maximum countries for one artist representation",
        "Observed Value": int(
            artist_market_counts.max()
        ),
    },
])

print()
print("Artist geographic breadth diagnostics")
print("=" * 100)

display(
    artist_breadth_diagnostics_df
)

# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

numeric_validation_columns = [
    "observations",
    "reporting_dates",
    "unique_tracks",
    "total_streams",
    "mean_streams",
    "median_streams",
    "maximum_streams",
    "minimum_chart_position",
    "mean_chart_position",
    "median_chart_position",
    "maximum_chart_position",
    "top_10_observations",
    "top_50_observations",
    "number_one_observations",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "streams_per_track",
    "artist_country_markets",
    "artist_market_coverage_pct",
]

finite_artist_country_metrics = bool(
    np.isfinite(
        artist_country_performance_df[
            numeric_validation_columns
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

expected_artist_country_pairs = int(
    source_df[
        [
            "artists_raw",
            "country",
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

section_4_3_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 4.2 completion",
        "Requirement": (
            "Country-level aggregation must be complete "
            "before artist-country preparation."
        ),
        "Observed Evidence": (
            f"Section 4.2 completion status: "
            f"{section_4_2_complete}"
        ),
        "Passed": bool(
            section_4_2_complete
        ),
    },
    {
        "Validation Area": "Artist-country pair completeness",
        "Requirement": (
            "Every observed artist-country combination "
            "must receive one summary."
        ),
        "Observed Evidence": (
            f"{len(artist_country_performance_df):,} summaries "
            f"for {expected_artist_country_pairs:,} observed pairs"
        ),
        "Passed": (
            len(
                artist_country_performance_df
            )
            == expected_artist_country_pairs
        ),
    },
    {
        "Validation Area": "Observation preservation",
        "Requirement": (
            "Artist-country grouped observation counts must "
            "reconcile with the validated source population."
        ),
        "Observed Evidence": (
            f"{int(artist_country_performance_df['observations'].sum()):,} "
            f"grouped observations from {len(source_df):,} source rows"
        ),
        "Passed": (
            int(
                artist_country_performance_df[
                    "observations"
                ].sum()
            )
            == len(
                source_df
            )
        ),
    },
    {
        "Validation Area": "Stream preservation",
        "Requirement": (
            "Artist-country total streams must reconcile "
            "with the validated source population."
        ),
        "Observed Evidence": (
            f"Grouped streams: "
            f"{artist_country_performance_df['total_streams'].sum():,.0f}; "
            f"source streams: {source_df['streams'].sum():,.0f}"
        ),
        "Passed": bool(
            np.isclose(
                artist_country_performance_df[
                    "total_streams"
                ].sum(),
                source_df[
                    "streams"
                ].sum(),
            )
        ),
    },
    {
        "Validation Area": "Finite artist-country metrics",
        "Requirement": (
            "Artist-country numerical metrics must remain finite."
        ),
        "Observed Evidence": (
            f"{len(numeric_validation_columns)} "
            f"numeric metric fields assessed"
        ),
        "Passed": finite_artist_country_metrics,
    },
    {
        "Validation Area": "Country safeguard preservation",
        "Requirement": (
            "Artist-country summaries must retain the "
            "validated country interpretation safeguards."
        ),
        "Observed Evidence": (
            f"{artist_country_performance_df['direct_comparison_caution'].notna().sum():,} "
            f"safeguard values retained"
        ),
        "Passed": bool(
            artist_country_performance_df[
                [
                    "historical_comparison_allowed",
                    "trend_interpretation_allowed",
                    "direct_comparison_caution",
                ]
            ]
            .notna()
            .all()
            .all()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Artist-country aggregation must exclude "
            "the global aggregate."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(artist_country_performance_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                artist_country_performance_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Artist-country aggregation validation")
print("=" * 100)

display(
    section_4_3_validation_df
)

section_4_3_complete = bool(
    section_4_3_validation_df[
        "Passed"
    ].all()
)

if not section_4_3_complete:
    failed_checks = (
        section_4_3_validation_df.loc[
            ~section_4_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 4.3 artist-country aggregation "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 4.3 artist-country aggregation "
    "validation checks passed."
)
print(
    f"Section 4.3 completion status: "
    f"{section_4_3_complete}"
)
print(
    f"Artist-country analytical table prepared: "
    f"{len(artist_country_performance_df):,} rows."
)
print(
    "Artist-country metrics remain descriptive preparation outputs "
    "and are not yet final PMIP geographic performance scores."
)
print(
    "The workflow is ready for "
    "Section 4.4 Track-Country Aggregation."
)
```

### Interpretation

Section 4.3 successfully creates the artist-country analytical layer required for PMIP geographic intelligence.

A total of **184,681 artist-country combinations** were prepared across:

- **49,843 unique artist representations**
- **76 country markets**
- **5,338,680 validated source observations**

The grouped artist-country population preserves the full validated streaming volume of **1,031,736,702,723 streams**, confirming that the aggregation does not lose or duplicate source performance information.

The median artist-country combination contains **8 observations**. This shows that artist-level geographic evidence is highly uneven: some artist-market relationships are represented by substantial historical data, while many others are supported by relatively few observations.

For this reason, the artist-country analytical table retains explicit support classifications. Artist-country combinations with stronger observation histories can support more confident comparison, while combinations with limited or very limited support should be interpreted cautiously.

The aggregation also preserves:

- streaming magnitude,
- chart-position performance,
- Top 10 and Top 50 frequencies,
- number-one chart appearances,
- reporting-date coverage,
- first and last observed dates,
- artist geographic breadth,
- country-level historical safeguards,
- country-level quality-review indicators.

The artist-country table therefore provides the structure required to investigate where particular artists perform most strongly and how broadly their observed performance extends across geographic markets.

However, these results remain descriptive preparation outputs. A high total-stream value, strong median chart position or broad observed market presence does not yet represent a final PMIP geographic performance score.

All Section 4.3 validation checks passed, and the artist-country analytical population is ready for track-country preparation.

## 4.4 Track-Country Aggregation

### Purpose

This subsection creates one validated analytical summary for every observed track-country combination.

The objective is to measure how individual tracks perform within specific country markets while retaining the geographic coverage and interpretation safeguards established earlier in the workflow.

The aggregation will capture:

- track-country observation support,
- reporting-date coverage,
- total, mean and median streaming performance,
- chart-position performance,
- Top 10 and Top 50 frequency,
- number-one chart frequency,
- first and last observed reporting dates,
- track geographic breadth,
- country-level comparison safeguards.

These outputs will later support analysis of local track strength, international reach, market penetration and geographic expansion.

```python
import pandas as pd
import numpy as np

print("Preparing PMIP track-country performance aggregation")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

if "geographic_analysis_df" not in globals():
    raise RuntimeError(
        "geographic_analysis_df is not available. "
        "Run Section 4.1 before Section 4.4."
    )

if not globals().get(
    "section_4_3_complete",
    False
):
    raise RuntimeError(
        "Section 4.3 is not complete. "
        "Run and validate Section 4.3 before Section 4.4."
    )

source_df = geographic_analysis_df.copy()

print(
    f"Section 4.3 completion status: "
    f"{section_4_3_complete}"
)

# ------------------------------------------------------------
# 2. Track-country aggregation
# ------------------------------------------------------------

track_country_performance_df = (
    source_df
    .groupby(
        [
            "track_id",
            "country",
        ],
        dropna=False
    )
    .agg(
        track_name=("track_name", "first"),
        artists_raw=("artists_raw", "first"),
        primary_artist=("primary_artist", "first"),
        observations=("track_id", "size"),
        reporting_dates=("date", "nunique"),
        total_streams=("streams", "sum"),
        mean_streams=("streams", "mean"),
        median_streams=("streams", "median"),
        maximum_streams=("streams", "max"),
        minimum_streams=("streams", "min"),
        minimum_chart_position=("position", "min"),
        mean_chart_position=("position", "mean"),
        median_chart_position=("position", "median"),
        maximum_chart_position=("position", "max"),
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        historical_comparison_allowed=(
            "historical_comparison_allowed",
            "first"
        ),
        trend_interpretation_allowed=(
            "trend_interpretation_allowed",
            "first"
        ),
        direct_comparison_caution=(
            "direct_comparison_caution",
            "first"
        ),
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. Additional chart indicators
# ------------------------------------------------------------

track_country_key = [
    "track_id",
    "country",
]

top_10_track_country = (
    source_df.loc[
        source_df["position"] <= 10
    ]
    .groupby(
        track_country_key
    )
    .size()
)

top_50_track_country = (
    source_df.loc[
        source_df["position"] <= 50
    ]
    .groupby(
        track_country_key
    )
    .size()
)

number_one_track_country = (
    source_df.loc[
        source_df["position"] == 1
    ]
    .groupby(
        track_country_key
    )
    .size()
)

track_country_index = pd.MultiIndex.from_frame(
    track_country_performance_df[
        track_country_key
    ]
)

track_country_performance_df[
    "top_10_observations"
] = (
    top_10_track_country
    .reindex(
        track_country_index,
        fill_value=0
    )
    .to_numpy()
)

track_country_performance_df[
    "top_50_observations"
] = (
    top_50_track_country
    .reindex(
        track_country_index,
        fill_value=0
    )
    .to_numpy()
)

track_country_performance_df[
    "number_one_observations"
] = (
    number_one_track_country
    .reindex(
        track_country_index,
        fill_value=0
    )
    .to_numpy()
)

# ------------------------------------------------------------
# 4. Track-country rates
# ------------------------------------------------------------

track_country_performance_df[
    "top_10_rate_pct"
] = (
    track_country_performance_df[
        "top_10_observations"
    ]
    / track_country_performance_df[
        "observations"
    ]
    * 100
)

track_country_performance_df[
    "top_50_rate_pct"
] = (
    track_country_performance_df[
        "top_50_observations"
    ]
    / track_country_performance_df[
        "observations"
    ]
    * 100
)

track_country_performance_df[
    "number_one_rate_pct"
] = (
    track_country_performance_df[
        "number_one_observations"
    ]
    / track_country_performance_df[
        "observations"
    ]
    * 100
)

# ------------------------------------------------------------
# 5. Observation-span calculation
# ------------------------------------------------------------

track_country_performance_df[
    "observation_span_days"
] = (
    track_country_performance_df[
        "last_reporting_date"
    ]
    - track_country_performance_df[
        "first_reporting_date"
    ]
).dt.days

# ------------------------------------------------------------
# 6. Track-country support classification
# ------------------------------------------------------------

track_country_performance_df[
    "support_classification"
] = np.select(
    [
        track_country_performance_df[
            "observations"
        ] >= 100,

        track_country_performance_df[
            "observations"
        ] >= 25,

        track_country_performance_df[
            "observations"
        ] >= 5,
    ],
    [
        "Strong support",
        "Moderate support",
        "Limited support",
    ],
    default="Very limited support"
)

# ------------------------------------------------------------
# 7. Track geographic breadth
# ------------------------------------------------------------

track_market_counts = (
    track_country_performance_df
    .groupby(
        "track_id"
    )["country"]
    .nunique()
)

track_country_performance_df[
    "track_country_markets"
] = (
    track_country_performance_df[
        "track_id"
    ]
    .map(
        track_market_counts
    )
)

track_country_performance_df[
    "track_market_coverage_pct"
] = (
    track_country_performance_df[
        "track_country_markets"
    ]
    / source_df[
        "country"
    ].nunique()
    * 100
)

# ------------------------------------------------------------
# 8. Country quality metadata
# ------------------------------------------------------------

if "country_quality_df" in globals():

    track_country_performance_df = (
        track_country_performance_df
        .merge(
            country_quality_df[
                [
                    "country",
                    "quality_review_priority",
                    "quality_risk_count",
                ]
            ],
            on="country",
            how="left",
            validate="many_to_one"
        )
    )

if "country_historical_coverage_df" in globals():

    track_country_performance_df = (
        track_country_performance_df
        .merge(
            country_historical_coverage_df[
                [
                    "country",
                    "date_coverage_pct",
                    "active_span_coverage_pct",
                    "historical_coverage_class",
                ]
            ],
            on="country",
            how="left",
            validate="many_to_one"
        )
    )

# ------------------------------------------------------------
# 9. Track-country overview
# ------------------------------------------------------------

print()
print("Track-country performance overview")
print("=" * 100)

print(
    f"Track-country summaries prepared: "
    f"{len(track_country_performance_df):,}"
)

print(
    f"Unique tracks represented: "
    f"{track_country_performance_df['track_id'].nunique():,}"
)

print(
    f"Country markets represented: "
    f"{track_country_performance_df['country'].nunique():,}"
)

print(
    f"Total grouped observations represented: "
    f"{int(track_country_performance_df['observations'].sum()):,}"
)

print(
    f"Total grouped streams represented: "
    f"{track_country_performance_df['total_streams'].sum():,.0f}"
)

print(
    f"Median observations per track-country pair: "
    f"{track_country_performance_df['observations'].median():,.1f}"
)

# ------------------------------------------------------------
# 10. Highest total-stream track-country combinations
# ------------------------------------------------------------

top_track_country_total_streams_df = (
    track_country_performance_df
    .sort_values(
        "total_streams",
        ascending=False
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Track-country combinations with highest total observed streams"
)
print("=" * 100)

display(
    top_track_country_total_streams_df[
        [
            "track_name",
            "artists_raw",
            "country",
            "total_streams",
            "observations",
            "reporting_dates",
            "median_streams",
            "median_chart_position",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 11. Supported median-stream combinations
# ------------------------------------------------------------

supported_track_country_df = (
    track_country_performance_df.loc[
        track_country_performance_df[
            "observations"
        ] >= 25
    ]
    .copy()
)

top_track_country_median_streams_df = (
    supported_track_country_df
    .sort_values(
        [
            "median_streams",
            "observations",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Supported track-country combinations with highest median streams"
)
print("=" * 100)

display(
    top_track_country_median_streams_df[
        [
            "track_name",
            "artists_raw",
            "country",
            "median_streams",
            "mean_streams",
            "total_streams",
            "observations",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Strongest supported chart profiles
# ------------------------------------------------------------

top_track_country_chart_df = (
    supported_track_country_df
    .sort_values(
        [
            "median_chart_position",
            "top_10_rate_pct",
            "observations",
        ],
        ascending=[
            True,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Supported track-country combinations with strongest chart profiles"
)
print("=" * 100)

display(
    top_track_country_chart_df[
        [
            "track_name",
            "artists_raw",
            "country",
            "median_chart_position",
            "mean_chart_position",
            "top_10_rate_pct",
            "top_50_rate_pct",
            "number_one_rate_pct",
            "observations",
            "support_classification",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 13. Support distribution
# ------------------------------------------------------------

track_country_support_summary_df = (
    track_country_performance_df[
        "support_classification"
    ]
    .value_counts()
    .rename_axis(
        "Support Classification"
    )
    .reset_index(
        name="Track-Country Pairs"
    )
)

track_country_support_summary_df[
    "Share (%)"
] = (
    track_country_support_summary_df[
        "Track-Country Pairs"
    ]
    / len(
        track_country_performance_df
    )
    * 100
)

print()
print("Track-country analytical support distribution")
print("=" * 100)

display(
    track_country_support_summary_df
)

# ------------------------------------------------------------
# 14. Track geographic breadth diagnostics
# ------------------------------------------------------------

track_breadth_diagnostics_df = pd.DataFrame([
    {
        "Diagnostic": "Unique tracks",
        "Observed Value": int(
            track_country_performance_df[
                "track_id"
            ].nunique()
        ),
    },
    {
        "Diagnostic": "Track-country combinations",
        "Observed Value": int(
            len(
                track_country_performance_df
            )
        ),
    },
    {
        "Diagnostic": "Tracks appearing in one country",
        "Observed Value": int(
            (
                track_market_counts == 1
            ).sum()
        ),
    },
    {
        "Diagnostic": "Tracks appearing in at least 5 countries",
        "Observed Value": int(
            (
                track_market_counts >= 5
            ).sum()
        ),
    },
    {
        "Diagnostic": "Tracks appearing in at least 10 countries",
        "Observed Value": int(
            (
                track_market_counts >= 10
            ).sum()
        ),
    },
    {
        "Diagnostic": "Tracks appearing in at least 25 countries",
        "Observed Value": int(
            (
                track_market_counts >= 25
            ).sum()
        ),
    },
    {
        "Diagnostic": "Tracks appearing in at least 50 countries",
        "Observed Value": int(
            (
                track_market_counts >= 50
            ).sum()
        ),
    },
    {
        "Diagnostic": "Maximum countries for one track",
        "Observed Value": int(
            track_market_counts.max()
        ),
    },
])

print()
print("Track geographic breadth diagnostics")
print("=" * 100)

display(
    track_breadth_diagnostics_df
)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

numeric_validation_columns = [
    "observations",
    "reporting_dates",
    "total_streams",
    "mean_streams",
    "median_streams",
    "maximum_streams",
    "minimum_streams",
    "minimum_chart_position",
    "mean_chart_position",
    "median_chart_position",
    "maximum_chart_position",
    "top_10_observations",
    "top_50_observations",
    "number_one_observations",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "track_country_markets",
    "track_market_coverage_pct",
]

finite_track_country_metrics = bool(
    np.isfinite(
        track_country_performance_df[
            numeric_validation_columns
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

expected_track_country_pairs = int(
    source_df[
        [
            "track_id",
            "country",
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

section_4_4_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 4.3 completion",
        "Requirement": (
            "Artist-country aggregation must be complete "
            "before track-country preparation."
        ),
        "Observed Evidence": (
            f"Section 4.3 completion status: "
            f"{section_4_3_complete}"
        ),
        "Passed": bool(
            section_4_3_complete
        ),
    },
    {
        "Validation Area": "Track-country pair completeness",
        "Requirement": (
            "Every observed track-country combination "
            "must receive one summary."
        ),
        "Observed Evidence": (
            f"{len(track_country_performance_df):,} summaries "
            f"for {expected_track_country_pairs:,} observed pairs"
        ),
        "Passed": (
            len(
                track_country_performance_df
            )
            == expected_track_country_pairs
        ),
    },
    {
        "Validation Area": "Observation preservation",
        "Requirement": (
            "Track-country grouped observation counts must "
            "reconcile with the validated source population."
        ),
        "Observed Evidence": (
            f"{int(track_country_performance_df['observations'].sum()):,} "
            f"grouped observations from {len(source_df):,} source rows"
        ),
        "Passed": (
            int(
                track_country_performance_df[
                    "observations"
                ].sum()
            )
            == len(
                source_df
            )
        ),
    },
    {
        "Validation Area": "Stream preservation",
        "Requirement": (
            "Track-country total streams must reconcile "
            "with the validated source population."
        ),
        "Observed Evidence": (
            f"Grouped streams: "
            f"{track_country_performance_df['total_streams'].sum():,.0f}; "
            f"source streams: {source_df['streams'].sum():,.0f}"
        ),
        "Passed": bool(
            np.isclose(
                track_country_performance_df[
                    "total_streams"
                ].sum(),
                source_df[
                    "streams"
                ].sum(),
            )
        ),
    },
    {
        "Validation Area": "Finite track-country metrics",
        "Requirement": (
            "Track-country numerical metrics must remain finite."
        ),
        "Observed Evidence": (
            f"{len(numeric_validation_columns)} "
            f"numeric metric fields assessed"
        ),
        "Passed": finite_track_country_metrics,
    },
    {
        "Validation Area": "Track identity preservation",
        "Requirement": (
            "Track IDs must retain a consistent track name "
            "and artist representation."
        ),
        "Observed Evidence": (
            f"{track_country_performance_df['track_id'].nunique():,} "
            f"unique track IDs represented"
        ),
        "Passed": bool(
            (
                track_country_performance_df
                .groupby("track_id")[
                    "track_name"
                ]
                .nunique()
                <= 1
            ).all()
            and
            (
                track_country_performance_df
                .groupby("track_id")[
                    "artists_raw"
                ]
                .nunique()
                <= 1
            ).all()
        ),
    },
    {
        "Validation Area": "Country safeguard preservation",
        "Requirement": (
            "Track-country summaries must retain validated "
            "country interpretation safeguards."
        ),
        "Observed Evidence": (
            f"{track_country_performance_df['direct_comparison_caution'].notna().sum():,} "
            f"safeguard values retained"
        ),
        "Passed": bool(
            track_country_performance_df[
                [
                    "historical_comparison_allowed",
                    "trend_interpretation_allowed",
                    "direct_comparison_caution",
                ]
            ]
            .notna()
            .all()
            .all()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Track-country aggregation must exclude "
            "the global aggregate."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(track_country_performance_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                track_country_performance_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Track-country aggregation validation")
print("=" * 100)

display(
    section_4_4_validation_df
)

section_4_4_complete = bool(
    section_4_4_validation_df[
        "Passed"
    ].all()
)

if not section_4_4_complete:
    failed_checks = (
        section_4_4_validation_df.loc[
            ~section_4_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 4.4 track-country aggregation "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 4.4 track-country aggregation "
    "validation checks passed."
)
print(
    f"Section 4.4 completion status: "
    f"{section_4_4_complete}"
)
print(
    f"Track-country analytical table prepared: "
    f"{len(track_country_performance_df):,} rows."
)
print(
    "Track-country metrics remain descriptive preparation outputs "
    "and are not yet final PMIP geographic performance scores."
)
print(
    "The workflow is ready for "
    "Section 4.5 Temporal Geographic Alignment."
)
```

### Interpretation

Section 4.4 successfully creates the validated track-country analytical layer for PMIP geographic intelligence.

A total of **353,282 track-country combinations** were prepared across:

- **110,198 unique tracks**
- **76 country markets**
- **5,338,680 validated source observations**

The aggregation preserves the complete validated streaming volume of **1,031,736,702,723 streams**, confirming that no source observations or streaming totals were lost during preparation.

Track-country evidence is substantially sparse. The median track-country combination contains only **5 observations**, showing that many tracks have limited historical evidence within individual markets.

The analytical support distribution confirms this imbalance:

- **174,041 track-country pairs (49.26%)** have Very Limited support.
- **112,416 pairs (31.82%)** have Limited support.
- **60,977 pairs (17.26%)** have Moderate support.
- only **5,848 pairs (1.66%)** have Strong support.

This means that approximately four-fifths of all observed track-country combinations have fewer than 25 observations. Later PMIP geographic performance indicators should therefore incorporate minimum-support requirements or confidence safeguards before treating individual market performance as strongly established.

Geographic breadth is also highly uneven. **81,082 tracks** appear in only one country market, while progressively fewer tracks achieve broad international coverage. Only **1,114 tracks** appear in at least 50 markets, and the widest observed track reaches **73 of the 76 validated country markets**.

The track-country table now retains:

- streaming magnitude,
- mean and median streaming performance,
- chart-position behaviour,
- Top 10 and Top 50 frequencies,
- number-one chart frequency,
- observation support,
- first and last observed reporting dates,
- geographic breadth,
- historical-comparison safeguards,
- trend-interpretation safeguards,
- direct-comparison cautions.

Some apparently strong track-country profiles occur in markets where direct-comparison caution is required. These values should therefore be interpreted together with their evidence support rather than as unconditional rankings.

All Section 4.4 validation checks passed. The track-country analytical table provides the required entity-market structure for later geographic performance intelligence, while remaining a descriptive preparation output rather than a final PMIP score.

## 4.5 Temporal Geographic Alignment

### Purpose

This subsection validates how country-market observations align across reporting dates before PMIP begins direct geographic performance comparisons.

The objective is to ensure that differences between markets are not incorrectly attributed to performance when they may instead result from different historical entry dates, reporting gaps or unequal temporal coverage.

The analysis will:

- construct country-by-date observation support,
- identify reporting dates shared across multiple markets,
- measure temporal overlap between countries,
- distinguish full-history and overlapping-period comparisons,
- identify markets with incomplete temporal alignment,
- preserve the historical and quality safeguards established in Section 3,
- establish rules for later cross-country and trend comparisons.

No final geographic performance score is calculated in this subsection.

```python
import pandas as pd
import numpy as np

print("Preparing PMIP temporal geographic alignment")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

if "geographic_analysis_df" not in globals():
    raise RuntimeError(
        "geographic_analysis_df is not available. "
        "Run Section 4.1 before Section 4.5."
    )

if not globals().get(
    "section_4_4_complete",
    False
):
    raise RuntimeError(
        "Section 4.4 is not complete. "
        "Run and validate Section 4.4 before Section 4.5."
    )

source_df = geographic_analysis_df.copy()

print(
    f"Section 4.4 completion status: "
    f"{section_4_4_complete}"
)

# ------------------------------------------------------------
# 2. Validate and standardise dates
# ------------------------------------------------------------

source_df["date"] = pd.to_datetime(
    source_df["date"],
    errors="coerce"
)

invalid_dates = int(
    source_df["date"].isna().sum()
)

if invalid_dates > 0:
    raise RuntimeError(
        f"Temporal alignment found {invalid_dates:,} "
        "invalid reporting dates."
    )

all_reporting_dates = (
    source_df["date"]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)

dataset_start_date = all_reporting_dates.min()
dataset_end_date = all_reporting_dates.max()
total_dataset_dates = len(all_reporting_dates)

print()
print("Validated reporting-date boundary")
print("=" * 100)

print(
    f"Dataset start date: "
    f"{dataset_start_date.date()}"
)

print(
    f"Dataset end date: "
    f"{dataset_end_date.date()}"
)

print(
    f"Unique reporting dates: "
    f"{total_dataset_dates:,}"
)

# ------------------------------------------------------------
# 3. Country-date support table
# ------------------------------------------------------------

country_date_support_df = (
    source_df
    .groupby(
        [
            "country",
            "date",
        ],
        dropna=False
    )
    .agg(
        observations=("track_id", "size"),
        unique_tracks=("track_id", "nunique"),
        unique_artist_representations=(
            "artists_raw",
            "nunique"
        ),
        total_streams=("streams", "sum"),
        median_streams=("streams", "median"),
        mean_streams=("streams", "mean"),
        median_chart_position=("position", "median"),
        mean_chart_position=("position", "mean"),
    )
    .reset_index()
)

print()
print("Country-date analytical support")
print("=" * 100)

print(
    f"Country-date summaries prepared: "
    f"{len(country_date_support_df):,}"
)

print(
    f"Countries represented: "
    f"{country_date_support_df['country'].nunique():,}"
)

print(
    f"Reporting dates represented: "
    f"{country_date_support_df['date'].nunique():,}"
)

# ------------------------------------------------------------
# 4. Country temporal boundaries
# ------------------------------------------------------------

country_temporal_alignment_df = (
    country_date_support_df
    .groupby(
        "country"
    )
    .agg(
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        reporting_dates=("date", "nunique"),
        total_observations=("observations", "sum"),
        total_streams=("total_streams", "sum"),
    )
    .reset_index()
)

country_temporal_alignment_df[
    "date_coverage_pct"
] = (
    country_temporal_alignment_df[
        "reporting_dates"
    ]
    / total_dataset_dates
    * 100
)

country_temporal_alignment_df[
    "entry_delay_days"
] = (
    country_temporal_alignment_df[
        "first_reporting_date"
    ]
    - dataset_start_date
).dt.days

country_temporal_alignment_df[
    "ending_gap_days"
] = (
    dataset_end_date
    - country_temporal_alignment_df[
        "last_reporting_date"
    ]
).dt.days

# ------------------------------------------------------------
# 5. Active-span expected reporting dates
# ------------------------------------------------------------

date_position_map = {
    date: position
    for position, date
    in enumerate(all_reporting_dates)
}

country_temporal_alignment_df[
    "first_date_position"
] = (
    country_temporal_alignment_df[
        "first_reporting_date"
    ]
    .map(date_position_map)
)

country_temporal_alignment_df[
    "last_date_position"
] = (
    country_temporal_alignment_df[
        "last_reporting_date"
    ]
    .map(date_position_map)
)

country_temporal_alignment_df[
    "possible_dates_within_active_span"
] = (
    country_temporal_alignment_df[
        "last_date_position"
    ]
    - country_temporal_alignment_df[
        "first_date_position"
    ]
    + 1
)

country_temporal_alignment_df[
    "missing_dates_within_active_span"
] = (
    country_temporal_alignment_df[
        "possible_dates_within_active_span"
    ]
    - country_temporal_alignment_df[
        "reporting_dates"
    ]
)

country_temporal_alignment_df[
    "active_span_coverage_pct"
] = (
    country_temporal_alignment_df[
        "reporting_dates"
    ]
    / country_temporal_alignment_df[
        "possible_dates_within_active_span"
    ]
    * 100
)

# ------------------------------------------------------------
# 6. Attach existing safeguards
# ------------------------------------------------------------

safeguard_columns = [
    "country",
    "historical_comparison_allowed",
    "trend_interpretation_allowed",
    "direct_comparison_caution",
]

available_safeguard_columns = [
    column
    for column in safeguard_columns
    if column in source_df.columns
]

country_safeguards = (
    source_df[
        available_safeguard_columns
    ]
    .drop_duplicates(
        subset="country"
    )
)

country_temporal_alignment_df = (
    country_temporal_alignment_df
    .merge(
        country_safeguards,
        on="country",
        how="left",
        validate="one_to_one"
    )
)

# ------------------------------------------------------------
# 7. Temporal-alignment classification
# ------------------------------------------------------------

country_temporal_alignment_df[
    "temporal_alignment_class"
] = np.select(
    [
        (
            country_temporal_alignment_df[
                "date_coverage_pct"
            ] >= 95
        )
        &
        (
            country_temporal_alignment_df[
                "active_span_coverage_pct"
            ] >= 95
        ),

        (
            country_temporal_alignment_df[
                "date_coverage_pct"
            ] >= 75
        )
        &
        (
            country_temporal_alignment_df[
                "active_span_coverage_pct"
            ] >= 90
        ),

        (
            country_temporal_alignment_df[
                "date_coverage_pct"
            ] >= 50
        ),
    ],
    [
        "Near-complete temporal alignment",
        "Strong temporal alignment",
        "Partial temporal alignment",
    ],
    default="Limited temporal alignment"
)

# ------------------------------------------------------------
# 8. Direct-comparison temporal rule
# ------------------------------------------------------------

country_temporal_alignment_df[
    "full_history_comparison_supported"
] = (
    (
        country_temporal_alignment_df[
            "date_coverage_pct"
        ] >= 50
    )
    &
    (
        country_temporal_alignment_df[
            "historical_comparison_allowed"
        ]
    )
)

country_temporal_alignment_df[
    "overlap_period_comparison_required"
] = (
    ~country_temporal_alignment_df[
        "full_history_comparison_supported"
    ]
    |
    country_temporal_alignment_df[
        "direct_comparison_caution"
    ]
)

# ------------------------------------------------------------
# 9. Reporting-date market availability
# ------------------------------------------------------------

reporting_date_market_support_df = (
    country_date_support_df
    .groupby(
        "date"
    )
    .agg(
        active_countries=("country", "nunique"),
        observations=("observations", "sum"),
        unique_tracks=("unique_tracks", "sum"),
        total_streams=("total_streams", "sum"),
    )
    .reset_index()
)

reporting_date_market_support_df[
    "market_coverage_pct"
] = (
    reporting_date_market_support_df[
        "active_countries"
    ]
    / source_df[
        "country"
    ].nunique()
    * 100
)

print()
print("Reporting-date market availability")
print("=" * 100)

print(
    f"Minimum countries available on one reporting date: "
    f"{reporting_date_market_support_df['active_countries'].min():,}"
)

print(
    f"Median countries available per reporting date: "
    f"{reporting_date_market_support_df['active_countries'].median():,.1f}"
)

print(
    f"Maximum countries available on one reporting date: "
    f"{reporting_date_market_support_df['active_countries'].max():,}"
)

# ------------------------------------------------------------
# 10. Temporal-alignment distribution
# ------------------------------------------------------------

temporal_alignment_summary_df = (
    country_temporal_alignment_df[
        "temporal_alignment_class"
    ]
    .value_counts()
    .rename_axis(
        "Temporal Alignment Class"
    )
    .reset_index(
        name="Countries"
    )
)

temporal_alignment_summary_df[
    "Share (%)"
] = (
    temporal_alignment_summary_df[
        "Countries"
    ]
    / len(
        country_temporal_alignment_df
    )
    * 100
)

print()
print("Country temporal-alignment classification")
print("=" * 100)

display(
    temporal_alignment_summary_df
)

# ------------------------------------------------------------
# 11. Earliest and strongest-aligned markets
# ------------------------------------------------------------

strongest_temporal_alignment_df = (
    country_temporal_alignment_df
    .sort_values(
        [
            "date_coverage_pct",
            "active_span_coverage_pct",
            "reporting_dates",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Country markets with strongest temporal alignment")
print("=" * 100)

display(
    strongest_temporal_alignment_df[
        [
            "country",
            "first_reporting_date",
            "last_reporting_date",
            "reporting_dates",
            "date_coverage_pct",
            "active_span_coverage_pct",
            "temporal_alignment_class",
            "direct_comparison_caution",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Weakest temporal-alignment markets
# ------------------------------------------------------------

weakest_temporal_alignment_df = (
    country_temporal_alignment_df
    .sort_values(
        [
            "date_coverage_pct",
            "active_span_coverage_pct",
            "reporting_dates",
        ],
        ascending=[
            True,
            True,
            True,
        ]
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print("Country markets with weakest temporal alignment")
print("=" * 100)

display(
    weakest_temporal_alignment_df[
        [
            "country",
            "first_reporting_date",
            "last_reporting_date",
            "reporting_dates",
            "date_coverage_pct",
            "entry_delay_days",
            "missing_dates_within_active_span",
            "active_span_coverage_pct",
            "temporal_alignment_class",
            "overlap_period_comparison_required",
        ]
    ]
)

# ------------------------------------------------------------
# 13. Comparison-rule summary
# ------------------------------------------------------------

temporal_comparison_policy_df = pd.DataFrame([
    {
        "Comparison Rule": (
            "Full-history comparison supported"
        ),
        "Countries": int(
            country_temporal_alignment_df[
                "full_history_comparison_supported"
            ].sum()
        ),
    },
    {
        "Comparison Rule": (
            "Overlapping-period comparison required"
        ),
        "Countries": int(
            country_temporal_alignment_df[
                "overlap_period_comparison_required"
            ].sum()
        ),
    },
    {
        "Comparison Rule": (
            "Near-complete or strong temporal alignment"
        ),
        "Countries": int(
            country_temporal_alignment_df[
                "temporal_alignment_class"
            ].isin(
                [
                    "Near-complete temporal alignment",
                    "Strong temporal alignment",
                ]
            ).sum()
        ),
    },
    {
        "Comparison Rule": (
            "Partial or limited temporal alignment"
        ),
        "Countries": int(
            country_temporal_alignment_df[
                "temporal_alignment_class"
            ].isin(
                [
                    "Partial temporal alignment",
                    "Limited temporal alignment",
                ]
            ).sum()
        ),
    },
])

print()
print("Temporal comparison policy summary")
print("=" * 100)

display(
    temporal_comparison_policy_df
)

# ------------------------------------------------------------
# 14. Core temporal validation
# ------------------------------------------------------------

finite_temporal_fields = [
    "reporting_dates",
    "total_observations",
    "total_streams",
    "date_coverage_pct",
    "entry_delay_days",
    "ending_gap_days",
    "possible_dates_within_active_span",
    "missing_dates_within_active_span",
    "active_span_coverage_pct",
]

finite_temporal_metrics = bool(
    np.isfinite(
        country_temporal_alignment_df[
            finite_temporal_fields
        ].to_numpy(
            dtype=float
        )
    ).all()
)

country_date_observation_total = int(
    country_date_support_df[
        "observations"
    ].sum()
)

country_date_stream_total = float(
    country_date_support_df[
        "total_streams"
    ].sum()
)

section_4_5_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 4.4 completion",
        "Requirement": (
            "Track-country aggregation must be complete "
            "before temporal alignment."
        ),
        "Observed Evidence": (
            f"Section 4.4 completion status: "
            f"{section_4_4_complete}"
        ),
        "Passed": bool(
            section_4_4_complete
        ),
    },
    {
        "Validation Area": "Country temporal completeness",
        "Requirement": (
            "Every validated country market must receive "
            "one temporal-alignment summary."
        ),
        "Observed Evidence": (
            f"{len(country_temporal_alignment_df):,} summaries "
            f"for {source_df['country'].nunique():,} countries"
        ),
        "Passed": (
            len(
                country_temporal_alignment_df
            )
            == source_df[
                "country"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Country-date observation preservation",
        "Requirement": (
            "Country-date aggregation must preserve all "
            "validated source observations."
        ),
        "Observed Evidence": (
            f"{country_date_observation_total:,} grouped observations "
            f"from {len(source_df):,} source rows"
        ),
        "Passed": (
            country_date_observation_total
            == len(
                source_df
            )
        ),
    },
    {
        "Validation Area": "Country-date stream preservation",
        "Requirement": (
            "Country-date aggregation must preserve total streams."
        ),
        "Observed Evidence": (
            f"Grouped streams: "
            f"{country_date_stream_total:,.0f}; "
            f"source streams: "
            f"{source_df['streams'].sum():,.0f}"
        ),
        "Passed": bool(
            np.isclose(
                country_date_stream_total,
                source_df[
                    "streams"
                ].sum()
            )
        ),
    },
    {
        "Validation Area": "Temporal metric validity",
        "Requirement": (
            "Temporal-alignment metrics must remain finite."
        ),
        "Observed Evidence": (
            f"{len(finite_temporal_fields)} "
            f"numeric temporal fields assessed"
        ),
        "Passed": finite_temporal_metrics,
    },
    {
        "Validation Area": "Coverage percentage bounds",
        "Requirement": (
            "Temporal coverage percentages must remain "
            "between 0 and 100."
        ),
        "Observed Evidence": (
            f"Date coverage range: "
            f"{country_temporal_alignment_df['date_coverage_pct'].min():.2f}% "
            f"to "
            f"{country_temporal_alignment_df['date_coverage_pct'].max():.2f}%"
        ),
        "Passed": bool(
            country_temporal_alignment_df[
                "date_coverage_pct"
            ].between(
                0,
                100
            ).all()
            and
            country_temporal_alignment_df[
                "active_span_coverage_pct"
            ].between(
                0,
                100
            ).all()
        ),
    },
    {
        "Validation Area": "Chronological boundary validity",
        "Requirement": (
            "Every country's first reporting date must not "
            "occur after its last reporting date."
        ),
        "Observed Evidence": (
            f"{len(country_temporal_alignment_df):,} "
            "country histories assessed"
        ),
        "Passed": bool(
            (
                country_temporal_alignment_df[
                    "first_reporting_date"
                ]
                <=
                country_temporal_alignment_df[
                    "last_reporting_date"
                ]
            ).all()
        ),
    },
    {
        "Validation Area": "Global aggregate exclusion",
        "Requirement": (
            "Temporal country analysis must exclude "
            "the global aggregate."
        ),
        "Observed Evidence": (
            f"'global' present: "
            f"{'global' in set(country_temporal_alignment_df['country'].astype(str))}"
        ),
        "Passed": (
            "global"
            not in set(
                country_temporal_alignment_df[
                    "country"
                ].astype(str)
            )
        ),
    },
])

print()
print("Temporal geographic alignment validation")
print("=" * 100)

display(
    section_4_5_validation_df
)

section_4_5_complete = bool(
    section_4_5_validation_df[
        "Passed"
    ].all()
)

section_4_overall_complete = bool(
    globals().get(
        "section_4_1_complete",
        False
    )
    and
    globals().get(
        "section_4_2_complete",
        False
    )
    and
    globals().get(
        "section_4_3_complete",
        False
    )
    and
    globals().get(
        "section_4_4_complete",
        False
    )
    and
    section_4_5_complete
)

if not section_4_5_complete:
    failed_checks = (
        section_4_5_validation_df.loc[
            ~section_4_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 4.5 temporal geographic alignment "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 4.5 temporal geographic alignment "
    "validation checks passed."
)

print(
    f"Section 4.5 completion status: "
    f"{section_4_5_complete}"
)

print(
    f"Section 4 overall completion status: "
    f"{section_4_overall_complete}"
)

print(
    f"Country-date analytical table prepared: "
    f"{len(country_date_support_df):,} rows."
)

print(
    f"Country temporal-alignment table prepared: "
    f"{len(country_temporal_alignment_df):,} rows."
)

print(
    "Temporal safeguards are now available for later "
    "cross-country performance comparison."
)

print(
    "Geographic Performance Preparation is complete."
)
```

### Interpretation

Section 4.5 confirms that geographic comparisons within PMIP must account for differences in when country markets entered the dataset and how consistently they were observed through time.

The validated geographic dataset spans from **28 April 2013 to 6 April 2023**, covering **515 unique reporting dates**.

A total of **30,863 country-date summaries** were created across all **76 validated country markets**.

Market availability is not constant across the historical period. Depending on the reporting date, between **27 and 73 countries** are represented, with a median of **61 countries per reporting date**.

This confirms that the geographic dataset is not a perfectly balanced panel where every country is observed at every point in time.

The country temporal-alignment classification shows:

- **32 countries (42.11%)** have Near-Complete Temporal Alignment.
- **22 countries (28.95%)** have Strong Temporal Alignment.
- **7 countries (9.21%)** have Partial Temporal Alignment.
- **15 countries (19.74%)** have Limited Temporal Alignment.

Therefore, **54 of the 76 countries** provide near-complete or strong temporal support, while **22 countries** require additional caution when making long-term geographic comparisons.

Some countries entered the dataset substantially later than others. For example, several markets first appear between 2019 and 2022, meaning that their available performance history is considerably shorter than markets represented from 2013 onward.

The analysis also distinguishes between missing history and internal reporting gaps. A market may have excellent continuity once it enters the dataset while still possessing limited total historical coverage because it entered late.

This distinction is important because late market entry should not automatically be interpreted as weak performance or missing data.

Based on the validated temporal safeguards:

- **61 countries** support full-history geographic comparison.
- **16 countries** require comparison using overlapping historical periods.
- **54 countries** have near-complete or strong temporal alignment.
- **22 countries** have partial or limited temporal alignment.

For markets requiring overlapping-period comparison, PMIP should compare performance only across dates where the relevant markets are simultaneously represented rather than comparing their full available histories directly.

All country-date observation counts and streaming totals reconcile exactly with the validated analytical population. Coverage percentages remain valid, chronological boundaries are consistent and the global aggregate remains excluded from country-level temporal analysis.

All Section 4.5 validation checks passed.

The geographic performance preparation stage is therefore complete, with country, artist-country, track-country and temporal analytical structures now prepared for downstream PMIP geographic performance analysis.

# 5. Geographic Feature Engineering

## Purpose

This section engineers the geographic features required for PMIP to describe how artists and tracks perform across country markets.

The objective is to transform the validated geographic analytical structures prepared in Section 4 into reusable numerical features that can later support geographic scoring, interpretation and PMIP software integration.

The section develops features for:

- country reach,
- market penetration,
- geographic concentration,
- cross-market performance,
- geographic growth and decline.

These features are descriptive analytical measures and are not yet final PMIP geographic performance scores.

## 5.1 Country Reach Features

### Purpose

This subsection creates country-reach features that describe how widely each artist and track is represented across the validated PMIP country markets.

The objective is to convert the geographic breadth information prepared in Section 4 into reusable numerical features.

The analysis measures:

- number of country markets reached,
- percentage of validated markets reached,
- number of markets with sufficient observation support,
- number of markets with moderate or strong analytical support,
- number of strongly supported markets,
- number of markets without direct-comparison caution,
- observed historical span,
- geographic reach classification.

These features measure observed geographic breadth rather than the strength or quality of geographic performance.

```python
import pandas as pd
import numpy as np

print("Engineering PMIP country-reach features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

required_objects = [
    "artist_country_performance_df",
    "track_country_performance_df",
    "country_temporal_alignment_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Section 5.1 requires the following Section 4 outputs: "
        + ", ".join(missing_objects)
    )

if not globals().get(
    "section_4_overall_complete",
    False
):
    raise RuntimeError(
        "Section 4 is not complete. "
        "Run all Section 4 subsections before Section 5.1."
    )

artist_country_df = artist_country_performance_df.copy()
track_country_df = track_country_performance_df.copy()
temporal_df = country_temporal_alignment_df.copy()

total_validated_markets = int(
    temporal_df["country"].nunique()
)

print(
    f"Section 4 overall completion status: "
    f"{section_4_overall_complete}"
)

print(
    f"Validated country markets available: "
    f"{total_validated_markets:,}"
)

# ------------------------------------------------------------
# 2. Artist country-reach features
# ------------------------------------------------------------

artist_country_reach_features_df = (
    artist_country_df
    .groupby(
        "artists_raw",
        dropna=False
    )
    .agg(
        country_markets_reached=("country", "nunique"),
        artist_country_pairs=("country", "size"),
        total_artist_country_observations=(
            "observations",
            "sum"
        ),
        total_artist_country_streams=(
            "total_streams",
            "sum"
        ),
        max_country_observations=(
            "observations",
            "max"
        ),
        median_country_observations=(
            "observations",
            "median"
        ),
        first_observed_date=(
            "first_reporting_date",
            "min"
        ),
        last_observed_date=(
            "last_reporting_date",
            "max"
        ),
    )
    .reset_index()
)

artist_country_reach_features_df[
    "country_reach_pct"
] = (
    artist_country_reach_features_df[
        "country_markets_reached"
    ]
    / total_validated_markets
    * 100
)

# ------------------------------------------------------------
# 3. Artist market-support reach
# ------------------------------------------------------------

artist_support_counts = (
    artist_country_df
    .groupby(
        "artists_raw"
    )
    .agg(
        markets_with_at_least_limited_support=(
            "support_classification",
            lambda s: int(
                s.isin(
                    [
                        "Limited support",
                        "Moderate support",
                        "Strong support",
                    ]
                ).sum()
            )
        ),
        markets_with_moderate_or_strong_support=(
            "support_classification",
            lambda s: int(
                s.isin(
                    [
                        "Moderate support",
                        "Strong support",
                    ]
                ).sum()
            )
        ),
        markets_with_strong_support=(
            "support_classification",
            lambda s: int(
                (s == "Strong support").sum()
            )
        ),
        markets_without_direct_comparison_caution=(
            "direct_comparison_caution",
            lambda s: int(
                (~s.astype(bool)).sum()
            )
        ),
    )
    .reset_index()
)

artist_country_reach_features_df = (
    artist_country_reach_features_df
    .merge(
        artist_support_counts,
        on="artists_raw",
        how="left",
        validate="one_to_one"
    )
)

artist_country_reach_features_df[
    "moderate_or_strong_reach_pct"
] = (
    artist_country_reach_features_df[
        "markets_with_moderate_or_strong_support"
    ]
    / total_validated_markets
    * 100
)

artist_country_reach_features_df[
    "strong_support_reach_pct"
] = (
    artist_country_reach_features_df[
        "markets_with_strong_support"
    ]
    / total_validated_markets
    * 100
)

artist_country_reach_features_df[
    "comparison_safe_reach_pct"
] = (
    artist_country_reach_features_df[
        "markets_without_direct_comparison_caution"
    ]
    / total_validated_markets
    * 100
)

artist_country_reach_features_df[
    "observed_span_days"
] = (
    artist_country_reach_features_df[
        "last_observed_date"
    ]
    - artist_country_reach_features_df[
        "first_observed_date"
    ]
).dt.days

# ------------------------------------------------------------
# 4. Artist reach classification
# ------------------------------------------------------------

artist_country_reach_features_df[
    "country_reach_class"
] = pd.cut(
    artist_country_reach_features_df[
        "country_markets_reached"
    ],
    bins=[
        0,
        1,
        4,
        9,
        24,
        49,
        np.inf,
    ],
    labels=[
        "Single-market",
        "Limited multi-market",
        "Moderate international",
        "Broad international",
        "Very broad international",
        "Near-global reach",
    ],
    include_lowest=True
)

# ------------------------------------------------------------
# 5. Track country-reach features
# ------------------------------------------------------------

track_country_reach_features_df = (
    track_country_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_name=("track_name", "first"),
        artists_raw=("artists_raw", "first"),
        primary_artist=("primary_artist", "first"),
        country_markets_reached=("country", "nunique"),
        track_country_pairs=("country", "size"),
        total_track_country_observations=(
            "observations",
            "sum"
        ),
        total_track_country_streams=(
            "total_streams",
            "sum"
        ),
        max_country_observations=(
            "observations",
            "max"
        ),
        median_country_observations=(
            "observations",
            "median"
        ),
        first_observed_date=(
            "first_reporting_date",
            "min"
        ),
        last_observed_date=(
            "last_reporting_date",
            "max"
        ),
    )
    .reset_index()
)

track_country_reach_features_df[
    "country_reach_pct"
] = (
    track_country_reach_features_df[
        "country_markets_reached"
    ]
    / total_validated_markets
    * 100
)

# ------------------------------------------------------------
# 6. Track market-support reach
# ------------------------------------------------------------

track_support_counts = (
    track_country_df
    .groupby(
        "track_id"
    )
    .agg(
        markets_with_at_least_limited_support=(
            "support_classification",
            lambda s: int(
                s.isin(
                    [
                        "Limited support",
                        "Moderate support",
                        "Strong support",
                    ]
                ).sum()
            )
        ),
        markets_with_moderate_or_strong_support=(
            "support_classification",
            lambda s: int(
                s.isin(
                    [
                        "Moderate support",
                        "Strong support",
                    ]
                ).sum()
            )
        ),
        markets_with_strong_support=(
            "support_classification",
            lambda s: int(
                (s == "Strong support").sum()
            )
        ),
        markets_without_direct_comparison_caution=(
            "direct_comparison_caution",
            lambda s: int(
                (~s.astype(bool)).sum()
            )
        ),
    )
    .reset_index()
)

track_country_reach_features_df = (
    track_country_reach_features_df
    .merge(
        track_support_counts,
        on="track_id",
        how="left",
        validate="one_to_one"
    )
)

track_country_reach_features_df[
    "moderate_or_strong_reach_pct"
] = (
    track_country_reach_features_df[
        "markets_with_moderate_or_strong_support"
    ]
    / total_validated_markets
    * 100
)

track_country_reach_features_df[
    "strong_support_reach_pct"
] = (
    track_country_reach_features_df[
        "markets_with_strong_support"
    ]
    / total_validated_markets
    * 100
)

track_country_reach_features_df[
    "comparison_safe_reach_pct"
] = (
    track_country_reach_features_df[
        "markets_without_direct_comparison_caution"
    ]
    / total_validated_markets
    * 100
)

track_country_reach_features_df[
    "observed_span_days"
] = (
    track_country_reach_features_df[
        "last_observed_date"
    ]
    - track_country_reach_features_df[
        "first_observed_date"
    ]
).dt.days

# ------------------------------------------------------------
# 7. Track reach classification
# ------------------------------------------------------------

track_country_reach_features_df[
    "country_reach_class"
] = pd.cut(
    track_country_reach_features_df[
        "country_markets_reached"
    ],
    bins=[
        0,
        1,
        4,
        9,
        24,
        49,
        np.inf,
    ],
    labels=[
        "Single-market",
        "Limited multi-market",
        "Moderate international",
        "Broad international",
        "Very broad international",
        "Near-global reach",
    ],
    include_lowest=True
)

# ------------------------------------------------------------
# 8. Artist reach summary
# ------------------------------------------------------------

print()
print("Artist country-reach overview")
print("=" * 100)

print(
    f"Artist reach records prepared: "
    f"{len(artist_country_reach_features_df):,}"
)

print(
    f"Median markets reached per artist representation: "
    f"{artist_country_reach_features_df['country_markets_reached'].median():,.1f}"
)

print(
    f"Mean markets reached per artist representation: "
    f"{artist_country_reach_features_df['country_markets_reached'].mean():,.2f}"
)

print(
    f"Maximum markets reached by one artist representation: "
    f"{artist_country_reach_features_df['country_markets_reached'].max():,}"
)

artist_reach_distribution_df = (
    artist_country_reach_features_df[
        "country_reach_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Country Reach Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_reach_distribution_df[
    "Share (%)"
] = (
    artist_reach_distribution_df[
        "Artist Representations"
    ]
    / len(
        artist_country_reach_features_df
    )
    * 100
)

print()
print("Artist country-reach classification")
print("=" * 100)

display(
    artist_reach_distribution_df
)

# ------------------------------------------------------------
# 9. Widest artist reach
# ------------------------------------------------------------

top_artist_reach_df = (
    artist_country_reach_features_df
    .sort_values(
        [
            "country_markets_reached",
            "markets_with_moderate_or_strong_support",
            "total_artist_country_observations",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with widest observed country reach")
print("=" * 100)

display(
    top_artist_reach_df[
        [
            "artists_raw",
            "country_markets_reached",
            "country_reach_pct",
            "markets_with_moderate_or_strong_support",
            "moderate_or_strong_reach_pct",
            "markets_with_strong_support",
            "comparison_safe_reach_pct",
            "total_artist_country_observations",
            "country_reach_class",
        ]
    ]
)

# ------------------------------------------------------------
# 10. Track reach summary
# ------------------------------------------------------------

print()
print("Track country-reach overview")
print("=" * 100)

print(
    f"Track reach records prepared: "
    f"{len(track_country_reach_features_df):,}"
)

print(
    f"Median markets reached per track: "
    f"{track_country_reach_features_df['country_markets_reached'].median():,.1f}"
)

print(
    f"Mean markets reached per track: "
    f"{track_country_reach_features_df['country_markets_reached'].mean():,.2f}"
)

print(
    f"Maximum markets reached by one track: "
    f"{track_country_reach_features_df['country_markets_reached'].max():,}"
)

track_reach_distribution_df = (
    track_country_reach_features_df[
        "country_reach_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Country Reach Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_reach_distribution_df[
    "Share (%)"
] = (
    track_reach_distribution_df[
        "Tracks"
    ]
    / len(
        track_country_reach_features_df
    )
    * 100
)

print()
print("Track country-reach classification")
print("=" * 100)

display(
    track_reach_distribution_df
)

# ------------------------------------------------------------
# 11. Widest track reach
# ------------------------------------------------------------

top_track_reach_df = (
    track_country_reach_features_df
    .sort_values(
        [
            "country_markets_reached",
            "markets_with_moderate_or_strong_support",
            "total_track_country_observations",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with widest observed country reach")
print("=" * 100)

display(
    top_track_reach_df[
        [
            "track_name",
            "artists_raw",
            "country_markets_reached",
            "country_reach_pct",
            "markets_with_moderate_or_strong_support",
            "moderate_or_strong_reach_pct",
            "markets_with_strong_support",
            "comparison_safe_reach_pct",
            "total_track_country_observations",
            "country_reach_class",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

artist_numeric_features = [
    "country_markets_reached",
    "artist_country_pairs",
    "total_artist_country_observations",
    "total_artist_country_streams",
    "max_country_observations",
    "median_country_observations",
    "country_reach_pct",
    "markets_with_at_least_limited_support",
    "markets_with_moderate_or_strong_support",
    "markets_with_strong_support",
    "markets_without_direct_comparison_caution",
    "moderate_or_strong_reach_pct",
    "strong_support_reach_pct",
    "comparison_safe_reach_pct",
    "observed_span_days",
]

track_numeric_features = [
    "country_markets_reached",
    "track_country_pairs",
    "total_track_country_observations",
    "total_track_country_streams",
    "max_country_observations",
    "median_country_observations",
    "country_reach_pct",
    "markets_with_at_least_limited_support",
    "markets_with_moderate_or_strong_support",
    "markets_with_strong_support",
    "markets_without_direct_comparison_caution",
    "moderate_or_strong_reach_pct",
    "strong_support_reach_pct",
    "comparison_safe_reach_pct",
    "observed_span_days",
]

artist_features_finite = bool(
    np.isfinite(
        artist_country_reach_features_df[
            artist_numeric_features
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

track_features_finite = bool(
    np.isfinite(
        track_country_reach_features_df[
            track_numeric_features
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

section_5_1_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 4 completion",
        "Requirement": (
            "Geographic preparation must be complete "
            "before feature engineering."
        ),
        "Observed Evidence": (
            f"Section 4 overall completion status: "
            f"{section_4_overall_complete}"
        ),
        "Passed": bool(
            section_4_overall_complete
        ),
    },
    {
        "Validation Area": "Artist reach completeness",
        "Requirement": (
            "Every validated artist representation must "
            "receive one country-reach record."
        ),
        "Observed Evidence": (
            f"{len(artist_country_reach_features_df):,} records "
            f"for {artist_country_df['artists_raw'].nunique():,} artists"
        ),
        "Passed": (
            len(
                artist_country_reach_features_df
            )
            ==
            artist_country_df[
                "artists_raw"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Track reach completeness",
        "Requirement": (
            "Every validated track must receive "
            "one country-reach record."
        ),
        "Observed Evidence": (
            f"{len(track_country_reach_features_df):,} records "
            f"for {track_country_df['track_id'].nunique():,} tracks"
        ),
        "Passed": (
            len(
                track_country_reach_features_df
            )
            ==
            track_country_df[
                "track_id"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Artist reach bounds",
        "Requirement": (
            "Artist country reach must remain between "
            "1 and the validated number of markets."
        ),
        "Observed Evidence": (
            f"Observed range: "
            f"{artist_country_reach_features_df['country_markets_reached'].min()} "
            f"to "
            f"{artist_country_reach_features_df['country_markets_reached'].max()}"
        ),
        "Passed": bool(
            artist_country_reach_features_df[
                "country_markets_reached"
            ]
            .between(
                1,
                total_validated_markets
            )
            .all()
        ),
    },
    {
        "Validation Area": "Track reach bounds",
        "Requirement": (
            "Track country reach must remain between "
            "1 and the validated number of markets."
        ),
        "Observed Evidence": (
            f"Observed range: "
            f"{track_country_reach_features_df['country_markets_reached'].min()} "
            f"to "
            f"{track_country_reach_features_df['country_markets_reached'].max()}"
        ),
        "Passed": bool(
            track_country_reach_features_df[
                "country_markets_reached"
            ]
            .between(
                1,
                total_validated_markets
            )
            .all()
        ),
    },
    {
        "Validation Area": "Artist feature validity",
        "Requirement": (
            "All engineered artist country-reach "
            "features must remain finite."
        ),
        "Observed Evidence": (
            f"{len(artist_numeric_features)} "
            f"numeric artist reach features assessed"
        ),
        "Passed": artist_features_finite,
    },
    {
        "Validation Area": "Track feature validity",
        "Requirement": (
            "All engineered track country-reach "
            "features must remain finite."
        ),
        "Observed Evidence": (
            f"{len(track_numeric_features)} "
            f"numeric track reach features assessed"
        ),
        "Passed": track_features_finite,
    },
    {
        "Validation Area": "Reach percentage bounds",
        "Requirement": (
            "All country-reach percentages must "
            "remain between 0 and 100."
        ),
        "Observed Evidence": (
            "Artist and track reach percentage fields assessed"
        ),
        "Passed": bool(
            artist_country_reach_features_df[
                [
                    "country_reach_pct",
                    "moderate_or_strong_reach_pct",
                    "strong_support_reach_pct",
                    "comparison_safe_reach_pct",
                ]
            ]
            .apply(
                lambda s: s.between(
                    0,
                    100
                ).all()
            )
            .all()
            and
            track_country_reach_features_df[
                [
                    "country_reach_pct",
                    "moderate_or_strong_reach_pct",
                    "strong_support_reach_pct",
                    "comparison_safe_reach_pct",
                ]
            ]
            .apply(
                lambda s: s.between(
                    0,
                    100
                ).all()
            )
            .all()
        ),
    },
])

print()
print("Country-reach feature validation")
print("=" * 100)

display(
    section_5_1_validation_df
)

section_5_1_complete = bool(
    section_5_1_validation_df[
        "Passed"
    ].all()
)

if not section_5_1_complete:
    failed_checks = (
        section_5_1_validation_df.loc[
            ~section_5_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.1 country-reach feature "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 5.1 country-reach feature "
    "validation checks passed."
)

print(
    f"Section 5.1 completion status: "
    f"{section_5_1_complete}"
)

print(
    f"Artist country-reach feature table prepared: "
    f"{len(artist_country_reach_features_df):,} rows."
)

print(
    f"Track country-reach feature table prepared: "
    f"{len(track_country_reach_features_df):,} rows."
)

print(
    "Country-reach features describe observed geographic breadth "
    "and evidence support, not geographic performance quality."
)

print(
    "The workflow is ready for "
    "Section 5.2 Market Penetration Features."
)
```

### Interpretation

Section 5.1 successfully transforms the validated geographic structures from Section 4 into reusable country-reach features for both artists and tracks.

The artist feature table contains **49,843 artist representations**, while the track feature table contains **110,198 tracks**, confirming complete preservation of the validated entity populations.

Artist geographic reach remains highly uneven. The median artist representation appears in only **1 country**, while the mean reach is approximately **3.71 markets**. However, the widest observed artist representation reaches all **76 validated country markets**.

The artist reach classifications show that:

- **34,830 artist representations (69.88%)** are Single-Market.
- **8,871 (17.80%)** have Limited Multi-Market reach.
- **2,062 (4.14%)** have Moderate International reach.
- **2,347 (4.71%)** have Broad International reach.
- **1,069 (2.14%)** have Very Broad International reach.
- **664 (1.33%)** have Near-Global reach.

Track reach is even more geographically concentrated. The median track appears in only **1 market**, while the mean track reach is approximately **3.21 markets**. The widest observed track reaches **73 of the 76 validated country markets**.

The track reach classifications show that:

- **81,082 tracks (73.58%)** are Single-Market.
- **17,859 (16.21%)** have Limited Multi-Market reach.
- **3,969 (3.60%)** have Moderate International reach.
- **4,140 (3.76%)** have Broad International reach.
- **2,034 (1.85%)** have Very Broad International reach.
- **1,114 (1.01%)** have Near-Global reach.

These results confirm that broad international presence is relatively uncommon within the historical PMIP dataset.

The engineered support features also reveal an important distinction between **geographic presence** and **analytical evidence strength**. A track may appear across many countries while still having limited observation depth within individual track-country combinations.

For example, some of the widest-reaching tracks contain little or no Moderate or Strong analytical support despite appearing in more than 70 markets. This does not invalidate their observed reach; instead, it means that geographic breadth and evidence depth describe different properties and must remain separate.

Conversely, several widely represented artists and tracks retain Moderate or Strong support across a large proportion of their observed markets, providing stronger evidence for later geographic comparison.

The comparison-safe reach features further preserve the country-level safeguards developed in Sections 3 and 4, preventing wide geographic presence from automatically implying that every market comparison is equally reliable.

All Section 5.1 validation checks passed:

- every validated artist representation received one country-reach feature record,
- every validated track received one country-reach feature record,
- artist reach remained within the validated 1–76 market range,
- track reach remained within the validated 1–73 observed market range,
- all engineered numerical features remained finite,
- all percentage features remained within valid 0–100% boundaries.

Country-reach features therefore provide a validated description of **how widely artists and tracks are represented across PMIP markets**, while intentionally remaining separate from geographic performance quality.

The workflow is ready for Section 5.2 Market Penetration Features.

## 5.2 Market Penetration Features

### Purpose

This subsection engineers market-penetration features that describe how strongly artists and tracks are represented within the country markets they have reached.

Country reach measures geographic breadth, while market penetration measures the depth of an entity's observed presence within those markets.

The analysis will engineer features based on:

- an artist or track's share of observations within each country,
- an artist or track's share of observed streams within each country,
- relative streaming strength compared with the wider country market,
- chart-presence depth within individual markets,
- number and percentage of markets where an entity achieves meaningful penetration,
- strongest observed market penetration,
- typical market penetration across an entity's reached markets.

The features will retain the analytical-support and comparison safeguards established in earlier sections.

Market-penetration features remain descriptive engineered indicators. They do not yet represent final PMIP geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Engineering PMIP market-penetration features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

required_objects = [
    "country_performance_df",
    "artist_country_performance_df",
    "track_country_performance_df",
    "artist_country_reach_features_df",
    "track_country_reach_features_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Section 5.2 requires the following previous outputs: "
        + ", ".join(missing_objects)
    )

if not globals().get(
    "section_5_1_complete",
    False
):
    raise RuntimeError(
        "Section 5.1 is not complete. "
        "Run and validate Section 5.1 before Section 5.2."
    )

country_df = country_performance_df.copy()
artist_country_df = artist_country_performance_df.copy()
track_country_df = track_country_performance_df.copy()

print(
    f"Section 5.1 completion status: "
    f"{section_5_1_complete}"
)

print(
    f"Country markets available: "
    f"{country_df['country'].nunique():,}"
)

# ------------------------------------------------------------
# 2. Validate required fields
# ------------------------------------------------------------

required_country_fields = [
    "country",
    "observations",
    "total_streams",
]

required_artist_country_fields = [
    "artists_raw",
    "country",
    "observations",
    "total_streams",
    "median_streams",
    "median_chart_position",
    "support_classification",
    "direct_comparison_caution",
]

required_track_country_fields = [
    "track_id",
    "track_name",
    "artists_raw",
    "country",
    "observations",
    "total_streams",
    "median_streams",
    "median_chart_position",
    "support_classification",
    "direct_comparison_caution",
]

for label, df, required_fields in [
    (
        "country_performance_df",
        country_df,
        required_country_fields,
    ),
    (
        "artist_country_performance_df",
        artist_country_df,
        required_artist_country_fields,
    ),
    (
        "track_country_performance_df",
        track_country_df,
        required_track_country_fields,
    ),
]:
    missing_fields = [
        field
        for field in required_fields
        if field not in df.columns
    ]

    if missing_fields:
        raise RuntimeError(
            f"{label} is missing required fields: "
            + ", ".join(missing_fields)
        )

# ------------------------------------------------------------
# 3. Prepare country market denominators
# ------------------------------------------------------------

country_market_denominators_df = (
    country_df[
        [
            "country",
            "observations",
            "total_streams",
        ]
    ]
    .rename(
        columns={
            "observations": "country_total_observations",
            "total_streams": "country_total_streams",
        }
    )
    .copy()
)

if (
    country_market_denominators_df[
        "country_total_observations"
    ] <= 0
).any():
    raise RuntimeError(
        "Country observation denominators must be positive."
    )

if (
    country_market_denominators_df[
        "country_total_streams"
    ] <= 0
).any():
    raise RuntimeError(
        "Country stream denominators must be positive."
    )

# ------------------------------------------------------------
# 4. Artist-country penetration features
# ------------------------------------------------------------

artist_market_penetration_df = (
    artist_country_df
    .merge(
        country_market_denominators_df,
        on="country",
        how="left",
        validate="many_to_one",
    )
)

artist_market_penetration_df[
    "country_observation_share_pct"
] = (
    artist_market_penetration_df[
        "observations"
    ]
    / artist_market_penetration_df[
        "country_total_observations"
    ]
    * 100
)

artist_market_penetration_df[
    "country_stream_share_pct"
] = (
    artist_market_penetration_df[
        "total_streams"
    ]
    / artist_market_penetration_df[
        "country_total_streams"
    ]
    * 100
)

# Relative stream intensity:
# > 1 means the entity's average streams per observation
# are above the overall country average.

artist_market_penetration_df[
    "artist_average_streams_per_observation"
] = (
    artist_market_penetration_df[
        "total_streams"
    ]
    / artist_market_penetration_df[
        "observations"
    ]
)

artist_market_penetration_df[
    "country_average_streams_per_observation"
] = (
    artist_market_penetration_df[
        "country_total_streams"
    ]
    / artist_market_penetration_df[
        "country_total_observations"
    ]
)

artist_market_penetration_df[
    "relative_stream_intensity"
] = (
    artist_market_penetration_df[
        "artist_average_streams_per_observation"
    ]
    / artist_market_penetration_df[
        "country_average_streams_per_observation"
    ]
)

# ------------------------------------------------------------
# 5. Track-country penetration features
# ------------------------------------------------------------

track_market_penetration_df = (
    track_country_df
    .merge(
        country_market_denominators_df,
        on="country",
        how="left",
        validate="many_to_one",
    )
)

track_market_penetration_df[
    "country_observation_share_pct"
] = (
    track_market_penetration_df[
        "observations"
    ]
    / track_market_penetration_df[
        "country_total_observations"
    ]
    * 100
)

track_market_penetration_df[
    "country_stream_share_pct"
] = (
    track_market_penetration_df[
        "total_streams"
    ]
    / track_market_penetration_df[
        "country_total_streams"
    ]
    * 100
)

track_market_penetration_df[
    "track_average_streams_per_observation"
] = (
    track_market_penetration_df[
        "total_streams"
    ]
    / track_market_penetration_df[
        "observations"
    ]
)

track_market_penetration_df[
    "country_average_streams_per_observation"
] = (
    track_market_penetration_df[
        "country_total_streams"
    ]
    / track_market_penetration_df[
        "country_total_observations"
    ]
)

track_market_penetration_df[
    "relative_stream_intensity"
] = (
    track_market_penetration_df[
        "track_average_streams_per_observation"
    ]
    / track_market_penetration_df[
        "country_average_streams_per_observation"
    ]
)

# ------------------------------------------------------------
# 6. Define support-qualified market penetration
# ------------------------------------------------------------

supported_classes = [
    "Moderate support",
    "Strong support",
]

artist_market_penetration_df[
    "support_qualified_market"
] = (
    artist_market_penetration_df[
        "support_classification"
    ].isin(
        supported_classes
    )
)

track_market_penetration_df[
    "support_qualified_market"
] = (
    track_market_penetration_df[
        "support_classification"
    ].isin(
        supported_classes
    )
)

artist_market_penetration_df[
    "comparison_safe_market"
] = (
    ~artist_market_penetration_df[
        "direct_comparison_caution"
    ].astype(bool)
)

track_market_penetration_df[
    "comparison_safe_market"
] = (
    ~track_market_penetration_df[
        "direct_comparison_caution"
    ].astype(bool)
)

# ------------------------------------------------------------
# 7. Artist-level penetration summary
# ------------------------------------------------------------

artist_penetration_features_df = (
    artist_market_penetration_df
    .groupby(
        "artists_raw",
        dropna=False
    )
    .agg(
        markets_observed=(
            "country",
            "nunique"
        ),
        median_country_observation_share_pct=(
            "country_observation_share_pct",
            "median"
        ),
        mean_country_observation_share_pct=(
            "country_observation_share_pct",
            "mean"
        ),
        maximum_country_observation_share_pct=(
            "country_observation_share_pct",
            "max"
        ),
        median_country_stream_share_pct=(
            "country_stream_share_pct",
            "median"
        ),
        mean_country_stream_share_pct=(
            "country_stream_share_pct",
            "mean"
        ),
        maximum_country_stream_share_pct=(
            "country_stream_share_pct",
            "max"
        ),
        median_relative_stream_intensity=(
            "relative_stream_intensity",
            "median"
        ),
        mean_relative_stream_intensity=(
            "relative_stream_intensity",
            "mean"
        ),
        maximum_relative_stream_intensity=(
            "relative_stream_intensity",
            "max"
        ),
        median_market_chart_position=(
            "median_chart_position",
            "median"
        ),
        best_market_chart_position=(
            "median_chart_position",
            "min"
        ),
        support_qualified_markets=(
            "support_qualified_market",
            "sum"
        ),
        comparison_safe_markets=(
            "comparison_safe_market",
            "sum"
        ),
    )
    .reset_index()
)

artist_penetration_features_df[
    "support_qualified_market_pct"
] = (
    artist_penetration_features_df[
        "support_qualified_markets"
    ]
    / artist_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

artist_penetration_features_df[
    "comparison_safe_market_pct"
] = (
    artist_penetration_features_df[
        "comparison_safe_markets"
    ]
    / artist_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

# ------------------------------------------------------------
# 8. Track-level penetration summary
# ------------------------------------------------------------

track_penetration_features_df = (
    track_market_penetration_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_name=(
            "track_name",
            "first"
        ),
        artists_raw=(
            "artists_raw",
            "first"
        ),
        markets_observed=(
            "country",
            "nunique"
        ),
        median_country_observation_share_pct=(
            "country_observation_share_pct",
            "median"
        ),
        mean_country_observation_share_pct=(
            "country_observation_share_pct",
            "mean"
        ),
        maximum_country_observation_share_pct=(
            "country_observation_share_pct",
            "max"
        ),
        median_country_stream_share_pct=(
            "country_stream_share_pct",
            "median"
        ),
        mean_country_stream_share_pct=(
            "country_stream_share_pct",
            "mean"
        ),
        maximum_country_stream_share_pct=(
            "country_stream_share_pct",
            "max"
        ),
        median_relative_stream_intensity=(
            "relative_stream_intensity",
            "median"
        ),
        mean_relative_stream_intensity=(
            "relative_stream_intensity",
            "mean"
        ),
        maximum_relative_stream_intensity=(
            "relative_stream_intensity",
            "max"
        ),
        median_market_chart_position=(
            "median_chart_position",
            "median"
        ),
        best_market_chart_position=(
            "median_chart_position",
            "min"
        ),
        support_qualified_markets=(
            "support_qualified_market",
            "sum"
        ),
        comparison_safe_markets=(
            "comparison_safe_market",
            "sum"
        ),
    )
    .reset_index()
)

track_penetration_features_df[
    "support_qualified_market_pct"
] = (
    track_penetration_features_df[
        "support_qualified_markets"
    ]
    / track_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

track_penetration_features_df[
    "comparison_safe_market_pct"
] = (
    track_penetration_features_df[
        "comparison_safe_markets"
    ]
    / track_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

# ------------------------------------------------------------
# 9. Penetration-strength indicators
# ------------------------------------------------------------

artist_penetration_features_df[
    "markets_above_country_stream_average"
] = (
    artist_market_penetration_df
    .assign(
        above_average=(
            artist_market_penetration_df[
                "relative_stream_intensity"
            ] > 1
        )
    )
    .groupby(
        "artists_raw"
    )[
        "above_average"
    ]
    .sum()
    .reindex(
        artist_penetration_features_df[
            "artists_raw"
        ]
    )
    .to_numpy()
)

artist_penetration_features_df[
    "above_country_average_market_pct"
] = (
    artist_penetration_features_df[
        "markets_above_country_stream_average"
    ]
    / artist_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

track_penetration_features_df[
    "markets_above_country_stream_average"
] = (
    track_market_penetration_df
    .assign(
        above_average=(
            track_market_penetration_df[
                "relative_stream_intensity"
            ] > 1
        )
    )
    .groupby(
        "track_id"
    )[
        "above_average"
    ]
    .sum()
    .reindex(
        track_penetration_features_df[
            "track_id"
        ]
    )
    .to_numpy()
)

track_penetration_features_df[
    "above_country_average_market_pct"
] = (
    track_penetration_features_df[
        "markets_above_country_stream_average"
    ]
    / track_penetration_features_df[
        "markets_observed"
    ]
    * 100
)

# ------------------------------------------------------------
# 10. Artist penetration overview
# ------------------------------------------------------------

print()
print("Artist market-penetration overview")
print("=" * 100)

print(
    f"Artist penetration records prepared: "
    f"{len(artist_penetration_features_df):,}"
)

print(
    f"Median observed markets per artist: "
    f"{artist_penetration_features_df['markets_observed'].median():,.1f}"
)

print(
    "Median artist typical country stream share: "
    f"{artist_penetration_features_df['median_country_stream_share_pct'].median():.6f}%"
)

print(
    "Median artist relative stream intensity: "
    f"{artist_penetration_features_df['median_relative_stream_intensity'].median():.4f}"
)

# ------------------------------------------------------------
# 11. Strongest artist market penetration
# ------------------------------------------------------------

top_artist_penetration_df = (
    artist_penetration_features_df
    .sort_values(
        [
            "maximum_country_stream_share_pct",
            "maximum_relative_stream_intensity",
            "markets_observed",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with strongest observed market penetration")
print("=" * 100)

display(
    top_artist_penetration_df[
        [
            "artists_raw",
            "markets_observed",
            "median_country_stream_share_pct",
            "maximum_country_stream_share_pct",
            "median_relative_stream_intensity",
            "maximum_relative_stream_intensity",
            "support_qualified_market_pct",
            "above_country_average_market_pct",
            "comparison_safe_market_pct",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Track penetration overview
# ------------------------------------------------------------

print()
print("Track market-penetration overview")
print("=" * 100)

print(
    f"Track penetration records prepared: "
    f"{len(track_penetration_features_df):,}"
)

print(
    f"Median observed markets per track: "
    f"{track_penetration_features_df['markets_observed'].median():,.1f}"
)

print(
    "Median track typical country stream share: "
    f"{track_penetration_features_df['median_country_stream_share_pct'].median():.6f}%"
)

print(
    "Median track relative stream intensity: "
    f"{track_penetration_features_df['median_relative_stream_intensity'].median():.4f}"
)

# ------------------------------------------------------------
# 13. Strongest track market penetration
# ------------------------------------------------------------

top_track_penetration_df = (
    track_penetration_features_df
    .sort_values(
        [
            "maximum_country_stream_share_pct",
            "maximum_relative_stream_intensity",
            "markets_observed",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with strongest observed market penetration")
print("=" * 100)

display(
    top_track_penetration_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "median_country_stream_share_pct",
            "maximum_country_stream_share_pct",
            "median_relative_stream_intensity",
            "maximum_relative_stream_intensity",
            "support_qualified_market_pct",
            "above_country_average_market_pct",
            "comparison_safe_market_pct",
        ]
    ]
)

# ------------------------------------------------------------
# 14. Feature validation
# ------------------------------------------------------------

artist_numeric_features = [
    "markets_observed",
    "median_country_observation_share_pct",
    "mean_country_observation_share_pct",
    "maximum_country_observation_share_pct",
    "median_country_stream_share_pct",
    "mean_country_stream_share_pct",
    "maximum_country_stream_share_pct",
    "median_relative_stream_intensity",
    "mean_relative_stream_intensity",
    "maximum_relative_stream_intensity",
    "median_market_chart_position",
    "best_market_chart_position",
    "support_qualified_markets",
    "comparison_safe_markets",
    "support_qualified_market_pct",
    "comparison_safe_market_pct",
    "markets_above_country_stream_average",
    "above_country_average_market_pct",
]

track_numeric_features = [
    "markets_observed",
    "median_country_observation_share_pct",
    "mean_country_observation_share_pct",
    "maximum_country_observation_share_pct",
    "median_country_stream_share_pct",
    "mean_country_stream_share_pct",
    "maximum_country_stream_share_pct",
    "median_relative_stream_intensity",
    "mean_relative_stream_intensity",
    "maximum_relative_stream_intensity",
    "median_market_chart_position",
    "best_market_chart_position",
    "support_qualified_markets",
    "comparison_safe_markets",
    "support_qualified_market_pct",
    "comparison_safe_market_pct",
    "markets_above_country_stream_average",
    "above_country_average_market_pct",
]

artist_features_finite = bool(
    np.isfinite(
        artist_penetration_features_df[
            artist_numeric_features
        ].to_numpy(
            dtype=float
        )
    ).all()
)

track_features_finite = bool(
    np.isfinite(
        track_penetration_features_df[
            track_numeric_features
        ].to_numpy(
            dtype=float
        )
    ).all()
)

artist_percentage_columns = [
    "median_country_observation_share_pct",
    "mean_country_observation_share_pct",
    "maximum_country_observation_share_pct",
    "median_country_stream_share_pct",
    "mean_country_stream_share_pct",
    "maximum_country_stream_share_pct",
    "support_qualified_market_pct",
    "comparison_safe_market_pct",
    "above_country_average_market_pct",
]

track_percentage_columns = [
    "median_country_observation_share_pct",
    "mean_country_observation_share_pct",
    "maximum_country_observation_share_pct",
    "median_country_stream_share_pct",
    "mean_country_stream_share_pct",
    "maximum_country_stream_share_pct",
    "support_qualified_market_pct",
    "comparison_safe_market_pct",
    "above_country_average_market_pct",
]

artist_percentages_valid = bool(
    artist_penetration_features_df[
        artist_percentage_columns
    ]
    .apply(
        lambda series: series.between(
            0,
            100
        ).all()
    )
    .all()
)

track_percentages_valid = bool(
    track_penetration_features_df[
        track_percentage_columns
    ]
    .apply(
        lambda series: series.between(
            0,
            100
        ).all()
    )
    .all()
)

section_5_2_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 5.1 completion",
        "Requirement": (
            "Country-reach features must be complete "
            "before market-penetration engineering."
        ),
        "Observed Evidence": (
            f"Section 5.1 completion status: "
            f"{section_5_1_complete}"
        ),
        "Passed": bool(
            section_5_1_complete
        ),
    },
    {
        "Validation Area": "Artist penetration completeness",
        "Requirement": (
            "Every validated artist representation must "
            "receive one penetration-feature record."
        ),
        "Observed Evidence": (
            f"{len(artist_penetration_features_df):,} records "
            f"for {artist_country_df['artists_raw'].nunique():,} artists"
        ),
        "Passed": (
            len(
                artist_penetration_features_df
            )
            ==
            artist_country_df[
                "artists_raw"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Track penetration completeness",
        "Requirement": (
            "Every validated track must receive "
            "one penetration-feature record."
        ),
        "Observed Evidence": (
            f"{len(track_penetration_features_df):,} records "
            f"for {track_country_df['track_id'].nunique():,} tracks"
        ),
        "Passed": (
            len(
                track_penetration_features_df
            )
            ==
            track_country_df[
                "track_id"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Artist feature validity",
        "Requirement": (
            "All engineered artist penetration features "
            "must remain finite."
        ),
        "Observed Evidence": (
            f"{len(artist_numeric_features)} "
            "numeric artist penetration features assessed"
        ),
        "Passed": artist_features_finite,
    },
    {
        "Validation Area": "Track feature validity",
        "Requirement": (
            "All engineered track penetration features "
            "must remain finite."
        ),
        "Observed Evidence": (
            f"{len(track_numeric_features)} "
            "numeric track penetration features assessed"
        ),
        "Passed": track_features_finite,
    },
    {
        "Validation Area": "Artist percentage bounds",
        "Requirement": (
            "Artist penetration percentage features must "
            "remain between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(artist_percentage_columns)} "
            "artist percentage fields assessed"
        ),
        "Passed": artist_percentages_valid,
    },
    {
        "Validation Area": "Track percentage bounds",
        "Requirement": (
            "Track penetration percentage features must "
            "remain between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(track_percentage_columns)} "
            "track percentage fields assessed"
        ),
        "Passed": track_percentages_valid,
    },
    {
        "Validation Area": "Artist reach reconciliation",
        "Requirement": (
            "Artist market counts must reconcile with "
            "the validated Section 5.1 reach table."
        ),
        "Observed Evidence": (
            "Artist market penetration counts compared "
            "with Section 5.1 country reach"
        ),
        "Passed": bool(
            artist_penetration_features_df[
                [
                    "artists_raw",
                    "markets_observed",
                ]
            ]
            .merge(
                artist_country_reach_features_df[
                    [
                        "artists_raw",
                        "country_markets_reached",
                    ]
                ],
                on="artists_raw",
                how="inner",
                validate="one_to_one",
            )
            .eval(
                "markets_observed == country_markets_reached"
            )
            .all()
        ),
    },
    {
        "Validation Area": "Track reach reconciliation",
        "Requirement": (
            "Track market counts must reconcile with "
            "the validated Section 5.1 reach table."
        ),
        "Observed Evidence": (
            "Track market penetration counts compared "
            "with Section 5.1 country reach"
        ),
        "Passed": bool(
            track_penetration_features_df[
                [
                    "track_id",
                    "markets_observed",
                ]
            ]
            .merge(
                track_country_reach_features_df[
                    [
                        "track_id",
                        "country_markets_reached",
                    ]
                ],
                on="track_id",
                how="inner",
                validate="one_to_one",
            )
            .eval(
                "markets_observed == country_markets_reached"
            )
            .all()
        ),
    },
])

print()
print("Market-penetration feature validation")
print("=" * 100)

display(
    section_5_2_validation_df
)

section_5_2_complete = bool(
    section_5_2_validation_df[
        "Passed"
    ].all()
)

if not section_5_2_complete:
    failed_checks = (
        section_5_2_validation_df.loc[
            ~section_5_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.2 market-penetration feature "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 5.2 market-penetration feature "
    "validation checks passed."
)

print(
    f"Section 5.2 completion status: "
    f"{section_5_2_complete}"
)

print(
    f"Artist market-penetration feature table prepared: "
    f"{len(artist_penetration_features_df):,} rows."
)

print(
    f"Track market-penetration feature table prepared: "
    f"{len(track_penetration_features_df):,} rows."
)

print(
    "Market-penetration features describe depth of observed "
    "presence within reached markets and do not yet represent "
    "a final PMIP geographic score."
)

print(
    "The workflow is ready for "
    "Section 5.3 Geographic Concentration Features."
)
```

### Interpretation

Section 5.2 successfully engineers market-penetration features for both artists and tracks using the validated 76-country PMIP market structure.

The resulting artist penetration table contains **49,843 artist representations**, while the track penetration table contains **110,198 tracks**, confirming complete preservation of the validated populations.

The median artist is observed in only **1 market**, with a median typical country stream share of approximately **0.004467%** and a median relative stream intensity of approximately **0.7669**.

This means that, for the typical artist representation, observed streaming intensity is below the wider average streaming intensity of the country market in which the artist appears.

The median track is also observed in only **1 market**, with a median typical country stream share of approximately **0.002844%** and a median relative stream intensity of approximately **0.7419**.

This again indicates that most tracks occupy only a very small share of their observed market and typically perform below the country-wide average streaming intensity.

However, the strongest market-penetration cases show a very different pattern.

Some artists and tracks achieve substantially larger shares of total country streams in particular markets, with the strongest observed artist cases exceeding **10% of a country's total observed streams** and some track-level cases exceeding **5%**.

Relative stream intensity further captures this local strength. Values above **1.0** indicate that an artist or track receives more streams per observation than the wider average observation in that country.

Several high-penetration tracks reach relative stream intensities well above this baseline, showing that their streaming activity is unusually strong within particular markets even when their wider geographic reach may be limited.

This distinction is important for PMIP:

- **Country reach** measures how many markets an entity reaches.
- **Market penetration** measures how deeply the entity performs within those reached markets.

An artist or track can therefore have narrow reach but very strong local penetration, or broad reach but relatively weak penetration within individual markets.

The support-qualified market features also preserve the analytical reliability rules established in Section 4. Market penetration measured from short or weakly supported track-country histories is kept separate from penetration supported by Moderate or Strong observation depth.

Similarly, comparison-safe market percentages retain the previously validated country-level comparison safeguards.

All Section 5.2 validation checks passed:

- all **49,843 artist representations** received penetration-feature records,
- all **110,198 tracks** received penetration-feature records,
- all engineered numerical features are finite,
- all percentage features remain within the valid **0–100%** range,
- artist market counts reconcile with the Section 5.1 country-reach features,
- track market counts reconcile with the Section 5.1 country-reach features.

The market-penetration feature layer is therefore validated and ready for downstream geographic intelligence.

These features remain descriptive and do not yet constitute a final PMIP geographic performance score.

The workflow is ready for **Section 5.3 Geographic Concentration Features**.

## 5.3 Geographic Concentration Features

### Purpose

This subsection engineers geographic-concentration features that measure whether an artist or track's observed streaming performance is broadly distributed across country markets or concentrated within a small number of markets.

The objective is to complement the country-reach and market-penetration features created in Sections 5.1 and 5.2.

Two artists may reach the same number of countries but still have very different geographic profiles. One may receive relatively balanced streaming support across many markets, while another may depend heavily on one or two dominant countries.

The analysis therefore develops features including:

- share of total observed streams generated by the strongest market,
- share generated by the top three markets,
- share generated by the top five markets,
- concentration of observations across markets,
- concentration of streams across markets,
- Herfindahl-Hirschman Index (HHI) for geographic stream concentration,
- effective number of markets implied by the concentration profile,
- concentration classifications for easier interpretation,
- support-aware concentration indicators.

These features describe how geographically diversified or concentrated an artist or track's observed performance is.

They remain descriptive engineered indicators and do not yet represent final PMIP geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Engineering PMIP geographic-concentration features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

required_objects = [
    "artist_market_penetration_df",
    "track_market_penetration_df",
    "artist_penetration_features_df",
    "track_penetration_features_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Section 5.3 requires the following Section 5.2 outputs: "
        + ", ".join(missing_objects)
    )

if not globals().get(
    "section_5_2_complete",
    False
):
    raise RuntimeError(
        "Section 5.2 is not complete. "
        "Run and validate Section 5.2 before Section 5.3."
    )

artist_market_df = artist_market_penetration_df.copy()
track_market_df = track_market_penetration_df.copy()

print(
    f"Section 5.2 completion status: "
    f"{section_5_2_complete}"
)

print(
    f"Artist-country rows available: "
    f"{len(artist_market_df):,}"
)

print(
    f"Track-country rows available: "
    f"{len(track_market_df):,}"
)

# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def top_n_share(values, n):
    values = pd.to_numeric(
        values,
        errors="coerce"
    )

    values = values[
        np.isfinite(values)
    ]

    total = values.sum()

    if total <= 0:
        return 0.0

    return float(
        values
        .nlargest(
            min(
                n,
                len(values)
            )
        )
        .sum()
        / total
        * 100
    )


def hhi_from_values(values):
    values = pd.to_numeric(
        values,
        errors="coerce"
    )

    values = values[
        np.isfinite(values)
    ]

    total = values.sum()

    if total <= 0:
        return 0.0

    shares = (
        values
        / total
    )

    return float(
        np.square(
            shares
        ).sum()
    )


def effective_market_count(hhi_value):
    if (
        not np.isfinite(hhi_value)
        or hhi_value <= 0
    ):
        return 0.0

    return float(
        1.0
        / hhi_value
    )


def concentration_class(hhi_value):
    if hhi_value >= 0.50:
        return "Very high concentration"

    if hhi_value >= 0.25:
        return "High concentration"

    if hhi_value >= 0.15:
        return "Moderate concentration"

    return "Low concentration"


# ------------------------------------------------------------
# 3. Artist stream-concentration features
# ------------------------------------------------------------

artist_stream_concentration_df = (
    artist_market_df
    .groupby(
        "artists_raw",
        dropna=False
    )
    .agg(
        markets_observed=(
            "country",
            "nunique"
        ),
        total_streams_across_markets=(
            "total_streams",
            "sum"
        ),
        total_observations_across_markets=(
            "observations",
            "sum"
        ),
        strongest_market_streams=(
            "total_streams",
            "max"
        ),
        strongest_market_observations=(
            "observations",
            "max"
        ),
    )
    .reset_index()
)

artist_stream_concentration_df[
    "top_1_stream_share_pct"
] = (
    artist_stream_concentration_df[
        "strongest_market_streams"
    ]
    / artist_stream_concentration_df[
        "total_streams_across_markets"
    ]
    * 100
)

artist_stream_concentration_df[
    "top_1_observation_share_pct"
] = (
    artist_stream_concentration_df[
        "strongest_market_observations"
    ]
    / artist_stream_concentration_df[
        "total_observations_across_markets"
    ]
    * 100
)

artist_top3_stream_share = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "total_streams"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            3
        )
    )
)

artist_top5_stream_share = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "total_streams"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            5
        )
    )
)

artist_top3_observation_share = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "observations"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            3
        )
    )
)

artist_top5_observation_share = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "observations"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            5
        )
    )
)

artist_stream_hhi = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "total_streams"
    ]
    .apply(
        hhi_from_values
    )
)

artist_observation_hhi = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "observations"
    ]
    .apply(
        hhi_from_values
    )
)

artist_stream_concentration_df[
    "top_3_stream_share_pct"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_top3_stream_share
    )
)

artist_stream_concentration_df[
    "top_5_stream_share_pct"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_top5_stream_share
    )
)

artist_stream_concentration_df[
    "top_3_observation_share_pct"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_top3_observation_share
    )
)

artist_stream_concentration_df[
    "top_5_observation_share_pct"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_top5_observation_share
    )
)

artist_stream_concentration_df[
    "stream_concentration_hhi"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_stream_hhi
    )
)

artist_stream_concentration_df[
    "observation_concentration_hhi"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_observation_hhi
    )
)

artist_stream_concentration_df[
    "effective_stream_markets"
] = (
    artist_stream_concentration_df[
        "stream_concentration_hhi"
    ]
    .apply(
        effective_market_count
    )
)

artist_stream_concentration_df[
    "effective_observation_markets"
] = (
    artist_stream_concentration_df[
        "observation_concentration_hhi"
    ]
    .apply(
        effective_market_count
    )
)

artist_stream_concentration_df[
    "geographic_concentration_class"
] = (
    artist_stream_concentration_df[
        "stream_concentration_hhi"
    ]
    .apply(
        concentration_class
    )
)

# ------------------------------------------------------------
# 4. Artist support-aware concentration
# ------------------------------------------------------------

artist_supported_df = (
    artist_market_df.loc[
        artist_market_df[
            "support_qualified_market"
        ]
    ]
    .copy()
)

artist_supported_hhi = (
    artist_supported_df
    .groupby(
        "artists_raw"
    )[
        "total_streams"
    ]
    .apply(
        hhi_from_values
    )
)

artist_supported_market_count = (
    artist_supported_df
    .groupby(
        "artists_raw"
    )[
        "country"
    ]
    .nunique()
)

artist_stream_concentration_df[
    "supported_market_count"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_supported_market_count
    )
    .fillna(0)
    .astype(int)
)

artist_stream_concentration_df[
    "supported_stream_concentration_hhi"
] = (
    artist_stream_concentration_df[
        "artists_raw"
    ]
    .map(
        artist_supported_hhi
    )
    .fillna(0.0)
)

# ------------------------------------------------------------
# 5. Track stream-concentration features
# ------------------------------------------------------------

track_stream_concentration_df = (
    track_market_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_name=(
            "track_name",
            "first"
        ),
        artists_raw=(
            "artists_raw",
            "first"
        ),
        markets_observed=(
            "country",
            "nunique"
        ),
        total_streams_across_markets=(
            "total_streams",
            "sum"
        ),
        total_observations_across_markets=(
            "observations",
            "sum"
        ),
        strongest_market_streams=(
            "total_streams",
            "max"
        ),
        strongest_market_observations=(
            "observations",
            "max"
        ),
    )
    .reset_index()
)

track_stream_concentration_df[
    "top_1_stream_share_pct"
] = (
    track_stream_concentration_df[
        "strongest_market_streams"
    ]
    / track_stream_concentration_df[
        "total_streams_across_markets"
    ]
    * 100
)

track_stream_concentration_df[
    "top_1_observation_share_pct"
] = (
    track_stream_concentration_df[
        "strongest_market_observations"
    ]
    / track_stream_concentration_df[
        "total_observations_across_markets"
    ]
    * 100
)

track_top3_stream_share = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "total_streams"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            3
        )
    )
)

track_top5_stream_share = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "total_streams"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            5
        )
    )
)

track_top3_observation_share = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "observations"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            3
        )
    )
)

track_top5_observation_share = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "observations"
    ]
    .apply(
        lambda s: top_n_share(
            s,
            5
        )
    )
)

track_stream_hhi = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "total_streams"
    ]
    .apply(
        hhi_from_values
    )
)

track_observation_hhi = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "observations"
    ]
    .apply(
        hhi_from_values
    )
)

track_stream_concentration_df[
    "top_3_stream_share_pct"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_top3_stream_share
    )
)

track_stream_concentration_df[
    "top_5_stream_share_pct"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_top5_stream_share
    )
)

track_stream_concentration_df[
    "top_3_observation_share_pct"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_top3_observation_share
    )
)

track_stream_concentration_df[
    "top_5_observation_share_pct"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_top5_observation_share
    )
)

track_stream_concentration_df[
    "stream_concentration_hhi"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_stream_hhi
    )
)

track_stream_concentration_df[
    "observation_concentration_hhi"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_observation_hhi
    )
)

track_stream_concentration_df[
    "effective_stream_markets"
] = (
    track_stream_concentration_df[
        "stream_concentration_hhi"
    ]
    .apply(
        effective_market_count
    )
)

track_stream_concentration_df[
    "effective_observation_markets"
] = (
    track_stream_concentration_df[
        "observation_concentration_hhi"
    ]
    .apply(
        effective_market_count
    )
)

track_stream_concentration_df[
    "geographic_concentration_class"
] = (
    track_stream_concentration_df[
        "stream_concentration_hhi"
    ]
    .apply(
        concentration_class
    )
)

# ------------------------------------------------------------
# 6. Track support-aware concentration
# ------------------------------------------------------------

track_supported_df = (
    track_market_df.loc[
        track_market_df[
            "support_qualified_market"
        ]
    ]
    .copy()
)

track_supported_hhi = (
    track_supported_df
    .groupby(
        "track_id"
    )[
        "total_streams"
    ]
    .apply(
        hhi_from_values
    )
)

track_supported_market_count = (
    track_supported_df
    .groupby(
        "track_id"
    )[
        "country"
    ]
    .nunique()
)

track_stream_concentration_df[
    "supported_market_count"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_supported_market_count
    )
    .fillna(0)
    .astype(int)
)

track_stream_concentration_df[
    "supported_stream_concentration_hhi"
] = (
    track_stream_concentration_df[
        "track_id"
    ]
    .map(
        track_supported_hhi
    )
    .fillna(0.0)
)

# ------------------------------------------------------------
# 7. Artist concentration overview
# ------------------------------------------------------------

print()
print("Artist geographic-concentration overview")
print("=" * 100)

print(
    f"Artist concentration records prepared: "
    f"{len(artist_stream_concentration_df):,}"
)

print(
    "Median strongest-market stream share: "
    f"{artist_stream_concentration_df['top_1_stream_share_pct'].median():.4f}%"
)

print(
    "Median top-three market stream share: "
    f"{artist_stream_concentration_df['top_3_stream_share_pct'].median():.4f}%"
)

print(
    "Median stream-concentration HHI: "
    f"{artist_stream_concentration_df['stream_concentration_hhi'].median():.4f}"
)

print(
    "Median effective stream markets: "
    f"{artist_stream_concentration_df['effective_stream_markets'].median():.2f}"
)

artist_concentration_distribution_df = (
    artist_stream_concentration_df[
        "geographic_concentration_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Concentration Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_concentration_distribution_df[
    "Share (%)"
] = (
    artist_concentration_distribution_df[
        "Artist Representations"
    ]
    / len(
        artist_stream_concentration_df
    )
    * 100
)

print()
print("Artist geographic-concentration classification")
print("=" * 100)

display(
    artist_concentration_distribution_df
)

# ------------------------------------------------------------
# 8. Most geographically diversified artists
# ------------------------------------------------------------

most_diversified_artists_df = (
    artist_stream_concentration_df
    .loc[
        artist_stream_concentration_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "markets_observed",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with most geographically diversified streaming")
print("=" * 100)

display(
    most_diversified_artists_df[
        [
            "artists_raw",
            "markets_observed",
            "top_1_stream_share_pct",
            "top_3_stream_share_pct",
            "top_5_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "supported_market_count",
            "supported_stream_concentration_hhi",
            "geographic_concentration_class",
        ]
    ]
)

# ------------------------------------------------------------
# 9. Most geographically concentrated artists
# ------------------------------------------------------------

most_concentrated_artists_df = (
    artist_stream_concentration_df
    .loc[
        artist_stream_concentration_df[
            "markets_observed"
        ] >= 2
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "top_1_stream_share_pct",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with strongest geographic concentration")
print("=" * 100)

display(
    most_concentrated_artists_df[
        [
            "artists_raw",
            "markets_observed",
            "top_1_stream_share_pct",
            "top_3_stream_share_pct",
            "top_5_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "supported_market_count",
            "supported_stream_concentration_hhi",
            "geographic_concentration_class",
        ]
    ]
)

# ------------------------------------------------------------
# 10. Track concentration overview
# ------------------------------------------------------------

print()
print("Track geographic-concentration overview")
print("=" * 100)

print(
    f"Track concentration records prepared: "
    f"{len(track_stream_concentration_df):,}"
)

print(
    "Median strongest-market stream share: "
    f"{track_stream_concentration_df['top_1_stream_share_pct'].median():.4f}%"
)

print(
    "Median top-three market stream share: "
    f"{track_stream_concentration_df['top_3_stream_share_pct'].median():.4f}%"
)

print(
    "Median stream-concentration HHI: "
    f"{track_stream_concentration_df['stream_concentration_hhi'].median():.4f}"
)

print(
    "Median effective stream markets: "
    f"{track_stream_concentration_df['effective_stream_markets'].median():.2f}"
)

track_concentration_distribution_df = (
    track_stream_concentration_df[
        "geographic_concentration_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Concentration Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_concentration_distribution_df[
    "Share (%)"
] = (
    track_concentration_distribution_df[
        "Tracks"
    ]
    / len(
        track_stream_concentration_df
    )
    * 100
)

print()
print("Track geographic-concentration classification")
print("=" * 100)

display(
    track_concentration_distribution_df
)

# ------------------------------------------------------------
# 11. Most geographically diversified tracks
# ------------------------------------------------------------

most_diversified_tracks_df = (
    track_stream_concentration_df
    .loc[
        track_stream_concentration_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "markets_observed",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with most geographically diversified streaming")
print("=" * 100)

display(
    most_diversified_tracks_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "top_1_stream_share_pct",
            "top_3_stream_share_pct",
            "top_5_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "supported_market_count",
            "supported_stream_concentration_hhi",
            "geographic_concentration_class",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Most geographically concentrated tracks
# ------------------------------------------------------------

most_concentrated_tracks_df = (
    track_stream_concentration_df
    .loc[
        track_stream_concentration_df[
            "markets_observed"
        ] >= 2
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "top_1_stream_share_pct",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with strongest geographic concentration")
print("=" * 100)

display(
    most_concentrated_tracks_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "top_1_stream_share_pct",
            "top_3_stream_share_pct",
            "top_5_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "supported_market_count",
            "supported_stream_concentration_hhi",
            "geographic_concentration_class",
        ]
    ]
)

# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

artist_numeric_features = [
    "markets_observed",
    "total_streams_across_markets",
    "total_observations_across_markets",
    "strongest_market_streams",
    "strongest_market_observations",
    "top_1_stream_share_pct",
    "top_1_observation_share_pct",
    "top_3_stream_share_pct",
    "top_5_stream_share_pct",
    "top_3_observation_share_pct",
    "top_5_observation_share_pct",
    "stream_concentration_hhi",
    "observation_concentration_hhi",
    "effective_stream_markets",
    "effective_observation_markets",
    "supported_market_count",
    "supported_stream_concentration_hhi",
]

track_numeric_features = [
    "markets_observed",
    "total_streams_across_markets",
    "total_observations_across_markets",
    "strongest_market_streams",
    "strongest_market_observations",
    "top_1_stream_share_pct",
    "top_1_observation_share_pct",
    "top_3_stream_share_pct",
    "top_5_stream_share_pct",
    "top_3_observation_share_pct",
    "top_5_observation_share_pct",
    "stream_concentration_hhi",
    "observation_concentration_hhi",
    "effective_stream_markets",
    "effective_observation_markets",
    "supported_market_count",
    "supported_stream_concentration_hhi",
]

artist_features_finite = bool(
    np.isfinite(
        artist_stream_concentration_df[
            artist_numeric_features
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

track_features_finite = bool(
    np.isfinite(
        track_stream_concentration_df[
            track_numeric_features
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

artist_share_columns = [
    "top_1_stream_share_pct",
    "top_1_observation_share_pct",
    "top_3_stream_share_pct",
    "top_5_stream_share_pct",
    "top_3_observation_share_pct",
    "top_5_observation_share_pct",
]

track_share_columns = [
    "top_1_stream_share_pct",
    "top_1_observation_share_pct",
    "top_3_stream_share_pct",
    "top_5_stream_share_pct",
    "top_3_observation_share_pct",
    "top_5_observation_share_pct",
]

artist_share_bounds_valid = bool(
    artist_stream_concentration_df[
        artist_share_columns
    ]
    .apply(
        lambda s: s.between(
            0,
            100
        ).all()
    )
    .all()
)

track_share_bounds_valid = bool(
    track_stream_concentration_df[
        track_share_columns
    ]
    .apply(
        lambda s: s.between(
            0,
            100
        ).all()
    )
    .all()
)

artist_hhi_bounds_valid = bool(
    artist_stream_concentration_df[
        [
            "stream_concentration_hhi",
            "observation_concentration_hhi",
            "supported_stream_concentration_hhi",
        ]
    ]
    .apply(
        lambda s: s.between(
            0,
            1
        ).all()
    )
    .all()
)

track_hhi_bounds_valid = bool(
    track_stream_concentration_df[
        [
            "stream_concentration_hhi",
            "observation_concentration_hhi",
            "supported_stream_concentration_hhi",
        ]
    ]
    .apply(
        lambda s: s.between(
            0,
            1
        ).all()
    )
    .all()
)

section_5_3_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 5.2 completion",
        "Requirement": (
            "Market-penetration features must be complete "
            "before geographic-concentration engineering."
        ),
        "Observed Evidence": (
            f"Section 5.2 completion status: "
            f"{section_5_2_complete}"
        ),
        "Passed": bool(
            section_5_2_complete
        ),
    },
    {
        "Validation Area": "Artist concentration completeness",
        "Requirement": (
            "Every validated artist representation must "
            "receive one concentration-feature record."
        ),
        "Observed Evidence": (
            f"{len(artist_stream_concentration_df):,} records "
            f"for "
            f"{artist_market_df['artists_raw'].nunique():,} artists"
        ),
        "Passed": (
            len(
                artist_stream_concentration_df
            )
            ==
            artist_market_df[
                "artists_raw"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Track concentration completeness",
        "Requirement": (
            "Every validated track must receive "
            "one concentration-feature record."
        ),
        "Observed Evidence": (
            f"{len(track_stream_concentration_df):,} records "
            f"for "
            f"{track_market_df['track_id'].nunique():,} tracks"
        ),
        "Passed": (
            len(
                track_stream_concentration_df
            )
            ==
            track_market_df[
                "track_id"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Artist feature validity",
        "Requirement": (
            "All artist concentration features "
            "must remain finite."
        ),
        "Observed Evidence": (
            f"{len(artist_numeric_features)} "
            "numeric artist concentration features assessed"
        ),
        "Passed": artist_features_finite,
    },
    {
        "Validation Area": "Track feature validity",
        "Requirement": (
            "All track concentration features "
            "must remain finite."
        ),
        "Observed Evidence": (
            f"{len(track_numeric_features)} "
            "numeric track concentration features assessed"
        ),
        "Passed": track_features_finite,
    },
    {
        "Validation Area": "Artist share bounds",
        "Requirement": (
            "Artist concentration-share features must "
            "remain between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(artist_share_columns)} "
            "artist share fields assessed"
        ),
        "Passed": artist_share_bounds_valid,
    },
    {
        "Validation Area": "Track share bounds",
        "Requirement": (
            "Track concentration-share features must "
            "remain between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(track_share_columns)} "
            "track share fields assessed"
        ),
        "Passed": track_share_bounds_valid,
    },
    {
        "Validation Area": "Artist HHI bounds",
        "Requirement": (
            "Artist HHI concentration values must "
            "remain between 0 and 1."
        ),
        "Observed Evidence": (
            "Artist stream, observation and supported "
            "HHI fields assessed"
        ),
        "Passed": artist_hhi_bounds_valid,
    },
    {
        "Validation Area": "Track HHI bounds",
        "Requirement": (
            "Track HHI concentration values must "
            "remain between 0 and 1."
        ),
        "Observed Evidence": (
            "Track stream, observation and supported "
            "HHI fields assessed"
        ),
        "Passed": track_hhi_bounds_valid,
    },
])

print()
print("Geographic-concentration feature validation")
print("=" * 100)

display(
    section_5_3_validation_df
)

section_5_3_complete = bool(
    section_5_3_validation_df[
        "Passed"
    ].all()
)

if not section_5_3_complete:
    failed_checks = (
        section_5_3_validation_df.loc[
            ~section_5_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.3 geographic-concentration feature "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 5.3 geographic-concentration feature "
    "validation checks passed."
)

print(
    f"Section 5.3 completion status: "
    f"{section_5_3_complete}"
)

print(
    f"Artist geographic-concentration feature table prepared: "
    f"{len(artist_stream_concentration_df):,} rows."
)

print(
    f"Track geographic-concentration feature table prepared: "
    f"{len(track_stream_concentration_df):,} rows."
)

print(
    "Geographic-concentration features describe how strongly "
    "observed streaming activity is concentrated across markets."
)

print(
    "The workflow is ready for "
    "Section 5.4 Cross-Market Performance Features."
)
```

### Interpretation

Section 5.3 successfully engineers geographic-concentration features for both artists and tracks.

The artist concentration table contains **49,843 artist representations**, while the track concentration table contains **110,198 tracks**, confirming that the full validated entity populations were preserved.

The artist-level results show a strongly concentrated geographic structure.

The median strongest-market stream share is **100.00%**, the median top-three market stream share is also **100.00%**, the median stream-concentration HHI is **1.0000**, and the median effective number of stream markets is **1.00**.

This pattern is consistent with the earlier geographic-coverage findings, where most artists were observed in only one market.

The concentration classification confirms this strongly:

- **45,216 artist representations (90.72%)** are classified as Very High Concentration.
- **2,439 (4.89%)** are classified as High Concentration.
- **1,357 (2.72%)** are classified as Moderate Concentration.
- **831 (1.67%)** are classified as Low Concentration.

This means that the overwhelming majority of artist representations in the historical PMIP dataset are geographically dependent on a very small number of markets.

An HHI close to **1.0** indicates that nearly all observed streaming activity is concentrated in one market, while lower HHI values indicate that streaming activity is distributed more evenly across several markets.

The effective-market feature provides the same idea in a more intuitive form. An effective market count of **1.00** means that the observed streaming profile behaves as though almost all activity comes from one equally weighted market.

These results should not be interpreted as evidence that most artists are genuinely local-only in the real world. They describe the geographic structure present in the historical PMIP dataset and are influenced by the strong single-market representation identified earlier in Sections 3 and 5.1.

The concentration features therefore add a different analytical dimension from reach and penetration:

- **Country reach** describes how many markets an artist or track appears in.
- **Market penetration** describes how strongly the entity performs within those markets.
- **Geographic concentration** describes how dependent the entity's total observed activity is on its strongest markets.

This distinction is important because two entities with similar geographic reach may still have very different concentration profiles. One may distribute streaming activity relatively evenly across several markets, while another may depend heavily on one dominant country.

Support-aware concentration features also retain the analytical-support safeguards from Section 4, allowing PMIP to distinguish concentration measured across all observed markets from concentration supported by stronger observation histories.

All Section 5.3 validation checks passed, confirming that:

- all validated artist and track populations received concentration-feature records,
- all engineered concentration values remained finite,
- all market-share features remained within valid **0–100%** boundaries,
- all HHI values remained within the valid **0–1** range.

The geographic-concentration feature layer is therefore validated and ready for downstream analysis.

These concentration features remain descriptive engineered indicators and do not yet constitute a final PMIP geographic performance score.

The workflow is ready for **Section 5.4 Cross-Market Performance Features**.

## 5.4 Cross-Market Performance Features

### Purpose

This subsection engineers cross-market performance features that compare how consistently an artist or track performs across the different country markets in which it appears.

The objective is to move beyond simple reach, penetration and concentration by measuring the degree of variation between markets.

The analysis will engineer features describing:

- variation in streaming performance across markets,
- variation in chart-position performance across markets,
- strongest versus weakest market performance,
- best-market and worst-market differences,
- cross-market consistency,
- cross-market performance spread,
- proportion of markets performing above an entity's own typical level,
- support-aware comparison features,
- comparison-safe market coverage.

The previously validated geographic safeguards remain in force.

Cross-market comparisons are therefore restricted by:

- market support quality,
- direct-comparison caution flags,
- historical and temporal alignment limitations.

These engineered features remain descriptive and do not yet represent final PMIP geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Engineering PMIP cross-market performance features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Prerequisite validation
# ------------------------------------------------------------

required_objects = [
    "artist_market_penetration_df",
    "track_market_penetration_df",
    "artist_stream_concentration_df",
    "track_stream_concentration_df",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Section 5.4 requires the following previous outputs: "
        + ", ".join(missing_objects)
    )

if not globals().get(
    "section_5_3_complete",
    False
):
    raise RuntimeError(
        "Section 5.3 is not complete. "
        "Run and validate Section 5.3 before Section 5.4."
    )

artist_market_df = artist_market_penetration_df.copy()
track_market_df = track_market_penetration_df.copy()

print(
    f"Section 5.3 completion status: "
    f"{section_5_3_complete}"
)

print(
    f"Artist-country rows available: "
    f"{len(artist_market_df):,}"
)

print(
    f"Track-country rows available: "
    f"{len(track_market_df):,}"
)

# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def safe_std(series):
    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    series = series[
        np.isfinite(series)
    ]

    if len(series) <= 1:
        return 0.0

    return float(
        series.std(
            ddof=0
        )
    )


def safe_cv(series):
    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    series = series[
        np.isfinite(series)
    ]

    if len(series) <= 1:
        return 0.0

    mean_value = float(
        series.mean()
    )

    if mean_value == 0:
        return 0.0

    return float(
        series.std(
            ddof=0
        )
        / abs(
            mean_value
        )
    )


def above_own_median_pct(series):
    series = pd.to_numeric(
        series,
        errors="coerce"
    )

    series = series[
        np.isfinite(series)
    ]

    if len(series) == 0:
        return 0.0

    median_value = float(
        series.median()
    )

    return float(
        (
            series > median_value
        ).mean()
        * 100
    )


def performance_consistency_class(cv_value):
    if cv_value <= 0.25:
        return "Very consistent"

    if cv_value <= 0.50:
        return "Consistent"

    if cv_value <= 1.00:
        return "Variable"

    return "Highly variable"


# ------------------------------------------------------------
# 3. Comparison-safe flags
# ------------------------------------------------------------

artist_market_df[
    "comparison_eligible_market"
] = (
    artist_market_df[
        "support_qualified_market"
    ].astype(bool)
    &
    ~artist_market_df[
        "direct_comparison_caution"
    ].astype(bool)
)

track_market_df[
    "comparison_eligible_market"
] = (
    track_market_df[
        "support_qualified_market"
    ].astype(bool)
    &
    ~track_market_df[
        "direct_comparison_caution"
    ].astype(bool)
)

# ------------------------------------------------------------
# 4. Artist cross-market features
# ------------------------------------------------------------

artist_cross_market_features_df = (
    artist_market_df
    .groupby(
        "artists_raw",
        dropna=False
    )
    .agg(
        markets_observed=(
            "country",
            "nunique"
        ),
        mean_market_streams=(
            "median_streams",
            "mean"
        ),
        median_market_streams=(
            "median_streams",
            "median"
        ),
        strongest_market_streams=(
            "median_streams",
            "max"
        ),
        weakest_market_streams=(
            "median_streams",
            "min"
        ),
        mean_chart_position=(
            "median_chart_position",
            "mean"
        ),
        median_chart_position=(
            "median_chart_position",
            "median"
        ),
        best_market_chart_position=(
            "median_chart_position",
            "min"
        ),
        worst_market_chart_position=(
            "median_chart_position",
            "max"
        ),
        comparison_eligible_markets=(
            "comparison_eligible_market",
            "sum"
        ),
    )
    .reset_index()
)

artist_stream_std = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "median_streams"
    ]
    .apply(
        safe_std
    )
)

artist_stream_cv = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "median_streams"
    ]
    .apply(
        safe_cv
    )
)

artist_chart_std = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "median_chart_position"
    ]
    .apply(
        safe_std
    )
)

artist_chart_cv = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "median_chart_position"
    ]
    .apply(
        safe_cv
    )
)

artist_above_median_stream_pct = (
    artist_market_df
    .groupby(
        "artists_raw"
    )[
        "median_streams"
    ]
    .apply(
        above_own_median_pct
    )
)

artist_cross_market_features_df[
    "stream_std_across_markets"
] = (
    artist_cross_market_features_df[
        "artists_raw"
    ]
    .map(
        artist_stream_std
    )
)

artist_cross_market_features_df[
    "stream_cv_across_markets"
] = (
    artist_cross_market_features_df[
        "artists_raw"
    ]
    .map(
        artist_stream_cv
    )
)

artist_cross_market_features_df[
    "chart_position_std_across_markets"
] = (
    artist_cross_market_features_df[
        "artists_raw"
    ]
    .map(
        artist_chart_std
    )
)

artist_cross_market_features_df[
    "chart_position_cv_across_markets"
] = (
    artist_cross_market_features_df[
        "artists_raw"
    ]
    .map(
        artist_chart_cv
    )
)

artist_cross_market_features_df[
    "above_own_median_stream_market_pct"
] = (
    artist_cross_market_features_df[
        "artists_raw"
    ]
    .map(
        artist_above_median_stream_pct
    )
)

artist_cross_market_features_df[
    "stream_performance_range"
] = (
    artist_cross_market_features_df[
        "strongest_market_streams"
    ]
    -
    artist_cross_market_features_df[
        "weakest_market_streams"
    ]
)

artist_cross_market_features_df[
    "chart_position_range"
] = (
    artist_cross_market_features_df[
        "worst_market_chart_position"
    ]
    -
    artist_cross_market_features_df[
        "best_market_chart_position"
    ]
)

artist_cross_market_features_df[
    "comparison_eligible_market_pct"
] = (
    artist_cross_market_features_df[
        "comparison_eligible_markets"
    ]
    / artist_cross_market_features_df[
        "markets_observed"
    ]
    * 100
)

artist_cross_market_features_df[
    "stream_consistency_class"
] = (
    artist_cross_market_features_df[
        "stream_cv_across_markets"
    ]
    .apply(
        performance_consistency_class
    )
)

# ------------------------------------------------------------
# 5. Track cross-market features
# ------------------------------------------------------------

track_cross_market_features_df = (
    track_market_df
    .groupby(
        "track_id",
        dropna=False
    )
    .agg(
        track_name=(
            "track_name",
            "first"
        ),
        artists_raw=(
            "artists_raw",
            "first"
        ),
        markets_observed=(
            "country",
            "nunique"
        ),
        mean_market_streams=(
            "median_streams",
            "mean"
        ),
        median_market_streams=(
            "median_streams",
            "median"
        ),
        strongest_market_streams=(
            "median_streams",
            "max"
        ),
        weakest_market_streams=(
            "median_streams",
            "min"
        ),
        mean_chart_position=(
            "median_chart_position",
            "mean"
        ),
        median_chart_position=(
            "median_chart_position",
            "median"
        ),
        best_market_chart_position=(
            "median_chart_position",
            "min"
        ),
        worst_market_chart_position=(
            "median_chart_position",
            "max"
        ),
        comparison_eligible_markets=(
            "comparison_eligible_market",
            "sum"
        ),
    )
    .reset_index()
)

track_stream_std = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "median_streams"
    ]
    .apply(
        safe_std
    )
)

track_stream_cv = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "median_streams"
    ]
    .apply(
        safe_cv
    )
)

track_chart_std = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "median_chart_position"
    ]
    .apply(
        safe_std
    )
)

track_chart_cv = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "median_chart_position"
    ]
    .apply(
        safe_cv
    )
)

track_above_median_stream_pct = (
    track_market_df
    .groupby(
        "track_id"
    )[
        "median_streams"
    ]
    .apply(
        above_own_median_pct
    )
)

track_cross_market_features_df[
    "stream_std_across_markets"
] = (
    track_cross_market_features_df[
        "track_id"
    ]
    .map(
        track_stream_std
    )
)

track_cross_market_features_df[
    "stream_cv_across_markets"
] = (
    track_cross_market_features_df[
        "track_id"
    ]
    .map(
        track_stream_cv
    )
)

track_cross_market_features_df[
    "chart_position_std_across_markets"
] = (
    track_cross_market_features_df[
        "track_id"
    ]
    .map(
        track_chart_std
    )
)

track_cross_market_features_df[
    "chart_position_cv_across_markets"
] = (
    track_cross_market_features_df[
        "track_id"
    ]
    .map(
        track_chart_cv
    )
)

track_cross_market_features_df[
    "above_own_median_stream_market_pct"
] = (
    track_cross_market_features_df[
        "track_id"
    ]
    .map(
        track_above_median_stream_pct
    )
)

track_cross_market_features_df[
    "stream_performance_range"
] = (
    track_cross_market_features_df[
        "strongest_market_streams"
    ]
    -
    track_cross_market_features_df[
        "weakest_market_streams"
    ]
)

track_cross_market_features_df[
    "chart_position_range"
] = (
    track_cross_market_features_df[
        "worst_market_chart_position"
    ]
    -
    track_cross_market_features_df[
        "best_market_chart_position"
    ]
)

track_cross_market_features_df[
    "comparison_eligible_market_pct"
] = (
    track_cross_market_features_df[
        "comparison_eligible_markets"
    ]
    / track_cross_market_features_df[
        "markets_observed"
    ]
    * 100
)

track_cross_market_features_df[
    "stream_consistency_class"
] = (
    track_cross_market_features_df[
        "stream_cv_across_markets"
    ]
    .apply(
        performance_consistency_class
    )
)

# ------------------------------------------------------------
# 6. Artist cross-market overview
# ------------------------------------------------------------

print()
print("Artist cross-market performance overview")
print("=" * 100)

print(
    f"Artist records prepared: "
    f"{len(artist_cross_market_features_df):,}"
)

print(
    "Median stream CV across markets: "
    f"{artist_cross_market_features_df['stream_cv_across_markets'].median():.4f}"
)

print(
    "Median stream performance range: "
    f"{artist_cross_market_features_df['stream_performance_range'].median():,.2f}"
)

print(
    "Median chart-position range: "
    f"{artist_cross_market_features_df['chart_position_range'].median():.2f}"
)

print(
    "Median comparison-eligible market share: "
    f"{artist_cross_market_features_df['comparison_eligible_market_pct'].median():.2f}%"
)

artist_consistency_distribution_df = (
    artist_cross_market_features_df[
        "stream_consistency_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Consistency Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_consistency_distribution_df[
    "Share (%)"
] = (
    artist_consistency_distribution_df[
        "Artist Representations"
    ]
    / len(
        artist_cross_market_features_df
    )
    * 100
)

print()
print("Artist stream-consistency classification")
print("=" * 100)

display(
    artist_consistency_distribution_df
)

# ------------------------------------------------------------
# 7. Most consistent multi-market artists
# ------------------------------------------------------------

most_consistent_artists_df = (
    artist_cross_market_features_df
    .loc[
        artist_cross_market_features_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_cv_across_markets",
            "markets_observed",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with most consistent cross-market streaming")
print("=" * 100)

display(
    most_consistent_artists_df[
        [
            "artists_raw",
            "markets_observed",
            "median_market_streams",
            "strongest_market_streams",
            "weakest_market_streams",
            "stream_performance_range",
            "stream_cv_across_markets",
            "chart_position_range",
            "comparison_eligible_market_pct",
            "stream_consistency_class",
        ]
    ]
)

# ------------------------------------------------------------
# 8. Most variable multi-market artists
# ------------------------------------------------------------

most_variable_artists_df = (
    artist_cross_market_features_df
    .loc[
        artist_cross_market_features_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_cv_across_markets",
            "stream_performance_range",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Artist representations with greatest cross-market variation")
print("=" * 100)

display(
    most_variable_artists_df[
        [
            "artists_raw",
            "markets_observed",
            "median_market_streams",
            "strongest_market_streams",
            "weakest_market_streams",
            "stream_performance_range",
            "stream_cv_across_markets",
            "chart_position_range",
            "comparison_eligible_market_pct",
            "stream_consistency_class",
        ]
    ]
)

# ------------------------------------------------------------
# 9. Track cross-market overview
# ------------------------------------------------------------

print()
print("Track cross-market performance overview")
print("=" * 100)

print(
    f"Track records prepared: "
    f"{len(track_cross_market_features_df):,}"
)

print(
    "Median stream CV across markets: "
    f"{track_cross_market_features_df['stream_cv_across_markets'].median():.4f}"
)

print(
    "Median stream performance range: "
    f"{track_cross_market_features_df['stream_performance_range'].median():,.2f}"
)

print(
    "Median chart-position range: "
    f"{track_cross_market_features_df['chart_position_range'].median():.2f}"
)

print(
    "Median comparison-eligible market share: "
    f"{track_cross_market_features_df['comparison_eligible_market_pct'].median():.2f}%"
)

track_consistency_distribution_df = (
    track_cross_market_features_df[
        "stream_consistency_class"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "Consistency Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_consistency_distribution_df[
    "Share (%)"
] = (
    track_consistency_distribution_df[
        "Tracks"
    ]
    / len(
        track_cross_market_features_df
    )
    * 100
)

print()
print("Track stream-consistency classification")
print("=" * 100)

display(
    track_consistency_distribution_df
)

# ------------------------------------------------------------
# 10. Most consistent multi-market tracks
# ------------------------------------------------------------

most_consistent_tracks_df = (
    track_cross_market_features_df
    .loc[
        track_cross_market_features_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_cv_across_markets",
            "markets_observed",
        ],
        ascending=[
            True,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with most consistent cross-market streaming")
print("=" * 100)

display(
    most_consistent_tracks_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "median_market_streams",
            "strongest_market_streams",
            "weakest_market_streams",
            "stream_performance_range",
            "stream_cv_across_markets",
            "chart_position_range",
            "comparison_eligible_market_pct",
            "stream_consistency_class",
        ]
    ]
)

# ------------------------------------------------------------
# 11. Most variable multi-market tracks
# ------------------------------------------------------------

most_variable_tracks_df = (
    track_cross_market_features_df
    .loc[
        track_cross_market_features_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_cv_across_markets",
            "stream_performance_range",
        ],
        ascending=[
            False,
            False,
        ]
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with greatest cross-market variation")
print("=" * 100)

display(
    most_variable_tracks_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "median_market_streams",
            "strongest_market_streams",
            "weakest_market_streams",
            "stream_performance_range",
            "stream_cv_across_markets",
            "chart_position_range",
            "comparison_eligible_market_pct",
            "stream_consistency_class",
        ]
    ]
)

# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

artist_numeric_features = [
    "markets_observed",
    "mean_market_streams",
    "median_market_streams",
    "strongest_market_streams",
    "weakest_market_streams",
    "mean_chart_position",
    "median_chart_position",
    "best_market_chart_position",
    "worst_market_chart_position",
    "comparison_eligible_markets",
    "stream_std_across_markets",
    "stream_cv_across_markets",
    "chart_position_std_across_markets",
    "chart_position_cv_across_markets",
    "above_own_median_stream_market_pct",
    "stream_performance_range",
    "chart_position_range",
    "comparison_eligible_market_pct",
]

track_numeric_features = [
    "markets_observed",
    "mean_market_streams",
    "median_market_streams",
    "strongest_market_streams",
    "weakest_market_streams",
    "mean_chart_position",
    "median_chart_position",
    "best_market_chart_position",
    "worst_market_chart_position",
    "comparison_eligible_markets",
    "stream_std_across_markets",
    "stream_cv_across_markets",
    "chart_position_std_across_markets",
    "chart_position_cv_across_markets",
    "above_own_median_stream_market_pct",
    "stream_performance_range",
    "chart_position_range",
    "comparison_eligible_market_pct",
]

artist_features_finite = bool(
    np.isfinite(
        artist_cross_market_features_df[
            artist_numeric_features
        ].to_numpy(
            dtype=float
        )
    ).all()
)

track_features_finite = bool(
    np.isfinite(
        track_cross_market_features_df[
            track_numeric_features
        ].to_numpy(
            dtype=float
        )
    ).all()
)

artist_percentage_columns = [
    "above_own_median_stream_market_pct",
    "comparison_eligible_market_pct",
]

track_percentage_columns = [
    "above_own_median_stream_market_pct",
    "comparison_eligible_market_pct",
]

artist_percentages_valid = bool(
    artist_cross_market_features_df[
        artist_percentage_columns
    ]
    .apply(
        lambda s: s.between(
            0,
            100
        ).all()
    )
    .all()
)

track_percentages_valid = bool(
    track_cross_market_features_df[
        track_percentage_columns
    ]
    .apply(
        lambda s: s.between(
            0,
            100
        ).all()
    )
    .all()
)

artist_ranges_valid = bool(
    (
        artist_cross_market_features_df[
            "stream_performance_range"
        ] >= 0
    ).all()
    and
    (
        artist_cross_market_features_df[
            "chart_position_range"
        ] >= 0
    ).all()
)

track_ranges_valid = bool(
    (
        track_cross_market_features_df[
            "stream_performance_range"
        ] >= 0
    ).all()
    and
    (
        track_cross_market_features_df[
            "chart_position_range"
        ] >= 0
    ).all()
)

section_5_4_validation_df = pd.DataFrame([
    {
        "Validation Area": "Section 5.3 completion",
        "Requirement": (
            "Geographic concentration features must be "
            "complete before cross-market feature engineering."
        ),
        "Observed Evidence": (
            f"Section 5.3 completion status: "
            f"{section_5_3_complete}"
        ),
        "Passed": bool(
            section_5_3_complete
        ),
    },
    {
        "Validation Area": "Artist cross-market completeness",
        "Requirement": (
            "Every validated artist representation must "
            "receive one cross-market feature record."
        ),
        "Observed Evidence": (
            f"{len(artist_cross_market_features_df):,} records "
            f"for {artist_market_df['artists_raw'].nunique():,} artists"
        ),
        "Passed": (
            len(
                artist_cross_market_features_df
            )
            ==
            artist_market_df[
                "artists_raw"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Track cross-market completeness",
        "Requirement": (
            "Every validated track must receive one "
            "cross-market feature record."
        ),
        "Observed Evidence": (
            f"{len(track_cross_market_features_df):,} records "
            f"for {track_market_df['track_id'].nunique():,} tracks"
        ),
        "Passed": (
            len(
                track_cross_market_features_df
            )
            ==
            track_market_df[
                "track_id"
            ].nunique()
        ),
    },
    {
        "Validation Area": "Artist feature validity",
        "Requirement": (
            "All artist cross-market features must remain finite."
        ),
        "Observed Evidence": (
            f"{len(artist_numeric_features)} "
            "numeric artist cross-market features assessed"
        ),
        "Passed": artist_features_finite,
    },
    {
        "Validation Area": "Track feature validity",
        "Requirement": (
            "All track cross-market features must remain finite."
        ),
        "Observed Evidence": (
            f"{len(track_numeric_features)} "
            "numeric track cross-market features assessed"
        ),
        "Passed": track_features_finite,
    },
    {
        "Validation Area": "Artist percentage bounds",
        "Requirement": (
            "Artist percentage features must remain "
            "between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(artist_percentage_columns)} "
            "artist percentage fields assessed"
        ),
        "Passed": artist_percentages_valid,
    },
    {
        "Validation Area": "Track percentage bounds",
        "Requirement": (
            "Track percentage features must remain "
            "between 0 and 100."
        ),
        "Observed Evidence": (
            f"{len(track_percentage_columns)} "
            "track percentage fields assessed"
        ),
        "Passed": track_percentages_valid,
    },
    {
        "Validation Area": "Artist performance-range validity",
        "Requirement": (
            "Artist cross-market performance ranges "
            "must remain non-negative."
        ),
        "Observed Evidence": (
            "Artist stream and chart-position ranges assessed"
        ),
        "Passed": artist_ranges_valid,
    },
    {
        "Validation Area": "Track performance-range validity",
        "Requirement": (
            "Track cross-market performance ranges "
            "must remain non-negative."
        ),
        "Observed Evidence": (
            "Track stream and chart-position ranges assessed"
        ),
        "Passed": track_ranges_valid,
    },
])

print()
print("Cross-market performance feature validation")
print("=" * 100)

display(
    section_5_4_validation_df
)

section_5_4_complete = bool(
    section_5_4_validation_df[
        "Passed"
    ].all()
)

if not section_5_4_complete:
    failed_checks = (
        section_5_4_validation_df.loc[
            ~section_5_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.4 cross-market performance feature "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 5.4 cross-market performance "
    "feature validation checks passed."
)

print(
    f"Section 5.4 completion status: "
    f"{section_5_4_complete}"
)

print(
    f"Artist cross-market feature table prepared: "
    f"{len(artist_cross_market_features_df):,} rows."
)

print(
    f"Track cross-market feature table prepared: "
    f"{len(track_cross_market_features_df):,} rows."
)

print(
    "Cross-market features describe consistency and variation "
    "between observed country markets."
)

print(
    "The workflow is ready for "
    "Section 5.5 Geographic Growth and Decline Features."
)
```

### Interpretation

Section 5.4 successfully engineers cross-market performance features for both artist and track populations.

The resulting feature tables contain:

- **49,843 artist representations**
- **110,198 tracks**

All validation checks passed, confirming that the complete validated artist and track populations were preserved and that the engineered numerical and percentage-based features remain within valid ranges.

A key result is that the median cross-market stream coefficient of variation is **0.0000** for both artists and tracks. The median stream-performance range and chart-position range are also **0.00**.

This does not mean that the PMIP dataset contains no geographic variation. Instead, it reflects the strong single-market structure identified earlier. Because a large share of artists and tracks appear in only one country market, those entities have no cross-market difference to measure and therefore naturally receive zero variation.

This pattern is visible in the consistency classifications.

For artists:

- **35,446 artist representations (71.12%)** are classified as Very Consistent.
- **964 (1.93%)** are classified as Consistent.
- **5,558 (11.15%)** are classified as Variable.
- **7,875 (15.80%)** are classified as Highly Variable.

For tracks:

- **82,257 tracks (74.64%)** are classified as Very Consistent.
- **1,823 (1.65%)** are classified as Consistent.
- **10,971 (9.96%)** are classified as Variable.
- **15,147 (13.75%)** are classified as Highly Variable.

The large Very Consistent category must therefore be interpreted carefully. For many entities, this classification reflects limited geographic breadth rather than evidence of truly stable performance across many international markets.

For this reason, the dedicated multi-market analyses provide the more meaningful cross-market interpretation.

Among artists observed in at least five markets, clear differences in geographic behaviour emerge. Some artist representations have relatively small cross-market stream coefficients of variation, indicating broadly similar streaming levels across their observed markets.

Other artists show very large differences between their strongest and weakest markets. For example, the most variable artist representations have stream coefficients of variation above **3.4**, with some exceeding **4.3**. Their strongest-market streaming levels can also be several million streams greater than their weakest-market levels.

A similar structure is visible at track level.

The most consistent multi-market tracks have comparatively low cross-market coefficients of variation, generally around **0.16–0.26** among the displayed examples. These tracks show relatively balanced streaming performance across the markets in which they are observed.

By contrast, highly variable tracks show much larger differences. Several displayed tracks have cross-market stream coefficients of variation above **3.5**, while the most variable example exceeds **4.7**. This indicates that some internationally distributed tracks perform extremely strongly in particular markets while remaining much weaker elsewhere.

The chart-position ranges provide a second dimension of this variation. Large chart-position ranges show that some tracks or artists can perform near the top of a chart in one market while appearing much lower in another.

The engineered features therefore allow PMIP to distinguish between different geographic performance structures that simple reach alone cannot identify.

For example, two tracks may both appear in 50 countries, but:

- one may perform at a similar level across most of those countries;
- another may depend heavily on only a few exceptionally strong markets.

The cross-market feature layer captures this distinction using:

- stream-performance ranges,
- stream coefficients of variation,
- chart-position ranges,
- strongest and weakest market values,
- above-own-median market shares,
- comparison-eligible market coverage.

The **comparison-eligible market percentage** is particularly important because PMIP does not assume that every observed market can be compared equally. The feature preserves the support and temporal safeguards established earlier in the geographic workflow.

The overall median comparison-eligible market percentage is **0.00%**, again largely because many entities are represented in only one market or do not satisfy the stricter support requirements for direct comparison. This reinforces the need to separate raw geographic presence from genuinely comparison-ready geographic evidence.

Therefore, the results from this subsection should not be interpreted as universal evidence of geographic consistency or inconsistency.

Instead, the strongest conclusions should be drawn from entities with:

- multiple observed markets,
- sufficient analytical support,
- adequate temporal overlap,
- no major direct-comparison caution.

Section 5.4 therefore adds an important new dimension to PMIP geographic intelligence:

**reach tells us where an entity appears, penetration tells us how strongly it appears, concentration tells us how dependent it is on particular markets, and cross-market performance tells us how consistent that performance is between markets.**

The cross-market performance features remain descriptive engineered indicators and do not yet constitute a final PMIP geographic performance score.

The workflow is ready for **Section 5.5 Geographic Growth and Decline Features**.

## 5.5 Geographic Growth and Decline Features

### Purpose

This subsection engineers time-aware geographic growth and decline features for artists and tracks across country markets.

The aim is to identify whether an entity's observed geographic performance is expanding, stable or declining over time.

The analysis will create features describing:

- change in streaming performance between earlier and later observations,
- change in chart-position performance over time,
- market-entry growth,
- market-exit or contraction patterns,
- recent-versus-historical streaming differences,
- recent-versus-historical chart-position differences,
- direction of geographic performance change,
- magnitude of growth or decline,
- support-aware temporal trend eligibility.

Because country markets entered the historical dataset at different times, the analysis preserves the temporal safeguards established in Sections 3.3, 3.5 and 4.5.

Growth or decline is therefore treated as a descriptive trend only when sufficient chronological support exists.

These features do not prove commercial growth, decline, popularity change or causal real-world events. They describe changes observed within the historical PMIP geographic dataset.

```python
import pandas as pd
import numpy as np

print("Engineering PMIP geographic growth and decline features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous-section outputs
# ------------------------------------------------------------

if not globals().get("section_5_4_complete", False):
    raise RuntimeError(
        "Section 5.5 requires Section 5.4 to be completed first."
    )

if "geographic_analysis_df" not in globals():
    raise RuntimeError(
        "Section 5.5 requires geographic_analysis_df from "
        "the validated geographic analytical population."
    )

print(f"Section 5.4 completion status: {section_5_4_complete}")

# ------------------------------------------------------------
# 2. Recover validated geographic analytical population
# ------------------------------------------------------------

geo_df = geographic_analysis_df.copy()

print(
    f"Validated geographic analytical observations: "
    f"{len(geo_df):,}"
)

print(
    f"Validated country markets: "
    f"{geo_df['country'].nunique():,}"
)

print(
    f"Validated tracks: "
    f"{geo_df['track_id'].nunique():,}"
)

# ------------------------------------------------------------
# 3. Validate required columns
# ------------------------------------------------------------

required_columns = [
    "date",
    "country",
    "track_id",
    "track_name",
    "artists_raw",
    "streams",
    "position",
]

missing_columns = [
    column
    for column in required_columns
    if column not in geo_df.columns
]

if missing_columns:
    raise RuntimeError(
        "Section 5.5 geographic analytical source is missing: "
        + ", ".join(missing_columns)
    )

# ------------------------------------------------------------
# 4. Standardise working fields
# ------------------------------------------------------------

geo_df["date"] = pd.to_datetime(
    geo_df["date"],
    errors="coerce",
)

geo_df["streams"] = pd.to_numeric(
    geo_df["streams"],
    errors="coerce",
)

geo_df["position"] = pd.to_numeric(
    geo_df["position"],
    errors="coerce",
)

if geo_df["date"].isna().any():
    raise RuntimeError(
        "Section 5.5 found invalid reporting dates."
    )

if geo_df["streams"].isna().any():
    raise RuntimeError(
        "Section 5.5 found invalid stream values."
    )

if geo_df["position"].isna().any():
    raise RuntimeError(
        "Section 5.5 found invalid chart-position values."
    )

# ------------------------------------------------------------
# 5. Sort observations chronologically
# ------------------------------------------------------------

geo_df = (
    geo_df
    .sort_values(
        [
            "artists_raw",
            "country",
            "track_id",
            "date",
        ]
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 6. Helper functions
# ------------------------------------------------------------

def safe_percentage_change(first_value, last_value):
    """
    Calculate percentage change while safely handling zero
    or non-finite starting values.
    """

    first_value = float(first_value)
    last_value = float(last_value)

    if not np.isfinite(first_value):
        return 0.0

    if not np.isfinite(last_value):
        return 0.0

    if first_value == 0:
        return 0.0

    return (
        (last_value - first_value)
        / abs(first_value)
    ) * 100


def classify_growth_direction(value, tolerance=5.0):
    """
    Classify an observed geographic streaming trend.
    """

    if value > tolerance:
        return "Growth"

    if value < -tolerance:
        return "Decline"

    return "Stable"


def classify_growth_strength(value):
    """
    Describe the magnitude of the observed percentage change.
    """

    magnitude = abs(float(value))

    if magnitude < 5:
        return "Minimal"

    if magnitude < 25:
        return "Moderate"

    if magnitude < 100:
        return "Strong"

    return "Very strong"

# ------------------------------------------------------------
# 7. Artist-country chronological histories
# ------------------------------------------------------------

artist_country_history_df = (
    geo_df
    .groupby(
        [
            "artists_raw",
            "country",
        ],
        dropna=False,
    )
    .agg(
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        reporting_dates=("date", "nunique"),
        observations=("date", "size"),
        first_streams=("streams", "first"),
        last_streams=("streams", "last"),
        median_streams=("streams", "median"),
        first_chart_position=("position", "first"),
        last_chart_position=("position", "last"),
        median_chart_position=("position", "median"),
    )
    .reset_index()
)

artist_country_history_df[
    "stream_change_pct"
] = artist_country_history_df.apply(
    lambda row: safe_percentage_change(
        row["first_streams"],
        row["last_streams"],
    ),
    axis=1,
)

artist_country_history_df[
    "chart_position_change"
] = (
    artist_country_history_df["first_chart_position"]
    -
    artist_country_history_df["last_chart_position"]
)

# Minimum historical support required for trend interpretation
artist_country_history_df[
    "trend_eligible"
] = (
    artist_country_history_df["reporting_dates"] >= 4
)

# ------------------------------------------------------------
# 8. Track-country chronological histories
# ------------------------------------------------------------

track_country_history_df = (
    geo_df
    .groupby(
        [
            "track_id",
            "track_name",
            "artists_raw",
            "country",
        ],
        dropna=False,
    )
    .agg(
        first_reporting_date=("date", "min"),
        last_reporting_date=("date", "max"),
        reporting_dates=("date", "nunique"),
        observations=("date", "size"),
        first_streams=("streams", "first"),
        last_streams=("streams", "last"),
        median_streams=("streams", "median"),
        first_chart_position=("position", "first"),
        last_chart_position=("position", "last"),
        median_chart_position=("position", "median"),
    )
    .reset_index()
)

track_country_history_df[
    "stream_change_pct"
] = track_country_history_df.apply(
    lambda row: safe_percentage_change(
        row["first_streams"],
        row["last_streams"],
    ),
    axis=1,
)

track_country_history_df[
    "chart_position_change"
] = (
    track_country_history_df["first_chart_position"]
    -
    track_country_history_df["last_chart_position"]
)

track_country_history_df[
    "trend_eligible"
] = (
    track_country_history_df["reporting_dates"] >= 4
)

# ------------------------------------------------------------
# 9. Artist geographic growth / decline features
# ------------------------------------------------------------

artist_growth_df = (
    artist_country_history_df
    .groupby(
        "artists_raw",
        dropna=False,
    )
    .agg(
        markets_observed=("country", "nunique"),
        trend_eligible_markets=("trend_eligible", "sum"),
        median_market_stream_change_pct=(
            "stream_change_pct",
            "median",
        ),
        mean_market_stream_change_pct=(
            "stream_change_pct",
            "mean",
        ),
        maximum_market_growth_pct=(
            "stream_change_pct",
            "max",
        ),
        maximum_market_decline_pct=(
            "stream_change_pct",
            "min",
        ),
        median_chart_position_change=(
            "chart_position_change",
            "median",
        ),
        mean_chart_position_change=(
            "chart_position_change",
            "mean",
        ),
        first_observed_date=(
            "first_reporting_date",
            "min",
        ),
        last_observed_date=(
            "last_reporting_date",
            "max",
        ),
    )
    .reset_index()
)

artist_market_trends_df = (
    artist_country_history_df
    .assign(
        growth_market=lambda df: (
            df["trend_eligible"]
            &
            (df["stream_change_pct"] > 5)
        ),
        decline_market=lambda df: (
            df["trend_eligible"]
            &
            (df["stream_change_pct"] < -5)
        ),
        stable_market=lambda df: (
            df["trend_eligible"]
            &
            df["stream_change_pct"].between(-5, 5)
        ),
    )
    .groupby(
        "artists_raw",
        dropna=False,
    )
    .agg(
        growth_markets=("growth_market", "sum"),
        decline_markets=("decline_market", "sum"),
        stable_markets=("stable_market", "sum"),
    )
    .reset_index()
)

artist_growth_df = artist_growth_df.merge(
    artist_market_trends_df,
    on="artists_raw",
    how="left",
)

artist_growth_df[
    "trend_eligible_market_pct"
] = np.where(
    artist_growth_df["markets_observed"] > 0,
    (
        artist_growth_df["trend_eligible_markets"]
        /
        artist_growth_df["markets_observed"]
    ) * 100,
    0.0,
)

artist_growth_df[
    "growth_market_pct"
] = np.where(
    artist_growth_df["trend_eligible_markets"] > 0,
    (
        artist_growth_df["growth_markets"]
        /
        artist_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

artist_growth_df[
    "decline_market_pct"
] = np.where(
    artist_growth_df["trend_eligible_markets"] > 0,
    (
        artist_growth_df["decline_markets"]
        /
        artist_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

artist_growth_df[
    "stable_market_pct"
] = np.where(
    artist_growth_df["trend_eligible_markets"] > 0,
    (
        artist_growth_df["stable_markets"]
        /
        artist_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

artist_growth_df[
    "growth_balance_pct"
] = (
    artist_growth_df["growth_market_pct"]
    -
    artist_growth_df["decline_market_pct"]
)

artist_growth_df[
    "geographic_trend_direction"
] = artist_growth_df[
    "median_market_stream_change_pct"
].apply(
    classify_growth_direction
)

artist_growth_df[
    "geographic_trend_strength"
] = artist_growth_df[
    "median_market_stream_change_pct"
].apply(
    classify_growth_strength
)

# ------------------------------------------------------------
# 10. Track geographic growth / decline features
# ------------------------------------------------------------

track_growth_df = (
    track_country_history_df
    .groupby(
        [
            "track_id",
            "track_name",
            "artists_raw",
        ],
        dropna=False,
    )
    .agg(
        markets_observed=("country", "nunique"),
        trend_eligible_markets=("trend_eligible", "sum"),
        median_market_stream_change_pct=(
            "stream_change_pct",
            "median",
        ),
        mean_market_stream_change_pct=(
            "stream_change_pct",
            "mean",
        ),
        maximum_market_growth_pct=(
            "stream_change_pct",
            "max",
        ),
        maximum_market_decline_pct=(
            "stream_change_pct",
            "min",
        ),
        median_chart_position_change=(
            "chart_position_change",
            "median",
        ),
        mean_chart_position_change=(
            "chart_position_change",
            "mean",
        ),
        first_observed_date=(
            "first_reporting_date",
            "min",
        ),
        last_observed_date=(
            "last_reporting_date",
            "max",
        ),
    )
    .reset_index()
)

track_market_trends_df = (
    track_country_history_df
    .assign(
        growth_market=lambda df: (
            df["trend_eligible"]
            &
            (df["stream_change_pct"] > 5)
        ),
        decline_market=lambda df: (
            df["trend_eligible"]
            &
            (df["stream_change_pct"] < -5)
        ),
        stable_market=lambda df: (
            df["trend_eligible"]
            &
            df["stream_change_pct"].between(-5, 5)
        ),
    )
    .groupby(
        "track_id",
        dropna=False,
    )
    .agg(
        growth_markets=("growth_market", "sum"),
        decline_markets=("decline_market", "sum"),
        stable_markets=("stable_market", "sum"),
    )
    .reset_index()
)

track_growth_df = track_growth_df.merge(
    track_market_trends_df,
    on="track_id",
    how="left",
)

track_growth_df[
    "trend_eligible_market_pct"
] = np.where(
    track_growth_df["markets_observed"] > 0,
    (
        track_growth_df["trend_eligible_markets"]
        /
        track_growth_df["markets_observed"]
    ) * 100,
    0.0,
)

track_growth_df[
    "growth_market_pct"
] = np.where(
    track_growth_df["trend_eligible_markets"] > 0,
    (
        track_growth_df["growth_markets"]
        /
        track_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

track_growth_df[
    "decline_market_pct"
] = np.where(
    track_growth_df["trend_eligible_markets"] > 0,
    (
        track_growth_df["decline_markets"]
        /
        track_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

track_growth_df[
    "stable_market_pct"
] = np.where(
    track_growth_df["trend_eligible_markets"] > 0,
    (
        track_growth_df["stable_markets"]
        /
        track_growth_df["trend_eligible_markets"]
    ) * 100,
    0.0,
)

track_growth_df[
    "growth_balance_pct"
] = (
    track_growth_df["growth_market_pct"]
    -
    track_growth_df["decline_market_pct"]
)

track_growth_df[
    "geographic_trend_direction"
] = track_growth_df[
    "median_market_stream_change_pct"
].apply(
    classify_growth_direction
)

track_growth_df[
    "geographic_trend_strength"
] = track_growth_df[
    "median_market_stream_change_pct"
].apply(
    classify_growth_strength
)

# ------------------------------------------------------------
# 11. Artist geographic trend overview
# ------------------------------------------------------------

print()
print("Artist geographic growth and decline overview")
print("=" * 100)

print(
    f"Artist growth records prepared: "
    f"{len(artist_growth_df):,}"
)

print(
    "Median trend-eligible market share: "
    f"{artist_growth_df['trend_eligible_market_pct'].median():.2f}%"
)

print(
    "Median observed market stream change: "
    f"{artist_growth_df['median_market_stream_change_pct'].median():.2f}%"
)

print(
    "Median growth-market share: "
    f"{artist_growth_df['growth_market_pct'].median():.2f}%"
)

print(
    "Median decline-market share: "
    f"{artist_growth_df['decline_market_pct'].median():.2f}%"
)

artist_trend_distribution_df = (
    artist_growth_df[
        "geographic_trend_direction"
    ]
    .value_counts()
    .rename_axis("Trend Direction")
    .reset_index(name="Artist Representations")
)

artist_trend_distribution_df[
    "Share (%)"
] = (
    artist_trend_distribution_df[
        "Artist Representations"
    ]
    / len(artist_growth_df)
    * 100
)

print()
print("Artist geographic trend classification")
print("=" * 100)

display(
    artist_trend_distribution_df
)

# ------------------------------------------------------------
# 12. Artist strongest observed growth
# ------------------------------------------------------------

strongest_artist_growth_df = (
    artist_growth_df
    .loc[
        artist_growth_df[
            "trend_eligible_markets"
        ] >= 3
    ]
    .sort_values(
        [
            "median_market_stream_change_pct",
            "growth_market_pct",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Artist representations with strongest observed "
    "geographic growth"
)
print("=" * 100)

display(
    strongest_artist_growth_df[
        [
            "artists_raw",
            "markets_observed",
            "trend_eligible_markets",
            "trend_eligible_market_pct",
            "median_market_stream_change_pct",
            "growth_market_pct",
            "decline_market_pct",
            "growth_balance_pct",
            "median_chart_position_change",
            "geographic_trend_direction",
            "geographic_trend_strength",
        ]
    ]
)

# ------------------------------------------------------------
# 13. Artist strongest observed decline
# ------------------------------------------------------------

strongest_artist_decline_df = (
    artist_growth_df
    .loc[
        artist_growth_df[
            "trend_eligible_markets"
        ] >= 3
    ]
    .sort_values(
        [
            "median_market_stream_change_pct",
            "decline_market_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print(
    "Artist representations with strongest observed "
    "geographic decline"
)
print("=" * 100)

display(
    strongest_artist_decline_df[
        [
            "artists_raw",
            "markets_observed",
            "trend_eligible_markets",
            "trend_eligible_market_pct",
            "median_market_stream_change_pct",
            "growth_market_pct",
            "decline_market_pct",
            "growth_balance_pct",
            "median_chart_position_change",
            "geographic_trend_direction",
            "geographic_trend_strength",
        ]
    ]
)

# ------------------------------------------------------------
# 14. Track geographic trend overview
# ------------------------------------------------------------

print()
print("Track geographic growth and decline overview")
print("=" * 100)

print(
    f"Track growth records prepared: "
    f"{len(track_growth_df):,}"
)

print(
    "Median trend-eligible market share: "
    f"{track_growth_df['trend_eligible_market_pct'].median():.2f}%"
)

print(
    "Median observed market stream change: "
    f"{track_growth_df['median_market_stream_change_pct'].median():.2f}%"
)

print(
    "Median growth-market share: "
    f"{track_growth_df['growth_market_pct'].median():.2f}%"
)

print(
    "Median decline-market share: "
    f"{track_growth_df['decline_market_pct'].median():.2f}%"
)

track_trend_distribution_df = (
    track_growth_df[
        "geographic_trend_direction"
    ]
    .value_counts()
    .rename_axis("Trend Direction")
    .reset_index(name="Tracks")
)

track_trend_distribution_df[
    "Share (%)"
] = (
    track_trend_distribution_df["Tracks"]
    / len(track_growth_df)
    * 100
)

print()
print("Track geographic trend classification")
print("=" * 100)

display(
    track_trend_distribution_df
)

# ------------------------------------------------------------
# 15. Tracks with strongest observed growth
# ------------------------------------------------------------

strongest_track_growth_df = (
    track_growth_df
    .loc[
        track_growth_df[
            "trend_eligible_markets"
        ] >= 3
    ]
    .sort_values(
        [
            "median_market_stream_change_pct",
            "growth_market_pct",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with strongest observed geographic growth")
print("=" * 100)

display(
    strongest_track_growth_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "trend_eligible_markets",
            "trend_eligible_market_pct",
            "median_market_stream_change_pct",
            "growth_market_pct",
            "decline_market_pct",
            "growth_balance_pct",
            "median_chart_position_change",
            "geographic_trend_direction",
            "geographic_trend_strength",
        ]
    ]
)

# ------------------------------------------------------------
# 16. Tracks with strongest observed decline
# ------------------------------------------------------------

strongest_track_decline_df = (
    track_growth_df
    .loc[
        track_growth_df[
            "trend_eligible_markets"
        ] >= 3
    ]
    .sort_values(
        [
            "median_market_stream_change_pct",
            "decline_market_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

print()
print("Tracks with strongest observed geographic decline")
print("=" * 100)

display(
    strongest_track_decline_df[
        [
            "track_name",
            "artists_raw",
            "markets_observed",
            "trend_eligible_markets",
            "trend_eligible_market_pct",
            "median_market_stream_change_pct",
            "growth_market_pct",
            "decline_market_pct",
            "growth_balance_pct",
            "median_chart_position_change",
            "geographic_trend_direction",
            "geographic_trend_strength",
        ]
    ]
)

# ------------------------------------------------------------
# 17. Feature validation
# ------------------------------------------------------------

artist_numeric_features = [
    "markets_observed",
    "trend_eligible_markets",
    "median_market_stream_change_pct",
    "mean_market_stream_change_pct",
    "maximum_market_growth_pct",
    "maximum_market_decline_pct",
    "median_chart_position_change",
    "mean_chart_position_change",
    "growth_markets",
    "decline_markets",
    "stable_markets",
    "trend_eligible_market_pct",
    "growth_market_pct",
    "decline_market_pct",
    "stable_market_pct",
    "growth_balance_pct",
]

track_numeric_features = [
    "markets_observed",
    "trend_eligible_markets",
    "median_market_stream_change_pct",
    "mean_market_stream_change_pct",
    "maximum_market_growth_pct",
    "maximum_market_decline_pct",
    "median_chart_position_change",
    "mean_chart_position_change",
    "growth_markets",
    "decline_markets",
    "stable_markets",
    "trend_eligible_market_pct",
    "growth_market_pct",
    "decline_market_pct",
    "stable_market_pct",
    "growth_balance_pct",
]

artist_features_finite = bool(
    np.isfinite(
        artist_growth_df[
            artist_numeric_features
        ].to_numpy(dtype=float)
    ).all()
)

track_features_finite = bool(
    np.isfinite(
        track_growth_df[
            track_numeric_features
        ].to_numpy(dtype=float)
    ).all()
)

artist_pct_fields = [
    "trend_eligible_market_pct",
    "growth_market_pct",
    "decline_market_pct",
    "stable_market_pct",
]

track_pct_fields = [
    "trend_eligible_market_pct",
    "growth_market_pct",
    "decline_market_pct",
    "stable_market_pct",
]

artist_percentages_valid = bool(
    artist_growth_df[
        artist_pct_fields
    ]
    .apply(
        lambda column: column.between(
            0,
            100,
        ).all()
    )
    .all()
)

track_percentages_valid = bool(
    track_growth_df[
        track_pct_fields
    ]
    .apply(
        lambda column: column.between(
            0,
            100,
        ).all()
    )
    .all()
)

artist_count_reconciliation = bool(
    (
        artist_growth_df["growth_markets"]
        +
        artist_growth_df["decline_markets"]
        +
        artist_growth_df["stable_markets"]
    )
    .eq(
        artist_growth_df[
            "trend_eligible_markets"
        ]
    )
    .all()
)

track_count_reconciliation = bool(
    (
        track_growth_df["growth_markets"]
        +
        track_growth_df["decline_markets"]
        +
        track_growth_df["stable_markets"]
    )
    .eq(
        track_growth_df[
            "trend_eligible_markets"
        ]
    )
    .all()
)

section_5_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 5.4 completion",

            "Requirement":
                "Cross-market performance features must "
                "already be complete.",

            "Observed Evidence":
                f"Section 5.4 completion status: "
                f"{section_5_4_complete}",

            "Passed":
                bool(section_5_4_complete),
        },
        {
            "Validation Area":
                "Validated source preservation",

            "Requirement":
                "Section 5.5 must use the validated "
                "country-market analytical population.",

            "Observed Evidence":
                f"{len(geo_df):,} validated observations",

            "Passed":
                len(geo_df) == 5_338_680,
        },
        {
            "Validation Area":
                "Artist growth completeness",

            "Requirement":
                "Every validated artist representation must "
                "receive a geographic growth record.",

            "Observed Evidence":
                f"{len(artist_growth_df):,} records for "
                f"{geo_df['artists_raw'].nunique():,} artists",

            "Passed":
                len(artist_growth_df)
                ==
                geo_df["artists_raw"].nunique(),
        },
        {
            "Validation Area":
                "Track growth completeness",

            "Requirement":
                "Every validated track must receive a "
                "geographic growth record.",

            "Observed Evidence":
                f"{len(track_growth_df):,} records for "
                f"{geo_df['track_id'].nunique():,} tracks",

            "Passed":
                len(track_growth_df)
                ==
                geo_df["track_id"].nunique(),
        },
        {
            "Validation Area":
                "Artist feature validity",

            "Requirement":
                "All engineered artist growth and decline "
                "features must remain finite.",

            "Observed Evidence":
                f"{len(artist_numeric_features)} numeric "
                "artist features assessed",

            "Passed":
                artist_features_finite,
        },
        {
            "Validation Area":
                "Track feature validity",

            "Requirement":
                "All engineered track growth and decline "
                "features must remain finite.",

            "Observed Evidence":
                f"{len(track_numeric_features)} numeric "
                "track features assessed",

            "Passed":
                track_features_finite,
        },
        {
            "Validation Area":
                "Artist percentage bounds",

            "Requirement":
                "Artist percentage features must remain "
                "between 0 and 100.",

            "Observed Evidence":
                f"{len(artist_pct_fields)} percentage "
                "fields assessed",

            "Passed":
                artist_percentages_valid,
        },
        {
            "Validation Area":
                "Track percentage bounds",

            "Requirement":
                "Track percentage features must remain "
                "between 0 and 100.",

            "Observed Evidence":
                f"{len(track_pct_fields)} percentage "
                "fields assessed",

            "Passed":
                track_percentages_valid,
        },
        {
            "Validation Area":
                "Artist trend reconciliation",

            "Requirement":
                "Growth, decline and stable artist-market "
                "counts must reconcile with eligible markets.",

            "Observed Evidence":
                "Artist market-trend counts reconciled",

            "Passed":
                artist_count_reconciliation,
        },
        {
            "Validation Area":
                "Track trend reconciliation",

            "Requirement":
                "Growth, decline and stable track-market "
                "counts must reconcile with eligible markets.",

            "Observed Evidence":
                "Track market-trend counts reconciled",

            "Passed":
                track_count_reconciliation,
        },
    ]
)

print()
print("Geographic growth and decline feature validation")
print("=" * 100)

display(
    section_5_5_validation_df
)

section_5_5_complete = bool(
    section_5_5_validation_df[
        "Passed"
    ].all()
)

if not section_5_5_complete:

    failed_checks = (
        section_5_5_validation_df
        .loc[
            ~section_5_5_validation_df["Passed"],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.5 geographic growth and decline "
        "feature validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 5.5 geographic growth and decline "
    "feature validation checks passed."
)

print(
    f"Section 5.5 completion status: "
    f"{section_5_5_complete}"
)

print(
    f"Artist geographic growth feature table prepared: "
    f"{len(artist_growth_df):,} rows."
)

print(
    f"Track geographic growth feature table prepared: "
    f"{len(track_growth_df):,} rows."
)

print(
    "Geographic growth and decline features describe "
    "observed historical changes only and do not establish causality."
)

print(
    "The workflow is ready for "
    "Section 5.6 Feature Validation."
)
```

### Interpretation

Section 5.5 successfully engineered geographic growth and decline features from the validated PMIP country-market analytical population of **5,338,680 observations across 76 country markets**.

At artist level, **49,843 artist representations** received geographic trend features, while **110,198 tracks** received corresponding track-level features. This preserves the complete validated artist and track populations established earlier in the workflow.

The median artist had sufficient historical evidence for trend analysis in approximately **71.43% of its observed markets**. However, the median growth-market and decline-market shares were both **0.00%**, showing that geographic change is highly uneven across the catalogue rather than being present consistently for the typical artist.

The artist-level trend classification identified three broad historical patterns:

- **22,264 artist representations (44.67%)** were classified as showing an observed geographic decline.
- **17,747 artist representations (35.61%)** were classified as relatively stable.
- **9,832 artist representations (19.73%)** were classified as showing observed geographic growth.

This indicates that historical geographic performance within the dataset is not uniformly expanding. A larger proportion of artist representations show declining rather than increasing stream patterns across their observed markets, while a substantial group remains comparatively stable.

The engineered features also retain the amount of historical support behind each trend through variables such as the number and percentage of trend-eligible markets. This is important because an apparent rise or decline supported by many repeated market observations should be interpreted differently from one based on limited historical coverage.

All Section 5.5 validation checks passed. Artist and track records were preserved, engineered numeric features remained valid, percentage measures remained within their expected bounds, and growth, decline and stable market counts reconciled with the number of markets eligible for trend analysis.

These features describe **observed historical geographic change only**. They do not explain why an artist or track grew or declined, and they should not be interpreted as evidence of causality, future performance, or commercial quality.

Section 5.5 is therefore complete, and the geographic feature-engineering workflow is ready for **Section 5.6 Feature Validation**.

### 5.6 Feature Validation

#### Purpose

Validate all geographic features engineered in Sections **5.1–5.5** before they are used in later PMIP scoring, modelling, comparison, or dashboard integration.

This subsection checks that the final geographic feature tables preserve the validated artist and track populations, contain no unexpected missing or infinite numerical values, keep percentage-based features within valid bounds, maintain unique artist and track records, and reconcile important measures across the different feature families.

The validation also confirms that the five geographic feature groups — **country reach, market penetration, geographic concentration, cross-market performance, and geographic growth/decline** — remain complete and internally consistent.

These checks are intended to ensure that the engineered geographic features are technically reliable analytical inputs before PMIP begins constructing geographic performance indicators or scores.

```python
import pandas as pd
import numpy as np

print("Validating engineered PMIP geographic features")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate completion of Sections 5.1–5.5
# ------------------------------------------------------------

required_section_flags = {
    "Section 5.1": globals().get("section_5_1_complete", False),
    "Section 5.2": globals().get("section_5_2_complete", False),
    "Section 5.3": globals().get("section_5_3_complete", False),
    "Section 5.4": globals().get("section_5_4_complete", False),
    "Section 5.5": globals().get("section_5_5_complete", False),
}

for section_name, status in required_section_flags.items():
    print(f"{section_name} completion status: {status}")

incomplete_sections = [
    section_name
    for section_name, status in required_section_flags.items()
    if not status
]

if incomplete_sections:
    raise RuntimeError(
        "Section 5.6 requires completion of: "
        + ", ".join(incomplete_sections)
    )

# ------------------------------------------------------------
# 2. Register the actual Section 5 feature tables
# ------------------------------------------------------------

required_feature_tables = {
    "artist_country_reach_features_df":
        "Artist country-reach features",

    "track_country_reach_features_df":
        "Track country-reach features",

    "artist_penetration_features_df":
        "Artist market-penetration features",

    "track_penetration_features_df":
        "Track market-penetration features",

    "artist_stream_concentration_df":
        "Artist geographic-concentration features",

    "track_stream_concentration_df":
        "Track geographic-concentration features",

    "artist_cross_market_features_df":
        "Artist cross-market performance features",

    "track_cross_market_features_df":
        "Track cross-market performance features",

    "artist_growth_df":
        "Artist geographic growth and decline features",

    "track_growth_df":
        "Track geographic growth and decline features",
}

missing_tables = [
    name
    for name in required_feature_tables
    if name not in globals()
]

if missing_tables:
    raise RuntimeError(
        "Section 5.6 is missing the following feature tables: "
        + ", ".join(missing_tables)
    )

print()
print("Recovered engineered geographic feature tables")
print("=" * 100)

feature_table_registry = []

for name, description in required_feature_tables.items():

    df = globals()[name]

    feature_table_registry.append(
        {
            "Feature Table": name,
            "Feature Family": description,
            "Rows": len(df),
            "Columns": len(df.columns),
        }
    )

feature_table_registry_df = pd.DataFrame(
    feature_table_registry
)

display(feature_table_registry_df)

# ------------------------------------------------------------
# 3. Validated entity populations
# ------------------------------------------------------------

EXPECTED_ARTISTS = 49_843
EXPECTED_TRACKS = 110_198

artist_tables = [
    "artist_country_reach_features_df",
    "artist_penetration_features_df",
    "artist_stream_concentration_df",
    "artist_cross_market_features_df",
    "artist_growth_df",
]

track_tables = [
    "track_country_reach_features_df",
    "track_penetration_features_df",
    "track_stream_concentration_df",
    "track_cross_market_features_df",
    "track_growth_df",
]

# ------------------------------------------------------------
# 4. Population-preservation validation
# ------------------------------------------------------------

population_rows = []

for name in artist_tables:

    df = globals()[name]

    population_rows.append(
        {
            "Feature Table": name,
            "Entity Type": "Artist representation",
            "Rows": len(df),
            "Expected Rows": EXPECTED_ARTISTS,
            "Passed": len(df) == EXPECTED_ARTISTS,
        }
    )

for name in track_tables:

    df = globals()[name]

    population_rows.append(
        {
            "Feature Table": name,
            "Entity Type": "Track",
            "Rows": len(df),
            "Expected Rows": EXPECTED_TRACKS,
            "Passed": len(df) == EXPECTED_TRACKS,
        }
    )

population_validation_df = pd.DataFrame(
    population_rows
)

print()
print("Feature population preservation")
print("=" * 100)

display(population_validation_df)

# ------------------------------------------------------------
# 5. Entity-key uniqueness
# ------------------------------------------------------------

entity_key_rows = []

for name in artist_tables:

    df = globals()[name]

    if "artists_raw" not in df.columns:

        entity_key_rows.append(
            {
                "Feature Table": name,
                "Entity Key": "artists_raw",
                "Unique Entities": np.nan,
                "Duplicate Entity Rows": np.nan,
                "Passed": False,
            }
        )

        continue

    unique_entities = int(
        df["artists_raw"].nunique()
    )

    duplicate_entities = int(
        df.duplicated(
            subset=["artists_raw"]
        ).sum()
    )

    entity_key_rows.append(
        {
            "Feature Table": name,
            "Entity Key": "artists_raw",
            "Unique Entities": unique_entities,
            "Duplicate Entity Rows": duplicate_entities,
            "Passed": (
                unique_entities == EXPECTED_ARTISTS
                and duplicate_entities == 0
            ),
        }
    )

for name in track_tables:

    df = globals()[name]

    if "track_id" not in df.columns:

        entity_key_rows.append(
            {
                "Feature Table": name,
                "Entity Key": "track_id",
                "Unique Entities": np.nan,
                "Duplicate Entity Rows": np.nan,
                "Passed": False,
            }
        )

        continue

    unique_entities = int(
        df["track_id"].nunique()
    )

    duplicate_entities = int(
        df.duplicated(
            subset=["track_id"]
        ).sum()
    )

    entity_key_rows.append(
        {
            "Feature Table": name,
            "Entity Key": "track_id",
            "Unique Entities": unique_entities,
            "Duplicate Entity Rows": duplicate_entities,
            "Passed": (
                unique_entities == EXPECTED_TRACKS
                and duplicate_entities == 0
            ),
        }
    )

entity_key_validation_df = pd.DataFrame(
    entity_key_rows
)

print()
print("Entity-key validation")
print("=" * 100)

display(entity_key_validation_df)

# ------------------------------------------------------------
# 6. Numeric missingness and finiteness
# ------------------------------------------------------------

numeric_rows = []

for name in required_feature_tables:

    df = globals()[name]

    numeric_columns = df.select_dtypes(
        include=[np.number]
    ).columns.tolist()

    if numeric_columns:

        numeric_values = (
            df[numeric_columns]
            .to_numpy(dtype=float)
        )

        missing_numeric_values = int(
            df[numeric_columns]
            .isna()
            .sum()
            .sum()
        )

        non_finite_values = int(
            (~np.isfinite(numeric_values)).sum()
        )

    else:

        missing_numeric_values = 0
        non_finite_values = 0

    numeric_rows.append(
        {
            "Feature Table": name,
            "Numeric Fields": len(numeric_columns),
            "Missing Numeric Values":
                missing_numeric_values,
            "Non-Finite Numeric Values":
                non_finite_values,
            "Passed": (
                missing_numeric_values == 0
                and non_finite_values == 0
            ),
        }
    )

numeric_validation_df = pd.DataFrame(
    numeric_rows
)

print()
print("Numeric feature validity")
print("=" * 100)

display(numeric_validation_df)

# ------------------------------------------------------------
# 7. Validate percentage fields that must remain 0–100
# ------------------------------------------------------------

def identify_bounded_percentage_fields(df):
    """
    Percentage fields representing shares, coverage,
    reach or eligible proportions must remain in [0, 100].

    Percentage-change and balance fields are deliberately
    excluded because they may validly exceed these bounds.
    """

    bounded_fields = []

    for column in df.columns:

        lower_name = column.lower()

        if not lower_name.endswith("_pct"):
            continue

        # Exclude signed/unbounded change measures
        excluded_terms = [
            "change",
            "balance",
            "growth_pct",
            "decline_pct",
        ]

        if any(
            term in lower_name
            for term in excluded_terms
        ):
            continue

        # Include clearly bounded proportion/share fields
        bounded_terms = [
            "share",
            "reach",
            "coverage",
            "support",
            "eligible",
            "market_pct",
        ]

        if any(
            term in lower_name
            for term in bounded_terms
        ):
            bounded_fields.append(column)

    return bounded_fields


percentage_rows = []

for name in required_feature_tables:

    df = globals()[name]

    bounded_fields = (
        identify_bounded_percentage_fields(df)
    )

    invalid_values = 0

    for column in bounded_fields:

        values = pd.to_numeric(
            df[column],
            errors="coerce",
        )

        invalid_mask = (
            values.isna()
            |
            ~np.isfinite(values)
            |
            (values < 0)
            |
            (values > 100)
        )

        invalid_values += int(
            invalid_mask.sum()
        )

    percentage_rows.append(
        {
            "Feature Table": name,
            "Bounded Percentage Fields":
                len(bounded_fields),
            "Invalid Percentage Values":
                invalid_values,
            "Passed":
                invalid_values == 0,
        }
    )

percentage_validation_df = pd.DataFrame(
    percentage_rows
)

print()
print("Bounded percentage-feature validation")
print("=" * 100)

display(percentage_validation_df)

# ------------------------------------------------------------
# 8. Cross-family market-count reconciliation
# ------------------------------------------------------------

def reconcile_artist_market_counts():

    reference = (
        artist_country_reach_features_df[
            [
                "artists_raw",
                "country_markets_reached",
            ]
        ]
        .copy()
    )

    comparisons = {
        "Market penetration":
            artist_penetration_features_df,

        "Geographic concentration":
            artist_stream_concentration_df,

        "Cross-market performance":
            artist_cross_market_features_df,

        "Growth and decline":
            artist_growth_df,
    }

    results = []

    for family, df in comparisons.items():

        if "markets_observed" not in df.columns:

            results.append(
                {
                    "Feature Family": family,
                    "Compared Entities": 0,
                    "Mismatched Market Counts": np.nan,
                    "Passed": False,
                }
            )

            continue

        check_df = reference.merge(
            df[
                [
                    "artists_raw",
                    "markets_observed",
                ]
            ],
            on="artists_raw",
            how="inner",
        )

        mismatch_count = int(
            (
                check_df[
                    "country_markets_reached"
                ]
                !=
                check_df[
                    "markets_observed"
                ]
            ).sum()
        )

        results.append(
            {
                "Feature Family": family,
                "Compared Entities":
                    len(check_df),
                "Mismatched Market Counts":
                    mismatch_count,
                "Passed": (
                    len(check_df)
                    == EXPECTED_ARTISTS
                    and mismatch_count == 0
                ),
            }
        )

    return pd.DataFrame(results)


def reconcile_track_market_counts():

    reference = (
        track_country_reach_features_df[
            [
                "track_id",
                "country_markets_reached",
            ]
        ]
        .copy()
    )

    comparisons = {
        "Market penetration":
            track_penetration_features_df,

        "Geographic concentration":
            track_stream_concentration_df,

        "Cross-market performance":
            track_cross_market_features_df,

        "Growth and decline":
            track_growth_df,
    }

    results = []

    for family, df in comparisons.items():

        if "markets_observed" not in df.columns:

            results.append(
                {
                    "Feature Family": family,
                    "Compared Entities": 0,
                    "Mismatched Market Counts": np.nan,
                    "Passed": False,
                }
            )

            continue

        check_df = reference.merge(
            df[
                [
                    "track_id",
                    "markets_observed",
                ]
            ],
            on="track_id",
            how="inner",
        )

        mismatch_count = int(
            (
                check_df[
                    "country_markets_reached"
                ]
                !=
                check_df[
                    "markets_observed"
                ]
            ).sum()
        )

        results.append(
            {
                "Feature Family": family,
                "Compared Entities":
                    len(check_df),
                "Mismatched Market Counts":
                    mismatch_count,
                "Passed": (
                    len(check_df)
                    == EXPECTED_TRACKS
                    and mismatch_count == 0
                ),
            }
        )

    return pd.DataFrame(results)


artist_market_reconciliation_df = (
    reconcile_artist_market_counts()
)

track_market_reconciliation_df = (
    reconcile_track_market_counts()
)

print()
print("Artist market-count reconciliation")
print("=" * 100)

display(
    artist_market_reconciliation_df
)

print()
print("Track market-count reconciliation")
print("=" * 100)

display(
    track_market_reconciliation_df
)

# ------------------------------------------------------------
# 9. Growth and decline reconciliation
# ------------------------------------------------------------

artist_growth_reconciliation = bool(
    (
        artist_growth_df["growth_markets"]
        +
        artist_growth_df["decline_markets"]
        +
        artist_growth_df["stable_markets"]
    )
    .eq(
        artist_growth_df[
            "trend_eligible_markets"
        ]
    )
    .all()
)

track_growth_reconciliation = bool(
    (
        track_growth_df["growth_markets"]
        +
        track_growth_df["decline_markets"]
        +
        track_growth_df["stable_markets"]
    )
    .eq(
        track_growth_df[
            "trend_eligible_markets"
        ]
    )
    .all()
)

print()
print("Growth and decline reconciliation")
print("=" * 100)

growth_reconciliation_df = pd.DataFrame(
    [
        {
            "Entity Type":
                "Artist representations",

            "Requirement":
                "Growth + decline + stable markets must "
                "equal trend-eligible markets.",

            "Passed":
                artist_growth_reconciliation,
        },
        {
            "Entity Type":
                "Tracks",

            "Requirement":
                "Growth + decline + stable markets must "
                "equal trend-eligible markets.",

            "Passed":
                track_growth_reconciliation,
        },
    ]
)

display(
    growth_reconciliation_df
)

# ------------------------------------------------------------
# 10. Feature-family completeness
# ------------------------------------------------------------

feature_family_validation_df = pd.DataFrame(
    [
        {
            "Feature Family":
                "Country reach",

            "Artist Rows":
                len(
                    artist_country_reach_features_df
                ),

            "Track Rows":
                len(
                    track_country_reach_features_df
                ),

            "Passed":
                (
                    len(
                        artist_country_reach_features_df
                    )
                    == EXPECTED_ARTISTS
                    and
                    len(
                        track_country_reach_features_df
                    )
                    == EXPECTED_TRACKS
                ),
        },
        {
            "Feature Family":
                "Market penetration",

            "Artist Rows":
                len(
                    artist_penetration_features_df
                ),

            "Track Rows":
                len(
                    track_penetration_features_df
                ),

            "Passed":
                (
                    len(
                        artist_penetration_features_df
                    )
                    == EXPECTED_ARTISTS
                    and
                    len(
                        track_penetration_features_df
                    )
                    == EXPECTED_TRACKS
                ),
        },
        {
            "Feature Family":
                "Geographic concentration",

            "Artist Rows":
                len(
                    artist_stream_concentration_df
                ),

            "Track Rows":
                len(
                    track_stream_concentration_df
                ),

            "Passed":
                (
                    len(
                        artist_stream_concentration_df
                    )
                    == EXPECTED_ARTISTS
                    and
                    len(
                        track_stream_concentration_df
                    )
                    == EXPECTED_TRACKS
                ),
        },
        {
            "Feature Family":
                "Cross-market performance",

            "Artist Rows":
                len(
                    artist_cross_market_features_df
                ),

            "Track Rows":
                len(
                    track_cross_market_features_df
                ),

            "Passed":
                (
                    len(
                        artist_cross_market_features_df
                    )
                    == EXPECTED_ARTISTS
                    and
                    len(
                        track_cross_market_features_df
                    )
                    == EXPECTED_TRACKS
                ),
        },
        {
            "Feature Family":
                "Growth and decline",

            "Artist Rows":
                len(
                    artist_growth_df
                ),

            "Track Rows":
                len(
                    track_growth_df
                ),

            "Passed":
                (
                    len(
                        artist_growth_df
                    )
                    == EXPECTED_ARTISTS
                    and
                    len(
                        track_growth_df
                    )
                    == EXPECTED_TRACKS
                ),
        },
    ]
)

print()
print("Geographic feature-family completeness")
print("=" * 100)

display(
    feature_family_validation_df
)

# ------------------------------------------------------------
# 11. Consolidated Section 5.6 validation
# ------------------------------------------------------------

section_5_6_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Sections 5.1–5.5 completion",

            "Requirement":
                "All previous geographic feature-engineering "
                "subsections must be complete.",

            "Observed Evidence":
                "Sections 5.1–5.5 completion statuses are True",

            "Passed":
                all(
                    required_section_flags.values()
                ),
        },
        {
            "Validation Area":
                "Feature-table availability",

            "Requirement":
                "All ten artist and track feature tables "
                "must be available.",

            "Observed Evidence":
                f"{len(required_feature_tables)} "
                "feature tables recovered",

            "Passed":
                len(required_feature_tables) == 10,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All artist feature families must preserve "
                "49,843 validated artist representations.",

            "Observed Evidence":
                "Five artist feature tables assessed",

            "Passed":
                bool(
                    population_validation_df
                    .loc[
                        population_validation_df[
                            "Entity Type"
                        ]
                        == "Artist representation",
                        "Passed",
                    ]
                    .all()
                ),
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All track feature families must preserve "
                "110,198 validated tracks.",

            "Observed Evidence":
                "Five track feature tables assessed",

            "Passed":
                bool(
                    population_validation_df
                    .loc[
                        population_validation_df[
                            "Entity Type"
                        ]
                        == "Track",
                        "Passed",
                    ]
                    .all()
                ),
        },
        {
            "Validation Area":
                "Entity-key uniqueness",

            "Requirement":
                "Every artist and track must appear once "
                "in each entity-level feature table.",

            "Observed Evidence":
                f"{len(entity_key_validation_df)} "
                "entity-level tables assessed",

            "Passed":
                bool(
                    entity_key_validation_df[
                        "Passed"
                    ].all()
                ),
        },
        {
            "Validation Area":
                "Numeric feature validity",

            "Requirement":
                "Engineered numeric features must contain "
                "no missing or non-finite values.",

            "Observed Evidence":
                f"{len(numeric_validation_df)} "
                "feature tables assessed",

            "Passed":
                bool(
                    numeric_validation_df[
                        "Passed"
                    ].all()
                ),
        },
        {
            "Validation Area":
                "Bounded percentage validity",

            "Requirement":
                "Share, coverage, reach and support "
                "percentage features must remain within "
                "0–100.",

            "Observed Evidence":
                "Bounded percentage fields assessed separately "
                "from signed percentage-change features",

            "Passed":
                bool(
                    percentage_validation_df[
                        "Passed"
                    ].all()
                ),
        },
        {
            "Validation Area":
                "Artist market-count reconciliation",

            "Requirement":
                "Artist market counts must reconcile across "
                "the engineered feature families.",

            "Observed Evidence":
                "Reach counts compared with four downstream "
                "artist feature families",

            "Passed":
                bool(
                    artist_market_reconciliation_df[
                        "Passed"
                    ].all()
                ),
        },
        {
            "Validation Area":
                "Track market-count reconciliation",

            "Requirement":
                "Track market counts must reconcile across "
                "the engineered feature families.",

            "Observed Evidence":
                "Reach counts compared with four downstream "
                "track feature families",

            "Passed":
                bool(
                    track_market_reconciliation_df[
                        "Passed"
                    ].all()
                ),
        },
        {
            "Validation Area":
                "Artist trend reconciliation",

            "Requirement":
                "Artist growth, decline and stable counts "
                "must reconcile with trend-eligible markets.",

            "Observed Evidence":
                "Artist trend-market counts reconciled",

            "Passed":
                artist_growth_reconciliation,
        },
        {
            "Validation Area":
                "Track trend reconciliation",

            "Requirement":
                "Track growth, decline and stable counts "
                "must reconcile with trend-eligible markets.",

            "Observed Evidence":
                "Track trend-market counts reconciled",

            "Passed":
                track_growth_reconciliation,
        },
        {
            "Validation Area":
                "Feature-family completeness",

            "Requirement":
                "All five planned geographic feature families "
                "must be complete for artists and tracks.",

            "Observed Evidence":
                "Country reach, penetration, concentration, "
                "cross-market performance and growth/decline assessed",

            "Passed":
                bool(
                    feature_family_validation_df[
                        "Passed"
                    ].all()
                ),
        },
    ]
)

print()
print("Section 5.6 geographic feature validation")
print("=" * 100)

display(
    section_5_6_validation_df
)

# ------------------------------------------------------------
# 12. Final completion status
# ------------------------------------------------------------

section_5_6_complete = bool(
    section_5_6_validation_df[
        "Passed"
    ].all()
)

section_5_complete = section_5_6_complete

if not section_5_6_complete:

    failed_checks = (
        section_5_6_validation_df
        .loc[
            ~section_5_6_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 5.6 geographic feature validation "
        "failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 5.6 geographic feature validation "
    "checks passed."
)

print(
    f"Section 5.6 completion status: "
    f"{section_5_6_complete}"
)

print(
    f"Section 5 overall completion status: "
    f"{section_5_complete}"
)

print(
    f"Validated artist feature population: "
    f"{EXPECTED_ARTISTS:,}"
)

print(
    f"Validated track feature population: "
    f"{EXPECTED_TRACKS:,}"
)

print(
    "Five geographic feature families are validated: "
    "country reach, market penetration, geographic "
    "concentration, cross-market performance, and "
    "geographic growth/decline."
)

print(
    "The geographic features remain analytical inputs "
    "and have not yet been converted into final PMIP "
    "geographic performance scores or rankings."
)
```

### Interpretation

Section 5.6 successfully validated the full PMIP geographic feature-engineering workflow.

All five geographic feature families were confirmed as complete and internally consistent:

- **Country Reach**
- **Market Penetration**
- **Geographic Concentration**
- **Cross-Market Performance**
- **Geographic Growth and Decline**

The validation preserved the full analytical populations of **49,843 artist representations** and **110,198 tracks** across the engineered feature tables.

The checks also confirmed that entity keys remained unique, numeric feature values were valid and finite, bounded percentage fields remained within acceptable limits, and market counts reconciled correctly across related feature families.

Growth, decline and stable market counts were also successfully reconciled against the number of trend-eligible markets, confirming that the historical trend features created in Section 5.5 are structurally consistent.

Importantly, these engineered variables remain **analytical inputs rather than final PMIP geographic performance scores**. They describe different dimensions of geographic behaviour, such as reach, penetration, concentration, variation and historical change, but they have not yet been combined into a ranking or composite intelligence score.

Section 5 is therefore fully complete and the validated geographic features are ready for the next stage of the PMIP geographic intelligence workflow.

The purpose of this section is to describe country-level performance patterns before any final PMIP geographic ranking or composite score is created.


### 6.1 Streaming Performance by Country

#### Purpose

Evaluate streaming performance across the validated PMIP country markets using the country-level analytical outputs prepared earlier in the workflow.

This subsection examines total streams, typical streaming intensity, stream share, observation depth, and related market context so that stronger and weaker observed streaming markets can be identified without confusing data coverage or market size with performance quality.

The resulting measures remain descriptive country-level performance indicators and are not yet final PMIP country rankings.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP streaming performance by country")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous-section completion
# ------------------------------------------------------------

section_5_status = globals().get("section_5_complete", False)

print(
    f"Section 5 overall completion status: "
    f"{section_5_status}"
)

if not section_5_status:
    raise RuntimeError(
        "Section 6.1 requires Section 5 geographic "
        "feature engineering to be complete."
    )

# ------------------------------------------------------------
# 2. Recover the validated country-level table
# ------------------------------------------------------------

country_candidates = [
    "country_performance_df",
    "country_level_df",
    "country_summary_df",
    "country_aggregation_df",
    "country_features_df",
]

country_performance_source_df = None
country_source_name = None

for name in country_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    required_columns = {
        "country",
        "total_streams",
        "observations",
    }

    if required_columns.issubset(candidate.columns):

        country_performance_source_df = (
            candidate.copy()
        )

        country_source_name = name

        break

# ------------------------------------------------------------
# 3. Fallback recovery from notebook memory
# ------------------------------------------------------------

if country_performance_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 76:
            continue

        required_columns = {
            "country",
            "total_streams",
            "observations",
        }

        if required_columns.issubset(candidate.columns):

            country_performance_source_df = (
                candidate.copy()
            )

            country_source_name = name

            break

if country_performance_source_df is None:
    raise RuntimeError(
        "Section 6.1 could not recover the validated "
        "country-level performance aggregation from "
        "previous notebook sections."
    )

print()
print("Validated country-level analytical source")
print("=" * 100)

print(
    f"Source dataframe: {country_source_name}"
)

print(
    f"Country markets: "
    f"{len(country_performance_source_df):,}"
)

# ------------------------------------------------------------
# 4. Validate required streaming fields
# ------------------------------------------------------------

required_stream_fields = [
    "country",
    "total_streams",
    "observations",
]

optional_stream_fields = [
    "median_streams",
    "mean_streams",
    "reporting_dates",
    "unique_tracks",
    "unique_artist_representations",
    "date_coverage_pct",
    "direct_comparison_caution",
]

missing_required_fields = [
    field
    for field in required_stream_fields
    if field not in country_performance_source_df.columns
]

if missing_required_fields:
    raise RuntimeError(
        "Section 6.1 is missing required country streaming "
        "fields: "
        + ", ".join(missing_required_fields)
    )

# ------------------------------------------------------------
# 5. Prepare analysis table
# ------------------------------------------------------------

country_streaming_analysis_df = (
    country_performance_source_df.copy()
)

numeric_fields = [
    field
    for field in [
        "total_streams",
        "observations",
        "median_streams",
        "mean_streams",
        "reporting_dates",
        "unique_tracks",
        "unique_artist_representations",
        "date_coverage_pct",
    ]
    if field in country_streaming_analysis_df.columns
]

for field in numeric_fields:
    country_streaming_analysis_df[field] = (
        pd.to_numeric(
            country_streaming_analysis_df[field],
            errors="coerce",
        )
    )

# ------------------------------------------------------------
# 6. Engineer streaming context metrics
# ------------------------------------------------------------

country_streaming_analysis_df[
    "streams_per_observation"
] = (
    country_streaming_analysis_df[
        "total_streams"
    ]
    /
    country_streaming_analysis_df[
        "observations"
    ].replace(0, np.nan)
)

if "unique_tracks" in country_streaming_analysis_df.columns:

    country_streaming_analysis_df[
        "streams_per_track"
    ] = (
        country_streaming_analysis_df[
            "total_streams"
        ]
        /
        country_streaming_analysis_df[
            "unique_tracks"
        ].replace(0, np.nan)
    )

else:

    country_streaming_analysis_df[
        "streams_per_track"
    ] = np.nan

total_all_streams = (
    country_streaming_analysis_df[
        "total_streams"
    ].sum()
)

country_streaming_analysis_df[
    "stream_share_pct"
] = (
    country_streaming_analysis_df[
        "total_streams"
    ]
    /
    total_all_streams
    * 100
)

# ------------------------------------------------------------
# 7. Descriptive streaming classification
# ------------------------------------------------------------

median_market_streams = (
    country_streaming_analysis_df[
        "streams_per_observation"
    ].median()
)

q25_streams = (
    country_streaming_analysis_df[
        "streams_per_observation"
    ].quantile(0.25)
)

q75_streams = (
    country_streaming_analysis_df[
        "streams_per_observation"
    ].quantile(0.75)
)

def classify_streaming_strength(value):

    if pd.isna(value):
        return "Unavailable"

    if value >= q75_streams:
        return "High observed streaming intensity"

    if value >= median_market_streams:
        return "Above-median observed streaming intensity"

    if value >= q25_streams:
        return "Below-median observed streaming intensity"

    return "Low observed streaming intensity"


country_streaming_analysis_df[
    "streaming_intensity_class"
] = (
    country_streaming_analysis_df[
        "streams_per_observation"
    ]
    .apply(
        classify_streaming_strength
    )
)

# ------------------------------------------------------------
# 8. Country streaming overview
# ------------------------------------------------------------

print()
print("Country streaming-performance overview")
print("=" * 100)

print(
    f"Country markets assessed: "
    f"{len(country_streaming_analysis_df):,}"
)

print(
    f"Total observed streams represented: "
    f"{int(total_all_streams):,}"
)

print(
    f"Median streams per observation across countries: "
    f"{median_market_streams:,.2f}"
)

print(
    f"25th percentile streams per observation: "
    f"{q25_streams:,.2f}"
)

print(
    f"75th percentile streams per observation: "
    f"{q75_streams:,.2f}"
)

# ------------------------------------------------------------
# 9. Highest total-stream markets
# ------------------------------------------------------------

display_columns = [
    "country",
    "total_streams",
    "stream_share_pct",
    "observations",
    "streams_per_observation",
]

for field in [
    "median_streams",
    "mean_streams",
    "unique_tracks",
    "date_coverage_pct",
    "direct_comparison_caution",
]:
    if field in country_streaming_analysis_df.columns:
        display_columns.append(field)

display_columns.append(
    "streaming_intensity_class"
)

print()
print(
    "Country markets with highest total observed streams"
)
print("=" * 100)

display(
    country_streaming_analysis_df[
        display_columns
    ]
    .sort_values(
        "total_streams",
        ascending=False,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 10. Highest streaming-intensity markets
# ------------------------------------------------------------

print()
print(
    "Country markets with highest streams per observation"
)
print("=" * 100)

display(
    country_streaming_analysis_df[
        display_columns
    ]
    .sort_values(
        "streams_per_observation",
        ascending=False,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 11. Lowest supported streaming-intensity markets
# ------------------------------------------------------------

lowest_streaming_df = (
    country_streaming_analysis_df[
        display_columns
    ]
    .sort_values(
        "streams_per_observation",
        ascending=True,
    )
    .head(20)
    .reset_index(drop=True)
)

print()
print(
    "Country markets with lowest observed "
    "streams per observation"
)
print("=" * 100)

display(
    lowest_streaming_df
)

# ------------------------------------------------------------
# 12. Streaming-intensity classification summary
# ------------------------------------------------------------

streaming_classification_summary_df = (
    country_streaming_analysis_df[
        "streaming_intensity_class"
    ]
    .value_counts()
    .rename_axis(
        "Streaming Intensity Class"
    )
    .reset_index(
        name="Countries"
    )
)

streaming_classification_summary_df[
    "Share (%)"
] = (
    streaming_classification_summary_df[
        "Countries"
    ]
    /
    len(country_streaming_analysis_df)
    * 100
)

print()
print(
    "Country streaming-intensity classification"
)
print("=" * 100)

display(
    streaming_classification_summary_df
)

# ------------------------------------------------------------
# 13. Streaming concentration diagnostics
# ------------------------------------------------------------

stream_share_sorted = (
    country_streaming_analysis_df[
        "stream_share_pct"
    ]
    .sort_values(
        ascending=False
    )
    .reset_index(drop=True)
)

top_1_share = float(
    stream_share_sorted.head(1).sum()
)

top_5_share = float(
    stream_share_sorted.head(5).sum()
)

top_10_share = float(
    stream_share_sorted.head(10).sum()
)

streaming_concentration_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Largest single-country stream share",

            "Observed Value":
                top_1_share,
        },
        {
            "Diagnostic":
                "Top-5 country stream share",

            "Observed Value":
                top_5_share,
        },
        {
            "Diagnostic":
                "Top-10 country stream share",

            "Observed Value":
                top_10_share,
        },
    ]
)

print()
print(
    "Country streaming-concentration diagnostics"
)
print("=" * 100)

display(
    streaming_concentration_df
)

# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

finite_metric_fields = [
    "total_streams",
    "observations",
    "streams_per_observation",
    "stream_share_pct",
]

country_stream_numeric_valid = bool(
    np.isfinite(
        country_streaming_analysis_df[
            finite_metric_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)

stream_share_reconciles = bool(
    np.isclose(
        country_streaming_analysis_df[
            "stream_share_pct"
        ].sum(),
        100.0,
        atol=1e-6,
    )
)

country_count_valid = bool(
    len(
        country_streaming_analysis_df
    )
    == 76
)

country_unique_valid = bool(
    country_streaming_analysis_df[
        "country"
    ].nunique()
    == 76
)

section_6_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 5 completion",

            "Requirement":
                "Geographic feature engineering must "
                "already be complete.",

            "Observed Evidence":
                f"Section 5 completion status: "
                f"{section_5_status}",

            "Passed":
                section_5_status,
        },
        {
            "Validation Area":
                "Country population preservation",

            "Requirement":
                "All 76 validated country markets must "
                "be represented.",

            "Observed Evidence":
                f"{len(country_streaming_analysis_df)} "
                "country markets assessed",

            "Passed":
                country_count_valid,
        },
        {
            "Validation Area":
                "Country uniqueness",

            "Requirement":
                "Each country market must appear once "
                "in the country streaming table.",

            "Observed Evidence":
                f"{country_streaming_analysis_df['country'].nunique()} "
                "unique country values",

            "Passed":
                country_unique_valid,
        },
        {
            "Validation Area":
                "Streaming metric validity",

            "Requirement":
                "Country streaming metrics must remain "
                "finite.",

            "Observed Evidence":
                f"{len(finite_metric_fields)} "
                "core numeric streaming fields assessed",

            "Passed":
                country_stream_numeric_valid,
        },
        {
            "Validation Area":
                "Stream-share reconciliation",

            "Requirement":
                "Country stream shares must reconcile "
                "to approximately 100%.",

            "Observed Evidence":
                f"{country_streaming_analysis_df['stream_share_pct'].sum():.6f}%",

            "Passed":
                stream_share_reconciles,
        },
        {
            "Validation Area":
                "Global aggregate exclusion",

            "Requirement":
                "The global aggregate must not be treated "
                "as a country market.",

            "Observed Evidence":
                "'global' present: "
                f"{'global' in set(country_streaming_analysis_df['country'].astype(str).str.lower())}",

            "Passed":
                (
                    "global"
                    not in set(
                        country_streaming_analysis_df[
                            "country"
                        ]
                        .astype(str)
                        .str.lower()
                    )
                ),
        },
    ]
)

print()
print(
    "Country streaming-performance validation"
)
print("=" * 100)

display(
    section_6_1_validation_df
)

# ------------------------------------------------------------
# 15. Final completion status
# ------------------------------------------------------------

section_6_1_complete = bool(
    section_6_1_validation_df[
        "Passed"
    ].all()
)

if not section_6_1_complete:

    failed_checks = (
        section_6_1_validation_df
        .loc[
            ~section_6_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 6.1 country streaming-performance "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 6.1 country streaming-performance "
    "validation checks passed."
)

print(
    f"Section 6.1 completion status: "
    f"{section_6_1_complete}"
)

print(
    "Country streaming measures remain descriptive "
    "performance indicators and are not yet final "
    "PMIP country rankings."
)
```

### Interpretation

Section 6.1 successfully analysed streaming performance across all **76 validated PMIP country markets**.

The country-level source preserved the validated market population and represented approximately **1.03 trillion observed streams** across the full analytical period.

The median streaming intensity was approximately **46,040 streams per observation**, while the 25th and 75th percentile values were approximately **22,063** and **184,914 streams per observation** respectively. This confirms that streaming intensity varies substantially between country markets.

The analysis therefore provides a more useful view than total streams alone, because markets with larger observation volumes do not automatically represent stronger streaming performance. Measures such as streams per observation and stream share help separate market activity from simple dataset size.

All Section 6.1 validation checks passed, confirming that the country population, numeric streaming metrics, stream-share reconciliation and global-market exclusion remained valid.

These results remain descriptive country-level streaming indicators and have not yet been converted into final PMIP country rankings or geographic performance scores.

### 6.2 Chart Performance by Country

#### Purpose

Analyse chart performance across the validated PMIP country markets using country-level chart-position metrics prepared earlier in the workflow.

This subsection examines typical chart positions, top-10 performance, top-50 performance, number-one rates, and related country-level chart indicators to identify markets where observed chart performance is stronger or weaker.

The analysis continues to preserve the geographic comparison safeguards established in earlier sections so that countries with limited historical coverage or weaker analytical support are not interpreted too aggressively.

These measures remain descriptive chart-performance indicators and are not yet final PMIP country rankings or geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP chart performance by country")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 6.1 completion
# ------------------------------------------------------------

section_6_1_status = globals().get(
    "section_6_1_complete",
    False
)

print(
    f"Section 6.1 completion status: "
    f"{section_6_1_status}"
)

if not section_6_1_status:
    raise RuntimeError(
        "Section 6.2 requires Section 6.1 "
        "Streaming Performance by Country to be complete."
    )

# ------------------------------------------------------------
# 2. Recover validated country-level performance source
# ------------------------------------------------------------

country_source_candidates = [
    "country_performance_df",
    "country_streaming_analysis_df",
    "country_level_df",
    "country_summary_df",
]

country_chart_source_df = None
country_chart_source_name = None

required_chart_source_fields = {
    "country",
    "observations",
}

for name in country_source_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if required_chart_source_fields.issubset(
        candidate.columns
    ):
        country_chart_source_df = (
            candidate.copy()
        )

        country_chart_source_name = name
        break

# Fallback search
if country_chart_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 76:
            continue

        if required_chart_source_fields.issubset(
            candidate.columns
        ):
            country_chart_source_df = (
                candidate.copy()
            )

            country_chart_source_name = name
            break

if country_chart_source_df is None:
    raise RuntimeError(
        "Section 6.2 could not recover a validated "
        "country-level performance dataframe."
    )

print()
print("Validated country chart-performance source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{country_chart_source_name}"
)

print(
    f"Country markets: "
    f"{len(country_chart_source_df):,}"
)

# ------------------------------------------------------------
# 3. Identify chart-performance fields
# ------------------------------------------------------------

candidate_chart_fields = [
    "median_chart_position",
    "mean_chart_position",
    "best_chart_position",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "observations",
    "reporting_dates",
    "date_coverage_pct",
    "direct_comparison_caution",
]

available_chart_fields = [
    field
    for field in candidate_chart_fields
    if field in country_chart_source_df.columns
]

core_required_chart_fields = [
    "median_chart_position",
    "mean_chart_position",
]

missing_core_fields = [
    field
    for field in core_required_chart_fields
    if field not in country_chart_source_df.columns
]

if missing_core_fields:
    raise RuntimeError(
        "Section 6.2 is missing required chart-performance "
        "fields: "
        + ", ".join(missing_core_fields)
    )

# ------------------------------------------------------------
# 4. Prepare analysis dataframe
# ------------------------------------------------------------

country_chart_analysis_df = (
    country_chart_source_df.copy()
)

numeric_chart_fields = [
    field
    for field in [
        "median_chart_position",
        "mean_chart_position",
        "best_chart_position",
        "top_10_rate_pct",
        "top_50_rate_pct",
        "number_one_rate_pct",
        "observations",
        "reporting_dates",
        "date_coverage_pct",
    ]
    if field in country_chart_analysis_df.columns
]

for field in numeric_chart_fields:

    country_chart_analysis_df[field] = (
        pd.to_numeric(
            country_chart_analysis_df[field],
            errors="coerce",
        )
    )

# ------------------------------------------------------------
# 5. Engineer chart-performance context features
# ------------------------------------------------------------

if "top_10_rate_pct" in country_chart_analysis_df.columns:
    top10_median = (
        country_chart_analysis_df[
            "top_10_rate_pct"
        ].median()
    )
else:
    top10_median = np.nan

if "top_50_rate_pct" in country_chart_analysis_df.columns:
    top50_median = (
        country_chart_analysis_df[
            "top_50_rate_pct"
        ].median()
    )
else:
    top50_median = np.nan

median_position_reference = (
    country_chart_analysis_df[
        "median_chart_position"
    ].median()
)

q25_position = (
    country_chart_analysis_df[
        "median_chart_position"
    ].quantile(0.25)
)

q75_position = (
    country_chart_analysis_df[
        "median_chart_position"
    ].quantile(0.75)
)

def classify_chart_strength(value):

    if pd.isna(value):
        return "Unavailable"

    # Lower chart position is stronger
    if value <= q25_position:
        return "Strong observed chart performance"

    if value <= median_position_reference:
        return "Above-median observed chart performance"

    if value <= q75_position:
        return "Below-median observed chart performance"

    return "Weak observed chart performance"


country_chart_analysis_df[
    "chart_performance_class"
] = (
    country_chart_analysis_df[
        "median_chart_position"
    ]
    .apply(
        classify_chart_strength
    )
)

# ------------------------------------------------------------
# 6. Overview
# ------------------------------------------------------------

print()
print("Country chart-performance overview")
print("=" * 100)

print(
    f"Country markets assessed: "
    f"{len(country_chart_analysis_df):,}"
)

print(
    f"Median country median-chart position: "
    f"{median_position_reference:.2f}"
)

print(
    f"25th percentile country median-chart position: "
    f"{q25_position:.2f}"
)

print(
    f"75th percentile country median-chart position: "
    f"{q75_position:.2f}"
)

if np.isfinite(top10_median):
    print(
        f"Median country top-10 rate: "
        f"{top10_median:.2f}%"
    )

if np.isfinite(top50_median):
    print(
        f"Median country top-50 rate: "
        f"{top50_median:.2f}%"
    )

# ------------------------------------------------------------
# 7. Strongest observed chart markets
# ------------------------------------------------------------

display_columns = [
    "country",
    "median_chart_position",
    "mean_chart_position",
]

for field in [
    "best_chart_position",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "observations",
    "date_coverage_pct",
    "direct_comparison_caution",
]:
    if field in country_chart_analysis_df.columns:
        display_columns.append(field)

display_columns.append(
    "chart_performance_class"
)

print()
print(
    "Country markets with strongest observed "
    "chart-position profiles"
)
print("=" * 100)

display(
    country_chart_analysis_df[
        display_columns
    ]
    .sort_values(
        [
            "median_chart_position",
            "mean_chart_position",
        ],
        ascending=True,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 8. Weakest observed chart markets
# ------------------------------------------------------------

print()
print(
    "Country markets with weakest observed "
    "chart-position profiles"
)
print("=" * 100)

display(
    country_chart_analysis_df[
        display_columns
    ]
    .sort_values(
        [
            "median_chart_position",
            "mean_chart_position",
        ],
        ascending=False,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 9. Highest top-10-rate markets
# ------------------------------------------------------------

if "top_10_rate_pct" in country_chart_analysis_df.columns:

    print()
    print(
        "Country markets with highest observed "
        "top-10 rates"
    )
    print("=" * 100)

    display(
        country_chart_analysis_df[
            display_columns
        ]
        .sort_values(
            "top_10_rate_pct",
            ascending=False,
        )
        .head(20)
        .reset_index(drop=True)
    )

# ------------------------------------------------------------
# 10. Highest number-one-rate markets
# ------------------------------------------------------------

if "number_one_rate_pct" in country_chart_analysis_df.columns:

    print()
    print(
        "Country markets with highest observed "
        "number-one rates"
    )
    print("=" * 100)

    display(
        country_chart_analysis_df[
            display_columns
        ]
        .sort_values(
            "number_one_rate_pct",
            ascending=False,
        )
        .head(20)
        .reset_index(drop=True)
    )

# ------------------------------------------------------------
# 11. Chart-performance classification summary
# ------------------------------------------------------------

chart_classification_summary_df = (
    country_chart_analysis_df[
        "chart_performance_class"
    ]
    .value_counts()
    .rename_axis(
        "Chart Performance Class"
    )
    .reset_index(
        name="Countries"
    )
)

chart_classification_summary_df[
    "Share (%)"
] = (
    chart_classification_summary_df[
        "Countries"
    ]
    /
    len(country_chart_analysis_df)
    * 100
)

print()
print(
    "Country chart-performance classification"
)
print("=" * 100)

display(
    chart_classification_summary_df
)

# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

finite_fields = [
    "median_chart_position",
    "mean_chart_position",
]

if "top_10_rate_pct" in country_chart_analysis_df.columns:
    finite_fields.append(
        "top_10_rate_pct"
    )

if "top_50_rate_pct" in country_chart_analysis_df.columns:
    finite_fields.append(
        "top_50_rate_pct"
    )

if "number_one_rate_pct" in country_chart_analysis_df.columns:
    finite_fields.append(
        "number_one_rate_pct"
    )

chart_numeric_valid = bool(
    np.isfinite(
        country_chart_analysis_df[
            finite_fields
        ]
        .to_numpy(dtype=float)
    ).all()
)

position_bounds_valid = bool(
    (
        country_chart_analysis_df[
            "median_chart_position"
        ] > 0
    ).all()
    and
    (
        country_chart_analysis_df[
            "mean_chart_position"
        ] > 0
    ).all()
)

percentage_fields = [
    field
    for field in [
        "top_10_rate_pct",
        "top_50_rate_pct",
        "number_one_rate_pct",
    ]
    if field in country_chart_analysis_df.columns
]

percentage_bounds_valid = True

for field in percentage_fields:

    valid = (
        country_chart_analysis_df[
            field
        ]
        .between(
            0,
            100,
            inclusive="both",
        )
        .all()
    )

    percentage_bounds_valid = (
        percentage_bounds_valid
        and bool(valid)
    )

country_population_valid = bool(
    len(country_chart_analysis_df) == 76
)

country_uniqueness_valid = bool(
    country_chart_analysis_df[
        "country"
    ].nunique()
    == 76
)

global_present = (
    "global"
    in set(
        country_chart_analysis_df[
            "country"
        ]
        .astype(str)
        .str.lower()
    )
)

section_6_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 6.1 completion",

            "Requirement":
                "Streaming performance by country must "
                "already be complete.",

            "Observed Evidence":
                f"Section 6.1 completion status: "
                f"{section_6_1_status}",

            "Passed":
                section_6_1_status,
        },
        {
            "Validation Area":
                "Country population preservation",

            "Requirement":
                "All 76 validated country markets must "
                "be represented.",

            "Observed Evidence":
                f"{len(country_chart_analysis_df)} "
                "country markets assessed",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Country uniqueness",

            "Requirement":
                "Each country market must appear once.",

            "Observed Evidence":
                f"{country_chart_analysis_df['country'].nunique()} "
                "unique countries",

            "Passed":
                country_uniqueness_valid,
        },
        {
            "Validation Area":
                "Chart metric validity",

            "Requirement":
                "Chart-performance metrics must remain finite.",

            "Observed Evidence":
                f"{len(finite_fields)} "
                "numeric chart fields assessed",

            "Passed":
                chart_numeric_valid,
        },
        {
            "Validation Area":
                "Chart-position bounds",

            "Requirement":
                "Chart positions must remain positive.",

            "Observed Evidence":
                "Median and mean chart positions assessed",

            "Passed":
                position_bounds_valid,
        },
        {
            "Validation Area":
                "Chart-rate bounds",

            "Requirement":
                "Top-10, top-50 and number-one rates "
                "must remain between 0 and 100.",

            "Observed Evidence":
                f"{len(percentage_fields)} "
                "chart-rate fields assessed",

            "Passed":
                percentage_bounds_valid,
        },
        {
            "Validation Area":
                "Global aggregate exclusion",

            "Requirement":
                "The global aggregate must not be treated "
                "as a country market.",

            "Observed Evidence":
                f"'global' present: {global_present}",

            "Passed":
                not global_present,
        },
    ]
)

print()
print(
    "Country chart-performance validation"
)
print("=" * 100)

display(
    section_6_2_validation_df
)

# ------------------------------------------------------------
# 13. Final completion status
# ------------------------------------------------------------

section_6_2_complete = bool(
    section_6_2_validation_df[
        "Passed"
    ].all()
)

if not section_6_2_complete:

    failed_checks = (
        section_6_2_validation_df
        .loc[
            ~section_6_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 6.2 country chart-performance "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 6.2 country chart-performance "
    "validation checks passed."
)

print(
    f"Section 6.2 completion status: "
    f"{section_6_2_complete}"
)

print(
    "Country chart measures remain descriptive "
    "performance indicators and are not yet final "
    "PMIP country rankings."
)
```

### Interpretation

Section 6.2 successfully analysed chart performance across all **76 validated PMIP country markets**.

This shows that most country markets have fairly similar typical chart-position profiles, although some markets still perform noticeably better or worse than the overall distribution.

### 6.3 Market Size and Observation Context

#### Purpose

Assess the size and analytical context of each validated PMIP country market so that later country-performance comparisons can distinguish genuine performance differences from differences caused by observation volume, reporting coverage, or market depth.

This subsection examines measures such as observation count, reporting-date coverage, unique tracks, unique artist representations, total streams, and related support indicators.

The goal is to identify which country markets have stronger analytical support and which require more cautious interpretation because of limited historical coverage or smaller observed populations.

These context measures are not performance scores themselves. They provide the evidence needed to interpret streaming and chart results more responsibly.

```python
import pandas as pd
import numpy as np

print("Assessing PMIP market size and observation context")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_6_2_status = globals().get(
    "section_6_2_complete",
    False
)

print(
    f"Section 6.2 completion status: "
    f"{section_6_2_status}"
)

if not section_6_2_status:
    raise RuntimeError(
        "Section 6.3 requires Section 6.2 "
        "Chart Performance by Country to be complete."
    )

# ------------------------------------------------------------
# 2. Recover validated country-level source
# ------------------------------------------------------------

country_context_candidates = [
    "country_performance_df",
    "country_chart_analysis_df",
    "country_streaming_analysis_df",
]

country_context_df = None
country_context_source_name = None

required_context_fields = {
    "country",
    "observations",
}

for name in country_context_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if required_context_fields.issubset(candidate.columns):

        country_context_df = candidate.copy()
        country_context_source_name = name
        break

if country_context_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 76:
            continue

        if required_context_fields.issubset(candidate.columns):

            country_context_df = candidate.copy()
            country_context_source_name = name
            break

if country_context_df is None:
    raise RuntimeError(
        "Section 6.3 could not recover the validated "
        "country-level analytical dataframe."
    )

print()
print("Validated country market-context source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{country_context_source_name}"
)

print(
    f"Country markets: "
    f"{len(country_context_df):,}"
)

# ------------------------------------------------------------
# 3. Validate and convert context fields
# ------------------------------------------------------------

candidate_numeric_fields = [
    "observations",
    "reporting_dates",
    "unique_tracks",
    "unique_artist_representations",
    "total_streams",
    "median_streams",
    "mean_streams",
    "date_coverage_pct",
]

available_numeric_fields = [
    field
    for field in candidate_numeric_fields
    if field in country_context_df.columns
]

for field in available_numeric_fields:

    country_context_df[field] = pd.to_numeric(
        country_context_df[field],
        errors="coerce",
    )

# ------------------------------------------------------------
# 4. Engineer market-size context measures
# ------------------------------------------------------------

if "reporting_dates" in country_context_df.columns:

    max_reporting_dates = (
        country_context_df[
            "reporting_dates"
        ].max()
    )

    country_context_df[
        "reporting_date_support_pct"
    ] = (
        country_context_df[
            "reporting_dates"
        ]
        /
        max_reporting_dates
        * 100
    )

else:

    country_context_df[
        "reporting_date_support_pct"
    ] = np.nan

if (
    "unique_tracks"
    in country_context_df.columns
    and
    "observations"
    in country_context_df.columns
):

    country_context_df[
        "observations_per_track"
    ] = (
        country_context_df[
            "observations"
        ]
        /
        country_context_df[
            "unique_tracks"
        ].replace(0, np.nan)
    )

else:

    country_context_df[
        "observations_per_track"
    ] = np.nan

if (
    "unique_artist_representations"
    in country_context_df.columns
    and
    "observations"
    in country_context_df.columns
):

    country_context_df[
        "observations_per_artist"
    ] = (
        country_context_df[
            "observations"
        ]
        /
        country_context_df[
            "unique_artist_representations"
        ].replace(0, np.nan)
    )

else:

    country_context_df[
        "observations_per_artist"
    ] = np.nan

# ------------------------------------------------------------
# 5. Observation-support classification
# ------------------------------------------------------------

obs_q25 = (
    country_context_df[
        "observations"
    ].quantile(0.25)
)

obs_median = (
    country_context_df[
        "observations"
    ].median()
)

obs_q75 = (
    country_context_df[
        "observations"
    ].quantile(0.75)
)

def classify_observation_support(value):

    if pd.isna(value):
        return "Unavailable"

    if value >= obs_q75:
        return "High observation support"

    if value >= obs_median:
        return "Above-median observation support"

    if value >= obs_q25:
        return "Below-median observation support"

    return "Low observation support"


country_context_df[
    "observation_support_class"
] = (
    country_context_df[
        "observations"
    ]
    .apply(
        classify_observation_support
    )
)

# ------------------------------------------------------------
# 6. Analytical-context classification
# ------------------------------------------------------------

def classify_context(row):

    low_observation = (
        row["observations"]
        < obs_q25
    )

    low_date_coverage = False

    if "date_coverage_pct" in row.index:
        low_date_coverage = (
            pd.notna(row["date_coverage_pct"])
            and row["date_coverage_pct"] < 50
        )

    caution = False

    if "direct_comparison_caution" in row.index:
        caution = bool(
            row["direct_comparison_caution"]
        )

    issue_count = sum(
        [
            low_observation,
            low_date_coverage,
            caution,
        ]
    )

    if issue_count >= 2:
        return "High interpretation caution"

    if issue_count == 1:
        return "Moderate interpretation caution"

    return "Standard interpretation context"


country_context_df[
    "market_context_class"
] = (
    country_context_df.apply(
        classify_context,
        axis=1,
    )
)

# ------------------------------------------------------------
# 7. Market-context overview
# ------------------------------------------------------------

print()
print("Country market-size and context overview")
print("=" * 100)

print(
    f"Country markets assessed: "
    f"{len(country_context_df):,}"
)

print(
    f"Minimum observations in one country: "
    f"{int(country_context_df['observations'].min()):,}"
)

print(
    f"Median observations per country: "
    f"{country_context_df['observations'].median():,.1f}"
)

print(
    f"Maximum observations in one country: "
    f"{int(country_context_df['observations'].max()):,}"
)

print(
    f"25th percentile observations: "
    f"{obs_q25:,.1f}"
)

print(
    f"75th percentile observations: "
    f"{obs_q75:,.1f}"
)

# ------------------------------------------------------------
# 8. Largest observed country markets
# ------------------------------------------------------------

display_columns = [
    "country",
    "observations",
]

for field in [
    "reporting_dates",
    "date_coverage_pct",
    "unique_tracks",
    "unique_artist_representations",
    "total_streams",
    "observations_per_track",
    "observations_per_artist",
    "direct_comparison_caution",
]:
    if field in country_context_df.columns:
        display_columns.append(field)

display_columns.extend(
    [
        "observation_support_class",
        "market_context_class",
    ]
)

print()
print("Largest observed country markets")
print("=" * 100)

display(
    country_context_df[
        display_columns
    ]
    .sort_values(
        "observations",
        ascending=False,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 9. Smallest observed country markets
# ------------------------------------------------------------

print()
print("Smallest observed country markets")
print("=" * 100)

display(
    country_context_df[
        display_columns
    ]
    .sort_values(
        "observations",
        ascending=True,
    )
    .head(20)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 10. Observation support summary
# ------------------------------------------------------------

observation_support_summary_df = (
    country_context_df[
        "observation_support_class"
    ]
    .value_counts()
    .rename_axis(
        "Observation Support Class"
    )
    .reset_index(
        name="Countries"
    )
)

observation_support_summary_df[
    "Share (%)"
] = (
    observation_support_summary_df[
        "Countries"
    ]
    /
    len(country_context_df)
    * 100
)

print()
print("Country observation-support classification")
print("=" * 100)

display(
    observation_support_summary_df
)

# ------------------------------------------------------------
# 11. Interpretation-context summary
# ------------------------------------------------------------

market_context_summary_df = (
    country_context_df[
        "market_context_class"
    ]
    .value_counts()
    .rename_axis(
        "Market Context Class"
    )
    .reset_index(
        name="Countries"
    )
)

market_context_summary_df[
    "Share (%)"
] = (
    market_context_summary_df[
        "Countries"
    ]
    /
    len(country_context_df)
    * 100
)

print()
print("Country interpretation-context classification")
print("=" * 100)

display(
    market_context_summary_df
)

# ------------------------------------------------------------
# 12. Countries requiring strongest caution
# ------------------------------------------------------------

caution_priority = {
    "High interpretation caution": 0,
    "Moderate interpretation caution": 1,
    "Standard interpretation context": 2,
}

country_context_df[
    "_context_order"
] = (
    country_context_df[
        "market_context_class"
    ]
    .map(caution_priority)
)

print()
print("Country markets requiring strongest interpretation caution")
print("=" * 100)

display(
    country_context_df[
        display_columns + ["_context_order"]
    ]
    .sort_values(
        [
            "_context_order",
            "observations",
        ],
        ascending=[
            True,
            True,
        ],
    )
    .drop(
        columns=["_context_order"]
    )
    .head(20)
    .reset_index(drop=True)
)

country_context_df = (
    country_context_df.drop(
        columns=["_context_order"]
    )
)

# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(country_context_df) == 76
)

country_uniqueness_valid = bool(
    country_context_df[
        "country"
    ].nunique()
    == 76
)

observation_values_valid = bool(
    np.isfinite(
        country_context_df[
            "observations"
        ].to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        country_context_df[
            "observations"
        ] > 0
    ).all()
)

context_classes_valid = bool(
    country_context_df[
        "market_context_class"
    ].notna().all()
)

global_present = (
    "global"
    in set(
        country_context_df[
            "country"
        ]
        .astype(str)
        .str.lower()
    )
)

section_6_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 6.2 completion",

            "Requirement":
                "Chart performance by country must "
                "already be complete.",

            "Observed Evidence":
                f"Section 6.2 completion status: "
                f"{section_6_2_status}",

            "Passed":
                section_6_2_status,
        },
        {
            "Validation Area":
                "Country population preservation",

            "Requirement":
                "All 76 validated country markets must "
                "be represented.",

            "Observed Evidence":
                f"{len(country_context_df)} "
                "country markets assessed",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Country uniqueness",

            "Requirement":
                "Each validated country must appear once.",

            "Observed Evidence":
                f"{country_context_df['country'].nunique()} "
                "unique countries",

            "Passed":
                country_uniqueness_valid,
        },
        {
            "Validation Area":
                "Observation-count validity",

            "Requirement":
                "Observation counts must remain finite "
                "and positive.",

            "Observed Evidence":
                f"Observed range: "
                f"{int(country_context_df['observations'].min()):,} "
                f"to "
                f"{int(country_context_df['observations'].max()):,}",

            "Passed":
                observation_values_valid,
        },
        {
            "Validation Area":
                "Context classification completeness",

            "Requirement":
                "Every country must receive an "
                "interpretation-context classification.",

            "Observed Evidence":
                f"{country_context_df['market_context_class'].notna().sum()} "
                "context classifications assigned",

            "Passed":
                context_classes_valid,
        },
        {
            "Validation Area":
                "Global aggregate exclusion",

            "Requirement":
                "The global aggregate must remain outside "
                "country-level context analysis.",

            "Observed Evidence":
                f"'global' present: {global_present}",

            "Passed":
                not global_present,
        },
    ]
)

print()
print("Market-size and observation-context validation")
print("=" * 100)

display(
    section_6_3_validation_df
)

# ------------------------------------------------------------
# 14. Final completion status
# ------------------------------------------------------------

section_6_3_complete = bool(
    section_6_3_validation_df[
        "Passed"
    ].all()
)

if not section_6_3_complete:

    failed_checks = (
        section_6_3_validation_df
        .loc[
            ~section_6_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 6.3 market-size and observation-context "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 6.3 market-size and observation-context "
    "validation checks passed."
)

print(
    f"Section 6.3 completion status: "
    f"{section_6_3_complete}"
)

print(
    "Market-size and observation-context measures are "
    "interpretation safeguards and are not themselves "
    "PMIP performance scores."
)
```

### Interpretation

Section 6.3 successfully assessed market size and observation context across all **76 validated PMIP country markets**.

The analysis shows substantial differences in observation depth between countries. The smallest observed market contained **1,321 observations**, while the largest contained **90,360 observations**. The median market contained approximately **85,507 observations**, with the 25th and 75th percentiles at approximately **54,823** and **89,111 observations** respectively.

These differences matter because a country with more observations has more evidence supporting its streaming and chart-performance estimates. Smaller or less historically complete markets may therefore require greater interpretation caution even when their raw performance indicators appear strong.

All Section 6.3 validation checks passed. The full 76-country population was preserved, each country appeared once, observation counts remained valid and positive, every market received an interpretation-context classification, and the global aggregate remained excluded from country-level analysis.

The resulting market-size and observation-context measures should therefore be treated as **interpretation safeguards rather than performance scores**. They provide the context needed to judge how much confidence should be placed in later country, artist-market and track-market comparisons.

### 6.4 Strongest Artist Markets

#### Purpose

Identify the country markets in which each validated PMIP artist representation performs most strongly using the artist-country aggregation prepared earlier in the workflow.

This subsection compares artist performance across observed country markets using streaming strength, chart-position quality, observation support, and geographic comparison safeguards.

The goal is to identify the strongest observed artist-market relationships while avoiding misleading conclusions from markets with weak historical coverage, limited observation depth, or direct-comparison cautions.

These results describe observed artist-market strength and do not yet represent final PMIP artist geographic rankings.

```python
import pandas as pd
import numpy as np

print("Identifying strongest PMIP artist markets")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_6_3_status = globals().get(
    "section_6_3_complete",
    False
)

print(
    f"Section 6.3 completion status: "
    f"{section_6_3_status}"
)

if not section_6_3_status:
    raise RuntimeError(
        "Section 6.4 requires Section 6.3 "
        "Market Size and Observation Context to be complete."
    )

# ------------------------------------------------------------
# 2. Recover artist-country aggregation
# ------------------------------------------------------------

artist_country_candidates = [
    "artist_country_performance_df",
    "artist_country_df",
    "artist_country_aggregation_df",
]

artist_country_source_df = None
artist_country_source_name = None

required_fields = {
    "artists_raw",
    "country",
    "observations",
}

for name in artist_country_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if required_fields.issubset(candidate.columns):
        artist_country_source_df = candidate.copy()
        artist_country_source_name = name
        break

# Fallback notebook-memory search
if artist_country_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 184_681:
            continue

        if required_fields.issubset(candidate.columns):
            artist_country_source_df = candidate.copy()
            artist_country_source_name = name
            break

if artist_country_source_df is None:
    raise RuntimeError(
        "Section 6.4 could not recover the validated "
        "artist-country aggregation from previous sections."
    )

print()
print("Validated artist-country source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{artist_country_source_name}"
)

print(
    f"Artist-country rows: "
    f"{len(artist_country_source_df):,}"
)

# ------------------------------------------------------------
# 3. Validate and convert core fields
# ------------------------------------------------------------

numeric_fields = [
    field
    for field in [
        "total_streams",
        "observations",
        "reporting_dates",
        "median_streams",
        "mean_streams",
        "median_chart_position",
        "mean_chart_position",
        "top_10_rate_pct",
        "top_50_rate_pct",
        "number_one_rate_pct",
    ]
    if field in artist_country_source_df.columns
]

for field in numeric_fields:

    artist_country_source_df[field] = pd.to_numeric(
        artist_country_source_df[field],
        errors="coerce",
    )

required_numeric_fields = [
    "observations",
]

if "total_streams" in artist_country_source_df.columns:
    required_numeric_fields.append("total_streams")

if "median_streams" in artist_country_source_df.columns:
    required_numeric_fields.append("median_streams")

if "median_chart_position" in artist_country_source_df.columns:
    required_numeric_fields.append("median_chart_position")

missing_numeric_fields = [
    field
    for field in required_numeric_fields
    if field not in artist_country_source_df.columns
]

if missing_numeric_fields:
    raise RuntimeError(
        "Section 6.4 is missing required artist-country "
        "performance fields: "
        + ", ".join(missing_numeric_fields)
    )

# ------------------------------------------------------------
# 4. Prepare support-aware artist-market analysis
# ------------------------------------------------------------

artist_market_analysis_df = (
    artist_country_source_df.copy()
)

# Primary streaming-strength measure
if "median_streams" in artist_market_analysis_df.columns:

    artist_market_analysis_df[
        "artist_market_stream_strength"
    ] = (
        artist_market_analysis_df[
            "median_streams"
        ]
    )

else:

    artist_market_analysis_df[
        "artist_market_stream_strength"
    ] = (
        artist_market_analysis_df[
            "total_streams"
        ]
        /
        artist_market_analysis_df[
            "observations"
        ].replace(0, np.nan)
    )

# Chart strength:
# lower chart position means stronger chart performance
if "median_chart_position" in artist_market_analysis_df.columns:

    artist_market_analysis_df[
        "artist_market_chart_strength"
    ] = (
        1
        /
        artist_market_analysis_df[
            "median_chart_position"
        ].replace(0, np.nan)
    )

else:

    artist_market_analysis_df[
        "artist_market_chart_strength"
    ] = np.nan

# ------------------------------------------------------------
# 5. Artist-relative market performance
# ------------------------------------------------------------

artist_market_analysis_df[
    "artist_median_stream_strength"
] = (
    artist_market_analysis_df
    .groupby("artists_raw")[
        "artist_market_stream_strength"
    ]
    .transform("median")
)

artist_market_analysis_df[
    "relative_stream_strength"
] = (
    artist_market_analysis_df[
        "artist_market_stream_strength"
    ]
    /
    artist_market_analysis_df[
        "artist_median_stream_strength"
    ].replace(0, np.nan)
)

if "median_chart_position" in artist_market_analysis_df.columns:

    artist_market_analysis_df[
        "artist_median_chart_position"
    ] = (
        artist_market_analysis_df
        .groupby("artists_raw")[
            "median_chart_position"
        ]
        .transform("median")
    )

    artist_market_analysis_df[
        "relative_chart_strength"
    ] = (
        artist_market_analysis_df[
            "artist_median_chart_position"
        ]
        /
        artist_market_analysis_df[
            "median_chart_position"
        ].replace(0, np.nan)
    )

else:

    artist_market_analysis_df[
        "relative_chart_strength"
    ] = np.nan

# ------------------------------------------------------------
# 6. Support classification
# ------------------------------------------------------------

def classify_artist_market_support(row):

    observations = row["observations"]

    if observations >= 25:
        support = "Strong support"

    elif observations >= 10:
        support = "Moderate support"

    elif observations >= 5:
        support = "Limited support"

    else:
        support = "Very limited support"

    caution = False

    if "direct_comparison_caution" in row.index:
        caution = bool(
            row["direct_comparison_caution"]
        )

    if caution and support == "Strong support":
        return "Moderate support"

    if caution and support == "Moderate support":
        return "Limited support"

    return support


artist_market_analysis_df[
    "artist_market_support_class"
] = (
    artist_market_analysis_df.apply(
        classify_artist_market_support,
        axis=1,
    )
)

support_rank_map = {
    "Strong support": 4,
    "Moderate support": 3,
    "Limited support": 2,
    "Very limited support": 1,
}

artist_market_analysis_df[
    "support_rank"
] = (
    artist_market_analysis_df[
        "artist_market_support_class"
    ]
    .map(support_rank_map)
)

# ------------------------------------------------------------
# 7. Composite descriptive artist-market strength
# ------------------------------------------------------------

stream_component = (
    artist_market_analysis_df[
        "relative_stream_strength"
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
)

chart_component = (
    artist_market_analysis_df[
        "relative_chart_strength"
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
)

artist_market_analysis_df[
    "artist_market_strength_index"
] = (
    stream_component.fillna(1.0)
    *
    chart_component.fillna(1.0)
)

# This is a descriptive comparison index only.
# It is not yet a final PMIP geographic score.

# ------------------------------------------------------------
# 8. Determine strongest observed market per artist
# ------------------------------------------------------------

artist_market_ranked_df = (
    artist_market_analysis_df
    .sort_values(
        [
            "artists_raw",
            "support_rank",
            "artist_market_strength_index",
            "observations",
        ],
        ascending=[
            True,
            False,
            False,
            False,
        ],
    )
)

strongest_artist_markets_df = (
    artist_market_ranked_df
    .groupby(
        "artists_raw",
        as_index=False,
    )
    .first()
)

print()
print("Strongest artist-market overview")
print("=" * 100)

print(
    f"Artist representations assessed: "
    f"{strongest_artist_markets_df['artists_raw'].nunique():,}"
)

print(
    f"Artist-country combinations assessed: "
    f"{len(artist_market_analysis_df):,}"
)

print(
    f"Country markets represented: "
    f"{artist_market_analysis_df['country'].nunique():,}"
)

# ------------------------------------------------------------
# 9. Prepare display fields
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "country",
    "artist_market_stream_strength",
    "relative_stream_strength",
]

for field in [
    "median_chart_position",
    "relative_chart_strength",
    "total_streams",
    "observations",
    "reporting_dates",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "direct_comparison_caution",
]:
    if field in strongest_artist_markets_df.columns:
        display_columns.append(field)

display_columns.extend(
    [
        "artist_market_support_class",
        "artist_market_strength_index",
    ]
)

# ------------------------------------------------------------
# 10. Artists with strongest supported observed markets
# ------------------------------------------------------------

print()
print(
    "Artist representations with strongest "
    "supported observed markets"
)
print("=" * 100)

strongest_supported_display_df = (
    strongest_artist_markets_df
    .sort_values(
        [
            "support_rank",
            "artist_market_strength_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

display(
    strongest_supported_display_df[
        display_columns
    ]
)

# ------------------------------------------------------------
# 11. Strongest artist markets by streaming strength
# ------------------------------------------------------------

print()
print(
    "Artist-market combinations with strongest "
    "observed streaming intensity"
)
print("=" * 100)

strongest_streaming_artist_market_df = (
    artist_market_analysis_df
    .sort_values(
        [
            "support_rank",
            "artist_market_stream_strength",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

display(
    strongest_streaming_artist_market_df[
        display_columns
    ]
)

# ------------------------------------------------------------
# 12. Strongest artist markets by chart performance
# ------------------------------------------------------------

if "median_chart_position" in artist_market_analysis_df.columns:

    print()
    print(
        "Artist-market combinations with strongest "
        "observed chart performance"
    )
    print("=" * 100)

    strongest_chart_artist_market_df = (
        artist_market_analysis_df
        .sort_values(
            [
                "support_rank",
                "median_chart_position",
                "artist_market_strength_index",
            ],
            ascending=[
                False,
                True,
                False,
            ],
        )
        .head(25)
        .reset_index(drop=True)
    )

    display(
        strongest_chart_artist_market_df[
            display_columns
        ]
    )

# ------------------------------------------------------------
# 13. Strongest-market country distribution
# ------------------------------------------------------------

artist_strongest_country_summary_df = (
    strongest_artist_markets_df[
        "country"
    ]
    .value_counts()
    .rename_axis(
        "Country"
    )
    .reset_index(
        name="Artists for Whom Market Is Strongest"
    )
)

artist_strongest_country_summary_df[
    "Share (%)"
] = (
    artist_strongest_country_summary_df[
        "Artists for Whom Market Is Strongest"
    ]
    /
    len(strongest_artist_markets_df)
    * 100
)

print()
print(
    "Countries most frequently identified as "
    "an artist's strongest observed market"
)
print("=" * 100)

display(
    artist_strongest_country_summary_df
    .head(20)
)

# ------------------------------------------------------------
# 14. Strongest-market support distribution
# ------------------------------------------------------------

artist_market_support_summary_df = (
    strongest_artist_markets_df[
        "artist_market_support_class"
    ]
    .value_counts()
    .rename_axis(
        "Support Classification"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_market_support_summary_df[
    "Share (%)"
] = (
    artist_market_support_summary_df[
        "Artist Representations"
    ]
    /
    len(strongest_artist_markets_df)
    * 100
)

print()
print(
    "Strongest artist-market support distribution"
)
print("=" * 100)

display(
    artist_market_support_summary_df
)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    strongest_artist_markets_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

artist_market_rows_valid = bool(
    len(artist_market_analysis_df)
    == 184_681
)

country_bounds_valid = bool(
    artist_market_analysis_df[
        "country"
    ].nunique()
    <= 76
)

strength_values_valid = bool(
    np.isfinite(
        artist_market_analysis_df[
            "artist_market_strength_index"
        ]
        .to_numpy(dtype=float)
    ).all()
)

one_strongest_market_per_artist = bool(
    len(strongest_artist_markets_df)
    == 49_843
)

support_complete = bool(
    strongest_artist_markets_df[
        "artist_market_support_class"
    ].notna().all()
)

global_present = (
    "global"
    in set(
        artist_market_analysis_df[
            "country"
        ]
        .astype(str)
        .str.lower()
    )
)

section_6_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 6.3 completion",

            "Requirement":
                "Market-size and observation-context analysis "
                "must already be complete.",

            "Observed Evidence":
                f"Section 6.3 completion status: "
                f"{section_6_3_status}",

            "Passed":
                section_6_3_status,
        },
        {
            "Validation Area":
                "Artist-country population preservation",

            "Requirement":
                "All validated artist-country combinations "
                "must be represented.",

            "Observed Evidence":
                f"{len(artist_market_analysis_df):,} "
                "artist-country rows assessed",

            "Passed":
                artist_market_rows_valid,
        },
        {
            "Validation Area":
                "Artist population completeness",

            "Requirement":
                "Every validated artist representation must "
                "receive a strongest observed market.",

            "Observed Evidence":
                f"{strongest_artist_markets_df['artists_raw'].nunique():,} "
                "artist representations assigned",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "One strongest market per artist",

            "Requirement":
                "Each artist representation must receive "
                "exactly one strongest observed market.",

            "Observed Evidence":
                f"{len(strongest_artist_markets_df):,} "
                "strongest-market records created",

            "Passed":
                one_strongest_market_per_artist,
        },
        {
            "Validation Area":
                "Market-strength validity",

            "Requirement":
                "Artist-market strength indicators must "
                "remain finite.",

            "Observed Evidence":
                "Artist-market strength index assessed "
                "across all artist-country combinations",

            "Passed":
                strength_values_valid,
        },
        {
            "Validation Area":
                "Support classification completeness",

            "Requirement":
                "Every strongest artist market must retain "
                "an analytical-support classification.",

            "Observed Evidence":
                f"{strongest_artist_markets_df['artist_market_support_class'].notna().sum():,} "
                "support labels assigned",

            "Passed":
                support_complete,
        },
        {
            "Validation Area":
                "Country-market bounds",

            "Requirement":
                "Artist-market analysis must remain within "
                "the validated 76-country universe.",

            "Observed Evidence":
                f"{artist_market_analysis_df['country'].nunique()} "
                "country markets represented",

            "Passed":
                country_bounds_valid,
        },
        {
            "Validation Area":
                "Global aggregate exclusion",

            "Requirement":
                "The global aggregate must not be treated "
                "as an artist country market.",

            "Observed Evidence":
                f"'global' present: {global_present}",

            "Passed":
                not global_present,
        },
    ]
)

print()
print(
    "Strongest artist-market validation"
)
print("=" * 100)

display(
    section_6_4_validation_df
)

# ------------------------------------------------------------
# 16. Final completion status
# ------------------------------------------------------------

section_6_4_complete = bool(
    section_6_4_validation_df[
        "Passed"
    ].all()
)

if not section_6_4_complete:

    failed_checks = (
        section_6_4_validation_df
        .loc[
            ~section_6_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 6.4 strongest artist-market "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 6.4 strongest artist-market "
    "validation checks passed."
)

print(
    f"Section 6.4 completion status: "
    f"{section_6_4_complete}"
)

print(
    f"Strongest observed market assigned to "
    f"{len(strongest_artist_markets_df):,} "
    "artist representations."
)

print(
    "Artist-market strength remains a descriptive "
    "comparison output and is not yet a final PMIP "
    "artist geographic ranking."
)
```

### Interpretation

Section 6.4 successfully identified the strongest observed country market for each of the **49,843 validated artist representations** using the full set of **184,681 artist-country combinations** across **76 country markets**.

The analysis combined artist-relative streaming strength, chart-position performance and analytical support so that a market was not considered strong purely because of raw stream totals or a small number of observations.

Each artist representation received exactly one strongest observed market, and every selected market retained an analytical-support classification. This helps distinguish well-supported artist-market relationships from cases where the evidence is more limited.

All Section 6.4 validation checks passed. The artist-country population was preserved, all 49,843 artist representations received a strongest-market assignment, market-strength indicators remained finite, the analysis stayed within the validated 76-country universe, and the global aggregate remained excluded.

The resulting strongest-market assignments should therefore be interpreted as **descriptive artist-market intelligence** rather than final PMIP geographic rankings. They provide a foundation for later identifying where artists appear strongest geographically while keeping observation quality and comparison safeguards visible.

### 6.5 Strongest Track Markets

#### Purpose

Identify the country markets in which each validated PMIP track performs most strongly using the track-country aggregation prepared earlier in the workflow.

This subsection compares track performance across observed country markets using streaming intensity, chart-position quality, observation support, and the geographic comparison safeguards established in previous sections.

The goal is to determine the strongest observed track-market relationship for each validated track while avoiding misleading conclusions from markets with limited support or weak comparison context.

These results describe observed track-market strength and do not yet represent final PMIP track geographic rankings.

```python
import pandas as pd
import numpy as np

print("Identifying strongest PMIP track markets")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_6_4_status = globals().get(
    "section_6_4_complete",
    False
)

print(
    f"Section 6.4 completion status: "
    f"{section_6_4_status}"
)

if not section_6_4_status:
    raise RuntimeError(
        "Section 6.5 requires Section 6.4 "
        "Strongest Artist Markets to be complete."
    )

# ------------------------------------------------------------
# 2. Recover track-country aggregation
# ------------------------------------------------------------

track_country_candidates = [
    "track_country_performance_df",
    "track_country_df",
    "track_country_aggregation_df",
]

track_country_source_df = None
track_country_source_name = None

required_fields = {
    "track_id",
    "country",
    "observations",
}

for name in track_country_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if required_fields.issubset(candidate.columns):
        track_country_source_df = candidate.copy()
        track_country_source_name = name
        break

# Fallback notebook-memory search
if track_country_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 353_282:
            continue

        if required_fields.issubset(candidate.columns):
            track_country_source_df = candidate.copy()
            track_country_source_name = name
            break

if track_country_source_df is None:
    raise RuntimeError(
        "Section 6.5 could not recover the validated "
        "track-country aggregation from previous sections."
    )

print()
print("Validated track-country source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_country_source_name}"
)

print(
    f"Track-country rows: "
    f"{len(track_country_source_df):,}"
)

# ------------------------------------------------------------
# 3. Validate and convert core fields
# ------------------------------------------------------------

numeric_fields = [
    field
    for field in [
        "total_streams",
        "observations",
        "reporting_dates",
        "median_streams",
        "mean_streams",
        "median_chart_position",
        "mean_chart_position",
        "top_10_rate_pct",
        "top_50_rate_pct",
        "number_one_rate_pct",
    ]
    if field in track_country_source_df.columns
]

for field in numeric_fields:
    track_country_source_df[field] = pd.to_numeric(
        track_country_source_df[field],
        errors="coerce",
    )

required_numeric_fields = [
    "observations",
]

if "total_streams" in track_country_source_df.columns:
    required_numeric_fields.append("total_streams")

if "median_streams" in track_country_source_df.columns:
    required_numeric_fields.append("median_streams")

if "median_chart_position" in track_country_source_df.columns:
    required_numeric_fields.append("median_chart_position")

missing_numeric_fields = [
    field
    for field in required_numeric_fields
    if field not in track_country_source_df.columns
]

if missing_numeric_fields:
    raise RuntimeError(
        "Section 6.5 is missing required track-country "
        "performance fields: "
        + ", ".join(missing_numeric_fields)
    )

# ------------------------------------------------------------
# 4. Prepare track-market analysis
# ------------------------------------------------------------

track_market_analysis_df = (
    track_country_source_df.copy()
)

# Primary streaming-strength measure
if "median_streams" in track_market_analysis_df.columns:

    track_market_analysis_df[
        "track_market_stream_strength"
    ] = (
        track_market_analysis_df[
            "median_streams"
        ]
    )

else:

    track_market_analysis_df[
        "track_market_stream_strength"
    ] = (
        track_market_analysis_df[
            "total_streams"
        ]
        /
        track_market_analysis_df[
            "observations"
        ].replace(0, np.nan)
    )

# Chart strength:
# lower chart position means stronger performance
if "median_chart_position" in track_market_analysis_df.columns:

    track_market_analysis_df[
        "track_market_chart_strength"
    ] = (
        1
        /
        track_market_analysis_df[
            "median_chart_position"
        ].replace(0, np.nan)
    )

else:

    track_market_analysis_df[
        "track_market_chart_strength"
    ] = np.nan

# ------------------------------------------------------------
# 5. Track-relative market performance
# ------------------------------------------------------------

track_market_analysis_df[
    "track_median_stream_strength"
] = (
    track_market_analysis_df
    .groupby("track_id")[
        "track_market_stream_strength"
    ]
    .transform("median")
)

track_market_analysis_df[
    "relative_stream_strength"
] = (
    track_market_analysis_df[
        "track_market_stream_strength"
    ]
    /
    track_market_analysis_df[
        "track_median_stream_strength"
    ].replace(0, np.nan)
)

if "median_chart_position" in track_market_analysis_df.columns:

    track_market_analysis_df[
        "track_median_chart_position"
    ] = (
        track_market_analysis_df
        .groupby("track_id")[
            "median_chart_position"
        ]
        .transform("median")
    )

    track_market_analysis_df[
        "relative_chart_strength"
    ] = (
        track_market_analysis_df[
            "track_median_chart_position"
        ]
        /
        track_market_analysis_df[
            "median_chart_position"
        ].replace(0, np.nan)
    )

else:

    track_market_analysis_df[
        "relative_chart_strength"
    ] = np.nan

# ------------------------------------------------------------
# 6. Support classification
# ------------------------------------------------------------

def classify_track_market_support(row):

    observations = row["observations"]

    if observations >= 25:
        support = "Strong support"

    elif observations >= 10:
        support = "Moderate support"

    elif observations >= 5:
        support = "Limited support"

    else:
        support = "Very limited support"

    caution = False

    if "direct_comparison_caution" in row.index:
        caution = bool(
            row["direct_comparison_caution"]
        )

    if caution and support == "Strong support":
        return "Moderate support"

    if caution and support == "Moderate support":
        return "Limited support"

    return support


track_market_analysis_df[
    "track_market_support_class"
] = (
    track_market_analysis_df.apply(
        classify_track_market_support,
        axis=1,
    )
)

support_rank_map = {
    "Strong support": 4,
    "Moderate support": 3,
    "Limited support": 2,
    "Very limited support": 1,
}

track_market_analysis_df[
    "support_rank"
] = (
    track_market_analysis_df[
        "track_market_support_class"
    ]
    .map(support_rank_map)
)

# ------------------------------------------------------------
# 7. Composite descriptive track-market strength
# ------------------------------------------------------------

stream_component = (
    track_market_analysis_df[
        "relative_stream_strength"
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
)

chart_component = (
    track_market_analysis_df[
        "relative_chart_strength"
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
)

track_market_analysis_df[
    "track_market_strength_index"
] = (
    stream_component.fillna(1.0)
    *
    chart_component.fillna(1.0)
)

# Descriptive comparison index only.
# Not yet a final PMIP production score.

# ------------------------------------------------------------
# 8. Determine strongest observed market per track
# ------------------------------------------------------------

track_market_ranked_df = (
    track_market_analysis_df
    .sort_values(
        [
            "track_id",
            "support_rank",
            "track_market_strength_index",
            "observations",
        ],
        ascending=[
            True,
            False,
            False,
            False,
        ],
    )
)

strongest_track_markets_df = (
    track_market_ranked_df
    .groupby(
        "track_id",
        as_index=False,
    )
    .first()
)

print()
print("Strongest track-market overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{strongest_track_markets_df['track_id'].nunique():,}"
)

print(
    f"Track-country combinations assessed: "
    f"{len(track_market_analysis_df):,}"
)

print(
    f"Country markets represented: "
    f"{track_market_analysis_df['country'].nunique():,}"
)

# ------------------------------------------------------------
# 9. Prepare display fields
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

for field in [
    "track_name",
    "artists_raw",
]:
    if field in strongest_track_markets_df.columns:
        display_columns.append(field)

display_columns.extend(
    [
        "country",
        "track_market_stream_strength",
        "relative_stream_strength",
    ]
)

for field in [
    "median_chart_position",
    "relative_chart_strength",
    "total_streams",
    "observations",
    "reporting_dates",
    "top_10_rate_pct",
    "top_50_rate_pct",
    "number_one_rate_pct",
    "direct_comparison_caution",
]:
    if field in strongest_track_markets_df.columns:
        display_columns.append(field)

display_columns.extend(
    [
        "track_market_support_class",
        "track_market_strength_index",
    ]
)

# ------------------------------------------------------------
# 10. Tracks with strongest supported observed markets
# ------------------------------------------------------------

print()
print(
    "Tracks with strongest supported observed markets"
)
print("=" * 100)

strongest_supported_track_display_df = (
    strongest_track_markets_df
    .sort_values(
        [
            "support_rank",
            "track_market_strength_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

display(
    strongest_supported_track_display_df[
        display_columns
    ]
)

# ------------------------------------------------------------
# 11. Strongest track markets by streaming strength
# ------------------------------------------------------------

print()
print(
    "Track-market combinations with strongest "
    "observed streaming intensity"
)
print("=" * 100)

strongest_streaming_track_market_df = (
    track_market_analysis_df
    .sort_values(
        [
            "support_rank",
            "track_market_stream_strength",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

display(
    strongest_streaming_track_market_df[
        display_columns
    ]
)

# ------------------------------------------------------------
# 12. Strongest track markets by chart performance
# ------------------------------------------------------------

if "median_chart_position" in track_market_analysis_df.columns:

    print()
    print(
        "Track-market combinations with strongest "
        "observed chart performance"
    )
    print("=" * 100)

    strongest_chart_track_market_df = (
        track_market_analysis_df
        .sort_values(
            [
                "support_rank",
                "median_chart_position",
                "track_market_strength_index",
            ],
            ascending=[
                False,
                True,
                False,
            ],
        )
        .head(25)
        .reset_index(drop=True)
    )

    display(
        strongest_chart_track_market_df[
            display_columns
        ]
    )

# ------------------------------------------------------------
# 13. Strongest-market country distribution
# ------------------------------------------------------------

track_strongest_country_summary_df = (
    strongest_track_markets_df[
        "country"
    ]
    .value_counts()
    .rename_axis(
        "Country"
    )
    .reset_index(
        name="Tracks for Which Market Is Strongest"
    )
)

track_strongest_country_summary_df[
    "Share (%)"
] = (
    track_strongest_country_summary_df[
        "Tracks for Which Market Is Strongest"
    ]
    /
    len(strongest_track_markets_df)
    * 100
)

print()
print(
    "Countries most frequently identified as "
    "a track's strongest observed market"
)
print("=" * 100)

display(
    track_strongest_country_summary_df
    .head(20)
)

# ------------------------------------------------------------
# 14. Support distribution
# ------------------------------------------------------------

track_market_support_summary_df = (
    strongest_track_markets_df[
        "track_market_support_class"
    ]
    .value_counts()
    .rename_axis(
        "Support Classification"
    )
    .reset_index(
        name="Tracks"
    )
)

track_market_support_summary_df[
    "Share (%)"
] = (
    track_market_support_summary_df[
        "Tracks"
    ]
    /
    len(strongest_track_markets_df)
    * 100
)

print()
print(
    "Strongest track-market support distribution"
)
print("=" * 100)

display(
    track_market_support_summary_df
)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

track_population_valid = bool(
    strongest_track_markets_df[
        "track_id"
    ].nunique()
    == 110_198
)

track_market_rows_valid = bool(
    len(track_market_analysis_df)
    == 353_282
)

one_strongest_market_per_track = bool(
    len(strongest_track_markets_df)
    == 110_198
)

country_bounds_valid = bool(
    track_market_analysis_df[
        "country"
    ].nunique()
    <= 76
)

strength_values_valid = bool(
    np.isfinite(
        track_market_analysis_df[
            "track_market_strength_index"
        ]
        .to_numpy(dtype=float)
    ).all()
)

support_complete = bool(
    strongest_track_markets_df[
        "track_market_support_class"
    ].notna().all()
)

global_present = (
    "global"
    in set(
        track_market_analysis_df[
            "country"
        ]
        .astype(str)
        .str.lower()
    )
)

section_6_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 6.4 completion",

            "Requirement":
                "Strongest Artist Markets must "
                "already be complete.",

            "Observed Evidence":
                f"Section 6.4 completion status: "
                f"{section_6_4_status}",

            "Passed":
                section_6_4_status,
        },
        {
            "Validation Area":
                "Track-country population preservation",

            "Requirement":
                "All validated track-country combinations "
                "must be represented.",

            "Observed Evidence":
                f"{len(track_market_analysis_df):,} "
                "track-country rows assessed",

            "Passed":
                track_market_rows_valid,
        },
        {
            "Validation Area":
                "Track population completeness",

            "Requirement":
                "Every validated track must receive "
                "a strongest observed market.",

            "Observed Evidence":
                f"{strongest_track_markets_df['track_id'].nunique():,} "
                "tracks assigned",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "One strongest market per track",

            "Requirement":
                "Each track must receive exactly one "
                "strongest observed market.",

            "Observed Evidence":
                f"{len(strongest_track_markets_df):,} "
                "strongest-market records created",

            "Passed":
                one_strongest_market_per_track,
        },
        {
            "Validation Area":
                "Market-strength validity",

            "Requirement":
                "Track-market strength indicators must "
                "remain finite.",

            "Observed Evidence":
                "Track-market strength index assessed "
                "across all track-country combinations",

            "Passed":
                strength_values_valid,
        },
        {
            "Validation Area":
                "Support classification completeness",

            "Requirement":
                "Every strongest track market must retain "
                "an analytical-support classification.",

            "Observed Evidence":
                f"{strongest_track_markets_df['track_market_support_class'].notna().sum():,} "
                "support labels assigned",

            "Passed":
                support_complete,
        },
        {
            "Validation Area":
                "Country-market bounds",

            "Requirement":
                "Track-market analysis must remain within "
                "the validated 76-country universe.",

            "Observed Evidence":
                f"{track_market_analysis_df['country'].nunique()} "
                "country markets represented",

            "Passed":
                country_bounds_valid,
        },
        {
            "Validation Area":
                "Global aggregate exclusion",

            "Requirement":
                "The global aggregate must not be treated "
                "as a track country market.",

            "Observed Evidence":
                f"'global' present: {global_present}",

            "Passed":
                not global_present,
        },
    ]
)

print()
print(
    "Strongest track-market validation"
)
print("=" * 100)

display(
    section_6_5_validation_df
)

# ------------------------------------------------------------
# 16. Final completion status
# ------------------------------------------------------------

section_6_5_complete = bool(
    section_6_5_validation_df[
        "Passed"
    ].all()
)

if not section_6_5_complete:

    failed_checks = (
        section_6_5_validation_df
        .loc[
            ~section_6_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 6.5 strongest track-market "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 6.5 strongest track-market "
    "validation checks passed."
)

print(
    f"Section 6.5 completion status: "
    f"{section_6_5_complete}"
)

print(
    f"Strongest observed market assigned to "
    f"{len(strongest_track_markets_df):,} tracks."
)

print(
    "Track-market strength remains a descriptive "
    "comparison output and is not yet a final PMIP "
    "track geographic ranking."
)
```

### Interpretation

Section 6.5 successfully identified the strongest observed country market for each of the **110,198 validated tracks** using the complete set of **353,282 track-country combinations** across the validated **76-country market universe**.

Each track received exactly one strongest observed market based on its relative streaming strength, chart-position performance, observation support, and the geographic comparison safeguards carried forward from earlier sections.

All Section 6.5 validation checks passed. The complete track-country population was preserved, every validated track received a strongest-market assignment, all track-market strength indicators remained finite, support classifications were retained, and the global aggregate remained excluded from country-level track analysis.

The resulting strongest-track-market assignments should be interpreted as **descriptive geographic intelligence rather than final PMIP track rankings**. They provide a structured way to identify where individual tracks appear strongest while still accounting for differences in observation depth and comparison reliability.

With Section 6.5 complete, the full **Section 6 — Country-Level Performance Analysis** is now complete.

## 7. Artist Geographic Intelligence

### Purpose

Develop artist-level geographic intelligence by combining the validated country-reach, market-penetration, geographic-concentration, cross-market performance, and growth-related features engineered in earlier sections.

This section focuses specifically on how individual artists perform across different country markets. It evaluates whether an artist has broad international reach, how deeply they penetrate the markets they enter, whether their performance is concentrated in only a few countries, and whether they perform more strongly in local or international markets.

The final goal is to build interpretable artist geographic performance profiles that PMIP can later use for artist discovery, market analysis, geographic comparison, backend services, and dashboard visualisations.

The outputs in this section remain analytical intelligence and should not yet be treated as final PMIP production rankings unless explicitly validated in later scoring sections.

### 7.1 Artist International Reach

#### Purpose

Assess the international geographic reach of each validated artist representation using the country-reach features created in Section 5.1.

This subsection measures how widely each artist has been observed across the validated PMIP country-market universe and distinguishes artists with local, limited multi-market, broad international, or near-global reach.

The analysis also considers evidence support so that wide geographic presence is not automatically interpreted as strong international performance when observations are sparse or weakly supported.

The resulting indicators describe the breadth of an artist's observed international presence and provide the foundation for the deeper market-penetration and geographic-profile analysis that follows.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP artist international reach")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_6_5_status = globals().get(
    "section_6_5_complete",
    False
)

print(
    f"Section 6.5 completion status: "
    f"{section_6_5_status}"
)

if not section_6_5_status:
    raise RuntimeError(
        "Section 7.1 requires Section 6.5 "
        "Strongest Track Markets to be complete."
    )

# ------------------------------------------------------------
# 2. Recover validated artist country-reach feature table
# ------------------------------------------------------------

artist_reach_candidates = [
    "artist_country_reach_features_df",
    "artist_geographic_coverage_df",
    "artist_reach_features_df",
]

artist_reach_source_df = None
artist_reach_source_name = None

required_fields = {
    "artists_raw",
}

for name in artist_reach_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if len(candidate) != 49_843:
        continue

    if required_fields.issubset(candidate.columns):

        artist_reach_source_df = candidate.copy()
        artist_reach_source_name = name
        break

# Fallback notebook-memory search
if artist_reach_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 49_843:
            continue

        if "artists_raw" not in candidate.columns:
            continue

        candidate_reach_fields = {
            "country_markets_reached",
            "country_reach_pct",
        }

        if candidate_reach_fields.intersection(
            candidate.columns
        ):
            artist_reach_source_df = candidate.copy()
            artist_reach_source_name = name
            break

if artist_reach_source_df is None:
    raise RuntimeError(
        "Section 7.1 could not recover the validated "
        "artist country-reach feature table from Section 5.1."
    )

print()
print("Validated artist international-reach source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{artist_reach_source_name}"
)

print(
    f"Artist representations: "
    f"{len(artist_reach_source_df):,}"
)

# ------------------------------------------------------------
# 3. Resolve core reach columns
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


markets_reached_col = first_existing_column(
    artist_reach_source_df,
    [
        "country_markets_reached",
        "country_markets",
        "markets_reached",
        "markets_observed",
    ],
)

reach_pct_col = first_existing_column(
    artist_reach_source_df,
    [
        "country_reach_pct",
        "market_coverage_pct",
        "market_reach_pct",
    ],
)

moderate_support_col = first_existing_column(
    artist_reach_source_df,
    [
        "markets_with_moderate_or_strong_support",
        "moderate_or_strong_markets",
    ],
)

moderate_support_pct_col = first_existing_column(
    artist_reach_source_df,
    [
        "moderate_or_strong_reach_pct",
        "moderate_or_strong_market_pct",
    ],
)

strong_support_col = first_existing_column(
    artist_reach_source_df,
    [
        "markets_with_strong_support",
        "strong_support_markets",
    ],
)

comparison_safe_pct_col = first_existing_column(
    artist_reach_source_df,
    [
        "comparison_safe_reach_pct",
        "comparison_safe_market_pct",
    ],
)

if markets_reached_col is None:
    raise RuntimeError(
        "Section 7.1 requires an artist market-count field "
        "such as 'country_markets_reached'."
    )

# ------------------------------------------------------------
# 4. Prepare artist international-reach dataframe
# ------------------------------------------------------------

artist_international_reach_df = (
    artist_reach_source_df.copy()
)

numeric_candidates = [
    markets_reached_col,
    reach_pct_col,
    moderate_support_col,
    moderate_support_pct_col,
    strong_support_col,
    comparison_safe_pct_col,
]

for field in numeric_candidates:

    if field is None:
        continue

    artist_international_reach_df[field] = (
        pd.to_numeric(
            artist_international_reach_df[field],
            errors="coerce",
        )
    )

# ------------------------------------------------------------
# 5. Create canonical international-reach fields
# ------------------------------------------------------------

artist_international_reach_df[
    "international_markets_reached"
] = (
    artist_international_reach_df[
        markets_reached_col
    ]
)

if reach_pct_col is not None:

    artist_international_reach_df[
        "international_reach_pct"
    ] = (
        artist_international_reach_df[
            reach_pct_col
        ]
    )

else:

    artist_international_reach_df[
        "international_reach_pct"
    ] = (
        artist_international_reach_df[
            "international_markets_reached"
        ]
        / 76
        * 100
    )

if moderate_support_col is not None:

    artist_international_reach_df[
        "supported_markets_reached"
    ] = (
        artist_international_reach_df[
            moderate_support_col
        ]
    )

else:

    artist_international_reach_df[
        "supported_markets_reached"
    ] = np.nan

if strong_support_col is not None:

    artist_international_reach_df[
        "strong_support_markets_reached"
    ] = (
        artist_international_reach_df[
            strong_support_col
        ]
    )

else:

    artist_international_reach_df[
        "strong_support_markets_reached"
    ] = np.nan

if moderate_support_pct_col is not None:

    artist_international_reach_df[
        "supported_reach_pct"
    ] = (
        artist_international_reach_df[
            moderate_support_pct_col
        ]
    )

else:

    artist_international_reach_df[
        "supported_reach_pct"
    ] = np.nan

if comparison_safe_pct_col is not None:

    artist_international_reach_df[
        "comparison_safe_reach_pct"
    ] = (
        artist_international_reach_df[
            comparison_safe_pct_col
        ]
    )

else:

    artist_international_reach_df[
        "comparison_safe_reach_pct"
    ] = np.nan

# ------------------------------------------------------------
# 6. International-reach classification
# ------------------------------------------------------------

def classify_artist_international_reach(markets):

    if pd.isna(markets):
        return "Unavailable"

    if markets <= 1:
        return "Single-market"

    if markets <= 4:
        return "Limited multi-market"

    if markets <= 14:
        return "Moderate international"

    if markets <= 39:
        return "Broad international"

    if markets <= 59:
        return "Very broad international"

    return "Near-global reach"


artist_international_reach_df[
    "international_reach_class"
] = (
    artist_international_reach_df[
        "international_markets_reached"
    ]
    .apply(
        classify_artist_international_reach
    )
)

# ------------------------------------------------------------
# 7. International evidence-strength classification
# ------------------------------------------------------------

def classify_reach_evidence(row):

    markets = row[
        "international_markets_reached"
    ]

    supported_pct = row[
        "supported_reach_pct"
    ]

    safe_pct = row[
        "comparison_safe_reach_pct"
    ]

    if markets <= 1:
        return "Local-only evidence"

    evidence_score = 0

    if pd.notna(supported_pct):

        if supported_pct >= 75:
            evidence_score += 2

        elif supported_pct >= 40:
            evidence_score += 1

    if pd.notna(safe_pct):

        if safe_pct >= 75:
            evidence_score += 2

        elif safe_pct >= 40:
            evidence_score += 1

    if markets >= 10:
        evidence_score += 1

    if markets >= 25:
        evidence_score += 1

    if evidence_score >= 5:
        return "Strong international evidence"

    if evidence_score >= 3:
        return "Moderate international evidence"

    return "Limited international evidence"


artist_international_reach_df[
    "international_evidence_class"
] = (
    artist_international_reach_df.apply(
        classify_reach_evidence,
        axis=1,
    )
)

# ------------------------------------------------------------
# 8. Descriptive international-reach score
# ------------------------------------------------------------

artist_international_reach_df[
    "reach_breadth_component"
] = (
    artist_international_reach_df[
        "international_reach_pct"
    ]
    / 100
)

supported_component = (
    artist_international_reach_df[
        "supported_reach_pct"
    ]
    / 100
)

safe_component = (
    artist_international_reach_df[
        "comparison_safe_reach_pct"
    ]
    / 100
)

# If support fields are unavailable,
# use neutral values rather than dropping artists.
supported_component = (
    supported_component.fillna(
        artist_international_reach_df[
            "reach_breadth_component"
        ]
    )
)

safe_component = (
    safe_component.fillna(
        artist_international_reach_df[
            "reach_breadth_component"
        ]
    )
)

artist_international_reach_df[
    "international_reach_index"
] = (
    0.50
    * artist_international_reach_df[
        "reach_breadth_component"
    ]
    +
    0.30
    * supported_component
    +
    0.20
    * safe_component
)

artist_international_reach_df[
    "international_reach_index"
] = (
    artist_international_reach_df[
        "international_reach_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)

# Descriptive analytical index only.
# Not yet a final PMIP production score.

# ------------------------------------------------------------
# 9. International reach overview
# ------------------------------------------------------------

print()
print("Artist international-reach overview")
print("=" * 100)

print(
    f"Artists assessed: "
    f"{len(artist_international_reach_df):,}"
)

print(
    f"Minimum markets reached: "
    f"{int(artist_international_reach_df['international_markets_reached'].min()):,}"
)

print(
    f"Median markets reached: "
    f"{artist_international_reach_df['international_markets_reached'].median():.1f}"
)

print(
    f"Mean markets reached: "
    f"{artist_international_reach_df['international_markets_reached'].mean():.2f}"
)

print(
    f"Maximum markets reached: "
    f"{int(artist_international_reach_df['international_markets_reached'].max()):,}"
)

print(
    f"Median international reach: "
    f"{artist_international_reach_df['international_reach_pct'].median():.2f}%"
)

# ------------------------------------------------------------
# 10. Reach classification summary
# ------------------------------------------------------------

reach_class_order = [
    "Single-market",
    "Limited multi-market",
    "Moderate international",
    "Broad international",
    "Very broad international",
    "Near-global reach",
]

artist_reach_class_summary_df = (
    artist_international_reach_df[
        "international_reach_class"
    ]
    .value_counts()
    .reindex(
        reach_class_order,
        fill_value=0,
    )
    .rename_axis(
        "International Reach Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_reach_class_summary_df[
    "Share (%)"
] = (
    artist_reach_class_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_international_reach_df)
    * 100
)

print()
print("Artist international-reach classification")
print("=" * 100)

display(
    artist_reach_class_summary_df
)

# ------------------------------------------------------------
# 11. Widest observed artist reach
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "international_markets_reached",
    "international_reach_pct",
]

for field in [
    "supported_markets_reached",
    "supported_reach_pct",
    "strong_support_markets_reached",
    "comparison_safe_reach_pct",
]:
    if field in artist_international_reach_df.columns:
        display_columns.append(field)

display_columns.extend(
    [
        "international_reach_class",
        "international_evidence_class",
        "international_reach_index",
    ]
)

print()
print(
    "Artist representations with widest "
    "observed international reach"
)
print("=" * 100)

display(
    artist_international_reach_df[
        display_columns
    ]
    .sort_values(
        [
            "international_markets_reached",
            "international_reach_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 12. Strongest supported international reach
# ------------------------------------------------------------

print()
print(
    "Artist representations with strongest "
    "supported international reach"
)
print("=" * 100)

display(
    artist_international_reach_df[
        display_columns
    ]
    .sort_values(
        [
            "international_reach_index",
            "international_markets_reached",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 13. International evidence summary
# ------------------------------------------------------------

artist_international_evidence_summary_df = (
    artist_international_reach_df[
        "international_evidence_class"
    ]
    .value_counts()
    .rename_axis(
        "International Evidence Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_international_evidence_summary_df[
    "Share (%)"
] = (
    artist_international_evidence_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_international_reach_df)
    * 100
)

print()
print(
    "Artist international-evidence classification"
)
print("=" * 100)

display(
    artist_international_evidence_summary_df
)

# ------------------------------------------------------------
# 14. Geographic breadth diagnostics
# ------------------------------------------------------------

breadth_diagnostics = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Single-market artist representations",

            "Observed Value":
                int(
                    (
                        artist_international_reach_df[
                            "international_markets_reached"
                        ]
                        == 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists appearing in at least 5 markets",

            "Observed Value":
                int(
                    (
                        artist_international_reach_df[
                            "international_markets_reached"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists appearing in at least 10 markets",

            "Observed Value":
                int(
                    (
                        artist_international_reach_df[
                            "international_markets_reached"
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists appearing in at least 25 markets",

            "Observed Value":
                int(
                    (
                        artist_international_reach_df[
                            "international_markets_reached"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists appearing in at least 50 markets",

            "Observed Value":
                int(
                    (
                        artist_international_reach_df[
                            "international_markets_reached"
                        ]
                        >= 50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Maximum observed country reach",

            "Observed Value":
                int(
                    artist_international_reach_df[
                        "international_markets_reached"
                    ].max()
                ),
        },
    ]
)

print()
print("Artist international-reach diagnostics")
print("=" * 100)

display(
    breadth_diagnostics
)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    len(artist_international_reach_df)
    == 49_843
)

artist_identity_valid = bool(
    artist_international_reach_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

reach_bounds_valid = bool(
    artist_international_reach_df[
        "international_markets_reached"
    ]
    .between(
        1,
        76,
        inclusive="both",
    )
    .all()
)

reach_pct_bounds_valid = bool(
    artist_international_reach_df[
        "international_reach_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

index_valid = bool(
    np.isfinite(
        artist_international_reach_df[
            "international_reach_index"
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    artist_international_reach_df[
        "international_reach_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

classification_complete = bool(
    artist_international_reach_df[
        "international_reach_class"
    ].notna().all()
)

evidence_complete = bool(
    artist_international_reach_df[
        "international_evidence_class"
    ].notna().all()
)

section_7_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 6.5 completion",

            "Requirement":
                "Country-level performance analysis must "
                "already be complete.",

            "Observed Evidence":
                f"Section 6.5 completion status: "
                f"{section_6_5_status}",

            "Passed":
                section_6_5_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All validated artist representations must "
                "be retained.",

            "Observed Evidence":
                f"{len(artist_international_reach_df):,} "
                "artist records assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must receive "
                "one international-reach record.",

            "Observed Evidence":
                f"{artist_international_reach_df['artists_raw'].nunique():,} "
                "unique artist representations",

            "Passed":
                artist_identity_valid,
        },
        {
            "Validation Area":
                "International reach bounds",

            "Requirement":
                "Observed market reach must remain between "
                "1 and 76 validated markets.",

            "Observed Evidence":
                f"Observed range: "
                f"{int(artist_international_reach_df['international_markets_reached'].min())} "
                f"to "
                f"{int(artist_international_reach_df['international_markets_reached'].max())}",

            "Passed":
                reach_bounds_valid,
        },
        {
            "Validation Area":
                "Reach percentage bounds",

            "Requirement":
                "International reach percentages must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "International reach percentages assessed "
                "for all artist representations",

            "Passed":
                reach_pct_bounds_valid,
        },
        {
            "Validation Area":
                "Reach-index validity",

            "Requirement":
                "International reach indices must remain "
                "finite and bounded between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_international_reach_df):,} "
                "reach indices assessed",

            "Passed":
                index_valid,
        },
        {
            "Validation Area":
                "Reach classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "an international-reach class.",

            "Observed Evidence":
                f"{artist_international_reach_df['international_reach_class'].notna().sum():,} "
                "reach classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Evidence classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "an international-evidence classification.",

            "Observed Evidence":
                f"{artist_international_reach_df['international_evidence_class'].notna().sum():,} "
                "evidence classes assigned",

            "Passed":
                evidence_complete,
        },
    ]
)

print()
print("Artist international-reach validation")
print("=" * 100)

display(
    section_7_1_validation_df
)

# ------------------------------------------------------------
# 16. Final completion status
# ------------------------------------------------------------

section_7_1_complete = bool(
    section_7_1_validation_df[
        "Passed"
    ].all()
)

if not section_7_1_complete:

    failed_checks = (
        section_7_1_validation_df
        .loc[
            ~section_7_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 7.1 artist international-reach "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 7.1 artist international-reach "
    "validation checks passed."
)

print(
    f"Section 7.1 completion status: "
    f"{section_7_1_complete}"
)

print(
    f"Artist international-reach intelligence prepared "
    f"for {len(artist_international_reach_df):,} "
    "artist representations."
)

print(
    "International reach measures geographic breadth "
    "and evidence support; it is not yet a final PMIP "
    "artist geographic performance ranking."
)
```

Section 7.1 successfully assessed international geographic reach for all **49,843 validated artist representations** across the **76-country PMIP market universe**.

The results show that artist geographic presence is highly concentrated. **34,830 artist representations (69.88%)** appear in only one country market, meaning that most artists in the dataset currently have local rather than international observed reach.

A further **8,871 artists (17.80%)** have limited multi-market reach, while progressively smaller groups demonstrate broader international presence. Only **289 artist representations (0.58%)** achieved near-global reach, showing that very wide geographic coverage is relatively rare within the PMIP dataset.

The evidence-support analysis also shows that geographic breadth and analytical confidence are not the same thing. Most artists fall into either **local-only evidence** or **limited international evidence**, while only **210 artist representations (0.42%)** received a strong international-evidence classification.

All Section 7.1 validation checks passed. The complete artist population was preserved, every artist received one reach record and classification, market counts remained within the validated 1–76 range, reach percentages remained valid, and all international-reach indices were finite and bounded.

The resulting international-reach measures should therefore be interpreted as **artist geographic breadth intelligence rather than final artist performance rankings**. They describe how widely an artist is observed and how strongly that international presence is supported by the available data.

### 7.2 Artist Market Penetration

#### Purpose

Assess how deeply each validated artist penetrates the country markets in which they are observed using the market-penetration features engineered in Section 5.2.

This subsection evaluates whether an artist's presence in a market is shallow or substantial by considering relative streaming intensity, country stream share, support-qualified market coverage, and above-average market performance.

The goal is to distinguish artists who merely appear across several markets from artists who achieve stronger and more meaningful performance within those markets.

The resulting penetration indicators describe the depth of an artist's observed market presence and provide the basis for later geographic-concentration and artist-profile analysis.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP artist market penetration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_7_1_status = globals().get(
    "section_7_1_complete",
    False
)

print(
    f"Section 7.1 completion status: "
    f"{section_7_1_status}"
)

if not section_7_1_status:
    raise RuntimeError(
        "Section 7.2 requires Section 7.1 "
        "Artist International Reach to be complete."
    )

# ------------------------------------------------------------
# 2. Recover validated artist penetration feature table
# ------------------------------------------------------------

artist_penetration_candidates = [
    "artist_penetration_features_df",
    "artist_market_penetration_df",
    "artist_penetration_df",
]

artist_penetration_source_df = None
artist_penetration_source_name = None

required_fields = {
    "artists_raw",
}

for name in artist_penetration_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if len(candidate) != 49_843:
        continue

    if required_fields.issubset(candidate.columns):

        artist_penetration_source_df = candidate.copy()
        artist_penetration_source_name = name
        break

# Fallback notebook-memory search
if artist_penetration_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 49_843:
            continue

        if "artists_raw" not in candidate.columns:
            continue

        penetration_field_candidates = {
            "median_country_stream_share_pct",
            "maximum_country_stream_share_pct",
            "median_relative_stream_intensity",
            "maximum_relative_stream_intensity",
        }

        if penetration_field_candidates.intersection(
            candidate.columns
        ):
            artist_penetration_source_df = candidate.copy()
            artist_penetration_source_name = name
            break

if artist_penetration_source_df is None:
    raise RuntimeError(
        "Section 7.2 could not recover the validated "
        "artist market-penetration feature table from Section 5.2."
    )

print()
print("Validated artist market-penetration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{artist_penetration_source_name}"
)

print(
    f"Artist representations: "
    f"{len(artist_penetration_source_df):,}"
)

# ------------------------------------------------------------
# 3. Resolve core penetration fields
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


markets_observed_col = first_existing_column(
    artist_penetration_source_df,
    [
        "markets_observed",
        "country_markets_reached",
        "country_markets",
    ],
)

median_share_col = first_existing_column(
    artist_penetration_source_df,
    [
        "median_country_stream_share_pct",
        "median_market_stream_share_pct",
    ],
)

maximum_share_col = first_existing_column(
    artist_penetration_source_df,
    [
        "maximum_country_stream_share_pct",
        "max_country_stream_share_pct",
        "maximum_market_stream_share_pct",
    ],
)

median_intensity_col = first_existing_column(
    artist_penetration_source_df,
    [
        "median_relative_stream_intensity",
        "median_stream_intensity",
    ],
)

maximum_intensity_col = first_existing_column(
    artist_penetration_source_df,
    [
        "maximum_relative_stream_intensity",
        "max_relative_stream_intensity",
    ],
)

support_qualified_pct_col = first_existing_column(
    artist_penetration_source_df,
    [
        "support_qualified_market_pct",
        "supported_market_pct",
    ],
)

above_average_pct_col = first_existing_column(
    artist_penetration_source_df,
    [
        "above_country_average_market_pct",
        "above_average_market_pct",
    ],
)

if median_intensity_col is None:
    raise RuntimeError(
        "Section 7.2 requires the artist "
        "median_relative_stream_intensity field."
    )

# ------------------------------------------------------------
# 4. Prepare artist market-penetration dataframe
# ------------------------------------------------------------

artist_market_penetration_intelligence_df = (
    artist_penetration_source_df.copy()
)

numeric_fields = [
    markets_observed_col,
    median_share_col,
    maximum_share_col,
    median_intensity_col,
    maximum_intensity_col,
    support_qualified_pct_col,
    above_average_pct_col,
]

for field in numeric_fields:

    if field is None:
        continue

    artist_market_penetration_intelligence_df[field] = (
        pd.to_numeric(
            artist_market_penetration_intelligence_df[field],
            errors="coerce",
        )
    )

# ------------------------------------------------------------
# 5. Create canonical penetration fields
# ------------------------------------------------------------

if markets_observed_col is not None:

    artist_market_penetration_intelligence_df[
        "markets_observed"
    ] = (
        artist_market_penetration_intelligence_df[
            markets_observed_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "markets_observed"
    ] = np.nan

if median_share_col is not None:

    artist_market_penetration_intelligence_df[
        "median_market_stream_share_pct"
    ] = (
        artist_market_penetration_intelligence_df[
            median_share_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "median_market_stream_share_pct"
    ] = np.nan

if maximum_share_col is not None:

    artist_market_penetration_intelligence_df[
        "maximum_market_stream_share_pct"
    ] = (
        artist_market_penetration_intelligence_df[
            maximum_share_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "maximum_market_stream_share_pct"
    ] = np.nan

artist_market_penetration_intelligence_df[
    "median_relative_stream_intensity"
] = (
    artist_market_penetration_intelligence_df[
        median_intensity_col
    ]
)

if maximum_intensity_col is not None:

    artist_market_penetration_intelligence_df[
        "maximum_relative_stream_intensity"
    ] = (
        artist_market_penetration_intelligence_df[
            maximum_intensity_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "maximum_relative_stream_intensity"
    ] = (
        artist_market_penetration_intelligence_df[
            "median_relative_stream_intensity"
        ]
    )

if support_qualified_pct_col is not None:

    artist_market_penetration_intelligence_df[
        "support_qualified_market_pct"
    ] = (
        artist_market_penetration_intelligence_df[
            support_qualified_pct_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "support_qualified_market_pct"
    ] = np.nan

if above_average_pct_col is not None:

    artist_market_penetration_intelligence_df[
        "above_average_market_pct"
    ] = (
        artist_market_penetration_intelligence_df[
            above_average_pct_col
        ]
    )

else:

    artist_market_penetration_intelligence_df[
        "above_average_market_pct"
    ] = np.nan

# ------------------------------------------------------------
# 6. Penetration classification
# ------------------------------------------------------------

def classify_artist_penetration(row):

    intensity = row[
        "median_relative_stream_intensity"
    ]

    above_avg = row[
        "above_average_market_pct"
    ]

    support_pct = row[
        "support_qualified_market_pct"
    ]

    score = 0

    if pd.notna(intensity):

        if intensity >= 2.0:
            score += 3

        elif intensity >= 1.25:
            score += 2

        elif intensity >= 0.90:
            score += 1

    if pd.notna(above_avg):

        if above_avg >= 75:
            score += 2

        elif above_avg >= 40:
            score += 1

    if pd.notna(support_pct):

        if support_pct >= 75:
            score += 2

        elif support_pct >= 40:
            score += 1

    if score >= 6:
        return "Very strong penetration"

    if score >= 4:
        return "Strong penetration"

    if score >= 2:
        return "Moderate penetration"

    return "Limited penetration"


artist_market_penetration_intelligence_df[
    "market_penetration_class"
] = (
    artist_market_penetration_intelligence_df.apply(
        classify_artist_penetration,
        axis=1,
    )
)

# ------------------------------------------------------------
# 7. Penetration index
# ------------------------------------------------------------

intensity_component = (
    artist_market_penetration_intelligence_df[
        "median_relative_stream_intensity"
    ]
    .clip(lower=0)
)

# Compress large values to reduce outlier domination
intensity_component = (
    intensity_component
    /
    (1 + intensity_component)
)

support_component = (
    artist_market_penetration_intelligence_df[
        "support_qualified_market_pct"
    ]
    / 100
)

above_average_component = (
    artist_market_penetration_intelligence_df[
        "above_average_market_pct"
    ]
    / 100
)

support_component = (
    support_component.fillna(
        intensity_component
    )
)

above_average_component = (
    above_average_component.fillna(
        intensity_component
    )
)

artist_market_penetration_intelligence_df[
    "market_penetration_index"
] = (
    0.50 * intensity_component
    +
    0.25 * support_component
    +
    0.25 * above_average_component
)

artist_market_penetration_intelligence_df[
    "market_penetration_index"
] = (
    artist_market_penetration_intelligence_df[
        "market_penetration_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)

# Descriptive intelligence index only.
# Not yet a final PMIP geographic score.

# ------------------------------------------------------------
# 8. Penetration overview
# ------------------------------------------------------------

print()
print("Artist market-penetration overview")
print("=" * 100)

print(
    f"Artists assessed: "
    f"{len(artist_market_penetration_intelligence_df):,}"
)

if markets_observed_col is not None:
    print(
        f"Median markets observed: "
        f"{artist_market_penetration_intelligence_df['markets_observed'].median():.1f}"
    )

if median_share_col is not None:
    print(
        f"Median market stream share: "
        f"{artist_market_penetration_intelligence_df['median_market_stream_share_pct'].median():.6f}%"
    )

print(
    f"Median relative stream intensity: "
    f"{artist_market_penetration_intelligence_df['median_relative_stream_intensity'].median():.4f}"
)

if support_qualified_pct_col is not None:
    print(
        f"Median support-qualified market share: "
        f"{artist_market_penetration_intelligence_df['support_qualified_market_pct'].median():.2f}%"
    )

if above_average_pct_col is not None:
    print(
        f"Median above-average market share: "
        f"{artist_market_penetration_intelligence_df['above_average_market_pct'].median():.2f}%"
    )

# ------------------------------------------------------------
# 9. Penetration classification summary
# ------------------------------------------------------------

penetration_order = [
    "Limited penetration",
    "Moderate penetration",
    "Strong penetration",
    "Very strong penetration",
]

artist_penetration_class_summary_df = (
    artist_market_penetration_intelligence_df[
        "market_penetration_class"
    ]
    .value_counts()
    .reindex(
        penetration_order,
        fill_value=0,
    )
    .rename_axis(
        "Market Penetration Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_penetration_class_summary_df[
    "Share (%)"
] = (
    artist_penetration_class_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_market_penetration_intelligence_df)
    * 100
)

print()
print("Artist market-penetration classification")
print("=" * 100)

display(
    artist_penetration_class_summary_df
)

# ------------------------------------------------------------
# 10. Strongest observed artist penetration
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "markets_observed",
    "median_market_stream_share_pct",
    "maximum_market_stream_share_pct",
    "median_relative_stream_intensity",
    "maximum_relative_stream_intensity",
    "support_qualified_market_pct",
    "above_average_market_pct",
    "market_penetration_class",
    "market_penetration_index",
]

display_columns = [
    field
    for field in display_columns
    if field in artist_market_penetration_intelligence_df.columns
]

print()
print(
    "Artist representations with strongest "
    "observed market penetration"
)
print("=" * 100)

display(
    artist_market_penetration_intelligence_df[
        display_columns
    ]
    .sort_values(
        [
            "market_penetration_index",
            "median_relative_stream_intensity",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 11. Highest relative stream-intensity artists
# ------------------------------------------------------------

print()
print(
    "Artist representations with highest "
    "relative stream intensity"
)
print("=" * 100)

display(
    artist_market_penetration_intelligence_df[
        display_columns
    ]
    .sort_values(
        [
            "median_relative_stream_intensity",
            "market_penetration_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 12. Penetration diagnostics
# ------------------------------------------------------------

penetration_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Artists with median relative intensity >= 1.0",

            "Observed Value":
                int(
                    (
                        artist_market_penetration_intelligence_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 1.0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with median relative intensity >= 1.5",

            "Observed Value":
                int(
                    (
                        artist_market_penetration_intelligence_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 1.5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with median relative intensity >= 2.0",

            "Observed Value":
                int(
                    (
                        artist_market_penetration_intelligence_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 2.0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists classified strong or very strong",

            "Observed Value":
                int(
                    artist_market_penetration_intelligence_df[
                        "market_penetration_class"
                    ]
                    .isin(
                        [
                            "Strong penetration",
                            "Very strong penetration",
                        ]
                    )
                    .sum()
                ),
        },
    ]
)

print()
print("Artist market-penetration diagnostics")
print("=" * 100)

display(
    penetration_diagnostics_df
)

# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    len(
        artist_market_penetration_intelligence_df
    )
    == 49_843
)

artist_identity_valid = bool(
    artist_market_penetration_intelligence_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

intensity_valid = bool(
    np.isfinite(
        artist_market_penetration_intelligence_df[
            "median_relative_stream_intensity"
        ]
        .to_numpy(dtype=float)
    ).all()
)

penetration_index_valid = bool(
    np.isfinite(
        artist_market_penetration_intelligence_df[
            "market_penetration_index"
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    artist_market_penetration_intelligence_df[
        "market_penetration_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

classification_complete = bool(
    artist_market_penetration_intelligence_df[
        "market_penetration_class"
    ].notna().all()
)

section_7_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 7.1 completion",

            "Requirement":
                "Artist International Reach must already "
                "be complete.",

            "Observed Evidence":
                f"Section 7.1 completion status: "
                f"{section_7_1_status}",

            "Passed":
                section_7_1_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All validated artist representations must "
                "be retained.",

            "Observed Evidence":
                f"{len(artist_market_penetration_intelligence_df):,} "
                "artist penetration records assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must receive "
                "one market-penetration record.",

            "Observed Evidence":
                f"{artist_market_penetration_intelligence_df['artists_raw'].nunique():,} "
                "unique artist representations",

            "Passed":
                artist_identity_valid,
        },
        {
            "Validation Area":
                "Relative stream-intensity validity",

            "Requirement":
                "Median relative stream-intensity values "
                "must remain finite.",

            "Observed Evidence":
                f"{len(artist_market_penetration_intelligence_df):,} "
                "intensity values assessed",

            "Passed":
                intensity_valid,
        },
        {
            "Validation Area":
                "Penetration-index validity",

            "Requirement":
                "Market-penetration indices must remain "
                "finite and bounded between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_market_penetration_intelligence_df):,} "
                "penetration indices assessed",

            "Passed":
                penetration_index_valid,
        },
        {
            "Validation Area":
                "Penetration classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "a market-penetration class.",

            "Observed Evidence":
                f"{artist_market_penetration_intelligence_df['market_penetration_class'].notna().sum():,} "
                "penetration classes assigned",

            "Passed":
                classification_complete,
        },
    ]
)

print()
print("Artist market-penetration validation")
print("=" * 100)

display(
    section_7_2_validation_df
)

# ------------------------------------------------------------
# 14. Final completion status
# ------------------------------------------------------------

section_7_2_complete = bool(
    section_7_2_validation_df[
        "Passed"
    ].all()
)

if not section_7_2_complete:

    failed_checks = (
        section_7_2_validation_df
        .loc[
            ~section_7_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 7.2 artist market-penetration "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 7.2 artist market-penetration "
    "validation checks passed."
)

print(
    f"Section 7.2 completion status: "
    f"{section_7_2_complete}"
)

print(
    f"Artist market-penetration intelligence prepared "
    f"for {len(artist_market_penetration_intelligence_df):,} "
    "artist representations."
)

print(
    "Market-penetration measures the depth of observed "
    "artist performance within reached markets and is not "
    "yet a final PMIP artist geographic ranking."
)
```

### Interpretation

Section 7.2 successfully evaluated market-penetration depth for all **49,843 validated artist representations** using the artist-level penetration features engineered in Section 5.2.

A smaller but still important group demonstrates stronger market depth. **7,963 artists (15.98%)** were classified as having strong penetration and **2,369 (4.75%)** achieved very strong penetration. The median support-qualified market share and median above-average market share are both **0%**, which reflects the large number of artists with single-market or weakly supported geographic observations. This reinforces the need to interpret penetration together with international reach and observation support rather than in isolation.

The resulting market-penetration measures describe the **depth of an artist's observed performance within reached markets**. They should therefore be combined with geographic reach and concentration measures before making broader conclusions about an artist's international geographic strength.

### 7.3 Artist Geographic Concentration

#### Purpose

Assess how strongly each validated artist's observed streaming performance is concentrated within a small number of country markets using the geographic-concentration features engineered in Section 5.3.

This subsection evaluates whether an artist depends heavily on one or a few markets or whether their performance is distributed more evenly across multiple countries.

The analysis uses measures such as strongest-market stream share, top-three market stream share, stream-concentration HHI, effective market count, and the concentration classifications created earlier in the workflow.

The goal is to distinguish artists with highly localised or concentrated geographic performance from artists with more diversified international market presence.

The resulting concentration indicators will later be combined with international reach and market penetration to support richer artist geographic performance profiles.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP artist geographic concentration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_7_2_status = globals().get(
    "section_7_2_complete",
    False
)

print(
    f"Section 7.2 completion status: "
    f"{section_7_2_status}"
)

if not section_7_2_status:
    raise RuntimeError(
        "Section 7.3 requires Section 7.2 "
        "Artist Market Penetration to be complete."
    )

# ------------------------------------------------------------
# 2. Recover validated artist concentration feature table
# ------------------------------------------------------------

artist_concentration_candidates = [
    "artist_stream_concentration_df",
    "artist_concentration_features_df",
    "artist_geographic_concentration_df",
    "artist_concentration_df",
]

artist_concentration_source_df = None
artist_concentration_source_name = None

for name in artist_concentration_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if len(candidate) != 49_843:
        continue

    required_fields = {
        "artists_raw",
        "stream_concentration_hhi",
        "effective_stream_markets",
    }

    if required_fields.issubset(candidate.columns):

        artist_concentration_source_df = candidate.copy()
        artist_concentration_source_name = name
        break

# Fallback notebook-memory search
if artist_concentration_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 49_843:
            continue

        required_fields = {
            "artists_raw",
            "stream_concentration_hhi",
            "effective_stream_markets",
        }

        if required_fields.issubset(candidate.columns):

            artist_concentration_source_df = candidate.copy()
            artist_concentration_source_name = name
            break

if artist_concentration_source_df is None:
    raise RuntimeError(
        "Section 7.3 could not recover the validated "
        "artist geographic-concentration feature table "
        "from Section 5.3."
    )

print()
print("Validated artist geographic-concentration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{artist_concentration_source_name}"
)

print(
    f"Artist representations: "
    f"{len(artist_concentration_source_df):,}"
)

# ------------------------------------------------------------
# 3. Prepare artist concentration intelligence dataframe
# ------------------------------------------------------------

artist_geographic_concentration_intelligence_df = (
    artist_concentration_source_df.copy()
)

numeric_fields = [
    "markets_observed",
    "total_streams_across_markets",
    "total_observations_across_markets",
    "strongest_market_streams",
    "strongest_market_observations",
    "top_1_stream_share_pct",
    "top_1_observation_share_pct",
    "top_3_stream_share_pct",
    "top_5_stream_share_pct",
    "top_3_observation_share_pct",
    "top_5_observation_share_pct",
    "stream_concentration_hhi",
    "observation_concentration_hhi",
    "effective_stream_markets",
    "effective_observation_markets",
    "supported_market_count",
    "supported_stream_concentration_hhi",
]

for field in numeric_fields:

    if field not in artist_geographic_concentration_intelligence_df.columns:
        continue

    artist_geographic_concentration_intelligence_df[field] = (
        pd.to_numeric(
            artist_geographic_concentration_intelligence_df[field],
            errors="coerce",
        )
    )

# ------------------------------------------------------------
# 4. Create canonical concentration fields
# ------------------------------------------------------------

artist_geographic_concentration_intelligence_df[
    "strongest_market_stream_share_pct"
] = (
    artist_geographic_concentration_intelligence_df[
        "top_1_stream_share_pct"
    ]
)

artist_geographic_concentration_intelligence_df[
    "top_three_market_stream_share_pct"
] = (
    artist_geographic_concentration_intelligence_df[
        "top_3_stream_share_pct"
    ]
)

# ------------------------------------------------------------
# 5. Canonical concentration classification
# ------------------------------------------------------------

def classify_artist_concentration(hhi):

    if pd.isna(hhi):
        return "Unavailable"

    if hhi >= 0.75:
        return "Very high concentration"

    if hhi >= 0.50:
        return "High concentration"

    if hhi >= 0.25:
        return "Moderate concentration"

    return "Low concentration"


artist_geographic_concentration_intelligence_df[
    "geographic_concentration_class"
] = (
    artist_geographic_concentration_intelligence_df[
        "stream_concentration_hhi"
    ]
    .apply(
        classify_artist_concentration
    )
)

# ------------------------------------------------------------
# 6. Geographic diversification index
# ------------------------------------------------------------

artist_geographic_concentration_intelligence_df[
    "geographic_diversification_index"
] = (
    1
    -
    artist_geographic_concentration_intelligence_df[
        "stream_concentration_hhi"
    ]
)

artist_geographic_concentration_intelligence_df[
    "geographic_diversification_index"
] = (
    artist_geographic_concentration_intelligence_df[
        "geographic_diversification_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)

# ------------------------------------------------------------
# 7. Market dependency classification
# ------------------------------------------------------------

def classify_market_dependency(row):

    strongest_share = row[
        "strongest_market_stream_share_pct"
    ]

    top_three_share = row[
        "top_three_market_stream_share_pct"
    ]

    hhi = row[
        "stream_concentration_hhi"
    ]

    markets = row[
        "markets_observed"
    ]

    # Explicit single-market case
    if pd.notna(markets) and markets == 1:
        return "Single-market dependent"

    # Strong reliance on the top market
    if strongest_share >= 75:
        return "Single-market dependent"

    # Most performance contained within only a few markets
    if top_three_share >= 85:
        return "Few-market dependent"

    # Still materially concentrated
    if (
        strongest_share >= 40
        or hhi >= 0.25
    ):
        return "Moderately concentrated"

    return "Diversified"


artist_geographic_concentration_intelligence_df[
    "market_dependency_class"
] = (
    artist_geographic_concentration_intelligence_df.apply(
        classify_market_dependency,
        axis=1,
    )
)

# ------------------------------------------------------------
# 8. Geographic concentration overview
# ------------------------------------------------------------

print()
print("Artist geographic-concentration overview")
print("=" * 100)

print(
    f"Artists assessed: "
    f"{len(artist_geographic_concentration_intelligence_df):,}"
)

print(
    f"Median strongest-market stream share: "
    f"{artist_geographic_concentration_intelligence_df['strongest_market_stream_share_pct'].median():.4f}%"
)

print(
    f"Median top-three market stream share: "
    f"{artist_geographic_concentration_intelligence_df['top_three_market_stream_share_pct'].median():.4f}%"
)

print(
    f"Median stream-concentration HHI: "
    f"{artist_geographic_concentration_intelligence_df['stream_concentration_hhi'].median():.4f}"
)

print(
    f"Median effective stream markets: "
    f"{artist_geographic_concentration_intelligence_df['effective_stream_markets'].median():.2f}"
)

print(
    f"Median diversification index: "
    f"{artist_geographic_concentration_intelligence_df['geographic_diversification_index'].median():.4f}"
)

# ------------------------------------------------------------
# 9. Geographic concentration classification summary
# ------------------------------------------------------------

concentration_order = [
    "Very high concentration",
    "High concentration",
    "Moderate concentration",
    "Low concentration",
]

artist_concentration_class_summary_df = (
    artist_geographic_concentration_intelligence_df[
        "geographic_concentration_class"
    ]
    .value_counts()
    .reindex(
        concentration_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Concentration Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_concentration_class_summary_df[
    "Share (%)"
] = (
    artist_concentration_class_summary_df[
        "Artist Representations"
    ]
    /
    len(
        artist_geographic_concentration_intelligence_df
    )
    * 100
)

print()
print("Artist geographic-concentration classification")
print("=" * 100)

display(
    artist_concentration_class_summary_df
)

# ------------------------------------------------------------
# 10. Market dependency classification summary
# ------------------------------------------------------------

dependency_order = [
    "Single-market dependent",
    "Few-market dependent",
    "Moderately concentrated",
    "Diversified",
]

artist_market_dependency_summary_df = (
    artist_geographic_concentration_intelligence_df[
        "market_dependency_class"
    ]
    .value_counts()
    .reindex(
        dependency_order,
        fill_value=0,
    )
    .rename_axis(
        "Market Dependency Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_market_dependency_summary_df[
    "Share (%)"
] = (
    artist_market_dependency_summary_df[
        "Artist Representations"
    ]
    /
    len(
        artist_geographic_concentration_intelligence_df
    )
    * 100
)

print()
print("Artist market-dependency classification")
print("=" * 100)

display(
    artist_market_dependency_summary_df
)

# ------------------------------------------------------------
# 11. Market dependency reconciliation
# ------------------------------------------------------------

dependency_total = int(
    artist_market_dependency_summary_df[
        "Artist Representations"
    ].sum()
)

print()
print("Market-dependency reconciliation")
print("=" * 100)

print(
    f"Dependency records classified: "
    f"{dependency_total:,}"
)

print(
    f"Expected artist population: "
    f"{len(artist_geographic_concentration_intelligence_df):,}"
)

print(
    f"Population reconciled: "
    f"{dependency_total == len(artist_geographic_concentration_intelligence_df)}"
)

# ------------------------------------------------------------
# 12. Most geographically concentrated artists
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "markets_observed",
    "strongest_market_stream_share_pct",
    "top_three_market_stream_share_pct",
    "stream_concentration_hhi",
    "effective_stream_markets",
    "geographic_concentration_class",
    "market_dependency_class",
    "geographic_diversification_index",
]

print()
print(
    "Artist representations with highest "
    "geographic concentration"
)
print("=" * 100)

display(
    artist_geographic_concentration_intelligence_df[
        display_columns
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "strongest_market_stream_share_pct",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 13. Most geographically diversified artists
# ------------------------------------------------------------

print()
print(
    "Artist representations with greatest "
    "geographic diversification"
)
print("=" * 100)

display(
    artist_geographic_concentration_intelligence_df[
        display_columns
    ]
    .sort_values(
        [
            "stream_concentration_hhi",
            "effective_stream_markets",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 14. Geographic concentration diagnostics
# ------------------------------------------------------------

concentration_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Artists with HHI >= 0.75",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "stream_concentration_hhi"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with HHI < 0.50",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "stream_concentration_hhi"
                        ]
                        < 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with effective market count >= 5",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "effective_stream_markets"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Single-market dependent artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "market_dependency_class"
                        ]
                        == "Single-market dependent"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Few-market dependent artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "market_dependency_class"
                        ]
                        == "Few-market dependent"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Diversified artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_concentration_intelligence_df[
                            "market_dependency_class"
                        ]
                        == "Diversified"
                    ).sum()
                ),
        },
    ]
)

print()
print("Artist geographic-concentration diagnostics")
print("=" * 100)

display(
    concentration_diagnostics_df
)

# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    len(
        artist_geographic_concentration_intelligence_df
    )
    == 49_843
)

artist_identity_valid = bool(
    artist_geographic_concentration_intelligence_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

hhi_valid = bool(
    np.isfinite(
        artist_geographic_concentration_intelligence_df[
            "stream_concentration_hhi"
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    artist_geographic_concentration_intelligence_df[
        "stream_concentration_hhi"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

effective_markets_valid = bool(
    np.isfinite(
        artist_geographic_concentration_intelligence_df[
            "effective_stream_markets"
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    (
        artist_geographic_concentration_intelligence_df[
            "effective_stream_markets"
        ]
        >= 1
    ).all()
)

stream_share_valid = bool(
    np.isfinite(
        artist_geographic_concentration_intelligence_df[
            [
                "strongest_market_stream_share_pct",
                "top_three_market_stream_share_pct",
            ]
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    artist_geographic_concentration_intelligence_df[
        "strongest_market_stream_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
    and
    artist_geographic_concentration_intelligence_df[
        "top_three_market_stream_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

diversification_valid = bool(
    np.isfinite(
        artist_geographic_concentration_intelligence_df[
            "geographic_diversification_index"
        ]
        .to_numpy(dtype=float)
    ).all()
    and
    artist_geographic_concentration_intelligence_df[
        "geographic_diversification_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

concentration_class_complete = bool(
    artist_geographic_concentration_intelligence_df[
        "geographic_concentration_class"
    ].notna().all()
)

dependency_class_complete = bool(
    artist_geographic_concentration_intelligence_df[
        "market_dependency_class"
    ].notna().all()
)

dependency_population_reconciled = bool(
    dependency_total
    ==
    len(
        artist_geographic_concentration_intelligence_df
    )
)

section_7_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 7.2 completion",

            "Requirement":
                "Artist Market Penetration must already "
                "be complete.",

            "Observed Evidence":
                f"Section 7.2 completion status: "
                f"{section_7_2_status}",

            "Passed":
                section_7_2_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All validated artist representations "
                "must be retained.",

            "Observed Evidence":
                f"{len(artist_geographic_concentration_intelligence_df):,} "
                "artist concentration records assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must receive "
                "one concentration record.",

            "Observed Evidence":
                f"{artist_geographic_concentration_intelligence_df['artists_raw'].nunique():,} "
                "unique artist representations",

            "Passed":
                artist_identity_valid,
        },
        {
            "Validation Area":
                "Stream-share validity",

            "Requirement":
                "Top-1 and top-3 market stream shares must "
                "remain finite and between 0 and 100.",

            "Observed Evidence":
                "Top-1 and top-3 stream-share fields assessed",

            "Passed":
                stream_share_valid,
        },
        {
            "Validation Area":
                "HHI validity",

            "Requirement":
                "Stream-concentration HHI values must remain "
                "finite and between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_geographic_concentration_intelligence_df):,} "
                "HHI values assessed",

            "Passed":
                hhi_valid,
        },
        {
            "Validation Area":
                "Effective-market validity",

            "Requirement":
                "Effective market counts must remain finite "
                "and at least 1.",

            "Observed Evidence":
                f"{len(artist_geographic_concentration_intelligence_df):,} "
                "effective-market values assessed",

            "Passed":
                effective_markets_valid,
        },
        {
            "Validation Area":
                "Diversification-index validity",

            "Requirement":
                "Diversification indices must remain finite "
                "and between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_geographic_concentration_intelligence_df):,} "
                "diversification indices assessed",

            "Passed":
                diversification_valid,
        },
        {
            "Validation Area":
                "Concentration classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "a concentration class.",

            "Observed Evidence":
                f"{artist_geographic_concentration_intelligence_df['geographic_concentration_class'].notna().sum():,} "
                "concentration classes assigned",

            "Passed":
                concentration_class_complete,
        },
        {
            "Validation Area":
                "Market-dependency classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "a dependency class.",

            "Observed Evidence":
                f"{artist_geographic_concentration_intelligence_df['market_dependency_class'].notna().sum():,} "
                "dependency classes assigned",

            "Passed":
                dependency_class_complete,
        },
        {
            "Validation Area":
                "Dependency population reconciliation",

            "Requirement":
                "Market-dependency classifications must "
                "reconcile with the complete artist population.",

            "Observed Evidence":
                f"{dependency_total:,} classified records "
                f"for {len(artist_geographic_concentration_intelligence_df):,} artists",

            "Passed":
                dependency_population_reconciled,
        },
    ]
)

print()
print("Artist geographic-concentration validation")
print("=" * 100)

display(
    section_7_3_validation_df
)

# ------------------------------------------------------------
# 16. Final completion status
# ------------------------------------------------------------

section_7_3_complete = bool(
    section_7_3_validation_df[
        "Passed"
    ].all()
)

if not section_7_3_complete:

    failed_checks = (
        section_7_3_validation_df
        .loc[
            ~section_7_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 7.3 artist geographic-concentration "
        "validation failed for: "
        + ", ".join(failed_checks)
    )

print()
print(
    "All Section 7.3 artist geographic-concentration "
    "validation checks passed."
)

print(
    f"Section 7.3 completion status: "
    f"{section_7_3_complete}"
)

print(
    f"Artist geographic-concentration intelligence "
    f"prepared for "
    f"{len(artist_geographic_concentration_intelligence_df):,} "
    "artist representations."
)

print(
    "Geographic concentration measures how strongly "
    "artist performance is concentrated across markets "
    "and is not yet a final PMIP artist geographic ranking."
)
```

Section 7.3 successfully evaluated geographic concentration for all **49,843 validated artist representations** using the concentration features engineered in Section 5.3.

This means the typical artist representation in the dataset is heavily dependent on a very small number of markets, often just one.

The geographic-concentration classification confirms this pattern. **42,058 artist representations (84.38%)** were classified as having very high concentration, while **3,158 (6.34%)** showed high concentration. A smaller proportion displayed moderate concentration, with **2,439 artists (4.89%)**, while only **2,188 (4.39%)** were classified as having low concentration.

The concentration diagnostics support the same conclusion. **42,058 artists** have a stream-concentration HHI of at least **0.75**, while only **4,627 artists** have an HHI below **0.50**. These findings are consistent with the earlier international-reach analysis, where most artists were observed in only one market. However, geographic reach and geographic concentration represent different ideas. Reach measures **how many markets an artist appears in**, while concentration measures **how dependent the artist's performance is on particular markets**.

All Section 7.3 validation checks passed. The resulting concentration intelligence provides an important measure of geographic dependence and diversification. It will be used alongside international reach and market penetration in the next stages of the PMIP artist geographic intelligence workflow.

### 7.4 Local versus International Performance

#### Purpose

Compare each artist's observed geographic performance pattern to determine whether their streaming activity is primarily concentrated within a local or narrow set of markets, or whether meaningful performance is distributed across multiple international markets.

Because the PMIP dataset does not contain a verified artist home-country field, **local performance is not interpreted as performance inside an artist's nationality or country of origin**. Instead, this subsection uses an evidence-based geographic definition: artists whose activity is concentrated in one or a small number of markets are treated as locally concentrated, while artists with broader, supported and less concentrated multi-market activity are treated as internationally distributed.

The analysis combines the international-reach intelligence from Section 7.1, market-penetration intelligence from Section 7.2, and geographic-concentration intelligence from Section 7.3.

The objective is to distinguish localised, emerging international, internationally established and broadly diversified artist performance patterns without making unsupported claims about an artist's actual home market.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP local versus international artist performance")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_7_3_status = globals().get(
    "section_7_3_complete",
    False
)

print(
    f"Section 7.3 completion status: "
    f"{section_7_3_status}"
)

if not section_7_3_status:
    raise RuntimeError(
        "Section 7.4 requires Section 7.3 "
        "Artist Geographic Concentration to be complete."
    )

# ------------------------------------------------------------
# 2. Recover Section 7 artist intelligence tables
# ------------------------------------------------------------

def recover_dataframe(
    candidate_names,
    required_columns,
    expected_rows=49_843
):

    for name in candidate_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if set(required_columns).issubset(
            candidate.columns
        ):
            return candidate.copy(), name

    # Fallback notebook-memory search
    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if set(required_columns).issubset(
            candidate.columns
        ):
            return candidate.copy(), name

    return None, None


# ------------------------------------------------------------
# 2A. International reach source
# ------------------------------------------------------------

artist_reach_df, artist_reach_source_name = (
    recover_dataframe(
        [
            "artist_international_reach_df",
            "artist_international_reach_intelligence_df",
            "artist_country_reach_features_df",
        ],
        [
            "artists_raw",
        ],
    )
)

if artist_reach_df is None:
    raise RuntimeError(
        "Section 7.4 could not recover the artist "
        "international-reach dataframe."
    )


# ------------------------------------------------------------
# 2B. Market penetration source
# ------------------------------------------------------------

artist_penetration_df, artist_penetration_source_name = (
    recover_dataframe(
        [
            "artist_market_penetration_intelligence_df",
            "artist_penetration_features_df",
            "artist_market_penetration_df",
        ],
        [
            "artists_raw",
        ],
    )
)

if artist_penetration_df is None:
    raise RuntimeError(
        "Section 7.4 could not recover the artist "
        "market-penetration dataframe."
    )


# ------------------------------------------------------------
# 2C. Geographic concentration source
# ------------------------------------------------------------

artist_concentration_df, artist_concentration_source_name = (
    recover_dataframe(
        [
            "artist_geographic_concentration_intelligence_df",
            "artist_stream_concentration_df",
        ],
        [
            "artists_raw",
            "stream_concentration_hhi",
        ],
    )
)

if artist_concentration_df is None:
    raise RuntimeError(
        "Section 7.4 could not recover the artist "
        "geographic-concentration dataframe."
    )


print()
print("Validated Section 7 artist intelligence sources")
print("=" * 100)

print(
    f"International reach source: "
    f"{artist_reach_source_name}"
)

print(
    f"Market penetration source: "
    f"{artist_penetration_source_name}"
)

print(
    f"Geographic concentration source: "
    f"{artist_concentration_source_name}"
)

print(
    f"Artist population: "
    f"{len(artist_concentration_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve useful intelligence fields
# ------------------------------------------------------------

def first_existing_column(
    df,
    candidates
):

    for column in candidates:

        if column in df.columns:
            return column

    return None


# Reach
reach_count_col = first_existing_column(
    artist_reach_df,
    [
        "country_markets_reached",
        "markets_observed",
        "country_markets",
    ]
)

reach_pct_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_pct",
        "country_reach_pct",
        "market_reach_pct",
    ]
)

reach_class_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_class",
        "country_reach_class",
    ]
)

evidence_class_col = first_existing_column(
    artist_reach_df,
    [
        "international_evidence_class",
    ]
)


# Penetration
relative_intensity_col = first_existing_column(
    artist_penetration_df,
    [
        "median_relative_stream_intensity",
        "relative_stream_intensity",
    ]
)

penetration_class_col = first_existing_column(
    artist_penetration_df,
    [
        "market_penetration_class",
        "penetration_class",
    ]
)

support_share_col = first_existing_column(
    artist_penetration_df,
    [
        "support_qualified_market_pct",
        "support_qualified_market_share_pct",
    ]
)

above_average_share_col = first_existing_column(
    artist_penetration_df,
    [
        "above_country_average_market_pct",
        "above_average_market_pct",
        "above_average_market_share_pct",
    ]
)


# Concentration
strongest_share_col = first_existing_column(
    artist_concentration_df,
    [
        "strongest_market_stream_share_pct",
        "top_1_stream_share_pct",
    ]
)

top_three_share_col = first_existing_column(
    artist_concentration_df,
    [
        "top_three_market_stream_share_pct",
        "top_3_stream_share_pct",
    ]
)

dependency_class_col = first_existing_column(
    artist_concentration_df,
    [
        "market_dependency_class",
    ]
)

concentration_class_col = first_existing_column(
    artist_concentration_df,
    [
        "geographic_concentration_class",
    ]
)

effective_market_col = first_existing_column(
    artist_concentration_df,
    [
        "effective_stream_markets",
    ]
)


if reach_count_col is None:
    raise RuntimeError(
        "Section 7.4 requires an artist market-reach count."
    )

if strongest_share_col is None:
    raise RuntimeError(
        "Section 7.4 requires the strongest-market "
        "stream-share field."
    )

if effective_market_col is None:
    raise RuntimeError(
        "Section 7.4 requires effective stream-market counts."
    )


# ------------------------------------------------------------
# 4. Build canonical local/international dataframe
# ------------------------------------------------------------

artist_local_international_df = pd.DataFrame(
    {
        "artists_raw":
            artist_concentration_df[
                "artists_raw"
            ].astype("string")
    }
)


# ------------------------------------------------------------
# 5. Merge reach intelligence
# ------------------------------------------------------------

reach_columns = [
    "artists_raw",
    reach_count_col,
]

for optional_col in [
    reach_pct_col,
    reach_class_col,
    evidence_class_col,
]:

    if (
        optional_col is not None
        and optional_col not in reach_columns
    ):
        reach_columns.append(
            optional_col
        )


artist_local_international_df = (
    artist_local_international_df.merge(
        artist_reach_df[
            reach_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 6. Merge penetration intelligence
# ------------------------------------------------------------

penetration_columns = [
    "artists_raw",
]

for optional_col in [
    relative_intensity_col,
    penetration_class_col,
    support_share_col,
    above_average_share_col,
]:

    if (
        optional_col is not None
        and optional_col not in penetration_columns
    ):
        penetration_columns.append(
            optional_col
        )


artist_local_international_df = (
    artist_local_international_df.merge(
        artist_penetration_df[
            penetration_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 7. Merge concentration intelligence
# ------------------------------------------------------------

concentration_columns = [
    "artists_raw",
    "stream_concentration_hhi",
    effective_market_col,
    strongest_share_col,
]

for optional_col in [
    top_three_share_col,
    dependency_class_col,
    concentration_class_col,
]:

    if (
        optional_col is not None
        and optional_col not in concentration_columns
    ):
        concentration_columns.append(
            optional_col
        )


artist_local_international_df = (
    artist_local_international_df.merge(
        artist_concentration_df[
            concentration_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 8. Create canonical field names
# ------------------------------------------------------------

artist_local_international_df[
    "markets_reached"
] = pd.to_numeric(
    artist_local_international_df[
        reach_count_col
    ],
    errors="coerce",
)

artist_local_international_df[
    "strongest_market_stream_share_pct"
] = pd.to_numeric(
    artist_local_international_df[
        strongest_share_col
    ],
    errors="coerce",
)

artist_local_international_df[
    "effective_stream_markets"
] = pd.to_numeric(
    artist_local_international_df[
        effective_market_col
    ],
    errors="coerce",
)

artist_local_international_df[
    "stream_concentration_hhi"
] = pd.to_numeric(
    artist_local_international_df[
        "stream_concentration_hhi"
    ],
    errors="coerce",
)


if relative_intensity_col is not None:

    artist_local_international_df[
        "median_relative_stream_intensity"
    ] = pd.to_numeric(
        artist_local_international_df[
            relative_intensity_col
        ],
        errors="coerce",
    )

else:

    artist_local_international_df[
        "median_relative_stream_intensity"
    ] = np.nan


if support_share_col is not None:

    artist_local_international_df[
        "support_qualified_market_pct"
    ] = pd.to_numeric(
        artist_local_international_df[
            support_share_col
        ],
        errors="coerce",
    )

else:

    artist_local_international_df[
        "support_qualified_market_pct"
    ] = np.nan


# ------------------------------------------------------------
# 9. Calculate international distribution measures
# ------------------------------------------------------------

TOTAL_VALIDATED_MARKETS = 76

artist_local_international_df[
    "international_market_reach_pct"
] = (
    artist_local_international_df[
        "markets_reached"
    ]
    /
    TOTAL_VALIDATED_MARKETS
    * 100
)

artist_local_international_df[
    "non_dominant_market_stream_share_pct"
] = (
    100
    -
    artist_local_international_df[
        "strongest_market_stream_share_pct"
    ]
)

artist_local_international_df[
    "non_dominant_market_stream_share_pct"
] = (
    artist_local_international_df[
        "non_dominant_market_stream_share_pct"
    ]
    .clip(
        lower=0,
        upper=100,
    )
)

artist_local_international_df[
    "geographic_diversification_index"
] = (
    1
    -
    artist_local_international_df[
        "stream_concentration_hhi"
    ]
)

artist_local_international_df[
    "geographic_diversification_index"
] = (
    artist_local_international_df[
        "geographic_diversification_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 10. Local versus international classification
# ------------------------------------------------------------

def classify_local_international(row):

    markets = row[
        "markets_reached"
    ]

    strongest_share = row[
        "strongest_market_stream_share_pct"
    ]

    effective_markets = row[
        "effective_stream_markets"
    ]

    hhi = row[
        "stream_concentration_hhi"
    ]

    if any(
        pd.isna(value)
        for value in [
            markets,
            strongest_share,
            effective_markets,
            hhi,
        ]
    ):
        return "Insufficient evidence"

    # One-market or overwhelmingly one-market performance
    if (
        markets == 1
        or strongest_share >= 85
    ):
        return "Localised performance"

    # Multi-market but still heavily concentrated
    if (
        markets < 5
        or effective_markets < 2
        or hhi >= 0.65
    ):
        return "Emerging international"

    # Meaningful international breadth with some concentration
    if (
        markets >= 5
        and effective_markets >= 2
        and hhi < 0.65
    ):
        return "Internationally established"

    return "Emerging international"


artist_local_international_df[
    "local_international_profile"
] = (
    artist_local_international_df.apply(
        classify_local_international,
        axis=1,
    )
)


# ------------------------------------------------------------
# 11. Broad international diversification refinement
# ------------------------------------------------------------

broad_international_mask = (
    (
        artist_local_international_df[
            "markets_reached"
        ]
        >= 25
    )
    &
    (
        artist_local_international_df[
            "effective_stream_markets"
        ]
        >= 5
    )
    &
    (
        artist_local_international_df[
            "stream_concentration_hhi"
        ]
        < 0.35
    )
)

artist_local_international_df.loc[
    broad_international_mask,
    "local_international_profile",
] = "Broadly diversified international"


# ------------------------------------------------------------
# 12. Local/international index
# ------------------------------------------------------------

reach_component = (
    artist_local_international_df[
        "international_market_reach_pct"
    ]
    / 100
)

diversification_component = (
    artist_local_international_df[
        "geographic_diversification_index"
    ]
)

effective_market_component = (
    artist_local_international_df[
        "effective_stream_markets"
    ]
    /
    TOTAL_VALIDATED_MARKETS
)

effective_market_component = (
    effective_market_component.clip(
        lower=0,
        upper=1,
    )
)

artist_local_international_df[
    "international_distribution_index"
] = (
    (
        0.40
        * reach_component
    )
    +
    (
        0.40
        * diversification_component
    )
    +
    (
        0.20
        * effective_market_component
    )
)

artist_local_international_df[
    "international_distribution_index"
] = (
    artist_local_international_df[
        "international_distribution_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 13. Overview
# ------------------------------------------------------------

print()
print("Artist local versus international performance overview")
print("=" * 100)

print(
    f"Artists assessed: "
    f"{len(artist_local_international_df):,}"
)

print(
    f"Median markets reached: "
    f"{artist_local_international_df['markets_reached'].median():.1f}"
)

print(
    f"Median strongest-market stream share: "
    f"{artist_local_international_df['strongest_market_stream_share_pct'].median():.2f}%"
)

print(
    f"Median effective stream markets: "
    f"{artist_local_international_df['effective_stream_markets'].median():.2f}"
)

print(
    f"Median international distribution index: "
    f"{artist_local_international_df['international_distribution_index'].median():.4f}"
)


# ------------------------------------------------------------
# 14. Profile distribution
# ------------------------------------------------------------

profile_order = [
    "Localised performance",
    "Emerging international",
    "Internationally established",
    "Broadly diversified international",
    "Insufficient evidence",
]

artist_local_international_summary_df = (
    artist_local_international_df[
        "local_international_profile"
    ]
    .value_counts()
    .reindex(
        profile_order,
        fill_value=0,
    )
    .rename_axis(
        "Local / International Profile"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_local_international_summary_df[
    "Share (%)"
] = (
    artist_local_international_summary_df[
        "Artist Representations"
    ]
    /
    len(
        artist_local_international_df
    )
    * 100
)

print()
print("Artist local versus international classification")
print("=" * 100)

display(
    artist_local_international_summary_df
)


# ------------------------------------------------------------
# 15. Strongest international-distribution profiles
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "markets_reached",
    "international_market_reach_pct",
    "strongest_market_stream_share_pct",
    "non_dominant_market_stream_share_pct",
    "stream_concentration_hhi",
    "effective_stream_markets",
    "geographic_diversification_index",
    "international_distribution_index",
    "local_international_profile",
]

if penetration_class_col is not None:
    display_columns.insert(
        -1,
        penetration_class_col,
    )

if reach_class_col is not None:
    display_columns.insert(
        -1,
        reach_class_col,
    )


print()
print(
    "Artist representations with strongest "
    "international distribution"
)
print("=" * 100)

display(
    artist_local_international_df[
        display_columns
    ]
    .sort_values(
        [
            "international_distribution_index",
            "markets_reached",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 16. Most localised artist profiles
# ------------------------------------------------------------

print()
print(
    "Artist representations with strongest "
    "localised performance"
)
print("=" * 100)

display(
    artist_local_international_df[
        display_columns
    ]
    .sort_values(
        [
            "international_distribution_index",
            "strongest_market_stream_share_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 17. Diagnostics
# ------------------------------------------------------------

local_international_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Single-market artist representations",

            "Observed Value":
                int(
                    (
                        artist_local_international_df[
                            "markets_reached"
                        ]
                        == 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists reaching at least 5 markets",

            "Observed Value":
                int(
                    (
                        artist_local_international_df[
                            "markets_reached"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists reaching at least 25 markets",

            "Observed Value":
                int(
                    (
                        artist_local_international_df[
                            "markets_reached"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with effective stream markets >= 5",

            "Observed Value":
                int(
                    (
                        artist_local_international_df[
                            "effective_stream_markets"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with strongest market share >= 85%",

            "Observed Value":
                int(
                    (
                        artist_local_international_df[
                            "strongest_market_stream_share_pct"
                        ]
                        >= 85
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Internationally established or broader",

            "Observed Value":
                int(
                    artist_local_international_df[
                        "local_international_profile"
                    ]
                    .isin(
                        [
                            "Internationally established",
                            "Broadly diversified international",
                        ]
                    )
                    .sum()
                ),
        },
    ]
)

print()
print("Local versus international performance diagnostics")
print("=" * 100)

display(
    local_international_diagnostics_df
)


# ------------------------------------------------------------
# 18. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    len(
        artist_local_international_df
    )
    == 49_843
)

artist_identity_valid = bool(
    artist_local_international_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

core_metrics_complete = bool(
    artist_local_international_df[
        [
            "markets_reached",
            "strongest_market_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
        ]
    ]
    .notna()
    .all()
    .all()
)

reach_bounds_valid = bool(
    artist_local_international_df[
        "markets_reached"
    ]
    .between(
        1,
        TOTAL_VALIDATED_MARKETS,
        inclusive="both",
    )
    .all()
)

strongest_share_valid = bool(
    artist_local_international_df[
        "strongest_market_stream_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

hhi_valid = bool(
    artist_local_international_df[
        "stream_concentration_hhi"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

distribution_index_valid = bool(
    np.isfinite(
        artist_local_international_df[
            "international_distribution_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    artist_local_international_df[
        "international_distribution_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

profile_complete = bool(
    artist_local_international_df[
        "local_international_profile"
    ]
    .notna()
    .all()
)

profile_total = int(
    artist_local_international_summary_df[
        "Artist Representations"
    ].sum()
)

profile_population_reconciled = bool(
    profile_total
    ==
    len(
        artist_local_international_df
    )
)


section_7_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 7.3 completion",

            "Requirement":
                "Artist Geographic Concentration must "
                "already be complete.",

            "Observed Evidence":
                f"Section 7.3 completion status: "
                f"{section_7_3_status}",

            "Passed":
                section_7_3_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All validated artist representations "
                "must be retained.",

            "Observed Evidence":
                f"{len(artist_local_international_df):,} "
                "artist records assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must receive "
                "one local/international record.",

            "Observed Evidence":
                f"{artist_local_international_df['artists_raw'].nunique():,} "
                "unique artist representations",

            "Passed":
                artist_identity_valid,
        },
        {
            "Validation Area":
                "Core geographic metric completeness",

            "Requirement":
                "Reach, concentration and effective-market "
                "measures must remain available.",

            "Observed Evidence":
                "Reach and concentration fields assessed "
                "for all artist records",

            "Passed":
                core_metrics_complete,
        },
        {
            "Validation Area":
                "Market-reach bounds",

            "Requirement":
                "Markets reached must remain between "
                "1 and 76.",

            "Observed Evidence":
                f"Observed range: "
                f"{artist_local_international_df['markets_reached'].min():.0f} "
                f"to "
                f"{artist_local_international_df['markets_reached'].max():.0f}",

            "Passed":
                reach_bounds_valid,
        },
        {
            "Validation Area":
                "Strongest-market share validity",

            "Requirement":
                "Strongest-market stream shares must "
                "remain between 0 and 100.",

            "Observed Evidence":
                "Strongest-market stream shares assessed",

            "Passed":
                strongest_share_valid,
        },
        {
            "Validation Area":
                "Concentration validity",

            "Requirement":
                "Stream-concentration HHI must remain "
                "between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_local_international_df):,} "
                "HHI values assessed",

            "Passed":
                hhi_valid,
        },
        {
            "Validation Area":
                "International-distribution index validity",

            "Requirement":
                "Distribution indices must remain finite "
                "and between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_local_international_df):,} "
                "distribution indices assessed",

            "Passed":
                distribution_index_valid,
        },
        {
            "Validation Area":
                "Profile classification completeness",

            "Requirement":
                "Every artist must receive a local versus "
                "international profile.",

            "Observed Evidence":
                f"{artist_local_international_df['local_international_profile'].notna().sum():,} "
                "profiles assigned",

            "Passed":
                profile_complete,
        },
        {
            "Validation Area":
                "Profile population reconciliation",

            "Requirement":
                "Profile classifications must reconcile "
                "with the full artist population.",

            "Observed Evidence":
                f"{profile_total:,} classified records "
                f"for {len(artist_local_international_df):,} artists",

            "Passed":
                profile_population_reconciled,
        },
    ]
)

print()
print("Artist local versus international performance validation")
print("=" * 100)

display(
    section_7_4_validation_df
)


# ------------------------------------------------------------
# 19. Final completion status
# ------------------------------------------------------------

section_7_4_complete = bool(
    section_7_4_validation_df[
        "Passed"
    ].all()
)

if not section_7_4_complete:

    failed_checks = (
        section_7_4_validation_df
        .loc[
            ~section_7_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 7.4 local versus international "
        "performance validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 7.4 local versus international "
    "performance validation checks passed."
)

print(
    f"Section 7.4 completion status: "
    f"{section_7_4_complete}"
)

print(
    f"Local versus international intelligence "
    f"prepared for "
    f"{len(artist_local_international_df):,} "
    "artist representations."
)

print(
    "The local/international distinction describes "
    "observed geographic distribution and must not be "
    "interpreted as an artist's nationality or verified "
    "home-country performance."
)
```

### Interpretation

Section 7.4 successfully compared localised and internationally distributed artist performance for all **49,843 validated artist representations** by combining international reach, market penetration and geographic concentration intelligence from Sections 7.1 to 7.3.

The results show that the majority of artists in the PMIP dataset have strongly localised geographic performance patterns. The median artist reaches only **1 market**, has a strongest-market stream share of **100%**, an effective stream-market count of **1.00**, and a median international-distribution index of only **0.0079**. This indicates that the typical artist representation is concentrated almost entirely within a single observed market rather than being broadly distributed internationally.

The classification results reinforce this pattern. **42,207 artist representations (84.68%)** were classified as having **Localised performance**, while **3,532 (7.09%)** were classified as **Emerging international**. A further **3,158 artists (6.34%)** were identified as **Internationally established**, and **946 artists (1.90%)** were classified as **Broadly diversified international**.

The supporting diagnostics show that **34,830 artists** appear in only one market, while only **6,142 artists** reach at least five markets and **1,733 artists** reach at least twenty-five markets. Only **1,592 artists** have at least five effective stream markets, confirming that broad geographic reach does not always translate into evenly distributed performance across those markets.

A total of **42,207 artists** have at least **85% of their observed streams concentrated in their strongest market**, which helps explain why the localised-performance category dominates the dataset. In contrast, only **4,104 artists** were classified as internationally established or broadly diversified, representing the subset of artists with both wider geographic reach and more distributed performance.

The international-distribution index provides a combined view of geographic breadth and diversification by considering market reach, stream concentration and effective market count together. Higher values indicate that an artist's observed performance is distributed across a broader and less concentrated set of markets, while lower values indicate stronger dependence on a limited geographic footprint.

It is important to note that the term **localised performance** does not refer to an artist's nationality, country of origin or verified home market. The dataset does not contain a reliable artist home-country field. Instead, the classification describes how concentrated or distributed each artist's observed performance is across the available PMIP country markets.

All Section 7.4 validation checks passed. The full artist population was preserved, geographic metrics remained complete and correctly bounded, all distribution indices were valid, every artist received a local-versus-international profile, and the classifications reconciled exactly to all **49,843 artist representations**.

The resulting local-versus-international intelligence is now ready to be combined with the earlier artist geographic measures in Section **7.5 Artist Geographic Performance Profiles**.

### 7.5 Artist Geographic Performance Profiles

#### Purpose

Combine the artist-level geographic intelligence developed in Sections 7.1 to 7.4 into a single interpretable profile for each validated artist representation.

This subsection brings together international reach, market penetration, geographic concentration, and local-versus-international distribution patterns to describe the overall geographic performance structure of each artist.

The aim is not to create a final PMIP ranking yet. Instead, the analysis assigns descriptive artist geographic profiles such as locally concentrated, emerging international, internationally established, diversified international, and globally distributed, while preserving the evidence limitations identified earlier in the notebook.

The resulting profile table will provide a compact artist-level geographic intelligence layer that can later support scoring, ranking, dashboard presentation, and backend integration.

```python
import pandas as pd
import numpy as np

print("Building PMIP artist geographic performance profiles")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_7_4_status = globals().get(
    "section_7_4_complete",
    False
)

print(
    f"Section 7.4 completion status: "
    f"{section_7_4_status}"
)

if not section_7_4_status:
    raise RuntimeError(
        "Section 7.5 requires Section 7.4 "
        "Local versus International Performance to be complete."
    )


# ------------------------------------------------------------
# 2. Recover Section 7 artist intelligence tables
# ------------------------------------------------------------

def recover_artist_dataframe(
    candidate_names,
    required_columns,
    expected_rows=49_843
):

    for name in candidate_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if set(required_columns).issubset(
            candidate.columns
        ):
            return candidate.copy(), name

    # Fallback search through notebook memory
    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if set(required_columns).issubset(
            candidate.columns
        ):
            return candidate.copy(), name

    return None, None


# ------------------------------------------------------------
# 2A. International reach intelligence
# ------------------------------------------------------------

artist_reach_df, artist_reach_source_name = (
    recover_artist_dataframe(
        [
            "artist_international_reach_df",
            "artist_international_reach_intelligence_df",
            "artist_country_reach_features_df",
        ],
        [
            "artists_raw",
        ],
    )
)

if artist_reach_df is None:
    raise RuntimeError(
        "Section 7.5 could not recover the "
        "artist international-reach intelligence."
    )


# ------------------------------------------------------------
# 2B. Market penetration intelligence
# ------------------------------------------------------------

artist_penetration_df, artist_penetration_source_name = (
    recover_artist_dataframe(
        [
            "artist_market_penetration_intelligence_df",
            "artist_penetration_features_df",
            "artist_market_penetration_df",
        ],
        [
            "artists_raw",
        ],
    )
)

if artist_penetration_df is None:
    raise RuntimeError(
        "Section 7.5 could not recover the "
        "artist market-penetration intelligence."
    )


# ------------------------------------------------------------
# 2C. Geographic concentration intelligence
# ------------------------------------------------------------

artist_concentration_df, artist_concentration_source_name = (
    recover_artist_dataframe(
        [
            "artist_geographic_concentration_intelligence_df",
            "artist_stream_concentration_df",
        ],
        [
            "artists_raw",
            "stream_concentration_hhi",
        ],
    )
)

if artist_concentration_df is None:
    raise RuntimeError(
        "Section 7.5 could not recover the "
        "artist geographic-concentration intelligence."
    )


# ------------------------------------------------------------
# 2D. Local versus international intelligence
# ------------------------------------------------------------

artist_local_int_df, artist_local_int_source_name = (
    recover_artist_dataframe(
        [
            "artist_local_international_df",
        ],
        [
            "artists_raw",
            "international_distribution_index",
            "local_international_profile",
        ],
    )
)

if artist_local_int_df is None:
    raise RuntimeError(
        "Section 7.5 could not recover the "
        "local versus international artist intelligence."
    )


print()
print("Validated Section 7 artist intelligence sources")
print("=" * 100)

print(
    f"International reach source: "
    f"{artist_reach_source_name}"
)

print(
    f"Market penetration source: "
    f"{artist_penetration_source_name}"
)

print(
    f"Geographic concentration source: "
    f"{artist_concentration_source_name}"
)

print(
    f"Local/international source: "
    f"{artist_local_int_source_name}"
)

print(
    f"Artist population: "
    f"{len(artist_local_int_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve reusable field names
# ------------------------------------------------------------

def first_existing_column(
    df,
    candidates
):

    for column in candidates:

        if column in df.columns:
            return column

    return None


reach_count_col = first_existing_column(
    artist_reach_df,
    [
        "country_markets_reached",
        "markets_observed",
        "country_markets",
    ]
)

reach_pct_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_pct",
        "country_reach_pct",
        "market_reach_pct",
    ]
)

reach_class_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_class",
        "country_reach_class",
    ]
)

evidence_class_col = first_existing_column(
    artist_reach_df,
    [
        "international_evidence_class",
    ]
)

penetration_class_col = first_existing_column(
    artist_penetration_df,
    [
        "market_penetration_class",
        "penetration_class",
    ]
)

relative_intensity_col = first_existing_column(
    artist_penetration_df,
    [
        "median_relative_stream_intensity",
        "relative_stream_intensity",
    ]
)

support_share_col = first_existing_column(
    artist_penetration_df,
    [
        "support_qualified_market_pct",
        "support_qualified_market_share_pct",
    ]
)

dependency_class_col = first_existing_column(
    artist_concentration_df,
    [
        "market_dependency_class",
    ]
)

concentration_class_col = first_existing_column(
    artist_concentration_df,
    [
        "geographic_concentration_class",
    ]
)

strongest_share_col = first_existing_column(
    artist_concentration_df,
    [
        "strongest_market_stream_share_pct",
        "top_1_stream_share_pct",
    ]
)

effective_market_col = first_existing_column(
    artist_concentration_df,
    [
        "effective_stream_markets",
    ]
)


if reach_count_col is None:
    raise RuntimeError(
        "Section 7.5 requires the artist market-reach count."
    )

if penetration_class_col is None:
    raise RuntimeError(
        "Section 7.5 requires the artist "
        "market-penetration classification."
    )

if concentration_class_col is None:
    raise RuntimeError(
        "Section 7.5 requires the artist "
        "geographic-concentration classification."
    )

if dependency_class_col is None:
    raise RuntimeError(
        "Section 7.5 requires the artist "
        "market-dependency classification."
    )

if strongest_share_col is None:
    raise RuntimeError(
        "Section 7.5 requires the strongest-market "
        "stream-share feature."
    )

if effective_market_col is None:
    raise RuntimeError(
        "Section 7.5 requires the effective "
        "stream-market feature."
    )


# ------------------------------------------------------------
# 4. Build artist geographic profile table
# ------------------------------------------------------------

artist_geographic_profile_df = (
    artist_local_int_df[
        [
            "artists_raw",
            "markets_reached",
            "international_market_reach_pct",
            "strongest_market_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "geographic_diversification_index",
            "international_distribution_index",
            "local_international_profile",
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 5. Merge reach classifications
# ------------------------------------------------------------

reach_columns = [
    "artists_raw",
]

for column in [
    reach_count_col,
    reach_pct_col,
    reach_class_col,
    evidence_class_col,
]:

    if (
        column is not None
        and column not in reach_columns
    ):
        reach_columns.append(
            column
        )


artist_geographic_profile_df = (
    artist_geographic_profile_df.merge(
        artist_reach_df[
            reach_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
        suffixes=(
            "",
            "_reach",
        ),
    )
)


# ------------------------------------------------------------
# 6. Merge penetration intelligence
# ------------------------------------------------------------

penetration_columns = [
    "artists_raw",
]

for column in [
    penetration_class_col,
    relative_intensity_col,
    support_share_col,
]:

    if (
        column is not None
        and column not in penetration_columns
    ):
        penetration_columns.append(
            column
        )


artist_geographic_profile_df = (
    artist_geographic_profile_df.merge(
        artist_penetration_df[
            penetration_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 7. Merge concentration classifications
# ------------------------------------------------------------

concentration_columns = [
    "artists_raw",
    concentration_class_col,
    dependency_class_col,
]

for column in [
    strongest_share_col,
    effective_market_col,
]:

    if (
        column is not None
        and column not in concentration_columns
    ):
        concentration_columns.append(
            column
        )


artist_geographic_profile_df = (
    artist_geographic_profile_df.merge(
        artist_concentration_df[
            concentration_columns
        ],
        on="artists_raw",
        how="left",
        validate="one_to_one",
        suffixes=(
            "",
            "_concentration",
        ),
    )
)


# ------------------------------------------------------------
# 8. Canonical classification aliases
# ------------------------------------------------------------

artist_geographic_profile_df[
    "international_reach_class"
] = (
    artist_geographic_profile_df[
        reach_class_col
    ]
    if reach_class_col is not None
    else pd.Series(
        pd.NA,
        index=artist_geographic_profile_df.index,
        dtype="string",
    )
)

artist_geographic_profile_df[
    "international_evidence_class"
] = (
    artist_geographic_profile_df[
        evidence_class_col
    ]
    if evidence_class_col is not None
    else pd.Series(
        pd.NA,
        index=artist_geographic_profile_df.index,
        dtype="string",
    )
)

artist_geographic_profile_df[
    "market_penetration_class"
] = (
    artist_geographic_profile_df[
        penetration_class_col
    ]
)

artist_geographic_profile_df[
    "geographic_concentration_class"
] = (
    artist_geographic_profile_df[
        concentration_class_col
    ]
)

artist_geographic_profile_df[
    "market_dependency_class"
] = (
    artist_geographic_profile_df[
        dependency_class_col
    ]
)


# ------------------------------------------------------------
# 9. Profile score components
# ------------------------------------------------------------

artist_geographic_profile_df[
    "reach_score_component"
] = (
    artist_geographic_profile_df[
        "international_market_reach_pct"
    ]
    / 100
)

artist_geographic_profile_df[
    "distribution_score_component"
] = (
    artist_geographic_profile_df[
        "international_distribution_index"
    ]
)

artist_geographic_profile_df[
    "diversification_score_component"
] = (
    artist_geographic_profile_df[
        "geographic_diversification_index"
    ]
)


if relative_intensity_col is not None:

    relative_intensity = pd.to_numeric(
        artist_geographic_profile_df[
            relative_intensity_col
        ],
        errors="coerce",
    )

    upper_intensity = float(
        relative_intensity.quantile(
            0.99
        )
    )

    if (
        not np.isfinite(
            upper_intensity
        )
        or upper_intensity <= 0
    ):
        upper_intensity = 1.0

    artist_geographic_profile_df[
        "penetration_score_component"
    ] = (
        relative_intensity
        .clip(
            lower=0,
            upper=upper_intensity,
        )
        /
        upper_intensity
    )

else:

    penetration_mapping = {
        "Limited penetration": 0.25,
        "Moderate penetration": 0.50,
        "Strong penetration": 0.75,
        "Very strong penetration": 1.00,
    }

    artist_geographic_profile_df[
        "penetration_score_component"
    ] = (
        artist_geographic_profile_df[
            "market_penetration_class"
        ]
        .map(
            penetration_mapping
        )
        .fillna(
            0
        )
    )


# ------------------------------------------------------------
# 10. Composite geographic profile index
# ------------------------------------------------------------

artist_geographic_profile_df[
    "geographic_profile_index"
] = (
    (
        0.30
        * artist_geographic_profile_df[
            "reach_score_component"
        ]
    )
    +
    (
        0.25
        * artist_geographic_profile_df[
            "penetration_score_component"
        ]
    )
    +
    (
        0.25
        * artist_geographic_profile_df[
            "distribution_score_component"
        ]
    )
    +
    (
        0.20
        * artist_geographic_profile_df[
            "diversification_score_component"
        ]
    )
)

artist_geographic_profile_df[
    "geographic_profile_index"
] = (
    artist_geographic_profile_df[
        "geographic_profile_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 11. Final artist geographic profile classification
# ------------------------------------------------------------

def classify_artist_geographic_profile(row):

    local_profile = row[
        "local_international_profile"
    ]

    penetration = row[
        "market_penetration_class"
    ]

    reach = row[
        "international_reach_class"
    ]

    dependency = row[
        "market_dependency_class"
    ]

    distribution_index = row[
        "international_distribution_index"
    ]

    effective_markets = row[
        "effective_stream_markets"
    ]

    # Local / highly dependent
    if (
        local_profile == "Localised performance"
        or dependency == "Single-market dependent"
    ):
        return "Locally concentrated"

    # International signals emerging but still narrow
    if (
        local_profile == "Emerging international"
    ):
        return "Emerging international"

    # Broad and relatively diversified
    if (
        local_profile
        == "Broadly diversified international"
        and effective_markets >= 5
        and distribution_index >= 0.25
    ):
        return "Diversified international"

    # Strongest international profile
    if (
        (
            reach
            in [
                "Very broad international",
                "Near-global reach",
            ]
        )
        and
        (
            penetration
            in [
                "Strong penetration",
                "Very strong penetration",
            ]
        )
        and dependency != "Single-market dependent"
        and distribution_index >= 0.20
    ):
        return "Globally distributed"

    # Established international presence
    if (
        local_profile
        in [
            "Internationally established",
            "Broadly diversified international",
        ]
    ):
        return "Internationally established"

    return "Emerging international"


artist_geographic_profile_df[
    "artist_geographic_profile"
] = (
    artist_geographic_profile_df.apply(
        classify_artist_geographic_profile,
        axis=1,
    )
)


# ------------------------------------------------------------
# 12. Profile distribution summary
# ------------------------------------------------------------

profile_order = [
    "Locally concentrated",
    "Emerging international",
    "Internationally established",
    "Diversified international",
    "Globally distributed",
]

artist_geographic_profile_summary_df = (
    artist_geographic_profile_df[
        "artist_geographic_profile"
    ]
    .value_counts()
    .reindex(
        profile_order,
        fill_value=0,
    )
    .rename_axis(
        "Artist Geographic Profile"
    )
    .reset_index(
        name="Artist Representations"
    )
)

artist_geographic_profile_summary_df[
    "Share (%)"
] = (
    artist_geographic_profile_summary_df[
        "Artist Representations"
    ]
    /
    len(
        artist_geographic_profile_df
    )
    * 100
)


# ------------------------------------------------------------
# 13. Overview
# ------------------------------------------------------------

print()
print("Artist geographic performance profile overview")
print("=" * 100)

print(
    f"Artists assessed: "
    f"{len(artist_geographic_profile_df):,}"
)

print(
    f"Median geographic profile index: "
    f"{artist_geographic_profile_df['geographic_profile_index'].median():.4f}"
)

print(
    f"Mean geographic profile index: "
    f"{artist_geographic_profile_df['geographic_profile_index'].mean():.4f}"
)

print()
print("Artist geographic profile classification")
print("=" * 100)

display(
    artist_geographic_profile_summary_df
)


# ------------------------------------------------------------
# 14. Display columns
# ------------------------------------------------------------

display_columns = [
    "artists_raw",
    "markets_reached",
    "international_market_reach_pct",
    "strongest_market_stream_share_pct",
    "market_penetration_class",
    "geographic_concentration_class",
    "market_dependency_class",
    "local_international_profile",
    "effective_stream_markets",
    "geographic_diversification_index",
    "international_distribution_index",
    "geographic_profile_index",
    "artist_geographic_profile",
]


# ------------------------------------------------------------
# 15. Strongest overall geographic profiles
# ------------------------------------------------------------

print()
print(
    "Artist representations with strongest "
    "overall geographic profiles"
)
print("=" * 100)

strongest_geographic_profiles_df = (
    artist_geographic_profile_df
    .sort_values(
        [
            "geographic_profile_index",
            "markets_reached",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_geographic_profiles_df
)


# ------------------------------------------------------------
# 16. Most locally concentrated profiles
# ------------------------------------------------------------

print()
print(
    "Artist representations with strongest "
    "local concentration"
)
print("=" * 100)

strongest_local_profiles_df = (
    artist_geographic_profile_df
    .sort_values(
        [
            "geographic_profile_index",
            "strongest_market_stream_share_pct",
        ],
        ascending=[
            True,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_local_profiles_df
)


# ------------------------------------------------------------
# 17. Profile diagnostics
# ------------------------------------------------------------

artist_profile_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Locally concentrated artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "artist_geographic_profile"
                        ]
                        == "Locally concentrated"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Emerging international artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "artist_geographic_profile"
                        ]
                        == "Emerging international"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Internationally established artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "artist_geographic_profile"
                        ]
                        == "Internationally established"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Diversified international artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "artist_geographic_profile"
                        ]
                        == "Diversified international"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Globally distributed artists",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "artist_geographic_profile"
                        ]
                        == "Globally distributed"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with profile index >= 0.50",

            "Observed Value":
                int(
                    (
                        artist_geographic_profile_df[
                            "geographic_profile_index"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)

print()
print("Artist geographic profile diagnostics")
print("=" * 100)

display(
    artist_profile_diagnostics_df
)


# ------------------------------------------------------------
# 18. Validation
# ------------------------------------------------------------

artist_population_valid = bool(
    len(
        artist_geographic_profile_df
    )
    == 49_843
)

artist_identity_valid = bool(
    artist_geographic_profile_df[
        "artists_raw"
    ].nunique()
    == 49_843
)

required_profile_fields = [
    "international_market_reach_pct",
    "market_penetration_class",
    "geographic_concentration_class",
    "market_dependency_class",
    "local_international_profile",
    "international_distribution_index",
    "geographic_profile_index",
    "artist_geographic_profile",
]

profile_fields_complete = bool(
    artist_geographic_profile_df[
        required_profile_fields
    ]
    .notna()
    .all()
    .all()
)

profile_index_valid = bool(
    np.isfinite(
        artist_geographic_profile_df[
            "geographic_profile_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    artist_geographic_profile_df[
        "geographic_profile_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

profile_class_complete = bool(
    artist_geographic_profile_df[
        "artist_geographic_profile"
    ]
    .notna()
    .all()
)

profile_total = int(
    artist_geographic_profile_summary_df[
        "Artist Representations"
    ].sum()
)

profile_population_reconciled = bool(
    profile_total
    ==
    len(
        artist_geographic_profile_df
    )
)


section_7_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 7.4 completion",

            "Requirement":
                "Local versus International Performance "
                "must already be complete.",

            "Observed Evidence":
                f"Section 7.4 completion status: "
                f"{section_7_4_status}",

            "Passed":
                section_7_4_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All validated artist representations "
                "must be retained.",

            "Observed Evidence":
                f"{len(artist_geographic_profile_df):,} "
                "artist profile records assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must receive "
                "one geographic profile.",

            "Observed Evidence":
                f"{artist_geographic_profile_df['artists_raw'].nunique():,} "
                "unique artist representations",

            "Passed":
                artist_identity_valid,
        },
        {
            "Validation Area":
                "Profile input completeness",

            "Requirement":
                "All geographic intelligence inputs "
                "required for profiling must be available.",

            "Observed Evidence":
                "Reach, penetration, concentration and "
                "distribution fields assessed",

            "Passed":
                profile_fields_complete,
        },
        {
            "Validation Area":
                "Geographic-profile index validity",

            "Requirement":
                "Profile indices must remain finite "
                "and between 0 and 1.",

            "Observed Evidence":
                f"{len(artist_geographic_profile_df):,} "
                "profile indices assessed",

            "Passed":
                profile_index_valid,
        },
        {
            "Validation Area":
                "Profile classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "a geographic profile class.",

            "Observed Evidence":
                f"{artist_geographic_profile_df['artist_geographic_profile'].notna().sum():,} "
                "profile classes assigned",

            "Passed":
                profile_class_complete,
        },
        {
            "Validation Area":
                "Profile population reconciliation",

            "Requirement":
                "Profile classifications must reconcile "
                "with the complete artist population.",

            "Observed Evidence":
                f"{profile_total:,} classified records "
                f"for {len(artist_geographic_profile_df):,} artists",

            "Passed":
                profile_population_reconciled,
        },
    ]
)

print()
print("Artist geographic performance profile validation")
print("=" * 100)

display(
    section_7_5_validation_df
)


# ------------------------------------------------------------
# 19. Final completion status
# ------------------------------------------------------------

section_7_5_complete = bool(
    section_7_5_validation_df[
        "Passed"
    ].all()
)

if not section_7_5_complete:

    failed_checks = (
        section_7_5_validation_df
        .loc[
            ~section_7_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 7.5 artist geographic profile "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

section_7_complete = bool(
    section_7_5_complete
)

print()
print(
    "All Section 7.5 artist geographic performance "
    "profile validation checks passed."
)

print(
    f"Section 7.5 completion status: "
    f"{section_7_5_complete}"
)

print(
    f"Section 7 overall completion status: "
    f"{section_7_complete}"
)

print(
    f"Artist geographic performance profiles prepared "
    f"for {len(artist_geographic_profile_df):,} "
    "artist representations."
)

print(
    "These profiles combine geographic reach, penetration, "
    "concentration and distribution intelligence. They remain "
    "descriptive intelligence outputs and are not yet final "
    "PMIP artist rankings."
)
```

#### Interpretation

Section 7.5 successfully combined the artist-level geographic intelligence developed across Sections 7.1 to 7.4 into a unified geographic performance profile for all **49,843 artist representations**.

The resulting profile distribution shows that the artist population is heavily concentrated toward local or narrow-market performance. **43,775 artists (87.83%)** were classified as **Locally concentrated**, indicating that most observed artist activity remains strongly dependent on one market or a very small number of markets.

A smaller group of **1,964 artists (3.94%)** were classified as **Emerging international**, while **3,144 artists (6.31%)** were identified as **Internationally established**. These groups represent artists with increasing geographic reach and stronger evidence of cross-market performance.

Only **946 artists (1.90%)** were classified as **Diversified international**, meaning their observed performance was distributed across several markets with relatively lower dependence on a single dominant market.

The most selective category, **Globally distributed**, contained only **14 artist representations (0.03%)**. These artists satisfied the strongest combination of international reach, market penetration, distribution, and diversification conditions defined in this subsection.

The geographic profile index also confirms that strong international geographic performance is relatively uncommon within the dataset. The **median profile index is 0.0722** and the **mean is 0.1117**, while only **976 artists** achieved an index of at least **0.50**.

All Section 7.5 validation checks passed. The complete artist population was preserved, artist identities remained unique, all required profile inputs were available, geographic profile indices remained finite and within the expected range, and every artist received exactly one geographic profile classification.

Overall, Section 7 converts the individual artist geographic measures into an interpretable intelligence layer. The profiles describe how broadly, deeply, and evenly an artist performs across observed markets, but they should still be treated as **descriptive geographic intelligence rather than final PMIP artist rankings**.

## 8. Track Geographic Intelligence

Section 8 develops track-level geographic intelligence from the validated geographic feature-engineering and country-performance outputs produced earlier in the notebook.

The analysis focuses on how individual tracks spread across country markets, how strongly they penetrate those markets, how consistently they perform across countries, how concentrated their streaming activity is geographically, and how these signals combine into interpretable track geographic performance profiles.

The section contains:

- **8.1 Track Country Reach**
- **8.2 Track Market Penetration**
- **8.3 Cross-Market Track Performance**
- **8.4 Track Geographic Concentration**
- **8.5 Track Geographic Performance Profiles**




### 8.1 Track Country Reach

#### Purpose

Assess the observed international geographic reach of each validated PMIP track.

This subsection uses the country-reach features prepared in Section 5.1 to measure how many of the **76 validated country markets** each track appears in and what proportion of the available geographic market space this represents.

Tracks are grouped into interpretable reach classes ranging from single-market presence to near-global reach. Additional evidence categories are created to distinguish tracks with only local or limited geographic evidence from tracks supported across a wider international market base.

The purpose is to establish the geographic breadth of each track before later sections evaluate market penetration, cross-market performance consistency, geographic concentration, and final track geographic profiles.

These measures describe **observed geographic reach only** and should not yet be interpreted as final PMIP track performance rankings.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP track country reach")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous major section completion
# ------------------------------------------------------------

section_7_status = globals().get(
    "section_7_complete",
    False
)

print(
    f"Section 7 overall completion status: "
    f"{section_7_status}"
)

if not section_7_status:
    raise RuntimeError(
        "Section 8.1 requires Section 7 "
        "Artist Geographic Intelligence to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track country-reach source
# ------------------------------------------------------------

def recover_track_reach_dataframe():

    preferred_names = [
        "track_country_reach_features_df",
        "track_geographic_coverage_df",
    ]

    required_identity = {
        "track_id",
    }

    for name in preferred_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if not required_identity.issubset(
            candidate.columns
        ):
            continue

        return candidate.copy(), name

    # Fallback search through notebook memory
    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if not required_identity.issubset(
            candidate.columns
        ):
            continue

        reach_candidates = {
            "country_markets_reached",
            "markets_observed",
            "country_markets",
            "country_reach_pct",
            "market_coverage_pct",
        }

        if any(
            column in candidate.columns
            for column in reach_candidates
        ):
            return candidate.copy(), name

    return None, None


track_reach_source_df, track_reach_source_name = (
    recover_track_reach_dataframe()
)

if track_reach_source_df is None:
    raise RuntimeError(
        "Section 8.1 could not recover the validated "
        "track country-reach feature table from previous "
        "notebook sections."
    )


print()
print("Validated track country-reach source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_reach_source_name}"
)

print(
    f"Tracks: "
    f"{len(track_reach_source_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve source columns
# ------------------------------------------------------------

def first_existing_column(
    df,
    candidates
):

    for column in candidates:

        if column in df.columns:
            return column

    return None


markets_reached_col = first_existing_column(
    track_reach_source_df,
    [
        "country_markets_reached",
        "markets_observed",
        "country_markets",
    ]
)

reach_pct_col = first_existing_column(
    track_reach_source_df,
    [
        "country_reach_pct",
        "market_coverage_pct",
        "international_reach_pct",
        "market_reach_pct",
    ]
)

track_name_col = first_existing_column(
    track_reach_source_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    track_reach_source_df,
    [
        "artists_raw",
        "artists",
    ]
)

existing_reach_class_col = first_existing_column(
    track_reach_source_df,
    [
        "country_reach_class",
        "coverage_classification",
    ]
)


if markets_reached_col is None:
    raise RuntimeError(
        "Section 8.1 requires a track market-count field."
    )


# ------------------------------------------------------------
# 4. Prepare canonical track international-reach table
# ------------------------------------------------------------

track_international_reach_df = (
    track_reach_source_df.copy()
)

track_international_reach_df[
    "markets_reached"
] = pd.to_numeric(
    track_international_reach_df[
        markets_reached_col
    ],
    errors="coerce",
)


# ------------------------------------------------------------
# 5. Calculate international reach percentage
# ------------------------------------------------------------

validated_country_markets = 76

if reach_pct_col is not None:

    track_international_reach_df[
        "international_market_reach_pct"
    ] = pd.to_numeric(
        track_international_reach_df[
            reach_pct_col
        ],
        errors="coerce",
    )

else:

    track_international_reach_df[
        "international_market_reach_pct"
    ] = (
        track_international_reach_df[
            "markets_reached"
        ]
        /
        validated_country_markets
        * 100
    )


# ------------------------------------------------------------
# 6. International reach index
# ------------------------------------------------------------

track_international_reach_df[
    "international_reach_index"
] = (
    track_international_reach_df[
        "markets_reached"
    ]
    /
    validated_country_markets
)

track_international_reach_df[
    "international_reach_index"
] = (
    track_international_reach_df[
        "international_reach_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 7. Track international-reach classification
# ------------------------------------------------------------

def classify_track_international_reach(
    markets
):

    if markets <= 1:
        return "Single-market"

    if markets <= 4:
        return "Limited multi-market"

    if markets <= 9:
        return "Moderate international"

    if markets <= 24:
        return "Broad international"

    if markets <= 49:
        return "Very broad international"

    return "Near-global reach"


track_international_reach_df[
    "international_reach_class"
] = (
    track_international_reach_df[
        "markets_reached"
    ]
    .apply(
        classify_track_international_reach
    )
)


# ------------------------------------------------------------
# 8. International evidence classification
# ------------------------------------------------------------

def classify_track_international_evidence(
    markets
):

    if markets <= 1:
        return "Local-only evidence"

    if markets <= 4:
        return "Limited international evidence"

    if markets <= 24:
        return "Moderate international evidence"

    return "Strong international evidence"


track_international_reach_df[
    "international_evidence_class"
] = (
    track_international_reach_df[
        "markets_reached"
    ]
    .apply(
        classify_track_international_evidence
    )
)


# ------------------------------------------------------------
# 9. Overview
# ------------------------------------------------------------

print()
print("Track international-reach overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_international_reach_df):,}"
)

print(
    f"Minimum markets reached: "
    f"{int(track_international_reach_df['markets_reached'].min())}"
)

print(
    f"Median markets reached: "
    f"{track_international_reach_df['markets_reached'].median():.1f}"
)

print(
    f"Mean markets reached: "
    f"{track_international_reach_df['markets_reached'].mean():.2f}"
)

print(
    f"Maximum markets reached: "
    f"{int(track_international_reach_df['markets_reached'].max())}"
)

print(
    f"Median international reach: "
    f"{track_international_reach_df['international_market_reach_pct'].median():.2f}%"
)


# ------------------------------------------------------------
# 10. Reach classification summary
# ------------------------------------------------------------

reach_order = [
    "Single-market",
    "Limited multi-market",
    "Moderate international",
    "Broad international",
    "Very broad international",
    "Near-global reach",
]

track_international_reach_summary_df = (
    track_international_reach_df[
        "international_reach_class"
    ]
    .value_counts()
    .reindex(
        reach_order,
        fill_value=0,
    )
    .rename_axis(
        "International Reach Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_international_reach_summary_df[
    "Share (%)"
] = (
    track_international_reach_summary_df[
        "Tracks"
    ]
    /
    len(track_international_reach_df)
    * 100
)

print()
print("Track international-reach classification")
print("=" * 100)

display(
    track_international_reach_summary_df
)


# ------------------------------------------------------------
# 11. Evidence classification summary
# ------------------------------------------------------------

evidence_order = [
    "Local-only evidence",
    "Limited international evidence",
    "Moderate international evidence",
    "Strong international evidence",
]

track_international_evidence_summary_df = (
    track_international_reach_df[
        "international_evidence_class"
    ]
    .value_counts()
    .reindex(
        evidence_order,
        fill_value=0,
    )
    .rename_axis(
        "International Evidence Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_international_evidence_summary_df[
    "Share (%)"
] = (
    track_international_evidence_summary_df[
        "Tracks"
    ]
    /
    len(track_international_reach_df)
    * 100
)

print()
print("Track international-evidence classification")
print("=" * 100)

display(
    track_international_evidence_summary_df
)


# ------------------------------------------------------------
# 12. Tracks with widest observed international reach
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_reached",
    "international_market_reach_pct",
    "international_reach_index",
    "international_reach_class",
    "international_evidence_class",
]


print()
print(
    "Tracks with widest observed international reach"
)
print("=" * 100)

widest_track_reach_df = (
    track_international_reach_df
    .sort_values(
        [
            "markets_reached",
            "international_reach_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    widest_track_reach_df
)


# ------------------------------------------------------------
# 13. Reach diagnostics
# ------------------------------------------------------------

track_reach_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Single-market tracks",

            "Observed Value":
                int(
                    (
                        track_international_reach_df[
                            "markets_reached"
                        ]
                        == 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks appearing in at least 5 markets",

            "Observed Value":
                int(
                    (
                        track_international_reach_df[
                            "markets_reached"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks appearing in at least 10 markets",

            "Observed Value":
                int(
                    (
                        track_international_reach_df[
                            "markets_reached"
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks appearing in at least 25 markets",

            "Observed Value":
                int(
                    (
                        track_international_reach_df[
                            "markets_reached"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks appearing in at least 50 markets",

            "Observed Value":
                int(
                    (
                        track_international_reach_df[
                            "markets_reached"
                        ]
                        >= 50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Maximum observed country reach",

            "Observed Value":
                int(
                    track_international_reach_df[
                        "markets_reached"
                    ].max()
                ),
        },
    ]
)

print()
print("Track international-reach diagnostics")
print("=" * 100)

display(
    track_reach_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

track_population_valid = bool(
    len(
        track_international_reach_df
    )
    == 110_198
)

track_identity_valid = bool(
    track_international_reach_df[
        "track_id"
    ].nunique()
    == 110_198
)

market_reach_valid = bool(
    track_international_reach_df[
        "markets_reached"
    ]
    .between(
        1,
        validated_country_markets,
        inclusive="both",
    )
    .all()
)

reach_pct_valid = bool(
    np.isfinite(
        track_international_reach_df[
            "international_market_reach_pct"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_international_reach_df[
        "international_market_reach_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

reach_index_valid = bool(
    np.isfinite(
        track_international_reach_df[
            "international_reach_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_international_reach_df[
        "international_reach_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

reach_class_complete = bool(
    track_international_reach_df[
        "international_reach_class"
    ]
    .notna()
    .all()
)

evidence_class_complete = bool(
    track_international_reach_df[
        "international_evidence_class"
    ]
    .notna()
    .all()
)

reach_population_reconciled = bool(
    int(
        track_international_reach_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_international_reach_df
    )
)


section_8_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 7 completion",

            "Requirement":
                "Artist Geographic Intelligence must "
                "already be complete.",

            "Observed Evidence":
                f"Section 7 overall completion status: "
                f"{section_7_status}",

            "Passed":
                section_7_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_international_reach_df):,} "
                "track records assessed",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one international "
                "reach record.",

            "Observed Evidence":
                f"{track_international_reach_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                track_identity_valid,
        },
        {
            "Validation Area":
                "International reach bounds",

            "Requirement":
                "Observed market reach must remain between "
                "1 and 76 validated country markets.",

            "Observed Evidence":
                f"Observed range: "
                f"{int(track_international_reach_df['markets_reached'].min())} "
                f"to "
                f"{int(track_international_reach_df['markets_reached'].max())}",

            "Passed":
                market_reach_valid,
        },
        {
            "Validation Area":
                "Reach percentage bounds",

            "Requirement":
                "International reach percentages must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "International reach percentages assessed "
                "for all tracks",

            "Passed":
                reach_pct_valid,
        },
        {
            "Validation Area":
                "Reach-index validity",

            "Requirement":
                "International reach indices must remain "
                "finite and between 0 and 1.",

            "Observed Evidence":
                f"{len(track_international_reach_df):,} "
                "reach indices assessed",

            "Passed":
                reach_index_valid,
        },
        {
            "Validation Area":
                "Reach classification completeness",

            "Requirement":
                "Every track must receive an international "
                "reach classification.",

            "Observed Evidence":
                f"{track_international_reach_df['international_reach_class'].notna().sum():,} "
                "reach classes assigned",

            "Passed":
                reach_class_complete,
        },
        {
            "Validation Area":
                "Evidence classification completeness",

            "Requirement":
                "Every track must receive an international "
                "evidence classification.",

            "Observed Evidence":
                f"{track_international_reach_df['international_evidence_class'].notna().sum():,} "
                "evidence classes assigned",

            "Passed":
                evidence_class_complete,
        },
        {
            "Validation Area":
                "Reach population reconciliation",

            "Requirement":
                "Reach classifications must reconcile with "
                "the complete validated track population.",

            "Observed Evidence":
                f"{int(track_international_reach_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_international_reach_df):,} tracks",

            "Passed":
                reach_population_reconciled,
        },
    ]
)

print()
print("Track international-reach validation")
print("=" * 100)

display(
    section_8_1_validation_df
)


# ------------------------------------------------------------
# 15. Final completion status
# ------------------------------------------------------------

section_8_1_complete = bool(
    section_8_1_validation_df[
        "Passed"
    ].all()
)

if not section_8_1_complete:

    failed_checks = (
        section_8_1_validation_df
        .loc[
            ~section_8_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 8.1 track country-reach validation "
        "failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 8.1 track country-reach "
    "validation checks passed."
)

print(
    f"Section 8.1 completion status: "
    f"{section_8_1_complete}"
)

print(
    f"Track international-reach intelligence prepared "
    f"for {len(track_international_reach_df):,} tracks."
)

print(
    "International reach measures geographic breadth "
    "and evidence support; it is not yet a final PMIP "
    "track geographic performance ranking."
)
```

#### Interpretation

Section 8.1 successfully evaluates the observed geographic reach of all **110,198 validated tracks** across the **76 PMIP country markets**.

The results show that track geographic coverage is highly uneven. The typical track has a very narrow observed footprint: the **median track appears in only 1 market**, while the mean is **3.21 markets**. The widest-observed tracks reach **73 of the 76 available country markets**, equivalent to approximately **96.05% geographic coverage**.

Most tracks remain geographically localised. **81,082 tracks (73.58%)** are classified as **Single-market**, while another **17,859 tracks (16.21%)** have only **Limited multi-market** reach. Together, these groups account for almost 90% of the track population, showing that broad international presence is relatively uncommon within the dataset.

Only **3,969 tracks (3.60%)** fall into the **Moderate international** category, while **4,140 tracks (3.76%)** achieve **Broad international** reach. A further **2,034 tracks (1.85%)** are classified as **Very broad international**, and just **1,114 tracks (1.01%)** achieve **Near-global reach**.

The international-evidence classification reinforces this pattern. **73.58%** of tracks contain only local-market evidence, while **16.21%** have limited international evidence. Only **2.86% of tracks** are supported by strong international evidence across at least 25 markets.

The reach diagnostics provide further context. Although **11,257 tracks** appear in at least five markets, this falls to **7,288 tracks** at ten or more markets, **3,148 tracks** at twenty-five or more markets, and only **1,114 tracks** at fifty or more markets.

Several tracks nevertheless demonstrate exceptionally broad geographic exposure. The widest-observed tracks include titles reaching between **70 and 73 markets**, corresponding to more than **92% of the validated PMIP market universe**. These tracks therefore provide particularly strong evidence for later cross-market analysis.

All Section 8.1 validation checks passed. Track identities remain unique, geographic reach remains within the valid 1-to-76 market boundary, percentage and index measures remain within their expected ranges, and every validated track received both a reach classification and an international-evidence classification.

Overall, Section 8.1 establishes the **geographic breadth** of track performance. It distinguishes locally observed tracks from tracks with substantial international exposure, but these measures represent geographic availability and evidence breadth rather than a final PMIP track-performance ranking.

### 8.2 Track Market Penetration

#### Purpose

Assess how strongly each validated track performs within the country markets where it is observed.

This subsection uses the track market-penetration features created in Section 5.2 to measure the depth of track performance across reached markets. The analysis considers market-level stream share, relative stream intensity, support-qualified market presence, and above-average performance across countries.

The purpose is to separate tracks that simply appear in several markets from tracks that demonstrate stronger and more meaningful penetration within those markets.

These measures remain descriptive track-level geographic intelligence and are not yet final PMIP track geographic performance rankings.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP track market penetration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_8_1_status = globals().get(
    "section_8_1_complete",
    False
)

print(
    f"Section 8.1 completion status: "
    f"{section_8_1_status}"
)

if not section_8_1_status:
    raise RuntimeError(
        "Section 8.2 requires Section 8.1 "
        "Track Country Reach to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track market-penetration source
# ------------------------------------------------------------

def recover_track_penetration_dataframe():

    preferred_names = [
        "track_penetration_features_df",
        "track_market_penetration_df",
    ]

    required_identity = {
        "track_id",
    }

    for name in preferred_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if not required_identity.issubset(
            candidate.columns
        ):
            continue

        return candidate.copy(), name

    # Fallback search through notebook memory
    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if "track_id" not in candidate.columns:
            continue

        penetration_candidates = {
            "median_country_stream_share_pct",
            "maximum_country_stream_share_pct",
            "median_relative_stream_intensity",
            "maximum_relative_stream_intensity",
            "support_qualified_market_pct",
            "above_country_average_market_pct",
        }

        if any(
            column in candidate.columns
            for column in penetration_candidates
        ):
            return candidate.copy(), name

    return None, None


track_penetration_source_df, track_penetration_source_name = (
    recover_track_penetration_dataframe()
)

if track_penetration_source_df is None:
    raise RuntimeError(
        "Section 8.2 could not recover the validated "
        "track market-penetration feature table from "
        "previous notebook sections."
    )


print()
print("Validated track market-penetration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_penetration_source_name}"
)

print(
    f"Tracks: "
    f"{len(track_penetration_source_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve source columns
# ------------------------------------------------------------

def first_existing_column(
    df,
    candidates
):

    for column in candidates:

        if column in df.columns:
            return column

    return None


track_name_col = first_existing_column(
    track_penetration_source_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    track_penetration_source_df,
    [
        "artists_raw",
        "artists",
    ]
)

markets_observed_col = first_existing_column(
    track_penetration_source_df,
    [
        "markets_observed",
        "country_markets_reached",
        "country_markets",
    ]
)

median_stream_share_col = first_existing_column(
    track_penetration_source_df,
    [
        "median_country_stream_share_pct",
        "median_market_stream_share_pct",
    ]
)

maximum_stream_share_col = first_existing_column(
    track_penetration_source_df,
    [
        "maximum_country_stream_share_pct",
        "max_country_stream_share_pct",
        "maximum_market_stream_share_pct",
    ]
)

median_relative_intensity_col = first_existing_column(
    track_penetration_source_df,
    [
        "median_relative_stream_intensity",
        "relative_stream_intensity",
    ]
)

maximum_relative_intensity_col = first_existing_column(
    track_penetration_source_df,
    [
        "maximum_relative_stream_intensity",
        "max_relative_stream_intensity",
    ]
)

support_qualified_col = first_existing_column(
    track_penetration_source_df,
    [
        "support_qualified_market_pct",
        "support_qualified_market_share_pct",
    ]
)

above_average_col = first_existing_column(
    track_penetration_source_df,
    [
        "above_country_average_market_pct",
        "above_average_market_pct",
    ]
)


required_resolved = {
    "markets_observed": markets_observed_col,
    "median_relative_stream_intensity": median_relative_intensity_col,
}

missing_required = [
    name
    for name, column in required_resolved.items()
    if column is None
]

if missing_required:
    raise RuntimeError(
        "Section 8.2 requires the following track "
        "market-penetration fields: "
        + ", ".join(missing_required)
    )


# ------------------------------------------------------------
# 4. Prepare canonical track market-penetration table
# ------------------------------------------------------------

track_market_penetration_df = (
    track_penetration_source_df.copy()
)

track_market_penetration_df[
    "markets_observed"
] = pd.to_numeric(
    track_market_penetration_df[
        markets_observed_col
    ],
    errors="coerce",
)

track_market_penetration_df[
    "median_relative_stream_intensity"
] = pd.to_numeric(
    track_market_penetration_df[
        median_relative_intensity_col
    ],
    errors="coerce",
)


if median_stream_share_col is not None:
    track_market_penetration_df[
        "median_country_stream_share_pct"
    ] = pd.to_numeric(
        track_market_penetration_df[
            median_stream_share_col
        ],
        errors="coerce",
    )
else:
    track_market_penetration_df[
        "median_country_stream_share_pct"
    ] = np.nan


if maximum_stream_share_col is not None:
    track_market_penetration_df[
        "maximum_country_stream_share_pct"
    ] = pd.to_numeric(
        track_market_penetration_df[
            maximum_stream_share_col
        ],
        errors="coerce",
    )
else:
    track_market_penetration_df[
        "maximum_country_stream_share_pct"
    ] = np.nan


if maximum_relative_intensity_col is not None:
    track_market_penetration_df[
        "maximum_relative_stream_intensity"
    ] = pd.to_numeric(
        track_market_penetration_df[
            maximum_relative_intensity_col
        ],
        errors="coerce",
    )
else:
    track_market_penetration_df[
        "maximum_relative_stream_intensity"
    ] = track_market_penetration_df[
        "median_relative_stream_intensity"
    ]


if support_qualified_col is not None:
    track_market_penetration_df[
        "support_qualified_market_pct"
    ] = pd.to_numeric(
        track_market_penetration_df[
            support_qualified_col
        ],
        errors="coerce",
    )
else:
    track_market_penetration_df[
        "support_qualified_market_pct"
    ] = 0.0


if above_average_col is not None:
    track_market_penetration_df[
        "above_country_average_market_pct"
    ] = pd.to_numeric(
        track_market_penetration_df[
            above_average_col
        ],
        errors="coerce",
    )
else:
    track_market_penetration_df[
        "above_country_average_market_pct"
    ] = 0.0


# ------------------------------------------------------------
# 5. Build penetration index
# ------------------------------------------------------------

relative_intensity = (
    track_market_penetration_df[
        "median_relative_stream_intensity"
    ]
)

upper_intensity = float(
    relative_intensity.quantile(0.99)
)

if (
    not np.isfinite(upper_intensity)
    or upper_intensity <= 0
):
    upper_intensity = 1.0


track_market_penetration_df[
    "relative_intensity_component"
] = (
    relative_intensity
    .clip(
        lower=0,
        upper=upper_intensity,
    )
    /
    upper_intensity
)


track_market_penetration_df[
    "support_component"
] = (
    track_market_penetration_df[
        "support_qualified_market_pct"
    ]
    .clip(
        lower=0,
        upper=100,
    )
    / 100
)


track_market_penetration_df[
    "above_average_component"
] = (
    track_market_penetration_df[
        "above_country_average_market_pct"
    ]
    .clip(
        lower=0,
        upper=100,
    )
    / 100
)


track_market_penetration_df[
    "market_penetration_index"
] = (
    (
        0.50
        * track_market_penetration_df[
            "relative_intensity_component"
        ]
    )
    +
    (
        0.25
        * track_market_penetration_df[
            "support_component"
        ]
    )
    +
    (
        0.25
        * track_market_penetration_df[
            "above_average_component"
        ]
    )
)

track_market_penetration_df[
    "market_penetration_index"
] = (
    track_market_penetration_df[
        "market_penetration_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 6. Market-penetration classification
# ------------------------------------------------------------

def classify_track_market_penetration(index_value):

    if index_value < 0.25:
        return "Limited penetration"

    if index_value < 0.50:
        return "Moderate penetration"

    if index_value < 0.75:
        return "Strong penetration"

    return "Very strong penetration"


track_market_penetration_df[
    "market_penetration_class"
] = (
    track_market_penetration_df[
        "market_penetration_index"
    ]
    .apply(
        classify_track_market_penetration
    )
)


# ------------------------------------------------------------
# 7. Overview
# ------------------------------------------------------------

print()
print("Track market-penetration overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_market_penetration_df):,}"
)

print(
    f"Median markets observed: "
    f"{track_market_penetration_df['markets_observed'].median():.1f}"
)

print(
    f"Median track market stream share: "
    f"{track_market_penetration_df['median_country_stream_share_pct'].median():.6f}%"
)

print(
    f"Median relative stream intensity: "
    f"{track_market_penetration_df['median_relative_stream_intensity'].median():.4f}"
)

print(
    f"Median support-qualified market share: "
    f"{track_market_penetration_df['support_qualified_market_pct'].median():.2f}%"
)

print(
    f"Median above-average market share: "
    f"{track_market_penetration_df['above_country_average_market_pct'].median():.2f}%"
)

print(
    f"Median market-penetration index: "
    f"{track_market_penetration_df['market_penetration_index'].median():.4f}"
)


# ------------------------------------------------------------
# 8. Penetration classification summary
# ------------------------------------------------------------

penetration_order = [
    "Limited penetration",
    "Moderate penetration",
    "Strong penetration",
    "Very strong penetration",
]

track_market_penetration_summary_df = (
    track_market_penetration_df[
        "market_penetration_class"
    ]
    .value_counts()
    .reindex(
        penetration_order,
        fill_value=0,
    )
    .rename_axis(
        "Market Penetration Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_market_penetration_summary_df[
    "Share (%)"
] = (
    track_market_penetration_summary_df[
        "Tracks"
    ]
    /
    len(track_market_penetration_df)
    * 100
)

print()
print("Track market-penetration classification")
print("=" * 100)

display(
    track_market_penetration_summary_df
)


# ------------------------------------------------------------
# 9. Tracks with strongest observed market penetration
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_observed",
    "median_country_stream_share_pct",
    "maximum_country_stream_share_pct",
    "median_relative_stream_intensity",
    "maximum_relative_stream_intensity",
    "support_qualified_market_pct",
    "above_country_average_market_pct",
    "market_penetration_index",
    "market_penetration_class",
]


print()
print(
    "Tracks with strongest observed market penetration"
)
print("=" * 100)

strongest_track_penetration_df = (
    track_market_penetration_df
    .sort_values(
        [
            "market_penetration_index",
            "median_relative_stream_intensity",
            "markets_observed",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_track_penetration_df
)


# ------------------------------------------------------------
# 10. Diagnostics
# ------------------------------------------------------------

track_penetration_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with median relative intensity >= 1.0",

            "Observed Value":
                int(
                    (
                        track_market_penetration_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 1.0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with median relative intensity >= 1.5",

            "Observed Value":
                int(
                    (
                        track_market_penetration_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 1.5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with median relative intensity >= 2.0",

            "Observed Value":
                int(
                    (
                        track_market_penetration_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 2.0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks classified strong or very strong",

            "Observed Value":
                int(
                    track_market_penetration_df[
                        "market_penetration_class"
                    ]
                    .isin(
                        [
                            "Strong penetration",
                            "Very strong penetration",
                        ]
                    )
                    .sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with >= 50% support-qualified markets",

            "Observed Value":
                int(
                    (
                        track_market_penetration_df[
                            "support_qualified_market_pct"
                        ]
                        >= 50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with >= 50% above-average markets",

            "Observed Value":
                int(
                    (
                        track_market_penetration_df[
                            "above_country_average_market_pct"
                        ]
                        >= 50
                    ).sum()
                ),
        },
    ]
)

print()
print("Track market-penetration diagnostics")
print("=" * 100)

display(
    track_penetration_diagnostics_df
)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

track_population_valid = bool(
    len(
        track_market_penetration_df
    )
    == 110_198
)

track_identity_valid = bool(
    track_market_penetration_df[
        "track_id"
    ].nunique()
    == 110_198
)

markets_observed_valid = bool(
    np.isfinite(
        track_market_penetration_df[
            "markets_observed"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_market_penetration_df[
            "markets_observed"
        ]
        >= 1
    ).all()
)

relative_intensity_valid = bool(
    np.isfinite(
        track_market_penetration_df[
            "median_relative_stream_intensity"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_market_penetration_df[
            "median_relative_stream_intensity"
        ]
        >= 0
    ).all()
)

penetration_index_valid = bool(
    np.isfinite(
        track_market_penetration_df[
            "market_penetration_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_market_penetration_df[
        "market_penetration_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

percentage_columns = [
    "support_qualified_market_pct",
    "above_country_average_market_pct",
]

percentage_bounds_valid = bool(
    all(
        track_market_penetration_df[
            column
        ]
        .between(
            0,
            100,
            inclusive="both",
        )
        .all()
        for column in percentage_columns
    )
)

classification_complete = bool(
    track_market_penetration_df[
        "market_penetration_class"
    ]
    .notna()
    .all()
)

population_reconciled = bool(
    int(
        track_market_penetration_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_market_penetration_df
    )
)


section_8_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 8.1 completion",

            "Requirement":
                "Track Country Reach must already be complete.",

            "Observed Evidence":
                f"Section 8.1 completion status: "
                f"{section_8_1_status}",

            "Passed":
                section_8_1_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_market_penetration_df):,} "
                "track penetration records assessed",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one market-penetration record.",

            "Observed Evidence":
                f"{track_market_penetration_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                track_identity_valid,
        },
        {
            "Validation Area":
                "Observed-market validity",

            "Requirement":
                "Observed market counts must remain finite "
                "and greater than or equal to one.",

            "Observed Evidence":
                f"Observed range: "
                f"{int(track_market_penetration_df['markets_observed'].min())} "
                f"to "
                f"{int(track_market_penetration_df['markets_observed'].max())}",

            "Passed":
                markets_observed_valid,
        },
        {
            "Validation Area":
                "Relative stream-intensity validity",

            "Requirement":
                "Relative stream-intensity values must remain "
                "finite and non-negative.",

            "Observed Evidence":
                f"{len(track_market_penetration_df):,} "
                "relative intensity values assessed",

            "Passed":
                relative_intensity_valid,
        },
        {
            "Validation Area":
                "Penetration-index validity",

            "Requirement":
                "Market-penetration indices must remain finite "
                "and between 0 and 1.",

            "Observed Evidence":
                f"{len(track_market_penetration_df):,} "
                "penetration indices assessed",

            "Passed":
                penetration_index_valid,
        },
        {
            "Validation Area":
                "Percentage-field bounds",

            "Requirement":
                "Support-qualified and above-average market "
                "percentages must remain between 0 and 100.",

            "Observed Evidence":
                "2 track percentage fields assessed",

            "Passed":
                percentage_bounds_valid,
        },
        {
            "Validation Area":
                "Penetration classification completeness",

            "Requirement":
                "Every track must receive a market-penetration class.",

            "Observed Evidence":
                f"{track_market_penetration_df['market_penetration_class'].notna().sum():,} "
                "penetration classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Penetration population reconciliation",

            "Requirement":
                "Penetration classifications must reconcile "
                "with the complete track population.",

            "Observed Evidence":
                f"{int(track_market_penetration_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_market_penetration_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)

print()
print("Track market-penetration validation")
print("=" * 100)

display(
    section_8_2_validation_df
)


# ------------------------------------------------------------
# 12. Final completion status
# ------------------------------------------------------------

section_8_2_complete = bool(
    section_8_2_validation_df[
        "Passed"
    ].all()
)

if not section_8_2_complete:

    failed_checks = (
        section_8_2_validation_df
        .loc[
            ~section_8_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 8.2 track market-penetration validation "
        "failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 8.2 track market-penetration "
    "validation checks passed."
)

print(
    f"Section 8.2 completion status: "
    f"{section_8_2_complete}"
)

print(
    f"Track market-penetration intelligence prepared "
    f"for {len(track_market_penetration_df):,} tracks."
)

print(
    "Market penetration measures the depth of observed "
    "track performance within reached markets and is not "
    "yet a final PMIP track geographic ranking."
)
```

#### Interpretation

Section 8.2 successfully evaluates the **market penetration depth** of all **110,198 validated tracks** across the country markets in which they are observed.

The results show that most tracks have relatively limited penetration. The **median track appears in only 1 market**, has a **median relative stream intensity of 0.7419**, a **median support-qualified market share of 0.00%**, and a **median above-average market share of 0.00%**. The resulting **median market-penetration index is 0.1137**, indicating that the typical track has relatively weak depth of performance within its observed market footprint.

The penetration classification confirms this pattern. **70,355 tracks (63.84%)** are classified as **Limited penetration**, while **28,813 tracks (26.15%)** fall into the **Moderate penetration** category. Together, these two groups account for almost **90% of all validated tracks**.

A smaller group demonstrates stronger market penetration. **8,624 tracks (7.83%)** are classified as **Strong penetration**, while only **2,406 tracks (2.18%)** achieve **Very strong penetration**. This shows that strong within-market performance is considerably less common than basic market presence.

The relative stream-intensity diagnostics provide further evidence. **31,866 tracks** have a median relative stream intensity of at least **1.0**, meaning their typical observed market performance meets or exceeds the market-relative reference level. This falls to **11,681 tracks** at an intensity of at least **1.5**, and only **5,213 tracks** at an intensity of at least **2.0**.

A total of **11,030 tracks** are classified as either **Strong** or **Very strong penetration**. In addition, **14,748 tracks** have at least half of their observed markets supported by sufficient analytical evidence, while **33,318 tracks** perform above the relevant country average in at least half of their observed markets.

The strongest-observed penetration examples achieve a **market-penetration index of 1.0**. These tracks combine very high relative stream intensity with **100% support-qualified market coverage** and **100% above-average market coverage** within the markets where they are observed. However, several of these tracks appear in only one or two markets, so high penetration should not automatically be interpreted as broad international success.

This distinction is important for PMIP: **Section 8.1 measures geographic breadth, whereas Section 8.2 measures geographic depth**. A track can therefore have very strong penetration in a small number of countries while still having limited international reach.

All Section 8.2 validation checks passed. The complete **110,198-track population** was preserved, track identities remained unique, observed-market counts and relative-intensity measures remained valid, penetration indices stayed within the expected **0-to-1 range**, and every track received exactly one market-penetration classification.

Overall, Section 8.2 establishes how deeply each track performs within its reached markets. These measures provide an important input for later cross-market and geographic-profile analysis, but they remain **descriptive geographic intelligence rather than final PMIP track rankings**.

### 8.3 Cross-Market Track Performance

#### Purpose

Evaluate how consistently each validated track performs across the country markets in which it is observed.

This subsection uses the cross-market track features prepared in Section 5.4 to assess variation in streaming performance, chart performance, and market-to-market consistency. It focuses on whether a track performs at a similar level across several countries or whether its performance is highly uneven between markets.

The analysis considers measures such as stream-performance range, coefficient of variation across markets, chart-position range, comparison-eligible market share, and overall consistency classification.

The purpose is to distinguish tracks with stable cross-market performance from tracks whose geographic performance depends heavily on a small number of stronger markets.

These measures remain descriptive track-level geographic intelligence and are not yet final PMIP track geographic performance rankings.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP cross-market track performance")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_8_2_status = globals().get(
    "section_8_2_complete",
    False
)

print(
    f"Section 8.2 completion status: "
    f"{section_8_2_status}"
)

if not section_8_2_status:
    raise RuntimeError(
        "Section 8.3 requires Section 8.2 "
        "Track Market Penetration to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated cross-market track source
# ------------------------------------------------------------

def recover_track_cross_market_dataframe():

    preferred_names = [
        "track_cross_market_features_df",
        "track_cross_market_performance_df",
    ]

    for name in preferred_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if "track_id" not in candidate.columns:
            continue

        return candidate.copy(), name

    # Fallback search through notebook memory
    cross_market_fields = {
        "stream_performance_range",
        "stream_cv_across_markets",
        "chart_position_range",
        "comparison_eligible_market_pct",
    }

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if "track_id" not in candidate.columns:
            continue

        if any(
            field in candidate.columns
            for field in cross_market_fields
        ):
            return candidate.copy(), name

    return None, None


track_cross_market_source_df, track_cross_market_source_name = (
    recover_track_cross_market_dataframe()
)

if track_cross_market_source_df is None:
    raise RuntimeError(
        "Section 8.3 could not recover the validated "
        "track cross-market feature table from previous "
        "notebook sections."
    )


print()
print("Validated track cross-market source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_cross_market_source_name}"
)

print(
    f"Tracks: "
    f"{len(track_cross_market_source_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve source columns
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


track_name_col = first_existing_column(
    track_cross_market_source_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    track_cross_market_source_df,
    [
        "artists_raw",
        "artists",
    ]
)

markets_observed_col = first_existing_column(
    track_cross_market_source_df,
    [
        "markets_observed",
        "country_markets_reached",
        "country_markets",
    ]
)

median_market_streams_col = first_existing_column(
    track_cross_market_source_df,
    [
        "median_market_streams",
        "median_streams_across_markets",
    ]
)

strongest_market_streams_col = first_existing_column(
    track_cross_market_source_df,
    [
        "strongest_market_streams",
        "maximum_market_streams",
    ]
)

weakest_market_streams_col = first_existing_column(
    track_cross_market_source_df,
    [
        "weakest_market_streams",
        "minimum_market_streams",
    ]
)

stream_range_col = first_existing_column(
    track_cross_market_source_df,
    [
        "stream_performance_range",
        "stream_range",
    ]
)

stream_cv_col = first_existing_column(
    track_cross_market_source_df,
    [
        "stream_cv_across_markets",
        "stream_coefficient_of_variation",
    ]
)

chart_range_col = first_existing_column(
    track_cross_market_source_df,
    [
        "chart_position_range",
        "chart_range",
    ]
)

comparison_eligible_col = first_existing_column(
    track_cross_market_source_df,
    [
        "comparison_eligible_market_pct",
        "comparison_safe_market_pct",
    ]
)

existing_consistency_col = first_existing_column(
    track_cross_market_source_df,
    [
        "stream_consistency_class",
        "consistency_class",
    ]
)


required_fields = {
    "markets_observed": markets_observed_col,
    "stream_performance_range": stream_range_col,
    "stream_cv_across_markets": stream_cv_col,
    "chart_position_range": chart_range_col,
}

missing_fields = [
    name
    for name, column in required_fields.items()
    if column is None
]

if missing_fields:
    raise RuntimeError(
        "Section 8.3 requires the following cross-market "
        "fields: "
        + ", ".join(missing_fields)
    )


# ------------------------------------------------------------
# 4. Prepare canonical cross-market table
# ------------------------------------------------------------

track_cross_market_performance_df = (
    track_cross_market_source_df.copy()
)

track_cross_market_performance_df[
    "markets_observed"
] = pd.to_numeric(
    track_cross_market_performance_df[
        markets_observed_col
    ],
    errors="coerce",
)

track_cross_market_performance_df[
    "stream_performance_range"
] = pd.to_numeric(
    track_cross_market_performance_df[
        stream_range_col
    ],
    errors="coerce",
)

track_cross_market_performance_df[
    "stream_cv_across_markets"
] = pd.to_numeric(
    track_cross_market_performance_df[
        stream_cv_col
    ],
    errors="coerce",
)

track_cross_market_performance_df[
    "chart_position_range"
] = pd.to_numeric(
    track_cross_market_performance_df[
        chart_range_col
    ],
    errors="coerce",
)


if comparison_eligible_col is not None:

    track_cross_market_performance_df[
        "comparison_eligible_market_pct"
    ] = pd.to_numeric(
        track_cross_market_performance_df[
            comparison_eligible_col
        ],
        errors="coerce",
    )

else:

    track_cross_market_performance_df[
        "comparison_eligible_market_pct"
    ] = 0.0


if median_market_streams_col is not None:

    track_cross_market_performance_df[
        "median_market_streams"
    ] = pd.to_numeric(
        track_cross_market_performance_df[
            median_market_streams_col
        ],
        errors="coerce",
    )

else:

    track_cross_market_performance_df[
        "median_market_streams"
    ] = np.nan


if strongest_market_streams_col is not None:

    track_cross_market_performance_df[
        "strongest_market_streams"
    ] = pd.to_numeric(
        track_cross_market_performance_df[
            strongest_market_streams_col
        ],
        errors="coerce",
    )

else:

    track_cross_market_performance_df[
        "strongest_market_streams"
    ] = np.nan


if weakest_market_streams_col is not None:

    track_cross_market_performance_df[
        "weakest_market_streams"
    ] = pd.to_numeric(
        track_cross_market_performance_df[
            weakest_market_streams_col
        ],
        errors="coerce",
    )

else:

    track_cross_market_performance_df[
        "weakest_market_streams"
    ] = np.nan


# ------------------------------------------------------------
# 5. Cross-market consistency classification
# ------------------------------------------------------------

if existing_consistency_col is not None:

    track_cross_market_performance_df[
        "cross_market_consistency_class"
    ] = (
        track_cross_market_performance_df[
            existing_consistency_col
        ]
        .astype("string")
    )

else:

    def classify_cross_market_consistency(row):

        markets = row["markets_observed"]
        cv = row["stream_cv_across_markets"]

        if markets <= 1:
            return "Very consistent"

        if cv <= 0.50:
            return "Very consistent"

        if cv <= 1.00:
            return "Consistent"

        if cv <= 2.00:
            return "Variable"

        return "Highly variable"

    track_cross_market_performance_df[
        "cross_market_consistency_class"
    ] = (
        track_cross_market_performance_df
        .apply(
            classify_cross_market_consistency,
            axis=1,
        )
    )


# ------------------------------------------------------------
# 6. Build cross-market consistency index
# ------------------------------------------------------------

cv_values = (
    track_cross_market_performance_df[
        "stream_cv_across_markets"
    ]
)

cv_upper = float(
    cv_values.quantile(0.99)
)

if (
    not np.isfinite(cv_upper)
    or cv_upper <= 0
):
    cv_upper = 1.0


chart_values = (
    track_cross_market_performance_df[
        "chart_position_range"
    ]
)

chart_upper = float(
    chart_values.quantile(0.99)
)

if (
    not np.isfinite(chart_upper)
    or chart_upper <= 0
):
    chart_upper = 1.0


track_cross_market_performance_df[
    "stream_consistency_component"
] = (
    1
    -
    (
        cv_values
        .clip(
            lower=0,
            upper=cv_upper,
        )
        /
        cv_upper
    )
)


track_cross_market_performance_df[
    "chart_consistency_component"
] = (
    1
    -
    (
        chart_values
        .clip(
            lower=0,
            upper=chart_upper,
        )
        /
        chart_upper
    )
)


track_cross_market_performance_df[
    "comparison_support_component"
] = (
    track_cross_market_performance_df[
        "comparison_eligible_market_pct"
    ]
    .clip(
        lower=0,
        upper=100,
    )
    / 100
)


track_cross_market_performance_df[
    "cross_market_consistency_index"
] = (
    (
        0.50
        * track_cross_market_performance_df[
            "stream_consistency_component"
        ]
    )
    +
    (
        0.25
        * track_cross_market_performance_df[
            "chart_consistency_component"
        ]
    )
    +
    (
        0.25
        * track_cross_market_performance_df[
            "comparison_support_component"
        ]
    )
)

track_cross_market_performance_df[
    "cross_market_consistency_index"
] = (
    track_cross_market_performance_df[
        "cross_market_consistency_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 7. Overview
# ------------------------------------------------------------

print()
print("Track cross-market performance overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_cross_market_performance_df):,}"
)

print(
    f"Median markets observed: "
    f"{track_cross_market_performance_df['markets_observed'].median():.1f}"
)

print(
    f"Median stream CV across markets: "
    f"{track_cross_market_performance_df['stream_cv_across_markets'].median():.4f}"
)

print(
    f"Median stream performance range: "
    f"{track_cross_market_performance_df['stream_performance_range'].median():,.2f}"
)

print(
    f"Median chart-position range: "
    f"{track_cross_market_performance_df['chart_position_range'].median():.2f}"
)

print(
    f"Median comparison-eligible market share: "
    f"{track_cross_market_performance_df['comparison_eligible_market_pct'].median():.2f}%"
)

print(
    f"Median cross-market consistency index: "
    f"{track_cross_market_performance_df['cross_market_consistency_index'].median():.4f}"
)


# ------------------------------------------------------------
# 8. Consistency classification summary
# ------------------------------------------------------------

consistency_order = [
    "Very consistent",
    "Consistent",
    "Variable",
    "Highly variable",
]

track_cross_market_summary_df = (
    track_cross_market_performance_df[
        "cross_market_consistency_class"
    ]
    .value_counts()
    .reindex(
        consistency_order,
        fill_value=0,
    )
    .rename_axis(
        "Consistency Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_cross_market_summary_df[
    "Share (%)"
] = (
    track_cross_market_summary_df[
        "Tracks"
    ]
    /
    len(track_cross_market_performance_df)
    * 100
)

print()
print("Track cross-market consistency classification")
print("=" * 100)

display(
    track_cross_market_summary_df
)


# ------------------------------------------------------------
# 9. Most consistent multi-market tracks
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_observed",
    "median_market_streams",
    "strongest_market_streams",
    "weakest_market_streams",
    "stream_performance_range",
    "stream_cv_across_markets",
    "chart_position_range",
    "comparison_eligible_market_pct",
    "cross_market_consistency_index",
    "cross_market_consistency_class",
]


print()
print(
    "Tracks with most consistent supported cross-market performance"
)
print("=" * 100)

most_consistent_tracks_df = (
    track_cross_market_performance_df
    .loc[
        track_cross_market_performance_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "cross_market_consistency_index",
            "stream_cv_across_markets",
            "markets_observed",
        ],
        ascending=[
            False,
            True,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    most_consistent_tracks_df
)


# ------------------------------------------------------------
# 10. Tracks with greatest cross-market variation
# ------------------------------------------------------------

print()
print(
    "Tracks with greatest observed cross-market variation"
)
print("=" * 100)

most_variable_tracks_df = (
    track_cross_market_performance_df
    .loc[
        track_cross_market_performance_df[
            "markets_observed"
        ] >= 5
    ]
    .sort_values(
        [
            "stream_cv_across_markets",
            "stream_performance_range",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    most_variable_tracks_df
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

track_cross_market_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Single-market tracks",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "markets_observed"
                        ]
                        == 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks observed in at least 5 markets",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "markets_observed"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks classified very consistent",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "cross_market_consistency_class"
                        ]
                        == "Very consistent"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks classified highly variable",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "cross_market_consistency_class"
                        ]
                        == "Highly variable"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with stream CV <= 0.50",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "stream_cv_across_markets"
                        ]
                        <= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with comparison-eligible market share >= 50%",

            "Observed Value":
                int(
                    (
                        track_cross_market_performance_df[
                            "comparison_eligible_market_pct"
                        ]
                        >= 50
                    ).sum()
                ),
        },
    ]
)

print()
print("Track cross-market performance diagnostics")
print("=" * 100)

display(
    track_cross_market_diagnostics_df
)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_cross_market_performance_df
    )
    == 110_198
)

identity_valid = bool(
    track_cross_market_performance_df[
        "track_id"
    ].nunique()
    == 110_198
)

markets_valid = bool(
    np.isfinite(
        track_cross_market_performance_df[
            "markets_observed"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_cross_market_performance_df[
            "markets_observed"
        ]
        >= 1
    ).all()
)

stream_range_valid = bool(
    np.isfinite(
        track_cross_market_performance_df[
            "stream_performance_range"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_cross_market_performance_df[
            "stream_performance_range"
        ]
        >= 0
    ).all()
)

stream_cv_valid = bool(
    np.isfinite(
        track_cross_market_performance_df[
            "stream_cv_across_markets"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_cross_market_performance_df[
            "stream_cv_across_markets"
        ]
        >= 0
    ).all()
)

chart_range_valid = bool(
    np.isfinite(
        track_cross_market_performance_df[
            "chart_position_range"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_cross_market_performance_df[
            "chart_position_range"
        ]
        >= 0
    ).all()
)

comparison_pct_valid = bool(
    track_cross_market_performance_df[
        "comparison_eligible_market_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

consistency_index_valid = bool(
    np.isfinite(
        track_cross_market_performance_df[
            "cross_market_consistency_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_cross_market_performance_df[
        "cross_market_consistency_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

classification_complete = bool(
    track_cross_market_performance_df[
        "cross_market_consistency_class"
    ]
    .notna()
    .all()
)

population_reconciled = bool(
    int(
        track_cross_market_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_cross_market_performance_df
    )
)


section_8_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 8.2 completion",

            "Requirement":
                "Track Market Penetration must already be complete.",

            "Observed Evidence":
                f"Section 8.2 completion status: "
                f"{section_8_2_status}",

            "Passed":
                section_8_2_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_cross_market_performance_df):,} "
                "cross-market records assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one cross-market record.",

            "Observed Evidence":
                f"{track_cross_market_performance_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Observed-market validity",

            "Requirement":
                "Observed market counts must remain finite "
                "and greater than or equal to one.",

            "Observed Evidence":
                f"Observed range: "
                f"{int(track_cross_market_performance_df['markets_observed'].min())} "
                f"to "
                f"{int(track_cross_market_performance_df['markets_observed'].max())}",

            "Passed":
                markets_valid,
        },
        {
            "Validation Area":
                "Stream-range validity",

            "Requirement":
                "Cross-market stream-performance ranges must "
                "remain finite and non-negative.",

            "Observed Evidence":
                f"{len(track_cross_market_performance_df):,} "
                "stream ranges assessed",

            "Passed":
                stream_range_valid,
        },
        {
            "Validation Area":
                "Stream-CV validity",

            "Requirement":
                "Cross-market stream coefficients of variation "
                "must remain finite and non-negative.",

            "Observed Evidence":
                f"{len(track_cross_market_performance_df):,} "
                "stream CV values assessed",

            "Passed":
                stream_cv_valid,
        },
        {
            "Validation Area":
                "Chart-range validity",

            "Requirement":
                "Chart-position ranges must remain finite "
                "and non-negative.",

            "Observed Evidence":
                f"{len(track_cross_market_performance_df):,} "
                "chart ranges assessed",

            "Passed":
                chart_range_valid,
        },
        {
            "Validation Area":
                "Comparison percentage bounds",

            "Requirement":
                "Comparison-eligible market percentages must "
                "remain between 0 and 100.",

            "Observed Evidence":
                "Comparison-eligible percentages assessed "
                "for all tracks",

            "Passed":
                comparison_pct_valid,
        },
        {
            "Validation Area":
                "Consistency-index validity",

            "Requirement":
                "Cross-market consistency indices must remain "
                "finite and between 0 and 1.",

            "Observed Evidence":
                f"{len(track_cross_market_performance_df):,} "
                "consistency indices assessed",

            "Passed":
                consistency_index_valid,
        },
        {
            "Validation Area":
                "Consistency classification completeness",

            "Requirement":
                "Every track must receive a cross-market "
                "consistency classification.",

            "Observed Evidence":
                f"{track_cross_market_performance_df['cross_market_consistency_class'].notna().sum():,} "
                "consistency classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Cross-market population reconciliation",

            "Requirement":
                "Consistency classifications must reconcile "
                "with the complete validated track population.",

            "Observed Evidence":
                f"{int(track_cross_market_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_cross_market_performance_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)

print()
print("Track cross-market performance validation")
print("=" * 100)

display(
    section_8_3_validation_df
)


# ------------------------------------------------------------
# 13. Final completion status
# ------------------------------------------------------------

section_8_3_complete = bool(
    section_8_3_validation_df[
        "Passed"
    ].all()
)

if not section_8_3_complete:

    failed_checks = (
        section_8_3_validation_df
        .loc[
            ~section_8_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 8.3 cross-market track performance "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )

print()
print(
    "All Section 8.3 cross-market track performance "
    "validation checks passed."
)

print(
    f"Section 8.3 completion status: "
    f"{section_8_3_complete}"
)

print(
    f"Cross-market track performance intelligence prepared "
    f"for {len(track_cross_market_performance_df):,} tracks."
)

print(
    "Cross-market measures describe consistency and "
    "variation between observed country markets and are "
    "not yet final PMIP track geographic rankings."
)
```

#### Interpretation

Section 8.3 successfully evaluates how consistently the **110,198 validated tracks** perform across the country markets in which they are observed.

The results show that the typical track has very limited cross-market evidence. The **median track is observed in only 1 market**, which explains why the median **stream coefficient of variation is 0.0000**, the median **stream-performance range is 0.00**, and the median **chart-position range is 0.00**. These zero values do not necessarily indicate genuinely stable international performance; for many tracks, there is simply no second market available for comparison.

The classification distribution reflects this structure. **82,257 tracks (74.64%)** are classified as **Very consistent**, while **1,823 tracks (1.65%)** are classified as **Consistent**. However, this result must be interpreted carefully because **81,082 tracks are single-market tracks**. A single-market track has no observed cross-market variation, so its apparent consistency mainly reflects limited geographic evidence rather than confirmed stability across multiple countries.

Among tracks with broader geographic presence, more meaningful variation becomes visible. **10,971 tracks (9.96%)** are classified as **Variable**, while **15,147 tracks (13.75%)** are classified as **Highly variable**. These tracks show stronger differences in streaming or chart behaviour between markets, suggesting that their success is more geographically uneven.

The diagnostics show that **11,257 tracks** appear in at least **5 country markets**, providing a much stronger basis for true cross-market comparison. These multi-market tracks are particularly important for interpreting geographic consistency because their stream and chart variation reflects behaviour across several distinct markets rather than a single observation context.

A total of **84,080 tracks** have a stream coefficient of variation of **0.50 or lower**, indicating relatively low stream variation according to the engineered metric. However, this figure should again be interpreted alongside market breadth, since the large single-market population naturally contributes many zero-variation observations.

Only **13,618 tracks** have at least **50% comparison-eligible market coverage**. This highlights an important analytical safeguard: even when a track is observed in several countries, not all those markets necessarily provide equally reliable temporal or historical support for direct comparison.

The median **cross-market consistency index is 0.7500**. This relatively high median is strongly influenced by the large number of single-market tracks, which receive zero observed stream and chart variation. Therefore, the consistency index should be interpreted together with `markets_observed` and comparison eligibility rather than used alone as evidence of international stability.

Overall, Section 8.3 establishes the distinction between **observed consistency** and **well-supported cross-market consistency**. Tracks with several observed markets and low variation provide stronger evidence of genuinely stable geographic performance, while single-market tracks should not be interpreted as internationally consistent simply because no variation can be observed.

All Section 8.3 validation checks passed. The full **110,198-track population** was preserved, track identities remained unique, stream and chart variation measures remained valid and non-negative, consistency indices remained within the expected **0-to-1 range**, and every track received a cross-market consistency classification.

These results provide the cross-market stability and variation intelligence required for the next stage of Track Geographic Intelligence, while remaining descriptive analytical outputs rather than final PMIP track rankings.

### 8.4 Track Geographic Concentration

#### Purpose

Assess how strongly each validated track's observed streaming activity is concentrated within a small number of country markets.

This subsection uses the track geographic-concentration features engineered in Section 5.3 to evaluate whether a track's streams are dominated by one or a few markets or are distributed more evenly across its geographic footprint.

The analysis considers measures such as strongest-market stream share, top-three market stream share, stream-concentration HHI, effective stream-market count, supported market count, and the previously prepared geographic-concentration classification.

The purpose is to distinguish highly market-dependent tracks from tracks with broader and more diversified geographic performance.

These concentration measures describe the distribution of observed streaming activity across markets and remain descriptive geographic intelligence rather than final PMIP track geographic performance rankings.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP track geographic concentration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_8_3_status = globals().get(
    "section_8_3_complete",
    False
)

print(
    f"Section 8.3 completion status: "
    f"{section_8_3_status}"
)

if not section_8_3_status:
    raise RuntimeError(
        "Section 8.4 requires Section 8.3 "
        "Cross-Market Track Performance to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track concentration source
# ------------------------------------------------------------

def recover_track_concentration_dataframe():

    preferred_names = [
        "track_stream_concentration_df",
        "track_concentration_features_df",
        "track_geographic_concentration_df",
    ]

    for name in preferred_names:

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if "track_id" not in candidate.columns:
            continue

        return candidate.copy(), name

    # Fallback search through notebook memory
    concentration_fields = {
        "top_1_stream_share_pct",
        "top_3_stream_share_pct",
        "stream_concentration_hhi",
        "effective_stream_markets",
        "geographic_concentration_class",
    }

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if "track_id" not in candidate.columns:
            continue

        if any(
            field in candidate.columns
            for field in concentration_fields
        ):
            return candidate.copy(), name

    return None, None


track_concentration_source_df, track_concentration_source_name = (
    recover_track_concentration_dataframe()
)

if track_concentration_source_df is None:
    raise RuntimeError(
        "Section 8.4 could not recover the validated "
        "track geographic-concentration feature table "
        "from previous notebook sections."
    )


print()
print("Validated track geographic-concentration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_concentration_source_name}"
)

print(
    f"Tracks: "
    f"{len(track_concentration_source_df):,}"
)


# ------------------------------------------------------------
# 3. Resolve source columns
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


track_name_col = first_existing_column(
    track_concentration_source_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    track_concentration_source_df,
    [
        "artists_raw",
        "artists",
    ]
)

markets_observed_col = first_existing_column(
    track_concentration_source_df,
    [
        "markets_observed",
        "country_markets_reached",
        "country_markets",
    ]
)

top_1_stream_share_col = first_existing_column(
    track_concentration_source_df,
    [
        "top_1_stream_share_pct",
        "strongest_market_stream_share_pct",
    ]
)

top_3_stream_share_col = first_existing_column(
    track_concentration_source_df,
    [
        "top_3_stream_share_pct",
    ]
)

top_5_stream_share_col = first_existing_column(
    track_concentration_source_df,
    [
        "top_5_stream_share_pct",
    ]
)

stream_hhi_col = first_existing_column(
    track_concentration_source_df,
    [
        "stream_concentration_hhi",
    ]
)

effective_stream_markets_col = first_existing_column(
    track_concentration_source_df,
    [
        "effective_stream_markets",
    ]
)

supported_market_count_col = first_existing_column(
    track_concentration_source_df,
    [
        "supported_market_count",
        "markets_with_moderate_or_strong_support",
    ]
)

supported_stream_hhi_col = first_existing_column(
    track_concentration_source_df,
    [
        "supported_stream_concentration_hhi",
    ]
)

existing_class_col = first_existing_column(
    track_concentration_source_df,
    [
        "geographic_concentration_class",
        "concentration_class",
    ]
)


required_fields = {
    "top_1_stream_share_pct":
        top_1_stream_share_col,

    "stream_concentration_hhi":
        stream_hhi_col,

    "effective_stream_markets":
        effective_stream_markets_col,
}

missing_fields = [
    name
    for name, column in required_fields.items()
    if column is None
]

if missing_fields:
    raise RuntimeError(
        "Section 8.4 requires the following track "
        "concentration fields: "
        + ", ".join(missing_fields)
    )


# ------------------------------------------------------------
# 4. Prepare canonical concentration table
# ------------------------------------------------------------

track_geographic_concentration_df = (
    track_concentration_source_df.copy()
)


if markets_observed_col is not None:

    track_geographic_concentration_df[
        "markets_observed"
    ] = pd.to_numeric(
        track_geographic_concentration_df[
            markets_observed_col
        ],
        errors="coerce",
    )

else:

    track_geographic_concentration_df[
        "markets_observed"
    ] = np.nan


track_geographic_concentration_df[
    "top_1_stream_share_pct"
] = pd.to_numeric(
    track_geographic_concentration_df[
        top_1_stream_share_col
    ],
    errors="coerce",
)


if top_3_stream_share_col is not None:

    track_geographic_concentration_df[
        "top_3_stream_share_pct"
    ] = pd.to_numeric(
        track_geographic_concentration_df[
            top_3_stream_share_col
        ],
        errors="coerce",
    )

else:

    track_geographic_concentration_df[
        "top_3_stream_share_pct"
    ] = np.nan


if top_5_stream_share_col is not None:

    track_geographic_concentration_df[
        "top_5_stream_share_pct"
    ] = pd.to_numeric(
        track_geographic_concentration_df[
            top_5_stream_share_col
        ],
        errors="coerce",
    )

else:

    track_geographic_concentration_df[
        "top_5_stream_share_pct"
    ] = np.nan


track_geographic_concentration_df[
    "stream_concentration_hhi"
] = pd.to_numeric(
    track_geographic_concentration_df[
        stream_hhi_col
    ],
    errors="coerce",
)


track_geographic_concentration_df[
    "effective_stream_markets"
] = pd.to_numeric(
    track_geographic_concentration_df[
        effective_stream_markets_col
    ],
    errors="coerce",
)


if supported_market_count_col is not None:

    track_geographic_concentration_df[
        "supported_market_count"
    ] = pd.to_numeric(
        track_geographic_concentration_df[
            supported_market_count_col
        ],
        errors="coerce",
    )

else:

    track_geographic_concentration_df[
        "supported_market_count"
    ] = 0


if supported_stream_hhi_col is not None:

    track_geographic_concentration_df[
        "supported_stream_concentration_hhi"
    ] = pd.to_numeric(
        track_geographic_concentration_df[
            supported_stream_hhi_col
        ],
        errors="coerce",
    )

else:

    track_geographic_concentration_df[
        "supported_stream_concentration_hhi"
    ] = np.nan


# ------------------------------------------------------------
# 5. Concentration classification
# ------------------------------------------------------------

if existing_class_col is not None:

    track_geographic_concentration_df[
        "geographic_concentration_class"
    ] = (
        track_geographic_concentration_df[
            existing_class_col
        ]
        .astype("string")
    )

else:

    def classify_track_concentration(hhi):

        if hhi >= 0.75:
            return "Very high concentration"

        if hhi >= 0.50:
            return "High concentration"

        if hhi >= 0.25:
            return "Moderate concentration"

        return "Low concentration"

    track_geographic_concentration_df[
        "geographic_concentration_class"
    ] = (
        track_geographic_concentration_df[
            "stream_concentration_hhi"
        ]
        .apply(
            classify_track_concentration
        )
    )


# ------------------------------------------------------------
# 6. Build diversification index
# ------------------------------------------------------------

track_geographic_concentration_df[
    "geographic_diversification_index"
] = (
    1
    -
    track_geographic_concentration_df[
        "stream_concentration_hhi"
    ]
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 7. Market-dependency classification
# ------------------------------------------------------------

def classify_market_dependency(row):

    top_1_share = row[
        "top_1_stream_share_pct"
    ]

    effective_markets = row[
        "effective_stream_markets"
    ]

    hhi = row[
        "stream_concentration_hhi"
    ]

    if (
        top_1_share >= 85
        or
        effective_markets < 1.5
    ):
        return "Single-market dependent"

    if (
        top_1_share >= 60
        or
        effective_markets < 2.5
    ):
        return "Few-market dependent"

    if (
        hhi >= 0.35
        or
        effective_markets < 5
    ):
        return "Moderately concentrated"

    return "Diversified"


track_geographic_concentration_df[
    "market_dependency_class"
] = (
    track_geographic_concentration_df
    .apply(
        classify_market_dependency,
        axis=1,
    )
)


# ------------------------------------------------------------
# 8. Overview
# ------------------------------------------------------------

print()
print("Track geographic-concentration overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_concentration_df):,}"
)

print(
    f"Median strongest-market stream share: "
    f"{track_geographic_concentration_df['top_1_stream_share_pct'].median():.4f}%"
)

if (
    track_geographic_concentration_df[
        "top_3_stream_share_pct"
    ]
    .notna()
    .any()
):

    print(
        f"Median top-three market stream share: "
        f"{track_geographic_concentration_df['top_3_stream_share_pct'].median():.4f}%"
    )

print(
    f"Median stream-concentration HHI: "
    f"{track_geographic_concentration_df['stream_concentration_hhi'].median():.4f}"
)

print(
    f"Median effective stream markets: "
    f"{track_geographic_concentration_df['effective_stream_markets'].median():.2f}"
)

print(
    f"Median geographic diversification index: "
    f"{track_geographic_concentration_df['geographic_diversification_index'].median():.4f}"
)


# ------------------------------------------------------------
# 9. Concentration classification summary
# ------------------------------------------------------------

concentration_order = [
    "Very high concentration",
    "High concentration",
    "Moderate concentration",
    "Low concentration",
]

track_concentration_summary_df = (
    track_geographic_concentration_df[
        "geographic_concentration_class"
    ]
    .value_counts()
    .reindex(
        concentration_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Concentration Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_concentration_summary_df[
    "Share (%)"
] = (
    track_concentration_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_concentration_df)
    * 100
)


print()
print("Track geographic-concentration classification")
print("=" * 100)

display(
    track_concentration_summary_df
)


# ------------------------------------------------------------
# 10. Market-dependency summary
# ------------------------------------------------------------

dependency_order = [
    "Single-market dependent",
    "Few-market dependent",
    "Moderately concentrated",
    "Diversified",
]

track_dependency_summary_df = (
    track_geographic_concentration_df[
        "market_dependency_class"
    ]
    .value_counts()
    .reindex(
        dependency_order,
        fill_value=0,
    )
    .rename_axis(
        "Market Dependency Class"
    )
    .reset_index(
        name="Tracks"
    )
)

track_dependency_summary_df[
    "Share (%)"
] = (
    track_dependency_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_concentration_df)
    * 100
)


print()
print("Track market-dependency classification")
print("=" * 100)

display(
    track_dependency_summary_df
)


# ------------------------------------------------------------
# 11. Tracks with strongest geographic concentration
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_observed",
    "top_1_stream_share_pct",
    "top_3_stream_share_pct",
    "stream_concentration_hhi",
    "effective_stream_markets",
    "supported_market_count",
    "geographic_diversification_index",
    "geographic_concentration_class",
    "market_dependency_class",
]


print()
print(
    "Tracks with strongest observed geographic concentration"
)
print("=" * 100)

strongest_concentration_tracks_df = (
    track_geographic_concentration_df
    .sort_values(
        [
            "stream_concentration_hhi",
            "top_1_stream_share_pct",
            "effective_stream_markets",
        ],
        ascending=[
            False,
            False,
            True,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_concentration_tracks_df
)


# ------------------------------------------------------------
# 12. Most geographically diversified tracks
# ------------------------------------------------------------

print()
print(
    "Tracks with most diversified observed geographic performance"
)
print("=" * 100)

most_diversified_tracks_df = (
    track_geographic_concentration_df
    .loc[
        track_geographic_concentration_df[
            "effective_stream_markets"
        ] >= 2
    ]
    .sort_values(
        [
            "geographic_diversification_index",
            "effective_stream_markets",
            "top_1_stream_share_pct",
        ],
        ascending=[
            False,
            False,
            True,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    most_diversified_tracks_df
)


# ------------------------------------------------------------
# 13. Diagnostics
# ------------------------------------------------------------

track_concentration_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with HHI >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "stream_concentration_hhi"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with HHI < 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "stream_concentration_hhi"
                        ]
                        < 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with effective market count >= 5",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "effective_stream_markets"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Single-market dependent tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "market_dependency_class"
                        ]
                        == "Single-market dependent"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Few-market dependent tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "market_dependency_class"
                        ]
                        == "Few-market dependent"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Diversified tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_df[
                            "market_dependency_class"
                        ]
                        == "Diversified"
                    ).sum()
                ),
        },
    ]
)


print()
print("Track geographic-concentration diagnostics")
print("=" * 100)

display(
    track_concentration_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_concentration_df
    )
    == 110_198
)

identity_valid = bool(
    track_geographic_concentration_df[
        "track_id"
    ].nunique()
    == 110_198
)

top_1_share_valid = bool(
    np.isfinite(
        track_geographic_concentration_df[
            "top_1_stream_share_pct"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_geographic_concentration_df[
        "top_1_stream_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

hhi_valid = bool(
    np.isfinite(
        track_geographic_concentration_df[
            "stream_concentration_hhi"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_geographic_concentration_df[
        "stream_concentration_hhi"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

effective_markets_valid = bool(
    np.isfinite(
        track_geographic_concentration_df[
            "effective_stream_markets"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        track_geographic_concentration_df[
            "effective_stream_markets"
        ]
        >= 1
    ).all()
)

diversification_valid = bool(
    np.isfinite(
        track_geographic_concentration_df[
            "geographic_diversification_index"
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
    and
    track_geographic_concentration_df[
        "geographic_diversification_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)

concentration_class_complete = bool(
    track_geographic_concentration_df[
        "geographic_concentration_class"
    ]
    .notna()
    .all()
)

dependency_class_complete = bool(
    track_geographic_concentration_df[
        "market_dependency_class"
    ]
    .notna()
    .all()
)

concentration_population_reconciled = bool(
    int(
        track_concentration_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_concentration_df
    )
)

dependency_population_reconciled = bool(
    int(
        track_dependency_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_concentration_df
    )
)


section_8_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 8.3 completion",

            "Requirement":
                "Cross-Market Track Performance must already "
                "be complete.",

            "Observed Evidence":
                f"Section 8.3 completion status: "
                f"{section_8_3_status}",

            "Passed":
                section_8_3_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_concentration_df):,} "
                "track concentration records assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one concentration record.",

            "Observed Evidence":
                f"{track_geographic_concentration_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Strongest-market stream-share validity",

            "Requirement":
                "Strongest-market stream shares must remain "
                "finite and between 0 and 100.",

            "Observed Evidence":
                "Top-1 stream shares assessed for all tracks",

            "Passed":
                top_1_share_valid,
        },
        {
            "Validation Area":
                "HHI validity",

            "Requirement":
                "Stream-concentration HHI values must remain "
                "finite and between 0 and 1.",

            "Observed Evidence":
                f"{len(track_geographic_concentration_df):,} "
                "HHI values assessed",

            "Passed":
                hhi_valid,
        },
        {
            "Validation Area":
                "Effective-market validity",

            "Requirement":
                "Effective stream-market counts must remain "
                "finite and at least one.",

            "Observed Evidence":
                f"{len(track_geographic_concentration_df):,} "
                "effective-market values assessed",

            "Passed":
                effective_markets_valid,
        },
        {
            "Validation Area":
                "Diversification-index validity",

            "Requirement":
                "Geographic diversification indices must "
                "remain finite and between 0 and 1.",

            "Observed Evidence":
                f"{len(track_geographic_concentration_df):,} "
                "diversification indices assessed",

            "Passed":
                diversification_valid,
        },
        {
            "Validation Area":
                "Concentration classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "concentration class.",

            "Observed Evidence":
                f"{track_geographic_concentration_df['geographic_concentration_class'].notna().sum():,} "
                "concentration classes assigned",

            "Passed":
                concentration_class_complete,
        },
        {
            "Validation Area":
                "Dependency classification completeness",

            "Requirement":
                "Every track must receive a market-dependency class.",

            "Observed Evidence":
                f"{track_geographic_concentration_df['market_dependency_class'].notna().sum():,} "
                "dependency classes assigned",

            "Passed":
                dependency_class_complete,
        },
        {
            "Validation Area":
                "Concentration population reconciliation",

            "Requirement":
                "Concentration classifications must reconcile "
                "with the complete track population.",

            "Observed Evidence":
                f"{int(track_concentration_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_concentration_df):,} tracks",

            "Passed":
                concentration_population_reconciled,
        },
        {
            "Validation Area":
                "Dependency population reconciliation",

            "Requirement":
                "Dependency classifications must reconcile "
                "with the complete track population.",

            "Observed Evidence":
                f"{int(track_dependency_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_concentration_df):,} tracks",

            "Passed":
                dependency_population_reconciled,
        },
    ]
)


print()
print("Track geographic-concentration validation")
print("=" * 100)

display(
    section_8_4_validation_df
)


# ------------------------------------------------------------
# 15. Final completion status
# ------------------------------------------------------------

section_8_4_complete = bool(
    section_8_4_validation_df[
        "Passed"
    ].all()
)

if not section_8_4_complete:

    failed_checks = (
        section_8_4_validation_df
        .loc[
            ~section_8_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 8.4 track geographic-concentration "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 8.4 track geographic-concentration "
    "validation checks passed."
)

print(
    f"Section 8.4 completion status: "
    f"{section_8_4_complete}"
)

print(
    f"Track geographic-concentration intelligence prepared "
    f"for {len(track_geographic_concentration_df):,} tracks."
)

print(
    "Geographic concentration measures how strongly track "
    "streaming activity is concentrated across observed "
    "markets and is not yet a final PMIP track geographic ranking."
)
```

#### Interpretation

Section 8.4 successfully evaluates the **geographic concentration of streaming performance** for all **110,198 validated tracks**.

The results show that track performance in the dataset is overwhelmingly concentrated in a small number of country markets. The **median strongest-market stream share is 100.00%**, the **median top-three market stream share is also 100.00%**, and the **median stream-concentration HHI is 1.0000**. The median number of **effective stream markets is 1.00**, while the median geographic diversification index is **0.0000**.

This means the typical track in the dataset is effectively dependent on one market rather than having evenly distributed performance across several countries.

The concentration classification confirms this pattern. **101,494 tracks (92.10%)** are classified as having **Very high concentration**. A further **4,877 tracks (4.43%)** have **High concentration**, while only **2,427 tracks (2.20%)** fall into the **Moderate concentration** category and just **1,400 tracks (1.27%)** have **Low concentration**.

The market-dependency classification shows a similar result. **97,238 tracks (88.24%)** are classified as **Single-market dependent**, while **6,173 tracks (5.60%)** are **Few-market dependent**. Another **4,066 tracks (3.69%)** are moderately concentrated, and only **2,721 tracks (2.47%)** are classified as **Diversified**.

The diagnostics reinforce this interpretation. **95,091 tracks** have an HHI of at least **0.75**, showing very strong concentration of streams within a small geographic footprint. By contrast, only **8,704 tracks** have an HHI below **0.50**, indicating a more balanced distribution of streams across markets.

Only **2,721 tracks** have an effective stream-market count of at least **5**. This is a particularly important group because these tracks demonstrate genuinely broader geographic distribution rather than merely appearing in multiple markets with most streams still concentrated in one country.

These findings are consistent with the earlier Track Country Reach results. Section 8.1 showed that **81,082 tracks are single-market tracks**, so it is expected that concentration measures are heavily skewed toward one-market dependence. Therefore, very high concentration should not automatically be interpreted as weak performance. In many cases, it simply means the track has strong but geographically localised success.

The distinction between **reach** and **concentration** is important. A track may reach several markets but still generate most of its streams in one country, while another track may have fewer total streams but distribute them more evenly across multiple markets. Section 8.4 captures this difference.

All Section 8.4 validation checks passed. The full **110,198-track population** was preserved, track identities remained unique, strongest-market stream shares stayed within valid bounds, HHI and diversification measures remained within the expected **0-to-1 range**, and every track received both a geographic-concentration classification and a market-dependency classification.

Overall, Section 8.4 establishes how dependent each track is on a small number of markets and identifies the relatively small subset of tracks with genuinely diversified geographic performance. These measures remain descriptive intelligence and will be combined with reach, penetration, and cross-market consistency in the next subsection.

### 8.5 Track Geographic Performance Profiles

#### Purpose

Combine the validated track-level geographic intelligence produced in Sections 8.1 to 8.4 into a single geographic performance profile for every track.

This subsection brings together four complementary dimensions of track geography: international reach, market penetration, cross-market consistency, and geographic diversification.

The purpose is to distinguish tracks that are primarily localised from tracks showing emerging international presence, established international performance, diversified geographic distribution, or exceptionally broad and well-supported geographic performance.

A composite geographic profile index is also created to summarise the overall strength of each track's geographic footprint while preserving the individual underlying indicators for interpretation.

The resulting profiles provide a consolidated descriptive view of track geographic behaviour. They remain PMIP intelligence outputs and are not yet final ranked geographic performance scores.

```python
import pandas as pd
import numpy as np

print("Building PMIP track geographic performance profiles")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_8_4_status = globals().get(
    "section_8_4_complete",
    False
)

print(
    f"Section 8.4 completion status: "
    f"{section_8_4_status}"
)

if not section_8_4_status:
    raise RuntimeError(
        "Section 8.5 requires Section 8.4 "
        "Track Geographic Concentration to be complete."
    )


# ------------------------------------------------------------
# 2. Helper: recover the BEST matching dataframe
# ------------------------------------------------------------

def recover_best_dataframe(
    preferred_names,
    required_any,
    expected_rows=110_198,
):
    """
    Recover the most appropriate dataframe.

    A dataframe must:
    - contain track_id
    - contain the expected number of rows
    - contain at least one required analytical field

    Preferred names are checked first, followed by all
    dataframes currently available in notebook memory.
    """

    candidates = []

    # Preferred candidates first
    for priority, name in enumerate(preferred_names):

        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if "track_id" not in candidate.columns:
            continue

        matched_fields = [
            field
            for field in required_any
            if field in candidate.columns
        ]

        if not matched_fields:
            continue

        candidates.append(
            (
                len(matched_fields),
                -priority,
                name,
                candidate,
            )
        )

    # Search all notebook-memory dataframes
    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != expected_rows:
            continue

        if "track_id" not in candidate.columns:
            continue

        matched_fields = [
            field
            for field in required_any
            if field in candidate.columns
        ]

        if not matched_fields:
            continue

        # Avoid adding exact preferred candidates twice
        if name in preferred_names:
            continue

        candidates.append(
            (
                len(matched_fields),
                -1000,
                name,
                candidate,
            )
        )

    if not candidates:
        return None, None

    # Prefer dataframe containing the greatest number
    # of the requested analytical fields
    candidates.sort(
        key=lambda item: (
            item[0],
            item[1],
        ),
        reverse=True,
    )

    _, _, best_name, best_df = candidates[0]

    return best_df.copy(), best_name


# ------------------------------------------------------------
# 3. Recover Section 8.1 international-reach intelligence
# ------------------------------------------------------------

track_reach_df, track_reach_name = (
    recover_best_dataframe(
        preferred_names=[
            "track_international_reach_df",
            "track_country_reach_intelligence_df",
            "track_country_reach_features_df",
        ],
        required_any=[
            "international_reach_index",
            "international_market_reach_pct",
            "country_reach_pct",
            "markets_reached",
            "country_markets_reached",
        ],
    )
)


# ------------------------------------------------------------
# 4. Recover Section 8.2 market-penetration intelligence
# ------------------------------------------------------------

track_penetration_df, track_penetration_name = (
    recover_best_dataframe(
        preferred_names=[
            "track_market_penetration_intelligence_df",
            "track_market_penetration_df",
            "track_penetration_intelligence_df",
            "track_penetration_features_df",
        ],
        required_any=[
            "market_penetration_index",
            "market_penetration_class",
            "median_relative_stream_intensity",
            "support_qualified_market_pct",
            "above_country_average_market_pct",
        ],
    )
)


# ------------------------------------------------------------
# 5. Recover Section 8.3 cross-market intelligence
# ------------------------------------------------------------

track_cross_market_df, track_cross_market_name = (
    recover_best_dataframe(
        preferred_names=[
            "track_cross_market_performance_df",
            "track_cross_market_features_df",
        ],
        required_any=[
            "cross_market_consistency_index",
            "cross_market_consistency_class",
            "stream_consistency_class",
            "comparison_eligible_market_pct",
        ],
    )
)


# ------------------------------------------------------------
# 6. Recover Section 8.4 geographic-concentration intelligence
# ------------------------------------------------------------

track_concentration_df, track_concentration_name = (
    recover_best_dataframe(
        preferred_names=[
            "track_geographic_concentration_df",
            "track_stream_concentration_df",
            "track_concentration_features_df",
        ],
        required_any=[
            "geographic_diversification_index",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "market_dependency_class",
        ],
    )
)


# ------------------------------------------------------------
# 7. Validate recovered Section 8 sources
# ------------------------------------------------------------

missing_sources = []

if track_reach_df is None:
    missing_sources.append(
        "Section 8.1 international reach"
    )

if track_penetration_df is None:
    missing_sources.append(
        "Section 8.2 market penetration"
    )

if track_cross_market_df is None:
    missing_sources.append(
        "Section 8.3 cross-market performance"
    )

if track_concentration_df is None:
    missing_sources.append(
        "Section 8.4 geographic concentration"
    )

if missing_sources:
    raise RuntimeError(
        "Section 8.5 could not recover: "
        + ", ".join(missing_sources)
    )


print()
print("Validated Section 8 track intelligence sources")
print("=" * 100)

print(
    f"International reach source: "
    f"{track_reach_name}"
)

print(
    f"Market penetration source: "
    f"{track_penetration_name}"
)

print(
    f"Cross-market performance source: "
    f"{track_cross_market_name}"
)

print(
    f"Geographic concentration source: "
    f"{track_concentration_name}"
)

print(
    f"Track population: "
    f"{len(track_reach_df):,}"
)


# ------------------------------------------------------------
# 8. Helper: resolve available columns
# ------------------------------------------------------------

def first_existing_column(
    df,
    candidates,
):

    for column in candidates:

        if column in df.columns:
            return column

    return None


# ------------------------------------------------------------
# 9. Resolve identity fields
# ------------------------------------------------------------

track_name_col = first_existing_column(
    track_reach_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    track_reach_df,
    [
        "artists_raw",
        "artists",
    ]
)


# ------------------------------------------------------------
# 10. Resolve reach fields
# ------------------------------------------------------------

reach_pct_col = first_existing_column(
    track_reach_df,
    [
        "international_market_reach_pct",
        "country_reach_pct",
    ]
)

reach_index_col = first_existing_column(
    track_reach_df,
    [
        "international_reach_index",
        "country_reach_index",
    ]
)

markets_reached_col = first_existing_column(
    track_reach_df,
    [
        "markets_reached",
        "country_markets_reached",
        "country_markets",
        "markets_observed",
    ]
)

reach_class_col = first_existing_column(
    track_reach_df,
    [
        "international_reach_class",
        "country_reach_class",
    ]
)

evidence_class_col = first_existing_column(
    track_reach_df,
    [
        "international_evidence_class",
    ]
)


# ------------------------------------------------------------
# 11. Resolve penetration fields
# ------------------------------------------------------------

penetration_index_col = first_existing_column(
    track_penetration_df,
    [
        "market_penetration_index",
    ]
)

penetration_class_col = first_existing_column(
    track_penetration_df,
    [
        "market_penetration_class",
    ]
)

relative_intensity_col = first_existing_column(
    track_penetration_df,
    [
        "median_relative_stream_intensity",
    ]
)

support_qualified_pct_col = first_existing_column(
    track_penetration_df,
    [
        "support_qualified_market_pct",
    ]
)

above_average_pct_col = first_existing_column(
    track_penetration_df,
    [
        "above_country_average_market_pct",
    ]
)


# ------------------------------------------------------------
# 12. Recover or reconstruct market-penetration index
# ------------------------------------------------------------

if penetration_index_col is None:

    print()
    print(
        "Direct market_penetration_index column not found; "
        "reconstructing Section 8.2 penetration index."
    )

    penetration_work_df = (
        track_penetration_df.copy()
    )

    if relative_intensity_col is None:
        raise RuntimeError(
            "Section 8.5 could not recover or reconstruct "
            "market penetration because "
            "median_relative_stream_intensity is unavailable."
        )

    relative_intensity = pd.to_numeric(
        penetration_work_df[
            relative_intensity_col
        ],
        errors="coerce",
    )

    # Normalize intensity using the 95th percentile
    intensity_upper = float(
        relative_intensity.quantile(0.95)
    )

    if (
        not np.isfinite(intensity_upper)
        or intensity_upper <= 0
    ):
        intensity_upper = 1.0

    penetration_work_df[
        "_penetration_intensity_component"
    ] = (
        relative_intensity
        .clip(
            lower=0,
            upper=intensity_upper,
        )
        /
        intensity_upper
    )

    # Support-qualified market component
    if support_qualified_pct_col is not None:

        penetration_work_df[
            "_penetration_support_component"
        ] = (
            pd.to_numeric(
                penetration_work_df[
                    support_qualified_pct_col
                ],
                errors="coerce",
            )
            .fillna(0)
            .clip(
                lower=0,
                upper=100,
            )
            / 100
        )

    else:

        penetration_work_df[
            "_penetration_support_component"
        ] = 0.0

    # Above-country-average component
    if above_average_pct_col is not None:

        penetration_work_df[
            "_penetration_above_average_component"
        ] = (
            pd.to_numeric(
                penetration_work_df[
                    above_average_pct_col
                ],
                errors="coerce",
            )
            .fillna(0)
            .clip(
                lower=0,
                upper=100,
            )
            / 100
        )

    else:

        penetration_work_df[
            "_penetration_above_average_component"
        ] = 0.0

    penetration_work_df[
        "market_penetration_index"
    ] = (
        (
            0.50
            * penetration_work_df[
                "_penetration_intensity_component"
            ]
        )
        +
        (
            0.25
            * penetration_work_df[
                "_penetration_support_component"
            ]
        )
        +
        (
            0.25
            * penetration_work_df[
                "_penetration_above_average_component"
            ]
        )
    ).clip(
        lower=0,
        upper=1,
    )

    track_penetration_df = (
        penetration_work_df.copy()
    )

    penetration_index_col = (
        "market_penetration_index"
    )


# ------------------------------------------------------------
# 13. Resolve cross-market fields
# ------------------------------------------------------------

consistency_index_col = first_existing_column(
    track_cross_market_df,
    [
        "cross_market_consistency_index",
    ]
)

consistency_class_col = first_existing_column(
    track_cross_market_df,
    [
        "cross_market_consistency_class",
        "stream_consistency_class",
        "consistency_class",
    ]
)

comparison_pct_col = first_existing_column(
    track_cross_market_df,
    [
        "comparison_eligible_market_pct",
        "comparison_safe_market_pct",
    ]
)


# ------------------------------------------------------------
# 14. Resolve concentration fields
# ------------------------------------------------------------

hhi_col = first_existing_column(
    track_concentration_df,
    [
        "stream_concentration_hhi",
    ]
)

effective_markets_col = first_existing_column(
    track_concentration_df,
    [
        "effective_stream_markets",
    ]
)

diversification_col = first_existing_column(
    track_concentration_df,
    [
        "geographic_diversification_index",
    ]
)

concentration_class_col = first_existing_column(
    track_concentration_df,
    [
        "geographic_concentration_class",
        "concentration_class",
    ]
)

dependency_class_col = first_existing_column(
    track_concentration_df,
    [
        "market_dependency_class",
    ]
)


# ------------------------------------------------------------
# 15. Validate essential profile inputs
# ------------------------------------------------------------

essential_inputs = {
    "international reach":
        (
            reach_index_col
            or reach_pct_col
        ),

    "market penetration":
        penetration_index_col,

    "cross-market consistency":
        consistency_index_col,

    "geographic diversification":
        diversification_col,

    "markets reached":
        markets_reached_col,

    "effective stream markets":
        effective_markets_col,
}

missing_inputs = [
    name
    for name, column
    in essential_inputs.items()
    if column is None
]

if missing_inputs:
    raise RuntimeError(
        "Section 8.5 requires the following profile inputs: "
        + ", ".join(missing_inputs)
    )


# ------------------------------------------------------------
# 16. Build canonical track population
# ------------------------------------------------------------

identity_columns = [
    "track_id",
]

if track_name_col is not None:
    identity_columns.append(
        track_name_col
    )

if artists_col is not None:
    identity_columns.append(
        artists_col
    )


track_geographic_profile_df = (
    track_reach_df[
        identity_columns
    ]
    .copy()
)


# ------------------------------------------------------------
# 17. Merge international-reach intelligence
# ------------------------------------------------------------

reach_merge = pd.DataFrame(
    {
        "track_id":
            track_reach_df[
                "track_id"
            ],

        "markets_reached":
            pd.to_numeric(
                track_reach_df[
                    markets_reached_col
                ],
                errors="coerce",
            ),
    }
)


if reach_index_col is not None:

    reach_merge[
        "international_reach_index"
    ] = pd.to_numeric(
        track_reach_df[
            reach_index_col
        ],
        errors="coerce",
    )

else:

    reach_merge[
        "international_reach_index"
    ] = (
        pd.to_numeric(
            track_reach_df[
                reach_pct_col
            ],
            errors="coerce",
        )
        /
        100
    )


if reach_pct_col is not None:

    reach_merge[
        "international_market_reach_pct"
    ] = pd.to_numeric(
        track_reach_df[
            reach_pct_col
        ],
        errors="coerce",
    )

else:

    reach_merge[
        "international_market_reach_pct"
    ] = (
        reach_merge[
            "international_reach_index"
        ]
        * 100
    )


if reach_class_col is not None:

    reach_merge[
        "international_reach_class"
    ] = (
        track_reach_df[
            reach_class_col
        ]
        .astype("string")
    )


if evidence_class_col is not None:

    reach_merge[
        "international_evidence_class"
    ] = (
        track_reach_df[
            evidence_class_col
        ]
        .astype("string")
    )


track_geographic_profile_df = (
    track_geographic_profile_df
    .merge(
        reach_merge,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 18. Merge market-penetration intelligence
# ------------------------------------------------------------

penetration_merge = pd.DataFrame(
    {
        "track_id":
            track_penetration_df[
                "track_id"
            ],

        "market_penetration_index":
            pd.to_numeric(
                track_penetration_df[
                    penetration_index_col
                ],
                errors="coerce",
            ),
    }
)


if penetration_class_col is not None:

    penetration_merge[
        "market_penetration_class"
    ] = (
        track_penetration_df[
            penetration_class_col
        ]
        .astype("string")
    )


if relative_intensity_col is not None:

    penetration_merge[
        "median_relative_stream_intensity"
    ] = pd.to_numeric(
        track_penetration_df[
            relative_intensity_col
        ],
        errors="coerce",
    )


track_geographic_profile_df = (
    track_geographic_profile_df
    .merge(
        penetration_merge,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 19. Merge cross-market performance intelligence
# ------------------------------------------------------------

cross_market_merge = pd.DataFrame(
    {
        "track_id":
            track_cross_market_df[
                "track_id"
            ],

        "cross_market_consistency_index":
            pd.to_numeric(
                track_cross_market_df[
                    consistency_index_col
                ],
                errors="coerce",
            ),
    }
)


if consistency_class_col is not None:

    cross_market_merge[
        "cross_market_consistency_class"
    ] = (
        track_cross_market_df[
            consistency_class_col
        ]
        .astype("string")
    )


if comparison_pct_col is not None:

    cross_market_merge[
        "comparison_eligible_market_pct"
    ] = pd.to_numeric(
        track_cross_market_df[
            comparison_pct_col
        ],
        errors="coerce",
    )

else:

    cross_market_merge[
        "comparison_eligible_market_pct"
    ] = 0.0


track_geographic_profile_df = (
    track_geographic_profile_df
    .merge(
        cross_market_merge,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 20. Merge geographic-concentration intelligence
# ------------------------------------------------------------

concentration_merge = pd.DataFrame(
    {
        "track_id":
            track_concentration_df[
                "track_id"
            ],

        "effective_stream_markets":
            pd.to_numeric(
                track_concentration_df[
                    effective_markets_col
                ],
                errors="coerce",
            ),

        "geographic_diversification_index":
            pd.to_numeric(
                track_concentration_df[
                    diversification_col
                ],
                errors="coerce",
            ),
    }
)


if hhi_col is not None:

    concentration_merge[
        "stream_concentration_hhi"
    ] = pd.to_numeric(
        track_concentration_df[
            hhi_col
        ],
        errors="coerce",
    )


if concentration_class_col is not None:

    concentration_merge[
        "geographic_concentration_class"
    ] = (
        track_concentration_df[
            concentration_class_col
        ]
        .astype("string")
    )


if dependency_class_col is not None:

    concentration_merge[
        "market_dependency_class"
    ] = (
        track_concentration_df[
            dependency_class_col
        ]
        .astype("string")
    )


track_geographic_profile_df = (
    track_geographic_profile_df
    .merge(
        concentration_merge,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 21. Evidence-adjusted cross-market consistency
# ------------------------------------------------------------

track_geographic_profile_df[
    "comparison_support_component"
] = (
    track_geographic_profile_df[
        "comparison_eligible_market_pct"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=100,
    )
    /
    100
)


track_geographic_profile_df[
    "evidence_adjusted_consistency"
] = (
    track_geographic_profile_df[
        "cross_market_consistency_index"
    ]
    *
    (
        0.50
        +
        (
            0.50
            * track_geographic_profile_df[
                "comparison_support_component"
            ]
        )
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 22. Geographic profile index
# ------------------------------------------------------------

track_geographic_profile_df[
    "geographic_profile_index"
] = (
    (
        0.35
        * track_geographic_profile_df[
            "international_reach_index"
        ]
    )
    +
    (
        0.25
        * track_geographic_profile_df[
            "market_penetration_index"
        ]
    )
    +
    (
        0.25
        * track_geographic_profile_df[
            "geographic_diversification_index"
        ]
    )
    +
    (
        0.15
        * track_geographic_profile_df[
            "evidence_adjusted_consistency"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 23. Geographic performance profile classification
# ------------------------------------------------------------

def classify_track_geographic_profile(row):

    markets = row[
        "markets_reached"
    ]

    reach = row[
        "international_reach_index"
    ]

    penetration = row[
        "market_penetration_index"
    ]

    diversification = row[
        "geographic_diversification_index"
    ]

    effective_markets = row[
        "effective_stream_markets"
    ]

    profile_index = row[
        "geographic_profile_index"
    ]

    comparison_support = row[
        "comparison_support_component"
    ]

    # --------------------------------------------------------
    # Globally distributed
    # --------------------------------------------------------

    if (
        markets >= 50
        and
        reach >= 0.65
        and
        diversification >= 0.50
        and
        penetration >= 0.40
        and
        comparison_support >= 0.50
    ):
        return "Globally distributed"

    # --------------------------------------------------------
    # Diversified international
    # --------------------------------------------------------

    if (
        markets >= 25
        and
        effective_markets >= 5
        and
        diversification >= 0.40
    ):
        return "Diversified international"

    # --------------------------------------------------------
    # Internationally established
    # --------------------------------------------------------

    if (
        markets >= 10
        and
        reach >= 0.10
        and
        profile_index >= 0.20
    ):
        return "Internationally established"

    # --------------------------------------------------------
    # Emerging international
    # --------------------------------------------------------

    if (
        markets >= 5
        or
        (
            markets >= 2
            and
            penetration >= 0.50
        )
    ):
        return "Emerging international"

    # --------------------------------------------------------
    # Localised profile
    # --------------------------------------------------------

    return "Locally concentrated"


track_geographic_profile_df[
    "track_geographic_profile"
] = (
    track_geographic_profile_df
    .apply(
        classify_track_geographic_profile,
        axis=1,
    )
)


# ------------------------------------------------------------
# 24. Profile overview
# ------------------------------------------------------------

print()
print("Track geographic performance profile overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_profile_df):,}"
)

print(
    f"Median geographic profile index: "
    f"{track_geographic_profile_df['geographic_profile_index'].median():.4f}"
)

print(
    f"Mean geographic profile index: "
    f"{track_geographic_profile_df['geographic_profile_index'].mean():.4f}"
)

print(
    f"Median international reach index: "
    f"{track_geographic_profile_df['international_reach_index'].median():.4f}"
)

print(
    f"Median market penetration index: "
    f"{track_geographic_profile_df['market_penetration_index'].median():.4f}"
)

print(
    f"Median diversification index: "
    f"{track_geographic_profile_df['geographic_diversification_index'].median():.4f}"
)


# ------------------------------------------------------------
# 25. Profile classification summary
# ------------------------------------------------------------

profile_order = [
    "Locally concentrated",
    "Emerging international",
    "Internationally established",
    "Diversified international",
    "Globally distributed",
]


track_geographic_profile_summary_df = (
    track_geographic_profile_df[
        "track_geographic_profile"
    ]
    .value_counts()
    .reindex(
        profile_order,
        fill_value=0,
    )
    .rename_axis(
        "Track Geographic Profile"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_profile_summary_df[
    "Share (%)"
] = (
    track_geographic_profile_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_profile_df)
    * 100
)


print()
print("Track geographic profile classification")
print("=" * 100)

display(
    track_geographic_profile_summary_df
)


# ------------------------------------------------------------
# 26. Strongest geographic profiles
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_reached",
    "international_reach_index",
    "market_penetration_index",
    "effective_stream_markets",
    "geographic_diversification_index",
    "cross_market_consistency_index",
    "comparison_eligible_market_pct",
    "geographic_profile_index",
    "track_geographic_profile",
]


print()
print(
    "Tracks with strongest observed geographic performance profiles"
)
print("=" * 100)


strongest_track_profiles_df = (
    track_geographic_profile_df
    .sort_values(
        [
            "geographic_profile_index",
            "international_reach_index",
            "geographic_diversification_index",
            "market_penetration_index",
        ],
        ascending=[
            False,
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)


display(
    strongest_track_profiles_df
)


# ------------------------------------------------------------
# 27. Diversified international tracks
# ------------------------------------------------------------

print()
print(
    "Tracks with strongest diversified international profiles"
)
print("=" * 100)


diversified_track_profiles_df = (
    track_geographic_profile_df
    .loc[
        track_geographic_profile_df[
            "track_geographic_profile"
        ]
        .isin(
            [
                "Diversified international",
                "Globally distributed",
            ]
        )
    ]
    .sort_values(
        [
            "geographic_profile_index",
            "markets_reached",
            "effective_stream_markets",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)


display(
    diversified_track_profiles_df
)


# ------------------------------------------------------------
# 28. Diagnostics
# ------------------------------------------------------------

track_profile_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Locally concentrated tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "track_geographic_profile"
                        ]
                        == "Locally concentrated"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Emerging international tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "track_geographic_profile"
                        ]
                        == "Emerging international"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Internationally established tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "track_geographic_profile"
                        ]
                        == "Internationally established"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Diversified international tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "track_geographic_profile"
                        ]
                        == "Diversified international"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Globally distributed tracks",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "track_geographic_profile"
                        ]
                        == "Globally distributed"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with profile index >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_profile_df[
                            "geographic_profile_index"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)


print()
print("Track geographic profile diagnostics")
print("=" * 100)

display(
    track_profile_diagnostics_df
)


# ------------------------------------------------------------
# 29. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_profile_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_profile_df[
        "track_id"
    ].nunique()
    == 110_198
)


core_profile_columns = [
    "international_reach_index",
    "market_penetration_index",
    "cross_market_consistency_index",
    "geographic_diversification_index",
    "geographic_profile_index",
]


profile_input_complete = bool(
    track_geographic_profile_df[
        core_profile_columns
    ]
    .notna()
    .all()
    .all()
)


profile_numeric_valid = bool(
    np.isfinite(
        track_geographic_profile_df[
            core_profile_columns
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


reach_valid = bool(
    track_geographic_profile_df[
        "international_reach_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)


penetration_valid = bool(
    track_geographic_profile_df[
        "market_penetration_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)


diversification_valid = bool(
    track_geographic_profile_df[
        "geographic_diversification_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)


profile_index_valid = bool(
    track_geographic_profile_df[
        "geographic_profile_index"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)


classification_complete = bool(
    track_geographic_profile_df[
        "track_geographic_profile"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_profile_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_profile_df
    )
)


section_8_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 8.4 completion",

            "Requirement":
                "Track Geographic Concentration must "
                "already be complete.",

            "Observed Evidence":
                f"Section 8.4 completion status: "
                f"{section_8_4_status}",

            "Passed":
                section_8_4_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_profile_df):,} "
                "track profile records assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive exactly one "
                "geographic profile.",

            "Observed Evidence":
                f"{track_geographic_profile_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Profile input completeness",

            "Requirement":
                "Reach, penetration, consistency and "
                "diversification inputs must be complete.",

            "Observed Evidence":
                "Four geographic intelligence dimensions "
                "assessed for all tracks",

            "Passed":
                profile_input_complete,
        },
        {
            "Validation Area":
                "Profile numeric validity",

            "Requirement":
                "Profile inputs and outputs must remain finite.",

            "Observed Evidence":
                f"{len(track_geographic_profile_df):,} "
                "profile records assessed",

            "Passed":
                profile_numeric_valid,
        },
        {
            "Validation Area":
                "Reach-index bounds",

            "Requirement":
                "International reach indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Reach indices assessed for all tracks",

            "Passed":
                reach_valid,
        },
        {
            "Validation Area":
                "Penetration-index bounds",

            "Requirement":
                "Market penetration indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Penetration indices assessed for all tracks",

            "Passed":
                penetration_valid,
        },
        {
            "Validation Area":
                "Diversification-index bounds",

            "Requirement":
                "Diversification indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Diversification indices assessed for all tracks",

            "Passed":
                diversification_valid,
        },
        {
            "Validation Area":
                "Geographic-profile index validity",

            "Requirement":
                "Geographic profile indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                f"{len(track_geographic_profile_df):,} "
                "profile indices assessed",

            "Passed":
                profile_index_valid,
        },
        {
            "Validation Area":
                "Profile classification completeness",

            "Requirement":
                "Every track must receive a geographic profile.",

            "Observed Evidence":
                f"{track_geographic_profile_df['track_geographic_profile'].notna().sum():,} "
                "profile classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Profile population reconciliation",

            "Requirement":
                "Profile classifications must reconcile "
                "with the validated track population.",

            "Observed Evidence":
                f"{int(track_geographic_profile_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_profile_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Track geographic performance profile validation")
print("=" * 100)

display(
    section_8_5_validation_df
)


# ------------------------------------------------------------
# 30. Final completion status
# ------------------------------------------------------------

section_8_5_complete = bool(
    section_8_5_validation_df[
        "Passed"
    ].all()
)


if not section_8_5_complete:

    failed_checks = (
        section_8_5_validation_df
        .loc[
            ~section_8_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 8.5 track geographic performance "
        "profile validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


section_8_complete = bool(
    section_8_5_complete
)


print()
print(
    "All Section 8.5 track geographic performance "
    "profile validation checks passed."
)

print(
    f"Section 8.5 completion status: "
    f"{section_8_5_complete}"
)

print(
    f"Section 8 overall completion status: "
    f"{section_8_complete}"
)

print(
    f"Track geographic performance profiles prepared "
    f"for {len(track_geographic_profile_df):,} tracks."
)

print(
    "These profiles combine international reach, "
    "market penetration, cross-market consistency "
    "and geographic diversification intelligence."
)

print(
    "They remain descriptive intelligence outputs "
    "and are not yet final PMIP track rankings."
)
```

Importantly, the geographic profile index does not rely on a single indicator. It combines four complementary dimensions:

- **International reach** measures how widely a track appears across country markets.
- **Market penetration** measures the depth of performance within the markets reached.
- **Cross-market consistency** measures how stable performance is across different markets, while accounting for available comparison evidence.
- **Geographic diversification** measures how evenly streaming activity is distributed rather than being concentrated in one dominant country.

This combined structure prevents broad reach alone from being interpreted as strong geographic performance. For example, a track may appear in many countries but still depend heavily on one major market, while another track may have a smaller overall reach but demonstrate stronger penetration and more balanced international distribution.

## 9. Geographic Expansion and Market Movement

Section 9 examines how artist and track performance changes geographically over time.

The focus moves from describing where performance currently exists to understanding **how geographic presence evolves** across reporting periods. This includes identifying entry into new country markets, expansion into a broader geographic footprint, contraction from previously observed markets, and changes in relative cross-market momentum.

The section uses the validated geographic growth, decline, reach, penetration, concentration, and temporal-alignment intelligence produced in earlier sections.

All expansion and contraction measures remain observational. They describe changes visible in the available PMIP data and do not establish why those changes occurred.


### 9.1 New-Market Entry

#### Purpose

Identify when validated artists and tracks begin appearing in country markets where they were not previously observed.

This subsection uses the chronological geographic observation history to detect first market appearances and distinguish genuine new-market entry from differences caused by staggered country coverage or incomplete historical reporting.

The analysis measures the number and share of newly entered markets, identifies the timing of those entries, and preserves the temporal safeguards developed earlier so that later-entry countries are not incorrectly treated as evidence of recent geographic expansion.

The resulting indicators describe observed market-entry behaviour and provide the foundation for the broader geographic expansion analysis in Section 9.2.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP new-market entry")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_8_status = globals().get(
    "section_8_complete",
    False
)

print(
    f"Section 8 overall completion status: "
    f"{section_8_status}"
)

if not section_8_status:
    raise RuntimeError(
        "Section 9.1 requires Section 8 "
        "Track Geographic Intelligence to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated geographic analytical population
# ------------------------------------------------------------

def recover_geographic_source():
    preferred_names = [
        "geographic_analysis_df",
        "country_geo_df",
        "coverage_source_df",
        "source_df",
        "canonical_geo_df",
    ]

    required_fields = {
        "date",
        "country",
        "track_id",
        "streams",
    }

    for name in preferred_names:
        if name not in globals():
            continue

        candidate = globals()[name]

        if not isinstance(candidate, pd.DataFrame):
            continue

        if not required_fields.issubset(candidate.columns):
            continue

        if len(candidate) < 5_000_000:
            continue

        return candidate.copy(), name

    for name, candidate in globals().copy().items():
        if not isinstance(candidate, pd.DataFrame):
            continue

        if not required_fields.issubset(candidate.columns):
            continue

        if len(candidate) < 5_000_000:
            continue

        return candidate.copy(), name

    return None, None


geo_df, geo_source_name = recover_geographic_source()

if geo_df is None:
    raise RuntimeError(
        "Section 9.1 could not recover the validated "
        "geographic analytical population."
    )


print()
print("Validated geographic source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{geo_source_name}"
)

print(
    f"Observations: "
    f"{len(geo_df):,}"
)

print(
    f"Country markets: "
    f"{geo_df['country'].nunique():,}"
)

print(
    f"Tracks: "
    f"{geo_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Resolve identity and safeguard columns
# ------------------------------------------------------------

def first_existing_column(df, candidates):
    for column in candidates:
        if column in df.columns:
            return column
    return None


track_name_col = first_existing_column(
    geo_df,
    [
        "track_name",
        "name",
    ]
)

artists_col = first_existing_column(
    geo_df,
    [
        "artists_raw",
        "artists",
    ]
)

primary_artist_col = first_existing_column(
    geo_df,
    [
        "primary_artist",
    ]
)

trend_allowed_col = first_existing_column(
    geo_df,
    [
        "trend_interpretation_allowed",
    ]
)

historical_allowed_col = first_existing_column(
    geo_df,
    [
        "historical_comparison_allowed",
    ]
)

direct_caution_col = first_existing_column(
    geo_df,
    [
        "direct_comparison_caution",
    ]
)


# ------------------------------------------------------------
# 4. Standardise temporal fields
# ------------------------------------------------------------

geo_df["date"] = pd.to_datetime(
    geo_df["date"],
    errors="coerce",
)

if geo_df["date"].isna().any():
    raise RuntimeError(
        "Section 9.1 found invalid reporting dates."
    )


# ------------------------------------------------------------
# 5. Prepare track-country first-observation table
# ------------------------------------------------------------

track_country_group_cols = [
    "track_id",
    "country",
]

track_country_agg = {
    "date": [
        "min",
        "max",
        "nunique",
    ],
    "streams": [
        "sum",
        "median",
    ],
}

if track_name_col is not None:
    track_country_agg[
        track_name_col
    ] = "first"

if artists_col is not None:
    track_country_agg[
        artists_col
    ] = "first"


track_country_entry_df = (
    geo_df
    .groupby(
        track_country_group_cols,
        as_index=False,
    )
    .agg(
        track_country_agg
    )
)


# Flatten multi-index columns
track_country_entry_df.columns = [
    "_".join(
        [
            str(part)
            for part in column
            if str(part) != ""
        ]
    ).strip("_")
    if isinstance(column, tuple)
    else column
    for column in track_country_entry_df.columns
]


# Rename canonical columns
rename_map = {
    "track_id_":
        "track_id",

    "country_":
        "country",

    "date_min":
        "first_observed_date",

    "date_max":
        "last_observed_date",

    "date_nunique":
        "reporting_dates",

    "streams_sum":
        "total_streams",

    "streams_median":
        "median_streams",
}

track_country_entry_df = (
    track_country_entry_df.rename(
        columns=rename_map
    )
)


# ------------------------------------------------------------
# 6. Attach country temporal-alignment safeguards
# ------------------------------------------------------------

country_temporal_df = None

for candidate_name in [
    "country_temporal_alignment_df",
    "temporal_country_alignment_df",
    "country_temporal_df",
]:
    if candidate_name in globals():
        candidate = globals()[candidate_name]

        if (
            isinstance(candidate, pd.DataFrame)
            and
            "country" in candidate.columns
            and
            len(candidate) == 76
        ):
            country_temporal_df = candidate.copy()
            break


if country_temporal_df is not None:

    safeguard_columns = [
        column
        for column in [
            "country",
            "first_reporting_date",
            "last_reporting_date",
            "date_coverage_pct",
            "active_span_coverage_pct",
            "temporal_alignment_class",
            "overlap_period_comparison_required",
            "direct_comparison_caution",
        ]
        if column in country_temporal_df.columns
    ]

    track_country_entry_df = (
        track_country_entry_df
        .merge(
            country_temporal_df[
                safeguard_columns
            ],
            on="country",
            how="left",
            validate="many_to_one",
        )
    )


# ------------------------------------------------------------
# 7. Determine first observed market entry per track
# ------------------------------------------------------------

track_country_entry_df = (
    track_country_entry_df
    .sort_values(
        [
            "track_id",
            "first_observed_date",
            "country",
        ]
    )
    .reset_index(drop=True)
)


track_country_entry_df[
    "track_first_market_date"
] = (
    track_country_entry_df
    .groupby(
        "track_id"
    )[
        "first_observed_date"
    ]
    .transform("min")
)


track_country_entry_df[
    "days_since_first_market"
] = (
    track_country_entry_df[
        "first_observed_date"
    ]
    -
    track_country_entry_df[
        "track_first_market_date"
    ]
).dt.days


track_country_entry_df[
    "is_initial_market"
] = (
    track_country_entry_df[
        "days_since_first_market"
    ]
    == 0
)


track_country_entry_df[
    "is_later_market_entry"
] = (
    track_country_entry_df[
        "days_since_first_market"
    ]
    > 0
)


# ------------------------------------------------------------
# 8. Build temporal-safeguarded market-entry flag
# ------------------------------------------------------------

track_country_entry_df[
    "market_entry_temporally_supported"
] = True


if "temporal_alignment_class" in track_country_entry_df.columns:

    limited_classes = [
        "Limited temporal alignment",
        "Partial temporal alignment",
    ]

    track_country_entry_df[
        "market_entry_temporally_supported"
    ] = ~(
        track_country_entry_df[
            "temporal_alignment_class"
        ]
        .isin(
            limited_classes
        )
        &
        track_country_entry_df[
            "is_later_market_entry"
        ]
    )


if "direct_comparison_caution" in track_country_entry_df.columns:

    track_country_entry_df[
        "market_entry_temporally_supported"
    ] = (
        track_country_entry_df[
            "market_entry_temporally_supported"
        ]
        &
        ~(
            track_country_entry_df[
                "direct_comparison_caution"
            ]
            .fillna(False)
            &
            track_country_entry_df[
                "is_later_market_entry"
            ]
        )
    )


track_country_entry_df[
    "validated_new_market_entry"
] = (
    track_country_entry_df[
        "is_later_market_entry"
    ]
    &
    track_country_entry_df[
        "market_entry_temporally_supported"
    ]
)


# ------------------------------------------------------------
# 9. Track-level market-entry intelligence
# ------------------------------------------------------------

track_entry_summary_df = (
    track_country_entry_df
    .groupby(
        "track_id",
        as_index=False,
    )
    .agg(
        markets_observed=(
            "country",
            "nunique",
        ),

        initial_markets=(
            "is_initial_market",
            "sum",
        ),

        later_market_entries=(
            "is_later_market_entry",
            "sum",
        ),

        validated_new_market_entries=(
            "validated_new_market_entry",
            "sum",
        ),

        first_market_date=(
            "first_observed_date",
            "min",
        ),

        latest_market_entry_date=(
            "first_observed_date",
            "max",
        ),
    )
)


track_entry_summary_df[
    "new_market_entry_share_pct"
] = np.where(
    track_entry_summary_df[
        "markets_observed"
    ] > 0,

    (
        track_entry_summary_df[
            "validated_new_market_entries"
        ]
        /
        track_entry_summary_df[
            "markets_observed"
        ]
        * 100
    ),

    0.0,
)


track_entry_summary_df[
    "market_expansion_span_days"
] = (
    track_entry_summary_df[
        "latest_market_entry_date"
    ]
    -
    track_entry_summary_df[
        "first_market_date"
    ]
).dt.days


# ------------------------------------------------------------
# 10. Restore track identity fields
# ------------------------------------------------------------

identity_cols = [
    "track_id",
]

if track_name_col is not None:
    identity_cols.append(
        track_name_col
    )

if artists_col is not None:
    identity_cols.append(
        artists_col
    )


track_identity_df = (
    geo_df[
        identity_cols
    ]
    .drop_duplicates(
        subset=[
            "track_id"
        ]
    )
)


track_entry_summary_df = (
    track_identity_df
    .merge(
        track_entry_summary_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 11. New-market entry classification
# ------------------------------------------------------------

def classify_new_market_entry(row):

    entries = row[
        "validated_new_market_entries"
    ]

    markets = row[
        "markets_observed"
    ]

    if entries == 0:
        return "No validated new-market entry"

    if entries <= 2:
        return "Limited new-market entry"

    if entries <= 9:
        return "Moderate new-market entry"

    if entries <= 24:
        return "Broad new-market entry"

    return "Very broad new-market entry"


track_entry_summary_df[
    "new_market_entry_class"
] = (
    track_entry_summary_df
    .apply(
        classify_new_market_entry,
        axis=1,
    )
)


# ------------------------------------------------------------
# 12. Overview
# ------------------------------------------------------------

print()
print("Track new-market entry overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_entry_summary_df):,}"
)

print(
    f"Track-country histories assessed: "
    f"{len(track_country_entry_df):,}"
)

print(
    f"Tracks with at least one later observed market: "
    f"{(track_entry_summary_df['later_market_entries'] > 0).sum():,}"
)

print(
    f"Tracks with at least one validated new-market entry: "
    f"{(track_entry_summary_df['validated_new_market_entries'] > 0).sum():,}"
)

print(
    f"Median validated new-market entries per track: "
    f"{track_entry_summary_df['validated_new_market_entries'].median():.1f}"
)

print(
    f"Maximum validated new-market entries for one track: "
    f"{track_entry_summary_df['validated_new_market_entries'].max():,}"
)


# ------------------------------------------------------------
# 13. Classification summary
# ------------------------------------------------------------

entry_class_order = [
    "No validated new-market entry",
    "Limited new-market entry",
    "Moderate new-market entry",
    "Broad new-market entry",
    "Very broad new-market entry",
]


track_entry_class_summary_df = (
    track_entry_summary_df[
        "new_market_entry_class"
    ]
    .value_counts()
    .reindex(
        entry_class_order,
        fill_value=0,
    )
    .rename_axis(
        "New-Market Entry Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_entry_class_summary_df[
    "Share (%)"
] = (
    track_entry_class_summary_df[
        "Tracks"
    ]
    /
    len(track_entry_summary_df)
    * 100
)


print()
print("Track new-market entry classification")
print("=" * 100)

display(
    track_entry_class_summary_df
)


# ------------------------------------------------------------
# 14. Tracks with broadest validated new-market entry
# ------------------------------------------------------------

display_columns = [
    "track_id",
]

if track_name_col is not None:
    display_columns.append(
        track_name_col
    )

if artists_col is not None:
    display_columns.append(
        artists_col
    )

display_columns += [
    "markets_observed",
    "initial_markets",
    "later_market_entries",
    "validated_new_market_entries",
    "new_market_entry_share_pct",
    "market_expansion_span_days",
    "new_market_entry_class",
]


print()
print(
    "Tracks with broadest validated new-market entry"
)
print("=" * 100)

broadest_track_entry_df = (
    track_entry_summary_df
    .sort_values(
        [
            "validated_new_market_entries",
            "new_market_entry_share_pct",
            "markets_observed",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    broadest_track_entry_df
)


# ------------------------------------------------------------
# 15. Country-level entry diagnostics
# ------------------------------------------------------------

country_entry_diagnostics_df = (
    track_country_entry_df
    .groupby(
        "country",
        as_index=False,
    )
    .agg(
        observed_track_histories=(
            "track_id",
            "count",
        ),

        later_market_entries=(
            "is_later_market_entry",
            "sum",
        ),

        validated_new_market_entries=(
            "validated_new_market_entry",
            "sum",
        ),
    )
)


country_entry_diagnostics_df[
    "validated_entry_share_pct"
] = np.where(
    country_entry_diagnostics_df[
        "later_market_entries"
    ] > 0,

    (
        country_entry_diagnostics_df[
            "validated_new_market_entries"
        ]
        /
        country_entry_diagnostics_df[
            "later_market_entries"
        ]
        * 100
    ),

    0.0,
)


print()
print("Country new-market entry diagnostics")
print("=" * 100)

display(
    country_entry_diagnostics_df
    .sort_values(
        [
            "validated_new_market_entries",
            "later_market_entries",
        ],
        ascending=False,
    )
    .head(25)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 16. Validation
# ------------------------------------------------------------

population_valid = bool(
    track_entry_summary_df[
        "track_id"
    ].nunique()
    == 110_198
)

identity_valid = bool(
    len(
        track_entry_summary_df
    )
    ==
    track_entry_summary_df[
        "track_id"
    ].nunique()
)

market_counts_valid = bool(
    (
        track_entry_summary_df[
            "markets_observed"
        ]
        >= 1
    ).all()
)

entry_counts_valid = bool(
    (
        track_entry_summary_df[
            "validated_new_market_entries"
        ]
        >= 0
    ).all()
    and
    (
        track_entry_summary_df[
            "validated_new_market_entries"
        ]
        <=
        track_entry_summary_df[
            "later_market_entries"
        ]
    ).all()
)

percentage_valid = bool(
    track_entry_summary_df[
        "new_market_entry_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)

chronology_valid = bool(
    (
        track_entry_summary_df[
            "latest_market_entry_date"
        ]
        >=
        track_entry_summary_df[
            "first_market_date"
        ]
    ).all()
)

classification_complete = bool(
    track_entry_summary_df[
        "new_market_entry_class"
    ]
    .notna()
    .all()
)

population_reconciled = bool(
    int(
        track_entry_class_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_entry_summary_df
    )
)


section_9_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 8 completion",

            "Requirement":
                "Track Geographic Intelligence must "
                "already be complete.",

            "Observed Evidence":
                f"Section 8 overall completion status: "
                f"{section_8_status}",

            "Passed":
                section_8_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be represented.",

            "Observed Evidence":
                f"{track_entry_summary_df['track_id'].nunique():,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one market-entry summary.",

            "Observed Evidence":
                f"{len(track_entry_summary_df):,} "
                "summary records for "
                f"{track_entry_summary_df['track_id'].nunique():,} tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Observed-market validity",

            "Requirement":
                "Each track must have at least one observed market.",

            "Observed Evidence":
                f"Observed market range: "
                f"{track_entry_summary_df['markets_observed'].min()} "
                f"to "
                f"{track_entry_summary_df['markets_observed'].max()}",

            "Passed":
                market_counts_valid,
        },
        {
            "Validation Area":
                "Validated entry-count validity",

            "Requirement":
                "Validated market entries must remain "
                "between zero and total later entries.",

            "Observed Evidence":
                "Validated market-entry counts reconciled "
                "with later observed market entries",

            "Passed":
                entry_counts_valid,
        },
        {
            "Validation Area":
                "Entry-percentage bounds",

            "Requirement":
                "New-market entry shares must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Entry-share percentages assessed "
                "for all tracks",

            "Passed":
                percentage_valid,
        },
        {
            "Validation Area":
                "Chronological validity",

            "Requirement":
                "Latest market-entry dates must not occur "
                "before first market dates.",

            "Observed Evidence":
                f"{len(track_entry_summary_df):,} "
                "track histories assessed",

            "Passed":
                chronology_valid,
        },
        {
            "Validation Area":
                "Entry classification completeness",

            "Requirement":
                "Every track must receive a new-market "
                "entry classification.",

            "Observed Evidence":
                f"{track_entry_summary_df['new_market_entry_class'].notna().sum():,} "
                "entry classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Entry population reconciliation",

            "Requirement":
                "New-market entry classifications must "
                "reconcile with the full track population.",

            "Observed Evidence":
                f"{int(track_entry_class_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_entry_summary_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Track new-market entry validation")
print("=" * 100)

display(
    section_9_1_validation_df
)


# ------------------------------------------------------------
# 17. Final completion status
# ------------------------------------------------------------

section_9_1_complete = bool(
    section_9_1_validation_df[
        "Passed"
    ].all()
)


if not section_9_1_complete:

    failed_checks = (
        section_9_1_validation_df
        .loc[
            ~section_9_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 9.1 new-market entry validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 9.1 new-market entry validation checks passed."
)

print(
    f"Section 9.1 completion status: "
    f"{section_9_1_complete}"
)

print(
    f"Track new-market entry intelligence prepared "
    f"for {len(track_entry_summary_df):,} tracks."
)

print(
    "New-market entry reflects observed chronological "
    "market appearance and is safeguarded against "
    "known temporal coverage limitations."
)

print(
    "The workflow is ready for "
    "Section 9.2 Geographic Expansion."
)
```

#### Interpretation

Section 9.1 successfully identifies and validates **new-market entry behaviour** for the full population of **110,198 tracks**.

The analysis is based on **353,282 track-country histories** across **76 country markets**, using the validated geographic analytical population of **5,338,680 observations**.

A total of **14,256 tracks** were observed in at least one country market after their initial market appearance. After applying the temporal safeguards, **12,564 tracks** retained at least one **validated new-market entry**. This means some apparent later market appearances were correctly excluded because they could have been caused by incomplete historical coverage or staggered country reporting rather than genuine geographic expansion.

The majority of tracks show no validated geographic entry beyond their initial observed markets. **97,634 tracks (88.60%)** are classified as having **No validated new-market entry**.

Among the tracks showing geographic movement:

- **6,464 tracks (5.87%)** have **Limited new-market entry**
- **3,353 tracks (3.04%)** have **Moderate new-market entry**
- **1,840 tracks (1.67%)** have **Broad new-market entry**
- **907 tracks (0.82%)** have **Very broad new-market entry**

The median number of validated new-market entries is **0**, which is consistent with the earlier geographic intelligence showing that most tracks in the dataset are locally concentrated or observed in only a small number of markets.

However, the maximum of **53 validated new-market entries for a single track** shows that a small subset of tracks experienced substantial geographic expansion across the available country markets.

The distinction between a **later observed market** and a **validated new-market entry** is important. A later appearance is not automatically interpreted as genuine expansion. Section 9.1 checks the temporal reliability of each market before treating that appearance as meaningful geographic movement.

This safeguard reduces the risk of incorrectly interpreting a country that simply entered the dataset later as a new market reached by the track.

All Section 9.1 validation checks passed. The complete **110,198-track population** was preserved, track identities remained unique, observed-market and entry counts remained valid, entry percentages stayed within the expected range, chronological ordering was preserved, and every track received exactly one new-market-entry classification.

Section 9.1 therefore establishes a validated chronological foundation for analysing geographic movement. The resulting indicators show **where and how strongly tracks entered additional markets after their initial geographic presence**, while accounting for known differences in historical market coverage.

The workflow is ready for **Section 9.2 Geographic Expansion**.

### 9.2 Geographic Expansion

#### Purpose

Measure how strongly each track expands from its initial observed market presence into additional validated country markets over time.

This subsection builds on the new-market entry intelligence from Section 9.1 by moving beyond simple entry counts and creating broader expansion indicators based on the number of validated new markets reached, the share of total observed markets entered later, the time taken for that expansion, and the breadth of the resulting geographic footprint.

The analysis distinguishes tracks with little or no expansion from tracks showing moderate, broad, or very broad geographic growth.

Temporal safeguards from earlier sections remain active so that apparent expansion caused by staggered market coverage is not treated as genuine geographic growth.

The resulting measures describe observed geographic expansion only and do not establish the causes behind that expansion.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP geographic expansion")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_9_1_status = globals().get(
    "section_9_1_complete",
    False
)

print(
    f"Section 9.1 completion status: "
    f"{section_9_1_status}"
)

if not section_9_1_status:
    raise RuntimeError(
        "Section 9.2 requires Section 9.1 "
        "New-Market Entry to be complete."
    )


# ------------------------------------------------------------
# 2. Recover Section 9.1 track-entry intelligence
# ------------------------------------------------------------

candidate_names = [
    "track_entry_summary_df",
    "track_new_market_entry_df",
    "track_market_entry_summary_df",
]

track_expansion_source_df = None
track_expansion_source_name = None

required_columns = {
    "track_id",
    "markets_observed",
    "validated_new_market_entries",
    "new_market_entry_share_pct",
    "market_expansion_span_days",
}

for name in candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_expansion_source_df = (
        candidate.copy()
    )

    track_expansion_source_name = name

    break


if track_expansion_source_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if len(candidate) != 110_198:
            continue

        if not required_columns.issubset(
            candidate.columns
        ):
            continue

        track_expansion_source_df = (
            candidate.copy()
        )

        track_expansion_source_name = name

        break


if track_expansion_source_df is None:
    raise RuntimeError(
        "Section 9.2 could not recover the validated "
        "Section 9.1 track new-market entry intelligence."
    )


print()
print("Validated geographic expansion source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_expansion_source_name}"
)

print(
    f"Tracks: "
    f"{len(track_expansion_source_df):,}"
)


# ------------------------------------------------------------
# 3. Prepare geographic expansion working table
# ------------------------------------------------------------

track_geographic_expansion_df = (
    track_expansion_source_df.copy()
)


# ------------------------------------------------------------
# 4. Ensure numeric fields
# ------------------------------------------------------------

numeric_columns = [
    "markets_observed",
    "initial_markets",
    "later_market_entries",
    "validated_new_market_entries",
    "new_market_entry_share_pct",
    "market_expansion_span_days",
]

for column in numeric_columns:

    if column in track_geographic_expansion_df.columns:

        track_geographic_expansion_df[
            column
        ] = pd.to_numeric(
            track_geographic_expansion_df[
                column
            ],
            errors="coerce",
        )


# ------------------------------------------------------------
# 5. Build expansion breadth indicators
# ------------------------------------------------------------

track_geographic_expansion_df[
    "validated_expansion_markets"
] = (
    track_geographic_expansion_df[
        "validated_new_market_entries"
    ]
    .fillna(0)
)


track_geographic_expansion_df[
    "expansion_market_share_pct"
] = (
    track_geographic_expansion_df[
        "new_market_entry_share_pct"
    ]
    .fillna(0)
    .clip(
        lower=0,
        upper=100,
    )
)


# ------------------------------------------------------------
# 6. Build expansion intensity
# ------------------------------------------------------------

max_market_count = float(
    track_geographic_expansion_df[
        "markets_observed"
    ].max()
)

if (
    not np.isfinite(max_market_count)
    or max_market_count <= 1
):
    max_market_count = 73.0


track_geographic_expansion_df[
    "expansion_breadth_index"
] = (
    track_geographic_expansion_df[
        "validated_expansion_markets"
    ]
    /
    max(
        max_market_count - 1,
        1,
    )
).clip(
    lower=0,
    upper=1,
)


track_geographic_expansion_df[
    "expansion_share_index"
] = (
    track_geographic_expansion_df[
        "expansion_market_share_pct"
    ]
    /
    100
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 7. Build expansion speed indicator
# ------------------------------------------------------------

positive_spans = (
    track_geographic_expansion_df
    .loc[
        track_geographic_expansion_df[
            "market_expansion_span_days"
        ] > 0,
        "market_expansion_span_days",
    ]
)

if len(positive_spans) > 0:

    span_reference = float(
        positive_spans.quantile(0.95)
    )

else:

    span_reference = 1.0


if (
    not np.isfinite(span_reference)
    or span_reference <= 0
):
    span_reference = 1.0


track_geographic_expansion_df[
    "expansion_speed_index"
] = np.where(
    track_geographic_expansion_df[
        "validated_expansion_markets"
    ] > 0,

    (
        1
        -
        (
            track_geographic_expansion_df[
                "market_expansion_span_days"
            ]
            .fillna(span_reference)
            .clip(
                lower=0,
                upper=span_reference,
            )
            /
            span_reference
        )
    ),

    0.0,
)


track_geographic_expansion_df[
    "expansion_speed_index"
] = (
    track_geographic_expansion_df[
        "expansion_speed_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 8. Build overall geographic expansion index
# ------------------------------------------------------------

track_geographic_expansion_df[
    "geographic_expansion_index"
] = (
    (
        0.50
        * track_geographic_expansion_df[
            "expansion_breadth_index"
        ]
    )
    +
    (
        0.35
        * track_geographic_expansion_df[
            "expansion_share_index"
        ]
    )
    +
    (
        0.15
        * track_geographic_expansion_df[
            "expansion_speed_index"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Expansion classification
# ------------------------------------------------------------

def classify_geographic_expansion(row):

    entries = row[
        "validated_expansion_markets"
    ]

    expansion_index = row[
        "geographic_expansion_index"
    ]

    share = row[
        "expansion_market_share_pct"
    ]

    if entries == 0:
        return "No validated expansion"

    if (
        entries >= 25
        and
        expansion_index >= 0.50
    ):
        return "Very broad expansion"

    if (
        entries >= 10
        or
        expansion_index >= 0.30
    ):
        return "Broad expansion"

    if (
        entries >= 3
        or
        expansion_index >= 0.15
    ):
        return "Moderate expansion"

    return "Limited expansion"


track_geographic_expansion_df[
    "geographic_expansion_class"
] = (
    track_geographic_expansion_df
    .apply(
        classify_geographic_expansion,
        axis=1,
    )
)


# ------------------------------------------------------------
# 10. Geographic expansion overview
# ------------------------------------------------------------

print()
print("Track geographic expansion overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_expansion_df):,}"
)

print(
    f"Tracks with validated geographic expansion: "
    f"{(track_geographic_expansion_df['validated_expansion_markets'] > 0).sum():,}"
)

print(
    f"Median validated expansion markets: "
    f"{track_geographic_expansion_df['validated_expansion_markets'].median():.1f}"
)

print(
    f"Maximum validated expansion markets: "
    f"{track_geographic_expansion_df['validated_expansion_markets'].max():.0f}"
)

print(
    f"Median expansion market share: "
    f"{track_geographic_expansion_df['expansion_market_share_pct'].median():.2f}%"
)

print(
    f"Median geographic expansion index: "
    f"{track_geographic_expansion_df['geographic_expansion_index'].median():.4f}"
)

print(
    f"Mean geographic expansion index: "
    f"{track_geographic_expansion_df['geographic_expansion_index'].mean():.4f}"
)


# ------------------------------------------------------------
# 11. Expansion classification summary
# ------------------------------------------------------------

expansion_class_order = [
    "No validated expansion",
    "Limited expansion",
    "Moderate expansion",
    "Broad expansion",
    "Very broad expansion",
]


track_geographic_expansion_summary_df = (
    track_geographic_expansion_df[
        "geographic_expansion_class"
    ]
    .value_counts()
    .reindex(
        expansion_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Expansion Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_expansion_summary_df[
    "Share (%)"
] = (
    track_geographic_expansion_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_expansion_df)
    * 100
)


print()
print("Track geographic expansion classification")
print("=" * 100)

display(
    track_geographic_expansion_summary_df
)


# ------------------------------------------------------------
# 12. Tracks with strongest geographic expansion
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "markets_observed",
        "validated_expansion_markets",
        "expansion_market_share_pct",
        "market_expansion_span_days",
        "expansion_breadth_index",
        "expansion_speed_index",
        "geographic_expansion_index",
        "geographic_expansion_class",
    ]
    if column in track_geographic_expansion_df.columns
]


print()
print("Tracks with strongest validated geographic expansion")
print("=" * 100)

strongest_geographic_expansion_df = (
    track_geographic_expansion_df
    .sort_values(
        [
            "geographic_expansion_index",
            "validated_expansion_markets",
            "expansion_market_share_pct",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_geographic_expansion_df
)


# ------------------------------------------------------------
# 13. Expansion diagnostics
# ------------------------------------------------------------

track_geographic_expansion_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with no validated expansion",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "validated_expansion_markets"
                        ]
                        == 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 1 validated expansion market",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "validated_expansion_markets"
                        ]
                        >= 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 5 validated expansion markets",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "validated_expansion_markets"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 10 validated expansion markets",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "validated_expansion_markets"
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 25 validated expansion markets",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "validated_expansion_markets"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with expansion index >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_df[
                            "geographic_expansion_index"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)


print()
print("Track geographic expansion diagnostics")
print("=" * 100)

display(
    track_geographic_expansion_diagnostics_df
)


# ------------------------------------------------------------
# 14. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_expansion_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_expansion_df[
        "track_id"
    ].nunique()
    == 110_198
)


expansion_count_valid = bool(
    (
        track_geographic_expansion_df[
            "validated_expansion_markets"
        ]
        >= 0
    ).all()
    and
    (
        track_geographic_expansion_df[
            "validated_expansion_markets"
        ]
        <=
        track_geographic_expansion_df[
            "markets_observed"
        ]
    ).all()
)


share_valid = bool(
    track_geographic_expansion_df[
        "expansion_market_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)


index_fields = [
    "expansion_breadth_index",
    "expansion_share_index",
    "expansion_speed_index",
    "geographic_expansion_index",
]


numeric_valid = bool(
    np.isfinite(
        track_geographic_expansion_df[
            index_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


indices_bounded = bool(
    track_geographic_expansion_df[
        index_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    track_geographic_expansion_df[
        "geographic_expansion_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_expansion_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_expansion_df
    )
)


section_9_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 9.1 completion",

            "Requirement":
                "New-Market Entry must already be complete.",

            "Observed Evidence":
                f"Section 9.1 completion status: "
                f"{section_9_1_status}",

            "Passed":
                section_9_1_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_expansion_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one expansion record.",

            "Observed Evidence":
                f"{track_geographic_expansion_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Expansion-count validity",

            "Requirement":
                "Validated expansion markets must remain "
                "within observed market bounds.",

            "Observed Evidence":
                "Expansion counts assessed against "
                "observed market totals",

            "Passed":
                expansion_count_valid,
        },
        {
            "Validation Area":
                "Expansion-share bounds",

            "Requirement":
                "Expansion market shares must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Expansion shares assessed for all tracks",

            "Passed":
                share_valid,
        },
        {
            "Validation Area":
                "Expansion metric validity",

            "Requirement":
                "Expansion indices must remain finite.",

            "Observed Evidence":
                f"{len(index_fields)} numeric expansion "
                "indices assessed",

            "Passed":
                numeric_valid,
        },
        {
            "Validation Area":
                "Expansion-index bounds",

            "Requirement":
                "All expansion indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Expansion breadth, share, speed and "
                "composite indices assessed",

            "Passed":
                indices_bounded,
        },
        {
            "Validation Area":
                "Expansion classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "expansion classification.",

            "Observed Evidence":
                f"{track_geographic_expansion_df['geographic_expansion_class'].notna().sum():,} "
                "expansion classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Expansion population reconciliation",

            "Requirement":
                "Expansion classifications must reconcile "
                "with the validated track population.",

            "Observed Evidence":
                f"{int(track_geographic_expansion_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_expansion_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Track geographic expansion validation")
print("=" * 100)

display(
    section_9_2_validation_df
)


# ------------------------------------------------------------
# 15. Final completion status
# ------------------------------------------------------------

section_9_2_complete = bool(
    section_9_2_validation_df[
        "Passed"
    ].all()
)


if not section_9_2_complete:

    failed_checks = (
        section_9_2_validation_df
        .loc[
            ~section_9_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 9.2 geographic expansion "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 9.2 geographic expansion "
    "validation checks passed."
)

print(
    f"Section 9.2 completion status: "
    f"{section_9_2_complete}"
)

print(
    f"Geographic expansion intelligence prepared "
    f"for {len(track_geographic_expansion_df):,} tracks."
)

print(
    "Geographic expansion measures validated "
    "growth in observed market presence over time."
)

print(
    "The workflow is ready for "
    "Section 9.3 Geographic Contraction."
)
```

#### Interpretation

Section 9.2 successfully converts the validated new-market entry results from Section 9.1 into a broader measure of **geographic expansion** for all **110,198 tracks**.

A total of **12,564 tracks** show at least one validated expansion into an additional country market, while **97,634 tracks (88.60%)** show **no validated geographic expansion**. This confirms that expansion beyond the initial observed market footprint is relatively uncommon across the full track population.

The median number of validated expansion markets is **0**, and the median geographic expansion index is also **0.0000**. This reflects the strong dominance of tracks with no validated expansion. However, the mean expansion index is higher at **0.0415**, indicating that a smaller subset of tracks experiences substantially stronger geographic movement.

The classification results show a clear split between non-expanding and actively expanding tracks:

- **97,634 tracks (88.60%)** have **No validated expansion**
- **126 tracks (0.11%)** show **Limited expansion**
- **3,240 tracks (2.94%)** show **Moderate expansion**
- **8,387 tracks (7.61%)** show **Broad expansion**
- **811 tracks (0.74%)** show **Very broad expansion**

Although the Limited Expansion category is small, the larger Broad Expansion group shows that many tracks which do expand tend to move beyond only one or two additional markets.

The diagnostics reinforce this pattern. **4,394 tracks** expanded into at least **5 validated markets**, while **2,747 tracks** expanded into at least **10 markets**. A smaller group of **907 tracks** reached at least **25 additional validated markets**, showing substantial international geographic movement.

The strongest observed track reached a maximum of **53 validated expansion markets**, demonstrating that a small number of tracks achieved very broad geographic diffusion across the available PMIP country coverage.

The geographic expansion index combines three complementary dimensions:

- **Expansion breadth**, representing how many validated new markets were entered.
- **Expansion share**, representing how much of the track's total observed geographic footprint consists of later validated market entries.
- **Expansion speed**, representing how quickly those additional markets were reached after the track's first observed market appearance.

This means the index does not reward market count alone. A track that expands into many markets gradually is distinguished from one that reaches a similarly broad footprint much more quickly.

The diagnostics also show that **1,761 tracks** achieve a geographic expansion index of at least **0.50**, representing the strongest expansion cases when breadth, share, and timing are considered together.

All Section 9.2 validation checks passed. The full **110,198-track population** was preserved, each track received exactly one expansion record, expansion counts remained within observed market bounds, percentage measures remained within valid limits, all expansion indices remained finite and between **0 and 1**, and all classifications reconciled with the complete track population.

Section 9.2 therefore establishes a validated measure of how strongly tracks expand geographically after their initial observed market presence.

The workflow is ready for **Section 9.3 Geographic Contraction**.

### 9.3 Geographic Contraction

#### Purpose

Identify tracks whose validated geographic presence contracts over time by detecting country markets that were previously observed but later disappear from the track's active market footprint.

This subsection complements Section 9.2 Geographic Expansion by measuring the opposite direction of movement. Rather than asking where a track enters new markets, it examines where previously established market presence is no longer observed.

The analysis uses chronological track-country histories to estimate market exits, contraction breadth, contraction share, and the timing of contraction. Temporal safeguards are retained so that apparent exits caused by incomplete market coverage or irregular reporting are not automatically treated as genuine geographic decline.

The resulting indicators describe observed geographic contraction only. They do not establish why a track disappeared from a market or prove that audience demand declined.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP geographic contraction")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_9_2_status = globals().get(
    "section_9_2_complete",
    False
)

print(
    f"Section 9.2 completion status: "
    f"{section_9_2_status}"
)

if not section_9_2_status:
    raise RuntimeError(
        "Section 9.3 requires Section 9.2 "
        "Geographic Expansion to be complete."
    )


# ------------------------------------------------------------
# 2. Recover Section 9.1 track-country history
# ------------------------------------------------------------

history_candidate_names = [
    "track_country_entry_df",
    "track_country_history_df",
    "track_market_history_df",
]

track_country_history_df = None
track_country_history_source_name = None

required_history_columns = {
    "track_id",
    "country",
    "first_observed_date",
    "last_observed_date",
}

for name in history_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_history_columns.issubset(
        candidate.columns
    ):
        continue

    track_country_history_df = candidate.copy()
    track_country_history_source_name = name
    break


if track_country_history_df is None:

    for name, candidate in globals().copy().items():

        if not isinstance(candidate, pd.DataFrame):
            continue

        if not required_history_columns.issubset(
            candidate.columns
        ):
            continue

        track_country_history_df = candidate.copy()
        track_country_history_source_name = name
        break


if track_country_history_df is None:
    raise RuntimeError(
        "Section 9.3 could not recover the validated "
        "track-country chronological history."
    )


# ------------------------------------------------------------
# 3. Recover Section 9.2 track-level expansion source
# ------------------------------------------------------------

expansion_candidate_names = [
    "track_geographic_expansion_df",
    "track_entry_summary_df",
]

track_level_source_df = None
track_level_source_name = None

for name in expansion_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if (
        len(candidate) == 110_198
        and
        "track_id" in candidate.columns
    ):
        track_level_source_df = candidate.copy()
        track_level_source_name = name
        break


if track_level_source_df is None:
    raise RuntimeError(
        "Section 9.3 could not recover the validated "
        "track-level geographic source."
    )


print()
print("Validated geographic contraction sources")
print("=" * 100)

print(
    f"Track-country history source: "
    f"{track_country_history_source_name}"
)

print(
    f"Track-level source: "
    f"{track_level_source_name}"
)

print(
    f"Track-country histories: "
    f"{len(track_country_history_df):,}"
)

print(
    f"Tracks: "
    f"{track_level_source_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 4. Standardise date fields
# ------------------------------------------------------------

track_country_history_df[
    "first_observed_date"
] = pd.to_datetime(
    track_country_history_df[
        "first_observed_date"
    ],
    errors="coerce",
)

track_country_history_df[
    "last_observed_date"
] = pd.to_datetime(
    track_country_history_df[
        "last_observed_date"
    ],
    errors="coerce",
)

if (
    track_country_history_df[
        [
            "first_observed_date",
            "last_observed_date",
        ]
    ]
    .isna()
    .any()
    .any()
):
    raise RuntimeError(
        "Section 9.3 found invalid chronological "
        "track-country dates."
    )


# ------------------------------------------------------------
# 5. Determine dataset end date
# ------------------------------------------------------------

dataset_end_date = (
    track_country_history_df[
        "last_observed_date"
    ]
    .max()
)

print()
print("Validated contraction reporting boundary")
print("=" * 100)

print(
    f"Dataset end date: "
    f"{dataset_end_date.date()}"
)


# ------------------------------------------------------------
# 6. Build market-exit timing features
# ------------------------------------------------------------

track_country_history_df[
    "days_before_dataset_end"
] = (
    dataset_end_date
    -
    track_country_history_df[
        "last_observed_date"
    ]
).dt.days


# ------------------------------------------------------------
# 7. Recover temporal safeguards
# ------------------------------------------------------------

temporal_class_col = None

for candidate_col in [
    "temporal_alignment_class",
]:
    if candidate_col in track_country_history_df.columns:
        temporal_class_col = candidate_col
        break


direct_caution_col = None

for candidate_col in [
    "direct_comparison_caution",
]:
    if candidate_col in track_country_history_df.columns:
        direct_caution_col = candidate_col
        break


# ------------------------------------------------------------
# 8. Define provisional observed market exit
# ------------------------------------------------------------

# A market must disappear sufficiently before the dataset end
# to avoid treating the final reporting period itself as an exit.
#
# 90 days provides a conservative buffer for irregular reporting.

EXIT_BUFFER_DAYS = 90

track_country_history_df[
    "is_provisional_market_exit"
] = (
    track_country_history_df[
        "days_before_dataset_end"
    ]
    >= EXIT_BUFFER_DAYS
)


# ------------------------------------------------------------
# 9. Apply temporal safeguard to market exits
# ------------------------------------------------------------

track_country_history_df[
    "market_exit_temporally_supported"
] = True


if temporal_class_col is not None:

    unreliable_temporal_classes = [
        "Limited temporal alignment",
        "Partial temporal alignment",
    ]

    track_country_history_df[
        "market_exit_temporally_supported"
    ] = ~(
        track_country_history_df[
            temporal_class_col
        ]
        .isin(
            unreliable_temporal_classes
        )
        &
        track_country_history_df[
            "is_provisional_market_exit"
        ]
    )


if direct_caution_col is not None:

    track_country_history_df[
        "market_exit_temporally_supported"
    ] = (
        track_country_history_df[
            "market_exit_temporally_supported"
        ]
        &
        ~(
            track_country_history_df[
                direct_caution_col
            ]
            .fillna(False)
            &
            track_country_history_df[
                "is_provisional_market_exit"
            ]
        )
    )


track_country_history_df[
    "validated_market_exit"
] = (
    track_country_history_df[
        "is_provisional_market_exit"
    ]
    &
    track_country_history_df[
        "market_exit_temporally_supported"
    ]
)


# ------------------------------------------------------------
# 10. Track-level contraction summary
# ------------------------------------------------------------

track_contraction_summary_df = (
    track_country_history_df
    .groupby(
        "track_id",
        as_index=False,
    )
    .agg(
        markets_observed=(
            "country",
            "nunique",
        ),

        provisional_market_exits=(
            "is_provisional_market_exit",
            "sum",
        ),

        validated_market_exits=(
            "validated_market_exit",
            "sum",
        ),

        first_market_date=(
            "first_observed_date",
            "min",
        ),

        latest_market_presence_date=(
            "last_observed_date",
            "max",
        ),

        earliest_market_exit_date=(
            "last_observed_date",
            "min",
        ),
    )
)


# ------------------------------------------------------------
# 11. Contraction share
# ------------------------------------------------------------

track_contraction_summary_df[
    "contraction_market_share_pct"
] = np.where(
    track_contraction_summary_df[
        "markets_observed"
    ] > 0,

    (
        track_contraction_summary_df[
            "validated_market_exits"
        ]
        /
        track_contraction_summary_df[
            "markets_observed"
        ]
        * 100
    ),

    0.0,
)


# ------------------------------------------------------------
# 12. Contraction recency
# ------------------------------------------------------------

track_contraction_summary_df[
    "days_since_earliest_market_exit"
] = (
    dataset_end_date
    -
    track_contraction_summary_df[
        "earliest_market_exit_date"
    ]
).dt.days


# ------------------------------------------------------------
# 13. Build contraction breadth index
# ------------------------------------------------------------

max_market_count = float(
    track_contraction_summary_df[
        "markets_observed"
    ].max()
)

if (
    not np.isfinite(max_market_count)
    or max_market_count <= 1
):
    max_market_count = 73.0


track_contraction_summary_df[
    "contraction_breadth_index"
] = (
    track_contraction_summary_df[
        "validated_market_exits"
    ]
    /
    max(
        max_market_count - 1,
        1,
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 14. Build contraction share index
# ------------------------------------------------------------

track_contraction_summary_df[
    "contraction_share_index"
] = (
    track_contraction_summary_df[
        "contraction_market_share_pct"
    ]
    /
    100
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 15. Build contraction recency index
# ------------------------------------------------------------

positive_exit_recency = (
    track_contraction_summary_df
    .loc[
        track_contraction_summary_df[
            "validated_market_exits"
        ] > 0,
        "days_since_earliest_market_exit",
    ]
)

if len(positive_exit_recency) > 0:

    recency_reference = float(
        positive_exit_recency.quantile(
            0.95
        )
    )

else:

    recency_reference = 1.0


if (
    not np.isfinite(recency_reference)
    or recency_reference <= 0
):
    recency_reference = 1.0


track_contraction_summary_df[
    "contraction_recency_index"
] = np.where(
    track_contraction_summary_df[
        "validated_market_exits"
    ] > 0,

    (
        1
        -
        (
            track_contraction_summary_df[
                "days_since_earliest_market_exit"
            ]
            .fillna(recency_reference)
            .clip(
                lower=0,
                upper=recency_reference,
            )
            /
            recency_reference
        )
    ),

    0.0,
)


track_contraction_summary_df[
    "contraction_recency_index"
] = (
    track_contraction_summary_df[
        "contraction_recency_index"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 16. Composite geographic contraction index
# ------------------------------------------------------------

track_contraction_summary_df[
    "geographic_contraction_index"
] = (
    (
        0.50
        * track_contraction_summary_df[
            "contraction_breadth_index"
        ]
    )
    +
    (
        0.35
        * track_contraction_summary_df[
            "contraction_share_index"
        ]
    )
    +
    (
        0.15
        * track_contraction_summary_df[
            "contraction_recency_index"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 17. Restore track identity fields
# ------------------------------------------------------------

identity_columns = [
    "track_id",
]

for column in [
    "track_name",
    "artists_raw",
]:
    if column in track_level_source_df.columns:
        identity_columns.append(
            column
        )


track_identity_df = (
    track_level_source_df[
        identity_columns
    ]
    .drop_duplicates(
        subset=[
            "track_id"
        ]
    )
)


track_geographic_contraction_df = (
    track_identity_df
    .merge(
        track_contraction_summary_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 18. Geographic contraction classification
# ------------------------------------------------------------

def classify_geographic_contraction(row):

    exits = row[
        "validated_market_exits"
    ]

    contraction_index = row[
        "geographic_contraction_index"
    ]

    if exits == 0:
        return "No validated contraction"

    if (
        exits >= 25
        and
        contraction_index >= 0.50
    ):
        return "Very broad contraction"

    if (
        exits >= 10
        or
        contraction_index >= 0.30
    ):
        return "Broad contraction"

    if (
        exits >= 3
        or
        contraction_index >= 0.15
    ):
        return "Moderate contraction"

    return "Limited contraction"


track_geographic_contraction_df[
    "geographic_contraction_class"
] = (
    track_geographic_contraction_df
    .apply(
        classify_geographic_contraction,
        axis=1,
    )
)


# ------------------------------------------------------------
# 19. Geographic contraction overview
# ------------------------------------------------------------

print()
print("Track geographic contraction overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_contraction_df):,}"
)

print(
    f"Tracks with validated geographic contraction: "
    f"{(track_geographic_contraction_df['validated_market_exits'] > 0).sum():,}"
)

print(
    f"Median validated market exits: "
    f"{track_geographic_contraction_df['validated_market_exits'].median():.1f}"
)

print(
    f"Maximum validated market exits: "
    f"{track_geographic_contraction_df['validated_market_exits'].max():.0f}"
)

print(
    f"Median contraction market share: "
    f"{track_geographic_contraction_df['contraction_market_share_pct'].median():.2f}%"
)

print(
    f"Median geographic contraction index: "
    f"{track_geographic_contraction_df['geographic_contraction_index'].median():.4f}"
)

print(
    f"Mean geographic contraction index: "
    f"{track_geographic_contraction_df['geographic_contraction_index'].mean():.4f}"
)


# ------------------------------------------------------------
# 20. Contraction classification summary
# ------------------------------------------------------------

contraction_class_order = [
    "No validated contraction",
    "Limited contraction",
    "Moderate contraction",
    "Broad contraction",
    "Very broad contraction",
]


track_geographic_contraction_summary_df = (
    track_geographic_contraction_df[
        "geographic_contraction_class"
    ]
    .value_counts()
    .reindex(
        contraction_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Contraction Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_contraction_summary_df[
    "Share (%)"
] = (
    track_geographic_contraction_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_contraction_df)
    * 100
)


print()
print("Track geographic contraction classification")
print("=" * 100)

display(
    track_geographic_contraction_summary_df
)


# ------------------------------------------------------------
# 21. Tracks with strongest geographic contraction
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "markets_observed",
        "validated_market_exits",
        "contraction_market_share_pct",
        "days_since_earliest_market_exit",
        "contraction_breadth_index",
        "contraction_recency_index",
        "geographic_contraction_index",
        "geographic_contraction_class",
    ]
    if column in track_geographic_contraction_df.columns
]


print()
print("Tracks with strongest validated geographic contraction")
print("=" * 100)

strongest_geographic_contraction_df = (
    track_geographic_contraction_df
    .sort_values(
        [
            "geographic_contraction_index",
            "validated_market_exits",
            "contraction_market_share_pct",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_geographic_contraction_df
)


# ------------------------------------------------------------
# 22. Contraction diagnostics
# ------------------------------------------------------------

track_geographic_contraction_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with no validated contraction",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "validated_market_exits"
                        ]
                        == 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 1 validated market exit",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "validated_market_exits"
                        ]
                        >= 1
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 5 validated market exits",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "validated_market_exits"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 10 validated market exits",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "validated_market_exits"
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with at least 25 validated market exits",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "validated_market_exits"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with contraction index >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_contraction_df[
                            "geographic_contraction_index"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)


print()
print("Track geographic contraction diagnostics")
print("=" * 100)

display(
    track_geographic_contraction_diagnostics_df
)


# ------------------------------------------------------------
# 23. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_contraction_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_contraction_df[
        "track_id"
    ].nunique()
    == 110_198
)


exit_count_valid = bool(
    (
        track_geographic_contraction_df[
            "validated_market_exits"
        ]
        >= 0
    ).all()
    and
    (
        track_geographic_contraction_df[
            "validated_market_exits"
        ]
        <=
        track_geographic_contraction_df[
            "markets_observed"
        ]
    ).all()
)


share_valid = bool(
    track_geographic_contraction_df[
        "contraction_market_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)


index_fields = [
    "contraction_breadth_index",
    "contraction_share_index",
    "contraction_recency_index",
    "geographic_contraction_index",
]


numeric_valid = bool(
    np.isfinite(
        track_geographic_contraction_df[
            index_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


indices_bounded = bool(
    track_geographic_contraction_df[
        index_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


chronology_valid = bool(
    (
        track_geographic_contraction_df[
            "latest_market_presence_date"
        ]
        >=
        track_geographic_contraction_df[
            "first_market_date"
        ]
    ).all()
)


classification_complete = bool(
    track_geographic_contraction_df[
        "geographic_contraction_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_contraction_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_contraction_df
    )
)


section_9_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 9.2 completion",

            "Requirement":
                "Geographic Expansion must already be complete.",

            "Observed Evidence":
                f"Section 9.2 completion status: "
                f"{section_9_2_status}",

            "Passed":
                section_9_2_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_contraction_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one contraction record.",

            "Observed Evidence":
                f"{track_geographic_contraction_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Market-exit count validity",

            "Requirement":
                "Validated market exits must remain "
                "within observed market bounds.",

            "Observed Evidence":
                "Validated exit counts assessed against "
                "observed market totals",

            "Passed":
                exit_count_valid,
        },
        {
            "Validation Area":
                "Contraction-share bounds",

            "Requirement":
                "Contraction market shares must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Contraction shares assessed for all tracks",

            "Passed":
                share_valid,
        },
        {
            "Validation Area":
                "Contraction metric validity",

            "Requirement":
                "Contraction indices must remain finite.",

            "Observed Evidence":
                f"{len(index_fields)} numeric contraction "
                "indices assessed",

            "Passed":
                numeric_valid,
        },
        {
            "Validation Area":
                "Contraction-index bounds",

            "Requirement":
                "All contraction indices must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Contraction breadth, share, recency and "
                "composite indices assessed",

            "Passed":
                indices_bounded,
        },
        {
            "Validation Area":
                "Chronological validity",

            "Requirement":
                "Latest market presence must not occur "
                "before the first observed market date.",

            "Observed Evidence":
                f"{len(track_geographic_contraction_df):,} "
                "track histories assessed",

            "Passed":
                chronology_valid,
        },
        {
            "Validation Area":
                "Contraction classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "contraction classification.",

            "Observed Evidence":
                f"{track_geographic_contraction_df['geographic_contraction_class'].notna().sum():,} "
                "contraction classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Contraction population reconciliation",

            "Requirement":
                "Contraction classifications must reconcile "
                "with the validated track population.",

            "Observed Evidence":
                f"{int(track_geographic_contraction_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_contraction_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Track geographic contraction validation")
print("=" * 100)

display(
    section_9_3_validation_df
)


# ------------------------------------------------------------
# 24. Final completion status
# ------------------------------------------------------------

section_9_3_complete = bool(
    section_9_3_validation_df[
        "Passed"
    ].all()
)


if not section_9_3_complete:

    failed_checks = (
        section_9_3_validation_df
        .loc[
            ~section_9_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 9.3 geographic contraction "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 9.3 geographic contraction "
    "validation checks passed."
)

print(
    f"Section 9.3 completion status: "
    f"{section_9_3_complete}"
)

print(
    f"Geographic contraction intelligence prepared "
    f"for {len(track_geographic_contraction_df):,} tracks."
)

print(
    "Geographic contraction measures validated "
    "reductions in observed market presence while "
    "preserving temporal coverage safeguards."
)

print(
    "The workflow is ready for "
    "Section 9.4 Cross-Market Momentum."
)
```

#### Interpretation

Section 9.3 successfully identifies and validates **geographic contraction patterns** for the complete population of **110,198 tracks** using **353,282 track-country histories**.

The analysis uses **2023-04-06** as the validated dataset end date and treats a market disappearance as a potential contraction only when the track's final observation in that country occurs sufficiently before this boundary. Temporal safeguards are then applied before the disappearance is retained as a validated market exit.

A total of **88,450 tracks** show at least one validated geographic contraction, while **21,748 tracks (19.74%)** show **No validated contraction**.

The median track has **1 validated market exit**, while the maximum observed contraction reaches **53 markets** for a single track. The median contraction-market share is **100.00%**, indicating that for a large proportion of tracks—particularly tracks observed in only one or a small number of markets—their historical market presence ends before the final period of the dataset.

The geographic contraction classification shows:

- **21,748 tracks (19.74%)** have **No validated contraction**
- **4 tracks (<0.01%)** show **Limited contraction**
- **1,489 tracks (1.35%)** show **Moderate contraction**
- **84,326 tracks (76.52%)** show **Broad contraction**
- **2,631 tracks (2.39%)** show **Very broad contraction**

The unusually large **Broad contraction** category should be interpreted carefully. Earlier PMIP analysis showed that most tracks have highly concentrated geographic footprints and many appear in only one market. Therefore, when such a track stops appearing before the end of the historical dataset, the disappearance can represent **100% of its observed geographic footprint** even though only one country was involved.

For this reason, the contraction measure should not automatically be interpreted as evidence that a track suddenly lost international popularity. It captures the **end of observed market presence within the available chart history**. Such disappearance may reflect declining performance, a normal chart lifecycle, falling below the chart threshold, or another unobserved process.

The diagnostics provide additional context. **9,905 tracks** experience at least **5 validated market exits**, **6,665 tracks** experience at least **10**, and **2,644 tracks** experience at least **25 validated exits**. These broader contraction cases provide stronger evidence of meaningful geographic withdrawal than a single-market disappearance.

A total of **8,174 tracks** achieve a geographic contraction index of at least **0.50**, identifying the strongest contraction cases when several dimensions are considered together.

The geographic contraction index combines:

- **Contraction breadth**, representing how many previously observed markets disappear.
- **Contraction share**, representing the proportion of the track's observed geographic footprint affected.
- **Contraction recency**, providing temporal context for when those market exits occurred.

The median geographic contraction index is **0.4247**, while the mean is **0.3544**. The higher median reflects the large population of geographically concentrated tracks for which the disappearance of a single market represents a substantial proportion of their total observed market footprint.

Importantly, Section 9.3 measures **observed geographic disappearance rather than causal decline**. A market exit means that the track is no longer observed in that country's available chart history after a particular point; it does not prove that listeners stopped streaming the track or that the track completely disappeared from that country's music market.

All Section 9.3 validation checks passed. The full **110,198-track population** was preserved, track identities remained unique, market-exit counts stayed within observed market bounds, contraction percentages remained valid, all four contraction indices were finite and bounded between **0 and 1**, chronological consistency was preserved, and every track received exactly one contraction classification.

Section 9.3 therefore provides the contraction side of PMIP's geographic movement intelligence and complements the expansion measures created in Section 9.2.

The workflow is ready for **Section 9.4 Cross-Market Momentum**.

### 9.4 Cross-Market Momentum

#### Purpose

Measure the direction and strength of geographic movement by combining validated expansion and contraction behaviour across country markets.

This subsection compares the number and intensity of new-market entries against validated market exits to determine whether each track is experiencing positive geographic momentum, relative stability, or geographic decline.

The analysis creates separate expansion and contraction components, calculates their net balance, and produces a cross-market momentum index that summarises whether a track's geographic footprint is strengthening or weakening over time.

Temporal safeguards from Sections 9.1 to 9.3 remain active, meaning that only validated market entries and exits are used. The resulting momentum indicators describe observed geographic movement and do not establish the causes of expansion or contraction.

```python
import pandas as pd
import numpy as np

print("Analysing PMIP cross-market momentum")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_9_3_status = globals().get(
    "section_9_3_complete",
    False
)

print(
    f"Section 9.3 completion status: "
    f"{section_9_3_status}"
)

if not section_9_3_status:
    raise RuntimeError(
        "Section 9.4 requires Section 9.3 "
        "Geographic Contraction to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated expansion intelligence
# ------------------------------------------------------------

expansion_candidates = [
    "track_geographic_expansion_df",
]

track_expansion_df = None
track_expansion_source_name = None

required_expansion_columns = {
    "track_id",
    "validated_expansion_markets",
    "expansion_market_share_pct",
    "geographic_expansion_index",
}

for name in expansion_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_expansion_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_expansion_df = candidate.copy()
    track_expansion_source_name = name
    break


if track_expansion_df is None:
    raise RuntimeError(
        "Section 9.4 could not recover the validated "
        "geographic expansion intelligence."
    )


# ------------------------------------------------------------
# 3. Recover validated contraction intelligence
# ------------------------------------------------------------

contraction_candidates = [
    "track_geographic_contraction_df",
]

track_contraction_df = None
track_contraction_source_name = None

required_contraction_columns = {
    "track_id",
    "validated_market_exits",
    "contraction_market_share_pct",
    "geographic_contraction_index",
}

for name in contraction_candidates:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_contraction_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_contraction_df = candidate.copy()
    track_contraction_source_name = name
    break


if track_contraction_df is None:
    raise RuntimeError(
        "Section 9.4 could not recover the validated "
        "geographic contraction intelligence."
    )


print()
print("Validated cross-market momentum sources")
print("=" * 100)

print(
    f"Expansion source: "
    f"{track_expansion_source_name}"
)

print(
    f"Contraction source: "
    f"{track_contraction_source_name}"
)

print(
    f"Tracks: "
    f"{track_expansion_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 4. Prepare identity columns
# ------------------------------------------------------------

identity_columns = [
    "track_id",
]

for column in [
    "track_name",
    "artists_raw",
]:
    if column in track_expansion_df.columns:
        identity_columns.append(
            column
        )


track_identity_df = (
    track_expansion_df[
        identity_columns
    ]
    .drop_duplicates(
        subset=[
            "track_id"
        ]
    )
)


# ------------------------------------------------------------
# 5. Select expansion features
# ------------------------------------------------------------

expansion_columns = [
    "track_id",
    "markets_observed",
    "validated_expansion_markets",
    "expansion_market_share_pct",
    "expansion_breadth_index",
    "expansion_share_index",
    "expansion_speed_index",
    "geographic_expansion_index",
    "geographic_expansion_class",
]

expansion_columns = [
    column
    for column in expansion_columns
    if column in track_expansion_df.columns
]


expansion_component_df = (
    track_expansion_df[
        expansion_columns
    ]
    .copy()
)


# ------------------------------------------------------------
# 6. Select contraction features
# ------------------------------------------------------------

contraction_columns = [
    "track_id",
    "validated_market_exits",
    "contraction_market_share_pct",
    "contraction_breadth_index",
    "contraction_share_index",
    "contraction_recency_index",
    "geographic_contraction_index",
    "geographic_contraction_class",
]

contraction_columns = [
    column
    for column in contraction_columns
    if column in track_contraction_df.columns
]


contraction_component_df = (
    track_contraction_df[
        contraction_columns
    ]
    .copy()
)


# ------------------------------------------------------------
# 7. Merge expansion and contraction intelligence
# ------------------------------------------------------------

track_cross_market_momentum_df = (
    track_identity_df
    .merge(
        expansion_component_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
    .merge(
        contraction_component_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 8. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    "markets_observed",
    "validated_expansion_markets",
    "expansion_market_share_pct",
    "expansion_breadth_index",
    "expansion_share_index",
    "expansion_speed_index",
    "geographic_expansion_index",
    "validated_market_exits",
    "contraction_market_share_pct",
    "contraction_breadth_index",
    "contraction_share_index",
    "contraction_recency_index",
    "geographic_contraction_index",
]

for column in numeric_columns:

    if column in track_cross_market_momentum_df.columns:

        track_cross_market_momentum_df[
            column
        ] = pd.to_numeric(
            track_cross_market_momentum_df[
                column
            ],
            errors="coerce",
        ).fillna(0.0)


# ------------------------------------------------------------
# 9. Net market movement
# ------------------------------------------------------------

track_cross_market_momentum_df[
    "net_market_movement"
] = (
    track_cross_market_momentum_df[
        "validated_expansion_markets"
    ]
    -
    track_cross_market_momentum_df[
        "validated_market_exits"
    ]
)


# ------------------------------------------------------------
# 10. Net movement share
# ------------------------------------------------------------

track_cross_market_momentum_df[
    "net_market_movement_share_pct"
] = (
    track_cross_market_momentum_df[
        "expansion_market_share_pct"
    ]
    -
    track_cross_market_momentum_df[
        "contraction_market_share_pct"
    ]
)


# ------------------------------------------------------------
# 11. Raw momentum balance
# ------------------------------------------------------------

track_cross_market_momentum_df[
    "raw_momentum_balance"
] = (
    track_cross_market_momentum_df[
        "geographic_expansion_index"
    ]
    -
    track_cross_market_momentum_df[
        "geographic_contraction_index"
    ]
)


# ------------------------------------------------------------
# 12. Convert balance to 0-to-1 momentum index
# ------------------------------------------------------------

track_cross_market_momentum_df[
    "cross_market_momentum_index"
] = (
    (
        track_cross_market_momentum_df[
            "raw_momentum_balance"
        ]
        + 1
    )
    /
    2
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 13. Directional strength
# ------------------------------------------------------------

track_cross_market_momentum_df[
    "momentum_strength"
] = (
    track_cross_market_momentum_df[
        "raw_momentum_balance"
    ]
    .abs()
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 14. Cross-market momentum classification
# ------------------------------------------------------------

def classify_cross_market_momentum(row):

    balance = row[
        "raw_momentum_balance"
    ]

    net_markets = row[
        "net_market_movement"
    ]

    expansion = row[
        "validated_expansion_markets"
    ]

    contraction = row[
        "validated_market_exits"
    ]

    if (
        expansion == 0
        and
        contraction == 0
    ):
        return "No validated movement"

    if (
        balance >= 0.30
        or
        net_markets >= 10
    ):
        return "Strong positive momentum"

    if (
        balance >= 0.10
        or
        net_markets >= 3
    ):
        return "Positive momentum"

    if (
        balance <= -0.30
        or
        net_markets <= -10
    ):
        return "Strong negative momentum"

    if (
        balance <= -0.10
        or
        net_markets <= -3
    ):
        return "Negative momentum"

    return "Stable / balanced movement"


track_cross_market_momentum_df[
    "cross_market_momentum_class"
] = (
    track_cross_market_momentum_df
    .apply(
        classify_cross_market_momentum,
        axis=1,
    )
)


# ------------------------------------------------------------
# 15. Momentum overview
# ------------------------------------------------------------

print()
print("Track cross-market momentum overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_cross_market_momentum_df):,}"
)

print(
    f"Median validated expansion markets: "
    f"{track_cross_market_momentum_df['validated_expansion_markets'].median():.1f}"
)

print(
    f"Median validated market exits: "
    f"{track_cross_market_momentum_df['validated_market_exits'].median():.1f}"
)

print(
    f"Median net market movement: "
    f"{track_cross_market_momentum_df['net_market_movement'].median():.1f}"
)

print(
    f"Median raw momentum balance: "
    f"{track_cross_market_momentum_df['raw_momentum_balance'].median():.4f}"
)

print(
    f"Median cross-market momentum index: "
    f"{track_cross_market_momentum_df['cross_market_momentum_index'].median():.4f}"
)

print(
    f"Mean cross-market momentum index: "
    f"{track_cross_market_momentum_df['cross_market_momentum_index'].mean():.4f}"
)


# ------------------------------------------------------------
# 16. Momentum classification summary
# ------------------------------------------------------------

momentum_class_order = [
    "Strong positive momentum",
    "Positive momentum",
    "Stable / balanced movement",
    "Negative momentum",
    "Strong negative momentum",
    "No validated movement",
]


track_cross_market_momentum_summary_df = (
    track_cross_market_momentum_df[
        "cross_market_momentum_class"
    ]
    .value_counts()
    .reindex(
        momentum_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Cross-Market Momentum Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_cross_market_momentum_summary_df[
    "Share (%)"
] = (
    track_cross_market_momentum_summary_df[
        "Tracks"
    ]
    /
    len(track_cross_market_momentum_df)
    * 100
)


print()
print("Track cross-market momentum classification")
print("=" * 100)

display(
    track_cross_market_momentum_summary_df
)


# ------------------------------------------------------------
# 17. Strongest positive momentum
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "markets_observed",
        "validated_expansion_markets",
        "validated_market_exits",
        "net_market_movement",
        "geographic_expansion_index",
        "geographic_contraction_index",
        "raw_momentum_balance",
        "cross_market_momentum_index",
        "cross_market_momentum_class",
    ]
    if column in track_cross_market_momentum_df.columns
]


print()
print("Tracks with strongest positive cross-market momentum")
print("=" * 100)

strongest_positive_momentum_df = (
    track_cross_market_momentum_df
    .sort_values(
        [
            "raw_momentum_balance",
            "net_market_movement",
            "geographic_expansion_index",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_positive_momentum_df
)


# ------------------------------------------------------------
# 18. Strongest negative momentum
# ------------------------------------------------------------

print()
print("Tracks with strongest negative cross-market momentum")
print("=" * 100)

strongest_negative_momentum_df = (
    track_cross_market_momentum_df
    .sort_values(
        [
            "raw_momentum_balance",
            "net_market_movement",
            "geographic_contraction_index",
        ],
        ascending=[
            True,
            True,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_negative_momentum_df
)


# ------------------------------------------------------------
# 19. Momentum diagnostics
# ------------------------------------------------------------

track_cross_market_momentum_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with net positive market movement",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "net_market_movement"
                        ]
                        > 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with net negative market movement",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "net_market_movement"
                        ]
                        < 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with balanced market counts",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "net_market_movement"
                        ]
                        == 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with positive raw momentum balance",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "raw_momentum_balance"
                        ]
                        > 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with negative raw momentum balance",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "raw_momentum_balance"
                        ]
                        < 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with momentum strength >= 0.50",

            "Observed Value":
                int(
                    (
                        track_cross_market_momentum_df[
                            "momentum_strength"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)


print()
print("Track cross-market momentum diagnostics")
print("=" * 100)

display(
    track_cross_market_momentum_diagnostics_df
)


# ------------------------------------------------------------
# 20. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_cross_market_momentum_df
    )
    == 110_198
)


identity_valid = bool(
    track_cross_market_momentum_df[
        "track_id"
    ].nunique()
    == 110_198
)


component_valid = bool(
    (
        track_cross_market_momentum_df[
            "validated_expansion_markets"
        ]
        >= 0
    ).all()
    and
    (
        track_cross_market_momentum_df[
            "validated_market_exits"
        ]
        >= 0
    ).all()
)


index_fields = [
    "geographic_expansion_index",
    "geographic_contraction_index",
    "cross_market_momentum_index",
    "momentum_strength",
]


numeric_valid = bool(
    np.isfinite(
        track_cross_market_momentum_df[
            index_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


component_indices_bounded = bool(
    track_cross_market_momentum_df[
        [
            "geographic_expansion_index",
            "geographic_contraction_index",
            "cross_market_momentum_index",
            "momentum_strength",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


balance_bounds_valid = bool(
    track_cross_market_momentum_df[
        "raw_momentum_balance"
    ]
    .between(
        -1,
        1,
        inclusive="both",
    )
    .all()
)


classification_complete = bool(
    track_cross_market_momentum_df[
        "cross_market_momentum_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_cross_market_momentum_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_cross_market_momentum_df
    )
)


section_9_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 9.3 completion",

            "Requirement":
                "Geographic Contraction must already be complete.",

            "Observed Evidence":
                f"Section 9.3 completion status: "
                f"{section_9_3_status}",

            "Passed":
                section_9_3_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_cross_market_momentum_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one momentum record.",

            "Observed Evidence":
                f"{track_cross_market_momentum_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Movement component validity",

            "Requirement":
                "Expansion and contraction counts must "
                "remain non-negative.",

            "Observed Evidence":
                "Validated expansion and contraction "
                "counts assessed",

            "Passed":
                component_valid,
        },
        {
            "Validation Area":
                "Momentum numeric validity",

            "Requirement":
                "Momentum components and indices must "
                "remain finite.",

            "Observed Evidence":
                f"{len(index_fields)} numeric momentum "
                "fields assessed",

            "Passed":
                numeric_valid,
        },
        {
            "Validation Area":
                "Momentum index bounds",

            "Requirement":
                "Expansion, contraction, momentum and "
                "strength indices must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded component and momentum indices assessed",

            "Passed":
                component_indices_bounded,
        },
        {
            "Validation Area":
                "Momentum balance bounds",

            "Requirement":
                "Raw momentum balance must remain "
                "between -1 and 1.",

            "Observed Evidence":
                "Raw momentum balances assessed for all tracks",

            "Passed":
                balance_bounds_valid,
        },
        {
            "Validation Area":
                "Momentum classification completeness",

            "Requirement":
                "Every track must receive a cross-market "
                "momentum classification.",

            "Observed Evidence":
                f"{track_cross_market_momentum_df['cross_market_momentum_class'].notna().sum():,} "
                "momentum classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Momentum population reconciliation",

            "Requirement":
                "Momentum classifications must reconcile "
                "with the validated track population.",

            "Observed Evidence":
                f"{int(track_cross_market_momentum_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_cross_market_momentum_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Track cross-market momentum validation")
print("=" * 100)

display(
    section_9_4_validation_df
)


# ------------------------------------------------------------
# 21. Final completion status
# ------------------------------------------------------------

section_9_4_complete = bool(
    section_9_4_validation_df[
        "Passed"
    ].all()
)


if not section_9_4_complete:

    failed_checks = (
        section_9_4_validation_df
        .loc[
            ~section_9_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 9.4 cross-market momentum "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 9.4 cross-market momentum "
    "validation checks passed."
)

print(
    f"Section 9.4 completion status: "
    f"{section_9_4_complete}"
)

print(
    f"Cross-market momentum intelligence prepared "
    f"for {len(track_cross_market_momentum_df):,} tracks."
)

print(
    "Cross-market momentum combines validated "
    "geographic expansion and contraction behaviour."
)

print(
    "The workflow is ready for "
    "Section 9.5 Temporal Validation of Geographic Change."
)
```

#### Interpretation

Section 9.4 successfully combines validated geographic expansion and contraction into a single **cross-market momentum framework** for all **110,198 tracks**.

The results show that geographic movement is strongly skewed toward contraction across the historical dataset. The median track records **0 validated expansion markets**, **1 validated market exit**, and a median **net market movement of -1**.

The median raw momentum balance is **-0.4074**, which corresponds to a median cross-market momentum index of **0.2963** on the 0-to-1 scale. Since **0.50 represents a neutral balance** between expansion and contraction, the median value below 0.50 indicates that contraction generally outweighs expansion across the track population.

The momentum classifications reinforce this pattern:

- **274 tracks (0.25%)** show **Strong positive momentum**
- **300 tracks (0.27%)** show **Positive momentum**
- **5,715 tracks (5.19%)** show **Stable / balanced movement**
- **5,623 tracks (5.10%)** show **Negative momentum**
- **76,898 tracks (69.78%)** show **Strong negative momentum**
- **21,388 tracks (19.41%)** show **No validated movement**

This means only a very small proportion of tracks show clearly positive geographic momentum, while the majority either contract geographically or show no validated movement.

The diagnostics provide further context. Only **392 tracks** have a net positive market count, while **87,849 tracks** have a net negative market count. A further **21,957 tracks** have balanced entry and exit counts.

However, the raw momentum balance provides a more nuanced picture than market counts alone. **3,055 tracks** have a positive raw momentum balance, while **85,753 tracks** have a negative balance. This difference occurs because the momentum calculation considers the strength of expansion and contraction, not only the number of markets involved.

A total of **4,586 tracks** have a momentum strength of at least **0.50**, identifying tracks where the directional difference between expansion and contraction is particularly pronounced.

The cross-market momentum framework combines two major components:

- **Geographic expansion**, representing validated entry into additional country markets.
- **Geographic contraction**, representing validated disappearance from previously observed country markets.

The raw momentum balance is calculated as the difference between these two geographic movement indices. Positive values indicate expansion outweighing contraction, values near zero indicate a relatively balanced geographic pattern, and negative values indicate contraction outweighing expansion.

The 0-to-1 cross-market momentum index then converts this balance into a consistent PMIP analytical scale, where values closer to **1.0** represent stronger positive movement, values around **0.50** represent geographic balance, and values closer to **0.0** represent stronger negative movement.

The dominance of negative momentum should still be interpreted carefully. Earlier sections established that many tracks have short or highly concentrated chart histories. Therefore, a market disappearance may reflect the normal end of a chart run rather than a permanent loss of audience or streaming activity.

For this reason, Section 9.4 should be interpreted as **observed geographic chart momentum**, not as direct evidence of commercial success or failure.

All Section 9.4 validation checks passed. The full **110,198-track population** was retained, track identities remained unique, expansion and contraction components remained valid, all momentum indices were finite and correctly bounded, raw momentum values remained between **-1 and 1**, and every track received exactly one momentum classification.

Section 9.4 therefore provides a unified representation of whether each track's observed geographic footprint is expanding, stable, or contracting over time.

The workflow is ready for **Section 9.5 Temporal Validation of Geographic Change**.

### 9.5 Temporal Validation of Geographic Change

#### Purpose

Validate whether the geographic expansion, contraction, and cross-market momentum signals created in Sections 9.1 to 9.4 are supported by sufficiently reliable temporal evidence.

This subsection checks whether observed geographic changes are based on country markets with adequate reporting history and temporal alignment, rather than being driven by late market entry, incomplete historical coverage, or irregular observation windows.

The analysis combines the geographic movement outputs with the validated country temporal-alignment safeguards developed earlier in the notebook. It then measures how much of each track's movement is supported by reliable temporal evidence and assigns a temporal-confidence classification.

The resulting validation does not change the underlying expansion or contraction events. Instead, it determines how confidently those geographic changes can be interpreted.

```python
import pandas as pd
import numpy as np

print("Validating PMIP temporal evidence for geographic change")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_9_4_status = globals().get(
    "section_9_4_complete",
    False
)

print(
    f"Section 9.4 completion status: "
    f"{section_9_4_status}"
)

if not section_9_4_status:
    raise RuntimeError(
        "Section 9.5 requires Section 9.4 "
        "Cross-Market Momentum to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated cross-market momentum source
# ------------------------------------------------------------

momentum_candidate_names = [
    "track_cross_market_momentum_df",
]

track_momentum_df = None
track_momentum_source_name = None

required_momentum_columns = {
    "track_id",
    "validated_expansion_markets",
    "validated_market_exits",
    "net_market_movement",
    "geographic_expansion_index",
    "geographic_contraction_index",
    "cross_market_momentum_index",
    "cross_market_momentum_class",
}

for name in momentum_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_momentum_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_momentum_df = candidate.copy()
    track_momentum_source_name = name
    break


if track_momentum_df is None:
    raise RuntimeError(
        "Section 9.5 could not recover the validated "
        "cross-market momentum intelligence."
    )


# ------------------------------------------------------------
# 3. Recover track-country chronological history
# ------------------------------------------------------------

history_candidate_names = [
    "track_country_entry_df",
    "track_country_history_df",
]

track_country_history_df = None
track_country_history_source_name = None

required_history_columns = {
    "track_id",
    "country",
    "first_observed_date",
    "last_observed_date",
}

for name in history_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_history_columns.issubset(
        candidate.columns
    ):
        continue

    track_country_history_df = candidate.copy()
    track_country_history_source_name = name
    break


if track_country_history_df is None:
    raise RuntimeError(
        "Section 9.5 could not recover the validated "
        "track-country chronological history."
    )


# ------------------------------------------------------------
# 4. Recover country temporal-alignment table
# ------------------------------------------------------------

temporal_candidate_names = [
    "country_temporal_alignment_df",
    "temporal_country_alignment_df",
    "country_temporal_df",
]

country_temporal_df = None
country_temporal_source_name = None

for name in temporal_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if "country" not in candidate.columns:
        continue

    if len(candidate) != 76:
        continue

    country_temporal_df = candidate.copy()
    country_temporal_source_name = name
    break


if country_temporal_df is None:
    raise RuntimeError(
        "Section 9.5 could not recover the validated "
        "country temporal-alignment table."
    )


print()
print("Validated temporal-validation sources")
print("=" * 100)

print(
    f"Momentum source: "
    f"{track_momentum_source_name}"
)

print(
    f"Track-country history source: "
    f"{track_country_history_source_name}"
)

print(
    f"Country temporal source: "
    f"{country_temporal_source_name}"
)

print(
    f"Tracks: "
    f"{track_momentum_df['track_id'].nunique():,}"
)

print(
    f"Track-country histories: "
    f"{len(track_country_history_df):,}"
)

print(
    f"Country markets: "
    f"{country_temporal_df['country'].nunique():,}"
)


# ------------------------------------------------------------
# 5. Standardise temporal-support fields
# ------------------------------------------------------------

temporal_class_col = None

for column in [
    "temporal_alignment_class",
]:
    if column in country_temporal_df.columns:
        temporal_class_col = column
        break


comparison_caution_col = None

for column in [
    "direct_comparison_caution",
]:
    if column in country_temporal_df.columns:
        comparison_caution_col = column
        break


if temporal_class_col is None:
    raise RuntimeError(
        "Section 9.5 requires the country "
        "temporal_alignment_class field."
    )


# ------------------------------------------------------------
# 6. Assign temporal reliability scores by country
# ------------------------------------------------------------

temporal_score_map = {
    "Near-complete temporal alignment": 1.00,
    "Strong temporal alignment": 0.85,
    "Partial temporal alignment": 0.55,
    "Limited temporal alignment": 0.25,
}


country_temporal_df[
    "temporal_reliability_score"
] = (
    country_temporal_df[
        temporal_class_col
    ]
    .map(
        temporal_score_map
    )
)


if country_temporal_df[
    "temporal_reliability_score"
].isna().any():

    unknown_classes = (
        country_temporal_df
        .loc[
            country_temporal_df[
                "temporal_reliability_score"
            ].isna(),
            temporal_class_col,
        ]
        .dropna()
        .unique()
        .tolist()
    )

    raise RuntimeError(
        "Section 9.5 found unrecognised temporal "
        "alignment classes: "
        + ", ".join(
            map(
                str,
                unknown_classes,
            )
        )
    )


# ------------------------------------------------------------
# 7. Apply direct-comparison caution adjustment
# ------------------------------------------------------------

country_temporal_df[
    "comparison_safe_temporal_score"
] = (
    country_temporal_df[
        "temporal_reliability_score"
    ]
)


if comparison_caution_col is not None:

    country_temporal_df[
        "comparison_safe_temporal_score"
    ] = np.where(
        country_temporal_df[
            comparison_caution_col
        ]
        .fillna(False),

        country_temporal_df[
            "temporal_reliability_score"
        ]
        * 0.75,

        country_temporal_df[
            "temporal_reliability_score"
        ],
    )


country_temporal_df[
    "comparison_safe_temporal_score"
] = (
    country_temporal_df[
        "comparison_safe_temporal_score"
    ]
    .clip(
        lower=0,
        upper=1,
    )
)


# ------------------------------------------------------------
# 8. Attach temporal reliability to track-country histories
# ------------------------------------------------------------

temporal_merge_columns = [
    "country",
    temporal_class_col,
    "temporal_reliability_score",
    "comparison_safe_temporal_score",
]

if comparison_caution_col is not None:
    temporal_merge_columns.append(
        comparison_caution_col
    )


track_temporal_history_df = (
    track_country_history_df
    .merge(
        country_temporal_df[
            temporal_merge_columns
        ],
        on="country",
        how="left",
        validate="many_to_one",
    )
)


if track_temporal_history_df[
    "comparison_safe_temporal_score"
].isna().any():

    raise RuntimeError(
        "Section 9.5 could not assign temporal "
        "reliability to all track-country histories."
    )


# ------------------------------------------------------------
# 9. Build track-level temporal support summary
# ------------------------------------------------------------

track_temporal_support_df = (
    track_temporal_history_df
    .groupby(
        "track_id",
        as_index=False,
    )
    .agg(
        markets_with_temporal_evidence=(
            "country",
            "nunique",
        ),

        mean_temporal_reliability=(
            "temporal_reliability_score",
            "mean",
        ),

        median_temporal_reliability=(
            "temporal_reliability_score",
            "median",
        ),

        mean_comparison_safe_temporal_score=(
            "comparison_safe_temporal_score",
            "mean",
        ),

        minimum_temporal_reliability=(
            "comparison_safe_temporal_score",
            "min",
        ),
    )
)


# ------------------------------------------------------------
# 10. Count strongly supported markets
# ------------------------------------------------------------

track_temporal_history_df[
    "strong_temporal_support"
] = (
    track_temporal_history_df[
        "comparison_safe_temporal_score"
    ]
    >= 0.75
)


track_temporal_history_df[
    "moderate_or_strong_temporal_support"
] = (
    track_temporal_history_df[
        "comparison_safe_temporal_score"
    ]
    >= 0.50
)


track_temporal_counts_df = (
    track_temporal_history_df
    .groupby(
        "track_id",
        as_index=False,
    )
    .agg(
        strongly_supported_markets=(
            "strong_temporal_support",
            "sum",
        ),

        moderate_or_strong_supported_markets=(
            "moderate_or_strong_temporal_support",
            "sum",
        ),
    )
)


track_temporal_support_df = (
    track_temporal_support_df
    .merge(
        track_temporal_counts_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 11. Calculate temporal support percentages
# ------------------------------------------------------------

track_temporal_support_df[
    "strong_temporal_support_pct"
] = np.where(
    track_temporal_support_df[
        "markets_with_temporal_evidence"
    ] > 0,

    (
        track_temporal_support_df[
            "strongly_supported_markets"
        ]
        /
        track_temporal_support_df[
            "markets_with_temporal_evidence"
        ]
        * 100
    ),

    0.0,
)


track_temporal_support_df[
    "moderate_or_strong_temporal_support_pct"
] = np.where(
    track_temporal_support_df[
        "markets_with_temporal_evidence"
    ] > 0,

    (
        track_temporal_support_df[
            "moderate_or_strong_supported_markets"
        ]
        /
        track_temporal_support_df[
            "markets_with_temporal_evidence"
        ]
        * 100
    ),

    0.0,
)


# ------------------------------------------------------------
# 12. Merge temporal support with momentum intelligence
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in track_momentum_df.columns
]


track_temporal_validation_df = (
    track_momentum_df[
        identity_columns
        + [
            "validated_expansion_markets",
            "validated_market_exits",
            "net_market_movement",
            "geographic_expansion_index",
            "geographic_contraction_index",
            "raw_momentum_balance",
            "cross_market_momentum_index",
            "cross_market_momentum_class",
        ]
    ]
    .merge(
        track_temporal_support_df,
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 13. Build geographic-change temporal confidence index
# ------------------------------------------------------------

track_temporal_validation_df[
    "geographic_change_temporal_confidence_index"
] = (
    (
        0.50
        * track_temporal_validation_df[
            "mean_comparison_safe_temporal_score"
        ]
    )
    +
    (
        0.30
        * (
            track_temporal_validation_df[
                "moderate_or_strong_temporal_support_pct"
            ]
            /
            100
        )
    )
    +
    (
        0.20
        * (
            track_temporal_validation_df[
                "strong_temporal_support_pct"
            ]
            /
            100
        )
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 14. Temporal-confidence classification
# ------------------------------------------------------------

def classify_temporal_confidence(value):

    if value >= 0.85:
        return "Very high temporal confidence"

    if value >= 0.70:
        return "High temporal confidence"

    if value >= 0.50:
        return "Moderate temporal confidence"

    return "Limited temporal confidence"


track_temporal_validation_df[
    "temporal_confidence_class"
] = (
    track_temporal_validation_df[
        "geographic_change_temporal_confidence_index"
    ]
    .apply(
        classify_temporal_confidence
    )
)


# ------------------------------------------------------------
# 15. Build interpretation policy
# ------------------------------------------------------------

def classify_geographic_change_policy(row):

    confidence = row[
        "geographic_change_temporal_confidence_index"
    ]

    movement_class = row[
        "cross_market_momentum_class"
    ]

    if movement_class == "No validated movement":
        return "No geographic movement interpretation required"

    if confidence >= 0.70:
        return "Direct geographic change interpretation supported"

    if confidence >= 0.50:
        return "Interpret with temporal caution"

    return "Geographic change interpretation limited"


track_temporal_validation_df[
    "geographic_change_interpretation_policy"
] = (
    track_temporal_validation_df
    .apply(
        classify_geographic_change_policy,
        axis=1,
    )
)


# ------------------------------------------------------------
# 16. Overview
# ------------------------------------------------------------

print()
print("Temporal validation of geographic change overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_temporal_validation_df):,}"
)

print(
    f"Median markets with temporal evidence: "
    f"{track_temporal_validation_df['markets_with_temporal_evidence'].median():.1f}"
)

print(
    f"Median temporal reliability: "
    f"{track_temporal_validation_df['median_temporal_reliability'].median():.4f}"
)

print(
    f"Median strong temporal support share: "
    f"{track_temporal_validation_df['strong_temporal_support_pct'].median():.2f}%"
)

print(
    f"Median geographic-change temporal confidence index: "
    f"{track_temporal_validation_df['geographic_change_temporal_confidence_index'].median():.4f}"
)

print(
    f"Mean geographic-change temporal confidence index: "
    f"{track_temporal_validation_df['geographic_change_temporal_confidence_index'].mean():.4f}"
)


# ------------------------------------------------------------
# 17. Temporal-confidence summary
# ------------------------------------------------------------

confidence_order = [
    "Very high temporal confidence",
    "High temporal confidence",
    "Moderate temporal confidence",
    "Limited temporal confidence",
]


track_temporal_confidence_summary_df = (
    track_temporal_validation_df[
        "temporal_confidence_class"
    ]
    .value_counts()
    .reindex(
        confidence_order,
        fill_value=0,
    )
    .rename_axis(
        "Temporal Confidence Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_temporal_confidence_summary_df[
    "Share (%)"
] = (
    track_temporal_confidence_summary_df[
        "Tracks"
    ]
    /
    len(track_temporal_validation_df)
    * 100
)


print()
print("Geographic change temporal-confidence classification")
print("=" * 100)

display(
    track_temporal_confidence_summary_df
)


# ------------------------------------------------------------
# 18. Interpretation-policy summary
# ------------------------------------------------------------

policy_order = [
    "Direct geographic change interpretation supported",
    "Interpret with temporal caution",
    "Geographic change interpretation limited",
    "No geographic movement interpretation required",
]


track_geographic_change_policy_summary_df = (
    track_temporal_validation_df[
        "geographic_change_interpretation_policy"
    ]
    .value_counts()
    .reindex(
        policy_order,
        fill_value=0,
    )
    .rename_axis(
        "Interpretation Policy"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_change_policy_summary_df[
    "Share (%)"
] = (
    track_geographic_change_policy_summary_df[
        "Tracks"
    ]
    /
    len(track_temporal_validation_df)
    * 100
)


print()
print("Geographic change interpretation policy")
print("=" * 100)

display(
    track_geographic_change_policy_summary_df
)


# ------------------------------------------------------------
# 19. Tracks with strongest temporally supported movement
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "net_market_movement",
        "cross_market_momentum_index",
        "cross_market_momentum_class",
        "mean_comparison_safe_temporal_score",
        "strong_temporal_support_pct",
        "geographic_change_temporal_confidence_index",
        "temporal_confidence_class",
        "geographic_change_interpretation_policy",
    ]
    if column in track_temporal_validation_df.columns
]


print()
print("Tracks with strongest temporally supported geographic movement")
print("=" * 100)

strongest_temporally_supported_change_df = (
    track_temporal_validation_df
    .loc[
        track_temporal_validation_df[
            "cross_market_momentum_class"
        ]
        !=
        "No validated movement"
    ]
    .sort_values(
        [
            "geographic_change_temporal_confidence_index",
            "cross_market_momentum_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    strongest_temporally_supported_change_df
)


# ------------------------------------------------------------
# 20. Diagnostics
# ------------------------------------------------------------

track_temporal_validation_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with temporal confidence >= 0.85",

            "Observed Value":
                int(
                    (
                        track_temporal_validation_df[
                            "geographic_change_temporal_confidence_index"
                        ]
                        >= 0.85
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with temporal confidence >= 0.70",

            "Observed Value":
                int(
                    (
                        track_temporal_validation_df[
                            "geographic_change_temporal_confidence_index"
                        ]
                        >= 0.70
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks requiring temporal caution",

            "Observed Value":
                int(
                    (
                        track_temporal_validation_df[
                            "geographic_change_interpretation_policy"
                        ]
                        ==
                        "Interpret with temporal caution"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with limited geographic-change interpretation",

            "Observed Value":
                int(
                    (
                        track_temporal_validation_df[
                            "geographic_change_interpretation_policy"
                        ]
                        ==
                        "Geographic change interpretation limited"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with direct geographic-change interpretation support",

            "Observed Value":
                int(
                    (
                        track_temporal_validation_df[
                            "geographic_change_interpretation_policy"
                        ]
                        ==
                        "Direct geographic change interpretation supported"
                    ).sum()
                ),
        },
    ]
)


print()
print("Temporal validation diagnostics")
print("=" * 100)

display(
    track_temporal_validation_diagnostics_df
)


# ------------------------------------------------------------
# 21. Final validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_temporal_validation_df
    )
    == 110_198
)


identity_valid = bool(
    track_temporal_validation_df[
        "track_id"
    ].nunique()
    == 110_198
)


temporal_scores_valid = bool(
    np.isfinite(
        track_temporal_validation_df[
            [
                "mean_temporal_reliability",
                "median_temporal_reliability",
                "mean_comparison_safe_temporal_score",
                "minimum_temporal_reliability",
                "geographic_change_temporal_confidence_index",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


temporal_scores_bounded = bool(
    track_temporal_validation_df[
        [
            "mean_temporal_reliability",
            "median_temporal_reliability",
            "mean_comparison_safe_temporal_score",
            "minimum_temporal_reliability",
            "geographic_change_temporal_confidence_index",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


percentage_valid = bool(
    track_temporal_validation_df[
        [
            "strong_temporal_support_pct",
            "moderate_or_strong_temporal_support_pct",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                100,
                inclusive="both",
            ).all()
    )
    .all()
)


confidence_complete = bool(
    track_temporal_validation_df[
        "temporal_confidence_class"
    ]
    .notna()
    .all()
)


policy_complete = bool(
    track_temporal_validation_df[
        "geographic_change_interpretation_policy"
    ]
    .notna()
    .all()
)


confidence_reconciled = bool(
    int(
        track_temporal_confidence_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_temporal_validation_df
    )
)


policy_reconciled = bool(
    int(
        track_geographic_change_policy_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_temporal_validation_df
    )
)


section_9_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 9.4 completion",

            "Requirement":
                "Cross-Market Momentum must already be complete.",

            "Observed Evidence":
                f"Section 9.4 completion status: "
                f"{section_9_4_status}",

            "Passed":
                section_9_4_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_temporal_validation_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one temporal-validation record.",

            "Observed Evidence":
                f"{track_temporal_validation_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Temporal metric validity",

            "Requirement":
                "Temporal reliability and confidence "
                "metrics must remain finite.",

            "Observed Evidence":
                "Temporal reliability and confidence "
                "metrics assessed for all tracks",

            "Passed":
                temporal_scores_valid,
        },
        {
            "Validation Area":
                "Temporal metric bounds",

            "Requirement":
                "Temporal reliability and confidence "
                "indices must remain between 0 and 1.",

            "Observed Evidence":
                "Temporal indices assessed for all tracks",

            "Passed":
                temporal_scores_bounded,
        },
        {
            "Validation Area":
                "Temporal percentage bounds",

            "Requirement":
                "Temporal support percentages must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Temporal support percentages assessed "
                "for all tracks",

            "Passed":
                percentage_valid,
        },
        {
            "Validation Area":
                "Confidence classification completeness",

            "Requirement":
                "Every track must receive a temporal "
                "confidence classification.",

            "Observed Evidence":
                f"{track_temporal_validation_df['temporal_confidence_class'].notna().sum():,} "
                "confidence classes assigned",

            "Passed":
                confidence_complete,
        },
        {
            "Validation Area":
                "Interpretation policy completeness",

            "Requirement":
                "Every track must receive a geographic-change "
                "interpretation policy.",

            "Observed Evidence":
                f"{track_temporal_validation_df['geographic_change_interpretation_policy'].notna().sum():,} "
                "interpretation policies assigned",

            "Passed":
                policy_complete,
        },
        {
            "Validation Area":
                "Confidence population reconciliation",

            "Requirement":
                "Temporal-confidence classifications must "
                "reconcile with the full track population.",

            "Observed Evidence":
                f"{int(track_temporal_confidence_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_temporal_validation_df):,} tracks",

            "Passed":
                confidence_reconciled,
        },
        {
            "Validation Area":
                "Policy population reconciliation",

            "Requirement":
                "Interpretation policies must reconcile "
                "with the full track population.",

            "Observed Evidence":
                f"{int(track_geographic_change_policy_summary_df['Tracks'].sum()):,} "
                f"policy records for "
                f"{len(track_temporal_validation_df):,} tracks",

            "Passed":
                policy_reconciled,
        },
    ]
)


print()
print("Temporal validation of geographic change")
print("=" * 100)

display(
    section_9_5_validation_df
)


# ------------------------------------------------------------
# 22. Final completion status
# ------------------------------------------------------------

section_9_5_complete = bool(
    section_9_5_validation_df[
        "Passed"
    ].all()
)

section_9_complete = bool(
    section_9_1_complete
    and
    section_9_2_complete
    and
    section_9_3_complete
    and
    section_9_4_complete
    and
    section_9_5_complete
)


if not section_9_5_complete:

    failed_checks = (
        section_9_5_validation_df
        .loc[
            ~section_9_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 9.5 temporal validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 9.5 temporal validation of "
    "geographic change checks passed."
)

print(
    f"Section 9.5 completion status: "
    f"{section_9_5_complete}"
)

print(
    f"Section 9 overall completion status: "
    f"{section_9_complete}"
)

print(
    f"Temporal geographic-change validation prepared "
    f"for {len(track_temporal_validation_df):,} tracks."
)

print(
    "Geographic movement signals now carry explicit "
    "temporal-confidence and interpretation safeguards."
)

print(
    "Geographic Expansion and Market Movement "
    "analysis is complete."
)
```

## 10. Geographic Intelligence Scoring and Classification

### Purpose

Convert the validated geographic intelligence developed in Sections 5 to 9 into structured PMIP scoring measures that can support consistent comparison between tracks.

This section transforms descriptive geographic indicators into normalized scores covering four major dimensions:

- geographic reach,
- market penetration,
- geographic concentration, and
- geographic expansion.

Each score is designed to remain interpretable on a common 0-to-1 scale, where higher values represent stronger performance within that specific geographic dimension.

The scoring framework retains the evidence-quality and temporal safeguards established earlier in the notebook. These scores therefore summarize validated geographic intelligence rather than replacing the underlying analytical measures.

The final subsection combines the individual dimensions into a geographic intelligence summary without treating the result as a causal explanation of commercial success.




### 10.1 Geographic Reach Score

#### Purpose

Create a normalized geographic reach score for each track by combining the breadth of observed country coverage with the strength of international evidence already validated in Section 8.1.

The score rewards tracks that appear across a larger proportion of the available PMIP country markets while also accounting for the quality of supporting international evidence.

The resulting **Geographic Reach Score** is bounded between 0 and 1, where values closer to 1 represent broader and better-supported international reach.

```python
import pandas as pd
import numpy as np

print("Scoring PMIP geographic reach")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_9_status = globals().get(
    "section_9_complete",
    False
)

print(
    f"Section 9 overall completion status: "
    f"{section_9_status}"
)

if not section_9_status:
    raise RuntimeError(
        "Section 10.1 requires Section 9 "
        "Geographic Expansion and Market Movement "
        "to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track international reach source
# ------------------------------------------------------------

reach_candidate_names = [
    "track_international_reach_df",
]

track_reach_df = None
track_reach_source_name = None

required_reach_columns = {
    "track_id",
    "markets_reached",
    "international_market_reach_pct",
    "international_reach_index",
    "international_reach_class",
    "international_evidence_class",
}

for name in reach_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_reach_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_reach_df = candidate.copy()
    track_reach_source_name = name
    break


if track_reach_df is None:
    raise RuntimeError(
        "Section 10.1 could not recover the validated "
        "track international-reach intelligence."
    )


print()
print("Validated geographic reach source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_reach_source_name}"
)

print(
    f"Tracks: "
    f"{track_reach_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    "markets_reached",
    "international_market_reach_pct",
    "international_reach_index",
]

for column in numeric_columns:

    track_reach_df[
        column
    ] = pd.to_numeric(
        track_reach_df[
            column
        ],
        errors="coerce",
    )


if track_reach_df[
    numeric_columns
].isna().any().any():

    raise RuntimeError(
        "Section 10.1 found missing or invalid "
        "numeric geographic reach values."
    )


# ------------------------------------------------------------
# 4. International evidence weighting
# ------------------------------------------------------------

evidence_weight_map = {
    "Local-only evidence": 0.25,
    "Limited international evidence": 0.50,
    "Moderate international evidence": 0.75,
    "Strong international evidence": 1.00,
}


track_reach_df[
    "international_evidence_weight"
] = (
    track_reach_df[
        "international_evidence_class"
    ]
    .map(
        evidence_weight_map
    )
)


if track_reach_df[
    "international_evidence_weight"
].isna().any():

    unknown_evidence_classes = (
        track_reach_df
        .loc[
            track_reach_df[
                "international_evidence_weight"
            ].isna(),
            "international_evidence_class",
        ]
        .dropna()
        .unique()
        .tolist()
    )

    raise RuntimeError(
        "Section 10.1 found unrecognised international "
        "evidence classes: "
        + ", ".join(
            map(
                str,
                unknown_evidence_classes,
            )
        )
    )


# ------------------------------------------------------------
# 5. Build geographic reach score
# ------------------------------------------------------------

# Reach breadth is the dominant component.
# Evidence strength moderates confidence in that breadth.

track_reach_df[
    "geographic_reach_score"
] = (
    (
        0.80
        * track_reach_df[
            "international_reach_index"
        ]
    )
    +
    (
        0.20
        * track_reach_df[
            "international_evidence_weight"
        ]
        * track_reach_df[
            "international_reach_index"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 6. Geographic reach score classification
# ------------------------------------------------------------

def classify_geographic_reach_score(value):

    if value >= 0.75:
        return "Exceptional geographic reach"

    if value >= 0.50:
        return "Strong geographic reach"

    if value >= 0.25:
        return "Moderate geographic reach"

    if value > 0:
        return "Limited geographic reach"

    return "Minimal geographic reach"


track_reach_df[
    "geographic_reach_score_class"
] = (
    track_reach_df[
        "geographic_reach_score"
    ]
    .apply(
        classify_geographic_reach_score
    )
)


# ------------------------------------------------------------
# 7. Final score table
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in track_reach_df.columns
]


track_geographic_reach_score_df = (
    track_reach_df[
        identity_columns
        + [
            "markets_reached",
            "international_market_reach_pct",
            "international_reach_index",
            "international_reach_class",
            "international_evidence_class",
            "international_evidence_weight",
            "geographic_reach_score",
            "geographic_reach_score_class",
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 8. Overview
# ------------------------------------------------------------

print()
print("Geographic reach scoring overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_reach_score_df):,}"
)

print(
    f"Median markets reached: "
    f"{track_geographic_reach_score_df['markets_reached'].median():.1f}"
)

print(
    f"Median international reach index: "
    f"{track_geographic_reach_score_df['international_reach_index'].median():.4f}"
)

print(
    f"Median geographic reach score: "
    f"{track_geographic_reach_score_df['geographic_reach_score'].median():.4f}"
)

print(
    f"Mean geographic reach score: "
    f"{track_geographic_reach_score_df['geographic_reach_score'].mean():.4f}"
)


# ------------------------------------------------------------
# 9. Score classification summary
# ------------------------------------------------------------

reach_score_class_order = [
    "Exceptional geographic reach",
    "Strong geographic reach",
    "Moderate geographic reach",
    "Limited geographic reach",
    "Minimal geographic reach",
]


track_geographic_reach_score_summary_df = (
    track_geographic_reach_score_df[
        "geographic_reach_score_class"
    ]
    .value_counts()
    .reindex(
        reach_score_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Reach Score Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_reach_score_summary_df[
    "Share (%)"
] = (
    track_geographic_reach_score_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_reach_score_df)
    * 100
)


print()
print("Geographic reach score classification")
print("=" * 100)

display(
    track_geographic_reach_score_summary_df
)


# ------------------------------------------------------------
# 10. Highest geographic reach scores
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "markets_reached",
        "international_market_reach_pct",
        "international_reach_index",
        "international_evidence_class",
        "geographic_reach_score",
        "geographic_reach_score_class",
    ]
    if column in track_geographic_reach_score_df.columns
]


print()
print("Tracks with highest geographic reach scores")
print("=" * 100)

highest_geographic_reach_scores_df = (
    track_geographic_reach_score_df
    .sort_values(
        [
            "geographic_reach_score",
            "markets_reached",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    highest_geographic_reach_scores_df
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

track_geographic_reach_score_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with reach score >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_reach_score_df[
                            "geographic_reach_score"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with reach score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_reach_score_df[
                            "geographic_reach_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks reaching at least 10 markets",

            "Observed Value":
                int(
                    (
                        track_geographic_reach_score_df[
                            "markets_reached"
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks reaching at least 25 markets",

            "Observed Value":
                int(
                    (
                        track_geographic_reach_score_df[
                            "markets_reached"
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with strong international evidence",

            "Observed Value":
                int(
                    (
                        track_geographic_reach_score_df[
                            "international_evidence_class"
                        ]
                        ==
                        "Strong international evidence"
                    ).sum()
                ),
        },
    ]
)


print()
print("Geographic reach score diagnostics")
print("=" * 100)

display(
    track_geographic_reach_score_diagnostics_df
)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_reach_score_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_reach_score_df[
        "track_id"
    ].nunique()
    == 110_198
)


score_valid = bool(
    np.isfinite(
        track_geographic_reach_score_df[
            [
                "international_reach_index",
                "international_evidence_weight",
                "geographic_reach_score",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


score_bounds_valid = bool(
    track_geographic_reach_score_df[
        [
            "international_reach_index",
            "international_evidence_weight",
            "geographic_reach_score",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    track_geographic_reach_score_df[
        "geographic_reach_score_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_reach_score_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_reach_score_df
    )
)


section_10_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 9 completion",

            "Requirement":
                "Geographic Expansion and Market Movement "
                "must already be complete.",

            "Observed Evidence":
                f"Section 9 overall completion status: "
                f"{section_9_status}",

            "Passed":
                section_9_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_reach_score_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one geographic "
                "reach score.",

            "Observed Evidence":
                f"{track_geographic_reach_score_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Reach score numeric validity",

            "Requirement":
                "Reach score inputs and outputs must "
                "remain finite.",

            "Observed Evidence":
                "Reach indices, evidence weights and scores "
                "assessed for all tracks",

            "Passed":
                score_valid,
        },
        {
            "Validation Area":
                "Reach score bounds",

            "Requirement":
                "Reach indices, weights and final scores "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded reach scoring fields assessed "
                "for all tracks",

            "Passed":
                score_bounds_valid,
        },
        {
            "Validation Area":
                "Reach classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "reach score classification.",

            "Observed Evidence":
                f"{track_geographic_reach_score_df['geographic_reach_score_class'].notna().sum():,} "
                "reach score classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Reach population reconciliation",

            "Requirement":
                "Reach score classifications must reconcile "
                "with the full track population.",

            "Observed Evidence":
                f"{int(track_geographic_reach_score_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_reach_score_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Geographic reach score validation")
print("=" * 100)

display(
    section_10_1_validation_df
)


# ------------------------------------------------------------
# 13. Final completion status
# ------------------------------------------------------------

section_10_1_complete = bool(
    section_10_1_validation_df[
        "Passed"
    ].all()
)


if not section_10_1_complete:

    failed_checks = (
        section_10_1_validation_df
        .loc[
            ~section_10_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 10.1 Geographic Reach Score "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 10.1 geographic reach "
    "score validation checks passed."
)

print(
    f"Section 10.1 completion status: "
    f"{section_10_1_complete}"
)

print(
    f"Geographic reach scores prepared "
    f"for {len(track_geographic_reach_score_df):,} tracks."
)

print(
    "The workflow is ready for "
    "Section 10.2 Market Penetration Score."
)
```

#### Interpretation

Section 10.1 successfully converts the validated international-reach intelligence into a normalized **Geographic Reach Score** for all **110,198 tracks**.

The results show that the overall track population remains heavily concentrated in a small number of country markets. The median track reaches only **1 market**, with a median international reach index of **0.0132** and a median geographic reach score of **0.0112**.

The mean geographic reach score is slightly higher at **0.0396**, showing that a relatively small group of tracks achieves substantially broader international reach and pulls the average upward.

The score classification shows:

- **577 tracks (0.52%)** have **Exceptional geographic reach**
- **1,257 tracks (1.14%)** have **Strong geographic reach**
- **2,007 tracks (1.82%)** have **Moderate geographic reach**
- **106,357 tracks (96.51%)** have **Limited geographic reach**
- **0 tracks** are classified as **Minimal geographic reach**

The dominance of the Limited Geographic Reach category is consistent with the earlier PMIP findings showing that most tracks are observed in only one or a small number of country markets.

The diagnostics provide additional context. A total of **7,288 tracks** reach at least **10 markets**, while **3,148 tracks** reach at least **25 markets**. The same **3,148 tracks** also carry **Strong international evidence**, indicating that the widest-reaching tracks generally have stronger supporting evidence across multiple markets.

Only **1,834 tracks** achieve a Geographic Reach Score of at least **0.50**, and just **577 tracks** achieve a score of at least **0.75**. This confirms that genuinely broad and strongly supported geographic reach is relatively rare in the overall track population.

The Geographic Reach Score combines two related elements:

- **International reach breadth**, which measures how much of the available country-market space a track reaches.
- **International evidence strength**, which adjusts the score according to the quality of supporting geographic evidence.

The reach component remains dominant in the scoring formula, while the evidence weighting acts as a supporting confidence adjustment. This means a track cannot obtain a high geographic reach score from evidence quality alone; it must first demonstrate genuinely broad market coverage.

All Section 10.1 validation checks passed. The complete **110,198-track population** was preserved, track identities remained unique, all scoring inputs and outputs were finite, every scoring field remained between **0 and 1**, every track received a geographic reach classification, and the classification totals reconciled correctly with the full population.

Section 10.1 therefore establishes the first normalized scoring dimension in PMIP's Geographic Intelligence Scoring framework.

The workflow is ready for **Section 10.2 Market Penetration Score**.

### 10.2 Market Penetration Score

#### Purpose

Create a normalized Market Penetration Score for each track using the validated market-penetration intelligence developed in Section 8.2.

The score measures how strongly a track performs within the markets it has already reached by combining relative stream intensity with the proportion of markets showing strong support and above-average performance.

The resulting Market Penetration Score is bounded between 0 and 1, where values closer to 1 represent stronger depth of performance within observed country markets.

This score measures market depth rather than geographic breadth, so it complements the Geographic Reach Score created in Section 10.1.

```python
import pandas as pd
import numpy as np

print("Scoring PMIP market penetration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_10_1_status = globals().get(
    "section_10_1_complete",
    False
)

print(
    f"Section 10.1 completion status: "
    f"{section_10_1_status}"
)

if not section_10_1_status:
    raise RuntimeError(
        "Section 10.2 requires Section 10.1 "
        "Geographic Reach Score to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track market-penetration source
# ------------------------------------------------------------

penetration_candidate_names = [
    "track_market_penetration_df",
    "track_penetration_features_df",
]

track_penetration_df = None
track_penetration_source_name = None

required_penetration_columns = {
    "track_id",
    "markets_observed",
    "median_relative_stream_intensity",
    "support_qualified_market_pct",
    "above_country_average_market_pct",
    "market_penetration_index",
    "market_penetration_class",
}

for name in penetration_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_penetration_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_penetration_df = candidate.copy()
    track_penetration_source_name = name
    break


if track_penetration_df is None:
    raise RuntimeError(
        "Section 10.2 could not recover the validated "
        "track market-penetration intelligence."
    )


print()
print("Validated market penetration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_penetration_source_name}"
)

print(
    f"Tracks: "
    f"{track_penetration_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    "markets_observed",
    "median_relative_stream_intensity",
    "support_qualified_market_pct",
    "above_country_average_market_pct",
    "market_penetration_index",
]

for column in numeric_columns:

    track_penetration_df[
        column
    ] = pd.to_numeric(
        track_penetration_df[
            column
        ],
        errors="coerce",
    )


if track_penetration_df[
    numeric_columns
].isna().any().any():

    raise RuntimeError(
        "Section 10.2 found missing or invalid "
        "numeric market-penetration values."
    )


# ------------------------------------------------------------
# 4. Normalize relative stream intensity
# ------------------------------------------------------------

# Use a capped transformation so very large intensity values
# do not dominate the score.
#
# 0.0  -> 0
# 1.0  -> 0.5
# 2.0+ -> approaches / reaches 1

track_penetration_df[
    "relative_intensity_score"
] = (
    track_penetration_df[
        "median_relative_stream_intensity"
    ]
    /
    (
        track_penetration_df[
            "median_relative_stream_intensity"
        ]
        + 1.0
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 5. Normalize percentage support fields
# ------------------------------------------------------------

track_penetration_df[
    "support_qualified_score"
] = (
    track_penetration_df[
        "support_qualified_market_pct"
    ]
    /
    100
).clip(
    lower=0,
    upper=1,
)


track_penetration_df[
    "above_average_market_score"
] = (
    track_penetration_df[
        "above_country_average_market_pct"
    ]
    /
    100
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 6. Build Market Penetration Score
# ------------------------------------------------------------

# Existing market_penetration_index is retained as the
# strongest component because it already summarizes validated
# penetration behaviour.
#
# Relative intensity adds depth.
# Support-qualified and above-average market shares add
# cross-market consistency of penetration.

track_penetration_df[
    "market_penetration_score"
] = (
    (
        0.50
        * track_penetration_df[
            "market_penetration_index"
        ]
    )
    +
    (
        0.25
        * track_penetration_df[
            "relative_intensity_score"
        ]
    )
    +
    (
        0.15
        * track_penetration_df[
            "support_qualified_score"
        ]
    )
    +
    (
        0.10
        * track_penetration_df[
            "above_average_market_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 7. Market penetration score classification
# ------------------------------------------------------------

def classify_market_penetration_score(value):

    if value >= 0.75:
        return "Exceptional market penetration"

    if value >= 0.50:
        return "Strong market penetration"

    if value >= 0.25:
        return "Moderate market penetration"

    if value > 0:
        return "Limited market penetration"

    return "Minimal market penetration"


track_penetration_df[
    "market_penetration_score_class"
] = (
    track_penetration_df[
        "market_penetration_score"
    ]
    .apply(
        classify_market_penetration_score
    )
)


# ------------------------------------------------------------
# 8. Final score table
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in track_penetration_df.columns
]


track_market_penetration_score_df = (
    track_penetration_df[
        identity_columns
        + [
            "markets_observed",
            "median_relative_stream_intensity",
            "support_qualified_market_pct",
            "above_country_average_market_pct",
            "market_penetration_index",
            "market_penetration_class",
            "relative_intensity_score",
            "support_qualified_score",
            "above_average_market_score",
            "market_penetration_score",
            "market_penetration_score_class",
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 9. Overview
# ------------------------------------------------------------

print()
print("Market penetration scoring overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_market_penetration_score_df):,}"
)

print(
    f"Median markets observed: "
    f"{track_market_penetration_score_df['markets_observed'].median():.1f}"
)

print(
    f"Median relative stream intensity: "
    f"{track_market_penetration_score_df['median_relative_stream_intensity'].median():.4f}"
)

print(
    f"Median existing penetration index: "
    f"{track_market_penetration_score_df['market_penetration_index'].median():.4f}"
)

print(
    f"Median Market Penetration Score: "
    f"{track_market_penetration_score_df['market_penetration_score'].median():.4f}"
)

print(
    f"Mean Market Penetration Score: "
    f"{track_market_penetration_score_df['market_penetration_score'].mean():.4f}"
)


# ------------------------------------------------------------
# 10. Score classification summary
# ------------------------------------------------------------

penetration_score_class_order = [
    "Exceptional market penetration",
    "Strong market penetration",
    "Moderate market penetration",
    "Limited market penetration",
    "Minimal market penetration",
]


track_market_penetration_score_summary_df = (
    track_market_penetration_score_df[
        "market_penetration_score_class"
    ]
    .value_counts()
    .reindex(
        penetration_score_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Market Penetration Score Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_market_penetration_score_summary_df[
    "Share (%)"
] = (
    track_market_penetration_score_summary_df[
        "Tracks"
    ]
    /
    len(track_market_penetration_score_df)
    * 100
)


print()
print("Market penetration score classification")
print("=" * 100)

display(
    track_market_penetration_score_summary_df
)


# ------------------------------------------------------------
# 11. Highest Market Penetration Scores
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "markets_observed",
        "median_relative_stream_intensity",
        "support_qualified_market_pct",
        "above_country_average_market_pct",
        "market_penetration_index",
        "market_penetration_score",
        "market_penetration_score_class",
    ]
    if column in track_market_penetration_score_df.columns
]


print()
print("Tracks with highest Market Penetration Scores")
print("=" * 100)

highest_market_penetration_scores_df = (
    track_market_penetration_score_df
    .sort_values(
        [
            "market_penetration_score",
            "median_relative_stream_intensity",
            "support_qualified_market_pct",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    highest_market_penetration_scores_df
)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

track_market_penetration_score_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with penetration score >= 0.75",

            "Observed Value":
                int(
                    (
                        track_market_penetration_score_df[
                            "market_penetration_score"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with penetration score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_market_penetration_score_df[
                            "market_penetration_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with median relative intensity >= 1.0",

            "Observed Value":
                int(
                    (
                        track_market_penetration_score_df[
                            "median_relative_stream_intensity"
                        ]
                        >= 1.0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with >= 50% support-qualified markets",

            "Observed Value":
                int(
                    (
                        track_market_penetration_score_df[
                            "support_qualified_market_pct"
                        ]
                        >= 50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with >= 50% above-average markets",

            "Observed Value":
                int(
                    (
                        track_market_penetration_score_df[
                            "above_country_average_market_pct"
                        ]
                        >= 50
                    ).sum()
                ),
        },
    ]
)


print()
print("Market penetration score diagnostics")
print("=" * 100)

display(
    track_market_penetration_score_diagnostics_df
)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_market_penetration_score_df
    )
    == 110_198
)


identity_valid = bool(
    track_market_penetration_score_df[
        "track_id"
    ].nunique()
    == 110_198
)


score_fields = [
    "market_penetration_index",
    "relative_intensity_score",
    "support_qualified_score",
    "above_average_market_score",
    "market_penetration_score",
]


score_valid = bool(
    np.isfinite(
        track_market_penetration_score_df[
            score_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


score_bounds_valid = bool(
    track_market_penetration_score_df[
        score_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


percentage_valid = bool(
    track_market_penetration_score_df[
        [
            "support_qualified_market_pct",
            "above_country_average_market_pct",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                100,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    track_market_penetration_score_df[
        "market_penetration_score_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_market_penetration_score_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_market_penetration_score_df
    )
)


section_10_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 10.1 completion",

            "Requirement":
                "Geographic Reach Score must already be complete.",

            "Observed Evidence":
                f"Section 10.1 completion status: "
                f"{section_10_1_status}",

            "Passed":
                section_10_1_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_market_penetration_score_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one market "
                "penetration score.",

            "Observed Evidence":
                f"{track_market_penetration_score_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Penetration score numeric validity",

            "Requirement":
                "Penetration scoring inputs and outputs "
                "must remain finite.",

            "Observed Evidence":
                f"{len(score_fields)} normalized penetration "
                "score fields assessed",

            "Passed":
                score_valid,
        },
        {
            "Validation Area":
                "Penetration score bounds",

            "Requirement":
                "All normalized penetration score fields "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded market penetration scoring fields "
                "assessed for all tracks",

            "Passed":
                score_bounds_valid,
        },
        {
            "Validation Area":
                "Penetration percentage bounds",

            "Requirement":
                "Market support percentages must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Support-qualified and above-average "
                "percentages assessed",

            "Passed":
                percentage_valid,
        },
        {
            "Validation Area":
                "Penetration classification completeness",

            "Requirement":
                "Every track must receive a market "
                "penetration score classification.",

            "Observed Evidence":
                f"{track_market_penetration_score_df['market_penetration_score_class'].notna().sum():,} "
                "penetration score classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Penetration population reconciliation",

            "Requirement":
                "Penetration score classifications must "
                "reconcile with the full track population.",

            "Observed Evidence":
                f"{int(track_market_penetration_score_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_market_penetration_score_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Market penetration score validation")
print("=" * 100)

display(
    section_10_2_validation_df
)


# ------------------------------------------------------------
# 14. Final completion status
# ------------------------------------------------------------

section_10_2_complete = bool(
    section_10_2_validation_df[
        "Passed"
    ].all()
)


if not section_10_2_complete:

    failed_checks = (
        section_10_2_validation_df
        .loc[
            ~section_10_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 10.2 Market Penetration Score "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 10.2 Market Penetration Score "
    "validation checks passed."
)

print(
    f"Section 10.2 completion status: "
    f"{section_10_2_complete}"
)

print(
    f"Market Penetration Scores prepared "
    f"for {len(track_market_penetration_score_df):,} tracks."
)

print(
    "The workflow is ready for "
    "Section 10.3 Geographic Concentration Score."
)
```

### 10.3 Geographic Concentration Score

#### Purpose

Create a normalized Geographic Concentration Score for each track using the validated geographic-concentration intelligence developed in Section 8.4.

The score measures how dependent a track's streaming activity is on a small number of country markets by combining stream concentration, strongest-market dominance, and effective market diversification.

Because high concentration indicates greater geographic dependency, the scoring direction is reversed so that higher values represent healthier geographic diversification and lower dependence on a small number of markets.

The resulting Geographic Concentration Score is bounded between 0 and 1, where values closer to 1 represent more geographically diversified performance and values closer to 0 represent highly concentrated market dependence.

```python
import pandas as pd
import numpy as np

print("Scoring PMIP geographic concentration")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_10_2_status = globals().get(
    "section_10_2_complete",
    False
)

print(
    f"Section 10.2 completion status: "
    f"{section_10_2_status}"
)

if not section_10_2_status:
    raise RuntimeError(
        "Section 10.3 requires Section 10.2 "
        "Market Penetration Score to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated track geographic-concentration source
# ------------------------------------------------------------

concentration_candidate_names = [
    "track_geographic_concentration_df",
    "track_stream_concentration_df",
]

track_concentration_df = None
track_concentration_source_name = None

required_concentration_columns = {
    "track_id",
    "top_1_stream_share_pct",
    "stream_concentration_hhi",
    "effective_stream_markets",
}

for name in concentration_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_concentration_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_concentration_df = candidate.copy()
    track_concentration_source_name = name
    break


if track_concentration_df is None:
    raise RuntimeError(
        "Section 10.3 could not recover the validated "
        "track geographic-concentration intelligence."
    )


print()
print("Validated geographic concentration source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{track_concentration_source_name}"
)

print(
    f"Tracks: "
    f"{track_concentration_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    "top_1_stream_share_pct",
    "stream_concentration_hhi",
    "effective_stream_markets",
]

for column in numeric_columns:

    track_concentration_df[
        column
    ] = pd.to_numeric(
        track_concentration_df[
            column
        ],
        errors="coerce",
    )


if track_concentration_df[
    numeric_columns
].isna().any().any():

    raise RuntimeError(
        "Section 10.3 found missing or invalid "
        "numeric geographic-concentration values."
    )


# ------------------------------------------------------------
# 4. Build inverse HHI diversification score
# ------------------------------------------------------------

track_concentration_df[
    "inverse_hhi_score"
] = (
    1
    -
    track_concentration_df[
        "stream_concentration_hhi"
    ]
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 5. Build strongest-market diversification score
# ------------------------------------------------------------

track_concentration_df[
    "strongest_market_diversification_score"
] = (
    1
    -
    (
        track_concentration_df[
            "top_1_stream_share_pct"
        ]
        /
        100
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 6. Normalize effective stream markets
# ------------------------------------------------------------

max_effective_markets = float(
    track_concentration_df[
        "effective_stream_markets"
    ]
    .quantile(
        0.99
    )
)


if (
    not np.isfinite(max_effective_markets)
    or
    max_effective_markets <= 1
):
    max_effective_markets = float(
        track_concentration_df[
            "effective_stream_markets"
        ]
        .max()
    )


if (
    not np.isfinite(max_effective_markets)
    or
    max_effective_markets <= 1
):
    max_effective_markets = 1.0


track_concentration_df[
    "effective_market_diversification_score"
] = (
    (
        track_concentration_df[
            "effective_stream_markets"
        ]
        - 1
    )
    /
    max(
        max_effective_markets - 1,
        1,
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 7. Optional existing diversification index
# ------------------------------------------------------------

diversification_source_col = None

for column in [
    "geographic_diversification_index",
    "diversification_index",
]:
    if column in track_concentration_df.columns:
        diversification_source_col = column
        break


if diversification_source_col is not None:

    track_concentration_df[
        "existing_diversification_score"
    ] = pd.to_numeric(
        track_concentration_df[
            diversification_source_col
        ],
        errors="coerce",
    ).fillna(0.0).clip(
        lower=0,
        upper=1,
    )

else:

    track_concentration_df[
        "existing_diversification_score"
    ] = (
        track_concentration_df[
            "inverse_hhi_score"
        ]
    )


# ------------------------------------------------------------
# 8. Build Geographic Concentration Score
# ------------------------------------------------------------

# Higher score = lower dependency / healthier diversification.
#
# HHI is the strongest component because it measures how
# concentrated the stream distribution is across markets.
#
# Strongest-market dominance and effective market breadth
# provide supporting diversification evidence.

track_concentration_df[
    "geographic_concentration_score"
] = (
    (
        0.40
        * track_concentration_df[
            "inverse_hhi_score"
        ]
    )
    +
    (
        0.25
        * track_concentration_df[
            "strongest_market_diversification_score"
        ]
    )
    +
    (
        0.20
        * track_concentration_df[
            "effective_market_diversification_score"
        ]
    )
    +
    (
        0.15
        * track_concentration_df[
            "existing_diversification_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Geographic concentration score classification
# ------------------------------------------------------------

def classify_geographic_concentration_score(value):

    if value >= 0.75:
        return "Highly diversified"

    if value >= 0.50:
        return "Diversified"

    if value >= 0.25:
        return "Moderately concentrated"

    if value > 0:
        return "Highly concentrated"

    return "Extremely concentrated"


track_concentration_df[
    "geographic_concentration_score_class"
] = (
    track_concentration_df[
        "geographic_concentration_score"
    ]
    .apply(
        classify_geographic_concentration_score
    )
)


# ------------------------------------------------------------
# 10. Final score table
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in track_concentration_df.columns
]


additional_columns = [
    column
    for column in [
        "geographic_concentration_class",
        "market_dependency_class",
    ]
    if column in track_concentration_df.columns
]


track_geographic_concentration_score_df = (
    track_concentration_df[
        identity_columns
        + [
            "top_1_stream_share_pct",
            "stream_concentration_hhi",
            "effective_stream_markets",
            "inverse_hhi_score",
            "strongest_market_diversification_score",
            "effective_market_diversification_score",
            "existing_diversification_score",
            "geographic_concentration_score",
            "geographic_concentration_score_class",
        ]
        + additional_columns
    ]
    .copy()
)


# ------------------------------------------------------------
# 11. Overview
# ------------------------------------------------------------

print()
print("Geographic concentration scoring overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_concentration_score_df):,}"
)

print(
    f"Median strongest-market stream share: "
    f"{track_geographic_concentration_score_df['top_1_stream_share_pct'].median():.2f}%"
)

print(
    f"Median stream-concentration HHI: "
    f"{track_geographic_concentration_score_df['stream_concentration_hhi'].median():.4f}"
)

print(
    f"Median effective stream markets: "
    f"{track_geographic_concentration_score_df['effective_stream_markets'].median():.2f}"
)

print(
    f"Median Geographic Concentration Score: "
    f"{track_geographic_concentration_score_df['geographic_concentration_score'].median():.4f}"
)

print(
    f"Mean Geographic Concentration Score: "
    f"{track_geographic_concentration_score_df['geographic_concentration_score'].mean():.4f}"
)


# ------------------------------------------------------------
# 12. Score classification summary
# ------------------------------------------------------------

concentration_score_class_order = [
    "Highly diversified",
    "Diversified",
    "Moderately concentrated",
    "Highly concentrated",
    "Extremely concentrated",
]


track_geographic_concentration_score_summary_df = (
    track_geographic_concentration_score_df[
        "geographic_concentration_score_class"
    ]
    .value_counts()
    .reindex(
        concentration_score_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Concentration Score Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_concentration_score_summary_df[
    "Share (%)"
] = (
    track_geographic_concentration_score_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_concentration_score_df)
    * 100
)


print()
print("Geographic concentration score classification")
print("=" * 100)

display(
    track_geographic_concentration_score_summary_df
)


# ------------------------------------------------------------
# 13. Most diversified tracks
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "top_1_stream_share_pct",
        "stream_concentration_hhi",
        "effective_stream_markets",
        "geographic_concentration_score",
        "geographic_concentration_score_class",
    ]
    if column in track_geographic_concentration_score_df.columns
]


print()
print("Tracks with highest Geographic Concentration Scores")
print("=" * 100)

highest_geographic_concentration_scores_df = (
    track_geographic_concentration_score_df
    .sort_values(
        [
            "geographic_concentration_score",
            "effective_stream_markets",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    highest_geographic_concentration_scores_df
)


# ------------------------------------------------------------
# 14. Diagnostics
# ------------------------------------------------------------

track_geographic_concentration_score_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with concentration score >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_score_df[
                            "geographic_concentration_score"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with concentration score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_score_df[
                            "geographic_concentration_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with HHI < 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_score_df[
                            "stream_concentration_hhi"
                        ]
                        < 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with effective stream markets >= 5",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_score_df[
                            "effective_stream_markets"
                        ]
                        >= 5
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with strongest-market share <= 50%",

            "Observed Value":
                int(
                    (
                        track_geographic_concentration_score_df[
                            "top_1_stream_share_pct"
                        ]
                        <= 50
                    ).sum()
                ),
        },
    ]
)


print()
print("Geographic concentration score diagnostics")
print("=" * 100)

display(
    track_geographic_concentration_score_diagnostics_df
)


# ------------------------------------------------------------
# 15. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_concentration_score_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_concentration_score_df[
        "track_id"
    ].nunique()
    == 110_198
)


score_fields = [
    "inverse_hhi_score",
    "strongest_market_diversification_score",
    "effective_market_diversification_score",
    "existing_diversification_score",
    "geographic_concentration_score",
]


score_valid = bool(
    np.isfinite(
        track_geographic_concentration_score_df[
            score_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


score_bounds_valid = bool(
    track_geographic_concentration_score_df[
        score_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


hhi_bounds_valid = bool(
    track_geographic_concentration_score_df[
        "stream_concentration_hhi"
    ]
    .between(
        0,
        1,
        inclusive="both",
    )
    .all()
)


share_bounds_valid = bool(
    track_geographic_concentration_score_df[
        "top_1_stream_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)


classification_complete = bool(
    track_geographic_concentration_score_df[
        "geographic_concentration_score_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_concentration_score_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_concentration_score_df
    )
)


section_10_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 10.2 completion",

            "Requirement":
                "Market Penetration Score must already be complete.",

            "Observed Evidence":
                f"Section 10.2 completion status: "
                f"{section_10_2_status}",

            "Passed":
                section_10_2_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_concentration_score_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one geographic "
                "concentration score.",

            "Observed Evidence":
                f"{track_geographic_concentration_score_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Concentration score numeric validity",

            "Requirement":
                "Concentration scoring inputs and outputs "
                "must remain finite.",

            "Observed Evidence":
                f"{len(score_fields)} normalized concentration "
                "score fields assessed",

            "Passed":
                score_valid,
        },
        {
            "Validation Area":
                "Concentration score bounds",

            "Requirement":
                "All normalized concentration score fields "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded geographic concentration scoring "
                "fields assessed",

            "Passed":
                score_bounds_valid,
        },
        {
            "Validation Area":
                "HHI bounds",

            "Requirement":
                "Stream-concentration HHI must remain "
                "between 0 and 1.",

            "Observed Evidence":
                "Stream-concentration HHI assessed "
                "for all tracks",

            "Passed":
                hhi_bounds_valid,
        },
        {
            "Validation Area":
                "Strongest-market percentage bounds",

            "Requirement":
                "Strongest-market stream share must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Strongest-market stream shares assessed "
                "for all tracks",

            "Passed":
                share_bounds_valid,
        },
        {
            "Validation Area":
                "Concentration classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "concentration score classification.",

            "Observed Evidence":
                f"{track_geographic_concentration_score_df['geographic_concentration_score_class'].notna().sum():,} "
                "concentration score classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Concentration population reconciliation",

            "Requirement":
                "Concentration score classifications must "
                "reconcile with the full track population.",

            "Observed Evidence":
                f"{int(track_geographic_concentration_score_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_concentration_score_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Geographic concentration score validation")
print("=" * 100)

display(
    section_10_3_validation_df
)


# ------------------------------------------------------------
# 16. Final completion status
# ------------------------------------------------------------

section_10_3_complete = bool(
    section_10_3_validation_df[
        "Passed"
    ].all()
)


if not section_10_3_complete:

    failed_checks = (
        section_10_3_validation_df
        .loc[
            ~section_10_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 10.3 Geographic Concentration Score "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 10.3 Geographic Concentration Score "
    "validation checks passed."
)

print(
    f"Section 10.3 completion status: "
    f"{section_10_3_complete}"
)

print(
    f"Geographic Concentration Scores prepared "
    f"for {len(track_geographic_concentration_score_df):,} tracks."
)

print(
    "Higher concentration scores represent stronger "
    "geographic diversification and lower market dependency."
)

print(
    "The workflow is ready for "
    "Section 10.4 Geographic Expansion Indicator."
)
```

#### Interpretation

Section 10.3 successfully converts the validated track-level geographic concentration intelligence into a normalized **Geographic Concentration Score** for all **110,198 tracks**.

The scoring direction is intentionally reversed so that a **higher score represents stronger geographic diversification and lower market dependency**, while a lower score represents heavier concentration in one or a small number of markets.

The overall distribution is strongly concentrated. The median track has a **100.00% strongest-market stream share**, a median **stream-concentration HHI of 1.0000**, and a median of only **1.00 effective stream market**. As a result, the median Geographic Concentration Score is **0.0000**.

The mean score is slightly higher at **0.0769**, showing that a relatively small group of tracks achieves meaningful geographic diversification and pulls the average upward.

The score classification shows:

- **2,431 tracks (2.21%)** are **Highly diversified**
- **4,162 tracks (3.78%)** are **Diversified**
- **6,346 tracks (5.76%)** are **Moderately concentrated**
- **16,177 tracks (14.68%)** are **Highly concentrated**
- **81,082 tracks (73.58%)** are **Extremely concentrated**

This distribution is consistent with the earlier PMIP geographic analysis, where most tracks were observed in a single market or had most of their streaming activity concentrated in one country.

The diagnostics reinforce this pattern. Only **6,593 tracks** achieve a concentration score of at least **0.50**, while just **2,431 tracks** reach the highly diversified threshold of **0.75 or above**.

A total of **8,704 tracks** have a stream-concentration HHI below **0.50**, indicating a more distributed streaming pattern across countries.

Only **2,721 tracks** have at least **5 effective stream markets**, while **5,635 tracks** have a strongest-market stream share of **50% or less**. These tracks provide the clearest evidence of genuinely diversified geographic performance.

The Geographic Concentration Score combines four complementary diversification signals:

- **inverse HHI**, which rewards more evenly distributed streaming across markets
- **strongest-market diversification**, which penalizes excessive dependence on a single country
- **effective market breadth**, which rewards tracks whose streaming activity is meaningfully distributed across several markets
- the existing validated **diversification measure**

Because the score is reversed relative to raw concentration, a track with a score near **1.0** is geographically diversified, while a track with a score near **0.0** is highly dependent on one dominant market.

This makes the score easier to combine later with the Geographic Reach Score and Market Penetration Score, because all three dimensions now follow the same general direction: **higher values represent stronger geographic intelligence characteristics**.

All Section 10.3 validation checks passed. The full **110,198-track population** was preserved, track identities remained unique, all five normalized concentration score fields remained finite and within the **0-to-1 range**, HHI values remained valid, strongest-market percentages remained within **0 to 100**, every track received a concentration classification, and all classification totals reconciled with the full population.

Section 10.3 therefore establishes the third normalized scoring dimension in PMIP's Geographic Intelligence Scoring framework.

The workflow is ready for **Section 10.4 Geographic Expansion Indicator**.

### 10.4 Geographic Expansion Indicator

#### Purpose

Create a normalized Geographic Expansion Indicator for each track using the validated geographic expansion, contraction, cross-market momentum, and temporal-confidence intelligence developed in Section 9.

The indicator measures whether a track's observed geographic footprint is currently strengthening or weakening by combining the scale of validated expansion with the balance between expansion and contraction.

Temporal-confidence safeguards are also included so that geographic movement supported by stronger reporting history receives greater interpretive weight.

The resulting Geographic Expansion Indicator is bounded between 0 and 1, where values closer to 1 represent stronger validated geographic expansion, values around 0.50 represent relatively balanced movement, and values closer to 0 represent stronger contraction or weak expansion evidence.

```python
import pandas as pd
import numpy as np

print("Scoring PMIP geographic expansion")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_10_3_status = globals().get(
    "section_10_3_complete",
    False
)

print(
    f"Section 10.3 completion status: "
    f"{section_10_3_status}"
)

if not section_10_3_status:
    raise RuntimeError(
        "Section 10.4 requires Section 10.3 "
        "Geographic Concentration Score to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated geographic expansion source
# ------------------------------------------------------------

expansion_candidate_names = [
    "track_geographic_expansion_df",
]

track_expansion_df = None
track_expansion_source_name = None

required_expansion_columns = {
    "track_id",
    "validated_expansion_markets",
    "expansion_market_share_pct",
    "geographic_expansion_index",
    "geographic_expansion_class",
}

for name in expansion_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_expansion_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_expansion_df = candidate.copy()
    track_expansion_source_name = name
    break


if track_expansion_df is None:
    raise RuntimeError(
        "Section 10.4 could not recover the validated "
        "geographic expansion intelligence."
    )


# ------------------------------------------------------------
# 3. Recover validated cross-market momentum source
# ------------------------------------------------------------

momentum_candidate_names = [
    "track_cross_market_momentum_df",
]

track_momentum_df = None
track_momentum_source_name = None

required_momentum_columns = {
    "track_id",
    "cross_market_momentum_index",
    "raw_momentum_balance",
    "cross_market_momentum_class",
}

for name in momentum_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_momentum_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_momentum_df = candidate.copy()
    track_momentum_source_name = name
    break


if track_momentum_df is None:
    raise RuntimeError(
        "Section 10.4 could not recover the validated "
        "cross-market momentum intelligence."
    )


# ------------------------------------------------------------
# 4. Recover temporal validation source
# ------------------------------------------------------------

temporal_candidate_names = [
    "track_temporal_validation_df",
]

track_temporal_df = None
track_temporal_source_name = None

required_temporal_columns = {
    "track_id",
    "geographic_change_temporal_confidence_index",
    "temporal_confidence_class",
    "geographic_change_interpretation_policy",
}

for name in temporal_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if not required_temporal_columns.issubset(
        candidate.columns
    ):
        continue

    if len(candidate) != 110_198:
        continue

    track_temporal_df = candidate.copy()
    track_temporal_source_name = name
    break


if track_temporal_df is None:
    raise RuntimeError(
        "Section 10.4 could not recover the validated "
        "temporal geographic-change intelligence."
    )


print()
print("Validated geographic expansion scoring sources")
print("=" * 100)

print(
    f"Expansion source: "
    f"{track_expansion_source_name}"
)

print(
    f"Momentum source: "
    f"{track_momentum_source_name}"
)

print(
    f"Temporal validation source: "
    f"{track_temporal_source_name}"
)

print(
    f"Tracks: "
    f"{track_expansion_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 5. Prepare identity columns
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in track_expansion_df.columns
]


identity_df = (
    track_expansion_df[
        identity_columns
    ]
    .drop_duplicates(
        subset=[
            "track_id"
        ]
    )
)


# ------------------------------------------------------------
# 6. Select required expansion fields
# ------------------------------------------------------------

expansion_columns = [
    column
    for column in [
        "track_id",
        "validated_expansion_markets",
        "expansion_market_share_pct",
        "expansion_breadth_index",
        "expansion_share_index",
        "expansion_speed_index",
        "geographic_expansion_index",
        "geographic_expansion_class",
    ]
    if column in track_expansion_df.columns
]


# ------------------------------------------------------------
# 7. Select required momentum fields
# ------------------------------------------------------------

momentum_columns = [
    column
    for column in [
        "track_id",
        "net_market_movement",
        "raw_momentum_balance",
        "cross_market_momentum_index",
        "cross_market_momentum_class",
    ]
    if column in track_momentum_df.columns
]


# ------------------------------------------------------------
# 8. Select required temporal fields
# ------------------------------------------------------------

temporal_columns = [
    column
    for column in [
        "track_id",
        "geographic_change_temporal_confidence_index",
        "temporal_confidence_class",
        "geographic_change_interpretation_policy",
    ]
    if column in track_temporal_df.columns
]


# ------------------------------------------------------------
# 9. Merge validated Section 9 intelligence
# ------------------------------------------------------------

track_geographic_expansion_indicator_df = (
    identity_df
    .merge(
        track_expansion_df[
            expansion_columns
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
    .merge(
        track_momentum_df[
            momentum_columns
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
    .merge(
        track_temporal_df[
            temporal_columns
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 10. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    "validated_expansion_markets",
    "expansion_market_share_pct",
    "geographic_expansion_index",
    "raw_momentum_balance",
    "cross_market_momentum_index",
    "geographic_change_temporal_confidence_index",
]

for column in numeric_columns:

    track_geographic_expansion_indicator_df[
        column
    ] = pd.to_numeric(
        track_geographic_expansion_indicator_df[
            column
        ],
        errors="coerce",
    )


if track_geographic_expansion_indicator_df[
    numeric_columns
].isna().any().any():

    raise RuntimeError(
        "Section 10.4 found missing or invalid "
        "numeric geographic expansion values."
    )


# ------------------------------------------------------------
# 11. Normalize expansion market share
# ------------------------------------------------------------

track_geographic_expansion_indicator_df[
    "expansion_market_share_score"
] = (
    track_geographic_expansion_indicator_df[
        "expansion_market_share_pct"
    ]
    /
    100
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 12. Build directional expansion score
# ------------------------------------------------------------

# geographic_expansion_index measures validated expansion
# cross_market_momentum_index places expansion in the context
# of contraction
#
# This means a track only achieves a strong directional score
# when expansion is meaningful relative to contraction.

track_geographic_expansion_indicator_df[
    "directional_expansion_score"
] = (
    (
        0.60
        * track_geographic_expansion_indicator_df[
            "geographic_expansion_index"
        ]
    )
    +
    (
        0.40
        * track_geographic_expansion_indicator_df[
            "cross_market_momentum_index"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 13. Apply temporal-confidence safeguard
# ------------------------------------------------------------

track_geographic_expansion_indicator_df[
    "temporally_supported_expansion_score"
] = (
    track_geographic_expansion_indicator_df[
        "directional_expansion_score"
    ]
    *
    track_geographic_expansion_indicator_df[
        "geographic_change_temporal_confidence_index"
    ]
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 14. Build final Geographic Expansion Indicator
# ------------------------------------------------------------

# Expansion strength remains the main component.
# Expansion breadth adds supporting information.
# Momentum provides directional context.
# Temporal confidence controls interpretation reliability.

track_geographic_expansion_indicator_df[
    "geographic_expansion_indicator"
] = (
    (
        0.45
        * track_geographic_expansion_indicator_df[
            "geographic_expansion_index"
        ]
    )
    +
    (
        0.20
        * track_geographic_expansion_indicator_df[
            "expansion_market_share_score"
        ]
    )
    +
    (
        0.20
        * track_geographic_expansion_indicator_df[
            "cross_market_momentum_index"
        ]
    )
    +
    (
        0.15
        * track_geographic_expansion_indicator_df[
            "temporally_supported_expansion_score"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 15. Geographic Expansion Indicator classification
# ------------------------------------------------------------

def classify_geographic_expansion_indicator(value):

    if value >= 0.75:
        return "Exceptional geographic expansion"

    if value >= 0.50:
        return "Strong geographic expansion"

    if value >= 0.25:
        return "Moderate geographic expansion"

    if value > 0:
        return "Limited geographic expansion"

    return "No geographic expansion"


track_geographic_expansion_indicator_df[
    "geographic_expansion_indicator_class"
] = (
    track_geographic_expansion_indicator_df[
        "geographic_expansion_indicator"
    ]
    .apply(
        classify_geographic_expansion_indicator
    )
)


# ------------------------------------------------------------
# 16. Overview
# ------------------------------------------------------------

print()
print("Geographic expansion indicator overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_expansion_indicator_df):,}"
)

print(
    f"Median validated expansion markets: "
    f"{track_geographic_expansion_indicator_df['validated_expansion_markets'].median():.1f}"
)

print(
    f"Median geographic expansion index: "
    f"{track_geographic_expansion_indicator_df['geographic_expansion_index'].median():.4f}"
)

print(
    f"Median cross-market momentum index: "
    f"{track_geographic_expansion_indicator_df['cross_market_momentum_index'].median():.4f}"
)

print(
    f"Median temporal confidence index: "
    f"{track_geographic_expansion_indicator_df['geographic_change_temporal_confidence_index'].median():.4f}"
)

print(
    f"Median Geographic Expansion Indicator: "
    f"{track_geographic_expansion_indicator_df['geographic_expansion_indicator'].median():.4f}"
)

print(
    f"Mean Geographic Expansion Indicator: "
    f"{track_geographic_expansion_indicator_df['geographic_expansion_indicator'].mean():.4f}"
)


# ------------------------------------------------------------
# 17. Classification summary
# ------------------------------------------------------------

expansion_indicator_class_order = [
    "Exceptional geographic expansion",
    "Strong geographic expansion",
    "Moderate geographic expansion",
    "Limited geographic expansion",
    "No geographic expansion",
]


track_geographic_expansion_indicator_summary_df = (
    track_geographic_expansion_indicator_df[
        "geographic_expansion_indicator_class"
    ]
    .value_counts()
    .reindex(
        expansion_indicator_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Expansion Indicator Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_expansion_indicator_summary_df[
    "Share (%)"
] = (
    track_geographic_expansion_indicator_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_expansion_indicator_df)
    * 100
)


print()
print("Geographic expansion indicator classification")
print("=" * 100)

display(
    track_geographic_expansion_indicator_summary_df
)


# ------------------------------------------------------------
# 18. Highest Geographic Expansion Indicators
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "validated_expansion_markets",
        "expansion_market_share_pct",
        "geographic_expansion_index",
        "cross_market_momentum_index",
        "cross_market_momentum_class",
        "geographic_change_temporal_confidence_index",
        "geographic_expansion_indicator",
        "geographic_expansion_indicator_class",
    ]
    if column in track_geographic_expansion_indicator_df.columns
]


print()
print("Tracks with highest Geographic Expansion Indicators")
print("=" * 100)

highest_geographic_expansion_indicator_df = (
    track_geographic_expansion_indicator_df
    .sort_values(
        [
            "geographic_expansion_indicator",
            "geographic_expansion_index",
            "validated_expansion_markets",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    highest_geographic_expansion_indicator_df
)


# ------------------------------------------------------------
# 19. Diagnostics
# ------------------------------------------------------------

track_geographic_expansion_indicator_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with expansion indicator >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_indicator_df[
                            "geographic_expansion_indicator"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with expansion indicator >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_indicator_df[
                            "geographic_expansion_indicator"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with validated expansion",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_indicator_df[
                            "validated_expansion_markets"
                        ]
                        > 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with positive cross-market momentum",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_indicator_df[
                            "raw_momentum_balance"
                        ]
                        > 0
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with temporal confidence >= 0.70",

            "Observed Value":
                int(
                    (
                        track_geographic_expansion_indicator_df[
                            "geographic_change_temporal_confidence_index"
                        ]
                        >= 0.70
                    ).sum()
                ),
        },
    ]
)


print()
print("Geographic expansion indicator diagnostics")
print("=" * 100)

display(
    track_geographic_expansion_indicator_diagnostics_df
)


# ------------------------------------------------------------
# 20. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_expansion_indicator_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_expansion_indicator_df[
        "track_id"
    ].nunique()
    == 110_198
)


score_fields = [
    "geographic_expansion_index",
    "expansion_market_share_score",
    "cross_market_momentum_index",
    "geographic_change_temporal_confidence_index",
    "directional_expansion_score",
    "temporally_supported_expansion_score",
    "geographic_expansion_indicator",
]


score_valid = bool(
    np.isfinite(
        track_geographic_expansion_indicator_df[
            score_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


score_bounds_valid = bool(
    track_geographic_expansion_indicator_df[
        score_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


expansion_percentage_valid = bool(
    track_geographic_expansion_indicator_df[
        "expansion_market_share_pct"
    ]
    .between(
        0,
        100,
        inclusive="both",
    )
    .all()
)


classification_complete = bool(
    track_geographic_expansion_indicator_df[
        "geographic_expansion_indicator_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        track_geographic_expansion_indicator_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_expansion_indicator_df
    )
)


section_10_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 10.3 completion",

            "Requirement":
                "Geographic Concentration Score must "
                "already be complete.",

            "Observed Evidence":
                f"Section 10.3 completion status: "
                f"{section_10_3_status}",

            "Passed":
                section_10_3_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_expansion_indicator_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one geographic "
                "expansion indicator.",

            "Observed Evidence":
                f"{track_geographic_expansion_indicator_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Expansion indicator numeric validity",

            "Requirement":
                "Expansion indicator inputs and outputs "
                "must remain finite.",

            "Observed Evidence":
                f"{len(score_fields)} normalized expansion "
                "indicator fields assessed",

            "Passed":
                score_valid,
        },
        {
            "Validation Area":
                "Expansion indicator bounds",

            "Requirement":
                "All normalized expansion indicator fields "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded expansion scoring fields assessed "
                "for all tracks",

            "Passed":
                score_bounds_valid,
        },
        {
            "Validation Area":
                "Expansion percentage bounds",

            "Requirement":
                "Expansion market shares must remain "
                "between 0 and 100.",

            "Observed Evidence":
                "Expansion market shares assessed "
                "for all tracks",

            "Passed":
                expansion_percentage_valid,
        },
        {
            "Validation Area":
                "Expansion classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "expansion indicator classification.",

            "Observed Evidence":
                f"{track_geographic_expansion_indicator_df['geographic_expansion_indicator_class'].notna().sum():,} "
                "expansion indicator classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Expansion population reconciliation",

            "Requirement":
                "Expansion indicator classifications must "
                "reconcile with the full track population.",

            "Observed Evidence":
                f"{int(track_geographic_expansion_indicator_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_expansion_indicator_df):,} tracks",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Geographic expansion indicator validation")
print("=" * 100)

display(
    section_10_4_validation_df
)


# ------------------------------------------------------------
# 21. Final completion status
# ------------------------------------------------------------

section_10_4_complete = bool(
    section_10_4_validation_df[
        "Passed"
    ].all()
)


if not section_10_4_complete:

    failed_checks = (
        section_10_4_validation_df
        .loc[
            ~section_10_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 10.4 Geographic Expansion Indicator "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 10.4 Geographic Expansion Indicator "
    "validation checks passed."
)

print(
    f"Section 10.4 completion status: "
    f"{section_10_4_complete}"
)

print(
    f"Geographic Expansion Indicators prepared "
    f"for {len(track_geographic_expansion_indicator_df):,} tracks."
)

print(
    "The workflow is ready for "
    "Section 10.5 Geographic Intelligence Summary."
)
```

#### Interpretation

Section 10.4 successfully converts the validated geographic expansion, cross-market momentum, and temporal-confidence intelligence into a normalized **Geographic Expansion Indicator** for all **110,198 tracks**.

The results show that geographic expansion is relatively uncommon across the overall track population. The median track has **0 validated expansion markets** and a median geographic expansion index of **0.0000**.

At the same time, the median cross-market momentum index is **0.2963**, which indicates that geographic contraction generally outweighs expansion for the typical track. The median temporal-confidence index is **1.0000**, showing that the underlying geographic movement evidence is generally supported by strong temporal coverage.

After combining these components, the median Geographic Expansion Indicator is **0.0768**, while the mean is slightly higher at **0.1199**. This suggests that a small group of tracks shows substantially stronger geographic expansion and raises the overall average.

The indicator classification shows:

- **76 tracks (0.07%)** have **Exceptional geographic expansion**
- **3,193 tracks (2.90%)** have **Strong geographic expansion**
- **8,125 tracks (7.37%)** have **Moderate geographic expansion**
- **98,804 tracks (89.66%)** have **Limited geographic expansion**
- **0 tracks** are classified as **No geographic expansion**

The dominance of the Limited Geographic Expansion category is consistent with the earlier Section 9 findings, where most tracks had either no validated market entry or stronger contraction than expansion.

The diagnostics provide additional evidence. A total of **12,564 tracks** show at least one validated geographic expansion event, while only **3,055 tracks** have positive cross-market momentum.

A total of **3,269 tracks** achieve a Geographic Expansion Indicator of at least **0.50**, and only **76 tracks** reach the exceptional threshold of **0.75 or above**. This confirms that strong geographic expansion is rare within the full track population.

The Geographic Expansion Indicator combines four complementary components:

- the validated **Geographic Expansion Index**
- the proportion of markets represented by validated expansion
- the **Cross-Market Momentum Index**
- the temporally supported expansion signal

The Geographic Expansion Index receives the greatest weight because it directly measures validated expansion behaviour. The cross-market momentum component adds directional context by considering expansion relative to contraction.

Temporal confidence then acts as an interpretation safeguard. Expansion behaviour supported by stronger reporting histories contributes more confidently to the final indicator, while movement based on weaker temporal evidence has less influence.

This means the indicator does not simply reward tracks that appear in new markets. A strong result requires a combination of meaningful expansion, favourable movement relative to contraction, and reliable temporal evidence.

An important point is that tracks with no validated expansion can still receive a small positive indicator value because the final score also includes the cross-market momentum component. Therefore, the indicator should be interpreted as a broader measure of **geographic expansion strength and directional movement**, rather than as a simple binary measure of whether a track entered a new market.

Section 10.4 therefore establishes the fourth normalized dimension of PMIP's Geographic Intelligence Scoring framework.

- **Geographic Reach Score** — breadth of international presence
- **Market Penetration Score** — depth of performance within reached markets
- **Geographic Concentration Score** — diversification and market dependency
- **Geographic Expansion Indicator** — direction and strength of geographic movement

### 10.5 Geographic Intelligence Summary

#### Purpose

Combine the four validated geographic scoring dimensions created in Sections 10.1 to 10.4 into a single PMIP Geographic Intelligence Summary for each track.

The summary integrates:

- Geographic Reach Score
- Market Penetration Score
- Geographic Concentration Score
- Geographic Expansion Indicator

Each component measures a different aspect of geographic performance, so the summary preserves the individual scores while also creating a combined Geographic Intelligence Score on a common 0-to-1 scale.

The final score is intended to support structured comparison and classification across tracks. It does not replace the underlying geographic indicators and should not be interpreted as a causal measure of commercial success.

```python
import pandas as pd
import numpy as np

print("Building PMIP geographic intelligence summary")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_10_4_status = globals().get(
    "section_10_4_complete",
    False
)

print(
    f"Section 10.4 completion status: "
    f"{section_10_4_status}"
)

if not section_10_4_status:
    raise RuntimeError(
        "Section 10.5 requires Section 10.4 "
        "Geographic Expansion Indicator to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated Section 10 scoring sources
# ------------------------------------------------------------

required_sources = {
    "reach": "track_geographic_reach_score_df",
    "penetration": "track_market_penetration_score_df",
    "concentration": "track_geographic_concentration_score_df",
    "expansion": "track_geographic_expansion_indicator_df",
}

recovered_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 10.5 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    if len(candidate) != 110_198:
        raise RuntimeError(
            f"{variable_name} does not contain the expected "
            f"110,198 tracks."
        )

    recovered_sources[source_key] = candidate.copy()


reach_df = recovered_sources["reach"]
penetration_df = recovered_sources["penetration"]
concentration_df = recovered_sources["concentration"]
expansion_df = recovered_sources["expansion"]


print()
print("Validated geographic intelligence scoring sources")
print("=" * 100)

print(
    f"Geographic reach source: "
    f"track_geographic_reach_score_df"
)

print(
    f"Market penetration source: "
    f"track_market_penetration_score_df"
)

print(
    f"Geographic concentration source: "
    f"track_geographic_concentration_score_df"
)

print(
    f"Geographic expansion source: "
    f"track_geographic_expansion_indicator_df"
)

print(
    f"Tracks: "
    f"{reach_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Validate required scoring columns
# ------------------------------------------------------------

required_score_columns = {
    "reach": "geographic_reach_score",
    "penetration": "market_penetration_score",
    "concentration": "geographic_concentration_score",
    "expansion": "geographic_expansion_indicator",
}

for source_key, score_column in required_score_columns.items():

    source_df = recovered_sources[source_key]

    if score_column not in source_df.columns:
        raise RuntimeError(
            f"Section 10.5 requires column "
            f"'{score_column}' in the {source_key} source."
        )


# ------------------------------------------------------------
# 4. Build identity base
# ------------------------------------------------------------

identity_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
    ]
    if column in reach_df.columns
]


track_geographic_intelligence_df = (
    reach_df[
        identity_columns
        + [
            "geographic_reach_score",
            "geographic_reach_score_class",
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 5. Merge Market Penetration Score
# ------------------------------------------------------------

track_geographic_intelligence_df = (
    track_geographic_intelligence_df
    .merge(
        penetration_df[
            [
                "track_id",
                "market_penetration_score",
                "market_penetration_score_class",
            ]
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 6. Merge Geographic Concentration Score
# ------------------------------------------------------------

track_geographic_intelligence_df = (
    track_geographic_intelligence_df
    .merge(
        concentration_df[
            [
                "track_id",
                "geographic_concentration_score",
                "geographic_concentration_score_class",
            ]
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 7. Merge Geographic Expansion Indicator
# ------------------------------------------------------------

track_geographic_intelligence_df = (
    track_geographic_intelligence_df
    .merge(
        expansion_df[
            [
                "track_id",
                "geographic_expansion_indicator",
                "geographic_expansion_indicator_class",
            ]
        ],
        on="track_id",
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 8. Numeric safety
# ------------------------------------------------------------

score_columns = [
    "geographic_reach_score",
    "market_penetration_score",
    "geographic_concentration_score",
    "geographic_expansion_indicator",
]

for column in score_columns:

    track_geographic_intelligence_df[
        column
    ] = pd.to_numeric(
        track_geographic_intelligence_df[
            column
        ],
        errors="coerce",
    )


if track_geographic_intelligence_df[
    score_columns
].isna().any().any():

    raise RuntimeError(
        "Section 10.5 found missing or invalid "
        "geographic scoring values."
    )


# ------------------------------------------------------------
# 9. Build Geographic Intelligence Score
# ------------------------------------------------------------

# Weighting logic:
#
# Reach              = 30%
# Penetration        = 30%
# Concentration      = 20%
# Expansion          = 20%
#
# Reach and penetration receive the largest weighting because
# they represent the core breadth and depth of geographic
# performance.
#
# Concentration adds diversification quality.
# Expansion adds movement and directional intelligence.

track_geographic_intelligence_df[
    "geographic_intelligence_score"
] = (
    (
        0.30
        * track_geographic_intelligence_df[
            "geographic_reach_score"
        ]
    )
    +
    (
        0.30
        * track_geographic_intelligence_df[
            "market_penetration_score"
        ]
    )
    +
    (
        0.20
        * track_geographic_intelligence_df[
            "geographic_concentration_score"
        ]
    )
    +
    (
        0.20
        * track_geographic_intelligence_df[
            "geographic_expansion_indicator"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 10. Build balance diagnostic
# ------------------------------------------------------------

track_geographic_intelligence_df[
    "geographic_dimension_mean"
] = (
    track_geographic_intelligence_df[
        score_columns
    ]
    .mean(
        axis=1
    )
)


track_geographic_intelligence_df[
    "geographic_dimension_std"
] = (
    track_geographic_intelligence_df[
        score_columns
    ]
    .std(
        axis=1,
        ddof=0,
    )
)


track_geographic_intelligence_df[
    "geographic_balance_index"
] = (
    1
    -
    track_geographic_intelligence_df[
        "geographic_dimension_std"
    ]
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 11. Identify strongest geographic dimension
# ------------------------------------------------------------

dimension_map = {
    "geographic_reach_score":
        "Geographic reach",

    "market_penetration_score":
        "Market penetration",

    "geographic_concentration_score":
        "Geographic diversification",

    "geographic_expansion_indicator":
        "Geographic expansion",
}


track_geographic_intelligence_df[
    "strongest_geographic_dimension"
] = (
    track_geographic_intelligence_df[
        score_columns
    ]
    .idxmax(
        axis=1
    )
    .map(
        dimension_map
    )
)


# ------------------------------------------------------------
# 12. Identify weakest geographic dimension
# ------------------------------------------------------------

track_geographic_intelligence_df[
    "weakest_geographic_dimension"
] = (
    track_geographic_intelligence_df[
        score_columns
    ]
    .idxmin(
        axis=1
    )
    .map(
        dimension_map
    )
)


# ------------------------------------------------------------
# 13. Geographic Intelligence classification
# ------------------------------------------------------------

def classify_geographic_intelligence_score(value):

    if value >= 0.75:
        return "Exceptional geographic intelligence"

    if value >= 0.50:
        return "Strong geographic intelligence"

    if value >= 0.30:
        return "Moderate geographic intelligence"

    if value >= 0.15:
        return "Developing geographic intelligence"

    return "Limited geographic intelligence"


track_geographic_intelligence_df[
    "geographic_intelligence_class"
] = (
    track_geographic_intelligence_df[
        "geographic_intelligence_score"
    ]
    .apply(
        classify_geographic_intelligence_score
    )
)


# ------------------------------------------------------------
# 14. Build geographic performance profile
# ------------------------------------------------------------

def build_geographic_performance_profile(row):

    score = row[
        "geographic_intelligence_score"
    ]

    reach = row[
        "geographic_reach_score"
    ]

    penetration = row[
        "market_penetration_score"
    ]

    diversification = row[
        "geographic_concentration_score"
    ]

    expansion = row[
        "geographic_expansion_indicator"
    ]

    if (
        score >= 0.65
        and
        reach >= 0.50
        and
        penetration >= 0.50
        and
        diversification >= 0.50
    ):
        return "Broad high-performance geographic profile"

    if (
        reach >= 0.50
        and
        penetration < 0.50
    ):
        return "Broad reach with developing penetration"

    if (
        reach < 0.50
        and
        penetration >= 0.50
    ):
        return "Concentrated reach with strong penetration"

    if (
        expansion >= 0.50
        and
        score < 0.50
    ):
        return "Emerging geographic growth profile"

    if (
        diversification >= 0.50
        and
        reach >= 0.25
    ):
        return "Diversified multi-market profile"

    return "Limited geographic footprint"


track_geographic_intelligence_df[
    "geographic_performance_profile"
] = (
    track_geographic_intelligence_df
    .apply(
        build_geographic_performance_profile,
        axis=1,
    )
)


# ------------------------------------------------------------
# 15. Overview
# ------------------------------------------------------------

print()
print("Geographic intelligence summary overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_geographic_intelligence_df):,}"
)

print(
    f"Median Geographic Reach Score: "
    f"{track_geographic_intelligence_df['geographic_reach_score'].median():.4f}"
)

print(
    f"Median Market Penetration Score: "
    f"{track_geographic_intelligence_df['market_penetration_score'].median():.4f}"
)

print(
    f"Median Geographic Concentration Score: "
    f"{track_geographic_intelligence_df['geographic_concentration_score'].median():.4f}"
)

print(
    f"Median Geographic Expansion Indicator: "
    f"{track_geographic_intelligence_df['geographic_expansion_indicator'].median():.4f}"
)

print(
    f"Median Geographic Intelligence Score: "
    f"{track_geographic_intelligence_df['geographic_intelligence_score'].median():.4f}"
)

print(
    f"Mean Geographic Intelligence Score: "
    f"{track_geographic_intelligence_df['geographic_intelligence_score'].mean():.4f}"
)

print(
    f"Median geographic balance index: "
    f"{track_geographic_intelligence_df['geographic_balance_index'].median():.4f}"
)


# ------------------------------------------------------------
# 16. Geographic Intelligence classification summary
# ------------------------------------------------------------

intelligence_class_order = [
    "Exceptional geographic intelligence",
    "Strong geographic intelligence",
    "Moderate geographic intelligence",
    "Developing geographic intelligence",
    "Limited geographic intelligence",
]


track_geographic_intelligence_summary_df = (
    track_geographic_intelligence_df[
        "geographic_intelligence_class"
    ]
    .value_counts()
    .reindex(
        intelligence_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Intelligence Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_intelligence_summary_df[
    "Share (%)"
] = (
    track_geographic_intelligence_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_intelligence_df)
    * 100
)


print()
print("Geographic intelligence classification")
print("=" * 100)

display(
    track_geographic_intelligence_summary_df
)


# ------------------------------------------------------------
# 17. Geographic performance profile summary
# ------------------------------------------------------------

profile_order = [
    "Broad high-performance geographic profile",
    "Broad reach with developing penetration",
    "Concentrated reach with strong penetration",
    "Emerging geographic growth profile",
    "Diversified multi-market profile",
    "Limited geographic footprint",
]


track_geographic_performance_profile_summary_df = (
    track_geographic_intelligence_df[
        "geographic_performance_profile"
    ]
    .value_counts()
    .reindex(
        profile_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Performance Profile"
    )
    .reset_index(
        name="Tracks"
    )
)


track_geographic_performance_profile_summary_df[
    "Share (%)"
] = (
    track_geographic_performance_profile_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_intelligence_df)
    * 100
)


print()
print("Geographic performance profile classification")
print("=" * 100)

display(
    track_geographic_performance_profile_summary_df
)


# ------------------------------------------------------------
# 18. Strongest dimension summary
# ------------------------------------------------------------

strongest_dimension_summary_df = (
    track_geographic_intelligence_df[
        "strongest_geographic_dimension"
    ]
    .value_counts()
    .rename_axis(
        "Strongest Geographic Dimension"
    )
    .reset_index(
        name="Tracks"
    )
)


strongest_dimension_summary_df[
    "Share (%)"
] = (
    strongest_dimension_summary_df[
        "Tracks"
    ]
    /
    len(track_geographic_intelligence_df)
    * 100
)


print()
print("Strongest geographic dimension")
print("=" * 100)

display(
    strongest_dimension_summary_df
)


# ------------------------------------------------------------
# 19. Highest Geographic Intelligence Scores
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "geographic_reach_score",
        "market_penetration_score",
        "geographic_concentration_score",
        "geographic_expansion_indicator",
        "geographic_intelligence_score",
        "geographic_balance_index",
        "strongest_geographic_dimension",
        "weakest_geographic_dimension",
        "geographic_intelligence_class",
        "geographic_performance_profile",
    ]
    if column in track_geographic_intelligence_df.columns
]


print()
print("Tracks with highest Geographic Intelligence Scores")
print("=" * 100)

highest_geographic_intelligence_scores_df = (
    track_geographic_intelligence_df
    .sort_values(
        [
            "geographic_intelligence_score",
            "geographic_balance_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)

display(
    highest_geographic_intelligence_scores_df
)


# ------------------------------------------------------------
# 20. Diagnostics
# ------------------------------------------------------------

track_geographic_intelligence_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks with Geographic Intelligence Score >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_intelligence_df[
                            "geographic_intelligence_score"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Geographic Intelligence Score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_intelligence_df[
                            "geographic_intelligence_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with all four component scores >= 0.50",

            "Observed Value":
                int(
                    (
                        track_geographic_intelligence_df[
                            score_columns
                        ]
                        >= 0.50
                    )
                    .all(
                        axis=1
                    )
                    .sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with geographic balance index >= 0.75",

            "Observed Value":
                int(
                    (
                        track_geographic_intelligence_df[
                            "geographic_balance_index"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with limited geographic footprint",

            "Observed Value":
                int(
                    (
                        track_geographic_intelligence_df[
                            "geographic_performance_profile"
                        ]
                        ==
                        "Limited geographic footprint"
                    ).sum()
                ),
        },
    ]
)


print()
print("Geographic intelligence diagnostics")
print("=" * 100)

display(
    track_geographic_intelligence_diagnostics_df
)


# ------------------------------------------------------------
# 21. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_geographic_intelligence_df
    )
    == 110_198
)


identity_valid = bool(
    track_geographic_intelligence_df[
        "track_id"
    ].nunique()
    == 110_198
)


final_score_fields = (
    score_columns
    + [
        "geographic_intelligence_score",
        "geographic_dimension_mean",
        "geographic_dimension_std",
        "geographic_balance_index",
    ]
)


score_valid = bool(
    np.isfinite(
        track_geographic_intelligence_df[
            final_score_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


component_bounds_valid = bool(
    track_geographic_intelligence_df[
        score_columns
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


final_score_bounds_valid = bool(
    track_geographic_intelligence_df[
        [
            "geographic_intelligence_score",
            "geographic_balance_index",
        ]
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    track_geographic_intelligence_df[
        "geographic_intelligence_class"
    ]
    .notna()
    .all()
)


profile_complete = bool(
    track_geographic_intelligence_df[
        "geographic_performance_profile"
    ]
    .notna()
    .all()
)


dimension_complete = bool(
    track_geographic_intelligence_df[
        [
            "strongest_geographic_dimension",
            "weakest_geographic_dimension",
        ]
    ]
    .notna()
    .all()
    .all()
)


classification_reconciled = bool(
    int(
        track_geographic_intelligence_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_intelligence_df
    )
)


profile_reconciled = bool(
    int(
        track_geographic_performance_profile_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_geographic_intelligence_df
    )
)


section_10_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 10.4 completion",

            "Requirement":
                "Geographic Expansion Indicator must "
                "already be complete.",

            "Observed Evidence":
                f"Section 10.4 completion status: "
                f"{section_10_4_status}",

            "Passed":
                section_10_4_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_geographic_intelligence_df):,} "
                "tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must receive one geographic "
                "intelligence summary.",

            "Observed Evidence":
                f"{track_geographic_intelligence_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_valid,
        },
        {
            "Validation Area":
                "Geographic scoring numeric validity",

            "Requirement":
                "All component and summary score fields "
                "must remain finite.",

            "Observed Evidence":
                f"{len(final_score_fields)} geographic "
                "numeric fields assessed",

            "Passed":
                score_valid,
        },
        {
            "Validation Area":
                "Component score bounds",

            "Requirement":
                "All four geographic component scores "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Four component scores assessed "
                "for all tracks",

            "Passed":
                component_bounds_valid,
        },
        {
            "Validation Area":
                "Summary score bounds",

            "Requirement":
                "Geographic Intelligence Score and "
                "balance index must remain between 0 and 1.",

            "Observed Evidence":
                "Summary scoring fields assessed "
                "for all tracks",

            "Passed":
                final_score_bounds_valid,
        },
        {
            "Validation Area":
                "Intelligence classification completeness",

            "Requirement":
                "Every track must receive a geographic "
                "intelligence classification.",

            "Observed Evidence":
                f"{track_geographic_intelligence_df['geographic_intelligence_class'].notna().sum():,} "
                "intelligence classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Performance profile completeness",

            "Requirement":
                "Every track must receive a geographic "
                "performance profile.",

            "Observed Evidence":
                f"{track_geographic_intelligence_df['geographic_performance_profile'].notna().sum():,} "
                "performance profiles assigned",

            "Passed":
                profile_complete,
        },
        {
            "Validation Area":
                "Dimension identification completeness",

            "Requirement":
                "Every track must receive strongest and "
                "weakest geographic dimensions.",

            "Observed Evidence":
                "Strongest and weakest geographic dimensions "
                "assigned for all tracks",

            "Passed":
                dimension_complete,
        },
        {
            "Validation Area":
                "Intelligence population reconciliation",

            "Requirement":
                "Geographic intelligence classifications "
                "must reconcile with the full population.",

            "Observed Evidence":
                f"{int(track_geographic_intelligence_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_geographic_intelligence_df):,} tracks",

            "Passed":
                classification_reconciled,
        },
        {
            "Validation Area":
                "Profile population reconciliation",

            "Requirement":
                "Geographic performance profiles must "
                "reconcile with the full population.",

            "Observed Evidence":
                f"{int(track_geographic_performance_profile_summary_df['Tracks'].sum()):,} "
                f"profile records for "
                f"{len(track_geographic_intelligence_df):,} tracks",

            "Passed":
                profile_reconciled,
        },
    ]
)


print()
print("Geographic intelligence summary validation")
print("=" * 100)

display(
    section_10_5_validation_df
)


# ------------------------------------------------------------
# 22. Final completion status
# ------------------------------------------------------------

section_10_5_complete = bool(
    section_10_5_validation_df[
        "Passed"
    ].all()
)


section_10_complete = bool(
    section_10_1_complete
    and
    section_10_2_complete
    and
    section_10_3_complete
    and
    section_10_4_complete
    and
    section_10_5_complete
)


if not section_10_5_complete:

    failed_checks = (
        section_10_5_validation_df
        .loc[
            ~section_10_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 10.5 Geographic Intelligence Summary "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 10.5 Geographic Intelligence Summary "
    "validation checks passed."
)

print(
    f"Section 10.5 completion status: "
    f"{section_10_5_complete}"
)

print(
    f"Section 10 overall completion status: "
    f"{section_10_complete}"
)

print(
    f"Geographic Intelligence Summaries prepared "
    f"for {len(track_geographic_intelligence_df):,} tracks."
)

print(
    "PMIP geographic reach, penetration, diversification "
    "and expansion intelligence are now integrated into "
    "a validated scoring framework."
)

print(
    "Geographic Intelligence Scoring and Classification "
    "is complete."
)
```

At the same time, the **12,180 tracks** classified as having concentrated reach with strong penetration are particularly interesting. These tracks may not have broad international coverage, but they perform strongly within the markets they do reach. This demonstrates why geographic reach and market penetration must be considered separately rather than treating them as the same concept.

- **30% Geographic Reach**
- **30% Market Penetration**
- **20% Geographic Concentration / Diversification**

The median geographic balance index is **0.9247**, and **101,208 tracks** have a balance index of at least **0.75**. This high balance value should not automatically be interpreted as strong overall geographic performance.

The balance index measures how similar the four component scores are to each other. Therefore, a track with four consistently low scores can still have a high balance index. The balance measure should therefore be interpreted alongside the overall Geographic Intelligence Score rather than independently.

### 11.1 Country-Level Findings

#### Purpose

Summarize the most important country-level geographic findings using the validated market intelligence created earlier in the notebook.

This subsection compares country markets using streaming performance, chart performance, observation coverage, and market context.

The analysis identifies high-performing, high-volume, and comparatively strong country markets while retaining the reporting-context safeguards established in Section 6.

The results provide a country-level view of where the strongest observed PMIP streaming and chart activity occurs.

```python
import pandas as pd
import numpy as np

print("Summarising PMIP country-level findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 10 completion
# ------------------------------------------------------------

section_10_status = globals().get(
    "section_10_complete",
    False
)

print(
    f"Section 10 overall completion status: "
    f"{section_10_status}"
)

if not section_10_status:
    raise RuntimeError(
        "Section 11.1 requires Section 10 "
        "Geographic Intelligence Scoring and Classification "
        "to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated country performance source
# ------------------------------------------------------------

country_candidate_names = [
    "country_performance_df",
]

country_findings_df = None
country_source_name = None

for name in country_candidate_names:

    if name not in globals():
        continue

    candidate = globals()[name]

    if not isinstance(candidate, pd.DataFrame):
        continue

    if "country" not in candidate.columns:
        continue

    if candidate["country"].nunique() != 76:
        continue

    country_findings_df = candidate.copy()
    country_source_name = name
    break


if country_findings_df is None:
    raise RuntimeError(
        "Section 11.1 could not recover the validated "
        "country-level performance intelligence."
    )


print()
print("Validated country-level findings source")
print("=" * 100)

print(
    f"Source dataframe: "
    f"{country_source_name}"
)

print(
    f"Country markets: "
    f"{country_findings_df['country'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Detect validated country-level metrics
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


observations_col = first_existing_column(
    country_findings_df,
    [
        "observations",
        "observation_count",
        "total_observations",
    ]
)


total_streams_col = first_existing_column(
    country_findings_df,
    [
        "total_streams",
        "streams",
    ]
)


median_streams_col = first_existing_column(
    country_findings_df,
    [
        "median_streams",
        "median_streams_per_observation",
        "median_country_streams",
    ]
)


mean_streams_col = first_existing_column(
    country_findings_df,
    [
        "mean_streams",
        "mean_streams_per_observation",
    ]
)


median_position_col = first_existing_column(
    country_findings_df,
    [
        "median_position",
        "median_chart_position",
        "median_country_chart_position",
    ]
)


top_10_rate_col = first_existing_column(
    country_findings_df,
    [
        "top_10_rate_pct",
        "top_10_rate",
        "top10_rate_pct",
    ]
)


top_50_rate_col = first_existing_column(
    country_findings_df,
    [
        "top_50_rate_pct",
        "top_50_rate",
        "top50_rate_pct",
    ]
)


number_one_rate_col = first_existing_column(
    country_findings_df,
    [
        "number_one_rate_pct",
        "number_one_rate",
        "number_1_rate_pct",
    ]
)


# ------------------------------------------------------------
# 4. Validate required analytical fields
# ------------------------------------------------------------

required_findings = {
    "observations": observations_col,
    "median_streams": median_streams_col,
    "median_position": median_position_col,
    "top_10_rate": top_10_rate_col,
    "top_50_rate": top_50_rate_col,
}

missing_findings = [
    label
    for label, column
    in required_findings.items()
    if column is None
]


if missing_findings:

    raise RuntimeError(
        "Section 11.1 is missing required country-level "
        "analytical fields: "
        + ", ".join(
            missing_findings
        )
    )


# ------------------------------------------------------------
# 5. Convert key findings fields to numeric
# ------------------------------------------------------------

numeric_columns = [
    column
    for column in [
        observations_col,
        total_streams_col,
        median_streams_col,
        mean_streams_col,
        median_position_col,
        top_10_rate_col,
        top_50_rate_col,
        number_one_rate_col,
    ]
    if column is not None
]


for column in numeric_columns:

    country_findings_df[
        column
    ] = pd.to_numeric(
        country_findings_df[
            column
        ],
        errors="coerce",
    )


if country_findings_df[
    numeric_columns
].isna().all().any():

    raise RuntimeError(
        "Section 11.1 found unusable numeric country-level fields."
    )


# ------------------------------------------------------------
# 6. Remove global aggregate if present
# ------------------------------------------------------------

country_findings_df = (
    country_findings_df
    .loc[
        country_findings_df[
            "country"
        ]
        .astype(str)
        .str.lower()
        .ne("global")
    ]
    .copy()
)


# ------------------------------------------------------------
# 7. Overall country-level findings
# ------------------------------------------------------------

print()
print("Country-level findings overview")
print("=" * 100)

print(
    f"Country markets assessed: "
    f"{country_findings_df['country'].nunique():,}"
)

print(
    f"Median observations per country: "
    f"{country_findings_df[observations_col].median():,.1f}"
)

print(
    f"Median streams per observation: "
    f"{country_findings_df[median_streams_col].median():,.2f}"
)

print(
    f"Median country chart position: "
    f"{country_findings_df[median_position_col].median():.2f}"
)

print(
    f"Median country top-10 rate: "
    f"{country_findings_df[top_10_rate_col].median():.2f}%"
)

print(
    f"Median country top-50 rate: "
    f"{country_findings_df[top_50_rate_col].median():.2f}%"
)


# ------------------------------------------------------------
# 8. Highest streaming-intensity markets
# ------------------------------------------------------------

stream_display_columns = [
    column
    for column in [
        "country",
        observations_col,
        total_streams_col,
        median_streams_col,
        mean_streams_col,
    ]
    if column is not None
]


highest_streaming_countries_df = (
    country_findings_df
    .sort_values(
        [
            median_streams_col,
            observations_col,
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(15)
    [
        stream_display_columns
    ]
    .reset_index(drop=True)
)


print()
print("Countries with strongest observed streaming performance")
print("=" * 100)

display(
    highest_streaming_countries_df
)


# ------------------------------------------------------------
# 9. Strongest chart-performing markets
# ------------------------------------------------------------

chart_display_columns = [
    column
    for column in [
        "country",
        observations_col,
        median_position_col,
        top_10_rate_col,
        top_50_rate_col,
        number_one_rate_col,
    ]
    if column is not None
]


strongest_chart_countries_df = (
    country_findings_df
    .sort_values(
        [
            top_10_rate_col,
            median_position_col,
        ],
        ascending=[
            False,
            True,
        ],
    )
    .head(15)
    [
        chart_display_columns
    ]
    .reset_index(drop=True)
)


print()
print("Countries with strongest observed chart performance")
print("=" * 100)

display(
    strongest_chart_countries_df
)


# ------------------------------------------------------------
# 10. Largest observed market contexts
# ------------------------------------------------------------

largest_observed_markets_df = (
    country_findings_df
    .sort_values(
        observations_col,
        ascending=False,
    )
    .head(15)
    [
        [
            column
            for column in [
                "country",
                observations_col,
                median_streams_col,
                median_position_col,
                top_10_rate_col,
            ]
            if column is not None
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Country markets with largest observation coverage")
print("=" * 100)

display(
    largest_observed_markets_df
)


# ------------------------------------------------------------
# 11. Build normalized country comparison indicators
# ------------------------------------------------------------

country_findings_df[
    "streaming_strength_percentile"
] = (
    country_findings_df[
        median_streams_col
    ]
    .rank(
        pct=True,
        method="average",
    )
)


country_findings_df[
    "chart_strength_percentile"
] = (
    country_findings_df[
        top_10_rate_col
    ]
    .rank(
        pct=True,
        method="average",
    )
)


country_findings_df[
    "market_context_percentile"
] = (
    country_findings_df[
        observations_col
    ]
    .rank(
        pct=True,
        method="average",
    )
)


# ------------------------------------------------------------
# 12. Build country findings index
# ------------------------------------------------------------

country_findings_df[
    "country_findings_index"
] = (
    (
        0.40
        * country_findings_df[
            "streaming_strength_percentile"
        ]
    )
    +
    (
        0.35
        * country_findings_df[
            "chart_strength_percentile"
        ]
    )
    +
    (
        0.25
        * country_findings_df[
            "market_context_percentile"
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 13. Country findings classification
# ------------------------------------------------------------

def classify_country_finding(value):

    if value >= 0.80:
        return "High-priority country market"

    if value >= 0.60:
        return "Strong country market"

    if value >= 0.40:
        return "Moderate country market"

    return "Lower-priority country market"


country_findings_df[
    "country_findings_class"
] = (
    country_findings_df[
        "country_findings_index"
    ]
    .apply(
        classify_country_finding
    )
)


# ------------------------------------------------------------
# 14. Highest overall country findings
# ------------------------------------------------------------

country_summary_columns = [
    column
    for column in [
        "country",
        observations_col,
        median_streams_col,
        median_position_col,
        top_10_rate_col,
        top_50_rate_col,
        "streaming_strength_percentile",
        "chart_strength_percentile",
        "market_context_percentile",
        "country_findings_index",
        "country_findings_class",
    ]
    if column is not None
]


highest_country_findings_df = (
    country_findings_df
    .sort_values(
        "country_findings_index",
        ascending=False,
    )
    .head(20)
    [
        country_summary_columns
    ]
    .reset_index(drop=True)
)


print()
print("Highest-priority country-level findings")
print("=" * 100)

display(
    highest_country_findings_df
)


# ------------------------------------------------------------
# 15. Country findings classification summary
# ------------------------------------------------------------

country_class_order = [
    "High-priority country market",
    "Strong country market",
    "Moderate country market",
    "Lower-priority country market",
]


country_findings_summary_df = (
    country_findings_df[
        "country_findings_class"
    ]
    .value_counts()
    .reindex(
        country_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Country Findings Class"
    )
    .reset_index(
        name="Country Markets"
    )
)


country_findings_summary_df[
    "Share (%)"
] = (
    country_findings_summary_df[
        "Country Markets"
    ]
    /
    len(country_findings_df)
    * 100
)


print()
print("Country-level findings classification")
print("=" * 100)

display(
    country_findings_summary_df
)


# ------------------------------------------------------------
# 16. Diagnostics
# ------------------------------------------------------------

country_findings_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Country markets assessed",

            "Observed Value":
                int(
                    country_findings_df[
                        "country"
                    ].nunique()
                ),
        },
        {
            "Diagnostic":
                "High-priority country markets",

            "Observed Value":
                int(
                    (
                        country_findings_df[
                            "country_findings_class"
                        ]
                        ==
                        "High-priority country market"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Countries above median streaming strength",

            "Observed Value":
                int(
                    (
                        country_findings_df[
                            median_streams_col
                        ]
                        >
                        country_findings_df[
                            median_streams_col
                        ].median()
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Countries above median top-10 rate",

            "Observed Value":
                int(
                    (
                        country_findings_df[
                            top_10_rate_col
                        ]
                        >
                        country_findings_df[
                            top_10_rate_col
                        ].median()
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Countries above median observation coverage",

            "Observed Value":
                int(
                    (
                        country_findings_df[
                            observations_col
                        ]
                        >
                        country_findings_df[
                            observations_col
                        ].median()
                    ).sum()
                ),
        },
    ]
)


print()
print("Country-level findings diagnostics")
print("=" * 100)

display(
    country_findings_diagnostics_df
)


# ------------------------------------------------------------
# 17. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        country_findings_df
    )
    == 76
)


country_unique = bool(
    country_findings_df[
        "country"
    ].nunique()
    == 76
)


findings_numeric_fields = [
    "streaming_strength_percentile",
    "chart_strength_percentile",
    "market_context_percentile",
    "country_findings_index",
]


findings_numeric_valid = bool(
    np.isfinite(
        country_findings_df[
            findings_numeric_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


findings_bounds_valid = bool(
    country_findings_df[
        findings_numeric_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    country_findings_df[
        "country_findings_class"
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        country_findings_summary_df[
            "Country Markets"
        ].sum()
    )
    ==
    len(
        country_findings_df
    )
)


section_11_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 10 completion",

            "Requirement":
                "Geographic Intelligence Scoring and "
                "Classification must already be complete.",

            "Observed Evidence":
                f"Section 10 overall completion status: "
                f"{section_10_status}",

            "Passed":
                section_10_status,
        },
        {
            "Validation Area":
                "Country population preservation",

            "Requirement":
                "All 76 validated country markets "
                "must be retained.",

            "Observed Evidence":
                f"{len(country_findings_df):,} "
                "country markets assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Country uniqueness",

            "Requirement":
                "Each country market must appear once.",

            "Observed Evidence":
                f"{country_findings_df['country'].nunique():,} "
                "unique countries",

            "Passed":
                country_unique,
        },
        {
            "Validation Area":
                "Country findings numeric validity",

            "Requirement":
                "Country comparison indicators must "
                "remain finite.",

            "Observed Evidence":
                f"{len(findings_numeric_fields)} "
                "normalized country indicators assessed",

            "Passed":
                findings_numeric_valid,
        },
        {
            "Validation Area":
                "Country findings bounds",

            "Requirement":
                "Normalized country indicators must "
                "remain between 0 and 1.",

            "Observed Evidence":
                "Country findings indicators assessed "
                "for all markets",

            "Passed":
                findings_bounds_valid,
        },
        {
            "Validation Area":
                "Country classification completeness",

            "Requirement":
                "Every country must receive a findings "
                "classification.",

            "Observed Evidence":
                f"{country_findings_df['country_findings_class'].notna().sum():,} "
                "country classifications assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Country population reconciliation",

            "Requirement":
                "Country classifications must reconcile "
                "with the full country population.",

            "Observed Evidence":
                f"{int(country_findings_summary_df['Country Markets'].sum()):,} "
                f"classified records for "
                f"{len(country_findings_df):,} country markets",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Country-level findings validation")
print("=" * 100)

display(
    section_11_1_validation_df
)


# ------------------------------------------------------------
# 18. Completion status
# ------------------------------------------------------------

section_11_1_complete = bool(
    section_11_1_validation_df[
        "Passed"
    ].all()
)


if not section_11_1_complete:

    failed_checks = (
        section_11_1_validation_df
        .loc[
            ~section_11_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 11.1 Country-Level Findings "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 11.1 country-level findings "
    "validation checks passed."
)

print(
    f"Section 11.1 completion status: "
    f"{section_11_1_complete}"
)

print(
    f"Country-level findings prepared for "
    f"{len(country_findings_df):,} country markets."
)

print(
    "The workflow is ready for "
    "Section 11.2 Artist-Level Findings."
)
```

### 11.2 Artist-Level Findings

#### Purpose

Summarize the strongest artist-level geographic findings using the validated artist intelligence developed in Section 7.

This subsection examines how artist representations differ in international reach, market penetration, geographic concentration, local versus international performance, and overall geographic profile.

The aim is to identify the main artist-level geographic patterns in PMIP and highlight which types of artist representations demonstrate the strongest evidence of international breadth, market depth, diversification, and geographic maturity.

The findings remain descriptive and should not be interpreted as proof that geographic distribution directly causes artist success.

```python
import pandas as pd
import numpy as np

print("Summarising PMIP artist-level findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_11_1_status = globals().get(
    "section_11_1_complete",
    False
)

print(
    f"Section 11.1 completion status: "
    f"{section_11_1_status}"
)

if not section_11_1_status:
    raise RuntimeError(
        "Section 11.2 requires Section 11.1 "
        "Country-Level Findings to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated artist geographic intelligence sources
# ------------------------------------------------------------

required_artist_sources = {
    "reach": "artist_international_reach_df",
    "penetration": "artist_market_penetration_intelligence_df",
    "concentration": "artist_geographic_concentration_intelligence_df",
    "local_international": "artist_local_international_df",
    "profile": "artist_geographic_profile_df",
}

artist_sources = {}

for source_key, variable_name in required_artist_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 11.2 could not recover required artist source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    if len(candidate) != 49_843:
        raise RuntimeError(
            f"{variable_name} does not contain the expected "
            f"49,843 artist representations."
        )

    artist_sources[source_key] = candidate.copy()


artist_reach_df = artist_sources["reach"]
artist_penetration_df = artist_sources["penetration"]
artist_concentration_df = artist_sources["concentration"]
artist_local_international_df = artist_sources["local_international"]
artist_profile_df = artist_sources["profile"]


print()
print("Validated artist-level findings sources")
print("=" * 100)

print(
    "International reach source: "
    "artist_international_reach_df"
)

print(
    "Market penetration source: "
    "artist_market_penetration_intelligence_df"
)

print(
    "Geographic concentration source: "
    "artist_geographic_concentration_intelligence_df"
)

print(
    "Local/international source: "
    "artist_local_international_df"
)

print(
    "Geographic profile source: "
    "artist_geographic_profile_df"
)

print(
    f"Artist representations: "
    f"{len(artist_profile_df):,}"
)


# ------------------------------------------------------------
# 3. Detect artist identity column
# ------------------------------------------------------------

artist_identity_col = None

for column in [
    "artists_raw",
    "artist",
    "artist_name",
]:
    if column in artist_profile_df.columns:
        artist_identity_col = column
        break


if artist_identity_col is None:
    raise RuntimeError(
        "Section 11.2 could not identify the artist identity column."
    )


# ------------------------------------------------------------
# 4. Recover important artist intelligence fields
# ------------------------------------------------------------

def first_existing_column(df, candidates):

    for column in candidates:
        if column in df.columns:
            return column

    return None


reach_index_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_index",
        "artist_international_reach_index",
    ]
)


markets_reached_col = first_existing_column(
    artist_reach_df,
    [
        "markets_reached",
        "country_markets_reached",
    ]
)


reach_class_col = first_existing_column(
    artist_reach_df,
    [
        "international_reach_class",
    ]
)


penetration_index_col = first_existing_column(
    artist_penetration_df,
    [
        "market_penetration_index",
    ]
)


penetration_class_col = first_existing_column(
    artist_penetration_df,
    [
        "market_penetration_class",
    ]
)


concentration_hhi_col = first_existing_column(
    artist_concentration_df,
    [
        "stream_concentration_hhi",
    ]
)


effective_markets_col = first_existing_column(
    artist_concentration_df,
    [
        "effective_stream_markets",
    ]
)


dependency_class_col = first_existing_column(
    artist_concentration_df,
    [
        "market_dependency_class",
    ]
)


local_international_profile_col = first_existing_column(
    artist_local_international_df,
    [
        "local_international_profile",
        "local_vs_international_profile",
    ]
)


geographic_profile_index_col = first_existing_column(
    artist_profile_df,
    [
        "geographic_profile_index",
    ]
)


geographic_profile_class_col = first_existing_column(
    artist_profile_df,
    [
        "artist_geographic_profile",
        "geographic_profile_class",
    ]
)


required_fields = {
    "reach index": reach_index_col,
    "markets reached": markets_reached_col,
    "reach class": reach_class_col,
    "penetration index": penetration_index_col,
    "penetration class": penetration_class_col,
    "stream concentration HHI": concentration_hhi_col,
    "effective stream markets": effective_markets_col,
    "market dependency class": dependency_class_col,
    "local/international profile": local_international_profile_col,
    "geographic profile index": geographic_profile_index_col,
    "geographic profile class": geographic_profile_class_col,
}

missing_fields = [
    label
    for label, column
    in required_fields.items()
    if column is None
]


if missing_fields:
    raise RuntimeError(
        "Section 11.2 is missing required artist-level fields: "
        + ", ".join(missing_fields)
    )


# ------------------------------------------------------------
# 5. Build unified artist findings table
# ------------------------------------------------------------

artist_findings_df = (
    artist_profile_df[
        [
            artist_identity_col,
            geographic_profile_index_col,
            geographic_profile_class_col,
        ]
    ]
    .copy()
)


artist_findings_df = (
    artist_findings_df
    .merge(
        artist_reach_df[
            [
                artist_identity_col,
                markets_reached_col,
                reach_index_col,
                reach_class_col,
            ]
        ],
        on=artist_identity_col,
        how="left",
        validate="one_to_one",
    )
    .merge(
        artist_penetration_df[
            [
                artist_identity_col,
                penetration_index_col,
                penetration_class_col,
            ]
        ],
        on=artist_identity_col,
        how="left",
        validate="one_to_one",
    )
    .merge(
        artist_concentration_df[
            [
                artist_identity_col,
                concentration_hhi_col,
                effective_markets_col,
                dependency_class_col,
            ]
        ],
        on=artist_identity_col,
        how="left",
        validate="one_to_one",
    )
    .merge(
        artist_local_international_df[
            [
                artist_identity_col,
                local_international_profile_col,
            ]
        ],
        on=artist_identity_col,
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 6. Numeric safety
# ------------------------------------------------------------

numeric_columns = [
    markets_reached_col,
    reach_index_col,
    penetration_index_col,
    concentration_hhi_col,
    effective_markets_col,
    geographic_profile_index_col,
]


for column in numeric_columns:

    artist_findings_df[
        column
    ] = pd.to_numeric(
        artist_findings_df[
            column
        ],
        errors="coerce",
    )


if artist_findings_df[
    numeric_columns
].isna().any().any():

    raise RuntimeError(
        "Section 11.2 found missing or invalid "
        "numeric artist-level intelligence values."
    )


# ------------------------------------------------------------
# 7. Build diversification indicator
# ------------------------------------------------------------

artist_findings_df[
    "artist_diversification_index"
] = (
    1
    -
    artist_findings_df[
        concentration_hhi_col
    ]
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 8. Build combined artist findings index
# ------------------------------------------------------------

artist_findings_df[
    "artist_findings_index"
] = (
    (
        0.30
        * artist_findings_df[
            reach_index_col
        ]
    )
    +
    (
        0.30
        * artist_findings_df[
            penetration_index_col
        ]
    )
    +
    (
        0.20
        * artist_findings_df[
            "artist_diversification_index"
        ]
    )
    +
    (
        0.20
        * artist_findings_df[
            geographic_profile_index_col
        ]
    )
).clip(
    lower=0,
    upper=1,
)


# ------------------------------------------------------------
# 9. Artist-level findings classification
# ------------------------------------------------------------

def classify_artist_finding(value):

    if value >= 0.75:
        return "Exceptional artist geographic strength"

    if value >= 0.50:
        return "Strong artist geographic strength"

    if value >= 0.30:
        return "Moderate artist geographic strength"

    if value >= 0.15:
        return "Developing artist geographic strength"

    return "Limited artist geographic strength"


artist_findings_df[
    "artist_findings_class"
] = (
    artist_findings_df[
        "artist_findings_index"
    ]
    .apply(
        classify_artist_finding
    )
)


# ------------------------------------------------------------
# 10. Overview
# ------------------------------------------------------------

print()
print("Artist-level findings overview")
print("=" * 100)

print(
    f"Artist representations assessed: "
    f"{len(artist_findings_df):,}"
)

print(
    f"Median markets reached: "
    f"{artist_findings_df[markets_reached_col].median():.1f}"
)

print(
    f"Median international reach index: "
    f"{artist_findings_df[reach_index_col].median():.4f}"
)

print(
    f"Median market penetration index: "
    f"{artist_findings_df[penetration_index_col].median():.4f}"
)

print(
    f"Median stream concentration HHI: "
    f"{artist_findings_df[concentration_hhi_col].median():.4f}"
)

print(
    f"Median geographic profile index: "
    f"{artist_findings_df[geographic_profile_index_col].median():.4f}"
)

print(
    f"Median artist findings index: "
    f"{artist_findings_df['artist_findings_index'].median():.4f}"
)

print(
    f"Mean artist findings index: "
    f"{artist_findings_df['artist_findings_index'].mean():.4f}"
)


# ------------------------------------------------------------
# 11. Geographic profile distribution
# ------------------------------------------------------------

artist_geographic_profile_summary_df = (
    artist_findings_df[
        geographic_profile_class_col
    ]
    .value_counts()
    .rename_axis(
        "Artist Geographic Profile"
    )
    .reset_index(
        name="Artist Representations"
    )
)


artist_geographic_profile_summary_df[
    "Share (%)"
] = (
    artist_geographic_profile_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_findings_df)
    * 100
)


print()
print("Artist geographic profile distribution")
print("=" * 100)

display(
    artist_geographic_profile_summary_df
)


# ------------------------------------------------------------
# 12. Local versus international profile distribution
# ------------------------------------------------------------

artist_local_international_summary_df = (
    artist_findings_df[
        local_international_profile_col
    ]
    .value_counts()
    .rename_axis(
        "Local / International Profile"
    )
    .reset_index(
        name="Artist Representations"
    )
)


artist_local_international_summary_df[
    "Share (%)"
] = (
    artist_local_international_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_findings_df)
    * 100
)


print()
print("Artist local versus international profile distribution")
print("=" * 100)

display(
    artist_local_international_summary_df
)


# ------------------------------------------------------------
# 13. Highest artist-level geographic findings
# ------------------------------------------------------------

display_columns = [
    artist_identity_col,
    markets_reached_col,
    reach_index_col,
    reach_class_col,
    penetration_index_col,
    penetration_class_col,
    concentration_hhi_col,
    effective_markets_col,
    dependency_class_col,
    geographic_profile_index_col,
    geographic_profile_class_col,
    local_international_profile_col,
    "artist_findings_index",
    "artist_findings_class",
]


highest_artist_findings_df = (
    artist_findings_df
    .sort_values(
        [
            "artist_findings_index",
            reach_index_col,
            penetration_index_col,
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)


print()
print("Artist representations with strongest geographic findings")
print("=" * 100)

display(
    highest_artist_findings_df
)


# ------------------------------------------------------------
# 14. Artist findings classification summary
# ------------------------------------------------------------

artist_findings_class_order = [
    "Exceptional artist geographic strength",
    "Strong artist geographic strength",
    "Moderate artist geographic strength",
    "Developing artist geographic strength",
    "Limited artist geographic strength",
]


artist_findings_summary_df = (
    artist_findings_df[
        "artist_findings_class"
    ]
    .value_counts()
    .reindex(
        artist_findings_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Artist Findings Class"
    )
    .reset_index(
        name="Artist Representations"
    )
)


artist_findings_summary_df[
    "Share (%)"
] = (
    artist_findings_summary_df[
        "Artist Representations"
    ]
    /
    len(artist_findings_df)
    * 100
)


print()
print("Artist-level findings classification")
print("=" * 100)

display(
    artist_findings_summary_df
)


# ------------------------------------------------------------
# 15. Diagnostics
# ------------------------------------------------------------

artist_findings_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Artist representations assessed",

            "Observed Value":
                int(
                    len(
                        artist_findings_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Artists reaching at least 10 markets",

            "Observed Value":
                int(
                    (
                        artist_findings_df[
                            markets_reached_col
                        ]
                        >= 10
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists reaching at least 25 markets",

            "Observed Value":
                int(
                    (
                        artist_findings_df[
                            markets_reached_col
                        ]
                        >= 25
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with penetration index >= 0.50",

            "Observed Value":
                int(
                    (
                        artist_findings_df[
                            penetration_index_col
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with HHI < 0.50",

            "Observed Value":
                int(
                    (
                        artist_findings_df[
                            concentration_hhi_col
                        ]
                        < 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Artists with geographic profile index >= 0.50",

            "Observed Value":
                int(
                    (
                        artist_findings_df[
                            geographic_profile_index_col
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
    ]
)


print()
print("Artist-level findings diagnostics")
print("=" * 100)

display(
    artist_findings_diagnostics_df
)


# ------------------------------------------------------------
# 16. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        artist_findings_df
    )
    == 49_843
)


identity_unique = bool(
    artist_findings_df[
        artist_identity_col
    ].nunique()
    == 49_843
)


findings_numeric_fields = [
    reach_index_col,
    penetration_index_col,
    concentration_hhi_col,
    "artist_diversification_index",
    geographic_profile_index_col,
    "artist_findings_index",
]


numeric_valid = bool(
    np.isfinite(
        artist_findings_df[
            findings_numeric_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


bounded_fields = [
    reach_index_col,
    penetration_index_col,
    concentration_hhi_col,
    "artist_diversification_index",
    geographic_profile_index_col,
    "artist_findings_index",
]


bounds_valid = bool(
    artist_findings_df[
        bounded_fields
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    artist_findings_df[
        "artist_findings_class"
    ]
    .notna()
    .all()
)


profile_complete = bool(
    artist_findings_df[
        geographic_profile_class_col
    ]
    .notna()
    .all()
)


population_reconciled = bool(
    int(
        artist_findings_summary_df[
            "Artist Representations"
        ].sum()
    )
    ==
    len(
        artist_findings_df
    )
)


section_11_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 11.1 completion",

            "Requirement":
                "Country-Level Findings must already be complete.",

            "Observed Evidence":
                f"Section 11.1 completion status: "
                f"{section_11_1_status}",

            "Passed":
                section_11_1_status,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All 49,843 validated artist representations "
                "must be retained.",

            "Observed Evidence":
                f"{len(artist_findings_df):,} "
                "artist representations assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Artist identity uniqueness",

            "Requirement":
                "Each artist representation must appear once.",

            "Observed Evidence":
                f"{artist_findings_df[artist_identity_col].nunique():,} "
                "unique artist representations",

            "Passed":
                identity_unique,
        },
        {
            "Validation Area":
                "Artist findings numeric validity",

            "Requirement":
                "Artist findings indicators must remain finite.",

            "Observed Evidence":
                f"{len(findings_numeric_fields)} artist numeric "
                "fields assessed",

            "Passed":
                numeric_valid,
        },
        {
            "Validation Area":
                "Artist findings bounds",

            "Requirement":
                "Normalized artist findings indicators must "
                "remain between 0 and 1.",

            "Observed Evidence":
                "Bounded artist findings indicators assessed "
                "for all artist representations",

            "Passed":
                bounds_valid,
        },
        {
            "Validation Area":
                "Artist classification completeness",

            "Requirement":
                "Every artist representation must receive "
                "an artist findings classification.",

            "Observed Evidence":
                f"{artist_findings_df['artist_findings_class'].notna().sum():,} "
                "artist findings classes assigned",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Artist profile completeness",

            "Requirement":
                "Every artist representation must retain "
                "a geographic profile.",

            "Observed Evidence":
                f"{artist_findings_df[geographic_profile_class_col].notna().sum():,} "
                "artist geographic profiles retained",

            "Passed":
                profile_complete,
        },
        {
            "Validation Area":
                "Artist population reconciliation",

            "Requirement":
                "Artist findings classifications must reconcile "
                "with the full artist population.",

            "Observed Evidence":
                f"{int(artist_findings_summary_df['Artist Representations'].sum()):,} "
                f"classified records for "
                f"{len(artist_findings_df):,} artist representations",

            "Passed":
                population_reconciled,
        },
    ]
)


print()
print("Artist-level findings validation")
print("=" * 100)

display(
    section_11_2_validation_df
)


# ------------------------------------------------------------
# 17. Completion status
# ------------------------------------------------------------

section_11_2_complete = bool(
    section_11_2_validation_df[
        "Passed"
    ].all()
)


if not section_11_2_complete:

    failed_checks = (
        section_11_2_validation_df
        .loc[
            ~section_11_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 11.2 Artist-Level Findings "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 11.2 artist-level findings "
    "validation checks passed."
)

print(
    f"Section 11.2 completion status: "
    f"{section_11_2_complete}"
)

print(
    f"Artist-level findings prepared for "
    f"{len(artist_findings_df):,} artist representations."
)

print(
    "The workflow is ready for "
    "Section 11.3 Track-Level Findings."
)
```

#### Interpretation

Section 11.2 successfully summarizes the validated **artist-level geographic findings** across all **49,843 artist representations**.

The overall results show that artist geography is still heavily concentrated. The median artist reaches only **1 market**, has a median international reach index of **0.0092**, a median market penetration index of **0.2383**, a median stream-concentration HHI of **1.0000**, and a median geographic profile index of **0.0722**.

The median artist findings index is **0.1390**, while the mean is slightly higher at **0.1551**. This shows that most artists remain geographically limited, while a smaller group achieves much stronger international reach, market penetration, and diversification.

The geographic profile distribution shows:

- **43,775 artist representations (87.83%)** are **Locally concentrated**
- **3,144 (6.31%)** are **Internationally established**
- **1,964 (3.94%)** are **Emerging international**
- **946 (1.90%)** are **Diversified international**
- **14 (0.03%)** are **Globally distributed**

This confirms that broad international geographic presence is rare within the overall artist population.

The local versus international profile distribution tells a similar story:

- **42,207 artist representations (84.68%)** show **Localised performance**
- **3,532 (7.09%)** are **Emerging international**
- **3,158 (6.34%)** are **Internationally established**
- **946 (1.90%)** are **Broadly diversified international**

The close agreement between these two profile systems strengthens the interpretation that the majority of artists are observed mainly within a narrow geographic footprint.

The artist-level findings classification shows:

- **62 artist representations (0.12%)** have **Exceptional artist geographic strength**
- **927 (1.86%)** have **Strong artist geographic strength**
- **2,997 (6.01%)** have **Moderate artist geographic strength**
- **19,651 (39.43%)** have **Developing artist geographic strength**
- **26,206 (52.58%)** have **Limited artist geographic strength**

This means more than half of all artist representations remain in the limited category, while only a small proportion demonstrate genuinely strong geographic performance.

The diagnostics provide useful evidence about the strongest artist groups:

- **4,080 artists** reach at least **10 markets**
- **1,733 artists** reach at least **25 markets**
- **14,196 artists** have a market penetration index of at least **0.50**
- **4,627 artists** have an HHI below **0.50**, indicating lower concentration and stronger diversification
- **976 artists** achieve a geographic profile index of at least **0.50**

These results show an important difference between **reach** and **penetration**. A much larger number of artists achieve strong market penetration than broad international reach. This means many artists perform strongly within the markets they already reach without yet building a broad international footprint.

The combined Artist Findings Index reflects four dimensions:

- **international reach**
- **market penetration**
- **geographic diversification**
- **overall geographic profile**

Reach and penetration receive the largest weighting because they represent the breadth and depth of artist performance, while diversification and profile strength provide additional evidence about how mature and balanced that geographic presence is.

The artist findings should therefore not be interpreted as a simple popularity ranking. An artist can demonstrate strong local or regional penetration while still receiving a moderate overall geographic score because international reach remains narrow.

All Section 11.2 validation checks passed. The complete **49,843 artist-representation population** was preserved, artist identities remained unique, all six numeric findings fields remained finite, normalized indicators stayed within the expected **0-to-1 range**, every artist retained a geographic profile and findings classification, and the final classification totals reconciled with the full population.

Section 11.2 therefore establishes the main **artist-level geographic findings layer** of PMIP.

The workflow is ready for **Section 11.3 Track-Level Findings**.

### 11.3 Track-Level Findings

#### Purpose

Summarize the most important track-level geographic findings using the validated PMIP track intelligence created in Sections 8, 9, and 10.

This subsection brings together track-level reach, market penetration, geographic diversification, expansion behaviour, and the final Geographic Intelligence Score.

The analysis identifies how tracks differ in geographic breadth, depth, concentration, movement, and overall performance profile.

The aim is to highlight the strongest track-level geographic patterns while preserving the distinction between broad reach, strong local penetration, geographic diversification, and expansion momentum.

```python
import pandas as pd
import numpy as np

print("Summarising PMIP track-level findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_11_2_status = globals().get(
    "section_11_2_complete",
    False
)

print(
    f"Section 11.2 completion status: "
    f"{section_11_2_status}"
)

if not section_11_2_status:
    raise RuntimeError(
        "Section 11.3 requires Section 11.2 "
        "Artist-Level Findings to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated final track geographic intelligence
# ------------------------------------------------------------

if "track_geographic_intelligence_df" not in globals():
    raise RuntimeError(
        "Section 11.3 could not recover "
        "track_geographic_intelligence_df from Section 10.5."
    )


track_findings_df = (
    globals()[
        "track_geographic_intelligence_df"
    ]
    .copy()
)


if not isinstance(
    track_findings_df,
    pd.DataFrame,
):
    raise RuntimeError(
        "track_geographic_intelligence_df is not "
        "a valid dataframe."
    )


if len(track_findings_df) != 110_198:
    raise RuntimeError(
        "Section 11.3 expected 110,198 track records."
    )


print()
print("Validated track-level findings source")
print("=" * 100)

print(
    "Source dataframe: "
    "track_geographic_intelligence_df"
)

print(
    f"Tracks: "
    f"{track_findings_df['track_id'].nunique():,}"
)


# ------------------------------------------------------------
# 3. Validate required track findings fields
# ------------------------------------------------------------

required_columns = [
    "track_id",
    "geographic_reach_score",
    "market_penetration_score",
    "geographic_concentration_score",
    "geographic_expansion_indicator",
    "geographic_intelligence_score",
    "geographic_balance_index",
    "geographic_intelligence_class",
    "geographic_performance_profile",
    "strongest_geographic_dimension",
    "weakest_geographic_dimension",
]


missing_columns = [
    column
    for column in required_columns
    if column not in track_findings_df.columns
]


if missing_columns:
    raise RuntimeError(
        "Section 11.3 is missing required "
        "track findings fields: "
        + ", ".join(
            missing_columns
        )
    )


# ------------------------------------------------------------
# 4. Numeric safety
# ------------------------------------------------------------

score_columns = [
    "geographic_reach_score",
    "market_penetration_score",
    "geographic_concentration_score",
    "geographic_expansion_indicator",
    "geographic_intelligence_score",
    "geographic_balance_index",
]


for column in score_columns:

    track_findings_df[
        column
    ] = pd.to_numeric(
        track_findings_df[
            column
        ],
        errors="coerce",
    )


if track_findings_df[
    score_columns
].isna().any().any():

    raise RuntimeError(
        "Section 11.3 found missing or invalid "
        "track-level geographic score values."
    )


# ------------------------------------------------------------
# 5. Overall track findings overview
# ------------------------------------------------------------

print()
print("Track-level findings overview")
print("=" * 100)

print(
    f"Tracks assessed: "
    f"{len(track_findings_df):,}"
)

print(
    f"Median Geographic Reach Score: "
    f"{track_findings_df['geographic_reach_score'].median():.4f}"
)

print(
    f"Median Market Penetration Score: "
    f"{track_findings_df['market_penetration_score'].median():.4f}"
)

print(
    f"Median Geographic Concentration Score: "
    f"{track_findings_df['geographic_concentration_score'].median():.4f}"
)

print(
    f"Median Geographic Expansion Indicator: "
    f"{track_findings_df['geographic_expansion_indicator'].median():.4f}"
)

print(
    f"Median Geographic Intelligence Score: "
    f"{track_findings_df['geographic_intelligence_score'].median():.4f}"
)

print(
    f"Mean Geographic Intelligence Score: "
    f"{track_findings_df['geographic_intelligence_score'].mean():.4f}"
)

print(
    f"Median Geographic Balance Index: "
    f"{track_findings_df['geographic_balance_index'].median():.4f}"
)


# ------------------------------------------------------------
# 6. Geographic Intelligence classification distribution
# ------------------------------------------------------------

intelligence_class_order = [
    "Exceptional geographic intelligence",
    "Strong geographic intelligence",
    "Moderate geographic intelligence",
    "Developing geographic intelligence",
    "Limited geographic intelligence",
]


track_findings_class_summary_df = (
    track_findings_df[
        "geographic_intelligence_class"
    ]
    .value_counts()
    .reindex(
        intelligence_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Intelligence Class"
    )
    .reset_index(
        name="Tracks"
    )
)


track_findings_class_summary_df[
    "Share (%)"
] = (
    track_findings_class_summary_df[
        "Tracks"
    ]
    /
    len(track_findings_df)
    * 100
)


print()
print("Track Geographic Intelligence classification")
print("=" * 100)

display(
    track_findings_class_summary_df
)


# ------------------------------------------------------------
# 7. Geographic performance profile distribution
# ------------------------------------------------------------

profile_order = [
    "Broad high-performance geographic profile",
    "Broad reach with developing penetration",
    "Concentrated reach with strong penetration",
    "Emerging geographic growth profile",
    "Diversified multi-market profile",
    "Limited geographic footprint",
]


track_profile_summary_df = (
    track_findings_df[
        "geographic_performance_profile"
    ]
    .value_counts()
    .reindex(
        profile_order,
        fill_value=0,
    )
    .rename_axis(
        "Geographic Performance Profile"
    )
    .reset_index(
        name="Tracks"
    )
)


track_profile_summary_df[
    "Share (%)"
] = (
    track_profile_summary_df[
        "Tracks"
    ]
    /
    len(track_findings_df)
    * 100
)


print()
print("Track geographic performance profile distribution")
print("=" * 100)

display(
    track_profile_summary_df
)


# ------------------------------------------------------------
# 8. Strongest geographic dimension distribution
# ------------------------------------------------------------

strongest_dimension_summary_df = (
    track_findings_df[
        "strongest_geographic_dimension"
    ]
    .value_counts()
    .rename_axis(
        "Strongest Geographic Dimension"
    )
    .reset_index(
        name="Tracks"
    )
)


strongest_dimension_summary_df[
    "Share (%)"
] = (
    strongest_dimension_summary_df[
        "Tracks"
    ]
    /
    len(track_findings_df)
    * 100
)


print()
print("Strongest geographic dimension across tracks")
print("=" * 100)

display(
    strongest_dimension_summary_df
)


# ------------------------------------------------------------
# 9. Weakest geographic dimension distribution
# ------------------------------------------------------------

weakest_dimension_summary_df = (
    track_findings_df[
        "weakest_geographic_dimension"
    ]
    .value_counts()
    .rename_axis(
        "Weakest Geographic Dimension"
    )
    .reset_index(
        name="Tracks"
    )
)


weakest_dimension_summary_df[
    "Share (%)"
] = (
    weakest_dimension_summary_df[
        "Tracks"
    ]
    /
    len(track_findings_df)
    * 100
)


print()
print("Weakest geographic dimension across tracks")
print("=" * 100)

display(
    weakest_dimension_summary_df
)


# ------------------------------------------------------------
# 10. Highest overall Geographic Intelligence tracks
# ------------------------------------------------------------

display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "geographic_reach_score",
        "market_penetration_score",
        "geographic_concentration_score",
        "geographic_expansion_indicator",
        "geographic_intelligence_score",
        "geographic_balance_index",
        "geographic_intelligence_class",
        "geographic_performance_profile",
        "strongest_geographic_dimension",
        "weakest_geographic_dimension",
    ]
    if column in track_findings_df.columns
]


highest_track_findings_df = (
    track_findings_df
    .sort_values(
        [
            "geographic_intelligence_score",
            "geographic_balance_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(25)
    [
        display_columns
    ]
    .reset_index(drop=True)
)


print()
print("Tracks with strongest overall geographic intelligence")
print("=" * 100)

display(
    highest_track_findings_df
)


# ------------------------------------------------------------
# 11. Highest reach tracks
# ------------------------------------------------------------

highest_reach_tracks_df = (
    track_findings_df
    .sort_values(
        [
            "geographic_reach_score",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(15)
    [
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_reach_score",
                "market_penetration_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in track_findings_df.columns
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Tracks with strongest geographic reach")
print("=" * 100)

display(
    highest_reach_tracks_df
)


# ------------------------------------------------------------
# 12. Highest penetration tracks
# ------------------------------------------------------------

highest_penetration_tracks_df = (
    track_findings_df
    .sort_values(
        [
            "market_penetration_score",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(15)
    [
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "market_penetration_score",
                "geographic_reach_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in track_findings_df.columns
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Tracks with strongest market penetration")
print("=" * 100)

display(
    highest_penetration_tracks_df
)


# ------------------------------------------------------------
# 13. Highest diversification tracks
# ------------------------------------------------------------

highest_diversification_tracks_df = (
    track_findings_df
    .sort_values(
        [
            "geographic_concentration_score",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(15)
    [
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_concentration_score",
                "geographic_reach_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in track_findings_df.columns
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Tracks with strongest geographic diversification")
print("=" * 100)

display(
    highest_diversification_tracks_df
)


# ------------------------------------------------------------
# 14. Highest expansion tracks
# ------------------------------------------------------------

highest_expansion_tracks_df = (
    track_findings_df
    .sort_values(
        [
            "geographic_expansion_indicator",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(15)
    [
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_expansion_indicator",
                "geographic_reach_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in track_findings_df.columns
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Tracks with strongest geographic expansion")
print("=" * 100)

display(
    highest_expansion_tracks_df
)


# ------------------------------------------------------------
# 15. Track-level findings diagnostics
# ------------------------------------------------------------

track_findings_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Tracks assessed",

            "Observed Value":
                int(
                    len(
                        track_findings_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Tracks with Geographic Intelligence Score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "geographic_intelligence_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Geographic Intelligence Score >= 0.75",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "geographic_intelligence_score"
                        ]
                        >= 0.75
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Reach Score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "geographic_reach_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Penetration Score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "market_penetration_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Diversification Score >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "geographic_concentration_score"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with Expansion Indicator >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            "geographic_expansion_indicator"
                        ]
                        >= 0.50
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Tracks with all four geographic dimensions >= 0.50",

            "Observed Value":
                int(
                    (
                        track_findings_df[
                            [
                                "geographic_reach_score",
                                "market_penetration_score",
                                "geographic_concentration_score",
                                "geographic_expansion_indicator",
                            ]
                        ]
                        >= 0.50
                    )
                    .all(
                        axis=1
                    )
                    .sum()
                ),
        },
    ]
)


print()
print("Track-level findings diagnostics")
print("=" * 100)

display(
    track_findings_diagnostics_df
)


# ------------------------------------------------------------
# 16. Validation
# ------------------------------------------------------------

population_valid = bool(
    len(
        track_findings_df
    )
    == 110_198
)


identity_unique = bool(
    track_findings_df[
        "track_id"
    ].nunique()
    == 110_198
)


numeric_valid = bool(
    np.isfinite(
        track_findings_df[
            score_columns
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


bounds_valid = bool(
    track_findings_df[
        score_columns
    ]
    .apply(
        lambda column:
            column.between(
                0,
                1,
                inclusive="both",
            ).all()
    )
    .all()
)


classification_complete = bool(
    track_findings_df[
        "geographic_intelligence_class"
    ]
    .notna()
    .all()
)


profile_complete = bool(
    track_findings_df[
        "geographic_performance_profile"
    ]
    .notna()
    .all()
)


dimension_complete = bool(
    track_findings_df[
        [
            "strongest_geographic_dimension",
            "weakest_geographic_dimension",
        ]
    ]
    .notna()
    .all()
    .all()
)


classification_reconciled = bool(
    int(
        track_findings_class_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_findings_df
    )
)


profile_reconciled = bool(
    int(
        track_profile_summary_df[
            "Tracks"
        ].sum()
    )
    ==
    len(
        track_findings_df
    )
)


section_11_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 11.2 completion",

            "Requirement":
                "Artist-Level Findings must already be complete.",

            "Observed Evidence":
                f"Section 11.2 completion status: "
                f"{section_11_2_status}",

            "Passed":
                section_11_2_status,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All 110,198 validated tracks must be retained.",

            "Observed Evidence":
                f"{len(track_findings_df):,} tracks assessed",

            "Passed":
                population_valid,
        },
        {
            "Validation Area":
                "Track identity uniqueness",

            "Requirement":
                "Each track must appear once.",

            "Observed Evidence":
                f"{track_findings_df['track_id'].nunique():,} "
                "unique tracks",

            "Passed":
                identity_unique,
        },
        {
            "Validation Area":
                "Track findings numeric validity",

            "Requirement":
                "Track findings scores must remain finite.",

            "Observed Evidence":
                f"{len(score_columns)} "
                "track score fields assessed",

            "Passed":
                numeric_valid,
        },
        {
            "Validation Area":
                "Track findings bounds",

            "Requirement":
                "All normalized track findings scores "
                "must remain between 0 and 1.",

            "Observed Evidence":
                "Bounded track geographic scores assessed "
                "for all tracks",

            "Passed":
                bounds_valid,
        },
        {
            "Validation Area":
                "Track classification completeness",

            "Requirement":
                "Every track must retain a Geographic "
                "Intelligence classification.",

            "Observed Evidence":
                f"{track_findings_df['geographic_intelligence_class'].notna().sum():,} "
                "track intelligence classes retained",

            "Passed":
                classification_complete,
        },
        {
            "Validation Area":
                "Track profile completeness",

            "Requirement":
                "Every track must retain a geographic "
                "performance profile.",

            "Observed Evidence":
                f"{track_findings_df['geographic_performance_profile'].notna().sum():,} "
                "track geographic profiles retained",

            "Passed":
                profile_complete,
        },
        {
            "Validation Area":
                "Track dimension completeness",

            "Requirement":
                "Every track must retain strongest and "
                "weakest geographic dimensions.",

            "Observed Evidence":
                "Strongest and weakest dimensions retained "
                "for all tracks",

            "Passed":
                dimension_complete,
        },
        {
            "Validation Area":
                "Track classification reconciliation",

            "Requirement":
                "Track classifications must reconcile "
                "with the full track population.",

            "Observed Evidence":
                f"{int(track_findings_class_summary_df['Tracks'].sum()):,} "
                f"classified records for "
                f"{len(track_findings_df):,} tracks",

            "Passed":
                classification_reconciled,
        },
        {
            "Validation Area":
                "Track profile reconciliation",

            "Requirement":
                "Track profiles must reconcile with "
                "the full track population.",

            "Observed Evidence":
                f"{int(track_profile_summary_df['Tracks'].sum()):,} "
                f"profile records for "
                f"{len(track_findings_df):,} tracks",

            "Passed":
                profile_reconciled,
        },
    ]
)


print()
print("Track-level findings validation")
print("=" * 100)

display(
    section_11_3_validation_df
)


# ------------------------------------------------------------
# 17. Completion status
# ------------------------------------------------------------

section_11_3_complete = bool(
    section_11_3_validation_df[
        "Passed"
    ].all()
)


if not section_11_3_complete:

    failed_checks = (
        section_11_3_validation_df
        .loc[
            ~section_11_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 11.3 Track-Level Findings "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 11.3 track-level findings "
    "validation checks passed."
)

print(
    f"Section 11.3 completion status: "
    f"{section_11_3_complete}"
)

print(
    f"Track-level findings prepared for "
    f"{len(track_findings_df):,} tracks."
)

print(
    "The workflow is ready for "
    "Section 11.4 Geographic Comparisons."
)
```

#### Interpretation

Section 11.3 successfully summarizes the validated **track-level geographic findings** across all **110,198 tracks**.

The overall results confirm that most tracks have limited geographic strength when all four dimensions are considered together.

The median scores are:

- Geographic Reach Score: **0.0112**
- Market Penetration Score: **0.1700**
- Geographic Concentration Score: **0.0000**
- Geographic Expansion Indicator: **0.0768**
- Geographic Intelligence Score: **0.0909**
- Geographic Balance Index: **0.9247**

The mean Geographic Intelligence Score is slightly higher at **0.1297**, which again suggests that a relatively small number of stronger tracks pull the average above the median.

The Geographic Intelligence classification shows:

- **13 tracks (0.01%)** have **Exceptional geographic intelligence**
- **1,225 tracks (1.11%)** have **Strong geographic intelligence**
- **4,557 tracks (4.14%)** have **Moderate geographic intelligence**
- **34,166 tracks (31.00%)** have **Developing geographic intelligence**
- **70,237 tracks (63.74%)** have **Limited geographic intelligence**

This confirms that the majority of tracks do not yet combine broad reach, strong penetration, diversification, and expansion at a high level.

The geographic performance profile distribution reinforces this pattern:

- **109 tracks (0.10%)** have a **Broad high-performance geographic profile**
- **1,613 tracks (1.46%)** show **Broad reach with developing penetration**
- **12,180 tracks (11.05%)** show **Concentrated reach with strong penetration**
- **2,413 tracks (2.19%)** show an **Emerging geographic growth profile**
- **923 tracks (0.84%)** show a **Diversified multi-market profile**
- **92,960 tracks (84.36%)** have a **Limited geographic footprint**

The largest notable non-limited group is **Concentrated reach with strong penetration**. This shows that many tracks perform strongly within a relatively narrow geographic footprint even when they do not have broad international reach.

The strongest-dimension analysis provides another important insight.

For **79,825 tracks (72.44%)**, the strongest geographic dimension is **Market Penetration**. This suggests that depth of performance within reached markets is the most common source of geographic strength.

For **18,883 tracks (17.14%)**, the strongest dimension is **Geographic Expansion**.

For **10,610 tracks (9.63%)**, the strongest dimension is **Geographic Diversification**.

Only **880 tracks (0.80%)** have **Geographic Reach** as their strongest dimension.

This shows that broad international reach is comparatively rare, while strong performance within existing markets is much more common.

The weakest-dimension analysis shows the opposite pattern:

- **84,856 tracks (77.00%)** have **Geographic Diversification** as their weakest dimension
- **17,402 tracks (15.79%)** have **Geographic Reach** as their weakest dimension
- **4,170 tracks (3.78%)** have **Market Penetration** as their weakest dimension
- **3,770 tracks (3.42%)** have **Geographic Expansion** as their weakest dimension

The fact that diversification is the weakest dimension for more than three-quarters of tracks is consistent with the earlier finding that most tracks are heavily concentrated in one market or a small number of markets.

The diagnostics further support the overall pattern:

- **1,238 tracks** have a Geographic Intelligence Score of at least **0.50**
- only **13 tracks** reach **0.75 or above**
- **1,834 tracks** have a Reach Score of at least **0.50**
- **12,401 tracks** have a Penetration Score of at least **0.50**
- **6,593 tracks** have a Diversification Score of at least **0.50**
- **3,269 tracks** have an Expansion Indicator of at least **0.50**
- only **58 tracks** score at least **0.50 across all four geographic dimensions**

This final group of **58 tracks** represents the most complete form of geographic strength in PMIP because these tracks combine:

- broad reach,
- strong market penetration,
- meaningful geographic diversification,
- and strong geographic expansion.

The track-level findings therefore show that geographic success is not one-dimensional.

A track may be strong because it performs deeply in a few markets, because it is expanding into new ones, because it is well diversified, or because it already has broad international reach.

PMIP keeps these dimensions separate before combining them into the final Geographic Intelligence Score, which makes the results more interpretable than a single raw popularity measure.

All Section 11.3 validation checks passed. The full **110,198-track population** was preserved, track identities remained unique, all six geographic score fields remained finite and within the **0-to-1 range**, every track retained a geographic intelligence class, performance profile, strongest dimension, and weakest dimension, and all classifications reconciled with the full track population.

Section 11.3 therefore establishes the main **track-level findings layer** of PMIP.

The workflow is ready for **Section 11.4 Geographic Comparisons**.

### 11.4 Geographic Comparisons

#### Purpose

Compare the validated country-, artist-, and track-level geographic findings to identify how geographic performance differs across levels of the PMIP analysis.

This subsection examines whether the same general patterns appear consistently across markets, artists, and tracks, with particular attention to international reach, market penetration, geographic concentration, and expansion behaviour.

The aim is to show where the geographic intelligence layers agree, where they differ, and what those differences reveal about how music performance develops across countries, artists, and individual tracks.

```python
import pandas as pd
import numpy as np

print("Comparing PMIP geographic findings across analytical levels")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_11_3_status = globals().get(
    "section_11_3_complete",
    False
)

print(
    f"Section 11.3 completion status: "
    f"{section_11_3_status}"
)

if not section_11_3_status:
    raise RuntimeError(
        "Section 11.4 requires Section 11.3 "
        "Track-Level Findings to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated findings sources
# ------------------------------------------------------------

required_sources = {
    "country": "country_findings_df",
    "artist": "artist_findings_df",
    "track": "track_findings_df",
}

comparison_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 11.4 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    comparison_sources[source_key] = candidate.copy()


country_comparison_df = comparison_sources["country"]
artist_comparison_df = comparison_sources["artist"]
track_comparison_df = comparison_sources["track"]


print()
print("Validated geographic comparison sources")
print("=" * 100)

print(
    f"Country markets: "
    f"{len(country_comparison_df):,}"
)

print(
    f"Artist representations: "
    f"{len(artist_comparison_df):,}"
)

print(
    f"Tracks: "
    f"{len(track_comparison_df):,}"
)


# ------------------------------------------------------------
# 3. Build country-level comparison metrics
# ------------------------------------------------------------

country_streaming_median = float(
    country_comparison_df[
        "streaming_strength_percentile"
    ].median()
)

country_chart_median = float(
    country_comparison_df[
        "chart_strength_percentile"
    ].median()
)

country_context_median = float(
    country_comparison_df[
        "market_context_percentile"
    ].median()
)

country_findings_median = float(
    country_comparison_df[
        "country_findings_index"
    ].median()
)

country_findings_mean = float(
    country_comparison_df[
        "country_findings_index"
    ].mean()
)


# ------------------------------------------------------------
# 4. Build artist-level comparison metrics
# ------------------------------------------------------------

artist_reach_median = float(
    artist_comparison_df[
        "international_reach_index"
    ].median()
)

artist_penetration_median = float(
    artist_comparison_df[
        "market_penetration_index"
    ].median()
)

artist_diversification_median = float(
    artist_comparison_df[
        "artist_diversification_index"
    ].median()
)

artist_profile_median = float(
    artist_comparison_df[
        "geographic_profile_index"
    ].median()
)

artist_findings_median = float(
    artist_comparison_df[
        "artist_findings_index"
    ].median()
)

artist_findings_mean = float(
    artist_comparison_df[
        "artist_findings_index"
    ].mean()
)


# ------------------------------------------------------------
# 5. Build track-level comparison metrics
# ------------------------------------------------------------

track_reach_median = float(
    track_comparison_df[
        "geographic_reach_score"
    ].median()
)

track_penetration_median = float(
    track_comparison_df[
        "market_penetration_score"
    ].median()
)

track_diversification_median = float(
    track_comparison_df[
        "geographic_concentration_score"
    ].median()
)

track_expansion_median = float(
    track_comparison_df[
        "geographic_expansion_indicator"
    ].median()
)

track_intelligence_median = float(
    track_comparison_df[
        "geographic_intelligence_score"
    ].median()
)

track_intelligence_mean = float(
    track_comparison_df[
        "geographic_intelligence_score"
    ].mean()
)


# ------------------------------------------------------------
# 6. Summary comparison table
# ------------------------------------------------------------

geographic_level_comparison_df = pd.DataFrame(
    [
        {
            "Analytical Level": "Country",
            "Population": len(country_comparison_df),
            "Reach / Breadth": np.nan,
            "Penetration / Depth": np.nan,
            "Diversification": np.nan,
            "Expansion / Movement": np.nan,
            "Overall Median Index": country_findings_median,
            "Overall Mean Index": country_findings_mean,
        },
        {
            "Analytical Level": "Artist",
            "Population": len(artist_comparison_df),
            "Reach / Breadth": artist_reach_median,
            "Penetration / Depth": artist_penetration_median,
            "Diversification": artist_diversification_median,
            "Expansion / Movement": np.nan,
            "Overall Median Index": artist_findings_median,
            "Overall Mean Index": artist_findings_mean,
        },
        {
            "Analytical Level": "Track",
            "Population": len(track_comparison_df),
            "Reach / Breadth": track_reach_median,
            "Penetration / Depth": track_penetration_median,
            "Diversification": track_diversification_median,
            "Expansion / Movement": track_expansion_median,
            "Overall Median Index": track_intelligence_median,
            "Overall Mean Index": track_intelligence_mean,
        },
    ]
)


print()
print("Geographic findings comparison by analytical level")
print("=" * 100)

display(
    geographic_level_comparison_df
)


# ------------------------------------------------------------
# 7. Strong versus limited population comparison
# ------------------------------------------------------------

country_high_priority_count = int(
    (
        country_comparison_df[
            "country_findings_class"
        ]
        ==
        "High-priority country market"
    ).sum()
)

country_strong_count = int(
    (
        country_comparison_df[
            "country_findings_class"
        ]
        ==
        "Strong country market"
    ).sum()
)

country_high_or_strong_count = (
    country_high_priority_count
    +
    country_strong_count
)


artist_strong_count = int(
    artist_comparison_df[
        "artist_findings_class"
    ]
    .isin(
        [
            "Exceptional artist geographic strength",
            "Strong artist geographic strength",
        ]
    )
    .sum()
)


artist_limited_count = int(
    (
        artist_comparison_df[
            "artist_findings_class"
        ]
        ==
        "Limited artist geographic strength"
    ).sum()
)


track_strong_count = int(
    track_comparison_df[
        "geographic_intelligence_class"
    ]
    .isin(
        [
            "Exceptional geographic intelligence",
            "Strong geographic intelligence",
        ]
    )
    .sum()
)


track_limited_count = int(
    (
        track_comparison_df[
            "geographic_intelligence_class"
        ]
        ==
        "Limited geographic intelligence"
    ).sum()
)


strength_comparison_df = pd.DataFrame(
    [
        {
            "Analytical Level": "Country",
            "Strong / High-Priority Count":
                country_high_or_strong_count,

            "Strong / High-Priority Share (%)":
                (
                    country_high_or_strong_count
                    /
                    len(country_comparison_df)
                    * 100
                ),

            "Limited / Lower-Priority Count":
                int(
                    (
                        country_comparison_df[
                            "country_findings_class"
                        ]
                        ==
                        "Lower-priority country market"
                    ).sum()
                ),

            "Limited / Lower-Priority Share (%)":
                (
                    (
                        country_comparison_df[
                            "country_findings_class"
                        ]
                        ==
                        "Lower-priority country market"
                    ).sum()
                    /
                    len(country_comparison_df)
                    * 100
                ),
        },
        {
            "Analytical Level": "Artist",
            "Strong / High-Priority Count":
                artist_strong_count,

            "Strong / High-Priority Share (%)":
                (
                    artist_strong_count
                    /
                    len(artist_comparison_df)
                    * 100
                ),

            "Limited / Lower-Priority Count":
                artist_limited_count,

            "Limited / Lower-Priority Share (%)":
                (
                    artist_limited_count
                    /
                    len(artist_comparison_df)
                    * 100
                ),
        },
        {
            "Analytical Level": "Track",
            "Strong / High-Priority Count":
                track_strong_count,

            "Strong / High-Priority Share (%)":
                (
                    track_strong_count
                    /
                    len(track_comparison_df)
                    * 100
                ),

            "Limited / Lower-Priority Count":
                track_limited_count,

            "Limited / Lower-Priority Share (%)":
                (
                    track_limited_count
                    /
                    len(track_comparison_df)
                    * 100
                ),
        },
    ]
)


print()
print("Strong versus limited geographic findings")
print("=" * 100)

display(
    strength_comparison_df
)


# ------------------------------------------------------------
# 8. Reach comparison
# ------------------------------------------------------------

artist_broad_reach_count = int(
    (
        artist_comparison_df[
            "international_reach_index"
        ]
        >= 0.50
    ).sum()
)

track_broad_reach_count = int(
    (
        track_comparison_df[
            "geographic_reach_score"
        ]
        >= 0.50
    ).sum()
)


reach_comparison_df = pd.DataFrame(
    [
        {
            "Analytical Level": "Artist",
            "Population": len(artist_comparison_df),
            "Median Reach Index": artist_reach_median,
            "Records with Reach >= 0.50":
                artist_broad_reach_count,

            "Share with Reach >= 0.50 (%)":
                (
                    artist_broad_reach_count
                    /
                    len(artist_comparison_df)
                    * 100
                ),
        },
        {
            "Analytical Level": "Track",
            "Population": len(track_comparison_df),
            "Median Reach Index": track_reach_median,
            "Records with Reach >= 0.50":
                track_broad_reach_count,

            "Share with Reach >= 0.50 (%)":
                (
                    track_broad_reach_count
                    /
                    len(track_comparison_df)
                    * 100
                ),
        },
    ]
)


print()
print("Artist versus track geographic reach")
print("=" * 100)

display(
    reach_comparison_df
)


# ------------------------------------------------------------
# 9. Penetration comparison
# ------------------------------------------------------------

artist_strong_penetration_count = int(
    (
        artist_comparison_df[
            "market_penetration_index"
        ]
        >= 0.50
    ).sum()
)

track_strong_penetration_count = int(
    (
        track_comparison_df[
            "market_penetration_score"
        ]
        >= 0.50
    ).sum()
)


penetration_comparison_df = pd.DataFrame(
    [
        {
            "Analytical Level": "Artist",
            "Population": len(artist_comparison_df),
            "Median Penetration Index":
                artist_penetration_median,

            "Records with Penetration >= 0.50":
                artist_strong_penetration_count,

            "Share with Penetration >= 0.50 (%)":
                (
                    artist_strong_penetration_count
                    /
                    len(artist_comparison_df)
                    * 100
                ),
        },
        {
            "Analytical Level": "Track",
            "Population": len(track_comparison_df),
            "Median Penetration Index":
                track_penetration_median,

            "Records with Penetration >= 0.50":
                track_strong_penetration_count,

            "Share with Penetration >= 0.50 (%)":
                (
                    track_strong_penetration_count
                    /
                    len(track_comparison_df)
                    * 100
                ),
        },
    ]
)


print()
print("Artist versus track market penetration")
print("=" * 100)

display(
    penetration_comparison_df
)


# ------------------------------------------------------------
# 10. Diversification comparison
# ------------------------------------------------------------

artist_diversified_count = int(
    (
        artist_comparison_df[
            "artist_diversification_index"
        ]
        >= 0.50
    ).sum()
)

track_diversified_count = int(
    (
        track_comparison_df[
            "geographic_concentration_score"
        ]
        >= 0.50
    ).sum()
)


diversification_comparison_df = pd.DataFrame(
    [
        {
            "Analytical Level": "Artist",
            "Population": len(artist_comparison_df),
            "Median Diversification Index":
                artist_diversification_median,

            "Records with Diversification >= 0.50":
                artist_diversified_count,

            "Share with Diversification >= 0.50 (%)":
                (
                    artist_diversified_count
                    /
                    len(artist_comparison_df)
                    * 100
                ),
        },
        {
            "Analytical Level": "Track",
            "Population": len(track_comparison_df),
            "Median Diversification Index":
                track_diversification_median,

            "Records with Diversification >= 0.50":
                track_diversified_count,

            "Share with Diversification >= 0.50 (%)":
                (
                    track_diversified_count
                    /
                    len(track_comparison_df)
                    * 100
                ),
        },
    ]
)


print()
print("Artist versus track geographic diversification")
print("=" * 100)

display(
    diversification_comparison_df
)


# ------------------------------------------------------------
# 11. Build comparison findings summary
# ------------------------------------------------------------

comparison_findings_summary_df = pd.DataFrame(
    [
        {
            "Finding":
                "Market penetration is stronger than "
                "international reach at artist level.",

            "Evidence":
                (
                    f"Median artist penetration "
                    f"{artist_penetration_median:.4f} "
                    f"vs reach "
                    f"{artist_reach_median:.4f}"
                ),
        },
        {
            "Finding":
                "Market penetration is also stronger than "
                "international reach at track level.",

            "Evidence":
                (
                    f"Median track penetration "
                    f"{track_penetration_median:.4f} "
                    f"vs reach "
                    f"{track_reach_median:.4f}"
                ),
        },
        {
            "Finding":
                "Geographic diversification remains weak "
                "for both artists and tracks.",

            "Evidence":
                (
                    f"Median artist diversification "
                    f"{artist_diversification_median:.4f}; "
                    f"median track diversification "
                    f"{track_diversification_median:.4f}"
                ),
        },
        {
            "Finding":
                "Strong overall geographic intelligence "
                "is rare at both artist and track level.",

            "Evidence":
                (
                    f"{artist_strong_count:,} strong/exceptional "
                    f"artist representations; "
                    f"{track_strong_count:,} strong/exceptional tracks"
                ),
        },
        {
            "Finding":
                "Country-level market strength is more widely "
                "distributed than artist- or track-level strength.",

            "Evidence":
                (
                    f"{country_high_or_strong_count:,} of "
                    f"{len(country_comparison_df):,} countries are "
                    f"high-priority or strong markets"
                ),
        },
    ]
)


print()
print("Cross-level geographic comparison findings")
print("=" * 100)

display(
    comparison_findings_summary_df
)


# ------------------------------------------------------------
# 12. Diagnostics
# ------------------------------------------------------------

geographic_comparison_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Country markets compared",

            "Observed Value":
                int(
                    len(
                        country_comparison_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Artist representations compared",

            "Observed Value":
                int(
                    len(
                        artist_comparison_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Tracks compared",

            "Observed Value":
                int(
                    len(
                        track_comparison_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Strong/high-priority country markets",

            "Observed Value":
                country_high_or_strong_count,
        },
        {
            "Diagnostic":
                "Strong/exceptional artist representations",

            "Observed Value":
                artist_strong_count,
        },
        {
            "Diagnostic":
                "Strong/exceptional tracks",

            "Observed Value":
                track_strong_count,
        },
    ]
)


print()
print("Geographic comparison diagnostics")
print("=" * 100)

display(
    geographic_comparison_diagnostics_df
)


# ------------------------------------------------------------
# 13. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(
        country_comparison_df
    )
    == 76
)


artist_population_valid = bool(
    len(
        artist_comparison_df
    )
    == 49_843
)


track_population_valid = bool(
    len(
        track_comparison_df
    )
    == 110_198
)


comparison_numeric_fields = [
    "Overall Median Index",
    "Overall Mean Index",
]


summary_numeric_valid = bool(
    np.isfinite(
        geographic_level_comparison_df[
            comparison_numeric_fields
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


reach_comparison_valid = bool(
    np.isfinite(
        reach_comparison_df[
            [
                "Median Reach Index",
                "Share with Reach >= 0.50 (%)",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


penetration_comparison_valid = bool(
    np.isfinite(
        penetration_comparison_df[
            [
                "Median Penetration Index",
                "Share with Penetration >= 0.50 (%)",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


diversification_comparison_valid = bool(
    np.isfinite(
        diversification_comparison_df[
            [
                "Median Diversification Index",
                "Share with Diversification >= 0.50 (%)",
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


findings_complete = bool(
    comparison_findings_summary_df[
        [
            "Finding",
            "Evidence",
        ]
    ]
    .notna()
    .all()
    .all()
)


section_11_4_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 11.3 completion",

            "Requirement":
                "Track-Level Findings must already be complete.",

            "Observed Evidence":
                f"Section 11.3 completion status: "
                f"{section_11_3_status}",

            "Passed":
                section_11_3_status,
        },
        {
            "Validation Area":
                "Country comparison population",

            "Requirement":
                "All 76 validated country markets "
                "must be available for comparison.",

            "Observed Evidence":
                f"{len(country_comparison_df):,} "
                "country markets compared",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Artist comparison population",

            "Requirement":
                "All 49,843 artist representations "
                "must be available for comparison.",

            "Observed Evidence":
                f"{len(artist_comparison_df):,} "
                "artist representations compared",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Track comparison population",

            "Requirement":
                "All 110,198 tracks must be "
                "available for comparison.",

            "Observed Evidence":
                f"{len(track_comparison_df):,} "
                "tracks compared",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "Summary comparison validity",

            "Requirement":
                "Overall comparison indices must remain finite.",

            "Observed Evidence":
                "Country-, artist-, and track-level "
                "summary indices assessed",

            "Passed":
                summary_numeric_valid,
        },
        {
            "Validation Area":
                "Reach comparison validity",

            "Requirement":
                "Artist and track reach comparisons "
                "must remain finite.",

            "Observed Evidence":
                "Artist and track reach comparisons assessed",

            "Passed":
                reach_comparison_valid,
        },
        {
            "Validation Area":
                "Penetration comparison validity",

            "Requirement":
                "Artist and track penetration comparisons "
                "must remain finite.",

            "Observed Evidence":
                "Artist and track penetration comparisons assessed",

            "Passed":
                penetration_comparison_valid,
        },
        {
            "Validation Area":
                "Diversification comparison validity",

            "Requirement":
                "Artist and track diversification comparisons "
                "must remain finite.",

            "Observed Evidence":
                "Artist and track diversification comparisons assessed",

            "Passed":
                diversification_comparison_valid,
        },
        {
            "Validation Area":
                "Comparison findings completeness",

            "Requirement":
                "Each cross-level comparison finding "
                "must contain supporting evidence.",

            "Observed Evidence":
                f"{len(comparison_findings_summary_df):,} "
                "cross-level findings prepared",

            "Passed":
                findings_complete,
        },
    ]
)


print()
print("Geographic comparisons validation")
print("=" * 100)

display(
    section_11_4_validation_df
)


# ------------------------------------------------------------
# 14. Completion status
# ------------------------------------------------------------

section_11_4_complete = bool(
    section_11_4_validation_df[
        "Passed"
    ].all()
)


if not section_11_4_complete:

    failed_checks = (
        section_11_4_validation_df
        .loc[
            ~section_11_4_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 11.4 Geographic Comparisons "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 11.4 geographic comparison "
    "validation checks passed."
)

print(
    f"Section 11.4 completion status: "
    f"{section_11_4_complete}"
)

print(
    "Country-, artist-, and track-level geographic "
    "findings have been compared successfully."
)

print(
    "The workflow is ready for "
    "Section 11.5 High-Priority Geographic Insights."
)
```

#### Interpretation

Section 11.4 successfully compares PMIP geographic findings across the **country, artist, and track levels**, giving a clearer picture of how geographic strength behaves at different levels of the analysis.

The comparison covers:

- **76 country markets**
- **49,843 artist representations**
- **110,198 tracks**

The first major finding is that **country-level strength is more widely distributed** than artist- or track-level strength.

At country level, **26 of 76 markets (34.21%)** are classified as either **High-priority** or **Strong**. By comparison, only **989 artist representations (1.98%)** are classified as **Strong or Exceptional**, while only **1,238 tracks (1.12%)** reach the equivalent strong or exceptional range.

This shows that strong market environments are much more common than strong artist- or track-level geographic performance.

A second major finding is that **market penetration is consistently stronger than geographic reach**.

At artist level:

- Median market penetration index: **0.2383**
- Median international reach index: **0.0092**

At track level:

- Median market penetration score: **0.1700**
- Median geographic reach score: **0.0112**

This suggests that artists and tracks are generally more successful at performing strongly within the markets they already reach than they are at expanding into a broad international footprint.

The same pattern is visible when using the **0.50 threshold**.

At artist level:

- **14,196 artist representations (28.48%)** achieve penetration of at least **0.50**
- only **660 artist representations (1.32%)** achieve reach of at least **0.50**

At track level:

- **12,401 tracks (11.25%)** achieve penetration of at least **0.50**
- only **1,834 tracks (1.66%)** achieve reach of at least **0.50**

This is one of the clearest geographic patterns in the notebook: **depth develops much more often than breadth**.

Geographic diversification is also weak across both artists and tracks.

The median diversification index is **0.0000** at both analytical levels.

However:

- **4,627 artist representations (9.28%)** achieve diversification of at least **0.50**
- **6,593 tracks (5.98%)** achieve diversification of at least **0.50**

This means most artists and tracks remain geographically concentrated even when they perform strongly within individual markets.

The overall comparison indices also reflect this difference in geographic maturity.

The median overall index is:

- **0.4908** at country level
- **0.1390** at artist level
- **0.0909** at track level

The country-level index is structurally different from the artist and track indices, so these numbers should not be treated as directly equivalent performance scores. However, they still reinforce the broader finding that relatively strong market environments are common, while broad and balanced geographic strength at the artist and track levels is much rarer.

The strong-versus-limited comparison highlights the same pattern.

At country level:

- **34.21%** are High-priority or Strong
- **31.58%** are Lower-priority

At artist level:

- **1.98%** are Strong or Exceptional
- **52.58%** are Limited

At track level:

- **1.12%** are Strong or Exceptional
- **63.74%** are Limited

This widening difference shows that geographic strength becomes increasingly selective when moving from the market level to individual artists and then to individual tracks.

The five cross-level findings generated by the analysis are therefore consistent with one another:

- market penetration is stronger than reach at artist level;
- market penetration is also stronger than reach at track level;
- geographic diversification remains weak at both artist and track level;
- strong overall geographic intelligence is rare for both artists and tracks;
- country-level market strength is more widely distributed than artist- or track-level geographic strength.

Together, these findings suggest a practical PMIP interpretation: **strong country markets provide opportunity, but only a small proportion of artists and tracks convert that opportunity into broad, diversified, and internationally strong geographic performance**.

This distinction is important for A&R and music intelligence use. A strong market should not automatically imply that every artist or track performing there has strong international potential. PMIP therefore keeps country opportunity, artist-level strength, and track-level strength as separate analytical layers.

All Section 11.4 validation checks passed. The full country, artist, and track populations were preserved, summary comparisons remained finite, artist-versus-track reach, penetration, and diversification comparisons were valid, and all five cross-level findings retained explicit supporting evidence.

Section 11.4 therefore establishes the **cross-level comparison layer** of PMIP.

The workflow is ready for **Section 11.5 High-Priority Geographic Insights**.

### 11.5 High-Priority Geographic Insights

#### Purpose

Identify the most important geographic intelligence signals produced by PMIP and convert the validated findings into a concise set of high-priority insights.

This subsection combines evidence from country-, artist-, and track-level analysis to highlight the patterns most relevant for practical music intelligence and A&R decision support.

The focus is on distinguishing broad market opportunity from actual artist and track geographic strength, identifying where strong penetration exists without broad reach, and isolating the relatively small number of artist and track records that demonstrate balanced international performance.

The insights remain descriptive and should be used as decision-support evidence rather than as automatic recommendations or proof of future commercial success.

```python
import pandas as pd
import numpy as np

print("Extracting PMIP high-priority geographic insights")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_11_4_status = globals().get(
    "section_11_4_complete",
    False
)

print(
    f"Section 11.4 completion status: "
    f"{section_11_4_status}"
)

if not section_11_4_status:
    raise RuntimeError(
        "Section 11.5 requires Section 11.4 "
        "Geographic Comparisons to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated Section 11 findings sources
# ------------------------------------------------------------

required_sources = {
    "country": "country_findings_df",
    "artist": "artist_findings_df",
    "track": "track_findings_df",
    "comparison": "comparison_findings_summary_df",
}

insight_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 11.5 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    insight_sources[source_key] = candidate.copy()


country_insight_df = insight_sources["country"]
artist_insight_df = insight_sources["artist"]
track_insight_df = insight_sources["track"]
comparison_insight_df = insight_sources["comparison"]


print()
print("Validated high-priority insight sources")
print("=" * 100)

print(
    f"Country markets: "
    f"{len(country_insight_df):,}"
)

print(
    f"Artist representations: "
    f"{len(artist_insight_df):,}"
)

print(
    f"Tracks: "
    f"{len(track_insight_df):,}"
)

print(
    f"Cross-level findings: "
    f"{len(comparison_insight_df):,}"
)


# ------------------------------------------------------------
# 3. High-priority country markets
# ------------------------------------------------------------

high_priority_country_df = (
    country_insight_df
    .loc[
        country_insight_df[
            "country_findings_class"
        ]
        .isin(
            [
                "High-priority country market",
                "Strong country market",
            ]
        )
    ]
    .sort_values(
        "country_findings_index",
        ascending=False,
    )
    .copy()
)


print()
print("High-priority and strong country markets")
print("=" * 100)

display(
    high_priority_country_df[
        [
            column
            for column in [
                "country",
                "country_findings_index",
                "country_findings_class",
                "streaming_strength_percentile",
                "chart_strength_percentile",
                "market_context_percentile",
            ]
            if column in high_priority_country_df.columns
        ]
    ]
    .head(20)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 4. High-priority artist representations
# ------------------------------------------------------------

high_priority_artist_df = (
    artist_insight_df
    .loc[
        artist_insight_df[
            "artist_findings_class"
        ]
        .isin(
            [
                "Exceptional artist geographic strength",
                "Strong artist geographic strength",
            ]
        )
    ]
    .sort_values(
        "artist_findings_index",
        ascending=False,
    )
    .copy()
)


artist_identity_col = None

for column in [
    "artists_raw",
    "artist",
    "artist_name",
]:
    if column in high_priority_artist_df.columns:
        artist_identity_col = column
        break


print()
print("High-priority artist geographic findings")
print("=" * 100)

artist_display_columns = [
    column
    for column in [
        artist_identity_col,
        "markets_reached",
        "international_reach_index",
        "market_penetration_index",
        "artist_diversification_index",
        "geographic_profile_index",
        "artist_findings_index",
        "artist_findings_class",
    ]
    if column is not None
    and column in high_priority_artist_df.columns
]

display(
    high_priority_artist_df[
        artist_display_columns
    ]
    .head(25)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. High-priority tracks
# ------------------------------------------------------------

high_priority_track_df = (
    track_insight_df
    .loc[
        track_insight_df[
            "geographic_intelligence_class"
        ]
        .isin(
            [
                "Exceptional geographic intelligence",
                "Strong geographic intelligence",
            ]
        )
    ]
    .sort_values(
        "geographic_intelligence_score",
        ascending=False,
    )
    .copy()
)


print()
print("High-priority track geographic findings")
print("=" * 100)

track_display_columns = [
    column
    for column in [
        "track_id",
        "track_name",
        "artists_raw",
        "geographic_reach_score",
        "market_penetration_score",
        "geographic_concentration_score",
        "geographic_expansion_indicator",
        "geographic_intelligence_score",
        "geographic_performance_profile",
    ]
    if column in high_priority_track_df.columns
]

display(
    high_priority_track_df[
        track_display_columns
    ]
    .head(25)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Broad-reach but under-penetrated tracks
# ------------------------------------------------------------

broad_reach_developing_penetration_df = (
    track_insight_df
    .loc[
        (
            track_insight_df[
                "geographic_reach_score"
            ]
            >= 0.50
        )
        &
        (
            track_insight_df[
                "market_penetration_score"
            ]
            < 0.50
        )
    ]
    .sort_values(
        [
            "geographic_reach_score",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .copy()
)


print()
print("Broad-reach tracks with developing penetration")
print("=" * 100)

display(
    broad_reach_developing_penetration_df[
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_reach_score",
                "market_penetration_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in broad_reach_developing_penetration_df.columns
        ]
    ]
    .head(20)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. Strong penetration but narrow reach tracks
# ------------------------------------------------------------

strong_penetration_narrow_reach_df = (
    track_insight_df
    .loc[
        (
            track_insight_df[
                "market_penetration_score"
            ]
            >= 0.50
        )
        &
        (
            track_insight_df[
                "geographic_reach_score"
            ]
            < 0.50
        )
    ]
    .sort_values(
        [
            "market_penetration_score",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .copy()
)


print()
print("Strong-penetration tracks with narrow geographic reach")
print("=" * 100)

display(
    strong_penetration_narrow_reach_df[
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "market_penetration_score",
                "geographic_reach_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in strong_penetration_narrow_reach_df.columns
        ]
    ]
    .head(20)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. Emerging expansion opportunities
# ------------------------------------------------------------

expansion_opportunity_df = (
    track_insight_df
    .loc[
        (
            track_insight_df[
                "geographic_expansion_indicator"
            ]
            >= 0.50
        )
        &
        (
            track_insight_df[
                "geographic_intelligence_score"
            ]
            < 0.50
        )
    ]
    .sort_values(
        [
            "geographic_expansion_indicator",
            "geographic_intelligence_score",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .copy()
)


print()
print("Emerging geographic expansion opportunities")
print("=" * 100)

display(
    expansion_opportunity_df[
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_expansion_indicator",
                "geographic_reach_score",
                "market_penetration_score",
                "geographic_intelligence_score",
                "geographic_performance_profile",
            ]
            if column in expansion_opportunity_df.columns
        ]
    ]
    .head(20)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 9. Balanced high-performing tracks
# ------------------------------------------------------------

balanced_high_performance_df = (
    track_insight_df
    .loc[
        (
            track_insight_df[
                [
                    "geographic_reach_score",
                    "market_penetration_score",
                    "geographic_concentration_score",
                    "geographic_expansion_indicator",
                ]
            ]
            >= 0.50
        )
        .all(
            axis=1
        )
    ]
    .sort_values(
        [
            "geographic_intelligence_score",
            "geographic_balance_index",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .copy()
)


print()
print("Balanced high-performing geographic tracks")
print("=" * 100)

display(
    balanced_high_performance_df[
        [
            column
            for column in [
                "track_id",
                "track_name",
                "artists_raw",
                "geographic_reach_score",
                "market_penetration_score",
                "geographic_concentration_score",
                "geographic_expansion_indicator",
                "geographic_intelligence_score",
                "geographic_balance_index",
            ]
            if column in balanced_high_performance_df.columns
        ]
    ]
    .head(20)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 10. Build high-priority insight statements
# ------------------------------------------------------------

high_priority_insights_df = pd.DataFrame(
    [
        {
            "Priority": 1,

            "Insight":
                "Strong country markets are more common than "
                "strong artist or track geographic performance.",

            "Evidence":
                (
                    f"{len(high_priority_country_df):,} of "
                    f"{len(country_insight_df):,} country markets "
                    f"are high-priority or strong, compared with "
                    f"{len(high_priority_artist_df):,} of "
                    f"{len(artist_insight_df):,} artists and "
                    f"{len(high_priority_track_df):,} of "
                    f"{len(track_insight_df):,} tracks."
                ),

            "PMIP Interpretation":
                "Market opportunity and entity-level geographic "
                "strength must be evaluated separately.",
        },
        {
            "Priority": 2,

            "Insight":
                "Market penetration develops more frequently "
                "than broad international reach.",

            "Evidence":
                (
                    f"{int((track_insight_df['market_penetration_score'] >= 0.50).sum()):,} "
                    f"tracks have penetration >= 0.50, while only "
                    f"{int((track_insight_df['geographic_reach_score'] >= 0.50).sum()):,} "
                    f"tracks have reach >= 0.50."
                ),

            "PMIP Interpretation":
                "Strong local or regional performance can exist "
                "well before broad international distribution.",
        },
        {
            "Priority": 3,

            "Insight":
                "Geographic diversification is the most common "
                "weakness across tracks.",

            "Evidence":
                (
                    f"{int((track_insight_df['weakest_geographic_dimension'] == 'Geographic diversification').sum()):,} "
                    f"tracks identify diversification as their "
                    f"weakest geographic dimension."
                ),

            "PMIP Interpretation":
                "Many tracks remain dependent on a narrow set "
                "of markets even when other signals are strong.",
        },
        {
            "Priority": 4,

            "Insight":
                "A small set of tracks shows strong geographic "
                "expansion before achieving high overall intelligence.",

            "Evidence":
                (
                    f"{len(expansion_opportunity_df):,} tracks have "
                    f"expansion >= 0.50 while their overall "
                    f"Geographic Intelligence Score remains below 0.50."
                ),

            "PMIP Interpretation":
                "These tracks may represent emerging geographic "
                "growth signals that deserve further monitoring.",
        },
        {
            "Priority": 5,

            "Insight":
                "Balanced high-level geographic strength is rare.",

            "Evidence":
                (
                    f"Only {len(balanced_high_performance_df):,} tracks "
                    f"score at least 0.50 across reach, penetration, "
                    f"diversification and expansion simultaneously."
                ),

            "PMIP Interpretation":
                "Balanced international performance is a highly "
                "selective signal within the full track population.",
        },
    ]
)


print()
print("PMIP high-priority geographic insight summary")
print("=" * 100)

display(
    high_priority_insights_df
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

high_priority_insights_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "High-priority or strong country markets",

            "Observed Value":
                int(
                    len(
                        high_priority_country_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Strong or exceptional artist representations",

            "Observed Value":
                int(
                    len(
                        high_priority_artist_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Strong or exceptional tracks",

            "Observed Value":
                int(
                    len(
                        high_priority_track_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Broad-reach tracks with developing penetration",

            "Observed Value":
                int(
                    len(
                        broad_reach_developing_penetration_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Strong-penetration tracks with narrow reach",

            "Observed Value":
                int(
                    len(
                        strong_penetration_narrow_reach_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Emerging expansion opportunities",

            "Observed Value":
                int(
                    len(
                        expansion_opportunity_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Balanced high-performing geographic tracks",

            "Observed Value":
                int(
                    len(
                        balanced_high_performance_df
                    )
                ),
        },
    ]
)


print()
print("High-priority geographic insight diagnostics")
print("=" * 100)

display(
    high_priority_insights_diagnostics_df
)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(
        country_insight_df
    )
    == 76
)

artist_population_valid = bool(
    len(
        artist_insight_df
    )
    == 49_843
)

track_population_valid = bool(
    len(
        track_insight_df
    )
    == 110_198
)

priority_insights_complete = bool(
    len(
        high_priority_insights_df
    )
    == 5
)

priority_evidence_complete = bool(
    high_priority_insights_df[
        [
            "Insight",
            "Evidence",
            "PMIP Interpretation",
        ]
    ]
    .notna()
    .all()
    .all()
)

priority_unique = bool(
    high_priority_insights_df[
        "Priority"
    ].nunique()
    == len(
        high_priority_insights_df
    )
)

subset_counts_valid = bool(
    len(high_priority_country_df)
    <= len(country_insight_df)
    and
    len(high_priority_artist_df)
    <= len(artist_insight_df)
    and
    len(high_priority_track_df)
    <= len(track_insight_df)
    and
    len(broad_reach_developing_penetration_df)
    <= len(track_insight_df)
    and
    len(strong_penetration_narrow_reach_df)
    <= len(track_insight_df)
    and
    len(expansion_opportunity_df)
    <= len(track_insight_df)
    and
    len(balanced_high_performance_df)
    <= len(track_insight_df)
)


section_11_5_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 11.4 completion",

            "Requirement":
                "Geographic Comparisons must already be complete.",

            "Observed Evidence":
                f"Section 11.4 completion status: "
                f"{section_11_4_status}",

            "Passed":
                section_11_4_status,
        },
        {
            "Validation Area":
                "Country findings population",

            "Requirement":
                "All 76 country markets must remain available.",

            "Observed Evidence":
                f"{len(country_insight_df):,} country markets assessed",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Artist findings population",

            "Requirement":
                "All 49,843 artist representations must remain available.",

            "Observed Evidence":
                f"{len(artist_insight_df):,} artist representations assessed",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Track findings population",

            "Requirement":
                "All 110,198 tracks must remain available.",

            "Observed Evidence":
                f"{len(track_insight_df):,} tracks assessed",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "Priority insight completeness",

            "Requirement":
                "Five high-priority geographic insights "
                "must be prepared.",

            "Observed Evidence":
                f"{len(high_priority_insights_df):,} "
                "priority insights prepared",

            "Passed":
                priority_insights_complete,
        },
        {
            "Validation Area":
                "Priority evidence completeness",

            "Requirement":
                "Every priority insight must contain evidence "
                "and a PMIP interpretation.",

            "Observed Evidence":
                "Evidence and interpretation retained "
                "for all priority insights",

            "Passed":
                priority_evidence_complete,
        },
        {
            "Validation Area":
                "Priority ordering uniqueness",

            "Requirement":
                "Each high-priority insight must have "
                "a unique priority rank.",

            "Observed Evidence":
                f"{high_priority_insights_df['Priority'].nunique():,} "
                "unique priority ranks",

            "Passed":
                priority_unique,
        },
        {
            "Validation Area":
                "Insight subset validity",

            "Requirement":
                "All high-priority subsets must remain "
                "within their validated source populations.",

            "Observed Evidence":
                "Country, artist and track insight subsets "
                "validated against source populations",

            "Passed":
                subset_counts_valid,
        },
    ]
)


print()
print("High-priority geographic insights validation")
print("=" * 100)

display(
    section_11_5_validation_df
)


# ------------------------------------------------------------
# 13. Completion status
# ------------------------------------------------------------

section_11_5_complete = bool(
    section_11_5_validation_df[
        "Passed"
    ].all()
)


section_11_complete = bool(
    globals().get(
        "section_11_1_complete",
        False
    )
    and
    globals().get(
        "section_11_2_complete",
        False
    )
    and
    globals().get(
        "section_11_3_complete",
        False
    )
    and
    globals().get(
        "section_11_4_complete",
        False
    )
    and
    section_11_5_complete
)


if not section_11_5_complete:

    failed_checks = (
        section_11_5_validation_df
        .loc[
            ~section_11_5_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 11.5 High-Priority Geographic Insights "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 11.5 High-Priority Geographic Insights "
    "validation checks passed."
)

print(
    f"Section 11.5 completion status: "
    f"{section_11_5_complete}"
)

print(
    f"Section 11 overall completion status: "
    f"{section_11_complete}"
)

print(
    "Five high-priority geographic insights have been "
    "prepared from the validated PMIP findings."
)

print(
    "Results and Interpretation is complete."
)
```

#### Interpretation

Section 11.5 successfully converts the validated PMIP geographic analysis into a concise set of **five high-priority geographic insights**.

The first major insight is that **strong country markets are much more common than strong artist- or track-level geographic performance**.

PMIP identifies:

- **26 of 76 country markets** as high-priority or strong
- **989 of 49,843 artist representations** as strong or exceptional
- **1,238 of 110,198 tracks** as strong or exceptional

This confirms that a favourable market environment does not automatically translate into strong geographic performance for individual artists or tracks.

The highest-priority country markets are led by:

- **United States**
- **Italy**
- **France**
- **Australia**
- **Germany**
- **Mexico**

These markets combine strong streaming performance, chart strength, and substantial observational support. Other strong markets include the United Kingdom, Spain, Poland, Norway, Sweden, Netherlands, Denmark, Malaysia, New Zealand, Ireland, Finland, Singapore, Belgium, and Austria.

The second major insight is that **market penetration develops more frequently than broad geographic reach**.

At track level:

- **12,401 tracks** have a Market Penetration Score of at least **0.50**
- only **1,834 tracks** have a Geographic Reach Score of at least **0.50**

This means tracks commonly develop strong performance inside reached markets before becoming broadly distributed across many markets.

PMIP therefore treats **depth** and **breadth** as separate forms of geographic intelligence.

The diagnostics reinforce this distinction.

There are:

- **1,613 broad-reach tracks with developing penetration**
- **12,180 strong-penetration tracks with narrow reach**

The much larger second group shows that strong local or regional traction is considerably more common than broad international strength.

This is a useful A&R signal because tracks with strong penetration but narrow reach may represent candidates for further geographic development, while broad-reach tracks with weaker penetration may require deeper engagement within the markets they have already entered.

The third major insight is that **geographic diversification remains the most common weakness**.

A total of **84,856 tracks** identify Geographic Diversification as their weakest geographic dimension.

This means that most tracks remain dependent on a relatively narrow set of markets even where other dimensions such as penetration or expansion are stronger.

Geographic strength should therefore not be interpreted only through market count or streaming volume. A track can appear successful while still being highly dependent on one or a few markets.

The fourth major insight identifies **2,643 emerging expansion opportunities**.

These tracks have a Geographic Expansion Indicator of at least **0.50** while their overall Geographic Intelligence Score remains below **0.50**.

This is an important distinction.

These tracks do not yet demonstrate mature overall geographic strength, but they show meaningful evidence of geographic movement. They may therefore represent developing international opportunities that would be useful to monitor over time.

The fifth and most selective finding is that **balanced high-level geographic strength is extremely rare**.

Only **58 tracks** score at least **0.50 simultaneously across all four major geographic dimensions**:

- Geographic Reach
- Market Penetration
- Geographic Diversification
- Geographic Expansion

These tracks represent the strongest balanced geographic cases in the dataset because their performance does not depend on only one geographic dimension.

This small population demonstrates why PMIP benefits from multi-dimensional analysis. A track may rank highly on penetration, expansion, or reach individually, but balanced international performance requires strength across several dimensions at once.

Together, the five high-priority insights establish a clear geographic interpretation of the PMIP dataset:

1. **Strong markets are more common than strong individual artists or tracks.**
2. **Market depth develops more frequently than international breadth.**
3. **Geographic diversification is the main structural weakness.**
4. **Some tracks show expansion momentum before achieving mature overall geographic strength.**
5. **Balanced international strength is highly selective.**

These findings make the geographic intelligence framework useful for practical decision support.

Rather than simply asking whether a track is popular, PMIP can distinguish whether that popularity comes from:

- strong performance in a small number of markets,
- broad but shallow international distribution,
- meaningful geographic diversification,
- active expansion into new markets,
- or balanced strength across all of these dimensions.

All Section 11.5 validation checks passed.

The analysis retained:

- all **76 country markets**
- all **49,843 artist representations**
- all **110,198 tracks**
- all **5 cross-level findings**

All five priority insights contain explicit supporting evidence and PMIP interpretations, all priority ranks are unique, and all extracted high-priority subsets remain within their validated source populations.

Section 11.5 is therefore complete.

With all subsections complete, **Section 11: Results and Interpretation is fully complete**.

PMIP now has a validated geographic results layer that translates the earlier feature engineering, movement analysis, scoring, and classification work into interpretable country-, artist-, and track-level intelligence.

## 12. Responsible Use and Limitations

### Purpose

Define the interpretation boundaries, data limitations, comparison safeguards, and human-review requirements that govern responsible use of the PMIP geographic intelligence outputs.

This section ensures that the geographic scores, classifications, profiles, and high-priority insights developed earlier in the notebook are not interpreted beyond what the underlying data can support.

The aim is to make clear what PMIP can reasonably indicate, what it cannot establish, and where human judgement remains necessary before geographic intelligence is used for A&R, market-development, or other music-industry decisions.

### 12.1 Interpretation Boundaries

#### Purpose

Define the boundaries for interpreting PMIP geographic intelligence so that descriptive patterns are not incorrectly treated as causal, predictive, or commercially certain conclusions.

This subsection distinguishes between what the validated geographic framework measures and what cannot be inferred directly from the available data.

The analysis establishes explicit safeguards around popularity, artist nationality, commercial success, causality, future performance, and automatic decision-making.

```python
import pandas as pd
import numpy as np

print("Defining PMIP geographic intelligence interpretation boundaries")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate Section 11 completion
# ------------------------------------------------------------

section_11_status = globals().get(
    "section_11_complete",
    False
)

print(
    f"Section 11 overall completion status: "
    f"{section_11_status}"
)

if not section_11_status:
    raise RuntimeError(
        "Section 12.1 requires Section 11 "
        "Results and Interpretation to be complete."
    )


# ------------------------------------------------------------
# 2. Recover final validated geographic intelligence
# ------------------------------------------------------------

required_sources = {
    "country": "country_findings_df",
    "artist": "artist_findings_df",
    "track": "track_findings_df",
    "priority_insights": "high_priority_insights_df",
}

responsible_use_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 12.1 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    responsible_use_sources[source_key] = candidate.copy()


country_boundary_df = responsible_use_sources["country"]
artist_boundary_df = responsible_use_sources["artist"]
track_boundary_df = responsible_use_sources["track"]
priority_boundary_df = responsible_use_sources["priority_insights"]


print()
print("Validated interpretation-boundary sources")
print("=" * 100)

print(
    f"Country markets represented: "
    f"{len(country_boundary_df):,}"
)

print(
    f"Artist representations represented: "
    f"{len(artist_boundary_df):,}"
)

print(
    f"Tracks represented: "
    f"{len(track_boundary_df):,}"
)

print(
    f"High-priority insights represented: "
    f"{len(priority_boundary_df):,}"
)


# ------------------------------------------------------------
# 3. Define PMIP interpretation boundaries
# ------------------------------------------------------------

interpretation_boundaries_df = pd.DataFrame(
    [
        {
            "Boundary Area":
                "Descriptive intelligence",

            "PMIP Can Support":
                "Describe observed geographic reach, penetration, "
                "concentration, diversification, expansion and "
                "market-performance patterns.",

            "PMIP Must Not Be Interpreted As":
                "Proof that geography caused the observed music "
                "performance.",

            "Required Safeguard":
                "Use causal language only when supported by a "
                "separate causal research design.",
        },
        {
            "Boundary Area":
                "Artist nationality",

            "PMIP Can Support":
                "Describe the country markets in which an artist "
                "representation was observed.",

            "PMIP Must Not Be Interpreted As":
                "Verified nationality, citizenship, birthplace "
                "or home country of an artist.",

            "Required Safeguard":
                "Refer to observed markets rather than inferred "
                "artist nationality.",
        },
        {
            "Boundary Area":
                "Track origin",

            "PMIP Can Support":
                "Identify markets where a track has observed "
                "streaming or chart evidence.",

            "PMIP Must Not Be Interpreted As":
                "The country where the track originated, was "
                "recorded, released, or first became popular.",

            "Required Safeguard":
                "Use 'observed market' terminology unless origin "
                "is verified from an external source.",
        },
        {
            "Boundary Area":
                "Commercial success",

            "PMIP Can Support":
                "Compare relative geographic streaming and chart "
                "strength within the analysed dataset.",

            "PMIP Must Not Be Interpreted As":
                "Complete commercial value, revenue, profitability, "
                "touring potential or label return on investment.",

            "Required Safeguard":
                "Combine PMIP outputs with commercial, audience "
                "and business evidence before investment decisions.",
        },
        {
            "Boundary Area":
                "Future performance",

            "PMIP Can Support":
                "Identify historical and observed expansion, "
                "contraction and momentum patterns.",

            "PMIP Must Not Be Interpreted As":
                "A guaranteed forecast of future geographic "
                "growth or future chart success.",

            "Required Safeguard":
                "Treat expansion and momentum indicators as "
                "historical decision-support signals.",
        },
        {
            "Boundary Area":
                "Popularity",

            "PMIP Can Support":
                "Measure geographic breadth and depth using the "
                "available streaming and chart evidence.",

            "PMIP Must Not Be Interpreted As":
                "A universal measure of an artist's or track's "
                "overall popularity.",

            "Required Safeguard":
                "Interpret geographic intelligence separately "
                "from social, cultural and platform-wide popularity.",
        },
        {
            "Boundary Area":
                "Geographic Intelligence Score",

            "PMIP Can Support":
                "Summarise multiple validated geographic dimensions "
                "into a comparable PMIP score.",

            "PMIP Must Not Be Interpreted As":
                "An objective universal ranking of music quality "
                "or artistic merit.",

            "Required Safeguard":
                "Use the score as a structured analytical aid and "
                "inspect its component dimensions.",
        },
        {
            "Boundary Area":
                "High-priority classifications",

            "PMIP Can Support":
                "Highlight markets, artists and tracks that meet "
                "defined PMIP analytical thresholds.",

            "PMIP Must Not Be Interpreted As":
                "Automatic recommendations to sign, promote, "
                "invest in, or reject an artist or track.",

            "Required Safeguard":
                "Require human review and supporting evidence "
                "before real-world action.",
        },
        {
            "Boundary Area":
                "Geographic balance",

            "PMIP Can Support":
                "Measure similarity between component geographic "
                "scores.",

            "PMIP Must Not Be Interpreted As":
                "Evidence of high geographic performance by itself.",

            "Required Safeguard":
                "Interpret balance together with the overall "
                "Geographic Intelligence Score and components.",
        },
        {
            "Boundary Area":
                "Cross-country comparison",

            "PMIP Can Support":
                "Compare markets within the validated PMIP "
                "observation framework.",

            "PMIP Must Not Be Interpreted As":
                "Perfectly equivalent country markets with identical "
                "platform coverage, population or industry context.",

            "Required Safeguard":
                "Retain observation context and coverage limitations "
                "when comparing countries.",
        },
    ]
)


print()
print("PMIP interpretation boundary framework")
print("=" * 100)

display(
    interpretation_boundaries_df
)


# ------------------------------------------------------------
# 4. Define permitted and restricted interpretation language
# ------------------------------------------------------------

interpretation_language_df = pd.DataFrame(
    [
        {
            "Use Type": "Preferred",
            "Example":
                "The track shows strong observed market penetration."
        },
        {
            "Use Type": "Preferred",
            "Example":
                "The artist has evidence of broad international reach."
        },
        {
            "Use Type": "Preferred",
            "Example":
                "The track demonstrates validated geographic expansion."
        },
        {
            "Use Type": "Preferred",
            "Example":
                "The market ranks highly within the PMIP dataset."
        },
        {
            "Use Type": "Avoid",
            "Example":
                "This country caused the track to become successful."
        },
        {
            "Use Type": "Avoid",
            "Example":
                "This is the artist's home country."
        },
        {
            "Use Type": "Avoid",
            "Example":
                "This track will definitely become internationally successful."
        },
        {
            "Use Type": "Avoid",
            "Example":
                "PMIP proves this artist should be signed."
        },
    ]
)


print()
print("Responsible interpretation language examples")
print("=" * 100)

display(
    interpretation_language_df
)


# ------------------------------------------------------------
# 5. Summarise interpretation policy
# ------------------------------------------------------------

interpretation_policy_summary_df = pd.DataFrame(
    [
        {
            "Policy Principle":
                "Observed evidence",

            "Application":
                "Describe what is present in the validated data "
                "without extending claims beyond the evidence.",
        },
        {
            "Policy Principle":
                "No causal inference",

            "Application":
                "Do not convert associations or geographic patterns "
                "into causal explanations.",
        },
        {
            "Policy Principle":
                "No automatic commercial decisions",

            "Application":
                "Scores and classifications must support rather "
                "than replace professional judgement.",
        },
        {
            "Policy Principle":
                "Component-level interpretation",

            "Application":
                "Inspect reach, penetration, diversification and "
                "expansion alongside summary scores.",
        },
        {
            "Policy Principle":
                "Context preservation",

            "Application":
                "Retain data coverage, temporal and country-context "
                "limitations when communicating results.",
        },
        {
            "Policy Principle":
                "Human review",

            "Application":
                "Require domain expertise before PMIP intelligence "
                "is converted into real-world decisions.",
        },
    ]
)


print()
print("Interpretation policy summary")
print("=" * 100)

display(
    interpretation_policy_summary_df
)


# ------------------------------------------------------------
# 6. Diagnostics
# ------------------------------------------------------------

interpretation_boundary_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Interpretation boundaries defined",

            "Observed Value":
                int(
                    len(
                        interpretation_boundaries_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Responsible-use policy principles defined",

            "Observed Value":
                int(
                    len(
                        interpretation_policy_summary_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Preferred language examples defined",

            "Observed Value":
                int(
                    (
                        interpretation_language_df[
                            "Use Type"
                        ]
                        ==
                        "Preferred"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Restricted language examples defined",

            "Observed Value":
                int(
                    (
                        interpretation_language_df[
                            "Use Type"
                        ]
                        ==
                        "Avoid"
                    ).sum()
                ),
        },
        {
            "Diagnostic":
                "Validated PMIP analytical populations protected",

            "Observed Value":
                (
                    f"{len(country_boundary_df):,} countries / "
                    f"{len(artist_boundary_df):,} artists / "
                    f"{len(track_boundary_df):,} tracks"
                ),
        },
    ]
)


print()
print("Interpretation-boundary diagnostics")
print("=" * 100)

display(
    interpretation_boundary_diagnostics_df
)


# ------------------------------------------------------------
# 7. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(
        country_boundary_df
    )
    == 76
)

artist_population_valid = bool(
    len(
        artist_boundary_df
    )
    == 49_843
)

track_population_valid = bool(
    len(
        track_boundary_df
    )
    == 110_198
)

boundary_framework_complete = bool(
    interpretation_boundaries_df[
        [
            "Boundary Area",
            "PMIP Can Support",
            "PMIP Must Not Be Interpreted As",
            "Required Safeguard",
        ]
    ]
    .notna()
    .all()
    .all()
)

policy_complete = bool(
    interpretation_policy_summary_df[
        [
            "Policy Principle",
            "Application",
        ]
    ]
    .notna()
    .all()
    .all()
)

language_guidance_complete = bool(
    set(
        interpretation_language_df[
            "Use Type"
        ].unique()
    )
    ==
    {
        "Preferred",
        "Avoid",
    }
)

critical_boundaries_present = bool(
    {
        "Descriptive intelligence",
        "Artist nationality",
        "Commercial success",
        "Future performance",
        "Geographic Intelligence Score",
        "High-priority classifications",
        "Cross-country comparison",
    }
    .issubset(
        set(
            interpretation_boundaries_df[
                "Boundary Area"
            ]
        )
    )
)


section_12_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 11 completion",

            "Requirement":
                "Results and Interpretation must already "
                "be complete.",

            "Observed Evidence":
                f"Section 11 overall completion status: "
                f"{section_11_status}",

            "Passed":
                section_11_status,
        },
        {
            "Validation Area":
                "Country population preservation",

            "Requirement":
                "All 76 validated country markets must "
                "remain represented.",

            "Observed Evidence":
                f"{len(country_boundary_df):,} "
                "country markets represented",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Artist population preservation",

            "Requirement":
                "All 49,843 artist representations must "
                "remain represented.",

            "Observed Evidence":
                f"{len(artist_boundary_df):,} "
                "artist representations represented",

            "Passed":
                artist_population_valid,
        },
        {
            "Validation Area":
                "Track population preservation",

            "Requirement":
                "All 110,198 tracks must remain represented.",

            "Observed Evidence":
                f"{len(track_boundary_df):,} "
                "tracks represented",

            "Passed":
                track_population_valid,
        },
        {
            "Validation Area":
                "Interpretation boundary completeness",

            "Requirement":
                "Every interpretation boundary must define "
                "supported use, prohibited interpretation "
                "and required safeguard.",

            "Observed Evidence":
                f"{len(interpretation_boundaries_df):,} "
                "interpretation boundaries defined",

            "Passed":
                boundary_framework_complete,
        },
        {
            "Validation Area":
                "Critical boundary coverage",

            "Requirement":
                "Core causal, geographic, commercial and "
                "decision-making limitations must be explicit.",

            "Observed Evidence":
                "Critical PMIP interpretation boundaries "
                "are explicitly represented",

            "Passed":
                critical_boundaries_present,
        },
        {
            "Validation Area":
                "Responsible-use policy completeness",

            "Requirement":
                "Responsible interpretation principles "
                "must include explicit applications.",

            "Observed Evidence":
                f"{len(interpretation_policy_summary_df):,} "
                "policy principles defined",

            "Passed":
                policy_complete,
        },
        {
            "Validation Area":
                "Interpretation language guidance",

            "Requirement":
                "Both preferred and restricted interpretation "
                "language must be documented.",

            "Observed Evidence":
                "Preferred and avoid-language examples defined",

            "Passed":
                language_guidance_complete,
        },
    ]
)


print()
print("Interpretation boundary validation")
print("=" * 100)

display(
    section_12_1_validation_df
)


# ------------------------------------------------------------
# 8. Completion status
# ------------------------------------------------------------

section_12_1_complete = bool(
    section_12_1_validation_df[
        "Passed"
    ].all()
)


if not section_12_1_complete:

    failed_checks = (
        section_12_1_validation_df
        .loc[
            ~section_12_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 12.1 Interpretation Boundaries "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 12.1 Interpretation Boundaries "
    "validation checks passed."
)

print(
    f"Section 12.1 completion status: "
    f"{section_12_1_complete}"
)

print(
    "PMIP geographic intelligence interpretation "
    "boundaries and responsible-use safeguards "
    "have been defined."
)

print(
    "The workflow is ready for "
    "Section 12.2 Data-Coverage Limitations."
)
```

#### Interpretation

Section 12.1 successfully defines the **responsible interpretation boundaries** for PMIP geographic intelligence.

The framework protects all validated analytical populations:

- **76 country markets**
- **49,843 artist representations**
- **110,198 tracks**
- **5 high-priority geographic insights**

A total of **10 interpretation boundaries** were established to make clear what PMIP can support and what must not be inferred from its outputs.

The first major boundary is that PMIP provides **descriptive geographic intelligence**, not causal proof.

The system can describe observed geographic reach, market penetration, concentration, diversification, expansion, and performance patterns, but it cannot prove that geography itself caused an artist or track to become successful.

The second boundary concerns **artist nationality and track origin**.

PMIP identifies where artists and tracks were observed in the dataset, but those observations must not be interpreted as verified nationality, citizenship, birthplace, home country, recording location, or country of origin.

This distinction is especially important because observed performance geography and actual artist origin are different concepts.

The third boundary concerns **commercial success**.

PMIP can compare relative streaming, chart, and geographic strength, but it does not directly measure:

- revenue,
- profitability,
- touring potential,
- label return on investment,
- total audience value,
- or complete commercial performance.

Therefore, geographic intelligence should be combined with additional commercial and audience evidence before business decisions are made.

The fourth boundary concerns **future performance**.

Expansion, contraction, and momentum indicators describe historical or observed geographic movement. They must not be treated as guaranteed forecasts of future international growth or chart success.

This means a strong expansion signal should be interpreted as evidence worth monitoring rather than a promise that future growth will continue.

The framework also establishes that PMIP geographic scores are **not universal popularity or artistic-quality measures**.

The Geographic Intelligence Score summarizes validated geographic dimensions, but it does not determine whether music is objectively good, culturally important, or universally popular.

Similarly, high-priority classifications are not automatic instructions to sign, promote, invest in, or reject an artist or track.

Human judgement remains necessary.

The Geographic Balance Index also receives an explicit interpretation safeguard.

A high balance score only shows that component geographic scores are relatively similar. It does not automatically mean the track has strong geographic performance. A track could be consistently weak across several dimensions and still appear balanced.

Balance must therefore be interpreted alongside the overall Geographic Intelligence Score and its underlying components.

Cross-country comparisons also require caution.

PMIP can compare the **76 validated markets within the dataset**, but those markets should not be treated as perfectly equivalent environments. Countries differ in population size, streaming-platform usage, chart structure, observation coverage, industry development, and other contextual factors.

Country rankings therefore represent comparative PMIP evidence rather than universal measures of national music-market value.

Section 12.1 also defines responsible interpretation language.

Preferred wording focuses on what the evidence shows, such as:

- a track showing strong **observed market penetration**;
- an artist having **evidence of broad international reach**;
- a track demonstrating **validated geographic expansion**;
- or a market ranking highly **within the PMIP dataset**.

Language that implies causality, certainty, verified nationality, or automatic commercial decisions should be avoided.

Six responsible-use principles were formally established:

- **Observed evidence**
- **No causal inference**
- **No automatic commercial decisions**
- **Component-level interpretation**
- **Context preservation**
- **Human review**

Together, these principles establish an important rule for the whole PMIP system: **the platform supports human decision-making rather than replacing it**.

All Section 12.1 validation checks passed. All analytical populations were preserved, all **10 interpretation boundaries** contain supported uses, prohibited interpretations, and safeguards, all critical boundary areas are represented, all **6 policy principles** were defined, and both preferred and restricted interpretation language were documented.

Section 12.1 therefore establishes the **responsible interpretation framework** for PMIP geographic intelligence.

The workflow is ready for **Section 12.2 Data-Coverage Limitations**.

### 12.2 Data-Coverage Limitations

#### Purpose

Assess the main limitations created by the geographic and temporal coverage of the PMIP dataset.

This subsection examines whether differences in observation volume, market representation, temporal availability, and entity-level coverage could influence the geographic intelligence produced earlier in the notebook.

The aim is to make clear that PMIP results are based on the evidence available within the analysed dataset and may not represent complete global music activity across every country, artist, track, platform, or time period.

```python
import pandas as pd
import numpy as np

print("Assessing PMIP data-coverage limitations")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_12_1_status = globals().get(
    "section_12_1_complete",
    False
)

print(
    f"Section 12.1 completion status: "
    f"{section_12_1_status}"
)

if not section_12_1_status:
    raise RuntimeError(
        "Section 12.2 requires Section 12.1 "
        "Interpretation Boundaries to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated coverage-related sources
# ------------------------------------------------------------

required_sources = {
    "country": "country_performance_df",
    "geographic": "geographic_analysis_df",
    "track_history": "track_country_entry_df",
    "country_temporal": "country_temporal_alignment_df",
}

coverage_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 12.2 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    coverage_sources[source_key] = candidate.copy()


country_coverage_df = coverage_sources["country"]
geographic_coverage_df = coverage_sources["geographic"]
track_history_coverage_df = coverage_sources["track_history"]
country_temporal_coverage_df = coverage_sources["country_temporal"]


print()
print("Validated data-coverage sources")
print("=" * 100)

print(
    f"Country markets represented: "
    f"{len(country_coverage_df):,}"
)

print(
    f"Geographic observations represented: "
    f"{len(geographic_coverage_df):,}"
)

print(
    f"Track-country histories represented: "
    f"{len(track_history_coverage_df):,}"
)

print(
    f"Country temporal records represented: "
    f"{len(country_temporal_coverage_df):,}"
)


# ------------------------------------------------------------
# 3. Recover key country coverage fields
# ------------------------------------------------------------

country_col = None

for column in [
    "country",
    "market",
    "country_code",
]:
    if column in country_coverage_df.columns:
        country_col = column
        break


if country_col is None:
    raise RuntimeError(
        "Section 12.2 could not identify the country column."
    )


observation_col = None

for column in [
    "observations",
    "observation_count",
    "total_observations",
]:
    if column in country_coverage_df.columns:
        observation_col = column
        break


if observation_col is None:
    raise RuntimeError(
        "Section 12.2 could not identify the country "
        "observation-count column."
    )


country_coverage_df[
    observation_col
] = pd.to_numeric(
    country_coverage_df[
        observation_col
    ],
    errors="coerce",
)


if country_coverage_df[
    observation_col
].isna().any():
    raise RuntimeError(
        "Section 12.2 found invalid country observation counts."
    )


# ------------------------------------------------------------
# 4. Country observation coverage overview
# ------------------------------------------------------------

country_obs_min = float(
    country_coverage_df[
        observation_col
    ].min()
)

country_obs_q1 = float(
    country_coverage_df[
        observation_col
    ].quantile(0.25)
)

country_obs_median = float(
    country_coverage_df[
        observation_col
    ].median()
)

country_obs_q3 = float(
    country_coverage_df[
        observation_col
    ].quantile(0.75)
)

country_obs_max = float(
    country_coverage_df[
        observation_col
    ].max()
)

country_obs_ratio = (
    country_obs_max
    /
    country_obs_min
    if country_obs_min > 0
    else np.nan
)


print()
print("Country observation coverage overview")
print("=" * 100)

print(
    f"Country markets assessed: "
    f"{len(country_coverage_df):,}"
)

print(
    f"Minimum observations in one market: "
    f"{country_obs_min:,.0f}"
)

print(
    f"25th percentile observations: "
    f"{country_obs_q1:,.1f}"
)

print(
    f"Median observations per market: "
    f"{country_obs_median:,.1f}"
)

print(
    f"75th percentile observations: "
    f"{country_obs_q3:,.1f}"
)

print(
    f"Maximum observations in one market: "
    f"{country_obs_max:,.0f}"
)

print(
    f"Maximum-to-minimum observation ratio: "
    f"{country_obs_ratio:,.2f}"
)


# ------------------------------------------------------------
# 5. Country coverage bands
# ------------------------------------------------------------

def classify_country_coverage(value):

    if value >= country_obs_q3:
        return "High observation coverage"

    if value >= country_obs_median:
        return "Above-median observation coverage"

    if value >= country_obs_q1:
        return "Below-median observation coverage"

    return "Limited observation coverage"


country_coverage_df[
    "data_coverage_class"
] = (
    country_coverage_df[
        observation_col
    ]
    .apply(
        classify_country_coverage
    )
)


coverage_class_order = [
    "High observation coverage",
    "Above-median observation coverage",
    "Below-median observation coverage",
    "Limited observation coverage",
]


country_coverage_summary_df = (
    country_coverage_df[
        "data_coverage_class"
    ]
    .value_counts()
    .reindex(
        coverage_class_order,
        fill_value=0,
    )
    .rename_axis(
        "Data Coverage Class"
    )
    .reset_index(
        name="Country Markets"
    )
)


country_coverage_summary_df[
    "Share (%)"
] = (
    country_coverage_summary_df[
        "Country Markets"
    ]
    /
    len(country_coverage_df)
    * 100
)


print()
print("Country observation-coverage classification")
print("=" * 100)

display(
    country_coverage_summary_df
)


# ------------------------------------------------------------
# 6. Most and least observed country markets
# ------------------------------------------------------------

most_observed_countries_df = (
    country_coverage_df[
        [
            country_col,
            observation_col,
            "data_coverage_class",
        ]
    ]
    .sort_values(
        observation_col,
        ascending=False,
    )
    .head(15)
    .reset_index(drop=True)
)


least_observed_countries_df = (
    country_coverage_df[
        [
            country_col,
            observation_col,
            "data_coverage_class",
        ]
    ]
    .sort_values(
        observation_col,
        ascending=True,
    )
    .head(15)
    .reset_index(drop=True)
)


print()
print("Countries with highest observation coverage")
print("=" * 100)

display(
    most_observed_countries_df
)


print()
print("Countries with lowest observation coverage")
print("=" * 100)

display(
    least_observed_countries_df
)


# ------------------------------------------------------------
# 7. Track-country history coverage
# ------------------------------------------------------------

track_id_col = None

for column in [
    "track_id",
    "id",
]:
    if column in track_history_coverage_df.columns:
        track_id_col = column
        break


history_country_col = None

for column in [
    "country",
    "market",
    "country_code",
]:
    if column in track_history_coverage_df.columns:
        history_country_col = column
        break


if track_id_col is None or history_country_col is None:
    raise RuntimeError(
        "Section 12.2 could not identify track-country "
        "history keys."
    )


track_market_count_df = (
    track_history_coverage_df
    .groupby(
        track_id_col
    )[
        history_country_col
    ]
    .nunique()
    .rename(
        "markets_observed"
    )
    .reset_index()
)


print()
print("Track geographic coverage overview")
print("=" * 100)

print(
    f"Tracks represented in track-country history: "
    f"{track_market_count_df[track_id_col].nunique():,}"
)

print(
    f"Median observed markets per track: "
    f"{track_market_count_df['markets_observed'].median():.1f}"
)

print(
    f"Mean observed markets per track: "
    f"{track_market_count_df['markets_observed'].mean():.2f}"
)

print(
    f"Maximum observed markets for one track: "
    f"{track_market_count_df['markets_observed'].max():.0f}"
)


single_market_tracks = int(
    (
        track_market_count_df[
            "markets_observed"
        ]
        == 1
    ).sum()
)

multi_market_tracks = int(
    (
        track_market_count_df[
            "markets_observed"
        ]
        > 1
    ).sum()
)


# ------------------------------------------------------------
# 8. Temporal coverage inspection
# ------------------------------------------------------------

date_columns = [
    column
    for column in geographic_coverage_df.columns
    if any(
        token in column.lower()
        for token in [
            "date",
            "time",
            "timestamp",
        ]
    )
]


parsed_date_info = []

for column in date_columns:

    parsed = pd.to_datetime(
        geographic_coverage_df[
            column
        ],
        errors="coerce",
    )

    valid_count = int(
        parsed.notna().sum()
    )

    if valid_count > 0:

        parsed_date_info.append(
            {
                "Date Field": column,
                "Valid Dates": valid_count,
                "Earliest Date": parsed.min(),
                "Latest Date": parsed.max(),
            }
        )


temporal_coverage_summary_df = pd.DataFrame(
    parsed_date_info
)


print()
print("Temporal coverage fields detected")
print("=" * 100)

if len(
    temporal_coverage_summary_df
) > 0:

    display(
        temporal_coverage_summary_df
    )

else:

    print(
        "No directly parseable date field was detected "
        "in geographic_analysis_df."
    )


# ------------------------------------------------------------
# 9. Define explicit PMIP data-coverage limitations
# ------------------------------------------------------------

data_coverage_limitations_df = pd.DataFrame(
    [
        {
            "Limitation Area":
                "Unequal country observation coverage",

            "Observed Evidence":
                (
                    f"Country observations range from "
                    f"{country_obs_min:,.0f} to "
                    f"{country_obs_max:,.0f}."
                ),

            "Potential Effect":
                "Markets with greater observation volume may "
                "produce more stable geographic estimates than "
                "markets with limited coverage.",

            "Required Interpretation":
                "Retain market observation context when comparing "
                "country-level performance.",
        },
        {
            "Limitation Area":
                "Incomplete global market representation",

            "Observed Evidence":
                (
                    f"PMIP currently represents "
                    f"{len(country_coverage_df):,} country markets."
                ),

            "Potential Effect":
                "The dataset does not necessarily capture all "
                "music markets worldwide.",

            "Required Interpretation":
                "Treat geographic reach as reach within the "
                "validated PMIP market universe.",
        },
        {
            "Limitation Area":
                "Sparse track geographic histories",

            "Observed Evidence":
                (
                    f"{single_market_tracks:,} tracks appear in "
                    f"only one observed market."
                ),

            "Potential Effect":
                "Single-market tracks cannot demonstrate broad "
                "cross-market behaviour within the available data.",

            "Required Interpretation":
                "Do not equate limited observed reach with proof "
                "of true real-world geographic limitation.",
        },
        {
            "Limitation Area":
                "Platform and source dependency",

            "Observed Evidence":
                "PMIP geographic features are derived from the "
                "streaming and chart sources available in the "
                "integrated dataset.",

            "Potential Effect":
                "Activity occurring on unrepresented platforms "
                "or sources may be absent.",

            "Required Interpretation":
                "Interpret PMIP as source-bounded intelligence "
                "rather than complete global audience measurement.",
        },
        {
            "Limitation Area":
                "Temporal coverage boundary",

            "Observed Evidence":
                "Geographic expansion, contraction and momentum "
                "are calculated within the available observation window.",

            "Potential Effect":
                "Market activity before the dataset starts or after "
                "it ends cannot be observed directly.",

            "Required Interpretation":
                "Treat movement indicators as window-specific "
                "historical evidence.",
        },
        {
            "Limitation Area":
                "Unequal entity histories",

            "Observed Evidence":
                (
                    f"Median markets observed per track: "
                    f"{track_market_count_df['markets_observed'].median():.1f}."
                ),

            "Potential Effect":
                "Tracks with longer or broader histories provide "
                "more evidence for geographic comparison than "
                "tracks with sparse histories.",

            "Required Interpretation":
                "Interpret confidence together with available "
                "entity-level observation history.",
        },
        {
            "Limitation Area":
                "Missing external market context",

            "Observed Evidence":
                "PMIP does not currently incorporate every external "
                "country-level factor such as population, streaming "
                "penetration, revenue or economic size.",

            "Potential Effect":
                "Raw geographic performance may reflect structural "
                "market differences not fully represented in PMIP.",

            "Required Interpretation":
                "Use PMIP geographic scores alongside external "
                "market research for commercial decisions.",
        },
        {
            "Limitation Area":
                "Historical observation bias",

            "Observed Evidence":
                "Only activity represented in the integrated "
                "historical dataset contributes to the analysis.",

            "Potential Effect":
                "Artists and tracks with stronger historical data "
                "availability may appear better supported.",

            "Required Interpretation":
                "Distinguish stronger evidence support from "
                "necessarily stronger underlying real-world performance.",
        },
    ]
)


print()
print("PMIP data-coverage limitation framework")
print("=" * 100)

display(
    data_coverage_limitations_df
)


# ------------------------------------------------------------
# 10. Coverage safeguards
# ------------------------------------------------------------

coverage_safeguards_df = pd.DataFrame(
    [
        {
            "Coverage Safeguard":
                "Observation context retention",

            "Application":
                "Keep country observation counts available "
                "alongside performance findings.",
        },
        {
            "Coverage Safeguard":
                "Validated market universe",

            "Application":
                "Describe international reach relative to the "
                "76 observed PMIP markets.",
        },
        {
            "Coverage Safeguard":
                "Temporal confidence",

            "Application":
                "Use the temporal validation framework from "
                "Section 9.5 when interpreting geographic movement.",
        },
        {
            "Coverage Safeguard":
                "Sparse-history caution",

            "Application":
                "Avoid strong cross-market conclusions for tracks "
                "with minimal observed market history.",
        },
        {
            "Coverage Safeguard":
                "Source-bounded interpretation",

            "Application":
                "Avoid presenting the integrated PMIP dataset "
                "as complete global platform coverage.",
        },
        {
            "Coverage Safeguard":
                "External-context review",

            "Application":
                "Supplement PMIP with external market and "
                "commercial information when decisions require it.",
        },
    ]
)


print()
print("Data-coverage safeguards")
print("=" * 100)

display(
    coverage_safeguards_df
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

coverage_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Country markets assessed",

            "Observed Value":
                int(
                    len(
                        country_coverage_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Minimum country observations",

            "Observed Value":
                int(
                    country_obs_min
                ),
        },
        {
            "Diagnostic":
                "Maximum country observations",

            "Observed Value":
                int(
                    country_obs_max
                ),
        },
        {
            "Diagnostic":
                "Single-market tracks",

            "Observed Value":
                single_market_tracks,
        },
        {
            "Diagnostic":
                "Multi-market tracks",

            "Observed Value":
                multi_market_tracks,
        },
        {
            "Diagnostic":
                "Data-coverage limitations defined",

            "Observed Value":
                int(
                    len(
                        data_coverage_limitations_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Coverage safeguards defined",

            "Observed Value":
                int(
                    len(
                        coverage_safeguards_df
                    )
                ),
        },
    ]
)


print()
print("Data-coverage diagnostics")
print("=" * 100)

display(
    coverage_diagnostics_df
)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(
        country_coverage_df
    )
    == 76
)


observation_counts_valid = bool(
    np.isfinite(
        country_coverage_df[
            observation_col
        ].to_numpy(
            dtype=float
        )
    ).all()
    and
    (
        country_coverage_df[
            observation_col
        ]
        > 0
    ).all()
)


coverage_class_complete = bool(
    country_coverage_df[
        "data_coverage_class"
    ]
    .notna()
    .all()
)


track_history_valid = bool(
    len(
        track_market_count_df
    )
    > 0
    and
    (
        track_market_count_df[
            "markets_observed"
        ]
        >= 1
    ).all()
)


limitations_complete = bool(
    data_coverage_limitations_df[
        [
            "Limitation Area",
            "Observed Evidence",
            "Potential Effect",
            "Required Interpretation",
        ]
    ]
    .notna()
    .all()
    .all()
)


safeguards_complete = bool(
    coverage_safeguards_df[
        [
            "Coverage Safeguard",
            "Application",
        ]
    ]
    .notna()
    .all()
    .all()
)


critical_coverage_limits_present = bool(
    {
        "Unequal country observation coverage",
        "Incomplete global market representation",
        "Sparse track geographic histories",
        "Platform and source dependency",
        "Temporal coverage boundary",
    }
    .issubset(
        set(
            data_coverage_limitations_df[
                "Limitation Area"
            ]
        )
    )
)


section_12_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 12.1 completion",

            "Requirement":
                "Interpretation Boundaries must already be complete.",

            "Observed Evidence":
                f"Section 12.1 completion status: "
                f"{section_12_1_status}",

            "Passed":
                section_12_1_status,
        },
        {
            "Validation Area":
                "Country coverage population",

            "Requirement":
                "All 76 validated country markets must "
                "remain available.",

            "Observed Evidence":
                f"{len(country_coverage_df):,} "
                "country markets assessed",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Observation-count validity",

            "Requirement":
                "Country observation counts must remain "
                "finite and positive.",

            "Observed Evidence":
                (
                    f"Observed range: "
                    f"{country_obs_min:,.0f} to "
                    f"{country_obs_max:,.0f}"
                ),

            "Passed":
                observation_counts_valid,
        },
        {
            "Validation Area":
                "Coverage classification completeness",

            "Requirement":
                "Every country market must receive "
                "a data-coverage classification.",

            "Observed Evidence":
                f"{country_coverage_df['data_coverage_class'].notna().sum():,} "
                "coverage classes assigned",

            "Passed":
                coverage_class_complete,
        },
        {
            "Validation Area":
                "Track history coverage validity",

            "Requirement":
                "Track-country histories must contain at least "
                "one observed market per represented track.",

            "Observed Evidence":
                f"{len(track_market_count_df):,} "
                "track histories assessed",

            "Passed":
                track_history_valid,
        },
        {
            "Validation Area":
                "Data-coverage limitation completeness",

            "Requirement":
                "Each data-coverage limitation must define "
                "evidence, effect and interpretation guidance.",

            "Observed Evidence":
                f"{len(data_coverage_limitations_df):,} "
                "coverage limitations defined",

            "Passed":
                limitations_complete,
        },
        {
            "Validation Area":
                "Critical coverage limitation presence",

            "Requirement":
                "Country, global, track-history, source and "
                "temporal limitations must be explicit.",

            "Observed Evidence":
                "Critical PMIP coverage limitations represented",

            "Passed":
                critical_coverage_limits_present,
        },
        {
            "Validation Area":
                "Coverage safeguard completeness",

            "Requirement":
                "Each coverage safeguard must contain "
                "an explicit application.",

            "Observed Evidence":
                f"{len(coverage_safeguards_df):,} "
                "coverage safeguards defined",

            "Passed":
                safeguards_complete,
        },
    ]
)


print()
print("Data-coverage limitation validation")
print("=" * 100)

display(
    section_12_2_validation_df
)


# ------------------------------------------------------------
# 13. Completion status
# ------------------------------------------------------------

section_12_2_complete = bool(
    section_12_2_validation_df[
        "Passed"
    ].all()
)


if not section_12_2_complete:

    failed_checks = (
        section_12_2_validation_df
        .loc[
            ~section_12_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 12.2 Data-Coverage Limitations "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 12.2 Data-Coverage Limitations "
    "validation checks passed."
)

print(
    f"Section 12.2 completion status: "
    f"{section_12_2_complete}"
)

print(
    "PMIP geographic data-coverage limitations "
    "and safeguards have been documented."
)

print(
    "The workflow is ready for "
    "Section 12.3 Cross-Country Comparison Limitations."
)
```

#### Interpretation

Section 12.2 successfully documents the main **data-coverage limitations** that affect how PMIP geographic intelligence should be interpreted.

The validated coverage sources include:

- **76 country markets**
- **5,338,680 geographic observations**
- **353,282 track-country histories**
- **76 country temporal records**

The first important limitation is **unequal country observation coverage**.

Country observation counts range from only **1,321 observations** in the least-covered market to **90,360 observations** in the most-covered market.

The median is **85,506.5 observations per market**, while the maximum-to-minimum observation ratio is **68.40**.

This is a substantial difference.

It means some countries are represented with far more evidence than others. Markets with richer observation histories may produce more stable and reliable geographic estimates, while low-coverage markets should be interpreted with greater caution.

To make this visible, the 76 markets were divided into four coverage groups:

- **19 markets (25%)** have High observation coverage
- **19 markets (25%)** have Above-median observation coverage
- **19 markets (25%)** have Below-median observation coverage
- **19 markets (25%)** have Limited observation coverage

These classes are relative coverage bands rather than quality labels. A market in the limited group is not necessarily unimportant; it simply has less observational support within the available PMIP dataset.

A second major limitation is that PMIP represents a **validated market universe of 76 countries**, not every music market globally.

Therefore, an artist or track described as having broad international reach is broad **within the PMIP market universe**. The result should not be interpreted as complete worldwide coverage.

Track-level coverage is even more important.

The diagnostics show:

- **81,082 tracks** appear in only **one observed market**
- **29,116 tracks** appear in more than one market

This means the majority of tracks have relatively sparse geographic histories.

A track observed in only one country cannot demonstrate broad cross-market behaviour inside this dataset, but that does not prove that the track had no activity elsewhere in the real world.

Limited observed reach and true geographic limitation are therefore not the same thing.

This sparse-history pattern also explains why earlier PMIP results showed:

- low median international reach,
- high geographic concentration,
- low diversification,
- and a large number of limited geographic profiles.

Those patterns are genuine within the observed dataset, but they are also influenced by the amount of available evidence for each track.

Another key limitation is **platform and source dependency**.

PMIP geographic intelligence is constructed from the streaming and chart information available in the integrated dataset. Activity occurring on platforms, charts, territories, or sources not represented in those datasets cannot contribute to the analysis.

PMIP should therefore be interpreted as **source-bounded intelligence**, rather than a complete measurement of every listening or commercial interaction worldwide.

Temporal coverage creates another boundary.

Geographic expansion, contraction, and momentum are calculated only within the available historical observation window.

Activity that occurred:

- before the dataset begins,
- after the dataset ends,
- or during periods with missing coverage

cannot be directly observed.

This is especially important when interpreting market entry or contraction. A market appearing later in the dataset does not always mean that this was the track's first-ever real-world appearance there, and disappearance from the observation window does not automatically prove permanent market exit.

The temporal-confidence safeguards developed earlier in Section 9 therefore remain important when using geographic movement intelligence.

The analysis also identifies **unequal entity histories** as a limitation.

Tracks with long, multi-market observation histories provide much more evidence for geographic comparison than tracks represented by only a small number of observations or markets.

Confidence in the interpretation should therefore increase with evidence support rather than being based solely on a final score.

PMIP also currently lacks every form of **external market context**.

Country performance is not fully adjusted for all possible structural differences such as:

- population size,
- streaming adoption,
- platform market share,
- music-industry revenue,
- economic size,
- local chart methodology,
- language,
- cultural consumption patterns,
- or touring and radio activity.

As a result, two countries with similar PMIP scores may still represent very different commercial environments.

External market research remains necessary where the analysis is being used for real business decisions.

A final limitation concerns **historical observation bias**.

Artists and tracks with stronger historical data availability naturally provide PMIP with more evidence. Strong evidence support should therefore not automatically be interpreted as stronger underlying real-world popularity.

Section 12.2 formalizes **8 data-coverage limitations** and **6 safeguards**.

The main safeguards are:

- retaining observation context,
- interpreting reach relative to the validated market universe,
- preserving temporal confidence,
- applying caution to sparse histories,
- treating results as source-bounded,
- and supplementing PMIP with external market information when required.

All validation checks passed.

The full **76-country market population** was preserved, country observation counts remained valid and positive, all markets received a coverage classification, all **110,198 track histories** retained at least one observed market, all critical data-coverage limitations were explicitly documented, and all six safeguards contain clear applications.

Section 12.2 therefore establishes the **data-coverage caution layer** for PMIP.

The workflow is ready for **Section 12.3 Cross-Country Comparison Limitations**.

### 12.3 Cross-Country Comparison Limitations

#### Purpose

Assess the main limitations involved when comparing geographic performance across different country markets in PMIP.

This subsection focuses on differences in observation volume, market size, chart structure, streaming behaviour, temporal coverage, and other contextual factors that can make direct country-to-country comparisons imperfect.

The aim is to ensure that country rankings and geographic comparisons are interpreted as relative evidence within the PMIP dataset rather than as universally equivalent measures of national music-market strength.

```python
import pandas as pd
import numpy as np

print("Assessing PMIP cross-country comparison limitations")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_12_2_status = globals().get(
    "section_12_2_complete",
    False
)

print(
    f"Section 12.2 completion status: "
    f"{section_12_2_status}"
)

if not section_12_2_status:
    raise RuntimeError(
        "Section 12.3 requires Section 12.2 "
        "Data-Coverage Limitations to be complete."
    )


# ------------------------------------------------------------
# 2. Recover validated country-level sources
# ------------------------------------------------------------

required_sources = {
    "country_performance": "country_performance_df",
    "country_findings": "country_findings_df",
    "country_coverage": "country_coverage_df",
    "country_temporal": "country_temporal_alignment_df",
}

country_comparison_sources = {}

for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 12.3 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    country_comparison_sources[source_key] = candidate.copy()


country_performance_limit_df = (
    country_comparison_sources["country_performance"]
)

country_findings_limit_df = (
    country_comparison_sources["country_findings"]
)

country_coverage_limit_df = (
    country_comparison_sources["country_coverage"]
)

country_temporal_limit_df = (
    country_comparison_sources["country_temporal"]
)


print()
print("Validated cross-country limitation sources")
print("=" * 100)

print(
    f"Country performance markets: "
    f"{len(country_performance_limit_df):,}"
)

print(
    f"Country findings markets: "
    f"{len(country_findings_limit_df):,}"
)

print(
    f"Country coverage markets: "
    f"{len(country_coverage_limit_df):,}"
)

print(
    f"Country temporal markets: "
    f"{len(country_temporal_limit_df):,}"
)


# ------------------------------------------------------------
# 3. Recover country identity field
# ------------------------------------------------------------

country_col = None

for column in [
    "country",
    "market",
    "country_code",
]:
    if column in country_findings_limit_df.columns:
        country_col = column
        break


if country_col is None:
    raise RuntimeError(
        "Section 12.3 could not identify the country field."
    )


# ------------------------------------------------------------
# 4. Validate required country findings fields
# ------------------------------------------------------------

required_country_fields = [
    "country_findings_index",
    "streaming_strength_percentile",
    "chart_strength_percentile",
    "market_context_percentile",
]


missing_country_fields = [
    column
    for column in required_country_fields
    if column not in country_findings_limit_df.columns
]


if missing_country_fields:
    raise RuntimeError(
        "Section 12.3 is missing required country "
        "comparison fields: "
        + ", ".join(
            missing_country_fields
        )
    )


# ------------------------------------------------------------
# 5. Recover country coverage identity field
# ------------------------------------------------------------

coverage_country_col = None

for column in [
    country_col,
    "country",
    "market",
    "country_code",
]:
    if column in country_coverage_limit_df.columns:
        coverage_country_col = column
        break


if coverage_country_col is None:
    raise RuntimeError(
        "Section 12.3 could not identify the country field "
        "inside country_coverage_df."
    )


# ------------------------------------------------------------
# 6. Recover observation-count field
# ------------------------------------------------------------

source_observation_col = None

for column in [
    "observations",
    "observation_count",
    "total_observations",
]:
    if column in country_coverage_limit_df.columns:
        source_observation_col = column
        break


if source_observation_col is None:
    raise RuntimeError(
        "Section 12.3 could not identify country "
        "observation-count field."
    )


country_coverage_limit_df[
    source_observation_col
] = pd.to_numeric(
    country_coverage_limit_df[
        source_observation_col
    ],
    errors="coerce",
)


if country_coverage_limit_df[
    source_observation_col
].isna().any():

    raise RuntimeError(
        "Section 12.3 found invalid country observation counts."
    )


# ------------------------------------------------------------
# 7. Prepare a collision-safe coverage dataframe
# ------------------------------------------------------------

coverage_merge_df = (
    country_coverage_limit_df[
        [
            coverage_country_col,
            source_observation_col,
            "data_coverage_class",
        ]
    ]
    .copy()
)


coverage_merge_df = coverage_merge_df.rename(
    columns={
        coverage_country_col: country_col,
        source_observation_col: "coverage_observations",
    }
)


observation_col = "coverage_observations"


# ------------------------------------------------------------
# 8. Merge country findings with coverage context
# ------------------------------------------------------------

# Remove any old coverage fields from findings before merging.
# This prevents _x / _y suffixes if the notebook is rerun.

fields_to_remove_before_merge = [
    "coverage_observations",
    "data_coverage_class",
]

country_findings_merge_df = (
    country_findings_limit_df
    .drop(
        columns=[
            column
            for column in fields_to_remove_before_merge
            if column in country_findings_limit_df.columns
        ],
        errors="ignore",
    )
    .copy()
)


country_context_limit_df = (
    country_findings_merge_df
    .merge(
        coverage_merge_df,
        on=country_col,
        how="left",
        validate="one_to_one",
    )
)


if country_context_limit_df[
    observation_col
].isna().any():

    missing_countries = (
        country_context_limit_df
        .loc[
            country_context_limit_df[
                observation_col
            ].isna(),
            country_col,
        ]
        .astype(str)
        .tolist()
    )

    raise RuntimeError(
        "Section 12.3 could not align country observation "
        "coverage with country findings for: "
        + ", ".join(
            missing_countries
        )
    )


if country_context_limit_df[
    "data_coverage_class"
].isna().any():

    raise RuntimeError(
        "Section 12.3 found countries without a "
        "data-coverage classification."
    )


print()
print("Country findings and coverage context aligned")
print("=" * 100)

print(
    f"Aligned country markets: "
    f"{len(country_context_limit_df):,}"
)

print(
    f"Coverage observation field: "
    f"{observation_col}"
)


# ------------------------------------------------------------
# 9. Quantify country comparison dispersion
# ------------------------------------------------------------

country_metric_summary_df = pd.DataFrame(
    [
        {
            "Country Comparison Metric":
                "Streaming strength percentile",

            "Minimum":
                float(
                    country_context_limit_df[
                        "streaming_strength_percentile"
                    ].min()
                ),

            "Median":
                float(
                    country_context_limit_df[
                        "streaming_strength_percentile"
                    ].median()
                ),

            "Maximum":
                float(
                    country_context_limit_df[
                        "streaming_strength_percentile"
                    ].max()
                ),
        },
        {
            "Country Comparison Metric":
                "Chart strength percentile",

            "Minimum":
                float(
                    country_context_limit_df[
                        "chart_strength_percentile"
                    ].min()
                ),

            "Median":
                float(
                    country_context_limit_df[
                        "chart_strength_percentile"
                    ].median()
                ),

            "Maximum":
                float(
                    country_context_limit_df[
                        "chart_strength_percentile"
                    ].max()
                ),
        },
        {
            "Country Comparison Metric":
                "Market context percentile",

            "Minimum":
                float(
                    country_context_limit_df[
                        "market_context_percentile"
                    ].min()
                ),

            "Median":
                float(
                    country_context_limit_df[
                        "market_context_percentile"
                    ].median()
                ),

            "Maximum":
                float(
                    country_context_limit_df[
                        "market_context_percentile"
                    ].max()
                ),
        },
        {
            "Country Comparison Metric":
                "Country findings index",

            "Minimum":
                float(
                    country_context_limit_df[
                        "country_findings_index"
                    ].min()
                ),

            "Median":
                float(
                    country_context_limit_df[
                        "country_findings_index"
                    ].median()
                ),

            "Maximum":
                float(
                    country_context_limit_df[
                        "country_findings_index"
                    ].max()
                ),
        },
        {
            "Country Comparison Metric":
                "Observation count",

            "Minimum":
                float(
                    country_context_limit_df[
                        observation_col
                    ].min()
                ),

            "Median":
                float(
                    country_context_limit_df[
                        observation_col
                    ].median()
                ),

            "Maximum":
                float(
                    country_context_limit_df[
                        observation_col
                    ].max()
                ),
        },
    ]
)


print()
print("Country comparison metric ranges")
print("=" * 100)

display(
    country_metric_summary_df
)


# ------------------------------------------------------------
# 10. Compare findings against coverage bands
# ------------------------------------------------------------

coverage_findings_summary_df = (
    country_context_limit_df
    .groupby(
        "data_coverage_class",
        dropna=False,
    )
    .agg(
        country_markets=(
            country_col,
            "count",
        ),
        median_country_findings_index=(
            "country_findings_index",
            "median",
        ),
        mean_country_findings_index=(
            "country_findings_index",
            "mean",
        ),
        median_streaming_strength=(
            "streaming_strength_percentile",
            "median",
        ),
        median_chart_strength=(
            "chart_strength_percentile",
            "median",
        ),
        median_market_context=(
            "market_context_percentile",
            "median",
        ),
        median_observations=(
            observation_col,
            "median",
        ),
    )
    .reset_index()
)


coverage_class_order = [
    "High observation coverage",
    "Above-median observation coverage",
    "Below-median observation coverage",
    "Limited observation coverage",
]


coverage_findings_summary_df[
    "data_coverage_class"
] = pd.Categorical(
    coverage_findings_summary_df[
        "data_coverage_class"
    ],
    categories=coverage_class_order,
    ordered=True,
)


coverage_findings_summary_df = (
    coverage_findings_summary_df
    .sort_values(
        "data_coverage_class"
    )
    .reset_index(drop=True)
)


print()
print("Country findings by observation-coverage class")
print("=" * 100)

display(
    coverage_findings_summary_df
)


# ------------------------------------------------------------
# 11. Rank sensitivity to coverage
# ------------------------------------------------------------

country_context_limit_df[
    "country_findings_rank"
] = (
    country_context_limit_df[
        "country_findings_index"
    ]
    .rank(
        method="min",
        ascending=False,
    )
)


country_context_limit_df[
    "coverage_rank"
] = (
    country_context_limit_df[
        observation_col
    ]
    .rank(
        method="min",
        ascending=False,
    )
)


rank_correlation = float(
    country_context_limit_df[
        [
            "country_findings_rank",
            "coverage_rank",
        ]
    ]
    .corr(
        method="spearman"
    )
    .iloc[
        0,
        1,
    ]
)


print()
print("Country findings versus coverage ranking")
print("=" * 100)

print(
    f"Spearman rank correlation between "
    f"country findings rank and observation-coverage rank: "
    f"{rank_correlation:.4f}"
)


# ------------------------------------------------------------
# 12. Identify higher-performing lower-coverage markets
# ------------------------------------------------------------

findings_median = float(
    country_context_limit_df[
        "country_findings_index"
    ].median()
)


low_coverage_labels = [
    "Below-median observation coverage",
    "Limited observation coverage",
]


high_performance_low_coverage_df = (
    country_context_limit_df
    .loc[
        (
            country_context_limit_df[
                "country_findings_index"
            ]
            >= findings_median
        )
        &
        (
            country_context_limit_df[
                "data_coverage_class"
            ]
            .isin(
                low_coverage_labels
            )
        )
    ]
    .sort_values(
        "country_findings_index",
        ascending=False,
    )
    [
        [
            country_col,
            "country_findings_index",
            "country_findings_class",
            observation_col,
            "data_coverage_class",
            "streaming_strength_percentile",
            "chart_strength_percentile",
            "market_context_percentile",
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Higher-performing markets with lower observation coverage")
print("=" * 100)

display(
    high_performance_low_coverage_df
)


# ------------------------------------------------------------
# 13. Identify lower-performing higher-coverage markets
# ------------------------------------------------------------

high_coverage_labels = [
    "High observation coverage",
    "Above-median observation coverage",
]


lower_performance_high_coverage_df = (
    country_context_limit_df
    .loc[
        (
            country_context_limit_df[
                "country_findings_index"
            ]
            < findings_median
        )
        &
        (
            country_context_limit_df[
                "data_coverage_class"
            ]
            .isin(
                high_coverage_labels
            )
        )
    ]
    .sort_values(
        "country_findings_index",
        ascending=True,
    )
    [
        [
            country_col,
            "country_findings_index",
            "country_findings_class",
            observation_col,
            "data_coverage_class",
            "streaming_strength_percentile",
            "chart_strength_percentile",
            "market_context_percentile",
        ]
    ]
    .reset_index(drop=True)
)


print()
print("Lower-performing markets with higher observation coverage")
print("=" * 100)

display(
    lower_performance_high_coverage_df
)


# ------------------------------------------------------------
# 14. Define cross-country comparison limitations
# ------------------------------------------------------------

cross_country_limitations_df = pd.DataFrame(
    [
        {
            "Limitation Area":
                "Unequal observation volume",

            "Observed Evidence":
                (
                    f"Country observation counts range from "
                    f"{country_context_limit_df[observation_col].min():,.0f} "
                    f"to "
                    f"{country_context_limit_df[observation_col].max():,.0f}."
                ),

            "Why It Matters":
                "Countries with very different evidence volumes "
                "may not have equally stable estimates.",

            "Responsible Comparison Rule":
                "Inspect observation coverage alongside country scores.",
        },
        {
            "Limitation Area":
                "Different market sizes",

            "Observed Evidence":
                "PMIP compares observed performance but does not "
                "fully standardise for national population or "
                "music-market size.",

            "Why It Matters":
                "A similar streaming level can imply different "
                "market penetration in differently sized countries.",

            "Responsible Comparison Rule":
                "Do not interpret equal PMIP performance as equal "
                "commercial market value.",
        },
        {
            "Limitation Area":
                "Different streaming ecosystems",

            "Observed Evidence":
                "Streaming-platform adoption and platform market "
                "share are not identical across all 76 countries.",

            "Why It Matters":
                "Observed streaming activity can reflect platform "
                "availability and adoption as well as music demand.",

            "Responsible Comparison Rule":
                "Treat streaming comparisons as platform-bounded.",
        },
        {
            "Limitation Area":
                "Different chart systems",

            "Observed Evidence":
                "Country chart behaviour may differ in methodology, "
                "ranking depth and eligibility rules.",

            "Why It Matters":
                "A chart position may not represent exactly the same "
                "competitive environment in every market.",

            "Responsible Comparison Rule":
                "Use chart percentiles as relative PMIP evidence, "
                "not universal chart equivalence.",
        },
        {
            "Limitation Area":
                "Different temporal coverage",

            "Observed Evidence":
                (
                    f"{len(country_temporal_limit_df):,} country "
                    f"temporal records are represented, but historical "
                    f"availability may differ by market."
                ),

            "Why It Matters":
                "Countries with shorter or irregular histories may "
                "have less evidence for change over time.",

            "Responsible Comparison Rule":
                "Retain temporal-confidence context for movement claims.",
        },
        {
            "Limitation Area":
                "Language and cultural differences",

            "Observed Evidence":
                "PMIP does not directly model local language preference, "
                "genre culture or cultural consumption patterns.",

            "Why It Matters":
                "Performance differences may partly reflect local "
                "cultural preferences rather than general market quality.",

            "Responsible Comparison Rule":
                "Use cultural and genre context alongside PMIP findings.",
        },
        {
            "Limitation Area":
                "External economic context",

            "Observed Evidence":
                "PMIP country scores do not directly include GDP, "
                "music revenue, disposable income or ticketing value.",

            "Why It Matters":
                "High observed music activity does not automatically "
                "mean greater commercial value.",

            "Responsible Comparison Rule":
                "Supplement PMIP with economic and industry research.",
        },
        {
            "Limitation Area":
                "Rank sensitivity",

            "Observed Evidence":
                (
                    f"Observed Spearman relationship between findings "
                    f"rank and coverage rank: {rank_correlation:.4f}."
                ),

            "Why It Matters":
                "Country rankings may partly reflect differences in "
                "the quantity of available evidence.",

            "Responsible Comparison Rule":
                "Treat ranks as analytical summaries rather than "
                "absolute country league tables.",
        },
    ]
)


print()
print("PMIP cross-country comparison limitation framework")
print("=" * 100)

display(
    cross_country_limitations_df
)


# ------------------------------------------------------------
# 15. Define cross-country comparison safeguards
# ------------------------------------------------------------

cross_country_safeguards_df = pd.DataFrame(
    [
        {
            "Comparison Safeguard":
                "Coverage-aware comparison",

            "Application":
                "Review observation counts and coverage classes "
                "before drawing conclusions between markets.",
        },
        {
            "Comparison Safeguard":
                "Percentile-based interpretation",

            "Application":
                "Use streaming, chart and market-context percentiles "
                "as relative measures inside PMIP.",
        },
        {
            "Comparison Safeguard":
                "No universal equivalence",

            "Application":
                "Do not assume equal scores imply equal country "
                "population, audience size or commercial value.",
        },
        {
            "Comparison Safeguard":
                "Temporal context preservation",

            "Application":
                "Use temporal alignment and confidence evidence "
                "when comparing geographic change.",
        },
        {
            "Comparison Safeguard":
                "External contextualisation",

            "Application":
                "Combine PMIP country findings with market size, "
                "economics, language and industry research.",
        },
        {
            "Comparison Safeguard":
                "Rank caution",

            "Application":
                "Treat country rankings as prioritisation aids rather "
                "than definitive measures of national market quality.",
        },
    ]
)


print()
print("Cross-country comparison safeguards")
print("=" * 100)

display(
    cross_country_safeguards_df
)


# ------------------------------------------------------------
# 16. Diagnostics
# ------------------------------------------------------------

cross_country_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Country markets compared",

            "Observed Value":
                int(
                    len(
                        country_context_limit_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Higher-performing lower-coverage markets",

            "Observed Value":
                int(
                    len(
                        high_performance_low_coverage_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Lower-performing higher-coverage markets",

            "Observed Value":
                int(
                    len(
                        lower_performance_high_coverage_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Findings-versus-coverage rank correlation",

            "Observed Value":
                round(
                    rank_correlation,
                    4,
                ),
        },
        {
            "Diagnostic":
                "Cross-country limitations defined",

            "Observed Value":
                int(
                    len(
                        cross_country_limitations_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Cross-country safeguards defined",

            "Observed Value":
                int(
                    len(
                        cross_country_safeguards_df
                    )
                ),
        },
    ]
)


print()
print("Cross-country comparison diagnostics")
print("=" * 100)

display(
    cross_country_diagnostics_df
)


# ------------------------------------------------------------
# 17. Validation
# ------------------------------------------------------------

country_population_valid = bool(
    len(
        country_context_limit_df
    )
    == 76
)


country_unique_valid = bool(
    country_context_limit_df[
        country_col
    ].nunique()
    == 76
)


comparison_numeric_valid = bool(
    np.isfinite(
        country_context_limit_df[
            [
                "country_findings_index",
                "streaming_strength_percentile",
                "chart_strength_percentile",
                "market_context_percentile",
                observation_col,
            ]
        ]
        .to_numpy(
            dtype=float
        )
    ).all()
)


coverage_context_complete = bool(
    country_context_limit_df[
        [
            observation_col,
            "data_coverage_class",
        ]
    ]
    .notna()
    .all()
    .all()
)


rank_correlation_valid = bool(
    np.isfinite(
        rank_correlation
    )
)


limitations_complete = bool(
    cross_country_limitations_df[
        [
            "Limitation Area",
            "Observed Evidence",
            "Why It Matters",
            "Responsible Comparison Rule",
        ]
    ]
    .notna()
    .all()
    .all()
)


safeguards_complete = bool(
    cross_country_safeguards_df[
        [
            "Comparison Safeguard",
            "Application",
        ]
    ]
    .notna()
    .all()
    .all()
)


critical_limitations_present = bool(
    {
        "Unequal observation volume",
        "Different market sizes",
        "Different streaming ecosystems",
        "Different chart systems",
        "Different temporal coverage",
        "Rank sensitivity",
    }
    .issubset(
        set(
            cross_country_limitations_df[
                "Limitation Area"
            ]
        )
    )
)


section_12_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 12.2 completion",

            "Requirement":
                "Data-Coverage Limitations must already be complete.",

            "Observed Evidence":
                f"Section 12.2 completion status: "
                f"{section_12_2_status}",

            "Passed":
                section_12_2_status,
        },
        {
            "Validation Area":
                "Country comparison population",

            "Requirement":
                "All 76 validated country markets must be retained.",

            "Observed Evidence":
                f"{len(country_context_limit_df):,} "
                "country markets compared",

            "Passed":
                country_population_valid,
        },
        {
            "Validation Area":
                "Country identity uniqueness",

            "Requirement":
                "Each country market must appear once.",

            "Observed Evidence":
                f"{country_context_limit_df[country_col].nunique():,} "
                "unique countries",

            "Passed":
                country_unique_valid,
        },
        {
            "Validation Area":
                "Country comparison numeric validity",

            "Requirement":
                "Country comparison indicators must remain finite.",

            "Observed Evidence":
                "Country findings, streaming, chart, context "
                "and observation indicators assessed",

            "Passed":
                comparison_numeric_valid,
        },
        {
            "Validation Area":
                "Coverage context completeness",

            "Requirement":
                "Each country must retain observation coverage context.",

            "Observed Evidence":
                f"{country_context_limit_df['data_coverage_class'].notna().sum():,} "
                "coverage contexts retained",

            "Passed":
                coverage_context_complete,
        },
        {
            "Validation Area":
                "Rank sensitivity validity",

            "Requirement":
                "Country findings-versus-coverage rank relationship "
                "must be measurable.",

            "Observed Evidence":
                f"Spearman rank correlation: "
                f"{rank_correlation:.4f}",

            "Passed":
                rank_correlation_valid,
        },
        {
            "Validation Area":
                "Cross-country limitation completeness",

            "Requirement":
                "Each limitation must define evidence, impact "
                "and a responsible comparison rule.",

            "Observed Evidence":
                f"{len(cross_country_limitations_df):,} "
                "cross-country limitations defined",

            "Passed":
                limitations_complete,
        },
        {
            "Validation Area":
                "Critical comparison limitation presence",

            "Requirement":
                "Coverage, size, platform, chart, temporal and "
                "ranking limitations must be explicit.",

            "Observed Evidence":
                "Critical cross-country comparison limitations represented",

            "Passed":
                critical_limitations_present,
        },
        {
            "Validation Area":
                "Comparison safeguard completeness",

            "Requirement":
                "Every comparison safeguard must contain "
                "an explicit application.",

            "Observed Evidence":
                f"{len(cross_country_safeguards_df):,} "
                "comparison safeguards defined",

            "Passed":
                safeguards_complete,
        },
    ]
)


print()
print("Cross-country comparison limitation validation")
print("=" * 100)

display(
    section_12_3_validation_df
)


# ------------------------------------------------------------
# 18. Completion status
# ------------------------------------------------------------

section_12_3_complete = bool(
    section_12_3_validation_df[
        "Passed"
    ].all()
)


if not section_12_3_complete:

    failed_checks = (
        section_12_3_validation_df
        .loc[
            ~section_12_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 12.3 Cross-Country Comparison Limitations "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 12.3 Cross-Country Comparison Limitations "
    "validation checks passed."
)

print(
    f"Section 12.3 completion status: "
    f"{section_12_3_complete}"
)

print(
    "PMIP cross-country comparison limitations "
    "and safeguards have been documented."
)

print(
    "The workflow is ready for "
    "Section 12.4 Human Interpretation Requirements."
)
```

#### Interpretation

Section 12.3 successfully evaluates the main limitations involved in comparing country markets within PMIP.

All **76 validated country markets** were retained and successfully aligned with their country-level findings, observation-coverage information, and temporal context.

The analysis confirms that country comparison within PMIP must be treated carefully because the underlying markets do not have identical levels of evidence or equivalent commercial environments.

The first major limitation is **unequal observation volume**.

Country observation counts range from:

- **1,321 observations**
- to **90,360 observations**

with a median of **85,506.5 observations**.

This means that some country markets are supported by substantially more evidence than others.

Markets with larger observation histories are likely to produce more stable geographic estimates, while markets with lower coverage should be interpreted with greater caution.

The comparison by coverage class reinforces this pattern.

Countries in the **High observation coverage** group have a median country findings index of approximately **0.7704**, while countries in the **Limited observation coverage** group have a median index of approximately **0.3026**.

A similar pattern appears across the component measures:

- streaming strength,
- chart strength,
- market context,
- and overall country findings.

This suggests that stronger observational support is associated with stronger country findings within the PMIP dataset.

The strongest evidence of this relationship is the **Spearman rank correlation of 0.8489** between country findings rank and observation-coverage rank.

A correlation of this size indicates a strong positive relationship.

In practical terms, country markets with higher observation coverage tend to also appear higher in PMIP country rankings.

This is an important limitation because it means country rankings may partially reflect differences in the amount of available evidence rather than only differences in underlying music-market strength.

However, coverage does not fully determine country performance.

The analysis identified:

- **3 higher-performing markets with lower observation coverage**
- **3 lower-performing markets with higher observation coverage**

This demonstrates that high PMIP performance is not simply a result of having more observations.

Country rankings therefore contain meaningful performance information, but observation coverage remains an important contextual factor that should always accompany interpretation.

A second limitation is **different market size**.

PMIP does not fully standardise country comparisons for factors such as:

- national population,
- number of music consumers,
- streaming-market size,
- or total commercial music value.

As a result, two markets with similar PMIP scores may represent very different real-world audience or revenue opportunities.

Equal PMIP performance must therefore not be interpreted as equal commercial market value.

A third limitation concerns **different streaming ecosystems**.

Streaming adoption, platform availability, and platform market share can vary significantly between countries.

Observed streaming performance may therefore reflect not only audience demand, but also differences in:

- platform usage,
- digital infrastructure,
- consumer behaviour,
- and source availability.

PMIP streaming comparisons should therefore be treated as platform-bounded evidence.

Another important limitation is **different chart systems**.

Country charts can vary in:

- chart methodology,
- eligibility rules,
- chart depth,
- reporting frequency,
- and the way streaming or sales activity contributes to rankings.

This means the same chart position may not represent exactly the same competitive environment across two different countries.

PMIP therefore uses chart evidence relatively, but chart percentiles should still not be interpreted as perfectly universal equivalents.

Temporal differences also matter.

All **76 country temporal records** are represented, but historical depth or regularity may differ between markets.

Countries with shorter or more irregular histories may provide weaker evidence for geographic movement, expansion, or contraction.

For this reason, temporal-confidence information from the earlier geographic-movement analysis should continue to be preserved when interpreting country-level change.

Language and cultural differences provide another important boundary.

PMIP does not directly model:

- local-language preference,
- genre traditions,
- cultural listening behaviour,
- regional music scenes,
- or local promotional structures.

Performance differences may therefore reflect cultural fit rather than a universal difference in country-market quality.

External economic factors are also not fully included.

The country findings do not directly incorporate factors such as:

- GDP,
- disposable income,
- recorded-music revenue,
- ticketing value,
- advertising value,
- or local music-industry investment.

A market showing strong observed streaming and chart activity is therefore not automatically the most commercially valuable market.

The final major limitation is **ranking sensitivity**.

Country rankings are useful for prioritisation, but they should not be treated as an absolute league table of national music-market quality.

They summarise evidence within the PMIP analytical framework and are influenced by:

- observation volume,
- market context,

Together, these safeguards ensure that country comparisons remain useful without overstating what the data can prove.

#### Interpretation

Section 12.4 successfully defines the **human interpretation requirements** for PMIP geographic intelligence and completes the Responsible Use and Limitations framework.

The analysis confirms that PMIP should operate as a **decision-support system**, not as an automated decision-maker.

A total of **10 human interpretation requirements** were defined, covering the situations where human judgement is necessary before PMIP outputs are acted upon.

The first and most important requirement concerns **signing and investment decisions**.

PMIP geographic scores do not directly measure:

- artistic quality,
- strategic label fit,
- contract value,
- revenue potential,
- long-term artist development,
- management capability,
- marketing cost,
- or broader commercial opportunity.

For this reason, PMIP must not automatically determine whether an artist should be signed, rejected, promoted, funded, or prioritised.

Instead, geographic intelligence should contribute evidence to a wider professional review.

A second requirement concerns **high-priority geographic signals**.

High-priority classifications can identify interesting markets, artists, or tracks, but they should not be interpreted in isolation.

A high score may result from:

- strong penetration but narrow reach,
- strong expansion but weak diversification,
- strong performance in only a small number of markets,
- or high values supported by uneven observation coverage.

Human reviewers should therefore inspect the underlying geographic dimensions before acting on a final classification.

The framework also requires human review for **country prioritisation**.

Country rankings must be considered together with:

- observation coverage,
- temporal evidence,
- market size,
- cultural context,
- streaming behaviour,
- chart structure,
- and external commercial information.

This reinforces the finding from Section 12.3 that PMIP country rankings are useful prioritisation aids, but not absolute measures of country value.

Another important requirement concerns **artist nationality and track origin**.

Geographic performance does not establish an artist's nationality, citizenship, home country, ethnicity, or cultural identity.

Similarly, a track's strongest observed market does not establish where that track originated, was recorded, was released, or first became successful.

Such information must come from verified external metadata when required.

Human review is also mandatory when interpreting **expansion and momentum**.

These indicators describe observed historical movement within the available data window.

They are not guaranteed forecasts.

Reviewers should therefore consider:

- temporal confidence,
- observation history,
- recent evidence,
- market-entry timing,
- and potential gaps in coverage

before interpreting geographic movement as meaningful.

The framework explicitly identifies **sparse-data cases** as another human-review requirement.

Tracks and artists with limited geographic histories provide less evidence for comparison.

Low observed reach must therefore not automatically be interpreted as true real-world geographic weakness.

Human judgement is needed to distinguish between genuinely limited performance and limited available evidence.

A particularly important requirement is the interpretation of **conflicting component scores**.

PMIP combines several different geographic dimensions:

- reach,
- penetration,
- diversification,
- expansion,
- concentration,
- and related temporal indicators.

These dimensions can disagree.

For example, a track may show strong penetration in a small number of countries while having weak international reach.

Another track may have broad reach but relatively shallow performance within each market.

The final summary score should therefore never replace inspection of the underlying components.

Human reviewers must interpret the structure of the geographic profile rather than relying only on one aggregate number.

The framework also requires human expertise for **cultural and genre context**.

PMIP does not fully model:

- language preference,
- genre communities,
- local music scenes,
- cultural consumption patterns,
- regional media behaviour,
- or territory-specific audience expectations.

Market and territory expertise therefore remains important when interpreting why an artist or track performs differently across countries.

External commercial context is also required.

Geographic intelligence alone does not fully include:

- revenue,
- ticketing potential,
- touring feasibility,
- marketing costs,
- local partnerships,
- media opportunities,
- population size,
- or economic value.

These factors must be considered separately before PMIP evidence is converted into commercial action.

Five human review roles were formally defined:

- **A&R / Music Strategy**
- **Data / Analytics**
- **Market / Territory Expert**
- **Commercial / Business**
- **Legal / Responsible Use**

These roles reflect the fact that responsible interpretation requires both analytical and domain-specific knowledge.

The framework also defines **7 mandatory human-review triggers**.

Human review is required when PMIP produces:

- strong or exceptional classifications,
- high expansion or momentum signals,
- low-coverage evidence,
- commercial cross-country rankings,
- conflicting geographic dimensions,
- identity or origin interpretations,
- or recommendations involving signing, promotion, or investment.

Five final responsible-decision principles were established:

- **Decision support, not decision replacement**
- **Evidence before action**
- **Uncertainty must remain visible**
- **Context-sensitive interpretation**
- **No unsupported inference**

These principles provide the final operational rule for the PMIP geographic intelligence system.

```python
from pathlib import Path
import pandas as pd
import json
import os

print("Saving PMIP geographic intelligence results")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_12_status = globals().get(
    "section_12_complete",
    False
)

print(
    f"Section 12 overall completion status: "
    f"{section_12_status}"
)

if not section_12_status:
    raise RuntimeError(
        "Section 13.1 requires Section 12 "
        "Responsible Use and Limitations to be complete."
    )


# ------------------------------------------------------------
# 2. Define artifact directory
# ------------------------------------------------------------

project_root = Path.cwd()

artifact_dir = (
    project_root
    / "models"
    / "geographic_intelligence"
)

artifact_dir.mkdir(
    parents=True,
    exist_ok=True
)


print()
print("Geographic intelligence artifact directory")
print("=" * 100)

print(
    f"Artifact directory: "
    f"{artifact_dir.resolve()}"
)


# ------------------------------------------------------------
# 3. Define validated result artifacts
# ------------------------------------------------------------

artifact_registry = {
    # Country-level outputs
    "country_performance":
        "country_performance_df",

    "country_findings":
        "country_findings_df",

    "country_coverage":
        "country_coverage_df",

    "country_temporal_alignment":
        "country_temporal_alignment_df",

    # Artist-level outputs
    "artist_international_reach":
        "artist_international_reach_df",

    "artist_market_penetration":
        "artist_market_penetration_intelligence_df",

    "artist_geographic_concentration":
        "artist_geographic_concentration_intelligence_df",

    "artist_local_international":
        "artist_local_international_df",

    "artist_geographic_profile":
        "artist_geographic_profile_df",

    "artist_findings":
        "artist_findings_df",

    # Track-level geographic intelligence
    "track_international_reach":
        "track_international_reach_df",

    "track_market_penetration":
        "track_market_penetration_df",

    "track_cross_market_performance":
        "track_cross_market_performance_df",

    "track_geographic_concentration":
        "track_geographic_concentration_df",

    "track_geographic_profile":
        "track_geographic_profile_df",

    # Geographic movement
    "track_country_entry":
        "track_country_entry_df",

    "track_geographic_expansion":
        "track_geographic_expansion_df",

    "track_geographic_contraction":
        "track_geographic_contraction_df",

    "track_cross_market_momentum":
        "track_cross_market_momentum_df",

    "track_temporal_validation":
        "track_temporal_validation_df",

    # Geographic scoring
    "track_geographic_reach_score":
        "track_geographic_reach_score_df",

    "track_market_penetration_score":
        "track_market_penetration_score_df",

    "track_geographic_concentration_score":
        "track_geographic_concentration_score_df",

    "track_geographic_expansion_indicator":
        "track_geographic_expansion_indicator_df",

    "track_geographic_intelligence":
        "track_geographic_intelligence_df",

    # Results and interpretation
    "geographic_comparison":
        "geographic_comparison_df",

    "cross_level_geographic_findings":
        "cross_level_geographic_findings_df",

    "high_priority_insights":
        "high_priority_insights_df",

    # Responsible use
    "interpretation_boundaries":
        "interpretation_boundaries_df",

    "responsible_use_policy":
        "responsible_use_policy_df",

    "data_coverage_limitations":
        "data_coverage_limitations_df",

    "coverage_safeguards":
        "coverage_safeguards_df",

    "cross_country_limitations":
        "cross_country_limitations_df",

    "cross_country_safeguards":
        "cross_country_safeguards_df",

    "human_interpretation_requirements":
        "human_interpretation_requirements_df",

    "human_review_roles":
        "human_review_roles_df",

    "human_review_triggers":
        "human_review_triggers_df",

    "responsible_decision_policy":
        "responsible_decision_policy_df",
}


# ------------------------------------------------------------
# 4. Recover only artifacts that exist and are dataframes
# ------------------------------------------------------------

available_artifacts = {}
missing_artifacts = []
invalid_artifacts = []

for artifact_name, variable_name in artifact_registry.items():

    if variable_name not in globals():
        missing_artifacts.append(
            {
                "Artifact":
                    artifact_name,

                "Expected Variable":
                    variable_name,
            }
        )
        continue

    candidate = globals()[variable_name]

    if not isinstance(candidate, pd.DataFrame):
        invalid_artifacts.append(
            {
                "Artifact":
                    artifact_name,

                "Expected Variable":
                    variable_name,

                "Observed Type":
                    type(candidate).__name__,
            }
        )
        continue

    available_artifacts[
        artifact_name
    ] = candidate.copy()


print()
print("Artifact source recovery")
print("=" * 100)

print(
    f"Registered artifacts: "
    f"{len(artifact_registry):,}"
)

print(
    f"Available valid dataframe artifacts: "
    f"{len(available_artifacts):,}"
)

print(
    f"Missing registered artifacts: "
    f"{len(missing_artifacts):,}"
)

print(
    f"Invalid registered artifacts: "
    f"{len(invalid_artifacts):,}"
)


if invalid_artifacts:
    invalid_artifacts_df = pd.DataFrame(
        invalid_artifacts
    )

    display(
        invalid_artifacts_df
    )


# ------------------------------------------------------------
# 5. Define required core artifacts
# ------------------------------------------------------------

required_core_artifacts = [
    "country_performance",
    "country_findings",
    "artist_geographic_profile",
    "artist_findings",
    "track_geographic_intelligence",
    "track_temporal_validation",
    "high_priority_insights",
    "interpretation_boundaries",
    "data_coverage_limitations",
    "cross_country_limitations",
    "human_interpretation_requirements",
]


missing_required_core = [
    artifact_name
    for artifact_name in required_core_artifacts
    if artifact_name not in available_artifacts
]


if missing_required_core:
    raise RuntimeError(
        "Section 13.1 could not recover required core artifacts: "
        + ", ".join(
            missing_required_core
        )
    )


# ------------------------------------------------------------
# 6. Save dataframe artifacts as CSV
# ------------------------------------------------------------

saved_artifact_records = []

for artifact_name, dataframe in available_artifacts.items():

    output_path = (
        artifact_dir
        / f"{artifact_name}.csv"
    )

    dataframe.to_csv(
        output_path,
        index=False
    )

    saved_artifact_records.append(
        {
            "Artifact":
                artifact_name,

            "Source Variable":
                artifact_registry[
                    artifact_name
                ],

            "File Name":
                output_path.name,

            "Rows":
                int(
                    len(dataframe)
                ),

            "Columns":
                int(
                    dataframe.shape[1]
                ),

            "File Size (Bytes)":
                int(
                    output_path.stat().st_size
                ),

            "Saved":
                bool(
                    output_path.exists()
                    and
                    output_path.stat().st_size > 0
                ),
        }
    )


saved_artifacts_df = pd.DataFrame(
    saved_artifact_records
)


print()
print("Saved geographic intelligence artifacts")
print("=" * 100)

display(
    saved_artifacts_df
)


# ------------------------------------------------------------
# 7. Save artifact manifest
# ------------------------------------------------------------

artifact_manifest = {
    "artifact_group":
        "PMIP Geographic Intelligence",

    "artifact_directory":
        str(
            artifact_dir.resolve()
        ),

    "artifact_count":
        int(
            len(saved_artifacts_df)
        ),

    "artifacts":
        saved_artifacts_df[
            [
                "Artifact",
                "Source Variable",
                "File Name",
                "Rows",
                "Columns",
                "File Size (Bytes)",
            ]
        ]
        .to_dict(
            orient="records"
        ),
}


manifest_path = (
    artifact_dir
    / "geographic_intelligence_artifact_manifest.json"
)


with open(
    manifest_path,
    "w",
    encoding="utf-8"
) as manifest_file:

    json.dump(
        artifact_manifest,
        manifest_file,
        indent=4
    )


print()
print("Artifact manifest")
print("=" * 100)

print(
    f"Manifest saved: "
    f"{manifest_path.name}"
)


# ------------------------------------------------------------
# 8. Diagnostics
# ------------------------------------------------------------

total_rows_saved = int(
    saved_artifacts_df[
        "Rows"
    ].sum()
)

total_bytes_saved = int(
    saved_artifacts_df[
        "File Size (Bytes)"
    ].sum()
)

required_core_saved = bool(
    set(
        required_core_artifacts
    )
    .issubset(
        set(
            saved_artifacts_df[
                "Artifact"
            ]
        )
    )
)


artifact_saving_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Artifacts saved",

            "Observed Value":
                int(
                    len(
                        saved_artifacts_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Required core artifacts saved",

            "Observed Value":
                int(
                    len(
                        required_core_artifacts
                    )
                ),
        },
        {
            "Diagnostic":
                "Total dataframe rows saved",

            "Observed Value":
                total_rows_saved,
        },
        {
            "Diagnostic":
                "Total artifact bytes saved",

            "Observed Value":
                total_bytes_saved,
        },
        {
            "Diagnostic":
                "Manifest files saved",

            "Observed Value":
                int(
                    manifest_path.exists()
                ),
        },
    ]
)


print()
print("Artifact-saving diagnostics")
print("=" * 100)

display(
    artifact_saving_diagnostics_df
)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

all_artifacts_saved = bool(
    saved_artifacts_df[
        "Saved"
    ].all()
)


all_artifact_files_exist = bool(
    all(
        (
            artifact_dir
            / file_name
        ).exists()
        for file_name
        in saved_artifacts_df[
            "File Name"
        ]
    )
)


all_artifact_files_nonempty = bool(
    all(
        (
            artifact_dir
            / file_name
        ).stat().st_size > 0
        for file_name
        in saved_artifacts_df[
            "File Name"
        ]
    )
)


manifest_valid = bool(
    manifest_path.exists()
    and
    manifest_path.stat().st_size > 0
)


artifact_population_valid = bool(
    len(
        saved_artifacts_df
    )
    >= len(
        required_core_artifacts
    )
)


section_13_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 12 completion",

            "Requirement":
                "Responsible Use and Limitations must "
                "already be complete.",

            "Observed Evidence":
                f"Section 12 overall completion status: "
                f"{section_12_status}",

            "Passed":
                section_12_status,
        },
        {
            "Validation Area":
                "Required core artifact preservation",

            "Requirement":
                "All required geographic intelligence "
                "artifacts must be available for saving.",

            "Observed Evidence":
                f"{len(required_core_artifacts):,} "
                "required core artifacts saved",

            "Passed":
                required_core_saved,
        },
        {
            "Validation Area":
                "Artifact saving completeness",

            "Requirement":
                "Every recovered artifact must be "
                "written successfully.",

            "Observed Evidence":
                f"{len(saved_artifacts_df):,} "
                "dataframe artifacts saved",

            "Passed":
                all_artifacts_saved,
        },
        {
            "Validation Area":
                "Artifact file existence",

            "Requirement":
                "Every saved artifact file must exist "
                "on disk.",

            "Observed Evidence":
                f"{len(saved_artifacts_df):,} "
                "artifact files verified",

            "Passed":
                all_artifact_files_exist,
        },
        {
            "Validation Area":
                "Artifact file integrity",

            "Requirement":
                "Every saved artifact file must contain "
                "non-zero content.",

            "Observed Evidence":
                f"{len(saved_artifacts_df):,} "
                "non-empty artifact files verified",

            "Passed":
                all_artifact_files_nonempty,
        },
        {
            "Validation Area":
                "Artifact population validity",

            "Requirement":
                "Saved artifact population must contain "
                "at least all required core artifacts.",

            "Observed Evidence":
                f"{len(saved_artifacts_df):,} "
                "artifacts in saved population",

            "Passed":
                artifact_population_valid,
        },
        {
            "Validation Area":
                "Artifact manifest validity",

            "Requirement":
                "The geographic intelligence artifact "
                "manifest must exist and be non-empty.",

            "Observed Evidence":
                f"Manifest file: "
                f"{manifest_path.name}",

            "Passed":
                manifest_valid,
        },
    ]
)


print()
print("Geographic intelligence artifact saving validation")
print("=" * 100)

display(
    section_13_1_validation_df
)


# ------------------------------------------------------------
# 10. Completion status
# ------------------------------------------------------------

section_13_1_complete = bool(
    section_13_1_validation_df[
        "Passed"
    ].all()
)


if not section_13_1_complete:

    failed_checks = (
        section_13_1_validation_df
        .loc[
            ~section_13_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 13.1 Save Geographic Intelligence Results "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 13.1 Save Geographic Intelligence Results "
    "validation checks passed."
)

print(
    f"Section 13.1 completion status: "
    f"{section_13_1_complete}"
)

print(
    f"Geographic intelligence artifacts saved to: "
    f"{artifact_dir.resolve()}"
)

print(
    "The workflow is ready for "
    "Section 13.2 Save Configuration and Metadata."
)
```

```python
from pathlib import Path
import shutil
import json

print("Correcting PMIP geographic intelligence artifact location")
print("=" * 100)

# ------------------------------------------------------------
# 1. Recover current notebook directory
# ------------------------------------------------------------

current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
else:
    project_root = current_dir


# ------------------------------------------------------------
# 2. Define incorrect and correct artifact locations
# ------------------------------------------------------------

old_artifact_dir = (
    current_dir
    / "models"
    / "geographic_intelligence"
)

artifact_dir = (
    project_root
    / "models"
    / "geographic_intelligence"
)

artifact_dir.mkdir(
    parents=True,
    exist_ok=True
)


print()
print("Artifact location correction")
print("=" * 100)

print(
    f"Previous location: "
    f"{old_artifact_dir.resolve()}"
)

print(
    f"Correct location: "
    f"{artifact_dir.resolve()}"
)


# ------------------------------------------------------------
# 3. Move saved artifacts to project-level models folder
# ------------------------------------------------------------

moved_files = []

if (
    old_artifact_dir.exists()
    and
    old_artifact_dir.resolve()
    != artifact_dir.resolve()
):

    for source_path in old_artifact_dir.iterdir():

        destination_path = (
            artifact_dir
            / source_path.name
        )

        if destination_path.exists():
            destination_path.unlink()

        shutil.move(
            str(source_path),
            str(destination_path)
        )

        moved_files.append(
            source_path.name
        )

    # Remove now-empty incorrect directory
    old_artifact_dir.rmdir()

    # Remove notebooks/models if it is now empty
    old_models_dir = old_artifact_dir.parent

    if (
        old_models_dir.exists()
        and
        not any(old_models_dir.iterdir())
    ):
        old_models_dir.rmdir()


# ------------------------------------------------------------
# 4. Update manifest with corrected path
# ------------------------------------------------------------

manifest_path = (
    artifact_dir
    / "geographic_intelligence_artifact_manifest.json"
)


if not manifest_path.exists():
    raise RuntimeError(
        "Artifact manifest could not be recovered "
        "after location correction."
    )


with open(
    manifest_path,
    "r",
    encoding="utf-8"
) as manifest_file:

    artifact_manifest = json.load(
        manifest_file
    )


artifact_manifest[
    "artifact_directory"
] = str(
    artifact_dir.resolve()
)


with open(
    manifest_path,
    "w",
    encoding="utf-8"
) as manifest_file:

    json.dump(
        artifact_manifest,
        manifest_file,
        indent=4
    )


# ------------------------------------------------------------
# 5. Verify saved artifact population
# ------------------------------------------------------------

expected_files = (
    saved_artifacts_df[
        "File Name"
    ]
    .tolist()
)


artifact_files_exist = bool(
    all(
        (
            artifact_dir
            / file_name
        ).exists()
        for file_name in expected_files
    )
)


artifact_files_nonempty = bool(
    all(
        (
            artifact_dir
            / file_name
        ).stat().st_size > 0
        for file_name in expected_files
    )
)


incorrect_directory_removed = bool(
    not old_artifact_dir.exists()
    or
    old_artifact_dir.resolve()
    == artifact_dir.resolve()
)


manifest_location_valid = bool(
    manifest_path.exists()
    and
    artifact_manifest[
        "artifact_directory"
    ]
    == str(
        artifact_dir.resolve()
    )
)


# ------------------------------------------------------------
# 6. Validation
# ------------------------------------------------------------

artifact_location_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Project-level artifact directory",

            "Requirement":
                "Geographic intelligence artifacts must reside "
                "under the project-level models directory.",

            "Observed Evidence":
                str(
                    artifact_dir.resolve()
                ),

            "Passed":
                artifact_dir.parent
                == project_root / "models",
        },
        {
            "Validation Area":
                "Saved artifact preservation",

            "Requirement":
                "All previously saved dataframe artifacts "
                "must remain available.",

            "Observed Evidence":
                f"{len(expected_files):,} "
                "artifact files verified",

            "Passed":
                artifact_files_exist,
        },
        {
            "Validation Area":
                "Saved artifact integrity",

            "Requirement":
                "All relocated artifact files must remain non-empty.",

            "Observed Evidence":
                f"{len(expected_files):,} "
                "non-empty artifact files verified",

            "Passed":
                artifact_files_nonempty,
        },
        {
            "Validation Area":
                "Incorrect nested directory removal",

            "Requirement":
                "The notebooks/models/geographic_intelligence "
                "directory must no longer hold the artifacts.",

            "Observed Evidence":
                f"Old artifact directory exists: "
                f"{old_artifact_dir.exists()}",

            "Passed":
                incorrect_directory_removed,
        },
        {
            "Validation Area":
                "Manifest location update",

            "Requirement":
                "The artifact manifest must reference "
                "the corrected project-level directory.",

            "Observed Evidence":
                artifact_manifest[
                    "artifact_directory"
                ],

            "Passed":
                manifest_location_valid,
        },
    ]
)


print()
print("Artifact location correction validation")
print("=" * 100)

display(
    artifact_location_validation_df
)


artifact_location_corrected = bool(
    artifact_location_validation_df[
        "Passed"
    ].all()
)


if not artifact_location_corrected:

    failed_checks = (
        artifact_location_validation_df
        .loc[
            ~artifact_location_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Artifact location correction failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "PMIP geographic intelligence artifact "
    "location corrected successfully."
)

print(
    f"Final artifact directory: "
    f"{artifact_dir.resolve()}"
)

print(
    f"Artifacts preserved: "
    f"{len(expected_files):,}"
)

print(
    f"Section 13.1 completion status: "
    f"{section_13_1_complete}"
)
```

```python
from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import numpy as np
import json
import sys
import platform

print("Saving PMIP geographic intelligence configuration and metadata")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_13_1_status = globals().get(
    "section_13_1_complete",
    False
)

print(
    f"Section 13.1 completion status: "
    f"{section_13_1_status}"
)

if not section_13_1_status:
    raise RuntimeError(
        "Section 13.2 requires Section 13.1 "
        "Save Geographic Intelligence Results to be complete."
    )


# ------------------------------------------------------------
# 2. Recover project and artifact paths
# ------------------------------------------------------------

current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
else:
    project_root = current_dir


artifact_dir = (
    project_root
    / "models"
    / "geographic_intelligence"
)

if not artifact_dir.exists():
    raise RuntimeError(
        "Section 13.2 could not recover the geographic "
        "intelligence artifact directory."
    )


print()
print("Validated geographic intelligence artifact location")
print("=" * 100)

print(
    f"Project root: "
    f"{project_root.resolve()}"
)

print(
    f"Artifact directory: "
    f"{artifact_dir.resolve()}"
)


# ------------------------------------------------------------
# 3. Recover existing artifact manifest
# ------------------------------------------------------------

manifest_path = (
    artifact_dir
    / "geographic_intelligence_artifact_manifest.json"
)

if not manifest_path.exists():
    raise RuntimeError(
        "Section 13.2 requires the Section 13.1 "
        "artifact manifest."
    )


with open(
    manifest_path,
    "r",
    encoding="utf-8"
) as manifest_file:

    artifact_manifest = json.load(
        manifest_file
    )


manifest_artifact_count = int(
    artifact_manifest.get(
        "artifact_count",
        0
    )
)


print()
print("Recovered artifact manifest")
print("=" * 100)

print(
    f"Manifest artifact count: "
    f"{manifest_artifact_count:,}"
)


# ------------------------------------------------------------
# 4. Define workflow completion metadata
# ------------------------------------------------------------

workflow_completion_status = {
    "section_7_artist_geographic_intelligence":
        bool(
            globals().get(
                "section_7_complete",
                False
            )
        ),

    "section_8_track_geographic_intelligence":
        bool(
            globals().get(
                "section_8_complete",
                False
            )
        ),

    "section_9_geographic_expansion_and_movement":
        bool(
            globals().get(
                "section_9_complete",
                False
            )
        ),

    "section_10_geographic_scoring_and_classification":
        bool(
            globals().get(
                "section_10_complete",
                False
            )
        ),

    "section_11_results_and_interpretation":
        bool(
            globals().get(
                "section_11_complete",
                False
            )
        ),

    "section_12_responsible_use_and_limitations":
        bool(
            globals().get(
                "section_12_complete",
                False
            )
        ),

    "section_13_1_artifact_saving":
        bool(
            section_13_1_status
        ),
}


# ------------------------------------------------------------
# 5. Recover validated population metadata
# ------------------------------------------------------------

country_population = None
artist_population = None
track_population = None
track_country_histories = None


if (
    "country_performance_df" in globals()
    and
    isinstance(
        globals()["country_performance_df"],
        pd.DataFrame
    )
):
    country_population = int(
        len(
            globals()["country_performance_df"]
        )
    )


if (
    "artist_findings_df" in globals()
    and
    isinstance(
        globals()["artist_findings_df"],
        pd.DataFrame
    )
):
    artist_population = int(
        len(
            globals()["artist_findings_df"]
        )
    )


if (
    "track_geographic_intelligence_df" in globals()
    and
    isinstance(
        globals()["track_geographic_intelligence_df"],
        pd.DataFrame
    )
):
    track_population = int(
        len(
            globals()["track_geographic_intelligence_df"]
        )
    )


if (
    "track_country_entry_df" in globals()
    and
    isinstance(
        globals()["track_country_entry_df"],
        pd.DataFrame
    )
):
    track_country_histories = int(
        len(
            globals()["track_country_entry_df"]
        )
    )


validated_populations = {
    "country_markets":
        country_population,

    "artist_representations":
        artist_population,

    "tracks":
        track_population,

    "track_country_histories":
        track_country_histories,
}


# ------------------------------------------------------------
# 6. Define geographic scoring configuration
# ------------------------------------------------------------

geographic_scoring_configuration = {
    "geographic_reach_score_range":
        [0.0, 1.0],

    "market_penetration_score_range":
        [0.0, 1.0],

    "geographic_concentration_score_range":
        [0.0, 1.0],

    "geographic_expansion_indicator_range":
        [0.0, 1.0],

    "geographic_intelligence_score_range":
        [0.0, 1.0],

    "higher_geographic_concentration_score_meaning":
        (
            "Greater geographic diversification "
            "and lower market dependency"
        ),

    "score_interpretation":
        (
            "Relative descriptive geographic intelligence "
            "within the validated PMIP dataset"
        ),

    "automatic_commercial_decision":
        False,

    "human_review_required":
        True,
}


# ------------------------------------------------------------
# 7. Define geographic intelligence dimensions
# ------------------------------------------------------------

geographic_dimensions = {
    "country_level":
        [
            "streaming strength",
            "chart strength",
            "market context",
            "data coverage",
            "temporal context",
        ],

    "artist_level":
        [
            "international reach",
            "market penetration",
            "geographic concentration",
            "local versus international distribution",
            "geographic performance profile",
        ],

    "track_level":
        [
            "country reach",
            "market penetration",
            "cross-market performance",
            "geographic concentration",
            "geographic performance profile",
        ],

    "movement":
        [
            "new-market entry",
            "geographic expansion",
            "geographic contraction",
            "cross-market momentum",
            "temporal validation",
        ],

    "summary_scoring":
        [
            "geographic reach score",
            "market penetration score",
            "geographic concentration score",
            "geographic expansion indicator",
            "geographic intelligence score",
        ],
}


# ------------------------------------------------------------
# 8. Define responsible-use metadata
# ------------------------------------------------------------

responsible_use_metadata = {
    "interpretation_boundary_count":
        int(
            len(
                globals().get(
                    "interpretation_boundaries_df",
                    []
                )
            )
        ),

    "data_coverage_limitation_count":
        int(
            len(
                globals().get(
                    "data_coverage_limitations_df",
                    []
                )
            )
        ),

    "cross_country_limitation_count":
        int(
            len(
                globals().get(
                    "cross_country_limitations_df",
                    []
                )
            )
        ),

    "human_interpretation_requirement_count":
        int(
            len(
                globals().get(
                    "human_interpretation_requirements_df",
                    []
                )
            )
        ),

    "human_review_role_count":
        int(
            len(
                globals().get(
                    "human_review_roles_df",
                    []
                )
            )
        ),

    "human_review_trigger_count":
        int(
            len(
                globals().get(
                    "human_review_triggers_df",
                    []
                )
            )
        ),

    "decision_support_only":
        True,

    "causal_inference_supported":
        False,

    "automatic_signing_decisions_supported":
        False,

    "automatic_investment_decisions_supported":
        False,
}


# ------------------------------------------------------------
# 9. Define temporal metadata
# ------------------------------------------------------------

dataset_end_date = None

if "dataset_end_date" in globals():

    candidate_date = globals()[
        "dataset_end_date"
    ]

    dataset_end_date = str(
        candidate_date
    )


temporal_metadata = {
    "dataset_end_date":
        dataset_end_date,

    "movement_interpretation":
        (
            "Observed historical geographic movement "
            "within the validated PMIP temporal window"
        ),

    "future_prediction":
        False,

    "temporal_confidence_required":
        True,
}


# ------------------------------------------------------------
# 10. Runtime environment metadata
# ------------------------------------------------------------

runtime_metadata = {
    "python_version":
        sys.version.split()[0],

    "python_implementation":
        platform.python_implementation(),

    "operating_system":
        platform.system(),

    "platform_release":
        platform.release(),

    "pandas_version":
        pd.__version__,

    "numpy_version":
        np.__version__,
}


# ------------------------------------------------------------
# 11. Build complete configuration metadata
# ------------------------------------------------------------

configuration_metadata = {
    "artifact_group":
        "PMIP Geographic Intelligence",

    "metadata_version":
        "1.0",

    "saved_at_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "project_root":
        str(
            project_root.resolve()
        ),

    "artifact_directory":
        str(
            artifact_dir.resolve()
        ),

    "artifact_manifest":
        manifest_path.name,

    "artifact_count":
        manifest_artifact_count,

    "validated_populations":
        validated_populations,

    "workflow_completion_status":
        workflow_completion_status,

    "geographic_scoring_configuration":
        geographic_scoring_configuration,

    "geographic_dimensions":
        geographic_dimensions,

    "responsible_use_metadata":
        responsible_use_metadata,

    "temporal_metadata":
        temporal_metadata,

    "runtime_metadata":
        runtime_metadata,
}


# ------------------------------------------------------------
# 12. Save configuration metadata JSON
# ------------------------------------------------------------

configuration_path = (
    artifact_dir
    / "geographic_intelligence_configuration.json"
)


with open(
    configuration_path,
    "w",
    encoding="utf-8"
) as configuration_file:

    json.dump(
        configuration_metadata,
        configuration_file,
        indent=4
    )


print()
print("Geographic intelligence configuration")
print("=" * 100)

print(
    f"Configuration saved: "
    f"{configuration_path.name}"
)


# ------------------------------------------------------------
# 13. Build flat metadata summary
# ------------------------------------------------------------

metadata_summary_df = pd.DataFrame(
    [
        {
            "Metadata Field":
                "Artifact group",

            "Observed Value":
                "PMIP Geographic Intelligence",
        },
        {
            "Metadata Field":
                "Metadata version",

            "Observed Value":
                "1.0",
        },
        {
            "Metadata Field":
                "Artifact count",

            "Observed Value":
                manifest_artifact_count,
        },
        {
            "Metadata Field":
                "Country markets",

            "Observed Value":
                country_population,
        },
        {
            "Metadata Field":
                "Artist representations",

            "Observed Value":
                artist_population,
        },
        {
            "Metadata Field":
                "Tracks",

            "Observed Value":
                track_population,
        },
        {
            "Metadata Field":
                "Track-country histories",

            "Observed Value":
                track_country_histories,
        },
        {
            "Metadata Field":
                "Interpretation boundaries",

            "Observed Value":
                responsible_use_metadata[
                    "interpretation_boundary_count"
                ],
        },
        {
            "Metadata Field":
                "Data-coverage limitations",

            "Observed Value":
                responsible_use_metadata[
                    "data_coverage_limitation_count"
                ],
        },
        {
            "Metadata Field":
                "Cross-country limitations",

            "Observed Value":
                responsible_use_metadata[
                    "cross_country_limitation_count"
                ],
        },
        {
            "Metadata Field":
                "Human interpretation requirements",

            "Observed Value":
                responsible_use_metadata[
                    "human_interpretation_requirement_count"
                ],
        },
        {
            "Metadata Field":
                "Python version",

            "Observed Value":
                runtime_metadata[
                    "python_version"
                ],
        },
        {
            "Metadata Field":
                "Pandas version",

            "Observed Value":
                runtime_metadata[
                    "pandas_version"
                ],
        },
        {
            "Metadata Field":
                "NumPy version",

            "Observed Value":
                runtime_metadata[
                    "numpy_version"
                ],
        },
    ]
)


print()
print("Configuration and metadata summary")
print("=" * 100)

display(
    metadata_summary_df
)


# ------------------------------------------------------------
# 14. Save metadata summary CSV
# ------------------------------------------------------------

metadata_summary_path = (
    artifact_dir
    / "geographic_intelligence_metadata_summary.csv"
)


metadata_summary_df.to_csv(
    metadata_summary_path,
    index=False
)


# ------------------------------------------------------------
# 15. Save artifact schema metadata
# ------------------------------------------------------------

schema_records = []


for artifact_record in artifact_manifest.get(
    "artifacts",
    []
):

    file_name = artifact_record.get(
        "File Name"
    )

    artifact_name = artifact_record.get(
        "Artifact"
    )

    artifact_file_path = (
        artifact_dir
        / file_name
    )

    if not artifact_file_path.exists():
        raise RuntimeError(
            f"Section 13.2 could not inspect schema for "
            f"missing artifact: {file_name}"
        )

    sample_df = pd.read_csv(
        artifact_file_path,
        nrows=50
    )

    for column_name, dtype in sample_df.dtypes.items():

        schema_records.append(
            {
                "Artifact":
                    artifact_name,

                "File Name":
                    file_name,

                "Column":
                    column_name,

                "Reloaded Dtype":
                    str(
                        dtype
                    ),
            }
        )


artifact_schema_metadata_df = pd.DataFrame(
    schema_records
)


schema_metadata_path = (
    artifact_dir
    / "geographic_intelligence_schema_metadata.csv"
)


artifact_schema_metadata_df.to_csv(
    schema_metadata_path,
    index=False
)


print()
print("Artifact schema metadata")
print("=" * 100)

print(
    f"Schema records saved: "
    f"{len(artifact_schema_metadata_df):,}"
)

print(
    f"Schema metadata file: "
    f"{schema_metadata_path.name}"
)


# ------------------------------------------------------------
# 16. Diagnostics
# ------------------------------------------------------------

configuration_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Configuration JSON files saved",

            "Observed Value":
                int(
                    configuration_path.exists()
                ),
        },
        {
            "Diagnostic":
                "Metadata summary files saved",

            "Observed Value":
                int(
                    metadata_summary_path.exists()
                ),
        },
        {
            "Diagnostic":
                "Schema metadata files saved",

            "Observed Value":
                int(
                    schema_metadata_path.exists()
                ),
        },
        {
            "Diagnostic":
                "Artifact schemas documented",

            "Observed Value":
                int(
                    artifact_schema_metadata_df[
                        "Artifact"
                    ].nunique()
                ),
        },
        {
            "Diagnostic":
                "Schema column records documented",

            "Observed Value":
                int(
                    len(
                        artifact_schema_metadata_df
                    )
                ),
        },
        {
            "Diagnostic":
                "Workflow status records documented",

            "Observed Value":
                int(
                    len(
                        workflow_completion_status
                    )
                ),
        },
    ]
)


print()
print("Configuration and metadata diagnostics")
print("=" * 100)

display(
    configuration_diagnostics_df
)


# ------------------------------------------------------------
# 17. Validation
# ------------------------------------------------------------

configuration_file_valid = bool(
    configuration_path.exists()
    and
    configuration_path.stat().st_size > 0
)


metadata_summary_valid = bool(
    metadata_summary_path.exists()
    and
    metadata_summary_path.stat().st_size > 0
)


schema_metadata_valid = bool(
    schema_metadata_path.exists()
    and
    schema_metadata_path.stat().st_size > 0
)


artifact_count_consistent = bool(
    manifest_artifact_count
    == int(
        artifact_schema_metadata_df[
            "Artifact"
        ].nunique()
    )
)


required_population_metadata_present = bool(
    all(
        value is not None
        for value in [
            country_population,
            artist_population,
            track_population,
            track_country_histories,
        ]
    )
)


responsible_use_metadata_valid = bool(
    responsible_use_metadata[
        "interpretation_boundary_count"
    ] > 0
    and
    responsible_use_metadata[
        "data_coverage_limitation_count"
    ] > 0
    and
    responsible_use_metadata[
        "cross_country_limitation_count"
    ] > 0
    and
    responsible_use_metadata[
        "human_interpretation_requirement_count"
    ] > 0
)


configuration_path_valid = bool(
    configuration_metadata[
        "artifact_directory"
    ]
    == str(
        artifact_dir.resolve()
    )
)


section_13_2_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13.1 completion",

            "Requirement":
                "Geographic intelligence results must "
                "already be saved.",

            "Observed Evidence":
                f"Section 13.1 completion status: "
                f"{section_13_1_status}",

            "Passed":
                section_13_1_status,
        },
        {
            "Validation Area":
                "Configuration metadata validity",

            "Requirement":
                "Configuration metadata JSON must exist "
                "and contain non-zero content.",

            "Observed Evidence":
                configuration_path.name,

            "Passed":
                configuration_file_valid,
        },
        {
            "Validation Area":
                "Metadata summary validity",

            "Requirement":
                "Metadata summary CSV must exist "
                "and contain non-zero content.",

            "Observed Evidence":
                metadata_summary_path.name,

            "Passed":
                metadata_summary_valid,
        },
        {
            "Validation Area":
                "Artifact schema metadata validity",

            "Requirement":
                "Artifact schema metadata must be saved "
                "for the saved artifact population.",

            "Observed Evidence":
                (
                    f"{artifact_schema_metadata_df['Artifact'].nunique():,} "
                    "artifact schemas documented"
                ),

            "Passed":
                schema_metadata_valid,
        },
        {
            "Validation Area":
                "Artifact count consistency",

            "Requirement":
                "Manifest and schema metadata must represent "
                "the same artifact population.",

            "Observed Evidence":
                (
                    f"{manifest_artifact_count:,} manifest artifacts / "
                    f"{artifact_schema_metadata_df['Artifact'].nunique():,} "
                    "schema artifacts"
                ),

            "Passed":
                artifact_count_consistent,
        },
        {
            "Validation Area":
                "Population metadata completeness",

            "Requirement":
                "Country, artist, track and track-country "
                "population metadata must be preserved.",

            "Observed Evidence":
                (
                    f"{country_population} countries / "
                    f"{artist_population} artists / "
                    f"{track_population} tracks / "
                    f"{track_country_histories} track-country histories"
                ),

            "Passed":
                required_population_metadata_present,
        },
        {
            "Validation Area":
                "Responsible-use metadata preservation",

            "Requirement":
                "Responsible-use boundaries and limitations "
                "must remain represented in metadata.",

            "Observed Evidence":
                (
                    f"{responsible_use_metadata['interpretation_boundary_count']} "
                    "boundaries / "
                    f"{responsible_use_metadata['data_coverage_limitation_count']} "
                    "coverage limits / "
                    f"{responsible_use_metadata['cross_country_limitation_count']} "
                    "country limits / "
                    f"{responsible_use_metadata['human_interpretation_requirement_count']} "
                    "human-review requirements"
                ),

            "Passed":
                responsible_use_metadata_valid,
        },
        {
            "Validation Area":
                "Artifact directory metadata validity",

            "Requirement":
                "Configuration metadata must point to the "
                "project-level geographic intelligence directory.",

            "Observed Evidence":
                configuration_metadata[
                    "artifact_directory"
                ],

            "Passed":
                configuration_path_valid,
        },
    ]
)


print()
print("Configuration and metadata validation")
print("=" * 100)

display(
    section_13_2_validation_df
)


# ------------------------------------------------------------
# 18. Completion status
# ------------------------------------------------------------

section_13_2_complete = bool(
    section_13_2_validation_df[
        "Passed"
    ].all()
)


if not section_13_2_complete:

    failed_checks = (
        section_13_2_validation_df
        .loc[
            ~section_13_2_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 13.2 Save Configuration and Metadata "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 13.2 Save Configuration and Metadata "
    "validation checks passed."
)

print(
    f"Section 13.2 completion status: "
    f"{section_13_2_complete}"
)

print(
    "PMIP geographic intelligence configuration, metadata "
    "and artifact schema information have been saved."
)

print(
    "The workflow is ready for "
    "Section 13.3 Reload and Verify Saved Artifacts."
)
```

### 13.3 Reload and Verify Saved Artifacts

#### Purpose

Reload the saved PMIP geographic intelligence artifacts from disk and verify that they can be recovered correctly without relying on the live notebook variables.

This subsection checks that the saved CSV artifacts, artifact manifest, configuration metadata, metadata summary, and schema metadata remain readable and structurally consistent after saving.

The goal is to confirm that the geographic intelligence outputs are genuinely reusable for future PMIP software integration, reporting, auditing, and reproducibility.

```python
from pathlib import Path
import pandas as pd
import json
import numpy as np

print("Reloading and verifying PMIP geographic intelligence artifacts")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous subsection completion
# ------------------------------------------------------------

section_13_2_status = globals().get(
    "section_13_2_complete",
    False
)

print(
    f"Section 13.2 completion status: "
    f"{section_13_2_status}"
)

if not section_13_2_status:
    raise RuntimeError(
        "Section 13.3 requires Section 13.2 "
        "Save Configuration and Metadata to be complete."
    )


# ------------------------------------------------------------
# 2. Recover project and artifact paths
# ------------------------------------------------------------

current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
else:
    project_root = current_dir


artifact_dir = (
    project_root
    / "models"
    / "geographic_intelligence"
)


if not artifact_dir.exists():
    raise RuntimeError(
        "Section 13.3 could not recover the "
        "geographic intelligence artifact directory."
    )


print()
print("Validated geographic intelligence artifact directory")
print("=" * 100)

print(
    f"Artifact directory: "
    f"{artifact_dir.resolve()}"
)


# ------------------------------------------------------------
# 3. Recover saved control files
# ------------------------------------------------------------

manifest_path = (
    artifact_dir
    / "geographic_intelligence_artifact_manifest.json"
)

configuration_path = (
    artifact_dir
    / "geographic_intelligence_configuration.json"
)

metadata_summary_path = (
    artifact_dir
    / "geographic_intelligence_metadata_summary.csv"
)

schema_metadata_path = (
    artifact_dir
    / "geographic_intelligence_schema_metadata.csv"
)


required_control_files = {
    "artifact_manifest":
        manifest_path,

    "configuration_metadata":
        configuration_path,

    "metadata_summary":
        metadata_summary_path,

    "schema_metadata":
        schema_metadata_path,
}


missing_control_files = [
    name
    for name, path in required_control_files.items()
    if not path.exists()
]


if missing_control_files:
    raise RuntimeError(
        "Section 13.3 could not recover required control files: "
        + ", ".join(
            missing_control_files
        )
    )


# ------------------------------------------------------------
# 4. Reload manifest and configuration
# ------------------------------------------------------------

with open(
    manifest_path,
    "r",
    encoding="utf-8"
) as manifest_file:

    reloaded_manifest = json.load(
        manifest_file
    )


with open(
    configuration_path,
    "r",
    encoding="utf-8"
) as configuration_file:

    reloaded_configuration = json.load(
        configuration_file
    )


reloaded_metadata_summary_df = pd.read_csv(
    metadata_summary_path
)

reloaded_schema_metadata_df = pd.read_csv(
    schema_metadata_path
)


print()
print("Reloaded control metadata")
print("=" * 100)

print(
    f"Manifest artifacts: "
    f"{reloaded_manifest.get('artifact_count', 0):,}"
)

print(
    f"Configuration artifact count: "
    f"{reloaded_configuration.get('artifact_count', 0):,}"
)

print(
    f"Metadata summary records: "
    f"{len(reloaded_metadata_summary_df):,}"
)

print(
    f"Schema metadata records: "
    f"{len(reloaded_schema_metadata_df):,}"
)


# ------------------------------------------------------------
# 5. Reload all saved dataframe artifacts
# ------------------------------------------------------------

reloaded_artifacts = {}
reload_records = []


manifest_artifacts = reloaded_manifest.get(
    "artifacts",
    []
)


for artifact_record in manifest_artifacts:

    artifact_name = artifact_record[
        "Artifact"
    ]

    file_name = artifact_record[
        "File Name"
    ]

    expected_rows = int(
        artifact_record[
            "Rows"
        ]
    )

    expected_columns = int(
        artifact_record[
            "Columns"
        ]
    )

    artifact_path = (
        artifact_dir
        / file_name
    )


    if not artifact_path.exists():

        reload_records.append(
            {
                "Artifact":
                    artifact_name,

                "File Name":
                    file_name,

                "Expected Rows":
                    expected_rows,

                "Reloaded Rows":
                    None,

                "Expected Columns":
                    expected_columns,

                "Reloaded Columns":
                    None,

                "File Exists":
                    False,

                "Shape Match":
                    False,

                "Reloaded":
                    False,
            }
        )

        continue


    try:
        reloaded_df = pd.read_csv(
            artifact_path
        )

        reloaded_artifacts[
            artifact_name
        ] = reloaded_df

        reload_records.append(
            {
                "Artifact":
                    artifact_name,

                "File Name":
                    file_name,

                "Expected Rows":
                    expected_rows,

                "Reloaded Rows":
                    int(
                        reloaded_df.shape[0]
                    ),

                "Expected Columns":
                    expected_columns,

                "Reloaded Columns":
                    int(
                        reloaded_df.shape[1]
                    ),

                "File Exists":
                    True,

                "Shape Match":
                    bool(
                        reloaded_df.shape
                        ==
                        (
                            expected_rows,
                            expected_columns,
                        )
                    ),

                "Reloaded":
                    True,
            }
        )

    except Exception:

        reload_records.append(
            {
                "Artifact":
                    artifact_name,

                "File Name":
                    file_name,

                "Expected Rows":
                    expected_rows,

                "Reloaded Rows":
                    None,

                "Expected Columns":
                    expected_columns,

                "Reloaded Columns":
                    None,

                "File Exists":
                    True,

                "Shape Match":
                    False,

                "Reloaded":
                    False,
            }
        )


artifact_reload_verification_df = pd.DataFrame(
    reload_records
)


print()
print("Artifact reload verification")
print("=" * 100)

display(
    artifact_reload_verification_df
)


# ------------------------------------------------------------
# 6. Validate artifact schema structure
# ------------------------------------------------------------

schema_validation_records = []


for artifact_name, reloaded_df in reloaded_artifacts.items():

    expected_schema = (
        reloaded_schema_metadata_df
        .loc[
            reloaded_schema_metadata_df[
                "Artifact"
            ]
            ==
            artifact_name
        ]
        .copy()
    )


    expected_columns = (
        expected_schema[
            "Column"
        ]
        .astype(
            str
        )
        .tolist()
    )


    reloaded_columns = (
        reloaded_df.columns
        .astype(
            str
        )
        .tolist()
    )


    missing_columns = [
        column
        for column in expected_columns
        if column not in reloaded_columns
    ]


    unexpected_columns = [
        column
        for column in reloaded_columns
        if column not in expected_columns
    ]


    schema_validation_records.append(
        {
            "Artifact":
                artifact_name,

            "Expected Columns":
                len(
                    expected_columns
                ),

            "Reloaded Columns":
                len(
                    reloaded_columns
                ),

            "Missing Columns":
                len(
                    missing_columns
                ),

            "Unexpected Columns":
                len(
                    unexpected_columns
                ),

            "Column Set Match":
                bool(
                    not missing_columns
                    and
                    not unexpected_columns
                ),
        }
    )


artifact_schema_verification_df = pd.DataFrame(
    schema_validation_records
)


print()
print("Artifact schema verification")
print("=" * 100)

display(
    artifact_schema_verification_df
)


# ------------------------------------------------------------
# 7. Verify required core artifacts
# ------------------------------------------------------------

required_core_artifacts = [
    "country_performance",
    "country_findings",
    "artist_geographic_profile",
    "artist_findings",
    "track_geographic_intelligence",
    "track_temporal_validation",
    "high_priority_insights",
    "interpretation_boundaries",
    "data_coverage_limitations",
    "cross_country_limitations",
    "human_interpretation_requirements",
]


required_core_reload_status = {
    artifact_name:
        artifact_name in reloaded_artifacts

    for artifact_name
    in required_core_artifacts
}


required_core_reload_df = pd.DataFrame(
    [
        {
            "Required Core Artifact":
                artifact_name,

            "Reloaded":
                reloaded
        }

        for artifact_name, reloaded
        in required_core_reload_status.items()
    ]
)


print()
print("Required core artifact verification")
print("=" * 100)

display(
    required_core_reload_df
)


# ------------------------------------------------------------
# 8. Verify important analytical populations
# ------------------------------------------------------------

population_checks = []


population_expectations = {
    "country_performance":
        reloaded_configuration[
            "validated_populations"
        ][
            "country_markets"
        ],

    "artist_findings":
        reloaded_configuration[
            "validated_populations"
        ][
            "artist_representations"
        ],

    "track_geographic_intelligence":
        reloaded_configuration[
            "validated_populations"
        ][
            "tracks"
        ],

    "track_country_entry":
        reloaded_configuration[
            "validated_populations"
        ][
            "track_country_histories"
        ],
}


for artifact_name, expected_population in population_expectations.items():

    if artifact_name not in reloaded_artifacts:

        population_checks.append(
            {
                "Artifact":
                    artifact_name,

                "Expected Population":
                    expected_population,

                "Reloaded Population":
                    None,

                "Population Match":
                    False,
            }
        )

        continue


    observed_population = int(
        len(
            reloaded_artifacts[
                artifact_name
            ]
        )
    )


    population_checks.append(
        {
            "Artifact":
                artifact_name,

            "Expected Population":
                expected_population,

            "Reloaded Population":
                observed_population,

            "Population Match":
                bool(
                    observed_population
                    ==
                    expected_population
                ),
        }
    )


population_reload_verification_df = pd.DataFrame(
    population_checks
)


print()
print("Population reload verification")
print("=" * 100)

display(
    population_reload_verification_df
)


# ------------------------------------------------------------
# 9. Verify responsible-use artifact preservation
# ------------------------------------------------------------

responsible_use_artifacts = [
    "interpretation_boundaries",
    "data_coverage_limitations",
    "coverage_safeguards",
    "cross_country_limitations",
    "cross_country_safeguards",
    "human_interpretation_requirements",
    "human_review_roles",
    "human_review_triggers",
    "responsible_decision_policy",
]


responsible_use_reload_df = pd.DataFrame(
    [
        {
            "Responsible Use Artifact":
                artifact_name,

            "Reloaded":
                artifact_name
                in reloaded_artifacts,
        }

        for artifact_name
        in responsible_use_artifacts
    ]
)


print()
print("Responsible-use artifact verification")
print("=" * 100)

display(
    responsible_use_reload_df
)


# ------------------------------------------------------------
# 10. Verify configuration consistency
# ------------------------------------------------------------

manifest_count = int(
    reloaded_manifest[
        "artifact_count"
    ]
)

configuration_count = int(
    reloaded_configuration[
        "artifact_count"
    ]
)

reloaded_count = int(
    len(
        reloaded_artifacts
    )
)

schema_artifact_count = int(
    reloaded_schema_metadata_df[
        "Artifact"
    ].nunique()
)


configuration_consistency_df = pd.DataFrame(
    [
        {
            "Consistency Check":
                "Manifest versus configuration artifact count",

            "Expected":
                manifest_count,

            "Observed":
                configuration_count,

            "Passed":
                bool(
                    manifest_count
                    ==
                    configuration_count
                ),
        },
        {
            "Consistency Check":
                "Manifest versus reloaded artifact count",

            "Expected":
                manifest_count,

            "Observed":
                reloaded_count,

            "Passed":
                bool(
                    manifest_count
                    ==
                    reloaded_count
                ),
        },
        {
            "Consistency Check":
                "Manifest versus schema artifact count",

            "Expected":
                manifest_count,

            "Observed":
                schema_artifact_count,

            "Passed":
                bool(
                    manifest_count
                    ==
                    schema_artifact_count
                ),
        },
        {
            "Consistency Check":
                "Artifact directory path",

            "Expected":
                str(
                    artifact_dir.resolve()
                ),

            "Observed":
                reloaded_configuration[
                    "artifact_directory"
                ],

            "Passed":
                bool(
                    reloaded_configuration[
                        "artifact_directory"
                    ]
                    ==
                    str(
                        artifact_dir.resolve()
                    )
                ),
        },
    ]
)


print()
print("Configuration consistency verification")
print("=" * 100)

display(
    configuration_consistency_df
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

all_files_reloaded = bool(
    artifact_reload_verification_df[
        "Reloaded"
    ].all()
)


all_shapes_match = bool(
    artifact_reload_verification_df[
        "Shape Match"
    ].all()
)


all_schemas_match = bool(
    artifact_schema_verification_df[
        "Column Set Match"
    ].all()
)


all_required_core_reloaded = bool(
    required_core_reload_df[
        "Reloaded"
    ].all()
)


all_population_checks_passed = bool(
    population_reload_verification_df[
        "Population Match"
    ].all()
)


all_responsible_use_reloaded = bool(
    responsible_use_reload_df[
        "Reloaded"
    ].all()
)


all_configuration_checks_passed = bool(
    configuration_consistency_df[
        "Passed"
    ].all()
)


reload_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Artifacts expected",

            "Observed Value":
                manifest_count,
        },
        {
            "Diagnostic":
                "Artifacts reloaded",

            "Observed Value":
                reloaded_count,
        },
        {
            "Diagnostic":
                "Artifacts with matching shapes",

            "Observed Value":
                int(
                    artifact_reload_verification_df[
                        "Shape Match"
                    ].sum()
                ),
        },
        {
            "Diagnostic":
                "Artifacts with matching schemas",

            "Observed Value":
                int(
                    artifact_schema_verification_df[
                        "Column Set Match"
                    ].sum()
                ),
        },
        {
            "Diagnostic":
                "Required core artifacts reloaded",

            "Observed Value":
                int(
                    required_core_reload_df[
                        "Reloaded"
                    ].sum()
                ),
        },
        {
            "Diagnostic":
                "Responsible-use artifacts reloaded",

            "Observed Value":
                int(
                    responsible_use_reload_df[
                        "Reloaded"
                    ].sum()
                ),
        },
        {
            "Diagnostic":
                "Population checks passed",

            "Observed Value":
                int(
                    population_reload_verification_df[
                        "Population Match"
                    ].sum()
                ),
        },
    ]
)


print()
print("Reload verification diagnostics")
print("=" * 100)

display(
    reload_diagnostics_df
)


# ------------------------------------------------------------
# 12. Final validation
# ------------------------------------------------------------

section_13_3_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13.2 completion",

            "Requirement":
                "Configuration and metadata must "
                "already be saved.",

            "Observed Evidence":
                f"Section 13.2 completion status: "
                f"{section_13_2_status}",

            "Passed":
                section_13_2_status,
        },
        {
            "Validation Area":
                "Artifact reload completeness",

            "Requirement":
                "Every manifest artifact must reload "
                "successfully from disk.",

            "Observed Evidence":
                f"{reloaded_count:,} of "
                f"{manifest_count:,} artifacts reloaded",

            "Passed":
                all_files_reloaded,
        },
        {
            "Validation Area":
                "Artifact shape preservation",

            "Requirement":
                "Reloaded artifact shapes must match "
                "the saved manifest dimensions.",

            "Observed Evidence":
                (
                    f"{artifact_reload_verification_df['Shape Match'].sum():,} "
                    f"of {manifest_count:,} shapes matched"
                ),

            "Passed":
                all_shapes_match,
        },
        {
            "Validation Area":
                "Artifact schema preservation",

            "Requirement":
                "Reloaded artifacts must preserve "
                "the documented column sets.",

            "Observed Evidence":
                (
                    f"{artifact_schema_verification_df['Column Set Match'].sum():,} "
                    f"of {schema_artifact_count:,} schemas matched"
                ),

            "Passed":
                all_schemas_match,
        },
        {
            "Validation Area":
                "Required core artifact recovery",

            "Requirement":
                "All required core geographic intelligence "
                "artifacts must reload successfully.",

            "Observed Evidence":
                (
                    f"{required_core_reload_df['Reloaded'].sum():,} "
                    f"of {len(required_core_artifacts):,} "
                    "required core artifacts reloaded"
                ),

            "Passed":
                all_required_core_reloaded,
        },
        {
            "Validation Area":
                "Population preservation",

            "Requirement":
                "Country, artist, track and track-country "
                "populations must match saved metadata.",

            "Observed Evidence":
                (
                    f"{population_reload_verification_df['Population Match'].sum():,} "
                    f"of {len(population_reload_verification_df):,} "
                    "population checks passed"
                ),

            "Passed":
                all_population_checks_passed,
        },
        {
            "Validation Area":
                "Responsible-use artifact preservation",

            "Requirement":
                "Responsible-use outputs must remain "
                "recoverable after reload.",

            "Observed Evidence":
                (
                    f"{responsible_use_reload_df['Reloaded'].sum():,} "
                    f"of {len(responsible_use_reload_df):,} "
                    "responsible-use artifacts reloaded"
                ),

            "Passed":
                all_responsible_use_reloaded,
        },
        {
            "Validation Area":
                "Configuration consistency",

            "Requirement":
                "Manifest, configuration, schema and "
                "artifact directory metadata must agree.",

            "Observed Evidence":
                (
                    f"{configuration_consistency_df['Passed'].sum():,} "
                    f"of {len(configuration_consistency_df):,} "
                    "configuration checks passed"
                ),

            "Passed":
                all_configuration_checks_passed,
        },
    ]
)


print()
print("Reloaded artifact validation")
print("=" * 100)

display(
    section_13_3_validation_df
)


# ------------------------------------------------------------
# 13. Completion status
# ------------------------------------------------------------

section_13_3_complete = bool(
    section_13_3_validation_df[
        "Passed"
    ].all()
)


section_13_complete = bool(
    globals().get(
        "section_13_1_complete",
        False
    )
    and
    globals().get(
        "section_13_2_complete",
        False
    )
    and
    section_13_3_complete
)


if not section_13_3_complete:

    failed_checks = (
        section_13_3_validation_df
        .loc[
            ~section_13_3_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 13.3 Reload and Verify Saved Artifacts "
        "validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 13.3 Reload and Verify Saved Artifacts "
    "validation checks passed."
)

print(
    f"Section 13.3 completion status: "
    f"{section_13_3_complete}"
)

print(
    f"Section 13 overall completion status: "
    f"{section_13_complete}"
)

print(
    f"Verified geographic intelligence artifacts: "
    f"{reloaded_count:,}"
)

print(
    "PMIP geographic intelligence artifacts can be "
    "successfully recovered from persistent storage."
)

print(
    "Artifact Saving and Reload Verification is complete."
)
```

### 14.1 Key Findings

#### Purpose

Summarise the strongest validated findings produced by the PMIP geographic intelligence workflow.

This subsection brings together evidence from country-level, artist-level, track-level, geographic movement, scoring, comparison, and responsible-use analysis.

The goal is to produce a concise final set of findings that explains what the geographic intelligence workflow has learned about observed market reach, penetration, diversification, expansion, and geographic performance within the PMIP dataset.

```python
import pandas as pd
import numpy as np

print("Summarising final PMIP geographic intelligence key findings")
print("=" * 100)

# ------------------------------------------------------------
# 1. Validate previous section completion
# ------------------------------------------------------------

section_13_status = globals().get(
    "section_13_complete",
    False
)

print(
    f"Section 13 overall completion status: "
    f"{section_13_status}"
)

if not section_13_status:
    raise RuntimeError(
        "Section 14.1 requires Section 13 "
        "Artifact Saving and Reload Verification to be complete."
    )


# ------------------------------------------------------------
# 2. Recover required validated sources
# ------------------------------------------------------------

required_sources = {
    "country_findings":
        "country_findings_df",

    "artist_findings":
        "artist_findings_df",

    "track_findings":
        "track_geographic_intelligence_df",

    "track_expansion":
        "track_geographic_expansion_indicator_df",

    "track_momentum":
        "track_cross_market_momentum_df",

    "high_priority_insights":
        "high_priority_insights_df",
}


final_finding_sources = {}


for source_key, variable_name in required_sources.items():

    if variable_name not in globals():
        raise RuntimeError(
            f"Section 14.1 could not recover required source: "
            f"{variable_name}"
        )

    candidate = globals()[variable_name]

    if not isinstance(
        candidate,
        pd.DataFrame
    ):
        raise RuntimeError(
            f"{variable_name} is not a valid dataframe."
        )

    final_finding_sources[
        source_key
    ] = candidate.copy()


country_findings_final_df = (
    final_finding_sources[
        "country_findings"
    ]
)

artist_findings_final_df = (
    final_finding_sources[
        "artist_findings"
    ]
)

track_findings_final_df = (
    final_finding_sources[
        "track_findings"
    ]
)

track_expansion_final_df = (
    final_finding_sources[
        "track_expansion"
    ]
)

track_momentum_final_df = (
    final_finding_sources[
        "track_momentum"
    ]
)

priority_insights_final_df = (
    final_finding_sources[
        "high_priority_insights"
    ]
)


print()
print("Validated final findings sources")
print("=" * 100)

print(
    f"Country markets: "
    f"{len(country_findings_final_df):,}"
)

print(
    f"Artist representations: "
    f"{len(artist_findings_final_df):,}"
)

print(
    f"Tracks: "
    f"{len(track_findings_final_df):,}"
)

print(
    f"Expansion records: "
    f"{len(track_expansion_final_df):,}"
)

print(
    f"Momentum records: "
    f"{len(track_momentum_final_df):,}"
)

print(
    f"High-priority insights: "
    f"{len(priority_insights_final_df):,}"
)


# ------------------------------------------------------------
# 3. Helper for safe column recovery
# ------------------------------------------------------------

def first_existing_column(
    dataframe,
    candidates,
    required=True
):

    for column in candidates:

        if column in dataframe.columns:
            return column

    if required:
        raise RuntimeError(
            "Could not recover any required column from: "
            + ", ".join(
                candidates
            )
        )

    return None


# ------------------------------------------------------------
# 4. Recover important country indicators
# ------------------------------------------------------------

country_class_col = first_existing_column(
    country_findings_final_df,
    [
        "country_findings_class",
        "Country Findings Class",
    ]
)

country_index_col = first_existing_column(
    country_findings_final_df,
    [
        "country_findings_index",
        "Country Findings Index",
    ]
)


high_priority_country_count = int(
    country_findings_final_df[
        country_class_col
    ]
    .isin(
        [
            "High-priority country market",
            "Strong country market",
        ]
    )
    .sum()
)


high_priority_country_share = float(
    high_priority_country_count
    /
    len(
        country_findings_final_df
    )
    *
    100
)


median_country_findings_index = float(
    pd.to_numeric(
        country_findings_final_df[
            country_index_col
        ],
        errors="coerce"
    )
    .median()
)


# ------------------------------------------------------------
# 5. Recover important artist indicators
# ------------------------------------------------------------

artist_class_col = first_existing_column(
    artist_findings_final_df,
    [
        "artist_findings_class",
        "Artist Findings Class",
    ]
)

artist_index_col = first_existing_column(
    artist_findings_final_df,
    [
        "artist_findings_index",
        "Artist Findings Index",
    ]
)


strong_artist_count = int(
    artist_findings_final_df[
        artist_class_col
    ]
    .isin(
        [
            "Exceptional artist geographic strength",
            "Strong artist geographic strength",
        ]
    )
    .sum()
)


strong_artist_share = float(
    strong_artist_count
    /
    len(
        artist_findings_final_df
    )
    *
    100
)


median_artist_findings_index = float(
    pd.to_numeric(
        artist_findings_final_df[
            artist_index_col
        ],
        errors="coerce"
    )
    .median()
)


# ------------------------------------------------------------
# 6. Recover important track indicators
# ------------------------------------------------------------

track_intelligence_class_col = first_existing_column(
    track_findings_final_df,
    [
        "geographic_intelligence_class",
        "Geographic Intelligence Class",
    ]
)

track_intelligence_score_col = first_existing_column(
    track_findings_final_df,
    [
        "geographic_intelligence_score",
        "Geographic Intelligence Score",
    ]
)

track_profile_col = first_existing_column(
    track_findings_final_df,
    [
        "geographic_performance_profile",
        "Geographic Performance Profile",
    ]
)

strong_track_count = int(
    track_findings_final_df[
        track_intelligence_class_col
    ]
    .isin(
        [
            "Exceptional geographic intelligence",
            "Strong geographic intelligence",
        ]
    )
    .sum()
)


strong_track_share = float(
    strong_track_count
    /
    len(
        track_findings_final_df
    )
    *
    100
)


limited_footprint_count = int(
    (
        track_findings_final_df[
            track_profile_col
        ]
        ==
        "Limited geographic footprint"
    )
    .sum()
)


limited_footprint_share = float(
    limited_footprint_count
    /
    len(
        track_findings_final_df
    )
    *
    100
)


median_track_intelligence_score = float(
    pd.to_numeric(
        track_findings_final_df[
            track_intelligence_score_col
        ],
        errors="coerce"
    )
    .median()
)


# ------------------------------------------------------------
# 7. Recover geographic dimension findings
# ------------------------------------------------------------

strongest_dimension_col = first_existing_column(
    track_findings_final_df,
    [
        "strongest_geographic_dimension",
        "Strongest Geographic Dimension",
    ]
)

weakest_dimension_col = first_existing_column(
    track_findings_final_df,
    [
        "weakest_geographic_dimension",
        "Weakest Geographic Dimension",
    ]
)


strongest_dimension_distribution = (
    track_findings_final_df[
        strongest_dimension_col
    ]
    .value_counts(
        dropna=False
    )
)


weakest_dimension_distribution = (
    track_findings_final_df[
        weakest_dimension_col
    ]
    .value_counts(
        dropna=False
    )
)


dominant_strongest_dimension = str(
    strongest_dimension_distribution.index[
        0
    ]
)


dominant_strongest_dimension_count = int(
    strongest_dimension_distribution.iloc[
        0
    ]
)


dominant_strongest_dimension_share = float(
    dominant_strongest_dimension_count
    /
    len(
        track_findings_final_df
    )
    *
    100
)


dominant_weakest_dimension = str(
    weakest_dimension_distribution.index[
        0
    ]
)


dominant_weakest_dimension_count = int(
    weakest_dimension_distribution.iloc[
        0
    ]
)


dominant_weakest_dimension_share = float(
    dominant_weakest_dimension_count
    /
    len(
        track_findings_final_df
    )
    *
    100
)


# ------------------------------------------------------------
# 8. Recover expansion findings
# ------------------------------------------------------------

expansion_class_col = first_existing_column(
    track_expansion_final_df,
    [
        "geographic_expansion_indicator_class",
        "Geographic Expansion Indicator Class",
    ]
)


strong_expansion_count = int(
    track_expansion_final_df[
        expansion_class_col
    ]
    .isin(
        [
            "Exceptional geographic expansion",
            "Strong geographic expansion",
        ]
    )
    .sum()
)


strong_expansion_share = float(
    strong_expansion_count
    /
    len(
        track_expansion_final_df
    )
    *
    100
)


# ------------------------------------------------------------
# 9. Recover momentum findings
# ------------------------------------------------------------

momentum_class_col = first_existing_column(
    track_momentum_final_df,
    [
        "cross_market_momentum_class",
        "Cross-Market Momentum Class",
    ]
)


positive_momentum_count = int(
    track_momentum_final_df[
        momentum_class_col
    ]
    .isin(
        [
            "Strong positive momentum",
            "Positive momentum",
        ]
    )
    .sum()
)


negative_momentum_count = int(
    track_momentum_final_df[
        momentum_class_col
    ]
    .isin(
        [
            "Negative momentum",
            "Strong negative momentum",
        ]
    )
    .sum()
)


positive_momentum_share = float(
    positive_momentum_count
    /
    len(
        track_momentum_final_df
    )
    *
    100
)


negative_momentum_share = float(
    negative_momentum_count
    /
    len(
        track_momentum_final_df
    )
    *
    100
)


# ------------------------------------------------------------
# 10. Build final key findings table
# ------------------------------------------------------------

final_geographic_key_findings_df = pd.DataFrame(
    [
        {
            "Priority":
                1,

            "Key Finding":
                (
                    "Country-level geographic strength is more widely "
                    "distributed than strong artist- or track-level "
                    "geographic intelligence."
                ),

            "Evidence":
                (
                    f"{high_priority_country_count:,} of "
                    f"{len(country_findings_final_df):,} country markets "
                    f"({high_priority_country_share:.2f}%) are high-priority "
                    "or strong, compared with only "
                    f"{strong_artist_count:,} strong/exceptional artist "
                    f"representations ({strong_artist_share:.2f}%) and "
                    f"{strong_track_count:,} strong/exceptional tracks "
                    f"({strong_track_share:.2f}%)."
                ),

            "PMIP Meaning":
                (
                    "Strong market opportunity can exist even when only "
                    "a relatively small subset of artists or tracks "
                    "demonstrate broad geographic strength."
                ),
        },
        {
            "Priority":
                2,

            "Key Finding":
                (
                    "Market penetration is the dominant geographic "
                    "strength across the track population."
                ),

            "Evidence":
                (
                    f"{dominant_strongest_dimension_count:,} tracks "
                    f"({dominant_strongest_dimension_share:.2f}%) identify "
                    f"{dominant_strongest_dimension} as their strongest "
                    "geographic dimension."
                ),

            "PMIP Meaning":
                (
                    "Many tracks demonstrate meaningful performance depth "
                    "inside reached markets before developing broad "
                    "international reach."
                ),
        },
        {
            "Priority":
                3,

            "Key Finding":
                (
                    "Geographic diversification is the most common "
                    "weakness across tracks."
                ),

            "Evidence":
                (
                    f"{dominant_weakest_dimension_count:,} tracks "
                    f"({dominant_weakest_dimension_share:.2f}%) identify "
                    f"{dominant_weakest_dimension} as their weakest "
                    "geographic dimension."
                ),

            "PMIP Meaning":
                (
                    "A large part of the catalogue remains dependent on "
                    "a relatively narrow group of observed country markets."
                ),
        },
        {
            "Priority":
                4,

            "Key Finding":
                (
                    "The majority of tracks retain a limited "
                    "geographic footprint."
                ),

            "Evidence":
                (
                    f"{limited_footprint_count:,} tracks "
                    f"({limited_footprint_share:.2f}%) are classified "
                    "with a Limited geographic footprint."
                ),

            "PMIP Meaning":
                (
                    "Broad international distribution is uncommon, "
                    "making geographic breadth a useful differentiating "
                    "signal for discovery and prioritisation."
                ),
        },
        {
            "Priority":
                5,

            "Key Finding":
                (
                    "Strong geographic expansion exists only within "
                    "a relatively small subset of tracks."
                ),

            "Evidence":
                (
                    f"{strong_expansion_count:,} tracks "
                    f"({strong_expansion_share:.2f}%) are classified as "
                    "strong or exceptional geographic expansion."
                ),

            "PMIP Meaning":
                (
                    "Expansion indicators may be especially useful for "
                    "surfacing emerging cross-market opportunities rather "
                    "than describing the majority of tracks."
                ),
        },
        {
            "Priority":
                6,

            "Key Finding":
                (
                    "Observed cross-market momentum is substantially "
                    "more negative than positive."
                ),

            "Evidence":
                (
                    f"{positive_momentum_count:,} tracks "
                    f"({positive_momentum_share:.2f}%) show positive or "
                    "strong positive momentum, while "
                    f"{negative_momentum_count:,} tracks "
                    f"({negative_momentum_share:.2f}%) show negative or "
                    "strong negative momentum."
                ),

            "PMIP Meaning":
                (
                    "Market contraction is common in the historical "
                    "observation window, so momentum should be interpreted "
                    "with temporal confidence rather than as a simple "
                    "growth ranking."
                ),
        },
        {
            "Priority":
                7,

            "Key Finding":
                (
                    "Balanced high-level geographic intelligence is rare."
                ),

            "Evidence":
                (
                    f"The median track Geographic Intelligence Score is "
                    f"{median_track_intelligence_score:.4f}, while only "
                    f"{strong_track_count:,} tracks "
                    f"({strong_track_share:.2f}%) achieve strong or "
                    "exceptional overall geographic intelligence."
                ),

            "PMIP Meaning":
                (
                    "Entities that perform strongly across multiple "
                    "geographic dimensions represent a small and potentially "
                    "high-value analytical subset."
                ),
        },
        {
            "Priority":
                8,

            "Key Finding":
                (
                    "Geographic intelligence should be interpreted "
                    "as a multi-level analytical framework rather than "
                    "a single universal ranking."
                ),

            "Evidence":
                (
                    f"Median country findings index: "
                    f"{median_country_findings_index:.4f}; "
                    f"median artist findings index: "
                    f"{median_artist_findings_index:.4f}; "
                    f"median track Geographic Intelligence Score: "
                    f"{median_track_intelligence_score:.4f}."
                ),

            "PMIP Meaning":
                (
                    "Country, artist and track outputs describe different "
                    "analytical levels and should be combined rather than "
                    "treated as directly interchangeable scores."
                ),
        },
    ]
)


print()
print("Final PMIP geographic intelligence key findings")
print("=" * 100)

display(
    final_geographic_key_findings_df
)


# ------------------------------------------------------------
# 11. Key findings diagnostics
# ------------------------------------------------------------

key_findings_diagnostics_df = pd.DataFrame(
    [
        {
            "Diagnostic":
                "Final key findings prepared",

            "Observed Value":
                int(
                    len(
                        final_geographic_key_findings_df
                    )
                ),
        },
        {
            "Diagnostic":
                "High-priority or strong country markets",

            "Observed Value":
                high_priority_country_count,
        },
        {
            "Diagnostic":
                "Strong or exceptional artist representations",

            "Observed Value":
                strong_artist_count,
        },
        {
            "Diagnostic":
                "Strong or exceptional tracks",

            "Observed Value":
                strong_track_count,
        },
        {
            "Diagnostic":
                "Limited-footprint tracks",

            "Observed Value":
                limited_footprint_count,
        },
        {
            "Diagnostic":
                "Strong-expansion tracks",

            "Observed Value":
                strong_expansion_count,
        },
        {
            "Diagnostic":
                "Positive-momentum tracks",

            "Observed Value":
                positive_momentum_count,
        },
        {
            "Diagnostic":
                "Negative-momentum tracks",

            "Observed Value":
                negative_momentum_count,
        },
    ]
)


print()
print("Final key findings diagnostics")
print("=" * 100)

display(
    key_findings_diagnostics_df
)


# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------

finding_fields_complete = bool(
    final_geographic_key_findings_df[
        [
            "Priority",
            "Key Finding",
            "Evidence",
            "PMIP Meaning",
        ]
    ]
    .notna()
    .all()
    .all()
)


priority_unique = bool(
    final_geographic_key_findings_df[
        "Priority"
    ]
    .is_unique
)


priority_order_valid = bool(
    final_geographic_key_findings_df[
        "Priority"
    ]
    .tolist()
    ==
    list(
        range(
            1,
            len(
                final_geographic_key_findings_df
            )
            + 1
        )
    )
)


finding_count_valid = bool(
    len(
        final_geographic_key_findings_df
    )
    >= 5
)


cross_level_evidence_present = bool(
    (
        high_priority_country_count > 0
    )
    and
    (
        strong_artist_count > 0
    )
    and
    (
        strong_track_count > 0
    )
)


movement_evidence_present = bool(
    (
        strong_expansion_count >= 0
    )
    and
    (
        positive_momentum_count >= 0
    )
    and
    (
        negative_momentum_count >= 0
    )
)


section_14_1_validation_df = pd.DataFrame(
    [
        {
            "Validation Area":
                "Section 13 completion",

            "Requirement":
                "Artifact Saving and Reload Verification "
                "must already be complete.",

            "Observed Evidence":
                f"Section 13 overall completion status: "
                f"{section_13_status}",

            "Passed":
                section_13_status,
        },
        {
            "Validation Area":
                "Key finding completeness",

            "Requirement":
                "Every key finding must contain priority, "
                "finding, evidence and PMIP meaning.",

            "Observed Evidence":
                f"{len(final_geographic_key_findings_df):,} "
                "complete findings prepared",

            "Passed":
                finding_fields_complete,
        },
        {
            "Validation Area":
                "Key finding count validity",

            "Requirement":
                "At least five final geographic findings "
                "must be represented.",

            "Observed Evidence":
                f"{len(final_geographic_key_findings_df):,} "
                "findings prepared",

            "Passed":
                finding_count_valid,
        },
        {
            "Validation Area":
                "Priority uniqueness",

            "Requirement":
                "Each final key finding must receive a "
                "unique priority number.",

            "Observed Evidence":
                (
                    f"{final_geographic_key_findings_df['Priority'].nunique():,} "
                    "unique priority values"
                ),

            "Passed":
                priority_unique,
        },
        {
            "Validation Area":
                "Priority ordering validity",

            "Requirement":
                "Finding priorities must form a continuous "
                "ordered sequence.",

            "Observed Evidence":
                (
                    f"Priority range: "
                    f"{final_geographic_key_findings_df['Priority'].min()} "
                    f"to "
                    f"{final_geographic_key_findings_df['Priority'].max()}"
                ),

            "Passed":
                priority_order_valid,
        },
        {
            "Validation Area":
                "Cross-level evidence presence",

            "Requirement":
                "Country-, artist-, and track-level evidence "
                "must remain represented.",

            "Observed Evidence":
                (
                    f"{high_priority_country_count} strong country markets / "
                    f"{strong_artist_count} strong artist representations / "
                    f"{strong_track_count} strong tracks"
                ),

            "Passed":
                cross_level_evidence_present,
        },
        {
            "Validation Area":
                "Movement evidence presence",

            "Requirement":
                "Expansion and cross-market momentum findings "
                "must remain represented.",

            "Observed Evidence":
                (
                    f"{strong_expansion_count} strong-expansion tracks / "
                    f"{positive_momentum_count} positive-momentum tracks / "
                    f"{negative_momentum_count} negative-momentum tracks"
                ),

            "Passed":
                movement_evidence_present,
        },
    ]
)


print()
print("Final key findings validation")
print("=" * 100)

display(
    section_14_1_validation_df
)


# ------------------------------------------------------------
# 13. Completion status
# ------------------------------------------------------------

section_14_1_complete = bool(
    section_14_1_validation_df[
        "Passed"
    ].all()
)


if not section_14_1_complete:

    failed_checks = (
        section_14_1_validation_df
        .loc[
            ~section_14_1_validation_df[
                "Passed"
            ],
            "Validation Area",
        ]
        .tolist()
    )

    raise AssertionError(
        "Section 14.1 Key Findings validation failed for: "
        + ", ".join(
            failed_checks
        )
    )


print()
print(
    "All Section 14.1 Key Findings "
    "validation checks passed."
)

print(
    f"Section 14.1 completion status: "
    f"{section_14_1_complete}"
)

print(
    "Final PMIP geographic intelligence findings "
    "have been consolidated successfully."
)

print(
    "The workflow is ready for "
    "Section 14.2 Model and Analytical Limitations."
)
```

#### Interpretation

Section 14.1 successfully consolidates the final PMIP geographic intelligence findings into a single validated summary.

A total of **8 final key findings** were prepared.

These findings combine evidence from:

- country-level market intelligence,
- artist-level geographic intelligence,
- track-level geographic intelligence,
- geographic expansion,
- cross-market momentum,
- and the final PMIP scoring framework.

The first major finding is that **country-level geographic strength is considerably more common than strong artist- or track-level geographic intelligence**.

A total of **26 of 76 country markets** are classified as high-priority or strong.

By comparison:

- **989 artist representations** are classified as strong or exceptional,
- and **1,238 tracks** are classified as strong or exceptional geographically.

This suggests that a country can represent a strong market opportunity even when only a relatively small number of artists or tracks demonstrate broad geographic strength within the dataset.

The second major finding is that **market penetration is the dominant geographic strength across tracks**.

A total of **79,825 tracks**, representing approximately **72.44%** of the track population, identify Market Penetration as their strongest geographic dimension.

This indicates that many tracks achieve meaningful performance depth inside the markets they reach before achieving broad international distribution.

The third major finding is that **geographic diversification is the most common weakness across tracks**.

A total of **84,856 tracks**, representing approximately **77.00%** of the track population, identify Geographic Diversification as their weakest geographic dimension.

This means much of the catalogue remains dependent on a relatively narrow group of observed markets.

The fourth major finding reinforces this pattern.

A total of **92,960 tracks**, representing approximately **84.36%** of all analysed tracks, are classified as having a **Limited geographic footprint**.

Broad geographic distribution is therefore uncommon in the dataset.

This makes geographic breadth useful as a distinguishing analytical signal because tracks that achieve genuine multi-market reach represent a comparatively small part of the catalogue.

The fifth major finding concerns geographic expansion.

Only **3,269 tracks**, representing approximately **2.97%** of the track population, achieve strong or exceptional geographic expansion.

This suggests that validated expansion is relatively uncommon and may therefore be particularly useful for identifying tracks showing emerging cross-market development.

The sixth major finding concerns cross-market momentum.

Only **574 tracks**, approximately **0.52%**, show positive or strong positive momentum.

By comparison, **82,521 tracks** show negative or strong negative momentum.

This indicates that contraction is considerably more common than positive expansion within the observed historical window.

However, this should not automatically be interpreted as evidence of commercial decline.

The momentum measures describe observed geographic movement within the dataset and remain dependent on temporal coverage, observation history, and the safeguards introduced in Section 9.5.

The seventh major finding is that **balanced high-level geographic intelligence is rare**.

The median Geographic Intelligence Score remains relatively low, while only a small proportion of tracks achieve strong or exceptional overall geographic intelligence.

This supports the idea that tracks which perform strongly across reach, penetration, diversification, and expansion simultaneously represent an uncommon analytical group.

The eighth finding confirms that PMIP geographic intelligence should not be interpreted as a single universal ranking.

The median indices differ substantially across analytical levels:

- country-level findings are comparatively stronger,
- artist-level findings are more moderate,
- track-level geographic intelligence remains lower overall.

This is expected because country, artist, and track analyses answer different questions.

Country analysis describes market conditions.

Artist analysis describes geographic performance across an artist's observed catalogue representation.

Track analysis describes the geographic behaviour of individual tracks.

These levels should therefore be interpreted together rather than treated as directly interchangeable scores.

The final diagnostics also confirm the presence of:

- **26 strong or high-priority country markets**
- **989 strong or exceptional artist representations**
- **1,238 strong or exceptional tracks**
- **92,960 limited-footprint tracks**
- **3,269 strong-expansion tracks**
- **574 positive-momentum tracks**
- **82,521 negative-momentum tracks**

All Section 14.1 validation checks passed.

The final findings:

- contain complete evidence,
- have unique priorities,
- follow a continuous priority order,
- preserve country-, artist-, and track-level evidence,
- and retain geographic movement evidence.

Section 14.1 completion status is therefore **True**.

The PMIP geographic intelligence workflow now has a consolidated final evidence base ready for **Section 14.2 Model and Analytical Limitations**.

### 14.2 Model and Analytical Limitations

#### Purpose

Summarise the main limitations that affect how PMIP geographic intelligence should be interpreted.

This subsection brings together the constraints identified throughout the workflow, including data coverage, temporal coverage, country comparability, sparse market histories, score construction, geographic inference boundaries, and the absence of causal or fully predictive claims.

#### Interpretation

Section 14.2 successfully consolidates the principal model and analytical limitations affecting the PMIP geographic intelligence framework.

A total of **10 final limitations** were documented.

These limitations ensure that the geographic intelligence outputs are interpreted as structured analytical evidence rather than as absolute measures of artist quality, commercial value, nationality, market causality, or guaranteed future performance.

The first major limitation concerns **unequal country observation coverage**.

The PMIP dataset does not contain equal amounts of evidence for every country market.

Markets with larger observation populations may support more stable geographic estimates than markets represented by considerably fewer observations.

Country rankings and comparisons should therefore always be interpreted alongside their observation coverage and market context.

The second limitation concerns **single-market dominance**.

A total of **81,082 tracks** are represented by single-market evidence, compared with **29,116 multi-market tracks**.

This structure has a major effect on the geographic intelligence results because a track observed in only one market will naturally demonstrate:

- limited international reach,
- limited diversification,
- high geographic concentration,
- and restricted evidence for cross-market comparison.

Low geographic reach must therefore not automatically be interpreted as poor musical or commercial performance.

The third limitation is the **temporal observation boundary**.

Geographic expansion, contraction, new-market entry and cross-market momentum are calculated only from the historical observations available within the PMIP dataset.

These indicators describe what was observed during the available temporal window.

They do not establish what will happen in the future.

The dataset end-date variable was not recovered in the live notebook state during this final limitation cell, so no specific end date should be inferred from this subsection alone.

Temporal interpretations should instead remain linked to the validated temporal metadata and saved PMIP artifacts.

The fourth limitation concerns **cross-country comparability**.

Country markets differ substantially in:

- audience size,
- platform activity,
- chart structure,
- observation volume,
- market maturity,
- and local consumption behaviour.

Two countries receiving similar PMIP scores therefore do not necessarily represent commercially or culturally equivalent opportunities.

Cross-country results should remain context-sensitive.

The fifth limitation concerns the distinction between **observed market activity and verified origin**.

PMIP identifies where artists and tracks have observed streaming or chart evidence.

It does not independently verify:

- artist nationality,
- citizenship,
- birthplace,
- home country,
- or track origin.

A strong result in one country therefore cannot be interpreted as proof that an artist or track originates from that country.

Verified external metadata would be required for such claims.

The sixth limitation is that the framework is **non-causal**.

PMIP identifies geographic patterns, relationships, strengths, weaknesses and movement.

It does not demonstrate that one market caused another outcome.

For example, strong performance in a particular country cannot by itself prove that the country caused:

- international growth,
- artist popularity,
- commercial success,
- or later market expansion.

Geographic results should therefore remain descriptive unless supported by a separate causal research design.

The seventh limitation concerns **composite score construction**.

The final Geographic Intelligence Score combines several different geographic dimensions.

A summary score is useful for prioritisation, but it can hide meaningful differences between:

- geographic reach,
- market penetration,
- diversification,
- and expansion.

For this reason, PMIP users should inspect the component scores rather than relying exclusively on the final composite score.

The eighth limitation concerns **classification threshold sensitivity**.

Labels such as:

- exceptional,
- strong,
- moderate,
- developing,
- limited,
- concentrated,
- and diversified

depend on analytical thresholds defined within the PMIP framework.

Entities close to a classification boundary could change class if those thresholds were modified.

The labels should therefore support structured interpretation and prioritisation rather than be treated as permanent facts.

The ninth limitation is **commercial-context incompleteness**.

Geographic intelligence does not fully represent several factors that may matter to professional music decisions, including:

- revenue,
- profitability,
- touring demand,
- marketing expenditure,
- fan conversion,
- rights ownership,
- contractual conditions,
- strategic fit,
- and long-term artist development potential.

Strong PMIP geographic evidence is therefore not sufficient on its own to justify signing, investment, promotion or budget-allocation decisions.

The tenth limitation is the requirement for **human judgement**.

Automated geographic metrics cannot fully capture:

- cultural context,
- genre-specific behaviour,
- local market knowledge,
- artistic relevance,
- commercial strategy,
- or organisational priorities.

Qualified human interpretation therefore remains necessary before PMIP intelligence is converted into real-world decisions.

- **8 cross-country limitations**

- **7 human-review triggers**

All Section 14.2 validation checks passed.

#### Interpretation

Section 14.3 successfully converts the completed PMIP geographic intelligence workflow into a practical software-integration plan.

A total of **12 integration recommendations** were prepared.

These recommendations explain how the validated geographic intelligence outputs should move from the notebook environment into the wider PMIP platform.

The first recommendation is to create a **persistent geographic intelligence service**.

The validated outputs should be loaded from:

`models/geographic_intelligence`

rather than being read directly from live notebook variables.

This creates a clear separation between analytical processing and the application layer.

The PMIP website, dashboards, APIs, and future services can therefore access the same validated intelligence without depending on the notebook runtime.

The second recommendation is to provide an **artist geographic intelligence endpoint**.

Artist-level responses should expose geographic reach, market penetration, concentration, local-versus-international behaviour, and geographic profile information.

This allows A&R users to examine not only whether an artist performs strongly, but also how that performance is distributed geographically.

The third recommendation is to create a **track geographic intelligence endpoint**.

Track-level responses should preserve the individual component measures behind the final Geographic Intelligence Score.

These should include:

- Geographic Reach Score
- Market Penetration Score
- Geographic Concentration Score
- Geographic Expansion Indicator
- strongest geographic dimension
- weakest geographic dimension
- geographic performance profile

This prevents the final score from hiding meaningful differences between breadth, depth, diversification, and movement.

The fourth recommendation is to create a **country opportunity view**.

Country intelligence should present market strength together with contextual information such as:

- country findings index,
- country classification,
- streaming strength,
- chart strength,
- market context,
- and observation-coverage class.

This ensures that country prioritisation remains evidence-based and context-aware.

The fifth recommendation is to introduce **geographic opportunity filters**.

PMIP users should be able to search for artists or tracks using geographic criteria such as:

- broad reach,
- strong penetration,
- strong diversification,
- geographic expansion,
- emerging international performance,
- and limited geographic footprint.

This converts geographic intelligence from a static analytical output into an interactive discovery feature.

The sixth recommendation concerns the **high-priority insight workflow**.

High-priority geographic signals should be surfaced as candidates for human review rather than automatic recommendations.

Each high-priority result should retain:

- supporting evidence,
- component scores,
- limitations,
- and review status.

This preserves the responsible-use boundary established throughout the geographic intelligence workflow.

The seventh recommendation is to build a **responsible-use presentation layer** directly into the PMIP interface.

Coverage warnings, temporal confidence, cross-country comparison limitations, and human-review guidance should be visible beside the relevant analytical outputs.

These safeguards should not be hidden only in technical documentation.

Making uncertainty visible reduces the risk that descriptive PMIP intelligence is mistaken for causal, predictive, or definitive commercial evidence.

The eighth recommendation is to treat **artifact versioning and schema validation** as part of the production deployment process.

The PMIP application should verify:

- artifact availability,
- manifest consistency,
- expected schema,
- configuration version,
- and artifact counts

before loading a new geographic intelligence release.

This prevents incompatible or incomplete analytical outputs from silently entering the application.

The ninth recommendation is to establish a **scheduled analytical refresh process**.

The current saved artifacts should not be treated as permanently current.

When validated source datasets are updated, the complete geographic intelligence workflow should be rerun so that:

- features,
- scoring,
- movement analysis,
- findings,
- limitations,
- metadata,
- and artifact validation

are regenerated together.

This preserves both freshness and reproducibility.

The tenth recommendation is to implement **monitoring and data-quality checks**.

PMIP should detect:

- missing artifacts,
- reload failures,
- schema changes,
- unusual population differences,
- and major score-distribution shifts.

Updated analytical outputs should only be promoted into the application after these checks pass.

The eleventh recommendation is to preserve **explainable geographic intelligence**.

Every high-level score should be accompanied by evidence explaining why the result was produced.

This is particularly important for non-technical PMIP users.

Instead of showing only a Geographic Intelligence Score, the interface should also show which geographic dimensions are strong or weak.

The twelfth recommendation is to integrate geographic intelligence with the **wider PMIP intelligence framework**.

Geographic intelligence can later be used alongside other validated PMIP capabilities such as:

- artist momentum scoring,
- artist performance forecasting,
- streaming anomaly detection,
- and other future analytical modules.

However, the individual signals should remain separately inspectable.

PMIP should avoid collapsing every model and analytical output into one opaque recommendation.

3. **Geographic Intelligence Service**

5. **Frontend / Dashboard**

6. **Monitoring / Validation**
